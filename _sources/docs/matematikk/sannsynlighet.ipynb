# 3. Sannsynlighet

Sannsynlighet handler om å prøve å forutsi. Vi vet ikke hva en terning viser før den lander, eller hvilke molekyler som reagerer først i et system - men vi kan si noe presist om hvor ofte de ulike utfallene inntreffer i det lange løp. Det er verdt å legge merke til hva slags kunnskap dette er. Vi gir ikke opp og sier at noe er tilfeldig og dermed uten struktur. Tvert imot: tilfeldighet har sine egne lover, og de er like eksakte som alt annet i matematikken. Vi begynner derfor helt fra bunnen, med å beskrive nøyaktig hva som kan skje, før vi går videre til hvor sannsynlig hvert utfall er.

For en kjemiker er dette mer enn terningkast og kortstokker. Et gram vann inneholder rundt $10^{22}$ molekyler, og vi kan umulig følge hvert enkelt av dem. Vi kan heller ikke forutsi når et bestemt radioaktivt atom henfaller, eller nøyaktig hvilken energi ett molekyl har i neste øyeblikk.  Løsningen er å beskrive systemet med sannsynligheter i stedet. Det er nettopp dette som gjør statistisk termodynamikk mulig: vi gir opp å følge enkeltmolekyler, og regner i stedet på fordelinger. Underveis skal vi se at begreper du møter i kjemien - mikrotilstander, makrotilstander og Boltzmann-fordelingen - er sannsynlighetsteori i forkledning.

## 3.1 Utfall, utfallsrom og hendelser

Du triller en terning, og teller antall øyne terningen viser. Dette er et av mange eksempler på et statistisk eksperiment. For å kunne snakke om sannsynlighet og eksperimenter trenger vi noen definisjoner. Vi starter med de mest grunnleggende: **utfall**, **utfallsrom** og **hendelser**.

:::{admonition} Def. 3.1 - Utfall
:class: warning

Et **utfall** $i$ er et mulig resultat i et eksperiment. 
:::


:::{admonition} Def. 3.2 - Utfallsrom
:class: warning

**Utfallsrommet** $S$ er mengden av alle mulige utfall i et eksperiment.
:::


:::{admonition} Def. 3.3 - Hendelse
:class: warning

En **hendelse** $A$ er delmengde av utfallsrommet, altså ett eller flere utfall vi er interessert i.

Merk at hele $S$ er en hendelse (den som alltid skjer), og den tomme mengden $\{\}$ også er en hendelse (den som aldri skjer).
:::

:::{admonition} Husk!
:class: error
Mengder og delmengder er definert i delkapittel [1.2 Mengdelære](https://andreasdh.github.io/Bifrost/docs/matematikk/formler_og_enheter.html#mengdelaere).
:::

:::{admonition} OBS
:class: error
Legg merke til at en hendelse alltid må være en delmengde av utfallsrommet! Tenk deg at du trekker ett kort fra en kortstokk, og kun ser på symbolet til kortet. Utfallsrommet er da $S=\{\text{kløver},\text{ruter},\text{spar},\text{hjerter}\}$. Nå er ikke "trekke et bildekort (dame, knekt, konge eller ess)" en hendelse - den informasjonen finnes rett og slett ikke i utfallsrommet vårt. 
:::

Før vi går videre trenger vi å si noe om hva **sannsynligheten til et utfall** faktisk betyr. 

:::{admonition} Def. 3.4 - Den klassiske sannsynlighetsdefinisjonen
:class: warning

I et eksperiment der alle utfall er **like sannsynlige**, som et kast med en likevektet terning, er sannsynligheten for hendelse $A$ gitt ved

$P(A) = \frac{\text{antall utfall i } A}{\text{antall utfall i } S}$
:::

Def. 3.4 krever at alle utfall er like sannsynlige, men slik er det sjelden i virkeligheten. I [3.4 Aksiomene i sannsynlighet](https://andreasdh.github.io/Bifrost/docs/matematikk/sannsynlighet.html#aksiomene-i-sannsynlighet) generaliserer vi slik at vi også kan jobbe med slike sannsynligheter.

:::{admonition} Kobling til fysikalsk kjemi
:class: important

I _Grunnleggende begreper_ defineres [mikrotilstander og makrotilstander](https://andreasdh.github.io/Bifrost/docs/1_introduksjon.html#mikrotilstander-og-makrotilstander). Mikrotilstander kan sammenlignes med utfall, altså én bestemt fordeling av energi/posisjon over alle partiklene i systemet. Utfallsrommet $S$ blir da mengden av *alle* mikrotilstander systemet kan befinne seg i, og en hendelse $A$ tilsvarer en **makrotilstand** (f.eks. en gitt total energi, temperatur eller trykk) - altså alle mikrotilstander som gir samme makroskopiske observasjon.
:::

:::{card} 
**Eksempel**
^^^

Du triller en sekssidet terning.

a) Hva er utfallsrommet $S$ til eksperimentet?

Du kan enten få $1,2,3,4,5$ eller $6$, som betyr at utfallsrommet er $S=\{1,2,3,4,5,6\}$.

b) Hva er et eksempel på en hendelse i dette eksperimentet?

Et eksempel på en hendelse i dette eksperimentet er å trille et partall. La oss betegne denne hendelsen ved $A$, og da har vi at $A=\{2,4,6\}$. Et annet eksempel er å trille $1,2,3,4,5$ eller $6$, altså $S$, og denne hendelsen er garantert.

c) Hva er sannsynligheten for å trille partall?

Hendelsen er $A=\{2,4,6\}$, som inneholder $3$ utfall. Utfallsrommet $S=\{1,2,3,4,5,6\}$ inneholder $6$ utfall. Siden terningen er likevektet, er alle utfall like sannsynlige, og vi kan telle:

$P(A) = \frac{\text{antall utfall i } A}{\text{antall utfall i } S} = \frac{3}{6} = \frac{1}{2}$

Sannsynligheten for å trille et partall er altså $\frac{1}{2}$.
:::

## 3.2 Komplement, snitt og union

La oss nå se på hva som skjer utenfor en hendelse $A$, og hvordan vi kan kombinere to hendelser $A$ og $B$ til nye hendelser. Vi ser først på **komplementet** til en hendelse, og deretter **snitt** og **union**.

:::{admonition} Def. 3.5 - Komplement
:class: warning

**Komplementet **til en hendelse $A$, $A'$, er mengden av alle utfall i $S$ som ikke er i $A$.
:::


:::{admonition} Def. 3.6 - Snitt
:class: warning

**Snittet** av to hendelser $A$ og $B$, $A \cap B$, er hendelsen som består av alle utfall som både er i $A$ og i $B$.
:::


:::{admonition} Def. 3.7 - Union
:class: warning

**Unionen** av to hendelser $A$ og $B$, $A \cup B$, er hendelsen som består av alle utfall som er i $A$ eller i $B$.
:::

a) Hva er komplementet til $A$?

Komplementet til $A$ er alle utfallene som er i $S$ men ikke $A$, som tilsvarer å trille 3 eller høyere. Altså er $A'=\{3,4,5,6\}$. 

b) Hva er snittet av $A$ og $B$?

Snittet mellom $A$ og $B$ blir den tomme mengden $\{\}$, nettopp fordi det er ingen utfall som er i både $A$ og $B$.

c) Hva er unionen av $A$ og $B$?
 
Unionen mellom $A$ og $B$ blir $A \cup B = \{1,2,5,6\}$, altså mengden alle utfallene i $A$ og $B$ samlet.

## 3.2 Disjunkte og uavhengige hendelser

Med de ulike operasjonene på plass kan vi beskrive to viktige egenskaper et par hendelser kan ha. Den første, **disjunkthet**, handler bare om utfallene: hendelsene overlapper ikke. Den andre, **uavhengighet**, handler om sannsynlighet: den ene hendelsen sier ingenting om den andre. Det er lett å blande dem sammen, men de betyr noe helt forskjellig - vi skal se at uavhengighet blir svært sentralt i statistisk termodynamikk.

:::{admonition} Def. 3.8 - Disjunkte hendelser
:class: warning

To hendelser $A$ og $B$ kalles **disjunkte** dersom de ikke har noen felles utfall. Matematisk, $A \cap B = \{\}$.
:::

:::{admonition} Def. 3.9 - Uavhengige hendelser
:class: warning

To hendelser $A$ og $B$ kalles **uavhengige** dersom

$P(A \cap B) = P(A) \cdot P(B)$
:::

Forskjellen er verdt å dvele ved. Disjunkte hendelser kan _aldri_ skje samtidig - triller du en terning, kan du ikke få både $1$ og $6$ i samme kast. Uavhengige hendelser kan derimot godt skje samtidig; poenget er at den ene ikke påvirker sannsynligheten for den andre.

:::{card} 
**Eksempel**
^^^
Du triller en terning to ganger etter hverandre og noterer resultatet for hvert kast. La $A$ være hendelsen "første terning viser $6$", og $B$ hendelsen "andre terning viser $6$". Er $A$ og $B$ uavhengige?

Hver terning har $\frac{1}{6}$ sannsynlighet for å vise $6$, så $P(A)=P(B)=\frac{1}{6}$.

For at begge skal vise $6$, må vi få utfallet $(6,6)$. Til sammen finnes det $6 \cdot 6 = 36$ mulige utfall for to terninger, og bare ett av dem er $(6,6)$. Altså er $P(A \cap B)=\frac{1}{36}$.

Vi sjekker kravet fra Def. 3.8:

$P(A) \cdot P(B) = \frac{1}{6} \cdot \frac{1}{6} = \frac{1}{36} = P(A \cap B)$

Kravet er oppfylt, så hendelsene er uavhengige - som forventet, siden den første terningen umulig kan påvirke den andre.
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Uavhengighet er sentralt når vi skal beskrive et molekyls bevegelse: translasjon, rotasjon og vibrasjon regnes vanligvis som uavhengige av hverandre. Det er nettopp derfor vi senere kan bruke $P(A \cap B)=P(A)\cdot P(B)$ til å si at [partisjonsfunksjonen](https://andreasdh.github.io/Bifrost/docs/3_partisjonsfunksjonen.html#partisjonsfunksjonen) faktoriserer, $Q_{tot}=Q_{trans}\cdot Q_{rot}\cdot Q_{vib}$.
:::

:class: tip

a) Er $A$ og $B$ disjunkte? Begrunn.

b) Er $A$ og $C$ disjunkte? Begrunn.

c) Undersøk om $A$ og $C$ er uavhengige.
:::

:::{admonition} Løsning 3.4
:class: tip, dropdown

a) $A \cap B = \{\}$, siden ingen utfall er med i begge. Hendelsene er derfor disjunkte.

b) $A \cap C = \{2\}$, som ikke er tom. Hendelsene er altså **ikke** disjunkte.

c) Vi regner ut hver sannsynlighet ved opptelling. $A$ har $2$ av $6$ utfall, så $P(A)=\frac{2}{6}=\frac{1}{3}$. $C$ har $3$ av $6$ utfall, så $P(C)=\frac{3}{6}=\frac{1}{2}$. Snittet $A \cap C = \{2\}$ har ett utfall, så $P(A \cap C)=\frac{1}{6}$.

Vi sjekker kravet:

$P(A) \cdot P(C) = \frac{1}{3} \cdot \frac{1}{2} = \frac{1}{6} = P(A \cap C)$

Kravet er oppfylt, så $A$ og $C$ er uavhengige. Legg merke til at dette skjer selv om begge hendelsene gjelder *samme* terningkast - uavhengighet krever ikke separate eksperimenter, men sier heller at sannsynligheten for den ene hendelsen ikke påvirker sannsynligheten for den andre.
:::

## 3.4 Aksiomene i sannsynlighet

Enten du spiller sjakk eller ikke, vet du at spillet har visse regler, som at hver brikke kan flyttes på en spesiell måte, at hvit starter osv. Disse reglene er ikke noe som er _bevist_, men de er fastsatt som et utgangspunkt for spillet. Alt annet du kan si om sjakk, for eksempel hvordan man vinner, taper eller om en åpning er god eller dårlig, er utledet fra disse reglene. 

Sånn er også matematikken! Som reglene i sjakk, har matematikken grunnleggende byggesteiner som kalles **aksiomer**. Teoremer og andre utledninger er bygget logisk fra aksiomene. Kolmogorov presenterte aksiomene i sannsynlighetsteori, og til tross for at disse kan virke åpenbare, er de utgangspunktet til alt vi skal se på - også flere utledninger i statistisk termodynamikk.

:::{admonition} Aksiomer 3.1 - Kolmogorovs sannsynlighetsaksiomer
:class: warning

For et eksperiment med utfallsrom $S$ har vi


**Ikke-negativitet**

Sannsynligheten for enhver hendelse $A$ er ikke-negativ

1. $P(A) \ge 0$ 

**Normering**

Sannsynligheten for $S$ er lik $1$, altså

2. $P(S)=1$

**Additivitet**

Sannsynligheten for unionen av flere disjunkte hendelser $A_1, A_2,\dots$ er summen av sannsynligheten for hver av hendelsene

3. $P(A_1 \cup A_2 \cup \dots)=P(A_1)+P(A_2)+\dots$
:::

Legg merke til at aksiomene ikke sier noe om *hvor stor* sannsynligheten for et bestemt utfall er - de sier bare hvilke regler sannsynlighetene må følge. Det er nettopp derfor de også gjelder for vektede terninger og for mikrotilstander i et fysisk system, ikke bare for rettferdige spill. Et viktig følge av aksiom 2 og 3 er følgende:

:::{admonition} Teorem 3.1 - Summen av alle utfalls sannsynligheter
:class: note

For et vilkårlig eksperiment med utfallsrom $S$ har vi

$\sum_{i}P(i)=1$

for alle utfall $i \in S$.
:::

Utledningen er kort: de enkelte utfallene er disjunkte hendelser - du kan ikke få både $3$ og $5$ i samme terningkast - og til sammen utgjør de hele utfallsrommet. Ved aksiom 3 kan vi da summere sannsynlighetene deres, og ved aksiom 2 må summen bli $P(S)=1$.

:::{admonition} Teorem 3.2 - Komplementregelen
:class: note

For enhver hendelse $A$ gjelder

$P(A') = 1 - P(A)$
:::

Også denne følger av aksiomene. En hendelse $A$ og komplementet $A'$ er disjunkte etter Def. 1.4, og sammen dekker de hele $S$. Aksiom 3 gir da $P(A)+P(A')=P(S)$, og aksiom 2 sier at $P(S)=1$. Flytter vi $P(A)$ over, får vi resultatet.

Regelen er ofte det raskeste verktøyet i praksis: skal du finne sannsynligheten for "minst én", er det som regel lettere å regne ut sannsynligheten for "ingen" og trekke fra $1$.

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Aksiomene over gjelder for *ethvert* utfallsrom, også for mikrotilstandene til et termodynamisk system. Sannsynligheten for at et system i termisk likevekt ved temperatur $T$ befinner seg i mikrotilstand $i$ med energi $E_i$ er gitt ved [Boltzmann-fordelingen](https://andreasdh.github.io/Bifrost/docs/2_entropi_boltzmann.html#fra-entropimaksimum-til-boltzmann-fordelingen)

$P(i) = \dfrac{e^{-E_i/k_BT}}{Q}$

der $k_B$ er Boltzmanns konstant og $Q=\sum_i e^{-E_i/k_BT}$ er [partisjonsfunksjonen](https://andreasdh.github.io/Bifrost/docs/3_partisjonsfunksjonen.html#partisjonsfunksjonen). Merk at $Q$ er akkurat den normeringskonstanten som gjør at $\sum_i P(i)=1$, i tråd med Teorem 2.1 over.
:::

:::{card} 
**Eksempel**
^^^
Et lite system med 2 partikler som hver kan ha energi $0$ eller $\varepsilon$ har mikrotilstandene $S=\{(0,0),(0,\varepsilon),(\varepsilon,0),(\varepsilon,\varepsilon)\}$ - helt analogt til å kaste to mynter. Makrotilstanden "total energi $=\varepsilon$" er hendelsen $A=\{(0,\varepsilon),(\varepsilon,0)\}$. Er alle mikrotilstander like sannsynlige (høy temperatur), blir $P(A)=2/4=1/2$, akkurat som å få én kron og ett mynt.
:::

$P(\text{ingen seksere}) = \frac{5}{6} \cdot \frac{5}{6} = \frac{25}{36}$

$P(\text{minst én sekser}) = 1 - \frac{25}{36} = \frac{11}{36}$

## 3.5 Betinget sannsynlighet 

Så langt har vi sett på sannsynligheten for hendelser hver for seg. I mange situasjoner sitter vi imidlertid med *delvis informasjon: vi vet allerede at noe har skjedd, og spør hva det gjør med sannsynligheten for noe annet. Trekker du et kort og får vite at det er rødt, endrer det sannsynligheten for at det er hjerter. Slike spørsmål besvares av betinget sannsynlighet.

:::{admonition} Def. 3.10 - Betinget sannsynlighet
:class: warning

Sannsynligheten for at hendelse $A$ skjer, gitt at hendelse $B$ allerede har skjedd, er

$P(A|B) = \dfrac{P(A \cap B)}{P(B)}$

forutsatt at $P(B)>0$.
:::

Formelen gir mening om du tenker på den som en innsnevring av utfallsrommet. Når vi vet at $B$ har skjedd, er alle utfall utenfor $B$ ikke lenger aktuelle - $B$ har i praksis blitt vårt nye utfallsrom. Telleren $P(A \cap B)$ er den delen av $A$ som fortsatt er mulig, og vi deler på $P(B)$ for å normere, slik at det nye utfallsrommet igjen har total sannsynlighet $1$, i tråd med aksiom 2.

Legg merke til sammenhengen med uavhengighet fra Def. 3.8. Er $A$ og $B$ uavhengige, er $P(A \cap B)=P(A)\cdot P(B)$, og da blir

$P(A|B) = \frac{P(A)\cdot P(B)}{P(B)} = P(A)$

Å vite at $B$ har skjedd endrer altså ingenting - som er nøyaktig det uavhengighet betyr.

:::{card} 
**Eksempel**
^^^
Du trekker ett kort fra en vanlig kortstokk. La $A$ være hendelsen "kortet er hjerter" og $B$ hendelsen "kortet er rødt". Hva er $P(A|B)$?

Det er $13$ hjerterkort, så $P(A)=\frac{13}{52}=\frac{1}{4}$. De røde kortene er hjerter og ruter, til sammen $26$ kort, så $P(B)=\frac{26}{52}=\frac{1}{2}$.

Alle hjerterkort er røde, så snittet $A \cap B$ er rett og slett alle hjerterkort: $P(A \cap B)=\frac{13}{52}=\frac{1}{4}$.

Ved Def. 3.9 får vi

$P(A|B)=\frac{P(A \cap B)}{P(B)}=\frac{1/4}{1/2}=\frac{1}{2}$

Vet vi at kortet er rødt, er sannsynligheten for hjerter altså $\frac{1}{2}$ - opp fra $\frac{1}{4}$ uten den informasjonen. Det stemmer med intuisjonen: blant de $26$ røde kortene er $13$ hjerter.
:::

Ganger vi opp med $P(B)$ på begge sider av definisjonen, får vi en nyttig omskriving som lar oss regne ut sannsynligheten for at *begge* hendelsene skjer.

:::{admonition} Teorem 3.3 - Produktregelen
:class: note

For to hendelser $A$ og $B$ gjelder

$P(A \cap B) = P(A|B) \cdot P(B)$

Dersom $A$ og $B$ er uavhengige, forenkles dette til $P(A \cap B)=P(A)\cdot P(B)$, som i Def. 3.8.
:::

Snittet $A \cap B$ er det samme som $B \cap A$, så vi kan skrive produktregelen på to måter: $P(A|B)P(B)$ og $P(B|A)P(A)$. Setter vi disse lik hverandre og løser for $P(A|B)$, får vi et av de mest brukte resultatene i all sannsynlighetsregning.

:::{admonition} Teorem 3.4 - Bayes' teorem
:class: note

For to hendelser $A$ og $B$ med $P(B)>0$ gjelder

$P(A|B) = \dfrac{P(B|A) \, P(A)}{P(B)}$
:::

Bayes' teorem lar oss snu en betingelse. Ofte kjenner vi $P(B|A)$, men ønsker egentlig $P(A|B)$. Et typisk eksempel er medisinsk testing: vi vet hvor ofte en test slår ut når pasienten faktisk er syk, $P(\text{positiv}|\text{syk})$, men det pasienten lurer på er det motsatte - hvor sannsynlig det er at hun er syk gitt at testen var positiv, $P(\text{syk}|\text{positiv})$. Disse to er ikke det samme, og forskjellen kan være dramatisk når sykdommen er sjelden.

:::{card} 
**Eksempel**
^^^
En sjelden sykdom rammer $1$ av $1000$ personer. En test gir positivt utslag hos $99\%$ av de syke, men gir også falskt positivt hos $5\%$ av de friske. Du tester positivt - hva er sannsynligheten for at du faktisk er syk?

La $A$ være "syk" og $B$ "positiv test". Vi kjenner

$P(A)=0{,}001, \qquad P(B|A)=0{,}99, \qquad P(B|A')=0{,}05$

Først finner vi $P(B)$, altså hvor ofte testen slår ut i det hele tatt. En positiv test kan komme fra en syk eller en frisk person:

$P(B)=P(B|A)P(A)+P(B|A')P(A')=0{,}99 \cdot 0{,}001 + 0{,}05 \cdot 0{,}999 = 0{,}00099+0{,}04995=0{,}05094$

Ved Bayes' teorem er

$P(A|B)=\frac{P(B|A)P(A)}{P(B)}=\frac{0{,}99 \cdot 0{,}001}{0{,}05094} \approx 0{,}019$

Sannsynligheten for å være syk er altså bare rundt $2\%$, til tross for en test som er $99\%$ treffsikker på syke. Grunnen er at sykdommen er så sjelden at de falske positive blant de mange friske langt overgår de ekte positive.
:::

Til slutt tar vi med en regel for unioner. Aksiom 3 lot oss summere sannsynligheter for disjunkte hendelser, men hva om hendelsene overlapper?

:::{admonition} Teorem 3.5 - Additiv lov
:class: note

For to hendelser $A$ og $B$ som ikke nødvendigvis er disjunkte, gjelder

$P(A \cup B) = P(A) + P(B) - P(A \cap B)$

Dersom $A$ og $B$ er disjunkte er $P(A \cap B)=0$, og uttrykket blir $P(A \cup B)=P(A)+P(B)$, som i aksiom 3.
:::

Grunnen til at vi trekker fra $P(A \cap B)$ er at utfallene i snittet ellers ville blitt talt to ganger - én gang i $P(A)$ og én gang i $P(B)$. Se gjerne tilbake på Venn-diagrammet i [3.2 Komplement, snitt og union](https://andreasdh.github.io/Bifrost/docs/matematikk/sannsynlighet.html#komplement-snitt-og-union): overlappen skal bare telles én gang.

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Betinget sannsynlighet dukker opp så snart vi har delvis informasjon om et system. Vet vi at et molekyl befinner seg i en bestemt elektronisk tilstand, endrer det sannsynlighetsfordelingen over vibrasjonsnivåene - vi regner da på $P(\text{vibrasjonsnivå} \mid \text{elektronisk tilstand})$.

Produktregelen i Teorem 3.3 er også grunnen til at [Boltzmann-fordelingen](https://andreasdh.github.io/Bifrost/docs/2_entropi_boltzmann.html#fra-entropimaksimum-til-boltzmann-fordelingen) kan faktoriseres når frihetsgradene er uavhengige. Er translasjon, rotasjon og vibrasjon uavhengige, er den betingede sannsynligheten lik den ubetingede, og sannsynlighetene kan ganges sammen direkte - noe som gir $Q_{tot}=Q_{trans}\cdot Q_{rot}\cdot Q_{vib}$ fra delkapittel 3.3.
:::

Du triller én sekssidet terning. La $A=\{2,4,6\}$ være hendelsen "partall" og $B=\{4,5,6\}$ hendelsen "fire eller høyere".

a) Finn $P(A)$, $P(B)$ og $P(A \cap B)$.

b) Finn $P(A|B)$, og forklar svaret med ord.

c) Er $A$ og $B$ uavhengige?

d) Bruk additiv lov til å finne $P(A \cup B)$.

:::{admonition} Løsning 3.6
:class: tip, dropdown

a) $A$ har $3$ av $6$ utfall, så $P(A)=\frac{1}{2}$. $B$ har $3$ av $6$ utfall, så $P(B)=\frac{1}{2}$. Snittet er $A \cap B=\{4,6\}$ med $2$ utfall, så $P(A \cap B)=\frac{2}{6}=\frac{1}{3}$.

b) Ved Def. 2.9:

$P(A|B)=\frac{P(A \cap B)}{P(B)}=\frac{1/3}{1/2}=\frac{2}{3}$

Med ord: vet vi at terningen viser $4$, $5$ eller $6$, er to av disse tre utfallene partall. Sannsynligheten for partall stiger derfor fra $\frac{1}{2}$ til $\frac{2}{3}$.

c) Vi sjekker kravet:

$P(A)\cdot P(B)=\frac{1}{2}\cdot\frac{1}{2}=\frac{1}{4} \neq \frac{1}{3}=P(A \cap B)$

Kravet er ikke oppfylt, så $A$ og $B$ er **ikke** uavhengige. Det stemmer med b): informasjon om $B$ endret sannsynligheten for $A$.

d) Ved Teorem 3.5:

$P(A \cup B)=P(A)+P(B)-P(A \cap B)=\frac{1}{2}+\frac{1}{2}-\frac{1}{3}=\frac{2}{3}$

Vi kan sjekke ved opptelling: $A \cup B=\{2,4,5,6\}$, altså $4$ av $6$ utfall, som gir $\frac{4}{6}=\frac{2}{3}$.

## 3.6 Kontinuerlige utfallsrom

Så langt har alle utfallsrommene våre vært **diskrete**: terningkast, kortstokk, myntkast. Felles for dem er at vi kan liste opp utfallene ett etter ett, og at hvert utfall har sin egen sannsynlighet. I fysikalsk kjemi er mange størrelser derimot **kontinuerlige**. Farten til et gassmolekyl kan være hvilken som helst verdi over null, og energien til en partikkel kan ligge hvor som helst i et intervall. Slike utfallsrom kan vi ikke telle oss gjennom.

:::{admonition} Def. 3.11 - Kontinuerlig utfallsrom
:class: warning

Et utfallsrom $S$ kalles **kontinuerlig** dersom utfallene utgjør et sammenhengende intervall av tall, slik at de ikke kan listes opp enkeltvis.
:::

Overgangen til kontinuerlige utfallsrom har en konsekvens som er verdt å stoppe ved. I et diskret utfallsrom hadde hvert enkelt utfall en positiv sannsynlighet - hver side på terningen hadde $\frac{1}{6}$. I et kontinuerlig utfallsrom er det umulig. Det finnes uendelig mange verdier i et intervall, og skulle hver av dem hatt en positiv sannsynlighet, ville summen sprengt $1$ og brutt aksiom 2.

Løsningen er at hvert **enkeltutfall** har sannsynlighet $0$. Sannsynligheten for at et molekyl har farten nøyaktig $400$ m/s, uten noe avvik overhodet, er null. Dette betyr ikke at utfallet er umulig - molekylet har tross alt en fart - men at det er én verdi blant uendelig mange.

::::{grid}
:gutter: 2

:::{grid-item-card} Diskret

_Utfallene kan listes én og én_

* Terningkast
* Trekke kort fra en kortstokk
* Elektroniske energinivåer

:::

:::{grid-item-card} Kontinuerlig

_Utfallene utgjør et sammenhengdene intervall av tall_

* Høyden til en tilfeldig person
* Hvor på en dartskive pilen treffer
* Tiden en reaksjon tar

:::

::::

Hendelser fungerer fortsatt som før: de er delmengder av $S$, helt i tråd med Def. 3.3. Forskjellen er at de meningsfulle hendelsene nå er **intervaller** i stedet for enkeltutfall. Vi spør ikke "hva er sannsynligheten for at farten er nøyaktig $400$ m/s?", men "hva er sannsynligheten for at farten ligger mellom $400$ og $500$ m/s?".

En følge av at enkeltutfall har sannsynlighet $0$ er at endepunktene ikke spiller noen rolle. Hendelsene $400 \leq v \leq 500$ og $400 < v < 500$ har nøyaktig samme sannsynlighet, siden de to endepunktene hver bidrar med $0$.

:::{card} 
**Eksempel**
^^^
Du snurrer en viser på en sirkulær skive, og noterer vinkelen $\theta$ visesen stopper på, målt i grader.

a) Hva er utfallsrommet?

Viseren kan stoppe på hvilken som helst vinkel fra $0$ til $360$ grader, så utfallsrommet er det kontinuerlige intervallet $S=[0,360)$.

b) Hva er sannsynligheten for at viseren stopper på nøyaktig $90$ grader?

Siden $S$ er kontinuerlig, har hvert enkeltutfall sannsynlighet $0$. Altså er $P(\theta = 90)=0$. Det finnes uendelig mange vinkler viseren kan stoppe på, og nøyaktig $90{,}000\ldots$ grader er bare én av dem.

c) Hva er sannsynligheten for at viseren stopper i første kvadrant, altså mellom $0$ og $90$ grader?

Dette er en hendelse på intervallform, $A=[0,90]$. Er skiven jevnt balansert, er ingen del av sirkelen favorisert, og sannsynligheten svarer til hvor stor andel av sirkelen intervallet dekker:

$P(A)=\frac{90}{360}=\frac{1}{4}$
:::

Legg merke til hva vi gjorde i c). I det diskrete tilfellet fant vi sannsynlighet ved å telle utfall og dele på antallet i $S$. Her kunne vi ikke telle, men vi kunne måle: vi sammenliknet **lengden** av intervallet med lengden av hele utfallsrommet. Så lenge ingen del av utfallsrommet er favorisert, er dette den kontinuerlige parallellen til opptelling.

De fleste kontinuerlige størrelser i kjemien er imidlertid *ikke* jevnt fordelt. Molekylfarter i en gass er tettere samlet rundt noen verdier enn andre, og da holder det ikke å sammenlikne lengder. For å håndtere slike tilfeller trenger vi et nytt verktøy, som vi kommer tilbake til.

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Skillet mellom diskrete og kontinuerlige utfallsrom går rett inn i kvantefysikkens kjerne. Energinivåene til et elektron i et atom er **diskrete** - elektronet kan bare ha bestemte energier, og utfallsrommet kan listes opp. Det er nettopp derfor vi ser skarpe linjer i et atomspektrum.

Translasjonsenergien til et molekyl i en gassbeholder er derimot i praksis **kontinuerlig**. Nivåene er strengt tatt kvantiserte også her, men de ligger så tett at vi behandler dem som et sammenhengende intervall. Denne tilnærmingen er helt sentral i statistisk termodynamikk, og er grunnen til at vi senere kan erstatte summer med integraler når vi regner på translasjonsbidraget til partisjonsfunksjonen.
:::

En buss kommer til holdeplassen på et tilfeldig tidspunkt mellom klokken $10{:}00$ og $10{:}30$, uten at noen tidspunkter er mer sannsynlige enn andre.

a) Skriv opp utfallsrommet, målt i minutter etter $10{:}00$.
