Denne modulen er ikke pensum, men heller et oppslagsverk for matematikken som brukes i de andre kapitlene. Bruk den som en støtte hvis du trenger det.

Delkapitlene innenfor hvert hovedtema presenteres i en logisk rekkefølge, slik at man kan gå videre fra et delkapittel til neste dersom stoffet virker kjent.

Boksene i denne modulen representerer følgende:

:::{admonition} Definisjoner og Aksiomer
:class: warning
Definisjoner er eksakte utsagn som forteller oss hva et matematisk ord eller symbol betyr.

Aksiomer (eller postulater) er grunnleggende regler eller startantakelser som er sanne uten behov for bevis. 
:::

:::{admonition} Teoremer
:class: note
Teoremer er matematiske utsagn som har blitt bevist å være sanne. Disse er utledet fra definisjoner og aksiomer. 
:::

:::{admonition} Koblinger til fysikalsk kjemi 
:class: important
Eksempler på hvordan matematikken ligger til grunn for fysikalsk kjemi. 
:::
