## Måledata og usikkerhet (KJM1101, KJM1130, KJM2400, IN-KJM1900)

Alle målinger vi gjør i laboratoriet, har en usikkerhet og en feil. Vi gjør flere grep i design av forsøk for å minimere denne, men den vil alltid være til stede. Jo flere målinger vi gjør, jo større sannsynlighet er det for at gjennomsnittet nærmer seg en sann verdi, så lenge eksperimentet ikke inneholder systematiske feil (se under).

**Nøyaktighet** («riktighet» eller «accuracy») er et begrep som betegner hvor nærme verdien av en eller flere observasjoner er den aksepterte verdien. Nøyaktigheten uttrykkes som en feil.  Hvis vi vet hva svaret skal bli (f.eks. hvis vi skal kalibrere glassutstyr, som pipetter) kan vi regne ut den absolutte feilen $E$:


$E = x_i - x_\text{true}$


- $E$ er absolutt feil  
- $x_i$ er målt verdi  
- $x_\text{true}$ er sann akseptert verdi  

Ofte uttrykkes feil som **relativ feil**:


$E_R = \frac{x_i - x_\text{true}}{x_\text{true}} \times 100\% $


Vi deler ofte eksperimentelle feil inn i **systematiske feil**, **tilfeldige feil** og **grove feil**. For å oppdage disse feilene er det vanlig å lage flere målereplikater når vi gjennomfører et forsøk.

### Tilfeldige feil

Tilfeldige feil kommer av ukontrollerbare variabler i målingen. Det er like stor sannsynlighet for at disse er positive eller negative. Tilfeldige feil kan aldri fjernes helt, men kan reduseres ved å øve på teknikk (pipettering, titrering etc.). Tilfeldige feil viser seg i presisjonen til forsøket og angis ofte som standardavvik $s$:

$s =\sqrt{\frac{\sum (x_i - \bar{x})^2}{n - 1}}$

- $s$ er standardavvik  
- $x_i$ er observert verdi  
- $\bar{x}$ er gjennomsnitt  
- $n$ er antall observasjoner  

Standardavvik oppgis alltid med ett gjeldende siffer.

For å kunne beregne standardavviket må du ha 3 eller flere målepunkt. Standardavvik er relatert til gjennomsnittet. Det må derfor oppgis sammen med gjennomsnittet for å gi mening. Det har også samme benevning som gjennomsnittet. Et standardavvik på 1 kan være lite om gjennomsnittet er 100, men veldig stort om gjennomsnittet er 3. Vi oppgir derfor som regel **relativt standardavvik (RSD)** (også kalt «coefficient of variation») i %:


$\text{RSD} = \frac{s}{\bar{x}} \times 100\% $


### Systematiske feil

Systematiske feil kommer av feil i utstyr eller forsøksdesignet. Dette kan f.eks. være feil i kalibreringen. Om du gjentar forsøket på samme måte, vil disse feilene være reproduserbare, det vil si at de ikke er synlige i presisjonen i forsøket. Systematiske feil vil føre til at alle målereplikatene enten er for høye eller for lave.

Systematiske feil er derfor viktig å unngå og kan oppdages ved å:

- analysere en kjent prøve («kontrollprøve», eller sertifisert referansemateriale) og sørge for at resultatet ditt ikke er forskjellig fra kjent verdi (se hypotesetesting)
- bruke en annen analysemetode og sammenligne resultatene, enten i din egen lab eller ved å sende prøven til en annen lab.

### Grove feil

Grove feil kan skyldes at man ubevisst har avviket fra prosedyren. Disse kan være:

- beregningsfeil  
- overtitrering  
- fortynningsfeil  
- instrumentfeil  
- andre feil  

Hvis du gjør flere målereplikater kan disse oppdages og forkastes. Hvis de ikke forkastes, medfører de at gjennomsnittet av målingene blir feil. Mistenker du grove feil, kan statistiske tester (som Grubbs test) hjelpe deg å forkaste et målepunkt. Men de krever ofte at du har 4 målereplikater eller mer. Om du ikke har dette og mistenker en grov feil, må du gjennomføre forsøket på nytt.

### Nøyaktighet og presisjon

Når du har en måleserie og samtidig har litt informasjon om hva den aksepterte verdien ($x_\text{true}$) skal være, kan du vurdere nøyaktigheten og presisjonen i måleteknikken. Innstillingen på utstyret styrer nøyaktigheten, mens alle tilfeldige feil styrer presisjonen.

Med dette som bakgrunn er det viktig å kunne formidle hva presisjonen i resultatet ditt er. Dette gjør vi med bruk av **signifikante (gjeldende) sifre**.

## Signifikante (gjeldende) siffer (KJM1101, KJM1130, KJM2400, IN-KJM1900)

Når man skal oppgi et resultat av et eksperiment, skal dette angis med korrekt antall signifikante (eller gjeldende) siffer. Resultater oppgis derfor som:


$\bar{x} \pm s$

Fordi standardavviket sier noe om presisjonen og usikkerheten i målingene, bestemmer det hvor usikkerheten ligger i resultatet ditt. Standardavvik oppgis alltid med ett signifikant (gjeldende) siffer.

**Signifikante (gjeldende) siffer** er det antallet sifre et tall har når alle nuller i begynnelsen (men ikke i slutten) av tallet fjernes.

### Eksempler

- $0{,}001$ (ett gjeldende siffer)  
- $0{,}0010$ (to gjeldende siffer)  
- $1000$ (fire gjeldende siffer)  
- $1000{,}00$ (seks gjeldende siffer)  
- $0{,}1$ (ett gjeldende siffer)  
- $0{,}10$ (to gjeldende siffer)  
- $0{,}1000$ (fire gjeldende siffer)  

### Sifrene 1 til og med 9

Disse sifrene er alltid gjeldende uansett hvor i tallet de står.

Eksempler:

- Tallet $23$ har to gjeldende siffer  
- Tallet $23{,}7$ har tre gjeldende siffer  

### Sifferet null

Hvis sifferet null (eller flere nuller) står **inni** et tall eller **ved slutten** av et tall, er det (eller de) gjeldende.  
Hvis sifferet null (eller flere nuller) står **i begynnelsen** av et tall, er det ikke gjeldende.

Eksempler:

- Tallet $30{,}8$ har tre gjeldende siffer  
- Tallet $2{,}230$ har fire gjeldende siffer  
- Tallet $0{,}03$ har bare ett gjeldende siffer, sifferet 3  

### Heltall

Hvis tallet null, eller flere nuller, står mellom andre siffer som ikke er null, er de gjeldende.

Eksempler:

- Tallet $509$ har tre gjeldende siffer  
- Tallet $2\,000\,501$ har syv gjeldende siffer  

### Eksponenter

Ved bruk av eksponenter i vitenskapelig fremstilling av tall, teller ikke eksponenten.

Eksempel:

- $1{,}38 \times 10^{-23}$ har tre signifikante siffer (1, 3 og 8)

Antall signifikante siffer er gitt av oppløsningen / feilen i måleteknikken.

### Bruk i analytisk og fysikalsk kjemi

I analytisk kjemi utfører man ofte flere målereplikater for å kunne si noe om presisjonen til et forsøk. I et godkjent analyselaboratorium har man ofte analysert så mange prøver at man ikke trenger flere målereplikater for å si noe om presisjonen, men man bruker heller metodens standardavvik for å angi presisjonen. Analyseresultatet skal oppgis med samme antall siffer etter desimaltegnet som standardavviket.

**Eksempel:**  
Analyseresultatet ditt har et gjennomsnitt på $101{,}167\ \mu\text{g/mL}$ og et standardavvik på $0{,}2\ \mu\text{g/mL}$. Usikkerheten ligger da allerede i det første tallet etter desimaltegnet og analyseresultatet oppgis som:


$101{,}2 \pm 0{,}2\ \mu\text{g/mL}$

I fysikalsk kjemi foretar man ikke alltid en rekke gjentatte målinger av en og samme størrelse for å øke presisjonen. I stedet legger man vekt på forbedring av eksperimentets design og eventuelt forbedring av utstyret og målemetodene.

Å vurdere en eksperimentell usikkerhet krever:

- kjennskap til instrumentell presisjon og nøyaktighet  
- analyse av den eksperimentelle metode  
- bestemmelse av indre sammenheng av eksperimentelle data  
- undersøkelse av hvordan feil i eksperimentelle data innvirker på beregnede resultater  

En eksperimentell fysikalsk‑kjemisk størrelse vil nesten alltid framkomme ved et kombinert måle- og regnearbeid hvor man benytter en teoretisk sammenheng, f.eks. en formel hvor de målte størrelsene inngår. I slike tilfeller må man beregne resultatets presisjon fra usikkerheten til de størrelsene som er målt direkte.

### Feilmargin og avrunding

Om feilmarginen har et gitt antall desimaler, kan ikke presisjonen i resultatet være bedre enn dette.

**Eksempel:**  
En røntgenteknikk har en oppgitt feilmargin på $0{,}1\ \text{nm}$. Målingen gir $10{,}21\ \text{nm}$.  
Tallet rundes ned til nærmeste desimal:


$10{,}2 \pm 0{,}1\ \text{nm}$

**Avrundinger avhenger av presisjonen (på målingen/instrumentet):**

- Tall fra 5–9 rundes oppover  
- Tall fra 1–4 rundes nedover  

Eksempler:

- Usikkerhet i andre desimal: $108{,}675$ blir $108{,}7$  
- Usikkerhet i første desimal: $109{,}9$ blir $110$  
- $101{,}3$ blir $101$

### Regler ved aritmetiske operasjoner

**Regel 1 (addisjon og subtraksjon):**  
Antall desimaler bak komma i resultatet skal være det samme som det laveste antall desimaler i utgangstallene.


$10{,}21 - 0{,}1 = 10{,}1$


**Regel 2 (multiplikasjon):**  
Antall signifikante siffer i resultatet skal være det samme som det minste antallet signifikante siffer i utgangstallene.

$503{,}29 \times 6{,}177 \approx 3109
10{,}2 \times 0{,}2 \approx 2 $


**Regel 3 (divisjon):**  
Antall signifikante siffer i resultatet skal være det samme som det minste antallet i utgangstallene.


$\frac{10{,}21}{0{,}2} \approx 51 $


**Regel 4 (logaritmer):**  
Antall desimaler i $\log x$ er det samme som antall signifikante tall i $x$. Samme regel gjelder for $\ln x$.


$\ln(100) = 4{,}605 $

**Regel 5 (potenser av 10, eksponentialer):**  
Antall signifikante tall i potensen av $10^x$ er det samme som i $x$. Samme regel gjelder for eksponentialer $\exp(x)$.

## Feilopphopningsloven (error propagation) (KJM1130)

Som nevnt for fysikalsk kjemi, er det mulig å estimere feil for en metode ut fra feilene på hver del av målingene som inngår. I praksis er dette anvendt Pythagoras i et flerdimensjonalt rom.

La oss kalle verdien vi er interessert i $R$. Den kan være en eksplisitt funksjon av et sett uavhengig målte størrelser $x_1, x_2, x_3, \ldots, x_n$:


$R = f(x_1, x_2, \ldots, x_n)$


Generelt vil vi ha feil i de målte størrelsene som kan kalles $\delta x_1, \ldots, \delta x_n$.  
Feilen i $R$, $\delta R$, gis da av feilopphopningsregelen, hvor uttrykket må partialderiveres med hensyn på hver variabel/målestørrelse:

\delta R = \sqrt{\left(\frac{\partial R}{\partial x_1},\delta x_1\right)^2 + \left(\frac{\partial R}{\partial x_2},\delta x_2\right)^2 + \cdots + \left(\frac{\partial R}{\partial x_n},\delta x_n\right)^2}

### Nyttige eksempler

Her er noen få eksempler på typiske tilfeller man finner i kjemi.

#### 1. Addisjon og subtraksjon

Verdien vi er interessert i, er et resultat av addisjon og subtraksjon av forskjellige termer, hver med en usikkerhet i seg:


$R = a x + b y - c z$


Gir:

$(\delta R)^2 =(a\,\delta x)^2 +(b\,\delta y)^2 +(-c\,\delta z)^2$

$\delta R =\sqrt{(a\,\delta x)^2 +(b\,\delta y)^2 +(-c\,\delta z)^2}$


#### 2. Multiplikasjon / divisjon (potenser)

Verdien vi er interessert i er et resultat av multiplikasjon/divisjon:


$R = a x^n y^m z^{-k}$


Den relative feilen blir:


$\frac{\delta R}{R} =\sqrt{\left(n \frac{\delta x}{x}\right)^2+\left(m \frac{\delta y}{y}\right)^2+\left(k \frac{\delta z}{z}\right)^2}$


#### 3. Logaritmisk avhengighet

Et resultat med logaritmisk avhengighet:


$R = -b \cdot \log(x)$


Gir:


$\delta R = b \frac{\delta x}{x}$

## Hypotesetesting (t-test) (KJM2400, IN-KJM1900)

En t-test er en statistisk metode vi kan bruke for å teste om det er signifikant forskjell mellom gjennomsnittet av to datasett. Det finnes i hovedsak tre ulike t-tester, avhengig av hva slags data vi har:

- **Uavhengig t-test:** Vi har to serier med målinger fra ulike prøver/utvalg som vi skal sammenlikne.  
  Nullhypotese:
 
 $ H_0 : \bar{x}_1 = \bar{x}_2$
  

- **Paret t-test:** Vi har to serier med målinger fra samme prøve/utvalg.

- **T-test for ett utvalg:** Vi har en serie med målinger som vi skal sammenlikne med et kjent gjennomsnitt.  
  Nullhypotese:
  
  $H_0 : \bar{x} = \mu$
  

I disse testene regner vi ut en t-verdi basert på de dataene vi har. Da får vi et tall som vi sammenlikner med den forventede verdien i en sannsynlighetsfordeling som kalles **t-fordelingen**. Denne finner du i tabeller. Denne sammenligningen hjelper oss med å avgjøre hvor sannsynlig det er at den observerte forskjellen (eller en mer ekstrem forskjell) kunne ha oppstått ved en tilfeldighet.

Vi kan sammenlikne denne sannsynligheten med det vi kaller et **signifikansnivå** $p$. Dette nivået setter vi som oftest til å være $p = 0{,}05$. Det vil si at dersom vi får en sannsynlighet under $5\%$ ($< 0{,}05$) når vi utfører en t-test, er det en signifikant forskjell mellom målingene.

### 1. Uavhengig t-test (to utvalg)

Denne testen sammenligner gjennomsnittet av to uavhengige grupper for å se om de er signifikant forskjellige.

#### F-test (likhet i varians / standardavvik)

For å kunne utføre en uavhengig t-test må vi først undersøke om datasettene er fra populasjoner med samme standardavvik eller ikke. Dette gjøres med en **F-test**:


$F = \frac{s_1^2}{s_2^2}$


- $s_1$ er standardavvik fra det ene datasettet  
- $s_2$ er standardavvik fra det andre datasettet  
- velg $s_1 > s_2$ slik at $F \ge 1$

$F$ som du regner ut skal sammenlignes med en $F$ fra en tosidig F-tabell. Hvis $F > F_\text{tabell}$, er standardavvikene signifikant forskjellige. Antall frihetsgrader til tabellen er:

- for $s_1$: $n_1 - 1$  
- for $s_2$: $n_2 - 1$

#### Uavhengig t-test (like standardavvik, «pooled»)

Hvis F-testen viser at datasettene er fra populasjoner med antatt samme standardavvik, skal du bruke følgende formel:


$t =\frac{\left|\bar{x}_1 - \bar{x}_2\right|}{s_\text{pooled}}\sqrt{\frac{n_1 n_2}{n_1 + n_2}}$


- $\bar{x}_1$ og $\bar{x}_2$ er gjennomsnittet i de to gruppene  
- $s_\text{pooled}$ er samlet standardavvik  
- $n_1$ og $n_2$ er antallet observasjoner i de to gruppene  

Samlet standardavvik $s_\text{pooled}$:


$s_\text{pooled} =\sqrt{\frac{s_1^2 (n_1 - 1) + s_2^2 (n_2 - 1)}{n_1 + n_2 - 2}}$


Antall frihetsgrader:


$\text{d.f.} = n_1 + n_2 - 2$


- Hvis $t > t_\text{tabell}$ ved $95\%$ konfidensnivå kan $H_0$ forkastes: de to datasettene er signifikant forskjellige.  
- Hvis $t < t_\text{tabell}$ ved $95\%$ konfidensnivå kan ikke $H_0$ forkastes: de to datasettene er ikke signifikant forskjellige.

#### Uavhengig t-test (ulike standardavvik, «Welch»)

Hvis F-testen viser at datasettene ikke er fra populasjoner med antatt samme standardavvik, skal du bruke følgende formel:


$t =\frac{\left|\bar{x}_1 - \bar{x}_2\right|}{\sqrt{\frac{s_1^2}{n_1} +\frac{s_2^2}{n_2}}}$


- $\bar{x}_1$ og $\bar{x}_2$ er gjennomsnittet i de to gruppene  
- $s_1^2$ og $s_2^2$ er variansene  
- $n_1$ og $n_2$ er antallet observasjoner  

Antall frihetsgrader (Welch–Satterthwaite-approksimasjon):

$\text{d.f.} =\frac{\left(\frac{s_1^2}{n_1} +\frac{s_2^2}{n_2}\right)^2}{\frac{\left(\frac{s_1^2}{n_1}\right)^2}{n_1 - 1}+\frac{\left(\frac{s_2^2}{n_2}\right)^2}{n_2 - 1}}$

- Hvis $t > t_\text{tabell}$ ved $95\%$ konfidensnivå kan $H_0$ forkastes: de to datasettene er signifikant forskjellige.  
- Hvis $t < t_\text{tabell}$ ved $95\%$ konfidensnivå kan ikke $H_0$ forkastes: de to datasettene er ikke signifikant forskjellige.

### 2. Paret t-test

Denne testen sammenligner gjennomsnittet av forskjellene mellom par av observasjoner fra samme utvalg for å se om det er en signifikant forskjell. Et eksempel er målinger på renheten i vann før og etter en renseteknikk er utført.

Definisjon:


$t =\frac{\bar{d}}{s_d / \sqrt{n}}$

- $\bar{d}$ er gjennomsnittsdifferansen mellom parrene  
- $n$ er antall par  
- $s_d$ er standardavviket til differansene, beregnet fra:


$s_d =\sqrt{\frac{\sum (d_i - \bar{d})^2}{n - 1}}$


Antall frihetsgrader:


$\text{d.f.} = n - 1$


- Hvis $t > t_\text{tabell}$ ved $95\%$ konfidensnivå kan $H_0$ forkastes: de to datasettene er signifikant forskjellige.  
- Hvis $t < t_\text{tabell}$ ved $95\%$ konfidensnivå kan ikke $H_0$ forkastes: de to datasettene er ikke signifikant forskjellige.

### 3. T-test for et utvalg

Denne testen sammenligner gjennomsnittet av et utvalg med et kjent gjennomsnitt for å se om de er signifikant forskjellige. Et eksempel er hvis vi vil sjekke om et nytt instrument gir nøyaktige målinger av en kjent standard.

Definisjon:


$t =\frac{\bar{x} - \mu}{s / \sqrt{n}}$


- $\bar{x}$ er gjennomsnitt  
- $\mu$ er sann, akseptert verdi  
- $n$ er antall observasjoner  
- $s$ er standardavviket for observasjonene  

Antall frihetsgrader:


$\text{d.f.} = n - 1$


- Hvis $t > t_\text{tabell}$ ved $95\%$ konfidensnivå kan $H_0$ forkastes: de to datasettene er signifikant forskjellige.  
- Hvis $t < t_\text{tabell}$ ved $95\%$ konfidensnivå kan ikke $H_0$ forkastes: de to datasettene er ikke signifikant forskjellige.
