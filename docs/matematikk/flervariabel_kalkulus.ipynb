## 5.1 Flervariable funksjoner

Så langt har alle funksjonene våre hatt én $x$, men det er mange eksempler hvor én størrelse avhenger av flere enn én annen størrelse. I kjemien avhenger de fleste størrelser av *flere* ting samtidig. Trykket i en gass avhenger både av volumet og temperaturen. Løseligheten til et stoff avhenger av både temperatur og trykk. Reaksjonshastigheten avhenger av konsentrasjonen til flere reaktanter på én gang. For å beskrive slike sammenhenger trenger vi funksjoner som tar inn flere variabler. Heldigvis er ikke ideen så forskjellig fra det vi allerede kan - vi utvider bare antallet $x$-er.

### 5.1.1 Funksjoner av n-variabler

Vi starter med å utvide selve funksjonsbegrepet. En **funksjon av flere variabler** fungerer akkurat som en funksjon av én variabel, bortsett fra at vi mater inn flere tall for å få ut ett tall.

:::{admonition} Def. 5.1 - Funksjon av flere variabler
:class: warning

En **funksjon av to variabler** $f$ er en regel som tildeler hvert tallpar $(x_1,x_2)$ nøyaktig ett tall $f(x_1,x_2)$. Vi skriver

$f:\mathbb{R}^2 \to \mathbb{R}$

Mer generelt tildeler en **funksjon av $n$ variabler** hvert punkt $(x_1, x_2, \ldots, x_n)$ nøyaktig ett tall $f(x_1, x_2, \ldots, x_n)$, og vi skriver 

$f:\mathbb{R}^n \to \mathbb{R}$.
:::

Legg merke til at funksjonsverdien fortsatt er ett enkelt tall - det er bare inputen som består av flere tall. Kravet fra [Def. 4.1 - Funksjon av én variabel](https://andreasdh.github.io/Bifrost/docs/matematikk/kalkulus.html#funksjon-definisjonmengde-og-verdiomrade), at hver input gir nøyaktig én output, er akkurat det samme. Vi setter inn verdier på nøyaktig samme måte som før, bare med flere tall om gangen.

:::{card} 
**Eksempel**
^^^
La $f(x,y)=x^2+3y$. Regn ut $f(2,1)$ og $f(0,-4)$.

Vi setter inn verdiene for $x$ og $y$, i den rekkefølgen de står.

For $f(2,1)$ setter vi $x=2$ og $y=1$:

$f(2,1)=2^2+3\cdot 1=4+3=7$

For $f(0,-4)$ setter vi $x=0$ og $y=-4$:

$f(0,-4)=0^2+3\cdot(-4)=0-12=-12$

Merk at rekkefølgen betyr noe: $f(2,1)$ er ikke det samme som $f(1,2)$.
:::

Definisjonsmengden til en flervariabel funksjon er mengden av alle tallpar $(x,y)$ vi har lov til å sette inn - altså alle par der regneuttrykket gir mening. Prinsippet er det samme som i én variabel: vi må unngå å dele på null, ta kvadratrot av negative tall, eller ta logaritmen av noe som ikke er positivt.

:::{card} 
**Eksempel**
^^^
Hva er definisjonsmengden til $f(x,y)=\dfrac{1}{x-y}$?

Uttrykket gir mening for alle tallpar $(x,y)$ så lenge vi ikke deler på null. Nevneren blir null når $x-y=0$, altså når $x=y$.

Definisjonsmengden er derfor alle tallpar $(x,y)$ der $x \neq y$. Geometrisk betyr det hele planet unntatt den rette linjen $y=x$.
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Idealgassloven er et av de mest kjente eksemplene på en flervariabel funksjon i kjemien. Løser vi $PV=nRT$ for trykket, får vi

$P(V,T)=\frac{nRT}{V}$

Her er trykket $P$ en funksjon av *to* variabler: volumet $V$ og temperaturen $T$ (mens $n$ og $R$ er konstanter). Definisjonsmengden er fysisk begrenset: volumet må være positivt ($V>0$, ellers deler vi på null), og temperaturen målt i Kelvin kan ikke være negativ ($T \geq 0$). Setter vi inn et bestemt volum og en bestemt temperatur, får vi ut nøyaktig ett trykk - helt i tråd med Def. 5.1.
:::

Nå som vi kan beskrive funksjoner av flere variabler, dukker et naturlig spørsmål opp: hvordan deriverer vi dem? Hvis trykket avhenger av både volum og temperatur, hva mener vi da med "endringsraten"? Svaret er at vi må derivere med hensyn på én variabel om gangen, og holde de andre konstante. Dette leder oss til partiellderiverte i neste delkapittel.

La $f(x,y)=2x+y^2$.

b) Er $f(1,2)$ det samme som $f(2,1)$? Regn ut begge.

c) Hva er definisjonsmengden til $g(x,y)=\dfrac{1}{\sqrt{x}}+y$?
:::

$f(3,2)=2\cdot 3+2^2=6+4=10$

b) Vi regner ut begge:

$f(1,2)=2\cdot 1+2^2=2+4=6$

$f(2,1)=2\cdot 2+1^2=4+1=5$

Siden $6 \neq 5$, er $f(1,2)$ **ikke** det samme som $f(2,1)$. Rekkefølgen på variablene betyr noe.

c) For at $g(x,y)=\dfrac{1}{\sqrt{x}}+y$ skal gi mening, må uttrykket under kvadratroten være positivt (vi kan verken ta rot av et negativt tall eller dele på null, og $\sqrt{0}=0$ gir null i nevneren). Vi trenger derfor $x>0$, mens $y$ kan være hvilket som helst reelt tall. Definisjonsmengden er alle tallpar $(x,y)$ med $x>0$.

## 5.2 Partiellderiverte

I forrige delkapittel så vi at trykket i en gass kan avhenge av både volum og temperatur gjennom $P(V,T)$. Men da dukker et problem opp: hva mener vi egentlig med gjennomsnittlig endringsrate til en funksjon av flere variable? For en funksjon $f(x)$ av én variabel var svaret entydig - den deriverte $f'(x)$ målte hvor raskt $f$ endret seg ([4.4.1 Gjennomsnittlig endringsrate](https://andreasdh.github.io/Bifrost/docs/matematikk/kalkulus.html#gjennomsnittlig-endringsrate)). For $f(x,y)$ er det ikke like klart: endrer vi $x$, eller $y$, eller begge samtidig? Løsningen er enkel og elegant: vi ser på én variabel om gangen, og holder de andre helt i ro. Dette kaller vi å **derivere partielt**.

### 5.2.1 Definisjonen av de partiellderiverte

Ideen er å velge én variabel å derivere med hensyn på, og late som om alle de andre variablene bare er konstanter. Deriverer vi med hensyn på $x$, behandler vi $y$ som et fast tall - og bruker ellers akkurat de samme derivasjonsreglene som i ([4.4.5 De deriverte til kjente funksjoner](https://andreasdh.github.io/Bifrost/docs/matematikk/kalkulus.html#derivasjonsregler)).

:::{admonition} Def. 5.2 - Partiellderivert
:class: warning

La $f(x,y)$ være en funksjon av to variabler. Den **partiellderiverte av $f$ med hensyn på $x$** er

$\frac{\partial f}{\partial x} = \lim_{h \to 0} \frac{f(x+h,\,y)-f(x,\,y)}{h}$

der vi holder $y$ konstant. Tilsvarende er den **partiellderiverte med hensyn på $y$**

$\frac{\partial f}{\partial y} = \lim_{h \to 0} \frac{f(x,\,y+h)-f(x,\,y)}{h}$

der vi holder $x$ konstant.
:::

Legg merke til det krøllete symbolet $\partial$ ("del" eller "partiell-d"). Vi bruker det i stedet for det vanlige $d$ nettopp for å understreke at funksjonen har flere variabler, og at vi bare deriverer med hensyn på én av dem. Sammenlign gjerne med definisjonen av den vanlige deriverte i [Def. 4.9 - Den deriverte](https://andreasdh.github.io/Bifrost/docs/matematikk/kalkulus.html#definisjonen-av-den-deriverte) - den eneste forskjellen er at den andre variabelen står fast som en tilskuer. Vi bruker også skrivemåtene $f_x$ og $f_y$ som kortform for $\frac{\partial f}{\partial x}$ og $\frac{\partial f}{\partial y}$.

:::{card} 
**Eksempel**
^^^
Finn begge de partiellderiverte til $f(x,y)=x^2 y + 3x y^3$.

**Med hensyn på $x$** behandler vi $y$ som en konstant. Da er $x^2 y$ et konstant-tall ($y$) ganget med $x^2$, og $3xy^3$ er en konstant ($3y^3$) ganget med $x$:

$\frac{\partial f}{\partial x} = 2x y + 3y^3$

**Med hensyn på $y$** behandler vi $x$ som en konstant. Nå er $x^2 y$ en konstant ($x^2$) ganget med $y$, og $3xy^3$ er en konstant ($3x$) ganget med $y^3$:

$\frac{\partial f}{\partial y} = x^2 + 9x y^2$

Legg merke til at leddet $x^2 y$ ga ulikt bidrag i de to tilfellene, alt etter hvilken variabel vi så på som konstant.
:::

:::{card} 
**Eksempel**
^^^
Finn de partiellderiverte til $g(x,y)=\sin{(xy)}$.

**Med hensyn på $x$** holder vi $y$ konstant. Den indre funksjonen er $xy$, som deriveres til $y$ (siden $y$ er konstant). Ved kjerneregelen:

$\frac{\partial g}{\partial x} = \cos{(xy)} \cdot y = y\cos{(xy)}$

**Med hensyn på $y$** holder vi $x$ konstant. Nå deriveres den indre funksjonen $xy$ til $x$:

$\frac{\partial g}{\partial y} = \cos{(xy)} \cdot x = x\cos{(xy)}$
:::

### 5.2.2 Blandede partiellderiverte

Akkurat som vi kunne derivere flere ganger i én variabel i [4.4.4 Høyere-ordens deriverte](https://andreasdh.github.io/Bifrost/docs/matematikk/kalkulus.html#hoyere-ordens-deriverte), kan vi derivere partielt flere ganger. Nå får vi flere muligheter: vi kan derivere to ganger med hensyn på $x$, to ganger med hensyn på $y$, eller én gang med hver. De to siste kalles **blandede partiellderiverte**.

:::{admonition} Def. 5.3 - Andreordens partiellderiverte
:class: warning

De **andreordens partiellderiverte** til $f(x,y)$ er

$\frac{\partial^2 f}{\partial x^2} = \frac{\partial}{\partial x}\left(\frac{\partial f}{\partial x}\right), \qquad \frac{\partial^2 f}{\partial y^2} = \frac{\partial}{\partial y}\left(\frac{\partial f}{\partial y}\right)$

De **blandede partiellderiverte** er

$\frac{\partial^2 f}{\partial x\, \partial y} = \frac{\partial}{\partial x}\left(\frac{\partial f}{\partial y}\right), \qquad \frac{\partial^2 f}{\partial y\, \partial x} = \frac{\partial}{\partial y}\left(\frac{\partial f}{\partial x}\right)$
:::

:::{card} 
**Eksempel**
^^^
For $f(x,y)=x^2 y + 3x y^3$ fant vi $\frac{\partial f}{\partial x}=2xy+3y^3$ og $\frac{\partial f}{\partial y}=x^2+9xy^2$. Finn de to blandede partiellderiverte.

For $\frac{\partial^2 f}{\partial x\, \partial y}$ deriverer vi $\frac{\partial f}{\partial y}=x^2+9xy^2$ med hensyn på $x$ (hold $y$ konstant):

$\frac{\partial^2 f}{\partial x\, \partial y} = 2x + 9y^2$

For $\frac{\partial^2 f}{\partial y\, \partial x}$ deriverer vi $\frac{\partial f}{\partial x}=2xy+3y^3$ med hensyn på $y$ (hold $x$ konstant):

$\frac{\partial^2 f}{\partial y\, \partial x} = 2x + 9y^2$

Merk at de to blandede ble like! Dette er ingen tilfeldighet.
:::

At de blandede partiellderiverte ble like i eksempelet over, er et generelt resultat: for de aller fleste funksjoner vi møter i praksis, spiller rekkefølgen vi deriverer i ingen rolle, så lenge de blandede partiellderiverte er kontinuerlige. For oss betyr det en grei sjekk: regner du ut begge blandede og får ulikt svar, har du sannsynligvis gjort en regnefeil.

### 5.2.3 Gradienten

De to partiellderiverte forteller hver for seg hvor bratt flaten heller i $x$-retningen og i $y$-retningen. Ofte er det nyttig å samle dem i ett objekt. Vi setter dem sammen til en vektor, som vi kaller **gradienten**.

:::{admonition} Def. 5.4 - Gradienten
:class: warning

**Gradienten** til en funksjon $f(x,y)$ er vektoren av de partiellderiverte,

$\nabla f = \left( \frac{\partial f}{\partial x}, \; \frac{\partial f}{\partial y} \right)$

Symbolet $\nabla$ leses "nabla". For en funksjon av $n$ variabler har gradienten tilsvarende én komponent per variabel.
:::

Gradienten har en vakker geometrisk tolkning: i et gitt punkt peker $\nabla f$ i den retningen der funksjonen vokser raskest, og lengden av vektoren forteller hvor bratt den vokser. Står du i fjellandskapet fra introduksjonsspørsmålene, peker gradienten rett oppover den bratteste stigningen. Et beslektet poeng, som er nyttig i kjemien: gradienten står alltid vinkelrett på nivåkurvene.

:::{card} 
**Eksempel**
^^^
Finn gradienten til $f(x,y)=x^2+y^2$, og regn den ut i punktet $(1,2)$.

Vi finner først de to partiellderiverte:

$\frac{\partial f}{\partial x} = 2x, \qquad \frac{\partial f}{\partial y} = 2y$

Gradienten er dermed

$\nabla f = (2x, \; 2y)$

I punktet $(1,2)$ blir den

$\nabla f(1,2) = (2\cdot 1, \; 2\cdot 2) = (2, \; 4)$

Denne vektoren peker i retningen der $f$ vokser raskest fra punktet $(1,2)$ - altså rett bort fra origo, som stemmer med at nivåkurvene her er sirkler om origo.
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Partiellderiverte er selve språket i termodynamikken, men kjemikere skriver dem litt annerledes enn matematikere. Siden en størrelse kan avhenge av mange variabler, er det avgjørende å presisere *hvilke* variabler som holdes konstante. Derfor skriver man den konstante variabelen som en senket indeks utenfor en parentes.

For idealgassen $P(V,T)=\dfrac{nRT}{V}$ blir de to partiellderiverte

$\left(\frac{\partial P}{\partial T}\right)_V = \frac{nR}{V} \qquad \text{og} \qquad \left(\frac{\partial P}{\partial V}\right)_T = -\frac{nRT}{V^2}$

Notasjonen $\left(\frac{\partial P}{\partial T}\right)_V$ leses "den partiellderiverte av $P$ med hensyn på $T$, med $V$ holdt konstant". Indeksen $V$ er bare en eksplisitt påminnelse om hva som er konstant. Slike uttrykk dukker opp overalt i termodynamikken, for eksempel i sammenhengen mellom varmekapasitet og indre energi, $C_V=\left(\frac{\partial U}{\partial T}\right)_V$.
:::

### 5.2.4 Den totale differensialen

Den deriverte $df$ til en funksjon av én variabel er en enkel sak, den avhenger jo bare av endringen i $dx$. Noe vi skal bruke mye i termodynamikken er **den totale differensialen** $df$ til en funksjon av $n$-variabler, $f(x_1, \dots, x_n)$. Her avhenger nemlig endringen i $f$ seg av _hvor mye_ hver variabel $x_i$ har endret seg, altså $dx_i$, i tillegg til hvor mye $f$ endrer seg når $x_i$ endrer seg litt, og alt annet er konstant, $\frac{\partial f}{\partial x_i}$. 


:::{admonition} Def. 5.5 - Den totale differensialen
:class: warning

**Den totale differensialen** til en funksjon $f(x_1, \dots, x_n)$ er 

$df = \frac{\partial f}{\partial x_1} \ dx_1 + \dots + \frac{\partial f}{\partial x_n} \ dx_n = \sum_{i=1}^n \ \frac{\partial f}{\partial x_i} \ dx_i$

Symbolet $\nabla$ leses "nabla". For en funksjon av $n$ variabler har gradienten tilsvarende én komponent per variabel.
:::

:::{card} 
**Eksempel**
^^^
Du selger is på sommeren og følger med på fortjenesten din $f$ som utelukkende er en funksjon av to variabler: prisen på isen $p$ og temperaturen ute $T$. For eksempel kan det tenkes at når det blir varmere ($T$ øker), så øker fortjenesten $f$. Hvordan kan vi beskrive endringen i fortjenesten, $df$?

Først og fremst har vi

$\frac{\partial f}{\partial T}$: ved konstant pris $p$, hvor mye øker fortjenesten per grad det blir varmere

$\frac{\partial f}{\partial p}$: ved konstant temperatur $T$, hvor mye øker fortjenesten per krone dyrere isen blir


For eksempel kan det tenkes at kundene dine er veldig sensitive for temperatur og bare kjøper is ved høy $T$, og da er $\frac{\partial f}{\partial T}$ stor og positiv. 

I tillegg må vi vite hvor mye $T$ og $p$ endrer seg, henholdvis $dT$ og $dp$. Vi får et bidrag for temperatur og ett for pris, og legger de sammen for å se hvordan $df$ endrer seg totalt:

$df = \frac{\partial f}{\partial T} \ dT + \frac{\partial f}{\partial p} \ dp$

Vi bruker ofte litt annen notasjon i kjemien, slik som denne

$df = \left( \frac{\partial f}{\partial T} \right)_p \ dT + \left( \frac{\partial f}{\partial p} \right)_T \ dp$

:::

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Gibbs-fri energi er en funksjon av temperatur $T$ og trykk $P$, samt $n_1, \dots, n_m$ som er stoffmengdene i systemet, når man antar at disse kan endres. Altså får vi

$G(T,P,n_1,\dots,n_m)$

Vi kan derfor skrive den totale differensialen som summen av bidraget fra hver variabel:

$dG = \left(\frac{\partial G}{\partial T}\right)_{P,n_i}dT + \left(\frac{\partial G}{\partial P}\right)_{T,n_i}dP + \sum_{i=1}^{m}\left(\frac{\partial G}{\partial n_i}\right)_{T,P,n_{j\neq i}}dn_i$

der hvert ledd forteller hvor mye $G$ endrer seg når den ene variabelen endres, mens de andre holdes konstant.

Anta at vi har konstant $T$ og $P$, slik at $dT = dP = 0$, og vi får

$dG = \sum_{i=1}^{m}\left(\frac{\partial G}{\partial n_i}\right)_{T,P,n_{j\neq i}}dn_i$

Kjemisk potensial $\mu_i$ er definert som

$\mu_i = \left(\frac{\partial G}{\partial n_i}\right)_{T,P,n_{j\neq i}}$

og vi har 

$dG = \sum_{i=1}^{m} \mu_i dn_i$


:::

Med partiellderiverte og gradienten på plass har vi verktøyene for å studere hvordan flervariable funksjoner endrer seg. Et naturlig neste spørsmål er hvordan vi finner topp- og bunnpunkter for slike funksjoner - særlig når de i tillegg er underlagt en betingelse, slik som å minimere energi under bevaring av masse. Det leder oss til [5.4 Lagrange multiplikatormetode](https://andreasdh.github.io/Bifrost/docs/matematikk/flervariabel_kalkulus.html#lagrange-multiplikatormetode), men først skal vi se på integraler for funksjoner av to og tre variabler i [5.3 Dobbel- og trippelintegraler](https://andreasdh.github.io/Bifrost/docs/matematikk/flervariabel_kalkulus.html#dobbel-og-trippelintegraler).

:::{admonition} Underveisoppgave 5.2
:class: tip

La $f(x,y)=x^3 y - 2x y^2$.

a) Finn $\dfrac{\partial f}{\partial x}$ og $\dfrac{\partial f}{\partial y}$.

b) Finn den blandede partiellderiverte $\dfrac{\partial^2 f}{\partial x\, \partial y}$.

c) Skriv opp gradienten $\nabla f$, og regn den ut i punktet $(1,1)$.

d) For idealgassen $P(V,T)=\dfrac{nRT}{V}$: regn ut $\left(\dfrac{\partial P}{\partial V}\right)_T$ og forklar hvorfor svaret er negativt.
:::

:::{admonition} Løsning 5.2
:class: tip, dropdown

a) Med hensyn på $x$ (hold $y$ konstant): $\dfrac{\partial f}{\partial x} = 3x^2 y - 2y^2$. Med hensyn på $y$ (hold $x$ konstant): $\dfrac{\partial f}{\partial y} = x^3 - 4xy$.

b) Vi deriverer $\dfrac{\partial f}{\partial y}=x^3-4xy$ med hensyn på $x$: $\dfrac{\partial^2 f}{\partial x\, \partial y} = 3x^2 - 4y$.

c) Gradienten er $\nabla f = \left(3x^2 y - 2y^2, \; x^3 - 4xy\right)$. I punktet $(1,1)$: $\nabla f(1,1) = (3\cdot 1 - 2, \; 1 - 4) = (1, \; -3)$.

d) Vi deriverer $P=\dfrac{nRT}{V}=nRT \cdot V^{-1}$ med hensyn på $V$ (hold $T$ konstant): $\left(\dfrac{\partial P}{\partial V}\right)_T = -nRT \cdot V^{-2} = -\dfrac{nRT}{V^2}$. Svaret er negativt fordi trykket *synker* når volumet øker ved konstant temperatur - akkurat som Boyles lov sier. En positiv $V$ i nevneren opphøyd i andre gjør hele uttrykket negativt.
:::

## 5.3 Dobbel-og trippelintegraler

I [4.5.4 Det bestemte integralet og Riemann-summer](https://andreasdh.github.io/Bifrost/docs/matematikk/kalkulus.html#det-bestemte-integralet-og-riemann-summer) så vi at det bestemte integralet $\int_a^b f(x)\,dx$ måler arealet under grafen til en funksjon av én variabel. Men nå kan vi funksjoner av flere variabler og et naturlig spørsmål melder seg: hva blir integralet aven funksjon av to variabler, $f(x,y)$? Der enkeltintegralet summerte opp arealet av uendelig mange tynne striper, skal vi nå summere opp volumet av uendelig mange tynne søyler - og resultatet blir volumet under en flate. Utvider vi til tre variabler, kan vi regne ut ting som massen eller ladningen fordelt i et romområde. 

Vi starter med to variabler. Akkurat som vi delte intervallet $[a,b]$ i tynne striper, deler vi nå et område i $xy$-planet inn i et rutenett av små rektangler, hvert med areal $\Delta A = \Delta x\,\Delta y$. Over hver rute reiser vi en søyle med høyde $f(x,y)$, altså volum $f(x,y)\,\Delta A$. Legger vi sammen alle søylene og lar rutene bli uendelig små, får vi **dobbeltintegralet**.

:::{admonition} Def. 5.6 - Dobbeltintegralet
:class: warning

**Dobbeltintegralet** av $f(x,y)$ over et område $D$ i $xy$-planet er grenseverdien av søylesummene når rutene blir uendelig små:

$\iint_D f(x,y)\,dA = \lim_{\Delta A \to 0} \sum_i f(x_i,y_i)\,\Delta A$

Når $f(x,y)\geq 0$, er dobbeltintegralet **volumet** mellom flaten $z=f(x,y)$ og området $D$ i $xy$-planet.
:::

Definisjonen er fin å forstå, men vi vil ikke summere uendelig mange søyler for hånd. Heldigvis finnes det en enkel oppskrift, helt analog til [Analysens fundamentalteorem](https://andreasdh.github.io/Bifrost/docs/matematikk/kalkulus.html#det-bestemte-integralet-og-riemann-summer): vi integrerer én variabel om gangen. Dette kalles et **iterert integral**. Vi holder den ene variabelen konstant (akkurat som ved partiellderivasjon), integrerer med hensyn på den andre, og integrerer så resultatet med hensyn på den første.

:::{admonition} Teorem 5.1 - Fubinis teorem
:class: note

La $f(x,y)$ være kontinuerlig på rektangelet $D=[a,b]\times[c,d]$. Da kan dobbeltintegralet regnes ut som et iterert integral, og rekkefølgen spiller ingen rolle:

$\iint_D f(x,y)\,dA = \int_a^b \left( \int_c^d f(x,y)\,dy \right) dx = \int_c^d \left( \int_a^b f(x,y)\,dx \right) dy$
:::

:::{card} 
**Eksempel**
^^^
Regn ut $\iint_D xy \, dA$ over rektangelet $D=[0,2]\times[0,1]$.

Vi regner det som et iterert integral, og starter innenfra med hensyn på $x$ (mens $y$ holdes konstant):

$\int_0^2 xy \, dx = y\int_0^2 x \, dx = y\Big[\frac{x^2}{2}\Big]_0^2 = y \cdot 2 = 2y$

Så integrerer vi resultatet med hensyn på $y$:

$\int_0^1 2y \, dy = \Big[y^2\Big]_0^1 = 1$

Altså er $\iint_D xy \, dA = 1$. Ifølge Fubini hadde vi fått samme svar om vi integrerte i motsatt rekkefølge - prøv gjerne selv.
:::

Steget videre til tre variabler er nå kort. I stedet for å dele opp et flatt område i ruter, deler vi opp et romområde $E$ i små bokser med volum $\Delta V = \Delta x\,\Delta y\,\Delta z$. Vi summerer $f(x,y,z)\,\Delta V$ over alle boksene og lar dem bli uendelig små.

:::{admonition} Def. 5.7 - Trippelintegralet
:class: warning

**Trippelintegralet** av $f(x,y,z)$ over et romområde $E$ er

$\iiint_E f(x,y,z)\,dV = \lim_{\Delta V \to 0} \sum_i f(x_i,y_i,z_i)\,\Delta V$

Setter vi $f=1$, gir integralet **volumet** av $E$. Er $f$ en tetthet (masse per volum), gir integralet den totale **massen**.
:::

Fubinis teorem gjelder også her: vi integrerer én variabel om gangen, tre ganger. Over en boks $[a_1,b_1]\times[a_2,b_2]\times[a_3,b_3]$ blir det tre nøstede integraler. Men mange av de viktigste anvendelsene i kjemien handler ikke om bokser - de handler om *kuler*, som når vi ser på et atom. Da lønner det seg å bytte koordinatsystem.

Et hydrogenatom er kulesymmetrisk, ikke boksformet. Å beskrive det med $x$, $y$ og $z$ blir tungvint. I stedet bruker vi **sfæriske koordinater**, der vi angir et punkt ved hvor langt det er fra origo og i hvilken retning: avstanden $r$ fra origo, vinkelen $\theta$ ned fra $z$-aksen (polarvinkelen), og vinkelen $\phi$ rundt $z$-aksen (asimutvinkelen).

:::{admonition} Def. 5.8 - Sfæriske koordinater
:class: warning

Et punkt $(x,y,z)$ kan beskrives ved sfæriske koordinater $(r,\theta,\phi)$ gjennom

$x = r\sin{\theta}\cos{\phi}, \qquad y = r\sin{\theta}\sin{\phi}, \qquad z = r\cos{\theta}$

der $r \geq 0$ er avstanden fra origo, $0 \leq \theta \leq \pi$ er vinkelen fra den positive $z$-aksen, og $0 \leq \phi < 2\pi$ er vinkelen i $xy$-planet.
:::

Når vi bytter koordinater, kan vi ikke bare bytte ut $dV = dx\,dy\,dz$ med $dr\,d\theta\,d\phi$ - det ville gitt feil svar. Grunnen er at en liten "boks" i sfæriske koordinater ikke har sidene $dr$, $d\theta$ og $d\phi$ i lengde. Bare $dr$ er en ekte lengde. De to vinklene må ganges opp med en radius for å bli lengder: en liten endring $d\theta$ svarer til en bue av lengde $r\,d\theta$, og en liten endring $d\phi$ svarer til en bue av lengde $r\sin{\theta}\,d\phi$ (faktoren $\sin\theta$ kommer av at sirklene rundt $z$-aksen er mindre nær polene enn ved ekvator). Volumet av den lille boksen blir derfor produktet av de tre lengdene.

:::{admonition} Def. 5.9- Volumelementet i sfæriske koordinater
:class: warning

I sfæriske koordinater er volumelementet

$dV = r^2 \sin{\theta} \, dr \, d\theta \, d\phi$

Faktoren $r^2\sin{\theta}$ kalles Jacobideterminanten for overgangen, og retter opp for at vinkelkoordinatene ikke er lengder.
:::

:::{card} 
**Eksempel**
^^^
Bruk sfæriske koordinater til å regne ut volumet av en kule med radius $R$.

Volumet er trippelintegralet av $f=1$ over hele kula. I sfæriske koordinater dekker vi kula ved å la $r$ gå fra $0$ til $R$, $\theta$ fra $0$ til $\pi$, og $\phi$ fra $0$ til $2\pi$. Vi bruker $dV=r^2\sin{\theta}\,dr\,d\theta\,d\phi$:

$V = \int_0^{2\pi}\int_0^{\pi}\int_0^{R} r^2\sin{\theta}\,dr\,d\theta\,d\phi$

De tre integralene kan skilles fra hverandre siden faktorene er uavhengige:

$V = \left(\int_0^{R} r^2\,dr\right)\left(\int_0^{\pi}\sin{\theta}\,d\theta\right)\left(\int_0^{2\pi}d\phi\right)$

Vi regner ut hver for seg:

$\int_0^{R} r^2\,dr = \frac{R^3}{3}, \qquad \int_0^{\pi}\sin{\theta}\,d\theta = \Big[-\cos{\theta}\Big]_0^{\pi} = 2, \qquad \int_0^{2\pi}d\phi = 2\pi$

Til sammen: $V = \dfrac{R^3}{3}\cdot 2 \cdot 2\pi = \dfrac{4}{3}\pi R^3$, som er den velkjente formelen for kulevolumet.
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Her kommer selve grunnen til at kjemistudenter trenger trippelintegraler i sfæriske koordinater: **normalisering av orbitaler**. En orbital $\psi$ er en bølgefunksjon, og $|\psi|^2$ er sannsynlighetstettheten for å finne elektronet. Siden elektronet må være *et eller annet sted*, må sannsynligheten summert over hele rommet bli nøyaktig $1$:

$\iiint_{\text{hele rommet}} |\psi|^2 \, dV = 1$

Ta hydrogenets 1s-orbital, som er kulesymmetrisk og bare avhenger av $r$:

$\psi_{1s} = \frac{1}{\sqrt{\pi a_0^3}}\, e^{-r/a_0}$

der $a_0$ er Bohr-radien. Vi sjekker normaliseringen i sfæriske koordinater. Siden $|\psi|^2$ ikke avhenger av vinklene, skiller integralene seg pent fra hverandre:

$\iiint |\psi_{1s}|^2 \, dV = \frac{1}{\pi a_0^3}\left(\int_0^{\infty} e^{-2r/a_0}\,r^2\,dr\right)\left(\int_0^{\pi}\sin{\theta}\,d\theta\right)\left(\int_0^{2\pi}d\phi\right)$

Vinkelintegralene kjenner vi igjen fra kule-eksempelet: de gir $2$ og $2\pi$. Radialintegralet er et standardintegral (som kan løses med delvis integrasjon fra 4.5, brukt to ganger):

$\int_0^{\infty} r^2 e^{-2r/a_0}\,dr = \frac{a_0^3}{4}$

Setter vi alt sammen:

$\frac{1}{\pi a_0^3}\cdot \frac{a_0^3}{4}\cdot 2 \cdot 2\pi = \frac{1}{\pi a_0^3}\cdot \pi a_0^3 = 1$

Integralet blir nøyaktig $1$ - orbitalen er riktig normalisert. Legg merke til hvor avgjørende faktoren $r^2$ fra volumelementet var: uten den ville integralet ikke gått opp. Nettopp derfor er $r^2\sin\theta$ ikke en teknisk detalj, men helt sentralt i kvantekjemien.
:::

Med dobbel- og trippelintegraler, og sfæriske koordinater, har vi fått verktøyene til å integrere over flater og romområder - og til å normalisere orbitaler, som er uunnværlig i kvantekjemien. Dette runder av kalkulusdelen: fra funksjoner og grenser, via derivasjon og integrasjon i én variabel, til flervariabel derivasjon og integrasjon.

b) Bruk et trippelintegral til å finne volumet av boksen $[0,2]\times[0,3]\times[0,1]$.

## 5.4 Lagrange multiplikatormetode

I [4.4.3 Ekstremalpunkter](https://andreasdh.github.io/Bifrost/docs/matematikk/kalkulus.html#ekstremalpunkter) så vi på topp-og bunnpunkter i funksjoner av én variabel, som for eksempel å maksimere et utbytte, minimere en energi. Vi skal fortsette å finne minimum og/eller maksimum i dette delkapittelet, men her er det to ting som er annerledes enn tidligere:

1. Funksjonen vår er av flere variabler 
2. Vi optimerer under én eller flere betingelser

Det første er kanskje opplagt siden i i hele kapittelet har sett på funksjoner av flere variabler, men punkt nummer 2 er nytt. I praksis er vi sjelden helt frie: vi optimerer under en **betingelse**. En kjemiker vil kanskje maksimere entropien, men bare blant tilstander som har en gitt total energi. Vi søker altså ekstremalverdien til en funksjon $f(x,y)$, men bare blant de punktene $(x,y)$ som oppfyller en betingelse på formen $g(x,y)=c$. **Lagranges multiplikatormetode** er et elegant verktøy for nettopp dette, og bygger direkte på gradienten fra delkapittel 4.7.

La oss tenke geometrisk. Betingelsen $g(x,y)=c$ beskriver en kurve i planet - den tillatte "stien" vi kan bevege oss langs. Funksjonen $f$ har sine egne nivåkurver, én for hver verdi $f$ kan ta. Når vi går langs betingelseskurven, krysser vi stadig nye nivåkurver for $f$, og verdien til $f$ endrer seg. Så lenge betingelseskurven *krysser* en nivåkurve, kan vi fortsette å gå og finne en enda høyere (eller lavere) verdi. Ekstremalverdien inntreffer først i det øyeblikket betingelseskurven bare *tangerer* en nivåkurve - da har vi presset $f$ så langt vi kan uten å forlate stien.

```python
vis_sim("../Simuleringer/lagrange_multiplikator.html")
```

I dette tangeringspunktet peker de to kurvenes normaler i samme retning. Fra [5.2.3 Gradienten](https://andreasdh.github.io/Bifrost/docs/matematikk/flervariabel_kalkulus.html#gradienten) vet vi at gradienten står vinkelrett på nivåkurvene. Både $\nabla f$ (normal på $f$ sin nivåkurve) og $\nabla g$ (normal på betingelseskurven) må derfor peke langs samme linje - de er parallelle. To vektorer er parallelle nøyaktig når den ene er et tall ganget med den andre, og det er hele kjernen i metoden.

:::{admonition} Teorem 5.2 - Lagranges multiplikatormetode
:class: note

La $f(x,y)$ og $g(x,y)$ være deriverbare funksjoner. Dersom $f$ har en ekstremalverdi i et punkt $(x,y)$ under betingelsen $g(x,y)=c$, så finnes det et tall $\lambda$ slik at

$\nabla f = \lambda \nabla g$

Tallet $\lambda$ kalles en **Lagrange-multiplikator**.
:::

Skriver vi ut vektorligningen $\nabla f = \lambda \nabla g$ komponent for komponent, og tar med selve betingelsen, får vi et system på tre ligninger med tre ukjente ($x$, $y$ og $\lambda$):

$\frac{\partial f}{\partial x} = \lambda \frac{\partial g}{\partial x}$

$\frac{\partial f}{\partial y} = \lambda \frac{\partial g}{\partial y}$

$g(x,y) = c$

Oppskriften blir dermed: regn ut de fire partiellderiverte, sett opp de tre ligningene, og løs systemet. Punktene vi finner er kandidatene til ekstremalverdi. Multiplikatoren $\lambda$ er som regel bare et mellomledd vi ikke bryr oss om til slutt - men i fysikk og kjemi viser den seg ofte å ha en dyp betydning helt selv.

:::{card} 
**Eksempel**
^^^
Finn det største mulige produktet $f(x,y)=xy$ når $x+y=10$.

Her er betingelsen $g(x,y)=x+y$ med $c=10$. Vi regner ut de partiellderiverte:

$\frac{\partial f}{\partial x}=y, \quad \frac{\partial f}{\partial y}=x, \quad \frac{\partial g}{\partial x}=1, \quad \frac{\partial g}{\partial y}=1$

Ligningssystemet $\nabla f = \lambda \nabla g$ sammen med betingelsen blir

$y = \lambda \cdot 1$

$x = \lambda \cdot 1$

$x+y=10$

Fra de to første ligningene ser vi at $x=\lambda$ og $y=\lambda$, altså $x=y$. Setter vi dette inn i betingelsen, får vi $x+x=10$, så $x=5$ og dermed $y=5$.

Det største produktet er derfor $f(5,5)=5 \cdot 5 = 25$.
:::

:::{card} 
**Eksempel**
^^^
Et rektangel skal ha omkrets $20$ cm. Hvilke sidelengder gir størst areal?

La sidene være $x$ og $y$. Vi maksimerer arealet $f(x,y)=xy$ under betingelsen $g(x,y)=2x+2y=20$.

De partiellderiverte er

$\frac{\partial f}{\partial x}=y, \quad \frac{\partial f}{\partial y}=x, \quad \frac{\partial g}{\partial x}=2, \quad \frac{\partial g}{\partial y}=2$

Ligningssystemet blir

$y = 2\lambda$

$x = 2\lambda$

$2x+2y=20$

De to første gir $x=y$. Setter vi inn i betingelsen: $2x+2x=20$, altså $x=5$ og $y=5$.

Rektangelet med størst areal er altså et **kvadrat** med sider $5$ cm og areal $25$ cm². Dette bekrefter intuisjonen fra introduksjonsspørsmålet - det jevneste valget er det beste.
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Lagranges metode er ikke bare et regnetriks - den er selve fundamentet for [Boltzmann-fordelingen](https://andreasdh.github.io/Bifrost/docs/statmek.html#boltzmanns-fordelingslov).

Tenk deg et system med partikler fordelt på energinivåer $\varepsilon_1, \varepsilon_2, \ldots$, med $n_i$ partikler på nivå $i$. Antall mikrotilstander som gir en bestemt fordeling er $W=\dfrac{N!}{n_1!\,n_2!\cdots}$. Naturen søker den fordelingen som maksimerer $W$ - eller, mer bekvemt, $\ln{W}$. Men vi kan ikke maksimere fritt: to betingelser må holdes fast.

$\sum_i n_i = N \qquad \text{(partikkeltallet er bevart)}$

$\sum_i n_i \varepsilon_i = E \qquad \text{(den totale energien er bevart)}$

Dette er nøyaktig et Lagrange-problem, bare med *to* betingelser i stedet for én - da trenger vi to multiplikatorer, som vi kaller $\alpha$ og $\beta$. Vi maksimerer altså

$\ln{W} - \alpha\left(\sum_i n_i - N\right) - \beta\left(\sum_i n_i \varepsilon_i - E\right)$

Deriverer vi med hensyn på hver $n_i$ og setter lik null (samme prinsipp som $\nabla f = \lambda \nabla g$), faller det ut at

$n_i \propto e^{-\beta \varepsilon_i}$

Dette er Boltzmann-fordelingen, og multiplikatoren $\beta$ - det "uinteressante mellomleddet" - viser seg å være $\beta=\dfrac{1}{k_B T}$. Slik dukker temperaturen opp helt av seg selv, som en Lagrange-multiplikator knyttet til energibevaring. Metoden du nettopp lærte på $f=xy$ er altså den samme som gir oss fordelingen bak all statistisk termodynamikk.
:::

Med Lagranges multiplikatormetode har vi rundet av kalkulusdelen. Vi har gått fra funksjoner av én variabel, via grenser, derivasjon og integrasjon, til funksjoner av flere variabler, partiellderiverte, flerdimensjonal integrasjon og til slutt optimering under betingelser. Underveis har vi sett hvordan hvert verktøy har en direkte anvendelse i kjemien - fra termodynamisk arbeid og normalisering av orbitaler til selve Boltzmann-fordelingen som styrer hvordan naturen fordeler energi.

$3 = 2\lambda y$

Største verdi er altså $13$ i punktet $(2,3)$, og minste verdi er $-13$ i punktet $(-2,-3)$.
