# 2. Kombinatorikk

Før vi giver løs på det sentrale tema sannsynlighet, skal vi se på ulike tellemåter for å si noe om antallet muligheter noe kan skje på. Dette er selve limet i statistisk termodynamikk: å telle antall mikrotilstander som gir samme makrotilstand ([multiplisiteten $\Omega$](https://andreasdh.github.io/Bifrost/docs/statmek.html#multiplisitet)) er nøyaktig et kombinatorikkproblem. Dette tallet dukker senere opp direkte i [Boltzmanns entropiformel](https://andreasdh.github.io/Bifrost/docs/entropi.html#boltzmanns-entropi), $S = k_B \ln \Omega$ - så verktøyene du lærer her er ikke bare telleteknikker, de er selve grunnlaget for hvordan vi definerer entropi.

## 2.1 OG-regelen og ELLER-regelen

OG- og ELLER-regelen er regler du mest sannsynlig har brukt før, enten bevisst eller ubevisst. De handler om å finne antall måter å organisere objekter på i to helt ulike situasjoner: OG-regelen brukes når vi gjør flere valg *etter hverandre*, mens ELLER-regelen brukes når vi velger *mellom* alternativer. Nøkkelen til å telle riktig er som regel å avgjøre hvilken av de to situasjonene man står i.

:::{admonition} Teorem 2.1 - OG-regelen
:class: note

Anta at en hendelse $A$ kan skje på $m$ måter og at hendelse $B$ kan skje uavhengig på $n$ måter. Da er antallet måter $A$ OG $B$ kan skje lik 

$m \cdot n$.
:::

:::{admonition} Teorem 2.2 - ELLER-regelen
:class: note

Anta at en hendelse $A$ kan skje på $m$ ulike måter, og hendelse $B$ kan skje på $n$ måter. Anta også at hendelsene er disjunkte. Da er antall måter hendelse $A$ ELLER $B$ skje på lik

$m + n$ 
:::

Legg merke til betingelsene i de to reglene. OG-regelen krever at valgene er uavhengige - antall muligheter for $B$ må ikke endre seg ut fra hva vi valgte for $A$. ELLER-regelen krever at hendelsene er disjunkte, altså at ingen mulighet telles i begge.

En grei huskeregel: skal du gjøre begge deler, ganger du. Skal du gjøre det ene eller det andre, legger du sammen.

:::{admonition} Kobling til fysikalsk kjemi
:class: important

OG-regelen er grunnen til at antall mikrotilstander vokser så voldsomt for store systemer. Har hver av $N$ partikler $k$ mulige tilstander, og partiklene er uavhengige, gir OG-regelen

$\Omega = k \cdot k \cdot \ldots \cdot k = k^N$

For ett mol partikler er $N \approx 6 \cdot 10^{23}$, og $\Omega$ blir da et tall så stort at det er umulig å skrive ut. Nettopp derfor tar vi logaritmen i [Boltzmanns entropiformel](https://andreasdh.github.io/Bifrost/docs/entropi.html#boltzmanns-entropi) $S = k_B \ln \Omega$ - logaritmen gjør de enorme produktene om til håndterbare summer.
:::

:::{card} 
**Eksempel**
^^^
Et gassmolekyl kan innta 4 ulike vibrasjonstilstander og 3 ulike rotasjonstilstander, og disse frihetsgradene er uavhengige av hverandre.

a) Hvor mange mikrotilstander der molekylet har en bestemt kombinasjon av vibrasjon og rotasjon finnes?

Antall mikrotilstander der molekylet har en bestemt kombinasjon av vibrasjon OG rotasjon blir dermed

$4 \cdot 3=12$

b) Hvor mange mikrotilstander der molekylet befinner seg i én bestemt virbasjonstilstand eller én bestemt rotasjonstilstand finnes?

Antallet mikrotilstander der molekylet befinner seg i én bestemt vibrasjonstilstand ELLER én bestemt rotasjonstilstand (disjunkte tilstander) blir

$4+3=7$
:::


:::{card} 
**Eksempel**
^^^
En kantine tilbyr $3$ varmretter, $4$ salater og $2$ desserter.

a) Hvor mange måter kan du sette sammen et måltid med én varmrett, én salat og én dessert?

Her tar du ett valg fra hver kategori, altså varmrett OG salat OG dessert. Ved OG-regelen ganger vi:

$3 \cdot 4 \cdot 2 = 24$

b) Hvor mange måter kan du velge én enkelt rett, uansett kategori?

Nå velger du enten en varmrett ELLER en salat ELLER en dessert. Ingen rett er i to kategorier, så hendelsene er disjunkte, og ved ELLER-regelen legger vi sammen:

$3 + 4 + 2 = 9$

Legg merke til hvor mye større svaret i a) er. Det illustrerer hvorfor antall muligheter vokser så raskt når valg kombineres.

## 2.2 Fakultet

Vi skal se at fakultet er svaret på et helt konkret spørsmål: på hvor mange måter kan vi stille opp $n$ forskjellige objekter i rekkefølge? Dette dukker opp overalt - fra hvor mange måter man kan stokke en kortstokk, til hvor mange måter partikler kan fordeles på energinivåer. I tillegg dukker fakultet opp i andre tellemåter vi skal se i [delkapittel 2.3](https://andreasdh.github.io/Bifrost/docs/matematikk/kombinatorikk.html#tellemetoder).

:::{admonition} Def. 2.1 - Fakultet
:class: warning

For et ikke-negativt heltall $n$, er fakultet definert slik

$n! = n \cdot (n-1) \cdot (n-2) \cdot \dots \cdot 1$ 

$0!$ er ved definisjon lik $1$.
:::

Vi har $n$-antall objekter som vi skal plassere i rekke - hvor mange ulike måter kan vi gjøre dette på? Første objekt har $n$-antall muligheter. Deretter har andre objekt $n-1$-antall muligheter. Slik følger det videre, helt til objekt $n$ kun har én mulighet. Ved bruk av OG-regelen fra tidligere, multipliseres alle mulighetene og gir opphav til fakultet.

```python
vis_sim("../Simuleringer/fakultet.html")
```

Fakultet av store tall kan ikke regnes ut direkte, verken for hånd eller på datamaskin. I statistisk termodynamikk, der vi opererer med $N \approx 10^{23}$ partikler, trenger vi derfor tilnærminger - vi kommer tilbake til dette i kjemi-koblingen under.

:::{card} 
**Eksempel**
^^^
Fire ulike molekyler skal plasseres på fire nummererte plasser i et krystallgitter, ett molekyl per plass.

a) På hvor mange måter kan dette gjøres?

Første plass kan fylles av hvilket som helst av de $4$ molekylene. Når det er gjort, er det $3$ igjen til andre plass, så $2$ til tredje, og til slutt $1$ molekyl til siste plass. Ved OG-regelen ganger vi mulighetene:

$4 \cdot 3 \cdot 2 \cdot 1 = 4! = 24$

Det finnes altså $24$ ulike måter.

b) Hva om ett av molekylene må stå på plass nummer 1?

Da er den plassen låst, og vi står igjen med $3$ molekyler som skal fordeles på $3$ plasser:

$3! = 6$ måter.
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Fakultet er selve byggesteinen i uttrykket for antall mikrotilstander. Skal $N$ partikler fordeles slik at $n_1$ havner på energinivå $1$, $n_2$ på nivå $2$ og så videre, er antall måter dette kan skje på

$\Omega = \dfrac{N!}{n_1! \, n_2! \cdots}$

Dette er [multiplisiteten](https://andreasdh.github.io/Bifrost/docs/statmek.html#multiplisitet), og den går rett inn i [Boltzmanns entropiformel](https://andreasdh.github.io/Bifrost/docs/entropi.html#boltzmanns-entropi) $S=k_B \ln \Omega$. Ikke vær redd for om denne virker komplisert, den kommer vi nemlig tilbake til i [delkapittel 2.3](https://andreasdh.github.io/Bifrost/docs/matematikk/kombinatorikk.html#tellemetoder).

Men her møter vi problemet fra avsnittet over: med $N \approx 10^{23}$ er $N!$ umulig å regne ut fordi tallet er så stort. Løsningen på problemet heter Stirlings tilnøærming og presenteres i [delkapittel 2.4](https://andreasdh.github.io/Bifrost/docs/matematikk/kombinatorikk.html#stirlings-tilnaerming).
:::

:::{admonition} Underveisoppgave 2.2
:class: tip

a) Regn ut $5!$ og $6!$. Hvor mange ganger større er $6!$ enn $5!$?

b) Fem ulike molekyler skal plasseres på fem plasser i et gitter. På hvor mange måter kan det gjøres?

c) Forklar hvorfor $\dfrac{8!}{6!} = 8 \cdot 7$, uten å regne ut hver fakultet for seg.

d) Tre atomer skal plasseres i rekke, men to av dem er identiske. Blir antall synlig forskjellige rekkefølger flere eller færre enn $3!$? Begrunn.
:::

:::{admonition} Løsning 2.2
:class: tip, dropdown

a) $5!=120$ og $6!=720$. Forholdet er $\frac{720}{120}=6$. Dette er ingen tilfeldighet: $6!=6 \cdot 5!$, så hvert steg opp ganger med det nye tallet.

b) Dette er antall måter å ordne $5$ ulike objekter i rekkefølge:

$5! = 5 \cdot 4 \cdot 3 \cdot 2 \cdot 1 = 120$ måter.

c) Vi skriver ut telleren:

$\frac{8!}{6!}=\frac{8 \cdot 7 \cdot 6 \cdot 5 \cdot 4 \cdot 3 \cdot 2 \cdot 1}{6 \cdot 5 \cdot 4 \cdot 3 \cdot 2 \cdot 1}$

Alt fra $6$ og nedover står i både teller og nevner, og forkortes bort. Igjen står $8 \cdot 7 = 56$. Dette trikset er verdt å merke seg - det er nettopp slik vi håndterer store fakulteter i praksis, og det brukes flittig i neste delkapittel.

d) Færre. Med tre *ulike* atomer finnes $3!=6$ rekkefølger. Er to av dem identiske, kan vi ikke se forskjell på de to rekkefølgene der de bytter plass med hverandre, så hver synlig rekkefølge telles dobbelt. Antallet blir da $\frac{3!}{2!}=\frac{6}{2}=3$.

Dette er nøyaktig samme idé som nevnerne i formelen for $\Omega$: vi deler bort de omstokkingene vi ikke kan skille fra hverandre.
:::

## 2.3 Tellemetoder

Før vi begynner å telle antall muligheter må vi vite hva vi har med å gjøre. Det er spesielt to egenskaper ved objektene vi teller vi må vite noe om:

**1) Gjentakelse**

Kan hvert objekt/plass brukes én gang, eller flere ganger? Vi sier at vi enten teller **med gjentakelse** eller **uten gjentakelse**.

**2) Rekkefølge**

Har rekkefølgen til objektene noe å si (er de skjelnbare), eller ikke? Vi sier at utvalget enten er **ordnet** hvis rekkefølgen har noe å si, eller **uordnet** ved motsatt tilfelle.

Alt i alt, gir dette fire ulike tellemåter!

Det kan være vanskelig å forstå hva vi mener med gjentakelse og rekkefølge, men de er lettere forstått ved noen eksempler:

:::{card} 
**Eksempel**
^^^

**Ordnet utvalg, med gjentakelse**

Du har en kodelås med 4 sifferplasser og hvert 10 valg av tall for hver plass (0-9). Alle 10 tallene kan velges ved hver plass, og derfor er dette et utvalg med gjentakelse. Du kan for eksempel velge "0000". Rekkefølgen betyr noe fordi koden "1234" er en annen enn "2314", selv om samme tall er brukt i begge kodene. Derfor er dette et ordnet utvalg.


**Ordnet utvalg, uten gjentakelse**

En klubb med 10 medlemmer skal velge leder, nestleder og økonomisjef. Et medlem kan maksimalt ha ett verv, så hvor mange muligheter har vi? Utvalget er uten gjentakelse, én person kan kun brukes én gang per verv. Rekkefølgen betyr noe fordi at Per er leder, Pål er nestleder og Askeladden er økonomisjef, er en annen ordning enn at Pål er leder, Askeladden er nestleder og Per er økonomisjef. Derfor er det et ordnet utvalg.


**Uordnet utvalg, uten gjentakelse**

Utvalget er uordnet fordi rekkefølgen på smakene ikke har noe å si. Når du skal telle antall muligheter er jordbær, sjokolade og vanlije, nøyaktig samme mulighet som sjokolade, vanilje og jordbær.


**Uordnet utvalg, med gjentakelse**

Utvalget er med gjentakelse fordi du kan velge en type bolle flere ganger. Rekkefølgen har ikke noe å si, bare totalen du har. For eksempel er sjokolade, vanilje, kanel og vanilje, nøyaktig samme mulighet som sjokolade, kanel, vanilje og vanilje. Derfor er utvalget uordnet.

Nå har vi en oversikt over de ulike tellemåtene, så hvordan regner man ut? Vi bruker de samme eksemplene for å vise. 

Det kan være vanskelig å forstå hva vi mener med gjentakelse og rekkefølge, men de er lettere forstått ved noen eksempler:

:::{card} 
**Eksempel**
^^^

**Ordnet utvalg, med gjentakelse**

På sifferplass 1 har du 10 valg av tall, det har du også ved sifferplass 2, 3 og 4. Ved [OG-regelen](https://andreasdh.github.io/Bifrost/docs/matematikk/kombinatorikk.html#og-regelen-og-eller-regelen) fra delkapittel 2.1 får vi dermed:

$10 \cdot 10 \cdot 10 \cdot 10 = 10^4 = 10000 \text{ muligheter}$


**Ordnet utvalg, uten gjentakelse**

Du har 10 medlemmer og ser på hvordan man kan fylle 3 ulike plasser. Her kommer [fakultet](https://andreasdh.github.io/Bifrost/docs/matematikk/kombinatorikk.html#fakultet) fra delkapittel 2.2 inn. På første plass har du $10$ valg av medlemmer, på andre plass har du $9$ valg av medlemmer siden ett har blitt valgt, og til slutt har du $8$ valg av medlemmer. Dette gir

$10 \cdot 9 \cdot 8 = 720 \text{ muligheter}$

Merk at dette er det samme som

$\frac{10 \cdot 9 \cdot 8 \cdot 7 \cdot 6 \cdot 5 \cdot 4 \cdot 3 \cdot 2 \cdot 1}{7 \cdot 6 \cdot 5 \cdot 4 \cdot 3 \cdot 2 \cdot 1} = \frac{10!}{7!}=\frac{10!}{(10-3)!}$

Dette blir viktig senere.


**Uordnet utvalg, uten gjentakelse**

Du ser på antall måter å velge ut 3 ulike iskremsmaker av totalt 10 mulige. Siden du velger ulike smaker er utvalget uten gjentakelse - du kan ikke ha sjokolade, sjokolade og jordbær, eller sjokolade, sjokolade og sjokolade. Utvalget er uordnet fordi rekkefølgen på smakene ikke har noe å si. Når du skal telle antall muligheter er jordbær, sjokolade og vanlije, nøyaktig samme mulighet som sjokolade, vanilje og jordbær.

Denne situasjonen skiller seg kun fra den forrige i det at vi ikke skal overtelle ulike kombinasjoner. Tidligere hadde rekkefølgen noe å si, det har den ikke nå lenger og vi må kompensere for det på noe måte.

Som i forrige oppgave får vi

$\frac{10!}{(10-3)!} \text{ muligheter}$

hvis rekkefølgen har noe å si. Det har den ikke nå! Vi har flere muligheter av 3 smaker som vi ser på som identiske, for eksempel jordbær, sjokolade og vanlije, nøyaktig samme mulighet som sjokolade, vanilje og jordbær. 

Hvor mange muligheter gir samme identiske kombinasjon? For eksempel kan smakene jordbær, vanilje og sjokolade organiseres med

$3 \cdot 2 \cdot 1 = 3! = 6 \text{ muligheter}$

Ved å dele på dette tallet i formelen som overteller, får vi det korrekte antallet, og dermed

$\frac{10!}{3! \cdot (10-3)!} = 120 \text{ muligheter}$



**Uordnet utvalg, med gjentakelse**

Dette er det mest kompliserte av alle de fire problemene, og løses gjennom "stjerner og streker"-metoden. I vårt tilfelle er bollene stjerner, og vi har $4-1=3$ vegger som tilsvarer til typene.

Du skal kjøpe 6 boller fra et bakeri og de har 4 typer. Du kan velge flere av type. Hvor mange ulike måter kan du velge boller på? Utvalget er med gjentakelse fordi du kan velge en type bolle flere ganger. Rekkefølgen har ikke noe å si, bare totalen du har. For eksempel er sjokolade, vanilje, kanel og vanilje, nøyaktig samme mulighet som sjokolade, kanel, vanilje og vanilje. Derfor er utvalget uordnet.

For eksempel betyr

⋆⋆ ∣ ⋆ ∣   ∣ ⋆⋆⋆

2 boller av sort 1, 1 bolle av sort 2, ingen boller av sort 3 og 3 boller av sort 3. Vi kan også ha 

⋆⋆⋆⋆⋆⋆ ∣   ∣   ∣ 

som betyr 6 boller av sort 1, og ingen av sort 2, 3 eller 4. 

Hvor mange måter kan vi ordne disse stjernene blant veggene? Spørsmålet brenner ned til at vi har totalt $6+3=9$ objekter (plasser), og skal velge 3 av de til stjernene. Dette blir nøyaktig det samme som vi gjorde i forrige oppgave, og vi får

$\frac{9!}{6! \cdot (9-6)!} = \frac{(6+4-1)!}{6! \cdot ((6+4-1)-6)!}$
:::

Vi skal nå generalisere dette og få noen ordentlige formler å jobbe med.

:::{admonition} Def. 2.2 - Permutasjon
:class: warning

Antall måter å velge $k$ elementer fra en mengde med $n$ elementer, uten gjentakelse og der rekkefølgen betyr noe er

$P(n,k) = \frac{n!}{(n-k)!}$

Hver slik måte kalles en **permutasjon**.
:::

:::{admonition} Def. 2.3 - Kombinasjon
:class: warning

Antall måter å velge $k$ elementer fra en mengde med $n$ elementer, uten gjentakelse og der rekkefølgen *ikke* betyr noe er

$\binom{n}{k} = \frac{n!}{k! \cdot (n-k)!}$

Hver slik måte kalles en **kombinasjon**, og tallet kalles **binomialkoeffisienten**.
:::

:::{admonition} Teorem 2.3 - Ordninger i et ordnet utvalg, med gjentakelse
:class: note

Antall måter å velge $k$ elementer fra en mengde med $n$ elementer, med gjentakelse og der rekkefølgen betyr noe er

$n^k$
:::

:::{admonition} Teorem 2.4 - Ordninger i et uordnet utvalg, med gjentakelse
:class: note

Antall måter å velge $k$ elementer fra en mengde med $n$ elementer, med gjentakelse og der rekkefølgen *ikke* betyr noe er

$\binom{n+k-1}{k}\frac{(n+k-1)!}{k! \cdot (n-1)!}$

:::

Vi følger på med flere eksempler!

:::{card} 
**Eksempler**
^^^
Et tripeptid bygges av tre aminosyrer i rekkefølge, valgt blant de 20 vanlige. Samme aminosyre kan brukes flere ganger, og rekkefølgen betyr noe. Hvor mange ulike tripeptider finnes? Dette er et ordnet utvalg med gjentakelse, slik at vi får:

$n^k = 20^3 = 8000$

Fem ulike ligander skal binde seg til et metallion, men bare tre bindingsplasser er ledige, og rekkefølgen de bindes i gir ulike mellomprodukter. Hvor mange rekkefølger finnes? Her velger vi uten gjentakelse og rekkefølgen teller, slik at vi får:

$P(5,3)=\frac{5!}{(5-3)!}=\frac{5!}{2!}=\frac{120}{2}=60$

En benzenring har seks posisjoner, og vi skal sette to like substituenter på to av dem. Rekkefølgen er likegyldig - de to substituentene er identiske. Vi får:

$\binom{6}{2}=\frac{6!}{2! \cdot 4!}=\frac{720}{2 \cdot 24}=15$

Fire like energikvanter skal fordeles på tre identiske oscillatorer. Kvantene kan ikke skilles fra hverandre (gjentakelse tillatt), og rekkefølgen betyr ingenting - bare hvor mange hver oscillator får. Vi får dermed:

$\binom{n+k-1}{k}=\binom{3+4-1}{4}=\binom{6}{4}=15$
:::

## 2.4 Stirlings tilnærming

**Stirlings tilnærming** brukes til å estimere fakultet av svært store tall. Vi skal se på hvordan vi kan estimere den naturlige logaritmen til fakultetet av et stort tall $n$, $\ln{n!}$, som blir nyttig i [Boltzmanns entropiformel](https://andreasdh.github.io/Bifrost/docs/entropi.html#boltzmanns-entropi).

La $n$ være et stort tall. Fra definisjonen av fakultet ([Def. 2.1](https://andreasdh.github.io/Bifrost/docs/matematikk/kombinatorikk.html#fakultet)) har vi

$n!=n \cdot (n-1) \cdot (n-2) \cdot \dots \cdot 1$

Vi kan ta den naturlige logartimen på begge sider, slik at vi får

$\ln{n!}=\ln{\left( n \cdot (n-1) \cdot (n-2) \cdot \dots \cdot 1 \right)}$

Ved [logartimereglene](https://andreasdh.github.io/Bifrost/docs/matematikk/formler_og_enheter.html#eksponenter-logaritmer-og-rotter) får vi

$\ln{n!}=\ln{(n)}+\ln{(n-1)}+\ln{(n-2)}+\dots+\ln{(1)}=\sum_{x=1}^{n} \ln{(x)}$

Vi kan approksimere denne summen til integralet av $\ln{(x)}$ fra $x=1$ til $x=n$, slik at

$\ln{n!}=\sum_{x=1}^{n} \ln{(x)}\approx \int_{1}^{n} \ln{(x)} \ dx$

Mellomregning:

$\int_{1}^{n} \ln{(x)} \ dx=\int_{1}^{n} 1 \cdot \ln{(x)} \ dx$

Vi bruker [delvis integrasjon](https://andreasdh.github.io/Bifrost/docs/matematikk/kalkulus.html#integrasjonsregler) ved å la 

$u=\ln{(x)}$, $v'=1$ $\Rightarrow$ $u'=\frac{1}{x}$, $v=x$

og får

$\int_{1}^{n} 1 \cdot \ln{(x)} \ dx = [x \ln{(x)}]_{1}^{n}-\int_{1}^{n} \left( \frac{1}{x} \cdot x \right) \  dx=[x \ln{(x)}]_{1}^{n}-\int_{1}^{n} \  dx=[x \ln{(x)} -x]_{1}^{n}$

Altså, har vi fått Stirlings tilnærming

$\ln{n!}\approx [x \ln{(x)} -x]_{1}^{n}$

:::{admonition} Teorem 2.5 - Stirlings tilnærming
:class: note

For store $n$ kan $\ln{n!}$ tilnærmes ved å erstatte summen $\sum_{k=1}^{n}\ln{k}$ med et integral:

$\ln{n!}=\sum_{k=1}^{n}\ln{k} \approx \int_1^n \ln{x}\,dx = \big[x\ln{x}-x\big]_1^n = n\ln{n}-n+1$

Siden $1$-leddet er neglisjerbart for store $n$, skrives resultatet vanligvis som

$\ln{n!} \approx n\ln{n}-n$
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Stirlings tilnærming er uunnværlig i statistisk termodynamikk. Antall mikrotilstander for et system er gitt ved uttrykk som

$\Omega = \frac{N!}{n_1!\,n_2!\cdots}$

og entropien er $S=k_B\ln{\Omega}$. For ett mol partikler er $N\approx 6\times 10^{23}$, og da er $N!$ et tall så enormt at det er fullstendig uregnbart - verken for hånd eller på datamaskin.

Men vi trenger aldri $N!$ selv, bare $\ln{N!}$, fordi entropien inneholder en logaritme. Der kommer Stirling til unnsetning:

$\ln{\Omega}=\ln{N!}-\sum_i \ln{n_i!} \approx \left(N\ln{N}-N\right)-\sum_i\left(n_i\ln{n_i}-n_i\right)$

Dette gjør uttrykket både regnbart og deriverbart. Nettopp fordi $\ln{\Omega}$ nå er en glatt funksjon, kan vi maksimere den med [Lagranges multiplikatormetode]() - og det er dét som gir Boltzmann-fordelingen.
