# Jupyter Book / MyST elements

This notebook is a compact reference for graphical and structural elements in **Jupyter Book 1.x** using **MyST Markdown**.

:::{note}
This version is updated for the Jupyter Book v1 / Sphinx-based toolchain. In particular, old `panels` syntax has been replaced by `grid`, `grid-item-card`, and `card`, and old `tabbed` syntax has been replaced by `tab-set` and `tab-item`.
:::

## Fence syntax

MyST directives can be written with backtick fences or colon fences.

````md
```{note}
A directive with backtick fences.
```
````

or:

```md
:::{note}
A directive with colon fences.
:::
```

Colon fences are often easier to read in notebooks, especially when the directive contains Markdown. If you nest directives inside directives, increase the number of outer fence characters.

## What changed since the 2021 version?

```{list-table}
:header-rows: 1
:name: old-new-syntax

* - Old / less recommended
  - Updated Jupyter Book 1.x syntax
  - Comment
* - `{panels}`
  - `{grid}` + `{grid-item-card}` or `{card}`
  - `panels` came from `sphinx-panels`; Jupyter Book v1 now uses Sphinx Design components.
* - `{tabbed}`
  - `{tab-set}` + `{tab-item}`
  - The newer tab syntax is more explicit and comes from Sphinx Design.
* - Generic `{admonition}` with `:class: note`
  - Prefer `{note}`, `{tip}`, `{warning}`, etc. when the title can be standard.
  - Use `{admonition}` when you need a custom title.
* - Only raw Markdown tables
  - Use `{list-table}` when you need captions, labels, and cross-references.
  - Plain Markdown tables are still fine for simple tables.
```

You can refer to the table above with:

```md
{numref}`old-new-syntax`
```

## Cards

Use `{card}` for a single card with optional header and footer.

````{card} Computational thinking

Card header
^^^
Card body. Use this for short, high-visibility content such as a concept, a task, or a link collection.
+++
Card footer
````

Card headers and footers are optional.

````{tab-set}
  return 0;

```
```

end
```
```

````


```md

```

## Hiding and removing code cells

For real notebook code cells, use cell tags rather than wrapping everything in Markdown.

Common Jupyter Book tags:

```json
{
  "tags": ["hide-input"]
}
```

```json
{
  "tags": ["hide-output"]
}
```

```json
{
  "tags": ["hide-cell"]
}
```

```json
{
  "tags": ["remove-input"]
}
```

```json
{
  "tags": ["remove-output"]
}
```

```json
{
  "tags": ["remove-cell"]
}
```

Use `hide-*` when the reader should be able to reveal the content. Use `remove-*` when the content should not appear in the built book.

Because Jupyter Book 2.x is now the current PyPI default, pin the version if you want the Sphinx-based v1 toolchain:

```bash
pip install "jupyter-book<2"
```

