I forrige kapittel kom vi fram til Boltzmann-fordelingen

$$p_i=\frac{1}{Q}e^{-E_i/k_\mathrm{B}T}.$$

Fordelingen sier at lavenergimikrotilstander vanligvis er mer sannsynlige enn høyenergimikrotilstander. Men for å beregne faktiske sannsynligheter må vi vite hva $Q$ er. Det viser seg at den samme summen også fungerer som en «regnskapsbok» over mikrotilstandene og energiene deres. Når vi kjenner den, kan vi beregne flere makroskopiske størrelser uten å følge systemets bevegelse i detalj.

Dette kapitlet åpner stoffet fra uke 35. Før vi definerer $Q$, må vi være tydelige på hvilken fysisk situasjon sannsynlighetsfordelingen beskriver.


## Hvilke betingelser gjelder?

Tenk deg et stort antall identiske kopier av det samme systemet, for eksempel like gassbeholdere i kontakt med hvert sitt store vannbad ved samme temperatur. Kopiene har samme partikkeltall og volum, men fordi de kan utveksle energi med vannbadet, trenger de ikke å være i samme mikrotilstand eller ha nøyaktig samme energi i et bestemt øyeblikk.

En slik tenkt samling av kopier kalles et **ensemble**. Vi bruker ikke ensemblet fordi vi faktisk må bygge tusen beholdere. Det er et verktøy for å uttrykke hvor ofte de forskjellige mikrotilstandene opptrer, og dermed for å beregne gjennomsnittet vi forventer å måle for ett makroskopisk system over tid.

Her arbeider vi i det **kanoniske ensemblet**:

- partikkeltallet $N$ er fast
- volumet $V$ er fast
- systemet kan utveksle energi med et stort reservoar
- reservoaret holder temperaturen $T$ tilnærmet konstant

Energien til én bestemt kopi kan variere, men sannsynlighetsfordelingen og gjennomsnittsenergien er stabile ved likevekt.

:::{admonition} Hvorfor trenger vi dette?
:class: important
En enkelt mikrotilstand forteller hva systemet gjør akkurat nå. Ensemblet og sannsynlighetsfordelingen gjør det mulig å forutsi stabile, målbare størrelser som indre energi og varmekapasitet.
:::

:::{admonition} Underveisoppgave
:class: tip
En lukket gassbeholder står i et stort vannbad med konstant temperatur. Hvilke av størrelsene $N$, $V$, $T$ og $E$ er faste, og hvilken kan fluktuere?
:::

:::{admonition} Løsningsforslag
:class: tip dropdown
$N$, $V$ og $T$ er faste i den kanoniske modellen. Systemets energi $E$ kan variere når energi utveksles med vannbadet.
:::


## Partisjonsfunksjonen

På molekylnivå er energi kvantisert: Et atom eller molekyl kan bare ha bestemte tillatte energier. Hver fullstendige kvantetilstand er en mikrotilstand. Flere forskjellige kvantetilstander kan ha samme energi; da sier vi at energinivået er degenerert. En mer utfyllende innføring finnes i appendikset [Grunnleggende begreper](grunnleggende_begreper.ipynb).

Boltzmann-faktoren $e^{-E_i/k_\mathrm{B}T}$ er en **vekt**, ikke en ferdig sannsynlighet. For å få sannsynligheter som summerer til 1 må vi dele hver vekt på summen av alle vektene:

:::{admonition} Kanonisk partisjonsfunksjon
:class: important

$$Q=\sum_i e^{-E_i/k_\mathrm{B}T}.$$

Dermed er

$$p_i=\frac{e^{-E_i/k_\mathrm{B}T}}{Q}.$$
:::

For ett molekyl eller én partikkel bruker vi ofte liten bokstav:

$$q=\sum_n e^{-\varepsilon_n/k_\mathrm{B}T},$$

der $\varepsilon_n$ er energien til partikkelens mikrotilstand $n$. Stor $Q$ brukes for mikrotilstandene til hele systemet.

Partisjonsfunksjonen summerer **unormaliserte Boltzmann-vekter**. Den er ikke i seg selv en sannsynlighet, og er heller ikke generelt et bokstavelig antall tilgjengelige tilstander.

Hvis flere mikrotilstander har samme energi $E_j$, kan vi gruppere dem. Dersom degenerasjonen er $g_j$, får vi

$$Q=\sum_j g_j e^{-E_j/k_\mathrm{B}T}.$$

Her konkurrerer energi og degenerasjon: Høy energi reduserer Boltzmann-vekten, mens mange mikrotilstander med samme energi øker det samlede bidraget fra energinivået.

## Gibbs-entropi

Boltzmanns uttrykk $S=k_\mathrm{B}\ln\Omega$ gjelder direkte når de $\Omega$ tilgjengelige mikrotilstandene behandles som like sannsynlige. I det kanoniske ensemblet har mikrotilstandene derimot forskjellige sannsynligheter. Vi trenger derfor en entropidefinisjon som bruker hele sannsynlighetsfordelingen:

:::{admonition} Gibbs-entropi
:class: important

$$S=-k_\mathrm{B}\sum_i p_i\ln p_i.$$
:::

Hvert ledd vekter informasjonsbidraget $-\ln p_i$ med sannsynligheten for den aktuelle mikrotilstanden. En svært konsentrert fordeling, der én mikrotilstand har sannsynlighet nær 1, har lav entropi. En fordeling som er spredt over mange mikrotilstander, har høyere entropi.

For like sannsynligheter er $p_i=1/\Omega$. Da får vi

$$
S=-k_\mathrm{B}\sum_{i=1}^{\Omega}\frac1\Omega\ln\frac1\Omega
=k_\mathrm{B}\ln\Omega.
$$

Gibbs-entropien er dermed en generalisering av Boltzmanns uttrykk, ikke en konkurrerende definisjon.

:::{admonition} Underveisoppgave
:class: tip
Sammenlikn to systemer med fire mikrotilstander:

- A: $p=(1/4,1/4,1/4,1/4)$
- B: $p=(0{,}70,0{,}10,0{,}10,0{,}10)$

1. Hvilket system har størst entropi før du regner?
2. Beregn $S/k_\mathrm{B}$ for begge.
3. Forklar resultatet med bredden på sannsynlighetsfordelingen.
:::

:::{admonition} Løsningsforslag
:class: tip dropdown
For A er $S_A/k_\mathrm{B}=\ln4\approx1{,}386$. For B er

$$\frac{S_B}{k_\mathrm{B}}=-[0{,}70\ln0{,}70+3(0{,}10\ln0{,}10)]\approx0{,}940.$$

A har størst entropi fordi sannsynligheten er fordelt jevnt over alle fire mikrotilstandene.
:::

:::{dropdown} Fordypning: kombinatorisk begrunnelse for Gibbs-entropien
Tenk deg $N$ gjentakelser av det samme forsøket. Utfallet $i$ forekommer $n_i$ ganger, med $p_i=n_i/N$. Multiplisiteten til hele utfallsfølgen er

$$\Omega=\frac{N!}{n_1!n_2!\cdots}.$$

Med Stirlings tilnærming, $\ln N!\approx N\ln N-N$, får vi

$$\ln\Omega\approx-N\sum_i p_i\ln p_i.$$

Boltzmanns entropi for de $N$ gjentakelsene er $S_N=k_\mathrm{B}\ln\Omega$. Entropien per gjentakelse blir derfor

$$\frac{S_N}{N}=-k_\mathrm{B}\sum_i p_i\ln p_i.$$
:::


## Indre energi fra partisjonsfunksjonen

En måling av indre energi gir ikke energien til én bestemt mikrotilstand. Den gir gjennomsnittsenergien til systemet ved likevekt:

$$U=\langle E\rangle=\sum_i p_iE_i.$$

Setter vi inn Boltzmann-fordelingen, får vi

$$U=\frac1Q\sum_i E_i e^{-E_i/k_\mathrm{B}T}.$$

Når energinivåene ikke selv avhenger av temperaturen, kan dette skrives

:::{admonition} Indre energi fra $Q$
:class: important

$$U=k_\mathrm{B}T^2\left(\frac{\partial\ln Q}{\partial T}\right)_{V,N}.$$
:::

Partisjonsfunksjonen er dermed mer enn en normaliseringsfaktor. Temperaturavhengigheten til $Q$ forteller hvordan systemets gjennomsnittsenergi endres.

## Entropi fra partisjonsfunksjonen

Gibbs-entropien kan også uttrykkes ved $Q$. Fra Boltzmann-fordelingen har vi

$$\ln p_i=-\frac{E_i}{k_\mathrm{B}T}-\ln Q.$$

Setter vi dette inn i $S=-k_\mathrm{B}\sum_i p_i\ln p_i$, bruker $\sum_i p_iE_i=U$ og $\sum_i p_i=1$, får vi

:::{admonition} Entropi fra $Q$
:class: important

$$S=k_\mathrm{B}\ln Q+\frac{U}{T}.$$
:::

Dette er koblingen mellom energinivåene, sannsynlighetene og den makroskopiske entropien i det kanoniske ensemblet.

:::{admonition} Underveisoppgave
:class: tip
Forklar hvor de to leddene i $S=k_\mathrm{B}\ln Q+U/T$ kommer fra når Gibbs-entropien kombineres med Boltzmann-fordelingen.
:::


## Fra én partikkel til mange

Hvis systemet består av uavhengige, nummererte delsystemer med samme lokale partisjonsfunksjon $q$, faktoriserer partisjonsfunksjonen:

$$Q=q^N.$$

Et eksempel er $N$ nummererte gitterplasser med uavhengige lokale vibrasjonsmoder. Det er plassene eller modene som faktoriserer i modellen, ikke at identiske atomer blir fundamentalt skjelnbare.

For en fortynnet klassisk idealgass av identiske, mobile partikler ville $q^N$ overtelle samme fysiske konfigurasjon for hver ombytting av partiklene. I Maxwell–Boltzmann-grensen korrigeres dette med

$$Q\approx\frac{q^N}{N!}.$$

:::{admonition} Viktig presisering
:class: note
Identiske partikler er alltid uskjelnbare. Forskjellen mellom uttrykkene skyldes hvilke delsystemer som faktoriserer, og hvordan de fysiske mikrotilstandene telles.
:::

:::{dropdown} Fordypning: når er $q^N/N!$ utilstrekkelig?
Uttrykket gjelder i den klassiske, fortynnede grensen. Ved svært lav temperatur eller høy tetthet blir kvanteeffekter ved partikkelutveksling viktige. Bosoner og fermioner må da behandles med henholdsvis Bose–Einstein- og Fermi–Dirac-statistikk.
:::


## Ekvipartisjon og varmekapasitet

Før vi går videre til kinetisk gassteori, ser vi kort på en viktig konsekvens av sammenhengen mellom $Q$ og $U$. Den mer detaljerte behandlingen av translasjon, rotasjon og vibrasjon kommer i gasskapitlet.

Varmekapasiteten ved konstant volum er

$$C_V=\left(\frac{\partial U}{\partial T}\right)_{V,N}.$$

Den forteller hvor mye energi systemet tar opp per temperaturøkning. I den klassiske grensen sier **ekvipartisjonsteoremet** at hvert uavhengige kvadratiske energiledd bidrar med $\tfrac12k_\mathrm{B}T$ til gjennomsnittsenergien.

For et diatomisk molekyl har vi i den klassiske stiv-rotor/harmonisk-oscillator-modellen:

- tre translasjonelle kvadratiske energiledd
- to rotasjonelle kvadratiske energiledd
- én vibrasjonsmode med ett kinetisk og ett potensielt kvadratisk ledd

Hvis alle bidragene er klassiske, blir

$$U_\mathrm{m} = \frac72RT,\qquad C_{V,\mathrm m}=\frac72R.$$

Kvantemekanikken begrenser når ekvipartisjon er gyldig:

- $\Delta\varepsilon\ll k_\mathrm{B}T$: nivåene ligger tett relativt til den termiske energiskalaen, og bidraget er nær den klassiske verdien
- $\Delta\varepsilon\gg k_\mathrm{B}T$: eksiterte nivåer har liten sannsynlighet, og bidraget er utfrosset

Vibrasjonelle energigap er ofte større enn rotasjonelle. For $N_2$ og $O_2$ ved romtemperatur er vibrasjonsbidraget derfor i stor grad utfrosset, slik at $C_{V,\mathrm m}$ ligger nær $5R/2$, ikke $7R/2$.

Virkelige sannsynligheter endres kontinuerlig med temperaturen, så «på» og «av» er bare en framstilling av grenseverdiene.

:::{admonition} Oppgave 2: Energi og degenerasjon
:class: tip
Et grunntilstandsnivå har $E_0=0$ og $g_0=1$. Et eksitert nivå har $E_1=2k_\mathrm{B}T$ og $g_1=4$.

1. Sammenlikn én bestemt eksitert mikrotilstand med grunntilstanden.
2. Beregn $Q$.
3. Beregn samlet sannsynlighet for det eksiterte nivået.
4. Forklar hvorfor degenerasjonen må tas med.
:::

:::{admonition} Løsningsforslag
:class: tip dropdown
For én bestemt mikrotilstand er forholdet $e^{-2}\approx0{,}135$. Partisjonsfunksjonen er $Q=1+4e^{-2}\approx1{,}541$, og samlet sannsynlighet for det eksiterte nivået er $4e^{-2}/Q\approx0{,}351$.
:::

:::{admonition} Oppgave 3: Gibbs-entropi
:class: tip
Et system har sannsynlighetene $p=(1/2,1/4,1/4)$.

1. Beregn $S/k_\mathrm{B}$.
2. Sammenlikn med en jevn fordeling over tre mikrotilstander.
3. Hvilken fordeling har størst entropi, og hvorfor?
:::

:::{admonition} Løsningsforslag
:class: tip dropdown

$$\frac{S}{k_\mathrm{B}}=-\left[\frac12\ln\frac12+2\left(\frac14\ln\frac14\right)\right]\approx1{,}040.$$

Den jevne fordelingen gir $S/k_\mathrm{B}=\ln3\approx1{,}099$ og har størst entropi fordi sannsynligheten er fordelt jevnere.
:::

:::{admonition} Oppgave 4: Fra $Q$ til $U$ og $S$
:class: tip
Et ikke-degenerert to-nivåsystem har $E_0=0$ og $E_1=\Delta E$.

1. Sett opp $Q$.
2. Finn $p_1$ og $U$.
3. Bruk $S=k_\mathrm{B}\ln Q+U/T$ til å finne entropien.
4. Undersøk grensene $T\to0$ og $T\to\infty$.
:::

:::{admonition} Løsningsforslag
:class: tip dropdown

$$Q=1+e^{-\Delta E/k_\mathrm{B}T},\qquad p_1=\frac{e^{-\Delta E/k_\mathrm{B}T}}{Q},\qquad U=p_1\Delta E.$$

Ved $T\to0$ går $p_1\to0$ og $S\to0$. Ved svært høy temperatur blir $p_0\to1/2$ og $p_1\to1/2$, slik at $S\to k_\mathrm{B}\ln2$.
:::

:::{admonition} Oppgave 5: Varmekapasitet
:class: tip
Sammenlikn helium og $N_2$ som ideelle gasser.

1. Finn $C_{V,\mathrm m}$ for helium.
2. Finn $C_{V,\mathrm m}$ for $N_2$ når rotasjon bidrar klassisk og vibrasjon er utfrosset.
3. Hva nærmer $C_{V,\mathrm m}$ for $N_2$ seg når vibrasjonen også blir klassisk?
4. Forklar hvorfor overgangen er gradvis.
:::

:::{admonition} Løsningsforslag
:class: tip dropdown
Helium har bare tre translasjonelle bidrag og får $C_{V,\mathrm m}=3R/2$. For $N_2$ bidrar også to rotasjonelle ledd, slik at $C_{V,\mathrm m}=5R/2$. Når vibrasjonen blir klassisk, nærmer verdien seg $7R/2$. Boltzmann-sannsynlighetene for de vibrasjonelle nivåene endres kontinuerlig med temperaturen.
:::
