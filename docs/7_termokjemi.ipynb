Nå skal vi **bruke** disse verktøyene på kjemiske prosesser.

Vi skal følge tre størrelser:

$$\Delta H$$

som beskriver varmeutvekslingen ved konstant trykk

$$\Delta S$$

som beskriver endringen i tilgjengelige mikrotilstander

og

$$\Delta G=\Delta H-T\Delta S$$

som forteller hvilken retning som er termodynamisk gunstig ved konstant temperatur og trykk.

Siden $H$, $S$ og $G$ er tilstandsfunksjoner, trenger termodynamikken bare start- og sluttilstanden. Selve reaksjonsveien blir viktig når vi senere studerer kinetikk.


## Entalpi: eksoterme og endoterme prosesser

I forrige kapittel fant vi at varmeutvekslingen ved konstant trykk er lik entalpiendringen når bare trykk–volum-arbeid er relevant:

$$\Delta H=q_P$$

Dette gjør entalpi særlig nyttig i kjemi, fordi mange reaksjoner og andre kjemiske prosesser foregår ved omtrent konstant atmosfærisk trykk.

Når systemet avgir varme til omgivelsene, er prosessen **eksoterm**:

$$\Delta H<0$$

Når systemet tar opp varme fra omgivelsene, er prosessen **endoterm**:

$$\Delta H>0$$

Fortegnet forteller altså hvilken vei varme overføres. Det forteller derimot ikke alene om prosessen er spontan. For det trenger vi også entropien, som vi kommer tilbake til senere i kapitlet.

### Hess' lov – vi kan velge en annen vei

Entalpi er en tilstandsfunksjon. Det betyr at $\Delta H$ bare avhenger av start- og sluttilstanden, ikke av veien mellom dem.

Dette er grunnlaget for **Hess' lov**. Dersom en prosess kan deles opp i flere delprosesser, er den totale entalpiendringen summen av entalpiendringene for delprosessene:

$$\Delta H=\Delta H_1+\Delta H_2+\Delta H_3+\cdots$$

Dette er svært nyttig fordi mange reaksjonsentalpier er vanskelige å måle direkte. Vi kan i stedet konstruere en alternativ vei mellom de samme tilstandene ved hjelp av prosesser vi kjenner entalpiendringen til.

### Standard dannelsesentalpi

Et særlig nyttig referansepunkt er **standard dannelsesentalpi**, $\Delta_f H^\circ$. Den beskriver entalpiendringen når ett mol av et stoff dannes fra grunnstoffene i deres standardtilstander.

For eksempel kan vann dannes fra hydrogen og oksygen:

$$\mathrm{H_2(g)+\frac{1}{2}O_2(g)\rightarrow H_2O(l)}$$

Ved $298{,}15\ \mathrm{K}$ er

$$\Delta_f H^\circ(\mathrm{H_2O(l)})\approx-285{,}8\ \mathrm{kJ\,mol^{-1}}$$

For et grunnstoff i sin standardtilstand defineres standard dannelsesentalpien som null. Dette betyr ikke at grunnstoffets entalpi i seg selv er null, men at vi har valgt denne tilstanden som referanse.

Standardtilstanden innebærer et standardtrykk på $1\ \mathrm{bar}$. Temperaturen må oppgis separat, og tabellverdier gis ofte ved $298{,}15\ \mathrm{K}$.

Når standard dannelsesentalpiene er kjent, kan standard reaksjonsentalpi beregnes fra

$$\Delta_r H^\circ=\sum_{\mathrm{produkter}}\nu_i\Delta_f H_i^\circ-\sum_{\mathrm{reaktanter}}\nu_i\Delta_f H_i^\circ$$

der $\nu_i$ er den støkiometriske koeffisienten til stoff $i$.

:::{dropdown} Vanlige typer entalpiendringer

I tillegg til dannelsesentalpi og reaksjonsentalpi møter vi flere navn på entalpiendringer. Matematikken er den samme, men prosessen er forskjellig.

**Fordampingsentalpi** beskriver overgangen fra væske til gass:

$$\mathrm{X(l)\rightarrow X(g)}$$

For vann ved kokepunktet er fordampingsentalpien omtrent

$$\Delta_\mathrm{vap}H\approx40{,}7\ \mathrm{kJ\,mol^{-1}}$$

**Smelteentalpi** beskriver overgangen fra fast stoff til væske.

**Forbrenningsentalpi** beskriver entalpiendringen når ett mol stoff forbrennes fullstendig i oksygen.

**Løsningsentalpi** beskriver entalpiendringen når et stoff løses i et løsemiddel.

**Nøytraliseringsentalpi** beskriver entalpiendringen ved en syre–base-reaksjon som danner vann.

:::

:::{admonition} Kjerneidé
:class: important

Når vi skal bestemme en entalpiendring, kan vi tenke i fire trinn:

1. **Definer start- og sluttilstanden.** Hvilke stoffer har vi, hvor mye har vi av dem, og hvilke faser er de i?
2. **Finn en egnet vei mellom tilstandene.** Kan prosessen deles opp i trinn med kjente entalpiendringer?
3. **Summer bidragene.** Bruk Hess' lov og husk støkiometriske koeffisienter.
4. **Tolk resultatet.** Er prosessen eksoterm eller endoterm, og hvilke praktiske konsekvenser kan varmeutvekslingen få?

:::


### Kan vi forutsi fortegnet til $\Delta H$ fra molekylære vekselvirkninger?

Tabellverdier gir oss kvantitative entalpiendringer, men ofte ønsker vi først en kvalitativ forklaring. Da kan vi spørre hva som skjer med bindinger og intermolekylære vekselvirkninger gjennom prosessen.

Et nyttig utgangspunkt er:

- det krever energi å bryte bindinger og attraktive vekselvirkninger
- energi frigjøres når nye bindinger og attraktive vekselvirkninger dannes

Den totale entalpiendringen bestemmes av **balansen mellom alle vekselvirkningene som brytes og dannes**.

Hvis dannelsen av nye, gunstige vekselvirkninger frigjør mer energi enn det krever å bryte de opprinnelige, blir prosessen vanligvis eksoterm. Hvis det kreves mer energi å bryte de opprinnelige vekselvirkningene enn det frigjøres når nye dannes, blir prosessen endoterm.

Dette er en bedre måte å tenke på enn å si at «systemet alltid vil mot lavest mulig energi». En kjemisk prosess bestemmes ikke av entalpi alene. Entropien kan gjøre en prosess spontan selv om den er endoterm.

Fordamping er et enkelt eksempel. Når en væske går over til gass, må molekylene delvis frigjøres fra de attraktive intermolekylære vekselvirkningene som holder væsken sammen. Derfor må systemet ta opp varme:

$$\Delta_\mathrm{vap}H>0$$

Samtidig får molekylene langt flere muligheter til å fordele seg i rommet. Fordamping er derfor et godt eksempel på at entalpi og entropi kan trekke i hver sin retning.

### Fra molekylstruktur til laboratoriepraksis

Entalpi er ikke bare et tall vi beregner etter at en reaksjon er ferdig. Varmeutvekslingen påvirker hvordan reaksjoner må gjennomføres i laboratoriet.

En sterkt eksoterm reaksjon kan utvikle varme raskere enn den rekker å transporteres bort. Da kan temperaturen stige, løsemiddel begynne å koke og reaksjonshastigheten øke ytterligere. I praksis kan kjemikeren derfor måtte kontrollere temperaturen og hvor raskt reagenser blandes.

Endoterme prosesser krever på sin side varme fra omgivelsene. I syntese kan det bety at reaksjonsblandingen må varmes. Hvis et flyktig løsemiddel brukes ved høy temperatur, kan en kjøler kondensere løsemiddeldampen og føre væsken tilbake til reaksjonskolben.

Det er viktig å skille to spørsmål:

**Hvor mye varme frigjøres eller tas opp?** Dette er et termodynamisk spørsmål knyttet til $\Delta H$.

**Hvor raskt frigjøres eller tas varmen opp?** Dette avhenger av kinetikken og varmeoverføringen.

### Eksempel fra organisk kjemi: Grignard-reaksjonen

Grignard-reaksjoner er et godt eksempel på hvordan molekylstruktur, termodynamikk og laboratoriepraksis møtes. Et Grignard-reagens inneholder en sterkt polarisert karbon–magnesium-binding. Karbonet får betydelig negativ karakter og kan reagere som en nukleofil mot blant annet karbonylforbindelser.

Dette gjør det mulig å danne nye karbon–karbon-bindinger, som er en av de viktigste operasjonene i organisk syntese.

Grignard-reagenset bør ikke beskrives som et fritt $Mg^{2+}$-ion mellom et $Br^-$-ion og et karbanion. Bindingene har både kovalent og ionisk karakter, og strukturen påvirkes sterkt av løsemiddelet. Den enkle polaritetsmodellen er likevel nyttig for å forstå hvorfor karbonet er nukleofilt.

Bildet under viser selve karbon–karbon-bindingsdannelsen:

![Grignard-reaksjon](bilder/grignard.png)

Det er fristende å forklare varmeutviklingen bare ved å sammenlikne C–Mg-bindingen og den nye C–C-bindingen. Men en reaksjonsentalpi gjelder **hele prosessen**. Flere bindinger og vekselvirkninger brytes og dannes, og løsemiddel og eventuell etterbehandling bidrar også til energiregnskapet.

Vi kan illustrere dette ved å betrakte en idealisert totalprosess som inkluderer dannelse av Grignard-reagenset, reaksjon med formaldehyd og vandig etterbehandling:

$$\mathrm{Mg(s)+CH_3Br(g)+H_2CO(g)+H_2O(l)\rightarrow CH_3CH_2OH(l)+Mg^{2+}(aq)+Br^-(aq)+OH^-(aq)}$$

For denne konstruerte totalprosessen kan vi bruke standard dannelsesentalpier:

| Stoff | $\Delta_f H^\circ$ / $\mathrm{kJ\,mol^{-1}}$ |
|---|---:|
| $\mathrm{CH_3CH_2OH(l)}$ | $-277{,}6$ |
| $\mathrm{Mg^{2+}(aq)}$ | $-466{,}9$ |
| $\mathrm{Br^-(aq)}$ | $-121{,}6$ |
| $\mathrm{OH^-(aq)}$ | $-230{,}0$ |
| $\mathrm{Mg(s)}$ | $0$ |
| $\mathrm{CH_3Br(g)}$ | $-37{,}7$ |
| $\mathrm{H_2CO(g)}$ | $-108{,}6$ |
| $\mathrm{H_2O(l)}$ | $-285{,}8$ |

Summen for produktene er

$$\sum\Delta_f H^\circ_\mathrm{produkter}=-1096{,}1\ \mathrm{kJ\,mol^{-1}}$$

og summen for reaktantene er

$$\sum\Delta_f H^\circ_\mathrm{reaktanter}=-432{,}1\ \mathrm{kJ\,mol^{-1}}$$

Dermed får vi

$$\Delta_r H^\circ=-1096{,}1-(-432{,}1)=-664{,}0\ \mathrm{kJ\,mol^{-1}}$$

For den idealiserte totalprosessen er entalpiendringen altså sterkt negativ. Det stemmer med at varmeutvikling er et viktig praktisk hensyn.

:::{admonition} Fra termodynamikk til lab
:class: note

At $\Delta H$ er sterkt negativ forteller hvor mye varme som kan frigjøres per omsatt stoffmengde. Det forteller ikke hvor raskt varmen frigjøres.

I en virkelig syntese kan temperaturkontroll og kontrollert tilsetting derfor være nødvendig for å hindre at varmeutviklingen blir raskere enn varme kan transporteres bort. Grignard-reagenser krever dessuten tørre betingelser fordi de reagerer med protiske stoffer som vann.

:::

:::{dropdown} Videre perspektiv: hvorfor studere Grignard-mekanismen?

Selv om termodynamikken bare trenger start- og sluttilstanden, er reaksjonsmekanismen avgjørende når vi ønsker å forstå *hvordan* reaksjonen faktisk skjer.

Grignard-kjemi er derfor også et eksempel på møtet mellom eksperimentell og teoretisk kjemi. Beregninger kan brukes til å sammenlikne energien til mulige mellomtrinn og overgangstilstander og dermed undersøke hvilke reaksjonsveier som er mest sannsynlige.

Dette er et kinetisk og mekanistisk spørsmål, og vi kommer tilbake til slike energibarrierer senere i kurset.

## Trender for entropiendring

Entalpien beskriver én del av den termodynamiske drivkraften. Men vi har allerede sett at entalpi alene ikke kan avgjøre om en prosess er spontan.

I kapitlet om entropi tok vi utgangspunkt i mikrotilstander. Entropien kan uttrykkes som et mål på hvor mange mikroskopiske måter en makroskopisk tilstand kan realiseres på:

$$S=k_\mathrm{B}\ln\Omega$$

Her skal vi ikke utvikle entropibegrepet på nytt. I stedet skal vi bruke det til å gjenkjenne noen kjemiske trender.

### Faste stoffer, væsker og gasser

For samme stoff er entropien vanligvis lavest i fast fase og høyest i gassfase:

$$S_\mathrm{fast}<S_\mathrm{væske}<S_\mathrm{gass}$$

Grunnen er ikke bare at en gass ser mer «uordnet» ut. I en gass kan molekylene fordele seg over et mye større volum og har tilgang til svært mange translasjonstilstander. Et krystallinsk fast stoff begrenser derimot partiklene til bestemte områder i strukturen.

Derfor gir en faseovergang til gass vanligvis en tydelig positiv entropiendring:

$$\Delta S_\mathrm{vap}>0$$

og kondensering gir en negativ entropiendring:

$$\Delta S_\mathrm{cond}<0$$

### Kjemiske reaksjoner

For reaksjoner er det ofte nyttig å se på hvor mange *gassmolekyler* som finnes før og etter reaksjonen.

Betrakt

$$\mathrm{N_2O_4(g)\rightarrow2NO_2(g)}$$

Her går ett gassmolekyl over til to. Det gir partiklene flere måter å fordele energi og posisjon på, og vi forventer derfor

$$\Delta S>0$$

Å telle alle atomer, ioner og forbindelser på hver side av en reaksjonslikning er derimot ikke en generell entropiregel. Entropien avhenger av mikrotilstandene til hele systemet, og løsemiddel, vekselvirkninger og molekylstruktur kan være viktige.

### Løsninger er mer sammensatte

Når et fast stoff løses, får de løste partiklene ofte flere muligheter til å fordele seg. Dette kan øke entropien.

Samtidig må løsemiddelmolekylene organisere seg rundt de løste partiklene. Sterkt ladde ioner kan binde løsemiddelmolekyler så kraftig at dette gir et betydelig negativt bidrag til entropien.

Derfor bør vi ikke bruke regelen «fast stoff $\rightarrow$ flere ioner = større entropi» ukritisk. Vi må vurdere både de løste partiklene *og løsemiddelet*.

:::{dropdown} Løsningsforslag

For én ball er det ni mulige plasseringer:

$$\Omega_1=9$$

For to skjelnbare baller har den første ni mulige plasser og den andre åtte:

$$\Omega_2=9\cdot8=72$$

Entropiendringen relativt til én ball blir

$$\Delta S=k_\mathrm{B}\ln\frac{\Omega_2}{\Omega_1}=k_\mathrm{B}\ln8$$

For to identiske baller teller vi hver plassering bare én gang:

$$\Omega_2=\binom{9}{2}=36$$

Da blir

$$\Delta S=k_\mathrm{B}\ln\frac{36}{9}=k_\mathrm{B}\ln4$$

Dette er en leketøysmodell, ikke en kvantitativ modell av en virkelig kjemisk reaksjon. Poenget er å vise hvorfor flere mulige fordelinger kan gi større entropi.

## Entalpi og entropi må vurderes sammen

Vi har nå sett hvordan kjemiske prosesser kan gi både entalpi- og entropibidrag. Disse bidragene kan trekke i samme retning eller motvirke hverandre.

I forrige kapittel viste vi at de samles i Gibbs-energien ved konstant temperatur og trykk:

$$\Delta G=\Delta H-T\Delta S$$

Vi skal ikke utlede denne sammenhengen på nytt her. Nå bruker vi den som et verktøy.

| $\Delta H$ | $\Delta S$ | Konsekvens |
|---|---|---|
| negativ | positiv | begge bidrag favoriserer prosessen |
| negativ | negativ | entalpi favoriserer, entropi motvirker |
| positiv | positiv | entropi favoriserer, entalpi motvirker |
| positiv | negativ | begge bidrag motvirker prosessen |

Begrepene **entalpidrevet** og **entropidrevet** brukes ofte som kortformer for hvilken størrelse som gjør prosessen spontan. Det er likevel mer presist å spørre hvilket bidrag som favoriserer prosessen, fordi begge leddene alltid inngår i $\Delta G$.

La oss bruke dette på noen kjemiske eksempler.

Simuleringen ovenfor viser et skjematisk eksempel der systemet blir mer begrenset når ligander binder seg til et metallion. Da kan entropien i systemet avta. Hvis kompleksdannelsen samtidig er eksoterm, overføres varme til omgivelsene.

I en virkelig kompleksdannelse kan også løsemiddelmolekyler frigjøres eller reorganiseres. Vi kan derfor ikke bestemme $\Delta S$ bare ved å telle frie ligander. Simuleringen illustrerer først og fremst at lokal orden ikke alene avgjør spontaniteten.

### Begge bidrag kan favorisere prosessen

Nedbrytning av hydrogenperoksid kan skrives

$$\mathrm{2H_2O_2(l)\rightarrow2H_2O(l)+O_2(g)}$$

Prosessen er sterkt eksoterm, og det dannes gass. Både et negativt $\Delta H$ og et positivt $\Delta S$ trekker derfor i retning av negativ $\Delta G$.

Likevel kan en flaske hydrogenperoksid stå lenge uten å reagere raskt. Dette blir et viktig eksempel når vi senere skiller mellom termodynamikk og kinetikk.

### Entalpien kan favorisere mens entropien motvirker

Når vann fryser, dannes en mer begrenset fast struktur:

$$\mathrm{H_2O(l)\rightarrow H_2O(s)}$$

Frysing er eksoterm:

$$\Delta H<0$$

men entropien i systemet avtar:

$$\Delta S<0$$

Ved lave temperaturer kan det negative entalpibidraget dominere. Ved høyere temperatur blir leddet $-T\Delta S$ stadig viktigere.

### Entropien kan favorisere mens entalpien motvirker

Ammoniumnitrat brukes i kuldepakker fordi oppløsningen i vann er endoterm:

$$\mathrm{NH_4NO_3(s)\rightarrow NH_4^+(aq)+NO_3^-(aq)}$$

Her er

$$\Delta H>0$$

Prosessen kan likevel være spontan fordi oppløsningen gir en tilstrekkelig positiv entropiendring:

$$\Delta S>0$$

Varmen som kreves til oppløsningen tas fra omgivelsene, og temperaturen synker.

### Temperaturen kan endre hvilken retning som er gunstig

Når $\Delta H$ og $\Delta S$ har samme fortegn, kan temperaturen avgjøre hvilket bidrag som dominerer.

Ved likevekt er

$$\Delta G=0$$

og dersom $\Delta H$ og $\Delta S$ kan behandles som omtrent konstante over temperaturintervallet, får vi

$$T_\mathrm{eq}\approx\frac{\Delta H}{\Delta S}$$

Dette blir særlig viktig når vi senere studerer faseoverganger og kjemiske likevekter.


:::{dropdown} Videre perspektiv: Hvordan kan en termodynamisk ugunstig prosess drives?

At en prosess har $\Delta G>0$ under bestemte betingelser betyr ikke at den er «forbudt». Prosessen kan drives dersom vi tilfører arbeid eller kobler den kjemisk til en annen prosess.

**Elektrisk arbeid**

Elektrolyse av vann kan drives ved å tilføre elektrisk arbeid:

$$\mathrm{2H_2O(l)\rightarrow2H_2(g)+O_2(g)}$$

**Lys**

Fotosyntesen bruker energi fra lys til å drive kjemiske omdannelser som ikke ville vært spontane i samme retning uten energitilførselen.

**Kobling til en annen reaksjon**

Hvis to prosesser er kjemisk koblet, summeres Gibbs-energiendringene:

$$\Delta G_\mathrm{tot}=\Delta G_1+\Delta G_2$$

En prosess med positiv $\Delta G$ kan derfor drives av en annen med tilstrekkelig negativ $\Delta G$. I biokjemi er ATP-hydrolyse et kjent eksempel.

:::

### En katalysator endrer ikke $\Delta G$

En katalysator gir en alternativ reaksjonsvei med lavere aktiveringsbarriere og kan derfor gjøre en reaksjon mye raskere.

Den endrer derimot ikke start- og sluttilstanden. Dermed endrer en katalysator heller ikke reaksjonens

$$\Delta G$$

$$\Delta H$$

eller

$$\Delta S$$

Katalyse handler altså om **reaksjonsveien og kinetikken**, ikke om å gjøre en termodynamisk ugunstig reaksjon spontan.

:::{dropdown} Eksempel fra katalyse: metan til metanol

Metan inneholder sterke C–H-bindinger og er kinetisk vanskelig å aktivere selektivt. Katalysatorforskning forsøker blant annet å finne reaksjonsveier som gjør det mulig å omdanne metan til mer nyttige produkter under mildere betingelser.

Utfordringen er ikke bare å aktivere metan, men også å unngå at ønskede produkter reagerer videre. Dette illustrerer hvorfor katalyse først og fremst handler om energibarrierer og reaksjonsveier.

:::


## Spontanitet betyr ikke at noe skjer raskt

Gjennom kapitlet har vi brukt Gibbs-energien til å vurdere hvilken retning en prosess er termodynamisk gunstig i. Det er viktig å være presis med hva dette betyr.

Når

$$\Delta G<0$$

har sluttilstanden lavere Gibbs-energi enn starttilstanden under de aktuelle betingelsene. Termodynamikken sier dermed hvilken retning som favoriseres.

Den sier *ikke hvor raskt systemet kommer dit*.

En kjemisk reaksjon kan måtte passere en energimessig ugunstig overgangstilstand før produktene kan dannes. Hvis energibarrieren er høy, kan reaksjonen være svært langsom selv om $\Delta G$ er sterkt negativ.

Dette forklarer hvorfor:

- hydrogenperoksid kan være termodynamisk ustabilt, men lagres over tid
- brennbare stoffer ikke nødvendigvis begynner å brenne av seg selv
- katalysatorer kan gjøre enorme forskjeller for reaksjonshastigheten uten å endre $\Delta G$

I dette kapitlet har vi hovedsakelig sammenliknet start- og sluttilstander. Når vi senere kommer til kinetikk, skal vi studere selve veien mellom dem.


## Oppsummering

I dette kapitlet har vi brukt termodynamikken på konkrete kjemiske prosesser.

Ved konstant trykk kan entalpiendringen bestemmes fra varmeutvekslingen, og fordi entalpi er en tilstandsfunksjon kan Hess' lov brukes til å finne reaksjonsentalpier fra alternative reaksjonsveier og standard dannelsesentalpier.

Vi så også hvordan fortegnet til $\Delta H$ kan forstås kvalitativt ved å vurdere balansen mellom bindinger og intermolekylære vekselvirkninger som brytes og dannes. Dette har direkte konsekvenser for hvordan reaksjoner må gjennomføres i laboratoriet.

Entropien knytter kjemiske prosesser tilbake til mikrotilstandene. Fase, antall gasspartikler, blanding, molekylstruktur og reorganisering av løsemiddelet kan alle påvirke $\Delta S$.

Ved konstant temperatur og trykk bruker vi

$$\Delta G=\Delta H-T\Delta S$$

til å vurdere hvilken retning som er termodynamisk gunstig. Temperaturen kan endre balansen mellom entalpi- og entropibidragene og dermed hvilken retning som favoriseres.

Til slutt skilte vi mellom termodynamikk og kinetikk. En negativ $\Delta G$ forteller hvilken retning som er gunstig, men ikke hvor raskt prosessen går. Katalysatorer påvirker reaksjonsveien og hastigheten, ikke reaksjonens $\Delta G$.

Dette blir utgangspunktet når vi videre skal studere faseoverganger, blandinger og kjemisk likevekt.


## Oppgaver

### Oppgave 1 – Hydrogenproduksjon ved dampreformering

En stor del av industrielt hydrogen produseres ved dampreformering av metan:

$$\mathrm{CH_4(g)+H_2O(g)\rightarrow CO(g)+3H_2(g)}$$

Ved $298\ \mathrm{K}$ kan du bruke følgende standard dannelsesentalpier:

| Stoff | $\Delta_f H^\circ$ / $\mathrm{kJ\,mol^{-1}}$ |
|---|---:|
| $\mathrm{CH_4(g)}$ | $-74{,}8$ |
| $\mathrm{H_2O(g)}$ | $-241{,}8$ |
| $\mathrm{CO(g)}$ | $-110{,}5$ |
| $\mathrm{H_2(g)}$ | $0$ |

a) Beregn $\Delta_r H^\circ$ for reaksjonen.

b) Er reaksjonen eksoterm eller endoterm?

c) Dampreformering drives industrielt ved høy temperatur. Forklar hvorfor den beregnede entalpiendringen gjør varmeforsyning til et viktig prosesshensyn.

d) Hvorfor kan du ikke konkludere fra $\Delta H$ alene om reaksjonen er spontan?

:::{dropdown} Løsningsforslag

a)

$$\Delta_r H^\circ=[-110{,}5+3(0)]-[-74{,}8+(-241{,}8)]$$

$$\Delta_r H^\circ=+206{,}1\ \mathrm{kJ\,mol^{-1}}$$

b) Reaksjonen er sterkt endoterm.

c) Systemet må tilføres betydelige mengder varme når reaksjonen går. I en industriell reaktor må denne varmen overføres effektivt til reaksjonsblandingen.

d) Spontanitet bestemmes av Gibbs-energien, som også inneholder entropibidraget:

$$\Delta G=\Delta H-T\Delta S$$

:::


### Oppgave 2 – Hvor varmt må en kalkovn være?

Ved produksjon av brent kalk spaltes kalsiumkarbonat:

$$\mathrm{CaCO_3(s)\rightarrow CaO(s)+CO_2(g)}$$

Anta at

$$\Delta H^\circ=178\ \mathrm{kJ\,mol^{-1}}$$

$$\Delta S^\circ=161\ \mathrm{J\,mol^{-1}\,K^{-1}}$$

og at verdiene kan behandles som omtrent konstante over temperaturintervallet.

a) Hvilket av bidragene $\Delta H$ og $\Delta S$ favoriserer reaksjonen?

b) Beregn temperaturen der $\Delta G^\circ=0$.

c) Er reaksjonen termodynamisk gunstig over eller under denne temperaturen?

d) Forklar molekylært hvorfor entropiendringen er positiv.

:::{dropdown} Løsningsforslag

a) $\Delta H>0$ motvirker reaksjonen, mens $\Delta S>0$ favoriserer den.

b)

$$T_\mathrm{eq}=\frac{\Delta H^\circ}{\Delta S^\circ}$$

$$T_\mathrm{eq}=\frac{178000\ \mathrm{J\,mol^{-1}}}{161\ \mathrm{J\,mol^{-1}\,K^{-1}}}\approx1106\ \mathrm{K}$$

Dette tilsvarer omtrent

$$833\ ^\circ\mathrm{C}$$

c) Ved høyere temperatur blir $-T\Delta S$ mer negativt, og reaksjonen blir termodynamisk gunstig.

d) Det dannes ett mol gass fra faste stoffer. Gassmolekylene får svært mange tilgjengelige translasjonstilstander, og entropien øker.

:::


### Oppgave 3 – Hvorfor blir en kuldepakke kald?

Oppløsning av ammoniumnitrat kan beskrives med de omtrentlige verdiene

$$\Delta H=+25{,}7\ \mathrm{kJ\,mol^{-1}}$$

$$\Delta S=+109\ \mathrm{J\,mol^{-1}\,K^{-1}}$$

a) Beregn $\Delta G$ ved $298\ \mathrm{K}$.

b) Er oppløsningen termodynamisk gunstig ved denne temperaturen?

c) Forklar hvorfor løsningen blir kaldere selv om prosessen er spontan.

d) Ved hvilken temperatur gir den enkle tilnærmingen $\Delta G=0$?

:::{dropdown} Løsningsforslag

a) Først omgjør vi entropien til $\mathrm{kJ\,mol^{-1}\,K^{-1}}$:

$$\Delta S=0{,}109\ \mathrm{kJ\,mol^{-1}\,K^{-1}}$$

Dermed

$$\Delta G=25{,}7-(298)(0{,}109)\approx-6{,}8\ \mathrm{kJ\,mol^{-1}}$$

b) Ja, fordi $\Delta G<0$.

c) Oppløsningen er endoterm og tar opp varme fra omgivelsene. Den positive entropiendringen er stor nok til å gjøre Gibbs-energien negativ.

d)

$$T_\mathrm{eq}=\frac{25{,}7}{0{,}109}\approx236\ \mathrm{K}$$

:::


### Oppgave 4 – Varmeutvikling i en organisk syntese

Anta at en idealisert organisk reaksjon har

$$\Delta_r H=-120\ \mathrm{kJ\,mol^{-1}}$$

I et forsøk planlegges det å omsette $0{,}050\ \mathrm{mol}$ av begrensende reaktant.

a) Hvor mye varme kan frigjøres dersom reaksjonen går fullstendig ved konstant trykk?

b) Hvorfor kan denne energimengden være et praktisk problem selv om $6\ \mathrm{kJ}$ ikke høres svært stort ut?

c) Forklar hvorfor kontrollert tilsetting og kjøling kan redusere risikoen uten å endre selve $\Delta_r H$.

d) En student sier: «Hvis vi kjøler reaksjonen, blir $\Delta H$ mindre negativ og derfor blir reaksjonen mindre spontan.» Hva er problemet med denne forklaringen?

:::{dropdown} Løsningsforslag

a)

$$q_P=n\Delta_r H=(0{,}050)(-120)=-6{,}0\ \mathrm{kJ}$$

b) Hvis energien frigjøres raskt i et lite volum, kan temperaturen øke kraftig. Det kan øke reaksjonshastigheten og føre til kraftig koking av løsemiddel.

c) Tiltakene påvirker først og fremst hvor raskt varme produseres og hvor raskt den transporteres bort. Reaksjonsentalpien er en tilstandsfunksjon for de valgte start- og sluttilstandene.

d) Spontanitet avgjøres av $\Delta G$, ikke av $\Delta H$ alene. Temperatur kan påvirke $\Delta G$, men kjøling er ikke det samme som å «gjøre entalpien mindre negativ».

:::


### Oppgave 5 – Hvorfor favoriseres mer brun $NO_2$ ved høyere temperatur?

Dinitrogentetraoksid og nitrogendioksid står i likevekt:

$$\mathrm{N_2O_4(g)\rightleftharpoons2NO_2(g)}$$

Spaltingen mot høyre er endoterm:

$$\Delta H>0$$

a) Hvilket fortegn forventer du for $\Delta S$ mot høyre? Begrunn svaret.

b) Hvilket av bidragene i $\Delta G=\Delta H-T\Delta S$ favoriserer spalting?

c) Forklar hvorfor økt temperatur gjør spalting mot $NO_2$ mer termodynamisk gunstig.

d) Hvorfor er «det blir flere molekyler» en bedre entropibegrunnelse her enn for en vilkårlig reaksjon i løsning?

:::{dropdown} Løsningsforslag

a) Vi forventer $\Delta S>0$ fordi ett gassmolekyl blir til to gassmolekyler.

b) Det positive $\Delta H$ motvirker spalting, mens det positive $\Delta S$ favoriserer den.

c) Når temperaturen øker, blir det negative bidraget $-T\Delta S$ større i absoluttverdi. Dermed blir $\Delta G$ for spaltingen mindre positiv eller mer negativ.

d) For gasser er antall molekyler tett knyttet til antallet mulige translasjonstilstander. I løsning kan reorganisering av løsemiddelet og sterke vekselvirkninger gi store bidrag som ikke fanges ved bare å telle specier.

:::


### Oppgave 6 – Kan et protein folde seg spontant når kjeden får færre mulige former?

Et protein er en lang molekylkjede. Du trenger ingen forkunnskaper i biokjemi for denne oppgaven.

Når kjeden er utfoldet, kan den anta svært mange ulike former. Når den folder seg til en bestemt tredimensjonal struktur, blir antallet mulige former for selve kjeden mye mindre. Samtidig kan det dannes gunstige vekselvirkninger i proteinet, og vannmolekyler som tidligere var organisert rundt upolare deler av kjeden kan frigjøres til vannet omkring.

a) Hvilket fortegn forventer du for entropiendringen til selve proteinkjeden ved folding?

b) Hvorfor er dette ikke nok til å avgjøre fortegnet til den totale $\Delta S$?

c) Hvordan kan dannelsen av gunstige vekselvirkninger påvirke $\Delta H$?

d) Hvordan kan frigjøring av vannmolekyler påvirke $\Delta S$?

e) Hvilken størrelse må til slutt brukes for å avgjøre om folding er termodynamisk gunstig ved konstant temperatur og trykk?

:::{dropdown} Løsningsforslag

a) Entropien til selve kjeden avtar fordi den får færre tilgjengelige konformasjoner.

b) Entropiendringen må vurderes for hele systemet. Vann og andre molekyler rundt proteinet kan få større entropi.

c) Dannelse av gunstige vekselvirkninger kan gi et negativt bidrag til $\Delta H$.

d) Frigjørte vannmolekyler får flere tilgjengelige mikrotilstander og kan derfor gi et positivt entropibidrag.

e) Vi må vurdere Gibbs-energien:

$$\Delta G=\Delta H-T\Delta S$$

:::
