## Molekylære interaksjoner: Hvorfor kondenserer noe i det hele tatt?

Avkjøl argongass nok, og den blir til en væske. Det virker kanskje helt naturlig; Mange stoffer kan jo gå fra gass til væske. Men hvis vi stopper opp og tenker etter, er det egentlig ganske merkelig. Argon består av elektrisk nøytrale atomer uten permanent dipolmoment. Ved første øyekast finnes det derfor ingen åpenbar grunn til at atomene skulle tiltrekke hverandre. Likevel gjør de det, sterkt nok til at gassen kan kondensere - en faseovergang slik vi så i [forrige delkapittel](https://andreasdh.github.io/Bifrost/docs/8_faser.html). 

Denne observasjonen peker på noe nytt. I tidligere kurs har du lært om kovalente og ioniske bindinger, bindinger som holder atomer sammen i molekyler og krystaller. I dette kapittelet skal vi se på en annen type interaksjoner: de svakere kreftene som virker mellom separate atomer og molekyler. Selv om disse kreftene er mye svakere enn kjemiske bindinger, er de avgjørende for at væsker og mange faste stoffer i det hele tatt kan eksistere.

Hvordan kan vi beskrive slike interaksjoner? I stedet for å starte med selve kraften, er det mer nyttig å beskrive den potensielle energien mellom to partikler. Denne energien avhenger av avstanden mellom dem og skrives som $u(r)$, der $r$ er avstanden mellom partiklene. Når partiklene beveger seg nærmere hverandre eller lenger fra hverandre, endrer den potensielle energien seg. Noen avstander er mer energetisk gunstige enn andre. 

Men den potensielle energien er bare én del av historien. Partiklene er hele tiden i bevegelse, hvor raskt de beveger seg bestemmes av temperaturen. Denne bevegelsen beskrives av den [kinetiske energien](https://andreasdh.github.io/Bifrost/docs/4_gasser.html#maxwell-boltzmann-fordelingen). Ved høy temperatur er bevegelsen stor, og partiklene kan lettere bevege seg bort fra hverandre. Ved lav temperatur er bevegelsen mindre, slik at de svake tiltrekkkende kreftene får større betydning. 

Om et stoff opptrer som en gass, en væske eller et fast stoff, bestemmes derfor av samspillet mellom potensiell og kinetisk energi. Den potensielle energien favoriserer bestemte avstander mellom partiklene, mens den kinetiske energien motvirker at de blir værende der. I dette modulen skal vi utvikle en enkel modell for denne balansen. Den skal gi oss et rammeverk for å forstå hvorfor stoffer kondenserer, hvorfor molekyler foretrekker bestemte avstander fra hverandre, og hvordan dette kan beskrives med blant annet **Lennard-Jones potensialet**.

## Fra Coulombs lov til parpotensial

**Coulombs lov** beskriver den elektriske kraften mellom to ladninger. Kraften øker når ladningene blir større, og når avstanden mellom dem blir mindre. Denne sammenhengen forklarer blant annet hvorfor små ioner danner sterkere ionebindinger enn store ioner:

$$F = \frac{q_1 q_2}{4\pi\varepsilon_0 r^2},$$

der $F$ er den elektriske kraften mellom to ladninger $q_1$ og $q_2$ og $r$ er avstanden mellom dem. Her må $\varepsilon_0$ ikke forveksles med grunntilstandsenergien i [Statistisk mekanikk](https://andreasdh.github.io/Bifrost/docs/1_introduksjon.html#mikrotilstander-og-energi) modulen, for her representerer den *permittiviteten* i tomt rom, noe vi ikke skal gå inn på i dette kurset. 

Coulombs lov kan ikke forklare alle intermolekylære vekselvirkninger. Mange molekyler er elektrisk nøytrale og har ikke permanente dipoler, men tiltrekker likevel hverandre. For eksempel kan nitrogen $N_2$ og edelgasser kondensere til væske ved lave temperaturer. 

For å beskrive slike vekselvirkninger trenger vi en mer generell modell. I stedet for å beskrive kraften mellom to punktladninger, beskriver vi den potensielle energien mellom to partikler som en funksjon av avstanden mellom dem. Denne beskrivelsen kalles et parpotensial $u(r)$, og danner grunnlaget for å forstå intermolekylære interaksjoner. 


:::{admonition} Underveisoppgave
:class: tip

Coulombs lov gir kraften mellom to ladninger: $F = \frac{q_1 q_2}{4\pi\varepsilon_0 r^2}$. Svar på følgende uten å regne - tenk bare på hvordan $F$ endrer seg.

**a)** Hva skjer med kraften hvis du dobler den ene ladningen $q_1$? Hva hvis du dobler begge ladningene samtidig?

**b)** Hva skjer med kraften hvis du dobler avstanden $r$ mellom ladningene? Blir endringen større eller mindre enn i a)? Forklar hvorfor.

**c)** Tenk deg at $q_1$ er positiv og $q_2$ negativ. Hva sier fortegnet til $F$ da, og hva betyr det fysisk for hvordan ladningene beveger seg i forhold til hverandre?

:::


:::{dropdown} Løsningsforslag

**a)** Kraften er proporsjonal med hver av ladningene. Dobler du $q_1$, blir kraften dobbelt så stor. Dobler du begge samtidig, blir kraften $2 \times 2 = 4$ ganger så stor, siden produktet $q_1 q_2$ firedobles.

**b)** Kraften avtar med kvadratet av avstanden ($F \propto 1/r^2$). Dobler du avstanden, blir kraften $\frac{1}{2^2} = \frac{1}{4}$ av det den var - altså fire ganger svakere. Endringen er kraftigere enn ladningseffekten i a), fordi avstanden står i annen potens: en liten økning i $r$ gir et stort fall i kraft.

**c)** Med én positiv og én negativ ladning blir produktet $q_1 q_2$ negativt, så $F$ blir negativ. Et negativt fortegn betyr at kraften er tiltrekkende - ladningene trekkes mot hverandre. (Er begge like ladet, blir $F$ positiv, altså frastøtende.)

:::

### Parpotensialet $u(r)$

Et parpotensial , $u(r)$, beskriver den potensielle energien mellom to partikler som en funksjon av avstanden $r$:

![ur](bilder/ur_graf.png)

- Når partiklene er langt fra hverandre, er vekselvirkningen neglisjerbar, og $u(r) \to 0$.
- Når de kommer nærmere, dominerer de tiltrekkende kreftene, slik at den potensielle energien avtar. Dersom partiklene presses enda tettere sammen, øker energien raskt fordi frastøtende krefter blir dominerende.

Denne frastøtningen skyldes både elektrostatisk frastøtning mellom like ladninger og at elektronskyene begynner å overlappe. Ved slik overlapp blir **Pauli-prinsippet** viktig: to elektroner kan ikke befinne seg i samme kvantetilstand, noe som gir opphav til en sterk, korttrekkende frastøtning. 

Ved en bestemt avstand, $r^*$, balanserer de tiltrekkende og frastøtende bidragene hverandre. Her har $u(r)$ et minimum, nettokraften mellom partiklene er null, og $r^*$ tilsvarer den gjennomsnittlige bindingslengden. 

Den generelle formen på parpotensialet er den samme for alle typer vekselvirkninger; Fra sterke kovalente bindinger til svake intermolekylære krefter. Det som skiller dem, er det matematiske uttrykket for $u(r)$ og hvor dyp potensialbrønnen er. 



:::{admonition} Underveisoppgave
:class: tip

For ioner som $Na^+$ og $Cl^-$ er svaret intuitivt. Coulombs love forklarer tiltrekningen direkte. Men hva med to nøytrale molekyler, som verken har postivt eller negativ nettoladning, hvordan kan slike molekyler likevel tiltrekke hverandre?
:::

:::{dropdown} Løsningsforslag

Selv om molekylet er nøytralt, er ladningen ikke jevnt fordelt inni det. Elektronene beveger seg, og kan i et øyeblikk være forskjøvet mot én side, så molekylet får en litt positiv og en litt negativ ende - en momentan dipol. Denne påvirker nabomolekylet og skaper en tilsvarende forskyvning der, så de trekkes svakt mot hverandre. Coulombs lov gjelder altså fortsatt, det er de små, ujevne ladningene inni molekylene som tiltrekker - ikke en nettoladning. Dette kalles van der Waals-krefter.

:::

### Ladningsfordeling og dipolmomenter

Vi har sett av nøytrale molekyler kan tiltrekke hverandre, selv om de ikke har en netto ladning. For å forstå hvorfor, må vi se nærmere på hvordan ladningen er fordelt i molekylet. 

Et molekyl kan være elektrisk nøytralt samtidig som den positive og negative ladningen er ujevnt fordelt. Denne romlige ladningsfordelingen bestemmes av både molekylets geometri og elektronfordelingen, og har stor betydning for hvilke intermolekylære vekselvirkninger molekylet kan danne og hvor sterke disse blir. 


![multipol](bilder/multipol.png)

Generelt kan enhver ladningsfordeling beskrives ved hjelp av en multipolutvikling. Den første termen er en monopol, som beskriver molekylets nettoladning. Deretter følger dipol, kvadrupol, oktupol og høyere multipolmomenter, som beskriver stadig mer detaljerte trekk ved ladningsfordelingen. Etter hvert som høyere multipolmomenter dominerer, blir vekselvirkningene både svakere og mer korttrekkende. 

I dette emnet skal vi hovedsakelig fokusere på dipolmomentet, siden dette er den viktigste egenskapen for å forstå de fleste intermolekylære vekselvirkningene. 

### Permanente og induserte dipolmomenter

Et dipolmoment beskrives hvor stor ladningsseparasjonene er i et molekyl. Vi skiller mellom to hovedtyper:

- Et permanent dipolmoment finnes i molekyler der den positive og negative ladningen er forskjøvet fra hverandre også uten påvirkning fra et ytre elektrisk felt. Dette skyldes molekylets geometri og elektronfordeling. 

- Et indusert dipolmoment oppstår når et elektrisk felt påvirker et atom eller molekyl og forskyver elektronskyen. Hvor lett dette skjer, beskrives av molekylets **polariserbarhet**. Molekyler med høy polariserbarhet får lettere indusert et dipolmoment enn molekyler med lav polariserbarhet. 

![krefter](bilder/krefter.svg)


Et molekyl kan ha et permanent dipolmoment, et indusert dipolmoment, eller begge deler. Dette danner grunnlaget for de tre viktigste van der Waals vekselvirkningnene: permanente dipoler gir opphav til Keesom-krefter, permanente dipoler som induserer dipoler gir Debye-krefter, og fluktuerende, induserte dipoler gir London-dispersjonskrefter. 

I gasser og væsker roterer molekylene kontinuerlig. Derfor må vekselvirkningene beskrives som et gjennomsnitt over mange orienteringer. Denne orienteringsmiddelverdien gjør at de attraktive vekselvirkningene blir svakere og mer korttrekkende enn de ville vært mellom molekyler med faste orienteringer. 

:::{card}
**Eksempel: dipol, kvadrupol, oktupol**
^^^
![Multipoleksempler](bilder/poler_eksempel.svg)

**Dipol – vann (H₂O).** Vannmolekylet er bøyd (~104,5°). Oksygen er mer elektronegativt og trekker elektrontettheten mot seg, slik at det blir delvis negativt (δ−) mens hydrogenatomene blir delvis positive (δ+). Ladningstyngdepunktene faller ikke sammen, og molekylet får et permanent dipolmoment.

**Kvadrupol – karbondioksid (CO₂).** CO₂ er lineært og symmetrisk. Hver C=O-binding er polar, men de to bindingsdipolene er like store og peker motsatt vei, så de kanselerer og nettodipolen blir null. Ladningsmønsteret δ− δ+ δ− langs molekylaksen gir likevel et kvadrupolmoment.

**Oktupol – metan (CH₄).** Metan er tetraedrisk. Den høye symmetrien gjør at både dipol- og kvadrupolmomentet forsvinner. Det første multipolmomentet som ikke er null, er derfor oktupolmomentet.
:::

## Lennard-Jones-potensialet


Å følge alle detaljene i elektronstrukturen til to atomer er upraktisk — og heldigvis unødvendig. I stedet bruker vi effektive modeller: enkle matematiske uttrykk som fanger den samlede effekten av de underliggende fysiske bidragene.

**Lennard-Jones-potensialet** er den mest brukte av disse. Det gir en enkel, men realistisk beskrivelse av hvordan to nøytrale atomer påvirker hverandre som funksjon av avstanden mellom dem:

$$U(r) = 4\varepsilon \left[ \left(\frac{\sigma}{r}\right)^{12} - \left(\frac{\sigma}{r}\right)^{6} \right].$$

Størrelsen $\sigma$ setter lengdeskalaen: den bestemmer hvor kurven krysser null, og kan tolkes som en effektiv atomstørrelse. Likevektsavstanden ligger litt lenger ut, gitt ved

$$r^* = 2^{1/6}\sigma.$$

Jo dypere minimum i funksjonen, desto sterkere binding. Denne dybden styres av $\varepsilon$, som gir et direkte mål på hvor «sterk» interaksjonen er.

Interaksjonen har alltid to motsatte sider: en attraktiv del som trekker atomene sammen på lengre avstand, og en frastøtende del som dominerer kraftig når de presses tett sammen.Det viktige er ikke detaljene i bidragene, men at de virker over ulik avstand: tiltrekningen strekker seg lenger, mens frastøtningen vokser ekstremt raskt når avstanden blir liten. Kombinert gir dette et potensial med en karakteristisk form: null ved store avstander, et minimum et stykke ut, og en bratt stigning når atomene kommer for nær.

:::{admonition} Underveisoppgave
:class: tip

Lennard-Jones-potensialet beskriver energien mellom to nøytrale atomer som funksjon av avstanden $r$:

$U(r) = 4\varepsilon \left[ \left(\frac{\sigma}{r}\right)^{12} - \left(\frac{\sigma}{r}\right)^{6} \right].$

For argon er $\varepsilon = 1.65 \times 10^{-21}\ \text{J}$ og $\sigma = 3.4 \times 10^{-10}\ \text{m}$.

**a)** Potensialet har to ledd: et positivt $\left(\frac{\sigma}{r}\right)^{12}$ og et negativt $\left(\frac{\sigma}{r}\right)^{6}$. Hvilket ledd dominerer ved svært små avstander, og hvilket dominerer ved store? Hva sier dette om at atomer frastøter hverandre på kort hold, men tiltrekker hverandre på litt lengre hold?

**b)** Regn ut likevektsavstanden $r^* = 2^{1/6}\sigma$ for argon. Dette er avstanden der energien er lavest.

**c)** Regn ut $U(r^*)$, altså energien ved likevektsavstanden. Sammenlign svaret med $\varepsilon$. Hva forteller dette deg om den fysiske betydningen av $\varepsilon$?

:::


:::{dropdown} Løsningsforslag

**a)** Ved små $r$ vokser $\left(\frac{\sigma}{r}\right)^{12}$ mye raskere enn $\left(\frac{\sigma}{r}\right)^{6}$, så det positive $r^{-12}$-leddet dominerer - energien blir stor og positiv, som gir en sterk frastøting når atomene kommer for nær. Ved store $r$ er begge ledd små, men $r^{-6}$ avtar saktere enn $r^{-12}$, så det negative leddet dominerer - energien er svakt negativ, som gir en tiltrekning. Atomene frastøter altså hverandre på kort hold og tiltrekker på litt lengre hold.

**b)** Vi setter inn:

$r^* = 2^{1/6} \cdot 3.4 \times 10^{-10}\ \text{m} = 1.122 \cdot 3.4 \times 10^{-10}\ \text{m} = 3.8 \times 10^{-10}\ \text{m}.$

**c)** Vi setter $r = r^* = 2^{1/6}\sigma$ inn i potensialet. Da er $\frac{\sigma}{r^*} = 2^{-1/6}$, så $\left(\frac{\sigma}{r^*}\right)^{6} = 2^{-1} = \frac{1}{2}$ og $\left(\frac{\sigma}{r^*}\right)^{12} = 2^{-2} = \frac{1}{4}$:

$U(r^*) = 4\varepsilon\left[\frac{1}{4} - \frac{1}{2}\right] = 4\varepsilon\left(-\frac{1}{4}\right) = -\varepsilon = -1.65 \times 10^{-21}\ \text{J}.$

Energien ved likevektsavstanden er nøyaktig $-\varepsilon$. Dette viser at $\varepsilon$ er dybden på potensialbrønnen - altså hvor sterkt bundet de to atomene er. En stor $\varepsilon$ betyr en dyp brønn og sterk binding.

:::

## Van der Waals gass-likningen


I de foregående avsnittene har vi beskrevet vekselvirkningen mellom to partikler ved hjelp av parpotensialet $u(r)$. Vi har sett at denne vekselvirkningen består av to konkurrerende bidrag: en sterkt frastøtende del ved korte avstander og en svakere attraktiv del ved større avstander. For mange nøytrale atomer og molekyler kan denne vekselvirkningen beskrives godt ved hjelp av Lennard-Jones potensialet. 

En virkelig gass består imidlertid ikke av bare to molekyler, men av et enormt antall partikler som alle vekselvirker med hverandre samtidig. Det naturlige spørsmålet blir derfor hvordan de intermolekylære kreftene mellom enkeltmolekyler påvirker makroskopiske størrelser som trykk, volum og temperatur. 

En av de enkleste modellene som kobler disse to beskrivelsesnivåene sammen er **van der Waals-likningen**:

$$\left(P+\frac{a}{V_m^2}\right)(V_m-b)=RT,$$

der $P$ er trykket til gassen, $V_m$ er molart volum av gassen og $R$ er den universelle gasskonstanten. Korreksjonsparametrene $a$ og $b$ kommer vi tilbake til straks. 

Van der Waals-likningen kan betraktes som en forbedring av den ideelle gassloven. Der den ideelle gassloven antar at molekylene er punktpartikler uten innbyrdes vekselvirkning, tar van der Waals-likningen hensyn til to viktige egenskaper ved virkelige gasser: Molekylene opptar plass, og de tiltrekker hverandre.

Disse to effektene beskrives av de to korreksjonsparametrene i likningen. Parameteren b korrigerer for at molekylene har en endelig størrelse, slik at det tilgjengelige volumet blir mindre enn beholdervolumet. Parameteren a korrigerer for de attraktive intermolekylære kreftene, som gjør at trykket blir lavere enn det den ideelle gassloven predikerer.

For å forstå hvorfor disse korreksjonene opptrer, er det nyttig å se nærmere på hva som faktisk bestemmer trykket i en gass.


### Parameteren b: Molekylenes endelige størrelse

Den ideelle gassloven antar at molekylene er punktpartikler, men virkelige molekyler har en endelig utstrekning. To molekyler kan derfor ikke oppta samme rom samtidig, noe som allerede kommer til uttrykk i Lennard–Jones-potensialet gjennom den kraftige frastøtningen ved små avstander.

Hvert molekyl ekskluderer dermed et lite område rundt seg som andre molekyler ikke kan trenge inn i. Det tilgjengelige volumet blir derfor mindre enn det totale beholdervolumet og skrives som $V_m-b$. Parameteren $b$ kan derfor tolkes som et **mål på molekylenes effektive størrelse** og er nært knyttet til den repulsive delen av parpotensialet.

### Parameteren a: Attraktive vekselvirkninger

Den andre korreksjonen skyldes at molekylene tiltrekker hverandre gjennom van der Waals-krefter. Tiltrekningen senker den potensielle energien til systemet og reduserer samtidig trykket, fordi molekylene trekkes svakt mot hverandre i stedet for å bevege seg fritt mot beholderveggene.

Styrken på denne effekten samles i parameteren $a$. **Jo sterkere de intermolekylære tiltrekningene er, desto større blir $a$**, og desto større blir korreksjonen til trykket.

Vi kan knytte denne parameteren direkte til parpotensialet. Dersom vi betrakter ett molekyl omgitt av alle de andre molekylene i gassen, kan den gjennomsnittlige potensielle energien beregnes ved å summere bidragene fra alle nabomolekylene,

$$\langle E_p \rangle=\int_0^\infty u(r)\,\rho\,4\pi r^2\,dr,$$

der $\rho$ er molekyltettheten. For et Lennard–Jones-lignende potensial viser denne integrasjonen at den totale attraktive energien er proporsjonal med $-N^2/V$. Konstanten $a$ kan derfor oppfattes som et mål på den samlede styrken til den attraktive delen av parpotensialet.


:::{admonition} Underveisoppgave
:class: tip

Van der Waals-likningen beskriver en reell gass bedre enn den ideelle gassloven:

$\left(P+\frac{a}{V_m^2}\right)(V_m-b)=RT,$

der $V_m$ er molart volum. Konstanten $a$ tar hensyn til tiltrekning mellom molekylene, og $b$ til at molekylene selv tar opp plass. For CO₂ er $a = 0.364\ \text{Pa}\,\text{m}^6/\text{mol}^2$ og $b = 4.27 \times 10^{-5}\ \text{m}^3/\text{mol}$.

**a)** Leddet $\frac{a}{V_m^2}$ legges til trykket, og $b$ trekkes fra volumet. Forklar med egne ord hvorfor tiltrekning mellom molekylene gjør at det *målte* trykket blir lavere enn for en ideell gass, og hvorfor molekylenes eget volum gjør at det tilgjengelige volumet blir mindre enn $V_m$.

**b)** Løs van der Waals-likningen for trykket $P$, og regn ut $P$ for CO₂ når $V_m = 0.50\ \text{L/mol}$ og $T = 300\ \text{K}$.

**c)** Regn ut trykket for samme $V_m$ og $T$ med den ideelle gassloven $P = \frac{RT}{V_m}$, og sammenlign med svaret i b). Hvilken av dem gir høyest trykk, og henger det sammen med rollen til $a$ og $b$?

:::


:::{dropdown} Løsningsforslag

**a)** Tiltrekningen mellom molekylene drar dem litt innover, så de treffer beholderveggen med litt mindre kraft enn de ellers ville gjort. Det målte trykket blir derfor lavere enn for en ideell gass, og korreksjonen $\frac{a}{V_m^2}$ legges til for å "kompensere" for dette. Molekylene har også et eget volum, så gassen kan ikke presses sammen til null - en del av rommet er allerede opptatt av molekylene selv. Derfor trekkes $b$ fra, slik at $V_m - b$ er det volumet som faktisk er ledig å bevege seg i.

**b)** Vi løser for $P$:

$P = \frac{RT}{V_m - b} - \frac{a}{V_m^2}.$

Vi gjør om volumet: $V_m = 0.50\ \text{L/mol} = 0.50 \times 10^{-3}\ \text{m}^3/\text{mol}$. Da blir

$P = \frac{8.314 \cdot 300}{0.50 \times 10^{-3} - 4.27 \times 10^{-5}} - \frac{0.364}{(0.50 \times 10^{-3})^2} = 4.0 \times 10^{6}\ \text{Pa} \approx 40\ \text{bar}.$

**c)** Med den ideelle gassloven:

$P = \frac{RT}{V_m} = \frac{8.314 \cdot 300}{0.50 \times 10^{-3}} = 5.0 \times 10^{6}\ \text{Pa} \approx 50\ \text{bar}.$

Den ideelle gassloven gir høyest trykk (50 mot 40 bar). Det henger sammen med $a$: tiltrekningen mellom CO₂-molekylene senker det reelle trykket, noe den ideelle loven ikke tar hensyn til. Ved denne høye komprimeringen er avviket stort - nettopp der reelle gasser skiller seg mest fra ideelle.

:::

:::{card} 
**Eksempel: trykk**
^^^

Det makroskopiske trykket i en gass kan forstås som resultatet av to konkurrerende effekter.

**Entropibidraget** skyldes molekylenes termiske bevegelse. Molekylene beveger seg tilfeldig og har en tendens til å spre seg utover for å maksimere antall tilgjengelige mikrotilstander. Dette er den eneste mekanismen som bidrar til trykket i en ideell gass, og leder til den velkjente sammenhengen:

 $$P=\frac{Nk_BT}{V}$$

**Energibidraget** skyldes de attraktive intermolekylære kreftene. Når molekylene tiltrekker hverandre, reduseres den potensielle energien dersom de befinner seg nær hverandre. Denne tiltrekningen trekker molekylene svakt bort fra beholderveggene og gjør at kollisjonene med veggen blir mindre kraftige. Resultatet er et lavere trykk enn for en ideell gass.

Van der Waals-likningen beskriver begge disse effektene ved hjelp av de to korreksjonsleddene $a$ og $b$.
:::


:::{admonition} Gittermodellen – en enkel fysisk forklaring
:class: important

Selv om sammenhengen ovenfor kan utledes direkte fra parpotensialet, finnes det en enklere og mer intuitiv måte å forstå van der Waals-likningen på.

Vi kan tenke oss at volumet er delt inn i et stort antall små celler, der hver celle er omtrent på størrelse med ett molekyl. Hver celle kan enten være tom eller inneholde ett molekyl. Denne modellen gjør det tydelig at molekylene opptar plass: Når et molekyl fyller en celle, kan ingen andre molekyler være der. Antallet tilgjengelige celler blir derfor mindre, og det effektive volumet som molekylene kan bevege seg i, reduseres. Dette er den fysiske betydningen av parameteren $b$, som representerer molekylenes endelige størrelse.

Videre antar vi at hvert molekyl vekselvirker attraktivt med sine nærmeste naboer. Antallet naboer et molekyl har, er proporsjonalt med tettheten $N/V$. Den gjennomsnittlige attraktive energien per molekyl blir derfor proporsjonal med $N/V$. Når vi summerer dette bidraget over alle $N$ molekylene i systemet, får vi et totalt energibidrag som skalerer som

$$N \cdot \frac{N}{V} = \frac{N^2}{V}$$

altså en attraktiv energi som er proporsjonal med $-N^2/V$. Dette er opphavet til parameteren $a$, som beskriver den gjennomsnittlige tiltrekningen mellom molekylene.

Gittermodellen viser dermed på en enkel måte at de to korreksjonene i van der Waals-likningen har ulik fysisk opprinnelse. Parameteren $b$ skyldes at molekylene har en endelig størrelse og derfor reduserer det tilgjengelige volumet, mens parameteren $a$ skyldes de attraktive kreftene mellom molekylene, som senker systemets energi.

![gittermodellen](bilder/gittermodell.png)

*Fotnote:* Bildet der «én celle tilsvarer ett molekyl» gir bare riktig størrelsesorden for $b$. En mer nøyaktig behandling av ekskludert volum for harde kuler viser at $b \approx 4$ ganger molekylvolumet. Koeffisienten i denne enkle modellen må derfor forstås som kvalitativ.

:::

## Parkorrelasjonsfunksjonen g(r)

Så langt har vi beskrevet vekselvirkningen mellom to partikler ved hjelp av parpotensialet $u(r)$. Parpotensialet forteller hvor energetisk gunstig eller ugunstig det er for to partikler å befinne seg i en bestemt avstand fra hverandre. Et virkelig materiale består imidlertid av et enormt antall partikler som alle vekselvirker samtidig. Da er det ikke lenger nok å beskrive ett enkelt partikkelpar, vi trenger også en måte å beskrive hvordan partiklene er fordelt i hele systemet.

Hvis du velger én partikkel som referanse, hvor sannsynlig er det da å finne en annen partikkel i en avstand $r$ fra denne? Svaret på dette spørsmålet beskrives av parkorrelasjonsfunksjonen, $g(r)$. Funksjonen forteller hvordan partiklene er fordelt rundt en referansepartikkel, og gir dermed et mål på hvor strukturert systemet er. Mer presist sammenligner $g(r)$ den lokale tettheten av partikler i avstanden $r$ med den gjennomsnittlige tettheten i systemet. Dette gir en enkel tolkning:

* $g(r)=1$ betyr at partiklene er fordelt som forventet ut fra gjennomsnittstettheten. Det finnes ingen spesiell preferanse for denne avstanden.

* $g(r)>1$ betyr at det er mer sannsynlig enn tilfeldig å finne en partikkel i denne avstanden. Partiklene har en tendens til å samle seg her.

* $g(r)=0$ betyr at det er umulig å finne en partikkel ved denne avstanden.

Den siste situasjonen kjenner vi allerede igjen fra parpotensialet. Ved svært korte avstander dominerer den repulsive delen av $u(r)$, og to partikler kan derfor ikke komme så nær hverandre. Dette gjenspeiles direkte ved at $g(r)=0$.

Vi ser dermed at $u(r)$ og $g(r)$ beskriver to sider av samme problem: parpotensialet forteller hvilke avstander som er energetisk gunstige, mens parkorrelasjonsfunksjonen viser hvordan partiklene faktisk fordeler seg som følge av disse vekselvirkningene.

:::{admonition} Underveisoppgaver
:class: tip

Parkorrelasjonsfunksjonen er $g(r) = 0$ ved svært små avstander. Hvilken del av parpotensialet $u(r)$ svarer dette til, og hvorfor gjør den at to partikler aldri finnes helt tett inntil hverandre? Koble de to beskrivelsene sammen.

:::



### Struktur i gass, væske og fast stoff

Formen til $g(r)$ avhenger av hvor mye orden det finnes i systemet, og gir derfor et enkelt bilde av forskjellen mellom gass, væske og fast stoff.

![gr_fase](bilder/gr.svg)

I en gass beveger partiklene seg nesten uavhengig av hverandre. Bortsett fra den korte avstanden der partiklene ikke kan overlappe, er sannsynligheten for å finne en partikkel omtrent den samme overalt. Derfor nærmer $g(r)$ seg raskt verdien $1$, noe som viser at gassen har svært liten struktur.

I en væske ligger partiklene tettere sammen og påvirker hverandre betydelig mer. Hver partikkel er omgitt av et første lag med naboer, noe som gir en tydelig topp i $g(r)$. Lenger unna ser vi flere, gradvis svakere topper. Dette viser at væsker har lokal orden: partiklene er organiserte over korte avstander, men denne ordenen forsvinner gradvis.

I et fast stoff er partiklene bundet til faste posisjoner i et krystallgitter. Da opptrer en serie skarpe topper i $g(r)$, som fortsetter over lange avstander. Dette gjenspeiler den langtrekkende ordenen som kjennetegner krystallinske materialer.

Parkorrelasjonsfunksjonen gir dermed en direkte kobling mellom de intermolekylære vekselvirkningene vi har beskrevet gjennom $u(r)$, og den strukturen som faktisk oppstår i et materiale.




:::{card}
**Eksempel: Røntgendiffraksjon (XRD)**
^^^

De tre kurvene over er ikke bare en teoretisk konstruksjon – de kan måles direkte. Røntgendiffraksjon (XRD) er et av våre viktigste verktøy for å avsløre materiens oppbygning på atomært nivå, og gir oss en eksperimentell vei til nettopp parkorrelasjonsfunksjonen $g(r)$.

Når røntgenstråler sendes mot et materiale, spres de av elektronene rundt atomene. Siden bølgelengden til røntgenstråling er sammenlignbar med avstanden mellom atomene, interfererer de spredte bølgene med hverandre. Måler vi hvordan intensiteten varierer med spredningsvinkelen, får vi ut strukturfaktoren $S(q)$, og en Fourier-transformasjon av denne gir oss tilbake $g(r)$. Diffraksjonsmønsteret er altså en direkte eksperimentell avbildning av den samme ordenen som $g(r)$ beskriver.

Koblingen til teorien blir spesielt tydelig for et krystallinsk fast stoff. Her ligger atomene i regelmessige plan med fast avstand $d$, og konstruktiv interferens oppstår bare for helt bestemte vinkler, gitt ved Braggs lov:

$$ n\lambda = 2d\sin\theta. $$

Dette er den enkleste modellen for et krystallgitter, men den er nok til å forutsi skarpe, veldefinerte topper i diffraksjonsmønsteret. Fourier-transformerer vi et slikt mønster, får vi en $g(r)$ med en serie skarpe topper som gjentar seg over lange avstander – akkurat den langtrekkende ordenen vi ser i panel (c). På denne måten kan vi gå hele veien fra en enkel geometrisk antakelse om gitteret til en målt kurve som stemmer med teorien.

Den samme metoden skiller fasene fra hverandre, fordi formen på mønsteret gjenspeiler graden av orden. En væske gir noen få brede, diffuse topper som svarer til den korttrekkende ordenen i panel (b), mens en gass gir et nesten strukturløst mønster der $g(r)$ raskt flater ut mot 1, slik som i panel (a). XRD lar oss dermed bekrefte eksperimentelt det teorien forutsier: at strukturen, og dermed $g(r)$, bestemmes av hvor mye orden systemet har.

![XRD](bilder/XRD.jpg)

#### Hva brukes XRD til?

* **Krystallinske faste stoffer:** identifisering av ukjente forbindelser, bestemmelse av krystallstruktur og kornstørrelser, samt kartlegging av gitterdefekter og spenninger i materialet.
* **Porøse og uorganiske rammeverk:** studier av indre hulrom og gitterstrukturer i uorganiske materialer som zeolitter og metall-organiske nettverk (MOF-er).
* **Myke materialer (soft matter):** karakterisering av pakking og orden i alt fra polymerer, såper og flytende krystaller til biologiske membraner og proteiner.
:::

:::{admonition} Underveisoppgaver
:class: tip

Ved svært små avstander er $g(r) = 0$. Hvilken del av parpotensialet $u(r)$ svarer dette til? Forklar hvorfor nettopp denne delen gjør at to partikler aldri kan ligge helt tett inntil hverandre, og koble slik de to beskrivelsene $u(r)$ og $g(r)$ sammen.

:::

## Kondensasjon - en konkurranse mellom energi og entropi

Nå har vi introdusert to viktige beskrivelser av et molekylært system. Parpotensialet $u(r)$ forteller hvilke avstander mellom partikler som er energetisk gunstige, mens parkorrelasjonsfunksjonen $g(r)$ beskriver hvordan partiklene faktisk organiserer seg. Det gjenstår imidlertid ett sentralt spørsmål:

*Hvorfor kondenserer en gass til en væske når temperaturen blir lav nok?*

Svaret ligger i en kontinuerlig konkurranse mellom energi og entropi.

De attraktive intermolekylære kreftene forsøker å trekke partiklene mot hverandre. Dersom partiklene befinner seg nær likevektsavstanden i den attraktive delen av parpotensialet, blir den potensielle energien lavere, og systemet blir mer stabilt.

Samtidig favoriserer entropien det motsatte. Når partiklene kan bevege seg fritt gjennom hele volumet, finnes det langt flere mulige måter å arrangere systemet på. En gass har derfor høyere entropi enn en væske, der partiklene er mer samlet.

Hvilken av disse effektene som dominerer bestemmes av temperaturen. Ved høy temperatur har partiklene stor kinetisk energi, og entropibidraget favoriserer en gassfase. Ved lav temperatur blir gevinsten ved å redusere den potensielle energien viktigere, og partiklene samler seg i den attraktive delen av parpotensialet. Da kan systemet kondensere til en væske.

Denne balansen beskrives av den frie energien,

$$F = U - TS$$

der $U$ er den indre energien (som inkluderer potensiell energi/parpotensialet) og $S$ er entropien. Systemet vil alltid utvikle seg mot den tilstanden som gir lavest fri energi.

Denne konkurransen mellom energi og entropi møtte vi allerede i van der Waals-likningen. Der representerer parameteren $a$ de attraktive intermolekylære kreftene som senker energien, mens molekylenes tendens til å spre seg utover gjenspeiler entropibidraget. I de neste avsnittene skal vi se nærmere på gittermodellen, og løsninger og blandinger.

:::{admonition} Fun fact
:class: note
Is er faktisk mindre tett enn flytende vann — et sjeldent unntak, siden det faste stoffet vanligvis er tettest av de to. Grunnen ligger i hydrogenbindingenes spesielle geometri i iskrystallen, noe du vil lære mer om senere. Poenget her er bare at balansen mellom energi og entropi ikke alltid gir den "opplagte" strukturen!
