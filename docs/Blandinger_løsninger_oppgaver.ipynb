## Uke X - Blandinger og løsninger

### Diskusjonsoppgaver

**Oppgave 1**

a) Som repetisjon, hva er en ideell gass? Hva ville ${\Delta}_{mix}H$ vært for blanding av to ideelle gasser? 

b) Hva er en ideell væske? Hva er ${\Delta }_{mix}H$ for en ideell væske? Hva står ΔH for? Hva betyr dette for interaksjonene og strukturen i væsken? 
	
c) Hva er drivkraften for miksing av to ideelle væsker?
	
d) Man kan regne på egenskaper ved ideelle væsker ved både Henry og Raoults lov. Hva skiller disse fra hverandre? Når bruker man hvilken, og hvorfor kan ikke begge brukes på samme system? 
	
e) Når sier man at en væske er ideell? Kom med to eksempler på systemer med to væsker som oppfører seg ideelt. (Evt. hva som skal til for at de mest sannsynlig vil oppføre seg ideelt.)
	
f) Se for dere et stort volum med vann ved $25 ^{\circ} C$ ($10 \, L$ for eksempel). Når $1 \, mol$ vann videre tilsettes så øker volumet med $18 \, cm^3$, og man sier derfor at det molare volumet til vann er $18 \, cm^3$. Men når man tilsetter $1 \, mol$ vann til et stort volum med etanol ($10 \, L$ for eksempel) ved $25 ^{\circ} C$ øker volumet kun med $14 \, cm^3$. Hvorfor er det slik? Hva kommer dette av? 
	
g) Er systemet i f) en ideell væske? 

h) Hva er partielt molart volum? Hvordan vil det være for en ideell væske?  

i) Hvorfor brukes damptrykk og partialtrykk så mye for å se på egenskaper til væsker?

c) Ved konstant trykk og temperatur er Gibbs-Duhem likningen: $\sum_{J} n_{J}{d\mu }_{J}=0$. Hva betyr dette?
