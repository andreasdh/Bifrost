# Gasser i bevegelse: Kinetisk teori, hastighetsfordelinger og idealgassloven

:::{admonition} Læringsutbytte
:class: note

Etter å ha arbeidet med dette temaet skal du kunne:
1.	Forklare hvorfor en gass behandles som et mange-partikkel-system, og gjør rede for antagelsene bak idealgass-modellen.
2.	Forklare hvorfor gassers egenskaper beskrives med sannsynglighetsfordelinger framfor egenskapene til enkeltmolekyler (kanskje skal ikke ha med - usikker)
3.	Skissere Maxwell-Boltzmann fordelingen og forklare den fysiske betydningen av $v^2$ -leddet og eksponensialleddet $e^{-Mv^2 /2RT}$
4. Gjøre rede for hvordan temperatur henger sammen med gjennomsnittlig kinetisk energi, og forklare hvordan temperatur og energi ikke er samme størrelse.
5. Forklare hvordan makroskopisk trykk oppstår fra molekylære støt mot beholderveggen, og skissere hvordan idealgassloven $PV=nRT$ følger av kinetisk teori. 
6. Forklare hvordan det samme $e^{-E/RT}$ -leddet knytter sammen hastighetsfordeling, kjemisk likevekt og reaksjonskinetikk. 
:::
