## Del 2 – Hvordan er molekylenes hastighet fordelt?

I forrige del så vi at en gass består av et enormt antall molekyler i konstant, tilfeldig bevegelse. Siden det er umulig å følge hvert enkelt molekyl, må vi bruke statistikk for å beskrive gassen som helhet. Det sentrale spørsmålet blir da: **Har alle molekylene i en gass samme fart?**

Det korte svaret er nei. Det er lett å se for seg at molekylene i en beholder suser av sted i et ganske jevnt tempo, men virkeligheten er langt mer kaotisk.

Hvis vi kunne ta et øyeblikksbilde av gassen og måle farten til hvert enkelt molekyl, ville vi funnet enorme forskjeller. Noen kryper av sted i sneglefart, andre suser av sted i ekstremhastighet, mens de fleste ligger et sted midt imellom.

Grunnen til dette spennet er at molekylene hele tiden kolliderer. Ved hvert eneste støt utveksler de energi. Et molekyl som nylig hadde enorm fart, kan miste nesten all energien sin i en kollisjon og bremses brutalt ned. Samtidig kan et langsomt molekyl bli truffet bakfra og skytes av sted som en biljardkule. Både fart og retning endrer seg altså kontinuerlig for det enkelte molekylet.

Likevel skjer det noe bemerkelsesverdig:

> Selv om enkeltmolekylene hele tiden bytter hastighet, forblir den *samlede fordelingen* av hastigheter fullstendig stabil.

Så lenge temperaturen holdes konstant, vil andelen molekyler som er langsomme, middels raske og ekstremt raske være nøyaktig den samme minutt etter minutt. Vi har et system som er i konstant, voldsom endring på mikronivå, men som fremstår fjellstøtt og uendret på makronivå. Det er denne stabile fordelingen vi ønsker å sette på formel.

---

### Tilfeldighet skaper mønster

Dette er ikke et unikt partikkelfenomen; det er en fundamental naturlov. Når nok tilfeldige hendelser virker sammen, oppstår det alltid et mønster.

Tenk på et enkelt eksperiment: Du kaster en mynt 60 ganger. Hvis du gjentar denne serien mange ganger, vil du sjelden få nøyaktig 30 kron hver gang. Noen ganger får du 28, andre ganger 33. Men ekstremt skjeve resultater, som å få bare 10 eller hele 55 kron, er forsvinnende sjeldne. Tegner du opp hvor ofte hvert resultat forekommer, får du en karakteristisk, symmetrisk klokkekurve – en **normalfordeling** (eller gaussfordeling).

![myntkast](bilder/myntkast.svg)

Fysikeren James Clerk Maxwell innså at de utallige kollisjonene i en gass følger akkurat det samme prinsippet. I stedet for å kaste bort tid på å spore enkeltpartikler, beskrev han sannsynligheten for at et tilfeldig molekyl har en bestemt fart.

Resultatet av dette arbeidet kalles **Maxwell-Boltzmann-fordelingen**:

*(Figur: Maxwell-Boltzmann-fordelingen)*

Det er verdt å merke seg at denne kurven har en fundamental forskjell fra den klassiske gausskurven: **Den er ikke symmetrisk.** Siden et molekyl aldri kan ha en negativ fart, stopper kurven kontant ved null på venstre side. Mot høyre har den derimot ingen fast grense, noe som gir kurven en lang, slank "hale" inn i de ekstremt høye hastighetene.

---

### Hva påvirker fordelingen?

Formen på denne kurven er ikke hugget i stein. Den styres i hovedsak av to fysiske faktorer: **temperatur** og **molekylmasse**.

#### 1. Temperatur

Når gassen varmes opp, tilføres molekylene mer kinetisk energi. Dette gjør at hele fordelingskurven flater ut og forskyver seg mot høyre. Gjennomsnittsfarten øker, spredningen i hastighet blir større, og en langt større andel av molekylene befinner seg nå ute i den raske "halen" av kurven.

#### 2. Molekylmasse

Ved én og samme temperatur har tunge og lette molekyler nøyaktig samme gjennomsnittlige *kinetiske energi*. Men siden kinetisk energi er gitt ved $E_k = \frac{1}{2}mv^2$, betyr det at masse og fart må balansere hverandre. Lette gasser (som helium) vil derfor ha en mye bredere fordeling som strekker seg mot langt høyere hastigheter enn tunge gasser (som nitrogen eller oksygen).

Selve det matematiske uttrykket som beskriver denne kurven ser slik ut:

$$f(v) = 4\pi \left(\frac{m}{2\pi k_B T}\right)^{3/2} v^2 e^{-\frac{mv^2}{2k_B T}}$$

Vi skal ikke utlede dette komplekse uttrykket matematisk her. Det viktige for oss er å forstå hva ligningen forteller oss i praksis: Ved en gitt temperatur har gassen en forutsigbar fordeling av energi, og denne mikroskopiske profilen dikterer nesten alle de makroskopiske egenskapene til gassen.

I neste del skal vi bruke akkurat denne fordelingen til å gi begrepet *temperatur* en solid, fysisk tolkning.

---

:::{admonition} Hvorfor er Maxwell-Boltzmann-fordelingen kjemiens ryggrad?
:class: tip

Maxwell-Boltzmann-fordelingen beskriver ikke bare gasspartikler i bevegelse; den er selve fundamentet for å forstå kjemisk oppførsel:

* **Aktiveringsenergi:** For at to molekyler skal reagere, må de kollidere med en viss minimumsenergi. Fordelingen viser oss nøyaktig hvor stor (eller forsvinnende liten) andel av molekylene som har høy nok fart til å overvinne denne barrieren. Dette er fundamentet for Arrhenius-ligningen og reaksjonshastigheter.
* **Separasjon og diffusjon:** Siden lette molekyler i snitt beveger seg raskere enn tunge molekyler ved samme temperatur, sprer de seg (diffunderer) i ulikt tempo. Det er denne fysiske forskjellen vi utnytter i separasjonsteknikker som gasskromatografi.
* **Temperaturfølsomhet:** Siden en liten temperaturøkning dytter en uforholdsmessig stor andel av partiklene ut i den energirike halen av kurven, forklarer fordelingen hvorfor kjemiske reaksjoner skyter voldsom fart når vi varmer opp kolben.

## Del 3 – Hva er egentlig temperatur?

I forrige del så vi at molekylene i en gass ikke beveger seg med samme fart, men følger Maxwell-Boltzmann-fordelingen. Dette leder naturlig til et nytt spørsmål: Hva er egentlig temperatur?

Vi bruker ordet temperatur hver dag og forbinder det med hvor varmt eller kaldt noe føles. I termodynamikken har begrepet en langt mer presis betydning. Mens vi senere i kurset skal se at den mest generelle definisjonen av temperatur handler om energi og statistisk uorden (entropi), kan vi for en idealgass forstå temperaturen på en langt mer intuitiv måte: **som ren molekylbevegelse.**



### Temperatur som molekylbevegelse

I en idealgass antar vi at molekylene beveger seg helt fritt, uten intermolekylære krefter som trekker eller skyver på dem mellom kollisjonene. Det betyr at gassens indre energi nesten utelukkende består av partiklenes egen bevegelsesenergi – altså kinetisk energi.

Ved å bruke Maxwell-Boltzmann-fordelingen kan man vise at den gjennomsnittlige kinetiske energien $\langle E_k \rangle$ til ett enkelt gassmolekyl er direkte proporsjonal med den absolutte temperaturen:

$$\langle E_k \rangle = \frac{3}{2} k_B T$$

Her er $k_B$ Boltzmanns konstant. Det er kritisk å merke seg at denne ligningen **ikke** betyr at alle molekyler har nøyaktig denne energien. Noen partikler suser av sted med voldsom fart, mens andre nesten står stille. Ligningen beskriver utelukkende *gjennomsnittet* for gassen.

Hvis vi i stedet zoomer ut og ser på en hel gassprøve med $n$ mol molekyler, får vi den makroskopiske tvillingligningen for den totale kinetiske energien ($K$):

$$K = \frac{3}{2} nRT$$

der $R$ er den universelle gasskonstanten. Mens den første ligningen beskriver ett enslig molekyl på mikronivå, beskriver den andre hele gassbeholderen på makronivå. Begge formidler akkurat den samme fysiske kjerneideen:

> **Kjerneidé:** Temperaturen i en gass er et direkte mål på den gjennomsnittlige kinetiske energien til molekylene. Den forteller oss hvor intenst partiklene i snitt vibrerer og spretter rundt.

---

### Temperatur er ikke det samme som energi

Siden temperatur og energi henger så tett sammen, er det lett å gå i fellen og tro at de er to ord for samme sak. Det er de ikke.

Det beste beviset på dette finner vi når et stoff skifter fase – for eksempel når en isbit smelter:

* Når isen når smeltepunktet ($0^\circ\text{C}$), fortsetter den å ta opp energi fra omgivelsene for å smelte.
* Til tross for at vi kontinuerlig tilfører energi, **stiger ikke temperaturen** en eneste grad før all isen har blitt til flytende vann.

Hvor blir det av energien? Den går ikke til å øke farten på molekylene (som ville økt temperaturen), men brukes i stedet til å bryte opp de låste bindingene i krystallstrukturen. Energien endrer altså hvordan vannmolekylene er *organisert*.

> **Innsikt til senere:** Dette viser at temperatur ikke bare måler hvor mye energi et system inneholder, men hvordan denne energien er fordelt. Senere i termodynamikken skal vi se at dette beskrives presist av **entropi** – som måler antall måter (mikrotilstander) et system kan fordele energien sin på.

---

### Temperaturen endrer hele fordelingen

Siden temperaturen bestemmer gjennomsnittsenergien, styrer den også formen på Maxwell-Boltzmann-fordelingen.

Når temperaturen øker, skjer det to ting med fordelingskurven:

1. **Den forskyves:** Hele kurven flytter seg mot høyre, mot høyere hastigheter og energier.
2. **Den flates ut:** Kurven blir bredere. Det betyr at spredningen i partiklenes fart blir større.

*(Se for deg kurven forskyve seg her)*

Dette har enorme praktiske konsekvenser, spesielt i kjemien. For at en kjemisk reaksjon skal skje, må molekylene kollidere med en viss minimumsenergi – kjent som *aktiveringsenergien*. Når temperaturen øker og fordelingen blir bredere, vil en dramatisk mye større andel av molekylene krysse denne energigrensen. Det er hovedgrunnen til at selv en liten temperaturøkning kan få kjemiske reaksjoner til å gå vanvittig mye raskere.

## 4 - Hvor kommer trykk fra?

I Del 3 fant vi ut hva temperatur faktisk er på molekylnivå. Nå skal vi gjøre noe som er selve kjernen i kinetisk gassteori: Vi skal utlede en makroskopisk, målbar størrelse utelukkende fra molekylær mekanikk.

Dette er verdt å stoppe opp ved et øyeblikk. Trykk er noe håndgripelig som du kan lese av på et manometer, og det er en størrelse ingeniører og forskere har brukt i århundrer. Vi skal nå vise at dette trykket følger direkte – uten noen nye antagelser – fra at gassen består av partikler som spretter rundt. Ingen empiri, ingen gjetting; bare partikler og Newtons lover.

### Hvorfor har en gass trykk?

Når du åpner en brusflaske, hører du det med én gang: et karakteristisk sus når gassen slipper ut. Før du åpnet flasken, presset gassen inni mot glasset, mot væskeoverflaten og mot korken – jevnt, i alle retninger, og helt uten pause. Det samme skjer inni et bildekk, i en ballong, og i lufta rundt deg akkurat nå.

![brus](bilder/brus.svg)

Men hva er egentlig dette presset? Hva er det som faktisk dytter?

> **Definisjon:** Trykk ($P$) er definert som kraft per areal:
> 
> $$P = \frac{F}{A}$$
> 
> 
> 
> SI-enheten for trykk er pascal ($\text{Pa}$), der $1\ \text{Pa} = 1\ \text{N/m}^2$.

### Svaret ligger på molekylnivå

For å forstå hvor denne kraften kommer fra, må vi zoome helt inn på partiklene:

> **Kjerneidé:** Trykk oppstår fordi et enormt antall molekyler kolliderer med veggene i beholderen.

Ett enkelt molekyl som treffer en vegg, gir fra seg et ekstremt lite dytt – så lite at det er umulig å måle for oss. Men en gass inneholder astronomiske mengder molekyler, og hvert av dem kolliderer med veggen utallige ganger per sekund. Når vi summerer alle disse ørsmå, tilfeldige støtene, opplever vi det makroskopisk som en jevn, konstant kraft mot veggen. Det er denne kollektive kraften vi kaller trykk.


![vegg](bilder/vegg.svg)


Dette er et kroneksempel på hvordan en makroskopisk egenskap oppstår fra kaotiske, mikroskopiske hendelser. Trykket "finnes" ikke i ett enkelt molekyl, like lite som temperatur gjør det. Det er en kollektiv, statistisk egenskap som krever et fellesskap av partikler for å eksistere.

### Litt om bevegelsesmengde
For å utlede gasslovene fra bunnen av trenger vi bare én enkelt størrelse fra mekanikken.

**Bevegelsesmengde** (momentum) er definert som $p = mv$ (masse ganger hastighet). Newtons andre lov sier egentlig at en kraft $F$ er endring i bevegelsesmengde per tid:

$$F = \frac{\Delta p}{\Delta t}$$

Dette er hele verktøykassen vår. Skal vi finne kraften gassen utøver på en vegg, trenger vi bare å finne ut to ting: **hvor mye** bevegelsesmengde som overføres i hvert støt, og **hvor ofte** støtene skjer.

---

### Utledningen, trinn for trinn

Vi ser på en rektangulær boks med lengde $a$ langs $x$-aksen, der veggen vi studerer (vegg 1) har areal $A$. Volumet til boksen er da $V = A \cdot a$. Boksen inneholder $N$ molekyler, hvert med masse $m$.

#### Trinn 1: Ett støt

Et molekyl treffer veggen med hastighetskomponenten $v_x$. Vi antar at støtet er elastisk. Det betyr at molekylet spretter tilbake med nøyaktig samme fart, men i motsatt retning: $v_x \rightarrow -v_x$. Bevegelsen parallelt med veggen ($v_y, v_z$) blir ikke påvirket – molekylet reflekteres akkurat som en lysstråle fra et speil.

Molekylets bevegelsesmengde langs $x$-aksen endrer seg dermed fra $+m v_x$ til $-m v_x$. Endringen for molekylet er $-2 m v_x$. Etter Newtons tredje lov (kraft og motkraft) får veggen en like stor dytt i motsatt retning:

$$\vert{}\Delta p\vert{} = 2 m v_x$$

#### Trinn 2: Hvor ofte skjer støtene?

Ett enkelt støt sier ingenting om en kontinuerlig kraft; vi må vite hvor hyppig støtene kommer. Etter å ha truffet veggen, farer molekylet over til den motsatte veggen, spretter der, og returnerer til vår vegg. Det har da reist strekningen $2a$ langs $x$-aksen. Tiden $\Delta t$ mellom to støt på samme vegg er derfor:

$$\Delta t = \frac{2a}{v_x}$$

> Raske molekyler slår både hardere (Trinn 1) og oftere (Trinn 2). Fordi farten bidrar i begge leddene, vil den om et øyeblikk dukke opp i andre potens ($v^2$).

#### Trinn 3: Kraften fra ett molekyl

Nå kombinerer vi de to første trinnene. Den tidsmidlede kraften $F_1$ fra ett enkelt molekyl på veggen blir:

$$F_1 = \frac{\vert{}\Delta p\vert{}}{\Delta t} = \frac{2 m v_x}{2a / v_x} = \frac{m v_x^2}{a}$$

Her dukket $v_x^2$ opp, akkurat som forventet.

#### Trinn 4: Summen over alle $N$ molekyler

Siden molekylene i en gass har ulike hastigheter, kan vi ikke bare gange med $N$. Vi må bruke gjennomsnittet av den kvadratiske hastigheten, $\langle v_x^2 \rangle$, for alle molekylene:

$$F = \frac{N m \langle v_x^2 \rangle}{a}$$

*Det er her Maxwell-Boltzmann-fordelingen ligger og ulmer i bakgrunnen: Det er den som sikrer at dette gjennomsnittet er en stabil og veldefinert størrelse.*

#### Trinn 5: Fra $v_x$ til den totale farten $v$

Vi vil helst uttrykke kraften ved den faktiske farten $v$, ikke bare $x$-komponenten. Fartvektoren er relatert til komponentene sine ved:

$$\langle v^2 \rangle = \langle v_x^2 \rangle + \langle v_y^2 \rangle + \langle v_z^2 \rangle$$

Her bruker vi et rent symmetriargument: Siden ingen retning i rommet er foretrukket i en tilfeldig gass, må bevegelsen i de tre retningene i snitt være helt like. Dermed er $\langle v_x^2 \rangle = \langle v_y^2 \rangle = \langle v_z^2 \rangle$, som gir:

$$\langle v_x^2 \rangle = \frac{1}{3}\langle v^2 \rangle$$

#### Trinn 6: Trykket oppstår

Trykk $P$ er kraft per areal ($P = F/A$). Siden volumet til boksen er $V = A \cdot a$, får vi:

$$P = \frac{F}{A} = \frac{N m \langle v_x^2 \rangle}{A \cdot a} = \frac{N m \langle v_x^2 \rangle}{V}$$

Vi setter inn $\langle v_x^2 \rangle = \frac{1}{3}\langle v^2 \rangle$ og flytter $V$ over på venstre side:

$$PV = \frac{1}{3} N m \langle v^2 \rangle$$

Dette er et rent mekanisk resultat. Legg merke til at vi ikke har nevnt ordet *temperatur* ennå.

Vi kan skrive dette om for å gjøre fysikken tydeligere. Siden den gjennomsnittlige kinetiske energien per molekyl er $\langle E_k \rangle = \frac{1}{2} m \langle v^2 \rangle$, kan vi erstatte $m\langle v^2\rangle$ med $2\langle E_k\rangle$:

$$PV = \frac{2}{3} N \langle E_k \rangle$$

Dette forteller oss noe fundamentalt: **Trykk er direkte proporsjonalt med molekylenes gjennomsnittlige kinetiske energi.**

#### Trinn 7: Temperaturen kobles på

Det er først nå vi henter inn broen mellom mikroskopisk bevegelse og makroskopisk temperatur:

$$\langle E_k \rangle = \frac{3}{2} k_B T$$

Dette er en fysisk definisjon som kommer utenfra mekanikken. Setter vi dette inn, faller alt på plass:

$$PV = \frac{2}{3} N \cdot \frac{3}{2} k_B T = N k_B T$$

Siden $N k_B = n R$ (der $n$ er antall mol og $R$ er den universelle gasskonstanten), står vi igjen med:

$$PV = n R T$$

Vi har utledet **idealgassloven** utelukkende ved å anta partikler i tilfeldig bevegelse.

:class: tip
### Hva vi egentlig har vist

Idealgassloven falt ikke ned fra himmelen. Historisk ble den oppdaget eksperimentelt av forskere som Boyle, Charles og Clapeyron gjennom nitidige målinger på gasser – lenge før man visste at molekyler fantes.

Det vi nettopp har gjort, er å bevise at loven er en direkte konsekvens av Newtons lover brukt på et enormt antall partikler. Symbolene i $PV = nRT$ har nå fått en konkret, mikroskopisk betydning:

| Symbol | Mikroskopisk betydning |
| --- | --- |
| **$P$ (Trykk)** | Hvor hardt og hvor ofte molekylene kolliderer med veggen. |
| **$V$ (Volum)** | Hvor stort rom partiklene har å bevege seg i (og dermed hvor lang tid det går mellom hver kollisjon). |
| **$N$ (Antall)** | Hvor mange molekyler som er med på å dytte. |
| **$T$ (Temperatur)** | Et direkte mål på partiklenes gjennomsnittlige kinetiske energi (hvor fort de beveger seg). |

:::

---

### Tre viktige innsikter å ta med videre

Før vi dykker dypere inn i termodynamikken, er det tre konsepter fra denne utledningen som er viktigere enn selve matematikken:

1. **Trykk er en statistisk størrelse.** Ett enkelt molekyl har ikke et "trykk", akkurat som det ikke har en "temperatur". Trykk er en makroskopisk egenskap som først oppstår og stabiliserer seg når vi midler over enorme mengder partikler ($\langle v_x^2 \rangle$).
2. **Trykk er en intensiv størrelse.** Hvis du slår sammen to identiske beholdere med gass, dobler du både volumet ($V$) og antall partikler ($N$). Men trykket ($P$) forblir det samme. Trykk måler den lokale energitettheten (kraft per areal), ikke summen av all energi.
3. **Mønsteret i tankegangen er kjernen i statistisk mekanikk.** Vi kommer til å bruke denne strategien igjen og igjen:

$$\text{Beskriv én partikkel med enkel fysikk \rightarrow Ta statistisk gjennomsnitt \rightarrow Få en målbar makroskopisk størrelse}$$

### En første kikk på reelle gasser

Utledningen over er elegant, men den hviler på noen strenge forenklinger. Vi har antatt at gassen er **ideal**. Det betyr at:

* Molekylene er uendelig små punkter uten eget volum.
* Det ikke finnes noen krefter mellom molekylene (de merker ikke hverandre før de kolliderer).

I virkeligheten er ingen gasser helt ideelle. Molekyler tar faktisk opp fysisk plass, og de tiltrekker hverandre svakt når de er nær hverandre. Vi kan se for oss avvikene slik:

* **Molekylenes eget volum:** Tenk på molekylene som små kuler. Siden de tar opp plass, har de andre molekylene litt mindre ledig rom å bevege seg på enn det totale volumet $V$ skulle tilsi. Dette fører til at de kolliderer med veggene *oftere*, noe som trekker trykket **opp**.
* **Intermolekylære tiltrekningskrefter:** Når et molekyl er på vei til å dundre inn i veggen, vil nabomolekylene bak trekke litt i det med svake krefter. Dette "holder igjen" molekylet litt, slik at støtet mot veggen blir svakere. Dette trekker trykket **ned**.

Ved romtemperatur og vanlig atmosfærisk trykk er disse avvikene så små at idealgassloven fungerer nesten perfekt. Men presser vi gassen sammen til høyt trykk (slik at de tvinges tett sammen) eller kjøler den ned (slik at de beveger seg sakte nok til at tiltrekningen får tid til å virke), begynner de reelle kreftene å dominere. Dette er starten på historien om intermolekylære krefter og faseoverganger, som vi skal se nærmere på senere.
:::
