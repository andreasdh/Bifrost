## 2 - Hvorfor beveger molekylene seg?

I forrige del så vi at en gass består av et enormt antall molekyler som beveger seg tilfeldig. Selv
om vi ikke kan følge hvert enkelt molekyl, kan vi beskrive hvordan hele gassen oppfører seg ved
hjelp av statistikk. Det sentrale spørsmålet blir derfor: **Hvordan er molekylenes hastighet
fordelt?**

Et naturlig delspørsmål er om alle molekylene har samme fart. Det korte svaret er nei.

Hvis vi kunne «fryse» bevegelsen i en gass et øyeblikk og måle farten til hvert eneste molekyl,
ville vi oppdaget at de beveger seg med mange forskjellige hastigheter. Noen molekyler beveger seg
sakte, andre svært raskt, mens de fleste har en fart et sted imellom. Grunnen er at molekylene hele
tiden kolliderer med hverandre. Ved hver kollisjon overføres energi: et raskt molekyl kan miste noe
av energien sin og bli langsommere, mens et annet mottar energien og får høyere fart. Både fart og
retning endrer seg altså kontinuerlig for det enkelte molekylet.

Likevel skjer det noe bemerkelsesverdig. Selv om de enkelte molekylene stadig bytter fart, forblir
den samlede fordelingen av hastigheter nesten uendret. Antallet langsomme, middels raske og
raske molekyler holder seg omtrent konstant så lenge temperaturen er den samme. Dette er et godt
eksempel på hvordan et system kan være i konstant endring på mikroskopisk nivå, samtidig som det
framstår stabilt på makroskopisk nivå. Det er denne stabile fordelingen vi ønsker å beskrive.

### Tilfeldighet gir mønster

Dette er ikke et fenomen som bare gjelder gasser. Mange tilfeldige prosesser i naturen viser seg å
følge regelmessige fordelinger.

Et enkelt eksempel er å kaste en mynt 60 ganger. Gjentar vi forsøket mange ganger, vil vi ikke
alltid få nøyaktig 30 kron. Noen ganger får vi 28, andre ganger 33. Resultater langt unna
gjennomsnittet, som 15 eller 45, er derimot svært sjeldne. Tegner vi hvor ofte hvert resultat
forekommer, får vi en karakteristisk klokkeformet kurve – en normalfordeling (gaussfordeling).
Denne typen fordeling dukker opp nettopp når mange uavhengige tilfeldige hendelser virker sammen.

(FIGURE myntkast og gauss)

James Clerk Maxwell innså at de utallige kollisjonene mellom molekylene i en gass er et slikt
system. Ved første øyekast kan det virke overraskende at noe så kaotisk som millioner av kolliderende
molekyler kan beskrives med en enkel matematisk funksjon – men det er nettopp et generelt trekk ved
store systemer. I stedet for å følge hvert enkelt molekyl, beskrev Maxwell sannsynligheten for
at et molekyl har en bestemt fart. Resultatet er at de fleste molekylene får en middels fart, mens
bare et lite antall beveger seg svært sakte eller svært raskt.

(SIMULERINGEN MED BOLTZMANN HER ALTSå)



Fordelingen er ikke symmetrisk. Det finnes ingen negative farter, så kurven får en lang hale mot
høyere hastigheter. Denne formen er karakteristisk for alle idealgasser.

### Hva påvirker fordelingen?

To faktorer har særlig stor betydning for hvordan Maxwell–Boltzmann-fordelingen ser ut:
temperatur og molekylmasse.

Når temperaturen øker, får molekylene mer kinetisk energi. Hele fordelingen flytter seg mot høyere
hastigheter, samtidig som den blir bredere. Flere molekyler får høy fart, og en større andel kan
delta i prosesser som krever høy energi, for eksempel kjemiske reaksjoner.

Molekylmassen spiller også en viktig rolle. Ved samme temperatur beveger lette molekyler seg i
gjennomsnitt raskere enn tunge. Derfor har helium en bredere fordeling med høyere typiske
hastigheter enn nitrogen eller oksygen. Disse sammenhengene blir viktige senere, når vi skal forstå
diffusjon, gasskromatografi og reaksjonshastigheter.

kan skrives som:

I neste del skal vi bruke fordelingen til å gi temperatur en fysisk tolkning. Vi skal se at

kinetiske energien til molekylene.


:::{admonition} Hvorfor er Maxwell–Boltzmann-fordelingen så viktig i kjemi?
:class: tip

Fordelingen beskriver ikke bare hvordan molekylene i en gass beveger seg. Den ligger også til
grunn for flere sentrale begreper i kjemi:


- Bare en liten andel av molekylene har høy nok energi til å reagere. Dette er grunnlaget for reaksjonshastigheter og Arrhenius-ligningen.
- Lette molekyler beveger seg raskere enn tunge, noe som påvirker diffusjon og separasjonsteknikker som gasskromatografi.
- Når temperaturen øker, forskyves hele fordelingen mot høyere energier, og sannsynligheten for at molekyler kan overvinne aktiveringsenergien i en reaksjon, øker.

Maxwell–Boltzmann-fordelingen er derfor en av de viktigste koblingene mellom molekylenes bevegelse
og gassens kjemiske oppførsel.

### Temperatur er ikke det samme som energi

Det er lett å tro at temperatur og energi er det samme, siden begge øker når vi varmer opp et stoff.
Likevel er de forskjellige størrelser.


Et godt eksempel er en isbit som smelter. Når isen når smeltepunktet, kan vi fortsette å tilføre
energi uten at temperaturen øker. Energien brukes i stedet til å endre hvordan vannmolekylene er
organisert, slik at isen smelter til væske. Dette viser at temperaturen ikke måler hvor mye energi
et system inneholder – den forteller hvordan energien er fordelt mellom molekylene.

Senere skal vi se at denne sammenhengen beskrives av entropi. Når et system får tilført energi,
øker vanligvis antallet tilgjengelige mikrotilstander, og temperaturen bestemmer hvor stor denne
økningen blir.


### Temperaturen endrer hele fordelingen
Siden temperaturen bestemmer den gjennomsnittlige kinetiske energien, påvirker den også
Maxwell–Boltzmann-fordelingen. Når temperaturen øker, forskyves hele fordelingen mot høyere
energier, samtidig som den blir bredere, slik at flere molekyler får høy nok energi til å delta i
kjemiske reaksjoner. Dette er en av de viktigste grunnene til at reaksjonshastigheter øker når
temperaturen stiger.
