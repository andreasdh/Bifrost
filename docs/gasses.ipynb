## Hvordan er molekylenes hastighet fordelt?

Siden det er umulig å følge hvert enkelt molekyl, må vi bruke statistikk for å beskrive gassen som helhet. Det sentrale spørsmålet er: **Har alle molekylene i en gass samme fart?**

Det korte svaret er nei. Det er lett å se for seg at molekylene i en beholder suser av sted i et ganske jevnt tempo, men virkeligheten er langt mer kaotisk.

Hvis vi kunne ta et øyeblikksbilde av gassen og måle farten til hvert enkelt molekyl, ville vi funnet enorme forskjeller. Noen kryper av sted i sneglefart, andre suser av sted i ekstremhastighet, mens de fleste ligger et sted midt imellom.

Grunnen til dette spennet er at molekylene hele tiden kolliderer. Ved hvert eneste støt utveksler de energi. Et molekyl som nylig hadde enorm fart, kan miste nesten all energien sin i en kollisjon og bremses brutalt ned. Samtidig kan et langsomt molekyl bli truffet bakfra og skytes av sted som en biljardkule. Både fart og retning endrer seg altså kontinuerlig for det enkelte molekylet.

Likevel skjer det noe bemerkelsesverdig:

*Selv om enkeltmolekylene hele tiden bytter hastighet, forblir den samlede fordelingen av hastigheter fullstendig stabil.*

Så lenge temperaturen holdes konstant, vil andelen molekyler som er langsomme, middels raske og ekstremt raske være nøyaktig den samme minutt etter minutt. Vi har et system som er i konstant, voldsom endring på mikronivå, men som fremstår fjellstøtt og uendret på makronivå. Vi kommer snart til å presentere en funksjon som beskriver denne stabile fordelingen.

### Maxwell-Boltzmann fordelingen

Dette er ikke et unikt partikkelfenomen; det er en fundamental naturlov at tilfeldighet skaper mønster. Når nok tilfeldige hendelser virker sammen, oppstår det alltid et mønster. Se [sentralgrenseteoremet](https://andreasdh.github.io/Bifrost/docs/mattehjelp.html#sentralgrenseteoremet). 

Tenk på et enkelt eksperiment: Du kaster en mynt 60 ganger. Hvis du gjentar denne serien mange ganger, vil du sjelden få nøyaktig 30 kron hver gang. Noen ganger får du 28, andre ganger 33. Men ekstremt skjeve resultater, som å få bare 10 eller hele 55 kron, er forsvinnende sjeldne. Tegner du opp hvor ofte hvert resultat forekommer, får du en karakteristisk, symmetrisk klokkekurve – en **normalfordeling** (eller gaussfordeling).

![myntkast](bilder/myntkast.svg)

Fysikeren James Clerk Maxwell innså at de utallige kollisjonene i en gass følger akkurat det samme prinsippet. I stedet for å kaste bort tid på å spore enkeltpartikler, beskrev han sannsynligheten for at et tilfeldig molekyl har en bestemt fart.

### Hva påvirker fordelingen?

Formen på denne kurven er ikke hugget i stein. Den styres i hovedsak av to fysiske faktorer: **temperatur** og **molekylmasse**.

#### 1. Temperatur

Når gassen varmes opp, tilføres molekylene mer kinetisk energi. Dette gjør at hele fordelingskurven flater ut og forskyver seg mot høyre. Gjennomsnittsfarten øker, spredningen i hastighet blir større, og en langt større andel av molekylene befinner seg nå ute i den raske "halen" av kurven.

#### 2. Molekylmasse

Ved én og samme temperatur har tunge og lette molekyler nøyaktig samme gjennomsnittlige *kinetiske energi*. Men siden kinetisk energi er gitt ved $E_k = \frac{1}{2}mv^2$, betyr det at masse og fart må balansere hverandre. Lette gasser (som helium) vil derfor ha en mye bredere fordeling som strekker seg mot langt høyere hastigheter enn tunge gasser (som nitrogen eller oksygen).

:::{admonition} Maxwell-Boltzmann-fordelingen
:class: important
Antallet partikler $f(v)$ med en hastighet $v$ er gitt ved:

$$f(v) = 4\pi \left(\frac{m}{2\pi k_B T}\right)^{3/2} v^2 e^{-\frac{mv^2}{2k_B T}},$$

der $m$ er massen til partiklene. 
:::

Vi skal ikke utlede dette komplekse uttrykket matematisk her. Det viktige for oss er å forstå hva ligningen forteller oss i praksis: Ved en gitt temperatur har gassen en forutsigbar fordeling av energi, og denne mikroskopiske profilen dikterer nesten alle de makroskopiske egenskapene til gassen.

Videre skal vi bruke akkurat denne fordelingen til å gi begrepet *temperatur* en solid, fysisk tolkning.

---

:::{admonition} Hvorfor er Maxwell-Boltzmann-fordelingen kjemiens ryggrad?
:class: note

Maxwell-Boltzmann-fordelingen beskriver ikke bare gasspartikler i bevegelse; den er selve fundamentet for å forstå kjemisk oppførsel:

* **Aktiveringsenergi:** For at to molekyler skal reagere, må de kollidere med en viss minimumsenergi. Fordelingen viser oss nøyaktig hvor stor (eller forsvinnende liten) andel av molekylene som har høy nok fart til å overvinne denne barrieren. Dette er fundamentet for Arrhenius-ligningen og reaksjonshastigheter.
* **Separasjon og diffusjon:** Siden lette molekyler i snitt beveger seg raskere enn tunge molekyler ved samme temperatur, sprer de seg (diffunderer) i ulikt tempo. Det er denne fysiske forskjellen vi utnytter i separasjonsteknikker som gasskromatografi.
* **Temperaturfølsomhet:** Siden en liten temperaturøkning dytter en uforholdsmessig stor andel av partiklene ut i den energirike halen av kurven, forklarer fordelingen hvorfor kjemiske reaksjoner skyter voldsom fart når vi varmer opp kolben.

## Hva er egentlig temperatur?

I forrige del så vi at molekylene i en gass ikke beveger seg med samme fart, men følger Maxwell-Boltzmann-fordelingen. Dette leder naturlig til et nytt spørsmål: Hva er egentlig temperatur?

Vi bruker ordet temperatur hver dag og forbinder det med hvor varmt eller kaldt noe føles. I termodynamikken har begrepet en langt mer presis betydning. Mens vi senere i kurset skal se at den mest generelle definisjonen av temperatur handler om energi og statistisk uorden (entropi), kan vi for en ideell gass forstå temperaturen på en langt mer intuitiv måte: **som ren molekylbevegelse.**



### Temperatur som molekylbevegelse

I en ideell gass antar vi at molekylene beveger seg helt fritt, uten intermolekylære krefter som trekker eller skyver på dem mellom kollisjonene. Det betyr at gassens indre energi nesten utelukkende består av partiklenes egen bevegelsesenergi – altså kinetisk energi.

Ved å bruke Maxwell-Boltzmann-fordelingen kan man vise at den gjennomsnittlige kinetiske energien $\langle E_k \rangle$ til ett enkelt gassmolekyl er direkte proporsjonal med den absolutte temperaturen:

$$\langle E_k \rangle = \frac{3}{2} k_B T$$

Det er kritisk å merke seg at denne ligningen **ikke** betyr at alle molekyler har nøyaktig denne energien. Noen partikler suser av sted med voldsom fart, mens andre nesten står stille. Ligningen beskriver utelukkende *gjennomsnittet* for gassen.

Hvis vi i stedet zoomer ut og ser på en hel gassprøve med $n$ mol molekyler, får vi den makroskopiske tvillingligningen for den totale kinetiske energien ($K$):

$$K = \frac{3}{2} nRT$$

der $R$ er den universelle gasskonstanten. Mens den første ligningen beskriver ett enslig molekyl på mikronivå, beskriver den andre hele gassbeholderen på makronivå. Begge formidler akkurat den samme fysiske kjerneideen:

:::{admonition} Kjerneidé
:class: important
Temperaturen i en gass er et direkte mål på den gjennomsnittlige kinetiske energien til molekylene. Den forteller oss hvor intenst partiklene i snitt vibrerer og spretter rundt.
:::

:::{admonition} OBS: per molekyl eller per mol?
:class: error

Gjennom kapitlet møter du de samme sammenhengene i to varianter: én som gjelder ett enkelt molekyl, og én som gjelder ett mol. 

På mikronivå, når vi ser på ett molekyl, bruker vi Boltzmanns konstant $k_B$ og molekylmassen $m$. Slik skriver vi den gjennomsnittlige kinetiske energien som $\langle E_k\rangle = \tfrac{3}{2}k_B T$ og Boltzmann-faktoren som $e^{-E/k_B T}$. På makronivå, når vi regner på et helt mol, bruker vi den universelle gasskonstanten $R$ og den molare massen $M$. Da blir de samme uttrykkene $E_k = \tfrac{3}{2}nRT$ og $e^{-E_a/RT}$.

De to nivåene henger sammen gjennom Avogadros tall $N_A \approx 6{,}022\times10^{23}$, som ganske enkelt forteller hvor mange molekyler det er i ett mol:

$$R = N_A \, k_B, \qquad M = N_A \, m.$$

Fysikken er nøyaktig den samme – du velger bare enheten som passer regnestykket. Kjenner du massen til ett molekyl, bruker $k_B$; jobber du med mol og gram, bruker $R$. Og går du fra det ene til det andre, ganger eller deler du med $N_A$. Derfor gir for eksempel $E/k_B T$ (energi per molekyl) og $E_a/RT$ (energi per mol) samme tall så lenge du er konsekvent: teller og nevner må høre til samme nivå.
:::

### Temperatur er ikke det samme som energi

Siden temperatur og energi henger så tett sammen, er det lett å gå i fellen og tro at de er to ord for samme sak. Det er de ikke.

Det beste beviset på dette finner vi når et stoff skifter fase – for eksempel når en isbit smelter:

* Når isen når smeltepunktet ($0^\circ\text{C}$), fortsetter den å ta opp energi fra omgivelsene for å smelte.
* Til tross for at vi kontinuerlig tilfører energi, **stiger ikke temperaturen** en eneste grad før all isen har blitt til flytende vann.

Hvor blir det av energien? Den går ikke til å øke farten på molekylene (som ville økt temperaturen), men brukes i stedet til å bryte opp de låste bindingene i krystallstrukturen. Energien endrer altså hvordan vannmolekylene er *organisert*.

---

### Termodynamikkens definisjon av temperatur
Temperatur handler ikke bare om mikroskopisk fart, men om samspillet mellom energi og tilgjengelige mikrotilstander.

Entropi gir oss et mål på antallet mikrotilstander systemet kan fordele energien sin på. Den fundamentale, termodynamiske definisjonen av temperatur forteller oss hvor følsomt systemet er for å ta imot ny energi:

$$\frac{1}{T} = \left(\frac{\partial S}{\partial E}\right)_{V,N}$$

Denne ligningen gir en dypere forståelse av hvorfor temperatur og energi oppfører seg ulikt:
- **Ved lav temperatur** har systemet få tilgjengelige mikrotilstander. Tilfører du litt energi her, åpnes det opp et voldsomt antall nye måter å organisere energien på (stor endring i $S$)
- **Ved høy temperatur** har systemet allerede så mange tilgjengelige mikrotilstander at en skvett ekstra energi utgjør en mye mindre relativ forskjell. 

For en ideell gass har vi ingen bindinger som skal brytes, og alle mikrotilstandene handler om ren bevegelse. Derfor kollapser denne dype statistiske definisjonen ned til det enkle bildet vi bruker her: **Temperatur er et direkte mål på molekylenes gjennomsnittlige kinetiske energi.**

### Temperaturen endrer Maxwell-Boltzmann-fordelingen

Siden temperaturen bestemmer gjennomsnittsenergien, styrer den også formen på Maxwell-Boltzmann-fordelingen.

Når temperaturen øker, skjer det to ting med fordelingskurven:

1. **Den forskyves:** Hele kurven flytter seg mot høyre, mot høyere hastigheter og energier.
2. **Den flates ut:** Kurven blir bredere. Det betyr at spredningen i partiklenes fart blir større.

Dette har enorme praktiske konsekvenser, spesielt i kjemien. For at en kjemisk reaksjon skal skje, må molekylene kollidere med en viss minimumsenergi – kjent som **aktiveringsenergien $E_a$**. Når temperaturen øker og fordelingen blir bredere, vil en dramatisk mye større andel av molekylene krysse denne energigrensen. Det er hovedgrunnen til at selv en liten temperaturøkning kan få kjemiske reaksjoner til å gå vanvittig mye raskere.

## Hvor kommer trykk fra?

[Tidligere](https://andreasdh.github.io/Bifrost/docs/gasses.html#hva-er-egentlig-temperatur) fant vi ut hva temperatur faktisk er på molekylnivå. Nå skal vi gjøre noe som er selve kjernen i kinetisk gassteori: Vi skal utlede en makroskopisk, målbar størrelse utelukkende fra molekylær mekanikk.

Dette er verdt å stoppe opp ved et øyeblikk. Trykk er noe håndgripelig som du kan lese av på et manometer, og det er en størrelse ingeniører og forskere har brukt i århundrer. Vi skal nå vise at dette trykket følger direkte – uten noen nye antagelser – fra at gassen består av partikler som spretter rundt. Ingen empiri, ingen gjetting; bare partikler og Newtons lover.

### Hvorfor har en gass trykk?

Når du åpner en brusflaske, hører du det med én gang: et karakteristisk sus når gassen slipper ut. Før du åpnet flasken, presset gassen inni mot glasset, mot væskeoverflaten og mot korken – jevnt, i alle retninger, og helt uten pause. Det samme skjer inni et bildekk, i en ballong, og i lufta rundt deg akkurat nå.

![brus](bilder/brus.svg)

Men hva er egentlig dette presset? Hva er det som faktisk dytter?

:::{admonition} Definisjon

### Svaret ligger på molekylnivå

For å forstå hvor denne kraften kommer fra, må vi zoome helt inn på partiklene:

:::{admonition} Kjerneidé
:class: important
Trykk oppstår fordi et enormt antall molekyler kolliderer med veggene i beholderen.
:::

Ett enkelt molekyl som treffer en vegg, gir fra seg et ekstremt lite dytt – så lite at det er umulig å måle for oss. Men en gass inneholder astronomiske mengder molekyler, og hvert av dem kolliderer med veggen utallige ganger per sekund. Når vi summerer alle disse ørsmå, tilfeldige støtene, opplever vi det makroskopisk som en jevn, konstant kraft mot veggen. Det er denne kollektive kraften vi kaller trykk.


![vegg](bilder/vegg.svg)


Dette er et kroneksempel på hvordan en makroskopisk egenskap oppstår fra kaotiske, mikroskopiske hendelser. Trykket "finnes" ikke i ett enkelt molekyl, like lite som temperatur gjør det. Det er en kollektiv, statistisk egenskap som krever et fellesskap av partikler for å eksistere.

### Litt om bevegelsesmengde
For å utlede gasslovene fra bunnen av trenger vi bare én enkelt størrelse fra mekanikken.

**Bevegelsesmengde** (momentum) er definert som $p = mv$ (masse ganger hastighet). Newtons andre lov sier egentlig at en kraft $F$ er endring i bevegelsesmengde per tid:

$$F = \frac{\Delta p}{\Delta t}$$

Dette er hele verktøykassen vår. Skal vi finne kraften gassen utøver på en vegg, trenger vi bare å finne ut to ting: **hvor mye** bevegelsesmengde som overføres i hvert støt, og **hvor ofte** støtene skjer.

---

### Utledningen, trinn for trinn

Vi ser på en rektangulær boks med lengde $a$ langs $x$-aksen, der veggen vi studerer (vegg 1) har areal $A$. Volumet til boksen er da $V = A \cdot a$. Boksen inneholder $N$ molekyler, hvert med masse $m$.

#### Trinn 1: Ett støt

Et molekyl treffer veggen med hastighetskomponenten $v_x$. Vi antar at støtet er elastisk. Det betyr at molekylet spretter tilbake med nøyaktig samme fart, men i motsatt retning: $v_x \rightarrow -v_x$. Bevegelsen parallelt med veggen ($v_y, v_z$) blir ikke påvirket – molekylet reflekteres akkurat som en lysstråle fra et speil.

Molekylets bevegelsesmengde langs $x$-aksen endrer seg dermed fra $+m v_x$ til $-m v_x$. Endringen for molekylet er $-2 m v_x$. Etter Newtons tredje lov (kraft og motkraft) får veggen en like stor dytt i motsatt retning:

$$\vert{}\Delta p\vert{} = 2 m v_x$$

#### Trinn 2: Hvor ofte skjer støtene?

Ett enkelt støt sier ingenting om en kontinuerlig kraft; vi må vite hvor hyppig støtene kommer. Etter å ha truffet veggen, farer molekylet over til den motsatte veggen, spretter der, og returnerer til vår vegg. Det har da reist strekningen $2a$ langs $x$-aksen. Tiden $\Delta t$ mellom to støt på samme vegg er derfor:

$$\Delta t = \frac{2a}{v_x}$$

Raske molekyler slår både hardere (Trinn 1) og oftere (Trinn 2). Fordi farten bidrar i begge leddene, vil den om et øyeblikk dukke opp i andre potens ($v^2$).

#### Trinn 3: Kraften fra ett molekyl

Nå kombinerer vi de to første trinnene. Den tidsmidlede kraften $F_1$ fra ett enkelt molekyl på veggen blir:

$$F_1 = \frac{\vert{}\Delta p\vert{}}{\Delta t} = \frac{2 m v_x}{2a / v_x} = \frac{m v_x^2}{a}$$

Her dukket $v_x^2$ opp, akkurat som forventet.

#### Trinn 4: Summen over alle $N$ molekyler

Siden molekylene i en gass har ulike hastigheter, kan vi ikke bare gange med $N$. Vi må bruke gjennomsnittet av den kvadratiske hastigheten, $\langle v_x^2 \rangle$, for alle molekylene:

$$F = \frac{N m \langle v_x^2 \rangle}{a}$$

*Det er her Maxwell-Boltzmann-fordelingen ligger og ulmer i bakgrunnen: Det er den som sikrer at dette gjennomsnittet er en stabil og veldefinert størrelse.*

#### Trinn 5: Fra $v_x$ til den totale farten $v$

Vi vil helst uttrykke kraften ved den faktiske farten $v$, ikke bare $x$-komponenten. Fartvektoren er relatert til komponentene sine ved:

$$\langle v^2 \rangle = \langle v_x^2 \rangle + \langle v_y^2 \rangle + \langle v_z^2 \rangle$$

Her bruker vi et rent symmetriargument: Siden ingen retning i rommet er foretrukket i en tilfeldig gass, må bevegelsen i de tre retningene i snitt være helt like. Dermed er $\langle v_x^2 \rangle = \langle v_y^2 \rangle = \langle v_z^2 \rangle$, som gir:

$$\langle v_x^2 \rangle = \frac{1}{3}\langle v^2 \rangle$$

#### Trinn 6: Trykket oppstår

Trykk $P$ er kraft per areal ($P = F/A$). Siden volumet til boksen er $V = A \cdot a$, får vi:

$$P = \frac{F}{A} = \frac{N m \langle v_x^2 \rangle}{A \cdot a} = \frac{N m \langle v_x^2 \rangle}{V}$$

Vi setter inn $\langle v_x^2 \rangle = \frac{1}{3}\langle v^2 \rangle$ og flytter $V$ over på venstre side:

$$PV = \frac{1}{3} N m \langle v^2 \rangle$$

Dette er et rent mekanisk resultat. Legg merke til at vi ikke har nevnt ordet *temperatur* ennå.

Vi kan skrive dette om for å gjøre fysikken tydeligere. Siden den gjennomsnittlige kinetiske energien per molekyl er $\langle E_k \rangle = \frac{1}{2} m \langle v^2 \rangle$, kan vi erstatte $m\langle v^2\rangle$ med $2\langle E_k\rangle$:

$$PV = \frac{2}{3} N \langle E_k \rangle$$

Dette forteller oss noe fundamentalt: **Trykk er direkte proporsjonalt med molekylenes gjennomsnittlige kinetiske energi.**

#### Trinn 7: Temperaturen kobles på

Det er først nå vi henter inn broen mellom mikroskopisk bevegelse og makroskopisk temperatur:

$$\langle E_k \rangle = \frac{3}{2} k_B T$$

Dette er en fysisk definisjon som kommer utenfra mekanikken. Setter vi dette inn, faller alt på plass:

$$PV = \frac{2}{3} N \cdot \frac{3}{2} k_B T = N k_B T$$

Siden $N k_B = n R$ (der $n$ er antall mol og $R$ er den universelle gasskonstanten), står vi igjen med:

$$PV = n R T$$

Vi har utledet **den ideelle gassloven** utelukkende ved å anta partikler i tilfeldig bevegelse.

:::{admonition} Den ideelle gassloven
:class: important
$$PV = n R T$$
:::

#### Introduksjon til Daltons lov

Hva skjer når en gass består av flere ulike stoffer?

Til å har vi stort sett sett for oss en gass med bare én type molekyler. Men går vi tilbake til selve utledningen av ideelle gassloven, ser vi at vi aldri faktisk brukte at molekylene måtte være like. Det vi gjorde, var å summere bidraget fra hvert enkelt molekyl som traff veggen - og det fungerer like greit uansett om molekylene er identiske eller ikke. 

Det innebærer at vi like gjerne kan gruppere molekylene etter hvilken gass de tilhører.

Anta at beholderen inneholder en blanding av nitrogen ($\mathrm{N_2}$) og oksygen ($\mathrm{O_2}$), omtrent som vanlig luft. Da kan trykket skrives som en sum av bidrag fra de to gassene:

$$
PV=\frac23\left(
N_{\mathrm{N_2}}\langle E_k\rangle_{\mathrm{N_2}}
+
N_{\mathrm{O_2}}\langle E_k\rangle_{\mathrm{O_2}}
\right).
$$

Fra forrige del vet vi at det er temperaturen som bestemmer gjennomsnittlig kinetisk energi. Ved samme temperatur har derfor nitrogen- og oksygenmolekylene lik gjennomsnittlig kinetisk energi, til tross for at massene er ulike.

Dette gjør at hver gass kan behandles som om den var alene i beholderen. For hver komponent $i$ får vi da

$$
P_i=\frac{n_iRT}{V},
$$

hvor $P_i$ kalles partialtrykket.

Partialtrykket kan tolkes som den delen av totaltrykket som én bestemt gass står for: dersom alle de andre gassene ble fjernet, mens temperatur og volum holdes uendret, ville denne gassen alene gitt nettopp dette trykket.

Fordi hver gass bidrar uavhengig av de andre, blir totaltrykket rett og slett summen av alle bidragene:

$$
P_{\text{tot}}=P_1+P_2+\cdots=\sum_iP_i.
$$

Dette er **Daltons lov om partialtrykk.**

Det er ofte praktisk å uttrykke partialtrykket ved hjelp av molfraksjonen

$$
x_i=\frac{n_i}{n_{\text{tot}}},
$$

som angir hvor stor andel av molekylene som tilhører komponent $i$. Da har vi at

$$
P_i=x_iP_{\text{tot}}.
$$

Med andre ord: utgjør nitrogen 80 % av molekylene i en beholder, står nitrogen også for omtrent 80 % av totaltrykket.

:::{card}
**Eksempel**
^^^

En beholder inneholder

* 1 mol $\mathrm{H_2}$
* 1 mol $\mathrm{CO_2}$

ved samme temperatur.

Siden begge gassene er representert med like mange mol, utgjør de også like store andeler av molekylene. Partialtrykkene blir derfor like:

$$
P_{\mathrm{H_2}}=P_{\mathrm{CO_2}}=\frac12P_{\text{tot}}.
$$

Dette kan virke overraskende, gitt at et $\mathrm{CO_2}$-molekyl veier omtrent 22 ganger så mye som et $\mathrm{H_2}$-molekyl.


![dalton](bilder/Dalton.svg)

Nøkkelen er at de tunge molekylene beveger seg saktere: de treffer veggen sjeldnere, men med større kraft per støt. De lette molekylene treffer oftere, men svakere. Ved samme temperatur veier disse to effektene nøyaktig opp for hverandre.

For en ideell gass er det derfor antall molekyler, ikke massen deres, som avgjør hvor mye de bidrar til trykket.
:::

:::{admonition} En kort oppsummering
:class: important
Den ideelle gassloven falt ikke ned fra himmelen. Historisk ble den oppdaget eksperimentelt av forskere som Boyle, Charles og Clapeyron gjennom nitidige målinger på gasser – lenge før man visste at molekyler fantes.

Det vi nettopp har gjort, er å bevise at loven er en direkte konsekvens av Newtons lover brukt på et enormt antall partikler. Symbolene i $PV = nRT$ har nå fått en konkret, mikroskopisk betydning:

| Symbol | Mikroskopisk betydning |
| --- | --- |
| **$P$ (Trykk)** | Hvor hardt og hvor ofte molekylene kolliderer med veggen. |
| **$V$ (Volum)** | Hvor stort rom partiklene har å bevege seg i (og dermed hvor lang tid det går mellom hver kollisjon). |
| **$N$ (Antall)** | Hvor mange molekyler som er med på å dytte. |
| **$T$ (Temperatur)** | Et direkte mål på partiklenes gjennomsnittlige kinetiske energi (hvor fort de beveger seg). |

:::

### Tre viktige innsikter å ta med videre

Før vi dykker dypere inn i termodynamikken, er det tre konsepter fra denne utledningen som er viktigere enn selve matematikken:

1. **Trykk er en statistisk størrelse.** Ett enkelt molekyl har ikke et "trykk", akkurat som det ikke har en "temperatur". Trykk er en makroskopisk egenskap som først oppstår og stabiliserer seg når vi midler over enorme mengder partikler ($\langle v_x^2 \rangle$).
2. **Trykk er en intensiv størrelse.** Hvis du slår sammen to identiske beholdere med gass, dobler du både volumet ($V$) og antall partikler ($N$). Men trykket ($P$) forblir det samme. Trykk måler den lokale energitettheten (kraft per areal), ikke summen av all energi.
3. **Mønsteret i tankegangen er kjernen i statistisk mekanikk.** Vi kommer til å bruke denne strategien igjen og igjen:

---

:::{admonition} Et lite blikk på reelle gasser
:class: note

Utledningen over er elegant, men den hviler på noen strenge forenklinger. Vi har antatt at gassen er **ideell**. Det betyr at:

* Molekylene er uendelig små punkter uten eget volum.
* Det ikke finnes noen krefter mellom molekylene (de merker ikke hverandre før de kolliderer).

I virkeligheten er ingen gasser helt ideelle. Molekyler tar faktisk opp fysisk plass, og de tiltrekker hverandre svakt når de er nær hverandre. Vi kan se for oss avvikene slik:

* **Molekylenes eget volum:** Tenk på molekylene som små kuler. Siden de tar opp plass, har de andre molekylene litt mindre ledig rom å bevege seg på enn det totale volumet $V$ skulle tilsi. Dette fører til at de kolliderer med veggene *oftere*, noe som trekker trykket **opp**.
* **Intermolekylære tiltrekningskrefter:** Når et molekyl er på vei til å dundre inn i veggen, vil nabomolekylene bak trekke litt i det med svake krefter. Dette "holder igjen" molekylet litt, slik at støtet mot veggen blir svakere. Dette trekker trykket **ned**.

Ved romtemperatur og vanlig atmosfærisk trykk er disse avvikene så små at den ideelle gassloven fungerer nesten perfekt. Men presser vi gassen sammen til høyt trykk (slik at de tvinges tett sammen) eller kjøler den ned (slik at de beveger seg sakte nok til at tiltrekningen får tid til å virke), begynner de reelle kreftene å dominere. Dette er starten på historien om intermolekylære krefter og faseoverganger, som vi skal se nærmere på senere.
:::

## Kinetisk gassteori som samlende rammeverk

Gjennom de forrige delene har vi bygget opp tre viktige biter av samme maskineri:

1. Molekylene har en **fartsfordeling**.
2. **Temperaturen** er et mål på molekylenes gjennomsnittlige kinetiske energi .
3. **Trykket** er summen av alle partikkelstøt mot veggene.

I denne siste delen skal vi se at dette samlede rammeverket – mange partikler i tilfeldig bevegelse – ikke bare forklarer rent fysiske gasslover, men også avdekker den mikroskopiske mekanismen bak **kjemisk likevekt**. Dette er et innblikk inn i videre pensum, og vi kommer tilbake til likevekter i mer detalj senere. Det som er viktig å huske, og som dette avsnittet illustrerer, er at *alt henger sammen*. Naturen bryr seg ikke om kapittelinndelingen i en lærebok, alle lover gjelder samtidig og samspiller med hverandre. 

Nå skal vi se hva som skjer når disse tilfeldige molekylbevegelsene driver en kjemisk reaksjon som kan gå begge veier (en reversibel reaksjon):

$$A \rightleftharpoons B$$

Det vakre er at vi ikke trenger noe nytt, magisk prinsipp og alt følger direkte fra bildet vi allerede har bygget opp.

---

### Dynamisk likevekt

Forestilling deg en gassblanding bestående av reaktant $A$ og produkt $B$. Molekylene er i konstant, kaotisk bevegelse.

* Noen $A$-molekyler kolliderer med nok energi til å overvinne aktiveringsbarrieren og bli til $B$.
* Samtidig vil noen $B$-molekyler kollidere hardt nok til å gå motsatt vei og bli til $A$ igjen.

I starten har vi kanskje bare $A$. Etter hvert som det dannes flere $B$-molekyler, vil reaksjonen bakover ($B \rightarrow A$) gå raskere og raskere fordi det blir stadig flere $B$-partikler som kan kollidere. Til slutt når systemet et punkt hvor reaksjonen forover og bakover skjer med **nøyaktig samme hastighet**.

:::{admonition} Kjerneidé
:class: important
Dynamisk likevekt kjennetegnes av **makroskopisk stillstand, og mikroskopisk febrilsk aktivitet**.
:::

Konsentrasjonene av $A$ og $B$ endrer seg ikke lenger, men molekylene fortsetter å omdannes i begge retninger. Likevekt betyr **ikke** at reaksjonen har stoppet – den går bare like fort begge veier. Det er en direkte konsekvens av den tilfeldige fartsfordelingen vi så [lenger opp](https://andreasdh.github.io/Bifrost/docs/gasses.html#hvordan-er-molekylenes-hastighet-fordelt).


![damp](bilder/damp.svg)

:::{admonition}
:class:important

I en lukket flaske med væske vil noen molekyler ved overflaten hele tiden ha nok energi til å rive seg løs og gå over i gassfase – de fordamper. Samtidig treffer damp­molekyler overflaten og fanges tilbake i væsken – de kondenserer. I begynnelsen fordamper det flest, men etter hvert som mengden damp øker, øker også kondensasjonen. Til slutt skjer de to prosessene like fort.

Da har systemet nådd dynamisk likevekt. Væskenivået og damptrykket står helt stille, men på molekylnivå er det full aktivitet: for hvert molekyl som forlater væsken, vender et annet tilbake. Likevekt betyr altså ikke at noe har stoppet – det betyr bare at fordampning og kondensasjon balanserer hverandre nøyaktig. Fjerner vi lokket, brytes balansen: dampen slipper ut, kondensasjonen kan aldri ta igjen fordampningen, og væsken tørker inn. Dynamisk likevekt krever et lukket system.
:::


### Hvorfor blir forholdet som det blir?

Hvorfor ender vi ikke alltid opp med nøyaktig $50\,\%$ $A$ og $50\,\%$ $B$?

Svaret ligger i energibarrierene. Sannsynligheten for at et molekyl har nok energi til å krysse en aktiveringsbarriere $E_a$, er styrt av Maxwell–Boltzmann-fordelingen og kan uttrykkes ved en Boltzmann-faktor:

$$\text{Sannsynlighet} \propto e^{-\frac{E_a}{RT}}$$

Hvis reaksjonen $A \rightarrow B$ krever *lavere* aktiveringsenergi enn $B \rightarrow A$, vil en større andel av $A$-molekylene ha nok fart til å reagere ved en gitt temperatur. For at reaksjonsratene skal bli like store, må systemet derfor ende opp med **flere $B$-molekyler enn $A$-molekyler**.

Setter vi ratene lik hverandre og deler dem på hverandre, ser vi at aktiveringsbarrierene kanselleres, og likevektsforholdet bestemmes av selve energiforskjellen $\Delta E = E_B - E_A$:

$$\frac{[B]}{[A]} \propto e^{-\frac{\Delta E}{RT}}$$

---

:::{admonition} Viktig presisering og blikk fremover
:class: error

1. **Barrieren bestemmer farten, energiforskjellen bestemmer posisjonen:**
Aktiveringsbarrieren ($E_a$) avgjør bare *hvor fort* likevekten innstiller seg (kinetikk), mens energiforskjellen ($\Delta E$) avgjør *hvor* likevekten havner (likevektsforholdet $[B]/[A]$).
2. **Fra $\Delta E$ til Gibbs fri energi ($\Delta G$):**
I mer realistiske systemer er det ikke bare energibarrieren $\Delta E$, men endringen i Gibbs fri energi $\Delta G$ – som også tar hensyn til entropi og ordning – som bestemmer det eksakte likevektsforholdet: $K_{\text{eq}} = e^{-\frac{\Delta G}{RT}}$. Dette skal vi se grundig på i senere kapitler!
:::

---

### Hvordan temperatur påvirker likevekten

Nå kan vi bruke dette bildet til å forstå **Le Chateliers prinsipp** på mikroskopisk nivå.

Fordi temperaturen $T$ står i eksponenten $e^{-\frac{\Delta E}{RT}}$, vil en endring i temperatur endre selve forholdet mellom $A$ og $B$.

Når du øker temperaturen, skyves fartsfordelingen mot høyere energier. Flere molekyler får nok energi til å komme over barrieren i *begge* retninger, men retningen som krever mest energi (den endoterme retningen) påvirkes sterkest:

* **Eksoterm reaksjon ($\Delta E < 0$):** $B$ ligger lavere i energi enn $A$. Oppvarming skyver likevekten **tilbake mot $A$**, fordi den energikrevende retningen ($B \rightarrow A$) får mest hjelp av ekstra varme.
* **Endoterm reaksjon ($\Delta E > 0$):** $B$ ligger høyere i energi enn $A$. Oppvarming skyver likevekten **framover mot $B$**.

Dette viser at Le Chateliers prinsipp ikke er et "mystisk svar" fra systemet, men ren statistikk: Fartsfordelingen flytter seg, og de to reaksjonsratene endrer seg ulikt.

---

### Hvordan trykk påvirker likevekten

Trykk virker på en helt annen måte enn temperatur.

Skal trykk forskyve en likevekt, må det totale antallet gassmolekyler endre seg i reaksjonen, for eksempel:

$$2A \rightleftharpoons B$$

Kobler vi inn den ideelle gassloven fra Del 4 ($PV = nRT$), ser vi at trykket forteller oss hvor mange partikler vi har per volum (partikkeltetthet):

* **Øker vi trykket (mindre volum):** Partiklene presses tettere sammen. Sjansene for at *to* $A$-molekyler kolliderer samtidig skyter i været, og reaksjonen forskyver seg mot **færre molekyler ($B$)**.
* **Minker vi trykket (større volum):** Reaksjonen forskyver seg mot **flere molekyler ($A$)**.

### Energien tar slutt som forklaring her

Legg merke til at denne trykkeffekten **ikke ligger i energi-eksponenten $e^{-\frac{\Delta E}{RT}}$ i det hele tatt!**

Temperatur virker gjennom *partiklenes energi*. Trykk virker gjennom *antall molekyler per volum*, altså hvor mange posisjoner molekylene har å fordele seg på. Dette er en **telle-effekt**, ikke en energieffekt.

For å forklare hvorfor et komprimert system foretrekker færre molekyler, må vi begynne å telle hvor mange måter molekylene kan organisere seg på i rommet – hvor mange mikrotilstander systemet har.

Dette er den direkte broen til [forrige modul](https://andreasdh.github.io/Bifrost/docs/entropi.html), der vi ga denne telle-logikken et navn og en ligning: **Entropi** ($S = k_B \ln \Omega$).
