## 2 - Hvorfor beveger molekylene seg?

I forrige del så vi at en gass består av et enormt antall molekyler som beveger seg tilfeldig. Selv
om vi ikke kan følge hvert enkelt molekyl, kan vi beskrive hvordan hele gassen oppfører seg ved
hjelp av statistikk. Det sentrale spørsmålet blir derfor: **Hvordan er molekylenes hastighet
fordelt?**

Et naturlig delspørsmål er om alle molekylene har samme fart. Det korte svaret er nei.

Hvis vi kunne «fryse» bevegelsen i en gass et øyeblikk og måle farten til hvert eneste molekyl,
ville vi oppdaget at de beveger seg med mange forskjellige hastigheter. Noen molekyler beveger seg
sakte, andre svært raskt, mens de fleste har en fart et sted imellom. Grunnen er at molekylene hele
tiden kolliderer med hverandre. Ved hver kollisjon overføres energi: et raskt molekyl kan miste noe
av energien sin og bli langsommere, mens et annet mottar energien og får høyere fart. Både fart og
retning endrer seg altså kontinuerlig for det enkelte molekylet.

Likevel skjer det noe bemerkelsesverdig. Selv om de enkelte molekylene stadig bytter fart, forblir
den samlede fordelingen av hastigheter nesten uendret. Antallet langsomme, middels raske og
raske molekyler holder seg omtrent konstant så lenge temperaturen er den samme. Dette er et godt
eksempel på hvordan et system kan være i konstant endring på mikroskopisk nivå, samtidig som det
framstår stabilt på makroskopisk nivå. Det er denne stabile fordelingen vi ønsker å beskrive.

### Tilfeldighet gir mønster

Dette er ikke et fenomen som bare gjelder gasser. Mange tilfeldige prosesser i naturen viser seg å
følge regelmessige fordelinger.

Et enkelt eksempel er å kaste en mynt 60 ganger. Gjentar vi forsøket mange ganger, vil vi ikke
alltid få nøyaktig 30 kron. Noen ganger får vi 28, andre ganger 33. Resultater langt unna
gjennomsnittet, som 15 eller 45, er derimot svært sjeldne. Tegner vi hvor ofte hvert resultat
forekommer, får vi en karakteristisk klokkeformet kurve – en normalfordeling (gaussfordeling).
Denne typen fordeling dukker opp nettopp når mange uavhengige tilfeldige hendelser virker sammen.

(FIGURE myntkast og gauss)

James Clerk Maxwell innså at de utallige kollisjonene mellom molekylene i en gass er et slikt
system. Ved første øyekast kan det virke overraskende at noe så kaotisk som millioner av kolliderende
molekyler kan beskrives med en enkel matematisk funksjon – men det er nettopp et generelt trekk ved
store systemer. I stedet for å følge hvert enkelt molekyl, beskrev Maxwell sannsynligheten for
at et molekyl har en bestemt fart. Resultatet er at de fleste molekylene får en middels fart, mens
bare et lite antall beveger seg svært sakte eller svært raskt.

(SIMULERINGEN MED BOLTZMANN HER ALTSå)



Fordelingen er ikke symmetrisk. Det finnes ingen negative farter, så kurven får en lang hale mot
høyere hastigheter. Denne formen er karakteristisk for alle idealgasser.

### Hva påvirker fordelingen?

To faktorer har særlig stor betydning for hvordan Maxwell–Boltzmann-fordelingen ser ut:
temperatur og molekylmasse.

Når temperaturen øker, får molekylene mer kinetisk energi. Hele fordelingen flytter seg mot høyere
hastigheter, samtidig som den blir bredere. Flere molekyler får høy fart, og en større andel kan
delta i prosesser som krever høy energi, for eksempel kjemiske reaksjoner.

Molekylmassen spiller også en viktig rolle. Ved samme temperatur beveger lette molekyler seg i
gjennomsnitt raskere enn tunge. Derfor har helium en bredere fordeling med høyere typiske
hastigheter enn nitrogen eller oksygen. Disse sammenhengene blir viktige senere, når vi skal forstå
diffusjon, gasskromatografi og reaksjonshastigheter.

kan skrives som:

I neste del skal vi bruke fordelingen til å gi temperatur en fysisk tolkning. Vi skal se at

### Temperatur er ikke det samme som energi

Det er lett å tro at temperatur og energi er det samme, siden begge øker når vi varmer opp et stoff.
Likevel er de forskjellige størrelser.


Et godt eksempel er en isbit som smelter. Når isen når smeltepunktet, kan vi fortsette å tilføre

et system inneholder – den forteller hvordan energien er fordelt mellom molekylene.

kjemiske reaksjoner. Dette er en av de viktigste grunnene til at reaksjonshastigheter øker når
temperaturen stiger.

## 4 - Hvor kommer trykk fra?

I Del 3 fant vi ut hva temperatur er på molekylnivå. Nå skal vi gjøre noe som er selve kjernen i kinetisk gassteori: utlede en makroskopisk, målbar størrelse fra ren molekylær mekanikk.
Dette er verdt å stoppe ved et øyeblikk. Trykk er noe du kan lese av på et manometer. Det er en størrelse ingeniører har brukt i århundrer. Og vi skal vise at den følger, uten noen ny antagelse utover det vi allerede har – fra at gassen er mange partikler som spretter rundt. Ingen empiri, bare partikler og Newtons lover.

Hvorfor har en gass trykk?

![brus](bilder/brus.svg)

Åpne en brusflaske, og du hører det med én gang: et sus når gassen slipper ut. Før du åpnet den, presset gassen inni mot glasset, mot væskeoverflaten og mot korken – jevnt, i alle retninger, uten pause. Det samme skjer i et bildekk, i en ballong, og i lufta rundt deg akkurat nå.
Men hva er dette presset? Hva er det egentlig som dytter?

Historisk var dette langt fra opplagt. Galileo oppdaget i 1594 at en sugepumpe ikke klarte å løfte vann mer enn omtrent 10 meter, uansett hvor godt den var bygget. Først femti år senere gjorde Evangelista Torricelli det begrepsmessige spranget: det er ikke pumpa som suger vannet opp – det er atmosfæren som dytter det opp, og vannet stopper når vekten av vannsøyla balanserer atmosfærens dytt. Torricelli målte dette dyttet med en kvikksølvsøyle, og oppfant dermed barometeret.
Definisjon. 
Trykk er kraft per areal: $P = F/A$. Enheten pascal er $1\ \text{Pa} = 1\ \text{N/m}^2$.

Torricelli visste at atmosfæren dyttet, men ikke hvorfor. Svaret er molekylært:
Kjerneidé: Trykk oppstår fordi et enormt antall molekyler kolliderer med veggene i beholderen.
Et enkelt molekyl som treffer en vegg, gir bare et ekstremt lite dytt – så lite at det er umulig å måle. Men en gass inneholder et enormt antall molekyler, og hvert av dem kolliderer med veggen utallige ganger per sekund. Summen av alle disse ørsmå, tilfeldig fordelte støtene blir en jevn, konstant kraft. Det er den vi kjenner som trykk.
Dette er et godt eksempel på hvordan en makroskopisk egenskap oppstår fra mange mikroskopiske hendelser – akkurat det mønsteret vi introduserte i Del 1. Trykket «finnes» ikke i ett enkelt molekyl, like lite som temperatur gjorde det. Det er en kollektiv, statistisk egenskap.

![vegg](bilder/vegg.svg)

### Litt om bevegelsesmengde
Vi trenger én størrelse fra mekanikken, og bare kort.
Bevegelsesmengde (engelsk: momentum) er $p = mv$ – masse ganger hastighet. Newtons andre lov sier egentlig at kraft er endring i bevegelsesmengde per tid:

$$ F = \frac{\Delta p}{\Delta t} $$
Det er hele verktøyet. Skal vi finne kraften på veggen, trenger vi bare to ting: hvor mye bevegelsesmengde som overføres i hvert støt, og hvor ofte støtene skjer.
 
Utledningen, trinn for trinn
Vi ser på en rektangulær boks med lengde $a$ langs $x$-aksen, der veggen vi studerer (kall den vegg 1) har areal $A$. Volumet er da $V = A \cdot a$. Boksen inneholder $N$ molekyler, hvert med masse $m$.

**Trinn 1 – Ett støt**

Et molekyl treffer vegg 1 med hastighetskomponenten $v_x$. Støtet er elastisk, så molekylet spretter tilbake med samme fart, men $x$-komponenten skifter fortegn: $v_x \rightarrow -v_x$. Bevegelsen parallelt med veggen ($v_y$, $v_z$) er uendret – molekylet reflekteres som en lysstråle fra et speil.
Molekylets bevegelsesmengde langs $x$ endrer seg dermed fra $+m v_x$ til $-m v_x$, altså en endring på $-2 m v_x$. Etter Newtons tredje lov får veggen en like stor dytt den andre veien:

$$|\Delta p| = 2 m v_x$$

**Trinn 2 – Hvor ofte skjer støtene?**

Ett støt alene sier ingenting om kraft; vi må vite hvor ofte de kommer. Etter å ha truffet vegg 1 farer molekylet over til motsatt vegg, spretter der, og kommer tilbake. Det har da reist strekningen $2a$ langs $x$-aksen. Tiden mellom to støt på samme vegg er derfor:

$ \Delta t = \frac{2a}{v_x}$$
Legg merke til noe elegant her: raske molekyler treffer både hardere (trinn 1) og oftere (trinn 2). Begge effektene trekker i samme retning, og det er derfor farten vil dukke opp i andre potens om et øyeblikk.

**Trinn 3 – Kraften fra ett molekyl**

Nå kombinerer vi de to. Den tidsmidlede kraften fra ett molekyl på vegg 1:

$$(F_1 = \frac{|\Delta p|}{\Delta t} = \frac{2 m v_x}{2a / v_x} = \frac{m v_x^2}{a})$$
Der er $v_x^2$ – akkurat som varslet.

**Trinn 4 – Summér over alle $N$ molekyler**

Molekylene har ikke samme fart (det vet vi godt fra Del 2!), så vi kan ikke bare gange med $N$. Vi må bruke gjennomsnittet $\langle v_x^2 \rangle$ over alle molekylene:

$$(F = \frac{N m \langle v_x^2 \rangle}{a})$$
Her er det Maxwell–Boltzmann-fordelingen kommer inn i praksis: den er det som gjør at $\langle v_x^2 \rangle$ i det hele tatt er en veldefinert, stabil størrelse.

**Trinn 5 – Fra $v_x$ til $v$**

Vi vil helst uttrykke dette med farten $v$, ikke bare $x$-komponenten. Farten henger sammen med komponentene slik:
$$(\langle v^2 \rangle = \langle v_x^2 \rangle + \langle v_y^2 \rangle + \langle v_z^2 \rangle)$$
Og her bruker vi det samme symmetriargumentet som i Del 2: ingen retning er spesiell. Det er ingenting som skiller bevegelse langs $x$ fra $y$ eller $z$, så de tre må være like:
$$(\langle v_x^2 \rangle = \langle v_y^2 \rangle = \langle v_z^2 \rangle \qquad \Longrightarrow \qquad \langle v_x^2 \rangle = \frac{1}{3}\langle v^2 \rangle)$$

**Trinn 6 – Trykket**

Nå deler vi kraften på arealet, og bruker at $V = A \cdot a$:
$$(P = \frac{F}{A} = \frac{N m \langle v_x^2 \rangle}{A \cdot a} = \frac{N m \langle v_x^2 \rangle}{V})$$
Setter vi inn $(\langle v_x^2 \rangle = \tfrac{1}{3}\langle v^2 \rangle)$ og ganger opp med $V$:
$$(\boxed{;PV = \tfrac{1}{3} N m \langle v^2 \rangle;})$$
Stopp opp her. Dette er et fullstendig resultat, og det er verdt å legge merke til hva som ikke står i det: det er ingen temperatur. Vi har ikke nevnt $T$ med ett ord. Alt vi har brukt, er Newtons lover og litt geometri. Dette er ren mekanikk.
Vi kan skrive det enda mer talende. Siden den gjennomsnittlige kinetiske energien per molekyl er $(\langle E_k \rangle = \tfrac{1}{2} m \langle v^2 \rangle)$, har vi $(m\langle v^2\rangle = 2\langle E_k\rangle)$:
$$(PV = \tfrac{2}{3} N \langle E_k \rangle)$$
Altså: trykk er direkte proporsjonalt med molekylenes gjennomsnittlige kinetiske energi. Det er alt mekanikken kan si oss. Ikke noe mer.

**Trinn 7 – Temperaturen kommer inn**

Og her, først nå, henter vi inn resultatet fra Del 3:
$$(\langle E_k \rangle = \tfrac{3}{2} k_B T)$$
Dette er det eneste stedet temperaturen kommer inn i hele utledningen, og den kommer utenfra – som en fysisk innsikt om hva temperatur er, ikke som noe mekanikken kunne gitt oss. Setter vi inn:
$$(PV = \tfrac{2}{3} N \cdot \tfrac{3}{2} k_B T = N k_B T)$$
Og siden $(N k_B = n N_A k_B = n R)$:
$$(\boxed{;PV = N k_B T = n R T;})$$

Idealgassloven.
 
Hva vi egentlig har vist
Legg merke til hvor denne loven kom fra. Den falt ikke ned fra himmelen, og den er ikke en empirisk regel man skal pugge.
Historisk var den riktignok empirisk: Boyle observerte sammenhengen mellom trykk og volum i 1662, Charles og Gay-Lussac studerte temperaturavhengigheten, og Clapeyron sydde det hele sammen til $pV = NkT$ i 1834 – alt sammen basert på målinger, uten noen anelse om molekyler. Poenget vårt er at vi nettopp har utledet den samme loven fra bunnen av, uten å måle noe som helst. Vi trengte bare partikler i tilfeldig bevegelse.
Hvert symbol i ligningen har nå en konkret molekylær betydning:

:::{admonition} 
:class: tip
Symbol	Molekylær betydning
$P$	Hvor hardt og hvor ofte molekylene dunker mot veggen
$V$	Hvor stort rom de har å bevege seg i (og dermed hvor lang tid mellom støtene)
$N$	Hvor mange som dunker
$T$	Hvor fort de dunker (gjennomsnittlig kinetisk energi)
:::

Idealgassloven er ikke en regel du skal pugge. Den følger av at gassen er mange partikler i tilfeldig bevegelse. Trykk = molekyler som dunker mot veggen. Temperatur = hvor fort de dunker. Alt henger sammen.
 

Hva er viktig å ta med videre
Før vi går løs på de makroskopiske utledningene i termodynamikken, er det tre innsikter som er viktigere enn algebraen over. Får du med deg disse, har du fått med deg Del 4:
1. Trykk er en statistisk størrelse, ikke en egenskap ved ett molekyl. Vi måtte bruke $(\langle v_x^2 \rangle)$ – et gjennomsnitt. Ett molekyl har ikke «trykk», like lite som ett molekyl har temperatur. Trykk er noe som oppstår først når du har svært mange partikler. Dette er samme poeng som i Del 3, og det kommer igjen og igjen.
2. Trykk er en intensiv størrelse. Slår du sammen to like beholdere med gass, dobler du volumet og antall molekyler, men ikke trykket. Trykk er et gjennomsnitt over partiklene, ikke en sum av dem – i motsetning til $V$, $N$ og energien, som er ekstensive. Dette skillet blir helt sentralt i termodynamikken.
3. Mønsteret i resonnementet er viktigere enn regnestykket. Vi gjorde noe her som vi kommer til å gjøre om og om igjen: beskriv én partikkel med enkel fysikk → ta gjennomsnittet over fordelingen → få en målbar, makroskopisk størrelse. Det er hele strategien i statistisk mekanikk. Klarer du å gjenfortelle utledningen som en historie – et molekyl treffer, spretter, kommer tilbake, mange sånne gir jevn kraft – så har du forstått det. Klarer du bare algebraen, har du ikke det.
Et blikk framover. I termodynamikken møter du en mer generell definisjon av trykk, der det ikke lenger er snakk om støt mot vegger i det hele tatt: $$(P = -\left(\frac{\partial U}{\partial V}\right){S,N})$$ Trykk blir da definert som systemets tendens til å endre volum. Det virker kanskje som en helt annen størrelse – men det er den ikke. Det finnes faktisk flere veier til samme lov: man kan også komme fram til $PV = nRT$ ved å telle mikrotilstander og bruke at $(p/T = (\partial S / \partial V){U,N})$, helt uten å snakke om kollisjoner. At tre helt ulike resonnementer lander på nøyaktig samme ligning, er et av de sterkeste argumentene for at bildet vårt av gasser stemmer. (Se Dill Eksempel 6.1 for gittermodell-veien og Cooksy kap. 2.5 for den statistiske. Vi kommer tilbake til dette når vi har entropi på plass.)

### Et lite blikk på reelle gasser

Utledningen viser også **hvor idealgassantagelsene brukes**:

* Molekylene antas å bevege seg fritt mellom kollisjonene, altså uten å vekselvirke med hverandre.
* Hele volumet (V) antas å være tilgjengelig, som om molekylene ikke har eget volum.
* Alle støt mot veggene antas å være fullstendig elastiske.

Ingen av disse antagelsene er helt oppfylt i en reell gass. Molekylene har både utstrekning og tiltrekker hverandre.

* **Tiltrekning** gjør at molekylene treffer veggen med litt lavere fart, slik at trykket blir lavere enn idealgassloven forutsier.
* **Eget volum** gjør det tilgjengelige volumet mindre, slik at molekylene treffer veggen oftere og trykket blir høyere.

Avviket fra idealgassoppførsel beskrives med **kompressibilitetsfaktoren**

$$(Z=\frac{PV}{nRT}.)$$

For en idealgass er (Z=1). Er (Z<1), dominerer tiltrekningskreftene, mens (Z>1) betyr at molekylenes eget volum har størst betydning.

Ved vanlige trykk og temperaturer er begge effektene små, og idealgassmodellen fungerer derfor svært godt. Først ved høyt trykk eller lav temperatur blir avvikene store, og gassen kan til slutt kondensere til væske.
:::
