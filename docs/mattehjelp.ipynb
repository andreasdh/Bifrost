LITEN TEKST OM MOTIVASJONEN RUNDT SANNSYNLIGHET I FYSIKALSK KJEMI.

### 1.1 Utfall, utfallsrom og hendelser

Vi definerer først utfall, utfallsrom og hendelser for statistiske eksperimenter. Et slikt eksperiment kan for eksempel være å trille en terning og se hvilket tall man får.

:::{admonition} Def. 1.1 - Utfall
:class: important

Et utfall $i$ er et mulig resultat i et eksperiment. 
:::


:::{admonition} Def. 1.2 - Utfallsrom $S$
:class: important

Utfallsrommet $S$ er mengden av alle mulige utfall i et eksperiment.
:::


:::{admonition} Def. 1.3 - Hendelse
:class: important

En hendelse $A$ er delmengde av utfallsrommet, altså ett eller flere utfall vi er interessert i.
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: note

I statistisk termodynamikk kaller vi et utfall for en **mikrotilstand**, altså én bestemt fordeling av energi/posisjon over alle partiklene i systemet. Utfallsrommet $S$ blir da mengden av *alle* mikrotilstander systemet kan befinne seg i, og en hendelse $A$ tilsvarer en **makrotilstand** (f.eks. en gitt total energi, temperatur eller trykk) - altså alle mikrotilstander som gir samme makroskopiske observasjon.
:::

_Eksempel 1.1_

_Du triller en likevektet, sekssidet terning._ 

_a) Hva er utfallsrommet $S$ til eksperimentet?_

_De mulige utfallene for eksperimentet er $1,2,3,4,5$ eller $6$, som betyr at utfallsrommet er $S=\{1,2,3,4,5,6\}$._



_b) Hva er et eksempel på en hendelse i dette eksperimentet?_

_Et eksempel på en hendelse i dette eksperimentet er å trille et partall. La oss betegne denne hendelsen ved $A$, og da har vi at $A=\{2,4,6\}$._

:::{admonition} Underveisoppgave 1.1
:class: tip

Du trekker ett kort fra en ordinær kortstokk med totalt 52 kort, og ser på kortets symbol (spar, hjerter, ruter eller kløver).

a) Hva er utfallsrommet $S$ til eksperimentet?

b) Hva er sannsynligheten for hvert av utfallene i utfallsrommet?

c) Gi eksempel på en mulig hendelse av dette eksperimentet.
::

:::{admonition} Løsning 1.1
:class: tip, dropdown

a) De mulige symbolene er spar, hjerter, ruter eller kløver, og dermed er utfallsrommet $S=\{spar, hjerter, ruter, kløver\}$.

b) Det er $13$ kort av hver symbolstype, slik at sannsynligheten for hvert av utfallene blir $\frac{13}{52}=\frac{1}{4}$.

c) Mulige hendelser for eksperimentet. Eksempler: trekke rødt kort ($\{ruter, hjerter\}$) eller trekke svart kort ($\{spar, kløver\}$). Merk at hendelser som å for eksempel trekke bildekort (konge, dame, knekt, ess) IKKE er hendelser i dette eksperimentet, nettopp fordi det ikke er en delmengde av utfallsrommet $S$.
:::

:::{admonition} Def. 1.4 - Komplement
:class: important

Komplementet til en hendelse $A$, $A'$, er mengden av alle utfall i $S$ som ikke er i $A$.
:::


:::{admonition} Def. 1.5 - Snitt
:class: important

Snittet av to hendelser $A$ og $B$, $A \cap B$, er hendelsen som består av alle utfall som både er i $A$ og i $B$.
:::


:::{admonition} Def. 1.6 - Union
:class: important

Unionen av to hendelser $A$ og $B$, $A \cup B$, er hendelsen som består av alle utfall som er i $A$ eller i $B$.
:::

![Komplement, snitt og union](bilder/venn_komplement_snitt_union.svg)

_Eksempel 1.2_

_Anta igjen at eksperimentet ditt er ett kast av en likevektet, sekssidet terning. Vi er interessert i å trille $2$ eller lavere, som kan beskrives ved hendelsen $A=\{1,2\}$. Vi er også interessert i å trille $5$ eller høyere, som kan beskrives ved hendelsen $B=\{5,6\}$._ 
 
_a) Hva er komplementet til $A$?_

_Komplementet til $A$ er alle utfallene som er i $S$ men ikke $A$, som tilsvarer å trille 3 eller høyere. Altså er $A'=\{3,4,5,6\}$._ 

_b) Hva er snittet av $A$ og $B$?_

_Snittet mellom $A$ og $B$ blir den tomme mengden $\{\}$, nettopp fordi det er ingen utfall som er i både $A$ og $B$._

_c) Hva er unionen av $A$ og $B$?_
 
_Unionen mellom $A$ og $B$ blir $A \cup B = \{1,2,5,6\}$, altså mengden alle utfallene i $A$ og $B$ samlet._


:::{admonition} Def. 1.5 - Disjunkte hendelser
:class: important

To hendelser $A$ og $B$ kalles disjunkte dersom de ikke har noen felles utfall. Matematisk, $A \cap B = \{\}$.
:::

:::{admonition} Def. 1.6 - Uavhengige hendelser
:class: important

To hendelser $A$ og $B$ kalles uavhengige dersom

$P(A \cap B) = P(A) \cdot P(B)$
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: note

Uavhengighet er sentralt når vi skal beskrive et molekyls bevegelse: translasjon, rotasjon og vibrasjon regnes vanligvis som uavhengige av hverandre. Det er nettopp derfor vi senere kan bruke $P(A \cap B)=P(A)\cdot P(B)$ til å si at partisjonsfunksjonen faktoriserer, $Q_{tot}=Q_{trans}\cdot Q_{rot}\cdot Q_{vib}$.
:::

### 1.2 Grunnleggende aksiomer i sannsynlighet

Nå som vi har noen definisjoner å ta utgangspunkt i, kan vi begynne å se på aksiomene, de grunnleggende byggesteinene, til sannsynlighet. Det virker kanskje åpenbart, men det første vi må notere oss er at sannsynligheten for utfallsrommet $S$ alltid er $1$. Hvorfor? Jo, nettopp fordi hvertfall ett av utfallene i $S$ er garantert. En annen måte å tenke på det på er at det er umulig å havne utenfor utfallsrommet $S$. I eksempelet med terningkast så vi at utfallsrommet var $S=\{1,2,3,4,5,6\}$, vi er altså garantert å få enten $1,2,3,4,5$ eller $6$, men det er umulig å få for eksempel $7$. 

:::{admonition} Sannsynligheten til utfallsrommet $S$
:class: important

Anta at vi har et eksperiment med utfallsrom $S$. Da er

$P(S)=1$
:::

La nå  $i \in S$ være et utfall, og la $P(i)$ være sannsynligheten for $i$. Da må $P(i)$ oppfylle følgende:

:::{admonition} Egenskaper ved sannsynligheten for et utfall
:class: important

Anta at vi har et eksperiment med utfallsrom $S$. For hver $i \in S$ har vi at

1. $0 \leq P(i) \leq 1$
2. $\sum_{i} P(i)=1$
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: note

Aksiomene over gjelder for *ethvert* utfallsrom, også for mikrotilstandene til et termodynamisk system. Sannsynligheten for at et system i termisk likevekt ved temperatur $T$ befinner seg i mikrotilstand $i$ med energi $E_i$ er gitt ved Boltzmann-fordelingen

$P(i) = \dfrac{e^{-E_i/k_BT}}{Q}$

der $k_B$ er Boltzmanns konstant og $Q=\sum_i e^{-E_i/k_BT}$ er **partisjonsfunksjonen**. Merk at $Q$ er akkurat den normeringskonstanten som gjør at $\sum_i P(i)=1$, i tråd med aksiom 2 over.
:::

_Eksempel 1.3_

_Et lite system med 2 partikler som hver kan ha energi $0$ eller $\varepsilon$ har mikrotilstandene $S=\{(0,0),(0,\varepsilon),(\varepsilon,0),(\varepsilon,\varepsilon)\}$ - helt analogt til å kaste to mynter. Makrotilstanden "total energi $=\varepsilon$" er hendelsen $A=\{(0,\varepsilon),(\varepsilon,0)\}$. Er alle mikrotilstander like sannsynlige (høy temperatur), blir $P(A)=2/4=1/2$, akkurat som å få én kron og ett mynt._

### 1.3 Betinget sannsynlighet og additiv lov

**Definisjon 1.7 - Betinget sannsynlighet**

Sannsynligheten for at hendelse $A$ skjer, gitt at hendelse $B$ allerede har skjedd, er

$P(A|B) = \dfrac{P(A \cap B)}{P(B)}$

**Bayes' teorem**

$P(A|B) = \dfrac{P(B|A) \, P(A)}{P(B)}$

:::{admonition} Additiv lov
:class: important

For to hendelser $A$ og $B$ som ikke nødvendigvis er disjunkte, gjelder

$P(A \cup B) = P(A) + P(B) - P(A \cap B)$

Dersom $A$ og $B$ er disjunkte er $P(A \cap B)=0$, og uttrykket blir $P(A \cup B)=P(A)+P(B)$, som i Eksempel 1.2.
:::

### 1.4 Kontinuerlige sannsynlighetsfordelinger

Så langt har vi sett på **diskrete** utfallsrom, som terningkast eller kortstokk, der utfallene kan telles opp én etter én. I fysikalsk kjemi er mange størrelser derimot **kontinuerlige**, for eksempel farten eller energien til et gassmolekyl, som kan anta en hvilken som helst verdi innenfor et intervall.

For en kontinuerlig variabel $x$ bruker vi en **sannsynlighetstetthet** $f(x)$ i stedet for $P(i)$. Sannsynligheten for at $x$ ligger mellom $a$ og $b$ er da

$P(a \leq x \leq b) = \displaystyle\int_a^b f(x)\,dx$

og normeringskravet $\sum_i P(i)=1$ blir i det kontinuerlige tilfellet

$\displaystyle\int_{-\infty}^{\infty} f(x)\,dx = 1$

## 2. Kombinatorikk

### OG-regelen

:::{admonition} OG-regelen
:class: important

Anta hendelse $A$  kan skje på $m$ måter og at hendelse $B$ kan skje uavhengig på $n$ måter. Da er antallet måter $A$ OG $B$ kan skje lik $m \cdot n$.
:::

### ELLER-regelen

:::{admonition} ELLER-regelen
:class: important

Anta at en hendelse $A$ kan skje på $m$ ulike måter, og hendelse $B$ kan skje på $n$ måter. Anta også at hendelsene er disjunkte. Da kan hendelse $A$ ELLER $B$ skje på $m + n$ måter.
:::

_Eksempel: Johan har 5 hatter og 3 solbriller. Antall måter å velge én hatt og et par solbriller blir dermed $5 \cdot 3=15$, mens antall måter å velge enten en hatt eller et par briller blir $5+3=8$._

### Fakultet

Hvor mange måter kan vi ordne $n$ forskjellige objekter i en rekke? Intuitivt vet vi at ved plass 1 i rekken, så har vi $n$ mulige valg av objekter. Deretter, på plass 2, har vi ett valg mindre, altså $n-1$ mulige valg av objekter. Slik følger det, til vi på plass $n$ kun har 1 valg igjen av objekt. Det er nettopp dette fakultet beskriver:

:::{admonition} Fakultet
:class: important

Antallet måter å ordne $n$ ulike objekter i en rekke er gitt ved


$n! = n \cdot (n-1) \cdot (n-2) \cdot \dots \cdot 1$
:::
