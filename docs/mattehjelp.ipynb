## 1. Sannsynlighet

LITEN TEKST OM MOTIVASJONEN RUNDT SANNSYNLIGHET I FYSIKALSK KJEMI.

**Definisjon 1.1 - Utfall**

Et utfall $i$ er et mulig resultat i et eksperiment. 

**Definisjon 1.2 - Utfallsrom**

Utfallsrommet $S$ er mengden av alle mulige utfall i et eksperiment.

**Definisjon 1.3 - Hendelse**

En hendelse $A$ er delmengde av utfallsrommet, altså ett eller flere utfall vi er interessert i.

_Eksempel 1.1 : Anta at eksperimentet ditt er ett kast av en likevektet, sekssidet terning. De mulige utfallene for eksperimentet er $1,2,3,4,5$ eller $6$, som betyr at utfallsrommet er $S=\{1,2,3,4,5,6\}$. Et eksempel på en hendelse i dette eksperimentet er å trille et partall. La oss betegne denne hendelsen ved $A$, og da har vi at $A=\{2,4,6\}$._

**Definisjon 1.4 - Komplement**

Komplementet til en hendelse $A$, $A'$, er mengden av alle utfall i $S$ som ikke er i $A$.

**Definisjon 1.5 - Snitt**

Snittet av to hendelser $A$ og $B$, $A \cap B$, er hendelsen som består av alle utfall som både er i $A$ og i $B$.

**Definisjon 1.6 - Union**

Unionen av to hendelser $A$ og $B$, $A \cup B$, er hendelsen som består av alle utfall som er i $A$ eller i $B$.

![Komplement, snitt og union](bilder/venn_komplement_snitt_union.svg)

_Eksempel 1.2 : Anta igjen at eksperimentet ditt er ett kast av en likevektet, sekssidet terning. Vi er interessert i å trille $2$ eller lavere, som kan beskrives ved hendelsen $A=\{1,2\}$. Komplementet til $A$ blir dermed alle utfallene som er i $S$ men ikke $A$, som tilsvarer å trille 3 eller høyere. Altså er $A'=\{3,4,5,6\}$. Vi er også interessert i å trille $5$ eller høyere, som kan beskrives ved hendelsen $B=\{5,6\}$. Snittet mellom $A$ og $B$ blir den tomme mengden $\{\}$, nettopp fordi det er ingen utfall som er i både $A$ og $B$. Unionen mellom $A$ og $B$ blir $A \cup B = \{1,2,5,6\}$, altså alle utfallene i $A$ og $B$._

### 1a Grunnleggende aksiomer i sannsynlighet

Nå som vi har noen definisjoner å ta utgangspunkt i, kan vi begynne å se på aksiomene, de grunnleggende byggesteinene, til sannsynlighet. Det virker kanskje åpenbart, men det første vi må notere oss er at sannsynligheten for utfallsrommet $S$ alltid er $1$. Hvorfor? Jo, nettopp fordi hvertfall ett av utfallene i $S$ er garantert. En annen måte å tenke på det på er at det er umulig å havne utenfor utfallsrommet $S$. I eksempelet med terningkast så vi at utfallsrommet var $S=\{1,2,3,4,5,6\}$, vi er altså garantert å få enten $1,2,3,4,5$ eller $6$, men det er umulig å få for eksempel $7$.

## Kombinatorikk

### OG-regelen

:::{admonition} OG-regelen
:class: important

Anta hendelse $A$  kan skje på $m$ måter og at hendelse $B$ kan skje uavhengig på $n$ måter. Da er antallet måter $A$ OG $B$ kan skje lik $m \cdot n$.
:::

### ELLER-regelen

:::{admonition} ELLER-regelen
:class: important

Anta at en hendelse $A$ kan skje på $m$ ulike måter, og hendelse $B$ kan skje på $n$ måter. Anta også at hendelsene er disjunkte. Da kan hendelse $A$ ELLER $B$ skje på $m + n$ måter.
:::

_Eksempel: Johan har 5 hatter og 3 solbriller. Antall måter å velge én hatt og et par solbriller blir dermed $5 \cdot 3=15$, mens antall måter å velge enten en hatt eller et par briller blir $5+3=8$._

### Fakultet

Hvor mange måter kan vi ordne $n$ forskjellige objekter i en rekke? Intuitivt vet vi at ved plass 1 i rekken, så har vi $n$ mulige valg av objekter. Deretter, på plass 2, har vi ett valg mindre, altså $n-1$ mulige valg av objekter. Slik følger det, til vi på plass $n$ kun har 1 valg igjen av objekt. Det er nettopp dette fakultet beskriver:

:::{admonition} Fakultet
:class: important

Antallet måter å ordne $n$ ulike objekter i en rekke er gitt ved

\begin{align*}
n! = n \cdot (n-1) \cdot (n-2) \cdot \dots \cdot 1
\end{align*}
:::
