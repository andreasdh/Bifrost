### 1.1 Utfall, utfallsrom og hendelser

Du triller en terning, og teller antall øyne terningen viser. Dette er et av mange eksempler på et statistisk eksperiment. For å kunne snakke om eksperimenter trenger vi noen definisjoner. Vi starter med de mest grunnleggende: utfall, utfallsrom og hendelser.

:::{admonition} Def. 1.1 - Utfall
:class: important

Et utfall $i$ er et mulig resultat i et eksperiment. 
:::


:::{admonition} Def. 1.2 - Utfallsrom $S$
:class: important

Utfallsrommet $S$ er mengden av alle mulige utfall i et eksperiment.
:::


:::{admonition} Def. 1.3 - Hendelse
:class: important

En hendelse $A$ er delmengde av utfallsrommet, altså ett eller flere utfall vi er interessert i.
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: note

I statistisk termodynamikk kaller vi et utfall for en **mikrotilstand**, altså én bestemt fordeling av energi/posisjon over alle partiklene i systemet. Utfallsrommet $S$ blir da mengden av *alle* mikrotilstander systemet kan befinne seg i, og en hendelse $A$ tilsvarer en **makrotilstand** (f.eks. en gitt total energi, temperatur eller trykk) - altså alle mikrotilstander som gir samme makroskopiske observasjon.
:::

:::{card} 
**Eksempel**
^^^
Du triller en sekssidet terning.

a) Hva er utfallsrommet $S$ til eksperimentet?

Du kan enten få $1,2,3,4,5$ eller $6$, som betyr at utfallsrommet er $S=\{1,2,3,4,5,6\}$.

b) Hva er et eksempel på en hendelse i dette eksperimentet?

Et eksempel på en hendelse i dette eksperimentet er å trille et partall. La oss betegne denne hendelsen ved $A$, og da har vi at $A=\{2,4,6\}$.
:::

:::{admonition} Underveisoppgave 1.1
:class: tip

Du trekker ett kort fra en ordinær kortstokk med totalt 52 kort, og ser på kortets symbol (spar, hjerter, ruter eller kløver).

a) Hva er utfallsrommet $S$ til eksperimentet?

b) Hva er sannsynligheten for hvert av utfallene i utfallsrommet?

c) Gi eksempel på en mulig hendelse av dette eksperimentet.
::

:::{admonition} Løsning 1.1
:class: tip, dropdown

a) De mulige symbolene er spar, hjerter, ruter eller kløver, og dermed er utfallsrommet $S=\{spar, hjerter, ruter, kløver\}$.

b) Det er $13$ kort av hver symbolstype, slik at sannsynligheten for hvert av utfallene blir $\frac{13}{52}=\frac{1}{4}$.

c) Mulige hendelser for eksperimentet. Eksempler: trekke rødt kort ($\{ruter, hjerter\}$) eller trekke svart kort ($\{spar, kløver\}$). Merk at hendelser som å for eksempel trekke bildekort (konge, dame, knekt, ess) IKKE er hendelser i dette eksperimentet, nettopp fordi det ikke er en delmengde av utfallsrommet $S$.
:::

### 1.2 Komplement, snitt og union

La oss nå se på hva som skjer utenfor utfallsrommet til en hendelse $A$, og hvordan vi kan beskrive sammenhengen mellom to hendelser $A$ og $B$. Vi ser først på komplementet til en hendelse, og deretter snitt og union.

:::{admonition} Def. 1.4 - Komplement
:class: important

Komplementet til en hendelse $A$, $A'$, er mengden av alle utfall i $S$ som ikke er i $A$.
:::


:::{admonition} Def. 1.5 - Snitt
:class: important

Snittet av to hendelser $A$ og $B$, $A \cap B$, er hendelsen som består av alle utfall som både er i $A$ og i $B$.
:::


:::{admonition} Def. 1.6 - Union
:class: important

Unionen av to hendelser $A$ og $B$, $A \cup B$, er hendelsen som består av alle utfall som er i $A$ eller i $B$.
:::

![Komplement, snitt og union](bilder/venn_komplement_snitt_union.svg)

:::{card} 
**Eksempel**
^^^
La igjen eksperimentet ditt være ett kast av en likevektet, sekssidet terning. Vi er interessert i å trille $2$ eller lavere, som kan beskrives ved hendelsen $A=\{1,2\}$. Vi er også interessert i å trille $5$ eller høyere, som kan beskrives ved hendelsen $B=\{5,6\}$.
 
a) Hva er komplementet til $A$?

Komplementet til $A$ er alle utfallene som er i $S$ men ikke $A$, som tilsvarer å trille 3 eller høyere. Altså er $A'=\{3,4,5,6\}$. 

b) Hva er snittet av $A$ og $B$?

Snittet mellom $A$ og $B$ blir den tomme mengden $\{\}$, nettopp fordi det er ingen utfall som er i både $A$ og $B$.

c) Hva er unionen av $A$ og $B$?
 
Unionen mellom $A$ og $B$ blir $A \cup B = \{1,2,5,6\}$, altså mengden alle utfallene i $A$ og $B$ samlet.
:::


I siste del av 1.1 definerer vi disjunkte hendelser og uavhengighet. Vi skal se at uavhengighet blir svært sentralt i statistisk termodynamikk.

:::{admonition} Def. 1.5 - Disjunkte hendelser
:class: important

To hendelser $A$ og $B$ kalles disjunkte dersom de ikke har noen felles utfall. Matematisk, $A \cap B = \{\}$.
:::

:::{admonition} Def. 1.6 - Uavhengige hendelser
:class: important

To hendelser $A$ og $B$ kalles uavhengige dersom

$P(A \cap B) = P(A) \cdot P(B)$
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: note

Uavhengighet er sentralt når vi skal beskrive et molekyls bevegelse: translasjon, rotasjon og vibrasjon regnes vanligvis som uavhengige av hverandre. Det er nettopp derfor vi senere kan bruke $P(A \cap B)=P(A)\cdot P(B)$ til å si at partisjonsfunksjonen faktoriserer, $Q_{tot}=Q_{trans}\cdot Q_{rot}\cdot Q_{vib}$.
:::

### 1.3 Kolmogorovs aksiomer for sannsynlighet

Enten du spiller sjakk eller ikke, vet du at spillet har visse regler, som at hver brikke kan flyttes på en spesiell måte, at hvit starter osv. Disse reglene er ikke bevist, men de er fastsatt som et utgangspunkt for spillet. Alt annet du kan si om sjakk, for eksempel hvordan man vinner, taper eller om en åpning er god eller dårlig, er utledet fra disse reglene. 

Sånn er også matematikken. Som reglene i sjakk, har matematikken grunnleggende byggesteiner som kalles **aksiomer**. Teoremer og alt annet bygges logisk på aksiomene. Kolmogorov presenterte aksiomene i sannsynlighetsteori, og til tross for at disse kan virke åpenbare, er de utgangspunktet til alt vi skal se på - også flere utledninger i statistisk termodynamikk.

:::{admonition} Kolmogorovs sannsynlighetsaksiomer
:class: note

For et eksperiment med utfallsrom $S$ har vi


**Ikke-negativitet**

Sannsynligheten for enhver hendelse $A$ er ikke-negativ

1. $P(A) \ge 0$ 

**Normering**

Sannsynligheten for $S$ er lik $1$, altså

2. $P(S)=1$

**Additivitet**

Sannsynligheten for unionen av flere disjunkte hendelser $A_1, A_2,\dots$ (hendelser som ikke overlapper i utfall) er summen av sannsynligheten for hver av hendelsene

3. $P(A_1 \cap A_2 \cap \dots)=P(A_1)+P(A_2)+\dots$
:::

:::{admonition} Kolmogorovs aksiomer lett forklart
:class: tip, dropdown

En hendelse $A$ vil enten skje eller ikke. Skjer den ikke, er $P(A)=0$, skjer den, må $P(A)>0.$

Sannsynligheten for $S$ må være $1$, fordi noe i utfallsrommet **må** skje, og noe utenfor $S$ kan ikke skje. Triller du en terning er du garantert å få enten $1,2,3,4,5$ eller $6$, men du kan aldri få for eksempel $7$, da $7$ ikke er i $S$ for terningkast.

Hvis du vil finne sannsynligheten for å få enten $A$ eller $B$, og de ikke har noen av de samme utfallene, kan du summere de individuelle sannsynlighetene for $A$ og $B$. Tenk på terningkast og hendelsene $A=\{1\}$ og $B=\{2\}$. Disse hendelsene er disjunkte fordi de ikke har noen felles utfall, og dermed vil

$P(A \cup B)=P(\{1,2\})=P(\{1\})+P(\{2\})=\frac{1}{6}+\frac{1}{6}=\frac{1}{3}$
:::

Som nevnt er aksiomene grunnlaget for utledninger og teoremer. Et viktig utledning av aksiom 2 og 3 er følgende:

:::{admonition} Teorem 1.1 - Summen av alle utfalls sannsynligheter
:class: note

For et vilkårlig eksperiment med utfallsrom $S$ har vi

$\sum_{i}P(i)=1$

for alle utfall $i \in S$.
:::

:::{admonition} Underveisoppgave 1.2
:class: tip

Vis at Teorem 1.1 stemmer for eksempelet om terningkast av en sekssidet terning.
:::

:::{admonition} Hint
:class: tip, dropdown

Vis at summen av sannsynlighetene for alle utfallene er lik $1$.
:::

:::{admonition} Løsning
:class: tip, dropdown

Utfallsrommet er $S=\{1,2,3,4,5,6\}$. Sannsynligheten for hvert av utfallene er $\frac{1}{6}$, slik at

$\sum_{i}P(i)=P(\{1\})+P(\{2\})+P(\{3\})+P(\{4\})+P(\{5\})+P(\{6\})=\frac{1}{6}+\frac{1}{6}+\frac{1}{6}+\frac{1}{6}+\frac{1}{6}+\frac{1}{6}=1$
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: note

Aksiomene over gjelder for *ethvert* utfallsrom, også for mikrotilstandene til et termodynamisk system. Sannsynligheten for at et system i termisk likevekt ved temperatur $T$ befinner seg i mikrotilstand $i$ med energi $E_i$ er gitt ved Boltzmann-fordelingen

$P(i) = \dfrac{e^{-E_i/k_BT}}{Q}$

der $k_B$ er Boltzmanns konstant og $Q=\sum_i e^{-E_i/k_BT}$ er **partisjonsfunksjonen**. Merk at $Q$ er akkurat den normeringskonstanten som gjør at $\sum_i P(i)=1$, i tråd med Teorem 1.1 over.
:::

:::{card} 
**Eksempel**
^^^
Et lite system med 2 partikler som hver kan ha energi $0$ eller $\varepsilon$ har mikrotilstandene $S=\{(0,0),(0,\varepsilon),(\varepsilon,0),(\varepsilon,\varepsilon)\}$ - helt analogt til å kaste to mynter. Makrotilstanden "total energi $=\varepsilon$" er hendelsen $A=\{(0,\varepsilon),(\varepsilon,0)\}$. Er alle mikrotilstander like sannsynlige (høy temperatur), blir $P(A)=2/4=1/2$, akkurat som å få én kron og ett mynt.
:::


### 1.4 Betinget sannsynlighet 

Så langt har vi sett på sannsynligheten for individuelle hendelser. I mange eksperimenter er vi imidlertid interessert i hvordan sannsynligheten for én hendelse påvirkes av at en annen hendelse allerede har skjedd, eller hvordan vi kan finne sannsynligheten for at minst én av flere hendelser skjer.

:::{admonition} Def. 1.7 - Betinget sannsynlighet
:class: important

Sannsynligheten for at hendelse $A$ skjer, gitt at hendelse $B$ allerede har skjedd, er

$P(A|B) = \dfrac{P(A \cap B)}{P(B)}$
:::


**Bayes' teorem**

$P(A|B) = \dfrac{P(B|A) \, P(A)}{P(B)}$

:::{admonition} Additiv lov
:class: important

For to hendelser $A$ og $B$ som ikke nødvendigvis er disjunkte, gjelder

$P(A \cup B) = P(A) + P(B) - P(A \cap B)$

Dersom $A$ og $B$ er disjunkte er $P(A \cap B)=0$, og uttrykket blir $P(A \cup B)=P(A)+P(B)$, som i Eksempel 1.2.
:::

### 1.5 Kontinuerlige sannsynlighetsfordelinger

Så langt har vi sett på **diskrete** utfallsrom, som terningkast eller kortstokk, der utfallene kan telles opp én etter én. I fysikalsk kjemi er mange størrelser derimot **kontinuerlige**, for eksempel farten eller energien til et gassmolekyl, som kan anta en hvilken som helst verdi innenfor et intervall.

For en kontinuerlig variabel $x$ bruker vi en **sannsynlighetstetthet** $f(x)$ i stedet for $P(i)$. Sannsynligheten for at $x$ ligger mellom $a$ og $b$ er da

$P(a \leq x \leq b) = \displaystyle\int_a^b f(x)\,dx$

og normeringskravet $\sum_i P(i)=1$ blir i det kontinuerlige tilfellet

$\displaystyle\int_{-\infty}^{\infty} f(x)\,dx = 1$

## 2 Kombinatorikk

Mens sannsynlighet handler om hvor stor sjanse et utfall har til å skje, handler kombinatorikk om noe mer grunnleggende: å telle *hvor mange* mulige utfall som faktisk finnes. Dette er ofte det første steget i et sannsynlighetsregnestykke - før vi kan si noe om sannsynligheten for en hendelse, må vi vite hvor mange måter den kan skje på, og hvor mange muligheter som finnes totalt.

I fysikalsk kjemi er dette limet i statistisk termodynamikk: å telle antall mikrotilstander som gir samme makrotilstand (den **statistiske vekten** $\Omega$) er nøyaktig et kombinatorikkproblem. Dette tallet dukker senere opp direkte i Boltzmanns entropiformel, $S = k_B \ln \Omega$ - så verktøyene du lærer her er ikke bare telleteknikker, de er selve grunnlaget for hvordan vi definerer entropi.

### 2.1 OG-regelen og ELLER-regelen

OG-og ELLER-regelen er garantert regler du har brukt enten bevisst eller ubevisst før. Her kommer definisjonen av disse.

:::{admonition} OG-regelen
:class: important

Anta at en hendelse $A$ kan skje på $m$ måter og at hendelse $B$ kan skje uavhengig på $n$ måter. Da er antallet måter $A$ OG $B$ kan skje lik 

$m \cdot n$.
:::

:::{admonition} ELLER-regelen
:class: important

Anta at en hendelse $A$ kan skje på $m$ ulike måter, og hendelse $B$ kan skje på $n$ måter. Anta også at hendelsene er disjunkte. Da er antall måter hendelse $A$ ELLER $B$ skje på lik

$m + n$ 
:::

:::{card} 
**Eksempel**
^^^
Et gassmolekyl kan innta 4 ulike vibrasjonstilstander og 3 ulike rotasjonstilstander, og disse frihetsgradene er uavhengige av hverandre.

a) Hvor mange mikrotilstander der molekylet har en bestemt kombinasjon av vibrasjon og rotasjon finnes?

Antall mikrotilstander der molekylet har en bestemt kombinasjon av vibrasjon OG rotasjon blir dermed

$4 \cdot 3=12$

b) Hvor mange mikrotilstander der molekylet befinner seg i én bestemt virbasjonstilstand eller én bestemt rotasjonstilstand finnes?

Antallet mikrotilstander der molekylet befinner seg i én bestemt vibrasjonstilstand ELLER én bestemt rotasjonstilstand (disjunkte tilstander) blir

$4+3=7$
:::


### 2.2 Fakultet

Fakultet ser skummelt ut, for hvorfor er det plutselig utropstegn i matten? Vi skal se at fakultet kan hjelpe oss å finne antall rekkefølger når vi har $n$-antall forskjellige objekter. Fakultet er definert slik

:::{admonition} Fakultet
:class: important

For et ikke-negativt heltall $n$, er fakultet definert slik

$n! = n \cdot (n-1) \cdot (n-2) \cdot \dots \cdot 1$ 

$0!$ er ved definisjon lik $1$.
:::

Vi har $n$-antall objekter som vi skal plassere i rekke - hvor mange ulike måter kan vi gjøre dette på? Første objekt har $n$-antall muligheter. Deretter har andre objekt $n-1$-antall muligheter. Slik følger det videre, helt til objekt $n$ kun har én mulighet. Ved bruk av OG-regelen fra tidligere, multipliseres alle mulighetene og gir opphav til fakultet.

Fakultet vokser utrolig fort, bare se her:

$1! = 1$

$2! = 2 \cdot 1 = 2$

$3! = 3 \cdot 2 \cdot 1 = 6$

$4! = 4 \cdot 3 \cdot 2 \cdot 1 = 24$

$5! = 5 \cdot 4 \cdot 3 \cdot 2 \cdot 1 = 120$

$6! = 6 \cdot 5 \cdot 4 \cdot 3 \cdot 2 \cdot 1 = 720$

$7! = 7 \cdot 6 \cdot 5 \cdot 4 \cdot 3 \cdot 2 \cdot 1 = 5040$

$8! = 8 \cdot 7 \cdot 6 \cdot 5 \cdot 4 \cdot 3 \cdot 2 \cdot 1 = 40320$

Før vi kan regne på antall mulige utfall, må vi vite hva vi har å gjøre med. Dette må vi vite

**1. Betyr rekkefølgen noe?**
**Ordnet** betyr at rekkefølgen betyr noe. Se for deg at du ser på pallplasseringer i fotball-VM. Hvis du er interessert i å se på ulike utfall der 1,2 og 3 plass betyr noe, altså er rekkefølgen ordnet, er følgende to forskjellige utfall:

1. Argentina
2. England
3. Frankrike
 
og

1. England
2. Argentina
3. Frankrike

Ser du derimot på kombinasjoner av land som kan havne på pallplassering uavhengig av rekkefølgen, altså uordnet, vil utfallene ovenfor være like. Ser du parallellen mellom dette og mikro-og makrotilstander?


**2. Kan objekter velges flere ganger?**
**Tilbakelegging** betyr at et element kan velges flere ganger. I eksemplene ovenfor kan vi ikke velge et land flere ganger (du har jo ikke flere Argentinaer). Siden vi i de fleste kjemieksempler er uten tilbakelegging,fokuserer vi ikke på dette nå.

### 4.1 Definisjonen av en funksjon

Funksjoner i matematikk er regler som forbinder hvert eneste tall i en mengde med ett og bare ett tall i en annen eller samme mengde. Du er allerede kjent med flere funksjoner, som for eksempel funksjonen for arealet av en sirkel gitt dens radius, pH-en til en løsning gitt dens syrekonsentrasjon eller antall H-atomer i et alkan gitt antallet C-atomer. 

:::{card} 
**Eksempel**
^^^
For et alkan med et heltall $n$ antall C-atomer, er antallet H-atomer $H$ gitt ved funksjonen

$H=2n+2$

Dette er en funksjon fordi for hver $n$ får vi nøyaktig én $H$. Vi kan også skrive $H(n)$ for $H$ for å spesifisere at $H$ er en funksjon av $n$:

$H(n)=2n+2$
:::


Som du nå ser kan funksjoner brukes til å si noe om hvordan en størrelse varierer med en annen størrelse. Derfor er det naturlig å finne ut nøyaktig hva vi kan si om disse såkalte funksjonene. Vi begynner med definisjonen av en funksjon, og dens definisjonsmengde og verdiområde.

:::{admonition} Def. 4.1 - Funksjon
:class: important

*Intuitivt:* En funksjon $f$ er en regel som tildeler hver verdi $x$ nøyaktig én verdi $f(x)$.

*Matematisk:* La $X$ og $Y$ være to mengder. En funksjon $f$ fra $X$ til $Y$, tildeler hvert element $x \in X$ nøyaktig ett element i $Y$, $f(x)$.

Vi bruker følgende notasjon:

$f:X \rightarrow Y$
:::

:::{admonition} Def. 4.2 - Definisjonsmengde og verdiområde
:class: important

La $f:X \to Y$ være en funksjon fra $X$ til $Y$. Da kalles $X$ for **definisjonsmengden** til $f$, mens $Y$ kalles for **verdiområdet**.
:::

:::{card} 
**Eksempel**
^^^
La $H:\{1,2,3,4,5\} \to N$ være funksjonen for antallet H-atomer $H(n)$ i et alkan med $n$-antall C-atomer, gitt ved

$H(n)=2n+2$

a) Hva er definisjonsområdet til $H$?

Definisjonsmengden er $\{1,2,3,4,5\}$ og sier noe om hvilke $n$-verdier vi ser på. Merk at dette er en diskret definisjonsmengde fordi vi ikke har alle tallene mellom $1$ og $5$, men spesifikt $1,2,3,4$ eller $5$. Med andre ord ser vi på metan, etan, propan, butan og pentan.

b) Hva er verdiområdet til $H$?

Verdiområdet er $N$, altså de naturlige tallene. Det betyr at antallet H-atomer $H(n)$ er et naturlig tall.
:::



Vi er ferdig med første steg, nemlig definisjonen av en funksjon. Allerede nå har vi fått to nye begreper å beskrive funksjoner med - definisjonsmengde og verdiområde. Før vi går videre, skal vi se på noen måter å representere funksjoner på. Du er allerede kjent med grafer og algebraisk representasjon (formelen for funksjonen). Det er i hovedsak disse representasjonene som er mest relevante for deg som kjemistudent, men det finnes også flere!

![Ulike representasjoner av funksjoner](bilder/funksjon_representasjoner.svg)

:::{admonition} Underveisoppgave 4.1
:class: tip

En rett sylinder har en fast høyde på $10$ cm. Volumet $V$ (i cm³) avhenger av radien $r$ (i cm) gjennom $V(r)=10 \cdot \pi r^2$.

a) Forklar hvorfor $V$ er en funksjon av $r$.

b) Hva er en naturlig definisjonsmengde for $V(r)$ når vi tenker på en fysisk sylinder?

c) Regn ut $V(2)$.
:::

:::{admonition} Løsning 4.1
:class: tip, dropdown

a) $V$ er en funksjon av $r$ fordi hver verdi av radien $r$ gir nøyaktig én verdi av volumet $V$. Dette er nettopp kravet fra Def. 4.1.

b) Radien til en fysisk sylinder må være positiv, så en naturlig definisjonsmengde er alle positive reelle tall, altså $r>0$. (Vi kan eventuelt ta med $r=0$ om vi tillater en "sylinder" uten utstrekning, men da blir volumet $0$).

c) Vi setter inn $r=2$: $V(2)=10 \cdot \pi \cdot 2^2 = 40\pi \approx 125{,}7$ cm³.
:::

### 4.2 Jevne og odde funksjoner            

To viktige typer funksjoner du kan komme på å møte i fysikalsk kjemi er **jevne og odde funksjoner**.

:::{admonition} Def. 4.3 - Jevne og odde funksjoner
:class: important

Anta at $f$ er definert ved både $x$ og $-x$.

Vi kaller $f$ for en **jevn funksjon** hvis

$f(-x)=f(x)$ for alle $x$ i definisjonsområdet til $f$.

Vi kaller $f$ for en **odde funksjon** hvis

$f(-x)=-f(x)$ for alle $x$ i definisjonsområdet til $f$.
:::

:::{card} 
**Eksempel**
^^^
Vis at $f(x)=x^2$ er en jevn funksjon.

Vi vet at $f$ er definert for alle reelle tall.

La $x$ være et vilkårlig tall i definisjonsområdet til $f$, altså $x \in R$. Da blir

$f(x)=x^2$

Samtidig blir

$f(-x)=(-x)^2=x^2$

Altså har vi vist at $f(x)=f(-x)$, og $f$ er en jevn funksjon.
:::

:::{card} 
**Eksempel**
^^^
Vis at $f(x)=x^3-2x$ er en odde funksjon.

Vi vet at $f$ er definert for alle reelle tall.

La $x$ være et vilkårlig tall i definisjonsområdet til $f$, altså $x \in R$. Da blir

$f(x)=x^3-2x$

Samtidig blir

$f(-x)=(-x)^3-2(-x)=-x^3+2x=-(x^3-2x)$

Altså har vi vist at $f(x)=-f(x)$, og $f$ er en odde funksjon.
:::

:::{caution}
En funksjon **må ikke** være enten jevn eller odde, den kan være ingen av delene.
:::

Det er derimot mye enklere å forstå hva jevne og odde funksjoner er ved å visualisere dem, fordi egenskapene beskrevet ovenfor gir opphav til en symmetri. Jevne funksjoner er symmetriske om $y$-aksen, mens odde funksjoner er symmetriske om origo. Nedenfor får du eksempel på en jevn og en odde funksjon, og kan endre $x$-verdiene for å se hvordan $f(x)$ og $f(-x)$ endrer seg.

```python
from IPython.display import display, HTML
import html

def vis_sim(sti, width="100%"):
    with open(sti, "r", encoding="utf-8") as f:
        innhold = f.read()
    escaped = html.escape(innhold, quote=True)
    display(HTML(
        f'<iframe srcdoc="{escaped}" width="{width}" height="200" '
        f'scrolling="no" style="border:none;display:block;" '
        f'onload="this.style.height='
        f"this.contentWindow.document.body.scrollHeight+'px'\"></iframe>"
    ))

vis_sim("Simuleringer/jevn_odde_funksjon.html");
```

### 4.3 Grenseverdier og kontinuitet

Vi innleder grenseverdier med et eksempel.

:::{card} 
**Eksempel**
^^^
Beskriv funksjonen $f(x)=\frac{x^2-1}{x-1}$ omkring $x=1$.

La oss se hva vi kan si om definisjonsmengden til $f$, altså hvilke $x$-verdier vi kan ha. Annet enn $x=1$ som gir $0$ i nevneren, er alle reelle $x$-verdier tillate. Verdiområdet er de reelle tallene.

Et observant øye ser at det er mulig å faktorisere uttrykket, slik at:

$f(x)=\frac{x^2-1}{x-1}=\frac{(x-1)(x+1)}{(x-1)}=x+1$ for $x\neq1$

Grafen til $f$ blir dermed linjen $y=x+1$ **uten** punktet i $(1,2)$, der blir det et hull! Selv om $f(1)$ ikke er definert, kan vi få $f(x)$ så nærme vi vil $2$ ved å velge en $x$ nærme nok $1$. Vi sier at $f(x)$ går mot grenseverdien $2$ når $x$ går mot $1$. Notasjonen for dette er:

$\lim_{x \to 1} f(x)=2$

eller

$\lim_{x \to 1} \frac{x^2-1}{x-1} = 2$
:::

![Grenseverdi eksempel med hull](bilder/funksjon_hull.svg)

Den formelle definisjonen av grenseverdi er komplisert, og muligens kjent for deg som har hatt MAT1100. Vi går for å presentere den uformelle definisjonen i stedet, da den ikke er særlig viktig for oss.

:::{admonition} Def. 4.4 - Grenseverdi (uformelt)
:class: important

La $f$ være en funksjon og $L$ et tall. Hvis $f(x) \to L$ når $x \to a$, kaller vi $L$ for **grenseverdien** til $f(x)$ når $x$ går mot $a$. Vi skriver:

$\lim_{x \to a} f(x) = L$
:::

Vi har nå nok verktøy til å gå løs på kontinuitet. Heldigvis har kontinuitet en veldig intuitiv forklaring: hvis du kan tegne grafen til en funksjon fra start til slutt uten å måtte løfte pennen din av papiret, vel da er den kontinuerlig. Men, matematikken er formell og vi må ha klare og tydelige definisjoner. 

:::{admonition} Def. 4.5 - Kontinuitiet i et punkt
:class: important

Vi sier at funksjonen $f$ er **kontinuerlig i et punkt** $c$ i definisjonsområdet hvis

$\lim_{x \to c} f(x) = f(c)$

Hvis 

$\lim_{x \to c} f(x)$ ikke eksisterer 

eller 
$\lim_{x \to c} f(x) \neq f(c)$ 

så er $f$ **diskontinuerlig** i $c$.
:::

![Kontinuitet i ett punkt](bilder/kontinuitet_tre_grafer.svg)

:::{admonition} Def. 4.6 - Kontinuitiet i et intervall
:class: important

Vi sier at funksjonen $f$ er **kontinuerlig på intervallet** $I$ hvis den er kontinuerlig i hver at punktene i $I$.
:::

:::{admonition} Def. 4.7 - Kontinuerlig funksjon
:class: important

Vi sier at funksjonen $f$ er **kontinuerlig** (på hele definisjonsområdet) hvis den er kontinuerlig i hvert at punktene i definisjonsområdet.
:::

:::{admonition} Underveisoppgave 4.3
:class: tip

a) Regn ut grenseverdien $\lim_{x \to 2} \dfrac{x^2-4}{x-2}$ ved å faktorisere telleren.

b) Funksjonen $g$ er definert ved $g(x)=\dfrac{x^2-4}{x-2}$ for $x \neq 2$, og $g(2)=3$. Er $g$ kontinuerlig i $x=2$? Begrunn med Def. 4.5.
:::

:::{admonition} Løsning 4.3
:class: tip, dropdown

a) Vi faktoriserer telleren: $x^2-4=(x-2)(x+2)$. For $x \neq 2$ er da

$\frac{x^2-4}{x-2}=\frac{(x-2)(x+2)}{x-2}=x+2$

slik at $\lim_{x \to 2} \dfrac{x^2-4}{x-2}=\lim_{x \to 2}(x+2)=4$.

b) For at $g$ skal være kontinuerlig i $x=2$, må $\lim_{x \to 2} g(x)=g(2)$. Vi fant at grenseverdien er $4$, men $g(2)=3$. Siden $\lim_{x \to 2} g(x)=4 \neq 3=g(2)$, er $g$ **diskontinuerlig** i $x=2$. (Hadde vi i stedet satt $g(2)=4$, ville funksjonen vært kontinuerlig.)
:::

### 4.4 Derivasjon

Derivasjon er et matematisk verktøy for å sette ord på noe du sannsynligvis allerede ser på grafen til en funksjon. Følger du en graf, kan du for eksempel observere at den stiger, når et toppunkt, synker litt og kanskje treffer et bunnpunkt. Husk at en av hovedgrunnene til at vi interesserer oss for funksjoner er at de kan modellere virkelige fenomener og sammenhengen mellom to størrelser. Ofte er vi derfor interessert i hvordan denne sammenhengen endrer seg, hvor mye den endrer seg, og i hvilken retning.

Vi begynner med å introdusere gjennomsnittlig endring mellom to punkter.

:::{admonition} Def. 4.8 - Gjennomsnittlig endringsrate mellom to punkter
:class: important

Den gjennomsnittlige endringsraten til en funksjon $f$ over et intervall $[a,b]$ er gitt ved 

$\frac{\Delta y}{\Delta x}=\frac{f(b)-f(a)}{b-a}$
:::

![Gjennomsnittlig endringsrate](bilder/gjennomsnittlig_endring.svg)

En rett linje som går gjennom to punkter, slik som den rette linjen som går gjennom $(a,f(a))$ og $(b,f(b))$, kalles en **sekant**. Den gjennomsnittlige endringsraten mellom disse to punktene er da stigningstallet til sekanten. Men hva med endringsraten i ett enkelt punkt? Vi introduserer definisjonen av den deriverte.

:::{card} 
**Eksempel**
^^^
Hva er den gjennomsnittlige endringsraten for funksjonen $f(x)=x^2+1$ i intervallet $[1,3]$?

Vi har at

$f(1)=1^2+1=2$

$f(3)=3^2+1=10$

Dermed får vi

$\frac{\Delta y}{\Delta x}=\frac{10-2}{3-1}=\frac{8}{2}=4$
:::

:::{admonition} Def. 4.8 - Den deriverte
:class: important

Den **deriverte** til en funksjon $f$ er en annen funksjon $f'$ definert ved

$f'(x)=\lim_{h \to 0} \frac{f(x+h)-f(x)}{h}$

for alle $x$ der grenseverdien eksisterer.

Hvis $f'(x)$ eksisterer, sier vi at $f$ er **deriverbar** i $x$.
:::

:::{card} 
**Eksempel**
^^^
Bruk definisjonen til å finne den deriverte av $f(x)=x^2$.

Vi setter inn i definisjonen:

$f'(x)=\lim_{h \to 0} \frac{f(x+h)-f(x)}{h}=\lim_{h \to 0} \frac{(x+h)^2-x^2}{h}$

Vi regner ut telleren:

$(x+h)^2-x^2=x^2+2xh+h^2-x^2=2xh+h^2$

Dermed får vi

$f'(x)=\lim_{h \to 0} \frac{2xh+h^2}{h}=\lim_{h \to 0} (2x+h)=2x$
:::

:::{card} 
**Eksempel**
^^^
Bruk definisjonen til å finne den deriverte av $f(x)=x^2-3x$.

Vi setter inn i definisjonen:

$f'(x)=\lim_{h \to 0} \frac{f(x+h)-f(x)}{h}=\lim_{h \to 0} \frac{\left[(x+h)^2-3(x+h)\right]-\left[x^2-3x\right]}{h}$

Vi regner ut telleren:

$(x+h)^2-3(x+h)-x^2+3x=x^2+2xh+h^2-3x-3h-x^2+3x=2xh+h^2-3h$

Dermed får vi

$f'(x)=\lim_{h \to 0} \frac{2xh+h^2-3h}{h}=\lim_{h \to 0} (2x+h-3)=2x-3$
:::

Definisjonen av den deriverte er ikke akkurat lett fordøyelig. Vi har punktene $(x, f(x))$ og tar et steg $h$ fra $x$ til punktet $(x+h, f(x+h))$. Så gjør vi $h$ mindre og mindre - vi finner altså grenseverdien av den gjennomsnittlige endringsraten mellom disse punktene når $h$ går mot $0$. Nedenfor kan du endre på $x$ og $h$ og se hvordan den gjennomsnittlige endringsraten og den deriverte endrer seg. Se spesielt på hva som skjer når $h$ går mot $0$. 

```python
vis_sim("Simuleringer/derivert_fra_sekant.html")
```

For å få en enda bedre forståelse for hva den deriverte er, se hvordan verdien til den deriverte endrer seg i simulasjonen nedenfor når vi følger grafen fra mindre til større $x$-verdier. Hva skjer med den deriverte ved topp-og bunnpunktene?

```python
vis_sim("Simuleringer/derivert_kritiske_punkter.html")
```

Vi har funnet den første-deriverte til $f$, men det er ingenting som stopper oss fra å finne høyere-ordens deriverte slik som den dobbelderiverte, tredje deriverte osv.

:::{admonition} Def. 4.8 - Høyere-ordens deriverte (n-te deriverte)
:class: important

Den $n$**-te deriverte** til en funksjon $f$ er en annen funksjon $f^{(n)}$ definert ved å ta den deriverte av $f$ $n$-antall ganger.
:::

Nå som vi har en idé av hva den deriverte er, skal vi lære oss å finne de deriverte for ulike funksjoner. I et av eksemplene ovenfor finner vi $f'$ for $f(x)=x^2$ direkte ut ifra definisjonen av den deriverte. Heldigvis for oss har mange flinke matematikere funnet ut av de deriverte for de vanligste, elementære funksjonene. De aller vanligste er gitt i tabellen nedenfor. 

| $f(x)$       | $f'(x)$      |  Betingelser |
|-----------   |-----------  |-------|
| $a$        | $0$           |  |
| $x^a$       | $ax^{a-1}$   | |
| $e^x$      | $e^x$   | |
| $a^x$   | $a^x \ln{(a)}$   | $a>0$ |
| $\ln{(x)}$ | $\frac{1}{x}$ | $x>0$ |
| $\log_{a}{(x)}$ | $\frac{1}{x \ln{(a)}}$ | $x,a>0$ |
| $\sin{(x)}$ | $\cos{(x)}$ | |
| $\cos{(x)}$ | $-\sin{(x)}$ | | 
| $\tan{(x)}$ | $\sec^2{(x)}=\frac{1}{\cos^2{(x)}}=1+\tan^2{(x)}$ | | 
| $\arcsin{(x)}$ | $\frac{1}{\sqrt{1-x^2}}$ | $-1 < x < 1$ |
| $\arccos{(x)}$ | -$\frac{1}{\sqrt{1-x^2}}$ | $-1 < x < 1$ |
| $\arctan{(x)}$ | $\frac{1}{1+x^2}$ |  |

For å finne de deriverte til sammensatte og mer kompliserte funksjoner, må vi ha noen derivasjonsregler. Først introduseres litt notasjon for å unngå forvirring!

| Lagrange       | Leipniz                |
|-----------     |-----------             |
| $f'(x)$        | $\frac{df}{dx}$        |
| $f''(x)$       | $\frac{d^2f}{dx^2}$   |
| $f'''(x)$      | $\frac{d^3f}{dx^3}$   |
| $f^{(n)}(x)$   | $\frac{d^nf}{dx^n}$   |

Videre presenteres de ulike derivasjonsreglene. 

:::{admonition} Teorem 4.1 - Faktorregelen
:class: note

La $g$ være en deriverbar funksjon i $x$ og la $a$ være et tall. Da er 

$\frac{d}{dx} [a \cdot g(x)]=a \cdot g'(x)$
:::

:::{admonition} Teorem 4.2 - Summeregelen
:class: note

La funksjonene $g$ og $h$ er deriverbare i $x$. Da er

$\frac{d}{dx} [g(x) + h(x)]=g'(x) + h'(x)$
:::

:::{admonition} Teorem 4.3 - Produktregelen
:class: note

La funksjonene $g$ og $h$ er deriverbare i $x$. Da er

$\frac{d}{dx} [g(x) \cdot h(x)]=g'(x)h(x)+g(x)h'(x)$
:::

:::{admonition} Teorem 4.4 - Kvotientregelen
:class: note

La funksjonene $g$ og $h$ er deriverbare i $x$. Da er

$\frac{d}{dx} [\frac{g(x)}{h(x)}]=\frac{g'(x)h(x)-g(x)h'(x)}{h(x)^2}$
:::

:::{admonition} Teorem 4.5 - Kjerneregelen
:class: note

La funksjonene $g$ og $h$ er deriverbare i $x$. Da er

$\frac{d}{dx} [(g \circ h)(x)]=(g(h(x)))'=g(h(x)) \cdot h'(x)$
:::

Derivasjon av funksjoner er mengdetrening. Prøv å forstå alle eksemplene nedenfor.

:::{card} 
**Eksempel 1**
^^^
Deriver $f(x)=7x^4$.

Vi kan faktorisere $7$ ut ved faktorregelen, og bruker potensregelen på $x^4$, slik at

$f'(x)=7 \cdot \frac{d}{dx}\left(x^4\right)=7 \cdot 4x^3=28x^3$
:::

:::{card} 
**Eksempel 2**
^^^
Deriver $f(x)=x^3+\sin{(x)}$.

Ved summeregelen deriverer vi hvert ledd for seg, slik at

$f'(x)=\frac{d}{dx}\left(x^3\right)+\frac{d}{dx}\left(\sin{(x)}\right)=3x^2+\cos{(x)}$
:::

:::{card} 
**Eksempel 3**
^^^
Deriver $f(x)=4x^2+3e^x-5\ln{(x)}$.

Vi bruker summeregelen på leddene og faktorregelen på hver konstant:

$f'(x)=4 \cdot 2x+3 \cdot e^x-5 \cdot \frac{1}{x}=8x+3e^x-\frac{5}{x}$
:::

:::{card} 
**Eksempel 4**
^^^
Deriver $f(x)=x^2\sin{(x)}$.

Dette er et produkt av $x^2$ og $\sin{(x)}$, så vi bruker produktregelen:

$f'(x)=2x\sin{(x)}+x^2\cos{(x)}$
:::

:::{card} 
**Eksempel 5**
^^^
Deriver $f(x)=e^x\cos{(x)}$.

Vi bruker produktregelen, og husker at $e^x$ er sin egen deriverte, slik at

$f'(x)=e^x\cos{(x)}+e^x\cdot(-\sin{(x)})=e^x\left(\cos{(x)}-\sin{(x)}\right)$
:::

:::{card} 
**Eksempel 6**
^^^
Deriver $f(x)=\dfrac{x}{\cos{(x)}}$.

Dette er en kvotient med teller $x$ og nevner $\cos{(x)}$, så vi bruker kvotientregelen, slik at

$f'(x)=\frac{1\cdot\cos{(x)}-x\cdot(-\sin{(x)})}{\cos^2{(x)}}=\frac{\cos{(x)}+x\sin{(x)}}{\cos^2{(x)}}$
:::

:::{card} 
**Eksempel 7**
^^^
Deriver $f(x)=\dfrac{\ln{(x)}}{x^2}$.

Vi bruker kvotientregelen med teller $\ln{(x)}$ og nevner $x^2$, slik at

$f'(x)=\frac{\frac{1}{x}\cdot x^2-\ln{(x)}\cdot 2x}{\left(x^2\right)^2}=\frac{x-2x\ln{(x)}}{x^4}=\frac{1-2\ln{(x)}}{x^3}$
:::

:::{card} 
**Eksempel 8**
^^^
Deriver $f(x)=\sin{(x^2)}$.

La $g(x)=\sin{(x)}$ være den ytre funksjonen og $h(x)=x^2$ være den indre funksjonen. Vi har at

$g'(x)=\cos{(x)}$

$h'(x)=2x$

Ved kjerneregelen får vi derfor

$f'(x)=g'(h(x)) \cdot h'(x)=\cos{(x^2)} \cdot 2x=2x\cos{(x^2)}$
:::

:::{card} 
**Eksempel 9**
^^^
Deriver $f(x)=e^{3x}$.

La $g(x)=e^{x}$ være den ytre funksjonen og $h(x)=3x$ være den indre funksjonen. Vi har at

$g'(x)=e^{x}$

$h'(x)=3$

Ved kjerneregelen får vi derfor

$f'(x)=g'(h(x)) \cdot h'(x)=e^{3x} \cdot 3=3e^{3x}$
:::

:::{card} 
**Eksempel 10**
^^^
Deriver $f(x)=x \cdot \ln{(x^2+1)}$.

Dette er et produkt av $u(x)=x$ og $v(x)=\ln{(x^2+1)}$, så vi bruker produktregelen. Først deriverer vi hver faktor.

For $u(x)=x$ har vi

$u'(x)=1$

Faktoren $v(x)=\ln{(x^2+1)}$ deriverer vi med kjerneregelen. La $g(x)=\ln{(x)}$ være den ytre funksjonen og $h(x)=x^2+1$ være den indre funksjonen. Vi har at

$g'(x)=\frac{1}{x}$

$h'(x)=2x$

slik at

$v'(x)=g'(h(x)) \cdot h'(x)=\frac{1}{x^2+1} \cdot 2x=\frac{2x}{x^2+1}$

Ved produktregelen får vi derfor

$f'(x)=u'(x)v(x)+u(x)v'(x)=1 \cdot \ln{(x^2+1)}+x \cdot \frac{2x}{x^2+1}=\ln{(x^2+1)}+\frac{2x^2}{x^2+1}$
:::

### 4.5 Integrasjon

Vi har til nå sett på derivasjon, som handler om endring: gitt en funksjon $f$, finner vi den momentane endringsraten $f'$. Integrasjon er på mange måter den motsatte operasjonen. I stedet for å spørre "hvor raskt endrer $f$ seg?", spør vi "hvor mye har $f$ samlet seg opp?". Vi skal se at integrasjon svarer på to spørsmål som i utgangspunktet virker helt ulike - arealet under en graf, og det å reversere en derivasjon - og at disse to henger tett sammen.

Vi starter med å reversere derivasjon. Vi har sett at å derivere $F(x)=x^2$ gir $f(x)=2x$. Men kan vi gå fra $f(x)=2x$ tilbake til $F(x)=x^2$? En slik funksjon $F$, som har $f$ som sin deriverte, kaller vi en antiderivert til $f$.

:::{admonition} Def. 4.9 - Antiderivert
:class: important

En funksjon $F$ kalles en **antiderivert** til $f$ dersom

$F'(x)=f(x)$

for alle $x$ i definisjonsmengden til $f$.
:::

Legg merke til at en antiderivert aldri er entydig. Siden den deriverte av en konstant er $0$, vil både $F(x)=x^2$, $F(x)=x^2+1$ og $F(x)=x^2-7$ alle ha $f(x)=2x$ som sin deriverte. Alle antideriverte til en funksjon skiller seg derfor bare med en konstant. Denne konstanten kaller vi integrasjonskonstanten $C$, og den må alltid tas med.

:::{admonition} Def. 4.10 - Det ubestemte integralet
:class: important

Det **ubestemte integralet** til en funksjon $f$ er mengden av alle antideriverte til $f$, skrevet

$\int f(x) \, dx = F(x) + C$

der $F$ er en antiderivert til $f$, og $C$ er en konstant.
:::

Symbolet $\int$ kalles integraltegnet, funksjonen $f(x)$ kalles integranden, og $dx$ forteller oss hvilken variabel vi integrerer med hensyn på. På samme måte som med derivasjon finnes det en tabell over de vanligste integralene. Legg merke til at tabellen bare er derivasjonstabellen "lest baklengs".

| $f(x)$          | $\int f(x)\,dx$                | Betingelser |
|-----------      |-----------                     |-------|
| $a$             | $ax + C$                       | |
| $x^n$           | $\frac{1}{n+1}x^{n+1} + C$      | $n \neq -1$ |
| $\frac{1}{x}$   | $\ln{\|x\|} + C$               | $x \neq 0$ |
| $e^x$           | $e^x + C$                      | |
| $a^x$           | $\frac{a^x}{\ln{(a)}} + C$     | $a>0,\ a\neq 1$ |
| $\sin{(x)}$     | $-\cos{(x)} + C$               | |
| $\cos{(x)}$     | $\sin{(x)} + C$                | |
| $\frac{1}{\cos^2{(x)}}$ | $\tan{(x)} + C$        | |
| $\frac{1}{\sqrt{1-x^2}}$ | $\arcsin{(x)} + C$    | $-1 < x < 1$ |
| $\frac{1}{1+x^2}$ | $\arctan{(x)} + C$           | |

Legg spesielt merke til graden for $x^n$: vi øker eksponenten med én og deler på den nye eksponenten. Dette er det motsatte av potensregelen for derivasjon. Regelen gjelder for alle $n \neq -1$, og tilfellet $n=-1$ (altså $\frac{1}{x}$) er nettopp derfor gitt sin egen rad, med $\ln{|x|}$ som svar.

Akkurat som for derivasjon finnes det regneregler som lar oss integrere summer og konstant-multipler ledd for ledd. Disse følger direkte av de tilsvarende reglene for derivasjon.

:::{admonition} Teorem 4.6 - Linearitet ved integrasjon
:class: note

La $f$ og $g$ være integrerbare funksjoner og la $a$ være et tall. Da er

$\int a \cdot f(x) \, dx = a \int f(x) \, dx$

$\int \left[ f(x) + g(x) \right] dx = \int f(x)\,dx + \int g(x)\,dx$
:::

:::{card} 
**Eksempel**
^^^
Regn ut $\int \left( 3x^2 + 2x \right) dx$.

Ved linearitet integrerer vi hvert ledd for seg, og trekker konstantene utenfor:

$\int \left( 3x^2 + 2x \right) dx = 3\int x^2 \, dx + 2\int x \, dx = 3 \cdot \frac{x^3}{3} + 2 \cdot \frac{x^2}{2} + C = x^3 + x^2 + C$

Merk at vi bare trenger én integrasjonskonstant $C$ til slutt, siden summen av flere vilkårlige konstanter fortsatt bare er en vilkårlig konstant.
:::

Vi bytter nå perspektiv, og spør: hva er arealet mellom grafen til $f$ og $x$-aksen, fra $x=a$ til $x=b$? For en rett linje er dette lett - da får vi et rektangel eller en trekant. Men for en krum graf trenger vi en ny idé: vi deler intervallet opp i mange smale striper, tilnærmer hver stripe med et rektangel, og legger sammen arealene. Jo smalere striper, jo bedre tilnærming.

Grenseverdien av disse rektangelsummene, når antall striper går mot uendelig og bredden mot null, kaller vi **det bestemte integralet**. Vi lar bredden på hver stripe være $\Delta x$ og høyden $f(x_i)$, og summerer.

Når vi tilnærmer arealet med rektangler, må vi bestemme hvor høyt hvert rektangel skal være - altså hvilket punkt i hver stripe vi leser av funksjonsverdien fra. Tre vanlige valg er å bruke venstre endepunkt, høyre endepunkt eller midtpunktet av hver stripe. Uansett hvilket vi velger, kaller vi en slik sum en **Riemann-sum**, etter matematikeren Bernhard Riemann. Velger vi venstre endepunkt for en stigende funksjon, blir rektanglene litt for lave og summen for liten; velger vi høyre endepunkt, blir den for stor. Midtpunktet treffer som regel best. Det avgjørende er at *alle* tre valgene gir samme grenseverdi når stripene blir uendelig smale.

:::{admonition} Def. 4.11 - Det bestemte integralet
:class: important

Det **bestemte integralet** til $f$ fra $a$ til $b$ er grenseverdien av rektangelsummene når antall striper $n$ går mot uendelig:

$\int_a^b f(x) \, dx = \lim_{n \to \infty} \sum_{i=1}^{n} f(x_i)\,\Delta x$

der $\Delta x = \frac{b-a}{n}$. 

Tallene $a$ og $b$ kalles henholdsvis nedre og øvre grense.
:::

Bruk simulasjonen nedenfor til å velge gradvis mindre rektangler for å finne arealet under kurven. Legg merke til hva som skjer når $n \to \infty$.

```python
vis_sim("Simuleringer/riemann_sum.html")
```

Til forskjell fra det ubestemte integralet, som er en funksjon (pluss en konstant), er det bestemte integralet et tall. Det er verdt å merke seg at dette tallet er et areal med fortegn: der grafen ligger under $x$-aksen, teller arealet negativt.

Vi har nå to helt forskjellige begreper som begge kalles "integral": det ubestemte (antiderivasjon) og det bestemte (areal som grense av rektangelsummer). Det store, og på mange måter overraskende, resultatet i kalkulus er at disse to henger sammen. For å regne ut et areal trenger vi ikke summere uendelig mange rektangler - det holder å finne en antiderivert. Dette er **analysens fundamentalteorem**.

:::{admonition} Teorem 4.7 - Analysens fundamentalteorem
:class: note

La $f$ være en kontinuerlig funksjon på $[a,b]$, og la $F$ være en antiderivert til $f$, altså $F'=f$. Da er

$\int_a^b f(x) \, dx = F(b) - F(a)$

Vi bruker gjerne en kort skrivemåte for uttrykket $F(b)-F(a)$, nemlig 

$\Big[F(x)\Big]_a^b$
:::

Fremgangsmåten for å regne ut et bestemt integral blir da: finn en antiderivert $F$, sett inn øvre grense, sett inn nedre grense, og trekk fra. Siden vi til slutt trekker $F(a)$ fra $F(b)$, faller integrasjonskonstanten $C$ alltid bort - derfor tar vi den ikke med i bestemte integraler.

:::{card} 
**Eksempel**
^^^
Regn ut arealet under grafen til $f(x)=x^2$ fra $x=0$ til $x=2$, altså $\int_0^2 x^2 \, dx$.

Vi finner først en antiderivert: $F(x)=\frac{x^3}{3}$. Ved fundamentalteoremet er

$\int_0^2 x^2 \, dx = \Big[\frac{x^3}{3}\Big]_0^2 = \frac{2^3}{3} - \frac{0^3}{3} = \frac{8}{3} - 0 = \frac{8}{3}$

Arealet er altså $\frac{8}{3} \approx 2{,}67$. Dette er nøyaktig den verdien rektangelsummene nærmet seg i simulasjonen ovenfor.
:::

:::{card} 
**Eksempel**
^^^
Regn ut $\int_1^3 (2x+1) \, dx$.

En antiderivert til $2x+1$ er $F(x)=x^2+x$. Da får vi

$\int_1^3 (2x+1) \, dx = \Big[x^2+x\Big]_1^3 = \left(3^2+3\right) - \left(1^2+1\right) = 12 - 2 = 10$
:::

:::{card} 
**Eksempel**
^^^
Regn ut $\int_0^{\pi} \sin{(x)} \, dx$.

En antiderivert til $\sin{(x)}$ er $F(x)=-\cos{(x)}$. Dermed er

$\int_0^{\pi} \sin{(x)} \, dx = \Big[-\cos{(x)}\Big]_0^{\pi} = -\cos{(\pi)} - \left(-\cos{(0)}\right) = -(-1) + 1 = 2$
:::

Så langt har vi kunnet slå opp antideriverte i tabellen, men hva med funksjoner som $x e^x$ eller $2x\cos{(x^2)}$, som ikke står i tabellen? For slike trenger vi teknikker. De to viktigste er substitusjon (som reverserer kjerneregelen) og delvis integrasjon (som reverserer produktregelen).

:::{admonition} Teorem 4.8 - Substitusjon
:class: note

La $f$ og $g$ være slik at $g$ er deriverbar. Da er

$\int f(g(x)) \cdot g'(x) \, dx = \int f(u) \, du$

der vi har satt $u=g(x)$ og $du = g'(x)\,dx$.
:::

Idéen bak substitusjon er å kjenne igjen en "indre funksjon" $g(x)$ og dens deriverte $g'(x)$ i integranden, og så bytte variabel fra $x$ til $u=g(x)$. Da forsvinner den ytre kompleksiteten, og vi sitter igjen med et integral vi kan slå opp.

:::{card} 
**Eksempel**
^^^
Regn ut $\int 2x \cos{(x^2)} \, dx$.

Vi ser at $x^2$ er en indre funksjon, og at dens deriverte $2x$ står som faktor i integranden. Vi setter derfor

$u = x^2 \quad \Rightarrow \quad \frac{du}{dx} = 2x \quad \Rightarrow \quad du = 2x \, dx$

Integralet blir da

$\int 2x \cos{(x^2)} \, dx = \int \cos{(u)} \, du = \sin{(u)} + C = \sin{(x^2)} + C$

Vi bytter altså tilbake til $x$ til slutt. Sjekk gjerne ved å derivere $\sin{(x^2)}$ med kjerneregelen - da får vi $2x\cos{(x^2)}$ tilbake.
:::

:::{admonition} Teorem 4.9 - Delvis integrasjon
:class: note

La $u$ og $v$ være deriverbare funksjoner av $x$. Da er

$\int u \, v' \, dx = u v - \int u' \, v \, dx$
:::

Delvis integrasjon er nyttig når integranden er et produkt av to funksjoner av ulik type - for eksempel et polynom ganget med en eksponential- eller trigonometrisk funksjon. Trikset er å velge den ene faktoren som $u$ (den vi deriverer) og den andre som $v'$ (den vi integrerer), slik at det nye integralet blir enklere enn det opprinnelige.

:::{card} 
**Eksempel**
^^^
Regn ut $\int x e^x \, dx$.

Vi velger $u=x$ (som blir enklere når vi deriverer) og $v'=e^x$ (som er lett å integrere). Da er

$u = x \quad \Rightarrow \quad u' = 1$

$v' = e^x \quad \Rightarrow \quad v = e^x$

Ved delvis integrasjon får vi

$\int x e^x \, dx = u v - \int u' v \, dx = x e^x - \int 1 \cdot e^x \, dx = x e^x - e^x + C = (x-1)e^x + C$

Legg merke til at det nye integralet $\int e^x \, dx$ var enklere enn det vi startet med - det er hele poenget med å velge $u$ og $v'$ smart.
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: note

Integrasjon er helt sentralt i termodynamikk. Når en gass utvider seg og utfører arbeid, er det totale arbeidet gitt ved integralet av trykket over volumendringen,

$W = -\int_{V_1}^{V_2} P \, dV$

altså arealet under kurven i et trykk-volum-diagram. På samme måte finner vi endring i entropi ved å integrere $\frac{dq_{rev}}{T}$, og sammenhengen mellom varmekapasitet og entalpi ved $\Delta H = \int_{T_1}^{T_2} C_p \, dT$. I alle disse tilfellene er idéen den samme som i Def. 4.11: vi summerer opp uendelig mange små bidrag til noe kontinuerlig endrer seg.
:::

Med integrasjon på plass har vi de to grunnpilarene i kalkulus - derivasjon og integrasjon - og fundamentalteoremet som binder dem sammen. I de neste delkapitlene utvider vi til funksjoner av flere variabler.

:::{admonition} Underveisoppgave 4.5
:class: tip

a) Regn ut det ubestemte integralet $\int \left(4x^3-2x\right)dx$.

b) Regn ut det bestemte integralet $\int_1^2 6x^2 \, dx$ med fundamentalteoremet.

c) Regn ut $\int_0^{\pi/2} \cos{(x)} \, dx$.

d) Regn ut $\int 3x^2 \sin{(x^3)} \, dx$.
:::

:::{admonition} Løsning 4.5
:class: tip, dropdown

a) Vi integrerer ledd for ledd: $\int \left(4x^3-2x\right)dx = 4 \cdot \frac{x^4}{4} - 2 \cdot \frac{x^2}{2} + C = x^4 - x^2 + C$.

b) En antiderivert til $6x^2$ er $F(x)=2x^3$. Da er

$\int_1^2 6x^2 \, dx = \Big[2x^3\Big]_1^2 = 2\cdot 8 - 2 \cdot 1 = 16-2 = 14$

c) En antiderivert til $\cos{(x)}$ er $F(x)=\sin{(x)}$. Da er

$\int_0^{\pi/2} \cos{(x)} \, dx = \Big[\sin{(x)}\Big]_0^{\pi/2} = \sin{\left(\tfrac{\pi}{2}\right)} - \sin{(0)} = 1 - 0 = 1$

d) Vi ser at $x^3$ er en indre funksjon med deriverte $3x^2$, som står som faktor. Vi setter $u=x^3$, slik at $du=3x^2\,dx$. Da blir

$\int 3x^2 \sin{(x^3)} \, dx = \int \sin{(u)} \, du = -\cos{(u)} + C = -\cos{(x^3)} + C$
:::

### 4.6 Flervariable funksjoner

Så langt har alle funksjonene våre hatt én input og én output - vi puttet inn ett tall $x$ og fikk ut ett tall $f(x)$. Men det er mange eksempler hvor en størrelse avhenger av flere enn én annen størrelse. I kjemien avhenger de fleste størrelser av *flere* ting samtidig. Trykket i en gass avhenger både av volumet og temperaturen. Løseligheten til et stoff avhenger av både temperatur og trykk. Reaksjonshastigheten avhenger av konsentrasjonen til flere reaktanter på én gang. For å beskrive slike sammenhenger trenger vi funksjoner som tar inn flere variabler. Heldigvis er ikke ideen så forskjellig fra det vi allerede kan - vi utvider bare antallet input.

Vi starter med å utvide selve funksjonsbegrepet. En funksjon av flere variabler fungerer akkurat som en funksjon av én variabel, bortsett fra at vi mater inn flere tall for å få ut ett tall.

:::{admonition} Def. 4.12 - Funksjon av flere variabler
:class: important

En **funksjon av to variabler** $f$ er en regel som tildeler hvert tallpar $(x_1,x_2)$ nøyaktig ett tall $f(x_1,x_2)$. Vi skriver

$f:\mathbb{R}^2 \to \mathbb{R}$

Mer generelt tildeler en **funksjon av $n$ variabler** hvert punkt $(x_1, x_2, \ldots, x_n)$ nøyaktig ett tall $f(x_1, x_2, \ldots, x_n)$, og vi skriver 

$f:\mathbb{R}^n \to \mathbb{R}$.
:::

Legg merke til at outputen fortsatt er ett enkelt tall - det er bare inputen som består av flere tall. Kravet fra Def. 4.1, at hver input gir nøyaktig én output, er akkurat det samme. Vi setter inn verdier på nøyaktig samme måte som før, bare med flere tall om gangen.

:::{card} 
**Eksempel**
^^^
La $f(x,y)=x^2+3y$. Regn ut $f(2,1)$ og $f(0,-4)$.

Vi setter inn verdiene for $x$ og $y$, i den rekkefølgen de står.

For $f(2,1)$ setter vi $x=2$ og $y=1$:

$f(2,1)=2^2+3\cdot 1=4+3=7$

For $f(0,-4)$ setter vi $x=0$ og $y=-4$:

$f(0,-4)=0^2+3\cdot(-4)=0-12=-12$

Merk at rekkefølgen betyr noe: $f(2,1)$ er ikke det samme som $f(1,2)$.
:::

Definisjonsmengden til en flervariabel funksjon er mengden av alle tallpar $(x,y)$ vi har lov til å sette inn - altså alle par der regneuttrykket gir mening. Prinsippet er det samme som i én variabel: vi må unngå å dele på null, ta kvadratrot av negative tall, eller ta logaritmen av noe som ikke er positivt.

:::{card} 
**Eksempel**
^^^
Hva er definisjonsmengden til $f(x,y)=\dfrac{1}{x-y}$?

Uttrykket gir mening for alle tallpar $(x,y)$ så lenge vi ikke deler på null. Nevneren blir null når $x-y=0$, altså når $x=y$.

Definisjonsmengden er derfor alle tallpar $(x,y)$ der $x \neq y$. Geometrisk betyr det hele planet unntatt den rette linjen $y=x$.
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: note

Idealgassloven er et av de mest kjente eksemplene på en flervariabel funksjon i kjemien. Løser vi $PV=nRT$ for trykket, får vi

$P(V,T)=\frac{nRT}{V}$

Her er trykket $P$ en funksjon av *to* variabler: volumet $V$ og temperaturen $T$ (mens $n$ og $R$ er konstanter). Definisjonsmengden er fysisk begrenset: volumet må være positivt ($V>0$, ellers deler vi på null), og temperaturen målt i kelvin kan ikke være negativ ($T \geq 0$). Setter vi inn et bestemt volum og en bestemt temperatur, får vi ut nøyaktig ett trykk - helt i tråd med Def. 4.12.
:::

En funksjon $f(x)$ av én variabel tegner vi som en kurve i planet: for hver $x$ langs den vannrette aksen markerer vi høyden $f(x)$. En funksjon $f(x,y)$ av to variabler trenger derimot to akser bare for inputen - én for $x$ og én for $y$ - og en tredje akse for outputen $f(x,y)$. Grafen blir derfor ikke en kurve, men en **flate** i det tredimensjonale rommet. Du kan tenke på den som et landskap: for hvert punkt $(x,y)$ på "bakken" forteller flaten hvor høyt over bakken funksjonen ligger.

```python
vis_sim("Simuleringer/flervariabel_flate.html")
```

Tredimensjonale flater er fine å se på, men vanskelige å tegne og lese av nøyaktig på et flatt ark. Derfor bruker kjemikere og matematikere ofte et triks som du kjenner igjen fra turkart: **nivåkurver**. I stedet for å tegne selve landskapet, tegner vi kurvene som forbinder alle punkter med samme høyde. På et turkart er dette høydekotene; for en funksjon $f(x,y)$ er en nivåkurve alle punkter $(x,y)$ der funksjonen har én bestemt verdi.

:::{card} 
**Eksempel**
^^^
Beskriv nivåkurvene til $f(x,y)=x^2+y^2$.

En nivåkurve består av alle punkter $(x,y)$ der funksjonen har én fast verdi $c$. Vi setter derfor $f(x,y)=c$:

$x^2+y^2=c$

For $c>0$ kjenner vi igjen dette som ligningen for en sirkel med sentrum i origo og radius $\sqrt{c}$. Nivåkurvene er altså konsentriske sirkler: jo større verdi $c$, jo større sirkel. For $c=0$ krymper sirkelen til det enkelte punktet $(0,0)$, og for $c<0$ finnes det ingen punkter (en sum av kvadrater kan ikke bli negativ).
:::

Nå som vi kan beskrive funksjoner av flere variabler, dukker et naturlig spørsmål opp: hvordan deriverer vi dem? Hvis trykket avhenger av både volum og temperatur, hva mener vi da med "endringsraten"? Svaret er at vi må derivere med hensyn på én variabel om gangen, og holde de andre konstante. Dette leder oss til partiellderiverte i neste delkapittel.

:::{admonition} Underveisoppgave 4.6
:class: tip

La $f(x,y)=2x+y^2$.

a) Regn ut $f(3,2)$ og $f(-1,4)$.

b) Er $f(1,2)$ det samme som $f(2,1)$? Regn ut begge.

c) Hva er definisjonsmengden til $g(x,y)=\dfrac{1}{\sqrt{x}}+y$?
:::

:::{admonition} Løsning 4.6
:class: tip, dropdown

a) Vi setter inn verdiene i rekkefølge.

$f(3,2)=2\cdot 3+2^2=6+4=10$

$f(-1,4)=2\cdot(-1)+4^2=-2+16=14$

b) Vi regner ut begge:

$f(1,2)=2\cdot 1+2^2=2+4=6$

$f(2,1)=2\cdot 2+1^2=4+1=5$

Siden $6 \neq 5$, er $f(1,2)$ **ikke** det samme som $f(2,1)$. Rekkefølgen på variablene betyr noe.

c) For at $g(x,y)=\dfrac{1}{\sqrt{x}}+y$ skal gi mening, må uttrykket under kvadratroten være positivt (vi kan verken ta rot av et negativt tall eller dele på null, og $\sqrt{0}=0$ gir null i nevneren). Vi trenger derfor $x>0$, mens $y$ kan være hvilket som helst reelt tall. Definisjonsmengden er alle tallpar $(x,y)$ med $x>0$.
:::

### 4.7 Partiellderiverte

I forrige delkapittel så vi at trykket i en gass kan avhenge av både volum og temperatur gjennom $P(V,T)$. Men da dukker et problem opp: hva mener vi egentlig med "endringsraten" til en funksjon av flere variabler? For en funksjon $f(x)$ av én variabel var svaret entydig - den deriverte $f'(x)$ målte hvor raskt $f$ endret seg. For $f(x,y)$ er det ikke like klart: endrer vi $x$, eller $y$, eller begge samtidig? Løsningen er enkel og elegant: vi ser på én variabel om gangen, og holder de andre helt i ro. Dette kaller vi å **derivere partielt**.

Ideen er å velge én variabel å derivere med hensyn på, og late som om alle de andre variablene bare er konstanter. Deriverer vi med hensyn på $x$, behandler vi $y$ som et fast tall - og bruker ellers akkurat de samme derivasjonsreglene som i delkapittel 4.4.

:::{admonition} Def. 4.13 - Partiellderivert
:class: important

La $f(x,y)$ være en funksjon av to variabler. Den **partiellderiverte av $f$ med hensyn på $x$** er

$\frac{\partial f}{\partial x} = \lim_{h \to 0} \frac{f(x+h,\,y)-f(x,\,y)}{h}$

der vi holder $y$ konstant. Tilsvarende er den **partiellderiverte med hensyn på $y$**

$\frac{\partial f}{\partial y} = \lim_{h \to 0} \frac{f(x,\,y+h)-f(x,\,y)}{h}$

der vi holder $x$ konstant.
:::

Legg merke til det krøllete symbolet $\partial$ ("del" eller "partiell-d"). Vi bruker det i stedet for det vanlige $d$ nettopp for å understreke at funksjonen har flere variabler, og at vi bare deriverer med hensyn på én av dem. Sammenlign gjerne med definisjonen av den vanlige deriverte i Def. 4.8 - den eneste forskjellen er at den andre variabelen står fast som en tilskuer. Vi bruker også skrivemåtene $f_x$ og $f_y$ som kortform for $\frac{\partial f}{\partial x}$ og $\frac{\partial f}{\partial y}$.

:::{card} 
**Eksempel**
^^^
Finn begge de partiellderiverte til $f(x,y)=x^2 y + 3x y^3$.

**Med hensyn på $x$** behandler vi $y$ som en konstant. Da er $x^2 y$ et konstant-tall ($y$) ganget med $x^2$, og $3xy^3$ er en konstant ($3y^3$) ganget med $x$:

$\frac{\partial f}{\partial x} = 2x y + 3y^3$

**Med hensyn på $y$** behandler vi $x$ som en konstant. Nå er $x^2 y$ en konstant ($x^2$) ganget med $y$, og $3xy^3$ er en konstant ($3x$) ganget med $y^3$:

$\frac{\partial f}{\partial y} = x^2 + 9x y^2$

Legg merke til at leddet $x^2 y$ ga ulikt bidrag i de to tilfellene, alt etter hvilken variabel vi så på som konstant.
:::

:::{card} 
**Eksempel**
^^^
Finn de partiellderiverte til $g(x,y)=\sin{(xy)}$.

**Med hensyn på $x$** holder vi $y$ konstant. Den indre funksjonen er $xy$, som deriveres til $y$ (siden $y$ er konstant). Ved kjerneregelen:

$\frac{\partial g}{\partial x} = \cos{(xy)} \cdot y = y\cos{(xy)}$

**Med hensyn på $y$** holder vi $x$ konstant. Nå deriveres den indre funksjonen $xy$ til $x$:

$\frac{\partial g}{\partial y} = \cos{(xy)} \cdot x = x\cos{(xy)}$
:::

Akkurat som vi kunne derivere flere ganger i én variabel (Def. 4.8 om høyere-ordens deriverte), kan vi derivere partielt flere ganger. Nå får vi flere muligheter: vi kan derivere to ganger med hensyn på $x$, to ganger med hensyn på $y$, eller én gang med hver. De to siste kalles **blandede partiellderiverte**.

:::{admonition} Def. 4.14 - Andreordens partiellderiverte
:class: important

De **andreordens partiellderiverte** til $f(x,y)$ er

$\frac{\partial^2 f}{\partial x^2} = \frac{\partial}{\partial x}\left(\frac{\partial f}{\partial x}\right), \qquad \frac{\partial^2 f}{\partial y^2} = \frac{\partial}{\partial y}\left(\frac{\partial f}{\partial y}\right)$

De **blandede partiellderiverte** er

$\frac{\partial^2 f}{\partial x\, \partial y} = \frac{\partial}{\partial x}\left(\frac{\partial f}{\partial y}\right), \qquad \frac{\partial^2 f}{\partial y\, \partial x} = \frac{\partial}{\partial y}\left(\frac{\partial f}{\partial x}\right)$
:::

:::{card} 
**Eksempel**
^^^
For $f(x,y)=x^2 y + 3x y^3$ fant vi $\frac{\partial f}{\partial x}=2xy+3y^3$ og $\frac{\partial f}{\partial y}=x^2+9xy^2$. Finn de to blandede partiellderiverte.

For $\frac{\partial^2 f}{\partial x\, \partial y}$ deriverer vi $\frac{\partial f}{\partial y}=x^2+9xy^2$ med hensyn på $x$ (hold $y$ konstant):

$\frac{\partial^2 f}{\partial x\, \partial y} = 2x + 9y^2$

For $\frac{\partial^2 f}{\partial y\, \partial x}$ deriverer vi $\frac{\partial f}{\partial x}=2xy+3y^3$ med hensyn på $y$ (hold $x$ konstant):

$\frac{\partial^2 f}{\partial y\, \partial x} = 2x + 9y^2$

Merk at de to blandede ble like! Dette er ingen tilfeldighet.
:::

At de blandede partiellderiverte ble like i eksempelet over, er et generelt resultat: for de aller fleste funksjoner vi møter i praksis, spiller rekkefølgen vi deriverer i ingen rolle. Dette kalles gjerne Clairauts teorem, og gjelder så lenge de blandede partiellderiverte er kontinuerlige. For oss betyr det en grei sjekk: regner du ut begge blandede og får ulikt svar, har du sannsynligvis gjort en regnefeil.

De to partiellderiverte forteller hver for seg hvor bratt flaten heller i $x$-retningen og i $y$-retningen. Ofte er det nyttig å samle dem i ett objekt. Vi setter dem sammen til en vektor, som vi kaller gradienten.

:::{admonition} Def. 4.15 - Gradienten
:class: important

**Gradienten** til en funksjon $f(x,y)$ er vektoren av de partiellderiverte,

$\nabla f = \left( \frac{\partial f}{\partial x}, \; \frac{\partial f}{\partial y} \right)$

Symbolet $\nabla$ leses "nabla". For en funksjon av $n$ variabler har gradienten tilsvarende én komponent per variabel.
:::

Gradienten har en vakker geometrisk tolkning: i et gitt punkt peker $\nabla f$ i den retningen der funksjonen vokser raskest, og lengden av vektoren forteller hvor bratt den vokser. Står du i fjellandskapet fra introduksjonsspørsmålene, peker gradienten rett oppover den bratteste stigningen. Et beslektet poeng, som er nyttig i kjemien: gradienten står alltid vinkelrett på nivåkurvene fra delkapittel 4.6.

:::{card} 
**Eksempel**
^^^
Finn gradienten til $f(x,y)=x^2+y^2$, og regn den ut i punktet $(1,2)$.

Vi finner først de to partiellderiverte:

$\frac{\partial f}{\partial x} = 2x, \qquad \frac{\partial f}{\partial y} = 2y$

Gradienten er dermed

$\nabla f = (2x, \; 2y)$

I punktet $(1,2)$ blir den

$\nabla f(1,2) = (2\cdot 1, \; 2\cdot 2) = (2, \; 4)$

Denne vektoren peker i retningen der $f$ vokser raskest fra punktet $(1,2)$ - altså rett bort fra origo, som stemmer med at nivåkurvene her er sirkler om origo.
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: note

Partiellderiverte er selve språket i termodynamikken, men kjemikere skriver dem litt annerledes enn matematikere. Siden en størrelse kan avhenge av mange variabler, er det avgjørende å presisere *hvilke* variabler som holdes konstante. Derfor skriver man den konstante variabelen som en senket indeks utenfor en parentes.

For idealgassen $P(V,T)=\dfrac{nRT}{V}$ blir de to partiellderiverte

$\left(\frac{\partial P}{\partial T}\right)_V = \frac{nR}{V} \qquad \text{og} \qquad \left(\frac{\partial P}{\partial V}\right)_T = -\frac{nRT}{V^2}$

Notasjonen $\left(\frac{\partial P}{\partial T}\right)_V$ leses "den partiellderiverte av $P$ med hensyn på $T$, med $V$ holdt konstant". Dette er nøyaktig samme operasjon som i Def. 4.13 - indeksen $V$ er bare en eksplisitt påminnelse om hva som står fast. Slike uttrykk dukker opp overalt i termodynamikken, for eksempel i sammenhengen mellom varmekapasitet og indre energi, $C_V=\left(\frac{\partial U}{\partial T}\right)_V$.
:::

Med partiellderiverte og gradienten på plass har vi verktøyene for å studere hvordan flervariable funksjoner endrer seg. Et naturlig neste spørsmål er hvordan vi finner topp- og bunnpunkter for slike funksjoner - særlig når de i tillegg er underlagt en betingelse, slik som å minimere energi under bevaring av masse. Det leder oss til Lagranges multiplikatormetode i neste delkapittel.

### 4.8 Dobbel-og trippelintegraler

I delkapittel 4.5 så vi at det bestemte integralet $\int_a^b f(x)\,dx$ måler arealet under grafen til en funksjon av én variabel. Men nå kan vi funksjoner av flere variabler (4.6), og et naturlig spørsmål melder seg: hva blir integralet av $f(x,y)$? Der enkeltintegralet summerte opp arealet av uendelig mange tynne striper, skal vi nå summere opp volumet av uendelig mange tynne søyler - og resultatet blir volumet under en flate. Utvider vi til tre variabler, kan vi regne ut ting som massen eller ladningen fordelt i et romområde. For kjemistudenter er dette selve verktøyet bak normalisering av orbitaler, som vi kommer til mot slutten.

Vi starter med to variabler. Akkurat som vi delte intervallet $[a,b]$ i tynne striper i 4.5, deler vi nå et område i $xy$-planet inn i et rutenett av små rektangler, hvert med areal $\Delta A = \Delta x\,\Delta y$. Over hver rute reiser vi en søyle med høyde $f(x,y)$, altså volum $f(x,y)\,\Delta A$. Legger vi sammen alle søylene og lar rutene bli uendelig små, får vi dobbeltintegralet.

:::{admonition} Def. 4.16 - Dobbeltintegralet
:class: important

**Dobbeltintegralet** av $f(x,y)$ over et område $D$ i $xy$-planet er grenseverdien av søylesummene når rutene blir uendelig små:

$\iint_D f(x,y)\,dA = \lim_{\Delta A \to 0} \sum_i f(x_i,y_i)\,\Delta A$

Når $f(x,y)\geq 0$, er dobbeltintegralet **volumet** mellom flaten $z=f(x,y)$ og området $D$ i $xy$-planet.
:::

Definisjonen er fin å forstå, men vi vil ikke summere uendelig mange søyler for hånd. Heldigvis finnes det en enkel oppskrift, helt analog til fundamentalteoremet: vi integrerer én variabel om gangen. Dette kalles et **iterert integral**. Vi holder den ene variabelen konstant (akkurat som ved partiellderivasjon i 4.7), integrerer med hensyn på den andre, og integrerer så resultatet med hensyn på den første.

:::{admonition} Teorem 4.10 - Fubinis teorem
:class: note

La $f(x,y)$ være kontinuerlig på rektangelet $D=[a,b]\times[c,d]$. Da kan dobbeltintegralet regnes ut som et iterert integral, og rekkefølgen spiller ingen rolle:

$\iint_D f(x,y)\,dA = \int_a^b \left( \int_c^d f(x,y)\,dy \right) dx = \int_c^d \left( \int_a^b f(x,y)\,dx \right) dy$
:::

:::{card} 
**Eksempel**
^^^
Regn ut $\iint_D xy \, dA$ over rektangelet $D=[0,2]\times[0,1]$.

Vi regner det som et iterert integral, og starter innenfra med hensyn på $x$ (mens $y$ holdes konstant):

$\int_0^2 xy \, dx = y\int_0^2 x \, dx = y\Big[\frac{x^2}{2}\Big]_0^2 = y \cdot 2 = 2y$

Så integrerer vi resultatet med hensyn på $y$:

$\int_0^1 2y \, dy = \Big[y^2\Big]_0^1 = 1$

Altså er $\iint_D xy \, dA = 1$. Ifølge Fubini hadde vi fått samme svar om vi integrerte i motsatt rekkefølge - prøv gjerne selv.
:::

Steget videre til tre variabler er nå kort. I stedet for å dele opp et flatt område i ruter, deler vi opp et romområde $E$ i små bokser med volum $\Delta V = \Delta x\,\Delta y\,\Delta z$. Vi summerer $f(x,y,z)\,\Delta V$ over alle boksene og lar dem bli uendelig små.

:::{admonition} Def. 4.17 - Trippelintegralet
:class: important

**Trippelintegralet** av $f(x,y,z)$ over et romområde $E$ er

$\iiint_E f(x,y,z)\,dV = \lim_{\Delta V \to 0} \sum_i f(x_i,y_i,z_i)\,\Delta V$

Setter vi $f=1$, gir integralet **volumet** av $E$. Er $f$ en tetthet (masse per volum), gir integralet den totale **massen**.
:::

Fubinis teorem gjelder også her: vi integrerer én variabel om gangen, tre ganger. Over en boks $[a_1,b_1]\times[a_2,b_2]\times[a_3,b_3]$ blir det tre nøstede integraler. Men mange av de viktigste anvendelsene i kjemien handler ikke om bokser - de handler om *kuler*, som når vi ser på et atom. Da lønner det seg å bytte koordinatsystem.

Et hydrogenatom er kulesymmetrisk, ikke boksformet. Å beskrive det med $x$, $y$ og $z$ blir tungvint. I stedet bruker vi **sfæriske koordinater**, der vi angir et punkt ved hvor langt det er fra origo og i hvilken retning: avstanden $r$ fra origo, vinkelen $\theta$ ned fra $z$-aksen (polarvinkelen), og vinkelen $\phi$ rundt $z$-aksen (asimutvinkelen).

:::{admonition} Def. 4.18 - Sfæriske koordinater
:class: important

Et punkt $(x,y,z)$ kan beskrives ved sfæriske koordinater $(r,\theta,\phi)$ gjennom

$x = r\sin{\theta}\cos{\phi}, \qquad y = r\sin{\theta}\sin{\phi}, \qquad z = r\cos{\theta}$

der $r \geq 0$ er avstanden fra origo, $0 \leq \theta \leq \pi$ er vinkelen fra den positive $z$-aksen, og $0 \leq \phi < 2\pi$ er vinkelen i $xy$-planet.
:::

Når vi bytter koordinater, kan vi ikke bare bytte ut $dV = dx\,dy\,dz$ med $dr\,d\theta\,d\phi$ - det ville gitt feil svar. Grunnen er at en liten "boks" i sfæriske koordinater ikke har sidene $dr$, $d\theta$ og $d\phi$ i lengde. Bare $dr$ er en ekte lengde. De to vinklene må ganges opp med en radius for å bli lengder: en liten endring $d\theta$ svarer til en bue av lengde $r\,d\theta$, og en liten endring $d\phi$ svarer til en bue av lengde $r\sin{\theta}\,d\phi$ (faktoren $\sin\theta$ kommer av at sirklene rundt $z$-aksen er mindre nær polene enn ved ekvator). Volumet av den lille boksen blir derfor produktet av de tre lengdene.

:::{admonition} Def. 4.19 - Volumelementet i sfæriske koordinater
:class: important

I sfæriske koordinater er volumelementet

$dV = r^2 \sin{\theta} \, dr \, d\theta \, d\phi$

Faktoren $r^2\sin{\theta}$ kalles Jacobideterminanten for overgangen, og retter opp for at vinkelkoordinatene ikke er lengder.
:::

:::{card} 
**Eksempel**
^^^
Bruk sfæriske koordinater til å regne ut volumet av en kule med radius $R$.

Volumet er trippelintegralet av $f=1$ over hele kula. I sfæriske koordinater dekker vi kula ved å la $r$ gå fra $0$ til $R$, $\theta$ fra $0$ til $\pi$, og $\phi$ fra $0$ til $2\pi$. Vi bruker $dV=r^2\sin{\theta}\,dr\,d\theta\,d\phi$:

$V = \int_0^{2\pi}\int_0^{\pi}\int_0^{R} r^2\sin{\theta}\,dr\,d\theta\,d\phi$

De tre integralene kan skilles fra hverandre siden faktorene er uavhengige:

$V = \left(\int_0^{R} r^2\,dr\right)\left(\int_0^{\pi}\sin{\theta}\,d\theta\right)\left(\int_0^{2\pi}d\phi\right)$

Vi regner ut hver for seg:

$\int_0^{R} r^2\,dr = \frac{R^3}{3}, \qquad \int_0^{\pi}\sin{\theta}\,d\theta = \Big[-\cos{\theta}\Big]_0^{\pi} = 2, \qquad \int_0^{2\pi}d\phi = 2\pi$

Til sammen: $V = \dfrac{R^3}{3}\cdot 2 \cdot 2\pi = \dfrac{4}{3}\pi R^3$, som er den velkjente formelen for kulevolumet.
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: note

Her kommer selve grunnen til at kjemistudenter trenger trippelintegraler i sfæriske koordinater: **normalisering av orbitaler**. En orbital $\psi$ er en bølgefunksjon, og $|\psi|^2$ er sannsynlighetstettheten for å finne elektronet. Siden elektronet må være *et eller annet sted*, må sannsynligheten summert over hele rommet bli nøyaktig $1$:

$\iiint_{\text{hele rommet}} |\psi|^2 \, dV = 1$

Ta hydrogenets 1s-orbital, som er kulesymmetrisk og bare avhenger av $r$:

$\psi_{1s} = \frac{1}{\sqrt{\pi a_0^3}}\, e^{-r/a_0}$

der $a_0$ er Bohr-radien. Vi sjekker normaliseringen i sfæriske koordinater. Siden $|\psi|^2$ ikke avhenger av vinklene, skiller integralene seg pent fra hverandre:

$\iiint |\psi_{1s}|^2 \, dV = \frac{1}{\pi a_0^3}\left(\int_0^{\infty} e^{-2r/a_0}\,r^2\,dr\right)\left(\int_0^{\pi}\sin{\theta}\,d\theta\right)\left(\int_0^{2\pi}d\phi\right)$

Vinkelintegralene kjenner vi igjen fra kule-eksempelet: de gir $2$ og $2\pi$. Radialintegralet er et standardintegral (som kan løses med delvis integrasjon fra 4.5, brukt to ganger):

$\int_0^{\infty} r^2 e^{-2r/a_0}\,dr = \frac{a_0^3}{4}$

Setter vi alt sammen:

$\frac{1}{\pi a_0^3}\cdot \frac{a_0^3}{4}\cdot 2 \cdot 2\pi = \frac{1}{\pi a_0^3}\cdot \pi a_0^3 = 1$

Integralet blir nøyaktig $1$ - orbitalen er riktig normalisert. Legg merke til hvor avgjørende faktoren $r^2$ fra volumelementet var: uten den ville integralet ikke gått opp. Nettopp derfor er $r^2\sin\theta$ ikke en teknisk detalj, men helt sentralt i kvantekjemien.
:::

Med dobbel- og trippelintegraler, og sfæriske koordinater, har vi fått verktøyene til å integrere over flater og romområder - og til å normalisere orbitaler, som er uunnværlig i kvantekjemien. Dette runder av kalkulusdelen: fra funksjoner og grenser, via derivasjon og integrasjon i én variabel, til flervariabel derivasjon og integrasjon.
