### 1.1 Å tolke kompliserte formler

I fysikalsk kjemi presenteres flere tunge formler som kan være vanskelige å fordøye. Vi skal derfor se et eksempel på hvordan du kan bryte ned og jobbe med formler, og vi tar utgangspunkt i **Maxwell-Boltzmann-fordelingen**.

$f(v)=4\pi \left( \frac{m}{2\pi k_B T} \right)^{\frac{3}{2}} v^2 e^{-\frac{mv^2}{2k_B T}}$

**1. Hvilke variable ser vi på?**

Det første vi kan gjøre er å se på *hva* funksjonen beskriver, og hvilke enheter det er snakk om.


| Variabel | Enhet | Forklaring |
| -------- | ----- | ---------- |
| $m$ | $[kg]$ | Massen til ett molekyl | 
| $v$ | $[m/s]$ | Fart |
| $k_B$ | $[J/K]=[kg \cdot m^2/s^2 \cdot K]$ | Boltzmann-konstanten |
| $T$ | $[K]$ | Temperatur |
| $f(v)$ | $[s/m]$ | Sansynnlighetstettheten for farten $v$ |

Nå vet vi at vi ser på sannsynligheten $f(v)$ som funksjon av fart $v$, der også masse $m$, Boltzmannkonstanten $k_B$ og temperaturen $T$ også er i funksjonen. Allerede her kan vi notere oss at sannsynligheten bare er en funksjon av $v$, det betyr at en slik funksjon holder de andre variablene konstante.

**2. Er det noen antakelser for formelen?**

Ofte er formlene i termodynamikk utledet fra andre og har forskjellige antakelser i grunn. Maxwell-Boltzmann-fordelingen har følgende antakelser i grunn:

1) Ideell gass - punktpartikler uten intermolekylære krefter mellom seg, med elastiske kollisjoner
2) Termisk likevekt ved temperatur $T$
3) Klassisk tilnærming - farten kan ta hvilken som helst verdi og molekylene har veldefinert posisjon og fart
4) Tilfeldig bevegelse - fordelingen av fart er lik i alle retninger i rommet, det er ingen retning som er favorisert

Nå har vi allerede styr på masse! Vi vet at Maxwell-Boltzmann-fordelingen beskriver sannsynlighetstettheten $f(v)$ som funksjon av farten $v$ til en ideell gass hvor molekylene har masse $m$, og ved termisk likevekt ved temperaturen $T$. Molekylene beveger seg tilfeldig, ingen retning er favorisert over en annen, og gassen vi beskriver er klassisk (heller enn kvantemekanisk).

**3. Hva er sammenhengen mellom formel og graf?**

Det kan være vanskelig å se hva som skjer når vi endrer de ulike variablene kun basert på formelen, men det er mye lettere visuelt. Simulasjonen nedenfor lar oss endre $T$ og $m$ og se hvordan fordelingen endrer seg.

**4. Hvordan henger formel og kjemi sammen?**

Dette er den viktigste delen. Vi har observert hvordan formelen endrer seg med de ulike variablene, men hva sier dette om kjemien?

**Temperatur.** Når du skrur opp $T$, flyttes kurven mot høyre og blir bredere og lavere. Molekylene beveger seg altså raskere, men fartene sprer seg også mer. Dette er grunnen til at reaksjoner går fortere ved høyere temperatur: en reaksjon krever som regel at molekylene har nok energi til å overkomme en aktiveringsbarriere, og andelen molekyler i den høye halen vokser kraftig med temperaturen. Legg merke til at det ikke er *gjennomsnittsfarten* som først og fremst avgjør dette - det er hvor mange som befinner seg helt ute i høyre hale.

**Masse.** Tyngre molekyler gir en smal, høy kurve nær lave farter, mens lette molekyler får en bred fordeling med langt høyere farter. Ved samme temperatur har alle molekyler like mye kinetisk energi i gjennomsnitt, men siden $E_k=\frac{1}{2}mv^2$, må de tunge kompensere med lavere fart. Dette forklarer hvorfor lette gasser diffunderer raskere enn tunge, og hvorfor helium lekker ut av en ballong mens nitrogen blir værende.

**Arealet er alltid $1$.** Uansett hvordan du stiller inn $T$ og $m$, forblir det totale arealet under kurven det samme. Det er normeringskravet: hvert molekyl må ha *en eller annen* fart. Blir kurven bredere, må den nødvendigvis bli lavere - dette er den samme sammenhengen vi så for normalfordelingen.

**Formen er alltid skjev.** Kurven starter i null, stiger til en topp, og faller av med en lang hale mot høyre. Grunnen ligger i de to faktorene i formelen: $v^2$ trekker kurven opp fra null, mens Boltzmann-faktoren $e^{-mv^2/2k_BT}$ trekker den ned igjen for høye farter. Produktet av en voksende og en avtagende faktor gir toppen. Uten $v^2$-leddet ville flest molekyler hatt fart null, noe som åpenbart er feil - i en gass i likevekt står ingen molekyler stille.

**Den lange halen betyr noe.** At kurven aldri når helt ned til null mot høyre, forteller at det alltid finnes noen få molekyler med svært høy fart. Det er disse som fordamper fra en væskeoverflate, som rømmer fra en planets atmosfære, og som klarer å reagere selv når temperaturen er lav. Selv om de utgjør en forsvinnende liten andel, er det ofte de som styrer hva som faktisk skjer.

Dette var et eksempel på hvordan man kan bryte ned en formel for å gjøre den lettere og mer intuitiv å forstå. Til det trenger vi både matte, kjemi og fysikk.

#### 1.2.1 Mengder, delmengder og elementer

:::{admonition} Def. 1.1 - Mengde
:class: warning

En **mengde** $A$ er en veldefinert samling av objekter som kalles **elementer**. Definisjonen av en mengde skrives ofte som et sett av klammeparenteser.

Den **tomme mengden** er en mengde uten elementer og skrives som $\emptyset$ eller $\{\}$.
:::

Merk at en mengde må være *veldefinert*, noe som betyr at det ikke kan være noe tvil om hvilke elementer som er med i mengden. Mengden av fine biler eller mengden av tall er ikke veldefinerte fordi det ikke er tydelig hvilke elementer som er med i mengdene. Mengden av alle partall eller mengden av alle røde biler er derimot mengder, fordi det er tydelig definert hva slags elementer som er i mengdene.

:::{admonition} Def. 1.2 - Delmengde
:class: warning

Hvis $A$ og $B$ er mengder, og ethvert element i $A$ også er et element i $B$, er $A$ en delmengde av $B$. Vi skriver

$A \subset B$
:::

#### 1.2.2 Mengdenotasjon

La $x$ være et element og la $A, B$ og $C$ være to mengder. Følgende notasjoner er verdt å vite:


| Matematisk notasjon | Forklaring | 
| ------------------- | ---------- |
| $x \in A$ | $x$ er et element i $A$. Ovenfor er for eksempel $1 \in A$. |
| $x \notin A$ | $x$ er ikke et element i $A$. Ovenfor er for eksempel $7 \notin A$. |
| $A=\{1,2,3,4,5\}$ | $A$ er en mengde som inneholder $5$ elementer, altså tallene $1,2,3,4$ og $5$. |
| $B=\{1,2,3,\dots,10\}$ | $B$ er en mengde som inneholder $10$ elementer, altså tallene fra $1$ til og med $10$. |
| $C=\{x \| P(x) \}$ | $C$ er en mengde av alle $x$ som oppfyller funksjonen $P(x)$. |
| $A \subset B$ | $A$ er en delmengde av $B$. Alle elementene i $A$ er også i $B$, men ikke alle elementene i $B$ er i $A$. |
| $A \subseteq B$ | $A$ er en delmengde av eller lik $B$. Alle elementene i $A$ er også i $B$, og alle elementene i $B$ _kan_ være i $A$. |
| $A \cup B$ | Mengden av alle elementer som er i $A$ eller $B$. |
| $A \cap B$ | Mengden av alle elementer som er i $A$ og $B$. |
| $A / B$ | Mengden av alle elementer som er i $A$ men ikke $B$. |

### 1.3 Sum-og multiplikasjonsnotasjon

Ofte summerer eller multipliserer vi flere tall, eller vil ha et generelt uttrykk for å forenkle operasjonene. 

:::{admonition} Def. 1.3 - Summetegnet
:class: warning

La $a_m, a_{m+1}, \dots, a_n$ være en rekke tall. Summen av disse skrives med den store greske bokstaven sigma,

$\sum_{i=m}^{n} a_i = a_m + a_{m+1} + \dots + a_n$

Her er $i$ en **indeks** som løper gjennom heltallene fra $m$ til og med $n$. Tallet $m$ under summetegnet er **startverdien**, tallet $n$ over er **sluttverdien**, og $a_i$ er **leddet** som summeres.
:::

:::{card} 
**Eksempel**
^^^
Regn ut $\sum_{i=1}^{4} i$.

Indeksen $i$ løper gjennom $1, 2, 3$ og $4$, og vi legger sammen leddene:

$\sum_{i=1}^{4} i = 1+2+3+4=10$
:::

:::{card} 
**Eksempel**
^^^
Regn ut $\sum_{i=1}^{4} i^2$.

Nå er leddet $a_i = i^2$, så vi kvadrerer hvert tall før vi summerer:

$\sum_{i=1}^{4} i^2 = 1^2+2^2+3^2+4^2=1+4+9+16=30$
:::

:::{card} 
**Eksempel**
^^^
Regn ut $\sum_{i=1}^{n} x_i$.

Nå er leddet $a_i = x_i$, så vi legger sammen leddene:

$\sum_{i=1}^{n} x_i = x_1+x_2+\dots+x_n$
:::

Navnet på indeksen har ingen betydning. Uttrykkene $\sum_{i=1}^{3} i$ og $\sum_{k=1}^{3} k$ gir begge $6$, siden indeksen forsvinner når summen er regnet ut. Den kan derfor aldri stå igjen i svaret.

Ofte møter du også skrivemåten $\sum_i a_i$ uten grenser. Det betyr at vi summerer over alle verdiene av $i$ som er aktuelle i sammenhengen. I del 1 skriver vi for eksempel $\sum_i P(i)=1$, altså at sannsynlighetene for alle utfallene til sammen blir $1$. Multiplikasjon har en tilsvarende operator.

:::{admonition} Def. 1.4 - Produkttegnet
:class: warning

La $a_m, a_{m+1}, \dots, a_n$ være en rekke tall. Produktet av disse skrives med den store greske bokstaven pi,

$\prod_{i=m}^{n} a_i = a_m \cdot a_{m+1} \cdot \ldots \cdot a_n$

Her er $i$ en **indeks** som løper gjennom heltallene fra $m$ til og med $n$. Tallet $m$ under produkttegnet er **startverdien**, tallet $n$ over er **sluttverdien**, og $a_i$ er **faktoren** som multipliseres.
:::

:::{card} 
**Eksempel**
^^^
Regn ut $\prod_{i=1}^{4} i$.

Indeksen $i$ løper gjennom $1, 2, 3$ og $4$, og vi multipliserer faktorene:

$\prod_{i=1}^{4} i = 1 \cdot 2 \cdot 3 \cdot 4 = 24$

Legg merke til at dette er nettopp $4!$, altså fakultet. Fakultet kan derfor skrives kompakt som $n!=\prod_{i=1}^{n} i$.
:::

:::{card} 
**Eksempel**
^^^
Regn ut $\prod_{i=1}^{3} (i+2)$.

Nå er faktoren $a_i = i+2$, så vi legger til $2$ før vi multipliserer:

$\prod_{i=1}^{3} (i+2) = 3 \cdot 4 \cdot 5 = 60$
:::

:::{card} 
**Eksempel**
^^^
Regn ut $\prod_{i=1}^{n} x_i$.

Nå er leddet $a_i = x_i$, så vi ganger sammen leddene:

$\prod_{i=1}^{n} x_i = x_1 \cdot x_2 \cdot \dots \cdot x_n$
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Produkttegnet dukker opp når mange bidrag skal ganges sammen i stedet for legges sammen. Et eksempel er likevektskonstanten for en reaksjon, der konsentrasjonene til produktene multipliseres:

$K = \dfrac{\prod_i [\text{produkt}_i]^{\nu_i}}{\prod_j [\text{reaktant}_j]^{\nu_j}}$

Notasjonen sier bare «gang sammen alle produktkonsentrasjonene, opphøyd i sine støkiometriske koeffisienter, og del på det samme for reaktantene».
:::

#### 1.4.2 Dimensjonsanalyse

Målet med dimensjonsanlyse er å kontrollere at enhetene i likningen stemmer, altså at den er konsistent. Dette gjøres ved å behandle enhetene i en formel som algebraiske størrelser som kan multipliseres, forkortes og settes inn i hverandre, akkurat slik man gjør med tallene. Det er enklest forstått med et eksempel.

:::{card} 
**Eksempel**
^^^
En beholder inneholder $2{.}0 \text{ mol}$ av en ideell gass ved temperatur $300 \text{ K}$ og trykk $1{.}0 \times 10^5 \text{ Pa}$. Finn volumet $V$.

Før vi regner ut $V$ ved bruk av idealgassloven, kan vi bruke den samme formelen til å sjekke hva enhetene til svaret blir:

$PV=nRT \Rightarrow V=\frac{nRT}{P} \Rightarrow [V]=\frac{[\text{mol}] \cdot [\text{J}/\text{mol} \cdot \text{K}]\cdot [\text{K}]}{[\text{N}/\text{m}^2]}$

Vi kan stryke både $\text{K}$ og $\text{mol}$ i telleren, slik at vi får:

$[V]=\frac{[\text{J}]}{[\text{N}/\text{m}^2]}$

Vi setter inn enhetene for $\text{N}$ og $\text{J}$, og kan forkorte $\text{m}$ i nevneren:

$[V]=\frac{[\text{kg} \cdot \text{m}^2/\text{s}^2]}{[[\text{kg} \cdot \text{m}/\text{s}^2]/\text{m}^2]}=\frac{[\text{kg} \cdot \text{m}^2/\text{s}^2]}{[\text{kg}/\text{m} \cdot \text{s}^2]}$

Til slutt stryker vi $\text{kg}$ og $\text{s}^2$, og forkorter $\text{m}$:

$[V]=\frac{[\text{m}^2]}{[1/\text{m}]}=[\text{m}^3]$

Altså; i denne oppgaven er enheten for volum $V$ lik $\text{m}^3$.
:::

### 1.5 Eksponenter, logaritmer og røtter

Eksponenter, logaritmer og røtter er tre sider av samme sak: eksponenten bygger opp, mens logaritmen og roten er to ulike måter å ta den fra hverandre igjen på. Alle tre dukker opp overalt i fysikalsk kjemi - fra Boltzmann-faktoren og Arrhenius-likningen til pH-skalaen og fartsfordelingen i en gass.

:::{admonition} Def. 1.5 - Potens og logaritme
:class: important

For et grunntall $a>0$ og et reelt tall $x$ er **potensen** $a^x$ resultatet av å opphøye $a$ i $x$.

**Logaritmen** er den omvendte operasjonen: $\log_a y$ er den eksponenten $a$ må opphøyes i for å gi $y$. De to er knyttet sammen ved

$\log_a y = x \quad \Longleftrightarrow \quad a^x = y$
:::

Det er et spesialtilfelle av logaritmer som er verdt å merke seg og som stadig dukker opp i kjemiske og fysiske beregninger, nemlig den **dnaturlige logaritmen**.

:::{admonition} Def. 1.6 - Naturlig logaritme
:class: important

Den **naturlige logaritmen** av et tall $x$, $\ln{x}$ er definert ved

$\ln x = \log_e x $

der $e \approx 2{.}718$ er Eulers tall.
:::

Røtter hører naturlig hjemme her, for en rot er egentlig bare en potens med brøkeksponent. 

:::{admonition} Def. 1.7 - n-te rot
:class: important

For et positivt heltall $n$ er den **$n$-te roten** av $a$, skrevet $\sqrt[n]{a}$, det tallet som opphøyd i $n$-te potens gir $a$:

$\sqrt[n]{a}=b \quad \Longleftrightarrow \quad b^n=a$

Tallet $n$ kalles **rotens orden**. 

En rot er det samme som en potens med brøkeksponent,

$\sqrt[n]{a}=a^{1/n}$
:::

Reglene for potenser lar oss forenkle uttrykk der samme grunntall opptrer flere ganger.

| Regel | Formel |
| ----- | ------ |
| Produkt | $a^x \cdot a^y = a^{x+y}$ |
| Kvotient | $\dfrac{a^x}{a^y} = a^{x-y}$ |
| Potens av potens | $(a^x)^y = a^{xy}$ |
| Null | $a^0 = 1$ |
| Negativ eksponent | $a^{-x} = \dfrac{1}{a^x}$ |
| Brøkeksponent | $a^{1/n} = \sqrt[n]{a}$ |

Logaritmereglene speiler eksponentreglene: der eksponenter gjør en sum om til et produkt, gjør logaritmen et produkt om til en sum.

| Regel | Formel |
| ----- | ------ |
| Produkt | $\ln(x \cdot y) = \ln x + \ln y$ |
| Kvotient | $\ln\left(\dfrac{x}{y}\right) = \ln x - \ln y$ |
| Potens | $\ln(x^n) = n \ln x$ |
| Grunntallet selv | $\ln e = 1$ |
| Én | $\ln 1 = 0$ |

:::{card} 
**Eksempler**
^^^
* Regn ut $2^3 \cdot 2^5$. Samme grunntall betyr at vi legger sammen eksponentene:

$2^3 \cdot 2^5 = 2^{3+5}=2^8=256$

* Forenkle $(3^2)^4$. Vi ganger eksponentene:

$(3^2)^4 = 3^{2 \cdot 4}=3^8=6561$

* Skriv $\sqrt{50}$ så enkelt som mulig. Vi trekker ut et perfekt kvadrat, og bruker at roten av et produkt er produktet av røttene:

$\sqrt{50}=\sqrt{25 \cdot 2}=\sqrt{25}\cdot\sqrt{2}=5\sqrt{2}$

* Regn ut $\log 1000 - \log 10$. Differansen av to logaritmer er logaritmen av kvotienten:

$\log 1000 - \log 10 = \log\frac{1000}{10}=\log 100 = 2$

* Regn ut $\log_2\sqrt{64}$. Vi skriver roten som brøkeksponent og flytter den ned med potensregelen:

$\log_2\sqrt{64}=\log_2\left(64^{1/2}\right)=\frac{1}{2}\log_2 64=\frac{1}{2}\cdot 6=3$

der vi brukte at $64=2^6$ gir $\log_2 64 = 6$.
:::

## 2 Kombinatorikk

Før vi giver løs på det sentrale tema sannsynlighet, skal vi se på ulike tellemåter for å si noe om antallet muligheter noe kan skje på. I fysikalsk kjemi er dette limet i statistisk termodynamikk: å telle antall mikrotilstander som gir samme makrotilstand (**multiplisiteten** $\Omega$) er nøyaktig et kombinatorikkproblem. Dette tallet dukker senere opp direkte i Boltzmanns entropiformel, $S = k_B \ln \Omega$ - så verktøyene du lærer her er ikke bare telleteknikker, de er selve grunnlaget for hvordan vi definerer entropi.

### 2.1 OG-regelen og ELLER-regelen

OG- og ELLER-regelen er regler du mest sannsynlig har brukt før, enten bevisst eller ubevisst. De handler om to helt ulike situasjoner: OG-regelen brukes når vi gjør flere valg *etter hverandre*, mens ELLER-regelen brukes når vi velger *mellom* alternativer. Nøkkelen til å telle riktig er som regel å avgjøre hvilken av de to situasjonene man står i.

:::{admonition} Def. 2.1 - OG-regelen
:class: warning

Anta at en hendelse $A$ kan skje på $m$ måter og at hendelse $B$ kan skje uavhengig på $n$ måter. Da er antallet måter $A$ OG $B$ kan skje lik 

$m \cdot n$.
:::

:::{admonition} Def. 2.2 - ELLER-regelen
:class: warning

Anta at en hendelse $A$ kan skje på $m$ ulike måter, og hendelse $B$ kan skje på $n$ måter. Anta også at hendelsene er disjunkte. Da er antall måter hendelse $A$ ELLER $B$ skje på lik

$m + n$ 
:::

Legg merke til betingelsene i de to reglene. OG-regelen krever at valgene er uavhengige - antall muligheter for $B$ må ikke endre seg ut fra hva vi valgte for $A$. ELLER-regelen krever at hendelsene er disjunkte, altså at ingen mulighet telles i begge.

En grei huskeregel: skal du gjøre begge deler, ganger du. Skal du gjøre det ene eller det andre, legger du sammen.

:::{admonition} Kobling til fysikalsk kjemi
:class: important

OG-regelen er grunnen til at antall mikrotilstander vokser så voldsomt for store systemer. Har hver av $N$ partikler $k$ mulige tilstander, og partiklene er uavhengige, gir OG-regelen

$\Omega = k \cdot k \cdot \ldots \cdot k = k^N$

For ett mol partikler er $N \approx 6 \cdot 10^{23}$, og $\Omega$ blir da et tall så stort at det er umulig å skrive ut. Nettopp derfor tar vi logaritmen i Boltzmanns entropiformel $S = k_B \ln \Omega$ - logaritmen gjør de enorme produktene om til håndterbare summer.
:::

a) Hvor mange mikrotilstander der molekylet har en bestemt kombinasjon av vibrasjon og rotasjon finnes?

Antall mikrotilstander der molekylet har en bestemt kombinasjon av vibrasjon OG rotasjon blir dermed

Antallet mikrotilstander der molekylet befinner seg i én bestemt vibrasjonstilstand ELLER én bestemt rotasjonstilstand (disjunkte tilstander) blir

### 2.2 Fakultet

Vi skal se at fakultet er svaret på et helt konkret spørsmål: på hvor mange måter kan vi stille opp $n$ forskjellige objekter i rekkefølge? Dette dukker opp overalt - fra hvor mange måter man kan stokke en kortstokk, til hvor mange måter partikler kan fordeles på energinivåer. I tillegg dukker fakultet opp i andre tellemetoder vi skal se i delkapittel 2.3.

:::{admonition} Def. 2.3 - Fakultet
:class: warning

For et ikke-negativt heltall $n$, er fakultet definert slik

$n! = n \cdot (n-1) \cdot (n-2) \cdot \dots \cdot 1$ 

$0!$ er ved definisjon lik $1$.
:::

Vi har $n$-antall objekter som vi skal plassere i rekke - hvor mange ulike måter kan vi gjøre dette på? Første objekt har $n$-antall muligheter. Deretter har andre objekt $n-1$-antall muligheter. Slik følger det videre, helt til objekt $n$ kun har én mulighet. Ved bruk av OG-regelen fra tidligere, multipliseres alle mulighetene og gir opphav til fakultet.

Fakultet av store tall kan ikke regnes ut direkte, verken for hånd eller på datamaskin. I statistisk termodynamikk, der vi opererer med $N \approx 10^{23}$ partikler, trenger vi derfor tilnærminger - vi kommer tilbake til dette i kjemi-koblingen under.

:::{card} 
**Eksempel**
^^^
Fire ulike molekyler skal plasseres på fire nummererte plasser i et krystallgitter, ett molekyl per plass.

a) På hvor mange måter kan dette gjøres?

Første plass kan fylles av hvilket som helst av de $4$ molekylene. Når det er gjort, er det $3$ igjen til andre plass, så $2$ til tredje, og til slutt $1$ molekyl til siste plass. Ved OG-regelen ganger vi mulighetene:

$4 \cdot 3 \cdot 2 \cdot 1 = 4! = 24$

Det finnes altså $24$ ulike måter.

b) Hva om ett av molekylene må stå på plass nummer 1?

Da er den plassen låst, og vi står igjen med $3$ molekyler som skal fordeles på $3$ plasser:

$3! = 6$ måter.
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Fakultet er selve byggesteinen i uttrykket for antall mikrotilstander. Skal $N$ partikler fordeles slik at $n_1$ havner på energinivå $1$, $n_2$ på nivå $2$ og så videre, er antall måter dette kan skje på

$\Omega = \dfrac{N!}{n_1! \, n_2! \cdots}$

Dette er den statistiske vekten fra introduksjonen til del 2, og den går rett inn i Boltzmanns entropiformel $S=k_B \ln \Omega$.

Men her møter vi problemet fra avsnittet over: med $N \approx 10^{23}$ er $N!$ umulig å regne ut. Løsningen er **Stirlings tilnærming**, som sier at for store $N$ er

$\ln N! \approx N \ln N - N$

Mer om denne utledningen finner du i delkapittel 4.10.
:::

:::{admonition} Underveisoppgave 2.2
:class: tip

a) Regn ut $5!$ og $6!$. Hvor mange ganger større er $6!$ enn $5!$?

b) Fem ulike molekyler skal plasseres på fem plasser i et gitter. På hvor mange måter kan det gjøres?

c) Forklar hvorfor $\dfrac{8!}{6!} = 8 \cdot 7$, uten å regne ut hver fakultet for seg.

d) Tre atomer skal plasseres i rekke, men to av dem er identiske. Blir antall synlig forskjellige rekkefølger flere eller færre enn $3!$? Begrunn.
:::

:::{admonition} Løsning 2.2
:class: tip, dropdown

a) $5!=120$ og $6!=720$. Forholdet er $\frac{720}{120}=6$. Dette er ingen tilfeldighet: $6!=6 \cdot 5!$, så hvert steg opp ganger med det nye tallet.

b) Dette er antall måter å ordne $5$ ulike objekter i rekkefølge:

$5! = 5 \cdot 4 \cdot 3 \cdot 2 \cdot 1 = 120$ måter.

c) Vi skriver ut telleren:

$\frac{8!}{6!}=\frac{8 \cdot 7 \cdot 6 \cdot 5 \cdot 4 \cdot 3 \cdot 2 \cdot 1}{6 \cdot 5 \cdot 4 \cdot 3 \cdot 2 \cdot 1}$

Alt fra $6$ og nedover står i både teller og nevner, og forkortes bort. Igjen står $8 \cdot 7 = 56$. Dette trikset er verdt å merke seg - det er nettopp slik vi håndterer store fakulteter i praksis, og det brukes flittig i neste delkapittel.

d) Færre. Med tre *ulike* atomer finnes $3!=6$ rekkefølger. Er to av dem identiske, kan vi ikke se forskjell på de to rekkefølgene der de bytter plass med hverandre, så hver synlig rekkefølge telles dobbelt. Antallet blir da $\frac{3!}{2!}=\frac{6}{2}=3$.

Dette er nøyaktig samme idé som nevnerne i formelen for $\Omega$: vi deler bort de omstokkingene vi ikke kan skille fra hverandre.
:::

### 2.3 Tellemetoder

Før vi begynner å telle antall muligheter må vi vite hva vi har med å gjøre. Det er spesielt to egenskaper ved objektene vi teller vi må vite noe om:

**1) Gjentakelse**

Kan hvert objekt/plass brukes én gang, eller flere ganger? Vi sier at vi enten teller **med gjentakelse** eller **uten gjentakelse**.

**2) Rekkefølge**

Har rekkefølgen til objektene noe å si (er de skjelnbare), eller ikke? Vi sier at utvalget enten er **ordnet** hvis rekkefølgen har noe å si, eller **uordnet** ved motsatt tilfelle.

Alt i alt, gir dette fire ulike tellemåter!

Som i forrige oppgave får vi

hvis rekkefølgen har noe å si. Det har den ikke nå! Vi har flere muligheter av 3 smaker som vi ser på som identiske, for eksempel jordbær, sjokolade og vanlije, nøyaktig samme mulighet som sjokolade, vanilje og jordbær.

$3 \cdot 2 \cdot 1 = 3! = 6 \text{ muligheter}$

Ved å dele på dette tallet i formelen som overteller, får vi det korrekte antallet, og dermed

$\frac{10!}{3! \cdot (10-3)!} = 120 \text{ muligheter}$

$\frac{9!}{6! \cdot (9-6)!} = \frac{(6+4-1)!}{6! \cdot ((6+4-1)-6)!}$

Vi skal nå generalisere dette og få noen ordentlige formler å jobbe med.

:::{admonition} Def. 2.4 - Ordninger i et ordnet utvalg, med gjentakelse

Antall måter å velge $k$ elementer fra en mengde med $n$ elementer, med gjentakelse og der rekkefølgen betyr noe er

$n^k$
:::

:::{admonition} Def. 2.5 - Permutasjon

Antall måter å velge $k$ elementer fra en mengde med $n$ elementer, uten gjentakelse og der rekkefølgen betyr noe er

$P(n,k) = \frac{n!}{(n-k)!}$

Hver slik måte kalles en **permutasjon**.
:::

:::{admonition} Def. 2.6 - Kombinasjon

Antall måter å velge $k$ elementer fra en mengde med $n$ elementer, uten gjentakelse og der rekkefølgen *ikke* betyr noe er

$\binom{n}{k} = \frac{n!}{k! \cdot (n-k)!}$

Hver slik måte kalles en **kombinasjon**, og tallet kalles **binomialkoeffisienten**.
:::

:::{admonition} Def. 2.7 - Ordninger i et uordnet utvalg, med gjentakelse

Antall måter å velge $k$ elementer fra en mengde med $n$ elementer, med gjentakelse og der rekkefølgen *ikke* betyr noe er

$\binom{n+k-1}{k}\frac{(n+k-1)!}{k! \cdot (n-1)!}$

:::

Vi følger på med flere eksempler!

:::{card} 
**Eksempler**
^^^
Et tripeptid bygges av tre aminosyrer i rekkefølge, valgt blant de 20 vanlige. Samme aminosyre kan brukes flere ganger, og rekkefølgen betyr noe. Hvor mange ulike tripeptider finnes? Dette er et ordnet utvalg med gjentakelse, så vi bruker Def. 2.4:

$n^k = 20^3 = 8000$

Fem ulike ligander skal binde seg til et metallion, men bare tre bindingsplasser er ledige, og rekkefølgen de bindes i gir ulike mellomprodukter. Hvor mange rekkefølger finnes? Her velger vi uten gjentakelse og rekkefølgen teller, altså en permutasjon (Def. 2.5):

$P(5,3)=\frac{5!}{(5-3)!}=\frac{5!}{2!}=\frac{120}{2}=60$

En benzenring har seks posisjoner, og vi skal sette to like substituenter på to av dem. Rekkefølgen er likegyldig - de to substituentene er identiske. Dette er en kombinasjon (Def. 2.6):

$\binom{6}{2}=\frac{6!}{2! \cdot 4!}=\frac{720}{2 \cdot 24}=15$

Fire like energikvanter skal fordeles på tre identiske oscillatorer. Kvantene kan ikke skilles fra hverandre (gjentakelse tillatt), og rekkefølgen betyr ingenting - bare hvor mange hver oscillator får. Dette er et uordnet utvalg med gjentakelse (Def. 2.7):

$\binom{n+k-1}{k}=\binom{3+4-1}{4}=\binom{6}{4}=15$

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Kombinasjoner er akkurat det vi trenger for å telle mikrotilstander. Se på et system med $N$ partikler der hver partikkel enten er i grunntilstanden eller i en eksitert tilstand. Spørsmålet "hvor mange mikrotilstander har nøyaktig $n$ eksiterte partikler?" er nøyaktig spørsmålet "hvor mange måter kan vi velge ut $n$ av $N$ partikler?":

$\Omega = \binom{N}{n} = \frac{N!}{n!\,(N-n)!}$

For $N=10$ partikler med $n=3$ eksiterte gir dette $\binom{10}{3}=120$ mikrotilstander.

Grunnen til at vi bruker kombinasjoner og ikke permutasjoner, er at partiklene er **uskillbare** i denne sammenhengen: makrotilstanden bryr seg bare om *hvor mange* som er eksiterte, ikke om hvilke. Dette er samme skille som mellom mikro- og makrotilstand fra delkapittel 1.1 - og legg merke til at $\Omega$ her er et spesialtilfelle av den generelle formelen $\Omega=\frac{N!}{n_1!\,n_2!\cdots}$ fra delkapittel 2.2, med bare to nivåer.
:::

## 3 Sannsynlighet

Sannsynlighet handler om å prøve å forutsi. Vi vet ikke hva en terning viser før den lander, eller hvilket molekyl som reagerer først - men vi kan si noe presist om hvor ofte de ulike utfallene inntreffer i det lange løp.

Det er verdt å legge merke til hva slags kunnskap dette er. Vi gir ikke opp og sier at noe er tilfeldig og dermed uten struktur. Tvert imot: tilfeldighet har sine egne lover, og de er like eksakte som alt annet i matematikken. Vi begynner derfor helt fra bunnen, med å beskrive nøyaktig hva som kan skje, før vi går videre til hvor sannsynlig hvert utfall er.

For en kjemiker er dette mer enn terningkast og kortstokker. Et gram vann inneholder rundt $10^{22}$ molekyler, og vi kan umulig følge hvert enkelt av dem. Vi kan heller ikke forutsi når et bestemt radioaktivt atom henfaller, eller nøyaktig hvilken energi ett molekyl har i neste øyeblikk.

Løsningen er å beskrive systemet med sannsynligheter i stedet. Det er nettopp dette som gjør statistisk termodynamikk mulig: vi gir opp å følge enkeltmolekyler, og regner i stedet på fordelinger. Underveis skal vi se at begreper du møter i kjemien - mikrotilstander, makrotilstander og Boltzmann-fordelingen - er sannsynlighetsteori i forkledning.

### 3.1 Utfall, utfallsrom og hendelser

Du triller en terning, og teller antall øyne terningen viser. Dette er et av mange eksempler på et statistisk eksperiment. For å kunne snakke om sannsynlighet og eksperimenter trenger vi noen definisjoner. Vi starter med de mest grunnleggende: **utfall**, **utfallsrom** og **hendelser**.

:::{admonition} Def. 3.1 - Utfall
:class: warning

Et **utfall** $i$ er et mulig resultat i et eksperiment. 
:::


:::{admonition} Def. 3.2 - Utfallsrom
:class: warning

**Utfallsrommet** $S$ er mengden av alle mulige utfall i et eksperiment.
:::


:::{admonition} Def. 3.3 - Hendelse
:class: warning

En **hendelse** $A$ er delmengde av utfallsrommet, altså ett eller flere utfall vi er interessert i.

Merk at hele $S$ er en hendelse (den som alltid skjer), og den tomme mengden $\{\}$ også er en hendelse (den som aldri skjer).
:::

:::{admonition} Husk
:class: error
Mengder og delmengder er definert i delkapittelet [1.2 Mengdelære](https://andreasdh.github.io/Bifrost/docs/mattehjelp.html#mengdelaere).
:::

:::{admonition} OBS
:class: error
Legg merke til at en hendelse alltid må være en delmengde av utfallsrommet! Tenk deg at du trekker ett kort fra en kortstokk, og kun ser på symbolet til kortet. Utfallsrommet er da $S=\{\text{kløver},\text{ruter},\text{spar},\text{hjerter}\}$. Nå er ikke "trekke et bildekort (dame, knekt, konge eller ess)" en hendelse - den informasjonen finnes rett og slett ikke i utfallsrommet vårt. 
:::

Før vi går videre trenger vi å si noe om hva sannsynligheten til et utfall faktisk betyr. 

:::{admonition} Def. 3.4 - Den klassiske sannsynlighetsdefinisjonen
:class: warning

I et eksperiment der alle utfall er **like sannsynlige**, som et kast med en likevektet terning, er sannsynligheten for hendelse $A$ gitt ved

$P(A) = \frac{\text{antall utfall i } A}{\text{antall utfall i } S}$
:::

Definisjon 3.4 krever at alle utfall er like sannsynlige, men slik er det sjelden i virkeligheten. I delkapittel 3.4 generaliserer vi slik at vi også kan jobbe med slike sannsynligheter.

:::{admonition} Kobling til fysikalsk kjemi
:class: important

I _Grunnleggende begreper_ defineres [mikrotilstander og makrotilstander](https://andreasdh.github.io/Bifrost/docs/intro.html#mikrotilstander-og-makrotilstander). Mikrotilstander kan sammenlignes med utfall, altså én bestemt fordeling av energi/posisjon over alle partiklene i systemet. Utfallsrommet $S$ blir da mengden av *alle* mikrotilstander systemet kan befinne seg i, og en hendelse $A$ tilsvarer en **makrotilstand** (f.eks. en gitt total energi, temperatur eller trykk) - altså alle mikrotilstander som gir samme makroskopiske observasjon.
:::

:::{card} 
**Eksempel**
^^^

Du triller en sekssidet terning.

a) Hva er utfallsrommet $S$ til eksperimentet?

Du kan enten få $1,2,3,4,5$ eller $6$, som betyr at utfallsrommet er $S=\{1,2,3,4,5,6\}$.

b) Hva er et eksempel på en hendelse i dette eksperimentet?

Et eksempel på en hendelse i dette eksperimentet er å trille et partall. La oss betegne denne hendelsen ved $A$, og da har vi at $A=\{2,4,6\}$. Et annet eksempel er å trille $1,2,3,4,5$ eller $6$, altså $S$, og denne hendelsen er garantert.

c) Hva er sannsynligheten for å trille partall?

Hendelsen er $A=\{2,4,6\}$, som inneholder $3$ utfall. Utfallsrommet $S=\{1,2,3,4,5,6\}$ inneholder $6$ utfall. Siden terningen er likevektet, er alle utfall like sannsynlige, og vi kan telle:

$P(A) = \frac{\text{antall utfall i } A}{\text{antall utfall i } S} = \frac{3}{6} = \frac{1}{2}$

Sannsynligheten for å trille et partall er altså $\frac{1}{2}$.
:::

### 3.2 Komplement, snitt og union

La oss nå se på hva som skjer utenfor en hendelse $A$, og hvordan vi kan kombinere to hendelser $A$ og $B$ til nye hendelser. Vi ser først på **komplementet** til en hendelse, og deretter **snitt** og **union**.

:::{admonition} Def. 3.5 - Komplement
:class: warning

**Komplementet **til en hendelse $A$, $A'$, er mengden av alle utfall i $S$ som ikke er i $A$.
:::


:::{admonition} Def. 3.6 - Snitt
:class: warning

**Snittet** av to hendelser $A$ og $B$, $A \cap B$, er hendelsen som består av alle utfall som både er i $A$ og i $B$.
:::


:::{admonition} Def. 3.7 - Union
:class: warning

**Unionen** av to hendelser $A$ og $B$, $A \cup B$, er hendelsen som består av alle utfall som er i $A$ eller i $B$.
:::

### 3.3 Disjunkte og uavhengige hendelser

Med operasjonene på plass kan vi beskrive to viktige egenskaper et par hendelser kan ha. Den første, **disjunkthet**, handler bare om utfallene: hendelsene overlapper ikke. Den andre, **uavhengighet**, handler om sannsynlighet: den ene hendelsen sier ingenting om den andre. Det er lett å blande dem sammen, men de betyr noe helt forskjellig - vi skal se at uavhengighet blir svært sentralt i statistisk termodynamikk.

:::{admonition} Def. 3.8 - Disjunkte hendelser
:class: warning

To hendelser $A$ og $B$ kalles **disjunkte** dersom de ikke har noen felles utfall. Matematisk, $A \cap B = \{\}$.
:::

:::{admonition} Def. 3.9 - Uavhengige hendelser
:class: warning

To hendelser $A$ og $B$ kalles **uavhengige** dersom

$P(A \cap B) = P(A) \cdot P(B)$
:::

Forskjellen er verdt å dvele ved. Disjunkte hendelser kan _aldri_ skje samtidig - triller du en terning, kan du ikke få både $1$ og $6$ i samme kast. Uavhengige hendelser kan derimot godt skje samtidig; poenget er at den ene ikke påvirker sannsynligheten for den andre.

:::{card} 
**Eksempel**
^^^
Du triller en terning to ganger etter hverandre og noterer resultatet for hvert kast. La $A$ være hendelsen "første terning viser $6$", og $B$ hendelsen "andre terning viser $6$". Er $A$ og $B$ uavhengige?

Hver terning har $\frac{1}{6}$ sannsynlighet for å vise $6$, så $P(A)=P(B)=\frac{1}{6}$.

For at begge skal vise $6$, må vi få utfallet $(6,6)$. Til sammen finnes det $6 \cdot 6 = 36$ mulige utfall for to terninger, og bare ett av dem er $(6,6)$. Altså er $P(A \cap B)=\frac{1}{36}$.

Vi sjekker kravet fra Def. 3.8:

$P(A) \cdot P(B) = \frac{1}{6} \cdot \frac{1}{6} = \frac{1}{36} = P(A \cap B)$

Kravet er oppfylt, så hendelsene er uavhengige - som forventet, siden den første terningen umulig kan påvirke den andre.
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Uavhengighet er sentralt når vi skal beskrive et molekyls bevegelse: translasjon, rotasjon og vibrasjon regnes vanligvis som uavhengige av hverandre. Det er nettopp derfor vi senere kan bruke $P(A \cap B)=P(A)\cdot P(B)$ til å si at partisjonsfunksjonen faktoriserer, $Q_{tot}=Q_{trans}\cdot Q_{rot}\cdot Q_{vib}$.
:::

:class: tip

b) Er $A$ og $C$ disjunkte? Begrunn.

c) Vi regner ut hver sannsynlighet ved opptelling. $A$ har $2$ av $6$ utfall, så $P(A)=\frac{2}{6}=\frac{1}{3}$. $C$ har $3$ av $6$ utfall, så $P(C)=\frac{3}{6}=\frac{1}{2}$. Snittet $A \cap C = \{2\}$ har ett utfall, så $P(A \cap C)=\frac{1}{6}$.

Vi sjekker kravet fra Def. 3.8:

$P(A) \cdot P(C) = \frac{1}{3} \cdot \frac{1}{2} = \frac{1}{6} = P(A \cap C)$

Kravet er oppfylt, så $A$ og $C$ er uavhengige. Legg merke til at dette skjer selv om begge hendelsene gjelder *samme* terningkast - uavhengighet krever ikke separate eksperimenter, men sier heller at sannsynligheten for den ene hendelsen ikke påvirker sannsynligheten for den andre.

### 3.4 Aksiomene i sannsynlighet

Enten du spiller sjakk eller ikke, vet du at spillet har visse regler, som at hver brikke kan flyttes på en spesiell måte, at hvit starter osv. Disse reglene er ikke noe som er _bevist_, men de er fastsatt som et utgangspunkt for spillet. Alt annet du kan si om sjakk, for eksempel hvordan man vinner, taper eller om en åpning er god eller dårlig, er utledet fra disse reglene. 

Sånn er også matematikken! Som reglene i sjakk, har matematikken grunnleggende byggesteiner som kalles **aksiomer**. Teoremer og andre utledninger er bygget logisk fra aksiomene. Kolmogorov presenterte aksiomene i sannsynlighetsteori, og til tross for at disse kan virke åpenbare, er de utgangspunktet til alt vi skal se på - også flere utledninger i statistisk termodynamikk.

:::{admonition} Aksiomer 3.1 - Kolmogorovs sannsynlighetsaksiomer
:class: warning

For et eksperiment med utfallsrom $S$ har vi


**Ikke-negativitet**

Sannsynligheten for enhver hendelse $A$ er ikke-negativ

1. $P(A) \ge 0$ 

**Normering**

Sannsynligheten for $S$ er lik $1$, altså

2. $P(S)=1$

**Additivitet**

Sannsynligheten for unionen av flere disjunkte hendelser $A_1, A_2,\dots$ er summen av sannsynligheten for hver av hendelsene

3. $P(A_1 \cup A_2 \cup \dots)=P(A_1)+P(A_2)+\dots$
:::

Legg merke til at aksiomene ikke sier noe om *hvor stor* sannsynligheten for et bestemt utfall er - de sier bare hvilke regler sannsynlighetene må følge. Det er nettopp derfor de også gjelder for vektede terninger og for mikrotilstander i et fysisk system, ikke bare for rettferdige spill. Et viktig følge av aksiom 2 og 3 er følgende:

:::{admonition} Teorem 3.1 - Summen av alle utfalls sannsynligheter
:class: note

For et vilkårlig eksperiment med utfallsrom $S$ har vi

$\sum_{i}P(i)=1$

for alle utfall $i \in S$.
:::

Utledningen er kort: de enkelte utfallene er disjunkte hendelser - du kan ikke få både $3$ og $5$ i samme terningkast - og til sammen utgjør de hele utfallsrommet. Ved aksiom 3 kan vi da summere sannsynlighetene deres, og ved aksiom 2 må summen bli $P(S)=1$.

:::{admonition} Teorem 3.2 - Komplementregelen
:class: note

For enhver hendelse $A$ gjelder

$P(A') = 1 - P(A)$
:::

Også denne følger av aksiomene. En hendelse $A$ og komplementet $A'$ er disjunkte etter Def. 1.4, og sammen dekker de hele $S$. Aksiom 3 gir da $P(A)+P(A')=P(S)$, og aksiom 2 sier at $P(S)=1$. Flytter vi $P(A)$ over, får vi resultatet.

Regelen er ofte det raskeste verktøyet i praksis: skal du finne sannsynligheten for "minst én", er det som regel lettere å regne ut sannsynligheten for "ingen" og trekke fra $1$.

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Aksiomene over gjelder for *ethvert* utfallsrom, også for mikrotilstandene til et termodynamisk system. Sannsynligheten for at et system i termisk likevekt ved temperatur $T$ befinner seg i mikrotilstand $i$ med energi $E_i$ er gitt ved Boltzmann-fordelingen

$P(i) = \dfrac{e^{-E_i/k_BT}}{Q}$

der $k_B$ er Boltzmanns konstant og $Q=\sum_i e^{-E_i/k_BT}$ er **partisjonsfunksjonen**. Merk at $Q$ er akkurat den normeringskonstanten som gjør at $\sum_i P(i)=1$, i tråd med Teorem 2.1 over.
:::

:::{card} 
**Eksempel**
^^^
Et lite system med 2 partikler som hver kan ha energi $0$ eller $\varepsilon$ har mikrotilstandene $S=\{(0,0),(0,\varepsilon),(\varepsilon,0),(\varepsilon,\varepsilon)\}$ - helt analogt til å kaste to mynter. Makrotilstanden "total energi $=\varepsilon$" er hendelsen $A=\{(0,\varepsilon),(\varepsilon,0)\}$. Er alle mikrotilstander like sannsynlige (høy temperatur), blir $P(A)=2/4=1/2$, akkurat som å få én kron og ett mynt.
:::

$\sum_{i}P(i)=P(\{1\})+P(\{2\})+P(\{3\})+P(\{4\})+P(\{5\})+P(\{6\})=\frac{1}{6}+\frac{1}{6}+\frac{1}{6}+\frac{1}{6}+\frac{1}{6}+\frac{1}{6}=1$

$P(\text{ingen seksere}) = \frac{5}{6} \cdot \frac{5}{6} = \frac{25}{36}$

$P(\text{minst én sekser}) = 1 - \frac{25}{36} = \frac{11}{36}$

### 3.5 Betinget sannsynlighet 

Så langt har vi sett på sannsynligheten for hendelser hver for seg. I mange situasjoner sitter vi imidlertid med **delvis informasjon**: vi vet allerede at noe har skjedd, og spør hva det gjør med sannsynligheten for noe annet. Trekker du et kort og får vite at det er rødt, endrer det sannsynligheten for at det er hjerter. Slike spørsmål besvares av betinget sannsynlighet.

:::{admonition} Def. 3.9 - Betinget sannsynlighet
:class: warning

Sannsynligheten for at hendelse $A$ skjer, gitt at hendelse $B$ allerede har skjedd, er

$P(A|B) = \dfrac{P(A \cap B)}{P(B)}$

forutsatt at $P(B)>0$.
:::

Formelen gir mening om du tenker på den som en **innsnevring av utfallsrommet**. Når vi vet at $B$ har skjedd, er alle utfall utenfor $B$ ikke lenger aktuelle - $B$ har i praksis blitt vårt nye utfallsrom. Telleren $P(A \cap B)$ er den delen av $A$ som fortsatt er mulig, og vi deler på $P(B)$ for å normere, slik at det nye utfallsrommet igjen har total sannsynlighet $1$, i tråd med aksiom 2.

Legg merke til sammenhengen med uavhengighet fra Def. 3.8. Er $A$ og $B$ uavhengige, er $P(A \cap B)=P(A)\cdot P(B)$, og da blir

$P(A|B) = \frac{P(A)\cdot P(B)}{P(B)} = P(A)$

Å vite at $B$ har skjedd endrer altså ingenting - som er nøyaktig det uavhengighet betyr.

:::{card} 
**Eksempel**
^^^
Du trekker ett kort fra en vanlig kortstokk. La $A$ være hendelsen "kortet er hjerter" og $B$ hendelsen "kortet er rødt". Hva er $P(A|B)$?

Det er $13$ hjerterkort, så $P(A)=\frac{13}{52}=\frac{1}{4}$. De røde kortene er hjerter og ruter, til sammen $26$ kort, så $P(B)=\frac{26}{52}=\frac{1}{2}$.

Alle hjerterkort er røde, så snittet $A \cap B$ er rett og slett alle hjerterkort: $P(A \cap B)=\frac{13}{52}=\frac{1}{4}$.

Ved Def. 3.9 får vi

$P(A|B)=\frac{P(A \cap B)}{P(B)}=\frac{1/4}{1/2}=\frac{1}{2}$

Vet vi at kortet er rødt, er sannsynligheten for hjerter altså $\frac{1}{2}$ - opp fra $\frac{1}{4}$ uten den informasjonen. Det stemmer med intuisjonen: blant de $26$ røde kortene er $13$ hjerter.
:::

Ganger vi opp med $P(B)$ på begge sider av definisjonen, får vi en nyttig omskriving som lar oss regne ut sannsynligheten for at *begge* hendelsene skjer.

:::{admonition} Teorem 3.3 - Produktregelen
:class: note

For to hendelser $A$ og $B$ gjelder

$P(A \cap B) = P(A|B) \cdot P(B)$

Dersom $A$ og $B$ er uavhengige, forenkles dette til $P(A \cap B)=P(A)\cdot P(B)$, som i Def. 3.8.
:::

Snittet $A \cap B$ er det samme som $B \cap A$, så vi kan skrive produktregelen på to måter: $P(A|B)P(B)$ og $P(B|A)P(A)$. Setter vi disse lik hverandre og løser for $P(A|B)$, får vi et av de mest brukte resultatene i all sannsynlighetsregning.

:::{admonition} Teorem 3.4 - Bayes' teorem
:class: note

For to hendelser $A$ og $B$ med $P(B)>0$ gjelder

$P(A|B) = \dfrac{P(B|A) \, P(A)}{P(B)}$
:::

Bayes' teorem lar oss **snu en betingelse**. Ofte kjenner vi $P(B|A)$, men ønsker egentlig $P(A|B)$. Et typisk eksempel er medisinsk testing: vi vet hvor ofte en test slår ut når pasienten faktisk er syk, $P(\text{positiv}|\text{syk})$, men det pasienten lurer på er det motsatte - hvor sannsynlig det er at hun er syk gitt at testen var positiv, $P(\text{syk}|\text{positiv})$. Disse to er ikke det samme, og forskjellen kan være dramatisk når sykdommen er sjelden.

:::{card} 
**Eksempel**
^^^
En sjelden sykdom rammer $1$ av $1000$ personer. En test gir positivt utslag hos $99\%$ av de syke, men gir også falskt positivt hos $5\%$ av de friske. Du tester positivt - hva er sannsynligheten for at du faktisk er syk?

La $A$ være "syk" og $B$ "positiv test". Vi kjenner

$P(A)=0{,}001, \qquad P(B|A)=0{,}99, \qquad P(B|A')=0{,}05$

Først finner vi $P(B)$, altså hvor ofte testen slår ut i det hele tatt. En positiv test kan komme fra en syk eller en frisk person:

$P(B)=P(B|A)P(A)+P(B|A')P(A')=0{,}99 \cdot 0{,}001 + 0{,}05 \cdot 0{,}999 = 0{,}00099+0{,}04995=0{,}05094$

Ved Bayes' teorem er

$P(A|B)=\frac{P(B|A)P(A)}{P(B)}=\frac{0{,}99 \cdot 0{,}001}{0{,}05094} \approx 0{,}019$

Sannsynligheten for å være syk er altså bare rundt $2\%$, til tross for en test som er $99\%$ treffsikker på syke. Grunnen er at sykdommen er så sjelden at de falske positive blant de mange friske langt overgår de ekte positive.
:::

Til slutt tar vi med en regel for unioner. Aksiom 3 lot oss summere sannsynligheter for disjunkte hendelser, men hva om hendelsene overlapper?

:::{admonition} Teorem 3.5 - Additiv lov
:class: note

For to hendelser $A$ og $B$ som ikke nødvendigvis er disjunkte, gjelder

$P(A \cup B) = P(A) + P(B) - P(A \cap B)$

Dersom $A$ og $B$ er disjunkte er $P(A \cap B)=0$, og uttrykket blir $P(A \cup B)=P(A)+P(B)$, som i aksiom 3.
:::

Grunnen til at vi trekker fra $P(A \cap B)$ er at utfallene i snittet ellers ville blitt talt to ganger - én gang i $P(A)$ og én gang i $P(B)$. Se gjerne tilbake på Venn-diagrammet i delkapittel 3.2: overlappen skal bare telles én gang.

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Betinget sannsynlighet dukker opp så snart vi har delvis informasjon om et system. Vet vi at et molekyl befinner seg i en bestemt elektronisk tilstand, endrer det sannsynlighetsfordelingen over vibrasjonsnivåene - vi regner da på $P(\text{vibrasjonsnivå} \mid \text{elektronisk tilstand})$.

Produktregelen i Teorem 3.3 er også grunnen til at Boltzmann-fordelingen kan faktoriseres når frihetsgradene er uavhengige. Er translasjon, rotasjon og vibrasjon uavhengige, er den betingede sannsynligheten lik den ubetingede, og sannsynlighetene kan ganges sammen direkte - noe som gir $Q_{tot}=Q_{trans}\cdot Q_{rot}\cdot Q_{vib}$ fra delkapittel 3.3.
:::

### 3.6 Kontinuerlige utfallsrom

Så langt har alle utfallsrommene våre vært **diskrete**: terningkast, kortstokk, myntkast. Felles for dem er at vi kan liste opp utfallene ett etter ett, og at hvert utfall har sin egen sannsynlighet. I fysikalsk kjemi er mange størrelser derimot **kontinuerlige**. Farten til et gassmolekyl kan være hvilken som helst verdi over null, og energien til en partikkel kan ligge hvor som helst i et intervall. Slike utfallsrom kan vi ikke telle oss gjennom.

:::{admonition} Def. 3.10 - Kontinuerlig utfallsrom
:class: warning

Et utfallsrom $S$ kalles **kontinuerlig** dersom utfallene utgjør et sammenhengende intervall av tall, slik at de ikke kan listes opp enkeltvis.
:::

Overgangen til kontinuerlige utfallsrom har en konsekvens som er verdt å stoppe ved. I et diskret utfallsrom hadde hvert enkelt utfall en positiv sannsynlighet - hver side på terningen hadde $\frac{1}{6}$. I et kontinuerlig utfallsrom er det umulig. Det finnes uendelig mange verdier i et intervall, og skulle hver av dem hatt en positiv sannsynlighet, ville summen sprengt $1$ og brutt aksiom 2.

Løsningen er at hvert **enkeltutfall** har sannsynlighet $0$. Sannsynligheten for at et molekyl har farten nøyaktig $400$ m/s, uten noe avvik overhodet, er null. Dette betyr ikke at utfallet er umulig - molekylet har tross alt en fart - men at det er én verdi blant uendelig mange.

::::{grid}
:gutter: 2

:::{grid-item-card} Diskret

_Utfallene kan listes én og én_

* Terningkast
* Trekke kort fra en kortstokk
* Elektroniske energinivåer

:::

:::{grid-item-card} Kontinuerlig

_Utfallene utgjør et sammenhengdene intervall av tall_

* Høyden til en tilfeldig person
* Hvor på en dartskive pilen treffer
* Tiden en reaksjon tar

:::

::::

Hendelser fungerer fortsatt som før: de er delmengder av $S$, helt i tråd med Def. 3.3. Forskjellen er at de meningsfulle hendelsene nå er **intervaller** i stedet for enkeltutfall. Vi spør ikke "hva er sannsynligheten for at farten er nøyaktig $400$ m/s?", men "hva er sannsynligheten for at farten ligger mellom $400$ og $500$ m/s?".

En følge av at enkeltutfall har sannsynlighet $0$ er at endepunktene ikke spiller noen rolle. Hendelsene $400 \leq v \leq 500$ og $400 < v < 500$ har nøyaktig samme sannsynlighet, siden de to endepunktene hver bidrar med $0$.

:::{card} 
**Eksempel**
^^^
Du snurrer en viser på en sirkulær skive, og noterer vinkelen $\theta$ visesen stopper på, målt i grader.

a) Hva er utfallsrommet?

Viseren kan stoppe på hvilken som helst vinkel fra $0$ til $360$ grader, så utfallsrommet er det kontinuerlige intervallet $S=[0,360)$.

b) Hva er sannsynligheten for at viseren stopper på nøyaktig $90$ grader?

Siden $S$ er kontinuerlig, har hvert enkeltutfall sannsynlighet $0$. Altså er $P(\theta = 90)=0$. Det finnes uendelig mange vinkler viseren kan stoppe på, og nøyaktig $90{,}000\ldots$ grader er bare én av dem.

c) Hva er sannsynligheten for at viseren stopper i første kvadrant, altså mellom $0$ og $90$ grader?

Dette er en hendelse på intervallform, $A=[0,90]$. Er skiven jevnt balansert, er ingen del av sirkelen favorisert, og sannsynligheten svarer til hvor stor andel av sirkelen intervallet dekker:

$P(A)=\frac{90}{360}=\frac{1}{4}$
:::

Legg merke til hva vi gjorde i c). I det diskrete tilfellet fant vi sannsynlighet ved å telle utfall og dele på antallet i $S$. Her kunne vi ikke telle, men vi kunne måle: vi sammenliknet **lengden** av intervallet med lengden av hele utfallsrommet. Så lenge ingen del av utfallsrommet er favorisert, er dette den kontinuerlige parallellen til opptelling.

De fleste kontinuerlige størrelser i kjemien er imidlertid *ikke* jevnt fordelt. Molekylfarter i en gass er tettere samlet rundt noen verdier enn andre, og da holder det ikke å sammenlikne lengder. For å håndtere slike tilfeller trenger vi et nytt verktøy, som vi kommer tilbake til i del 3.

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Skillet mellom diskrete og kontinuerlige utfallsrom går rett inn i kvantefysikkens kjerne. Energinivåene til et elektron i et atom er **diskrete** - elektronet kan bare ha bestemte energier, og utfallsrommet kan listes opp. Det er nettopp derfor vi ser skarpe linjer i et atomspektrum.

Translasjonsenergien til et molekyl i en gassbeholder er derimot i praksis **kontinuerlig**. Nivåene er strengt tatt kvantiserte også her, men de ligger så tett at vi behandler dem som et sammenhengende intervall. Denne tilnærmingen er helt sentral i statistisk termodynamikk, og er grunnen til at vi senere kan erstatte summer med integraler når vi regner på translasjonsbidraget til partisjonsfunksjonen.
:::

## 4 Kalkulus

Der de to foregående delene handlet om usikkerhet og telling, handler kalkulus om **endring**. Hvor raskt stiger temperaturen? Hvor mye arbeid utføres når en gass utvider seg? Hvordan finner vi det punktet der energien er lavest?

Kalkulus gir oss to verktøy som svarer på hver sin side av dette. Derivasjon måler hvor raskt noe endrer seg akkurat nå, og integrasjon summerer opp mange små bidrag til en helhet. Det som gjør kalkulus til noe mer enn to regneteknikker, er at disse to viser seg å være hverandres motsatte.

For en kjemiker er kalkulus arbeidsspråket i store deler av faget. Reaksjonskinetikk handler om den deriverte av konsentrasjon med hensyn på tid. Termodynamikk er full av partiellderiverte som $\left(\frac{\partial P}{\partial T}\right)_V$, der vi spør hvordan én størrelse endrer seg mens andre holdes fast. Arbeidet en gass utfører er et integral, $W=-\int P\,dV$, og normaliseringen av en orbital er et trippelintegral over hele rommet.

Verktøyene herfra brukes også direkte i neste del. Når vi der skal beskrive kontinuerlige sannsynlighetsfordelinger, er det integrasjon som gjør det mulig: sannsynligheten for et intervall blir arealet under en kurve.

Vi bygger derfor opp verktøykassen gradvis: først funksjoner og grenseverdier, deretter derivasjon og integrasjon for én variabel, og til slutt utvidelsen til funksjoner av flere variabler - som er nødvendig så snart en størrelse avhenger av både trykk og temperatur, eller av tre romkoordinater.

### 4.1 Definisjonen av en funksjon

Funksjoner i matematikk er regler som forbinder hvert eneste tall i en mengde med ett og bare ett tall i en annen eller samme mengde. Du er allerede kjent med flere funksjoner, som for eksempel funksjonen for arealet av en sirkel gitt dens radius, pH-en til en løsning gitt dens syrekonsentrasjon eller antall H-atomer i et alkan gitt antallet C-atomer. 

:::{card} 
**Eksempel**
^^^
For et alkan med et heltall $n$ antall C-atomer, er antallet H-atomer $H$ gitt ved funksjonen

$H=2n+2$

Dette er en funksjon fordi for hver $n$ får vi nøyaktig én $H$. Vi kan også skrive $H(n)$ for $H$ for å spesifisere at $H$ er en funksjon av $n$:

$H(n)=2n+2$
:::


Som du nå ser kan funksjoner brukes til å si noe om hvordan en størrelse varierer med en annen størrelse. Derfor er det naturlig å finne ut nøyaktig hva vi kan si om disse såkalte funksjonene. Vi begynner med definisjonen av en funksjon, og dens definisjonsmengde og verdiområde.

:::{admonition} Def. 4.1 - Funksjon
:class: warning

*Intuitivt:* En funksjon $f$ er en regel som tildeler hver verdi $x$ nøyaktig én verdi $f(x)$.

*Matematisk:* La $X$ og $Y$ være to mengder. En funksjon $f$ fra $X$ til $Y$, tildeler hvert element $x \in X$ nøyaktig ett element i $Y$, $f(x)$.

Vi bruker følgende notasjon:

$f:X \rightarrow Y$
:::

:::{admonition} Def. 4.2 - Definisjonsmengde og verdiområde
:class: warning

La $f:X \to Y$ være en funksjon fra $X$ til $Y$. Da kalles $X$ for **definisjonsmengden** til $f$, mens $Y$ kalles for **verdiområdet**.
:::

:::{card} 
**Eksempel**
^^^
La $H:\{1,2,3,4,5\} \to N$ være funksjonen for antallet H-atomer $H(n)$ i et alkan med $n$-antall C-atomer, gitt ved

$H(n)=2n+2$

a) Hva er definisjonsområdet til $H$?

Definisjonsmengden er $\{1,2,3,4,5\}$ og sier noe om hvilke $n$-verdier vi ser på. Merk at dette er en diskret definisjonsmengde fordi vi ikke har alle tallene mellom $1$ og $5$, men spesifikt $1,2,3,4$ eller $5$. Med andre ord ser vi på metan, etan, propan, butan og pentan.

b) Hva er verdiområdet til $H$?

Verdiområdet er $N$, altså de naturlige tallene. Det betyr at antallet H-atomer $H(n)$ er et naturlig tall.
:::



Vi er ferdig med første steg, nemlig definisjonen av en funksjon. Allerede nå har vi fått to nye begreper å beskrive funksjoner med - definisjonsmengde og verdiområde. Før vi går videre, skal vi se på noen måter å representere funksjoner på. Du er allerede kjent med grafer og algebraisk representasjon (formelen for funksjonen). Det er i hovedsak disse representasjonene som er mest relevante for deg som kjemistudent, men det finnes også flere!

### 4.2 Jevne og odde funksjoner            

To viktige typer funksjoner du kan komme på å møte i fysikalsk kjemi er **jevne og odde funksjoner**.

:::{admonition} Def. 4.3 - Jevne og odde funksjoner
:class: warning

Anta at $f$ er definert ved både $x$ og $-x$.

Vi kaller $f$ for en **jevn funksjon** hvis

$f(-x)=f(x)$ for alle $x$ i definisjonsområdet til $f$.

Vi kaller $f$ for en **odde funksjon** hvis

$f(-x)=-f(x)$ for alle $x$ i definisjonsområdet til $f$.
:::

:::{card} 
**Eksempel**
^^^
Vis at $f(x)=x^2$ er en jevn funksjon.

Vi vet at $f$ er definert for alle reelle tall.

La $x$ være et vilkårlig tall i definisjonsområdet til $f$, altså $x \in R$. Da blir

$f(x)=x^2$

Samtidig blir

$f(-x)=(-x)^2=x^2$

Altså har vi vist at $f(x)=f(-x)$, og $f$ er en jevn funksjon.
:::

:::{card} 
**Eksempel**
^^^
Vis at $f(x)=x^3-2x$ er en odde funksjon.

Vi vet at $f$ er definert for alle reelle tall.

La $x$ være et vilkårlig tall i definisjonsområdet til $f$, altså $x \in R$. Da blir

$f(x)=x^3-2x$

Samtidig blir

$f(-x)=(-x)^3-2(-x)=-x^3+2x=-(x^3-2x)$

Altså har vi vist at $f(x)=-f(x)$, og $f$ er en odde funksjon.
:::

:::{admonition} OBS
:class: error
En funksjon **må ikke** være enten jevn eller odde, den kan være ingen av delene.
:::

Det er derimot mye enklere å forstå hva jevne og odde funksjoner er ved å visualisere dem, fordi egenskapene beskrevet ovenfor gir opphav til en symmetri. Jevne funksjoner er symmetriske om $y$-aksen, mens odde funksjoner er symmetriske om origo. Nedenfor får du eksempel på en jevn og en odde funksjon, og kan endre $x$-verdiene for å se hvordan $f(x)$ og $f(-x)$ endrer seg.

### 4.3 Grenseverdier og kontinuitet

Vi innleder grenseverdier med et eksempel.

:::{card} 
**Eksempel**
^^^
Beskriv funksjonen $f(x)=\frac{x^2-1}{x-1}$ omkring $x=1$.

La oss se hva vi kan si om definisjonsmengden til $f$, altså hvilke $x$-verdier vi kan ha. Annet enn $x=1$ som gir $0$ i nevneren, er alle reelle $x$-verdier tillate. Verdiområdet er de reelle tallene.

Et observant øye ser at det er mulig å faktorisere uttrykket, slik at:

$f(x)=\frac{x^2-1}{x-1}=\frac{(x-1)(x+1)}{(x-1)}=x+1$ for $x\neq1$

Grafen til $f$ blir dermed linjen $y=x+1$ **uten** punktet i $(1,2)$, der blir det et hull! Selv om $f(1)$ ikke er definert, kan vi få $f(x)$ så nærme vi vil $2$ ved å velge en $x$ nærme nok $1$. Vi sier at $f(x)$ går mot grenseverdien $2$ når $x$ går mot $1$. Notasjonen for dette er:

$\lim_{x \to 1} f(x)=2$

eller

$\lim_{x \to 1} \frac{x^2-1}{x-1} = 2$
:::

![Grenseverdi eksempel med hull](bilder/funksjon_hull.svg)

Den formelle definisjonen av grenseverdi er komplisert, og muligens kjent for deg som har hatt MAT1100. Vi går for å presentere den uformelle definisjonen i stedet, da den ikke er særlig viktig for oss.

:::{admonition} Def. 4.4 - Grenseverdi (uformelt)
:class: warning

La $f$ være en funksjon og $L$ et tall. Hvis $f(x) \to L$ når $x \to a$, kaller vi $L$ for **grenseverdien** til $f(x)$ når $x$ går mot $a$. Vi skriver:

$\lim_{x \to a} f(x) = L$
:::

Vi har nå nok verktøy til å gå løs på kontinuitet. Heldigvis har kontinuitet en veldig intuitiv forklaring: hvis du kan tegne grafen til en funksjon fra start til slutt uten å måtte løfte pennen din av papiret, vel da er den kontinuerlig. Men, matematikken er formell og vi må ha klare og tydelige definisjoner. 

:::{admonition} Def. 4.5 - Kontinuitiet i et punkt
:class: warning

Vi sier at funksjonen $f$ er **kontinuerlig i et punkt** $c$ i definisjonsområdet hvis

$\lim_{x \to c} f(x) = f(c)$

Hvis 

$\lim_{x \to c} f(x)$ ikke eksisterer 

eller 
$\lim_{x \to c} f(x) \neq f(c)$ 

så er $f$ **diskontinuerlig** i $c$.
:::

![Kontinuitet i ett punkt](bilder/kontinuitet_tre_grafer.svg)

:::{admonition} Def. 4.6 - Kontinuitiet i et intervall
:class: warning

Vi sier at funksjonen $f$ er **kontinuerlig på intervallet** $I$ hvis den er kontinuerlig i hver at punktene i $I$.
:::

:::{admonition} Def. 4.7 - Kontinuerlig funksjon
:class: warning

Vi sier at funksjonen $f$ er **kontinuerlig** (på hele definisjonsområdet) hvis den er kontinuerlig i hvert at punktene i definisjonsområdet.
:::

:::{admonition} Underveisoppgave 4.3
:class: tip

a) Regn ut grenseverdien $\lim_{x \to 2} \dfrac{x^2-4}{x-2}$ ved å faktorisere telleren.

b) Funksjonen $g$ er definert ved $g(x)=\dfrac{x^2-4}{x-2}$ for $x \neq 2$, og $g(2)=3$. Er $g$ kontinuerlig i $x=2$? Begrunn med Def. 4.5.
:::

:::{admonition} Løsning 4.3
:class: tip, dropdown

a) Vi faktoriserer telleren: $x^2-4=(x-2)(x+2)$. For $x \neq 2$ er da

$\frac{x^2-4}{x-2}=\frac{(x-2)(x+2)}{x-2}=x+2$

slik at $\lim_{x \to 2} \dfrac{x^2-4}{x-2}=\lim_{x \to 2}(x+2)=4$.

b) For at $g$ skal være kontinuerlig i $x=2$, må $\lim_{x \to 2} g(x)=g(2)$. Vi fant at grenseverdien er $4$, men $g(2)=3$. Siden $\lim_{x \to 2} g(x)=4 \neq 3=g(2)$, er $g$ **diskontinuerlig** i $x=2$. (Hadde vi i stedet satt $g(2)=4$, ville funksjonen vært kontinuerlig.)
:::

### 4.4 Derivasjon

Derivasjon er et matematisk verktøy for å sette ord på noe du sannsynligvis allerede ser på grafen til en funksjon. Følger du en graf, kan du for eksempel observere at den stiger, når et toppunkt, synker litt og kanskje treffer et bunnpunkt. Husk at en av hovedgrunnene til at vi interesserer oss for funksjoner er at de kan modellere virkelige fenomener og sammenhengen mellom to størrelser. Ofte er vi derfor interessert i hvordan denne sammenhengen endrer seg, hvor mye den endrer seg, og i hvilken retning.

Vi begynner med å introdusere gjennomsnittlig endring mellom to punkter.

:::{admonition} Def. 4.8 - Gjennomsnittlig endringsrate mellom to punkter
:class: warning

Den gjennomsnittlige endringsraten til en funksjon $f$ over et intervall $[a,b]$ er gitt ved 

$\frac{\Delta y}{\Delta x}=\frac{f(b)-f(a)}{b-a}$
:::

![Gjennomsnittlig endringsrate](bilder/gjennomsnittlig_endring.svg)

En rett linje som går gjennom to punkter, slik som den rette linjen som går gjennom $(a,f(a))$ og $(b,f(b))$, kalles en **sekant**. Den gjennomsnittlige endringsraten mellom disse to punktene er da stigningstallet til sekanten. Men hva med endringsraten i ett enkelt punkt? Vi introduserer definisjonen av den deriverte.

:::{card} 
**Eksempel**
^^^
Hva er den gjennomsnittlige endringsraten for funksjonen $f(x)=x^2+1$ i intervallet $[1,3]$?

Vi har at

$f(1)=1^2+1=2$

$f(3)=3^2+1=10$

Dermed får vi

$\frac{\Delta y}{\Delta x}=\frac{10-2}{3-1}=\frac{8}{2}=4$
:::

:::{admonition} Def. 4.8 - Den deriverte
:class: warning

Den **deriverte** til en funksjon $f$ er en annen funksjon $f'$ definert ved

$f'(x)=\lim_{h \to 0} \frac{f(x+h)-f(x)}{h}$

for alle $x$ der grenseverdien eksisterer.

Hvis $f'(x)$ eksisterer, sier vi at $f$ er **deriverbar** i $x$.
:::

:::{card} 
**Eksempel**
^^^
Bruk definisjonen til å finne den deriverte av $f(x)=x^2$.

Vi setter inn i definisjonen:

$f'(x)=\lim_{h \to 0} \frac{f(x+h)-f(x)}{h}=\lim_{h \to 0} \frac{(x+h)^2-x^2}{h}$

Vi regner ut telleren:

$(x+h)^2-x^2=x^2+2xh+h^2-x^2=2xh+h^2$

Dermed får vi

$f'(x)=\lim_{h \to 0} \frac{2xh+h^2}{h}=\lim_{h \to 0} (2x+h)=2x$
:::

:::{card} 
**Eksempel**
^^^
Bruk definisjonen til å finne den deriverte av $f(x)=x^2-3x$.

Vi setter inn i definisjonen:

$f'(x)=\lim_{h \to 0} \frac{f(x+h)-f(x)}{h}=\lim_{h \to 0} \frac{\left[(x+h)^2-3(x+h)\right]-\left[x^2-3x\right]}{h}$

Vi regner ut telleren:

$(x+h)^2-3(x+h)-x^2+3x=x^2+2xh+h^2-3x-3h-x^2+3x=2xh+h^2-3h$

Dermed får vi

$f'(x)=\lim_{h \to 0} \frac{2xh+h^2-3h}{h}=\lim_{h \to 0} (2x+h-3)=2x-3$
:::

Definisjonen av den deriverte er ikke akkurat lett fordøyelig. Vi har punktene $(x, f(x))$ og tar et steg $h$ fra $x$ til punktet $(x+h, f(x+h))$. Så gjør vi $h$ mindre og mindre - vi finner altså grenseverdien av den gjennomsnittlige endringsraten mellom disse punktene når $h$ går mot $0$. Nedenfor kan du endre på $x$ og $h$ og se hvordan den gjennomsnittlige endringsraten og den deriverte endrer seg. Se spesielt på hva som skjer når $h$ går mot $0$.

Vi har funnet den første-deriverte til $f$, men det er ingenting som stopper oss fra å finne høyere-ordens deriverte slik som den dobbelderiverte, tredje deriverte osv.

:::{admonition} Def. 4.8 - Høyere-ordens deriverte (n-te deriverte)
:class: warning

Den $n$**-te deriverte** til en funksjon $f$ er en annen funksjon $f^{(n)}$ definert ved å ta den deriverte av $f$ $n$-antall ganger.
:::

Nå som vi har en idé av hva den deriverte er, skal vi lære oss å finne de deriverte for ulike funksjoner. I et av eksemplene ovenfor finner vi $f'$ for $f(x)=x^2$ direkte ut ifra definisjonen av den deriverte. Heldigvis for oss har mange flinke matematikere funnet ut av de deriverte for de vanligste, elementære funksjonene. De aller vanligste er gitt i tabellen nedenfor. 

| $f(x)$       | $f'(x)$      |  Betingelser |
|-----------   |-----------  |-------|
| $a$        | $0$           |  |
| $x^a$       | $ax^{a-1}$   | |
| $e^x$      | $e^x$   | |
| $a^x$   | $a^x \ln{(a)}$   | $a>0$ |
| $\ln{(x)}$ | $\frac{1}{x}$ | $x>0$ |
| $\log_{a}{(x)}$ | $\frac{1}{x \ln{(a)}}$ | $x,a>0$ |
| $\sin{(x)}$ | $\cos{(x)}$ | |
| $\cos{(x)}$ | $-\sin{(x)}$ | | 
| $\tan{(x)}$ | $\sec^2{(x)}=\frac{1}{\cos^2{(x)}}=1+\tan^2{(x)}$ | | 
| $\arcsin{(x)}$ | $\frac{1}{\sqrt{1-x^2}}$ | $-1 < x < 1$ |
| $\arccos{(x)}$ | -$\frac{1}{\sqrt{1-x^2}}$ | $-1 < x < 1$ |
| $\arctan{(x)}$ | $\frac{1}{1+x^2}$ |  |

For å finne de deriverte til sammensatte og mer kompliserte funksjoner, må vi ha noen derivasjonsregler. Først introduseres litt notasjon for å unngå forvirring!

| Lagrange       | Leipniz                |
|-----------     |-----------             |
| $f'(x)$        | $\frac{df}{dx}$        |
| $f''(x)$       | $\frac{d^2f}{dx^2}$   |
| $f'''(x)$      | $\frac{d^3f}{dx^3}$   |
| $f^{(n)}(x)$   | $\frac{d^nf}{dx^n}$   |

Videre presenteres de ulike derivasjonsreglene. 

:::{admonition} Teorem 4.1 - Faktorregelen
:class: note

La $g$ være en deriverbar funksjon i $x$ og la $a$ være et tall. Da er 

$\frac{d}{dx} [a \cdot g(x)]=a \cdot g'(x)$
:::

:::{admonition} Teorem 4.2 - Summeregelen
:class: note

La funksjonene $g$ og $h$ er deriverbare i $x$. Da er

$\frac{d}{dx} [g(x) + h(x)]=g'(x) + h'(x)$
:::

:::{admonition} Teorem 4.3 - Produktregelen
:class: note

La funksjonene $g$ og $h$ er deriverbare i $x$. Da er

$\frac{d}{dx} [g(x) \cdot h(x)]=g'(x)h(x)+g(x)h'(x)$
:::

:::{admonition} Teorem 4.4 - Kvotientregelen
:class: note

La funksjonene $g$ og $h$ er deriverbare i $x$. Da er

$\frac{d}{dx} [\frac{g(x)}{h(x)}]=\frac{g'(x)h(x)-g(x)h'(x)}{h(x)^2}$
:::

:::{admonition} Teorem 4.5 - Kjerneregelen
:class: note

La funksjonene $g$ og $h$ er deriverbare i $x$. Da er

$\frac{d}{dx} [(g \circ h)(x)]=(g(h(x)))'=g(h(x)) \cdot h'(x)$
:::

Derivasjon av funksjoner er mengdetrening. Prøv å forstå alle eksemplene nedenfor.

:::{card} 
**Eksempel 1**
^^^
Deriver $f(x)=7x^4$.

Vi kan faktorisere $7$ ut ved faktorregelen, og bruker potensregelen på $x^4$, slik at

$f'(x)=7 \cdot \frac{d}{dx}\left(x^4\right)=7 \cdot 4x^3=28x^3$
:::

:::{card} 
**Eksempel 2**
^^^
Deriver $f(x)=x^3+\sin{(x)}$.

Ved summeregelen deriverer vi hvert ledd for seg, slik at

$f'(x)=\frac{d}{dx}\left(x^3\right)+\frac{d}{dx}\left(\sin{(x)}\right)=3x^2+\cos{(x)}$
:::

:::{card} 
**Eksempel 3**
^^^
Deriver $f(x)=4x^2+3e^x-5\ln{(x)}$.

Vi bruker summeregelen på leddene og faktorregelen på hver konstant:

$f'(x)=4 \cdot 2x+3 \cdot e^x-5 \cdot \frac{1}{x}=8x+3e^x-\frac{5}{x}$
:::

:::{card} 
**Eksempel 4**
^^^
Deriver $f(x)=x^2\sin{(x)}$.

Dette er et produkt av $x^2$ og $\sin{(x)}$, så vi bruker produktregelen:

$f'(x)=2x\sin{(x)}+x^2\cos{(x)}$
:::

:::{card} 
**Eksempel 5**
^^^
Deriver $f(x)=e^x\cos{(x)}$.

Vi bruker produktregelen, og husker at $e^x$ er sin egen deriverte, slik at

$f'(x)=e^x\cos{(x)}+e^x\cdot(-\sin{(x)})=e^x\left(\cos{(x)}-\sin{(x)}\right)$
:::

:::{card} 
**Eksempel 6**
^^^
Deriver $f(x)=\dfrac{x}{\cos{(x)}}$.

Dette er en kvotient med teller $x$ og nevner $\cos{(x)}$, så vi bruker kvotientregelen, slik at

$f'(x)=\frac{1\cdot\cos{(x)}-x\cdot(-\sin{(x)})}{\cos^2{(x)}}=\frac{\cos{(x)}+x\sin{(x)}}{\cos^2{(x)}}$
:::

:::{card} 
**Eksempel 7**
^^^
Deriver $f(x)=\dfrac{\ln{(x)}}{x^2}$.

Vi bruker kvotientregelen med teller $\ln{(x)}$ og nevner $x^2$, slik at

$f'(x)=\frac{\frac{1}{x}\cdot x^2-\ln{(x)}\cdot 2x}{\left(x^2\right)^2}=\frac{x-2x\ln{(x)}}{x^4}=\frac{1-2\ln{(x)}}{x^3}$
:::

:::{card} 
**Eksempel 8**
^^^
Deriver $f(x)=\sin{(x^2)}$.

La $g(x)=\sin{(x)}$ være den ytre funksjonen og $h(x)=x^2$ være den indre funksjonen. Vi har at

$g'(x)=\cos{(x)}$

$h'(x)=2x$

Ved kjerneregelen får vi derfor

$f'(x)=g'(h(x)) \cdot h'(x)=\cos{(x^2)} \cdot 2x=2x\cos{(x^2)}$
:::

:::{card} 
**Eksempel 9**
^^^
Deriver $f(x)=e^{3x}$.

La $g(x)=e^{x}$ være den ytre funksjonen og $h(x)=3x$ være den indre funksjonen. Vi har at

$g'(x)=e^{x}$

$h'(x)=3$

Ved kjerneregelen får vi derfor

$f'(x)=g'(h(x)) \cdot h'(x)=e^{3x} \cdot 3=3e^{3x}$
:::

:::{card} 
**Eksempel 10**
^^^
Deriver $f(x)=x \cdot \ln{(x^2+1)}$.

Dette er et produkt av $u(x)=x$ og $v(x)=\ln{(x^2+1)}$, så vi bruker produktregelen. Først deriverer vi hver faktor.

For $u(x)=x$ har vi

$u'(x)=1$

Faktoren $v(x)=\ln{(x^2+1)}$ deriverer vi med kjerneregelen. La $g(x)=\ln{(x)}$ være den ytre funksjonen og $h(x)=x^2+1$ være den indre funksjonen. Vi har at

$g'(x)=\frac{1}{x}$

$h'(x)=2x$

slik at

$v'(x)=g'(h(x)) \cdot h'(x)=\frac{1}{x^2+1} \cdot 2x=\frac{2x}{x^2+1}$

Ved produktregelen får vi derfor

$f'(x)=u'(x)v(x)+u(x)v'(x)=1 \cdot \ln{(x^2+1)}+x \cdot \frac{2x}{x^2+1}=\ln{(x^2+1)}+\frac{2x^2}{x^2+1}$

### 4.5 Integrasjon

Vi har til nå sett på derivasjon, som handler om endring: gitt en funksjon $f$, finner vi den momentane endringsraten $f'$. Integrasjon er på mange måter den motsatte operasjonen. I stedet for å spørre "hvor raskt endrer $f$ seg?", spør vi "hvor mye har $f$ samlet seg opp?". Vi skal se at integrasjon svarer på to spørsmål som i utgangspunktet virker helt ulike - arealet under en graf, og det å reversere en derivasjon - og at disse to henger tett sammen.

Vi starter med å reversere derivasjon. Vi har sett at å derivere $F(x)=x^2$ gir $f(x)=2x$. Men kan vi gå fra $f(x)=2x$ tilbake til $F(x)=x^2$? En slik funksjon $F$, som har $f$ som sin deriverte, kaller vi en antiderivert til $f$.

:::{admonition} Def. 4.9 - Antiderivert
:class: warning

En funksjon $F$ kalles en **antiderivert** til $f$ dersom

$F'(x)=f(x)$

for alle $x$ i definisjonsmengden til $f$.
:::

Legg merke til at en antiderivert aldri er entydig. Siden den deriverte av en konstant er $0$, vil både $F(x)=x^2$, $F(x)=x^2+1$ og $F(x)=x^2-7$ alle ha $f(x)=2x$ som sin deriverte. Alle antideriverte til en funksjon skiller seg derfor bare med en konstant. Denne konstanten kaller vi integrasjonskonstanten $C$, og den må alltid tas med.

:::{admonition} Def. 4.10 - Det ubestemte integralet
:class: warning

Det **ubestemte integralet** til en funksjon $f$ er mengden av alle antideriverte til $f$, skrevet

$\int f(x) \, dx = F(x) + C$

der $F$ er en antiderivert til $f$, og $C$ er en konstant.
:::

Symbolet $\int$ kalles integraltegnet, funksjonen $f(x)$ kalles integranden, og $dx$ forteller oss hvilken variabel vi integrerer med hensyn på. På samme måte som med derivasjon finnes det en tabell over de vanligste integralene. Legg merke til at tabellen bare er derivasjonstabellen "lest baklengs".

| $f(x)$          | $\int f(x)\,dx$                | Betingelser |
|-----------      |-----------                     |-------|
| $a$             | $ax + C$                       | |
| $x^n$           | $\frac{1}{n+1}x^{n+1} + C$      | $n \neq -1$ |
| $\frac{1}{x}$   | $\ln{\|x\|} + C$               | $x \neq 0$ |
| $e^x$           | $e^x + C$                      | |
| $a^x$           | $\frac{a^x}{\ln{(a)}} + C$     | $a>0,\ a\neq 1$ |
| $\sin{(x)}$     | $-\cos{(x)} + C$               | |
| $\cos{(x)}$     | $\sin{(x)} + C$                | |
| $\frac{1}{\cos^2{(x)}}$ | $\tan{(x)} + C$        | |
| $\frac{1}{\sqrt{1-x^2}}$ | $\arcsin{(x)} + C$    | $-1 < x < 1$ |
| $\frac{1}{1+x^2}$ | $\arctan{(x)} + C$           | |

Legg spesielt merke til graden for $x^n$: vi øker eksponenten med én og deler på den nye eksponenten. Dette er det motsatte av potensregelen for derivasjon. Regelen gjelder for alle $n \neq -1$, og tilfellet $n=-1$ (altså $\frac{1}{x}$) er nettopp derfor gitt sin egen rad, med $\ln{|x|}$ som svar.

Akkurat som for derivasjon finnes det regneregler som lar oss integrere summer og konstant-multipler ledd for ledd. Disse følger direkte av de tilsvarende reglene for derivasjon.

:::{admonition} Teorem 4.6 - Linearitet ved integrasjon
:class: note

La $f$ og $g$ være integrerbare funksjoner og la $a$ være et tall. Da er

$\int a \cdot f(x) \, dx = a \int f(x) \, dx$

$\int \left[ f(x) + g(x) \right] dx = \int f(x)\,dx + \int g(x)\,dx$
:::

:::{card} 
**Eksempel**
^^^
Regn ut $\int \left( 3x^2 + 2x \right) dx$.

Ved linearitet integrerer vi hvert ledd for seg, og trekker konstantene utenfor:

$\int \left( 3x^2 + 2x \right) dx = 3\int x^2 \, dx + 2\int x \, dx = 3 \cdot \frac{x^3}{3} + 2 \cdot \frac{x^2}{2} + C = x^3 + x^2 + C$

Merk at vi bare trenger én integrasjonskonstant $C$ til slutt, siden summen av flere vilkårlige konstanter fortsatt bare er en vilkårlig konstant.
:::

Vi bytter nå perspektiv, og spør: hva er arealet mellom grafen til $f$ og $x$-aksen, fra $x=a$ til $x=b$? For en rett linje er dette lett - da får vi et rektangel eller en trekant. Men for en krum graf trenger vi en ny idé: vi deler intervallet opp i mange smale striper, tilnærmer hver stripe med et rektangel, og legger sammen arealene. Jo smalere striper, jo bedre tilnærming.

Grenseverdien av disse rektangelsummene, når antall striper går mot uendelig og bredden mot null, kaller vi **det bestemte integralet**. Vi lar bredden på hver stripe være $\Delta x$ og høyden $f(x_i)$, og summerer.

Når vi tilnærmer arealet med rektangler, må vi bestemme hvor høyt hvert rektangel skal være - altså hvilket punkt i hver stripe vi leser av funksjonsverdien fra. Tre vanlige valg er å bruke venstre endepunkt, høyre endepunkt eller midtpunktet av hver stripe. Uansett hvilket vi velger, kaller vi en slik sum en **Riemann-sum**, etter matematikeren Bernhard Riemann. Velger vi venstre endepunkt for en stigende funksjon, blir rektanglene litt for lave og summen for liten; velger vi høyre endepunkt, blir den for stor. Midtpunktet treffer som regel best. Det avgjørende er at *alle* tre valgene gir samme grenseverdi når stripene blir uendelig smale.

:::{admonition} Def. 4.11 - Det bestemte integralet
:class: warning

Det **bestemte integralet** til $f$ fra $a$ til $b$ er grenseverdien av rektangelsummene når antall striper $n$ går mot uendelig:

$\int_a^b f(x) \, dx = \lim_{n \to \infty} \sum_{i=1}^{n} f(x_i)\,\Delta x$

der $\Delta x = \frac{b-a}{n}$. 

Tallene $a$ og $b$ kalles henholdsvis nedre og øvre grense.
:::

Til forskjell fra det ubestemte integralet, som er en funksjon (pluss en konstant), er det bestemte integralet et tall. Det er verdt å merke seg at dette tallet er et areal med fortegn: der grafen ligger under $x$-aksen, teller arealet negativt.

Vi har nå to helt forskjellige begreper som begge kalles "integral": det ubestemte (antiderivasjon) og det bestemte (areal som grense av rektangelsummer). Det store, og på mange måter overraskende, resultatet i kalkulus er at disse to henger sammen. For å regne ut et areal trenger vi ikke summere uendelig mange rektangler - det holder å finne en antiderivert. Dette er **analysens fundamentalteorem**.

:::{admonition} Teorem 4.7 - Analysens fundamentalteorem
:class: note

La $f$ være en kontinuerlig funksjon på $[a,b]$, og la $F$ være en antiderivert til $f$, altså $F'=f$. Da er

$\int_a^b f(x) \, dx = F(b) - F(a)$

Vi bruker gjerne en kort skrivemåte for uttrykket $F(b)-F(a)$, nemlig 

$\Big[F(x)\Big]_a^b$
:::

Fremgangsmåten for å regne ut et bestemt integral blir da: finn en antiderivert $F$, sett inn øvre grense, sett inn nedre grense, og trekk fra. Siden vi til slutt trekker $F(a)$ fra $F(b)$, faller integrasjonskonstanten $C$ alltid bort - derfor tar vi den ikke med i bestemte integraler.

:::{card} 
**Eksempel**
^^^
Regn ut arealet under grafen til $f(x)=x^2$ fra $x=0$ til $x=2$, altså $\int_0^2 x^2 \, dx$.

Vi finner først en antiderivert: $F(x)=\frac{x^3}{3}$. Ved fundamentalteoremet er

$\int_0^2 x^2 \, dx = \Big[\frac{x^3}{3}\Big]_0^2 = \frac{2^3}{3} - \frac{0^3}{3} = \frac{8}{3} - 0 = \frac{8}{3}$

Arealet er altså $\frac{8}{3} \approx 2{,}67$. Dette er nøyaktig den verdien rektangelsummene nærmet seg i simulasjonen ovenfor.
:::

:::{card} 
**Eksempel**
^^^
Regn ut $\int_1^3 (2x+1) \, dx$.

En antiderivert til $2x+1$ er $F(x)=x^2+x$. Da får vi

$\int_1^3 (2x+1) \, dx = \Big[x^2+x\Big]_1^3 = \left(3^2+3\right) - \left(1^2+1\right) = 12 - 2 = 10$
:::

:::{card} 
**Eksempel**
^^^
Regn ut $\int_0^{\pi} \sin{(x)} \, dx$.

En antiderivert til $\sin{(x)}$ er $F(x)=-\cos{(x)}$. Dermed er

$\int_0^{\pi} \sin{(x)} \, dx = \Big[-\cos{(x)}\Big]_0^{\pi} = -\cos{(\pi)} - \left(-\cos{(0)}\right) = -(-1) + 1 = 2$
:::

Så langt har vi kunnet slå opp antideriverte i tabellen, men hva med funksjoner som $x e^x$ eller $2x\cos{(x^2)}$, som ikke står i tabellen? For slike trenger vi teknikker. De to viktigste er substitusjon (som reverserer kjerneregelen) og delvis integrasjon (som reverserer produktregelen).

:::{admonition} Teorem 4.8 - Substitusjon
:class: note

La $f$ og $g$ være slik at $g$ er deriverbar. Da er

$\int f(g(x)) \cdot g'(x) \, dx = \int f(u) \, du$

der vi har satt $u=g(x)$ og $du = g'(x)\,dx$.
:::

Idéen bak substitusjon er å kjenne igjen en "indre funksjon" $g(x)$ og dens deriverte $g'(x)$ i integranden, og så bytte variabel fra $x$ til $u=g(x)$. Da forsvinner den ytre kompleksiteten, og vi sitter igjen med et integral vi kan slå opp.

:::{card} 
**Eksempel**
^^^
Regn ut $\int 2x \cos{(x^2)} \, dx$.

Vi ser at $x^2$ er en indre funksjon, og at dens deriverte $2x$ står som faktor i integranden. Vi setter derfor

$u = x^2 \quad \Rightarrow \quad \frac{du}{dx} = 2x \quad \Rightarrow \quad du = 2x \, dx$

Integralet blir da

$\int 2x \cos{(x^2)} \, dx = \int \cos{(u)} \, du = \sin{(u)} + C = \sin{(x^2)} + C$

Vi bytter altså tilbake til $x$ til slutt. Sjekk gjerne ved å derivere $\sin{(x^2)}$ med kjerneregelen - da får vi $2x\cos{(x^2)}$ tilbake.
:::

:::{admonition} Teorem 4.9 - Delvis integrasjon
:class: note

La $u$ og $v$ være deriverbare funksjoner av $x$. Da er

$\int u \, v' \, dx = u v - \int u' \, v \, dx$
:::

Delvis integrasjon er nyttig når integranden er et produkt av to funksjoner av ulik type - for eksempel et polynom ganget med en eksponential- eller trigonometrisk funksjon. Trikset er å velge den ene faktoren som $u$ (den vi deriverer) og den andre som $v'$ (den vi integrerer), slik at det nye integralet blir enklere enn det opprinnelige.

:::{card} 
**Eksempel**
^^^
Regn ut $\int x e^x \, dx$.

Vi velger $u=x$ (som blir enklere når vi deriverer) og $v'=e^x$ (som er lett å integrere). Da er

$u = x \quad \Rightarrow \quad u' = 1$

$v' = e^x \quad \Rightarrow \quad v = e^x$

Ved delvis integrasjon får vi

$\int x e^x \, dx = u v - \int u' v \, dx = x e^x - \int 1 \cdot e^x \, dx = x e^x - e^x + C = (x-1)e^x + C$

Legg merke til at det nye integralet $\int e^x \, dx$ var enklere enn det vi startet med - det er hele poenget med å velge $u$ og $v'$ smart.
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Integrasjon er helt sentralt i termodynamikk. Når en gass utvider seg og utfører arbeid, er det totale arbeidet gitt ved integralet av trykket over volumendringen,

$W = -\int_{V_1}^{V_2} P \, dV$

altså arealet under kurven i et trykk-volum-diagram. På samme måte finner vi endring i entropi ved å integrere $\frac{dq_{rev}}{T}$, og sammenhengen mellom varmekapasitet og entalpi ved $\Delta H = \int_{T_1}^{T_2} C_p \, dT$. I alle disse tilfellene er idéen den samme som i Def. 4.11: vi summerer opp uendelig mange små bidrag til noe kontinuerlig endrer seg.
:::

Med integrasjon på plass har vi de to grunnpilarene i kalkulus - derivasjon og integrasjon - og fundamentalteoremet som binder dem sammen. I de neste delkapitlene utvider vi til funksjoner av flere variabler.

:::{admonition} Underveisoppgave 4.5
:class: tip

a) Regn ut det ubestemte integralet $\int \left(4x^3-2x\right)dx$.

b) Regn ut det bestemte integralet $\int_1^2 6x^2 \, dx$ med fundamentalteoremet.

c) Regn ut $\int_0^{\pi/2} \cos{(x)} \, dx$.

d) Regn ut $\int 3x^2 \sin{(x^3)} \, dx$.
:::

:::{admonition} Løsning 4.5
:class: tip, dropdown

a) Vi integrerer ledd for ledd: $\int \left(4x^3-2x\right)dx = 4 \cdot \frac{x^4}{4} - 2 \cdot \frac{x^2}{2} + C = x^4 - x^2 + C$.

b) En antiderivert til $6x^2$ er $F(x)=2x^3$. Da er

$\int_1^2 6x^2 \, dx = \Big[2x^3\Big]_1^2 = 2\cdot 8 - 2 \cdot 1 = 16-2 = 14$

c) En antiderivert til $\cos{(x)}$ er $F(x)=\sin{(x)}$. Da er

$\int_0^{\pi/2} \cos{(x)} \, dx = \Big[\sin{(x)}\Big]_0^{\pi/2} = \sin{\left(\tfrac{\pi}{2}\right)} - \sin{(0)} = 1 - 0 = 1$

d) Vi ser at $x^3$ er en indre funksjon med deriverte $3x^2$, som står som faktor. Vi setter $u=x^3$, slik at $du=3x^2\,dx$. Da blir

$\int 3x^2 \sin{(x^3)} \, dx = \int \sin{(u)} \, du = -\cos{(u)} + C = -\cos{(x^3)} + C$
:::

### 4.6 Flervariable funksjoner

Så langt har alle funksjonene våre hatt én input og én output - vi puttet inn ett tall $x$ og fikk ut ett tall $f(x)$. Men det er mange eksempler hvor en størrelse avhenger av flere enn én annen størrelse. I kjemien avhenger de fleste størrelser av *flere* ting samtidig. Trykket i en gass avhenger både av volumet og temperaturen. Løseligheten til et stoff avhenger av både temperatur og trykk. Reaksjonshastigheten avhenger av konsentrasjonen til flere reaktanter på én gang. For å beskrive slike sammenhenger trenger vi funksjoner som tar inn flere variabler. Heldigvis er ikke ideen så forskjellig fra det vi allerede kan - vi utvider bare antallet input.

Vi starter med å utvide selve funksjonsbegrepet. En funksjon av flere variabler fungerer akkurat som en funksjon av én variabel, bortsett fra at vi mater inn flere tall for å få ut ett tall.

:::{admonition} Def. 4.12 - Funksjon av flere variabler
:class: warning

En **funksjon av to variabler** $f$ er en regel som tildeler hvert tallpar $(x_1,x_2)$ nøyaktig ett tall $f(x_1,x_2)$. Vi skriver

$f:\mathbb{R}^2 \to \mathbb{R}$

Mer generelt tildeler en **funksjon av $n$ variabler** hvert punkt $(x_1, x_2, \ldots, x_n)$ nøyaktig ett tall $f(x_1, x_2, \ldots, x_n)$, og vi skriver 

$f:\mathbb{R}^n \to \mathbb{R}$.
:::

Legg merke til at outputen fortsatt er ett enkelt tall - det er bare inputen som består av flere tall. Kravet fra Def. 4.1, at hver input gir nøyaktig én output, er akkurat det samme. Vi setter inn verdier på nøyaktig samme måte som før, bare med flere tall om gangen.

:::{card} 
**Eksempel**
^^^
La $f(x,y)=x^2+3y$. Regn ut $f(2,1)$ og $f(0,-4)$.

Vi setter inn verdiene for $x$ og $y$, i den rekkefølgen de står.

For $f(2,1)$ setter vi $x=2$ og $y=1$:

$f(2,1)=2^2+3\cdot 1=4+3=7$

For $f(0,-4)$ setter vi $x=0$ og $y=-4$:

$f(0,-4)=0^2+3\cdot(-4)=0-12=-12$

Merk at rekkefølgen betyr noe: $f(2,1)$ er ikke det samme som $f(1,2)$.
:::

Definisjonsmengden til en flervariabel funksjon er mengden av alle tallpar $(x,y)$ vi har lov til å sette inn - altså alle par der regneuttrykket gir mening. Prinsippet er det samme som i én variabel: vi må unngå å dele på null, ta kvadratrot av negative tall, eller ta logaritmen av noe som ikke er positivt.

:::{card} 
**Eksempel**
^^^
Hva er definisjonsmengden til $f(x,y)=\dfrac{1}{x-y}$?

Uttrykket gir mening for alle tallpar $(x,y)$ så lenge vi ikke deler på null. Nevneren blir null når $x-y=0$, altså når $x=y$.

Definisjonsmengden er derfor alle tallpar $(x,y)$ der $x \neq y$. Geometrisk betyr det hele planet unntatt den rette linjen $y=x$.
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Idealgassloven er et av de mest kjente eksemplene på en flervariabel funksjon i kjemien. Løser vi $PV=nRT$ for trykket, får vi

$P(V,T)=\frac{nRT}{V}$

Her er trykket $P$ en funksjon av *to* variabler: volumet $V$ og temperaturen $T$ (mens $n$ og $R$ er konstanter). Definisjonsmengden er fysisk begrenset: volumet må være positivt ($V>0$, ellers deler vi på null), og temperaturen målt i kelvin kan ikke være negativ ($T \geq 0$). Setter vi inn et bestemt volum og en bestemt temperatur, får vi ut nøyaktig ett trykk - helt i tråd med Def. 4.12.
:::

En funksjon $f(x)$ av én variabel tegner vi som en kurve i planet: for hver $x$ langs den vannrette aksen markerer vi høyden $f(x)$. En funksjon $f(x,y)$ av to variabler trenger derimot to akser bare for inputen - én for $x$ og én for $y$ - og en tredje akse for outputen $f(x,y)$. Grafen blir derfor ikke en kurve, men en **flate** i det tredimensjonale rommet. Du kan tenke på den som et landskap: for hvert punkt $(x,y)$ på "bakken" forteller flaten hvor høyt over bakken funksjonen ligger.

```python
vis_sim("Simuleringer/flervariabel_flate.html")
```

Tredimensjonale flater er fine å se på, men vanskelige å tegne og lese av nøyaktig på et flatt ark. Derfor bruker kjemikere og matematikere ofte et triks som du kjenner igjen fra turkart: **nivåkurver**. I stedet for å tegne selve landskapet, tegner vi kurvene som forbinder alle punkter med samme høyde. På et turkart er dette høydekotene; for en funksjon $f(x,y)$ er en nivåkurve alle punkter $(x,y)$ der funksjonen har én bestemt verdi.

```python
vis_sim("Simuleringer/flervariabel_nivakurver.html")
```

:::{card} 
**Eksempel**
^^^
Beskriv nivåkurvene til $f(x,y)=x^2+y^2$.

En nivåkurve består av alle punkter $(x,y)$ der funksjonen har én fast verdi $c$. Vi setter derfor $f(x,y)=c$:

$x^2+y^2=c$

For $c>0$ kjenner vi igjen dette som ligningen for en sirkel med sentrum i origo og radius $\sqrt{c}$. Nivåkurvene er altså konsentriske sirkler: jo større verdi $c$, jo større sirkel. For $c=0$ krymper sirkelen til det enkelte punktet $(0,0)$, og for $c<0$ finnes det ingen punkter (en sum av kvadrater kan ikke bli negativ).
:::

Nå som vi kan beskrive funksjoner av flere variabler, dukker et naturlig spørsmål opp: hvordan deriverer vi dem? Hvis trykket avhenger av både volum og temperatur, hva mener vi da med "endringsraten"? Svaret er at vi må derivere med hensyn på én variabel om gangen, og holde de andre konstante. Dette leder oss til partiellderiverte i neste delkapittel.

:::{admonition} Underveisoppgave 4.6
:class: tip

La $f(x,y)=2x+y^2$.

a) Regn ut $f(3,2)$ og $f(-1,4)$.

b) Er $f(1,2)$ det samme som $f(2,1)$? Regn ut begge.

c) Hva er definisjonsmengden til $g(x,y)=\dfrac{1}{\sqrt{x}}+y$?
:::

:::{admonition} Løsning 4.6
:class: tip, dropdown

a) Vi setter inn verdiene i rekkefølge.

$f(3,2)=2\cdot 3+2^2=6+4=10$

$f(-1,4)=2\cdot(-1)+4^2=-2+16=14$

b) Vi regner ut begge:

$f(1,2)=2\cdot 1+2^2=2+4=6$

$f(2,1)=2\cdot 2+1^2=4+1=5$

Siden $6 \neq 5$, er $f(1,2)$ **ikke** det samme som $f(2,1)$. Rekkefølgen på variablene betyr noe.

c) For at $g(x,y)=\dfrac{1}{\sqrt{x}}+y$ skal gi mening, må uttrykket under kvadratroten være positivt (vi kan verken ta rot av et negativt tall eller dele på null, og $\sqrt{0}=0$ gir null i nevneren). Vi trenger derfor $x>0$, mens $y$ kan være hvilket som helst reelt tall. Definisjonsmengden er alle tallpar $(x,y)$ med $x>0$.
:::

### 4.7 Partiellderiverte

I forrige delkapittel så vi at trykket i en gass kan avhenge av både volum og temperatur gjennom $P(V,T)$. Men da dukker et problem opp: hva mener vi egentlig med "endringsraten" til en funksjon av flere variabler? For en funksjon $f(x)$ av én variabel var svaret entydig - den deriverte $f'(x)$ målte hvor raskt $f$ endret seg. For $f(x,y)$ er det ikke like klart: endrer vi $x$, eller $y$, eller begge samtidig? Løsningen er enkel og elegant: vi ser på én variabel om gangen, og holder de andre helt i ro. Dette kaller vi å **derivere partielt**.

Ideen er å velge én variabel å derivere med hensyn på, og late som om alle de andre variablene bare er konstanter. Deriverer vi med hensyn på $x$, behandler vi $y$ som et fast tall - og bruker ellers akkurat de samme derivasjonsreglene som i delkapittel 4.4.

:::{admonition} Def. 4.13 - Partiellderivert
:class: warning

La $f(x,y)$ være en funksjon av to variabler. Den **partiellderiverte av $f$ med hensyn på $x$** er

$\frac{\partial f}{\partial x} = \lim_{h \to 0} \frac{f(x+h,\,y)-f(x,\,y)}{h}$

der vi holder $y$ konstant. Tilsvarende er den **partiellderiverte med hensyn på $y$**

$\frac{\partial f}{\partial y} = \lim_{h \to 0} \frac{f(x,\,y+h)-f(x,\,y)}{h}$

der vi holder $x$ konstant.
:::

Legg merke til det krøllete symbolet $\partial$ ("del" eller "partiell-d"). Vi bruker det i stedet for det vanlige $d$ nettopp for å understreke at funksjonen har flere variabler, og at vi bare deriverer med hensyn på én av dem. Sammenlign gjerne med definisjonen av den vanlige deriverte i Def. 4.8 - den eneste forskjellen er at den andre variabelen står fast som en tilskuer. Vi bruker også skrivemåtene $f_x$ og $f_y$ som kortform for $\frac{\partial f}{\partial x}$ og $\frac{\partial f}{\partial y}$.

:::{card} 
**Eksempel**
^^^
Finn begge de partiellderiverte til $f(x,y)=x^2 y + 3x y^3$.

**Med hensyn på $x$** behandler vi $y$ som en konstant. Da er $x^2 y$ et konstant-tall ($y$) ganget med $x^2$, og $3xy^3$ er en konstant ($3y^3$) ganget med $x$:

$\frac{\partial f}{\partial x} = 2x y + 3y^3$

**Med hensyn på $y$** behandler vi $x$ som en konstant. Nå er $x^2 y$ en konstant ($x^2$) ganget med $y$, og $3xy^3$ er en konstant ($3x$) ganget med $y^3$:

$\frac{\partial f}{\partial y} = x^2 + 9x y^2$

Legg merke til at leddet $x^2 y$ ga ulikt bidrag i de to tilfellene, alt etter hvilken variabel vi så på som konstant.
:::

:::{card} 
**Eksempel**
^^^
Finn de partiellderiverte til $g(x,y)=\sin{(xy)}$.

**Med hensyn på $x$** holder vi $y$ konstant. Den indre funksjonen er $xy$, som deriveres til $y$ (siden $y$ er konstant). Ved kjerneregelen:

$\frac{\partial g}{\partial x} = \cos{(xy)} \cdot y = y\cos{(xy)}$

**Med hensyn på $y$** holder vi $x$ konstant. Nå deriveres den indre funksjonen $xy$ til $x$:

$\frac{\partial g}{\partial y} = \cos{(xy)} \cdot x = x\cos{(xy)}$
:::

Akkurat som vi kunne derivere flere ganger i én variabel (Def. 4.8 om høyere-ordens deriverte), kan vi derivere partielt flere ganger. Nå får vi flere muligheter: vi kan derivere to ganger med hensyn på $x$, to ganger med hensyn på $y$, eller én gang med hver. De to siste kalles **blandede partiellderiverte**.

:::{admonition} Def. 4.14 - Andreordens partiellderiverte
:class: warning

De **andreordens partiellderiverte** til $f(x,y)$ er

$\frac{\partial^2 f}{\partial x^2} = \frac{\partial}{\partial x}\left(\frac{\partial f}{\partial x}\right), \qquad \frac{\partial^2 f}{\partial y^2} = \frac{\partial}{\partial y}\left(\frac{\partial f}{\partial y}\right)$

De **blandede partiellderiverte** er

$\frac{\partial^2 f}{\partial x\, \partial y} = \frac{\partial}{\partial x}\left(\frac{\partial f}{\partial y}\right), \qquad \frac{\partial^2 f}{\partial y\, \partial x} = \frac{\partial}{\partial y}\left(\frac{\partial f}{\partial x}\right)$
:::

:::{card} 
**Eksempel**
^^^
For $f(x,y)=x^2 y + 3x y^3$ fant vi $\frac{\partial f}{\partial x}=2xy+3y^3$ og $\frac{\partial f}{\partial y}=x^2+9xy^2$. Finn de to blandede partiellderiverte.

For $\frac{\partial^2 f}{\partial x\, \partial y}$ deriverer vi $\frac{\partial f}{\partial y}=x^2+9xy^2$ med hensyn på $x$ (hold $y$ konstant):

$\frac{\partial^2 f}{\partial x\, \partial y} = 2x + 9y^2$

For $\frac{\partial^2 f}{\partial y\, \partial x}$ deriverer vi $\frac{\partial f}{\partial x}=2xy+3y^3$ med hensyn på $y$ (hold $x$ konstant):

$\frac{\partial^2 f}{\partial y\, \partial x} = 2x + 9y^2$

Merk at de to blandede ble like! Dette er ingen tilfeldighet.
:::

At de blandede partiellderiverte ble like i eksempelet over, er et generelt resultat: for de aller fleste funksjoner vi møter i praksis, spiller rekkefølgen vi deriverer i ingen rolle. Dette kalles gjerne Clairauts teorem, og gjelder så lenge de blandede partiellderiverte er kontinuerlige. For oss betyr det en grei sjekk: regner du ut begge blandede og får ulikt svar, har du sannsynligvis gjort en regnefeil.

De to partiellderiverte forteller hver for seg hvor bratt flaten heller i $x$-retningen og i $y$-retningen. Ofte er det nyttig å samle dem i ett objekt. Vi setter dem sammen til en vektor, som vi kaller gradienten.

:::{admonition} Def. 4.15 - Gradienten
:class: warning

**Gradienten** til en funksjon $f(x,y)$ er vektoren av de partiellderiverte,

$\nabla f = \left( \frac{\partial f}{\partial x}, \; \frac{\partial f}{\partial y} \right)$

Symbolet $\nabla$ leses "nabla". For en funksjon av $n$ variabler har gradienten tilsvarende én komponent per variabel.
:::

Gradienten har en vakker geometrisk tolkning: i et gitt punkt peker $\nabla f$ i den retningen der funksjonen vokser raskest, og lengden av vektoren forteller hvor bratt den vokser. Står du i fjellandskapet fra introduksjonsspørsmålene, peker gradienten rett oppover den bratteste stigningen. Et beslektet poeng, som er nyttig i kjemien: gradienten står alltid vinkelrett på nivåkurvene fra delkapittel 4.6.

:::{card} 
**Eksempel**
^^^
Finn gradienten til $f(x,y)=x^2+y^2$, og regn den ut i punktet $(1,2)$.

Vi finner først de to partiellderiverte:

$\frac{\partial f}{\partial x} = 2x, \qquad \frac{\partial f}{\partial y} = 2y$

Gradienten er dermed

$\nabla f = (2x, \; 2y)$

I punktet $(1,2)$ blir den

$\nabla f(1,2) = (2\cdot 1, \; 2\cdot 2) = (2, \; 4)$

Denne vektoren peker i retningen der $f$ vokser raskest fra punktet $(1,2)$ - altså rett bort fra origo, som stemmer med at nivåkurvene her er sirkler om origo.
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Partiellderiverte er selve språket i termodynamikken, men kjemikere skriver dem litt annerledes enn matematikere. Siden en størrelse kan avhenge av mange variabler, er det avgjørende å presisere *hvilke* variabler som holdes konstante. Derfor skriver man den konstante variabelen som en senket indeks utenfor en parentes.

For idealgassen $P(V,T)=\dfrac{nRT}{V}$ blir de to partiellderiverte

$\left(\frac{\partial P}{\partial T}\right)_V = \frac{nR}{V} \qquad \text{og} \qquad \left(\frac{\partial P}{\partial V}\right)_T = -\frac{nRT}{V^2}$

Notasjonen $\left(\frac{\partial P}{\partial T}\right)_V$ leses "den partiellderiverte av $P$ med hensyn på $T$, med $V$ holdt konstant". Dette er nøyaktig samme operasjon som i Def. 4.13 - indeksen $V$ er bare en eksplisitt påminnelse om hva som står fast. Slike uttrykk dukker opp overalt i termodynamikken, for eksempel i sammenhengen mellom varmekapasitet og indre energi, $C_V=\left(\frac{\partial U}{\partial T}\right)_V$.
:::

Med partiellderiverte og gradienten på plass har vi verktøyene for å studere hvordan flervariable funksjoner endrer seg. Et naturlig neste spørsmål er hvordan vi finner topp- og bunnpunkter for slike funksjoner - særlig når de i tillegg er underlagt en betingelse, slik som å minimere energi under bevaring av masse. Det leder oss til Lagranges multiplikatormetode i neste delkapittel.

:class: tip

La $f(x,y)=x^3 y - 2x y^2$.

a) Finn $\dfrac{\partial f}{\partial x}$ og $\dfrac{\partial f}{\partial y}$.

b) Finn den blandede partiellderiverte $\dfrac{\partial^2 f}{\partial x\, \partial y}$.

c) Skriv opp gradienten $\nabla f$, og regn den ut i punktet $(1,1)$.

:::{admonition} Løsning 4.7
:class: tip, dropdown

a) Med hensyn på $x$ (hold $y$ konstant): $\dfrac{\partial f}{\partial x} = 3x^2 y - 2y^2$. Med hensyn på $y$ (hold $x$ konstant): $\dfrac{\partial f}{\partial y} = x^3 - 4xy$.

b) Vi deriverer $\dfrac{\partial f}{\partial y}=x^3-4xy$ med hensyn på $x$: $\dfrac{\partial^2 f}{\partial x\, \partial y} = 3x^2 - 4y$.

c) Gradienten er $\nabla f = \left(3x^2 y - 2y^2, \; x^3 - 4xy\right)$. I punktet $(1,1)$: $\nabla f(1,1) = (3\cdot 1 - 2, \; 1 - 4) = (1, \; -3)$.

### 4.8 Dobbel-og trippelintegraler

I delkapittel 4.5 så vi at det bestemte integralet $\int_a^b f(x)\,dx$ måler arealet under grafen til en funksjon av én variabel. Men nå kan vi funksjoner av flere variabler (4.6), og et naturlig spørsmål melder seg: hva blir integralet av $f(x,y)$? Der enkeltintegralet summerte opp arealet av uendelig mange tynne striper, skal vi nå summere opp volumet av uendelig mange tynne søyler - og resultatet blir volumet under en flate. Utvider vi til tre variabler, kan vi regne ut ting som massen eller ladningen fordelt i et romområde. For kjemistudenter er dette selve verktøyet bak normalisering av orbitaler, som vi kommer til mot slutten.

Vi starter med to variabler. Akkurat som vi delte intervallet $[a,b]$ i tynne striper i 4.5, deler vi nå et område i $xy$-planet inn i et rutenett av små rektangler, hvert med areal $\Delta A = \Delta x\,\Delta y$. Over hver rute reiser vi en søyle med høyde $f(x,y)$, altså volum $f(x,y)\,\Delta A$. Legger vi sammen alle søylene og lar rutene bli uendelig små, får vi dobbeltintegralet.

:::{admonition} Def. 4.16 - Dobbeltintegralet
:class: warning

**Dobbeltintegralet** av $f(x,y)$ over et område $D$ i $xy$-planet er grenseverdien av søylesummene når rutene blir uendelig små:

$\iint_D f(x,y)\,dA = \lim_{\Delta A \to 0} \sum_i f(x_i,y_i)\,\Delta A$

Når $f(x,y)\geq 0$, er dobbeltintegralet **volumet** mellom flaten $z=f(x,y)$ og området $D$ i $xy$-planet.
:::

Definisjonen er fin å forstå, men vi vil ikke summere uendelig mange søyler for hånd. Heldigvis finnes det en enkel oppskrift, helt analog til fundamentalteoremet: vi integrerer én variabel om gangen. Dette kalles et **iterert integral**. Vi holder den ene variabelen konstant (akkurat som ved partiellderivasjon i 4.7), integrerer med hensyn på den andre, og integrerer så resultatet med hensyn på den første.

:::{admonition} Teorem 4.10 - Fubinis teorem
:class: note

La $f(x,y)$ være kontinuerlig på rektangelet $D=[a,b]\times[c,d]$. Da kan dobbeltintegralet regnes ut som et iterert integral, og rekkefølgen spiller ingen rolle:

$\iint_D f(x,y)\,dA = \int_a^b \left( \int_c^d f(x,y)\,dy \right) dx = \int_c^d \left( \int_a^b f(x,y)\,dx \right) dy$
:::

:::{card} 
**Eksempel**
^^^
Regn ut $\iint_D xy \, dA$ over rektangelet $D=[0,2]\times[0,1]$.

Vi regner det som et iterert integral, og starter innenfra med hensyn på $x$ (mens $y$ holdes konstant):

$\int_0^2 xy \, dx = y\int_0^2 x \, dx = y\Big[\frac{x^2}{2}\Big]_0^2 = y \cdot 2 = 2y$

Så integrerer vi resultatet med hensyn på $y$:

$\int_0^1 2y \, dy = \Big[y^2\Big]_0^1 = 1$

Altså er $\iint_D xy \, dA = 1$. Ifølge Fubini hadde vi fått samme svar om vi integrerte i motsatt rekkefølge - prøv gjerne selv.
:::

Steget videre til tre variabler er nå kort. I stedet for å dele opp et flatt område i ruter, deler vi opp et romområde $E$ i små bokser med volum $\Delta V = \Delta x\,\Delta y\,\Delta z$. Vi summerer $f(x,y,z)\,\Delta V$ over alle boksene og lar dem bli uendelig små.

:::{admonition} Def. 4.17 - Trippelintegralet
:class: warning

**Trippelintegralet** av $f(x,y,z)$ over et romområde $E$ er

$\iiint_E f(x,y,z)\,dV = \lim_{\Delta V \to 0} \sum_i f(x_i,y_i,z_i)\,\Delta V$

Setter vi $f=1$, gir integralet **volumet** av $E$. Er $f$ en tetthet (masse per volum), gir integralet den totale **massen**.
:::

Fubinis teorem gjelder også her: vi integrerer én variabel om gangen, tre ganger. Over en boks $[a_1,b_1]\times[a_2,b_2]\times[a_3,b_3]$ blir det tre nøstede integraler. Men mange av de viktigste anvendelsene i kjemien handler ikke om bokser - de handler om *kuler*, som når vi ser på et atom. Da lønner det seg å bytte koordinatsystem.

Et hydrogenatom er kulesymmetrisk, ikke boksformet. Å beskrive det med $x$, $y$ og $z$ blir tungvint. I stedet bruker vi **sfæriske koordinater**, der vi angir et punkt ved hvor langt det er fra origo og i hvilken retning: avstanden $r$ fra origo, vinkelen $\theta$ ned fra $z$-aksen (polarvinkelen), og vinkelen $\phi$ rundt $z$-aksen (asimutvinkelen).

:::{admonition} Def. 4.18 - Sfæriske koordinater
:class: warning

Et punkt $(x,y,z)$ kan beskrives ved sfæriske koordinater $(r,\theta,\phi)$ gjennom

$x = r\sin{\theta}\cos{\phi}, \qquad y = r\sin{\theta}\sin{\phi}, \qquad z = r\cos{\theta}$

der $r \geq 0$ er avstanden fra origo, $0 \leq \theta \leq \pi$ er vinkelen fra den positive $z$-aksen, og $0 \leq \phi < 2\pi$ er vinkelen i $xy$-planet.
:::

Når vi bytter koordinater, kan vi ikke bare bytte ut $dV = dx\,dy\,dz$ med $dr\,d\theta\,d\phi$ - det ville gitt feil svar. Grunnen er at en liten "boks" i sfæriske koordinater ikke har sidene $dr$, $d\theta$ og $d\phi$ i lengde. Bare $dr$ er en ekte lengde. De to vinklene må ganges opp med en radius for å bli lengder: en liten endring $d\theta$ svarer til en bue av lengde $r\,d\theta$, og en liten endring $d\phi$ svarer til en bue av lengde $r\sin{\theta}\,d\phi$ (faktoren $\sin\theta$ kommer av at sirklene rundt $z$-aksen er mindre nær polene enn ved ekvator). Volumet av den lille boksen blir derfor produktet av de tre lengdene.

:::{admonition} Def. 4.19 - Volumelementet i sfæriske koordinater
:class: warning

I sfæriske koordinater er volumelementet

$dV = r^2 \sin{\theta} \, dr \, d\theta \, d\phi$

Faktoren $r^2\sin{\theta}$ kalles Jacobideterminanten for overgangen, og retter opp for at vinkelkoordinatene ikke er lengder.
:::

:::{card} 
**Eksempel**
^^^
Bruk sfæriske koordinater til å regne ut volumet av en kule med radius $R$.

Volumet er trippelintegralet av $f=1$ over hele kula. I sfæriske koordinater dekker vi kula ved å la $r$ gå fra $0$ til $R$, $\theta$ fra $0$ til $\pi$, og $\phi$ fra $0$ til $2\pi$. Vi bruker $dV=r^2\sin{\theta}\,dr\,d\theta\,d\phi$:

$V = \int_0^{2\pi}\int_0^{\pi}\int_0^{R} r^2\sin{\theta}\,dr\,d\theta\,d\phi$

De tre integralene kan skilles fra hverandre siden faktorene er uavhengige:

$V = \left(\int_0^{R} r^2\,dr\right)\left(\int_0^{\pi}\sin{\theta}\,d\theta\right)\left(\int_0^{2\pi}d\phi\right)$

Vi regner ut hver for seg:

$\int_0^{R} r^2\,dr = \frac{R^3}{3}, \qquad \int_0^{\pi}\sin{\theta}\,d\theta = \Big[-\cos{\theta}\Big]_0^{\pi} = 2, \qquad \int_0^{2\pi}d\phi = 2\pi$

Til sammen: $V = \dfrac{R^3}{3}\cdot 2 \cdot 2\pi = \dfrac{4}{3}\pi R^3$, som er den velkjente formelen for kulevolumet.
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Her kommer selve grunnen til at kjemistudenter trenger trippelintegraler i sfæriske koordinater: **normalisering av orbitaler**. En orbital $\psi$ er en bølgefunksjon, og $|\psi|^2$ er sannsynlighetstettheten for å finne elektronet. Siden elektronet må være *et eller annet sted*, må sannsynligheten summert over hele rommet bli nøyaktig $1$:

$\iiint_{\text{hele rommet}} |\psi|^2 \, dV = 1$

Ta hydrogenets 1s-orbital, som er kulesymmetrisk og bare avhenger av $r$:

$\psi_{1s} = \frac{1}{\sqrt{\pi a_0^3}}\, e^{-r/a_0}$

der $a_0$ er Bohr-radien. Vi sjekker normaliseringen i sfæriske koordinater. Siden $|\psi|^2$ ikke avhenger av vinklene, skiller integralene seg pent fra hverandre:

$\iiint |\psi_{1s}|^2 \, dV = \frac{1}{\pi a_0^3}\left(\int_0^{\infty} e^{-2r/a_0}\,r^2\,dr\right)\left(\int_0^{\pi}\sin{\theta}\,d\theta\right)\left(\int_0^{2\pi}d\phi\right)$

Vinkelintegralene kjenner vi igjen fra kule-eksempelet: de gir $2$ og $2\pi$. Radialintegralet er et standardintegral (som kan løses med delvis integrasjon fra 4.5, brukt to ganger):

$\int_0^{\infty} r^2 e^{-2r/a_0}\,dr = \frac{a_0^3}{4}$

Setter vi alt sammen:

$\frac{1}{\pi a_0^3}\cdot \frac{a_0^3}{4}\cdot 2 \cdot 2\pi = \frac{1}{\pi a_0^3}\cdot \pi a_0^3 = 1$

Integralet blir nøyaktig $1$ - orbitalen er riktig normalisert. Legg merke til hvor avgjørende faktoren $r^2$ fra volumelementet var: uten den ville integralet ikke gått opp. Nettopp derfor er $r^2\sin\theta$ ikke en teknisk detalj, men helt sentralt i kvantekjemien.
:::

Med dobbel- og trippelintegraler, og sfæriske koordinater, har vi fått verktøyene til å integrere over flater og romområder - og til å normalisere orbitaler, som er uunnværlig i kvantekjemien. Dette runder av kalkulusdelen: fra funksjoner og grenser, via derivasjon og integrasjon i én variabel, til flervariabel derivasjon og integrasjon.

### 4.9 Lagrange multiplikatormetode

Gjennom kalkulusdelen har vi bygget opp verktøy for å beskrive hvordan funksjoner endrer seg. Ofte er målet vårt å finne det *største* eller *minste* mulige - maksimere et utbytte, minimere en energi. Men i praksis er vi sjelden helt frie: vi optimerer under en **betingelse**. En kjemiker vil kanskje maksimere entropien, men bare blant tilstander som har en gitt total energi. Vi søker altså ekstremalverdien til en funksjon $f(x,y)$, men bare blant de punktene $(x,y)$ som oppfyller en betingelse på formen $g(x,y)=c$. Lagranges multiplikatormetode er et elegant verktøy for nettopp dette, og bygger direkte på gradienten fra delkapittel 4.7.

La oss tenke geometrisk. Betingelsen $g(x,y)=c$ beskriver en kurve i planet - den tillatte "stien" vi kan bevege oss langs. Funksjonen $f$ har sine egne nivåkurver, én for hver verdi $f$ kan ta. Når vi går langs betingelseskurven, krysser vi stadig nye nivåkurver for $f$, og verdien til $f$ endrer seg. Så lenge betingelseskurven *krysser* en nivåkurve, kan vi fortsette å gå og finne en enda høyere (eller lavere) verdi. Ekstremalverdien inntreffer først i det øyeblikket betingelseskurven bare *tangerer* en nivåkurve - da har vi presset $f$ så langt vi kan uten å forlate stien.

I dette tangeringspunktet peker de to kurvenes normaler i samme retning. Fra delkapittel 4.7 vet vi at gradienten står vinkelrett på nivåkurvene. Både $\nabla f$ (normal på $f$ sin nivåkurve) og $\nabla g$ (normal på betingelseskurven) må derfor peke langs samme linje - de er parallelle. To vektorer er parallelle nøyaktig når den ene er et tall ganget med den andre, og det er hele kjernen i metoden.

:::{admonition} Teorem 4.11 - Lagranges multiplikatormetode
:class: note

La $f(x,y)$ og $g(x,y)$ være deriverbare funksjoner. Dersom $f$ har en ekstremalverdi i et punkt $(x,y)$ under betingelsen $g(x,y)=c$, så finnes det et tall $\lambda$ slik at

$\nabla f = \lambda \nabla g$

Tallet $\lambda$ kalles en **Lagrange-multiplikator**.
:::

Skriver vi ut vektorligningen $\nabla f = \lambda \nabla g$ komponent for komponent, og tar med selve betingelsen, får vi et system på tre ligninger med tre ukjente ($x$, $y$ og $\lambda$):

$\frac{\partial f}{\partial x} = \lambda \frac{\partial g}{\partial x}$

$\frac{\partial f}{\partial y} = \lambda \frac{\partial g}{\partial y}$

$g(x,y) = c$

Oppskriften blir dermed: regn ut de fire partiellderiverte, sett opp de tre ligningene, og løs systemet. Punktene vi finner er kandidatene til ekstremalverdi. Multiplikatoren $\lambda$ er som regel bare et mellomledd vi ikke bryr oss om til slutt - men i fysikk og kjemi viser den seg ofte å ha en dyp betydning helt selv.

:::{card} 
**Eksempel**
^^^
Finn det største mulige produktet $f(x,y)=xy$ når $x+y=10$.

Her er betingelsen $g(x,y)=x+y$ med $c=10$. Vi regner ut de partiellderiverte:

$\frac{\partial f}{\partial x}=y, \quad \frac{\partial f}{\partial y}=x, \quad \frac{\partial g}{\partial x}=1, \quad \frac{\partial g}{\partial y}=1$

Ligningssystemet $\nabla f = \lambda \nabla g$ sammen med betingelsen blir

$y = \lambda \cdot 1$

$x = \lambda \cdot 1$

$x+y=10$

Fra de to første ligningene ser vi at $x=\lambda$ og $y=\lambda$, altså $x=y$. Setter vi dette inn i betingelsen, får vi $x+x=10$, så $x=5$ og dermed $y=5$.

Det største produktet er derfor $f(5,5)=5 \cdot 5 = 25$.
:::

:::{card} 
**Eksempel**
^^^
Et rektangel skal ha omkrets $20$ cm. Hvilke sidelengder gir størst areal?

La sidene være $x$ og $y$. Vi maksimerer arealet $f(x,y)=xy$ under betingelsen $g(x,y)=2x+2y=20$.

De partiellderiverte er

$\frac{\partial f}{\partial x}=y, \quad \frac{\partial f}{\partial y}=x, \quad \frac{\partial g}{\partial x}=2, \quad \frac{\partial g}{\partial y}=2$

Ligningssystemet blir

$y = 2\lambda$

$x = 2\lambda$

$2x+2y=20$

De to første gir $x=y$. Setter vi inn i betingelsen: $2x+2x=20$, altså $x=5$ og $y=5$.

Rektangelet med størst areal er altså et **kvadrat** med sider $5$ cm og areal $25$ cm². Dette bekrefter intuisjonen fra introduksjonsspørsmålet - det jevneste valget er det beste.
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Lagranges metode er ikke bare et regnetriks - den er selve fundamentet for **Boltzmann-fordelingen**, som vi allerede møtte i delkapittel 1.

Tenk deg et system med partikler fordelt på energinivåer $\varepsilon_1, \varepsilon_2, \ldots$, med $n_i$ partikler på nivå $i$. Antall mikrotilstander som gir en bestemt fordeling er $W=\dfrac{N!}{n_1!\,n_2!\cdots}$ (kjenner du igjen fakultetene fra delkapittel 2?). Naturen søker den fordelingen som maksimerer $W$ - eller, mer bekvemt, $\ln{W}$. Men vi kan ikke maksimere fritt: to betingelser må holdes fast.

$\sum_i n_i = N \qquad \text{(partikkeltallet er bevart)}$

$\sum_i n_i \varepsilon_i = E \qquad \text{(den totale energien er bevart)}$

Dette er nøyaktig et Lagrange-problem, bare med *to* betingelser i stedet for én - da trenger vi to multiplikatorer, som vi kaller $\alpha$ og $\beta$. Vi maksimerer altså

$\ln{W} - \alpha\left(\sum_i n_i - N\right) - \beta\left(\sum_i n_i \varepsilon_i - E\right)$

Deriverer vi med hensyn på hver $n_i$ og setter lik null (samme prinsipp som $\nabla f = \lambda \nabla g$), faller det ut at

$n_i \propto e^{-\beta \varepsilon_i}$

Dette er Boltzmann-fordelingen. Og multiplikatoren $\beta$ - det "uinteressante mellomleddet" - viser seg å være $\beta=\dfrac{1}{k_B T}$. Slik dukker temperaturen opp helt av seg selv, som en Lagrange-multiplikator knyttet til energibevaring. Metoden du nettopp lærte på $f=xy$ er altså den samme som gir oss fordelingen bak all statistisk termodynamikk.
:::

Med Lagranges multiplikatormetode har vi rundet av kalkulusdelen. Vi har gått fra funksjoner av én variabel, via grenser, derivasjon og integrasjon, til funksjoner av flere variabler, partiellderiverte, flerdimensjonal integrasjon og til slutt optimering under betingelser. Underveis har vi sett hvordan hvert verktøy har en direkte anvendelse i kjemien - fra termodynamisk arbeid og normalisering av orbitaler til selve Boltzmann-fordelingen som styrer hvordan naturen fordeler energi.

b) Finn den største og minste verdien av $f(x,y)=2x+3y$ på sirkelen $x^2+y^2=13$.

$2x = \lambda \cdot 1$

$2y = \lambda \cdot 2$

Største verdi er altså $13$ i punktet $(2,3)$, og minste verdi er $-13$ i punktet $(-2,-3)$.
:::

### 4.10 Stirlings tilnærming

**Stirlings tilnærming** brukes til å estimere fakultet av store tall. Vi skal se på hvordan vi kan estimere den naturlige logaritmen til fakultetet av et stort tall $n$, altså $\ln{n!}$.

La $n$ være et stort tall. Fra definisjonen av fakultet (Def. 2.3) har vi

$n!=n \cdot (n-1) \cdot (n-2) \cdot \dots \cdot 1$

Vi kan ta den naturlige logartimen på begge sider, slik at vi får

$\ln{n!}=\ln{\left( n \cdot (n-1) \cdot (n-2) \cdot \dots \cdot 1 \right)}$

Ved å bruke logaritmereglen $\log{(a \cdot b)}=\ln{(a)}+\ln{(b)}$ får vi

$\ln{n!}=\ln{(n)}+\ln{(n-1)}+\ln{(n-2)}+\dots+\ln{(1)}=\sum_{x=1}^{n} \ln{(x)}$

Vi kan approksimere denne summen til integralet av $\ln{(x)}$ fra $x=1$ til $x=n$, slik at

$\ln{n!}=\sum_{x=1}^{n} \ln{(x)}\approx \int_{1}^{n} \ln{(x)} \ dx$

Mellomregning:

$\int_{1}^{n} \ln{(x)} \ dx=\int_{1}^{n} 1 \cdot \ln{(x)} \ dx$

Vi bruker delvis integrasjon ved å la 

$u=\ln{(x)}$, $v'=1$ $\Rightarrow$ $u'=\frac{1}{x}$, $v=x$

og får

$\int_{1}^{n} 1 \cdot \ln{(x)} \ dx = [x \ln{(x)}]_{1}^{n}-\int_{1}^{n} \left( \frac{1}{x} \cdot x \right) \  dx=[x \ln{(x)}]_{1}^{n}-\int_{1}^{n} \  dx=[x \ln{(x)} -x]_{1}^{n}$

Altså, har vi fått Stirlings tilnærming

$\ln{n!}\approx [x \ln{(x)} -x]_{1}^{n}$

:::{admonition} Teorem 4.12 - Stirlings tilnærming
:class: note

For store $n$ kan $\ln{n!}$ tilnærmes ved å erstatte summen $\sum_{k=1}^{n}\ln{k}$ med et integral:

$\ln{n!}=\sum_{k=1}^{n}\ln{k} \approx \int_1^n \ln{x}\,dx = \big[x\ln{x}-x\big]_1^n = n\ln{n}-n+1$

Siden $1$-leddet er neglisjerbart for store $n$, skrives resultatet vanligvis som

$\ln{n!} \approx n\ln{n}-n$
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Stirlings tilnærming er uunnværlig i statistisk termodynamikk. Antall mikrotilstander for et system er gitt ved uttrykk som

$\Omega = \frac{N!}{n_1!\,n_2!\cdots}$

og entropien er $S=k_B\ln{\Omega}$. For ett mol partikler er $N\approx 6\times 10^{23}$, og da er $N!$ et tall så enormt at det er fullstendig uregnbart - verken for hånd eller på datamaskin.

Men vi trenger aldri $N!$ selv, bare $\ln{N!}$, fordi entropien inneholder en logaritme. Der kommer Stirling til unnsetning:

$\ln{\Omega}=\ln{N!}-\sum_i \ln{n_i!} \approx \left(N\ln{N}-N\right)-\sum_i\left(n_i\ln{n_i}-n_i\right)$

Dette gjør uttrykket både regnbart og deriverbart. Nettopp fordi $\ln{\Omega}$ nå er en glatt funksjon, kan vi maksimere den med Lagranges multiplikatormetode - og det er dét som gir Boltzmann-fordelingen.

I del 1 lærte vi hva sannsynlighet er, og i del 2 hvordan vi teller muligheter. Nå setter vi det sammen: statistikk handler om å beskrive hvordan sannsynligheten *fordeler seg* over alle mulige utfall, og om å oppsummere en slik fordeling med noen få tall - hvor den ligger, og hvor spredt den er.

### 5.1 Sannsynlighetsfordelinger 

Så langt har vi regnet ut sannsynligheten for enkelthendelser, én om gangen. Men ofte vil vi ha hele oversikten: hvordan fordeler sannsynligheten seg over *alle* utfallene samtidig? En slik oversikt kaller vi en **sannsynlighetsfordeling**.

Fra delkapittel 1.6 husker vi skillet mellom diskrete og kontinuerlige utfallsrom. Begge typer har sannsynlighetsfordelinger, men de beskrives på hver sin måte - og det er nettopp forskjellen mellom å telle og å måle som går igjen.

:::{admonition} Def. 5.1 - Stokastisk variabel
:class: warning

En **stokastisk variabel** $X$ er en størrelse som får sin verdi bestemt av utfallet av et tilfeldig eksperiment.

Variabelen kalles **diskret** dersom den kan anta et tellbart antall verdier, og **kontinuerlig** dersom den kan anta alle verdier i et intervall.
:::

Navnet høres mer skremmende ut enn det er. Kaster du to terninger, er summen av øynene en stokastisk variabel - vi vet ikke på forhånd hva den blir, men vi vet hvilke verdier den kan ta. Vi bruker stor bokstav $X$ for selve variabelen, og liten bokstav $x$ for en konkret verdi den kan anta. Skrivemåten $P(X=x)$ betyr altså «sannsynligheten for at variabelen får verdien $x$».

:::{admonition} OBS
:class: error
Stokastisk er bare et fancy ord for "tilfeldig".
:::

:::{admonition} Def. 5.2 - Sannsynlighetsfordeling
:class: warning

En **sannsynlighetsfordeling** beskriver hvordan sannsynligheten fordeler seg over de mulige verdiene til en stokastisk variabel $X$.

**Diskret variabel:** Fordelingen gis ved **punktsannsynligheten** $P(X=x)$, som må oppfylle

$P(X=x) \geq 0 \qquad \text{og} \qquad \sum_x P(X=x) = 1$

**Kontinuerlig variabel:** Fordelingen gis ved **tettheten** $f(x)$, som må oppfylle

$f(x) \geq 0 \qquad \text{og} \qquad \int_{-\infty}^{\infty} f(x)\,dx = 1$

Sannsynligheten for et intervall er da

$P(a \leq X \leq b) = \int_a^b f(x)\,dx$
:::

:::{admonition} Husk
:class: error

Kravene om ikke-negativitet og normering for hver av sannsynlighetsfordelingene kommer fra Aksiomer 1.1 og Teorem 1.1, henholdsvis.
:::

:::{admonition} OBS
:class: error
For en diskret variabel er $P(X=x)$ en ekte sannsynlighet - vi kan spørre «hvor sannsynlig er det at summen blir 7?» og få et tall.

For en kontinuerlig variabel er $f(x)$ **ikke** en sannsynlighet. Fra delkapittel 1.6 vet vi at hvert enkeltutfall har sannsynlighet $0$, så $P(X=x)=0$ uansett hvilken $x$ vi velger. Tettheten $f(x)$ er i stedet en sannsynlighet *per enhet* av $x$, og det er først når vi integrerer over et intervall at vi får en sannsynlighet ut. Verdien $f(3)=0{.}5$ betyr altså ikke at $P(X=3)=0{.}5$.

Her får integrasjonen en helt konkret betydning: arealet under grafen til $f$ **er** sannsynligheten.
:::

### 5.2 Forventningsverdi

En sannsynlighetsfordeling gir full oversikt, men det er ofte mer informasjon enn vi trenger. Vil du vite «hvor mye energi har et molekyl i gjennomsnitt?», er ikke svaret en hel tabell eller en kurve - du vil ha ett tall.

**Forventningsverdien** er nettopp det: et tyngdepunkt for fordelingen, verdien vi i gjennomsnitt lander på når eksperimentet gjentas mange ganger.

:::{admonition} Def. 5.3 - Forventningsverdi
:class: warning

**Forventningsverdien** til en stokastisk variabel $X$, skrevet $E[X]$, $\mu$ eller $\left< X \right>$, er tyngdepunktet i sannsynlighetsfordelingen.

**Diskret variabel:** Med punktsannsynligheten $P(X=x)$  er forventingsverdien gitt ved

$E[X] = \sum_x x \cdot P(X=x)$

**Kontinuerlig variabel:** Med tettheten $f(x)$ er forventningsverdien gitt ved

$E[X] = \int_{-\infty}^{\infty} x \cdot f(x)\,dx$
:::

Begge formlene sier det samme: vi ganger hver verdi med hvor sannsynlig den er, og legger sammen. Verdier med stor sannsynlighet trekker resultatet mot seg, mens sjeldne verdier bidrar lite. Dette er et **vektet gjennomsnitt**, der sannsynlighetene er vektene.

Overgangen mellom de to formlene følger samme mønster som i delkapittel 3.1: summen $\sum_x$ er byttet med integralet $\int$, og punktsannsynligheten $P(X=x)$ med tettheten $f(x)\,dx$.

Forventningsverdi er ikke et helt nytt begrep, men en generalisering av gjennomsnittet du kjenner fra før. Er alle $n$ utfall like sannsynlige, har hvert av dem sannsynlighet $\frac{1}{n}$, og da blir

$E[X] = \sum_x x \cdot \frac{1}{n} = \frac{1}{n}\sum_x x$

altså summen delt på antallet - det vanlige gjennomsnittet. Forventningsverdien lar i tillegg utfallene ha ulik vekt, som er nettopp det vi trenger når fordelingen ikke er jevn.

:::{card} 
**Eksempel**
^^^
Du triller en likevektet sekssidet terning. Hva er forventningsverdien til antall øyne?

Hver av verdiene $1$ til $6$ har punktsannsynlighet $\frac{1}{6}$, så

$E[X]=1 \cdot \frac{1}{6}+2 \cdot \frac{1}{6}+3 \cdot \frac{1}{6}+4 \cdot \frac{1}{6}+5 \cdot \frac{1}{6}+6 \cdot \frac{1}{6}=\frac{21}{6}=3{,}5$

Legg merke til at forventningsverdien er $3{,}5$ - en verdi terningen **aldri** kan vise. Det er ikke en feil. Forventningsverdien er et tyngdepunkt, ikke en spådom om neste kast. Triller du terningen mange tusen ganger, vil gjennomsnittet av kastene nærme seg $3{.}5$.
:::

:::{card} 
**Eksempel**
^^^
En variabel $X$ har punktsannsynlighetene

| $x$ | 0 | 1 | 2 | 3 | 4 | 5 |
|---|---|---|---|---|---|---|
| $P(X=x)$ | 0.05 | 0.15 | 0.30 | 0.25 | 0.15 | 0.10 |

Finn $E[X]$.

Vi ganger hver verdi med sin punktsannsynlighet og summerer:

$E[X]=0 \cdot 0{.}05+1 \cdot 0{.}15+2 \cdot 0{.}30+3 \cdot 0{.}25+4 \cdot 0{.}15+5 \cdot 0{.}10$

$E[X]=0+0{.}15+0{.}60+0{.}75+0{.}60+0{.}50=2{.}6$

Tyngdepunktet ligger på $2{.}6$, altså litt til høyre for den mest sannsynlige verdien $x=2$. Grunnen er at halen mot høyre trekker gjennomsnittet oppover.
:::

:::{card} 
**Eksempel**
^^^
En variabel har tettheten $f(x)=2x$ for $0 \leq x \leq 1$, og $0$ ellers. Finn $E[X]$.

Vi setter inn i Def. 5.3, og integrerer bare der tettheten er ulik null:

$E[X]=\int_0^1 x \cdot 2x \, dx = \int_0^1 2x^2\,dx = \Big[\frac{2x^3}{3}\Big]_0^1=\frac{2}{3}$

Forventningsverdien er $\frac{2}{3} \approx 0{.}67$. At den ligger over midtpunktet $0{.}5$ gir mening: tettheten vokser mot høyre, så de store verdiene er mest sannsynlige.
:::

:::{admonition} Teorem 5.1 - Linearitet
:class: note

For en stokastisk variabel $X$ og konstanter $a$ og $b$ gjelder

$E[aX+b]=a\,E[X]+b$

For to stokastiske variabler $X$ og $Y$ gjelder

$E[X+Y]=E[X]+E[Y]$

:::

Linearitet er et av de mest brukte verktøyene i praksis, fordi det lar oss regne på summer uten å finne fordelingen til summen først.

Ta forventningsverdien for summen av to terninger. Vi kunne satt opp hele fordelingen fra delkapittel 5.1 og summert elleve ledd - eller vi kan bruke at hver terning har $E[X]=3{.}5$, slik at summen har $E=3{.}5+3{.}5=7$. Det siste er langt raskere, og gir samme svar.

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Forventningsverdi er selve broen mellom mikroskopisk og makroskopisk beskrivelse. Den gjennomsnittlige energien til et system er forventningsverdien av energien over Boltzmann-fordelingen:
$\langle E \rangle = \sum_i E_i \, P(i) = \frac{\sum_i E_i \, e^{-E_i/k_BT}}{Q}$

Legg merke til notasjonen: i fysikk og kjemi skrives forventningsverdi gjerne med vinkelparenteser, $\langle E \rangle$, i stedet for $E[X]$. Det er nøyaktig samme størrelse - et vektet gjennomsnitt over alle mikrotilstander, der Boltzmann-faktoren er vekten.

Målte størrelser som trykk, temperatur og indre energi er alle forventningsverdier over enormt mange mikrotilstander. Det er nettopp derfor de virker stabile: selv om enkeltmolekyler oppfører seg tilfeldig, er gjennomsnittet over $10^{23}$ av dem praktisk talt konstant.
:::

:::{admonition} Def. 5.4 - Forventningsverdi av en funksjon
:class: warning

La $g$ være en funksjon og $X$ en stokastisk variabel. Forventningsverdien til $g(X)$ er

**Diskret variabel.** Med punktsannsynligheten $P(X=x)$ er

$E[g(X)] = \sum_x g(x) \cdot P(X=x)$

**Kontinuerlig variabel.** Med tettheten $f(x)$ er

$E[g(X)] = \int_{-\infty}^{\infty} g(x) \cdot f(x)\,dx$
:::

Legg merke til hva som endrer seg, og hva som ikke gjør det. Vi vekter fortsatt med sannsynligheten for hver verdi $x$ - vektene er de samme som før. Det eneste nye er at vi ikke summerer verdien $x$ selv, men verdien $g(x)$ som funksjonen gir.

Dette er verdt å understreke, fordi det er lett å tro at man må finne fordelingen til $g(X)$ først. Det trenger man ikke: vi bruker den opprinnelige fordelingen til $X$, og setter bare $g(x)$ inn i stedet for $x$. Setter vi $g(x)=x$, får vi tilbake Def. 3.3.

Det viktigste spesialtilfellet er $g(x)=x^2$, som gir

$E[X^2] = \sum_x x^2 \cdot P(X=x) \qquad \text{eller} \qquad E[X^2]=\int_{-\infty}^{\infty} x^2 f(x)\,dx$

Denne størrelsen dukker opp igjen i neste delkapittel.

:::{card} 
**Eksempel**
^^^
En likevektet terning har punktsannsynlighet $\frac{1}{6}$ for hver av verdiene $1$ til $6$. Finn $E[X^2]$.

Vi bruker Def. 5.4 med $g(x)=x^2$, altså kvadrerer vi hver verdi før vi vekter:

$E[X^2]=1^2 \cdot \frac{1}{6}+2^2 \cdot \frac{1}{6}+3^2 \cdot \frac{1}{6}+4^2 \cdot \frac{1}{6}+5^2 \cdot \frac{1}{6}+6^2 \cdot \frac{1}{6}$

$E[X^2]=\frac{1+4+9+16+25+36}{6}=\frac{91}{6}\approx 15{.}17$

Legg merke til at dette **ikke** er det samme som $\left(E[X]\right)^2=3{.}5^2=12{.}25$. Gjennomsnittet av kvadratene er ikke kvadratet av gjennomsnittet - og forskjellen mellom dem er nettopp variansen, som vi ser i neste delkapittel.

### 5.3 Varians og standardavvik

Forventningsverdien forteller oss hvor fordelingen ligger, men ikke hvor samlet den er. To fordelinger kan ha nøyaktig samme tyngdepunkt og likevel se helt forskjellige ut - den ene tett samlet rundt midten, den andre spredt utover.

Tenk på to måleserier som begge har gjennomsnitt $3$: den ene gir verdiene $2, 3, 4$, den andre gir $0, 3, 6$. Samme forventningsverdi, men vidt forskjellig presisjon. Vi trenger derfor et tall til, som måler spredning. 

:::{admonition} Def. 5.5 - Varians og standardavvik
:class: warning

**Variansen** til en stokastisk variabel $X$, skrevet $\text{Var}(X)$ eller $\sigma^2$, er forventningsverdien av det kvadrerte avviket fra $\mu=E[X]$:

$\text{Var}(X) = E\left[(X-\mu)^2\right]$

**Diskret variabel.** Med punktsannsynligheten $P(X=x)$ er

$\text{Var}(X) = \sum_x (x-\mu)^2 \cdot P(X=x)$

**Kontinuerlig variabel.** Med tettheten $f(x)$ er

$\text{Var}(X) = \int_{-\infty}^{\infty} (x-\mu)^2 \cdot f(x)\,dx$

**Standardavviket** er kvadratroten av variansen,

$\sigma = \sqrt{\text{Var}(X)}$
:::

Husk også fra Def. 5.4:

\begin{align*}
    E[g(X)] = \sum g(x) \cdot P(X=x)
\end{align*}

slik at

\begin{align*}
    E\left[(x-\mu)^2\right]= \sum (x-\mu)^2 \cdot P(X=x)
\end{align*}

La oss prøve å få formelen til å virke mer intuitiv:

1. Legg merke til at vi kvadrerer avviket $x-\mu$. Grunnen er enkel: uten kvadratet ville positive og negative avvik utjevnet hverandre, og vi ville alltid fått null. Det er nettopp det forventningsverdien betyr - tyngdepunktet ligger der avvikene balanserer.

2. Kvadratet gjør alle avvik positive, og gir samtidig ekstra vekt til de store avvikene. En verdi som ligger dobbelt så langt fra midten bidrar fire ganger så mye.

3. Mange verdier $x$ langt fra forvetningsverdien $\mu$ $\to$ mange store kvadrerte avvik $(x-\mu)^2$ $\to$ stor $Var(x)$


Bruk simulasjonen nedenfor til å gå gradvis fra steg 1 til steg 4, for å se hvordan variansen er bygget opp.

```python
vis_sim("Simuleringer/varians.html")
```

Kvadreringen har én ulempe: variansen får feil enhet. Måler vi energi i joule, får variansen enheten joule i andre, som ikke er særlig meningsfullt. Derfor tar vi kvadratroten til slutt.

**Standardavviket $\sigma$ har samme enhet som variabelen selv**, og kan tolkes direkte som «hvor langt fra gjennomsnittet ligger en typisk måling». Det er derfor standardavviket som oftest rapporteres, mens variansen er det vi regner med underveis.

:::{card} 
**Eksempel**
^^^
To variabler har samme forventningsverdi, men ulik spredning:

$X$: verdiene $2, 3, 4$, hver med sannsynlighet $\frac{1}{3}$

$Y$: verdiene $0, 3, 6$, hver med sannsynlighet $\frac{1}{3}$

Begge har $\mu=3$. Finn variansen til hver.

For $X$ er avvikene $-1, 0, 1$:

$\text{Var}(X)=(-1)^2 \cdot \frac{1}{3}+0^2 \cdot \frac{1}{3}+1^2 \cdot \frac{1}{3}=\frac{2}{3} \approx 0{.}67$

For $Y$ er avvikene $-3, 0, 3$:

$\text{Var}(Y)=(-3)^2 \cdot \frac{1}{3}+0^2 \cdot \frac{1}{3}+3^2 \cdot \frac{1}{3}=\frac{18}{3}=6$

Standardavvikene blir $\sigma_X=\sqrt{2/3}\approx 0{.}82$ og $\sigma_Y=\sqrt{6}\approx 2{.}45$. Samme tyngdepunkt, men $Y$ er omtrent tre ganger så spredt.
:::

:::{admonition} Teorem 5.2 - Regnesnarvei for varians
:class: note

For enhver stokastisk variabel $X$ gjelder

$\text{Var}(X) = E[X^2] - \left(E[X]\right)^2$
:::

Denne omskrivingen er ofte raskere å regne med, fordi vi slipper å trekke fra $\mu$ i hvert eneste ledd. I stedet finner vi først $E[X^2]$ - forventningsverdien av den kvadrerte variabelen - og trekker så fra kvadratet av forventningsverdien.

Formelen leses gjerne som «gjennomsnittet av kvadratene minus kvadratet av gjennomsnittet». Merk rekkefølgen: de to leddene er ikke like, og forskjellen mellom dem er nøyaktig variansen.

:::{card} 
**Eksempel**
^^^
Finn variansen og standardavviket til antall øyne på en likevektet terning.

Fra delkapittel 5.2 vet vi at $\mu=3{.}5$. Vi bruker regnesnarveien i Teorem 5.2, og finner først $E[X^2]$:

$E[X^2]=1^2 \cdot \frac{1}{6}+2^2 \cdot \frac{1}{6}+3^2 \cdot \frac{1}{6}+4^2 \cdot \frac{1}{6}+5^2 \cdot \frac{1}{6}+6^2 \cdot \frac{1}{6}=\frac{91}{6}\approx 15{.}17$

Da er

$\text{Var}(X)=E[X^2]-\mu^2=\frac{91}{6}-3{.}5^2=15{.}17-12{.}25=\frac{35}{12}\approx 2{.}92$

Standardavviket blir

$\sigma=\sqrt{2{.}92}\approx 1{.}71$

Et typisk terningkast ligger altså rundt $1{.}7$ øyne unna gjennomsnittet på $3{.}5$.
:::

:::{card} 
**Eksempel**
^^^
En variabel har tettheten $f(x)=2x$ for $0 \leq x \leq 1$. Finn variansen.

Fra delkapittel 5.2 vet vi at $\mu=\frac{2}{3}$. Vi finner først $E[X^2]$ ved å integrere:

$E[X^2]=\int_0^1 x^2 \cdot 2x\,dx=\int_0^1 2x^3\,dx=\Big[\frac{x^4}{2}\Big]_0^1=\frac{1}{2}$

Ved Teorem 5.2 er da

$\text{Var}(X)=\frac{1}{2}-\left(\frac{2}{3}\right)^2=\frac{1}{2}-\frac{4}{9}=\frac{9-8}{18}=\frac{1}{18}\approx 0{.}056$

Standardavviket blir $\sigma=\sqrt{1/18}\approx 0{.}24$.
:::

:::{admonition} Teorem 5.3 - Varians ved skalering
:class: note

For en stokastisk variabel $X$ og konstanter $a$ og $b$ gjelder

$\text{Var}(aX+b)=a^2\,\text{Var}(X)$
:::

To ting er verdt å merke seg her, og begge gir god mening.

Konstanten $b$ forsvinner helt. Legger vi $5$ til alle verdiene, flytter hele fordelingen seg fem hakk til høyre - men den blir ikke mer eller mindre spredt. Spredning handler om avstander, og de er upåvirket av en forskyvning.

Faktoren $a$ kommer derimot inn i andre potens, fordi variansen måler *kvadrerte* avvik. Dobler vi alle verdiene, firedobles variansen. For standardavviket blir sammenhengen enklere, siden vi tar kvadratroten:

$\sigma_{aX+b}=|a| \cdot \sigma_X$

Standardavviket skalerer altså rett proporsjonalt, noe som er nok en grunn til at det ofte er mer intuitivt å bruke enn variansen.

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Spredning er ikke bare en teknisk detalj i kjemien - den bærer fysisk informasjon.

**Fluktuasjoner.** Energien til et system i termisk likevekt er ikke helt konstant; den svinger rundt middelverdien $\langle E \rangle$. Variansen i energien henger direkte sammen med varmekapasiteten:

$\text{Var}(E) = \langle E^2 \rangle - \langle E \rangle^2 = k_B T^2 C_V$

Legg merke til at venstresiden er nøyaktig regnesnarveien fra Teorem 5.2, bare skrevet med vinkelparenteser. Et system med stor varmekapasitet har altså større energifluktuasjoner.

**Hvorfor makroskopiske målinger er stabile.** For et system med $N$ partikler vokser standardavviket i energi som $\sqrt{N}$, mens energien selv vokser som $N$. Den *relative* spredningen går derfor som

$\frac{\sigma_E}{\langle E \rangle} \sim \frac{\sqrt{N}}{N} = \frac{1}{\sqrt{N}}$

For $N \approx 10^{23}$ blir dette rundt $10^{-11{.}5}$ - altså helt neglisjerbart. Det er derfor et termometer viser en stabil temperatur, selv om molekylene under overflaten beveger seg fullstendig kaotisk.
:::

:::{admonition} Underveisoppgave 5.3

### 5.4 Normalfordeling

Så langt har vi snakket om sannsynlighetsfordelinger helt generelt. I praksis finnes det en rekke navngitte fordelinger som dukker opp igjen og igjen, både diskrete og kontinuerlige. Blant de diskrete finner vi for eksempel binomisk fordeling, som beskriver antall treff i en serie forsøk, og poissonfordeling, som brukes til å telle sjeldne hendelser. Blant de kontinuerlige har vi uniform fordeling, som vi allerede har møtt, og eksponentialfordeling, som beskriver ventetider.

Én kontinuerlig fordeling skiller seg likevel ut i betydning: **normalfordelingen**. Den dukker opp overalt - i måleusikkerhet, i biologiske størrelser, i fysiske signaler - og er derfor den fordelingen du oftest vil møte i praktisk arbeid.

Grunnen til at normalfordelingen dukker opp så ofte er verdt å nevne allerede nå. Når en størrelse påvirkes av mange små, uavhengige bidrag som legges sammen, blir resultatet tilnærmet normalfordelt - nesten uansett hvordan de enkelte bidragene er fordelt.

Måler du samme prøve mange ganger, vil resultatene sprike litt. Avviket skyldes ikke én enkelt feilkilde, men summen av mange bittesmå: temperatursvingninger, avlesningsunøyaktighet, vibrasjoner, elektronisk støy. Nettopp derfor er målefeil så ofte normalfordelt. Vi kommer tilbake til dette resultatet i delkapittel 3.5.

:::{admonition} Def. 5.6 - Normalfordelingen
:class: warning

En kontinuerlig stokastisk variabel $X$ er **normalfordelt** med forventningsverdi $\mu$ og standardavvik $\sigma$ dersom tettheten er

$f(x)=\dfrac{1}{\sigma\sqrt{2\pi}}\,e^{-\frac{(x-\mu)^2}{2\sigma^2}}$

Vi skriver da $X \sim N(\mu, \sigma^2)$.

Tilfellet $\mu=0$ og $\sigma=1$ kalles **standard normalfordeling**.
:::

Uttrykket ser tungt ut, men de enkelte delene har hver sin klare rolle. Bruk simulasjonen nedenfor til å trykke på de ulike delene av formelen og se hva de representerer grafisk. Du kan også endre forventningen $\mu$ og standardavviket $\sigma$ og se hvordan grafen endres.

```python
vis_sim("Simuleringer/normalfordeling.html")
```

Normalfordelingen er fullstendig bestemt av bare to tall. Forventningsverdien $\mu$ sier hvor kurven ligger, og standardavviket $\sigma$ sier hvor bred den er. Kjenner du disse to, kjenner du hele fordelingen.

Kurven er symmetrisk om $\mu$, har sitt toppunkt der, og strekker seg i prinsippet uendelig langt i begge retninger - men halene blir så tynne at sannsynligheten for verdier langt unna er forsvinnende liten.

:::{admonition} Teorem 5.4 - Regelen om 68-95-99,7
:class: note

For en normalfordelt variabel $X \sim N(\mu,\sigma^2)$ gjelder

$P(\mu-\sigma \leq X \leq \mu+\sigma) \approx 0{,}68$

$P(\mu-2\sigma \leq X \leq \mu+2\sigma) \approx 0{,}95$

$P(\mu-3\sigma \leq X \leq \mu+3\sigma) \approx 0{,}997$
:::

Denne regelen er verdt å huske, fordi den gir en rask følelse for hva et standardavvik betyr i praksis.

Omtrent to tredjedeler av alle verdier ligger innenfor ett standardavvik fra gjennomsnittet. Nesten alt - $95\%$ - ligger innenfor to. Og innenfor tre standardavvik finner vi praktisk talt alt, med bare $0{.}3\%$ igjen ute i halene.

Legg merke til at regelen ikke avhenger av $\mu$ eller $\sigma$ i det hele tatt. Den gjelder for *enhver* normalfordeling, enten vi måler energier i joule eller høyder i centimeter. Det er nettopp derfor standardavviket er et så nyttig mål på spredning.

:::{card} 
**Eksempel**
^^^
En serie målinger av samme prøve er normalfordelt med $\mu=500$ og $\sigma=100$.

a) Hvor stor andel av målingene ligger mellom $400$ og $600$?

Vi ser at $400=\mu-\sigma$ og $600=\mu+\sigma$, altså nøyaktig ett standardavvik på hver side. Ved Teorem 5.4 er

$P(400 \leq X \leq 600) \approx 0{,}68$

Omtrent $68\%$ av målingene ligger altså i dette intervallet.

b) Hvor stor andel ligger mellom $300$ og $700$?

Her er $300=\mu-2\sigma$ og $700=\mu+2\sigma$, så

$P(300 \leq X \leq 700) \approx 0{,}95$
:::

:::{admonition} Def. 5.7 - Standardisering (z-verdi)
:class: warning

For en normalfordelt variabel $X \sim N(\mu,\sigma^2)$ er **z-verdien** til en observasjon $x$ gitt ved

$z=\dfrac{x-\mu}{\sigma}$

Den standardiserte variabelen $Z=\frac{X-\mu}{\sigma}$ er da standard normalfordelt, $Z \sim N(0,1)$.
:::

$z$-verdien svarer på et enkelt spørsmål: **hvor mange standardavvik unna gjennomsnittet ligger denne verdien?** En z-verdi på $2$ betyr at observasjonen ligger to standardavvik over $\mu$, og $z=-1{.}5$ at den ligger halvannet standardavvik under.

Nytten er at vi kan sammenligne på tvers av helt ulike fordelinger. Er én måling $z=2$ og en annen $z=0{.}3$, vet vi at den første er langt mer uvanlig - uansett hvilke enheter de er målt i.

Legg merke til at standardisering er nøyaktig operasjonen fra Teorem 5.1 og 5.3: vi trekker fra $\mu$ (som flytter tyngdepunktet til null) og deler på $\sigma$ (som skalerer standardavviket til én).

Sannsynligheter for intervaller i den standardiserte normalfordelingen finnes i ulike oppslag og på kalkulatorer. Derfor er det i praksis veldig nyttig å standardisere  normalfordelingen vi ser på til en standardisert normalfordeling, og bla opp i tabeller.

:::{card} 
**Eksempel**
^^^
En serie målinger er normalfordelt med $\mu=250$ og $\sigma=20$.

a) Hva er sannsynligheten for at en måling er mindre enn $290$?

Vi kan ikke regne ut dette integralet for hånd, så vi standardiserer først og slår deretter opp i en z-tabell. Fra Def. 5.7 er

$z=\frac{290-250}{20}=2{.}00$

Oppslag i z-tabellen gir $P(Z \leq 2{.}00)=0{.}9772$. Altså er

$P(X \leq 290)=0{.}9772$

Omtrent $98\%$ av målingene ligger under $290$.

b) Hva er sannsynligheten for at en måling er større enn $290$?

Dette er komplementet til a), så vi bruker Teorem 3.2:

$P(X > 290)=1-0{.}9772=0{.}0228$

Bare rundt $2{.}3\%$ av målingene ligger over $290$.

c) Hva er sannsynligheten for at en måling ligger mellom $230$ og $290$?

Vi standardiserer begge grensene:

$z_1=\frac{230-250}{20}=-1{.}00 \qquad z_2=\frac{290-250}{20}=2{.}00$

Sannsynligheten for et intervall finner vi ved å trekke fra:

$P(230 \leq X \leq 290)=P(Z \leq 2{.}00)-P(Z \leq -1{.}00)=0{.}9772-0{.}1587=0{.}8185$

Omtrent $82\%$ av målingene ligger i dette intervallet.
:::

:::{admonition} Kobling til fysikalsk kjemi
:class: important

Normalfordelingen dukker opp flere steder i kjemien.

**Måleusikkerhet.** Gjentatte målinger av samme størrelse er normalfordelt rundt den sanne verdien. Når et resultat oppgis som $c = 0{,}250 \pm 0{,}004$ mol/L, er tallet etter $\pm$ som regel ett standardavvik - og ved Teorem 3.4 betyr det at den sanne verdien ligger i dette intervallet med omtrent $68\%$ sannsynlighet.

**Fartsfordelingen i en gass.** Maxwell-Boltzmann-fordelingen for én enkelt fartskomponent, for eksempel $v_x$, er nøyaktig en normalfordeling med $\mu=0$ og $\sigma^2=k_BT/m$:

$f(v_x) \propto e^{-mv_x^2/(2k_BT)}$

Sammenlign eksponenten med Def. 5.6 - formen er den samme, med $\sigma^2=k_BT/m$. Dette forteller oss at fartsfordelingen blir bredere ved høyere temperatur og smalere for tunge molekyler.

**Spektrallinjer.** Molekyler i bevegelse gir dopplerforskyvning, og siden fartene er normalfordelte, får spektrallinjene en normalfordelt form. Bredden på linjen kan derfor brukes til å bestemme temperaturen i en gass.
:::

:::{admonition} Underveisoppgave 5.4
:class: tip

En serie målinger er normalfordelt med $\mu=250$ og $\sigma=20$.

a) Hvilket intervall inneholder omtrent $68\%$ av målingene?

b) Hvilket intervall inneholder omtrent $95\%$?

c) Finn z-verdien til målingen $x=290$, og forklar hva den betyr.

d) En annen måleserie har $\mu=8{.}0$ og $\sigma=0{.}5$. En måling gir $x=9{.}0$. Hvilken av de to målingene - $290$ i den første serien eller $9{.}0$ i den andre - er mest uvanlig?
:::

:::{admonition} Løsning 5.4
:class: tip, dropdown

a) Intervallet er $\mu \pm \sigma$, altså fra $250-20=230$ til $250+20=270$.

b) Intervallet er $\mu \pm 2\sigma$, altså fra $250-40=210$ til $250+40=290$.

c) Vi bruker Def. 5.7:

$z=\frac{290-250}{20}=\frac{40}{20}=2$

Målingen ligger altså nøyaktig to standardavvik over gjennomsnittet. Fra Teorem 5.4 vet vi at bare rundt $5\%$ av målingene ligger utenfor $\pm 2\sigma$, og siden fordelingen er symmetrisk, ligger omtrent $2{.}5\%$ over $290$.

d) Vi regner ut z-verdien for den andre målingen:

Begge målingene har $z=2$, og er derfor **like uvanlige**, selv om tallene og enhetene er helt forskjellige. Det er nettopp dette standardisering gjør mulig: vi kan sammenligne avvik på tvers av fordelinger.

### 5.5 Sentralgrenseteoremet

I delkapittel 5.2 sa vi at forventningsverdien til en terning er $3{.}5$, og at gjennomsnittet av mange kast vil nærme seg dette tallet. Det er en påstand vi ikke har begrunnet ennå. Kaster du terningen fem ganger, kan gjennomsnittet fint bli $2{.}4$ eller $4{.}8$ - så hvorfor skal vi tro at det stabiliserer seg?

Svaret er **store talls lov**, som er selve broen mellom teori og virkelighet i sannsynlighetsregning. Uten den ville forventningsverdi bare vært et regnestykke uten praktisk betydning.

:::{admonition} Def. 5.8 - Utvalgsgjennomsnitt
:class: warning

La $X_1, X_2, \dots, X_n$ være $n$ uavhengige målinger av samme stokastiske variabel. **Utvalgsgjennomsnittet** er

$\bar{X}_n = \dfrac{1}{n}\sum_{i=1}^{n} X_i$
:::

:::{admonition} OBS
:class: error
Legg merke til at $\bar{X}_n$ selv er en stokastisk variabel. Gjentar du hele måleserien, får du et nytt gjennomsnitt. Det er nettopp derfor det er meningsfullt å spørre hvordan gjennomsnittet oppfører seg - det har sin egen fordeling, med sin egen forventningsverdi og spredning.
:::

:::{admonition} Teorem 5.5 - Store talls lov
:class: note

La $X_1, X_2, \dots$ være uavhengige målinger av samme variabel med forventningsverdi $\mu$. Da vil utvalgsgjennomsnittet nærme seg $\mu$ når antall målinger vokser:

$\bar{X}_n \to \mu \qquad \text{når} \qquad n \to \infty$
:::

Loven sier at tilfeldigheten jevner seg ut. Enkeltmålinger spriker, men gjennomsnittet av mange blir stadig mer forutsigbart.

Det er verdt å være presis om hva loven **ikke** sier. Har du kastet mynt ti ganger og fått kron hver gang, sier ikke loven at mynten nå «skylder» deg noen mynter for å rette opp balansen. Mynten har ingen hukommelse, og hvert nytt kast er fortsatt $50/50$ - dette er uavhengighet fra delkapittel 2.3. Det som skjer, er at de ti første kastene rett og slett drukner i mengden når antallet vokser. Ubalansen forsvinner ikke; den blir bare ubetydelig i forhold til alt det andre.

:::{admonition} Teorem 5.6 - Standardfeilen
:class: note

For $n$ uavhengige målinger av en variabel med forventningsverdi $\mu$ og standardavvik $\sigma$ gjelder

$E[\bar{X}_n]=\mu \qquad \text{og} \qquad \text{Var}(\bar{X}_n)=\dfrac{\sigma^2}{n}$

Standardavviket til gjennomsnittet kalles **standardfeilen**, og er

$\sigma_{\bar{X}} = \dfrac{\sigma}{\sqrt{n}}$
:::

Dette teoremet gjør store talls lov konkret: det forteller ikke bare *at* gjennomsnittet stabiliserer seg, men *hvor fort*.

Legg merke til kvadratroten. Vil du halvere usikkerheten i gjennomsnittet, må du firedoble antall målinger. Vil du gjøre den ti ganger mindre, trenger du hundre ganger så mange. Dette er en hard begrensning i alt eksperimentelt arbeid - presisjon koster, og den koster kvadratisk.

For en terning med $\sigma \approx 1{.}71$ blir standardfeilen $0{.}54$ etter 10 kast, $0{.}17$ etter 100 kast og $0{.}054$ etter 1000. Gjennomsnittet krymper altså inn mot $3{.}5$, men stadig langsommere.

:::{card} 
**Eksempel**
^^^
En måling har standardavvik $\sigma=0{.}8$ enheter. Hvor mange målinger trengs for at standardfeilen skal bli mindre enn $0{.}1$?

Vi bruker Teorem 5.6 og krever

$\frac{\sigma}{\sqrt{n}} < 0{.}1 \quad \Rightarrow \quad \frac{0{.}8}{\sqrt{n}} < 0{.}1$

Vi løser for $\sqrt{n}$:

$\sqrt{n} > \frac{0{.}8}{0{.}1}=8 \quad \Rightarrow \quad n > 64$

Det trengs altså minst $65$ målinger. Vil vi i stedet ha standardfeil under $0{.}05$, blir kravet $\sqrt{n}>16$, altså $n>256$ - fire ganger så mange målinger for å halvere usikkerheten.
:::

:::{admonition} Teorem 5.7 - Sentralgrenseteoremet
:class: note

La $X_1, X_2, \dots, X_n$ være uavhengige målinger av samme variabel med forventningsverdi $\mu$ og standardavvik $\sigma$. Når $n$ blir stor, er utvalgsgjennomsnittet tilnærmet normalfordelt:

$\bar{X}_n \approx N\left(\mu, \dfrac{\sigma^2}{n}\right)$

Dette gjelder **uansett** hvilken fordeling de enkelte målingene har.
:::

Her er svaret på spørsmålet vi lot ligge i delkapittel 5.4: hvorfor dukker normalfordelingen opp overalt?

Det bemerkelsesverdige er den siste setningen i teoremet. De enkelte målingene kan være uniformt fordelt, skjevt fordelt, eller ha en helt merkelig form - gjennomsnittet blir likevel klokkeformet. Kaster du én terning, er fordelingen flat, med like stor sannsynlighet for alle seks verdiene. Kaster du to og tar gjennomsnittet, får du en trekantform. Med fem kast begynner klokkeformen å vise seg, og med tretti er den praktisk talt en normalfordeling.

Dette er grunnen til at måleusikkerhet så ofte er normalfordelt. En måling påvirkes av mange små uavhengige feilkilder som legges sammen, og summen av mange uavhengige bidrag er tilnærmet normalfordelt - uansett hvordan hver enkelt feilkilde oppfører seg.

:::{admonition} OBS
:class: error

De to teoremene handler om det samme gjennomsnittet $\bar{X}_n$, men svarer på hvert sitt spørsmål.

**Store talls lov** sier *hvor* gjennomsnittet havner: det konvergerer mot $\mu$. Svaret er ett enkelt tall.

**Sentralgrenseteoremet** sier *hvordan gjennomsnittet fordeler seg rundt* dette tallet: det er tilnærmet normalfordelt med spredning $\frac{\sigma}{\sqrt{n}}$. Svaret er en hel fordeling.

Tenk på en blink. Store talls lov forteller deg at pilene treffer nær midten. Sentralgrenseteoremet forteller deg hvordan treffene sprer seg rundt midten - at de danner en klokkeform, og hvor bred den er.

I praksis er sentralgrenseteoremet det mest brukte av de to. Store talls lov beroliger deg med at målingene konvergerer, men sier ingenting om hvor nær du er etter for eksempel $50$ målinger. Sentralgrenseteoremet lar deg faktisk regne på usikkerheten, og er derfor grunnlaget for feilmarginer og konfidensintervall.

Legg til slutt merke til at sentralgrenseteoremet er det sterkere resultatet: siden spredningen $\frac{\sigma}{\sqrt{n}}$ går mot null når $n$ vokser, følger store talls lov som en direkte konsekvens.
:::

*Sentralgrenseteoremet*

En medisinprodusert lager tabletter som skal veie $500$ mg. Tablettene blir sjelden nøyaktig 500 mg, men de varierer med et standardavvik på $\sigma=20$ mg. Enkelttabletter har derfor en vekt mellom $460$ mg og $540$ mg. Fabrikken plukker ut $25$ tabletter fra en batch, veier dem, og regner ut gjennomsnittsvekten. De får $501.2$ mg. Sentralgrenseteoremet sier at dette gjennomsnittet er normalfordelt med standardavvik $\frac{\sigma}{\sqrt{n}}=\frac{20}{\sqrt{25}}=\frac{20}{5}=4$ mg. Vekten på enkelttabletter spriker altså med $20$ mg, mens gjennomsnittet av $25$ tabletter bare spriker med $4$ mg. Hvorofr betyr dette noe i praksis? Finner du én tablett på $540$ mg, er dette normalt og 2 standardavvik unna gjennomsnittet. Får du derimot et *gjennomsnitt* på $540$ mg fra $25$ tabletter, er du 10 standardavvik unna. Sannsynligheten for dette er liten, og kan tyde på feil ved apparatet.

Store talls lov er selve grunnen til at termodynamikk fungerer.

Et enkelt molekyl beveger seg fullstendig kaotisk. Vi kan umulig følge hvert av dem, og vil heller ikke: det vi måler er gjennomsnitt over enormt mange partikler. Trykk er gjennomsnittlig kraft fra molekylkollisjoner, temperatur er knyttet til gjennomsnittlig kinetisk energi, og indre energi er en forventningsverdi over mikrotilstander.

Med $N \approx 10^{23}$ partikler blir standardfeilen fra Teorem 3.6 uendelig liten. Vi så i delkapittel 5.3 at den relative spredningen går som $1/\sqrt{N}$, altså omtrent $10^{-11{,}5}$ for et mol. Det er derfor et termometer viser en stabil verdi, og hvorfor gasslovene oppfører seg som eksakte sammenhenger selv om de bygger på tilfeldig molekylbevegelse.

Dette forklarer også hvorfor termodynamikken bryter sammen for svært små systemer. Har du bare noen få hundre molekyler, er $\sqrt{N}$ ikke lenger stor nok til å dempe svingningene, og fluktuasjonene blir målbare.

b) Vi krever $\frac{1{.}71}{\sqrt{n}}<0{.}1$. altså

Det trengs minst $293$ kast.
