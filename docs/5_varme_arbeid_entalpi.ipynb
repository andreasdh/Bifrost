## Fra mikroskopisk til makroskopisk termodynamikk

Mange av størrelsene vi trenger, har du allerede møtt. [Indre energi](https://andreasdh.github.io/Bifrost/docs/3_partisjonsfunksjonen.html#indre-energi-fra-partisjonsfunksjonen) er en tilstandsfunksjon, mens varme og arbeid beskriver måter energi kan overføres over systemgrensen. Vi skal derfor ikke begynne på nytt, men bruke disse ideene som utgangspunkt for den makroskopiske beskrivelsen.

### Varme i historien

Gjennom historien har varme blitt forklart på ulike måter. På 1700-tallet var det vanlig å tenke på varme som en vektløs væske, *calorique*, som kunne flyte fra et varmt stoff til et kaldt stoff. Joseph Black viste gjennom studier av blant annet vann at et stoff kan ta opp eller avgi varme uten at temperaturen nødvendigvis endres, det vi i dag forbinder med latent varme. Han observerte også at ulike stoffer får ulik temperaturendring når de tilføres samme mengde energi.

Senere ble koblingen mellom mekanisk arbeid og varme stadig tydeligere. Count Rumford observerte den store varmeutviklingen ved boring av kanoner, og James Joules eksperimenter viste kvantitativt at arbeid og varme er to måter å overføre energi på.

### Varme i dag

I moderne termodynamikk er varme $q$ ikke noe systemet *inneholder*. Varme er energi som overføres på grunn av en temperaturforskjell.

Når et system mottar energi som varme, kan den indre energien endres. På mikronivå kan energien fordeles mellom translasjon, rotasjon, vibrasjon og elektroniske tilstander, og den kan endre systemets potensielle energi gjennom endringer i molekylenes struktur og vekselvirkninger.

Når varmeoverføringen er ferdig, inngår energien i systemets **indre energi**. Kort sagt: varme beskriver *hvordan energi overføres*, mens indre energi beskriver *energien systemet har*.


## Arbeid og reversible prosesser

Arbeid er den andre grunnleggende måten energi kan overføres over systemgrensen på. Arbeid skjer når en kraft virker gjennom en forflytning eller, mer generelt, når systemet endrer en makroskopisk variabel mot en ytre påvirkning. I fysikken kan dette være alt fra å komprimere en gass og trekke i en fjær til å flytte en ladning i et elektrisk felt.

I kjemisk termodynamikk er vi særlig interessert i **trykk–volum-arbeid**, for eksempel når en gass ekspanderer mot et ytre trykk.

Vi bruker den vanlige kjemiske fortegnskonvensjonen:

- $w>0$ når omgivelsene gjør arbeid **på** systemet
- $w<0$ når systemet gjør arbeid **på** omgivelsene

Vi kan se for oss en gass i en sylinder med et bevegelig stempel. Stempelet har areal $A$, og et ytre trykk $P_\mathrm{ext}$ presser på stempelet. Kraften som virker på stempelet er

$$F=P_\mathrm{ext}A$$

Dersom stempelet flyttes en liten avstand $dx$, endres volumet med

$$dV=A\,dx$$

Arbeidet som gjøres **på systemet** ved denne lille volumendringen blir derfor

$$\delta w=-F\,dx=-P_\mathrm{ext}A\,dx=-P_\mathrm{ext}\,dV$$

For en endelig prosess får vi

$$w=-\int_{V_i}^{V_f}P_\mathrm{ext}\,dV$$

Dersom det ytre trykket er konstant, kan vi trekke det utenfor integralet:

$$w=-P_\mathrm{ext}\int_{V_i}^{V_f}dV=-P_\mathrm{ext}\Delta V$$

Ved ekspansjon er $\Delta V>0$, og dermed er $w<0$: systemet gjør arbeid på omgivelsene. En mer grundig beskrivelse av slike bestemte integraler finner du i [4.5 Det bestemte integralet og Riemann-summer](https://andreasdh.github.io/Bifrost/docs/matematikk/kalkulus.html#det-bestemte-integralet-og-riemann-summer).

### Hva betyr reversibel?

Men hvordan kan vi få mest mulig arbeid ut av en slik ekspansjon? Her kommer ideen om en **reversibel prosess** inn.

En reversibel prosess er en idealisert prosess som kan snus ved en uendelig liten endring i betingelsene. Systemet går gjennom en sammenhengende serie tilstander som ligger vilkårlig nær likevekt. For en gass som ekspanderer, betyr dette at det ytre trykket hele tiden er nesten lik gasstrykket, og at vi unngår dissipative effekter som friksjon.

Et klassisk bilde er å tenke seg sand som ligger oppå stempelet og presser gassen sammen. Dersom vi fjerner store mengder sand om gangen, endrer vi det ytre trykket brått og prosessen blir tydelig irreversibel. Dersom vi i stedet kunne fjerne ett og ett stadig mindre sandkorn, ville stempelet flytte seg bare litt mellom hvert trinn og systemet få tid til å innstille seg svært nær en ny likevekt.

En virkelig prosess kan aldri være perfekt reversibel. Den reversible prosessen er en idealisert **grenseverdi**, men en svært nyttig en: Ved ekspansjon gir den størst mulig arbeid fra systemet mellom to gitte tilstander.

## Indre energi og termodynamikkens første lov

Indre energi $U$ har vi allerede møtt fra et mikroskopisk perspektiv. Den beskriver den totale energien i systemet og kan inneholde bidrag fra molekylenes translasjon, rotasjon og vibrasjon, elektroniske tilstander og potensiell energi knyttet til struktur og vekselvirkninger.

Skjematisk kan vi skrive

$$U=E_\mathrm{trans}+E_\mathrm{rot}+E_\mathrm{vib}+E_\mathrm{elektronisk}+E_\mathrm{potensiell}$$

I statistisk termodynamikk beskrev vi den indre energien som den gjennomsnittlige energien over mikrotilstandene:

$$U=\langle E\rangle$$

På makroskopisk nivå trenger vi ikke kjenne de enkelte mikrotilstandene. Vi beskriver i stedet hvordan $U$ endres når energi krysser systemgrensen. For et lukket system kan dette skje ved varmeoverføring eller arbeid.

:::{admonition} Termodynamikkens første lov
:class: important

$$\Delta U=q+w$$

Varme og arbeid beskriver energioverføring under en prosess, mens $U$ er en tilstandsfunksjon.

:::

Dersom systemet går fra tilstand A til tilstand B, er

$$\Delta U=U_B-U_A$$

Verdiene av $q$ og $w$ kan avhenge av veien mellom A og B, men summen må være den samme så lenge start- og sluttilstanden er de samme.

Dette er forskjellen mellom [tilstandsfunksjoner og stifunksjoner](https://andreasdh.github.io/Bifrost/docs/A1_grunnleggende_begreper.html#stifunksjoner-og-tilstandsfunksjoner). Indre energi er en egenskap ved tilstanden. Varme og arbeid beskriver hva som skjer *på veien* mellom tilstandene.

Det skillet blir tydelig dersom vi tilfører den samme typen system energi på forskjellige måter.

## Entalpi – nyttig ved konstant trykk

I et kjemisk laboratorium foregår svært mange prosesser ved omtrent konstant trykk: i åpne begerglass og kolber eller i systemer som står i kontakt med atmosfæren. Systemet kan da både ta opp eller avgi varme og utføre volumarbeid dersom volumet endres, for eksempel når det dannes gass.

Hvis vi bare bruker den indre energien $U$, må vi holde styr på både varmeoverføringen og volumarbeidet. Ved konstant trykk finnes det en mer praktisk tilstandsfunksjon: **entalpi**.

$$H=U+PV$$

La oss se hvorfor.

Vi differensierer uttrykket:

$$dH=dU+P\,dV+V\,dP$$

Fra første lov har vi

$$dU=\delta q+\delta w$$

og dersom det eneste arbeidet er trykk–volum-arbeid,

$$\delta w=-P_\mathrm{ext}\,dV$$

Dermed får vi

$$dH=\delta q+(P-P_\mathrm{ext})\,dV+V\,dP$$

For de vanlige konstant-trykk-prosessene vi skal bruke dette på, er systemet i mekanisk likevekt med omgivelsene slik at $P=P_\mathrm{ext}$, og trykket holdes konstant slik at $dP=0$. Da står vi igjen med

$$dH=\delta q_P$$

For en endelig prosess ved konstant trykk får vi derfor

$$\Delta H=q_P$$

Dette er grunnen til at entalpi er så nyttig i kjemi: Under vanlige laboratoriebetingelser kan den målbare varmeoverføringen kobles direkte til endringen i en tilstandsfunksjon.

:::{admonition} Viktig
:class: important

Entalpi er **ikke varmeinnhold**. $H$ er en tilstandsfunksjon. Under betingelser med konstant trykk og bare trykk–volum-arbeid er

$$\Delta H=q_P$$

:::

Entalpi gjør altså energiregnskapet ved konstant trykk enklere. Men den svarer fortsatt bare på et energispørsmål: hvor mye varme utveksles? Den forteller ikke alene hvilken retning en prosess er termodynamisk drevet. Det blir utgangspunktet for neste kapittel.


## Varmekapasitet og kalorimetri

### Varmekapasitet i makroskopisk termodynamikk

To forskjellige stoffer kan få svært ulik temperaturendring selv om de tilføres samme mengde energi. Dette var blant observasjonene Joseph Black gjorde allerede på 1700-tallet. I dag beskriver vi dette ved hjelp av **varmekapasitet**.

Fra [statistisk termodynamikk](https://andreasdh.github.io/Bifrost/docs/3_partisjonsfunksjonen.html#ekvipartisjon-og-varmekapasitet) så vi hvordan varmekapasitet kan kobles til hvordan systemets energi endres med temperaturen. I makroskopisk termodynamikk skiller vi særlig mellom varmekapasitet ved konstant volum og ved konstant trykk:

$$C_V=\left(\frac{\partial U}{\partial T}\right)_V$$

$$C_P=\left(\frac{\partial H}{\partial T}\right)_P$$

Forskjellen har en klar fysisk betydning. Ved konstant volum kan systemet ikke ekspandere og utføre trykk–volum-arbeid. Ved konstant trykk kan en del av energiøkningen være knyttet til ekspansjon. For en gass er derfor $C_P$ vanligvis større enn $C_V$.

I praksis er varmekapasitet et viktig verktøy for å koble en målbar temperaturendring til energi som er overført som varme. Vi kan måle temperaturendringen $\Delta T$ og beregne varmen som er tilført eller avgitt ved hjelp av kalorimetri.

### Kalorimetri

**Kalorimetri** er en eksperimentell metode for å måle varmeoverføring. Dersom varmekapasiteten $C$ kan behandles som konstant over temperaturintervallet, har vi

$$q=C\Delta T$$

Det er nyttig å skille mellom to vanlige typer kalorimetri.

1. Kaffekoppkalorimetri

I et enkelt kaffekoppkalorimeter foregår prosessen ved omtrent konstant atmosfærisk trykk. Reaktanter blandes, og temperaturendringen måles. Siden trykket er konstant, er varmeoverføringen direkte knyttet til entalpiendringen:

$$q_P=\Delta H$$

Slike målinger er en viktig kilde til eksperimentelle reaksjonsentalpier.

```{figure} bilder/kaffekoppkalorimeter.svg
:name: fig-kaffekoppkal
:width: 480px

Kaffekoppkalorimeter. Prosessen skjer ved konstant trykk, slik at målt varme svarer til $\Delta H$.
```

2. Bombekalorimetri

I et bombekalorimeter foregår prosessen i en stiv, lukket beholder. Volumet er konstant, slik at trykk–volum-arbeidet er null. Da blir

$$q_V=\Delta U$$

```{figure} bilder/bombekalorimeter.svg
:name: fig-bombekal
:width: 480px

Bombekalorimeter. Prosessen skjer ved konstant volum, i en lukket beholder.
```

Kalorimetri viser dermed svært konkret hvorfor både $U$ og $H$ er nyttige: Hvilken av dem som kobles direkte til den målte varmen, avhenger av hvilke betingelser eksperimentet utføres under.

## Oppsummering

I dette kapitlet har vi gått fra den statistiske beskrivelsen av energi til et makroskopisk energiregnskap.

**Varme** og **arbeid** er to måter energi kan krysse systemgrensen på. Termodynamikkens første lov samler dem:

$$\Delta U=q+w$$

Indre energi er en tilstandsfunksjon, mens varme og arbeid avhenger av prosessveien.

For trykk–volum-arbeid fant vi

$$\delta w=-P_\mathrm{ext}\,dV$$

og så hvorfor den reversible grensen gir størst mulig arbeid fra en ekspanderende gass mellom to gitte tilstander.

Ved konstant trykk er entalpi en særlig nyttig tilstandsfunksjon:

$$H=U+PV$$

og under de vanlige betingelsene med bare trykk–volum-arbeid gjelder

$$\Delta H=q_P$$

Kalorimetri gjør denne koblingen eksperimentell. Et bombekalorimeter ved konstant volum gir direkte informasjon om $\Delta U$, mens et kaffekoppkalorimeter ved omtrent konstant trykk gir informasjon om $\Delta H$.

Vi kan dermed holde rede på **hvor mye energi som utveksles**. I neste kapittel skal vi ta det neste steget: Hvordan avgjør termodynamikken **hvilken retning** en prosess er drevet?


## Oppgaver


### Oppgave 1 – Når en kjemisk reaksjon gjør arbeid

Når kalsiumkarbonat reagerer med saltsyre, dannes karbondioksid:

$$\mathrm{CaCO_3(s)+2HCl(aq)\rightarrow CaCl_2(aq)+H_2O(l)+CO_2(g)}$$

Tenk deg at reaksjonen foregår i en sylinder med et bevegelig stempel. Under reaksjonen tar systemet opp $500\ \mathrm{J}$ varme fra omgivelsene, samtidig som gassen som dannes utvider seg og gjør $200\ \mathrm{J}$ arbeid på omgivelsene.

a) Finn $\Delta U$ for reaksjonen.

b) Den samme kjemiske endringen gjennomføres på en annen måte, mellom de samme start- og sluttilstandene. Denne gangen er $q=100\ \mathrm{J}$. Hva må $w$ være?

c) Hvorfor kan varme og arbeid være forskjellige i de to prosessene selv om den kjemiske start- og sluttilstanden er den samme?

:::{dropdown} Løsningsforslag

a) Når systemet gjør arbeid på omgivelsene, er arbeidet negativt:

$$w=-200\ \mathrm{J}$$

Fra første lov får vi

$$\Delta U=q+w$$

$$\Delta U=500\ \mathrm{J}-200\ \mathrm{J}=300\ \mathrm{J}$$

b) Siden start- og sluttilstanden er den samme, må $\Delta U$ fortsatt være $300\ \mathrm{J}$:

$$300\ \mathrm{J}=100\ \mathrm{J}+w$$

Dermed

$$w=200\ \mathrm{J}$$

I denne prosessen gjør omgivelsene netto arbeid på systemet.

c) Varme og arbeid beskriver hvordan energi overføres under selve prosessen og avhenger derfor av prosessveien. Den indre energien er derimot en tilstandsfunksjon og avhenger bare av start- og sluttilstanden.

:::

### Oppgave 2 – Hvor mye energi frigjøres i en nøytralisjonsreaksjon?

En student blander saltsyre og natriumhydroksid i et kaffekoppkalorimeter:

$$\mathrm{HCl(aq)+NaOH(aq)\rightarrow NaCl(aq)+H_2O(l)}$$

Reaksjonen danner $0{,}0625\ \mathrm{mol}$ vann. Kalorimeteret og løsningen har samlet varmekapasitet

$$C_P=850\ \mathrm{J\,K^{-1}}$$

Temperaturen øker med $4{,}2\ \mathrm{K}$.

a) Hvor mye varme tar løsningen og kalorimeteret opp?

b) Hva er $\Delta H$ for reaksjonen i forsøket?

c) Beregn entalpiendringen per mol dannet vann.

d) Hvorfor kan vi finne $\Delta H$ direkte fra varmen i dette forsøket?

e) Hva ville vært annerledes dersom reaksjonen i stedet foregikk i et bombekalorimeter?

:::{dropdown} Løsningsforslag

a)

$$q_\mathrm{kal}=C_P\Delta T$$

$$q_\mathrm{kal}=(850\ \mathrm{J\,K^{-1}})(4{,}2\ \mathrm{K})=3{,}57\ \mathrm{kJ}$$

b) Reaksjonen avgir den energien som kalorimeteret tar opp:

$$q_\mathrm{rxn}=-3{,}57\ \mathrm{kJ}$$

Ved konstant trykk gjelder

$$\Delta H=q_P$$

Dermed

$$\Delta H=-3{,}57\ \mathrm{kJ}$$

c)

$$\Delta H_\mathrm{mol}=\frac{-3{,}57\ \mathrm{kJ}}{0{,}0625\ \mathrm{mol}}=-57{,}1\ \mathrm{kJ\,mol^{-1}}$$

d) Kaffekoppkalorimeteret arbeider ved omtrent konstant atmosfærisk trykk. Ved konstant trykk er varmeutvekslingen lik entalpiendringen.

e) I et bombekalorimeter er volumet konstant. Da utføres ikke trykk–volum-arbeid, og den målte varmen gir direkte

$$q_V=\Delta U$$

:::

### Oppgave 3 – Forbrenning av etanol i et bombekalorimeter

Forbrenningsenergi kan bestemmes ved å brenne et stoff i et bombekalorimeter. En prøve med $0{,}500\ \mathrm{g}$ etanol forbrennes fullstendig i et kalorimeter med samlet varmekapasitet

$$C_V=6{,}90\ \mathrm{kJ\,K^{-1}}$$

Temperaturen øker med

$$\Delta T=2{,}15\ \mathrm{K}$$

Molar masse for etanol er $46{,}07\ \mathrm{g\,mol^{-1}}$.

a) Hvor mye varme tar kalorimeteret opp?

b) Hva er $q_V$ for etanolen som forbrennes?

c) Hvor mange mol etanol ble forbrent?

d) Beregn $\Delta U$ per mol etanol.

e) Hvorfor gir et bombekalorimeter $\Delta U$ direkte, mens et kaffekoppkalorimeter vanligvis gir $\Delta H$?

:::{dropdown} Løsningsforslag

a)

$$q_\mathrm{kal}=C_V\Delta T=(6{,}90)(2{,}15)=14{,}8\ \mathrm{kJ}$$

b) Reaksjonen avgir den varmen kalorimeteret tar opp:

$$q_V=-14{,}8\ \mathrm{kJ}$$

c)

$$n=\frac{0{,}500\ \mathrm{g}}{46{,}07\ \mathrm{g\,mol^{-1}}}=0{,}0109\ \mathrm{mol}$$

d) Ved konstant volum er $q_V=\Delta U$:

$$\Delta U_\mathrm{mol}=\frac{-14{,}8\ \mathrm{kJ}}{0{,}0109\ \mathrm{mol}}\approx-1{,}37\times10^3\ \mathrm{kJ\,mol^{-1}}$$

e) I bombekalorimeteret er volumet konstant, slik at trykk–volum-arbeidet er null og $q_V=\Delta U$. I kaffekoppkalorimeteret er trykket omtrent konstant, og da er $q_P=\Delta H$ når bare trykk–volum-arbeid er relevant.

:::
