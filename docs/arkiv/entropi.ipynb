## Fra multiplisitet til entropi

Vi repeterer kort utgangspunktet fra introduksjonen: En **makrotilstand** beskriver systemet med noen få makroskopiske størrelser, mens en **mikrotilstand** angir den fullstendige tilstanden innenfor modellen vi bruker. Antallet mikrotilstander som er forenlige med en bestemt makrotilstand, kalles **multiplisiteten** og betegnes $\Omega$.

For et isolert system med fast energi, volum og partikkeltall behandles de tilgjengelige mikrotilstandene som like sannsynlige. En makrotilstand med stor $\Omega$ kan da realiseres på flere måter og observeres oftere enn en makrotilstand med liten $\Omega$. Boltzmann knyttet denne tellingen til en termodynamisk tilstandsfunksjon:

:::{admonition} Boltzmanns entropi
:class: important

$$S=k_\mathrm{B}\ln\Omega,$$

der $k_\mathrm{B}=1{,}380\,649\times10^{-23}\ \mathrm{J\,K^{-1}}$ er Boltzmanns konstant.
:::

Logaritmen gjør entropien additiv. For to uavhengige delsystemer er

$$\Omega_\mathrm{tot}=\Omega_A\Omega_B,$$

og dermed

$$S_\mathrm{tot}=k_\mathrm{B}\ln(\Omega_A\Omega_B)=S_A+S_B.$$

Entropi er derfor ikke ganske enkelt «uorden». I denne sammenhengen er entropi et mål på hvor mange mikroskopiske muligheter som er forenlige med den makroskopiske tilstanden.

:::{admonition} Underveisoppgave
:class: tip
Et system kan realisere makrotilstand A på 10 måter og makrotilstand B på 1000 måter.

1. Hvilken makrotilstand har størst entropi?
2. Regn ut $S_B-S_A$ uttrykt ved $k_\mathrm{B}$.
3. Hvorfor er ikke entropiforskjellen proporsjonal med $1000-10$?
:::

:::{dropdown} Løsningsforslag
Makrotilstand B har størst entropi. Forskjellen er

$$S_B-S_A=k_\mathrm{B}\ln\!\left(\frac{1000}{10}\right)=k_\mathrm{B}\ln100\approx4{,}61k_\mathrm{B}.$$

Entropien avhenger av logaritmen til multiplisiteten. Det er forholdet mellom antall muligheter, ikke differansen mellom dem, som bestemmer entropiforskjellen.
:::


## Et kjemisk eksempel: gass i to kamre

Før vi fordeler energi mellom oscillatorer, begynner vi med en prosess vi allerede kjenner utfallet av.

Tenk deg en beholder delt i to like store kamre, $A$ og $B$, forbundet med en ventil. Til å begynne med er all gassen i kammer $A$, og ventilen er lukket. Så åpner vi ventilen. Vi vet alle hva som skjer: Gassen fyller hele beholderen. Men *hvorfor* skjer det helt av seg selv, og hvorfor ser vi aldri det motsatte – at gassen spontant trekker seg tilbake til $A$?

For en ideell gass beveger molekylene seg uavhengig av hverandre. Når de to kamrene er like store, er hvert molekyl til enhver tid like sannsynlig i $A$ som i $B$. Vi kan derfor sammenlikne hvert molekyl med et myntkast: kron $=$ kammer $A$, mynt $=$ kammer $B$.


![Fri ekspansjon av en gass mellom to kamre](bilder/gass_mikro_makro.svg)

Vi bruker begrepene fra introkapitlet: **Makrotilstanden** er antallet molekyler i $A$, som vi kaller $n_A$, mens en **mikrotilstand** angir *hvilke* molekyler som er i $A$. Antallet mikrotilstander er en binomialkoeffisient:

$$\Omega(n_A)=\binom{N}{n_A},$$

og det totale antallet mikrotilstander er $2^N$. Siden alle mikrotilstandene er like sannsynlige, blir sannsynligheten for å observere en gitt makrotilstand

$$p(n_A)=\frac{\Omega(n_A)}{2^N}=\frac{1}{2^N}\binom{N}{n_A}.$$

:::{admonition} Underveisoppgave
:class: tip
Sammenlikn modellen med mynteksemplet i introkapitlet, der makrotilstanden var «antall kron» og antallet mikrotilstander var $\binom{N}{k}$:

1. Hva svarer «kron» og «mynt» til?
2. Hvor mange mikrotilstander har makrotilstanden «alle molekyler i $A$»? Hva med «halvparten i hvert kammer» når $N$ er et partall?
3. Bruk $S=k_\mathrm{B}\ln\Omega$ til å forklare hvilken av disse makrotilstandene som har størst entropi.
:::

:::{dropdown} Løsningsforslag
«Kron» og «mynt» svarer til at et bestemt molekyl befinner seg i henholdsvis kammer $A$ og $B$. Tilstanden med alle molekylene i $A$ kan bare realiseres på én måte, mens en jevn fordeling kan realiseres på $\binom{N}{N/2}$ måter. Den jevne fordelingen har derfor langt større multiplisitet og dermed større entropi.
:::


```python
from math import comb
import numpy as np
import matplotlib.pyplot as plt

def omega_gass(N, nA):
    """Antall mikrotilstander for makrotilstanden 'nA molekyler i kammer A'."""
    return comb(N, nA)

def sannsynlighet_gass(N):
    """Analytisk fordeling p(nA) = C(N, nA) / 2**N."""
    nA = np.arange(0, N + 1)
    O  = np.array([omega_gass(N, int(k)) for k in nA], dtype=float)
    return nA, O / O.sum()          # summen er nettopp 2**N

def plott_gassfordeling(N):
    nA, p = sannsynlighet_gass(N)
    fig, ax = plt.subplots(figsize=(9, 4))
    ax.bar(nA, p * 100, color='tomato', edgecolor='white', linewidth=0.5)
    ax.axvline(N / 2, color='0.3', ls='--', lw=1.2)
    ax.text(N / 2 + 0.1, p.max() * 100 * 0.96,
            r'$n_A = N/2$' + '\nmaks entropi', fontsize=9, va='top')
    ax.annotate(f'p(alle tilbake i A)\n$= 1/2^{{{N}}} = {p[-1]:.2e}$',
                xy=(N, p[-1] * 100), xytext=(N * 0.55, p.max() * 100 * 0.5),
                fontsize=9, color='firebrick',
                arrowprops=dict(arrowstyle='->', color='firebrick', lw=1))
    ax.set_xlabel(r'Molekyler i kammer A  ($n_A$)', fontsize=11)
    ax.set_ylabel('Sannsynlighet (%)', fontsize=11)
    ax.set_title(f'Fordeling av $N = {N}$ molekyler mellom to like kamre')
    plt.tight_layout()
    plt.show()

plott_gassfordeling(10)
```

Fordelingen topper seg der molekylene er jevnt fordelt ($n_A = N/2$), og faller bratt mot kantene. Tilstanden «alle molekyler i $A$» ($n_A = N$) er bare *én* av $2^N$ mikrotilstander. Allerede med $N = 10$ molekyler er sannsynligheten for at gassen spontant samler seg i $A$ igjen under **0,1 %**. For et makroskopisk antall molekyler er sannsynligheten så liten at fri ekspansjon framstår som irreversibel — ikke fordi en kraft «drar» gassen utover, men på grunn av ren telling.

## Energi fordelt mellom oscillatorer

For å gjøre den statistiske definisjonen av entropi konkret trenger vi et system der vi faktisk kan telle mikrotilstandene. Vi bruker **Einstein-modellen** for et fast stoff. I modellen sitter atomene omtrent på faste plasser i et gitter, men de står ikke stille: De vibrerer rundt likevektsposisjonene sine. Vibrasjonene beskrives som uavhengige harmoniske oscillatorer.

Et tredimensjonalt fast stoff med $N_\mathrm{atom}$ atomer har $3N_\mathrm{atom}$ oscillatorer; her lar vi $N$ betegne antallet oscillatorer. Energien er kvantisert. Hver oscillator kan inneholde $0,1,2,\ldots$ energikvanter med energi $\varepsilon$. Vi kaller dem **energipakker**.

La $M$ være det totale antallet energipakker. En mikrotilstand angir nøyaktig hvor mange pakker hver oscillator har. Hvis oscillator $i$ har $m_i$ pakker, må

$$m_1+m_2+\cdots+m_N=M.$$

Med tre pakker og tre oscillatorer er for eksempel $(3,0,0)$, $(2,1,0)$ og $(1,1,1)$ tre forskjellige mikrotilstander. De har samme totalenergi, men energien er fordelt forskjellig. Nullpunktsenergien er den samme i alle fordelingene og påvirker derfor ikke tellingen.

Spørsmålet er nå enkelt å stille, men svaret vokser fort: På hvor mange måter kan vi fordele $M$ energipakker på $N$ oscillatorer?

$$\Omega(M,N)=\binom{M+N-1}{N-1}.$$

Dette er et «stjerner og skillevegger»-problem: De $M$ energipakkene er stjerner, og $N-1$ skillevegger deler dem mellom oscillatorene. Allerede med 100 pakker og 50 oscillatorer får vi et antall mikrotilstander som er langt utenfor det vi klarer å forestille oss. Det er nettopp denne eksplosive veksten som gjør statistisk termodynamikk nyttig: Når nesten alle mikrotilstandene tilhører et lite område av mulige makrotilstander, blir den makroskopiske oppførselen forutsigbar.

En mer detaljert gjennomgang av tellemetoden finnes i appendikset [Kombinatorikk](matematikk/kombinatorikk.ipynb).


```python
from math import comb, log, factorial
import numpy as np
import matplotlib.pyplot as plt

def omega(M, N):
    """Antall mikrotilstander: M energipakker fordelt på N oscillatorer."""
    return comb(M + N - 1, N - 1)

# Noen eksempler
eksempler = [(3, 3), (10, 8), (20, 16), (100, 50)]
print(f"{'M':>5}  {'N':>5}  {'Omega':>55}  {'ln Omega':>12}")
print("-" * 100)
for M, N in eksempler:
    o = omega(M, N)
    print(f"{M:>5}  {N:>5}  {o:>55,}  {log(o):>12.2f}")
```

## To systemer i termisk kontakt

Nå kommer det interessante: Hva skjer når to faste stoffer med ulik temperatur settes i kontakt slik at de kan utveksle energi?

Vi setter sammen to Einstein-faste stoffer, A og B. Stoff A starter med mer energi per oscillator enn stoff B. Energi kan flyttes mellom dem, men totalsystemet er isolert, slik at det totale antallet energipakker er bevart:

$$M_\mathrm{tot}=M_A+M_B=\text{konstant}.$$

For en bestemt energifordeling $(M_A,M_B)$ er multiplisiteten til totalsystemet

$$\Omega_\mathrm{tot}(M_A)=\Omega_A(M_A)\Omega_B(M_\mathrm{tot}-M_A).$$

Multiplikasjonen følger av at hver mikrotilstand i A kan kombineres med hver mikrotilstand i B. Fordelingen med størst $\Omega_\mathrm{tot}$ er derfor også fordelingen med størst total entropi:

$$S_\mathrm{tot}=k_\mathrm{B}\ln\Omega_\mathrm{tot}=S_A+S_B.$$

Simuleringen nedenfor viser energiutvekslingen. Hvert felt representerer en oscillator, og de glødende prikkene representerer energipakker. Stoff A starter med flere energipakker per oscillator enn stoff B. I hvert steg flyttes en tilfeldig pakke til en nabo. Ingen regel sier at pakken må flyttes fra A til B, og ingen av oscillatorene vet hvilken side som er varmest. Likevel ser vi en tydelig makroskopisk retning når systemet er stort nok.

Det er nettopp dette vi ønsker å forstå: Hvordan kan en serie tilfeldige, reversible energioverføringer gi en prosess som ser lovmessig og irreversibel ut?

2. La simuleringen gå lenge. Er $M_A$ konstant, eller fluktuerer den rundt en bestemt verdi?
3. Finn fordelingen der $\Omega_\mathrm{tot}$ er størst. Hvordan er energien da fordelt per oscillator?

```python
from math import comb

def omega(M, N):
    return comb(M + N - 1, N - 1)

NA, NB = 8, 8   # oscillatorer per stoff (4×4 grid, delt på midten)
M_tot  = 10     # totalt antall energipakker
init_MA = 7     # starttilstand for stoff A

omega_verdier = [omega(MA, NA) * omega(M_tot - MA, NB) for MA in range(M_tot + 1)]
total         = sum(omega_verdier)

p_varmere = sum(omega_verdier[MA] for MA in range(init_MA + 1, M_tot + 1)) / total
MA_maks = max(range(M_tot + 1), key=lambda MA: omega_verdier[MA])

print(f"p(stoff A varmere enn start) = {p_varmere*100:.2f} %")
print(f"Maksimal multiplisitet ved M_A = {MA_maks} og M_B = {M_tot-MA_maks}")
```

Koden ovenfor regner ut $\Omega_\mathrm{tot}$ for alle mulige verdier av $M_A$. Siden mikrotilstandene til det isolerte totalsystemet behandles som like sannsynlige, blir sannsynligheten for en bestemt energifordeling

$$p(M_A)=\frac{\Omega_A(M_A)\Omega_B(M_\mathrm{tot}-M_A)}{\sum_{M_A'=0}^{M_\mathrm{tot}}\Omega_A(M_A')\Omega_B(M_\mathrm{tot}-M_A')}.$$

Systemet «søker» ikke etter likevekt, og energipakkene prøver ikke å fordele seg rettferdig. Systemet beveger seg ganske enkelt mellom de tilgjengelige mikrotilstandene. Fordelingene nær maksimum av $\Omega_\mathrm{tot}$ observeres nesten hele tiden fordi et overveldende antall mikrotilstander tilhører disse makrotilstandene.

Likevekt er dermed ikke et sted systemet endelig stopper. Energipakkene fortsetter å flytte seg, og $M_A$ fluktuerer. Likevekt er det området av makrotilstander systemet statistisk sett befinner seg i nesten hele tiden.


## Temperatur måler hvordan entropien endres med energi

Temperatur er ikke det samme som total energi. Den termodynamiske definisjonen er

:::{admonition} Temperatur
:class: important

$$\frac{1}{T}=\left(\frac{\partial S}{\partial U}\right)_{V,N}.$$
:::

Uttrykket forteller hvor mye entropien øker når systemet får litt mer indre energi, mens volum og partikkeltall holdes konstante. Ved lav temperatur gir litt ekstra energi tilgang til relativt mange nye mikrotilstander, slik at $S$ øker mye. Ved høy temperatur gir den samme energimengden en mindre relativ økning i antallet tilgjengelige tilstander.

Vi kan nå vise hvorfor energi overføres fra varmt til kaldt. For de to systemene er

$$dU_\mathrm{tot}=dU_A+dU_B=0,$$

slik at $dU_B=-dU_A$. Endringen i total entropi er

$$
dS_\mathrm{tot}
=\frac{dU_A}{T_A}+\frac{dU_B}{T_B}
=\left(\frac{1}{T_A}-\frac{1}{T_B}\right)dU_A.
$$

Hvis $T_A>T_B$, er $(1/T_A-1/T_B)<0$. Total entropi øker derfor når $dU_A<0$: Energi overføres fra A til B, altså fra varmt til kaldt. Ved maksimal total entropi kan ingen liten energioverføring øke entropien videre:

$$dS_\mathrm{tot}=0\quad\Rightarrow\quad T_A=T_B.$$

:::{admonition} Kjerneidé
:class: important
Termisk likevekt kan beskrives på to ekvivalente måter:

- makroskopisk: Systemene har samme temperatur
- statistisk: Den totale entropien er maksimal med hensyn på energiutveksling
:::

:::{admonition} Underveisoppgave
:class: tip
System A har $T_A=400\ \mathrm K$, og system B har $T_B=300\ \mathrm K$. Anta at en liten energi $\delta U>0$ overføres fra A til B.

1. Sett $dU_A=-\delta U$ inn i uttrykket for $dS_\mathrm{tot}$.
2. Vis at $dS_\mathrm{tot}>0$.
3. Hva skjer med drivkraften når temperaturene nærmer seg hverandre?
:::

:::{dropdown} Løsningsforslag

$$dS_\mathrm{tot}=\left(\frac1{400}-\frac1{300}\right)(-\delta U)>0.$$

Parentesen er negativ, og $dU_A$ er negativ. Produktet blir derfor positivt. Når temperaturene nærmer seg hverandre, nærmer $(1/T_A-1/T_B)$ seg null. Ved lik temperatur er det ingen netto drivkraft for energiutveksling.
:::


## Termodynamikkens andre lov som statistisk lov

Det lille systemet viser oss prinsippet, men det som gjør termodynamikken *praktisk nyttig*, er at reelle systemer inneholder omtrent $10^{23}$ partikler. I et lite system kan energien av og til bevege seg «feil vei». I et makroskopisk system drukner slike fluktuasjoner fullstendig i antallet mulige mikrotilstander.

For et isolert system formuleres termodynamikkens andre lov som

$$\Delta S_\mathrm{tot}\geq0$$

for en spontan makroskopisk prosess. Den statistiske forklaringen er at makrotilstander med høy entropi svarer til langt flere mikrotilstander enn makrotilstander med lav entropi.

Når antallet partikler øker, vokser antallet mikrotilstander enormt, samtidig som sannsynlighetsfordelingen blir stadig smalere *relativt til systemets størrelse*. For en enkel tosidig modell av $N$ partikler avtar den relative bredden som

$$\frac{\sigma}{\mu}=\frac{1}{\sqrt{N}}.$$

For $N\approx10^{23}$ er dette omtrent $3\times10^{-12}$. Systemet skifter hele tiden mikrotilstand, men nesten alle mikrotilstandene vi observerer, tilhører makrotilstander svært nær entropimaksimum. En målbar, spontan reduksjon i total entropi er derfor ikke matematisk umulig, men så usannsynlig at den ikke observeres i et makroskopisk system.

:::{admonition} Viktig presisering
:class: note
Andre lov skyldes ikke en kraft som trekker systemet mot større entropi. Systemet utforsker tilgjengelige mikrotilstander, men et overveldende flertall av dem tilhører makrotilstander svært nær entropimaksimum.
:::

:::{admonition} Underveisoppgave
:class: tip
Forklar med egne ord hvorfor energi noen ganger kan overføres «feil vei» i et svært lite system, men ikke observeres å gjøre det som en vedvarende makroskopisk prosess.
:::

:::{dropdown} Løsningsforslag
I et lite system er sannsynlighetsfordelingen relativt bred, og fluktuasjoner bort fra entropimaksimum kan derfor observeres. Når systemet blir makroskopisk, blir den relative fordelingen svært smal. Da finnes fluktuasjonene fortsatt på molekylnivå, men en vedvarende makroskopisk energioverføring i retningen som reduserer total entropi, er overveldende usannsynlig.
:::


## Fra entropimaksimum til Boltzmann-fordelingen

Så langt har vi studert to deler av et isolert totalsystem. I laboratoriet er situasjonen ofte litt annerledes: Et relativt lite system står i kontakt med store omgivelser, for eksempel en prøve i et stort vannbad. Systemet kan da utveksle energi med omgivelsene, mens temperaturen i omgivelsene praktisk talt ikke endres.

Dette er viktig fordi systemet nå kan besøke mikrotilstander med forskjellig energi. De er ikke like sannsynlige. Hvis systemet er i mikrotilstand $i$ med energi $E_i$, må omgivelsene ha avgitt den samme energien. En høyere energi i systemet gir dermed færre tilgjengelige mikrotilstander i omgivelsene og en lavere samlet multiplisitet.

For et stort varmereservoar fører dette til den eksponentielle energivekten

$$e^{-E_i/k_\mathrm{B}T}.$$

Etter normalisering får vi Boltzmann-fordelingen:

:::{admonition} Boltzmann-fordelingen
:class: important

$$p_i=\frac{1}{Q}e^{-E_i/k_\mathrm{B}T},$$

der $Q$ er normaliseringsfaktoren som gjør at $\sum_i p_i=1$.
:::

Boltzmann-fordelingen er altså ikke en løsrevet regel. Den følger av at sannsynligheten for systemets mikrotilstand bestemmes av hvor mange mikrotilstander som samtidig er tilgjengelige for det store reservoaret. Forholdet mellom sannsynlighetene for to mikrotilstander er

$$\frac{p_j}{p_i}=e^{-(E_j-E_i)/k_\mathrm{B}T}.$$

Det er derfor energiforskjellen sammenliknet med $k_\mathrm{B}T$ som avgjør hvor ulikt tilstandene vektes.

:::{dropdown} Fordypning: fra reservoarets entropi til Boltzmann-faktoren
Når systemet mottar energi $E_i$, mister reservoaret den samme energien. For et stort reservoar kan entropien tilnærmes med

$$S_\mathrm{res}(U_\mathrm{tot}-E_i)\approx S_\mathrm{res}(U_\mathrm{tot})-\left(\frac{\partial S_\mathrm{res}}{\partial U_\mathrm{res}}\right)E_i=S_\mathrm{res}(U_\mathrm{tot})-\frac{E_i}{T}.$$

Siden $S=k_\mathrm{B}\ln\Omega$, er reservoarets multiplisitet proporsjonal med $e^{S_\mathrm{res}/k_\mathrm{B}}$. Den delen som avhenger av systemets mikrotilstand, blir dermed

$$e^{-E_i/k_\mathrm{B}T}.$$
:::

:class: tip
Et Einstein-faststoff har $N=4$ oscillatorer og $M=6$ energipakker.

1. Beregn $\Omega(6,4)$.
2. Beregn entropien uttrykt ved $k_\mathrm{B}$.
3. Beregn $\Omega(6,8)$ og forklar hvorfor entropien endres når antallet oscillatorer økes.

$$\Omega(6,4)=\binom93=84,\qquad S=k_\mathrm{B}\ln84\approx4{,}43k_\mathrm{B}.$$

For åtte oscillatorer er

$$\Omega(6,8)=\binom{13}{7}=1716,$$

og $S\approx7{,}45k_\mathrm{B}$. Flere oscillatorer gir flere måter å fordele den samme energien på.

:::{admonition} Oppgave 2: To systemer
:class: tip
To systemer har $N_A=4$ og $N_B=6$ oscillatorer og totalt åtte energipakker.

1. Beregn $\Omega_\mathrm{tot}(M_A)$ for $M_A=0,1,\ldots,8$.
2. Finn den mest sannsynlige energifordelingen.
3. Forklar hvorfor fordelingen ikke er symmetrisk rundt $M_A=4$.
:::

:::{dropdown} Løsningsforslag
Verdiene av $\Omega_\mathrm{tot}$ er $1287,3168,4620,5040,4410,3136,1764,720,165$. Maksimum ligger ved $M_A=3$ og $M_B=5$. Fordelingen er ikke symmetrisk fordi systemene har forskjellig antall oscillatorer.
:::

:::{admonition} Oppgave 3: Temperatur og entropi
:class: tip
To systemer kan utveksle en liten energimengde, men totalsystemet er isolert.

1. Utled $dS_\mathrm{tot}=(1/T_A-1/T_B)dU_A$.
2. Bruk fortegnet til å avgjøre retningen på energiutvekslingen når $T_A>T_B$.
3. Vis at entropimaksimum krever $T_A=T_B$.
:::

:::{dropdown} Løsningsforslag
Bruk $dS_A=dU_A/T_A$, $dS_B=dU_B/T_B$ og $dU_B=-dU_A$. Når $T_A>T_B$, er parentesen negativ, og $dS_\mathrm{tot}>0$ krever $dU_A<0$. Ved maksimum er den deriverte null, slik at $1/T_A=1/T_B$ og dermed $T_A=T_B$.
:::

:class: tip
To mikrotilstander har energiforskjellen $E_1-E_0=3k_\mathrm{B}T$.

1. Beregn $p_1/p_0$.
2. Hvordan endres forholdet dersom temperaturen dobles mens energigapet holdes fast?
:::

:::{dropdown} Løsningsforslag
Ved den første temperaturen er $p_1/p_0=e^{-3}\approx0{,}0498$. Når temperaturen dobles, blir energigapet $1{,}5k_\mathrm{B}T_\mathrm{ny}$, og forholdet blir $e^{-1{,}5}\approx0{,}223$. Den eksiterte tilstanden får større relativ sannsynlighet.
