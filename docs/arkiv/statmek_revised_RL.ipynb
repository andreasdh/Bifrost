## Mikro- og Makrotilstander: en kort repetisjon

Selv om vi kjenner de mulige mikrotilstandene for ett molekyl, finnes det et enormt antall måter tilstandene kan fordeles mellom mange molekyler. Én mikrotilstand for *hele systemet*, som er én realisering av systemets makrotilsdand, må angi tilstanden til samtlige partikler. Antallet mikrotilstander vokser derfor eksplosivt når vi går fra ett molekyl til et makroskopisk system.

Mange forskjellige mikrotilstander kan gi de *samme* makroskopiske egenskapene, altså samme makrotilstand. To systemer kan for eksempel ha samme temperatur, trykk og totale energi, selv om molekylene befinner seg i forskjellige kvantetilstander. **Antallet mikrotilstander som er forenlige med en bestemt makrotilstand, kaller vi multiplisiteten $\Omega$.**

Fra nå av bruker vi **partikkel** som et felles ord for atomer og molekyler. Atomer, molekyler, elektroner og protoner er alle eksempler på partikler; hvilken type vi mener, må framgå av modellen. Hvis ett energinivå har flere forskjellige mikrotilstander med samme energi, er nivået **degenerert**. Degenerasjonen betegner vi ofte med $g$.



## Multiplisitet for et 2 nivå system
2-nivå systemer er en god måte å illustrere multiplisitet på. Disse omfatter alle systemer hvor vi kan fordele ting inn i 2 kategorier. Dette kan for eksempel være at partikler kan befinne seg i en av 2 bokser A og B, eller at en partikkel kan være i en av 2 forskjellige energierilstander. 

Vi ser litt nærmere på to-nivåsystemet bestående av $N$ partikler, hvor hver kan være i en grunntilstand med energi $\varepsilon_0=0$ eller en eksitert tilstand med energi $\varepsilon$. Hvis $n$ av partiklene er eksiterte, er totalenergien $E=n\varepsilon$, uansett *hvilke* $n$ partikler det gjelder. Antallet måter å velge de $n$ eksiterte partiklene på er gitt ved multiplisiteten $\Omega(n)$:

$$\Omega(n) = \binom{N}{n} = \frac{N!}{n!(N-n)!}.$$

Dette er multiplisiteten, ikke en sannsynlighetsfordeling i seg selv. Dersom alle de $2^N$ mikrotilstandene var like sannsynlige, ville sannsynligheten for makrotilstanden med $n$ eksiterte partikler være:

 $$p(n)=\Omega(n)/2^N = \frac{N!}{n!\cdot (N-n)!}\cdot
\frac{1}{2}^n
\frac{1}{2}^{N-n}$$

Dette er et spesialtilfelle av den generelle **Binomialfordelingen**:  

 $$p(n)=\frac{N!}{n!\cdot (N-n)!}\cdot
p^n
(1-p)^{N-n}$$

der $p=\frac{1}{2}$, som følger av at alle mikrotilstander er like sansynilig. 

I simuleringen under kan du klikke på partiklene for å eksitere eller de-eksitere dem, eller trykke «ny mikrotilstand» for å se at forskjellige konfigurasjoner kan ha samme totalenergi. Legg særlig merke til forskjellen mellom *hvilken mikrotilstand systemet er i*, og *hvor mange mikrotilstander som gir den samme verdien av $n$*. 

Legg merke til at den totale energien i systemet her er gitt ved antallet partikler som er eksitert. Ofte så er det mer realistisk å se på systemer hvor den totale energien er gitt, og hvor vi da ønsker å hvordan det er mest sannsynlig at partiklene er fordelt mellom energinivåene. Svaret på dette er Boltzmann-fordelingen som vi kommer tilbake til.

## Multiplisitet og entropi (Utkast)

-- Boltzmann entropi (ln multiplisitet)

(Poengtere at entropi er ekstensiv, det er ikke multiplisitet, men ln av multiplisitet er det)

Kan finne endring i entropi fra endring i multiplisitet

For et [isolert system](https://andreasdh.github.io/Bifrost/docs/introduksjon.html#system-og-omgivelser) med fast energi, volum og partikkeltall (mikrokanionisk ensenble) behandles de tilgjengelige mikrotilstandene som like sannsynlige. En makrotilstand med stor $\Omega$ kan da realiseres på flere måter og observeres oftere enn en makrotilstand med liten $\Omega$. Boltzmann knyttet denne tellingen til en termodynamisk [tilstandsfunksjon](https://andreasdh.github.io/Bifrost/docs/introduksjon.html#prosesser-og-tilstandsfunksjoner):

:::{admonition} Boltzmanns entropi
:class: important

$$S=k_\mathrm{B}\ln\Omega,$$

der $k_\mathrm{B}=1{,}380\,649\times10^{-23}\ \mathrm{J\,K^{-1}}$ er Boltzmanns konstant.
:::

Logaritmen gjør entropien additiv. For to uavhengige delsystemer er

$$\Omega_\mathrm{tot}=\Omega_A\Omega_B,$$

og dermed

$$S_\mathrm{tot}=k_\mathrm{B}\ln(\Omega_A\Omega_B)=k_\mathrm{B}(\ln(\Omega_A)+\ln(\Omega_B))=k_\mathrm{B}\ln(\Omega_A)+k_\mathrm{B}\ln(\Omega_B)=S_A+S_B.$$

Og endrigen i entropi når et system går fra fra en tilstand A til en tilstand B kan utrykkes ved: 

$$\Delta S=k_\mathrm{B}\ln(\Omega_A)-k_\mathrm{B}\ln(\Omega_B)=S_A-S_B.$$

:::{admonition} Underveisoppgave
:class: tip
Et system kan realisere makrotilstand A på 10 måter og makrotilstand B på 1000 måter.

1. Hvilken makrotilstand har størst entropi?
2. Regn ut $S_B-S_A$ uttrykt ved $k_\mathrm{B}$.
3. Hvorfor er ikke entropiforskjellen proporsjonal med $1000-10$?
:::

:::{admonition} Løsningsforslag
:class: tip dropdown
Makrotilstand B har størst entropi. Forskjellen er

$$S_B-S_A=k_\mathrm{B}(\ln(\Omega_B)-\ln(\Omega_A))=k_\mathrm{B}\ln\!\left(\frac{1000}{10}\right)=k_\mathrm{B}\ln100\approx4{,}61k_\mathrm{B}.$$

Entropien avhenger av logaritmen til multiplisiteten. Det er forholdet mellom antall muligheter, ikke differansen mellom dem, som bestemmer entropiforskjellen.
:::





## Et kjemisk eksempel: gass i to kamre

Tenk deg en beholder delt i to like store kamre, $A$ og $B$, forbundet med en ventil. Til å begynne med er all gassen i kammer $A$, og ventilen er lukket. Så åpner vi ventilen. Vi vet alle hva som skjer: gassen fyller hele beholderen. Men *hvorfor* skjer det helt av seg selv, og hvorfor ser vi aldri det motsatte: at gassen spontant trekker seg tilbake til $A$?

For en ideell gass beveger molekylene seg uavhengig av hverandre, og når de to kamrene er like store, er hvert molekyl til enhver tid like sannsynlig i $A$ som i $B$. Man kan si at hvret molekyl er som et myntkast. Kron $=$ kammer $A$, mynt $=$ kammer $B$.

![Fri ekspansjon av en gass mellom to kamre](bilder/gass_mikro_makro.svg)

````python
from math import comb
import numpy as np
import matplotlib.pyplot as plt

def omega_gass(N, nA):
    """Antall mikrotilstander for makrotilstanden 'nA molekyler i kammer A'."""
    return comb(N, nA)

def sannsynlighet_gass(N):
    """Analytisk fordeling P(nA) = C(N, nA) / 2**N."""
    nA = np.arange(0, N + 1)
    O  = np.array([omega_gass(N, int(k)) for k in nA], dtype=float)
    return nA, O / O.sum()          # summen er nettopp 2**N

def plott_gassfordeling(N):
    nA, P = sannsynlighet_gass(N)
    fig, ax = plt.subplots(figsize=(9, 4))
    ax.bar(nA, P * 100, color='tomato', edgecolor='white', linewidth=0.5)
    ax.axvline(N / 2, color='0.3', ls='--', lw=1.2)
    ax.text(N / 2 + 0.1, P.max() * 100 * 0.96,
            r'$n_A = N/2$' + '\nmaks entropi', fontsize=9, va='top')
    ax.annotate(f'P(alle tilbake i A)\n$= 1/2^{{{N}}} = {P[-1]:.2e}$',
                xy=(N, P[-1] * 100), xytext=(N * 0.55, P.max() * 100 * 0.5),
                fontsize=9, color='firebrick',
                arrowprops=dict(arrowstyle='->', color='firebrick', lw=1))
    ax.set_xlabel(r'Molekyler i kammer A  ($n_A$)', fontsize=11)
    ax.set_ylabel('Sannsynlighet (%)', fontsize=11)
    ax.set_title(f'Fordeling av $N = {N}$ molekyler mellom to like kamre')
    plt.tight_layout()
    plt.show()

plott_gassfordeling(10)
````

Fordelingen topper seg der molekylene er jevnt fordelt ($n_A = N/2$), og faller bratt mot kantene. Tilstanden «alle molekyler i $A$» ($n_A = N$) er bare *én* av $2^N$ mikrotilstander. Allerede med $N = 10$ molekyler er sannsynligheten for at gassen spontant samler seg i $A$ igjen under **0,1 %**. Det er dette som gjør fri ekspansjon irreversibel — det er ikke en kraft som drar molekylene utover, men det er rett og slett mange flere muligheter for at det skal skje!

### Kvantetilstander og kvantisert energi

I klassisk mekanikk kan energien ofte endres kontinuerlig. For eksempel kan du løpe med økende fart, der alle verdier av farten din er tillatt (fram til du ikke klarer å løpe fortere, selvsagt). På atom- og molekylnivå er energien derimot **kvantisert**: Systemet kan bare ha bestemte tillatte energier:

$$\varepsilon_0,\varepsilon_1,\varepsilon_2,\ldots$$

Forskjellen mellom to tillatte energier er et **energigap**, $\Delta\varepsilon=\varepsilon_f-\varepsilon_i$. Et atom eller molekyl kan ta opp eller avgi energi når det går mellom to slike tilstander.

En **kvantetilstand** beskriver den mikroskopiske tilstanden til et kvantesystem, det vil her si små systemer på atom- og molekylnivå. I dette kapitlet bruker vi kvantetilstander med bestemt energi som mikrotilstander.

For et molekyl kan energien komme fra translasjon, rotasjon, vibrasjon og elektroniske tilstander. Alle bidragene er i prinsippet kvantiserte, men translasjonsnivåene ligger vanligvis så tett i et makroskopisk volum at de kan behandles som kontinuerlige. I statistisk mekanikk trenger vi først og fremst å kjenne de tillatte energiene; vi skal ikke løse de kvantemekaniske ligningene som gir dem.

:::{admonition} Det viktigste om kvantisering
:class: important
Kvantemekanikken bestemmer hvilke mikrotilstander og energier som er tillatt. Temperatur skaper ikke nye energinivåer, men påvirker hvor sannsynlig det er at systemet befinner seg i de ulike nivåene.
:::

## Ensembler: hvilke betingelser holder vi faste?

Tenk deg at vi vil forutsi varmekapasiteten til en gass fra energinivåene til molekylene. En eksperimentell måling gir ikke energien til én bestemt mikrotilstand, men en stabil makroskopisk verdi for svært mange molekyler som stadig skifter tilstand. For å kunne regne ut denne verdien må vi vite **hvilke betingelser forsøket foregår under**, **hvilke mikrotilstander systemet kan besøke**, og **hvordan vi skal ta gjennomsnittet over dem**. Et **ensemble** er et regneverktøy som presiserer nettopp dette.

Et ensemble er en tenkt samling av svært mange kopier av det samme systemet. Kopiene har de samme makroskopiske betingelsene, men kan befinne seg i forskjellige mikrotilstander. Ensemblet er altså et regneverktøy: I stedet for å følge ett system gjennom alle mikrotilstandene det besøker over tid, kan vi ta et gjennomsnitt over mange kopier på ett tidspunkt.

Hvilke mikrotilstander og sannsynligheter vi skal bruke, avhenger av hva systemet kan utveksle med omgivelsene. Vi må altså velge hvilket **ensemble** vi skal regne med ut fra hva slags system vi ønsker å studere. Navnene er mindre viktige enn den fysiske forskjellen mellom dem. I disse modulene trenger vi hovedsakelig to ensembler:

:::{list-table}
:header-rows: 1
:widths: 20 25 25 30

* - Ensemble
  - Faste størrelser
  - Fysisk modell
  - Konsekvens
* - Mikrokanonisk
  - $N$, $V$ og $E$
  - Isolert system
  - Bare mikrotilstander med den faste totalenergien er tilgjengelige. De behandles som like sannsynlige.
* - Kanonisk
  - $N$, $V$ og $T$
  - Lukket system i termisk kontakt med et stort varmereservoar
  - Systemet kan utveksle energi, slik at $E$ fluktuerer. Mikrotilstandene vektes med Boltzmann-faktoren.
:::

I den mikrokanoniske to-nivåmodellen ovenfor er totalenergien $E=n\varepsilon$ fast. Da er også $n$ fast, og $\Omega(n)$ teller mikrotilstandene som systemet kan besøke. De tilgjengelige mikrotilstandene behandles som like sannsynlige.

I det kanoniske ensemblet kan systemet motta og avgi energi. Det kan derfor besøke mikrotilstander med forskjellige totalenergier. Varmereservoaret (omgivelsene) antas å være så mye større enn systemet at temperaturen $T$ praktisk talt ikke endres når energi utveksles. Energiutvekslingen gjør at systemets energi varierer, men gjennomsnittsenergien er stabil ved likevekt.

:::{admonition} Ensemblet i resten av kapitlet
:class: important
Boltzmann-fordelingen, den kanoniske partisjonsfunksjonen og uttrykkene vi senere bruker for $U$ og $C_V$, gjelder for det **kanoniske ensemblet** med fast $N$, $V$ og $T$. Dette tilsvarer mange vanlige laboratorieforsøk, der en liten prøve kan utveksle varme med omgivelser som holder tilnærmet konstant temperatur. Andre systemer krever andre ensembler, som igjen krever andre sannsynlighetsfordelinger eller partisjonsfunksjoner.
:::

## Boltzmanns fordelingslov

Energien til et molekyl kan ligge i translasjon, rotasjon, vibrasjon eller elektronisk eksitasjon. Hvilke av disse energiformene som faktisk bidrar ved en gitt temperatur, avgjør blant annet stoffets indre energi, varmekapasitet og hvilke overganger vi kan observere spektroskopisk. For å svare på dette må vi vite hvor sannsynlig hver mikrotilstand er.

For et system i det kanoniske ensemblet gir **Boltzmanns fordelingslov** sannsynligheten for å finne systemet i mikrotilstand $i$ med energi $E_i$:

$$p_i = \frac{1}{Q}e^{-\frac{E_i}{k_BT}}.$$

Her er $k_B=1{,}380\,649\times10^{-23}\ \mathrm{J\,K^{-1}}$ Boltzmanns konstant, $T$ er reservoarets absolutte temperatur, og $Q$ er partisjonsfunksjonen som normaliserer sannsynlighetene slik at $\sum_i p_i=1$. Vi kommer tilbake til $Q$ i neste hovedavsnitt.

Hvis systemet er én partikkel eller ett molekyl med mikrotilstander $n$ og energier $\varepsilon_n$, bruker vi ofte små bokstaver:

$$p_n = \frac{1}{q}e^{-\frac{\varepsilon_n}{k_BT}}.$$

Denne énpartikkelformen kan også brukes for et stort system av uavhengige partikler når totalenergien kan skrives som en sum av enkeltpartiklenes energier. Hvis partiklene vekselvirker sterkt, må hele systemets mikrotilstander og energier behandles samlet.

:::{admonition} Termen $\: k_BT$
:class: important
Produktet $k_BT$ har enhet energi og setter den **termiske energiskalaen**. Det er ikke en generell formel for gjennomsnittsenergien til en partikkel; gjennomsnittsenergien avhenger av hvilke energinivåer og frihetsgrader systemet har. Forholdet $\Delta E/k_BT$ forteller hvor stort et energigap er sammenliknet med den termiske energiskalaen.
:::

I introkapitlet beskrev vi en sannsynlighetsfordeling som en oversikt over mulige utfall og sannsynligheten for hvert av dem. Her er utfallene systemets mikrotilstander. Fordelingen må oppfylle $0\leq p_i\leq1$ og $\sum_i p_i=1$. En mer systematisk innføring finnes i [Sannsynlighetsfordelinger og statistikk](matematikk/statistikk.ipynb).

Se på Boltzmann-faktoren $e^{-E_i/k_BT}$. Den styrer hvordan sannsynligheten for en *bestemt mikrotilstand* avtar med energien:

- Høyere $E_i$ $\rightarrow$ mer negativ eksponent $\rightarrow$ lavere sannsynlighet for den enkelte mikrotilstanden.
- Høyere $T$ $\rightarrow$ mindre forskjell mellom Boltzmann-faktorene $\rightarrow$ større sannsynlighet for eksiterte mikrotilstander.
- Lavere $T$ $\rightarrow$ større forskjell mellom Boltzmann-faktorene $\rightarrow$ sannsynligheten konsentreres mot mikrotilstandene med lavest energi.

Hvis et energinivå $E_j$ har degenerasjon $g_j$, er den samlede sannsynligheten for nivået proporsjonal med $g_j e^{-E_j/k_BT}$. Et høyere energinivå har altså lavere sannsynlighet *per mikrotilstand*, men mange degenererte mikrotilstander kan til sammen gi nivået en stor samlet sannsynlighet.

Det er energiforskjeller målt relativt til $k_BT$ som bestemmer forholdet mellom sannsynlighetene, ikke energien eller temperaturen alene.

I simuleringen over ser du seks ikke-degenererte energinivåer, jevnt fordelt med avstand $\Delta\varepsilon$ ($\varepsilon_n=n\Delta\varepsilon$). Glideren styrer $k_BT/\Delta\varepsilon$, altså den termiske energiskalaen sammenliknet med avstanden mellom nivåene. Bruk den til å se hva som skjer:
- Ved lav $k_BT/\Delta\varepsilon$ er nesten all sannsynlighet i grunntilstanden.
- Etter hvert som $k_BT/\Delta\varepsilon$ øker, får stadig flere av de høyere nivåene merkbar sannsynlighet.
- Ved svært høy $k_BT/\Delta\varepsilon$ nærmer fordelingen seg (nesten) jevn: alle nivåene blir omtrent like sannsynlige.

Dette forklarer hvorfor et system ved romtemperatur som regel *ikke* befinner seg i en høyt eksitert vibrasjonell eller elektronisk tilstand: Energitrinnene $\Delta\varepsilon$ er ofte store sammenliknet med $k_BT$.

For to bestemte mikrotilstander avhenger **forholdet mellom sannsynlighetene** bare av energiforskjellen og temperaturen, ikke av hvor vi velger nullpunktet for energi:

$$\frac{p(\varepsilon_1)}{p(\varepsilon_0)} = e^{-\frac{\varepsilon_1-\varepsilon_0}{k_BT}}$$

Dette forholdet kalles ofte **Boltzmann-faktoren**. Dersom vi sammenlikner den samlede sannsynligheten for to *energinivåer* med degenerasjoner $g_0$ og $g_1$, må degenerasjonen også tas med:

$$\frac{p(E_1)}{p(E_0)}=\frac{g_1}{g_0}e^{-\frac{E_1-E_0}{k_BT}}.$$

Boltzmanns fordelingslov er broen mellom mikro og makro. Når vi kjenner sannsynlighetene, kan vi regne ut makroskopiske størrelser ved å ta sannsynlighetsvektede gjennomsnitt over mikrotilstandene. Partisjonsfunksjonen gjør denne framgangsmåten systematisk.


:::{admonition} Hva er systemet i Boltzmann-fordelingen?
:class: important
Boltzmann-fordelingen gjelder mikrotilstandene til det systemet vi har valgt. Systemet kan være ett molekyl eller en hel gass, så lenge det er i termisk kontakt med et mye større reservoar ved temperatur $T$. Énpartikkelformen er særlig nyttig for ideelle eller svakt vekselvirkende systemer. Ved sterke vekselvirkninger må mikrotilstandene til hele systemet behandles samlet.
:::

:::{card}

**Eksempel på bruk av Boltzmanns fordelingslov: vibrasjon i et gassmolekyl**
^^^

Et gassmolekyl kan både translatere, rotere og vibrere. Én vibrasjonell normalmode kan modelleres som en kvantemekanisk harmonisk oscillator med energinivåene

$$E_v=\left(v+\frac12\right)h\nu, \qquad v=0,1,2,\dots$$

der $v$ er vibrasjonskvantetallet og $\nu$ er vibrasjonsfrekvensen. Nullpunktsenergien $\tfrac12h\nu$ er den samme i alle energiforskjeller. Hvis vi velger grunntilstanden som energireferanse, kan vi derfor skrive $\varepsilon_v=vh\nu$. I denne enkle modellen er nivåene ikke-degenererte.

La oss sammenligne sannsynligheten for at molekylet befinner seg i første eksiterte vibrasjonsnivå ($v=1$) med sannsynligheten for at det er i grunntilstanden ($v=0$).

```python
import numpy as np

E1 = 1  # vibrasjonskvantetall v=1, energi i enheter av h*nu
E0 = 0  # grunntilstand v=0

hv_over_kT = 1.0  # velg selv forholdet h*nu / (k_B T)

p_ratio = np.exp(-(E1 - E0) * hv_over_kT)
print(p_ratio)
```

Ved $h\nu/k_BT=1$ er forholdet $p(E_1)/p(E_0)\approx0{,}37$. Det er dermed mer sannsynlig å finne molekylet i grunntilstanden enn i hvert enkelt eksitert nivå. For en harmonisk oscillator med uendelig mange nivåer er den samlede sannsynligheten for grunntilstanden i dette tilfellet $p_0=1-e^{-1}\approx0{,}63$, så eksiterte nivåer gir fortsatt et betydelig samlet bidrag.

Jo høyere verdi av `hv_over_kT` (lavere temperatur, eller en stivere binding med høyere vibrasjonsfrekvens $\nu$), desto mindre blir sannsynligheten for de eksiterte vibrasjonsnivåene sammenliknet med grunntilstanden.
:::

## Partisjonsfunksjonen

Boltzmann-fordelingen forteller hvordan én mikrotilstand skal vektes, men et reelt system kan ha svært mange tilstander. Vi trenger derfor én størrelse som samler hele energinivåstrukturen. Når denne størrelsen er kjent, kan vi både normalisere sannsynlighetene og regne ut målbare egenskaper uten å behandle hvert makroskopisk spørsmål fra bunnen av. Denne størrelsen er partisjonsfunksjonen.

**Partisjonsfunksjonen** samler Boltzmann-faktorene til alle mikrotilstandene. Vi bruker ofte $q$ for ett molekyl eller én partikkel og $Q$ for hele systemet:

$$ q = \sum_n e^{-\frac{\varepsilon_n}{k_BT}} \text{,}$$

$$ Q = \sum_i e^{-\frac{E_i}{k_BT}} \text{.}$$

Definisjonene har samme form, men beskriver forskjellige systemer. Det er derfor viktig å være tydelig på om indeksene viser til mikrotilstander for ett molekyl eller for hele systemet. Fra nå av bruker vi $Q$ når vi omtaler de generelle egenskapene til partisjonsfunksjonen.

$Q$ er her den **kanoniske partisjonsfunksjonen** for et system med fast $N$, $V$ og $T$. Dersom systemet også kan utveksle partikler med omgivelsene, må vi bruke et annet ensemble og en annen partisjonsfunksjon.

Hvis vi sier at $t$ er antallet mikrotilstander, slik at $i=0,1,2,...,t-1$, og skriver ut $Q$, ser det slik ut:

$$
Q=\sum_{i=0}^{t-1} e^{-\frac{E_i}{k_BT}}
= e^{-\frac{E_0}{k_BT}}
+ e^{-\frac{E_1}{k_BT}}
+ e^{-\frac{E_2}{k_BT}}
+ \cdots
+ e^{-\frac{E_{t-1}}{k_BT}}.
$$

Partisjonsfunksjonen summerer altså de **unormaliserte Boltzmann-vektene** til alle mikrotilstandene. Den summerer ikke sannsynlighetene; sannsynlighetene får vi først når hver vekt deles på $Q$. Dermed er

$$p_i=\frac{e^{-E_i/k_BT}}{Q}.$$

Sammenhengen kan oppsummeres slik:
- Kvantemekanikken gir oss energinivåene.
- Statistisk mekanikk gir sannsynlighetene for disse.
- Partisjonsfunksjonen samler all informasjonen.

Vi må kjenne energinivåene for å regne ut $Q$, men det fysiske nullpunktet for energi kan velges fritt. Vi velger ofte grunntilstanden som referanse, $E_0=0$, og skriver funksjonen med energiforskjeller:

$$ Q = 1 + e^{-\frac{(E_1-E_0)}{k_BT}} + e^{-\frac{(E_2-E_0)}{k_BT}} + \cdots + e^{-\frac{(E_{t-1}-E_0)}{k_BT}} \text{.}$$

Vi kan også gruppere mikrotilstander med samme energi. Hvis energinivå $E_j$ har degenerasjon $g_j$, får vi

$$ Q=\sum_j g_j e^{-\frac{E_j}{k_BT}}.$$

Dette er den samme summen som før; vi har bare samlet like ledd. For diskrete nivåer bruker vi betegnelsen **degenerasjon**. **Tilstandstetthet** brukes vanligvis om antall tilstander per energiintervall når energiene behandles som kontinuerlige, og er derfor ikke helt det samme som $g_j$.

Når grunntilstanden er valgt som energireferanse, kan $Q$ gi en nyttig intuisjon for hvor mange tilstander som bidrar merkbart: Hver mikrotilstand bidrar med et tall mellom 0 og 1. $Q$ er likevel ikke bokstavelig talt et antall tilstander, og verdien av $Q$ endres dersom vi flytter energinullpunktet. Sannsynligheter, energiforskjeller og målbare energiendringer er uavhengige av dette valget.

Siden flere mikrotilstander deler samme energi, kan vi bruke $Q=\sum_j g_j e^{-E_j/k_BT}$ direkte, med bare to energinivåer:

$$g_0 = 1 \quad \text{(lukket)}, \qquad g_1 = 3 \quad \text{(åpen)}$$

$$Q = g_0e^{0} + g_1e^{-\Delta\varepsilon/k_BT} = 1+3e^{-\Delta\varepsilon/k_BT}\text{.}$$

Sannsynlighetene for hver makrotilstand blir da

$$p_{\text{lukket}} = \frac{1}{Q}\text{,} \qquad p_{\text{åpen}} = \frac{3e^{-\Delta\varepsilon/k_BT}}{Q}\text{.}$$

Ved lav temperatur ($\Delta\varepsilon/k_BT \to \infty$) domineres molekylet av den lukkede tilstanden, og $Q\to 1$. Merk at $p_{\text{lukket}}$ ikke går mot null her, men mot $1/4$, siden den lukkede tilstanden fortsatt er én av de fire like sannsynlige mikrotilstandene.

### Uavhengige delsystemer og identiske partikler

Når vi går fra énpartikkelfunksjonen $q$ til partisjonsfunksjonen $Q$ for $N$ partikler, må vi vite både om energien kan adderes, dvs om den er uavhengig, og hvordan mikrotilstandene skal telles. Identiske kvantepartikler er fundamentalt **uskillbare**, men i en krystall der partiklene vibrerer rundt faste gitterplasser kan man allikevel bruke en tilnærming om at de er skillbare og dermed telles unikt. 

 Hvis de lokale bevegelsene kan behandles som uavhengige, er totalenergien en sum:

 For $N$ ikke-vekselvirkende skillbare partikler er en mikrotilstand $(i_1,i_2,\ldots,i_N)$ med total energi $ E_i=\varepsilon_{i_1}+\varepsilon_{i_2}+\cdots+\varepsilon_{i_N}$.

Den kanoniske partisjonsfunksjonen faktoreiseres og blir da:

$Q = \sum_{i_1,\ldots,i_N} e^{-\beta(\varepsilon_{i_1}+\cdots+\varepsilon_{i_N})} = \left(\sum_i e^{-\beta\varepsilon_i}\right)^N$ og dermed:

$$Q_{\text{lokaliserte delsystemer}}=q^N.$$

Her er det **gitterplassene eller de lokale frihetsgradene** som er nummererte i modellen, ikke de identiske atomene som fundamentale partikler. Modellen er nyttig når utveksling mellom plassene kan neglisjeres, for eksempel for lokale vibrasjoner i et krystallgitter.

For en fortynnet ideell gass av $N$ identiske partikler gir $q^N$ derimot en klassisk telling der partiklene i praksis er nummerert. Da telles samme fysiske konfigurasjon $N!$ ganger. I Maxwell–Boltzmann-grensen korrigerer vi overtellingen med Gibbs-faktoren:

$$Q_{\text{klassisk idealgass}}\approx\frac{q^N}{N!}.$$

Tilnærmingen gjelder når gassen er tilstrekkelig fortynnet og temperaturen er høy nok til at kvanteeffekter knyttet til partikkelutveksling kan neglisjeres. Ved høy tetthet eller svært lav temperatur må identiske bosoner og fermioner behandles med henholdsvis Bose–Einstein- og Fermi–Dirac-statistikk; da er $q^N/N!$ ikke generelt riktig.

:::{admonition} Hva er det som egentlig endres?
:class: important
Identiske partikler er alltid uskillelige. Det som endres mellom modellene, er hvilke frihetsgrader vi beskriver og hvordan vi teller de fysiske mikrotilstandene. $Q=q^N$ beskriver her $N$ uavhengige, nummererte delsystemer. $Q\approx q^N/N!$ er den klassiske grensen for $N$ identiske, mobile partikler.
:::

## Indre energi fra partisjonsfunksjonen

I introkapitlet beskrev vi indre energi som en tilstandsfunksjon og varme og arbeid som energi overført under en prosess. Nå skal vi se hvordan indre energi kan regnes ut fra partisjonsfunksjonen. Vi bruker uttrykkene uten å utlede dem; utledningene er ikke pensum i Fysikalsk kjemi I.

Hovedpoenget er å se at en makroskopisk størrelse kan bygges fra mikrotilstandenes energier og sannsynligheter. Det er her den abstrakte konstruksjonen gir uttelling: Energistigen til molekylene, Boltzmann-fordelingen og partisjonsfunksjonen kan brukes til å forutsi størrelser vi måler i laboratoriet.

**Indre energi** er forventningsverdien, eller ensemblegjennomsnittet $\langle E\rangle$, av mikrotilstandsenergiene $E_i$, vektet med sannsynlighetene $p_i$:

$$U= \langle E \rangle = \sum_i E_i p_i=\frac{1}{Q}\sum_i E_i e^{-\frac{E_i}{k_BT}}\text{,}$$

der vi har brukt Boltzmanns fordelingslov. Når energinivåene ikke selv avhenger av temperaturen, kan det vises at dette er det samme som den deriverte av $\ln Q$ med hensyn på temperaturen:

$$ U = k_BT^2\left(\frac{\partial \ln Q}{\partial T}\right)_{V,N} \text{.}$$

Noen lærebøker skriver i stedet dette uttrykket ved hjelp av $\beta \equiv \frac{1}{k_BT}$, som gir en litt penere form:

$$ U = -\left(\frac{\partial \ln Q}{\partial \beta}\right)_{V,N}\text{.}$$

:::{admonition} Notasjon
:class: note
Notasjonen $\langle E\rangle$ betyr «gjennomsnittlig energi».

Et sannsynlighetsvektet gjennomsnitt betyr at en mikrotilstand med stor sannsynlighet bidrar mer til gjennomsnittet enn en mikrotilstand med liten sannsynlighet. Legg også merke til at symbolet $q$ brukes både for énpartikkel-partisjonsfunksjonen og for varme i termodynamikken. Betydningen må leses av sammenhengen.
:::

I en prosess går systemet fra en tilstand A til en tilstand B. Partisjonsfunksjonen gir den indre energien i hver tilstand:

$$ U_A = -\left(\frac{\partial \ln Q_A}{\partial \beta_A}\right)_{V,N},$$

$$ U_B = -\left(\frac{\partial \ln Q_B}{\partial \beta_B}\right)_{V,N}.$$

Forskjellen $\Delta U=U_B-U_A$ kan bestemmes eksperimentelt fra varme og arbeid ved hjelp av termodynamikkens første lov:

$$ \Delta U = U_B - U_A = q + w, $$

der $q$ er varme tilført systemet og $w$ er arbeid utført på systemet med den kjemiske fortegnskonvensjonen. Dermed beskriver $Q$ den mikroskopiske opprinnelsen til $U$, mens $q$ og $w$ beskriver hvordan $U$ endres under en makroskopisk prosess.

### Varmekapasitet

Varmekapasiteten ved konstant volum er rett og slett hvor mye $U$ endrer seg med temperaturen:

$$ C_V = \left(\frac{\partial U}{\partial T}\right)_{V,N}\text{.}$$

Siden vi allerede har $U$ uttrykt ved $Q$, kan $C_V$ i prinsippet regnes ut fullstendig fra partisjonsfunksjonen alene.

Partisjonsfunksjonen gir altså først $U(T)$. Når vi deretter undersøker hvor raskt $U$ endres med temperaturen, får vi $C_V$.

## Hvorfor varierer varmekapasitet?

### Ekvipartisjonsteoremet

**Ekvipartisjonsteoremet** sier at hvert uavhengige **kvadratiske energiledd** i den klassiske energifunksjonen bidrar med $\tfrac12k_BT$ til gjennomsnittsenergien. Det er mer presist enn å si at enhver frihetsgrad alltid bidrar med $\tfrac12k_BT$: En vibrasjonsmode har både et kvadratisk kinetisk og et kvadratisk potensielt energiledd og bidrar derfor med $k_BT$ i den klassiske grensen.

:::{card} 

**Eksempel: $\text{O}_2$ - et diatomisk molekyl**
^^^

Et oksygenmolekyl har i den klassiske stiv-rotor/harmonisk-oscillator-modellen:
- 3 translasjonelle frihetsgrader (beveger seg i tre romlige retninger).
- 2 rotasjonelle kvadratiske energiledd. Rotasjon om molekylaksen tas ikke med i stiv-rotor-modellen fordi treghetsmomentet om denne aksen er tilnærmet null.
- 1 vibrasjonell normalmode (bindingen strekkes og trekkes sammen).

Vibrasjonen bidrar med **to** kvadratiske energiledd: ett kinetisk og ett potensielt. Den ene vibrasjonsmoden gir derfor to bidrag i ekvipartisjonsregnskapet. Totalt får vi

$$3+2+2=7 \text{ kvadratiske ledd} \quad\Rightarrow\quad U_\mathrm{termisk}=\frac{7}{2}k_BT$$

per molekyl, eller $U_{\mathrm{m,termisk}}=\tfrac72RT$ per mol, relativt til det valgte energinullpunktet.
:::

Kvantemekanikken er den generelle teorien på mikroskopisk nivå. Ekvipartisjonsteoremet er en klassisk tilnærming som blir gyldig når energinivåene ligger tett sammenliknet med $k_BT$. Når et energigap blir stort sammenliknet med $k_BT$, er sannsynligheten for de eksiterte nivåene liten, og denne energiformen bidrar mindre enn den klassiske verdien. Vi sier ofte at frihetsgraden **fryses ut**. Ekvipartisjonsteoremet er derfor den klassiske grensen av den kvantemekaniske beskrivelsen.

La oss repetere frihetsgradene og se hvilke av dem som fryses ut først, og hvorfor. Dette er faktisk grunnen til at varmekapasiteten til reelle gasser avviker fra den enkle klassiske forventningen.

### Translasjon

Translasjonsnivåene ligger svært tett. Energigapet er så lite sammenliknet med $k_BT$ at translasjon bidrar tilnærmet klassisk ved praktisk talt alle temperaturer vi møter i laboratoriet.

### Rotasjon

Rotasjonsnivåene har små energigap sammenlignet med vibrasjon, og aktiveres derfor lett. Ved romtemperatur er $k_BT$ som regel god nok til at rotasjon oppfører seg tilnærmet klassisk for de fleste molekyler.

### Vibrasjon

Vibrasjonsnivåene har derimot store energigap. For mange molekyler ved romtemperatur er $\Delta\varepsilon_{\text{vib}} \gg k_BT$, så vibrasjonen fryses ut: molekylet sitter nesten utelukkende i vibrasjonens grunntilstand, og bidrar dermed ikke med sin fulle klassiske andel til $U$ eller $C_V$. Dette er nøyaktig grunnen til at oksygengass ($O_2$) og nitrogengass ($N_2$) ved romtemperatur har en varmekapasitet nær $\frac{5}{2}R$ (bare translasjon + rotasjon), ikke $\frac{7}{2}R$ som den fulle klassiske telningen over skulle tilsi. Vibrasjonen deres er rett og slett for "stiv" til å tine ved den temperaturen.

Simuleringen under viser nøyaktig denne utfrysningen på et enkelt diatomisk molekyl:

Trekk temperaturen ned, og legg merke til at vibrasjonsbidraget avtar først fordi vibrasjonsnivåene vanligvis har størst energigap. Deretter avtar rotasjonsbidraget. Translasjonsnivåene ligger så tett at translasjon vanligvis er i den klassiske grensen ved temperaturene vi møter i kjemilaboratoriet.

Simuleringen viser overgangen som idealiserte trinn, $\tfrac32R\to\tfrac52R\to\tfrac72R$. I et virkelig kvantemekanisk system endres populasjonene og $C_V$ **gradvis** med temperaturen. «På» og «av» er derfor en pedagogisk tilnærming til to grenseverdier:

- Hvis $\Delta\varepsilon \ll k_BT$: klassisk grense, ekvipartisjonen holder, frihetsgraden bidrar fullt.
- Hvis $\Delta\varepsilon \gg k_BT$: nivåene er frosset, ekvipartisjonen bryter sammen for denne frihetsgraden.

### Varmekapasitet

Varmekapasiteten ved konstant volum, $C_V=\left(\frac{\partial U}{\partial T}\right)_{V,N}$, måler hvor mye den indre energien øker per temperaturøkning. I de klassiske og utfrosne grenseverdiene kan $C_V$ brukes som en omtrentlig teller for hvilke energiformer som bidrar. Mellom grensene bidrar en kvantisert mode bare delvis. $C_V$ er derfor ikke et fast tall bestemt av molekylstrukturen alene.

To ting ved molekylstrukturen avgjør varmekapasiteten: hvor mange frihetsgrader molekylet har, og hvor stort energigapet er for hver av dem.

**Antall atomer og geometri** bestemmer hvordan molekylets $3N$ romlige koordinater fordeles mellom translasjon, rotasjon og vibrasjon:

- Lineære molekyler (som $O_2$ og $CO_2$): 3 translasjonskoordinater + 2 rotasjonskoordinater + $(3N-5)$ vibrasjonsmoder.
- Ikke-lineære molekyler (som $H_2O$): 3 translasjonskoordinater + 3 rotasjonskoordinater + $(3N-6)$ vibrasjonsmoder.

Et lineært molekyl har bare to rotasjonskoordinater i stiv-rotor-modellen, mens et ikke-lineært molekyl har tre. Derfor står det igjen $3N-5$ vibrasjonsmoder for et lineært molekyl og $3N-6$ for et ikke-lineært molekyl. En nærmere oversikt finnes under «Energitilstander og frihetsgrader» i appendikset [Grunnleggende begreper](grunnleggende_begreper.ipynb).

**Atommasse og bindingsstyrke** påvirker størrelsen på energigapene. Selv om to molekyler har like mange moder, kan bidragene deres til $C_V$ derfor bli merkbare ved forskjellige temperaturer. Molekylstrukturen bestemmer både hvor mange energiformer som finnes, og hvordan bidragene deres utvikler seg med temperaturen.

## Oppsummering

**Kvantemekanikk**
- Kvantemekanikken bestemmer hvilke energitilstander som er tillatt.
- Kvantetilstand = fullstendig beskrivelse av alle frihetsgrader (elektronisk, vibrasjon, rotasjon, translasjon).
- Translasjon er teknisk sett også kvantisert, men nivåene ligger så tett at de oppfører seg som kontinuerlige.

**Mikrotilstander og multiplisitet**
- Mikrotilstand $\leftrightarrow$ én kvantetilstand (for ett enkelt atom/molekyl).
- Degenererte tilstander: flere kvantetilstander deler samme energi, telles med $g$.
- For systemer av mange partikler: én mikrotilstand = kombinasjonen av alle partiklenes kvantetilstander samtidig.
- Multiplisitet $\Omega$ = antall mikrotilstander som er forenlige med en bestemt makrotilstand. Degenerasjon $g_j$ = antall mikrotilstander med samme diskrete energi $E_j$.

**Boltzmanns fordelingslov**
- $p_i=Q^{-1}e^{-E_i/k_BT}$: sannsynligheten for én mikrotilstand i det kanoniske ensemblet.
- En mikrotilstand med lav energi er mer sannsynlig enn én mikrotilstand med høy energi ved samme temperatur. Populasjonen til et degenerert energinivå avhenger også av $g_j$.
- Forholdet $\Delta\varepsilon/k_BT$ bestemmer Boltzmann-faktoren og dermed forholdet mellom sannsynlighetene.

**Partisjonsfunksjonen**
- $q=\sum_n e^{-\varepsilon_n/k_BT}$ (ett molekyl).
- $Q=\sum_i e^{-E_i/k_BT}$ (helt system).
- Kan også skrives med degenerasjon: $Q=\sum_j g_j e^{-E_j/k_BT}$.
- Når grunntilstanden er energireferanse, kan $Q$ tolkes som et mål på hvor mange tilstander som bidrar merkbart. $Q$ er ikke bokstavelig talt et antall tilstander.
- Uavhengige, nummererte delsystemer: $Q=q^N$. Identiske partikler i den klassiske, fortynnede grensen: $Q\approx q^N/N!$. Identiske partikler er fundamentalt uskjelnbare i begge tilfeller.

**Indre energi og varmekapasitet fra $Q$**
- $U=\langle E\rangle = k_BT^2\left(\frac{\partial \ln Q}{\partial T}\right)_{V,N}$: ensemble-gjennomsnitt over mikrotilstandsenergiene.
- $C_V=\left(\frac{\partial U}{\partial T}\right)_{V,N}$.

**Ekvipartisjon og varmekapasitet**
- Hvert uavhengige kvadratiske energiledd bidrar $\tfrac12k_BT$ til $U$ i den klassiske grensen. En harmonisk vibrasjonsmode har ett kinetisk og ett potensielt ledd.
- Ekvipartisjon = klassisk grense av kvantemekanikken, gyldig når $\Delta\varepsilon \ll k_BT$.
- Bidrag med store energigap avtar ved høyere temperatur enn bidrag med små energigap. For mange molekyler avtar derfor vibrasjonsbidraget før rotasjonsbidraget ved avkjøling.
- Molekylstruktur (antall atomer, lineær/bøyd geometri, atommasse, bindingsstivhet) bestemmer både *hvor mange* frihetsgrader som finnes og *ved hvilken temperatur* hver av dem våkner, derfor varierer $C_V$ fra molekyl til molekyl.

## Øvingsoppgaver

:::{admonition} Øvingsoppgave 1: Multiplisitet og ensemble
:class: tip

Fem partikler kan ha energien $0$ eller $\varepsilon$. Et isolert system har totalenergien $E=2\varepsilon$.

1. Hvor mange partikler er eksiterte?
2. Regn ut multiplisiteten $\Omega$.
3. Forklar hvorfor systemet ikke kan besøke en mikrotilstand med $E=3\varepsilon$.
4. Beskriv hva som endres dersom systemet settes i termisk kontakt med et stort reservoar.
:::

:::{admonition} Løsningsforslag
:class: tip dropdown

To partikler må være eksiterte fordi $E=n\varepsilon$, og multiplisiteten er $\Omega=\binom{5}{2}=10$. I det mikrokanoniske ensemblet er totalenergien fast, så en mikrotilstand med tre eksiterte partikler er ikke tilgjengelig. I det kanoniske ensemblet kan systemet utveksle energi med reservoaret. Da kan antallet eksiterte partikler variere, og sannsynlighetene bestemmes av Boltzmann-fordelingen.
:::

:::{admonition} Øvingsoppgave 2: Energi konkurrerer med degenerasjon
:class: tip

Et system har et grunntilstandsnivå med $E_0=0$ og $g_0=1$, og et eksitert nivå med degenerasjon $g_1=4$. Ved temperaturen vi undersøker, er $E_1-E_0=2k_BT$.

1. Sammenlikn sannsynligheten for én bestemt mikrotilstand i det eksiterte nivået med grunntilstanden.
2. Regn ut partisjonsfunksjonen.
3. Regn ut den samlede sannsynligheten for at systemet befinner seg i det eksiterte nivået.
:::

:::{admonition} Løsningsforslag
:class: tip dropdown

For én bestemt mikrotilstand er $p_1/p_0=e^{-2}\approx0{,}135$. Når degenerasjonen tas med, blir

$$Q=1+4e^{-2}\approx1{,}541,$$

og den samlede sannsynligheten for det eksiterte nivået er

$$p(E_1)=\frac{4e^{-2}}{1+4e^{-2}}\approx0{,}351.$$

Hver eksiterte mikrotilstand er mindre sannsynlig enn grunntilstanden, men de fire mikrotilstandene gir et betydelig samlet bidrag.
:::

:::{admonition} Øvingsoppgave 3: Partisjonsfunksjon og indre energi
:class: tip

Et ikke-degenerert to-nivåsystem har energiene $E_0=0$ og $E_1=\Delta E$. Temperaturen er slik at $\Delta E=k_BT$. Regn ut $Q$, sannsynlighetene $p_0$ og $p_1$, og den indre energien $U=\langle E\rangle$.
:::

:::{admonition} Løsningsforslag
:class: tip dropdown

Partisjonsfunksjonen er $Q=1+e^{-1}$. Dermed er

$$p_0=\frac{1}{1+e^{-1}}\approx0{,}731, \qquad p_1=\frac{e^{-1}}{1+e^{-1}}\approx0{,}269.$$

Gjennomsnittsenergien blir

$$U=E_0p_0+E_1p_1=\frac{\Delta E}{e+1}\approx0{,}269\,\Delta E.$$
:::

:::{admonition} Øvingsoppgave 4: Hvordan skal tilstandene telles?
:class: tip

Sammenlikn $N$ uavhengige, lokaliserte vibrasjonsmoder i et gitter med $N$ identiske atomer i en fortynnet idealgass. Forklar hvorfor vi kan bruke $Q=q^N$ i den første modellen, men $Q\approx q^N/N!$ i den andre. Drøft påstanden: «Atomene blir skjelnbare når gassen fryser til et fast stoff.»
:::

:::{admonition} Løsningsforslag
:class: tip dropdown

I gittermodellen er det de nummererte gitterplassene eller lokale vibrasjonsmodene som faktoriserer. Hver plass bidrar med den samme lokale partisjonsfunksjonen $q$, slik at $Q=q^N$. For den klassiske idealgassen ville $q^N$ telt en fysisk konfigurasjon én gang for hver ombytting av de $N$ identiske partiklene. Vi deler derfor på $N!$ i den fortynnede Maxwell–Boltzmann-grensen. Identiske atomer er fundamentalt uskjelnbare i begge faser. Det er modellen og tilstandstellingen som endres, ikke partiklenes identitet.
:::

Hva blir $C_{V,\mathrm m}$ for hver gass? Hva skjer med $C_{V,\mathrm m}$ for $N_2$ når temperaturen økes så mye at vibrasjonen nærmer seg den klassiske grensen?

I et virkelig molekyl er overgangen jevn, ikke et brått trinn.
