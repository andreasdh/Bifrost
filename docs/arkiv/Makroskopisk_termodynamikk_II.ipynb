Hvorfor må vi varme opp noen reaksjoner, mens andre må stå på isbad? Hva mener vi med at en reaksjon er spontan? Nå som vi har lært om de termodynamiske potensialene og termodynamiske drivkrefter er vi endelig klare for å studere den praktiske anvendelsen av disse størrelsene i kjemifaget: La oss gå løs på termokjemi!🧪

**Termokjemien** beskriver kjemiske prosesser ved hjelp av de termodynamiske formlene for entalpi, entropi og fri energi, som vi møtte i det [forrige kapittelet](https://andreasdh.github.io/Bifrost/docs/Makroskopisk_termodynamikk_1.html#hvorfor-trenger-vi-flere-termodynamiske-potensialer). Det er mange ulike prosesser som kan foregå innen kjemifaget, men de du kommer til å lære mest om  i Fysikalsk Kjemi I er _faseoverganger_, _løsninger og blandinger_ og ikke minst _kjemiske reaksjoner_! Ofte kan en prosess være svært sammensatt. Dette er særlig sant for kjemiske reaksjoner, som ofte går via én eller flere overgangstilstander mellom reagens og ønsket produkt. I første omgang skal vi derimot kun interessere oss for egenskapene til start- og sluttilstandene i de ulike prosessene. Ved å sammenlikne disse tilstandene med hverandre vil vi kunne avgjøre om en prosess er _spontan_ eller ikke og for hvilke temperaturer dette gjelder. Vi skal også se på hvordan ulike prosesser kan drives av entalpi eller entropi, begge samtidig eller ingen av dem! Vi skal også knytte dette opp mot den praktiske kjemien ved å diskutere labforhold og reaksjoner du kjenner igjen fra tidligere kurs.🥼🔬

## Entalpi: Ekstoterme og endoterme prosesser

```{admonition} Hess lov
:class: important 

I [tilstandsfunksjoner](https://andreasdh.github.io/Bifrost/docs/introduksjon.html#prosesser-og-tilstandsfunksjoner) kan veien mellom start- og sluttilstand velges fritt. Det å dele prosessen opp i delprosesser svarer til å endre veien men så lenge start- og sluttilstanden er de samme vil endringen i tilstandsfunksjonen _totalt_ være den samme. Det å finne termodynamiske størrelser eksperimentelt kan være utfordrende. Istedenfor holder det å vite start og slutt-tilstandene. Veien mellom tilstandene kan da konstrueres ved å legge sammen kjente reaksjoner og man kan regne ut mange termodynamiske størrelser basert på tabeller over eksperimentelt målte verdier. **Hess lov** gjelder for alle tilstandsfunksjoner, men uttrykkes for eksempel slik for entalpi: 

$$\Delta H = \sum \Delta H(produkter) - \sum \Delta^{\circ} _f H(reaktanter)$$

Dersom vi har delprosessene A, B, C, vil vi heller bruke loven slik: 

$\Delta H = \Delta H(A) + \Delta H(B) + \Delta H(C)$ 

Innad i disse delprosessene har allerede forskjellen i entalpi mellom start- og sluttilstand blitt beregnet eller målt i henhold til formelen over. I oppgaver om Hess lov videre vil du bli bedt om å finne uttrykk som dette basert på gitte delprosesser eller delprosesser du finner selv. 
```

### Entalpi beskriver varmeutvekslinger i løpet av en kjemisk prosess

Som vi har forklart tidligere gir den termodynamiske størrelsen entalpi et bilde på [varmeutveksling ved konstant trykk](https://andreasdh.github.io/Bifrost/docs/Makroskopisk_termodynamikk_1.html#entalpi-nyttig-ved-konstant-trykk). Siden de fleste kjemiske prosesser foregår under konstant trykk, kan altså entalpien fortelle oss hvordan varme flyter inn og ut av systemet vi er interessert i, i løpet av en prosess. Vi kan introdusere to nye begreper som er svært viktige i kjemien: Dersom systemet gir fra seg varme til omgivelsene, er prosessen _eksoterm_. Dersom det tar opp varme fra omgivelsene, er prosessen derimot _endoterm_. Ved en eksoterm prosess vil endringen i entalpi $\Delta H$ ha negativt fortegn og ved en endoterm prosess vil endringen i entalpi $\Delta H$ ha positivt fortegn. Men hvordan beregner vi egentlig en slik endring? For å forstå dette må vi introdusere de mange typene entalpi du kan møte på i fysikalsk kjemi.

**Grunnstoffers entalpi** 

Som et referansepunkt har rene grunnstoffer i sin mest stabile tilstand en entalpi lik 0. For eksempel er mange gasser i sin mest stabile tilstand som dimerer: $H_2$, $O_2$ eller $N_2$. 

**Dannelsesentalpi**

Dannelsesentalpi er entalpien (varmen ved konstant trykk) som må til for å danne $1 \text{ mol}$ av et stoff fra grunnstoffene det består av. For eksempel har dannelsen av vann fra oksygen og hydrogen i sine mest stabile tilstander ($O_2$ og $H_2$) en entalpi på $\Delta H_f = –285.84 kJ/mol$. Ofte snakker vi om _standard_ dannelsesentalpi, og dette betyr ved _standard_ forhold, altså ved $25^\circ\mathrm{C}$ og 1 atm trykk. Da betegner vi dannelsesentalpien slik: $\Delta^{\circ} H_f$ hvor $\Delta$ betegner endring i løpet av prosessen, $^\circ$ betegner standard forhold og subskript f betegner "formation" på engelsk. Disse dannelsesentalpiene er tabulerte verdier du kan benytte deg av for å regne videre på entalpier for faseoverganger og reaksjoner. De har blitt bestemt ved å gjøre kalorimetri-målinger under dannelsen av stoffer fra grunnstoffer. Dannelsesentalpi er altså basisen for alle entalpiberegninger, og er et nyttig verktøy kombinert med Hess lov!

**Entalpi knyttet til faseoverganger**

Noen typer entalpi er knyttet opp mot faseoverganger: De vanligste er fordampingsentalpi som betegnes med subskript "vap" ("vaporisation" på engelsk) og smelteentalpi som betegnes med subskript "fus" ("fusion" på engelsk). Her beskriver altså overganger mellom ulike faser av ett og samme stoff. For eksempel har vi for fordamping av vann, en fordampingsentalpi $\Delta H_{vap} \approx 40.7 kJ/mol$. Vi må altså tilføre varme for å fordampe vann, som stemmer overens med erfaringene våre både fra lab og kjøkkenet!

**Reaksjonsentalpi**

Reaksjonsentalpi beskriver varmeendring (ved konstant trykk) knyttet til kjemiske reaksjoner. Her er dannelsesentalpier veldig nyttige! Vi kan dele opp en reaksjon i flere trinn og vurdere entalpien for hvert av trinnene før vi setter dem sammen for å beregne den totale reaksjonsentalpien slik Hess lov tillater. Slik bestemmer vi om en reaksjon er endoterm eller ekstoterm. Det er viktig for en kjemiker eller materialviter å ta stilling til hvor mye varme som vil genereres eller brukes opp i løpet av en reaksjon da dette vil påvirke hvordan vi designer et eksperiment eller setter opp en reaksjon. Vi skal se på et konkret eksempel av dette i slutten av dette avsnittet.

**Andre entalpier**

Noen typer entalpier faller ikke innenfor kategoriene vi har nevnt ovenfor, men det er allikevel lure å ha hørt om noen av dem. _Forbrenningsentalpi_ er entalpiendringen når et mol stoff fullstendig forbrennes i oksygen og betegnes med $\Delta H_c$ ( c fra engelsk "combustion"). Løsningsentalpi er mye brukt i kjemi og er definert som entalpiendringen når ett mol av et stoff løses i en gitt mengde løsemiddel. Entalpien betegnes som regel med $\Delta H_{sol}$ (fra engelsk "solvation"). Et siste eksempel er nøytraliseringsentalpi som beskriver entalpiendringen når en base og en syre reagerer for å danne vann og et salt. Entalpien betegnes med $\Delta H_{neut}$ (fra engelsk "neutralization").

### Et steg tilbake: Kan vi forutse entalpiendring basert på molekylstruktur?

I løpet av en kjemisk prosess ser vi ofte en endring i sammensetningen til systemet vi er interesserte i: I kjemiske reaksjoner forvandles reagenser til produkter ved at kjemiske bindinger brytes og dannes. I faseoverganger endres de _intermolekylære_ kreftene som holder molekylene i systemet sammen. Og i løsninger kan for eksempel salter _solvatiseres_ ved at ionene omringes av løsemiddel molekyler. For enkelte prosesser er det ikke sikkert vi har tilgang på dannelsesentalpiene vi trenger for å regne på $\Delta H$. Da kan det være nyttig å kunne resonnere seg frem til hva slags entalpiendring vi forventer for prosessen basert på de molekylære kreftene i systemet. 

Kort fortalt kan vi gjøre dette ved å vurdere den relative styrken til bindingene som brytes og dannes i løpet av prosessen! _Svake bindinger_ gir atomene mye frihet til å vibrere, rotere og bevege seg rundt og ligger derfor høyt i energi. Fordi et system alltid tenderer mot å minimere energien sin, byttes gjerne svake bindinger ut med _sterke bindinger_ som ligger lavere i energi da systemet blir mer rigid og atomene har mindre frihet til vibrasjon, rotasjon og translasjon. Energigapet mellom de to tilstandene frigjøres som varme, og det er slik vi ender opp med **eksoterme prosesser**. Vi skal se på et konkret eksempel på en eksoterm reaksjon i organisk kjemi i slutten av avsnittet. Vi kan også gå motsatt vei: I faseoverganger fra væske til gass bytter vi ut _sterke interaksjoner_ som gjør at væsken holdes sammen med _svake interaksjoner_ som lar gassen flyte fritt i rommet. I dette tilfellet må systemet ta opp varme fra omgivelsene som det bruker til å "tette" energigapet mellom tilstandene. Slik får vi **endoterme prosesser** hvor vi må tilføre varme til systemet. Hva som driver de endoterme prosessene fremover kommer vi inn på i neste avsnitt! Intermolekylære krefter skal vi lære mer om i neste kapittel.

### Et steg frem: Hvilke hensyn må en kjemiker/materialviter ta i møte med entalpiendring?

Selv om vi som kjemikere og materialvitere har en tendens til å tenke på praktisk labarbeid som en helt annen verden enn fysikalsk og teoretisk kjemi, bruker vi hele tiden fysikalske konsepter for å drive prosessene vi ønsker å gjennomføre fremover og ikke minst for å bevare sikkerheten på laben. Å ha kontroll på varmeoverføring er et av de viktigste hensynene vi tar når vi designer eksperimenter og synteser. I noen analysemetoder er til og med varmeoverføring det sentrale parameteret vi måler for å studere systemets egenskaper! 

**Ekstoterme prosesser**

Eksoterme prosesser gir fra seg varme i ulik grad. Noen kan foregå i romtemperatur og man kan kun kjenne at glassutstyret blir varmt, mens andre vil generere så mye varme at glassutstyr kan eksplodere dersom vi ikke kjøler systemet ned samtidig som prosessen får gå. Dette gjøres gjerne med isbad som holder $0 ^{\circ} C$ eller tørris/aceton bad som holder $-78 ^{\circ} C$. Et annet hensyn vi ofte tar er å tilsette reagenser forsiktig slik at litt og litt varme genereres i stedet for alt på én gang. 

**Endoterme prosesser** 

Endoterme prosesser trenger ulik grad av varmetilførsel for å gå. Noen reaksjoner i organisk kjemi kan man varme til $50 ^{\circ} C$ for at de skal gå, mens andre reaksjoner, særlig i uorganisk materialkjemi krever nærmere $1000 ^{\circ} C$ for å skje slik vi ønsker. Dersom prosessen skal gjøres i løsemiddel som står i fare for å koke bort ved høye temperaturer bruker vi ofte _kjølere_ som hjelper løsemiddelet å kondensere og dryppe tilbake i reaksjonsblandingen. I andre tilfeller ønsker vi nettopp å fjerne organiske stoffer, og gjør derfor oppvarming i ventilerte ovner slik at de organiske stoffene kan fordampe som gass.

**Regne på total entalpiendring**

Når vi skal regne på den totale entalpiendringen $\Delta H$ er vi interesserte i å sammenlikne dannelsesentalpiene til speciene som er tilstede ved starten og slutten av reaksjonen. Vi jobber derfra utifra den totale reaksjonslikningen: 

$$Mg_{(s)} + CH_3Br_{(g)} + H_2CO_{(g)} + H_2O_{(l)} \longrightarrow CH_3CH_2OH_{(l)} + Mg^{2+}_{(aq)} + Br^-_{(aq)} + OH^-_{(aq)}$$

Formelen for reaksjonsentalpien (Hess lov) som sier at reaksjonentalpien $\Delta^{\circ} _r H$ er lik summen av de standard dannelsesentapiene $\Delta^{\circ}_f H$ til produktene minus summen av de standard dannelsesentalpiene til reaktantene: 

$$\Delta^{\circ} _r H = \sum \Delta^{\circ} _f H(produkter) - \sum \Delta^{\circ} _f H(reaktanter)$$

I eksempelet vårt er vi heldige: Vi kjenner til tabulerte standard dannelsesentalpier for alle speciene i reaksjonslikningen vår og slipper derfor å dele opp reaksjonen vår i flere fiktive trinn! Faktisk hadde dette gjort oppdraget vårt vanskeligere, da det ikke finnes etablerte tabellverdier for dannelsen av Grignard-reagenset på grunn av forbindelsens lave stabilitet! La oss nå sette opp uttrykket vårt: 

Produktside:

| Stoff       | ΔfH° (kJ/mol) |
|------------|---------------|
| CH₃CH₂OH(l) | -277,6        |
| Mg²⁺(aq)    | -466,9        |
| Br⁻(aq)     | -121,6        |
| OH⁻(aq)     | -230,0        |
| Sum         | -1096,1       |


Reaktantside:

| Stoff    | ΔfH° (kJ/mol) |
|----------|---------------|
| Mg(s)    | 0             |
| CH₃Br(g) | -37,7         |
| HCHO(g)  | -108,6        |
| H₂O(l)   | -285,8        |
| Sum      | -432,1        |

Totalt får vi: 

$$ \Delta^{\circ} _r H = -1096.1 kJ/mol - (-432.1 kJ/mol) = -664.0 kJ/mol $$

For hvert mol av produkt som dannes frigis altså 664 kJ med energi. Vi har altså en svært eksoterm reaksjon slik vi forutså ved hjelp av vår studie av molekylstruktur! Her ser vi et eksempel på hvordan mikroskopiske egenskaper og makroskopiske størrelser kan illustrere et konsept på ulike måter. 


:::{admonition} Underveisoppgave
:class: tip

Standard reaksjonsentalpi kan regnes ut fra dannelsesentalpiene til stoffene:

$\Delta^{\circ}_r H = \sum \Delta^{\circ}_f H(\text{produkter}) - \sum \Delta^{\circ}_f H(\text{reaktanter})$

Vi ser på forbrenning av metan:

$CH_4 + 2\,O_2 \rightarrow CO_2 + 2\,H_2O$

Standard dannelsesentalpier er $\Delta^{\circ}_f H(CH_4) = -74.6\ \text{kJ/mol}$, $\Delta^{\circ}_f H(O_2) = 0\ \text{kJ/mol}$, $\Delta^{\circ}_f H(CO_2) = -393.5\ \text{kJ/mol}$ og $\Delta^{\circ}_f H(H_2O) = -285.8\ \text{kJ/mol}$.

**a)** Summer dannelsesentalpiene for produktene. Husk å gange med de støkiometriske tallene.

**b)** Summer dannelsesentalpiene for reaktantene.

**c)** Bruk formelen til å finne $\Delta^{\circ}_r H$. Er reaksjonen eksoterm eller endoterm?

:::


:::{dropdown} Løsningsforslag

**a)** Produktene er $CO_2$ og $2\,H_2O$:

$\sum \Delta^{\circ}_f H(\text{produkter}) = \Delta^{\circ}_f H(CO_2) + 2\,\Delta^{\circ}_f H(H_2O) = -393.5 + 2(-285.8) = -965.1\ \text{kJ/mol}$

**b)** Reaktantene er $CH_4$ og $2\,O_2$:

$\sum \Delta^{\circ}_f H(\text{reaktanter}) = \Delta^{\circ}_f H(CH_4) + 2\,\Delta^{\circ}_f H(O_2) = -74.6 + 2(0) = -74.6\ \text{kJ/mol}$

(Grunnstoffer i sin stabile form, som $O_2$, har per definisjon $\Delta^{\circ}_f H = 0$.)

**c)** Vi setter inn i formelen:

$\Delta^{\circ}_r H = -965.1 - (-74.6) = -890.5\ \text{kJ/mol}$

Reaksjonsentalpien er negativ, så reaksjonen er eksoterm - den frigjør varme. Det stemmer godt med at metan brenner og avgir mye energi.

:::



**Labprosedyre**

Grignard-reaksjonens eksoterme karakter fører også til tilpasninger i prosedyren for reaksjonen! Først, hva ville skjedd dersom vi bare blandet alt av reagensene sammen i løsemiddelet på én gang? Vi vet at reaksjonen er svært eksoterm, så derfor ville en rask og plutselig blanding av reagenser føre til mye varmeutvikling. Løsemiddelet, som regel eter, har et ganske lavt kokepunkt og vil fosskoke. Dette fører til to problemer: Én, når løsemiddelet koker bort løses reagensene dårligere og dårligere og vi risikerer at reaksjonen ikke går som vi ønsker. To, og aller viktigst, trykket vil øke inne i glassutstyret etterhvert som flere og flere løsemiddelmolekyler går over i gassfase. Raskt vil trykket bli så høyt at korken på kolba vil skyte av og eksponere blandingen for luft. Dette kan være farlig da Grignard-reagenset reagerer kraftig med vannet som finnes i lufta. Vi risikerer rett og slett at kolba eksploderer på grunn av den raske varmeutviklingen, og dette kan være svært farlig for den som står på lab! Hvordan skal vi unngå dette? Vi gjør tre hovedgrep for å holde Grignard-reaksjonen under kontroll: 

1. **Reaksjonen gjøres i fullstendig tørt glassutstyr og under inert atmosfære** (som regel argon- eller nitrogen-atmosfære) for å hindre Grignard-reagensen i å reagere med vann på glasset eller i lufta. 
2. **Reaksjonsblandingen kjøles ned** til $0$ eller $-78^\circ\mathrm{C}$ før tilsettingen av Grignard-reagenset.
3. **Grignard-reagensen tilsettes litt og litt** så ikke alle molekylene reagerer samtidig og den totale varmen som genereres av gangen blir mindre. 
:::

## Trender for entropiendring

Nå som vi har snakket lenge om entalpi, er det tid for å introdusere den andre komponenten av Gibbs fri energi: Entropi. Dere har allerede fått en grundig gjennomgang av entropi som både mikroskopisk og makroskopisk termodynamisk størrelse, så i dette avsnittet skal vi heller introdusere entropien i lys av hvilke prossesser som øker og minsker entropien innad i et system. For dette er det nyttig å definere entropien med ord og ikke et matematisk uttrykk. Vi henter frem igjen et uttrykk fra kapitellet om entropi: **Entropi er et mål på antall mikroskopiske måter et system kan realisere sin makroskopiske tilstand på**. La oss først tenke på entropien i noen overordnede kjemiske systemer: 

**Faste stoffer**

Faste stoffer kjennetegnes ved at de er _ordnede_ systemer, i hvert fall når vi sammenlikner dem med væsker og gasser. Partiklene i et fast stoff sitter i gittere. Nesten alle posisjonene i gitteret er opptatt av partikler og derfor er det lite mulighet for hver enkelt partikkel å bevege seg rundt. Disse faste posisjonene gjør at systemet kan arrangeres på få måter, og derfor har det tilgang til få mikrotilstander for å realisere sin makroskopiske tilstand. Dette forklarer hvorfor faste stoffer generelt har lav entropi. Dette forklarer også de makroskopiske karakteristikkene til faste stoffer: De kan ofte være harde og flyter ikke utover slik væsker og gasser gjør. 

**Væsker og løsninger**

Væsker og løsninger eksisterer et sted mellom faste stoffer og gasser, det er verken fullstendig ordnede eller fullstendig uordnede. De flyter mye friere rundt enn et fast stoff gjør, men henger fremdeles sammen i motsetning til gasser som får spre seg i et åpent rom. I kjemi er vi særlig interesserte i løsninger. Et fast stoff som løses i en væske øker entropien sin, men løsemiddelet ordnes også i større grad enn før rundt det løste stoffet. Nyansene rundt entropi og løsninger forklarer vi nærmere senere i avsnittet. 

**Gasser**

Gasser kjennetegnes ved at de er _uordnede_ systemer i forhold til faste stoffer og væsker. I et åpent system får gass slippe ut og vil derfor oppta et stort volum, for eksempel et helt avtrekksskap. Dette gjør at gassen har utrolig mye plass å boltre seg på! Gasspartiklene har utallige plasser de kan være og derfor kan systemet arrangeres på svært mange måter. Med andre ord har det tilgang på mange mikrotilstander for å realisere sin makrotilstand og vil derfor ha en høy entropi! De fleste prosesser som innebærer dannelse av gass er entropisk drevne.

## Gibbs fri energi forutsier spontanitet

Mot slutten av forrige avsnitt begynte vi å beskrive samspillet mellom entalpisk og entropisk drevne reaksjoner. Gibbs fri energi er et uttrykk for denne balansen og blir dermed et slags regnskap som kan fortelle oss hvorvidt en prosess vil skje _spontant_ eller ikke. En **spontan** prosess er en prosess som skjer av seg selv, uten ytre påvirkning, ved en gitt temperatur T. Under minner vi om uttrykket for Gibbs fri energi:  

$$ \Delta G = \Delta H - T\Delta S$$

Vi kan også introdusere nye begreper: Når $\Delta G > 0$ har vi _endergone prosesser_ som tar opp fri energi fra omgivelsene og ikke er spontane. Når $\Delta G < 0$ har vi _eksergone prosesser_ som gir fra seg fri energi til omgivelsene og er spontane. _Entalpisk drevne_ prosesser defineres ved at de er så eksoterme at det negative $\Delta H$-leddet i uttrykket for fri energi er dominerende. _Entropisk drevne_ prosesser har så stor entropigevinst ved en gitt temperatur at $-T\Delta S$-leddet er dominerende. Videre i dette avsnittet forklarer vi balansegangen mellom entalpisk og entropisk drevne prosesser i kjemien ved hjelp av eksempler.

### Hva med prosesser som er verken eller? 

Noen reaksjoner er verken entropisk eller entalpisk gunstige og derfor ikke spontane. Dersom vi skal få disse prosessene til å skje, må vi utsette systemet for ytre faktorer for å hjelpe prosessene i gang. I dette avsnittet skal vi kort snakke om noen av de vanligste måtene å gjøre dette på!


**Koble prosessen til en sterkt eksergon prosess**

Hvis en ikke-spontan reaksjon ($\Delta G_1 > 0$) skjer samtidig med en sterkt spontan reaksjon ($\Delta G_2 < 0$, med $|\Delta G_2| > |\Delta G_1|$), blir summen $\Delta G_{tot} = \Delta G_1 + \Delta G_2 < 0$, og den koblede totalprosessen blir spontan. Dette er selve grunnprinsippet bak biologisk energiomsetning: hydrolyse av $ATP$ til $ADP + P_i$​ er sterkt eksergon ($\Delta G \approx -30 \text{ kJ/mol}$) og kobles til utallige ellers ikke-spontane prosesser i cellen: Muskelkontraksjon, aktiv transport over membraner, biosyntese av makromolekyler osv. Nøkkelen er at prosessene deler et felles mellomtrinn eller intermediat (f.eks. et fosforylert enzym), slik at de reelt er koblet og ikke bare skjer parallelt. 

**Tilføre arbeid utenfra**
Man kan drive en ikke-spontan prosess ved å tilføre energi i en annen form enn varme, for eksempel elektrisk arbeid eller lys:

Elektrolyse er et klassisk eksempel: nedbryting av vann til $H_2$​ og $O_2$​ er sterkt ikke-spontant ($\Delta G > 0$), men ved å påtrykke en ytre spenning (tilføre elektrisk arbeid $w_{elek}$​) kan vi likevel drive reaksjonen. 

Fotosyntese er naturens variant: lysenergi driver den sterkt ikke-spontane omdannelsen av $CO_2$​ og $H_2O$ til glukose og $O_2$.

**Katalyse**

En katalysator senker aktiveringsenergien og øker dermed reaksjonshastigheten — den endrer altså _kinetikken_, ikke termodynamikken. Siden $G$ er en tilstandsfunksjon (avhenger kun av start- og sluttilstand), påvirker en katalysator verken $\Delta G$, $\Delta H$ eller $\Delta S$ for reaksjonen i det hele tatt. En katalysator senker aktiveringsenergien slik at det blir lettere for prosessen å gå mellom start- og sluttilstanden: Veien endres. I praksis skjer dette ofte ved at katalysatoren bidrar til å stabilisere overgangstilstanden. Dette er helt sentrale temaer i kinetikk, som vi skal se på til slutt i kurset.

### OBS: Spontanitet betyr ikke at noe skjer raskt!

Helt til slutt i dette kapitellet om termokjemi er det viktig å understreke forskjellen mellom termodynamikk og kinetikk. Vi har snakket mye om hvorfor noen reaksjoner er spontane og andre ikke, men hva innebærer egentlig spontanitet i praksis? Som vi beskrev i introduksjon har vi i dette kapitellet vært mest opptatt av å studere forskjellen mellom start- og sluttilstanden til et system før og etter en prosess. Vi har ikke brydd oss om veien fra den ene tilstanden til den andre. Når vi skal forstå kjemiske reaksjoner må vi derimot også betrakte **hvordan** vi kommer oss fra reaktant til produkt. Spontane reaksjoner er reaksjoner som er termodynamisk gunstige: Sluttilstanden er mer stabil enn starttilstanden. Men spontantinet sier ingenting om hvor raskt en reaksjon vil skje. Noen ganger har vi en situasjon hvor **overgangstilstanden** mellom en start- og sluttilstand som er svært ugunstig, ofte på grunn av lav stabilitet. Da spiller det ingen rolle om sluttilstanden er mer stabil, for systemet må uansett ta til seg nok energi for å komme over energibarrieren til overgangstilstanden, hvis det i det hele tatt er mulig. Altså kan noen spontane reaksjoner for eksempel bruke hundrevis, tusenvis eller flere år på å skje. Satt på spissen har vi mange reaksjoner vi beregner at er spontane, men som vi ser på som umulige å gjennomføre fordi de **i praksis** aldri vil skje. Vi må altså passe oss for å ikke overtolke de termodynamiske resultatene våre, og alltid ta med **kinetikken**, studien av hvor raskt prosesser skjer, med i betrakningen.
