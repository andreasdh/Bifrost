### Kvantisert energi

**Kvantemekanikken** er en matematisk beskrivelse av oppførselen til enkelte molekyler og atomer eller partikler. Tidligere var det antatt at klassisk fysikk var nok til å beskrive disse systemene, men eksperimentelle funn (Ultrafiolett katastrofe, dobbeltspalteeksperimentet, fotoelektrisk effekt, Comptonspredning, Stern-Gerlach eksperimentet mm.) stemte ikke overens med fysikken slik vi kjente den, og en helt ny gren ble nødvendig for å forklare funnene. Felles for etterarbeidet med mange av disse eksperimentene var at **kvantisering** av energi ble forklaringen for å få observasjonene til å gi mening. Vi bruker her Franck-Hertz eksperimentet til å illustrere hva kvantisering er og det faktum at det er en iboende egenskap ved vårt univers. Eksperimentet selv er ikke del av pensum, men den konseptuelle forståelsen er det. 

Klassisk fysikk handler om kontinuerlige spektre av størrelser; kjører du oppover en bakke får du og bilen din (systemet) litt og litt mer potensiell energi mens du beveger deg oppover. Vår virkelighetsoppfattelse utfordres når dette ikke lenger stemmer i kvanteverdenen. **Franck-Hertz eksperimentet** ble utført i 1914 av James Franck og Gustav Hertz, og ga oss det første beviset på atomenes **diskrete**, kvantiserte energinivåer. 

:::{admonition} Diskret energi
:class: important
En diskret verdi er en spesifikk, adskilt verdi. Tenk på trinnene i en stige. Det finnes ikke noe trinn nummer $1.5$ eller $\frac{1}{4}$, kun $0, 1, 2$ osv, og det er umulig å flyte i luften mellom trinnene. Når vi snakker om diskret energi, snakker vi om helt bestemte verdier av en eller annen energi $E_0, E_1, E_2,...$ som et system kan ha, og det er ikke mulig for systemet å ha en energi med en verdi som ligger mellom disse. Dette har du trolig vært borti før når vi snakker om eksitasjon og deeksitasjon av elektroner i orbitaler. 
:::

I Franck-Hertz eksperimentet ble elektroner sendt gjennom en beholder fylt med kvikksølv ($\text{Hg}$) i gassform, dvs. kvikksølvatomer. Franck og Hertz kontrollerte hvor mye energi elektronene hadde før de kolliderte med kvikksølvatomene, $\epsilon_i$, for så å måle hvor mye energi elektronene hadde igjen etter de hadde kollidert, $\epsilon_f$. 

![Franck-Hertz oppsett](Simuleringer/franck_hertz_oppsett.png)

Basert på klassisk fysikk forventet de at elektronene skulle avgi energi til kvikksølvatomene på en gradvis måte; dersom elektronene fikk litt høyere startenergi, skulle de også miste litt mer energi i kollisjonene. De observerte derimot noe helt annet: 

1. Når elektronet hadde lavere enn én bestemt startenergi $\epsilon_i < \Delta \epsilon$, kom det uendret ut av kollisjonen.
2. Når elektronet hadde startenergi $\epsilon_i=\Delta \epsilon$ eller høyere, mistet det alltid *samme mengde* energi $\Delta \epsilon$ under kollisjonen, og dersom startenergien ble økt litt mer, endret det ikke måleverdien. 
3. Først ved enda høyere startenergier kunne elektronet miste *to ganger* denne "minimumsenergien", altså $2\Delta \epsilon$. 
4. Ved enda høyere startenergier enn dette igjen kunne elektronet miste $3\Delta \epsilon$, $4\Delta \epsilon$ osv.

Den grå og blå grafen under representerer henholdsvis forventede observasjoner og hva som faktisk ble observert.

Hvordan skulle dette tolkes? Den eneste forklaringen som passet observasjonene var at kvikksølvatomene *kun* kunne ta opp helt bestemte energimengder, eller **energipakker**, med verdi $n\Delta \epsilon$, der $n=1,2,3,...$ er et naturlig tall. Atomene kunne ikke absorbere vilkårlige energier, de absorberte helt bestemte, diskrete energimengder. De lovlige energitilstandene, eller nivåene, atomene kan eksistere i er dermed **kvantiserte**, dvs. de kan kun eksistere i tilstander av diskrete energier. Hvis energi ikke var kvantisert, ville et kvikksølvatom kunne ta opp så mye eller så lite energi som elektronet hadde å gi. Da ville elektronene gradvis kunne miste mer og mer energi etter hvert som startenergien økte, uten noen bestemte terskelverdier.

### Kvantetilstander

Vi har nå sett at atomer kun kan eksistere i bestemte energinivåer, men *hvor* i atomet ligger denne energien? Og når vi oppskalerer til molekyler, hvordan beskriver vi da alle de mulige energiene et system kan ha? For å svare på dette introduserer vi begrepet **kvantetilstand**. En kvantetilstand beskriver den fullstendige fysiske tilstanden til et kvantesystem (partikkel, atom eller molekyl). Hver kvantetilstand har en bestemt energi, og siden energien er kvantisert, finnes det bare bestemte kvantetilstander et system kan befinne seg i. De mulige energiene kan vi skrive som:

$$\epsilon = \epsilon_0,\epsilon_1,\epsilon_2,...\text{,}$$

der hvert energinivå tilsvarer én eller flere kvantetilstander. Vi velger å bruke $\epsilon$ i stedet for det vanlige symbolet for energi, $E$, fordi vi nå snakker om enkelte atomer eller molekyler. Dette er kun noe vi velger, vi kunne valgt en emoji i stedet🤠. Med tanke på at vi skal snakke om den totale energien til en samling av veldig mange atomer eller molekyler senere, som vi kaller $E$, er det rett og slett bare mer ryddig å bestemme oss for å bruke $\epsilon$ for veldig små systemer. Stor $E$, stort system. Liten $\epsilon$, lite system.

For et atom er energien i stor grad knyttet til elektronenes tillatte tilstander rundt atomkjernen. Den laveste mulige energien kalles **grunntilstanden**, og betegnes ofte med $\epsilon_0$. Grunntilstanden til et atom er nok kjent fra før, da er alle elektroner rundt atomkjernen i sine laveste mulige energinivåer. Alle tilstander med høyere energi kalles **eksiterte tilstander**. For eksempel hvis et elektron eksiteres, sier vi at *elektronet* er eksitert og at *atomet* er i en eksitert *tilstand*. Dersom et system går fra en kvantetilstand med energi $\epsilon_i$ til en annen kvantetilstand med energi $\epsilon_f$, er energiforskjellen

$$\Delta \epsilon=\epsilon_f-\epsilon_i\text{.}$$

Noen ganger kan to eller flere forskjellige kvantetilstander ha nøyaktig samme energi. Slike tilstander kalles **degenererte tilstander**. De er altså forskjellige fysiske tilstander, selv om energien er den samme. 

![Energier](Simuleringer/energi_stige_atom.png)

For molekyler blir situasjonen mer kompleks. I tillegg til elektronenes energi kan molekylet også bevege seg som en helhet, rotere, og vibrere. Disse uavhengige bevegelsesmåtene kalles **frihetsgrader**, og hver av dem representerer en måte molekylet kan lagre energi på. En frihetsgrad er altså en uavhengig type bevegelse eller dynamikk som bidrar til systemets totale energi. Grunnen til at vi ikke snakker om frihetsgrader for enkeltatomer, er at et atom verken har noen kjemisk binding som kan vibrere eller noen molekylstruktur som kan rotere, det eneste et enkeltatom kan gjøre er å bevege seg som en helhet (translasjon) og ha elektronene sine i ulike tilstander.

En kvantetilstand for et molekyl må derfor beskrive alle disse bidragene samtidig:

- **Den elektroniske tilstanden** beskriver hvilken kvantetilstand elektronene rundt atomkjernene befinner seg i (tenk orbitalokkupasjon), akkurat det samme prinsippet som for et enkeltatom, bare nå fordelt på flere atomkjerner.
- **Vibrasjonstilstanden** beskriver hvordan atomene i molekylet oscillerer rundt sine likevektsposisjoner i bindingene. Et molekyl med flere atomer kan vibrere på flere uavhengige måter samtidig (kalt normalmoder), og hver av disse har sitt eget sett med kvantiserte energinivåer.
- **Rotasjonstilstanden** beskriver hvordan molekylet som en helhet roterer om en akse gjennom tyngdepunktet sitt. Også rotasjonsenergien er kvantisert, molekylet kan bare rotere ved bestemte, tillatte "hastigheter".
- **Translasjonen** beskriver bevegelsen til hele molekylets massesenter gjennom rommet. 

Hvorfor er translasjon med, når den ikke er kvantisert? Molekyler kan vel bevege seg rundt i rommet, uten å gjøre hopp langs tillatte posisjoner? Strengt tatt *er* faktisk også translasjonsenergien kvantisert, det skal vi ikke beskrive så nøye nå. Et molekyl beveger seg typisk fritt i et makroskopisk volum (en beholder, et reagensrør, et rom), og da blir avstanden mellom de tillatte translasjonsenerginivåene enormt liten sammenlignet med termisk energi, så liten at den i praksis er umulig å skille fra en kontinuerlig fordeling av energier. Derfor sier vi gjerne at translasjon "ikke er kvantisert", selv om det mer presist betyr at kvantiseringen ikke er merkbar. Elektroniske, vibrasjons- og rotasjonsnivåer har derimot energiforskjeller som er store nok til at kvantiseringen faktisk merkes og det er derfor disse tre gir opphav til diskrete spektrallinjer, mens translasjon i praksis oppfører seg klassisk.

Vi kan derfor tenke på en kvantetilstand som en fullstendig beskrivelse av hvordan energien er fordelt mellom systemets ulike frihetsgrader. Den totale energien til et molekyl, $\epsilon_{tot}$, kan skrives som summen av bidragene fra hver frihetsgrad:

$$\epsilon_{tot}=\epsilon_{elec}+\epsilon_{vib}+\epsilon_{rot}+\epsilon_{trans}\text{.}$$

Rotasjon, vibrasjon og elektroniske tilstander kan bare anta bestemte energiverdier. Hver av disse frihetsgradene gir derfor opphav til sitt eget sett av kvantetilstander, som beskrives ved hjelp av ett eller flere **kvantetall**. For eksempel beskrives vibrasjonstilstanden ofte av vibrasjonskvantetallet $v$, mens rotasjonstilstanden beskrives av rotasjonskvantetallet $J$. En fullstendig kvantetilstand for et molekyl består av den samlede informasjonen om alle disse frihetsgradene.

### Hva er *egentlig* en mikrotilstand?

Nå kan vi knytte kvantetilstander til et begrep vi allerede har møtt: mikrotilstander. En mikrotilstand er én spesifikk måte et system kan være organisert på. I kvantemekanikk tilsvarer en mikrotilstand én bestemt kvantetilstand. Når vi sier at vi kjenner systemets mikrotilstand, mener vi derfor at vi kjenner alle kvantetallene som beskriver systemets frihetsgrader. Dersom flere forskjellige kvantetilstander har samme energi (degenererte tilstander), finnes det altså flere mikrotilstander som tilsvarer den samme energien. Vi kan tenke oss sammenhengen slik: 

$$\text{Mikrotilstand} \longleftrightarrow \text{Kvantetilstand} \longrightarrow \text{Energinivå}$$

Flere forskjellige kvantetilstander kan altså peke mot det samme energinivået, tilstandene er da degenererte. Vi betegner degenerasjonen, altså antallet tilstander med lik energi, av enkelte partikler, atomer eller molekyler med $g$.

Hvis energi ikke var kvantisert, ville hver frihetsgrad kunne anta kontinuerlig mange energier. Da ville antallet mulige kvantetilstander, og dermed mikrotilstander, vært uendelig selv innenfor et begrenset energiintervall. Kvantemekanikkens lover begrenser altså hvilke tilstander naturen faktisk tillater. Det er derfor mikrotilstander ikke bare er et matematisk hjelpemiddel. De er en direkte konsekvens av at naturen på mikroskopisk nivå består av diskrete kvantetilstander. 

## Overgangen mellom mikro og makro

:::{admonition} Quote
:class: important
The laws of thermodynamics, as empirically determined, express the approximate and probable behavior of systems of a great number of particles, or, more precisely, they express the laws of mechanics for such systems as they appear to beings who have not the fineness of perception to enable them to appreciate quantities of the order of magnitude of those which relate to single particles, and who cannot repeat their experiments often enough to obtain any but the most probable results. 

– J.W. Gibbs
:::

**Statistisk mekanikk** bygger på den kvantemekaniske beskrivelsen og lar oss koble de mikroskopiske kvantetilstandene til de makroskopiske størrelsene vi kan måle, som energi, entropi og varmekapasitet. Statistisk mekanikk er dermed den mikroskopiske tilnærmelsen til termodynamikken. Mens vi i den klassiske termodynamikken studerer makroskopiske variabler ved bruk av empiriske lover, utledes de samme egenskapene i statistisk mekanikk ved å analysere den tilfeldige, samlede oppførselen til mengder av atomer og molekyler ved å bruke sannsynligheter, som ble introdusert i entropi-modulen. 

Så hva hvis vi nå har 100 eller 1000 eller mange millioner molekyler?  Selv om vi kjenner alle mulige kvantetilstander for ett molekyl, finnes det et enormt antall måter energien kan fordeles mellom alle molekylene når vi har flere av dem. Hver slik fordeling er en mikrotilstand for *hele systemet*. Altså, antallet mikrotilstander eksploderer når vi går fra ett enkelt molekyl til et makroskopisk system.

Samtidig vil mange forskjellige mikrotilstander gi opphav til de *samme* makroskopiske egenskapene. To systemer kan for eksempel ha nøyaktig samme temperatur, trykk og total energi, selv om molekylene er fordelt på helt forskjellige kvantetilstander. Fra et makroskopisk perspektiv ser de identiske ut, selv om de er forskjellige på mikroskopisk nivå. Antallet mikrotilstander som tilsvarer samme makrotilstand for et system av flere atomer eller molekyler kaller vi multiplisiteten $\Omega$.

Vi skal nå se på hvilke mikroskopiske tilstander som faktisk er *befolkede*, dvs hvilke av dem eksisterer? Noen har høy energi, noen har lav, de kan vel ikke alle alltid være dominerende? Vi skal også introdusere partisjonsfunksjonen, som er et utrolig nyttig verktøy for å koble sammen mikro- og makro-definisjonene. 


### Boltzmanns fordelingslov

Hva er det som bestemmer hvilke mikrotilstander som faktisk er befolkede? Vi tar utgangspunkt i ett enkelt atom eller molekyl som er del av et mye større system (mange millioner andre atomer eller molekyler) ved en gitt temperatur $T$. Dette molekylet kan befinne seg i hvilken som helst av sine tillatte mikrotilstander (kvantetilstander), men ikke alle er like sannsynlige. **Boltzmanns fordelingslov** forteller oss nøyaktig hvor sannsynlig det er å finne molekylet i en mikrotilstand med energi $\epsilon_n$:

$$p(\epsilon_n) = \frac{1}{q}e^{-\frac{\epsilon_n}{k_BT}}$$

Her er $k_B$ Boltzmanns konstant, $T$ er den absolutte temperaturen til *hele* systemet molekylet er en del av, og $q$ er en normaliseringsfaktor som sørger for at alle sannsynlighetene til sammen blir 1 (vi kommer tilbake til nøyaktig hva $q$ er i neste avsnitt). Boltzmanns fordelingslov er en **sannsynlighetsfordeling** over energier. En sannsynlighetsfordeling er rett og slett en funksjon som forteller hvor sannsynlig hvert mulig utfall er. Du har allerede møtt dette i entropi-modulen. Her er utfallene "å befinne seg i mikrotilstand med energi $\epsilon_n$", og kravet er at summen av $p(\epsilon_n)$ over *alle* mikrotilstander molekylet kan innta, må være 1 fordi molekylet må jo befinne seg i én av dem.

!Demonstrasjon av hvordan det summes til 1, grafisk!

Se på leddet $e^{-\epsilon_n/k_BT}$. Det er dette som styrer hvordan sannsynligheten avtar med energien:

- Høyere $\epsilon_n$: $\rightarrow$ eksponenten blir mer negativ $\rightarrow$ $p(\epsilon_n)$ blir mindre. Høyenergitilstander er alltid mindre sannsynlige enn lavenergitilstander, ved en gitt temperatur.
- Høyere $T$: $\rightarrow$ forholdet $\epsilon_n/k_BT$ blir mindre for en gitt $\epsilon_n$ $\rightarrow$ eksponentialen avtar saktere $\rightarrow$ sannsynligheten "sprer seg" mer utover høyere energinivåer.
- Lavere $T$: $\rightarrow$ forholdet $\epsilon_n/k_BT$ blir stort selv for moderate energier $\rightarrow$ sannsynligheten faller brått, og nesten alt "samler seg" i grunntilstanden ($\epsilon_0=0$).

Det er altså *forholdet* mellom energien og $k_BT$ som avgjør hvor sannsynlig en tilstand er, ikke energien alene.

I simuleringen over ser du seks energinivåer for et molekyl, jevnt fordelt med avstand $\varepsilon$ mellom hvert nivå ($\epsilon_n = n\varepsilon$). Glideren styrer $k_BT/\varepsilon$, altså hvor stor den termiske energien er sammenlignet med avstanden mellom energinivåene. Bruk den til å se hva som skjer:
- Ved lav $k_BT/\varepsilon$ er nesten all sannsynlighet i grunntilstanden.
- Etter hvert som $k_BT/\varepsilon$ øker, blir stadig flere av de høyere nivåene befolket.
- Ved svært høy $k_BT/\varepsilon$ nærmer fordelingen seg (nesten) jevn — alle nivåene blir omtrent like sannsynlige.

Dette er nøyaktig den samme mekanismen som avgjør hvorfor et molekyl ved romtemperatur som regel *ikke* er i en høyt eksitert vibrasjons- eller elektrontilstand: energitrinnene $\varepsilon$ er ofte store sammenlignet med $k_BT$ ved romtemperatur.

Fra fordelingen kan vi trekke ut noen nyttige "tommelfingerregler".
**Lavenergitilstander er alltid mer befolket enn høyenergitilstander**, ved en gitt temperatur. Det er fordi at systemet må ha en total energi (makro) som er fordelt på hvert atom eller molekyl, og hvis flere partikler har lav energi, vil det være flere mulige fordelinger av total energi. Hvis ett atom eller molekyl har veldig høy energi, må alle de gjenstående ha ganske lav energi. Dette er veldig usannsynlig.**Forholdet mellom to sannsynligheter** avhenger bare av energiforskjellen og temperaturen, ikke av de absolutte energiverdiene:

$$\frac{p(\epsilon_1)}{p(\epsilon_0)} = e^{-\frac{\epsilon_1-\epsilon_0}{k_BT}}$$

Dette kalles gjerne **Boltzmann-faktoren**, og den dukker opp igjen i fysikalsk kjemi, for eksempel når vi senere skal se på likevektskonstanter(??). Temperatur avgjør hvor "spredt" fordelingen er, ikke hvilke tilstander som finnes. Hvilke mikrotilstander som er *tilgjengelige* bestemmes av kvantemekanikken, men hvor *befolket* de er, bestemmes av temperaturen.

Boltzmanns fordelingslov er selve broen mellom mikro og makro. Så lenge vi vet hvordan sannsynligheten fordeler seg over mikrotilstandene, kan vi i prinsippet regne oss frem til makroskopiske størrelser som gjennomsnittlig energi, varmekapasitet og entropi ved å veie hver mikrotilstands bidrag med dens sannsynlighet $p(\epsilon_n)$. Det er nettopp dette partisjonsfunksjonen gjør mulig å regne ut systematisk, og det er tema for neste avsnitt. 

Loven handler ikke nødvendigvis om én og én partikkel. Litt mer presist gjelder Boltzmann-fordelingen for et hvilket som helst system som er i termisk kontakt med et reservoar (omgivelse) ved temperatur $T$, og som kan utveksle energi med dette reservoaret. "Systemet" kan i prinsippet være et enkelt atom eller molekyl *eller* feks en hel gass, betraktet som ett stort system, hvis den selv er i kontakt med et enda større reservoar. Poenget er at loven beskriver sannsynligheten for å finne det systemet du har valgt å se på i en gitt mikrotilstand, gitt temperaturen til omgivelsene rundt. I kjemien blir det ofte én og én partikkel i praksis, fordi vi i kjemi som regel har med et stort antall ikke-vekselvirkende molekyler å gjøre (ideell gass-tilnærmingen). Da kan man vise at hele systemets energifordeling deles opp til at hvert enkelt molekyl uavhengig følger nøyaktig den samme Boltzmann-fordelingen over sine egne energinivåer. Det er derfor vi kan snakke om "hvor mange molekyler som er i grunntilstanden" osv. Det fungerer fordi molekylene ikke påvirker hverandre. Hvis partiklene vekselvirker sterkt med hverandre, holder ikke denne tilnærmingen lenger, da må du behandle hele systemet av $N$ partikler samlet som "systemet" i den kanoniske fordelingen, og det blir fort mye mer komplisert matematikk.

## Partisjonsfunksjonen

Partisjonsfunksjonen for et atom eller molekyl $q$ eller for et sammensatt system $Q$ er et matematisk verktøy som er veldig nyttig når vi vil bruke mikroskopiske egenskaper til å forutsi makroskopiske egenskaper. 

$$ q = \sum_n e^{-\frac{\epsilon_n}{k_BT}} \text{,}$$

$$ Q = \sum_i e^{-\frac{E_i}{k_BT}} \text{.}$$

Disse to er samme funksjon, så det er strengt tatt ikke nødvendig å inkludere begge. Her har vi inkludert begge, bare for å vise forskjellen i notasjon og for å presisere at det er viktig å vite hva slags system man studerer. Fra nå av vil vi bruke $Q$ for å forklare egenskaper ved partisjonsfunksjonen, men være klar over at disse gjelder for $q$ også. Hvis vi antar at $t+1$ er antallet mikrotilstander og skriver ut $Q$, ser det slik ut:

$$
Q=\sum_{i=0}^{t} e^{-\frac{E_i}{k_BT}}
= e^{-\frac{E_0}{k_BT}}
+ e^{-\frac{E_1}{k_BT}}
+ e^{-\frac{E_2}{k_BT}}
+ \cdots
+ e^{-\frac{E_t}{k_BT}}.
$$

Partisjonsfunksjonen summerer altså bidrag fra *alle* mikrotilstander, enten vi snakker små eller store systemer. Den adderer sannsynligheten for alle mikrotilstandene, og er dermed et lite beist av informasjon om systemet. 

La oss ta en liten oppsummering før vi går videre: 
- Kvantemekanikken gir oss energinivåene.
- Statistisk mekanikk gir sannsynlighetene for disse.
- Partisjonsfunksjonen samler all informasjonen.

Vi er nødt til å kjenne til energinivåene for å regne ut $Q$, men når vi måler energier eksperimentelt, måler vi som regel *energidifferanser* og ikke absolutte energier. Derfor skriver vi om partisjonsfunksjonen litt ved å definere grunntilstandsenergiene $E_0=0$ . Så kan vi skrive funksjonene uttrykt ved energidifferansene: 

$$ Q = 1 + e^{-\frac{(E_1-E_0)}{k_BT}} + e^{-\frac{(E_2-E_0)}{k_BT}} + \cdots + e^{-\frac{(E_{t}-E_0)}{k_BT}} \text{.}$$

Man kan også uttrykke partisjonsfunksjonen ved *makrotilstandene*. Vi bruker multiplisiteten av tilstander $\Omega$, men endrer litt på skrivemåten. Et system kan eksistere i energinivå $E$ på $\Omega(E)$ måter. Hvis $\Omega(E)>1$, er tilstanden degenerert. Noen steder kalles også multiplisiteten for "tilstandstettheten", kjært barn har mange navn. Si at vi har $l$ ulike energinivåer/makrotilstander, som kan være degenererte eller ei. Når vi utrykker partisjonsfunksjonen på denne måten, ser den slik ut: 

$$ Q=\sum_j^{l} \Omega(E_j) e^{-\frac{E_j}{k_BT}}.$$

Måten funksjonen skrives er opp til deg, det kommer an på hvordan systemet du vil studere beskrives best. For eksempel er $\sum_i$ over hver enkelt mikrotilstand mest naturlig når tilstandene ikke er degenererte, eller når degenerasjonen er upraktisk å gruppere. 

Dersom man kjenner energitilstandene til et system, er det ganske enkelt å regne ut $Q$. Men hvordan skal tallet man får tolkes? En intuitiv måte å tenke på hva partisjonsfunksjonen *er*, er å se på den som "antallet effektivt tilgjengelige tilstander". Husk, sannsynligheten for hver av mikrotilstandene er bakt inn i partisjonsfunksjonen. Derfor kan partisjonsfunksjonen gi oss svaret på hvor mange tilstander systemet i praksis kan eksistere i, gitt en temperatur og avstanden mellom energinivåene.

### Distinguishable vs indistinguishable?? KAN DROPPES???

Nå skal vi gi navn til to ulike måter å modellere atomer/molekyler, avhengig av hva slags system man utforsker. 

**Distinguishable systemer**: disse systemene består av molekyler eller atomer som har en eller annen egenskap som gjør at vi kan si at den ene er den ene og den andre er den andre. For eksempel atomene i en krystall; hvert atom har sin faste fysiske plassering i gitteret. Man kan med sikkerhet si at atom $A$ har plassering $x_A, y_A, z_A$ og atom $B$ har plassering $x_B, y_B, z_B$, uavhengig av om atom $A$ og $B$ er samme grunnstoff eller ikke. Dermed vet vi at "dette molekylet er $A$, og det andre er $B$". I dette tilfellet er det altså den romlige posisjonen som gjør molekylene mulige å skille fra hverandre. I slike systemer er den totale energien $E$ en sum av energiene til hvert molekyl $\epsilon_k$, og partisjonsfunksjonen for hele systemet er produktet av enkelte molekyl-partisjonsfunksjoner. 

**Indistinguishable systemer**: i indistinguishable systemer er alle molekyler/atomer identiske, og det er umulig å peke på dem og si at det molekylet er $A$ og det molekylet er $B$. Vi kan vite at de begge er til stede, men det er helt umulig å si noe om hvilken som er hvilken. Det er ikke sånn at "vi skulle ønske" vi kunne si hvilken som er hvilken, det er bare naturen av systemet at det ikke er mulig, og det er også derfor vi skiller mellom distinguishable og indistinguishable. Dette er viktig å vite om når vi setter opp modeller for systemene våre. 

OBS BOKS: hvis du ser på en løsning med øynene dine og den ser homogen ut, betyr IKKE det at dette er et indistinguishable system. Homogen er ikke lik indistinguishable. For å vite om noe er indistinguishable eller distinguishable, må vi vurdere hvordan systemet skal modelleres og hvilke fysiske egenskaper som skal inkluderes. Også hvordan det måles. distinguishable finnes ikke egntlgi, alt i alt. vi må velge ut ting

## Termodynamiske egenskaper forutsies av partisjonsfunksjonen

Nå skal vi se hvordan de termodynamiske størrelsene, feks indre energi, kan uttrykkes og dermed regnes ut ved bruk av partisjonsfunksjonen. Vi skal også se at selv om størrelsene er uttrykt på en ny måte, handler de om akkurat det samme som de empiriske uttrykkene. Vi tar ikke for oss noen detaljerte utledningenger her, og tar for gitt at matten stemmer, hvis du er interessert vil dette bli undervist i Fysikalsk Kjemi III. Det er *ikke* pensum å kunne utlede noen av disse uttrykkene i Fysikalsk Kjemi I. 

**Entropi** handler ikke om uorden, som vi har vært gjennom i entropi-modulen, men om hvor mange mikroskopiske muligheter som er kompatible med det som blir den makroskopiske observasjonen. En økning i entropi er det samme som at systemet går fra en mindre sannsynlig makroskopisk tilstand til en mer sannsynlig makroskopisk tilstand. Entropien er, som vist tidligere, definert som:  

$$ S=k_B \ln \Omega $$

og blir, uttrykt ved partisjonsfunksjonen (vis en VELDIG forenklet overgang her)

$$ S = k_B \ln Q + \frac{U}{T} .$$

**Indre energi** er forventningsverdien, eller ensemble-gjennomsnittet $\langle E \rangle$, over alle mikrotilstands-energiene $E_i$, vektet av sannsynlighetene deres $P_i$: 

$$U= \langle E \rangle = \sum_i E_i P_i=\frac{1}{Q}\sum_i E_i e^{-\frac{E_i}{k_BT}},$$

der vi har brukt Boltzmanns fordelingslov $P_i=\frac{1}{Q}e^{-\frac{E_i}{k_BT}}$. Videre kan det vises at:

$$ U = -\frac{\partial \ln Q}{\partial (k_BT)}= -k_B T^2\frac{\partial \ln Q}{\partial T}.$$

:::{admonition} Hjelpenotes
:class: important
Notasjonen $\langle E \rangle$, betyr bare "gjennomsnittlig energi".

Å "vekte" noe med sannsynligheten dens, betyr at dersom sannsynligheten for én energi $E_1$ er større enn en annen energi $E_2$, så må denne forskjellen i sannsynlighet inkluderes i uttrykket for at det skal bli riktig. Vi har jo sett at ulike mikrotilstander har ulik sannsynlighet, og dette kan vi ikke bare overse når vi regner ut gjennomsnittet over dem. 
:::

## Hvorfor varierer varmekapasitet?

### Ekvipartisjonsteoremet

**Ekvipartisjonsteoremet** sier at den **totale energien til et system er delt likt mellom dets frihetsgrader**. Hver frihetsgrad (måte å lagre energi), får tildelt et gjennomsnitt på $\frac{1}{2}k_B T$ av termisk energi. Se for deg en lab med mye aktivitet. Hvis det er 5 avtrekksskap og 5 kjemikere, får hvert avtrekksskap tildelt omtrent én kjemiker. Kjemikerne er temperatur og avtrekksskapene er frihetsgradene, og ekvipartisjonsteoremet sier at kjemikerne fordeles gjevnt. 

EKSEMPEL: DIATOMISK GASS
- bevege seg tre romlige retninger
- vibrere over båndet
- rotere to retninger
- totalt 7 frihetsgrader, som gir en energi $\frac{7}{2}k_B T$. 

Kvantemekanikken er den generelle teorien når vi beskriver systemer på mikroskala. Ekvipartisjonsteoremet er en klassisk tilnærming som blir gyldig når energinivåene er så tettliggende at de oppfører seg nesten kontinuerlig i forhold til den termiske energien $kT$. Ved veldig lave temperaturer, kan ikke lenger molekylene dele energien inn i kontinuerlige deler, så høyere-energi frihetsgrader fryses ut. Ekvipartisjonsteoremet beskrives derfor som den klassiske grensen av den kvantemekaniske beskrivelsen. Det er altså ikke et grunnleggende prinsipp, men et resultat som gjelder når kvantiseringen ikke lenger er viktig.

La oss repetere frihetsgradene og se hvilke av dem som fryses ut først og hvorfor. 

### Translasjon

nesten kontinuerlige nivåer, bidrar som ekvipartisjonsteoremet sier ved alle vanlige temperaturer

### Rotasjon

små nivåavstander, aktiveres lett, omtrent klassisk bidrag

### Vibrasjon

store nivåavstander, frosset ut ved romtemp, klassisk forventning bryter sammen.

## Oppsummering

Kvantemekanikken forteller oss at molekyler bare kan eksistere i bestemte energitilstander. Ved en gitt temperatur er ikke alle disse tilstandene like sannsynlige; fordelingen bestemmes av Boltzmann-faktoren. Statistisk mekanikk kobler derfor kvantetilstandene til målbare termodynamiske størrelser gjennom partisjonsfunksjonen. Entropi kan forstås som et mål på hvor mange mikroskopiske tilstander som er tilgjengelige for systemet, og er dermed en naturlig bro mellom mikro- og makroverden.

:::{admonition} Kort oppsummert
:class: important
Kvantemekanikk bestemmer hvilke mikrotilstander som finnes, statistisk mekanikk bestemmer hvor sannsynlige de er og termodynamikk beskriver resultatet vi måler.
:::
