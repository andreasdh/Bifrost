I fysikalsk kjemi ønsker vi å forstå hvordan makroskopiske størrelser som energi, entropi og varmekapasitet oppstår fra molekylers mikroskopiske oppførsel. I denne modulen skal vi bygge bro mellom den mikroskopiske og den makroskopiske verden. Vi starter med en kort introduksjon til kvantemekanikk, som beskriver hvordan atomer og molekyler kun kan eksistere i diskrete energitilstander. Disse kvantetilstandene danner grunnlaget for mikrotilstandene som brukes i statistisk mekanikk.

Et makroskopisk system består av et enormt antall molekyler, og det er umulig å følge hver enkelt av dem. I stedet beskriver statistisk mekanikk systemet ved hjelp av sannsynligheter og gjennomsnitt over alle mulige mikrotilstander. Dette gir en direkte kobling mellom molekylenes kvantemekaniske energinivåer og de termodynamiske størrelsene vi kan måle, som energi, entropi og varmekapasitet.

Gjennom modulen skal vi se hvordan kvantemekanikk gir opphav til mikrotilstander, hvordan disse beskrives statistisk ved hjelp av Boltzmann-fordelingen og partisjonsfunksjonen, og hvordan partisjonsfunksjonen kan brukes til å beregne termodynamiske tilstandsfunksjoner. Til slutt ser vi hvordan molekylers frihetsgrader bestemmer varmekapasiteten, og hvordan ekvipartisjonsteoremet kan forstås som den klassiske grensen av den kvantemekaniske beskrivelsen.

Kort oppsummert kan modulen beskrives slik:

> **Kvantemekanikk bestemmer hvilke mikrotilstander som finnes, statistisk mekanikk bestemmer hvor sannsynlige de er og termodynamikk beskriver resultatet vi måler.**

## Statistisk Entropi

Hvis mange forskjellige mikrotilstander gir opphav til samme makroskopiske observasjon (samme U, V, N), hvordan kan vi beskrive dette mangfoldet? Boltzmann kvantifiserte dette ved å innføre: Likning

Dermed får entropi en svært konkret betydning: Entropi er et mål på hvor mange kvantemekanisk tillatte mikrotilstander som er forenlige med en gitt makroskopisk tilstand.

et system kan realiseres på mange mikroskopiske måter. Jo flere mikrotilstander, desto høyere multiplisitet. Boltzmann koblet dette til entropi. 

Entropi handler ikke om uorden, men om hvor mange mikroskopiske muligheter som er kompatible med det som blir den makroskopiske observasjonen.

# Hvorfor varierer varmekapasitet?

## Ekvipartisjonsteoremet

Kvantemekanikken er den generelle teorien. Ekvipartisjonsteoremet er bare en tilnærming som blir gyldig når energinivåene er så tettliggende at de oppfører seg nesten kontinuerlig i forhold til den termiske energien $$kT$$. Kan beskrives som den klassiske grensen av den kvantemekaniske beskrivelsen. Det er altså ikke et grunnleggende prinsipp, men et resultat som gjelder når kvantiseringen ikke lenger er viktig.

Hva om energinivåene i systemet ligger veldig tett?
Kvante -> diskret energi -> energigap -> boltzmann bestemmer befolkning -> middelenergi -> varmekapasitet

Hvis $$\Delta E << kT$$ : klassisk grense -> ekvipartisjon. systemet oppfører seg klassisk. 

Hvis $$\Delta E >> kT$$ : nivåer er frosset -> ekvipartisjon bryter sammen

## Translasjon

nesten kontinuerlige nivåer, bidrar som ekvipartisjonsteoremet sier ved alle vanlige temperaturer

## Rotasjon

små nivåavstander, aktiveres lett, omtrent klassisk bidrag

## Vibrasjon

store nivåavstander, frosset ut ved romtemp, klassisk forventning bryter sammen. 

# Oppsummering

Kvantemekanikken forteller oss at molekyler bare kan eksistere i bestemte energitilstander. Ved en gitt temperatur er ikke alle disse tilstandene like sannsynlige; fordelingen bestemmes av Boltzmann-faktoren. Statistisk mekanikk kobler derfor kvantetilstandene til målbare termodynamiske størrelser gjennom partisjonsfunksjonen. Entropi kan forstås som et mål på hvor mange mikroskopiske tilstander som er tilgjengelige for systemet, og blir dermed den naturlige broen mellom mikro- og makroverden.

> **Kvantemekanikk bestemmer hvilke mikrotilstander som finnes, statistisk mekanikk bestemmer hvor sannsynlige de er og termodynamikk beskriver resultatet vi måler.**
