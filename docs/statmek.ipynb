## Kvantemekanikken ligger til grunn for våre observasjoner

Kjemikere arbeider som oftest med stoffmengder som inneholder $10^{23}$ molekyler eller mer. I laboratoriet observerer vi derfor alltid den samlede oppførselen til et enormt antall partikler. På denne skalaen beskrives systemene godt av termodynamikk og klassisk fysikk, men de makroskopiske egenskapene har sitt opphav i prosesser som foregår på atom- og molekylnivå. For å beskrive disse prosessene trenger vi kvantemekanikken. 

Ordet "kvante" eller "quantum" kan høres skremmende ut, kanskje takket være en interessant bruk av det i diverse sci-fi filmer. Det er egentlig ikke så hokus-pokus som man ofte har inntrykk av og i dette delkapitlet går vi gjennom det helt grunnleggende og dropper utledninger og tung matte. ### Kvantisert energi

**Kvantemekanikken** er en matematisk beskrivelse av oppførselen til enkelte molekyler og atomer eller partikler. Tidligere var det antatt at klassisk fysikk var nok til å beskrive disse systemene, men eksperimentelle funn (Ultrafiolett katasttrofe,Dobbeltspalteeksperimentet, fotoelektrisk effekt, Comptonspredning, Stern-Gerlach eksperimentet mm.) stemte ikke overens med fysikken slik vi kjente den, og en helt ny gren ble nødvendig for å forklare funnene. Felles for etterarbeidet med mange av disse eksperimentene var at **kvantisering** av energi ble forklaringen for å få observasjonene til å gi mening. Vi bruker her Franck-Hertz eksperimentet til å illustrere hva kvantisering er og det faktum at det er en iboende egenskap ved vårt univers. Eksperimentet selv er ikke del av pensum, men den konseptuelle forståelsen er det. 

Klassisk fysikk handler om kontinuerlige spektre av størrelser; kjører du oppover en bakke får du og bilen din (systemet) litt og litt mer potensiell energi for hver eneste mikrometer du beveger deg oppover. Vår virkelighetsoppfattelse utfordres når dette ikke lenger stemmer i kvanteverdenen. **Franck-Hertz eksperimentet** ble utført i 1914 av James Franck og Gustav Hertz, og ga oss det første beviset på atomenes **diskrete**, kvantiserte energinivåer. 

:::{admonition} Diskret energi
:class: important
En diskret verdi er en spesifikk, adskilt verdi. Tenk på trinnene i en stige. Det finnes ikke noe trinn nummer $1.5$ eller $\frac{1}{4}$, kun $0, 1, 2$ osv, og det er umulig å flyte i luften mellom trinnene. Når vi snakker om diskret energi, snakker vi om helt bestemte verdier av energi $E_0, E_1, E_2,...$ som et system kan ha, og det er ikke mulig for systemet å ha en energi med en verdi som ligger mellom disse. Dette har du trolig vært borti før når vi snakker om eksitasjon og deeksitasjon av elektroner i orbitaler. 
:::

I Franck-Hertz eksperimentet ble elektroner sendt gjennom en beholder fylt med kvikksølv ($\text{Hg}$) i gassform, dvs. kvikksølvatomer. Franck og Hertz kontrollerte hvor mye energi elektronene hadde før de kolliderte med kvikksølvatomene, for så å måle hvor mye energi elektronene hadde igjen etter de hadde kollidert. 

BILDE AV FORENKLET OPPSETT HER

Basert på klassisk fysikk forventet de at elektronene skulle avgi energi til kvikksølvatomene på en gradvis måte. Dersom elektronene fikk litt høyere startenergi, skulle de også miste litt mer energi i kollisjonene. De observerte derimot noe helt annet: 

1. Når elektronet hadde mindre enn én bestemt energiverdi $E$, kom det helt uendret ut av kollisjonen.
2. Når elektronet hadde energi $E$ eller høyere, mistet det alltid *samme mengde* energi under kollisjonen, og dersom startenergien ble økt litt mer, endret det ikke måleverdien. 
3. Først ved enda høyere startenergier kunne elektronet miste *to ganger* denne "minimumsenergien", altså $2E$. 
4. Ved enda høyere startenergier enn dette igjen kunne elektronet miste $3E$, $4E$ osv.

FORVENTET VS OBSERVERT GRAF HER

Hvordan skulle dette tolkes? Den eneste forklaringen som passet observasjonene var at kvikksølvatomene *kun* kunne ta opp helt bestemte energimengder, eller **energipakker**, med verdi $nE$, der $n=1,2,3,...$ er et naturlig tall. Atomene kunne ikke absorbere vilkårlige energier, de absorberte helt bestemte, diskrete energimengder. Atomenes energitilstander er dermed **kvantiserte**, dvs. de kan kun eksistere i tilstander av helt spesifikke energier. Hvis energi ikke var kvantisert, ville et kvikksølvatom kunne ta opp så mye eller så lite energi som elektronet hadde å gi. Da ville elektronene gradvis kunne miste mer og mer energi etter hvert som startenergien økte, uten noen bestemte terskelverdier.

### Kvantetilstander

Vi har nå sett at atomer kun kan eksistere i bestemte energinivåer, men *hvor* i atomet ligger denne energien? Og når vi oppskalerer til molekyler, hvordan beskriver vi da alle de mulige energiene et system kan ha? For å svare på dette introduserer vi begrepet **kvantetilstand**. En kvantetilstand beskriver den fullstendige fysiske tilstanden til et kvantesystem (partikkel, atom eller molekyl). Hver kvantetilstand har en bestemt energi, og siden energien er kvantisert, finnes det bare bestemte kvantetilstander et system kan befinne seg i. De mulige energiene kan vi skrive som:

$E = E_0,E_1,E_2,...$,

der hvert energinivå tilsvarer én eller flere kvantetilstander. 

For et atom er energien i stor grad knyttet til elektronenes tillatte tilstander rundt atomkjernen. Den laveste mulige energien kalles **grunntilstanden**, og betegnes ofte med $E_0$. Alle tilstander med høyere energi kalles **eksiterte tilstander**. For eksempel hvis et elektron eksiteres, sier vi at *atomet* er i en eksitert *tilstand*. Dersom et system går fra en kvantetilstand med energi $E_i$ til en annen kvantetilstand med energi $E_f$, er energiforskjellen

$\Delta E=E_f-E_i$.

Denne energien kan for eksempel absorbered eller emitteres som et foton. Noen ganger kan to eller flere forskjellige kvantetilstander ha nøyaktig samme energi. Slike tilstander kalles **degenererte tilstander**. De er altså forskjellige fysiske tilstander, selv om energien er den samme.

BILDE HER?

For molekyler blir situasjonen mer kompleks. I tillegg til elektronenes energi kan molekylet også bevege seg som en helhet, rotere og vibrere. Disse uavhengige bevegelsesmåtene kalles **frihetsgrader**, og hver av dem representerer en måte systemet kan lagre energi på. En frihetsgrad er med andre ord en uavhengig type bevegelse eller dynamikk som kan bidra til systemets totale energi. En kvantetilstand for et molekyl beskrives derfor av alle disse bidragene samtidig: 
- **Den elektroniske tilstanden**: elektronenes kvantetilstand rundt atomkjernene (tenk orbitalokkupasjon).
- **Vibrasjonstilstanden**: atomene oscillerer rundt sine likevektsposisjoner.
- **Rotasjonstilstanden**: molekylet roterer om én eller flere akser.
- **Translasjon**: hele molekylet beveger seg gjennom rommet. Dette er *ikke* kvantisert.

Vi kan derfor tenke på en kvantetilstand som en fullstendig beskrivelse av hvordan energien er fordelt mellom systemets ulike frihetsgrader, og den totale energien til et molekyl $E_{tot}$ kan skrives som summen av bidragene fra de ulike frihetsgradene: 

$E_{tot}=E_{elec}+E_{vib}+E_{rot}+E_{trans}$.

Hver av disse energiene er i seg selv kvantisert. Det betyr at både translasjon, rotasjon, vibrasjon og elektroniske tilstander bare kan anta bestemte energiverdier. Hver frihetsgrad gir derfor opphav til sitt eget sett av kvantetilstander, som beskrives ved hjelp av ett eller flere **kvantetall**. For eksempel beskrives vibrasjonstilstanden ofte av vibrasjonskvantetallet $v$, mens rotasjonstilstanden beskrives av rotasjonskvantetallet $J$. En fullstendig kvantetilstand for et molekyl består av den samlede informasjonen om alle disse frihetsgradene.


### Hva er *egentlig* en mikrotilstand?

Nå kan vi knytte kvantetilstander til et begrep vi allerede har møtt: mikrotilstander. En mikrotilstand er én spesifikk måte et system kan være organisert på. I kvantemekanikk tilsvarer en mikrotilstand én bestemt kvantetilstand. Når vi sier at vi kjenner systemets mikrotilstand, mener vi derfor at vi kjenner alle kvantetallene som beskriver systemets frihetsgrader. Dersom flere forskjellige kvantetilstander har samme energi (degenererte tilstander), finnes det altså flere mikrotilstander som tilsvarer den samme energien. Vi kan tenke oss sammenhengen slik: 

$\text{Mikrotilstand} \longleftrightarrow \text{Kvantetilstand} \longrightarrow \text{Energinivå}$

Flere forskjellige kvantetilstander kan altså peke mot det samme energinivået dersom tilstanden er degenerert.

Hvis energi ikke var kvantisert, ville hver frihetsgrad kunne anta kontinuerlig mange energier. Da ville antallet mulige kvantetilstander, og dermed mikrotilstander, vært uendelig selv innenfor et begrenset energiintervall. Kvantemekanikkens lover begrenser altså hvilke tilstander naturen faktisk tillater. Det er derfor mikrotilstander ikke bare er et matematisk hjelpemiddel. De er en direkte konsekvens av at naturen på mikroskopisk nivå består av diskrete kvantetilstander. 

## Overgangen mellom mikro og makro

**Statistisk mekanikk** bygger på den kvantemekaniske beskrivelsen og lar oss koble de mikroskopiske kvantetilstandene til de makroskopiske størrelsene vi kan måle, som energi, entropi og varmekapasitet. Statistisk mekanikk er dermed det mikroskopiske grunnlaget for termodynamikken. Mens vi i den klassiske termodynamikken studerer makroskopiske variabler ved bruk av empiriske lover, utledes de samme egenskapene i statistisk mekanikk ved å analysere den tilfeldige, samlede oppførselen til mengder av atomer og molekyler ved å bruke sannsynlighetsteori. 

Så hva hvis vi nå har 100 eller 1000 eller mange millioner molekyler?  Selv om vi kjenner alle mulige kvantetilstander for ett molekyl, finnes det et enormt antall måter energien kan fordeles mellom alle molekylene når vi har flere av dem. Hver slik fordeling er en mikrotilstand *for hele systemet*. Altså, antallet mikrotilstander eksploderer når vi går fra ett enkelt molekyl til et makroskopisk system.

Samtidig vil mange forskjellige mikrotilstander gi opphav til de *samme* makroskopiske egenskapene. To systemer kan for eksempel ha nøyaktig samme temperatur, trykk og totale energi, selv om molekylene er fordelt på helt forskjellige kvantetilstander. Fra et makroskopisk perspektiv ser de identiske ut, selv om de er forskjellige på mikroskopisk nivå. Antallet mikrotilstander som gir opphav til samme makrotilstand kaller vi **multiplisiteten** $\Omega$ til makrotilstanden. Hva sier multiplisiteten til to ulike makrotilstander oss? Hvis multiplisiteten til tilstand A er høyere enn multiplisiteten til tilstand B, har det noe å si for hvilken tilstand vi faktisk vil finne systemet i?


### Statistisk Entropi

Boltzmann entropien til et system er, som definert i LINK TIL ENTROPI-SIDEN:

$S=k_B \text{ln} \Omega$, 

der $k_B=1.381\dot 10^{-23} JK^{-1}$ er Boltzmannkonstanten og $\Omega$ er antallet mikrotilstander som tilsvarer den makroskopiske tilstanden til systemet. Entropi får en konkret betydning, den er et mål på hvor mange kvantemekanisk tillatte mikrotilstander som er forenlige med en gitt makroskopisk tilstand. Entropi handler ikke om uorden, men om hvor mange mikroskopiske muligheter som er kompatible med det som blir den makroskopiske observasjonen.



En økning i entropi er det samme som at systemet går fra en mindre sannsynlig makroskopisk tilstand til en mer sannsynlig makroskopisk tilstand. Vise dette??

### Boltzmanns sannsynlighetsfordeling

Konseptuell bro mellom kvantemek og statmek. kvantemek etablerer hvilke mikrotilstander som finnes, mens entropien gir dem fysisk betydning ved å knytte antallet av dem til en målbar termodynamisk størrelse.
Boltzmann sier hvilke mikrotilst som dominerer. 

Men alle disse mikrotilstandene kan vel ikke være like viktige?
Noen har høy energi, noen har lav energi. Ved en gitt temperatur er noen mer sannsynlige enn andre.

Den er det første verktøyet statistisk mekanikk bruker etter at vi har innsett at mange mikrotilstander eksisterer.

Hvordan favoriseres mikrotilstander, hvordan blir noen makrotilstander mer eller mindre sannsynlige? Hvis et molekyl har mange mulige energinivåer, hvordan vet vi hvilke nivåer som er faktisk er befolket ved temperatur T? temperaturen bestemmer hvor sannsynlige mikrotilstander er.

Middelenergien bestemmes ikke bare av energinivåene, men av temperaturen.

## Hvorfor varierer varmekapasitet?

### Ekvipartisjonsteoremet

Kvantemekanikken er den generelle teorien. Ekvipartisjonsteoremet er bare en tilnærming som blir gyldig når energinivåene er så tettliggende at de oppfører seg nesten kontinuerlig i forhold til den termiske energien $kT$. Kan beskrives som den klassiske grensen av den kvantemekaniske beskrivelsen. Det er altså ikke et grunnleggende prinsipp, men et resultat som gjelder når kvantiseringen ikke lenger er viktig.

Hva om energinivåene i systemet ligger veldig tett?
Kvante -> diskret energi -> energigap -> boltzmann bestemmer befolkning -> middelenergi -> varmekapasitet

Hvis $\Delta E << kT$ : klassisk grense -> ekvipartisjon. systemet oppfører seg klassisk. 

Hvis $\Delta E >> kT$ : nivåer er frosset -> ekvipartisjon bryter sammen

### Translasjon

nesten kontinuerlige nivåer, bidrar som ekvipartisjonsteoremet sier ved alle vanlige temperaturer

### Rotasjon

små nivåavstander, aktiveres lett, omtrent klassisk bidrag

### Vibrasjon

store nivåavstander, frosset ut ved romtemp, klassisk forventning bryter sammen. 

## Oppsummering

Kvantemekanikken forteller oss at molekyler bare kan eksistere i bestemte energitilstander. Ved en gitt temperatur er ikke alle disse tilstandene like sannsynlige; fordelingen bestemmes av Boltzmann-faktoren. Statistisk mekanikk kobler derfor kvantetilstandene til målbare termodynamiske størrelser gjennom partisjonsfunksjonen. Entropi kan forstås som et mål på hvor mange mikroskopiske tilstander som er tilgjengelige for systemet, og er dermed en naturlig bro mellom mikro- og makroverden.
