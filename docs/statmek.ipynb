### Multiplisitet

Så hva hvis vi nå har 100 eller 1000 eller mange millioner molekyler?  Selv om vi kjenner alle mulige kvantetilstander for ett molekyl, finnes det et enormt antall måter energien kan fordeles mellom alle molekylene når vi har flere av dem. Hver slik fordeling er en mikrotilstand for *hele systemet*. Altså, antallet mikrotilstander eksploderer når vi går fra ett enkelt molekyl til et makroskopisk system.

Samtidig vil mange forskjellige mikrotilstander gi opphav til de *samme* makroskopiske egenskapene. To systemer kan for eksempel ha nøyaktig samme temperatur, trykk og total energi, selv om molekylene er fordelt på helt forskjellige kvantetilstander. Fra et makroskopisk perspektiv ser de identiske ut, selv om de er forskjellige på mikroskopisk nivå. Antallet mikrotilstander som tilsvarer samme makrotilstand for et system av flere atomer eller molekyler kaller vi multiplisiteten $\Omega$.

Fra nå av vil vi bruke et felles ord for atomer og molekyler, og kaller bare begge to for **partikler**. Dette er litt synd, for en partikkel er jo egentlig et elektron, et proton eller liknende, men flere lærebøker bruker denne notasjonen så det er greit å bli kjent med den. Vi betegner multiplisiteten av energien til én partikkel med $g$.

La oss se på det enklest mulige tilfellet: $N$ partikler som hver bare kan være i én av to tilstander: en grunntilstand med energi $\varepsilon_0=0$ eller en eksitert tilstand med energi $\varepsilon$. Vi kaller en modell som dette for et to-nivå system. Selv med et så enkelt system finnes det mange forskjellige mikrotilstander som gir nøyaktig samme totale energi, altså samme makrotilstand. Hvis $n$ av $N$ partikler er eksiterte, er den totale energien $E = n\varepsilon$, uansett *hvilke* $n$ partikler det er snakk om. Antallet måter å velge ut disse $n$ partiklene på er multiplisiteten

$$\Omega(n) = \binom{N}{n} = \frac{N!}{n!(N-n)!},$$

som gir en **binomialfordeling**. I simuleringen under kan du klikke på partiklene for å eksitere eller de-eksitere dem for hånd, eller trykke "ny mikrotilstand" for å se at helt forskjellige konfigurasjoner kan gi samme energi og bidra til en $\Omega$. Legg merke til hvordan $\Omega(n)$ er mye større når omtrent halvparten av partiklene er eksiterte enn når nesten alle eller nesten ingen er det, det er nettopp dette som gjør at visse fordelinger av energi dominerer sterkt over andre når $N$ blir stor. Bruk litt tid på å undersøke og forstå alt som skjer i simuleringen.

### Gjennomsnittet over mikrotilstander

Et makroskopisk system er ikke fastlåst i én mikrotilstand. Partiklene kolliderer og utveksler energi hele tiden, slik at systemet i praksis hopper mellom enormt mange forskjellige mikrotilstander i løpet av svært kort tid, gjerne milliarder av ganger i sekundet. Når vi måler en makroskopisk størrelse som temperatur eller trykk, måler vi derfor aldri én enkelt mikrotilstand. Vi måler et **gjennomsnitt** tatt over alle mikrotilstandene systemet rekker å besøke i løpet av måletiden.

````python
from IPython.display import HTML

with open("Simuleringer/tidsgjennomsnitt_mikrotilstander.html", "r", encoding="utf-8") as f:
    html = f.read()

HTML(html)
````

I simuleringen over ser du et lite to-nivå (grunntilstand og ett eksitert nivå) system som stadig hopper mellom ulike mikrotilstander. Blå ruter viser partikler i grunntilstanden, oransje ruter viser partikler i eksitert tilstand. I grafen viser de oransje prikkene den momentane andelen eksiterte partikler, $n/N$, ved hvert tidssteg, mens den turkise linjen er det løpende gjennomsnittet av denne andelen. Den røde stiplede linjen markerer $p$, den sanne likevektssannsynligheten systemet svinger rundt.

Legg merke til hvordan enkeltmikrotilstandene ser tilfeldige og rotete ut hver for seg, mens gjennomsnittet (vist som en jevn linje eller verdi nederst) raskt stabiliserer seg og blir nesten helt konstant. Det er denne stabile gjennomsnittsverdien vi oppfatter som en makroskopisk egenskap. Jo flere partikler og jo flere mikrotilstander som bidrar til gjennomsnittet, jo mindre svinger denne verdien, og for et system med $\sim 10^{23}$ partikler er svingningene så små at vi rett og slett ikke klarer å måle dem. Det vi kaller en «makrotilstand» er altså ikke noe annet enn dette stabile gjennomsnittet over utrolig mange mikrotilstander.

En annen måte å tenke på dette gjennomsnittet er å forestille seg mange identiske kopier av samme system, alle med samme temperatur og volum, men der hver kopi tilfeldigvis befinner seg i en annen mikrotilstand akkurat nå. En slik samling kalles et **ensemble** (ikke det samme som [ensembler](https://andreasdh.github.io/Bifrost/docs/intro.html#ensembler) som beskriver hvilke termodynamiske størrelser som er konstante). Den makroskopiske egenskapen vi måler for det virkelige systemet, er den samme som gjennomsnittet av egenskapen over alle kopiene i ensemblet.

I simuleringen over kan du se et ensemble av mange like systemer side om side, hver i sin egen mikrotilstand. Hver rute er én kopi av det samme totale systemet, hver med sitt eget tilfeldige antall eksiterte partikler, $n$. Fargen glir fra blå til oransje etter hvor stor andel av partiklene i kopien som er eksiterte. Søylene i histogrammet viser hvor mange av de $M$ kopiene som har akkurat $n$ eksiterte partikler, og den turkise loddrette linjen markerer gjennomsnittet av $n$ på tvers av alle kopiene i ensemblet.

Selv om hvert enkelt system har en energi som avviker fra gjennomsnittet, samler fordelingen av energier seg tett rundt én bestemt verdi når antallet systemer (eller partikler i hvert system) blir stort. Denne gjennomsnittsverdien er det vi observerer som systemets makroskopiske energi. Poenget med denne tankegangen er at vi slipper å følge ett system over tid, vi kan i stedet regne ut gjennomsnittet direkte over alle mulige mikrotilstander, vektet med hvor sannsynlige de er. Det er nettopp dette Boltzmann-fordelingen, som vi skal se på snart, forteller oss.

Selv om vi måler et gjennomsnitt, er ikke alle mikrotilstandene som er inkludert i dette gjennomsnittet like sannsynlige. Vi skal nå se på *hvilke* mikroskopiske tilstander som faktisk er befolkede, dvs hvilke av dem finner vi i praksis? Noen har høy energi, noen har lav, de kan vel ikke alle alltid være dominerende? 

## Boltzmanns fordelingslov

Hva er det som bestemmer hvilke mikrotilstander som er befolkede? Vi tar utgangspunkt i én enkelt partikkel som er del av et mye større system (mange millioner andre partikler) ved en gitt temperatur $T$. Denne partikkelen kan befinne seg i hvilken som helst av sine tillatte mikrotilstander, men ikke alle er like sannsynlige. **Boltzmanns fordelingslov** forteller oss nøyaktig hvor sannsynlig det er å finne partikkelen i en mikrotilstand med energi $\varepsilon_n$:

$$p(\varepsilon_n) = \frac{1}{q}e^{-\frac{\varepsilon_n}{k_BT}}$$

Her er $k_B=1.38\times 10^{-23} \text{JK}^{-1}$ Boltzmanns konstant, $T$ er den absolutte temperaturen til *hele* systemet partikkelen er en del av, og $q$ er en normaliseringsfaktor som sørger for at alle sannsynlighetene til sammen blir 1 (vi kommer tilbake til nøyaktig hva $q$ er [senere](https://andreasdh.github.io/Bifrost/docs/statmek.html#partisjonsfunksjonen)). 

:::{admonition} Termen $\: k_BT$
:class: important
Produktet av Boltzmanns konstant og temperaturen, $k_BT$, måler den gjennomsnittlige termiske energien til en partikkel ved en gitt temperatur. Du kan tenke på $k_BT$ som et slags energibudsjett: jo høyere temperatur, desto større budsjett har hver partikkel å "bruke" på tilfeldig, termisk bevegelse. Det er dette budsjettet som konkurrerer mot energigapet $\Delta\varepsilon$ i uttrykkene vi har sett så langt. 
:::

Boltzmanns fordelingslov er en **sannsynlighetsfordeling** over energier. En sannsynlighetsfordeling er rett og slett en funksjon som forteller hvor sannsynlig hvert mulig utfall er. Her er utfallene "å befinne seg i mikrotilstandene med energi $\varepsilon_n$", og kravet er at summen av $p(\varepsilon_n)$ over *alle* mikrotilstander partikkelen kan innta, må være 1 fordi partikkelen må jo befinne seg i én av dem. Hvis du er usikker på hvordan dette gir mening, kan du lese om det i [Nyttige matematiske verktøy](https://andreasdh.github.io/Bifrost/docs/mattehjelp.html#sannsynlighet).

Se på leddet $e^{-\varepsilon_n/k_BT}$. Det er dette som styrer hvordan sannsynligheten avtar med energien:

- Høyere $\varepsilon_n$ $\rightarrow$ eksponenten blir mer negativ $\rightarrow$ $p(\varepsilon_n)$ blir mindre. **Høyenergitilstander er alltid mindre sannsynlige enn lavenergitilstander, ved en gitt temperatur**.
- Høyere $T$ $\rightarrow$ forholdet $\varepsilon_n/k_BT$ blir mindre for en gitt $\varepsilon_n$ $\rightarrow$ eksponentialen avtar saktere $\rightarrow$ **sannsynligheten "sprer seg" mer utover høyere energinivåer**.
- Lavere $T$ $\rightarrow$ forholdet $\varepsilon_n/k_BT$ blir stort selv for moderate energier $\rightarrow$ **sannsynligheten faller brått, og nesten alt "samler seg" i grunntilstanden ($\varepsilon_0=0$**).

Det er altså *forholdet* mellom energien til nivået $\varepsilon_n$ og $k_BT$ som avgjør hvor sannsynlig en tilstand er, ikke energien eller temperaturen alene.

````python

from IPython.display import HTML

with open("Simuleringer/boltzmann_sim.html", "r", encoding="utf-8") as f:
    html = f.read()

HTML(html)
````

I simuleringen over ser du seks energinivåer for et system, jevnt fordelt med avstand $\Delta\varepsilon$ mellom hvert nivå ($\varepsilon_n = n\Delta\varepsilon$). Glideren styrer $k_BT/\Delta\varepsilon$, altså hvor stor den termiske energien er sammenlignet med avstanden mellom energinivåene. Bruk den til å se hva som skjer:
- Ved lav $k_BT/\Delta\varepsilon$ er nesten all sannsynlighet i grunntilstanden.
- Etter hvert som $k_BT/\Delta\varepsilon$ øker, blir stadig flere av de høyere nivåene befolket.
- Ved svært høy $k_BT/\Delta\varepsilon$ nærmer fordelingen seg (nesten) jevn: alle nivåene blir omtrent like sannsynlige.

Dette er mekanismen som avgjør hvorfor et system ved romtemperatur som regel *ikke* er i en høyt eksitert vibrasjons- eller elektrontilstand: energitrinnene $\Delta\varepsilon$ er ofte store sammenlignet med $k_BT$ ved romtemperatur.

Fra fordelingen kan vi se de samme "tommelfingerregler" som listet tidligere. **Lavenergitilstander er alltid mer befolket enn høyenergitilstander**, ved en gitt temperatur. Det er fordi at systemet må ha en total energi (makro) som er fordelt på hver partikkel, og hvis flere partikler har lav energi, vil det være flere mulige fordelinger av total energi. Hvis én partikkel har veldig høy energi, må alle de gjenstående ha ganske lav energi. Dette er veldig usannsynlig.**Forholdet mellom to sannsynligheter** avhenger bare av energiforskjellen og temperaturen, ikke av de absolutte energiverdiene:

$$\frac{p(\varepsilon_1)}{p(\varepsilon_0)} = e^{-\frac{\varepsilon_1-\varepsilon_0}{k_BT}}$$

Dette kalles gjerne **Boltzmann-faktoren**, og den dukker opp igjen i fysikalsk kjemi. 

Boltzmanns fordelingslov er selve broen mellom mikro og makro. Så lenge vi vet hvordan sannsynligheten fordeler seg over mikrotilstandene, kan vi i prinsippet regne oss frem til makroskopiske størrelser som gjennomsnittlig energi, varmekapasitet og entropi ved å veie hver mikrotilstands bidrag med dens sannsynlighet $p(\varepsilon_n)$. Det er dette partisjonsfunksjonen gjør mulig å regne ut systematisk, og det er tema for neste avsnitt. 


:::{admonition} OBS
:class: error
Boltzmanns fordelingslov handler ikke nødvendigvis om én og én partikkel. Litt mer presist gjelder Boltzmann-fordelingen for et hvilket som helst system som er i termisk kontakt med et reservoar (omgivelse) ved temperatur $T$, og som kan utveksle energi med dette reservoaret, med andre ord for det **kanoniske ensemblet**. "Systemet" kan i prinsippet være en enkelt partikkel *eller* feks en hel gass, betraktet som ett stort system, hvis den selv er i kontakt med et enda større reservoar. Poenget er at loven beskriver sannsynligheten for å finne det systemet du har valgt å se på i en gitt mikrotilstand, gitt temperaturen til omgivelsene rundt. I kjemien blir det ofte én og én partikkel i praksis, fordi vi i kjemi som regel har med et stort antall ikke-vekselvirkende partikler å gjøre (ideell gass-tilnærmingen). Da kan man vise at hele systemets energifordeling deles opp til at hver enkelt partikkel uavhengig følger nøyaktig den samme Boltzmann-fordelingen over sine egne energinivåer. Det er derfor vi kan snakke om "hvor mange molekyler som er i grunntilstanden" osv. Det fungerer fordi partiklene ikke påvirker hverandre. Hvis de vekselvirker sterkt med hverandre, holder ikke denne tilnærmingen lenger, da må du behandle hele systemet av $N$ partikler samlet som "systemet" i den kanoniske fordelingen, og det blir fort mye mer komplisert matematikk.
:::

**Eksempel på bruk av Boltzmanns fordelingslov: vibrasjon i et gassmolekyl**
^^^

Et gassmolekyl har ikke bare translasjonsbevegelse (bevegelse i rommet i x- y- og z-retning) det kan også vibrere. Dette gir molekylet en **ekstra frihetsgrad** utover de tre translasjonelle. Vi kan modellere vibrasjonen som en enkel harmonisk oscillator, med kvantiserte energinivåer

$$E_v = v\,h\nu, \qquad v=0,1,2,\dots$$

der $v$ er vibrasjonskvantetallet og $\nu$ er vibrasjonsfrekvensen. Disse nivåene er ikke-degenererte: det finnes bare én mikrotilstand per $v$.

La oss sammenligne sannsynligheten for at molekylet befinner seg i første eksiterte vibrasjonsnivå ($v=1$) med sannsynligheten for at det er i grunntilstanden ($v=0$).

```python
import numpy as np

E1 = 1  # vibrasjonskvantetall v=1, energi i enheter av h*nu
E0 = 0  # grunntilstand v=0

hv_over_kT = 1.0  # velg selv forholdet h*nu / (k_B T)

p_ratio = np.exp(-(E1 - E0) * hv_over_kT)
print(p_ratio)
```

Ved $h\nu/k_BT=1$ er sannsynligheten for at molekylet er i første eksiterte vibrasjonsnivå omtrent $p(E_1)/p(E_0)\approx0{,}37$ ganger sannsynligheten for å være i grunntilstanden. Med andre ord: ved denne temperaturen befinner de aller fleste molekylene seg fortsatt i vibrasjonsgrunntilstanden.

Jo høyere verdi av `hv_over_kT` (lavere temperatur, eller en stivere binding med høyere vibrasjonsfrekvens $\nu$), desto sjeldnere blir de eksiterte vibrasjonsnivåene befolket sammenlignet med grunntilstanden.
:::

## Partisjonsfunksjonen

**Partisjonsfunksjonen** $q$ for en partikkel eller $Q$ for et sammensatt system er et matematisk verktøy som er veldig nyttig når vi vil bruke mikroskopiske muligheter til å forutsi makroskopiske egenskaper. 

$$ q = \sum_n e^{-\frac{\varepsilon_n}{k_BT}} \text{,}$$

$$ Q = \sum_i e^{-\frac{E_i}{k_BT}} \text{.}$$

Disse to er samme funksjon, så det er strengt tatt ikke nødvendig å inkludere begge. Her har vi inkludert begge, bare for å vise forskjellen i notasjon og for å presisere at det er viktig å vite hva slags system man studerer. Fra nå av vil vi bruke $Q$ for å forklare egenskaper ved partisjonsfunksjonen, men være klar over at disse gjelder for $q$ også. 

$Q$ slik vi har definert den her, er spesifikt den **kanoniske partisjonsfunksjonen**, gyldig for et system med fast $N, V, T$. Dersom antall partikler i systemet også kan variere (f.eks. molekyler som adsorberer og desorberer fra en overflate), trengs et annet ensemble og en annen partisjonsfunksjon, men dette dekker vi ikke her.

Hvis vi sier at $t$ er antallet mikrotilstander, slik at $i=0,1,2,...,t-1$, og skriver ut $Q$, ser det slik ut:

$$
Q=\sum_{i=0}^{t-1} e^{-\frac{E_i}{k_BT}}
= e^{-\frac{E_0}{k_BT}}
+ e^{-\frac{E_1}{k_BT}}
+ e^{-\frac{E_2}{k_BT}}
+ \cdots
+ e^{-\frac{E_{t-1}}{k_BT}}.
$$

Partisjonsfunksjonen summerer altså bidrag fra *alle* mikrotilstander, enten vi snakker små eller store systemer. Den adderer sannsynligheten for alle mikrotilstandene, og er derfor et lite beist av informasjon om systemet. 

La oss ta en liten oppsummering før vi går videre: 
- Kvantemekanikken gir oss energinivåene.
- Statistisk mekanikk gir sannsynlighetene for disse.
- Partisjonsfunksjonen samler all informasjonen.

Vi er nødt til å kjenne til energinivåene for å regne ut $Q$, men når vi måler energier eksperimentelt, måler vi som regel *energidifferanser* og ikke absolutte energier. Derfor skriver vi om partisjonsfunksjonen litt ved å definere grunntilstandsenergiene $E_0=0$ . Så kan vi skrive funksjonen uttrykt ved energidifferansene: 

$$ Q = 1 + e^{-\frac{(E_1-E_0)}{k_BT}} + e^{-\frac{(E_2-E_0)}{k_BT}} + \cdots + e^{-\frac{(E_{t-1}-E_0)}{k_BT}} \text{.}$$

Man kan også uttrykke partisjonsfunksjonen ved *makrotilstandene*. Vi bruker multiplisiteten av tilstander $\Omega$, men endrer litt på skrivemåten. Et system kan eksistere i energinivå $E_j$ på $\Omega(E_j)$ måter. Hvis $\Omega(E_j)>1$, er tilstanden degenerert. Noen steder kalles også multiplisiteten for "tilstandstettheten", kjært barn har mange navn. Si at vi har $l$ ulike energinivåer/makrotilstander, som kan være degenererte eller ei. Da har vi $j=0,1,2,...,l-1$. Når vi utrykker partisjonsfunksjonen på denne måten, ser den slik ut: 

$$ Q=\sum_{j=0}^{l-1} \Omega(E_j) e^{-\frac{E_j}{k_BT}}.$$

Måten funksjonen skrives er opp til deg, det kommer an på hvordan systemet du vil studere beskrives best. For eksempel er $\sum_i$ over hver enkelt mikrotilstand mest naturlig når tilstandene ikke er degenererte, eller når degenerasjonen er upraktisk å gruppere. 

Dersom man kjenner energitilstandene til et system, er det ganske enkelt å regne ut $Q$. Men hvordan skal tallet man får tolkes? En intuitiv måte å tenke på hva partisjonsfunksjonen *er*, er å se på den som "antallet effektivt tilgjengelige tilstander". Husk, sannsynligheten for hver av mikrotilstandene er bakt inn i partisjonsfunksjonen. Derfor kan partisjonsfunksjonen gi oss svaret på hvor mange tilstander systemet i praksis kan eksistere i, gitt en temperatur og avstanden mellom energinivåene.

- Ved stort gap er praktisk talt all sannsynlighet låst i grunntilstanden, og $Q\to 1$, selv om de fire andre nivåene fortsatt eksisterer og er fullt lovlige, bidrar de ikke lenger nevneverdig til $Q$.

**Eksempel: en enkel molekylbryter**
^^^

Dette eksempelet illustrerer oppsett og enkel bruk av partisjonsfunksjonen. 

Tenk deg et molekyl med én intern binding som enten er hel eller brutt. Når bindingen er hel, kalder vi molekylet **lukket**, og det finnes bare én mulig konfigurasjon i denne tilstanden: molekylet er stivt, og vi setter denne energien som referanse, $\varepsilon = 0$. Når bindingen brytes, kan den delen av molekylet som frigjøres rotere til $g$ forskjellige, like sannsynlige vinkler. Vi kaller denne makrotilstanden **åpen**, og alle disse $g$ mikrotilstandene har samme energi $\Delta\varepsilon$ over den lukkede tilstanden. La oss si $g=3$.

Siden dette er akkurat den situasjonen der flere mikrotilstander deler samme energi, kan vi bruke $Q=\sum_{j=0}^{l-1} \Omega(E_j)e^{-E_j/k_BT}$ direkte, med bare to ledd ($l=2$):

$$\Omega(0) = 1 \quad \text{(lukket)}, \qquad \Omega(\Delta\varepsilon) = g = 3 \quad \text{(åpen)}$$

$$Q = \Omega(0)e^{0} + \Omega(\Delta\varepsilon)e^{-\Delta\varepsilon/k_BT} = 1+3e^{-\Delta\varepsilon/k_BT}\text{.}$$

Sannsynlighetene for hver makrotilstand blir da

$$p_{\text{lukket}} = \frac{1}{Q}\text{,} \qquad p_{\text{åpen}} = \frac{3e^{-\Delta\varepsilon/k_BT}}{Q}\text{.}$$

Ved lav temperatur ($\Delta\varepsilon/k_BT \to \infty$) domineres molekylet av den lukkede tilstanden, og $Q\to 1$. Ved høy temperatur ($\Delta\varepsilon/k_BT \to 0$) blir alle $g+1=4$ mikrotilstandene like sannsynlige, og $Q \to 4$. Merk at $p_{\text{lukket}}$ ikke går mot null her, men mot $1/4$, siden den lukkede tilstanden fortsatt er én av de fire like sannsynlige mikrotilstandene.

![Sannsynligheten for lukket og åpen tilstand som funksjon av k_BT/Δε](Simuleringer/molekylbryter_populasjon.png)

Figuren viser krysningspunktet: ved lav $k_BT/\Delta\varepsilon$ (lav temperatur relativt til energigapet) er molekylet nesten utelukkende lukket, mens ved høy $k_BT/\Delta\varepsilon$ flater begge kurvene ut på sine respektive grenseverdier $1/4$ og $3/4$,  bestemt av degenerasjonen $g$, ikke av at den ene tilstanden "vinner" fullstendig.
:::

### Skjelnbare vs uskjelnbare partikler

Vi skal nå se på to ulike måter å modellere et system av flere partikler, avhengig av om vi kan spore de enkelte partiklene individuelt eller ikke. Disse to modellene krever ulike former av partisjonsfunksjonen. 

![Skjelnbare og uskjelnbare partikler](Simuleringer/skjelnbare_vs_uskjelnbare.png)

**Skjelnbare partikler** kan identifiseres individuelt, typisk fordi hver av dem har en fast, kjent posisjon. Tenk deg atomer festet i et krystallgitter: atom $A$ sitter alltid på posisjonen $(x_A,y_A,z_A)$, og atom $B$ på $(x_B,y_B,z_B)$, uavhengig av om $A$ og $B$ er samme grunnstoff. Siden vi alltid kan si "dette er $A$ sin posisjon, og det er $B$ sin", er partiklene skjelnbare, selv om de er kjemisk identiske.

Så lenge vi antar at partiklene ikke vekselvirker med hverandre, er den totale energien til systemet summen av enkeltbidragene, $E=\varepsilon_1+\varepsilon_2+\cdots+\varepsilon_N$. Denne antakelsen er avgjørende: uten den ville energien til én partikkel avhenge av tilstanden til de andre, og summen ville ikke faktorisere så enkelt. Så lenge antakelsen holder, kan vi telle opp alle kombinasjoner av kvantetilstander separat for hver partikkel, og partisjonsfunksjonen faktoriseres:

$$Q_{\text{skjelnbare}} = \sum_{\text{alle kombinasjoner}} e^{-(\varepsilon_1+\varepsilon_2+\cdots+\varepsilon_N)/k_BT} = \left(\sum_n e^{-\varepsilon_n/k_BT}\right)^N = q^N$$

der $q$ er partisjonsfunksjonen til én enkelt partikkel. Dette er en enkel modell for f.eks. atomer i et fast stoff, der hvert atom vibrerer om sin egen faste likevektsposisjon. 

**Uskjelnbare systemer** består av identiske partikler som *ikke* kan spores individuelt, det finnes ingen fysisk egenskap (som en fast posisjon) som lar oss si hvilken som er hvilken. Det klassiske eksempelet er atomene i en gass: de beveger seg fritt og bytter kontinuerlig plass med hverandre, så det gir ingen mening å spørre "hvor er atom nummer 7 akkurat nå, sammenlignet med i sta?".

Hvis vi naivt brukte $q^N$ for et slikt system, ville vi telle hver fysiske mikrotilstand flere ganger: én gang for hver måte å fordele "hvem er hvem" blant de $N$ identiske partiklene, altså $N!$ ganger for mye (du kan lese om fakultetet $!$ i matte-modulen). Riktig partisjonsfunksjon blir derfor

$$Q_{\text{uskjelnbare}} = \frac{q^N}{N!}\text{.}$$

Begge uttrykkene $Q_{\text{skjelnbare}}$ og $Q_{\text{uskjelnbare}}$ forutsetter i tillegg at antall partikler i systemet $N$, er fast. Igjen en forutsetning fra det kanoniske ensemblet.

:::{Admonition} OBS
:class: error
Om en løsning eller gass ser homogen ut for øyet, forteller det oss *ingenting* om hvorvidt systemet er uskjelnbart. Det avgjørende spørsmålet er ikke hvordan systemet ser ut, men om partiklene *kan spores*, for eksempel via en fast posisjon. Atomer i et krystallgitter er skjelnbare selv om krystallen er laget av kun ett grunnstoff og ser helt homogen ut, nettopp fordi hver posisjon er sporbar. Molekyler eller atomer i en gass av samme stoff er uskjelnbare, ikke fordi de "ser like ut", men fordi ingenting i systemet lar oss følge ett bestemt molekyl over tid.
:::

Om et system skal modelleres som skjelnbart eller uskjelnbart, er altså ikke en vilkårlig smakssak. Det er bestemt av om partiklene har en fysisk, varig egenskap som skiller dem fra hverandre eller ei. Det er faktisk nettopp derfor samme stoff behandles ulikt avhengig av fase: atomene i en krystall av et grunnstoff regnes som skjelnbare (faste posisjoner), mens det samme stoffet som gass regnes som uskjelnbare (fri bevegelse), og det er $N!$-faktoren som gjør at beregninger av entropi og frie energier stemmer riktig i det siste tilfellet.

## Termodynamiske egenskaper forutsies av partisjonsfunksjonen

Nå skal vi se hvordan de termodynamiske størrelsene, feks indre energi, kan uttrykkes og dermed regnes ut ved bruk av partisjonsfunksjonen. Vi skal også se at selv om størrelsene er uttrykt på en ny måte, handler de om akkurat det samme som de empiriske uttrykkene. Vi tar ikke for oss noen detaljerte utledninger her, og tar for gitt at matten stemmer, hvis du er interessert vil dette bli undervist i Fysikalsk Kjemi III. Det er *ikke* pensum å kunne utlede noen av disse uttrykkene i Fysikalsk Kjemi I.

Vi introduserer noen størrelser som ikke har blitt nevnt tidligere i Bifrost. For nå handler det kun om å *se at disse størrelsene kommer fra mikroskopiske egenskaper*. I neste modul og videre i pensum diskuteres disse størrelsene i mer detalj, fra et mer kjemi-rettet synspunkt. 

### Indre energi

**Indre energi** er forventningsverdien, eller ensemble-gjennomsnittet $\langle E \rangle$, over alle mikrotilstands-energiene $E_i$, vektet av sannsynlighetene deres $p_i$:

$$U= \langle E \rangle = \sum_i E_i p_i=\frac{1}{Q}\sum_i E_i e^{-\frac{E_i}{k_BT}}\text{,}$$

der vi har brukt Boltzmanns fordelingslov $p_i=\frac{1}{Q}e^{-\frac{E_i}{k_BT}}$. Videre kan det vises at dette er det samme som den deriverte av $\ln Q$ med hensyn på temperaturen:

$$ U = k_BT^2\left(\frac{\partial \ln Q}{\partial T}\right)_{V,N} \text{.}$$

Noen lærebøker skriver i stedet dette uttrykket ved hjelp av $\beta \equiv \frac{1}{k_BT}$, som gir en litt penere form:

$$ U = -\left(\frac{\partial \ln Q}{\partial \beta}\right)_{V,N}\text{.}$$

:::{admonition} OBS: hjelpenotes
:class: error
Notasjonen $\langle E \rangle$ betyr bare "gjennomsnittlig energi".

Å "vekte" noe med sannsynligheten dens, betyr at dersom sannsynligheten for én energi $E_1$ er større enn en annen energi $E_2$, så må denne forskjellen i sannsynlighet inkluderes i uttrykket for at det skal bli riktig. Vi har jo sett at ulike mikrotilstander har ulik sannsynlighet, og dette kan vi ikke bare overse når vi regner ut gjennomsnittet over dem. 
:::

På laben måler vi *endringen i indre energi*. Systemet vårt kan motta eller avgi energi (neste modul diskuterer hvordan) og gå fra en tilstand A til en tilstand B. Energiene i tilstand A og B kan, som vi har sett, uttrykkes som:

$$ U_A = -\left(\frac{\partial \ln Q_A}{\partial \beta_A}\right)_{V,N},$$

$$ U_B = -\left(\frac{\partial \ln Q_B}{\partial \beta_B}\right)_{V,N}.$$

Den empiriske formelen for $\Delta U$ er gitt ved:

$$ \Delta U = U_B - U_A = q + w, $$

der $q$ er varmen og $w$ er arbeidet. Vi kommer tilbake til disse størrelsene veldig snart. Husk, poenget nå er å se sammenhengen: mens $U_A$ og $U_B$ lar oss koble makroskopisk energi til mikroskopiske tilstander, gir sammenhengen med $q$ og $w$ oss en direkte overgang til hvordan disse størrelsene måles og observeres. Vi har dermed beveget oss fra det mikroskopiske, til målbare størrelser på laben. 

Gjennomsnittet over ensemblet, altså indre energi $U$, kan visualiseres slik:

### Entropi

**Entropi** handler ikke om uorden, som du kanskje har hørt før, men om hvor mange mikroskopiske muligheter som er kompatible med det som blir den makroskopiske observasjonen. En økning i entropi er det samme som at systemet går fra en mindre sannsynlig makroskopisk tilstand til en mer sannsynlig makroskopisk tilstand. Vi har koblet dette sammen med sannsynligheter for mikrotilstander i [Entropi-modulen](https://andreasdh.github.io/Bifrost/docs/entropi.html#likevekt-og-entropimaksimering). Uttrykket vi finner der, er Gibbs entropi, som er gitt ved:

$$ S = -k_B \sum_i p_i \ln p_i $$

der summen går over alle mikrotilstander $i$, og $p_i$ er sannsynligheten for å finne systemet i mikrotilstand $i$.

Vi kan koble denne definisjonen til partisjonsfunksjonen ved å sette inn Boltzmann-fordelingen $p_i = \frac{1}{Q}e^{-E_i/k_BT}$. Vi har at:

$$ \ln p_i = -\frac{E_i}{k_BT} - \ln Q $$

Så setter vi $\ln p_i$ inn i uttrykket for entropien:

$$ S = -k_B \sum_i p_i\left(-\frac{E_i}{k_BT} - \ln Q\right) = \frac{1}{T}\sum_i p_i E_i \; + \; k_B\ln Q \sum_i p_i $$

Her er $\sum_i p_i E_i = U$ gjennomsnittsenergien og $\sum_i p_i = 1$ (siden sannsynlighetene summerer til 1). Vi ender dermed opp med entropien uttrykt ved partisjonsfunksjonen:

$$ S = k_B \ln Q + \frac{U}{T} \text{.}$$

### Helmholtz fri energi

Kombinerer vi de to uttrykkene over ($S$ og $U$), får vi Helmholtz fri energi rett fram:

$$ F = U - TS = U - T\left(k_B\ln Q + \frac{U}{T}\right) = -k_BT\ln Q \text{.}$$

### Varmekapasitet

Varmekapasiteten ved konstant volum er rett og slett hvor mye $U$ endrer seg med temperaturen:

$$ C_V = \left(\frac{\partial U}{\partial T}\right)_{V,N}\text{.}$$

Siden vi allerede har $U$ uttrykt ved $Q$, kan $C_V$ i prinsippet regnes ut fullstendig fra partisjonsfunksjonen alene.

Her er en liste over uttrykkene vi har gått gjennom.

:::{list-table} Empiriske- vs. partisjonsfunksjon-uttrykk
:header-rows: 1
:widths: 20 30

* - Empirisk uttrykk
  - Uttrykt ved partisjonsfunksjonen
* - $ S = -\left(\frac{\partial F}{\partial T}\right)_V=\frac{U-F}{T}$
  - $ S = k_B \ln Q + \frac{U}{T}$
* - $ U = q + w $
  - $ U = k_BT^2\left(\frac{\partial \ln Q}{\partial T}\right)_{V,N} = -\left(\frac{\partial \ln Q}{\partial \beta}\right)_{V,N}$
* - $ F = U - TS$
  - $ F = -k_B T \ln Q$
* - $ C_V = \left(\frac{\partial U}{\partial T}\right)_{V,N}$
  - $ C_V = k_B \beta^2 \left(\frac{\partial^2 \ln Q}{\partial \beta^2}\right)_{V,N} = k_B T^2\left(\frac{\partial^2 \ln Q}{\partial T^2}\right)_{V,N} + 2k_BT\left(\frac{\partial \ln Q}{\partial T}\right)_{V,N}$
:::

## Hvorfor varierer varmekapasitet?

### Ekvipartisjonsteoremet

**Ekvipartisjonsteoremet** sier at den **totale termiske energien til et system er delt likt mellom dets frihetsgrader**. Hver frihetsgrad (måte å lagre energi på) får tildelt et gjennomsnitt på $\frac{1}{2}k_BT$.

Se for deg et laboratorium med mye aktivitet: hvis en fast mengde termisk energi (kjemikerne) skal fordeles likt mellom fem avtrekksskap (frihetsgradene), ender hvert avtrekksskap opp med omtrent like mange kjemikere. Kjemikerne representerer altså *energien* som skal fordeles, ikke temperaturen selv. Temperaturen bestemmer bare hvor mange kjemikere som totalt er tilgjengelige å fordele.

:::{card} 

**Eksempel: $\text{O}_2$ - et diatomisk molekyl**
^^^

Et oksygen-molekyl har:
- 3 translasjonelle frihetsgrader (beveger seg i tre romlige retninger).
- 2 rotasjonelle frihetsgrader (roterer om to akser, husk fra tidligere at et lineært molekyl mangler den tredje rotasjonsaksen).
- 1 vibrasjonell frihetsgrad (bindingen strekkes og trekkes sammen).

Her ligger en liten fallgruve, vibrasjon bidrar med **to** kvadratiske energiledd, ikke ett: én for kinetisk energi (atomene beveger seg) og én for potensiell energi (energien lagret i selve bindingen), siden en harmonisk oscillator alltid lagrer energi i begge former samtidig. Vibrasjon teller derfor som 2 frihetsgrader i ekvipartisjonsregnskapet, selv om det bare er én fysisk bevegelse. Dette kommer vi tilbake til i intermolekylære krefter modulen. Totalt:

$$3+2+2=7 \text{ frihetsgrader} \quad\Rightarrow\quad U=\frac{7}{2}k_BT$$
:::

Kvantemekanikken er den generelle teorien når vi beskriver systemer på mikroskala. Ekvipartisjonsteoremet er en klassisk tilnærming som blir gyldig når energinivåene er så tettliggende at de oppfører seg nesten kontinuerlig i forhold til den termiske energien $k_BT$. Ved lavere temperaturer kan ikke molekylene lenger dele energien inn i så små, kontinuerlige biter, og høyere-energi frihetsgrader **fryses ut**. Ekvipartisjonsteoremet er derfor ikke et grunnleggende prinsipp, men den klassiske grensen av den kvantemekaniske beskrivelsen, resultatet som gjelder når kvantiseringen ikke lenger er merkbar.

La oss repetere frihetsgradene og se hvilke av dem som fryses ut først, og hvorfor. Dette er faktisk grunnen til at varmekapasiteten til reelle gasser avviker fra den enkle klassiske forventningen.

### Translasjon

Translasjonsnivåene ligger nesten kontinuerlig tett. Energigapet er så lite sammenlignet med $k_BT$ at translasjon bidrar fullt ut ved praktisk talt alle temperaturer vi møter i laboratoriet, den fryser aldri ut under normale forhold.

### Rotasjon

Rotasjonsnivåene har små energigap sammenlignet med vibrasjon, og aktiveres derfor lett. Ved romtemperatur er $k_BT$ som regel god nok til at rotasjon oppfører seg tilnærmet klassisk for de fleste molekyler.

### Vibrasjon

Vibrasjonsnivåene har derimot store energigap. For mange molekyler ved romtemperatur er $\Delta\varepsilon_{\text{vib}} \gg k_BT$, så vibrasjonen fryses ut: molekylet sitter nesten utelukkende i vibrasjonens grunntilstand, og bidrar dermed ikke med sin fulle klassiske andel til $U$ eller $C_V$. Dette er nøyaktig grunnen til at oksygengass ($O_2$) og nitrogengass ($N_2$) ved romtemperatur har en varmekapasitet nær $\frac{5}{2}R$ (bare translasjon + rotasjon), ikke $\frac{7}{2}R$ som den fulle klassiske telningen over skulle tilsi. Vibrasjonen deres er rett og slett for "stiv" til å tine ved den temperaturen.

Legg også merke til hvordan $C_V$ hopper i tydelige steg, $\frac{3}{2}R \to \frac{5}{2}R \to \frac{7}{2}R$, i stedet for å øke jevnt, nettopp fordi hver frihetsgrad enten er "på" eller "av" avhengig av om $k_BT$ er stor nok til å konkurrere med energigapet.

- Hvis $\Delta\varepsilon \ll k_BT$: klassisk grense, ekvipartisjonen holder, frihetsgraden bidrar fullt.
- Hvis $\Delta\varepsilon \gg k_BT$: nivåene er frosset, ekvipartisjonen bryter sammen for denne frihetsgraden.

### Varmekapasitet

Varmekapasiteten ved konstant volum, $C_V=\left(\frac{\partial U}{\partial T}\right)_{V,N}$, måler rett og slett hvor mye ekstra energi et system klarer å ta opp per grad temperaturøkning. Siden hver frihetsgrad enten bidrar fullt (klassisk) eller ikke i det hele tatt (utfrosset), er $C_V$ i praksis en teller for *hvor mange frihetsgrader som faktisk er "våkne"* ved en gitt temperatur, ikke et fast tall bestemt av molekylets struktur *alene*. 

To ting ved molekylstrukturen avgjør varmekapasiteten: hvor mange frihetsgrader molekylet har, og hvor stort energigapet er for hver av dem.

**Antall atomer og geometri** bestemmer *hvor mange* frihetsgrader som finnes. For et molekyl med $N$ atomer er det alltid $3N$ frihetsgrader totalt, men hvordan disse fordeler seg mellom translasjon, rotasjon og vibrasjon avhenger av formen:

- Lineære molekyler: (som $O_2$, $CO_2$): 3 translasjon + 2 rotasjon + $(3N-5)$ vibrasjon.
- Ikke-lineære (bøyde) molekyler: (som $H_2O$): 3 translasjon + 3 rotasjon + $(3N-6)$ vibrasjon.

Dette er den samme forskjellen vi så [tidligere](https://andreasdh.github.io/Bifrost/docs/intro.html#kvantetilstander) mellom $O_2$ og $H_2O$: et lineært molekyl "mangler" én rotasjonsakse sammenlignet med et bøyd molekyl, og den frihetsgraden dukker i stedet opp som én ekstra vibrasjonsmode.

**Atommasse og bindingsstyrke** bestemmer *når* hver frihetsgrad våkner. Selv om to molekyler har like mange frihetsgrader, kan energigapene deres være helt forskjellige.Molekylstrukturen bestemmer altså ikke bare hvor mange ledd som *til slutt* kan bidra til $U$ og $C_V$ ved svært høy temperatur, men også nøyaktig ved hvilken temperatur hvert ledd faktisk begynner å telle med.

## Oppsummering

**Kvantemekanikk**
- Energi er kvantisert: kun diskrete verdier $\varepsilon_0,\varepsilon_1,\varepsilon_2,\dots$ er tillatt (Franck-Hertz eksperimentet: elektroner mister energi i hele multipler av $\Delta\varepsilon$, aldri kontinuerlig).
- Kvantetilstand = fullstendig beskrivelse av alle frihetsgrader (elektronisk, vibrasjon, rotasjon, translasjon).
- Translasjon er teknisk sett også kvantisert, men nivåene ligger så tett at de oppfører seg som kontinuerlige.

**Mikrotilstander og multiplisitet**
- Mikrotilstand $\leftrightarrow$ én kvantetilstand (for ett enkelt atom/molekyl).
- Degenererte tilstander: flere kvantetilstander deler samme energi, telles med $g$.
- For systemer av mange partikler: én mikrotilstand = kombinasjonen av alle partiklenes kvantetilstander samtidig.
- Multiplisitet/tilstandstetthet $\Omega(E)$ = antall mikrotilstander som gir samme energi $E$.

**Boltzmanns fordelingslov**
- $p(\varepsilon)=\frac{1}{q}e^{-\varepsilon/k_BT}$: sannsynligheten for én mikrotilstand ved en gitt energi.
- Lave energier mer sannsynlige enn høye, ved en gitt temperatur.
- Forholdet $\Delta\varepsilon/k_BT$ avgjør alt, ikke $\Delta\varepsilon$ eller $T$ hver for seg.

**Partisjonsfunksjonen**
- $q=\sum_n e^{-\varepsilon_n/k_BT}$ (ett molekyl).
- $Q=\sum_i e^{-E_i/k_BT}$ (helt system).
- Kan òg skrives med multiplisitet: $Q=\sum_l \Omega(E_l)e^{-E_l/k_BT}$.
- Tolkning: "antall effektivt tilgjengelige tilstander": $Q\to 1$ ved lav temperatur/stort gap, $Q\to$ (antall tilstander) ved høy temperatur/lite gap.
- Skjelnbare partikler: $Q=q^N$. Uskjelnbare partikler: $Q=q^N/N!$ (unngår overtelling av identiske ombyttinger).

**Termodynamiske størrelser fra $Q$**
- $U=\langle E\rangle = k_BT^2\left(\frac{\partial \ln Q}{\partial T}\right)_{V,N}$: ensemble-gjennomsnitt over mikrotilstandsenergiene.
- $S\approx k_B\ln Q + \dfrac{U}{T}$.
- $F=U-TS=-k_BT\ln Q$.
- $C_V=\left(\frac{\partial U}{\partial T}\right)_{V,N}$.

**Ekvipartisjon og varmekapasitet**
- Hver frihetsgrad bidrar $\frac{1}{2}k_BT$ til $U$ i klassisk grense (vibrasjon teller dobbelt: kinetisk + potensiell energi).
- Ekvipartisjon = klassisk grense av kvantemekanikken, gyldig når $\Delta\varepsilon \ll k_BT$.
- Frihetsgrader fryser ut i rekkefølge etter energigap: vibrasjon fryser først, så rotasjon, translasjon fryser (praktisk talt) aldri.
- Molekylstruktur (antall atomer, lineær/bøyd geometri, atommasse, bindingsstivhet) bestemmer både *hvor mange* frihetsgrader som finnes og *ved hvilken temperatur* hver av dem våkner, derfor varierer $C_V$ fra molekyl til molekyl.
