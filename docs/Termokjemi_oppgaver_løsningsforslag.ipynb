## Termokjemi

**Entalpi** 

1. Hess lov sier at dersom vi deler en prosess for et system opp i del-prosesser, så er entalpiendringen for systemet lik summen av entalpiendringene for delprosessene. Siden entalpi er en tilstandsfunksjon kan veien mellom start- og sluttilstand velges fritt. Det å dele prosessen opp i delprosesser svarer til å endre veien men så lenge start- og sluttilstanden er de samme vil entalpiendringen være den samme. Det å finne termodynamiske størrelser eksperimentelt kan være utfordrende. Istedenfor holder det å vite start og slutt-tilstandene. Veien mellom tilstandene kan da konstrueres ved å legge sammen kjente reaksjoner og man kan regne ut mange termodynamiske størrelser basert på tabeller.
