# Løsninger og blandinger kan beskrives av lattice-modeller

## Introduksjon til modellen

Vi har et antall $N_A$ av en partikkel A og et antall $N_B$ av partikkel B. Sammen fyller partiklene opp et gitter med $N$ plasser i en lattice-modell slik at vi har:

$$N = N_A + N_B$$

Antall tilgjengelige mikrotilstander vil da kunne beskrives som: 

$$W = \frac{N!}{N_A!N_B!}$$

som beskrevet i [lenke]. Dette blir utgangspunktet vårt til å utlede størrelser som entropi, indre energi, fri energi og kjemisk potensial for blandinger med to komponenter. La oss begynne med entropien.

## Blandingsentropi

Med utgangspunkt i Boltzmanns definisjon av entropi har vi følgende uttrykk for blandingsentropien: 

$$\Delta S_{sol}  = klnW$$

Ved hjelp av [Stirlings approksimasjon]lenke, kan vi derimot uttrykket antall tilgjengelige mikrotilstander på en litt annen måte:

$$ln\left(\frac{N!}{N_A!N_B!}\right) = NlnN - N_AlnN_A - N_BlnN_B$$

Da får vi for blandingsentropien: 

$$\Delta S_{sol} = k(NlnN - N_AlnN_A - N_BlnN_B)$$

Vi skriver om ved å sette $N_A = N \times \left(\frac{N_A}{N}\right)$ og $N_B = N \times \left(\frac{N_B}{N}\right)$ får vi:

$$\Delta S_{sol} = k(NlnN - N \times \left(\frac{N_A}{N}\right)lnN_A - N \times \left(\frac{N_B}{N}\right)lnN_B)$$

Vi kan trekke N utenfor parentesen: 

$$\Delta S_{sol} = Nk(lnN - \left(\frac{N_A}{N}\right)lnN_A - \left(\frac{N_B}{N}\right)lnN_B)$$

Videre kan vi skrive om $lnN_A$ on $lnN_B$ slik (husk [lenke] logaritmeregler):

$$lnN_A = lnN + ln\frac{N_A}{N}$$

$$lnN_B = lnN + ln\frac{N_B}{N}$$

Vi setter dette inn i uttrykket vårt: 

$$\Delta S_{sol} = Nk(lnN - \left(\frac{N_A}{N}\right)\left(lnN + ln\frac{N_A}{N}\right) - \left(\frac{N_B}{N}\right)\left(lnN + ln\frac{N_B}{N}\right))$$

Vi kan samle logaritmene hver for seg slik: 

$$\Delta S_{sol} = Nk(lnN - \left(\frac{N_A}{N}\right)lnN - \left(\frac{N_B}{N}\right)lnN - \left(\frac{N_A}{N}\right)ln\left(\frac{N_A}{N}\right) - \left(\frac{N_B}{N}\right)ln\left(\frac{N_B}{N}\right))$$

$$\Delta S_{sol} = Nk(lnN - \left(\left(\frac{N_A}{N}\right) + \left(\frac{N_B}{N}\right)\right)lnN - \left(\frac{N_A}{N}\right)ln\left(\frac{N_A}{N}\right) - \left(\frac{N_B}{N}\right)ln\left(\frac{N_B}{N}\right))$$

Vi kan observere at $\frac{N_A}{N} + \frac{N_B}{N} = \frac{N_A + N_B}{N} = \frac{N}{N} = 1$ og dermed forsvinner $lnN$ helt fra likningen!

$$\Delta S_{sol} = Nk(lnN - lnN - \left(\frac{N_A}{N}\right)ln\left(\frac{N_A}{N}\right) - \left(\frac{N_B}{N}\right)ln\left(\frac{N_B}{N}\right))$$

$$\Delta S_{sol} = Nk( - \left(\frac{N_A}{N}\right)ln\left(\frac{N_A}{N}\right) - \left(\frac{N_B}{N}\right)ln\left(\frac{N_B}{N}\right))$$

$$\Delta S_{sol} = -Nk(\left(\frac{N_A}{N}\right)ln\left(\frac{N_A}{N}\right) + \left(\frac{N_B}{N}\right)ln\left(\frac{N_B}{N}\right))$$

Videre ønsker vi heller å uttrykke entropien ved hjelp av de _relative_ konsentrasjonene til A og B. Vi kan beskrive konsentrasjoner som _molfraksjoner_ slik: 

$$x = \frac{N_A}{N}$$

$$1-x = \frac{N_B}{N}$$

Ved å sette dette inn i formelen vi fant finner vi blandingsentropien for et to-komponent system: 

$$\frac{\Delta S_{sol}}{Nk} = -xlnx - (1-x)ln(1-x)$$

A og B vil blande seg fordi multiplisiteten av måter å arrangere A og B sammen i blandingen vil gi en større entropi enn de to rene systemene hver for seg. Dette er drivkraften bak blanding. Som du vil se i oppgaven under kan vi bruke denne blandingsentropien utledet fra et mikroskopisk lattice-modell system til å regne på blandingsentalpi i makroskopiske systemer hvor vi kjenner fraksjonene av komponentene A og B!


:::{admonition} Oppgave
:class: tip

**Oppgave 1: Mikroskopisk** 

Du har 7 A-partikler og 3 B-partikler som skal blandes sammen. Se for deg at du plasserer dem i et gitter med 10 ledige plasser. Finn først molfraksjonene av hver type partikkel før du bruker uttrykket vi fant til å beregne blandingsentropien. 

**Oppgave 2: Makroskopisk**

Du skal lage en blanding bestående av 30% metanol i vann. Finn først molfraksjonene og regn deretter ut blandingsentropien til blandingen. 

:::

```{dropdown} Klikk for å se svarene på oppgavene

**Oppgave 1: Mikroskopisk**

Vi finner først molfraksjonene:

$$N = N_A + N_B = 7 + 3 = 10$$

$$x = \frac{N_A}{N} = \frac{7}{10} = 0.7$$

$$1 - x = \frac{N_B}{N} = 0.3$

Nå kan vi bruke formelen for blandingsentropi: 

$$\frac{\Delta S_{sol}}{Nk} = -0.7ln0.7 - (1-0.7)ln(1-0.7) = -0.7ln0.7 - 0.3ln0.3 \approx 1.2$$

**Oppgave 2: Makroskopisk**

Vi finner molfraksjonene:

Vi skal ha en løsning med 30% etanol, så molfraksjonen av etanol $\frac{N_{etanol}}{N}$ må være lik $0.3$. Resten skal være vann, så molfraksjonen av vann må være $1 - 0.3 = 0.7$.

Vi får akkurat samme situasjon som i forrige oppgave!

$$\frac{\Delta S_{sol}}{Nk} = -0.7ln0.7 - (1-0.7)ln(1-0.7) = -0.7ln0.7 - 0.3ln0.3 \approx 1.2$$ 
```

##
