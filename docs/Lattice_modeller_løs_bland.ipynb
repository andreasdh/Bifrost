## Introduksjon til modellen

Vi har et antall $N_A$ av en partikkel A og et antall $N_B$ av partikkel B. Sammen fyller partiklene opp et gitter med $N$ plasser i en lattice-modell slik at vi har:

$$N = N_A + N_B$$

Antall tilgjengelige mikrotilstander vil da kunne beskrives som: 

$$W = \frac{N!}{N_A!N_B!}$$

som beskrevet [her](https://andreasdh.github.io/Bifrost/docs/mattehjelp.html#kombinasjoner-og-permutasjoner). Utforsk gittermodellen og mikrotilstandene ved hjelp av animasjonen under!

## Blandingsentropi

Med utgangspunkt i Boltzmanns definisjon av entropi har vi følgende uttrykk for blandingsentropien: 

$$\Delta S_{sol}  = k_BlnW$$

Ved hjelp av [Stirlings approksimasjon](https://andreasdh.github.io/Bifrost/docs/mattehjelp.html#stirlings-tilnaerming) kan vi derimot uttrykket antall tilgjengelige mikrotilstander på en litt annen måte:

$$ln\left(\frac{N!}{N_A!N_B!}\right) = NlnN - N_AlnN_A - N_BlnN_B$$

Da får vi for blandingsentropien: 

$$\Delta S_{sol} = k_B(NlnN - N_AlnN_A - N_BlnN_B)$$

Vi skriver om ved å sette $N_A = N \times \left(\frac{N_A}{N}\right)$ og $N_B = N \times \left(\frac{N_B}{N}\right)$ får vi:

$$\Delta S_{sol} = k_B(NlnN - N \times \left(\frac{N_A}{N}\right)lnN_A - N \times \left(\frac{N_B}{N}\right)lnN_B)$$

Vi kan trekke N utenfor parentesen: 

$$\Delta S_{sol} = Nk_B(lnN - \left(\frac{N_A}{N}\right)lnN_A - \left(\frac{N_B}{N}\right)lnN_B)$$

Videre kan vi skrive om $lnN_A$ on $lnN_B$ slik (husk [logaritmereglene](https://andreasdh.github.io/Bifrost/docs/mattehjelp.html#logaritmeregler)):

$$lnN_A = lnN + ln\frac{N_A}{N}$$

$$lnN_B = lnN + ln\frac{N_B}{N}$$

Vi setter dette inn i uttrykket vårt: 

$$\Delta S_{sol} = Nk_B(lnN - \left(\frac{N_A}{N}\right)\left(lnN + ln\frac{N_A}{N}\right) - \left(\frac{N_B}{N}\right)\left(lnN + ln\frac{N_B}{N}\right))$$

Vi kan samle logaritmene hver for seg slik: 

$$\Delta S_{sol} = Nk_B(lnN - \left(\frac{N_A}{N}\right)lnN - \left(\frac{N_B}{N}\right)lnN - \left(\frac{N_A}{N}\right)ln\left(\frac{N_A}{N}\right) - \left(\frac{N_B}{N}\right)ln\left(\frac{N_B}{N}\right))$$

$$\Delta S_{sol} = Nk_B(lnN - \left(\left(\frac{N_A}{N}\right) + \left(\frac{N_B}{N}\right)\right)lnN - \left(\frac{N_A}{N}\right)ln\left(\frac{N_A}{N}\right) - \left(\frac{N_B}{N}\right)ln\left(\frac{N_B}{N}\right))$$

Vi kan observere at $\frac{N_A}{N} + \frac{N_B}{N} = \frac{N_A + N_B}{N} = \frac{N}{N} = 1$ og dermed forsvinner $lnN$ helt fra likningen!

$$\Delta S_{sol} = Nk_B(lnN - lnN - \left(\frac{N_A}{N}\right)ln\left(\frac{N_A}{N}\right) - \left(\frac{N_B}{N}\right)ln\left(\frac{N_B}{N}\right))$$

$$\Delta S_{sol} = Nk_B( - \left(\frac{N_A}{N}\right)ln\left(\frac{N_A}{N}\right) - \left(\frac{N_B}{N}\right)ln\left(\frac{N_B}{N}\right))$$

$$\Delta S_{sol} = -Nk_B(\frac{N_A}{N}ln\left(\frac{N_A}{N}\right) + \frac{N_B}{N}ln\left(\frac{N_B}{N}\right))$$

Videre ønsker vi heller å uttrykke entropien ved hjelp av de _relative_ konsentrasjonene til A og B. Vi kan beskrive konsentrasjoner som _molfraksjoner_ slik: 

$$x = \frac{N_A}{N}$$

$$1-x = \frac{N_B}{N}$$

Ved å sette dette inn i formelen vi fant finner vi blandingsentropien for et to-komponent system: 

$$\frac{\Delta S_{sol}}{Nk_B} = -xlnx - (1-x)ln(1-x)$$

A og B vil blande seg fordi multiplisiteten av måter å arrangere A og B sammen i blandingen vil gi en større entropi enn de to rene systemene hver for seg. Dette er drivkraften bak blanding. Som du vil se i oppgaven under kan vi bruke denne blandingsentropien utledet fra et mikroskopisk lattice-modell system til å regne på blandingsentalpi i makroskopiske systemer hvor vi kjenner fraksjonene av komponentene A og B!

## Idelle løsninger 

Vi husker fra tidligere kapitler at en ideell gass er en gass uten interaksjoner mellom molekylene. Siden væsker har interaksjoner mellom seg, kan man ikke si at intermolekylære interaksjoner er lik 0. I ideelle væsker sier vi heller at interaksjonene har en gjennomsnittlig styrke: Interaksjonene mellom stoffene A og B er lik interaksjonene mellom A og A, og B og B. $\{\Delta }_{mix}U$ er fortsatt 0 når man blander to ideelle væsker slik som når vi blander to ideelle gasser. Det betyr at det ikke er noen forskjell i intermolekylære bindinger mellom molekylene i væsken. En homogen, helt ren væske vil derfor alltid være ideell, fordi alle interaksjonene mellom molekylene er de samme. Men blir noe tilsatt som har et sterkere/svakere/endret bindingsmønster mellom molekylene vil dette gjøre væsken ikke-ideell. Vann og etanol er et eksempel på en ikke-ideell væske (mye hydrogenbindinger). Et eksempel på to væsker som oppfører seg ideelt er benzen og toluen (de ligner, og interaksjonene mellom dem vil derfor være ganske like). I en ideell løsning utveksles ingen energi i løpet av blandingen og entropien kommer kun av blandingen selv som beskrevet i avsnittet over, altså ikke av volumendring, strukturering eller ordning i løsningen. Den frie energien til en ideell løsning er gitt ved: 

$$\Delta F_{sol} = -T\Delta S_{sol}$$

hvor $\Delta S_{sol}$ er hentet fra likningen vi fant i første avsnitt. Videre skal vi diskutere mer realistiske modeller, hvor vi tar hensyn også til blandingsenergi. 

## Energien til en løsning

De fleste løsninger har en løsningsenergi som kommer fra interaksjoner mellom komponentene i løsningen. I lattice-modeller tar vi hensyn til interaksjonene en partikkel har med de nærmeste naboene sine. Det vil naboene over og under seg og på hver side av seg: Totalt fire interaksjoner (i 2D).

![Vi tar hensyn til de nærmeste naboene](naerste_naboer_lattice.png) 

Dersom vi fortsetter med modellen vi har beskrevet for komponentene A og B kan vi se for oss tre ulike interaksjoner i blandingen: Dersom en A partikkel er nabo med en annen A partikkel, får vi en AA interaksjon. Dersom en B partikkel er nabo med en annen B partikkel, får vi en BB interaksjon. Og dersom en A partikkel og en B partikkel er naboer får vi en AB interaksjon. 

![Vi har tre typer kontakter](bilder/aa_bb_ab_interaksjoner.png)

Den totale energien til systemet blir da summen over alle disse interaksjonene. Dersom vi ser for oss at $m_{AA}$, $m_{BB}$ og $m_{AB}$ beskriver antallet av hver interaksjon i systemet og $w_{AA}$, $w_{BB}$ og $w_{AB}$ de tilhørende kontaktenergiene, får vi: 

$$U = m_{AA}w_{AA} + m_{BB}w_{BB} + m_{AB}w_{AB}$$

 
Fordi det er vanskelig å vite hvor mange kontakter $m_{AA}$, $m_{BB}$ og $m_{AB}$ vi har, er vi interesserte i å uttrykke disse som funksjon av $N_A$ og $N_B$ som er tall vi kjenner. Vi sier at hver av plassene i lattice-modellen har z sider. I en 2D-modell (flat) vil en boks ha fire sider, z = 4, og i en 3D-modell, vil en boks ha seks flater, z = 6. Hver kontakt mellom to partikler innebærer to sider som vist på bildet tatt fra læreboka under: 

![En kontakt innebærer to sider](bilder/sites.png)

For partiklene av type A har vi totalt $zN_A$ sider og for partiklene av type B har vi totalt $zN_B$. Fordi systemet kun består av partiklene A og B, må hver side være assosiert med en kontakt:

$$zN_A = 2 \, m_{AA} + m_{AB}$$

Altså består sidene til partiklene A av kontaktene mellom to A partikler ($2 \, m_{AA}$) og kontakten mellom en A og en B partikkel (teller én gang, $m_{AB}$). Vi har tilsvarende for $zN_B$:

$$zN_B = 2 \, m_{BB} + m_{AB}$$

Nå kan vi skrive om disse uttrykkene for å finne $m_{AA}$ og $m_{BB}$:

$$m_{AA} = \frac{zN_A - m_{AB}}{2}$$

$$m_{BB} = \frac{zN_B - m_{AB}}{2}$$

Når vi setter dette inn i uttrykket for $U$ vi fant over står bare $m_{AB}$ igjen som ukjent: 

$$U = \frac{zN_A - m_{AB}}{2}w_{AA} + \frac{zN_B - m_{AB}}{2}w_{BB} + m_{AB}w_{AB}$$

Vi skriver om uttrykket vårt litt, først ved å utvide parentesene våre, så ved å omstrukturere: 

$$U = \frac{zN_Aw_{AA} - m_{AB}w_{AA}}{2} + \frac{zN_Bw_{BB} - m_{AB}w_{BB}}{2} + m_{AB}w_{AB}$$

Vi trekker $N_A$, $N_B$ og $m_{AB}$ utenfor parenteser: 

$$ U = \left(\frac{zw_{AA}}{2}\right)N_A + \left(\frac{zw_{BB}}{2}\right)N_B + m_{AB}\left(w_{AB} - \left(\frac{w_{AA} + w_{BB}}{2}\right)\right)$$


### Mean-field approksimasjon

I prinsippet kan $m_{AB}$ variere mye mellom de ulike måtene og arrangere systemet på. I teorien måtte vi ha tatt hensyn til hver enkelt konfigurasjon og vektet dem ved hjelp av Boltzmann fordelingen, men dette føler til modeller som er tunge å regne på. _Mean-field approksimasjon_ er derimot en enklere måte å se på systemet på, og lar oss fremdeles komme frem til sentrale konklusjoner om løsninger og blandinger. Approksimasjonen forutsetter at for alle mulige $N_A$ og $N_B$ vil partiklene blandes så tilfeldig og uniformt som mulig. Dette lar oss finne et estimat for $m_{AB}$. 

La oss tenke på en plass i gitteret ved siden av en A partikkel: Hva er sjansen for at en B partikkel er i denne posisjonen? I mean-field approksimasjonen er B partiklene tilfeldig spredt utover gitteret. Sannsynligheten $p_B$ for at en B partikkel sitter i hvilkens om helst gitt posisjon i gitteret er gitt ved fraksjonen av alle posisjonene som er opptatt av B partikler fordi ingen av posisjonene vil favoriseres over noen annen: 

$$p_B = \frac{N_B}{N} = x_B = 1 - x$$

Fordi vi har z naboplasser til enhver A-partikkel, vil det gjennomsnittlige antallet AB kontakter for en bestemt A partikkel være gitt ved:

$$zp_B = z\frac{N_B}{N} = z(1-x)$$

Siden dette gjelder for alle A-partikler kan vi gange dette med det totale antallet A $N_A$ og vi får da:

$$m_{AB} \approx z\frac{N_AN_B}{N} = z\left(\frac{N_A}{N} \times \frac{N_B}{N}\right) = zNx_Ax_B = zNx(x -1)$$

Vi kan nå sette dette nye resultatet inn i likningen vi fant for blandingsenergien:

$$U = \left(\frac{zw_{AA}}{2}\right)N_A + \left(\frac{zw_{BB}}{2}\right)N_B + z\left(\frac{N_A}{N} \times \frac{N_B}{N}\right)\left(w_{AB} - \left(\frac{w_{AA} + w_{BB}}{2}\right)\right)$$

Vi kan introdusere en ny, dimensjonsløs størrelse kalt _utvekslingsparameter_ eller _exchange parameter_ på engelsk: 

$$\chi _{AB} = \frac{z}{k_BT}\left(w_{AB} - \frac{w_{AA} + w_{BB}}{2}\right)$$

Og setter denne inn i likningen for blandingsenergi: 

$$U = \left(\frac{zw_{AA}}{2}\right)N_A + \left(\frac{zw_{BB}}{2}\right)N_B + k_BT\chi _{AB}\frac{N_AN_B}{N}$$

Nå har vi funnet et uttrykk for den indre energien til en blanding ved å bruke lattice-modellen! Men hva er begrensingene til dette utrykket, særlig med tanke på bruk av mean-field approksimasjon? I noen tilfeller vil AB interaksjoner være mer favorable enn AA eller BB interaksjoner, tenk for eksempel på positivt og negativt ladde partikler som vil tiltrekkes hverandre. Da vil mean-field approksimasjonen stemme dårlig fordi A og B molekyler vil foretrekke å være ved siden av hverandre oftere enn approksimasjonen tilsier. Vi kan også ha et motsatt tilfelle hvor A molekylene trives best ved siden av andre A molekyler, og det samme gjelder for B. Også her vil mean-field approksimasjonen gi et feil bilde. 

## En løsnings fri energi

La oss gå videre til å kombinere de to resultatene våre for å få et uttrykk for den frie energien: 

$$F = U - TS$$

$$\frac{F(N_A, N_B)}{k_BT} = N_Aln\left(\frac{N_A}{N}\right) + N_Bln\left(\frac{N_B}{N}\right) + \left(\frac{zw_{AA}}{2k_BT}\right)N_A + \left(\frac{zw_{BB}}{2k_BT}\right)N_B + \chi _{AB}\frac{N_AN_B}{N}$$

Som regel er vi interesserte i forskjellen i fri energi mellom sluttilstanden til systemet og starttilstanden til systemet. I dette tilfellet blir dette differansen i fri energi mellom blandingen av A og B partiklene og den frie energien til A og B partiklene hver for seg (før blanding): 

$$\Delta F_{sol} = F(N_A, N_B) - F(N_A, 0) - F(0, N_B)$$

La oss finne $F(N_A, 0)$ og $F(0, N_B)$ ved hjelp av formelen for fri energi vi fant over: 

$$F(N_A, 0) = N_Aln\left(\frac{N_A}{N}\right) + \left(\frac{zw_{AA}}{2k_BT}\right)N_A$$

Men fordi systemet bare består av $N_A$ partikler har vi $N = N_A$ og fordi $ln(1) = 0$ står vi bare igjen med det andre leddet av likningen: 

$$\frac{F(N_A, 0)}{k_BT} = \left(\frac{zw_{AA}}{2}\right)N_A$$

På samme måte finner vi:

$$\frac{F(0, N_B)}{k_BT} = \left(\frac{zw_{BB}}{2}\right)N_B$$

La oss sette dette sammen for å finne $\Delta S_{sol}$:

$$\frac{\Delta F_{sol}}{k_BT} = N_Aln\left(\frac{N_A}{N}\right) + N_Bln\left(\frac{N_B}{N}\right) + \left(\frac{zw_{AA}}{2k_BT}\right)N_A + \left(\frac{zw_{BB}}{2k_BT}\right)N_B + \chi _{AB}\frac{N_AN_B}{N} - \left(\frac{zw_{AA}}{2k_BT}\right)N_A - \left(\frac{zw_{BB}}{2k_BT}\right)N_B$$

$$\frac{\Delta F_{sol}}{k_BT} = N_Aln\left(\frac{N_A}{N}\right) + N_Bln\left(\frac{N_B}{N}\right) + \chi _{AB}\frac{N_AN_B}{N}$$

Vi kan dele på N på begge sider: 

$$\frac{\Delta F_{sol}}{Nk_BT} = \frac{N_A}{N}ln\left(\frac{N_A}{N}\right) + \frac{N_B}{N}ln\left(\frac{N_B}{N}\right) + \chi _{AB}\frac{N_AN_B}{N^2}$$

Vi kan igjen skrive om til molfraksjoner for å gjøre uttrykket enklere: $x =\frac{N_A}{N}$ og $1 - x = \frac{N_B}{N}$:

$$\frac{\Delta F_{sol}}{Nk_BT} = xlnx + (1 - x)ln(1 - x) + \chi _{AB}x(1 - x)$$

Vi har funnet et uttrykk for den frie energien vi vinner/taper på å blande de to stoffene! Modellen ble først beskrevet av JH Hildebrant i 1929 og kalles "Regular Solution Modell". I tillegg til entropien for ideelle løsninger vi fant til å begynne med, baseres den også på mean-field approksimasjonen av løsningsenergi. Modellen har noen begrensninger, særlig dersom molekylene det er snakk har en annerledes form (staver for eksempel) eller dersom A og B er svært forskjellige størrelser som i polymerløsninger.

## Kjemisk potensiale

Som introdusert i kapitellet om Makroskopisk Termodynamikk defineres [kjemisk potensiale](https://andreasdh.github.io/Bifrost/docs/Makroskopisk_termodynamikk_1.html#kjemisk-potensial) som endringen i fri energi når vi endrer sammensetningen til systemet vårt (N øker eller minskes): 

$$ \mu = \left(\frac{\partial F}{\partial N}\right)_{p,T}$$

For å finne det kjemiske potensialet til komponent A i lattice-modellen vår, holder vi $N_B$ konstant:

$$ \mu _A = \left[\frac{\partial}{\partial N_A}\left(\frac{F}{k_BT}\right)\right]_{N_B,T}$$

Derivasjonen som må gjøres er ganske kompleks fordi $N$ også avhenger av $N_A$ gjennom relasjonen $N = N_A + N_B$. Derfor velger vi her å skrive resultatet direkte, så kan du bryne deg på derivasjonen dersom du er spesielt interessert: 

$$ \mu _A = \left[\frac{\partial}{\partial N_A}\left(\frac{F}{k_BT}\right)\right]_{N_B,T} = ln\left(\frac{N_A}{N}\right) + 1 - \frac{N_A}{N} - \frac{N_B}{N} + \frac{zw_{AA}}{2k_BT} + \chi _{AB} \frac{(N_A + N_B)N_B - N_AN_B}{(N_A + N_B)^2}$$

Vi skriver igjen om ved hjelp av molfraksjoner: 

$$ \mu _A = \left[\frac{\partial}{\partial N_A}\left(\frac{F}{k_BT}\right)\right]_{N_B,T} = lnx_A - \frac{zw_{AA}}{2k_BT} + \chi _{AB}(1 - x_A)^2$$

På samme måte finner vi det kjemiske potensialet for komponent B: 

$$ \mu _B = \left[\frac{\partial}{\partial N_B}\left(\frac{F}{kT}\right)\right]_{N_A,T} = lnx_B - \frac{zw_{BB}}{2k_BT} + \chi _{AB}(1 - x_B)^2$$

Det kjemiske potensialet beskriver partiklenes tendens til å bevege seg fra en fase til en annen eller fra en tilstand til en annen. Partiklene drives av to tendenser: For det første har A-partikler en tendes til å forlate områder med høy konsentrasjon av A og bevege seg mot områder med lavere konsentrasjon av A for å øke blandingens entropi. For det andre vil A-partikler tiltrekkes områder eller faser de har høy kjemisk affinitet for. Mens den frie energien beskriver hvordan hele systemet tenderer mot likevekt, beskriver det kjemiske potensialet hvordan hver enkelt komponent i systemet tenderer mot likevekt. 

I animasjonen under kan du se hvordan A-molekylene vil spre seg utover gitteret slik at det kjemiske potensialet minimeres. Systemet tenderer mot likevekt! Dette kaller vi _diffusjon_ og dette er et viktig tema i kurset Fysikalsk Kjemi III. Hva beskriver utvekslingsparameteret $\chi _{AB}$ egentlig? Parameteret beskriver energien det vil koste oss å flytte én A-partikkel inn i en fase som består av ren B og vice versa. Vi kan beskrive utvekslingsprosessen slik: 

$$\frac{1}{2}z(AA) + \frac{1}{2}z(BB) \longrightarrow z(AB)$$

Hvis K er likevekstkonstanten til utvekslingen kan $\chi _{AB}$ beskrives slik: 

$$\chi _{AB} = -lnK$$

Jo større utvekslingsparameteret er, jo vanskeligere er det å blande de to stoffene. Når utvekslingsparameteret er lik 0, er det ingen favorisering av noen av tilstandene og blandingen skjer fritt. 

## Hvilke forenklinger har vi gjort ved å bruke Lattice-modellen? 

Latticemodellen vi har brukt for å beskrive løsninger er nyttig fordi den gir en enkel og intuitiv måte å forstå blandinger på, men vi huske på at den er forenklet. I slutten av kapittel 15 i læreboka, peker Dill og Bromberg på hva som ikke tas hensyn til: 

- Virkelige væsker er ikke gitter: Molekyler sitter ikke på faste plasser; de beveger seg kontinuerlig og har varierende avstander. Modellen ignorerer denne strukturelle kompleksiteten.

- Interaksjoner er mye mer detaljerte enn én kontaktenergi: I modellen har vi én energi for A–A, B–B og A–B‑kontakter. I virkeligheten finnes retning, polaritet, hydrogenbindinger, elektrostatikk og mange andre effekter.

- Entropien er sterkt forenklet: Vi tar kun med blandingsentropi (kombinatorikk) og utelater vibrasjons-, rotasjons- og konformasjonsentropi, samt volumendringer ved blanding.
