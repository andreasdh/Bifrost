# Løsninger og blandinger kan beskrives av lattice-modeller

## Introduksjon til modellen

Vi har et antall $N_A$ av en partikkel A og et antall $N_B$ av partikkel B. Sammen fyller partiklene opp et gitter med $N$ plasser i en lattice-modell slik at vi har:

$$N = N_A + N_B$$

Antall tilgjengelige mikrotilstander vil da kunne beskrives som: 

$$W = \frac{N!}{N_A!N_B!}$$

som beskrevet i [lenke]. Dette blir utgangspunktet vårt til å utlede størrelser som entropi, indre energi, fri energi og kjemisk potensial for blandinger med to komponenter. La oss begynne med entropien.

## Blandingsentropi

Med utgangspunkt i Boltzmanns definisjon av entropi har vi følgende uttrykk for blandingsentropien: 

$$\Delta S_{sol}  = klnW$$

Ved hjelp av [Stirlings approksimasjon]lenke, kan vi derimot uttrykket antall tilgjengelige mikrotilstander på en litt annen måte:

$$ln\left(\frac{N!}{N_A!N_B!}\right) = NlnN - N_AlnN_A - N_BlnN_B$$

Da får vi for blandingsentropien: 

$$\Delta S_{sol} = k(NlnN - N_AlnN_A - N_BlnN_B)$$

Vi skriver om ved å sette $N_A = N \times \left(\frac{N_A}{N}\right)$ og $N_B = N \times \left(\frac{N_B}{N}\right)$ får vi:

$$\Delta S_{sol} = k(NlnN - N \times \left(\frac{N_A}{N}\right)lnN_A - N \times \left(\frac{N_B}{N}\right)lnN_B)$$

Vi kan trekke N utenfor parentesen: 

$$\Delta S_{sol} = Nk(lnN - \left(\frac{N_A}{N}\right)lnN_A - \left(\frac{N_B}{N}\right)lnN_B)$$

Videre kan vi skrive om $lnN_A$ on $lnN_B$ slik (husk [lenke] logaritmeregler):

$$lnN_A = lnN + ln\frac{N_A}{N}$$

$$lnN_B = lnN + ln\frac{N_B}{N}$$

Vi setter dette inn i uttrykket vårt: 

$$\Delta S_{sol} = Nk(lnN - \left(\frac{N_A}{N}\right)\left(lnN + ln\frac{N_A}{N}\right) - \left(\frac{N_B}{N}\right)\left(lnN + ln\frac{N_B}{N}\right))$$

Vi kan samle logaritmene hver for seg slik: 

$$\Delta S_{sol} = Nk(lnN - \left(\frac{N_A}{N}\right)lnN - \left(\frac{N_B}{N}\right)lnN - \left(\frac{N_A}{N}\right)ln\left(\frac{N_A}{N}\right) - \left(\frac{N_B}{N}\right)ln\left(\frac{N_B}{N}\right))$$

$$\Delta S_{sol} = Nk(lnN - \left(\left(\frac{N_A}{N}\right) + \left(\frac{N_B}{N}\right)\right)lnN - \left(\frac{N_A}{N}\right)ln\left(\frac{N_A}{N}\right) - \left(\frac{N_B}{N}\right)ln\left(\frac{N_B}{N}\right))$$

Vi kan observere at $\frac{N_A}{N} + \frac{N_B}{N} = \frac{N_A + N_B}{N} = \frac{N}{N} = 1$ og dermed forsvinner $lnN$ helt fra likningen!

$$\Delta S_{sol} = Nk(lnN - lnN - \left(\frac{N_A}{N}\right)ln\left(\frac{N_A}{N}\right) - \left(\frac{N_B}{N}\right)ln\left(\frac{N_B}{N}\right))$$

$$\Delta S_{sol} = Nk( - \left(\frac{N_A}{N}\right)ln\left(\frac{N_A}{N}\right) - \left(\frac{N_B}{N}\right)ln\left(\frac{N_B}{N}\right))$$

$$\Delta S_{sol} = -Nk(\left(\frac{N_A}{N}\right)ln\left(\frac{N_A}{N}\right) + \left(\frac{N_B}{N}\right)ln\left(\frac{N_B}{N}\right))$$

Videre ønsker vi heller å uttrykke entropien ved hjelp av de _relative_ konsentrasjonene til A og B. Vi kan beskrive konsentrasjoner som _molfraksjoner_ slik: 

$$x = \frac{N_A}{N}$$

$$1-x = \frac{N_B}{N}$$

Ved å sette dette inn i formelen vi fant finner vi blandingsentropien for et to-komponent system: 

$$\frac{\Delta S_{sol}}{Nk} = -xlnx - (1-x)ln(1-x)$$

A og B vil blande seg fordi multiplisiteten av måter å arrangere A og B sammen i blandingen vil gi en større entropi enn de to rene systemene hver for seg. Dette er drivkraften bak blanding. Som du vil se i oppgaven under kan vi bruke denne blandingsentropien utledet fra et mikroskopisk lattice-modell system til å regne på blandingsentalpi i makroskopiske systemer hvor vi kjenner fraksjonene av komponentene A og B!


:::{admonition} Oppgave
:class: tip

**Oppgave 1: Mikroskopisk** 

Du har 7 A-partikler og 3 B-partikler som skal blandes sammen. Se for deg at du plasserer dem i et gitter med 10 ledige plasser. Finn først molfraksjonene av hver type partikkel før du bruker uttrykket vi fant til å beregne blandingsentropien. 

**Oppgave 2: Makroskopisk**

Du skal lage en blanding bestående av 30% metanol i vann. Finn først molfraksjonene og regn deretter ut blandingsentropien til blandingen. 

:::

```{dropdown} Klikk for å se svarene på oppgavene

**Oppgave 1: Mikroskopisk**

Vi finner først molfraksjonene:

$$N = N_A + N_B = 7 + 3 = 10$$

$$x = \frac{N_A}{N} = \frac{7}{10} = 0.7$$

$$1 - x = \frac{N_B}{N} = 0.3$$

Nå kan vi bruke formelen for blandingsentropi: 

$$\frac{\Delta S_{sol}}{Nk} = -0.7ln0.7 - (1-0.7)ln(1-0.7) = -0.7ln0.7 - 0.3ln0.3 \approx 1.2 \, J/molK$$

**Oppgave 2: Makroskopisk**

Vi finner molfraksjonene:

Vi skal ha en løsning med 30% etanol, så molfraksjonen av etanol $\frac{N_{etanol}}{N}$ må være lik $0.3$. Resten skal være vann, så molfraksjonen av vann må være $1 - 0.3 = 0.7$.

Vi får akkurat samme situasjon som i forrige oppgave!

$$\frac{\Delta S_{sol}}{Nk} = -0.7ln0.7 - (1-0.7)ln(1-0.7) = -0.7ln0.7 - 0.3ln0.3 \approx 1.2 \, J/molK$$ 
```

## Idelle løsninger 

I en ideell løsning utveksles ingen energi i løpet av blandingen og entropien kommer kun av blandingen selv som beskrevet i avsnittet over, altså ikke av volumendring, strukturering eller ordning i løsningen. Med andre ord er det ingen interaksjoner mellom de to stoffene som blandes slik som mellom gassmolekylene i ideell gass. Den frie energien til en ideell løsning er gitt ved: 

$$\Delta F_{sol} = -T\Delta S_{sol}$$

hvor $\Delta S_{sol}$ er hentet fra likningen vi fant i første avsnitt. Videre skal vi diskutere mer realistiske modeller. 

## Energien til en løsning

De fleste løsninger har en løsningsenergi som kommer fra interaksjoner mellom komponentene i løsningen. I lattice-modeller tar vi hensyn til interaksjonene en partikkel har med de nærmeste naboene sine. Det vil naboene over og under seg og på hver side av seg: Totalt fire interaksjoner. Dersom vi fortsetter med modellen vi har beskrevet for komponentene A og B kan vi se for oss tre ulike interaksjoner i blandingen: Dersom en A partikkel er nabo med en annen A partikkel, får vi en AA interaksjon. Dersom en B partikkel er nabo med en annen B partikkel, får vi en BB interaksjon. Og dersom en A partikkel og en B partikkel er naboer får vi en AB interaksjon. 

Den totale energien til systemet blir da summen over alle disse interaksjonene. Dersom vi ser for oss at $m_{AA}$, $m_{BB}$ og $m_{AB}$ beskriver antallet av hver interaksjon i systemet og $w_{AA}$, $w_{BB}$ og $w_{AB}$ de tilhørende kontaktenergiene, får vi: 

$$U = m_{AA}w_{AA} + m_{BB}w_{BB} + m_{AB}w_{AB}$$

Fordi det er vanskelig å vite hvor mange kontakter $m_{AA}$, $m_{BB}$ og $m_{AB}$ vi har, er vi interesserte i å uttrykke disse som funksjon av $N_A$ og $N_B$ som er tall vi kjenner. Vi sier at hver av plassene i lattice-modellen har z sider. I en 2D-modell (flat) vil en boks ha fire sider, z = 4, og i en 3D-modell, vil en boks ha seks flater, z = 6. Hver kontakt mellom to partikler innebærer to sider som vist på bildet tatt fra læreboka under: 

![En kontakt innebærer to sider](bilder/sites.png)

For partiklene av type A har vi totalt $zN_A$ sider og for partiklene av type B har vi totalt $zN_B$. Fordi systemet kun består av partiklene A og B, må hver side være assosiert med en kontakt:

$$zN_A = 2 \, m_{AA} + m_{AB}$$

Altså består sidene til partiklene A av kontaktene mellom to A partikler ($2 \, m_{AA}$) og kontakten mellom en A og en B partikkel (teller én gang, $m_{AB}$). Vi har tilsvarende for $zN_B$:

$$zN_B = 2 \, m_{BB} + m_{AB}$$

Nå kan vi skrive om disse uttrykkene for å finne $m_{AA}$ og $m_{BB}$:

$$m_{AA} = \frac{zN_A - m_{AB}}{2}$$

$$m_{BB} = \frac{zN_B - m_{AB}}{2}$$

Når vi setter dette inn i uttrykket for $U$ vi fant over står bare $m_{AB}$ igjen som ukjent: 

$$U = \frac{zN_A - m_{AB}}{2}w_{AA} + \frac{zN_B - m_{AB}}{2}w_{BB} + m_{AB}w_{AB}$$

Vi skriver om uttrykket vårt litt, først ved å utvide parentesene våre, så ved å omstrukturere: 

$$U = \frac{zN_Aw_{AA} - m_{AB}w_{AA}}{2} + \frac{zN_Bw_{BB} - m_{AB}w_{BB}}{2} + m_{AB}w_{AB}$$

Vi trekker $N_A$, $N_B$ og $m_{AB}$ utenfor parenteser: 

$$\left(\frac{zw_{AA}}{2}\right)N_A + \left(\frac{zw_{BB}}{2}\right)N_B + m_{AB}\left(w_{AB} - \left(\frac{w_{AA} + w_{BB}}{2}\right)\right)$$

### Mean-field approksimasjon

I prinsippet kan $m_{AB}$ variere mye mellom de ulike måtene og arrangere systemet på. I teorien måtte vi ha tatt hensyn til hver enkelt konfigurasjon og vektet dem ved hjelp av Boltzmann fordelingen, men dette føler til modeller som er tunge å regne på. Mean-field approksimasjon er derimot en enklere måte å se på systemet på, og lar oss fremdeles komme frem til sentrale konklusjoner om løsninger og blandinger. Approksimasjonen forutsetter at for alle mulige $N_A$ og $N_B$ vil partiklene blandes så tilfeldig og uniformt som mulig. Dette lar oss finne et estimat for $m_{AB}$. 

La oss tenke på en plass i gitteret ved siden av en A partikkel: Hva er sjansen for at en B partikkel er i denne posisjonen? I mean-field approksimasjonen er B partiklene tilfeldig spredt utover gitteret. Sannsynligheten $p_B$ for at en B partikkel sitter i hvilkens om helst gitt posisjon i gitteret er gitt ved fraksjonen av alle posisjonene som er opptatt av B partikler fordi ingen av posisjonene vil favoriseres over noen annen: 

$$p_B = \frac{N_B}{N} = x_B = 1 - x$$

Fordi vi har z naboplasser til enhver A-partikkel, vil det gjennomsnittlige antallet AB kontakter for en bestemt A partikkel være gitt ved:

$$zp_B = z\frac{N_B}{N} = z(1-x)$$

Siden dette gjelder for alle A-partikler kan vi gange dette med det totale antallet A $N_A$ og vi får da:

$$m_{AB} \approx z\frac{N_AN_B}{N} = z\left(\frac{N_A}{N} \times \frac{N_B}{N}\right) = zNx_Ax_B = zNx(x -1)$$

Vi kan nå sette dette nye resultatet inn i likningen vi fant for blandingsenergien:


$$\left(\frac{zw_{AA}}{2}\right)N_A + \left(\frac{zw_{BB}}{2}\right)N_B + z\left(\frac{N_A}{N} \times \frac{N_B}{N}\right)\left(w_{AB} - \left(\frac{w_{AA} + w_{BB}}{2}\right)\right)$$
