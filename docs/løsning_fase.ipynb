- b) 

 - Vi har én fase. Vi antar at løsningen vil være homogen siden vi forventer at $Na^{+}_{(aq)}$ og $Cl^-_{(aq)}$ ionene er spredt jevnt utover i løsningen. Dersom saltet ikke var løst hadde vi at hatt to faser. 
 - Vi har to faser. Is og vann har veldig forskjellig tetthet og danner hver sin fase. 
 - Vi har tre faser. Selv om både $CaCO_3$ og $CaO$ er i fast fase er det ingen grunn til å tro at de blandes til én fase der begge er spredt jevnt. Altså er $CaCO_3$, $CaO$ og $CO_2$ distinkte faser.

- c) c)	En superkritisk væske fås ved ekstreme betingelser når tettheten til væskeformen og gassformen til et stoff blir lik. Da viskes gresnen mellom de to formene ut. Dette skjer ved en kritisk temperatur og trykk. For vann er det ved $374 ^{\circ} C$ og $218 \, atm$. 

- d) Vi har uttrykket: 

$$ C = s - r$$

Der $C$ er antall komponenter, $s$ er antall specier og $r$ er antall relasjoner mellom dem. Gibbs faseregel gir oss antall frihetsgrader $F$: 

$$ F = C - P + 2 $$ 

Der $P$ er antall faser ("**p**hases"). For de tre første systemene i oppgaven over får vi:

1. Vi har fem specier $s$: $H_2O$, $H^+$, $OH^-$ i vannet og $Na^+$, $Cl-$ fra saltet. 
 
Vi har tre relasjoner $r$: 

$$ {H_2O}_{(l)} \leftrightharpoons H^+_{(aq)} + OH^-_{(aq)} $$

$$ NaCl_{(s)} \leftrightharpoons Na^+_{(aq)} + Cl^-_{(aq)} $$

Ladningsbalansen: Antall negative og positivt ladde specier må være balansert.

Satt sammen får vi antall komponenter: $C=5-3=2$.

Vi er klare for å bruke Gibbs faseregel: 

$$ F=2-1+2=3$$

Her er frihetsgradene sammensetning, temperatur og trykk. 

2. Vi har bare én specie $s$: Vann. 

Vi har ingen relasjoner mellom fasene, altså $r=0$ ¨

Satt sammen får vi antall komponenter: $C=1-0=1$.

Vi er klare for å bruke Gibbs faseregel: 

$$ F=1-2+2=1$$

Her er frihetsgraden temperatur eller trykk. 

3. Vi har tre specier $s$: $CaCO_3$, $CaO$ og $CO_2$

Det er kun én reaksjon som binder speciene sammen (reaksjonen i oppgaveteksten), så vi for $r = 1$. 

Satt sammen får vi antall komponenter: $C = 3 - 1 = 2$.

Vi er klare for å bruke Gibbs faseregel: 

$$ F=2-3+2=1$$

Her er frihetsgraden enten sammensetning, temperatur eller trykk. 

- e) Når $ΔH=TΔS$ er $ΔG=0$. Det betyr at faseovergangstemperaturen (T) er hvor $ΔH$ er lik størrelsen av $TΔS$. $ΔG$ er positiv under frysepunktet (i eksempelet med vann og is) og denne faseovergangen er ikke spontan. $ΔG$ er negativ over frysepunktet, og dette er en spontan prosess. Frysepunktet er punktet hvor det går fra å være en ikke-spontan prosess til en spontan prosess, og systemet kan gå frem og tilbake mellom de to fasene reversibelt. Det vil si at $∆G= 0$ ved likevekt. 

- f) Ved kokepunktet til vann vil temperaturen være konstant selv om man tilfører varme. Da vil all energi gå med på å endre fra en fase (væske) til en annen (gass). Selv når vi har dannet noe gass vil det fremdeles gjenstå vann i væskeform som må omdaness for at overgangen blir fullstendig, og varmen går nettopp til å gjøre om vann i væskefase til vanndamp. Varmekapasiteten er definert som:

$$C_{v}={\left(\frac{\partial U}{\partial T}\right)}_{V}$$

Når U øker imens T holdes konstant divergerer Cv mot uendelig. Dette kalles også for en _første ordens faseovergang_.

- g) Det finnes to typer faseoverganger som er relevante for kurset, _første_ og _andre ordens faseoverganger_. En første-ordens faseovergang er definert ved at den første deriverte til kjemisk potensial er diskontinuerlig. Det er kanskje kjent for deg at: 

$${\left(\frac{\partial \mathrm{\mu }}{\partial p}\right)}_{T}=V_{m}$$

og at  

$${\left(\frac{\partial \mathrm{\mu }}{\partial T}\right)}_{p}={-S}_{m}$$

Vi ser at Vm og Sm er diskontinuerlige i det midterste panelet som altså viser en første-ordens faseovergang. Fysisk forstår vi dette som at dersom fasen endres, får vi et plutselig hopp i molart volum og derfor i den molare entropien til stoffet. Tenk for eksempel på disse forskjellene i en væske versus en gassfase. Varmekapasiteten divergerer mot uendelig (se forklaring i oppgave f)). En andre-ordens faseovergang er definert ved at den andre-deriverte av kjemisk potensial er diskontinuerlig. Her er det ingen latent varme, entalpien er kontinuerlig og det er kun stigningstallet til entalpien som endrer seg. Dette fører til at varmekapasiteten er diskontinuerlig. En slik faseovergang er vist i det tredje panelet. Det øverste panelet viser ingen faseovergang.

**Oppgave 3**

- a) Kjemisk potensial er et mål på endring av fri energi på grunn av en endring i antall partikler. Kjemisk potensial er matematisk definert som den partielldervierte av den frie energien med hensyn på stoffmengde ved konstant volum ($C_V$)/trykk($C_p$) og temperatur: 

$$\mathrm{\mu }={\left(\frac{\partial G}{\partial N}\right)}_{V,T}$$

- b) 	For et én komponent system har vi 

$$µ=G_{m}=\frac{H}{N}-\frac{TS}{N}=H_{m}-TS_{m}$$

Grafen viser hvordan µ varierer med temperatur. Vi finner stigningstallet ved å derivere med hensyn på T:  

$${\left(\frac{\partial \mu }{\partial T}\right)}_{p}={\left(\frac{\partial G_{m}}{\partial T}\right)}_{p}=-S_{m}$$ 

Så jo høyere entropi, jo større negativt stigningstall vil vises i et plot av kjemisk potensial mot temperaturen. Siden entropien er høyere for en gass (energien er mer spredt utover, flere antall frihetsgrader, tilgjengelige energinivåer osv) er stigningstallet for gassfasen størst. Vi ser noe av den samme trenden ved overgangen til væske fra fast stoff, men denne er svakere enn for overgangen fra væske til damp. 

- c) Fra fundamentallikningen for Gibbs fri energi vet vi at:

$$dG=Vdp-SdT={\left(\frac{\partial G}{\partial p}\right)}_{T}dp+{\left(\frac{\partial G}{\partial T}\right)}_{p}dT$$

Altså har vi:
$${\left(\frac{\partial G}{\partial p}\right)}_{T}=V slik at {{\left(\frac{\partial G_{m}}{\partial p}\right)}_{T}=\left(\frac{\partial \mu }{\partial p}\right)}_{T}=\ V_{m}$$

Gass har størst molart volum. Så stigningstallet ville vært størst for gassfasen. Men i dette tilfelle ville stigningen vært motsatt av forrige eksempel da 

$${\left(\frac{\partial \mu }{\partial p}\right)}_{T}=\ V_{m}$$

Væsker har stort sett et større molart volum enn fast fase (bortsett fra vann), så stigningstallet for denne ville i de fleste tilfeller vært nest størst. Så grafen ville sett noe liknende ut som vist under: 

![Kjemisk potensial vs. trykk](/bilder/kjemisk_potensial_trykk_faser.png)

- d) For én-komponent stoffer har vi som sagt: 

$$\mu =G_{m}=V_{m}dp-S_{m}dT\$$

Avhengigheten av μ av trykk ved konstant temperatur er altså: 

$${\left(\frac{\partial G_{m}}{\partial p}\right)}_{T}={\left(\frac{\partial \mu }{\partial p}\right)}_{T}=\ V_{m}$$

Avhengigheten av μ av temperatur ved konstant trykk er: 

$${\left(\frac{\partial G_{m}}{\partial T}\right)}_{p}={\left(\frac{\partial \mu }{\partial T}\right)}_{p}=\ -S_{m}$$

- e) Figuren viser kjemisk potensial $\mu$ som funksjon av temperatur $T$ for fast og væske ved lavt og høyt trykk for to forskjellige stoffer. Helningen til hver kurve er gitt ved

$$\frac{d\mu}{dT} = -S$$

der $S$ er molarentropien. Væske har høyere entropi enn fast, og derfor blir kurven for væske brattere nedover enn kurven for fast.  

Smeltepunktet ved et gitt trykk er definert ved den temperaturen $T_f$ der fasene er i likevekt:

$$\mu_\text{fast}(T_f) = \mu_\text{væske}(T_f)$$

- i. Stoff (a) – «vanlig» oppførsel

I figur (a) ser vi at ved lavt trykk har vi et smeltepunkt $T_f$ der kurvene for fast og væske krysser. Ved høyt trykk ligger kurven for fast lavere enn kurven for væske (fast har lavere $\mu$ enn væske), og de to kurvene skjærer hverandre ved en høyere temperatur $T_f'$ slik at $T_f' > T_f$. Dette kan forklares ved at fastfasen har **mindre molart volum enn væsken**:

$$\Delta V_\text{fusjon} = V_\text{væske} - V_\text{fast} > 0$$

Endringen i kjemisk potensial med trykk kan skrives som:

$$d\mu \approx V\,dp$$

Når trykket øker, vil fasen med minst volum (her fast) få størst reduksjon i $\mu$ og dermed bli mest stabil. Økt trykk favoriserer derfor fastfasen, og man trenger høyere temperatur for at væsken skal bli like stabil som fast → smeltepunktet øker med trykk. Dette er typisk oppførsel for «vanlige» stoffer, som mange metaller og salter, f.eks. aluminium eller natriumklorid.

- ii. Stoff (b) – «anomal» oppførsel (vann-type)

I figur (b) har vi igjen et smeltepunkt $T_f$ ved lavt trykk, der $\mu_\text{fast} = \mu_\text{væske}$. Ved høyt trykk ser vi at kurven for væske nå ligger lavere enn kurven for fast, slik at væsken er mer stabil ved høyt trykk. Skjæringspunktet (smeltepunktet) flyttes til en lavere temperatur $T_f'$ slik at $T_f' < T_f$. Dette betyr at fastfasen har **større molart volum enn væsken**:

$$\Delta V_\text{fusjon} = V_\text{væske} - V_\text{fast} < 0$$

Siden $d\mu \approx V\,dp$, vil økt trykk senke det kjemiske potensialet mest for fasen med minst volum, altså væsken. Økende trykk favoriserer derfor væskefasen, og det blir «lettere» å smelte — smeltepunktet synker når trykket øker. Denne oppførselen er karakteristisk for vann, der is har lavere tetthet enn flytende vann (is har større volum enn vann). Derfor synker isens smeltepunkt med økende trykk.

Her er en kode du kan bruke for å konstruere fasediagrammet: 

```python
import matplotlib.pyplot as plt

t = [20, 25, 30, 35, 40, 45, 50, 55, 60, 65, 68.8]
vann = [8.40, 8.71, 8.92, 9.34, 9.78, 10.62, 12.08, 13.88, 17.10, 22.26, 35.9]
fenol = [72.24, 71.38, 69.95, 68.28, 66.81, 65.02, 62.83, 60.18, 56.10, 49.34, 35.9]

plt.plot(vann, t, 'o-', label='Vannlag')
plt.plot(fenol, t, 's-', label='Fenollag')
plt.fill_betweenx(t, vann, fenol, alpha=0.15, color='gray')
plt.xlabel('Massprosent fenol (%)')
plt.ylabel('Temperatur (°C)')
plt.title('Fasediagram: fenol–vann')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

- a) en fase med 8,9 % fenol og en fase med 70 % fenol

- b) Den kritiske temperaturen for systemet er $69 °C$. Ved temperaturer over $69 °C$ vil væskene alltid være blandbare. Altså foreligger én fase.

- c) c)	Når en blanding med 20 masse-% fenol kjøles ned fra $70 ^{\circ} C$  vil det dannes to faser når temperaturen når $63 ^{\circ}C$ (krysser over fra énfase- til to-fase området). Ved dette punktet vil $∆G_{mix}$ gå fra å være negativ til å være positiv. Det dannes en fase med 20 % fenol og en fase med ca 53 % fenol.

- d) Her kan vi bruke vekststangsregelen: 

$$\frac{n_{\alpha }}{n_{\beta }}=\frac{l_{\beta }}{l_{\alpha }}\$$

der α er vannlaget og β er fenollaget.

$$\frac{n_{\alpha }}{n_{\beta }}=\frac{0,70-0,30}{0,30-0,089}=1,9$$

Det er altså 1,9 ganger mer av vannlaget ved denne sammensetningen og temperaturen enn fenollag. Videre kan vi skrive:

$$n_{\alpha }+n_{\beta }=1$$ 

Siden det bare er de to fasene må summen blir 1 (100%). 

Vi har fra tidligere at: 

$$n_{\alpha }=1,9\cdot n_{\beta }$$

Setter inn i likningen hvor de summeres til 1: 

$$1,9n_{\beta }+n_{\beta }=1\ n_{\beta }=\frac{1}{2,9}=0,34=34\%$$

Det er altså 34% fenollag og (100%-34%) og 66% vannlag ved denne sammensetningen og temperaturen.
