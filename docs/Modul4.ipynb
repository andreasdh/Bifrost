## 4.1 Molekylære interaksjoner: Hvorfor kondenserer noe i det hele tatt?

Avkjøl argongass nok, og den blir til en væske. Det virker kanskje helt naturlig; Mange stoffer kan jo gå fra gass til væske. Men hvis vi stopper opp og tenker etter, er det egentlig ganske merkelig. Argon består av elektrisk nøytrale atomer uten permanent dipolmoment. Ved første øyekast finnes det derfor ingen åpenbar grunn til at atomene skulle tiltrekke hverandre. Likevel gjør de det, sterkt nok til at gassen kan kondensere. 

Denne observasjonen peker på noe nytt. Til nå har du lært om kovalente og ioniske bindinger, bindinger som holder atomer sammen i molekyler og krystaller. I denne modulen skal vi se på en annen type interaksjoner: de svakere kreftene som virker mellom separate atomer og molekyler. Selv om disse kreftene er mye svakere enn kjemiske bindinger, er de avgjørende for at væsker og mange faste stoffer i det hele tatt kan eksistere.

Hvordan kan vi beskrive slike interaksjoner? I stedeet for å starte med selve kraften, er det mer nyttig å beskrive den potensielle energien mellom to partikler. Denne energien avhenger av avstanden mellom dem og skrives som u(r), der (r) er avstanden mellom partiklene. Når partiklene beveger seg nærmere hverandre eller lenger fra hverandre, endrer den potensielle energien seg. Noen avstander er mer energetisk gusntige enn andre. 

Men den potensielle energien er bare én del av historien. Partiklene er hele tiden i bevegelse, hvor raskt de beveger seg bestemmes av temperaturen. Denne bevegelsen beskrives av den kinetiske energien. Ved høy temperatur er bevegelsen stor, og partiklene kan lettere bevege seg bort fra hverandre. Ved lav temperatur er bevegelsen mindre, slik at de svake tiltrekkkende kreftene får større betydning. 

Om et stoff opprer som en gass, en væske eller et fast stoff, bestemmesderfor av samspillet mellom potensiell og kinetisk energi. Den potensielle energien favoriserer bestemte avstander mellom partiklene, mens den kinetiske energien motvirker at de blir værende der. I dette modulen skal vi utvikle en enkel modell for denne balansen. Den skal gi oss et rammeverk for å forstå hvorfor stoffer kondenserer, hvorfor molekyler foretrekker bestemte avstander fra hverandre, og hvordan dette kan beskrives med blant annet Lennard-Jones potensialet.

## 4.2 - Fra coulombs lov til parpotensial

Fra tidligere emner kjenner vi Coulombs lov, som beskriver den elektriske kraften mellom to ladninger. Kraften øker når ladningene blir større, og når avstanden mellom dem blir mindre. Denne sammenhengen forklarer blant annet hvorfor små ioner danner sterkere ionebindinger enn store ioner, og den kan også brukes til å forstå dipol-dipol interaksjoner og hydrogenbindinger, der tiltrekningen skyldes delvis ladninger. 

$$F = \frac{q_1 q_2}{4\pi\varepsilon_0 r^2}$$

Coulombs lov kan imidlertid ikke forklare alle intermolekylære vekselsvirkninger. Mange molekyler er elektrisk nøytrale og har ikke permanente dipoler, men tiltrekker likevel hverandre. For eksempel kan nitrogen $N_2$ og edelgasser kondensere til væske ved lave temperaturer. 

For å beskrive slike vekselsvirkninger trenger vi en mer generell modell. I stedet for å beskrive kraften mellom to punktladninger, beskriver vi den potensielle energien mellom to partikler som en funksjon av avstanden mellom dem. Denne beskrivelsen kalles et parpotensial, og danner grunnlaget for å forstå intermolekylære interaksjoner. 

### Parpotensial u(r)

Et parpotensial , u(r), beskriver den potensielle energien mellom to partikler som en funksjon av avstanden r:

![ur](bilder/ur_graf.png)

- Når partiklene er langt fra hverandre, er vekselvirkningen neglisjerbar, og $u(r) \to 0$.
- Når de kommer nærmere, dominerer de tiltrekkende kreftene, slik at den potensielle energien avtar. Dersom partiklene presses enda tettere sammen, øker energien raskt fordi frastøtende krefter blir dominerende.

Denne frastøtningen skyldes både elektrostatisk frastøtning mellom like ladninger og at elektronskyene begynner å overlappe. Ved slik overlapp blir Pauli-prinsippet viktig: elektroner kan ikke befinne seg i samme kvantetilstand, noe som gir opphav til en sterk, korttrekkende frastøtning. 

Ved en bestemt avstand, $r^*$, balanserer de tiltrekkende og frastøtende bidragene hverandre. Her har $u(r)$ et minimum, nettokraften mellom partiklene er null, og $r^*$ tilsvarer den gjennomsnittlige bindingslengden. 

Den generelle formen på parpotensialet er den samme for alle typer vekselvirkninger; Fra sterke kovalente bindinger til svake intermolekylære krefter. Det som skiller dem, er det matematiske uttrykket for $u(r)$ og hvor dyp potensialbrønnen er. 



:::{admonition} Spørsmål: Hva tror du skaper den tiltrekkende delen av parpotensialet?
:class:important

For ioner som $Na^+$ og $Cl^-$ er svaret intuitivt. Coulombs love forklarer tiltrekningen direkte. Men hva med to nøytrale molekyler, som verken har postivt eller negativ nettoladning, hvordan kan slike molekyler likevel tiltrekke hverandre?
:::

### Ladningsfordeling og dipolmomenter

Vi har sett av nøytrale molekyler kan tiltrekke hverandre, selv om de ikke har en netto ladning. For å forstå hvorfor, må vi se nærmere på hvordan ladningen er fordelt i molekylet. 

Et molekyl kan være elektrisk nøytralt samtidig som den positive og negative ladningen er ujevnt fordelt. Denne romlige ladningsfordelingen bestemmes av både molekylets geometri og elektronfordelingen, og har stor betydning for hvilke intermolekylære vekselsvirkninger molekylet kan danne og hvor sterke disse blir. 


![multipol](bilder/multipol.png)

Generelt kan enhver ladningsfordeling beskrives ved hjelp av en multipolutvikling. Den første termen er en monopol, som beskriver molekylets nettoladning. Deretter følger dipol, kvadrupol, oktupol og høyere multipolmomenter, som beskriver stadig mer detaljerte trekk ved ladningsfordelingen. Etter hvert som høyere multipolmomenter dominerer, blir vekselsvirkningene både svakere og mer korttrekkende. 


:::{admonition} Tips
:class: important

I dette emnet skal vi hovedsakelig fokusere på dipolmomentet, siden dette er den viktigste egenskapen for å forstå de fleste intermolekylære vekselsvirkningene. 

:::

### Parmanente og induserte dipolmomenter

Et dipolmoment beskrives hvor stor ladningsseparasjonene er i et molekyl. Vi skiller mellom to hovedtyper:

- Et permanent dipolmoment finnes i molekyler der den positive og negative ladningen er forskjøvet fra hverandre også uten påvirkning fra et ytre elektrisk felt. Dette skyldes molekylets geometri og elektronfordeling. 

- Et indusert dipolmoment oppstår når et elektrisk felt påvirker et atom eller molekyl og forskyver elektronskyen. Hvor lett dette skjer, beskrives av molekylets **polariserbarhet**. Molekyler med høy polariserbarhet får lettere indusert et dipolmoment enn molekyler med lav polariserbarhet. 

![krefter](bilder/krefter.svg)


Et molekyl kan ha et permanent dipolmoment, et indusert dipolmoment, eller begge deler. Dette danner grunnlaget for de tre viktigste van der Waals vekselsvirkningnene: permanente dipoler gir opphav til Keesom-krefter, permanente dipoler som induserer dipoler gir Debye-krefter, og fluktuerende, induserte dipoler gir London-dispersjonskrefter. 

I gasser og væsker roterer molekylene kontinuerlig. Derfor må vekselsvirkningene beskrives som et gjennomsnitt over mange orienteringer. Denne orienteringsmiddelverdien gjør at de attraktive vekselsvirkningene blir svakere og mer korttrekkende enn de ville vært mellom molekyler med faste orienteringer. 

## 4.3 Lennard-Jones-potensialet


Å følge alle detaljene i elektronstrukturen til to atomer er upraktisk — og heldigvis unødvendig. I stedet bruker vi effektive modeller: enkle matematiske uttrykk som fanger den samlede effekten av de underliggende fysiske bidragene.

Lennard-Jones-potensialet er den mest brukte av disse. Det gir en enkel, men realistisk beskrivelse av hvordan to nøytrale atomer påvirker hverandre som funksjon av avstanden mellom dem.


Interaksjonen har alltid to motsatte sider: en attraktiv del som trekker atomene sammen på lengre avstand, og en frastøtende del som dominerer kraftig når de presses tett sammen.

Det viktige er ikke detaljene i bidragene, men at de virker over ulik avstand: tiltrekningen strekker seg lenger, mens frastøtningen vokser ekstremt raskt når avstanden blir liten. Kombinert gir dette et potensial med en karakteristisk form: null ved store avstander, et minimum et stykke ut, og en bratt stigning når atomene kommer for nær.


$$U(r) = 4\varepsilon \left[ \left(\frac{\sigma}{r}\right)^{12} - \left(\frac{\sigma}{r}\right)^{6} \right]$$

Denne formelen er ikke noe fysisk «fundament» — den er en kompakt måte å kode inn den typiske formen på intermolekylær interaksjon.

### Hva parameterne forteller oss

Jo dypere minimum, desto sterkere binding. Denne dybden styres av ε, som gir et direkte mål på hvor «sterk» interaksjonen er.

σ setter lengdeskalaen: den bestemmer hvor kurven krysser null, og kan tolkes som en effektiv atomstørrelse. Likevektsavstanden ligger litt lenger ut, gitt ved

$$r^* = 2^{1/6}\sigma$$

## 4.4 - Fra parpotensial til van der Waals gas-likningen


I de foregående avsnittene har vi beskrevet vekselvirkningen mellom to partikler ved hjelp av parpotensialet $u(r)$. Vi har sett at denne vekselvirkningen består av to konkurrerende bidrag: en sterkt frastøtende del ved korte avstander og en svakere attraktiv del ved større avstander. For mange nøytrale atomer og molekyler kan denne vekselvirkningen beskrives godt ved hjelp av Lennard-Jones potensialet. 

En virkelig gass består imidlertid ikke av bare to molekyler, men av et enormt antall partikler som alle vekselvirker med hverandre samtidig. Det naturlige spørsmålet blir derfor hvordan de intermolekylære kreftene mellom enkeltmolekyler påvirker makroskopiske størrelser som trykk, volum og temperatur. 

En av de emkleste modellene som kobler disse to beskrivelsesnivåene sammen er **van der Waals-likningen**:

$$\left(P+\frac{a}{V_m^2}\right)(V_m-b)=RT$$

Van der Waals-likningen kan betraktes som en forbedring av den ideelle gassloven. Der den ideelle gassloven antar at molekylene er punktpartikler uten innbyrdes vekselvirkning, tar van der Waals-likningen hensyn til to viktige egenskaper ved virkelige gasser: Molekylene opptar plass, og de tiltrekker hverandre.

Disse to effektene beskrives av de to korreksjonsparametrene i likningen. Parameteren b korrigerer for at molekylene har en endelig størrelse, slik at det tilgjengelige volumet blir mindre enn beholdervolumet. Parameteren a korrigerer for de attraktive intermolekylære kreftene, som gjør at trykket blir lavere enn det den ideelle gassloven predikerer.

For å forstå hvorfor disse korreksjonene opptrer, er det nyttig å se nærmere på hva som faktisk bestemmer trykket i en gass.


### Leddet b: Molekylenes endelige størrelse

Den ideelle gassloven antar at molekylene er punktpartikler, men virkelige molekyler har en endelig utstrekning. To molekyler kan derfor ikke oppta samme rom samtidig, noe som allerede kommer til uttrykk i Lennard–Jones-potensialet gjennom den kraftige frastøtningen ved små avstander.

Hvert molekyl ekskluderer dermed et lite område rundt seg som andre molekyler ikke kan trenge inn i. Det tilgjengelige volumet blir derfor mindre enn det totale beholdervolumet og skrives som $V_m-b$. Parameteren b kan derfor tolkes som et mål på molekylenes effektive størrelse og er nært knyttet til den repulsive delen av parpotensialet.

### Leddet a: Attraktive vekselvirkninger

Den andre korreksjonen skyldes at molekylene tiltrekker hverandre gjennom van der Waals-krefter. Tiltrekningen senker den potensielle energien til systemet og reduserer samtidig trykket, fordi molekylene trekkes svakt mot hverandre i stedet for å bevege seg fritt mot beholderveggene.

Styrken på denne effekten samles i parameteren a. Jo sterkere de intermolekylære tiltrekningene er, desto større blir a, og desto større blir korreksjonen til trykket.

Vi kan knytte denne parameteren direkte til parpotensialet. Dersom vi betrakter ett molekyl omgitt av alle de andre molekylene i gassen, kan den gjennomsnittlige potensielle energien beregnes ved å summere bidragene fra alle nabomolekylene,

$$U=\int_0^\infty u(r)\,\rho\,4\pi r^2\,dr,$$

der $\rho$ er molekyltettheten. For et Lennard–Jones-lignende potensial viser denne integrasjonen at den totale attraktive energien er proporsjonal med $-N^2/V$. Konstanten a kan derfor oppfattes som et mål på den samlede styrken til den attraktive delen av parpotensialet.


:::{admonition} To bidrag til trykket
:class: important

Det makroskopiske trykket i en gass kan forstås som resultatet av to konkurrerende effekter.

**Entropibidraget** skyldes molekylenes termiske bevegelse. Molekylene beveger seg tilfeldig og har en tendens til å spre seg utover for å maksimere antall tilgjengelige mikrotilstander. Dette er den eneste mekanismen som bidrar til trykket i en ideell gass, og leder til den velkjente sammenhengen:

 $$P=\frac{Nk_BT}{V}$$




**Energibidraget** skyldes de attraktive intermolekylære kreftene. Når molekylene tiltrekker hverandre, reduseres den potensielle energien dersom de befinner seg nær hverandre. Denne tiltrekningen trekker molekylene svakt bort fra beholderveggene og gjør at kollisjonene med veggen blir mindre kraftige. Resultatet er et lavere trykk enn for en ideell gass.

Van der Waals-likningen beskriver begge disse effektene ved hjelp av de to korreksjonsleddene $a$ og $b$.
:::

### Gittermodellen – en enkel fysisk forklaring (skal fix)

Selv om sammenhengen ovenfor kan utledes direkte fra parpotensialet, finnes det en enklere måte å forstå van der Waals-likningen på.

Vi tenker oss at volumet deles inn i et stort antall små celler, der hver celle omtrent har størrelsen til ett molekyl. Hver celle kan enten være tom eller inneholde ett molekyl. På denne måten blir det umiddelbart klart at molekylene opptar plass, og at ikke hele volumet er tilgjengelig for bevegelse. Dette er den fysiske opprinnelsen til parameteren b.

Deretter lar vi hvert molekyl vekselvirke med sine nærmeste naboer gjennom den attraktive delen av parpotensialet. Når disse bidragene summeres over hele systemet, får vi en gjennomsnittlig attraktiv energi som er proporsjonal med -N^2/V, og dette gir parameteren a.

Gittermodellen viser dermed at de to korreksjonene i van der Waals-likningen har ulik fysisk opprinnelse: b beskriver at molekylene har en endelig størrelse og reduserer det tilgjengelige volumet, mens a beskriver den gjennomsnittlige attraktive vekselvirkningen mellom molekylene.
