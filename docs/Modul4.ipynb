## 4.1 Molekylære interaksjoner: Hvorfor kondenserer noe i det hele tatt?

Avkjøl argongass nok, og den blir til en væske. Det virker kanskje helt naturlig; Mange stoffer kan jo gå fra gass til væske. Men hvis vi stopper opp og tenker etter, er det egentlig ganske merkelig. Argon består av elektrisk nøytrale atomer uten permanent dipolmoment. Ved første øyekast finnes det derfor ingen åpenbar grunn til at atomene skulle tiltrekke hverandre. Likevel gjør de det, sterkt nok til at gassen kan kondensere. 

Denne observasjonen peker på noe nytt. Til nå har du lært om kovalente og ioniske bindinger, bindinger som holder atomer sammen i molekyler og krystaller. I denne modulen skal vi se på en annen type interaksjoner: de svakere kreftene som virker mellom separate atomer og molekyler. Selv om disse kreftene er mye svakere enn kjemiske bindinger, er de avgjørende for at væsker og mange faste stoffer i det hele tatt kan eksistere.

Hvordan kan vi beskrive slike interaksjoner? I stedeet for å starte med selve kraften, er det mer nyttig å beskrive den potensielle energien mellom to partikler. Denne energien avhenger av avstanden mellom dem og skrives som u(r), der (r) er avstanden mellom partiklene. Når partiklene beveger seg nærmere hverandre eller lenger fra hverandre, endrer den potensielle energien seg. Noen avstander er mer energetisk gusntige enn andre. 

Men den potensielle energien er bare én del av historien. Partiklene er hele tiden i bevegelse, hvor raskt de beveger seg bestemmes av temperaturen. Denne bevegelsen beskrives av den kinetiske energien. Ved høy temperatur er bevegelsen stor, og partiklene kan lettere bevege seg bort fra hverandre. Ved lav temperatur er bevegelsen mindre, slik at de svake tiltrekkkende kreftene får større betydning. 

Om et stoff opprer som en gass, en væske eller et fast stoff, bestemmesderfor av samspillet mellom potensiell og kinetisk energi. Den potensielle energien favoriserer bestemte avstander mellom partiklene, mens den kinetiske energien motvirker at de blir værende der. Den skal gi oss et rammeverk for å forstå hvorfor stoffer kondenserer, hvorfor molekyler foretrekker bestemte avstander fra hverandre, og hvordan dette kan beskrives med blant annet Lennard-Jones potensialet.

## 4.2 - Fra coulombs lov til parpotensial

Fra tidligere emner kjenner vi Coulombs lov, som beskriver den elektriske kraften mellom to ladninger. Kraften øker når ladningene blir større, og når avstanden mellom dem blir mindre. Denne sammenhengen forklarer blant annet hvorfor små ioner danner sterkere ionebindinger enn store ioner, og den kan også brukes til å forstå dipol-dipol interaksjoner og hydrogenbindinger, der tiltrekningen skyldes delvis ladninger. 

Coulombs lov kan imidlertid ikke forklare alle intermolekylære vekselsvirkninger. Mange molekyler er elektrisk nøytrale og har ikke permanente dipoler, men tiltrekker likevel hverandre. For eksempel kan nitrogen $N_2$ og edelgasser kondensere til væske ved lave temperaturer. 

For å beskrive slike vekselsvirkninger trenger vi en mer generell modell. I stedet for å beskrive kraften mellom to punktladninger, beskriver vi den potensielle energien mellom to partikler som en funksjon av avstanden mellom dem. Denne beskrivelsen kalles et parpotensial, og danner grunnlaget for å forstå intermolekylære interaksjoner. 

### Parpotensial u(r)

Et parpotensial , u(r), beskriver den potensielle energien mellom to partikler som en funksjon av avstanden r:

```{figure} ./docs/bilder/ur_graf.png
:name: ur-figur
:alt: Graf av parpotensialet u(r)
:width: 70%
```

- Når partiklene er langt fra hverandre, er vekselvirkningen neglisjerbar, og $u(r) \to 0$.
- Når de kommer nærmere, dominerer de tiltrekkende kreftene, slik at den potensielle energien avtar. Dersom partiklene presses enda tettere sammen, øker energien raskt fordi frastøtende krefter blir dominerende.

Denne frastøtningen skyldes både elektrostatisk frastøtning mellom like ladninger og at elektronskyene begynner å overlappe. Ved slik overlapp blir Pauli-prinsippet viktig: elektroner kan ikke befinne seg i samme kvantetilstand, noe som gir opphav til en sterk, korttrekkende frastøtning. 

Ved en bestemt avstand, $r^*$, balanserer de tiltrekkende og frastøtende bidragene hverandre. Her har $u(r)$ et minimum, nettokraften mellom partiklene er null, og $r^*$ tilsvarer den gjennomsnittlige bindingslengden. 

Den generelle formen på parpotensialet er den samme for alle typer vekselvirkninger; Fra sterke kovalente bindinger til svake intermolekylære krefter. Det som skiller dem, er det matematiske uttrykket for $u(r)$ og hvor dyp potensialbrønnen er.

### Ladningsfordeling og dipolmomenter

Vi har sett av nøytrale molekyler kan tiltrekke hverandre, selv om de ikke har en netto ladning. For å forstå hvorfor, må vi se nærmere på hvordan ladningen er fordelt i molekylet. 

Et molekyl kan være elektrisk nøytralt samtidig som den positive og negative ladningen er ujevnt fordelt. Denne romlige ladningsfordelingen bestemmes av både molekylets geometri og elektronfordelingen, og har stor betydning for hvilke intermolekylære vekselsvirkninger molekylet kan danne og hvor sterke disse blir. 

```{figure} ./docs/bilder/multipol.png
:name: pol-figur
:alt: Multipolmoment illustrasjoner
:width: 70%
```

![pol-figure](multipol.png)

Generelt kan enhver ladningsfordeling beskrives ved hjelp av en multipolutvikling. Den første termen er en monopol, som beskriver molekylets nettoladning. Deretter følger dipol, kvadrupol, oktupol og høyere multipolmomenter, som beskriver stadig mer detaljerte trekk ved ladningsfordelingen. Etter hvert som høyere multipolmomenter dominerer, blir vekselsvirkningene både svakere og mer korttrekkende. 

I dette emnet skal vi hovedsakelig fokusere på dipolmomentet, siden dette er den viktigste egenskapen for å forstå de fleste intermolekylære vekselsvirkningene. 

### Parmanente og induserte dipolmomenter

Et dipolmoment beskrives hvor stor ladningsseparasjonene er i et molekyl. Vi skiller mellom to hovedtyper:

- Et permanent dipolmoment finnes i molekyler der den positive og negative ladningen er forskjøvet fra hverandre også uten påvirkning fra et ytre elektrisk felt. Dette skyldes molekylets geometri og elektronfordeling. 

- Et indusert dipolmoment oppstår når et elektrisk felt påvirker et atom eller molekyl og forskyver elektronskyen. Hvor lett dette skjer, beskrives av molekylets **polariserbarhet**. Molekyler med høy polariserbarhet får lettere indusert et dipolmoment enn molekyler med lav polariserbarhet. 

Et molekyl kan ha et permanent dipolmoment, et indusert dipolmoment, eller begge deler. Dette danner grunnlaget for de tre viktigste van der Waals vekselsvirkningnene: permanente dipoler gir opphav til Keesom-krefter, permanente dipoler som induserer dipoler gir Debye-krefter, og fluktuerende, induserte dipoler gir London-dispersjonskrefter. 

I gasser og væsker roterer molekylene kontinuerlig. Derfor må vekselsvirkningene beskrives som et gjennomsnitt over mange orienteringer. Denne orienteringsmiddelverdien gjør at de attraktive vekselsvirkningene blir svakere og mer korttrekkende enn de ville vært mellom molekyler med faste orienteringer.

## 4.3 - Lennard-Jones potensialet

Når vi skal beskrive intermolekylære krefter mellom nøytrale atomer, er det ikke praktisk å følge alle detaljer i den mikroskopiske elektronstrukturen. I stedet bruker vi effektive modeller som fanger den samlede effekten av de underliggende fysikalske bidragene. 

Lennard-Jones potensialet er en slik modell. Det er konstruert for å gi en enkel, men realistisk beskrivelse av hvordan to nøytrale atomer påvirker hverandre som funksjon av avstand mellom dem. 

Kjernen i modellen er at interaksjonen alltid har to motsatte tendenser. På den ene siden finnes en attraktiv del som gjør at atomene trekkes mot hverandre når de er relativt langt fra hverandre. På den andre siden finnes en sterkt frastøtende del som dominerer når atomene presses tett sammen. Det viktige er ikke detaljene i disse bidragene, men at de har svært ulik avstandsavhengighet: tiltrekningen virker over litt lengre avstander, mens frastøtningen vokser ekstremt raskt når avstanden blir liten.

Når disse to effektene kombineres, får vi et potensial som alltid har samme karakteristiske form: det går mot null ved store avstander, det får et minimum ved en bestemt mellomliggende avstand, og det stiger bratt når partiklene kommer for nær hverandre.

Dette er akkurat det Lennard–Jones-potensialet fanger i én funksjon:
(Put funksjonen her)

Det som er viktig å forstå, er at denne formelen ikke er “fundamentet” i seg selv, men en kompakt måte å kode inn den typiske formen på intermolekylær interaksjon.

Hvis vi ser på kurven til U(r), kan vi lese av hele fysikken direkte. Når atomene er langt fra hverandre, er U(r) nær null, og de påvirker hverandre nesten ikke. Når de kommer nærmere, begynner energien å synke, noe som betyr at systemet blir mer stabilt – atomene trekkes mot hverandre. Hvis de kommer enda nærmere, snur utviklingen brått, og energien øker kraftig fordi overlapping av elektronskyer gir sterk frastøtning.

Mellom disse to områdene finner vi et punkt der energien er lavest. Dette punktet er spesielt viktig fordi det representerer den mest stabile konfigurasjonen. Her balanserer tiltrekning og frastøtning hverandre, og netto kraft er null. Dette er ikke et tilfeldig punkt, men den naturlige likevektsavstanden mellom atomene.

Formen på kurven gjør det også mulig å lese av stabilitet direkte: jo dypere minimum, desto sterkere binding mellom atomene. Denne dybden styres av parameteren \varepsilon, som dermed gir et direkte mål på hvor “sterk” interaksjonen er.

Den andre parameteren, \sigma, setter lengdeskalaen i problemet. Den bestemmer hvor kurven krysser null, og kan tolkes som en effektiv størrelse på atomene i modellen. Likevektsavstanden ligger litt lenger ut enn dette punktet og er gitt ved r^* = 2^{1/6}\sigma.

Det som gjør Lennard–Jones-modellen spesielt nyttig, er at den oversetter komplisert kvante- og elektrodynamisk fysikk til en enkel energikurve som direkte kan brukes i beregninger. I molekylær simulering betyr dette at atomene ikke trenger å beskrives eksplisitt med elektroner og bølgefunksjoner – de “følger” bare formen til U(r), og dermed oppstår riktig makroskopisk oppførsel automatisk.

Samtidig er modellen en idealisering. Den fanger ikke retningsavhengige bindinger eller mer komplekse elektroniske effekter, men den gir likevel en svært god beskrivelse av mange systemer der svake intermolekylære krefter dominerer, som for eksempel edelgasser og ikke-polare molekyler.

## 4.3 - Fra parpotensial til van der Waals gas-likningen

Gjennom de foregående avsnittetne har vi beskrevet vekselvirkningen mellom to partikler ved hjelp av et parpotensial u(r). Vi har sett at potensialet består av en sterkt frastøtende del av korte avstander og en svakere attraktiv del ved større avstander. For mange systemer kan denne vekselvirkningen beskrives godt av Lennard-Jones potensialet.

I en virkelig gass består systemet imidlertid ikke av to molekyler, men av et enormt antall partikler som alle vekselvirker med hverandre samtidig. Det naturlige spørsmålet blir derfor hvordan de intermolekylære kreftene vi har beskrevet på molekylet nivå pårvirker makroskopiske størrelser som trykk,volum og temperatur. 

En av de enkleste modellene som kobler disse to beskrivelsesnivåene sammen er van der Waals likningen. 

(Sett in likningen her)

Som kan betraktes som en korrigert utgave av den ideelle gassloven. De to korreksjonsleddene, a og b, representerer hver sin fysiske mekanisme som vi allerede har møtt gjennom parpotensialet. 

### Molekylene har en endelig størrelse

Den ideelle gassloven behandler molekylene som punktpartikler uten utstrekning. Dette er en god tilnærming ved lave tettheter, men ved høyere tettheter blir molekylenes størrelse viktig. To molekyler kan ikke oppta samme rom samtidig, noe som allerede kommer til uttrykk i Lennard-Jones potensialet gjennom den kraftige frastøtningen ved små avstander. 

Konsekvensen er at ikke hele beholdervolumet er tilgjengelig for molekylenes bevegelse. RUndt hvert molekyl finnes et område som andre molekyler ikke kan trenge inn i. Det tilgjengelige volumet blir derfor mindre enn det totale volumet, og dette beskrives ved korreksjonsleddet b. I stedet for å bevege seg i volumet Vm, beveger molekylene seg effektivt i volumet Vm - b. 

Parameteren b er derfor et mål på molekylenes effektive størrelse, og kan knyttes direkte til den frastøtende delen av parpotensialet. 

### Molekylene tiltrekker hverandre
Den ideelle gassloven antar også at molekylene ikke påvirker hverandre mellom kollisjonene. Dette stemmer heller ikke for virkelige gasser. Som vi har sett tidligere, vil van der Waals-krefter gi opphav til en attraktiv vekselvirkning mellom molekylene.

Denne tiltrekningen reduserer trykket på beholderveggene. Et molekyl som nærmer seg veggen blir trukket svakt tilbake av nabomolekylene inne i gassen, slik at kollisjonen med veggen blir mindre kraftig enn i en ideell gass. Det målte trykket blir derfor lavere enn det den ideelle gassloven predikerer.

Styrken til denne effekten samles i parameteren a. Jo sterkere de intermolekylære tiltrekningene er, desto større blir a, og desto større blir korreksjonen til trykket.

### To bidrag til trykket

Trykket kan generelt tenkes på som summen av to effekter:

- Et **entropi-bidrag**: molekylene "vil" spre seg utover for å maksimere antall tilgjengelige mikrotilstander. Dette er det eneste bidraget en ideell gass har, og det gir det kjente P = NkBT/V.
- Et **energi-bidrag**: tiltrekningen mellom molekylene trekker dem sammen og reduserer trykket sammenlignet med det ideelle tilfellet, fordi molekylene «foretrekker» å være nær hverandre.

Van der Waals-likningen fanger opp begge disse effektene ved å gjøre to enkle justeringer av den ideelle gassloven.


### Leddet b: det ekskluderte volumet

Tenk deg at molekylene ikke er punktpartikler, men små harde kuler med en viss radius. Da kan ikke to molekyler okkupere samme sted samtidig — akkurat som i Lennard-Jones-modellen, der potensialet stiger brått mot uendelig når R nærmer seg RLJ (eller i hardkule-modellen, der u(R) = ∞ for R ≤ Rhs).

Konsekvensen er at hvert molekyl faktisk har **mindre plass til rådighet** enn det totale volumet V skulle tilsi, fordi et område rundt hvert av de andre molekylene er «forbudt». Dette forbudte volumet er nettopp b (multiplisert med Avogadros tall for å få det per mol). Legg merke til at b avhenger bare av *størrelsen* på molekylene (gjennom RLJ eller Rhs) — ikke av hvor sterkt de tiltrekker hverandre.

Fordi det effektive tilgjengelige volumet er (V − Nb) i stedet for V, blir entropien til gassen redusert i forhold til det ideelle tilfellet, og dette gir opphav til leddet (Vm − b) i van der Waals-likningen.

### Leddet a: tiltrekningen

Tiltrekningen mellom molekylene senker den totale potensielle energien til systemet. Jo tettere gassen er pakket (jo mindre Vm), desto flere naboer har hvert molekyl innenfor rekkevidde av tiltrekningskreftene, og desto mer negativ blir den gjennomsnittlige potensielle energien per molekyl.

Vi kan faktisk regne ut dette bidraget direkte fra u(R): energien til ett «test-molekyl» på grunn av alle naboene er integralet av u(R) vektet med tettheten av molekyler i en tynn kuleskall på avstand R,

$$U' = \int_0^\infty u(R)\,\rho\, 4\pi R^2\, dR,$$

og gjør vi dette for et Lennard-Jones-lignende potensial (eller det enklere «tiltrekning ~ −1/R⁶ utenfor en hard kjerne»-potensialet), finner vi at U er proporsjonal med −N²/V. Konstanten a samler dermed opp styrken på tiltrekningen (proporsjonal med brønndybden ε og rekkevidden RLJ³), på samme måte som b samler opp størrelsen. Jo større a, desto mer senker tiltrekningen trykket sammenlignet med det ideelle tilfellet — dette er nøyaktig leddet a/Vm² i likningen.

### Gittermodellen – en enkel fysisk forklaring

Selv om sammenhengen ovenfor kan utledes direkte fra parpotensialet, finnes det en enklere måte å forstå van der Waals-likningen på.

Vi tenker oss at volumet deles inn i et stort antall små celler, der hver celle omtrent har størrelsen til ett molekyl. Hver celle kan enten være tom eller inneholde ett molekyl. På denne måten blir det umiddelbart klart at molekylene opptar plass, og at ikke hele volumet er tilgjengelig for bevegelse. Dette er den fysiske opprinnelsen til parameteren b.

Deretter lar vi hvert molekyl vekselvirke med sine nærmeste naboer gjennom den attraktive delen av parpotensialet. Når disse bidragene summeres over hele systemet, får vi en gjennomsnittlig attraktiv energi som er proporsjonal med -N^2/V, og dette gir parameteren a.

Gittermodellen viser dermed at de to korreksjonene i van der Waals-likningen har ulik fysisk opprinnelse: b beskriver at molekylene har en endelig størrelse og reduserer det tilgjengelige volumet, mens a beskriver den gjennomsnittlige attraktive vekselvirkningen mellom molekylene.
