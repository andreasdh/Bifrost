## Oppgave 1

Lennard-Jones-potensialet modellerer van der Waals-energien mellom to nøytrale molekyler:

$$u(r) = 4\varepsilon\left[\left(\frac{\sigma}{r}\right)^{12} - \left(\frac{\sigma}{r}\right)^{6}\right]$$

For metan (CH₄) er parameterne $\sigma = 3{,}93\ \text{Å}$ og $\varepsilon/k_B = 148\ \text{K}$.

**a)** Skisser $u(r)$ som funksjon av $r$. Marker $\sigma$ (der kurven krysser null), likevektsavstanden $r^*$ (bunnen av brønnen) og brønndybden $\varepsilon$. Hva betyr det fysisk at kurven krysser null nettopp ved $\sigma$?

**b)** Bunnen av brønnen ligger litt *utenfor* $\sigma$ — faktisk ved $r^* = 2^{1/6}\sigma \approx 1{,}12\,\sigma$. Hvorfor gir det mening at likevektsavstanden er litt større enn $\sigma$, og ikke akkurat der kurven krysser null? *(Tenk på hvor tiltrekning og frastøtning balanserer.)*


![LJ_oppgave](../bilder/LJ_oppgave1.svg)

**c)** Bunnen av brønnen ligger $\varepsilon$ under null — det er nettopp derfor $\varepsilon$ kalles brønndybden. For metan tilsvarer $\varepsilon$ omtrent 148 K i temperaturenheter. Er dette en «sterk» eller «svak» binding sammenlignet med en kovalent binding (som er tusenvis av K)? Hva sier det om hvor lett to metanmolekyler rives fra hverandre?

**d)** Under energikurven, skisser hvilken vei *kraften* dytter på ulike avstander: utover (frastøtning) når molekylene er for nær, innover (tiltrekning) når de er litt for langt fra hverandre, og null i bunnen av brønnen. Marker det stabile likevektspunktet. *(Du trenger ikke regne — bare les av hvilken vei kurven heller.)*

**e)** Hvorfor er det naturlig at $\varepsilon$ oppgis i temperaturenheter? Hva «føler» to metanmolekyler når den termiske energien $k_B T$ er mye større enn $\varepsilon$ — og hva skjer når $k_B T \approx \varepsilon$?

## Oppgave 2

Lennard-Jones-parametre for tre nøytrale stoffer (Atkins, *Physical Chemistry*):

| Molekyl | $\varepsilon/k_B$ (Kelvin) | $\sigma$ (Å) |
|---|---|---|
| Helium (He) | 10,22 | 2,58 |
| Etan (C₂H₆) | 205 | 4,23 |
| Benzen (C₆H₆) | 440 | 5,27 |

**a)** Hvilket av stoffene er størst, og hvilket binder sterkest? Hvordan leser du dette direkte ut av de to kolonnene?

**b)** Ingen av de tre har et permanent dipolmoment. Hvorfor tiltrekker de likevel hverandre? Hvilken av van der Waals-kreftene (Keesom, Debye, London) er ansvarlig her?

**c)** Brønndybden øker kraftig fra He til etan til benzen. Hva er den fysiske grunnen til at $\varepsilon$ vokser i denne rekkefølgen? Knytt svaret til begrepet *polariserbarhet*.

**d)** Rangér stoffene etter forventet kokepunkt ut fra $\varepsilon$. Sammenlign så med de virkelige kokepunktene: He $-269\ ^\circ$C, etan $-89\ ^\circ$C, benzen $+80\ ^\circ$C. Stemmer rangeringen?

**e) Organisk kobling:** Gå oppover alkanrekka: metan, etan, propan, … , dekan. Kokepunktet stiger jevnt, og både $\sigma$ og $\varepsilon$ øker med kjedelengden. Forklar hvorfor det er *den samme* svake kraften som virker hele veien, men at den summerer seg opp langs kjeden. Bruk dette til å begrunne hvorfor de korteste alkanene er gasser, mellomlange er væsker og de lengste er voksaktige faste stoffer ved romtemperatur.

## Oppgave 3

Van der Waals-likningen:

$$\left(P + \frac{a}{V_m^2}\right)(V_m - b) = RT$$

Van der Waals-konstanter for noen ekte gasser ($a$ i L²·bar·mol⁻², $b$ i L·mol⁻¹):

| Stoff | $a$ | $b$ |
|---|---|---|
| Helium (He) | 0,035 | 0,0238 |
| Argon (Ar) | 1,355 | 0,0320 |
| Nitrogen (N₂) | 1,370 | 0,0387 |
| Metan (CH₄) | 2,30 | 0,0431 |
| Karbondioksid (CO₂) | 3,64 | 0,0427 |
| Benzen (C₆H₆) | 18,24 | 0,1193 |

**a)** Konstanten $a$ hører til den ene delen av parpotensialet $u(r)$, og $b$ til den andre. Hvilken er hvilken? Koble hver av dem til Lennard-Jones-parameterne $\varepsilon$ og $\sigma$ og forklar hvorfor.

**b)** Rangér stoffene etter *molekylstørrelse* (ut fra $b$) og etter *tiltrekningsstyrke* (ut fra $a$). Følger de to rangeringene hverandre? Pek på et sted der de spriker, og forklar hva det forteller (f.eks. He vs. N₂, eller CO₂ vs. CH₄).

**c)** To gasser — én med stor $a$ og liten $b$, én med liten $a$ og stor $b$. Hvilken avviker mest fra idealgassloven ved *lav temperatur*? Og ved *høyt trykk*? Hvorfor?

**d) Bonus** Ekskludert volum og "harde kuler":
Den aller enkleste modellen for et molekyl er en "hard kule": parpotensialet er uendelig stort så snart to molekyler prøver å overlappe $u(r) = \infty$ for $r < \sigma$ og null ellers, altså bare den harde frastøtningen, ingen tiltrekning. Hvert molekyl "stenges av" et lite volum rundt seg som ingen andre kan trenge inni; dette kalles *ekskludert volum*, og det er dette b måler. 
* Hva blir $a$ i denne modellen, når det ikke finnes noen tiltrekning? Og hva blir $b$, når molekylene fortsatt opptar plass?

 
## Oppgave 4

Parkorrelasjonsfunksjonen $g(r)$ forteller hvor sannsynlig det er å finne en partikkel i avstand $r$ fra en referansepartikkel, sammenlignet med gjennomsnittstettheten.

**a)** Skisser $g(r)$ for argon i tre tilstander: gass, væske (ved ~87 K) og fast stoff. Hva er hovedforskjellen mellom de tre kurvene?

**b)** For alle tre er $g(r) = 0$ ved svært små $r$. Koble dette direkte til formen på parpotensialet $u(r)$. Hvorfor kan ikke partikler komme nærmere enn en viss avstand?

**c)** I væsken har $g(r)$ en tydelig første topp, og deretter flere topper som gradvis dør ut. Hva forteller *posisjonen* til første topp? Og hva forteller det at toppene *dør ut* på litt lengre avstand?

**d) Eksperimentell kobling:** $g(r)$ kan måles med røntgendiffraksjon (XRD). Hva slags diffraksjonsmønster forventer du for *fast* argon kontra *flytende* argon — skarpe topper eller diffuse ringer? Knytt dette til Braggs lov, $n\lambda = 2d\sin\theta$.

**e) kobling til multipoler:** Sammenlign to parpotensialer:

$$(u_1(r) = -\frac{1}{r} \qquad \text{(ione-ione, Coulomb)} \qquad u_2(r) = -\frac{1}{r^{6}} \qquad \text{(dispersjon)})$$

Ved $r = 1$ og ved $r = 2$: hvilket potensial gir sterkest tiltrekning? Bruk dette til å forklare hvorfor ionekrystaller (NaCl) har *langtrekkende* orden, mens en edelgassvæske bare har *korttrekkende* orden. Knytt an til påstanden i kapittelet om at høyere multipolmomenter gir svakere og mer korttrekkende vekselvirkninger.

> **Hint (e):** Sett inn $r=1$ og $r=2$ i begge uttrykkene og sammenlign tallene — $1/r$ dør mye saktere med avstanden enn $1/r^6$. Det er nettopp derfor ione-vekselvirkninger (monopol) rekker langt og gir langtrekkende orden, mens dispersjon bare merkes mellom nærmeste naboer.


## Oppgave 5

Fri energi: $F = U - TS$. Systemet søker alltid lavest $F$.

**a)** Ved *høy* temperatur vinner det ene bidraget, ved *lav* temperatur det andre. Hvilket vinner når, og hvorfor er det nettopp temperaturen som avgjør? Bruk $F = U - TS$ i argumentet.

**b)** Hvorfor kondenserer en gass når du kjøler den ned? Argumenter fra *begge* sider — hva vil energien, hva vil entropien, og hvem vinner ved lav $T$? Koble energisiden til den attraktive brønnen i $u(r)$ og til $a$-leddet i van der Waals-likningen.

**c)** Kondensasjon er bare ett eksempel på en faseovergang. Nevn kort et par andre overganger (f.eks. tørris som sublimerer, is som smelter) og pek på hvilken vei energi- og entropibalansen tipper i hvert tilfelle. *(Selve fasediagrammene ser vi nærmere på i et senere kapittel — her holder det å tenke i termer av $U$ mot $TS$.)*

**d)** Is flyter på vann — det faste stoffet er *mindre* tett enn væsken, motsatt av det vanlige. Hva forteller dette unntaket oss om at «lavest energi» ikke alltid gir tettest pakking? *(Tenk på geometrien til hydrogenbindingene i iskrystallen.)*
