## Regneoppgaver 

### Oppgave 1 - utvid med diskusjon om likevekter og reversibelt/irreversibelt arbeid

Vi ser på en ideell gass som ekspanderer isotermt ved temperaturen $T=300 K$ fra en begynnelsestilstand med trykk $p_1 = 10.0$ atm og volum $V_1 = 1.0 \text{dm}^3$ til en slutt-tilstand hvor trykket er $p_2 = 1.0$ atm og volumet er $V_2 = 10.0 \text{dm}^3$.
Prosessen skjer på 3 ulike måter:  

1.	Reversibelt  
2.	Irreversibelt mot et konstant ytre trykk på $1$ atm  
3.	I to trinn: først irreversibelt mot konstant ytre trykk på $5.0$ atm frem til trykket i gassen er $p_{midtveis} = 5.0$ atm og volumet er $V_{midtveis} = 2.0 \text{dm}^3$, deretter til slutt-tilstanden under et konstant ytre trykk på $1.0$ atm. 

**a)** Beregn arbeidet $w$ i de tre tilfellene.  

**b)** Tegn pV-diagrammer hvor dere markerer de arealene som svarer til det utførte arbeidet for alle tre tilfellene.  

**c)** Hva er det maksimale arbeid denne prosessen kan gi?  


### Oppgave 2 - utvide med mer fokus på relasjonen mellom Cp og Cv?

Når energien $187 J$ tilføres som varme til $1.9$ mol gassmolekyler ved konstant trykk, øker prøvens temperatur med $1,78 K$.   

**a)** Beregne de molare varmekapasitetene for gassen ved hhv konstant volum og konstant trykk.  

**b)** Hva er relasjonen mellom varmekapasitetene ved henholdsvis konstant trykk og volum?   


### Oppgave 3 - finne dH fra dU 

$1,00$ mol av en ideell gass ved $27^{\circ}C$ får ekspandere isotermt fra et start-trykk på $3,00$ atm til et slutt-trykk på $1,00$ atm på to måter:  

1. Reversibelt  
2. Mot et konstant ytre trykk på $1,00$ atm. 

Beregne $q, w, \Delta U, \Delta H, \Delta S_{system}, \Delta S_{omgivelser}, \Delta S_{tot}$ for hver vei og kommenter på de forskjellige verdiene dere får for $\Delta S$.


### Oppgave 4

En jernbit blir plassert i kontakt med en aluminiumskloss hvor begge har samme volum på $50 cm^3$. Jernbiten har en temperatur på $0^{\circ}C$ og aluminiumsklossen $100^{\circ}C$. De to klossene bringes i termisk kontakt, og likevekt innstiller seg. Anta at systemet er termisk isolert. $C_{p,m}= 25.10 J K^{-1} mol^{-1}$ for jern og $C_{p,m}= 24,4 J K^{-1} mol^{-1}$ for aluminium. Tettheten er henholdsvis $2,698 g cm^{-3}$ og $7,874 g cm^{-3}$ for aluminium og jern.  

**a)** Beregne slutt-temperaturen. 

**b)** Beregne $\Delta S$ for hver av de to klossene, og for de to klossene til sammen. Kommenter kort fortegnene til de tre $\Delta S$-ene som er beregnet.
