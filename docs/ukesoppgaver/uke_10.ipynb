### Oppgave 7

For å illustrere kjemisk likevekt mikroskopisk kan vi modellere et ligand $L$ som binder til en reseptor $R$ og danner et kompleks $C$ (se figur under). Det er $N_R$, $N_L$ og $N_C$ molekyler av henholdsvis $R$, $L$ og $C$, plassert på et gitter med $V$ mulige plasseringer. Reaksjonen skjer ved lav konsentrasjon, slik at vi kan se bort fra at flere molekyler konkurrerer om samme plass.

Hvert molekyl har en pil som peker i én av $z$ mulige retninger ($z=6$ på et tredimensjonalt kubisk gitter). Når et ligand-molekyl er riktig orientert i en reseptor-plass, bindes det med energien $\varepsilon_0<0$. Ser vi bort fra feil-orienterte (ubundne) ligander, blir partisjonsfunksjonene til de tre molekyltypene

$$q_R=Vz,\qquad q_L=Vz,\qquad q_C=Vze^{-\varepsilon_0/k_BT}.$$

Her er $V$ den translasjonelle partisjonsfunksjonen (antall gitterplasser molekylet kan sentreres på), og $z$ den rotasjonelle partisjonsfunksjonen (antall retninger pilen kan peke). Bindingsenergien $\varepsilon_0$ gjelder kun når ligand- og reseptorpilene er riktig innrettet mot hverandre.

![Gittermodell for bindingslikevekt](../bilder/gitter_bindingslikevekt.svg)

**a)** Likevektskonstanten (bindingsaffiniteten) er definert som $K=\dfrac{N_C}{N_RN_L}$. Bruk partisjonsfunksjonene over til å vise at

$$K=\frac{e^{-\varepsilon_0/k_BT}}{Vz}\text{.}$$

**b)** Hva skjer med $K$ dersom vi fortynner systemet (øker $V$, med samme antall molekyler av hver sort)? Forklar hvorfor det gir fysisk mening at større fortynning gir mindre kompleks $C$ ved likevekt.

**c)** Hva skjer med $K$ dersom $z$ øker, altså om molekylet krever en strengere orientering for å kunne binde seg (flere mulige retninger, men bare én som gir binding)? Hvorfor gir det mening at flere "unyttige" orienteringsmuligheter reduserer bindingsaffiniteten?

**d)**  
$$K = \underbrace{e^{-\varepsilon_0/k_BT}}_{\text{energileddet}} \cdot \underbrace{\frac{1}{Vz}}_{\text{frihetsleddet}}$$

Kjemikere gir disse to hver sitt navn: **energileddet** knyttes til bindingsentalpien $\Delta h$, og **frihetsleddet** til bindingsentropien $\Delta s$. I denne modellen blir de

$$\Delta h = \varepsilon_0, \qquad \Delta s = -k_B\ln(Vz)$$
