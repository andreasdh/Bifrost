### Oppgave 1: Prosesser hvor vi holder en variabel konstant.

**a)** Forklar kort hva som kjennetegner henholdsvis en *isoterm*, *isokor*, *isobar* og *adiabatisk* prosess.

**b)** Skisser følgende prosesser i samme PV-diagram, anta idealgass i alle trinn: 

1) Isobar ekspansjon ved et trykk $P_1$ fra volum $V_1$ til $V_2$
2) Isokor nedkjøling ved $V_2$ av gassen til et trykk $P_2$
3) Isoterm kompresjon av gassen fra ($P_2, V_2$) til ($P_1$,$V_1$) 

**c)** I diagrammet fra oppgave *b*, skisser visuelt hva arbeidet utført av denne syklusen blir 

**d)** Anta at hvert steg gjøres uendelig langsomt (kvasi-statisk) slik at likevekt alltid er opprettet, og null friksjon.  

Finn utrykk for hvor mye *arbeid* gjøres av systemet på omgivelsene i hvert trinn 1-3? 

Bruk $V_1=1.0\ \mathrm{dm^3}$,  $V_2=3.0\ \mathrm{dm^3}$, $P_1=4.0\ \mathrm{atm}$, 
Finn først P2 (husk idealgass, og isoterm prosess), og bruk så dette til å finne totale arbeidet utført av syklusen?

**e)** Hva er entropiendringen til gassen for den isoterme kompresjonen (steg 3), og hva er gassens entropiendring totalt (bruk at T=300 K)?


```python
import numpy as np
import matplotlib.pyplot as plt

P1 = 4.0
V1 = 1.0
V2 = 3.0
P2 = P1 * V1 / V2
print(f"P2 = {P2:.2f} (for P1={P1}, V1={V1}, V2={V2})")

# Tilstandspunkter
A = (V1, P1)
B = (V2, P1)
C = (V2, P2)

# Isoterm: pV = P1*V1
V_iso = np.linspace(V1, V2, 300)
P_iso = P1 * V1 / V_iso

fig, ax = plt.subplots(figsize=(7, 5))

# 1) Isobar ekspansjon: A -> B
ax.plot([V1, V2], [P1, P1], color="tab:blue", lw=2)
ax.annotate(
    "", xy=B, xytext=A,
    arrowprops=dict(arrowstyle="->", color="tab:blue", lw=2)
)

# 2) Isokor nedkjøling: B -> C
ax.plot([V2, V2], [P1, P2], color="tab:orange", lw=2)
ax.annotate(
    "", xy=C, xytext=B,
    arrowprops=dict(arrowstyle="->", color="tab:orange", lw=2)
)

# 3) Isoterm kompresjon: C -> A
ax.plot(V_iso, P_iso, color="tab:green", lw=2)
"""
ax.annotate(
    "", xy=(V_iso[80], P_iso[80]),
    xytext=(V_iso[160], P_iso[160]),
    arrowprops=dict(arrowstyle="->", color="tab:green", lw=2)
)"""

# Tilstandspunkter og etiketter
ax.plot(*A, "ko")
ax.plot(*B, "ko")
ax.plot(*C, "ko")

ax.annotate(r"$A=(V_1,P_1)$", A, xytext=(7, 8), textcoords="offset points")
ax.annotate(r"$B=(V_2,P_1)$", B, xytext=(7, 8), textcoords="offset points")
ax.annotate(r"$C=(V_2,P_2)$", C, xytext=(7, -18), textcoords="offset points")

# Formatering
ax.set_xlabel(r"Volum, $V$")
ax.set_ylabel(r"Trykk, $p$")
ax.set_title(r"$pV$-diagram for syklusen")

ax.axvline(V1, color="gray", ls=":", alpha=0.5)
ax.axvline(V2, color="gray", ls=":", alpha=0.5)
ax.axhline(P1, color="gray", ls=":", alpha=0.5)
ax.axhline(P2, color="gray", ls=":", alpha=0.5)

ax.set_xlim(0.6, 3.4)
ax.set_ylim(0, 4.6)
ax.grid(alpha=0.25)

plt.tight_layout()
plt.show()
```
