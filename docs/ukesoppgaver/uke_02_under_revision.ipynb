## Diskusjonsoppgaver

Tema for denne ukas oppgaver er Boltzmann-fordelingen og den relaterte partisjonsfunsjonen. I forrige uke så vi at systemer drives mot maksimering av entropi, og som Boltzman viste er entropien, $S$, direkte knyttet til multiplisiteten, $\Omega$, av mikrotilstander som hver makrotilstand kan ha gjennom: 

$$ S=k_{B} \ln(\Omega). $$

Kun ved å se på hvilken tilstand som hadde høyest multiplisitet, og ved å anta at alle mikrotilstander er like sansynlige, kunne vi bruke enkle modeller til å forklare hvorfor en gass vil fylle rommet den er i, hvorfor polymerer trekker seg sammen, hvorfor vesker blandes, og hvorfor energi utgjevnes når to stoffer settes i termisk kontakt. (Diskusjonsoppgaver 2-5 i uke 1). 

Nå skal vi se på hva som sjer dersom det er begrensinger på hvilke mikrotilstander et system kan være i.

### Oppgave 2: Hva er sannsynligheten for å finne en partikkel i et gitt nivå når entropien er maksimert OG vi har en konstant totalenergi?

**c)** Hvis vi antar et konstant antall partikler N, og en konstant total energi $E$, hva kan vi si om gjennomsnitsenergien $\langle \varepsilon \rangle$ per partikkel? Bruk den statistiske definisjonen av gjennomsnitt til å finne et utrykk for $\langle \varepsilon \rangle$

$$\langle k \rangle = \sum_{i=1}^t k_i \cdot p_i $$

:::

**d)** Vi ønsker nå å finne sannsynlighetene, $p_i$ for hvert nivå som gir den maksimale entropien, altså den mest sannsynlige fordelingen av partikler når systemet er i likevekt. Dette kan vi gjøre ved bruke Gibbs definisjon av entropi per partikkel: 

$$ S(p)=-k_B\cdot\sum_{i=1}^t p_i \cdot \ln{p_i} $$

Merk at dette er en flervariabel funskjon, som avhenger $p_1,p_2,p_3,...,p_t$. Hva er den totalderiverte, dS, av denne funsjonen? Og hvordan bruker vi denne til å finne maksimum entropi? 

**e)** Fra deloppgave a) og c) har vi fått to betingelser, som vi videre kaller for g(p) og h(p), knyttet til sansynligheten, den første handler om summen av sannsynlighetene $p_i$ og den andre om gjennomsnittsenergien per partikkel. Vi antar nå at dere har funnet disse i henholdsvis deloppgave a) og deloppgave c), men dere kan dobbeltsjekke svarene under før dere går videre:

:::{dropdown} Delsvar:

a) Summen av alle sansynlihetene $p_i$ må være lik 1 

$$ g(p) = \sum_{i=1}^t p_i = 1 $$

c) Gjennomsnittsenergien per partikkel er gitt ved:

$$ h(p) = \langle \varepsilon \rangle = \frac{E}{N} = \sum_{i=1}^t \varepsilon_i \cdot p_i $$

:::

For å finne sansynlighetsfordelingen $p=[p_1,p_2,...,p_t]$ som maksimerer entropien $S(p)$ under disse betingelsene som vi kaller $g(p)$ og $h(p)$, kan vi bruke Lagrange multiplikatorer. Dette gir oss et sett av likninger på formen: 

$$ \frac{\partial S(p)}{\partial p_i} - \alpha \cdot \frac{\partial g(p)}{\partial p_i} -\beta \cdot \frac{\partial h(p)}{\partial p_i} = 0$$

hvor $\alpha$ og $\beta$ er Lagrange multiplikatorene. Vi har egentlig én ligning per verdi av $i$, men formen på disse ligningene og multiplikatorene er de samme. Dermed trenger vi bare å løse en av de med hensyn på en generell variabel $p_i$. 

**Finn hver av de tre partiellderiverte i lingningen over.**

**f)** Sett inn de partiellderiverte dere fant i likningen over og løs med hensyn på $p_i$. Hint: $\alpha$ og $\beta$ skal inngå i svaret.

**g)** Vi husker på at $\sum_{i=1}^t p_i = 1$ og dermed kan vi dele utrykket for $p_i$ på $\sum_{i=1}^t p_i$, bruk dette til å bli kvitt $\alpha$ og forenkle utrykket mest mulig. I hintet under får du se hvordan det endelige utrykke skal se ut til slutt.

$$ p_i = \frac{e^{-\beta\varepsilon_i}}{\sum_{i=1}^t e^{-\beta\varepsilon_i}} $$

:::

### Oppgave 3: Litt om partisjonsfunskjonen

Som vi så i oppgaven over er gir Boltzmann fordelingen sansynligheten $p_i$ for å finne én partikkel i energinivå $e_i$: 

$$ p_i = \frac{e^{-\frac{\varepsilon_i}{k_BT}}}{\sum_{i=1}^t e^{-\frac{\varepsilon_i}{k_BT}}} $$

Nevneren i dette utrykket kalles for den *molekylære partisjonsfunssjonen* $q$: 

$$ q = \sum_{i=1}^t e^{-\frac{\varepsilon_i}{k_BT}}$$

Da kan vi skrive: 

$$ p_i = \frac{e^{-\frac{\varepsilon_i}{k_BT}}}{q} $$

**a)** Hvilken "funskjon" har det rent matematisk å dele sansynlighetene på $q$?

**b)** Fra forelesning har dere kanskje hørt at vi kan tenke på partisjonsfunsjonen som et mål på hvor mange av energitilstandene er tilgjengelige ved en gitt temperatur. Hva menes med dette, og hvordan kan vi se det? 

Hint: Skriv ut et par ledd av $q$ og tenk på hvilken av variablene som er bestemt av systemet og som vi dermed ikke kan endre, og hvilke vi kan styre. 

**c)** Vi bruker symbolet $q$ når vi snakker om partisjonsfunsjonen for én partikkel, dette kan vi tenke på som sannsynlighetsfordelingen for at den partikkelen er i et gitt energinivå. Om vi har N partikler som er **uavhengige** og **skillbare** er den totale partisjonsfunskjonen gitt ved å multiplisere sammen partisjonsfunksjonene for hver partikkel:

$$ Q=q_1 \cdot q_2 \cdot q_3 \cdot ... \cdot q_N = q^N $$

Hva er den totale partisjonsfunsjonen $Q$ dersom partiklene er **uavhengige**, men **uskillbare**? Hvordor er dette slik?

### Oppgave 4: Fra partisjonfunsjon til termodynamiske størrelser

Partisjonsfunksjonen kan være svært nyttig ettersom at vi kan bruke den til å bestemme veldig mange makroskopiske termodynamiske størrelser for et system, slik som indre energi, entropi og varmekapasitet for å nevne noen. Her skal vi begynne med et eksempel fra vår kjente 2-nivå model, som i dette tilfelle også kalles for **Schottky-modellen**. 

Vi har N partikler som er uavhengige og skillbare, hver partikkel kan enten være i grunntilstant med energi $\varepsilon_0=0$ eller i en eksitert tilstand med energi $\varepsilon_1$. 

**a)** Hva er den molekylære partisjonsfunsjonen $q$ for dette systemet?

**b)** Hva er den gjennomsnittelige energien per partikkel $\langle \varepsilon \rangle$ utrykket med $\varepsilon_1$ og $k_BT$? Hvilken verdi går $\langle \varepsilon \rangle$ mot når $T \rightarrow \infty $?

**c)** Hva er den totale indre energien $U$ i systemet, utrykkt med $N$ og $\langle \varepsilon \rangle$

**d)** Bruk utrykket for $\langle \varepsilon \rangle$ fra b), $U$ fra oppgave c) og definisjonen av varmekapasitet: 

$$ C_v= \left( \frac{\partial U}{\partial T}  \right)_{N,V} $$

til å finne varmekapasiteten for dette systemet. 

:::{dropdown} Hint for derivasjonen:

$$C_v= \left( \frac{\partial U}{\partial T}  \right)_{N,V} = N \left( \frac{\partial \langle \varepsilon \rangle}{\partial T}  \right) = N \left( \frac{\partial \langle \varepsilon \rangle}{\partial \beta} \right)\left( \frac{\partial \beta}{\partial T} \right) $$

der $\beta = 1/k_BT$ og partisjonsfunsjonen utrykkes med $\beta$

bruk da derivasjonsregelen:



$$ f(x)=\frac{u(x)}{v(x)} \implies f'(x)=\frac{u'\cdot v - v'\cdot u}{v^2} $$

:::
