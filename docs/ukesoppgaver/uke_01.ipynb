# Uke 1 – Statistisk mekanikk og mikro–makro-sammenhengen
:::{admonition} Læringsutbytte
:class: note
Etter å ha jobbet med dette temaet skal du kunne:

- forklare forskjellen mellom en mikrotilstand og en makrotilstand
- forklare hvorfor statistiske beskrivelser er nødvendige for systemer som består av svært mange partikler
- beskrive energi som en egenskap ved mikrotilstander og som en størrelse som kan fordeles mellom molekyler på mange ulike måter
- forklare entropi som et mål på antall mikrotilstander som er forenlige med en gitt makrotilstand
- forklare hvorfor makrotilstander med mange mikrotilstander er mer sannsynlige enn makrotilstander med få mikrotilstander
:::

### Oppgave 3

I modellen under ser vi på et lite $2\times2$-gitter med fire plasser ($A$, $B$, $C$ og $D$), der to av plassene er okkupert av type 1-molekyler og to av type 2-molekyler. Molekylene av samme type regnes som uskjelnbare seg imellom: å bytte om de to type 1-molekylene gir ingen ny mikrotilstand. Dette gitteret kan du tenke på som et lite subsystem hentet ut fra et mye større gitter, vist til venstre i figuren.

En kontakt mellom et type 1- og et type 2-molekyl, markert med oransje streker, koster energien $\Delta\epsilon$, mens kontakt mellom to like molekyler ikke koster noe. Vi teller bare direkte naboer: opp, ned, venstre, høyre.

**a)** Bruk subsystemet over som eksempel: hva er én mikrotilstand i denne modellen, og hvorfor er det *ikke* det samme som én mikrotilstand for ett enkelt atom?

**d)** Antallet *ulike* energiverdier tilsvarer antallet makrotilstander. Hvor mange makrotilstander $E$ fant du blant de seks mikrotilstandene i c)? Tenk deg nå et større system med $N_A\approx6\cdot10^{23}$ molekyler i stedet for bare 4. Uten å regne det ut nøyaktig: hva tror du skjer med antall mikrotilstander sammenlignet med antall makrotilstander, når systemet blir så mye større? Bruk svaret ditt til å forklare hvorfor vi trenger en statistisk beskrivelse for systemer med svært mange partikler, i stedet for å bare telle opp alle mikrotilstandene direkte, slik du gjorde i c).

### Oppgave 4

Vi ser på $N$ uavhengige delsystemer med faste, identifiserbare plasser. **c)** Energien i systemet lagres ved at flere og flere delsystemer eksiteres. Blir det alltid flere tilgjengelige mikrotilstander av å øke energien? (Hint: hva skjer med $\Omega(n)$ når *nesten alle* delsystemene er eksiterte?)

Lek med glidebryterne (spesielt $N$) og sammenlign posisjonen til den røde søylen med toppen av fordelingen. Er de alltid den samme søylen? ### Oppgave 5

Sannsynligheten for å finne en mikrotilstand $i$ med energi $E_i$, ved temperatur $T$, er gitt av Boltzmann-fordelingen:

$$p_i=\frac{1}{Q}e^{-E_i/k_BT}$$

der $k_B$ er Boltzmanns konstant og $Q$ er partisjonsfunksjonen, som sørger for at alle sannsynlighetene til sammen blir 1.

**a)** Hva skjer med $p_i$ når energien $E_i$ øker, ved en fast temperatur $T$?

**b)** Hva skjer med $p_i$ for en gitt, fast energi $E_i$ dersom temperaturen $T$ øker?

**c)** Sammenlign to mikrotilstander med energi $E_1$ og $E_2$, der $E_2>E_1$. Hva tror dere skjer med forholdet $p_2/p_1$ når temperaturen blir svært høy? Og når temperaturen nærmer seg null?

**d)** Ved svært lav temperatur: hvilken tilstand vil systemet stort sett befinne seg i? Hvorfor?

**e)** Ved svært høy temperatur: hva skjer med forskjellen i sannsynlighet mellom lave og høye energitilstander?

**f)** Boltzmann-faktoren inneholder forholdet $E_i/(k_BT)$, ikke $E_i$ eller $T$ hver for seg. Diskuter: hvorfor tror dere det er nettopp dette *forholdet* som avgjør sannsynligheten, og ikke energien eller temperaturen alene?

### Oppgave 7

**a)** Bruk Stirlings approksimasjon på $\Omega(q,N)=\frac{(q+N-1)!}{q!(N-1)!}$ og vis at

$$\ln\Omega\approx(q+N)\ln(q+N)-q\ln q-N\ln N$$

(for store $q$ og $N$, slik at $q+N-1\approx q+N$ og $N-1\approx N$).

**b)** Bruk dette til å utlede at likevektsbetingelsen $\frac{d}{dq_A}\ln\Omega_\mathrm{tot}=0$ gir $\frac{q_A}{N_A}=\frac{q_B}{N_B}$, altså $T_A=T_B$ med $T=q/N$.
