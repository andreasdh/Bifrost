### Oppgave 9 - Ekstra utfordring

I denne oppgaven vil du få bruk for Stirlings approksimasjon, $\ln n! \approx n\ln n - n$, som brukes i statistisk mekanikk for å forenkle uttrykk med store fakulteter. Denne tilnærmingen kan også skrives som $n! \approx (n/e)^n$.

**b)** Bruk svaret fra oppgave a og b, og vis at vi kan uttrykke multiplisiteten som
