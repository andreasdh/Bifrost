### Oppgave 2 $^*$

Hvorfor fyller en gass hele det tilgjengelige volumet, og hvordan henger dette sammen med trykk?

### Oppgave 5 $^*$

I oppgave 3 så vi på omfordeling av partikler. Energi kan også overføres mellom systemer.

Vi ser på to delsystemer, A og B, med ti skjelnbare partikler hver. Hver partikkel kan befinne seg i grunntilstanden med energi $0$ eller i en eksitert tilstand med energi $\varepsilon$. Hvis $n_A$ og $n_B$ er antallet eksiterte partikler, er energiene $U_A=n_A\varepsilon$ og $U_B=n_B\varepsilon$.

Til å begynne med er $U_A=2\varepsilon$ og $U_B=4\varepsilon$. Når delsystemene settes i termisk kontakt, kan energien fordeles på nytt, mens totalenergien er konstant.

2. Vis at den mest sannsynlige makrotilstanden har $n_A=n_B=3$ og dermed $U_A=U_B=3\varepsilon$.
3. Betyr termisk likevekt at energien alltid må være fordelt helt likt mellom små delsystemer? Begrunn svaret.

### Oppgave 8: Likevekt i et to-nivåsystem

Vi ser på $N$ uavhengige delsystemer med faste, identifiserbare plasser. Hvert delsystem kan være i grunntilstanden med energi $0$ eller i en eksitert tilstand med energi $\Delta\varepsilon$. Vi antar at hvert nivå består av én mikrotilstand.

$$E=n\Delta\varepsilon$$

### Oppgave 9 – Ekstra utfordring: Fra sannsynlighet til entropi

Vi skal nå bruke en enkel modell til å knytte sammen multiplisitet, sannsynlighet og entropi.

Tenk deg overflaten til et fast stoff med $N$ faste, identifiserbare bindingsplasser for gassmolekyler. Slike bindingsplasser kalles adsorpsjonsplasser. Hver plass kan være tom eller inneholde ett molekyl, og nøyaktig $k$ av plassene er opptatt. Vi skiller ikke mellom molekylene, men ulike mønstre av opptatte og tomme plasser teller som ulike mikrotilstander. Multiplisiteten er derfor

$$\Omega(k,N)=\binom{N}{k}=\frac{N!}{k!(N-k)!}$$

**a)** Forklar hvorfor binomialkoeffisienten teller antallet mikrotilstander. Hvis du velger én tilfeldig plass, hva er sannsynligheten $p$ for at den er opptatt? Hva er sannsynligheten for at den er tom?

For store verdier av $N$ er det upraktisk å beregne fakultetene direkte. Vi bruker derfor Stirlings tilnærming

$$\ln n!\approx n\ln n-n$$

**b)** Ta logaritmen av uttrykket for $\Omega$, og bruk Stirlings tilnærming til å vise at

$$\ln\Omega\approx N\ln N-k\ln k-(N-k)\ln(N-k)$$

Sett deretter inn $p=k/N$ og vis at uttrykket kan skrives som

$$\ln\Omega\approx-N\left[p\ln p+(1-p)\ln(1-p)\right]$$

Vis også at dette tilsvarer

$$\Omega\approx\frac{1}{p^k(1-p)^{N-k}}$$

**c)** Bruk Boltzmanns entropidefinisjon $S=k_\mathrm{B}\ln\Omega$ til å vise at entropien kan skrives som

$$S\approx-Nk_\mathrm{B}\left[p\ln p+(1-p)\ln(1-p)\right]$$

Tolk uttrykket. Når er entropien størst? Hva skjer når nesten alle plassene er tomme eller nesten alle er opptatte?

**d)** Anta nå at hver plass kan være i én av $t$ mulige tilstander. Tilstand $i$ forekommer $n_i$ ganger, slik at $\sum_i n_i=N$, og vi definerer $p_i=n_i/N$. Multiplisiteten er

$$\Omega=\frac{N!}{\prod_i n_i!}$$

Bruk samme framgangsmåte til å vise at

$$S\approx-Nk_\mathrm{B}\sum_i p_i\ln p_i$$

Forklar med egne ord hvordan dette uttrykket knytter entropi til fordelingen av sannsynligheter mellom de mulige tilstandene.

**Ekstra:** For en halvfylt overflate er $p=1/2$, og tilnærmingen gir $\ln\Omega\approx N\ln 2$. Sammenlign denne verdien med den eksakte verdien $\ln\binom{N}{N/2}$ for $N=8$, $20$, $50$ og $100$. Du kan bruke funksjonene **math.comb** og **math.log** i Python. Hvordan endres den relative feilen når $N$ øker, og hvorfor er logaritmeformen spesielt nyttig for store systemer?
