### Oppgave 3

Vi tenker oss en generell galvanisk celle med de to halvreaksjonene:  

Anode: $A (s) \rightarrow A^{\nu+}(aq) + \nu {e^-} $

Katode $ B^{\nu+}(aq) + \nu {e^-} \rightarrow B(s)  $

Vi kan skrive opp endringen i raksjonens Gibbs frie energi ved "extent of reaction", $\xi$, på følgende måte: $\Delta_r G = (\frac{\partial G}{\partial \xi})_{p,T}$

Vis at dersom vi har en reaksjon $\nu$ elekroner overføres, vil endringen i fri energi være gitt ved $\Delta_r G= -\nu F E_{cell}$
