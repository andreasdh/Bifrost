## Faser og faseoverganger

### Hva er en fase?

:::{admonition} Definisjon - Fase
:class: important

En **fase** er en del av et system som er ensartet, både i kjemisk sammensetning og fysisk tilstand.

For et rent stoff skiller vi ofte mellom aggregattilstandene fast stoff, væske, gass og plasma. Men *fase* er et mer presist begrep: Et system kan ha flere faste faser med ulik struktur, og en fase kan inneholde flere kjemiske forbindelser.
:::

Legg merke til at *fase* ikke er det samme som *tilstand*. Is og vann er to faser, men en blanding av isbiter i vann er også to faser, selv om isbitene er adskilte. Det avgjørende er at hver fase er homogen i seg selv, med en tydelig grenseflate mot de andre.



### Faseoverganger

Når temperatur, trykk eller sammensetning endres, kan én fase gå over til en annen. Dette kaller vi en **faseovergang**.


:::{admonition} Definisjon - Faseovergang
:class: important

En **faseovergang** er en endring fra én fase til en annen. Overgangen kan drives av endringer i temperatur, trykk eller sammensetning.
:::


Faseoverganger kan opptre både i rene stoffer og i blandinger. To eksempler er strukturelle overganger og overganger i blandinger.


::::{grid}
:gutter: 2

:::{grid-item-card} Strukturelle faseoverganger
Et stoff endrer struktur uten at den kjemiske sammensetningen endres. For grunstoffer kalles dette **allotropi**, mens for forbindelser kalles dette **polymorfi**.


*Eksempel: Rent, fast karbon kan finnes i flere former - f.eks. grafitt og diamant.*
:::

:::{grid-item-card} Faseoverganger for blandinger
Faseoverganger som involverer mer enn én forbindelse, og som ofte er litt mer kompliserte enn strukturelle faseoverganger.

*Eksempel: Legeringen av kobber og nikkel - de to metallene har samme krystallstruktur og er fullstendig blandbare i fast tilstand der de danner en fast løsning uansett blandingsforhold. En slik legering smelter ikke ved én enkelt temperatur, men over et intervall: ved en gitt sammensetning begynner den å smelte ved solidustemperaturen, og er først fullstendig flytende ved den høyere liquidustemperaturen. I mellom eksisterer fast og flytende fase side om side i likevekt.*
:::
::::

## Kjemisk potensial

Vi introduserte kjemisk potensial i [kapitlet om termodynamiske potensialer](https://andreasdh.github.io/Bifrost/docs/6_termodynamiske_potensialer.html#kjemisk-potensial). Her skal vi ikke definere begrepet på nytt, men bruke det til å avgjøre hvilken fase som er stabil.

:::{admonition} Husk: kjemisk potensial
:class: important

Ved konstant temperatur og trykk er det kjemiske potensialet til stoff $i$

$$\mu_i=\left(\frac{\partial G}{\partial n_i}\right)_{T,P,n_{j\ne i}}$$

En forskjell i kjemisk potensial gir en drivkraft for stoffoverføring. Stoff overføres i den retningen som senker systemets Gibbs-energi.
:::



:::{admonition} Underveisoppgave
:class: tip

Et stoff finnes i to faser, væske (fase 1) og gass (fase 2), ved konstant $T$ og $P$. Flytter vi en liten mengde $dn$ fra én fase til den andre, endrer Gibbs-energien seg med

$dG = (\mu_2 - \mu_1)\,dn$

når $dn$ flyttes fra fase 1 til fase 2. Systemet søker alltid mot lavest mulig Gibbs-energi.

**a)** Anta at $\mu_1 = 5.0\ \text{kJ/mol}$ (væske) og $\mu_2 = 3.0\ \text{kJ/mol}$ (gass). Vi flytter $dn = 0.010\ \text{mol}$ fra væske til gass. Regn ut endringen $dG$.

**b)** Ble $dG$ positiv eller negativ? Vil denne forflytningen skje spontant? Hvilken vei strømmer stoffet — fra væske til gass, eller motsatt?

**c)** Anta i stedet at $\mu_1 = 3.0\ \text{kJ/mol}$ og $\mu_2 = 3.0\ \text{kJ/mol}$. Regn ut $dG$ for samme forflytning. Hva betyr dette resultatet for systemet, og hvilken tilstand er systemet da i?

:::


:::{dropdown} Løsningsforslag

**a)** Vi setter inn i formelen:

$dG = (\mu_2 - \mu_1)\,dn = (3.0 - 5.0)\ \text{kJ/mol} \times 0.010\ \text{mol} = -0.020\ \text{kJ} = -20\ \text{J}$

**b)** $dG$ er negativ. Siden systemet søker mot lavest Gibbs-energi, skjer forflytningen spontant - en negativ $dG$ betyr at flyttingen senker $G$. Stoffet strømmer fra væske (fase 1) til gass (fase 2), altså fra høyt til lavt kjemisk potensial.

**c)** Nå er

$dG = (3.0 - 3.0)\ \text{kJ/mol} \times 0.010\ \text{mol} = 0$

Når $dG = 0$, endrer ikke Gibbs-energien seg om vi flytter stoff den ene eller andre veien. Det er ingen netto drivkraft, så systemet er i likevekt. Det er nettopp det vi skal se på videre.

:::



### Likevekt

For at to faser skal kunne eksistere side om side i likevekt, må det ikke finnes noen netto drivkraft for å flytte stoff mellom dem.


:::{admonition} Definisjon - Faselikevekt
:class: important

To faser $\alpha$ og $\beta$ er i **faselikevekt** når det kjemiske potensialet til hvert stoff $i$ er likt i de to fasene:

$\mu_i^\alpha = \mu_i^\beta$
:::

Det er viktig å skille mellom to spørsmål. Termodynamikken forteller hvilken fase som *bør* være stabil - den med lavest $\mu$ - men ikke hvor fort overgangen skjer. Farten er kinetikkens domene, og noen ganger er den så langsom at en ustabil fase blir værende nesten uendelig lenge.


:::{admonition} Underveisoppgave
:class: tip

For stoff $i$ er $\mu_i^\alpha$ litt større enn $\mu_i^\beta$. Hvilken vei vil stoffet overføres? Hvilken betingelse må være oppfylt dersom begge fasene skal kunne eksistere side om side i likevekt?

:::


:::{dropdown} Løsningsforslag

Stoffet overføres fra fase $\alpha$ til fase $\beta$, altså fra høyt til lavt kjemisk potensial. En slik overføring senker Gibbs-energien. Dersom begge fasene skal eksistere side om side i likevekt, må $\mu_i^\alpha=\mu_i^\beta$. For et rent stoff ved gitt temperatur og trykk vil fasen med høyest kjemisk potensial ellers forsvinne til fordel for den stabile fasen.

:::



### Endringer i kjemisk potensial med temperatur og trykk

Siden faser kan skifte når $\mu$ endres, må vi vite hvordan det kjemiske potensialet avhenger av temperatur og trykk. Begge sammenhengene følger direkte av egenskapene til Gibbs-energien.

For temperatur $T$ gjelder

$$\left(\frac{\partial \mu}{\partial T}\right)_P = -S_m$$

der $S_m$ er den molare entropien.


Fordi entropien alltid er positiv, synker det kjemiske potensialet når temperaturen øker - og stigningstallet er nettopp $-S_m$. Her ligger nøkkelen til smelting og koking: gass har høyest molar entropi, væske lavere, og fast stoff lavest. Kurvene for $\mu$ mot $T$ heller derfor brattest nedover for gass og slakest for fast stoff.


For trykk $P$ gjelder tilsvarende

$$\left(\frac{\partial \mu}{\partial p}\right)_T = V_m$$

der $V_m$ er det molare volumet.


Siden molart volum alltid er positivt, øker det kjemiske potensialet når trykket øker. For de fleste stoffer har fast fase mindre molart volum enn væske, så økt trykk hever væskens $\mu$ mer enn faststoffets - og frysepunktet stiger med trykket.


:::{admonition} Underveisoppgave
:class: tip

**a)** For vann er $S_m = 70\ \text{J/(mol K)}$ for væske og $S_m = 189\ \text{J/(mol K)}$ for gass. Bruk $\left(\frac{\partial \mu}{\partial T}\right)_P = -S_m$ til å avgjøre hvilken fase som får sitt kjemiske potensial til å synke raskest når temperaturen øker. Knytt svaret til at gassfasen har flere tilgjengelige mikrotilstander enn væskefasen.

**b)** Tegn for deg $\mu$ mot $T$ for væske og gass som to rette linjer med stigningstall $-S_m$. Hvilken fase har lavest $\mu$ ved lav temperatur, og hvilken har lavest ved høy temperatur? Hva betyr krysningspunktet mellom de to linjene, og hvorfor koker vann når det blir varmt nok? (Husk at systemet alltid velger fasen med lavest $\mu$.)

**c)** Det molare volumet til is er $V_m = 19.7\ \text{cm}^3/\text{mol}$ og til væske $V_m = 18.0\ \text{cm}^3/\text{mol}$. Bruk $\left(\frac{\partial \mu}{\partial P}\right)_T = V_m$ til å avgjøre hvilken fase som får sitt kjemiske potensial til å øke mest når trykket øker. Hva betyr det for smeltepunktet til is?

:::


:::{dropdown} Løsningsforslag

**a)** Stigningstallet til $\mu$ mot $T$ er $-S_m$, så en større $S_m$ gir et brattere fall. Gass har størst molar entropi ($189$ mot $70$), så gassens kjemiske potensial synker raskest når temperaturen øker. Det henger sammen med at molekylene i gassfasen kan fordele seg over langt flere tilgjengelige mikrotilstander enn i væskefasen.

**b)** Ved lav temperatur ligger væskens $\mu$ lavest, så væske er den stabile fasen (systemet velger alltid lavest $\mu$). Fordi gasslinjen faller brattere, synker den under væskelinjen ved høy nok temperatur, og etter kryssingen har gass lavest $\mu$. Krysningspunktet er kokepunktet: der er $\mu$ lik for de to fasene, så de er i likevekt, og akkurat her skifter den stabile fasen fra væske til gass. Over denne temperaturen gir gassfasen lavest Gibbs-energi, og derfor koker vannet.

**c)** Stigningstallet til $\mu$ mot $P$ er $V_m$, så fasen med størst molart volum øker mest når trykket øker. Is har større molart volum enn væske ($19.7$ mot $18.0$), så isens $\mu$ øker mest. Det gjør is mindre stabil enn væske ved økt trykk, så smeltepunktet synker når trykket øker. Dette er spesielt for vann, siden is uvanlig nok har større volum enn væske - for de fleste andre stoffer stiger smeltepunktet med trykket.

:::



### Metastabil fase

Noen ganger er overgangen til den mest stabile fasen kinetisk hindret. Stoffet kan da bli værende i en fase med høyere kjemisk potensial. Slike faser kalles **metastabile faser**.


:::{admonition} Definisjon - Metastabil fase
:class: important

En **metastabil fase** er en fase som har høyere kjemisk potensial enn en annen tilgjengelig fase, men som likevel består fordi overgangen er kinetisk hindret. 
:::

Diamant er det klassiske eksempelet. Ved romtemperatur og normalt trykk har grafitt lavere kjemisk potensial enn diamant, så termodynamisk *burde* diamant omdannes til grafitt. At det ikke skjer, skyldes at omorganiseringen av karbonatomene krever en enorm aktiveringsenergi - overgangen er så langsom at den i praksis aldri finner sted. Diamant er altså termodynamisk ustabil, men kinetisk fastlåst.


Vi lar kjemisk potensial ligge for nå. Senere bruker vi det samme begrepet til å beskrive blandinger og kjemiske reaksjoner.

:::{admonition} Viktig presisering
:class: note

At trykk senker smeltepunktet til is, brukes noen ganger som hele forklaringen på både skøyting og isbrebevegelse. Trykksmelting kan bidra, men friksjonsvarme er også viktig for skøyter, og isbreer beveger seg hovedsakelig ved deformasjon av isen og glidning mot underlaget.
:::


## Fasediagrammer

### Hva er et fasediagram?

Som kjemikere er vi interessert i ulike egenskaper ved faseoverganger fordi de avgjør når og hvordan et stoff endrer seg. Det er viktig i alt fra separasjon til syntese og analyse. Derfor er **fasediagrammer** nyttige verktøy når vi studerer faser. For et rent stoff bruker vi ofte et $p$–$T$-diagram med temperatur på x-aksen og trykk på y-aksen. For blandinger kan en av aksene i stedet vise sammensetningen.



### Viktige egenskaper ved fasediagrammer

Det er noen spesielle trekk ved fasediagrammer som er verdt å merke seg:

* **Koeksistenslinjer** - en linje mellom to faser A og B, der fasene er i likevekt med hverandre

* **Trippelpunkt** - et punkt mellom tre faser A, B og C, der fasene er i likevekt med hverandre

* **Kritisk punkt** - punktet der skillet mellom væske og gass forsvinner


Et omtrentelig fasediagram for vann er vist nedenfor, med eksempel på koekeistenslinje, trippelpunkt og kritisk punkt.

![Fasediagram vann](bilder/fasediagram_vann.svg)

Det er også noen spesielle temperaturer og trykk man bør vite om, som man ofte møter på på laboratoriet:

| Navn | Symbol | Definisjon |
| ---- | ------ | ---------- |
| Overgangstemperatur | $T_{trs}$ | Temperaturen for når en faseovergang skjer, gitt et bestemt trykk. |
| Damptrykk | $P*$ | Trykket dampen utøver i likevekt med den kondenserte fasen, ved en gitt temperatur. |
| Kokepunkt | $T_b$ | Temperaturen der væskens damptrykk er lik det ytre trykket. |
| Smeltepunkt | $T_m$ | Temperaturen der fast og flytende fase er i likevekt, ved et gitt trykk. |
| Kritisk temperatur | $T_c$ | Den høyeste temperaturen der væske og gass kan eksistere som to atskilte faser. |

:::{admonition} Viktig
:class: note
Temperaturene beskrevet i tabellen ovenfor er som regel forskjellig for ulike trykk. Visste du for eksempel at temperaturen for når vann koker er lavere på fjellet enn i Oslo? Det er fordi lufttrykket er lavere jo høyere opp vi kommer, og i gjengjeld er koketemperaturen lavere. Derfor må vi spesifisere trykket vi snakker om for overgangstemperaturer.
:::

Fordamping er faseovergangen fra væske til gass. **Koking** er fordamping i hele væsken og skjer når damptrykket er lik det ytre trykket. Flere vanlige navn på faseoverganger er oppsummert i tabellen under:

| Fra/til | Gass | Væske | Fast stoff | Plasma |
| ------- | ---- | ----- | ---------- | ------ |
| **Gass** | | Kondensasjon | Deposisjon | Ionisering | 
| **Væske** | Fordamping | | Frysing | |
| **Fast stoff** | Sublimasjon | Smelting | | | 
| **Plasma** | Deionisering | | | |

:::{admonition} Underveisoppgave
:class: tip

Diamant "burde" termodynamisk bli til grafitt ved romtemperatur, men gjør det ikke. Forklar forskjellen mellom å si at en fase er *termodynamisk ustabil* og *kinetisk fastlåst*. Hvilken av de to sier noe om hva som *skjer*, og hvilken om hvor *fort* det skjer?
:::

:::{dropdown} Løsningsforslag

*Termodynamisk ustabil* betyr at fasen har høyere Gibbs-energi enn en annen fase, så den *burde* gå over til den - det sier hva som skjer (retningen). *Kinetisk fastlåst* betyr at omdanningen går uendelig sakte fordi en høy energibarriere står i veien - det sier hvor fort det skjer.

Diamant er termodynamisk ustabil (grafitt har lavere $\mu$), men kinetisk fastlåst: bindingene som må brytes gir en så høy barriere at omdanningen praktisk talt ikke skjer ved romtemperatur.

:::

## Faseregler

Vi har hittil sett at fasediagrammet fungerer som et "kart" over *hvilke* faser som opptrer. Vi skal nå se at **fasereglene** svarer på *hvor mange* av disse fasene som kan eksistere samtidig, og *hvor fritt* vi kan endre trykk og temperatur uten at fasene endrer seg.


### Frihetsgrader, konstituenter og komponenter

Før vi begynner å snakke om antallet faser som eksisterer samtidig, må vi ha noen begreper. Vi introduserer **frihetsgrader**, **konsitutenter** og **komponenter**. NB! Dette er ikke frihetsgradene du er kjent med fra tidligere.


:::{admonition} Definisjon - Frihetsgrader (varians)
:class: important

**Frihetsgradene** $F$ er antall intensive variabler - som trykk og temperatur - vi kan endre uavhengig av hverandre uten å forandre antall faser i likevekt.
:::


:::{admonition} Definisjon - Konstituent
:class: important

En **konstituent** er en hvilken som helst kjemisk art som er til stede i systemet.
:::


:::{admonition} Definisjon - Komponent
:class: important

En **komponent** er en kjemisk uavhengig konstituent. **Antall komponenter** $C$ er det minste antallet uavhengige arter som trengs for å angi sammensetningen av alle fasene i systemet.
:::



### Gibbs' faseregel

Gibbs' faseregel gir oss et enkelt regnestykke som knytter sammen antall komponenter i systemet, antall faser som er i likevekt, og antall frihetsgrader vi står igjen med - altså hvor mange av de [intensive variablene](https://andreasdh.github.io/Bifrost/docs/A1_grunnleggende_begreper.html#intensive-og-ekstensive-storrelser) som trykk og temperatur vi fritt kan endre uten å endre antall faser.

:::{admonition} Gibbs' faseregel
:class: important

For et system i likevekt gjelder

$$F = C - P + 2$$

der $F$ er antall frihetsgrader, $C$ antall komponenter og $P$ antall faser.

For et rent stoff er $C = 1$, og faseregelen forenkles til

$$F = 3 - P$$
:::


Dette forklarer nøyaktig strukturen i fasediagrammet:

* **Én fase** ($P=1$) gir $F=2$. Vi kan endre både trykk og temperatur fritt - fasen dekker et *område*.
* **To faser i likevekt** ($P=2$) gir $F=1$. Nå kan bare én variabel velges fritt; den andre følger med. Likevekten er en *linje*, en fasegrense.
* **Tre faser i likevekt** ($P=3$) gir $F=0$. Ingenting kan endres - dette skjer bare i ett eneste *punkt*, trippelpunktet.


:::{card} 
**Eksempel**
^^^
Vi bruker Gibbs' faseregel på rent vann ($C=1$) i tre ulike situasjoner i fasediagrammet.

**Vann som bare væske.** Her er én fase til stede, $P=1$, så

$F = C - P + 2 = 1 - 1 + 2 = 2$

Vi har to frihetsgrader: både trykk og temperatur kan endres fritt uten at vi forlater væskeområdet. Det er derfor væske dekker en hel *flate* i diagrammet.

**Vann som koker.** Nå er væske og damp i likevekt, altså $P=2$, og

$F = 1 - 2 + 2 = 1$

Bare én frihetsgrad. Velger vi temperaturen, er trykket samtidig låst - vi kan ikke endre begge uten å forlate koeksistensen. Derfor er væske-damp-likevekten en *linje* i diagrammet.

**Vann i trippelpunktet.** Her er alle tre faser i likevekt, $P=3$, så

$F = 1 - 3 + 2 = 0$

Ingen frihetsgrader. Verken trykk eller temperatur kan endres uten at én av fasene forsvinner. Trippelpunktet er derfor et enkelt *punkt*, fastlåst til én bestemt kombinasjon av trykk og temperatur - for vann er det $273{.}16\,\text{K}$ og $611\,\text{Pa}$.
:::



:::{admonition} Underveisoppgave
:class: tip

Bruk faseregelen på rent vann til å forklare hvorfor trippelpunktet er et *punkt*, fasegrensene er *linjer*, og faseområdene er *flater* i fasediagrammet. Hvorfor kan du ikke ha fire faser av et rent stoff i likevekt samtidig?

:::

:::{dropdown} Løsningsforslag

Faseregelen er $F = C - P + 2$, der $F$ er antall frihetsgrader, $C$ antall komponenter og $P$ antall faser. For rent vann er $C = 1$, så $F = 3 - P$.

- **Ett faseområde** ($P = 1$): $F = 2$. Vi kan variere både $T$ og $P$ fritt uten å forlate fasen - det gir en *flate* i diagrammet.
- **Fasegrense** ($P = 2$): $F = 1$. Nå henger $T$ og $P$ sammen: velger vi temperaturen, er trykket bestemt. Det ene frie valget gir en *linje*.
- **Trippelpunkt** ($P = 3$): $F = 0$. Alt er låst - både $T$ og $P$ har bestemte verdier der tre faser møtes. Ingen frihet gir et *punkt*.

Fire faser samtidig ville krevd $P = 4$, altså $F = 3 - 4 = -1$. Et negativt antall frihetsgrader er umulig, så et rent stoff kan ikke ha fire faser i likevekt på én gang.

:::

### Vektstangregelen

Et fasediagram forteller oss *hvilke* faser som er i likevekt ved en gitt temperatur og sammensetning. Men når systemet befinner seg i et tofaseområde, vil vi ofte også vite *hvor mye* det er av hver fase. Til det bruker vi **vektstangregelen**.

Se for deg en blanding av to komponenter A og B. Vi kan ha to faser, fast fase eller væskefase. Til tross for at det bare er to faser, har vi to énfaseområder: helt fast (S) eller helt flytende (L), men også et tofaseområde av en blanding (L+S). I tofaseområdet kan L og S eksistere side om side, for eksempel ved utkrystallisering på laboratoriet. Nøkkelspørsmålet vårt er:

_Når systemet ligger i tofaseområdet ved en gitt temperatur $T$ , hvor mye har vi av hver fase?_

Dette er svært relevant, spesielt hvis vi skal utkrystallisere så mye så mulig fast stoff. 

Fasediagrammet til systemet er illustrert nedenfor, og har med følgende:

* $x$-aksen : molfraksjonen til B $x$. Denne går fra $0$ til $1$, der $0$ betyr 0% B og 100% A, mens $1$ betyr 100% B og 0% A

* $y$-aksen : temperaturen $T$

* $x_{eks}$ : molfraksjonen (til B) som vi har i eksperimentet vårt

* $T_{eks}$ : temperaturen som vi har i eksperimentet vårt

* $x_l$ : molfraksjonen (til B) i væskefasen. Denne leser vi av der bindelinjen treffer væskekurven (venstre skjæringspunkt)

* $x_s$ : molfraksjonen (til B) i den faste fasen. Denne leser vi av der bindelinjen treffer fastkurven (høyre skjæringspunkt)


![Vektstangregelen](bilder/lever_rule_diagram.svg)



#### Utledning av vekststangregelen

Merk at vi ofte befinner oss i en situasjon på laboratoriet der vi vet $x_{eks}$ fordi vi har veid inn stoffmengdene til A og B, og derfor vet molfraksjonen til B, og at vi vet $T_{eks}$ fordi det er temperaturen vi jobber i. 

Det vi da er interessert i å finne ut er $x_l$ og $x_s$. La oss se hvordan:

Vi vet at stoffmengden må være bevart. La $n_s$ og $n_l$ være stoffmengdene i de to fasene, slik at 

$$n = n_s + n_l$$ 

er den totale mengden.

Mengden av komponent B må være den samme enten vi teller den i systemet som helhet, eller summerer bidraget fra hver fase:

$$n\,x_{eks} = n_s\,x_s + n_l\,x_l$$

Vi skriver $n = n_s + n_l$ på venstre side:

$$(n_s + n_l)\,x_{eks} = n_s\,x_s + n_l\,x_l$$

Vi ganger ut og samler leddene med $n_s$ på den ene siden og $n_l$ på den andre:

$$n_s\,x_{eks} - n_s\,x_s = n_l\,x_l - n_l\,x_{eks}$$

$$n_s\,(x_{eks} - x_s) = n_l\,(x_l - x_{eks})$$

Avstandene fra systempunktet til hver av fasegrensene er armene $l_s = x_{eks} - x_s$ og $l_l = x_l - x_{eks}$. Dermed får vi **vektstangregelen**:

$$n_s\,l_s = n_l\,l_l$$


![Vektstangregelen](bilder/lever_rule_diagram_2.svg)


:::{admonition} Vektstangregelen
:class: important

Der avstandene fra systempunktet til hver av fasegrensene er armene $l_s = x_{eks} - x_s$ og $l_l = x_l - x_{eks}$, gir **vektstangregelen** oss

$$n_s\,l_s = n_l\,l_l$$

:::


Dette gir en enkel huskeregel: **jo nærmere systempunktet ligger en fasegrense, desto mer er det av den fasen.** Ligger systempunktet nær væskegrensen, er armen til væsken kort, og da må mengden væske være stor for at balansen skal holde. Fasen du er nærmest, er det altså mest av - akkurat motsatt av hva armlengden alene kan antyde.

Løser vi for andelene, får vi

$$\frac{n_s}{n} = \frac{l_l}{l_s + l_l}, \qquad \frac{n_l}{n} = \frac{l_s}{l_s + l_l}$$

Legg merke til at telleren for hver fase er den *motsatte* armen.


:::{card} Eksempler

Vi har laget en blanding av A og B med samlet sammensetning $x_{eks} = 0.60$ (molfraksjon B), og holder den ved en temperatur $T_{eks}$ der systemet ligger i tofaseområdet. Fra fasediagrammet leser vi av bindelinjens endepunkter: væsken har $x_l = 0.40$ og den faste fasen har $x_s = 0.75$.

**Steg 1 — regn ut armene.** Armene er avstandene fra systempunktet til hver fasegrense:

$$l_l = x_{eks} - x_l = 0.60 - 0.40 = 0.20$$

$$l_s = x_s - x_{eks} = 0.75 - 0.60 = 0.15$$

**Steg 2 — bruk vektstangregelen.** Andelen av hver fase er den *motsatte* armen delt på hele bindelinjen, $l_l + l_s = 0.35$:

$$\frac{n_l}{n} = \frac{l_s}{l_l + l_s} = \frac{0.15}{0.35} = 0.43$$

$$\frac{n_s}{n} = \frac{l_l}{l_l + l_s} = \frac{0.20}{0.35} = 0.57$$

Systemet er altså omtrent 43 % væske og 57 % fast fase.

**Sjekk:** Systempunktet ligger nærmest den faste grensen ($x_s = 0.75$), og det er nettopp fast fase det er mest av (57 %). Den korteste armen hører til den fasen det er mest av.

:::

## Fasegrenser

Fasegrensene er linjene i fasediagrammet der to faser møtes og er i likevekt, og siden likevekt betyr likt kjemisk potensial i begge faser, kan vi bruke nettopp denne betingelsen til å regne ut hvor bratt en fasegrense løper - altså hvordan overgangstemperaturen endrer seg når vi endrer trykket. 


### Clapeyrons likning

**Clapeyrons likning** gir oss stigningstallet til en hvilken som helst fasegrense, altså nøyaktig hvor mye overgangstemperaturen endrer seg når trykket endres. Det anbefales sterkt å se på utledningen for likningen, gitt i boksen under formelen.


:::{admonition} Clapeyrons likning
:class: important

Langs en hvilken som helst fasegrense gjelder

$$\frac{dP}{dT} = \frac{\Delta_{trs}S}{\Delta_{trs}V}$$

der $\Delta_{trs}S$ og $\Delta_{trs}V$ er endringen i molar entropi og molart volum ved overgangen.
:::


:::{admonition} Utledning av Clapeyrons likning
:class: important, dropdown

**Antakelse 1: Vi beveger oss langs en fasegrense.** To faser $\alpha$ og $\beta$ er i likevekt, så de har likt kjemisk potensial:

$\mu_\alpha = \mu_\beta$

**Antakelse 2: Vi holder oss på grensen mens vi endrer $T$ og $p$ litt.** Beveger vi oss et lite stykke langs grensen, må potensialene fortsatt være like etterpå. Endringen i $\mu$ må derfor være den samme for begge faser:

$d\mu_\alpha = d\mu_\beta$

**Vi bruker hvordan $\mu$ endrer seg.** For et rent stoff er $\mu = G_m$, og fra $dG_m = -S_m\,dT + V_m\,dp$ har vi for hver fase

$d\mu = -S_m\,dT + V_m\,dp$

Setter vi dette inn på begge sider av $d\mu_\alpha = d\mu_\beta$:

$-S_{\alpha}\,dT + V_{\alpha}\,dp = -S_{\beta}\,dT + V_{\beta}\,dp$

**Vi samler leddene.** Flytter vi $dT$-leddene til én side og $dp$-leddene til den andre:

$(V_{\beta} - V_{\alpha})\,dp = (S_{\beta} - S_{\alpha})\,dT$

Med $\Delta_{trs}V = V_\beta - V_\alpha$ og $\Delta_{trs}S = S_\beta - S_\alpha$ gir dette

$\frac{dp}{dT} = \frac{\Delta_{trs}S}{\Delta_{trs}V}$

**Antakelse 3: Overgangen skjer ved likevekt.** Da er $\Delta_{trs}G = 0$, og siden $\Delta G = \Delta H - T\Delta S$ følger det at $\Delta_{trs}S = \Delta_{trs}H / T_{trs}$. Setter vi inn dette, får vi den formen som er lettest å bruke i praksis:

$\frac{dp}{dT} = \frac{\Delta_{trs}H}{T_{trs}\,\Delta_{trs}V}$

Merk at denne utledningen er **eksakt** - vi har ikke gjort noen tilnærminger, bare brukt at potensialene er like langs grensen. Det er først når vi går videre til Clausius-Clapeyron for væske-damp-grensen at vi antar ideell gass og neglisjerbart væskevolum.
:::


For **fast-væske-grensen** setter vi inn at $\Delta_{fus}S = \Delta_{fus}H / T$, siden overgangen skjer ved likevekt. Det gir

$$\frac{dP}{dT} = \frac{\Delta_{fus}H}{T\,\Delta_{fus}V}$$

[Smelteentalpien](https://andreasdh.github.io/Bifrost/docs/5_varme_arbeid_entalpi.html#entalpi-nyttig-ved-konstant-trykk) er stor og volumendringen ved smelting liten, så denne grensen er svært bratt - det skal enorme trykkendringer til for å flytte smeltepunktet merkbart. For de fleste stoffer er $\Delta_{fus}V > 0$ og grensen heller forover; for vann er $\Delta_{fus}V < 0$, og grensen heller bakover.


:::{admonition} Underveisoppgave
:class: tip

Vi ser på smeltegrensen mellom is og flytende vann. Ved smeltepunktet $T = 273.15\ \text{K}$ er smelteentalpien $\Delta_{fus}H = 6.01\ \text{kJ/mol}$, og de molare volumene er $V_m = 19.7\ \text{cm}^3/\text{mol}$ for is og $V_m = 18.0\ \text{cm}^3/\text{mol}$ for væske.

**a)** Entropiendringen ved smelting er $\Delta_{fus}S = \frac{\Delta_{fus}H}{T}$. Regn ut $\Delta_{fus}S$.

**b)** Regn ut volumendringen $\Delta_{fus}V = V_{m,\text{væske}} - V_{m,\text{is}}$. Hvilket fortegn får den, og hva sier det om vann sammenlignet med de fleste andre stoffer?

**c)** Sett inn i Clapeyrons likning $\frac{dP}{dT} = \frac{\Delta_{fus}S}{\Delta_{fus}V}$ og regn ut stigningstallet til smeltegrensen. Hva blir fortegnet, og hva betyr det for hvordan smeltepunktet endrer seg når trykket øker?

:::


:::{dropdown} Løsningsforslag

**a)** Vi gjør om entalpien: $\Delta_{fus}H = 6.01\ \text{kJ/mol} = 6010\ \text{J/mol}$. Da blir

$\Delta_{fus}S = \frac{6010\ \text{J/mol}}{273.15\ \text{K}} = 22.0\ \text{J/(mol K)}.$

Entropiendringen er positiv: Den flytende fasen har flere tilgjengelige mikrotilstander enn isen.

**b)** Vi gjør om volumene: $19.7\ \text{cm}^3/\text{mol} = 19.7 \times 10^{-6}\ \text{m}^3/\text{mol}$ og $18.0\ \text{cm}^3/\text{mol} = 18.0 \times 10^{-6}\ \text{m}^3/\text{mol}$. Da blir

$\Delta_{fus}V = (18.0 - 19.7) \times 10^{-6}\ \text{m}^3/\text{mol} = -1.7 \times 10^{-6}\ \text{m}^3/\text{mol}.$

Fortegnet er negativt: væske har mindre molart volum enn is. Det er uvanlig - for de fleste stoffer krymper stoffet når det størkner, men vann utvider seg når det fryser (derfor flyter is).

**c)** Vi setter inn:

$\frac{dP}{dT} = \frac{22.0\ \text{J/(mol K)}}{-1.7 \times 10^{-6}\ \text{m}^3/\text{mol}} = -1.3 \times 10^{7}\ \text{Pa/K}.$

Stigningstallet er negativt. Det betyr at smeltegrensen heller bakover: når trykket øker, synker smeltepunktet. Øker vi trykket, smelter altså isen ved en lavere temperatur - motsatt av de fleste stoffer, og en direkte følge av at $\Delta_{fus}V$ er negativ for vann.

:::




### Clausius-Clapeyrons likning

Clausius–Clapeyrons likning er en forenklet form av Clapeyrons likning for **væske–damp- og faststoff–damp-grensene**, der vi antar at dampen oppfører seg som en ideell gass og at den kondenserte fasens volum er neglisjerbart sammenlignet med dampens. Også utledningen til denne likningen ligger under formelen, se gjerne på den!


:::{admonition} Clausius–Clapeyrons likning
:class: important


For væske-damp-grensen gjelder

$$\frac{d\ln P}{dT} = \frac{\Delta_{vap}H}{RT^2}$$

Antar vi at $\Delta_{vap}H$ er konstant over temperaturintervallet, kan vi integrere og få

$P = P^* e^{-\chi}, \qquad \chi = \frac{\Delta_{vap}H}{R}\left(\frac{1}{T} - \frac{1}{T^*}\right)$

der $P^*$ er damptrykket ved en referansetemperatur $T^*$.
:::


:::{admonition} Utledning av Clausius–Clapeyrons likning
:class: note, dropdown

Vi tar utgangspunkt i Clapeyrons likning for væske-damp-grensen:

$\frac{dp}{dT} = \frac{\Delta_{vap}H}{T\,\Delta_{vap}V}$

**Antakelse 1: Væskens volum er neglisjerbart.** Damp har langt større molart volum enn væske - typisk rundt tusen ganger. Volumendringen ved fordamping er derfor tilnærmet lik dampens volum alene:

$\Delta_{vap}V = V_m(g) - V_m(l) \approx V_m(g)$

**Antakelse 2: Dampen er en ideell gass.** Da er $V_m(g) = RT/p$, og innsatt gir det

$\frac{dp}{dT} = \frac{\Delta_{vap}H}{T \cdot RT/p} = \frac{p\,\Delta_{vap}H}{RT^2}$

**Vi skriver om til logaritmisk form.** Deler vi begge sider på $p$ og bruker at $\frac{1}{p}\frac{dp}{dT} = \frac{d\ln p}{dT}$, får vi den kompakte formen

$\frac{d\ln p}{dT} = \frac{\Delta_{vap}H}{RT^2}$

**Antakelse 3: Fordampningsentalpien er konstant.** Over et ikke for stort temperaturintervall varierer $\Delta_{vap}H$ lite, så vi behandler den som konstant og integrerer fra en referansetilstand $(T^*, p^*)$ til $(T, p)$:

$\int_{p^*}^{p} d\ln p = \frac{\Delta_{vap}H}{R}\int_{T^*}^{T} \frac{dT}{T^2}$

Venstre side gir $\ln(p/p^*)$, og høyre side gir $-\frac{\Delta_{vap}H}{R}\left(\frac{1}{T} - \frac{1}{T^*}\right)$, siden den antideriverte av $1/T^2$ er $-1/T$. Altså

$\ln\frac{p}{p^*} = -\frac{\Delta_{vap}H}{R}\left(\frac{1}{T} - \frac{1}{T^*}\right)$

**Til slutt løser vi for $p$** ved å opphøye $e$ på begge sider:

$p = p^* e^{-\chi}, \qquad \chi = \frac{\Delta_{vap}H}{R}\left(\frac{1}{T} - \frac{1}{T^*}\right)$

Mens Clapeyrons likning var eksakt, hviler denne på tre tilnærminger: neglisjerbart væskevolum, ideell gass og konstant fordampningsentalpi. Alle tre er som regel gode et godt stykke under det kritiske punktet, men svikter når man nærmer seg det, der dampen blir tett og langt fra ideell.
:::

### Clapeyron og Clausius-Clapeyron likningene i praksis

Både Clapeyron og Clausius-Clapeyron er direkte måleverktøy, ikke bare teori.

For **væske-damp-grensen** er Clausius-Clapeyrons likning på logaritmisk form en rett linje: $\ln p = -\frac{\Delta_{vap}H}{R}\cdot\frac{1}{T} + \text{konstant}$. Måler du damptrykket ved noen få temperaturer og plotter $\ln p$ mot $\frac{1}{T}$, får du en linje med stigningstall $-\frac{\Delta_{vap}H}{R}$, og kan lese av **fordampningsentalpien** direkte - uten å måle varme. Ofte er trykket dessuten allerede kjent, for eksempel når du måler ved atmosfæretrykk, slik at du bare trenger å registrere temperaturene.

For **fast-væske-grensen** brukes Clapeyrons likning på samme måte. Måler du hvordan smeltepunktet forskyver seg når trykket endres, altså stigningstallet $\frac{dp}{dT}$ langs grensen, kan du regne ut **smelteentalpien** $\Delta_{fus}H$ - forutsatt at du kjenner volumendringen ved smelting.

:::{admonition} Underveisoppgaver
:class: tip

Plotter du ln $P$ mot $1/T$ for et fordampende stoff, får du en rett linje. Hva forteller stigningstallet deg om stoffet - og hvorfor er dette en smart måte å måle fordampningsentalpi på, uten å måle varme direkte?

:::

:::{dropdown} Løsningsforslag

Fra den integrerte Clausius–Clapeyron-likningen har vi

$$\ln P=-\frac{\Delta_{vap}H}{R}\frac{1}{T}+\text{konstant}$$

Stigningstallet er derfor $-\Delta_{vap}H/R$. Når vi har funnet stigningstallet fra måledata, kan vi beregne fordampningsentalpien uten å måle varme direkte.
:::


:::{card} 
**Eksempel**
^^^
Is og vann er i likevekt ved $273{.}15\,\text{K}$ når trykket er $1{.}0\,\text{bar}$. Øker vi trykket til $136\,\text{bar}$, synker smeltepunktet til $272{.}15\,\text{K}$. Molart volum er $18{.}02\,\text{cm}^3/\text{mol}$ for vann og $19{.}65\,\text{cm}^3/\text{mol}$ for is. Finn smelteentalpien $\Delta_{fus}H$.

Først finner vi stigningstallet til fast-væske-grensen fra de to punktene. Trykkendringen er $\Delta p = 135\,\text{bar} = 1{.}35 \times 10^7\,\text{Pa}$ og temperaturendringen $\Delta T = -1{.}0\,\text{K}$, så

$\frac{dp}{dT} \approx \frac{\Delta p}{\Delta T} = \frac{1{.}35 \times 10^7}{-1{.}0} = -1{.}35 \times 10^7\,\text{Pa/K}$

Så løser vi Clapeyrons likning for smelteentalpien:

$\frac{dp}{dT} = \frac{\Delta_{fus}H}{T\,\Delta_{fus}V} \quad \Rightarrow \quad \Delta_{fus}H = \frac{dp}{dT}\cdot T \cdot \Delta_{fus}V$

Volumendringen ved smelting er

$\Delta_{fus}V = V_m(\text{væske}) - V_m(\text{is}) = 18{.}02 - 19{.}65 = -1{.}63\,\text{cm}^3/\text{mol} = -1{.}63 \times 10^{-6}\,\text{m}^3/\text{mol}$

Legg merke til at volumendringen er **negativ** - is har større volum enn vann. Vi setter inn:

$\Delta_{fus}H = (-1{.}35 \times 10^7) \cdot 273{.}15 \cdot (-1{.}63 \times 10^{-6}) \approx 6{.}0 \times 10^3\,\text{J/mol}$

Smelteentalpien er altså $\Delta_{fus}H \approx 6{.}0\,\text{kJ/mol}$, som stemmer godt med den kjente verdien for vann.

Merk at de to negative fortegnene - fra stigningstallet og fra volumendringen - opphever hverandre, slik at smelteentalpien blir positiv. Det må den også være, siden smelting alltid krever tilført energi.
:::


:::{admonition} Underveisoppgaver
:class: tip

Clausius–Clapeyron bygger på tre antakelser: at væskens volum er neglisjerbart, at dampen er en ideell gass, og at $\Delta_{vap}H$ er konstant. Nær det kritiske punktet svikter alle tre. Hvorfor svikter de akkurat der? Hva skjer med forskjellen mellom væske og damp?

:::

:::{dropdown} Løsningsforslag

Nær det kritiske punktet blir dampen tett, mens væsken utvider seg. Forskjellen i tetthet og molart volum mellom fasene blir stadig mindre. Dampens volum er derfor ikke lenger mye større enn væskens, og dampen oppfører seg heller ikke som en ideell gass. Samtidig går fordampningsentalpien mot null når skillet mellom væske og gass forsvinner.

## Førsteordens og kontinuerlige faseoverganger

Alle faseovergangene vi har sett på så langt - smelting, koking og sublimasjon - krever latent varme. Men ikke alle faseoverganger oppfører seg slik. For å forklare forskjellen ser vi på hvordan det kjemiske potensialet og de deriverte av det endrer seg gjennom overgangen. Den klassiske inndelingen skyldes Ehrenfest og bygger på sammenhengene $\left(\frac{\partial \mu}{\partial T}\right)_p = -S_m$ og $\left(\frac{\partial \mu}{\partial p}\right)_T = V_m$.


### Førsteordens faseoverganger

:::{admonition} Definisjon - Førsteordens faseovergang
:class: important

En **førsteordens faseovergang** er en overgang der de førstederiverte av det kjemiske potensialet - molar entropi $S_m$ og molart volum $V_m$ - endrer seg diskontinuerlig, altså med et sprang, ved overgangstemperaturen.
:::

Fordi entropien hopper, er det knyttet en **latent varme** til overgangen: $\Delta_{trs}H = T_{trs}\,\Delta_{trs}S \neq 0$. Det er denne varmen som tilføres uten at temperaturen stiger mens overgangen pågår - nettopp den pausen vi bruker til å oppdage overgangen ved termisk analyse.

Smelting, fordamping og sublimasjon er alle førsteordens faseoverganger. Entropien og volumet endrer seg brått: is og vann har målbart forskjellig entropi og tetthet. Det er nettopp fordi $\Delta_{trs}S$ og $\Delta_{trs}V$ er forskjellige fra null at Clapeyrons likning kan beskrive disse grensene.


### Kontinuerlige faseoverganger

:::{admonition} Definisjon - Kontinuerlig faseovergang
:class: important

En **kontinuerlig faseovergang** er en overgang uten latent varme og uten et sprang i molar entropi eller molart volum. I Ehrenfests klassiske inndeling ble slike overganger omtalt som andreordens når en andrederivert, som varmekapasiteten, var diskontinuerlig.
:::

Siden entropien er kontinuerlig, er det **ingen latent varme** ved en kontinuerlig overgang, og heller ingen brå volumendring. Varmekapasiteten kan gjøre et sprang eller stige skarpt mot en topp.

Slike overganger er mer subtile enn smelting og koking. Eksempler er overgangen til superledende tilstand i visse metaller og overgangen til superflytende helium. Her ligger informasjonen ikke i en latent varme, men i hvordan varmekapasiteten oppfører seg - den gjør et sprang, eller stiger skarpt mot en topp, ved overgangen.


Vi kan oppsummere forskjellen slik:

* **Første orden:** $\mu$ er kontinuerlig ved likevekt, men den førstederiverte har et sprang. Entropi og volum endres brått, og overgangen har latent varme.
* **Kontinuerlig overgang:** $\mu$, entropien og volumet er kontinuerlige. Det finnes ingen latent varme, men en andrederivert som varmekapasiteten kan endre seg kraftig.

:::{admonition} Underveisoppgaver
:class: tip

Forklar forskjellen mellom en førsteordens og en kontinuerlig faseovergang. Knytt svaret til at den molare entropien $S_m$ har et sprang ved den ene overgangen, men er kontinuerlig ved den andre.

:::

:::{dropdown} Løsningsforslag

Ved en førsteordens overgang endres den molare entropien med et sprang. Da er $\Delta_{trs}S\ne0$, og overgangen krever latent varme: $\Delta_{trs}H=T_{trs}\Delta_{trs}S$. Energien som tilføres, går til faseovergangen mens temperaturen står stille. Ved en kontinuerlig overgang er entropien kontinuerlig, slik at det ikke finnes noen latent varme og dermed heller ingen tilsvarende temperaturpause.
:::

## Oppsummering

- En fase er en homogen del av et system med ensartede egenskaper.
- Ved faselikevekt har stoffet samme kjemiske potensial i fasene som sameksisterer.
- Et fasediagram viser hvilke faser som er stabile ved ulike temperaturer og trykk.
- Clapeyrons likning beskriver stigningstallet til en fasegrense. Clausius–Clapeyrons likning er en tilnærming for grenser mot dampfasen.
- Førsteordens faseoverganger har latent varme. Kontinuerlige faseoverganger har ikke latent varme.

## Oppgaver

**1. Faselikevekt**

Flytende vann og vanndamp er i likevekt. Hva kan du si om det kjemiske potensialet til vann i de to fasene? Hva skjer dersom de ikke er like?

:::{dropdown} Løsningsforslag

Ved likevekt er $\mu_{\mathrm{vann(l)}}=\mu_{\mathrm{vann(g)}}$. Hvis de ikke er like, vil vann gå fra fasen med høyest kjemisk potensial til fasen med lavest kjemisk potensial til likevekt er nådd eller én fase forsvinner.
:::

**2. Fasediagram**

Forklar forskjellen mellom et trippelpunkt og et kritisk punkt.

:::{dropdown} Løsningsforslag

I trippelpunktet er tre faser i likevekt samtidig. I det kritiske punktet ender væske–damp-grensen, og forskjellen mellom væske og gass forsvinner.
:::

**3. Clapeyrons likning**

Et stoff har positiv smelteentalpi og øker i volum ved smelting. Hvilket fortegn har stigningstallet $dp/dT$ til smeltegrensen?

:::{dropdown} Løsningsforslag

Både $\Delta_{fus}H$ og $\Delta_{fus}V$ er positive. Fra $dp/dT=\Delta_{fus}H/(T\Delta_{fus}V)$ følger det at stigningstallet er positivt.
:::

**4. Type faseovergang**

Forklar forskjellen mellom en førsteordens og en kontinuerlig faseovergang. Ta med entropi og latent varme.

:::{dropdown} Løsningsforslag

Ved en førsteordens overgang har den molare entropien et sprang, og overgangen har latent varme. Ved en kontinuerlig overgang er entropien kontinuerlig, og det er ingen latent varme.
:::
