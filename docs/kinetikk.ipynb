## Reaksjonshastighet

I en reaksjon forbrukes reagenser og dannes produkter, men ikke nødvendigvis i samme tempo: i $A + 2B \rightarrow 3C$ forsvinner $B$ dobbelt så fort som $A$, og $C$ dannes tre ganger så fort. Det ville være upraktisk å ha tre forskjellige hastigheter for samme reaksjon. Løsningen er å dele på det støkiometriske tallet $\nu_J$, slik at vi får ett entydig tall.

:::{admonition} Definisjon - Reaksjonshastighet
:class: important

**Reaksjonshastigheten** er endringen i konsentrasjon per tid, delt på stoffets støkiometriske tall:

$v = \frac{1}{\nu_J}\frac{d[J]}{dt}$
:::

Siden $\nu_J$ er negativ for reagenser og positiv for produkter, blir hastigheten $v$ positiv uansett hvilket stoff vi følger - de to fortegnene opphever hverandre for reagenser.

## Hastighetslover, orden og hastighetskonstant

Hvordan hastigheten avhenger av konsentrasjonene, kan vi ikke lese ut av reaksjonslikningen - det må bestemmes eksperimentelt. Resultatet uttrykkes som en **hastighetslov**.

:::{admonition} Definisjon - Hastighetslov
:class: important

En **hastighetslov** knytter reaksjonshastigheten til konsentrasjonene, ofte på formen

$v = k_r[A]^a[B]^b$
:::

Her er $k_r$ **hastighetskonstanten**, som er uavhengig av konsentrasjonene, men avhenger av temperaturen. Potensene forteller reaksjonens orden.

:::{admonition} Definisjon - Reaksjonsorden
:class: important

**Ordenen** med hensyn på et bestemt stoff er potensen stoffet er opphøyd i hastighetsloven. **Den totale ordenen** er summen av alle potensene.
:::

Merk at ordenen ikke trenger å stemme med de støkiometriske koeffisientene - den er en ren eksperimentell størrelse.


### Å bestemme hastighetsloven

Et vanlig triks er **isolasjonsmetoden**: man lar alle reagenser unntatt ett være i stort overskudd. Da endrer de seg nesten ikke underveis, og hastighetsloven forenkles til å avhenge bare av det ene stoffet vi følger.

Er for eksempel $B$ i stort overskudd, kan $[B]$ regnes som konstant og slås sammen med hastighetskonstanten:

$k_{r,\text{eff}} = k_r[B]_0$

Reaksjonen oppfører seg da som om den var av lavere orden - vi kaller den en *pseudoførsteordens* reaksjon. Slik kan en komplisert hastighetslov kartlegges ett stoff av gangen.




## Integrerte hastighetslover

En hastighetslov er en differensiallikning: den forteller hvor fort konsentrasjonen endrer seg akkurat nå. Integrerer vi den, får vi konsentrasjonen som funksjon av tiden - som er det vi faktisk måler i laben.


### Nullteordens reaksjoner

Er hastigheten uavhengig av konsentrasjonen, avtar reagensen med konstant fart:

$[A] = [A]_0 - k_r t$

Konsentrasjonen synker lineært med tiden, helt til reagensen er brukt opp.


### Førsteordens reaksjoner

Er hastigheten proporsjonal med $[A]$, får integrasjonen en eksponentiell løsning:

$\ln\frac{[A]}{[A]_0} = -k_r t \qquad \Longrightarrow \qquad [A] = [A]_0\,e^{-k_r t}$

Konsentrasjonen avtar altså eksponentielt. Dette gir en enkel test: plotter vi $\ln[A]$ mot tiden $t$, får vi en rett linje med stigningstall $-k_r$. Er linjen rett, er reaksjonen første orden, og stigningstallet gir hastighetskonstanten.


### Andreordens reaksjoner

Er hastigheten proporsjonal med $[A]^2$, gir integrasjonen i stedet

$\frac{1}{[A]} - \frac{1}{[A]_0} = k_r t$

Nå er det $1/[A]$ som er lineær i tiden, med stigningstall $+k_r$. Hvilket plott som gir en rett linje - $\ln[A]$ eller $1/[A]$ - avslører altså ordenen direkte.

## Halveringstid

Et intuitivt mål på hvor fort en reaksjon går, er tiden det tar før halvparten av reagensen er borte.

:::{admonition} Definisjon - Halveringstid
:class: important

**Halveringstiden** $t_{1/2}$ er tiden det tar for konsentrasjonen å synke til halvparten av utgangsverdien.
:::

For de to vanligste ordenene blir halveringstiden

$\text{første orden:} \quad t_{1/2} = \frac{\ln 2}{k_r} \qquad\qquad \text{andre orden:} \quad t_{1/2} = \frac{1}{k_r[A]_0}$

Forskjellen er verdt å merke seg. For en førsteordens reaksjon er halveringstiden *konstant* - den avhenger ikke av hvor mye man startet med (dette er grunnen til at radioaktiv nedbrytning har en fast halveringstid). For en andreordens reaksjon avhenger den derimot av startkonsentrasjonen, og blir *lengre* jo lavere konsentrasjonen er. Derfor kan stoffer som brytes ned ved andreordens kinetikk vedvare svært lenge når de først er blitt fortynnet.

## Temperaturavhengighet: Arrhenius-likningen

Nesten alle reaksjoner går fortere når temperaturen øker. Eksperimentelt finner man at et plott av $\ln k_r$ mot $1/T$ gir en rett linje, og dette er innholdet i Arrhenius-likningen.

:::{admonition} Arrhenius-likningen
:class: note

$k_r = A\,e^{-E_a/RT} \qquad \Longleftrightarrow \qquad \ln k_r = \ln A - \frac{E_a}{RT}$
:::

Likningen inneholder to parametre, som begge kan leses av fra Arrhenius-plottet.

:::{admonition} Definisjon - Aktiveringsenergi og frekvensfaktor
:class: important

**Aktiveringsenergien** $E_a$ er den minste energien et molekylært møte trenger for å føre til reaksjon. Den finnes fra stigningstallet til Arrhenius-plottet, som er $-E_a/R$.

**Frekvensfaktoren** $A$ er hastighetskonstanten i grensen der alle møter fører til reaksjon. Den finnes fra skjæringspunktet.
:::


Kjenner vi hastighetskonstanten ved én temperatur, kan vi finne den ved en annen. Skriver vi Arrhenius-likningen for to temperaturer og trekker fra hverandre, får vi

$\ln\frac{k_{r,2}}{k_{r,1}} = -\frac{E_a}{R}\left(\frac{1}{T_2} - \frac{1}{T_1}\right)$


:::{admonition} På laboratoriet
:class: tip

Arrhenius-likningen er et måleverktøy for aktiveringsenergien. Måler du hastighetskonstanten ved flere temperaturer og plotter $\ln k_r$ mot $1/T$, får du en rett linje. Stigningstallet er $-E_a/R$, så aktiveringsenergien leses rett av - og skjæringspunktet gir frekvensfaktoren $A$.
:::


:::{card} 
**Eksempel**
^^^
En reaksjon har aktiveringsenergi $E_a = 50 \text{ kJ/mol}$. Hvor mye raskere går den ved kroppstemperatur ($37\,^\circ\text{C} = 310 \text{ K}$) enn ved romtemperatur ($25\,^\circ\text{C} = 298 \text{ K}$)?

Vi bruker toparts-formen:

$\ln\frac{k_{r,2}}{k_{r,1}} = -\frac{50000}{8{,}314}\left(\frac{1}{310} - \frac{1}{298}\right) = 0{,}78$

Opphøyer vi, får vi $k_{r,2}/k_{r,1} = e^{0{,}78} \approx 2{,}2$. En temperaturøkning på bare $12\,^\circ\text{C}$ mer enn dobler altså hastigheten - et illustrerende eksempel på hvor følsomme reaksjoner er for temperatur.
:::


### Aktivert kompleks og overgangstilstand

Aktiveringsenergien har en anskuelig tolkning. Følger vi den potensielle energien mens reaktantene møtes, forvrenges og omdannes til produkter, stiger den til et maksimum før den faller mot produktenes nivå. Toppen kalles **overgangstilstanden**, og klyngen av atomer der kalles det **aktiverte komplekset**.

Aktiveringsenergien er nettopp høyden på denne barrieren. Molekyler som ikke har nok energi til å komme over, spretter tilbake som reagenser; bare de energirike møtene fører til reaksjon. Det forklarer hvorfor høyere temperatur - som gir flere energirike molekyler - øker hastigheten, og hvorfor en høy barriere gjør reaksjonen mer temperaturfølsom.

## Reaksjonsmekanismer

De fleste reaksjoner skjer ikke i ett sprang, men gjennom en rekke enklere trinn - en **mekanisme**. Å finne mekanismen er et av kinetikkens hovedmål, for den forteller hva som faktisk skjer mellom molekylene.

:::{admonition} Definisjon - Elementærreaksjon og molekylaritet
:class: important

En **elementærreaksjon** er ett enkelt molekylært trinn i mekanismen. **Molekylariteten** er antall molekyler som møtes i dette trinnet - ett i en unimolekylær, to i en bimolekylær reaksjon.
:::

Vær nøye med forskjellen: molekylaritet gjelder ett elementærtrinn og er en teoretisk størrelse, mens orden gjelder totalreaksjonen og er eksperimentell. De to trenger ikke være like.

:::{admonition} Definisjon - Mellomprodukt
:class: important

Et **reaksjonsmellomprodukt** dannes og forbrukes underveis i mekanismen, men står ikke i totallikningen.
:::


### Det hastighetsbestemmende trinnet

:::{admonition} Definisjon - Hastighetsbestemmende trinn
:class: important

Det **hastighetsbestemmende trinnet** er trinnet i mekanismen som styrer totalhastigheten - typisk det tregeste, som virker som en flaskehals produktene må gjennom.
:::

En vanlig, men upresis, formulering er «det første trinnet er tregt og det andre raskt, derfor er det første hastighetsbestemmende». Poenget er ikke at hastighetene er forskjellige - i en sekvens er de like store - men at det er *hastighetskonstantene* som skiller trinnene.




## Steady-state-tilnærmingen

Så snart en mekanisme har mellomprodukter og flere trinn, blir de eksakte likningene fort uløselige. Steady-state-tilnærmingen er hovedverktøyet for å komme rundt dette.

:::{admonition} Definisjon - Steady-state-tilnærmingen
:class: important

Et reaktivt mellomprodukt dannes og forbrukes så raskt at konsentrasjonen holder seg tilnærmet konstant og lav. Da kan nettodannelsen settes lik null:

$\frac{d[I]}{dt} \approx 0$
:::

Fremgangsmåten er alltid den samme: skriv opp nettohastigheten for hvert mellomprodukt, sett den lik null, løs for mellomproduktets konsentrasjon, og sett dette inn i uttrykket for produktdannelsen. Da sitter man igjen med en hastighetslov uttrykt bare gjennom reagensene.


### Utledning med steady-state

:::{admonition} Steady-state på et to-trinns skjema
:class: note, dropdown

Se på skjemaet $A \xrightarrow{k_a} I \xrightarrow{k_b} P$, der $I$ er et mellomprodukt. Mellomproduktet dannes i det første trinnet og forbrukes i det andre, så nettohastigheten er

$\frac{d[I]}{dt} = k_a[A] - k_b[I]$

Steady-state-tilnærmingen setter denne lik null:

$k_a[A] - k_b[I] \approx 0 \qquad \Longrightarrow \qquad [I] \approx \frac{k_a}{k_b}[A]$

Produktet dannes med hastigheten $\frac{d[P]}{dt} = k_b[I]$. Setter vi inn uttrykket for $[I]$:

$\frac{d[P]}{dt} = k_b \cdot \frac{k_a}{k_b}[A] = k_a[A]$

Den kompliserte to-trinns-mekanismen reduseres altså til en enkel førsteordens hastighetslov - styrt av det tregeste trinnet.
:::


### Prelikevekt

Noen ganger rekker et mellomprodukt å komme tilnærmet i likevekt med reagensene før det går videre til produkt. Da kan hastighetsloven skrives med en sammensatt hastighetskonstant:

$k_r = \frac{k_a k_b}{k_a'}$

der $k_a$ og $k_a'$ er fram- og tilbakekonstantene for det første trinnet, og $k_b$ hører til det andre.


## Enzymkinetikk

En viktig anvendelse av steady-state er enzymreaksjoner. Enzymer er biologiske katalysatorer - de framskynder reaksjoner uten selv å forbrukes, og er så effektive og spesifikke at ett enzym kan øke hastigheten mange tierpotenser.

:::{admonition} Definisjon - Katalysator
:class: important

En **katalysator** øker reaksjonshastigheten ved å tilby en reaksjonsvei med lavere aktiveringsenergi, uten selv å forbrukes netto.
:::

:::{admonition} Definisjon - Enzym, substrat og aktivt sete
:class: important

Et **enzym** binder reagensen - kalt **substratet** - i et bestemt område kalt det **aktive setet**, og omdanner det til produkt før det frigjøres og setet er klart igjen.
:::


### Michaelis–Menten-mekanismen

Enzymreaksjoner beskrives ofte av et to-trinns skjema: enzym $E$ og substrat $S$ danner først et kompleks $ES$ i en reversibel likevekt, og komplekset gir så produkt. Bruker vi steady-state på $[ES]$, får vi hastighetsloven.

:::{admonition} Definisjon - Michaelis-konstanten
:class: important

**Michaelis-konstanten** samler de tre hastighetskonstantene i mekanismen:

$K_M = \frac{k_a' + k_b}{k_a}$
:::

:::{admonition} Michaelis–Menten-likningen
:class: note

$v = \frac{k_b[E]_0}{1 + K_M/[S]_0}$
:::

Likningen fanger to grenser. Ved lav substratkonsentrasjon er hastigheten tilnærmet førsteordens i $[S]$. Ved høy substratkonsentrasjon blir enzymet *mettet* - alle aktive seter er opptatt - og hastigheten nærmer seg et maksimum, $v_{max} = k_b[E]_0$, uansett hvor mye mer substrat man tilsetter.


### Lineweaver–Burk-plottet

Michaelis–Menten-likningen er en kurve, og kurver er vanskelige å lese parametre av. Tar vi den inverse på begge sider, får vi i stedet en rett linje.

:::{admonition} Lineweaver–Burk-likningen
:class: note

$\frac{1}{v} = \frac{1}{v_{max}} + \frac{K_M}{v_{max}}\cdot\frac{1}{[S]_0}$
:::

Plotter vi $1/v$ mot $1/[S]_0$, får vi en rett linje. Skjæringspunktet med $y$-aksen er $1/v_{max}$, og stigningstallet er $K_M/v_{max}$. Fra disse to leser vi av både maksimalhastigheten og Michaelis-konstanten.


:::{card} 
**Eksempel**
^^^
Enzymet karboanhydrase katalyserer hydrering av $CO_2$. Følgende data ble målt:

| $[S]$ (mmol/dm³) | 1,25 | 2,5 | 5 | 20 |
|---|---|---|---|---|
| $v$ (mmol/dm³/s) | 0,0278 | 0,0500 | 0,0833 | 0,167 |

Vi lager et Lineweaver-Burk-plott ved å regne ut de inverse verdiene og tilpasse en rett linje. Den gir et skjæringspunkt på $4{,}00$ og et stigningstall på $40{,}0$.

Maksimalhastigheten er da

$v_{max} = \frac{1}{\text{skjæringspunkt}} = \frac{1}{4{,}00} = 0{,}250 \text{ mmol/dm}^3\text{/s}$

og Michaelis-konstanten er

$K_M = \text{stigningstall} \times v_{max} = 40{,}0 \times 0{,}250 = 10{,}0 \text{ mmol/dm}^3$
:::

## Kinetisk og termodynamisk kontroll

Til slutt knytter vi kinetikk og likevekt sammen. Kan en reaksjon gi to ulike produkter, avgjør omstendighetene hvilket som dominerer.

Før likevekt er nådd, styres produktforholdet av hastighetene - altså av forholdet mellom hastighetskonstantene:

$\frac{[P_2]}{[P_1]} = \frac{k_{r,2}}{k_{r,1}}$

Dette kalles **kinetisk kontroll**: det raskest dannede produktet vinner, uansett hva termodynamikken skulle tilsi. Lar man derimot reaksjonen nå likevekt, overtar **termodynamisk kontroll**, og produktforholdet bestemmes i stedet av standard reaksjons-Gibbs-energi. Om et produkt «vinner» kan altså avhenge av om man stopper reaksjonen tidlig eller lar den gå helt fram.
