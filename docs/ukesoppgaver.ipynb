### Litt samla oppgave idk

Sett opp et lite $2\times2$-gitter (4 plasser, nummerert 1–4), der to plasser er okkupert av type 1-molekyler og to av type 2-molekyler. Energien til en plassering er antall 1–2-kontakter (kun opp/ned/venstre/høyre, ikke diagonalt), ganget med $\Delta\epsilon$.
- List opp alle $\binom{4}{2}=6$ mulige plasseringer for hånd (tegn gjerne gitteret), og finn energien til hver.
- Bestem multiplisiteten $\Omega(E)$ for hver distinkte energiverdi du finner.
- Sett opp og regn ut $Q=\sum_E\Omega(E)e^{-E\Delta\epsilon/k_BT}$ ved $\Delta\epsilon/k_BT=1$.
- Bruk resultatet til å finne sannsynligheten for at systemet befinner seg i sin laveste energitilstand.
