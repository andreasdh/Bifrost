:::{admonition} Læringsutbytte
:class: note
Etter å ha jobbet med dette temaet skal du kunne: 

- forklare forskjellen mellom en mikrotilstand og en makrotilstand
- forklare hvorfor statistiske beskrivelser er nødvendige for systemer som består av 
svært mange partikler
- beskrive energi som en egenskap ved mikrotilstander og som en størrelse som kan 
fordeles mellom molekyler på mange ulike måter
- forklare entropi som et mål på antall mikrotilstander som er forenlige med en gitt 
makrotilstand
- forklare hvorfor makrotilstander med mange mikrotilstander er mer sannsynlige enn 
makrotilstander med få mikrotilstander
- beskrive hvordan sannsynlighet og antall mikrotilstander er knyttet til termodynamisk 
likevekt
- forklare temperatur som et mål på sammenhengen mellom energi og entropi.

### Oppgave 3 - mikrotilstander og makrotilstander

![Oppgave mikro, makro, statmek](Simuleringer/oppgave_gitter_composite.png)

I modellen over har vi brukt de samme antakelsene som i det ene eksempelet fra Statistisk mekanikk modulen. Vi ser på et lite $2\times2$-gitter med fire plasser ($A$, $B$, $C$ og $D$), der to av plassene er okkupert av type 1-molekyler og to av type 2-molekyler. Molekylene av samme type regnes som uskjelnbare seg imellom: å bytte om de to type 1-molekylene gir ingen ny mikrotilstand. Dette gitteret kan du tenke på som et lite subsystem hentet ut fra et mye større gitter, vist til venstre i figuren.

En kontakt mellom et type 1- og et type 2-molekyl, markert med oransje streker, koster energien $\Delta\epsilon$, mens kontakt mellom to like molekyler ikke koster noe. Vi teller bare direkte naboer: opp, ned, venstre, høyre.

En modell som dette kan representere flere reelle kjemiske fenomener, for eksempel to typer atomer i en legering eller en fast blanding (der "kontaktenergien" $\Delta\epsilon$ tilsvarer forskjellen mellom å ha like og ulike naboatomer), eller et løst stoff omgitt av løsemiddel, slik vi har sett tidligere.

- **a)** Forklar forskjellen mellom en kvantetilstand og en mikrotilstand for et system med mange molekyler. Bruk subsystemet over som eksempel: hva er én mikrotilstand i denne modellen, og hvorfor er det *ikke* det samme som én kvantetilstand for ett enkelt atom?
- **b)** Forklar med egne ord hva som er forskjellen på en mikrotilstand og en makrotilstand. Gjerne bruk modellen i forklaringen. 
- **c)** Det finnes $\binom{4}{2}=6$ mulige plasseringer av de to type 1-molekylene på subsystemet. List opp alle seks (f.eks. "type 1 på plasseringene $A$ og $B$"), og finn energien $\varepsilon_n$ til hver av dem ved å telle 1–2-kontakter for hånd. List opp energiene i økende rekkefølge. 
- **d)** Antallet ulike energiverdier tilsvarer antallet makrotilstander. Hvor mange *ulike* energiverdier $E$ fant du blant de seks mikrotilstandene i c)? Tenk deg nå et reelt system med $N_A\approx6\cdot10^{23}$ molekyler i stedet for bare 4. Uten å regne det ut nøyaktig: hva tror du skjer med antall mikrotilstander sammenlignet med antall makrotilstander, når systemet blir så mye større? Bruk svaret ditt til å forklare hvorfor vi trenger en statistisk beskrivelse for systemer med svært mange partikler, i stedet for å bare telle opp alle mikrotilstandene direkte, slik du gjorde i c).

### Oppgave 4 - Boltzmanns fordelingslov

### Oppgave 5 - partisjonsfunksjonen

- **a)** Forklar forskjellen på skjelnbare og uskjelnbare partikler. Hvorfor skiller vi mellom dem og hvorfor er dette viktig å være klar over når vi setter opp partisjonsfunksjonen?
- **b)** Hva er partisjonsfunksjonen og hva brukes den til?
- **c)** I praksis måler vi som regel *energiendringer* i et system, ikke absolutte energier. Tenk deg at vi har målt energiendringen $\Delta E$ til et lite system (for eksempel en liten mengde gass i en beholder) relativt til grunntilstanden, flere ganger, og funnet fire mulige verdier. Multiplisiteten $\Omega(\Delta E)$ for hver av dem er gitt i tabellen under:

  | $\Delta E$ (i enheter av $\Delta\epsilon$) | 0 | 1 | 2 | 3 |
  |---|---|---|---|---|
  | $\Omega(\Delta E)$ | 1 | 4 | 6 | 4 |

  - Sett $\Delta\epsilon/k_BT=1$, og bruk tabellen til å sette opp $Q$ uttrykt ved energidifferanser, altså på formen $Q=\Omega(0)+\Omega(\Delta E_1)e^{-\Delta E_1\Delta\epsilon/k_BT}+\cdots$, og regn den ut.
  - Bruk resultatet til å finne forholdet $p(\Delta E{=}1\Delta\epsilon)/p(\Delta E{=}0)$. (Husk at $p(\Delta E)=\Omega(\Delta E)e^{-\Delta E\Delta\epsilon/k_BT}/Q$, multiplisiteten skal være med.)
  - Forklar hvordan $Q$ oppfører seg i de to ekstremtilfellene: hva blir $Q$ dersom $\Delta\epsilon/k_BT\to0$, og hva blir $Q$ dersom $\Delta\epsilon/k_BT\to\infty$? Bruk tabellen til å begrunne svaret ditt.
  - Regn ut indre energi $U=\sum_{\Delta E}\Omega(\Delta E)\,\Delta E\,e^{-\Delta E\Delta\epsilon/k_BT}/Q$ (i enheter av $\Delta\epsilon$), og deretter entropien $S=k_B\ln Q+U/T$ (skriv svaret som $S/k_B$).

### Litt samla oppgave idk

Sett opp et lite $2\times2$-gitter (4 plasser, nummerert 1–4), der to plasser er okkupert av type 1-molekyler og to av type 2-molekyler. Energien til en plassering er antall 1–2-kontakter (kun opp/ned/venstre/høyre, ikke diagonalt), ganget med $\Delta\epsilon$.
- List opp alle $\binom{4}{2}=6$ mulige plasseringer for hånd (tegn gjerne gitteret), og finn energien til hver.
- Bestem multiplisiteten $\Omega(E)$ for hver distinkte energiverdi du finner.
- Sett opp og regn ut $Q=\sum_E\Omega(E)e^{-E\Delta\epsilon/k_BT}$ ved $\Delta\epsilon/k_BT=1$.
- Bruk resultatet til å finne sannsynligheten for at systemet befinner seg i sin laveste energitilstand.
