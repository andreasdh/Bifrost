# Termokjemi og likevekter

Hvorfor må vi varme opp noen reaksjoner, mens andre må stå på isbad? Hva mener vi med at en reaksjon er spontan? Nå som vi har lært om de termodynamiske potensialene og termodynamiske drivkrefter er vi endelig klare for å studere den praktiske anvendelsen av disse størrelsene i kjemifaget: La oss gå løs på termokjemi!🧪

Termokjemien beskriver kjemiske prosesser ved hjelp av de termodynamiske formlene for entalpi, entropi og fri energi. Det er mange ulike prosesser som kan foregå innen kjemifaget, men de du kommer til å lære mest om  i Fysikalsk Kjemi I er _faseoverganger_, _løsninger og blandinger_ og ikke minst _kjemiske reaksjoner_! Ofte kan en prosess være svært sammensatt. Dette er særlig sant for kjemiske reaksjoner, som ofte går via én eller flere overgangstilstander mellom reagens og ønsket produkt. I første omgang skal vi derimot kun interessere oss for egenskapene til start- og sluttilstandene i de ulike prosessene. Ved å sammenlikne disse tilstandene med hverandre vil vi kunne avgjøre om en prosess er _spontan_ eller ikke og for hvilke temperaturer dette gjelder. Vi skal også se på hvordan ulike reaksjoner kan drives av entalpi eller entropi, begge samtidig eller ingen av dem! Vi skal også knytte dette opp mot den praktiske kjemien ved å diskutere labforhold og reaksjoner du kjenner igjen fra tidligere kurs.🥼🔬

## Entalpi: Ekstoterme og endoterme prosesser

### Entalpi beskriver varmeutvekslinger i løpet av en kjemisk prosess

Som vi har forklart tidligere gir den termodynamiske størrelsen entalpi et bilde på varmeutveksling ved konstant trykk. Siden de fleste kjemiske prosesser foregår under konstant trykk, kan altså entalpien fortelle oss hvordan varme flyter inn og ut av systemet vi er interessert i, i løpet av en prosess. Vi kan introdusere to nye begreper som er svært viktige i kjemien: Dersom systemet gir fra seg varme til omgivelsene, er prosessen _eksoterm_. Dersom det tar opp varme fra omgivelsene, er prosessen derimot _endoterm_. Ved en eksoterm prosess vil endringen i entalpi $\Delta H$ ha negativt fortegn og ved en endoterm prosess vil endringen i entalpi $\Delta H$ ha positivt fortegn. Men hvordan beregner vi egentlig en slik endring? For å forstå dette må vi introdusere de mange typene entalpi du kan møte på i fysikalsk kjemi.

**Grunnstoffers entalpi** 

Som et referansepunkt har rene grunnstoffer i sin mest stabile tilstand en entalpi lik 0. For eksempel er mange gasser i sin mest stabile tilstand som dimerer: $H_2$, $O_2$ eller $N_2$. 

**Dannelsesentalpi**

Dannelsesentalpi er entalpien (varmen ved konstant trykk) som må til for å danne et mol av et stoff fra grunnstoffene det består av. For eksempel har dannelsen av vann fra oksygen og hydrogen i sine mest stabile tilstander ($O_2$ og $H_2$) en entalpi på $\Delta H_f = –285.84 kJ/mol$. Ofte snakker vi om _standard_ dannelsesentalpi, og dette betyr ved _standard_ forhold, altså ved $25 \degree C$ og $1 atm$ trykk. Da betegner vi dannelsesentalpien slik: $\Delta\degree H_f$ hvor $\Delta$ betegner endring i løpet av prosessen, $\degree$ betegner standard forhold og subskript f betegner "formation" på engelsk. Disse dannelsesentalpiene er tabulerte verdier du kan benytte deg av for å regne videre på entalpier for faseoverganger og reaksjoner. De har blitt bestemt ved å gjøre kalorimetri-målinger under dannelsen av stoffer fra grunnstoffer. Dannelsesentalpi er altså basisen for alle entalpiberegninger, og er et nyttig verktøy kombinert med Hess lov!

**Entalpi knyttet til faseoverganger**

Noen typer entalpi er knyttet opp mot faseoverganger: De vanligste er fordampingsentalpi som betegnes med subskript "vap" ("vaporisation" på engelsk) og smelteentalpi som betegnes med subskript "fus" ("fusion" på engelsk). Her beskriver altså overganger mellom ulike faser av ett og samme stoff. For eksempel har vi for fordamping av vann, en fordampingsentalpi $\Delta H_{vap} \approx 40.7 kJ/mol$. Vi må altså tilføre varme for å fordampe vann, som stemmer overens med erfaringene våre både fra lab og kjøkkenet!

**Reaksjonsentalpi**

Reaksjonsentalpi beskriver varmeendring (ved konstant trykk) knyttet til kjemiske reaksjoner. Her er dannelsesentalpier veldig nyttige! Vi kan dele opp en reaksjon i flere trinn og vurdere entalpien for hvert av trinnene før vi setter dem sammen for å beregne den totale reaksjonsentalpien slik Hess lov tillater. Slik bestemmer vi om en reaksjon er endoterm eller ekstoterm. 

[Eksempel] 

Det er viktig for en kjemiker eller materialviter å ta stilling til hvor mye varme som vil genereres eller brukes opp i løpet av en reaksjon da dette vil påvirke hvordan vi designer et eksperiment eller setter opp en reaksjon. Vi skal se på et konkret eksempel av dette i slutten av dette avsnittet.

**Andre entalpier**

Noen typer entalpier faller ikke innenfor kategoriene vi har nevnt ovenfor, men det er allikevel lure å ha hørt om noen av dem. _Forbrenningsentalpi_ er entalpiendringen når et mol stoff fullstendig forbrennes i oksygen og betegnes med $\Delta H_c$ ( c fra engelsk "combustion"). Løsningsentalpi er mye brukt i kjemi og er definert som entalpiendringen når ett mol av et stoff løses i en gitt mengde løsemiddel. Entalpien betegnes som regel med $\Delta H_{sol}$ (fra engelsk "solvation"). Et siste eksempel er nøytraliseringsentalpi som beskriver entalpiendringen når en base og en syre reagerer for å danne vann og et salt. Entalpien betegnes med $\Delta H_{neut}$ (fra engelsk "neutralization"). 

**I prinsippet beregnes alle typer $\Delta H$ likt:** 
1. Forstå hvor prosessen vi er interesserte i starter og slutter: Hvilke stoffer begynner vi med? Hvor mye av stoffene har vi (støkiometri)? Hvilke stoffer har vi igjen til slutt? Hvilke faser er stoffene i?
2. Dele opp prosessen for å bruke Hess lov: Hvilke mindre prosesser kan vi dele den totale prosessen i? Hvilke relevante dannelsesentalpier har vi tilgjengelig? 
3. Sett sammen for å beregne total endring: Summere over alle delprosesser og deres entalpier for å beregne total entalpiendring
4. Tolke resultater: Hvilket fortegn har entalpiendringen? I hvilken grad er prosessen endoterm/eksoterm? Hvilke tilpasninger må jeg gjøre i laben?
