# Termokjemi og likevekter

Hvorfor må vi varme opp noen reaksjoner, mens andre må stå på isbad? Hva mener vi med at en reaksjon er spontan? Nå som vi har lært om de termodynamiske potensialene og termodynamiske drivkrefter er vi endelig klare for å studere den praktiske anvendelsen av disse størrelsene i kjemifaget: La oss gå løs på termokjemi!🧪

Termokjemien beskriver kjemiske prosesser ved hjelp av de termodynamiske formlene for entalpi, entropi og fri energi. Det er mange ulike prosesser som kan foregå innen kjemifaget, men de du kommer til å lære mest om  i Fysikalsk Kjemi I er _faseoverganger_, _løsninger og blandinger_ og ikke minst _kjemiske reaksjoner_! Ofte kan en prosess være svært sammensatt. Dette er særlig sant for kjemiske reaksjoner, som ofte går via én eller flere overgangstilstander mellom reagens og ønsket produkt. I første omgang skal vi derimot kun interessere oss for egenskapene til start- og sluttilstandene i de ulike prosessene. Ved å sammenlikne disse tilstandene med hverandre vil vi kunne avgjøre om en prosess er _spontan_ eller ikke og for hvilke temperaturer dette gjelder. Vi skal også se på hvordan ulike prosesser kan drives av entalpi eller entropi, begge samtidig eller ingen av dem! Vi skal også knytte dette opp mot den praktiske kjemien ved å diskutere labforhold og reaksjoner du kjenner igjen fra tidligere kurs.🥼🔬

## Hess lov

Som du har lest om i introduksjonen og i forrige kapittel om makroskopisk termodynamikk er flere størrelser innen termodynamikken _tilstandsfunksjoner_: De avhenger kun av start- og sluttilstand. Hess lov baserer seg på nettopp dette! Dersom vi har en _prosess_ vi ønsker å beregne en total endring for, for eksempel i entalpi eller Gibbs fri energi som vi skal snakke mye om senere på denne siden, kan vi dele den inn i så mange ulike _delprosesser_ vi ønsker. Dette er ofte et lurt triks, da ikke alle start- og sluttilstander kommer til å ha de tabulerte verdiene vi trenger. 

## Entalpi: Ekstoterme og endoterme prosesser

### Entalpi beskriver varmeutvekslinger i løpet av en kjemisk prosess

Som vi har forklart tidligere gir den termodynamiske størrelsen entalpi et bilde på varmeutveksling ved konstant trykk. Siden de fleste kjemiske prosesser foregår under konstant trykk, kan altså entalpien fortelle oss hvordan varme flyter inn og ut av systemet vi er interessert i, i løpet av en prosess. Vi kan introdusere to nye begreper som er svært viktige i kjemien: Dersom systemet gir fra seg varme til omgivelsene, er prosessen _eksoterm_. Dersom det tar opp varme fra omgivelsene, er prosessen derimot _endoterm_. Ved en eksoterm prosess vil endringen i entalpi $\Delta H$ ha negativt fortegn og ved en endoterm prosess vil endringen i entalpi $\Delta H$ ha positivt fortegn. Men hvordan beregner vi egentlig en slik endring? For å forstå dette må vi introdusere de mange typene entalpi du kan møte på i fysikalsk kjemi.

**Grunnstoffers entalpi** 

Som et referansepunkt har rene grunnstoffer i sin mest stabile tilstand en entalpi lik 0. For eksempel er mange gasser i sin mest stabile tilstand som dimerer: $H_2$, $O_2$ eller $N_2$. 

**Dannelsesentalpi**

Dannelsesentalpi er entalpien (varmen ved konstant trykk) som må til for å danne et mol av et stoff fra grunnstoffene det består av. For eksempel har dannelsen av vann fra oksygen og hydrogen i sine mest stabile tilstander ($O_2$ og $H_2$) en entalpi på $\Delta H_f = –285.84 kJ/mol$. Ofte snakker vi om _standard_ dannelsesentalpi, og dette betyr ved _standard_ forhold, altså ved $25^\circ\mathrm{C}$ og 1 atm trykk. Da betegner vi dannelsesentalpien slik: $\Delta^{\circ} H_f$ hvor $\Delta$ betegner endring i løpet av prosessen, $^\circ$ betegner standard forhold og subskript f betegner "formation" på engelsk. Disse dannelsesentalpiene er tabulerte verdier du kan benytte deg av for å regne videre på entalpier for faseoverganger og reaksjoner. De har blitt bestemt ved å gjøre kalorimetri-målinger under dannelsen av stoffer fra grunnstoffer. Dannelsesentalpi er altså basisen for alle entalpiberegninger, og er et nyttig verktøy kombinert med Hess lov!

**Entalpi knyttet til faseoverganger**

Noen typer entalpi er knyttet opp mot faseoverganger: De vanligste er fordampingsentalpi som betegnes med subskript "vap" ("vaporisation" på engelsk) og smelteentalpi som betegnes med subskript "fus" ("fusion" på engelsk). Her beskriver altså overganger mellom ulike faser av ett og samme stoff. For eksempel har vi for fordamping av vann, en fordampingsentalpi $\Delta H_{vap} \approx 40.7 kJ/mol$. Vi må altså tilføre varme for å fordampe vann, som stemmer overens med erfaringene våre både fra lab og kjøkkenet!

**Reaksjonsentalpi**

Reaksjonsentalpi beskriver varmeendring (ved konstant trykk) knyttet til kjemiske reaksjoner. Her er dannelsesentalpier veldig nyttige! Vi kan dele opp en reaksjon i flere trinn og vurdere entalpien for hvert av trinnene før vi setter dem sammen for å beregne den totale reaksjonsentalpien slik Hess lov tillater. Slik bestemmer vi om en reaksjon er endoterm eller ekstoterm. Det er viktig for en kjemiker eller materialviter å ta stilling til hvor mye varme som vil genereres eller brukes opp i løpet av en reaksjon da dette vil påvirke hvordan vi designer et eksperiment eller setter opp en reaksjon. Vi skal se på et konkret eksempel av dette i slutten av dette avsnittet.

**Andre entalpier**

Noen typer entalpier faller ikke innenfor kategoriene vi har nevnt ovenfor, men det er allikevel lure å ha hørt om noen av dem. _Forbrenningsentalpi_ er entalpiendringen når et mol stoff fullstendig forbrennes i oksygen og betegnes med $\Delta H_c$ ( c fra engelsk "combustion"). Løsningsentalpi er mye brukt i kjemi og er definert som entalpiendringen når ett mol av et stoff løses i en gitt mengde løsemiddel. Entalpien betegnes som regel med $\Delta H_{sol}$ (fra engelsk "solvation"). Et siste eksempel er nøytraliseringsentalpi som beskriver entalpiendringen når en base og en syre reagerer for å danne vann og et salt. Entalpien betegnes med $\Delta H_{neut}$ (fra engelsk "neutralization").

### Et steg tilbake: Kan vi forutse entalpiendring basert på molekylstruktur?

I løpet av en kjemisk prosess ser vi ofte en endring i sammensetningen til systemet vi er interesserte i: I kjemiske reaksjoner forvandles reagenser til produkter ved at kjemiske bindinger brytes og dannes. I faseoverganger endres de _intermolekylære_ kreftene som holder molekylene i systemet sammen. Og i løsninger kan for eksempel salter _solvatiseres_ ved at ionene omringes av løsemiddel molekyler. For enkelte prosesser er det ikke sikkert vi har tilgang på dannelsesentalpiene vi trenger for å regne på $\Delta H$. Da kan det være nyttig å kunne resonnere seg frem til hva slags entalpiendring vi forventer for prosessen basert på de molekylære kreftene i systemet. 

Kort fortalt kan vi gjøre dette ved å vurdere den relative styrken til bindingene som brytes og dannes i løpet av prosessen! _Svake bindinger_ gir atomene mye frihet til å vibrere, rotere og bevege seg rundt og ligger derfor høyt i energi. Fordi et system alltid tenderer mot å minimere energien sin, byttes gjerne svake bindinger ut med _sterke bindinger_ som ligger lavere i energi da systemet blir mer rigid og atomene har mindre frihet til vibrasjon, rotasjon og translasjon. Energigapet mellom de to tilstandene frigjøres som varme, og det er slik vi ender opp med **eksoterme prosesser**. Vi skal se på et konkret eksempel på en eksoterm reaksjon i organisk kjemi i slutten av avsnittet. Vi kan også gå motsatt vei: I faseoverganger fra væske til gass bytter vi ut _sterke interaksjoner_ som gjør at væsken holdes sammen med _svake interaksjoner_ som lar gassen flyte fritt i rommet. I dette tilfellet må systemet ta opp varme fra omgivelsene som det bruker til å "tette" energigapet mellom tilstandene. Slik får vi **endoterme prosesser** hvor vi må tilføre varme til systemet. ### Konkret eksempel: Grignard-reaksjonen er en ekstoterm reaksjon: Fra molekylstruktur til labprosedyre

Eksempelet vi nå skal vise handler om en reaksjon kjemistudentene lærte om og gjorde på laben i kurset Organisk Kjemi I, nemlig Grignard-reaksjonen! Grignard-reaksjonen er en svært nyttig reaksjon i organisk kjemi da den lar oss danne nye karbon-karbon bindinger, noe som hjelper oss å bygge større og mer komplekse organiske og metallorganiske molekyler for diverse formål. På UiO blir reaksjonen blant annet benyttet både til legemiddel- og katalysatorsyntese⚗️ Grignard-reaksjonen starter med dannelsen av et Grignard-reagens, som vi lar reagere med et karbonyl for å danne det koblede produktet (ny C-C binding). Men hvorfor trekker vi akkurat denne reaksjonen frem som eksempel? Grignard-reaksjonen er svært eksoterm! Denne egenskapen er ikke bare direkte koblet til molekylstrukturen til Grignard-reagenset, men har også viktige implikasjoner for labprosedyren til reaksjonen. Målet med dette eksempelet er altså å vise dere hvordan både mikroskopisk og makroskopisk fysikalsk kjemi kan være relevant, selv i et eksperimentelt felt som organisk kjemi! La oss først tenke på hvorfor denne reaksjonen er så eksoterm utifra molekylstrukturen til reagensene og produktet, før vi går videre til å regne på $\Delta H$ ved hjelp av Hess lov og til slutt vurderer hvilke konsekvenser dette får for labprosedyren til reaksjonen🥰

**Labprosedyre**

Grignard-reaksjonens eksoterme karakter fører også til tilpasninger i prosedyren for reaksjonen. Først, hva ville skjedd dersom vi bare blandet alt av reagensene sammen i løsemiddelet på én gang? Vi vet at reaksjonen er svært eksoterm, så derfor ville en rask og plutselig blanding av reagenser føre til mye varmeutvikling. Løsemiddelet, som regel eter, vil fosskoke. Dette fører til to problemer: Én, når løsemiddelet koker bort løses reagensene dårligere og dårligere og vi risikerer at reaksjonen ikke går som vi ønsker. To, og aller viktigst, trykket vil øke inne i glassutstyret etterhvert som flere og flere løsemiddelmolekyler går over i gassfase. Raskt vil trykket bli så høyt at korken på kolba vil skyte av og eksponere blandingen for luft. Dette kan være farlig da Grignard-reagenset reagerer kraftig med vann som finnes i lufta. Vi risikerer rett og slett at kolba eksploderer på grunn av den raske varmeutviklingen, og dette kan være svært farlig for den som står på lab! Hvordan skal vi unngå dette? Vi gjør tre hovedgrep for å holde Grignard-reaksjonen under kontroll: 

1. *Reaksjonen gjøres i fullstendig tørt glassutstyr og under inert atmosfære* (som regel argon- eller nitrogen-atmosfære) for å hindre Grignard-reagensen i å reagere med vann på glasset eller i lufta. 
2. *Reaksjonsblandingen kjøles ned* til $0$ eller $-78^\circ\mathrm{C}$ før tilsettingen av Grignard-reagenset
3. *Grignardreagensen tilsettes litt og litt* så ikke alle molekylene reagerer samtidig og den totale varmen som genereres av gangen blir mindre.
