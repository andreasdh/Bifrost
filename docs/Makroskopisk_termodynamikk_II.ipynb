# Makroskopisk termodynamikk II

I forrige kapitell _Makroskopisk termodynamikk I_ lærte du om noen viktige størrelser i klassisk termodynamikk samt om de fire termodynamiske potensialene indre energi U, entalpi H, Helmholtz fri energi F og Gibbs fri energi G. I dette kapitellet skal vi ta anvendelsen av disse størrelsene et steg videre og lære om termodynamiske drivkrefter.
