## Termodynamiske drivkrefter matematisk og praktisk

Varme strømmer fra varmt til kaldt, gasser beveger seg fra høyt trykk til lavt trykk, stoffer diffunderer fra områder med høy konsentrasjon til områder med lav konsentrasjon. Dette virker kanskje intuitivt for oss, men hvorfor skjer disse prosessene egentlig? Hva har de til felles? Bak dem ligger de _termodynamiske drivkreftene_. Først skal vi se på hvordan vi matematisk kan definere de termodynamiske potensialene fra tilstandsfunksjonene. Dette gir opphav til _fundamentallikningene_, den matematiske definisjonen av de termodynamiske drivkreftene og _Maxwell-relasjonene_. Deretter skal vi gå videre til termokjemi. 

### Fundamentallikningen for indre energi i et lukket system

La oss starte fra definisjonen av indre energi igjen: 

$$\Delta U = q + w$$

I stedet for definisjonen som inneholder stifunksjonene arbeid og varme ønsker vi nå å finne ut av hvordan vi kan uttrykke den indre energien ved hjelp av tilstandsfunksjoner som temperatur, trykk, volum og entropi. Vi benytter oss av noen uttrykk fra de forrige kapitlene og finner et nytt uttrykk for dU:

$$dS = \frac{dq}{T} \leftrightarrow dq = TdS$$
$$dw = -pdV$$
$$dU = TdS - pdV$$

Dette er en *fundamentallikning* for den indre energien og gjør indre energi til en funksjon av entropi og volum. Vi kan nå trekke ut temperatur og volum som partiellderiverte! Dersom volumet er konstant vil dV være lik 0 og vi står igjen med:

$$\delta U = T\delta S \leftrightarrow T = \left(\frac{\delta U}{\delta S}\right)_V$$

Temperatur viser altså hvor mye den indre energien må øke når vi får økning i entropi ved fast volum, for eksempel ved varmetilførsel. Vi sier at temperatur er _drivkraften_ for varmeoverføring. 

Dersom vi nå setter entropien konstant, vil dS være lik 0 og vi står igjen med:

$$\delta U = -p\delta V \leftrightarrow -p = \left(\frac{\delta U}{\delta V}\right)_S$$

Trykk viser altså hvor mye den indre energien må senkes når vi får økning i volum ved fast entropi. Vi sier at trykk er _drivkraften_ for volumendring.

### Fra indre energi til entalpi og fri energi

På samme måte som for den indre energien ønsker vi også å definere entalpi H og Helmholtz og Gibbs fri energi F og G ved hjelp av tilstandsfunksjoner. Dette blir enklere siden vi nå kan ta utgangspunkt i uttrykket vi fant for indre energi! Se på utledningen av fundamentalllikningen til entalpi først, samt hvordan vi kan finne de termodynamiske drivkreftene knyttet til entalpi. Deretter skal du få prøve deg på egenhånd for de frie energiene🔥

*Entalpi* 
Fra forrige kapittel vet vi at entalpi er definert som den indre energien pluss pV. Vi starter utledningen vår med å differensiere uttrykket og bytte ut $dU$ med uttrykket vi fant i første del av kapittelet:

$$H = U + pV$$
$$dH = dU + d(pV) = (TdS - pdV) + (pdV + Vdp)$$
$$dH = TdS + Vdp$$

Dersom vi setter trykket konstant, blir dp lik null, og vi står igjen med:

$$\delta H = T\delta S \leftrightarrow T = \left(\frac{\delta H}{\delta S}\right)_p$$

Dersom vi setter entropien konstant, finner vi:

$$V = \left(\frac{\delta H}{\delta p}\right)_S$$

*Helmholtz fri energi* 
Fra forrige kapittel vet vi at Helmholtz fri energi er definert som indre energi minus TS. Vi starter utledningen vår med å differensiere uttrykket og bytte ut $dU$ med uttrykket vi fant i første del av dette kapittelet:

$$F = U - TS$$
$$dF = dU - d(TS) = (TdS - pdV) - (TdS + SdT)$$
$$dF = -pdV - SdT$$

Dersom vi setter trykket konstant, blir dp lik null, og vi står igjen med:

$$\delta H = T\delta S \leftrightarrow T = \left(\frac{\delta H}{\delta S}\right)_p$$

Dersom vi setter temperaturen konstant, finner vi:

$$-p = \left(\frac{\delta F}{\delta V}\right)_T$$

Dersom vi setter volumet konstant, finner vi:

$$-S = \left(\frac{\delta F}{\delta T}\right)_V$$

*Gibbs fri energi*
Fra forrige kapittel vet vi at Gibbs fri energi er definert som entalpi minus TS. Vi starter utledningen vår med å differensiere uttrykket og bytte ut $dH$ med uttrykket vi fant i første del av dette kapittelet:

$$G = H - TS$$
$$dG = dH - d(TS) = (TdS - pdV) - (TdS + SdT)$$
$$dF = -pdV - SdT$$

Dersom vi setter temperaturen konstant, finner vi:

$$-p = \left(\frac{\delta F}{\delta V}\right)_T$$

Dersom vi setter volumet konstant, finner vi:

$$-S = \left(\frac{\delta F}{\delta T}\right)_V$$
