## Termodynamiske drivkrefter

Varme strømmer fra varmt til kaldt, gasser beveger seg fra høyt trykk til lavt trykk, stoffer diffunderer fra områder med høy konsentrasjon til områder med lav konsentrasjon. Dette virker kanskje intuitivt for oss, men hvorfor skjer disse prosessene egentlig? Hva har de til felles? Bak dem ligger de _termodynamiske drivkreftene_. 

### Fundamentallikningen for indre energi i et lukket system

La oss starte fra definisjonen av indre energi igjen: 

$$\Delta U = q + w$$

I stedet for definisjonen som inneholder stifunksjonene arbeid og varme ønsker vi nå å finne ut av hvordan vi kan uttrykke den indre energien ved hjelp av tilstandsfunksjoner som temperatur, trykk, volum og entropi. Vi benytter oss av noen uttrykk fra de forrige kapitlene og finner et nytt uttrykk for dU:

$$dS = \frac{dq}{T} \leftrightarrow dq = TdS$$
$$dw = -pdV$$
$$dU = TdS - pdV$$

Dette er en *fundamentallikning* for den indre energien og gjør indre energi til en funksjon av entropi og volum. Vi kan nå trekke ut temperatur og volum som partiellderiverte! Dersom volumet er konstant vil dV være lik 0 og vi står igjen med:

$$\delta U = T\delta S \leftrightarrow T = (\frac{\delta U}{\delta S})_V$$

Temperatur viser altså hvor mye den indre energien må øke når vi får økning i entropi ved fast volum, for eksempel ved varmetilførsel. Vi sier at temperatur er _drivkraften_ for varmeoverføring. 

Dersom vi nå setter entropien konstant, vil dS være lik 0 og vi står igjen med:

$$\delta U = -p\delta V \leftrightarrow -p = (\frac{\delta U}{\delta V})_V$$

Trykk viser altså hvor mye den indre energien må senkes når vi får økning i volum ved fast entropi. Vi sier at trykk er _drivkraften_ for volumendring.
