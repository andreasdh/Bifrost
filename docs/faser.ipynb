## Faser og faseoverganger

### Hva er en fase?

:::{admonition} Definisjon - Fase
:class: important

En **fase** er en del av et system som er ensartet, både i kjemisk sammensetning og fysisk tilstand.

Ofte kan fase være synonymt med **aggregattilstand**, som er de fire tilstandene et stoff kan være i - fast stoff, væske, gass eller plasma. Men en fase kan også være sammensatt av flere kjemiske forbindelser.
:::

Legg merke til at *fase* ikke er det samme som *tilstand*. Is og vann er to faser, men en blanding av isbiter i vann er også to faser, selv om isbitene er adskilte. Det avgjørende er at hver fase er homogen i seg selv, med en tydelig grenseflate mot de andre.



### Faseoverganger

Ved endring i for eksempel temperatur eller trykk, kan én fase endres til en annen fase. Dette kaller vi en **faseovergang**.


:::{admonition} Definisjon - Faseovergang
:class: important

En **faseovergang** er en endring fra en fase til en annen, og oppstår ved endring i for eksempel trykk eller temperatur.
:::


Vi skal i all hovedsak se på to typer faseoverganger: strukturelle faseoverganger og faseoverganger for blandinger. 


::::{grid}
:gutter: 2

:::{grid-item-card} Strukturelle faseoverganger
Et stoff endrer struktur uten at den kjemiske sammensetningen endres. For grunstoffer kalles dette **allotropi**, mens for forbindelser kalles dette **polymorfi**.


*Eksempel: Rent, fast karbon kan finnes i flere former - f.eks. grafitt og diamant.*
:::

:::{grid-item-card} Faseoverganger for blandinger
Faseoverganger som involverer mer enn én forbindelse, og som ofte er litt mer kompliserte enn strukturelle faseoverganger.

*Eksempel: Legeringen av kobber og nikkel - de to metallene har samme krystallstruktur og er fullstendig blandbare i fast tilstand der de danner en fast løsning uansett blandingsforhold. En slik legering smelter ikke ved én enkelt temperatur, men over et intervall: ved en gitt sammensetning begynner den å smelte ved solidustemperaturen, og er først fullstendig flytende ved den høyere liquidustemperaturen. I mellom eksisterer fast og flytende fase side om side i likevekt.*
:::
::::

## Kjemisk potensial

### Hva er kjemisk potensial?

Ordet *potensial* beskriver muligheten for noe. I termodynamikk er derfor **kjemisk potensial** muligheten for at et stoff forandrer seg, og vi skal se at det er svært sentralt for å beskrive faser og faseoverganger.

:::{admonition} Definisjon - Kjemisk potensial
:class: important

Det **kjemiske potensialet** $\mu_i$ til et stoff $i$ er 

$$\mu_i = \left( \frac{\partial G}{\partial n_i} \right)_{T,P,n_j} = \left( \frac{\partial F}{\partial n_i} \right)_{T,V,n_J} = \left( \frac{\partial U}{\partial n_i} \right)_{S,V,n_j}$$

der $G$ er Gibbs fri energi, $F$ er Helmholtz fri energi, $U$ er indre energi, $n_i$ er stoffmengden til stoffet vårt $i$, $n_j$ er stoffmengden til et annet stoff, $T$ er temperaturen, $P$ er trykket, $S$ er entropien og $V$ er volumet.
:::

Ved første øyekast er formlene for kjemisk potensial lite intuitive, men de er bare   forskjellige sider av samme sak. Termodynamikkens 2.lov garanterer at *universets* totale entropi maksimeres, og universet er summen av systemet vårt og omgivelsene rundt. For å slippe å styre med omgivelsene og kun fokusere systemet vårt, har vi utledet ulike potensialer som vi ved å minimére, maksimerer entropi. Hvilket potensial som er relevant for vårt system, avhenger av hvilke andre størrelser vi holder konstante. For eksempel, ved konstant temperatur $T$ og trykk $P$, maksimerer vi entropi $S$ ved å minimere potensialet Gibbs fri energi $G$. En oppsummering av de andre potensialene og deres konstante størrelser er gitt i tabellen under.


| Størrelsen i minimerer | Konstant |
| ---------------------- | -------- |
| $U$ | $S, V$ |
| $H$ | $S,P$ |
| $F$ | $T,V$ |
| $G$ | $T,P$ |

Legg nå merke til at de størrelsene som er konstante for de ulike potensialene, tilsvarer de samme størrelsene som er konstante i de partiellderiverte for sitt potensial i formelen for kjemisk potensial. I tillegg er de andre stoffenes stoffmengder, $n_j$, konstante. Derfor blir kjemisk potensial et mål på hvor mye fri energi systemet endrer seg med når vi tilfører en liten mengde av stoff $i$, mens alt annet holdes fast. Det kan også forstås som *tendensen til å forlate fasen sin*. 

For eksempel for stoff $i$, ved konstant $T$, $P$ og $n_j$:

* Stoff $i$ vil bli i fasen sin $\to$ det koster lite fri energi å beholde stoffet $i$ i fasen sin $\to$ $\left( \frac{\partial G}{\partial n_i} \right)_{T,P,n_j}$ er liten $\to$ $\mu_i$ er liten

* Stoff $i$ vil forlate fasen sin $\to$ det koster mye fri energi å beholde stoffet $i$ i fasen sin $\to$ $\left( \frac{\partial G}{\partial n_i} \right)_{T,P,n_j}$ er stor $\to$ $\mu_i$ er stor



### Likevekt

I senere temaer skal vi se at kjemisk potensial gir opphav til et annet viktig konsept i termodynamikken, nemlig **likevekt**. Akkurat nå må du bare akseptere det! Tenk på det slik: hvis vi har to eller flere faser og alle stoffene er *fornøyde* i sin fase, tjener de ikke på noe ved å forlate fasen sin, og derfor er kjemisk potensial likt for alle fasene. 


:::{admonition} Definisjon - Likevekt
:class: important

To faser A og B er i **likevekt** når deres kjemiske potensialer er like, det vil si

$\mu_A = \mu_B$
:::

Det er viktig å skille mellom to spørsmål. Termodynamikken forteller hvilken fase som *bør* være stabil - den med lavest $\mu$ - men ikke hvor fort overgangen skjer. Farten er kinetikkens domene, og noen ganger er den så langsom at en ustabil fase blir værende nesten uendelig lenge.



### Endringer i kjemisk potensial med temperatur og trykk

Siden faser kan skifte når $\mu$ endres, må vi vite hvordan det kjemiske potensialet avhenger av temperatur og trykk. Begge sammenhengene følger direkte av egenskapene til Gibbs-energien.

For temperatur $T$ gjelder

$$\left(\frac{\partial \mu}{\partial T}\right)_P = -S_m$$

der $S_m$ er den molare entropien.


Fordi entropien alltid er positiv, synker det kjemiske potensialet når temperaturen øker - og stigningstallet er nettopp $-S_m$. Her ligger nøkkelen til smelting og koking: gass har høyest molar entropi, væske lavere, og fast stoff lavest. Kurvene for $\mu$ mot $T$ heller derfor brattest nedover for gass og slakest for fast stoff.


For trykk $P$ gjelder tilsvarende

$$\left(\frac{\partial \mu}{\partial p}\right)_T = V_m$$

der $V_m$ er det molare volumet.


Siden molart volum alltid er positivt, øker det kjemiske potensialet når trykket øker. For de fleste stoffer har fast fase mindre molart volum enn væske, så økt trykk hever væskens $\mu$ mer enn faststoffets - og frysepunktet stiger med trykket.


### Metastabil fase

Ved noen tilfeller, er det faktisk slik at en fase som har høyere kjemisk potensial er kinetisk hindret, og at stoffet vårt derfor er i en fase med lavere kjemisk potensial. Slike faser kalles **metastabile faser**.


:::{admonition} Definisjon - Metastabil fase
:class: important

En **metastabil fase** er en fase som har høyere kjemisk potensial enn en annen tilgjengelig fase, men som likevel består fordi overgangen er kinetisk hindret. 
:::

Diamant er det klassiske eksempelet. Ved romtemperatur og normalt trykk har grafitt lavere kjemisk potensial enn diamant, så termodynamisk *burde* diamant omdannes til grafitt. At det ikke skjer, skyldes at omorganiseringen av karbonatomene krever en enorm aktiveringsenergi - overgangen er så langsom at den i praksis aldri finner sted. Diamant er altså termodynamisk ustabil, men kinetisk fastlåst.


Vi lar kjemisk potensial og likevekt ligge for nå, og kommer tilbake til det i mer detalj i de neste kapitlene.

## Fasediagrammer

### Hva er et fasediagram?

Som kjemikere er vi interessert i ulike egenskaper ved faseoverganger fordi de avgjør når og hvordan et stoff endrer seg - noe som er avgjørende for både separasjon, syntese og analyse i praksis. Derfor er **fasediagrammer** nyttige verktøy når vi studerer faser. Et slikt diagram har temperatur $T$ på x-aksen, og trykk $P$ på y-aksen, og viser de ulike områdene der hver fase for et stoff er stabil. 



### Viktige egenskaper ved fasediagrammer

Det er noen spesielle trekk ved fasediagrammer som er verdt å merke seg:

* **Koeksistenslinjer** - en linje mellom to faser A og B, der fasene er i likevekt med hverandre

* **Trippelpunkt** - et punkt mellom tre faser A, B og C, der fasene er i likevekt med hverandre

* **Kritisk punkt** - punktet der skillet mellom væske og gass forsvinner


Et omtrentelig fasediagram for vann er vist nedenfor, med eksempel på koekeistenslinje, trippelpunkt og kritisk punkt.

![Fasediagram vann](bilder/fasediagram_vann.svg)

Det er også noen spesielle temperaturer og trykk man bør vite om, som man ofte møter på på laboratoriet:

| Navn | Symbol | Definisjon |
| ---- | ------ | ---------- |
| Overgangstemperatur | $T_{trs}$ | Temperaturen for når en faseovergang skjer, gitt et bestemt trykk. |
| Damptrykk | $P*$ | Trykket dampen utøver i likevekt med den kondenserte fasen, ved en gitt temperatur. |
| Kokepunkt | $T_b$ | Temperaturen der væskens damptrykk er lik det ytre trykket. |
| Smeltepunkt | $T_m$ | Temperaturen der fast og flytende fase er i likevekt, ved et gitt trykk. |
| Kritisk temperatur | $T_c$ | Temperaturen |

## Faseregler

Vi har hittil sett at fasediagrammet fungerer som et "kart" over *hvilke* faser som opptrer. Vi skal nå se at **fasereglene** svarer på *hvor mange* av disse fasene som kan eksistere samtidig, og *hvor fritt* vi kan endre trykk og temperatur uten at fasene endrer seg.


### Frihetsgrader, konstituenter og komponenter

Før vi begynner å snakke om antallet faser som eksisterer samtidig, må vi ha noen begreper. Vi introduserer **frihetsgrader**, **konsitutenter** og **komponenter**.


:::{admonition} Definisjon - Frihetsgrader (varians)
:class: important

**Frihetsgradene** $F$ er antall intensive variabler - som trykk og temperatur - vi kan endre uavhengig av hverandre uten å forandre antall faser i likevekt.
:::


:::{admonition} Definisjon - Konstituent
:class: important

En **konstituent** er en hvilken som helst kjemisk art som er til stede i systemet.
:::


:::{admonition} Definisjon - Komponent
:class: important

En **komponent** er en kjemisk uavhengig konstituent. **Antall komponenter** $C$ er det minste antallet uavhengige arter som trengs for å angi sammensetningen av alle fasene i systemet.
:::



### Gibbs' faseregel

Gibbs' faseregel gir oss et enkelt regnestykke som knytter sammen antall komponenter i systemet, antall faser som er i likevekt, og antall frihetsgrader vi står igjen med - altså hvor mange av de intensive variablene som trykk og temperatur vi fritt kan endre uten å endre antall faser.

:::{admonition} Gibbs' faseregel
:class: important

For et system i likevekt gjelder

$$F = C - P + 2$$

der $F$ er antall frihetsgrader, $C$ antall komponenter og $P$ antall faser.

For et rent stoff er $C = 1$, og faseregelen forenkles til

$$F = 3 - P$$
:::


Dette forklarer nøyaktig strukturen i fasediagrammet:

* **Én fase** ($P=1$) gir $F=2$. Vi kan endre både trykk og temperatur fritt - fasen dekker et *område*.
* **To faser i likevekt** ($P=2$) gir $F=1$. Nå kan bare én variabel velges fritt; den andre følger med. Likevekten er en *linje*, en fasegrense.
* **Tre faser i likevekt** ($P=3$) gir $F=0$. Ingenting kan endres - dette skjer bare i ett eneste *punkt*, trippelpunktet.


:::{card} 
**Eksempel**
^^^
Vi bruker Gibbs' faseregel på rent vann ($C=1$) i tre ulike situasjoner i fasediagrammet.

**Vann som bare væske.** Her er én fase til stede, $P=1$, så

$F = C - P + 2 = 1 - 1 + 2 = 2$

Vi har to frihetsgrader: både trykk og temperatur kan endres fritt uten at vi forlater væskeområdet. Det er derfor væske dekker en hel *flate* i diagrammet.

**Vann som koker.** Nå er væske og damp i likevekt, altså $P=2$, og

$F = 1 - 2 + 2 = 1$

Bare én frihetsgrad. Velger vi temperaturen, er trykket samtidig låst - vi kan ikke endre begge uten å forlate koeksistensen. Derfor er væske-damp-likevekten en *linje* i diagrammet.

**Vann i trippelpunktet.** Her er alle tre faser i likevekt, $P=3$, så

$F = 1 - 3 + 2 = 0$

Ingen frihetsgrader. Verken trykk eller temperatur kan endres uten at én av fasene forsvinner. Trippelpunktet er derfor et enkelt *punkt*, fastlåst til én bestemt kombinasjon av trykk og temperatur - for vann er det $273{.}16\,\text{K}$ og $611\,\text{Pa}$.
:::

## Fasegrenser

Fasegrensene er linjene i fasediagrammet der to faser møtes og er i likevekt, og siden likevekt betyr likt kjemisk potensial i begge faser, kan vi bruke nettopp denne betingelsen til å regne ut hvor bratt en fasegrense løper - altså hvordan overgangstemperaturen endrer seg når vi endrer trykket.


### Clapeyrons likning

**Clapeyrons likning** gir oss stigningstallet til en hvilken som helst fasegrense, altså nøyaktig hvor mye overgangstemperaturen endrer seg når trykket endres. Det anbefales sterkt å se på utledningen for likningen, gitt i boksen under formelen.


:::{admonition} Clapeyrons likning
:class: 

Langs en hvilken som helst fasegrense gjelder

$$\frac{dP}{dT} = \frac{\Delta_{trs}S}{\Delta_{trs}V}$$

der $\Delta_{trs}S$ og $\Delta_{trs}V$ er endringen i molar entropi og molart volum ved overgangen.
:::


:::{admonition} Utledning av Clapeyrons likning
:class: important
, dropdown

**Antakelse 1: Vi beveger oss langs en fasegrense.** To faser $\alpha$ og $\beta$ er i likevekt, så de har likt kjemisk potensial:

$\mu_\alpha = \mu_\beta$

**Antakelse 2: Vi holder oss på grensen mens vi endrer $T$ og $p$ litt.** Beveger vi oss et lite stykke langs grensen, må potensialene fortsatt være like etterpå. Endringen i $\mu$ må derfor være den samme for begge faser:

$d\mu_\alpha = d\mu_\beta$

**Vi bruker hvordan $\mu$ endrer seg.** For et rent stoff er $\mu = G_m$, og fra $dG_m = -S_m\,dT + V_m\,dp$ har vi for hver fase

$d\mu = -S_m\,dT + V_m\,dp$

Setter vi dette inn på begge sider av $d\mu_\alpha = d\mu_\beta$:

$-S_{\alpha}\,dT + V_{\alpha}\,dp = -S_{\beta}\,dT + V_{\beta}\,dp$

**Vi samler leddene.** Flytter vi $dT$-leddene til én side og $dp$-leddene til den andre:

$(V_{\beta} - V_{\alpha})\,dp = (S_{\beta} - S_{\alpha})\,dT$

Med $\Delta_{trs}V = V_\beta - V_\alpha$ og $\Delta_{trs}S = S_\beta - S_\alpha$ gir dette

$\frac{dp}{dT} = \frac{\Delta_{trs}S}{\Delta_{trs}V}$

**Antakelse 3: Overgangen skjer ved likevekt.** Da er $\Delta_{trs}G = 0$, og siden $\Delta G = \Delta H - T\Delta S$ følger det at $\Delta_{trs}S = \Delta_{trs}H / T_{trs}$. Setter vi inn dette, får vi den formen som er lettest å bruke i praksis:

$\frac{dp}{dT} = \frac{\Delta_{trs}H}{T_{trs}\,\Delta_{trs}V}$

Merk at denne utledningen er **eksakt** - vi har ikke gjort noen tilnærminger, bare brukt at potensialene er like langs grensen. Det er først når vi går videre til Clausius-Clapeyron for væske-damp-grensen at vi antar ideell gass og neglisjerbart væskevolum.
:::


For **fast-væske-grensen** setter vi inn at $\Delta_{fus}S = \Delta_{fus}H / T$, siden overgangen skjer ved likevekt. Det gir

$$\frac{dP}{dT} = \frac{\Delta_{fus}H}{T\,\Delta_{fus}V}$$

Smelteentalpien er stor og volumendringen ved smelting liten, så denne grensen er svært bratt - det skal enorme trykkendringer til for å flytte smeltepunktet merkbart. For de fleste stoffer er $\Delta_{fus}V > 0$ og grensen heller forover; for vann er $\Delta_{fus}V < 0$, og grensen heller bakover.




### Clausus-Clapeyrons likning

Clausius–Clapeyrons likning er en forenklet form av Clapeyrons likning for **væske–damp- og faststoff–damp-grensene**, der vi antar at dampen oppfører seg som en ideell gass og at den kondenserte fasens volum er neglisjerbart sammenlignet med dampens. Også utledningen til denne likningen ligger under formelen, se gjerne på den!


:::{admonition} Clausius–Clapeyrons likning
:class: important


For væske-damp-grensen gjelder

$$\frac{d\ln P}{dT} = \frac{\Delta_{vap}H}{RT^2}$$

Antar vi at $\Delta_{vap}H$ er konstant over temperaturintervallet, kan vi integrere og få

$P = P^* e^{-\chi}, \qquad \chi = \frac{\Delta_{vap}H}{R}\left(\frac{1}{T} - \frac{1}{T^*}\right)$

der $P^*$ er damptrykket ved en referansetemperatur $T^*$.
:::


:::{admonition} Utledning av Clausius–Clapeyrons likning
:class: note, dropdown

Vi tar utgangspunkt i Clapeyrons likning for væske-damp-grensen:

$\frac{dp}{dT} = \frac{\Delta_{vap}H}{T\,\Delta_{vap}V}$

**Antakelse 1: Væskens volum er neglisjerbart.** Damp har langt større molart volum enn væske - typisk rundt tusen ganger. Volumendringen ved fordamping er derfor tilnærmet lik dampens volum alene:

$\Delta_{vap}V = V_m(g) - V_m(l) \approx V_m(g)$

**Antakelse 2: Dampen er en ideell gass.** Da er $V_m(g) = RT/p$, og innsatt gir det

$\frac{dp}{dT} = \frac{\Delta_{vap}H}{T \cdot RT/p} = \frac{p\,\Delta_{vap}H}{RT^2}$

**Vi skriver om til logaritmisk form.** Deler vi begge sider på $p$ og bruker at $\frac{1}{p}\frac{dp}{dT} = \frac{d\ln p}{dT}$, får vi den kompakte formen

$\frac{d\ln p}{dT} = \frac{\Delta_{vap}H}{RT^2}$

**Antakelse 3: Fordampningsentalpien er konstant.** Over et ikke for stort temperaturintervall varierer $\Delta_{vap}H$ lite, så vi behandler den som konstant og integrerer fra en referansetilstand $(T^*, p^*)$ til $(T, p)$:

$\int_{p^*}^{p} d\ln p = \frac{\Delta_{vap}H}{R}\int_{T^*}^{T} \frac{dT}{T^2}$

Venstre side gir $\ln(p/p^*)$, og høyre side gir $-\frac{\Delta_{vap}H}{R}\left(\frac{1}{T} - \frac{1}{T^*}\right)$, siden den antideriverte av $1/T^2$ er $-1/T$. Altså

$\ln\frac{p}{p^*} = -\frac{\Delta_{vap}H}{R}\left(\frac{1}{T} - \frac{1}{T^*}\right)$

**Til slutt løser vi for $p$** ved å opphøye $e$ på begge sider:

$p = p^* e^{-\chi}, \qquad \chi = \frac{\Delta_{vap}H}{R}\left(\frac{1}{T} - \frac{1}{T^*}\right)$

Mens Clapeyrons likning var eksakt, hviler denne på tre tilnærminger: neglisjerbart væskevolum, ideell gass og konstant fordampningsentalpi. Alle tre er som regel gode et godt stykke under det kritiske punktet, men svikter når man nærmer seg det, der dampen blir tett og langt fra ideell.
:::



### Clapeyron og Clausius-Clapeyron likningene i praksis

Både Clapeyron og Clausius-Clapeyron er direkte måleverktøy, ikke bare teori.

For **væske-damp-grensen** er Clausius-Clapeyrons likning på logaritmisk form en rett linje: $\ln p = -\frac{\Delta_{vap}H}{R}\cdot\frac{1}{T} + \text{konstant}$. Måler du damptrykket ved noen få temperaturer og plotter $\ln p$ mot $\frac{1}{T}$, får du en linje med stigningstall $-\frac{\Delta_{vap}H}{R}$, og kan lese av **fordampningsentalpien** direkte - uten å måle varme. Ofte er trykket dessuten allerede kjent, for eksempel når du måler ved atmosfæretrykk, slik at du bare trenger å registrere temperaturene.

For **fast-væske-grensen** brukes Clapeyrons likning på samme måte. Måler du hvordan smeltepunktet forskyver seg når trykket endres, altså stigningstallet $\frac{dp}{dT}$ langs grensen, kan du regne ut **smelteentalpien** $\Delta_{fus}H$ - forutsatt at du kjenner volumendringen ved smelting.


:::{card} 
**Eksempel**
^^^
Is og vann er i likevekt ved $273{.}15\,\text{K}$ når trykket er $1{.}0\,\text{bar}$. Øker vi trykket til $136\,\text{bar}$, synker smeltepunktet til $272{.}15\,\text{K}$. Molart volum er $18{.}02\,\text{cm}^3/\text{mol}$ for vann og $19{.}65\,\text{cm}^3/\text{mol}$ for is. Finn smelteentalpien $\Delta_{fus}H$.

Først finner vi stigningstallet til fast-væske-grensen fra de to punktene. Trykkendringen er $\Delta p = 135\,\text{bar} = 1{.}35 \times 10^7\,\text{Pa}$ og temperaturendringen $\Delta T = -1{.}0\,\text{K}$, så

$\frac{dp}{dT} \approx \frac{\Delta p}{\Delta T} = \frac{1{.}35 \times 10^7}{-1{.}0} = -1{.}35 \times 10^7\,\text{Pa/K}$

Så løser vi Clapeyrons likning for smelteentalpien:

$\frac{dp}{dT} = \frac{\Delta_{fus}H}{T\,\Delta_{fus}V} \quad \Rightarrow \quad \Delta_{fus}H = \frac{dp}{dT}\cdot T \cdot \Delta_{fus}V$

Volumendringen ved smelting er

$\Delta_{fus}V = V_m(\text{væske}) - V_m(\text{is}) = 18{.}02 - 19{.}65 = -1{.}63\,\text{cm}^3/\text{mol} = -1{.}63 \times 10^{-6}\,\text{m}^3/\text{mol}$

Legg merke til at volumendringen er **negativ** - is har større volum enn vann. Vi setter inn:

$\Delta_{fus}H = (-1{.}35 \times 10^7) \cdot 273{.}15 \cdot (-1{.}63 \times 10^{-6}) \approx 6{.}0 \times 10^3\,\text{J/mol}$

Smelteentalpien er altså $\Delta_{fus}H \approx 6{.}0\,\text{kJ/mol}$, som stemmer godt med den kjente verdien for vann.

Merk at de to negative fortegnene - fra stigningstallet og fra volumendringen - opphever hverandre, slik at smelteentalpien blir positiv. Det må den også være, siden smelting alltid krever tilført energi.
:::


## Første- og andreordens faseoverganger

Alle faseovergangene vi har sett på så langt - smelting, koking, sublimasjon - deler ett trekk: de krever tilført varme, og temperaturen er konstant mens overgangen pågår. Men ikke alle faseoverganger oppfører seg slik. For å skille de ulike typene ser vi på hvordan det kjemiske potensialet endrer seg gjennom overgangen. Denne inndelingen skyldes Ehrenfest, og bygger direkte på de deriverte vi allerede kjenner: $\left(\frac{\partial \mu}{\partial T}\right)_p = -S_m$ og $\left(\frac{\partial \mu}{\partial p}\right)_T = V_m$.


### Førsteordens faseoverganger

:::{admonition} Definisjon - Førsteordens faseovergang
:class: important

En **førsteordens faseovergang** er en overgang der de førstederiverte av det kjemiske potensialet - molar entropi $S_m$ og molart volum $V_m$ - endrer seg diskontinuerlig, altså med et sprang, ved overgangstemperaturen.
:::

Fordi entropien hopper, er det knyttet en **latent varme** til overgangen: $\Delta_{trs}H = T_{trs}\,\Delta_{trs}S \neq 0$. Det er denne varmen som tilføres uten at temperaturen stiger mens overgangen pågår - nettopp den pausen vi bruker til å oppdage overgangen ved termisk analyse.

Smelting, fordamping og sublimasjon er alle førsteordens faseoverganger. Entropien og volumet endrer seg brått: is og vann har målbart forskjellig entropi og tetthet. Det er nettopp fordi $\Delta_{trs}S$ og $\Delta_{trs}V$ er forskjellige fra null at Clapeyrons likning kan beskrive disse grensene.


### Andreordens faseoverganger

:::{admonition} Definisjon - Andreordens faseovergang
:class: important

En **andreordens faseovergang** er en overgang der de førstederiverte av det kjemiske potensialet ($S_m$ og $V_m$) er kontinuerlige, mens de andrederiverte - som varmekapasitet - endrer seg diskontinuerlig.
:::

Konsekvensen er avgjørende: siden entropien er kontinuerlig, er det **ingen latent varme** ved en andreordens overgang, og heller ingen brå volumendring. Stoffet skifter karakter uten at vi trenger å tilføre eller fjerne varme i selve overgangspunktet.

Slike overganger er mer subtile enn smelting og koking. Eksempler er overgangen til superledende tilstand i visse metaller og overgangen til superflytende helium. Her ligger informasjonen ikke i en latent varme, men i hvordan varmekapasiteten oppfører seg - den gjør et sprang, eller stiger skarpt mot en topp, ved overgangen.


Ordenen forteller altså hvilken derivert av det kjemiske potensialet som først blir diskontinuerlig:

* **Første orden:** $\mu$ er kontinuerlig, men den førstederiverte hopper. Entropi og volum spretter, og det finnes latent varme.
* **Andre orden:** $\mu$ og førstederiverte er kontinuerlige, men den andrederiverte hopper. Ingen latent varme, men varmekapasiteten endrer seg brått.

Selve det kjemiske potensialet er alltid kontinuerlig gjennom en faseovergang - ellers ville stoffet hoppet mellom to ulike verdier av fri energi, noe som ikke gir fysisk mening. Det er hvor skarpt kurven bøyer, ikke selve verdien, som skiller de to typene.
