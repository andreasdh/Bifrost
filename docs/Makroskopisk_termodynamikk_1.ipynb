## Introduksjon til viktige størrelser i klassisk termodynamikk

I forbindelse med den industrielle revolusjonen ble samspillet mellom gass, varme og arbeid særlig viktig. Vitenskapsmenn og oppfinnere fikk en ny forståelse av hvordan energi utveksles mellom disse parameterene, og kunne utnytte dette til å gjøre transport og produksjon av varer langt mer effektivt! I denne første delen skal vi introdusere disse størrelsene som senere blir viktige for å forstå klassiske termodynamiske uttrykk. 

### Varme i historien

Gjennom historien har menneskeheten forklart varme på ulike måter. På 1700-tallet trodde man at varme var en vektløs væske som flyter fra et varmt stoff til et kaldt stoff (Calorique-teorien). Dette ble senere kritisert da Joseph Black i 1750 oppdaget at vann rundt $100^\circ\mathrm{C}$ kunne ta opp og avgi varme uten at temperaturen ble endret. Dette kalte han latent varme. Ti år senere oppdaget Black at vann og kvikksølv som ble varmet over samme flamme hadde ulik temperatur. I dag forklarer vi disse observasjonene ved hjelp av _varmekapasitet_, og dette skal vi se nærmere på senere i kapitellet. I slutten av 1700-tallet bemerket Count Rumford en sammenheng mellom mekanisk energi og varme i forbindelse med boring av kanoner. Dette ble studert videre av Joule som var den første til å tolke temperaturøkning som følge av friksjon som en overføring av kinetisk mekanisk energi til varme. 

### Varme i dag

I moderne termodynamikk ser vi varme $q$ som en måte energi overføres på, ikke som en substans. Når et system mottar varme, øker energien i de mikroskopiske frihetsgradene – translasjon, rotasjon, vibrasjon og elektroniske tilstander. Husk hva dere lærte om disse på forrige side! I statistisk mekanikk betyr dette at varme er energi som endrer sannsynlighetsfordelingen over mikrotilstandene systemet kan befinne seg i. Flere mikrotilstander blir tilgjengelige, og systemet “flytter seg” til en mer sannsynlig fordeling. Varme overføres kun når det er temperaturforskjell mellom to systemer, og eksisterer derfor bare mens energi faktisk flyter. Etter overføringen snakker vi ikke om “varmeinnhold”, men om indre energi – den totale mikroskopiske energien som nå ligger lagret i systemets frihetsgrader. Kort sagt: varme er prosessen som endrer den mikroskopiske energien i et system, ikke en egen form for energi. Når varme er overført, er alt som gjenstår den indre energien som nå er fordelt på molekylenes frihetsgrader. 

### Arbeid og reversibilitet

Arbeid er den energien som overføres når et system "tvinges" til å endre en makroskopisk variabel. I fysikken kan dette være alt fra å komprimere en gass (endre volum) og trekke i en fær (endre lengde) til å flytte en ladning ved hjelp av et elektrisk felt (endring i posisjon). Matematisk er arbeid definert som produktet av en kraft F og forflytningen i kraftens retning s. 

$$w = -F \times s$$

Legg merke til konvensjonen om fortegn! Dersom $w<0$, forlater energi systemet, altså har systemet gjort arbeid på omgivelsene. Dersom $w>0$ får systemet energi, altså har omgivelsene gjort arbeid på systemet.

I kjemi er vi særlige interessert i tilfellet hvor vi ekspanderer en gass mot et konstant ytre trykk. Vi kan se for oss en gass i en sylinder som har et bevegelig stempel slik at gassen kan få mindre eller større volum. Stempelet er rundt og har et areal og gassen kan utgjøre et trykk p for å skyve stempelet utover. Kraften fra gassen på stempelet er lik $F = pA$. Videre kan vi se for oss at stempelet beveger seg med en liten avstand dx får vi følgende uttrykk for arbeidet:

$$dw = -F \times dx = pA \times dx$$

Vi kan også beskrive hvordan volumet tilgjengelig for gassen i sylinderen endrer seg. Volumet er gitt ved å gange sammen stempelets areal og lengden til sylinderen. For volumendringen får vi altså: $dV = Adx \leftrightarrow dx = \frac{dV}{A}$. Vi kan sette dette inn i uttrykket vårt for arbeid:

$$dw = pA \times dx \leftrightarrow dw = pA \times \frac{dV}{A} \leftrightarrow dw = p \times dV$$

For å beskrive arbeidet som utføres i løpet av prosessen kan vi integrere fra start- til sluttvolumet på denne måten: 

$$w = - \int_{V_i}^{V_f} pdV$$

Dersom vi har et konstant ytre trykk $p_{ex}$ kan vi trekke trykket utenfor integralet og står igjen med et enklere uttrykk:

$$w = - \int_{V_i}^{V_f} p_{ex}dV \leftrightarrow w = - p_{ex} \int_{V_i}^{V_f} dV = -p_{ex}\Delta V$$

Men hvordan kan vi utnytte en slik prosess på best mulig måte? Hvordan kan vi bruke 100% av energien vår til å forflytte stempelet og unngå å miste energi som varme på grunn av friksjon? Svaret på disse spørsmålene er at prosessen må være _reversibel_, kort forklart at den må gjøres uendelig sakte, det vil si deles inn i uendelig mange små deler! For å visualisere dette i praksis kan du tenke deg at vi har sand som ligger oppå stempelet og presser gassen sammen i sylinderen. For å få til en reversibel prosess kan du plukke av ett og ett sandkorn med en pinsett slik at stempelet beveger seg bare bittelitt mellom hvert av dem. På denne måten blir friksjonen mellom stempelet og sylinderen minimal og prosessen blir derfor tilnærmet reversibel. Du kan utforske dette eksempelet ved hjelp av simuleringen du finner nedenfor!

## Indre energi og termodynamikkens første lov

Nå som vi kan litt mer om varme og arbeid, kan vi definere størrelsen vi kaller indre energi! Indre energi betegnes ved bokstaven U, og kan defineres både fra et mikroskopisk perspektiv ved å benytte statistisk mekanikk og fra et makroskopisk perspektiv gjennom termodynamikkens første lov. Vi tar en kort recap av den mikroskopiske definisjonen fra kapitellet om statistisk mekanikk, før vi går løs på indre energi som en klassisk termodynamisk størrelse🥰

På det mikroskopiske plan defineres den indre energien som summen av alle de mikroskopiske energiene som finnes i systemet: Kinetisk energi fra molekylenes translasjon (når de flytter seg fra en posisjon med koordinater (x, y, z) til en annen), rotasjonsenergi, vibrasjonsenergi, elektronisk energi og potensiell energi fra intermolekylære krefter:

$$U = E_{trans} + E_{rot} + E_{vib} + E_{elek} + E_{pot}$$

Legg merke til vi benytter oss av stor bokstav E for å betegne energien! Dette gjør vi fordi vi snakker om en stor mengde partikler og ikke én og én slik vi gjerne gjør i statistisk mekanikk. Den totale energien $E$ er lik summen over energiene ${\epsilon}_i$ fra de individuelle partiklene som utgjør systemet:

$$E = {\epsilon}_1 +  {\epsilon}_2 + ... +  {\epsilon}_i$$

I stastisk mekanikk er den indre energien definert som den gjennomsnittlige energien over alle mikrotilstander, som beskrevet av Boltzmann fordelingen: 

$$U = ⟨E⟩$$

Dette kan du lese mer om i det foregående kapitellet om statistisk mekanikk. Der vises det også hvordan man matematisk kan gå fra den mikroskopiske definisjonen til den makroskopiske definisjonen!

På det makroskopiske plan er den indre energien definert ved hjelp av termodynamikkens første lov: Energien i et lukket system er bevart, altså kan ikke skapes eller forsvinne men kun skifte form. 

```{admonition} Termodynamikkens første lov
:class: important

Energien i et lukket system er bevart. Den kan ikke skapes eller forsvinne, men kan bytte form via varmeoverføring eller arbeid. Den indre energien til systemet har derfor uttrykket:

$$U = q + w$$
```

Deersom vi systemet går fra en tilstand A til en tilstand B, får vi en endring i indre energi som er lik er summen av varme og arbeid som har blitt tilført systemet i løpet av prosessen:

$$\Delta U = U_B - U_A = q + w$$

Selv om q og w er såkalte stifunksjoner og endrer seg avhengig av veien fra start- til sluttilstand, er den indre energien derimot en tilstandsfunksjon: Den er kun avhengig av hvilket trykk og temperatur systemet befinner seg i start- og sluttilstanden, og hvor mange molekyler det består av. Energien kan altså flyte fritt mellom de to formene arbeid og varme, men summen av dem vil alltid gi den samme økningen/senkingen av energien. Mer arbeid betyr mindre varme og vice versa. Et intuitivt eksempel kan være en fjelltur: Vi kan ta mange ulike stier opp et fjell (arbeid), og spise en variende mengde Kvikk Lunsj for å komme oss opp fjellet (varme), men til slutt vil man uansett ha kommet seg fra A til B og det er dette som representerer den indre energien. ⛰️

Vi kan også finne eksempler i kjemien! Dersom vi har en reaksjon hvor vi ønsker å danne et produkt fra to reagenser er vi nødt til å få de to reagensene til å kollidere. Vi kan se for oss to måter å gjøre dette på: Vi kan mekanisk tvinge molekylene til sammen (arbeid) eller vi kan varme opp molekylene slik at de får økt bevegelse og større sjanse for å kollidere (varme). Dette korresponderer til to ulike metoder i laben! Selv om prosessene er svært ulike, vil de gi det samme ønskede produktet! Vi må allikevel huske på at dette er et svært enkelt eksempel, og at vi ofte må bruke en kombinasjon av mekaniske og termiske metoder for å få reaksjonene våre til å fungere🥼🧪

## Entalpi

I et kjemisk laboratorium skjer nesten alle reaksjoner ved konstant trykk: I åpne begerglass, kolber med en jevn strøm av inert gass eller i en reaktor. I en slik kontekst kan det kjemiske systemet kunne ta opp og gi fra seg varme, men også gjøre volumarbeid ved for eksempel dannelse av gass. Som sett i avsnittet om arbeid over kan arbeid ved konstant trykk uttrykkes som $w = -p\Delta V$. Når vi skal regne på ulike termokjemiske størrelser senere, kan det bli upraktisk å ta med dette leddet for arbeid hver gang vi ønsker å diskutere flyt av energi. Hvordan kan vi bli kvitt det? La oss se på uttrykket for størrelsen _entalpi_ H:

$$H = U + pV$$

Ved hjelp av litt algebra skjønner vi hvorfor entalpi er løsningen på problemet vårt:

$$H = U + pV$$
$$dH = dU + d(pV) = dU + Vdp + pdV$$
$$dH = dq + dw + Vdp + pdV = dq - pdV + Vdp + pdV$$
$$dH = dq + Vdp$$

Når trykket er konstant, er $dp=0$, og vi står igjen med: 

$$dH = dq_P$$

Med ord er entalpi varmen som overføres ved konstant trykk. Som vi diskuterte i starten av avsnittet foregår mange kjemiske prosesser ved konstant trykk og entalpi er derfor en svært nyttig størrelse! 

## Introduksjon til fri energi

Hvordan vet jeg om reaksjonen min kommer til å gå eller ikke? Hvor mye må jeg varme opp kolben min for at reaksjonen skal gå? Når vil likevekt innstilles? Nå har vi endelig alle verktøyene vi trenger for å diskutere fri energi! Fri energi er en størrelse som tar hensyn til både energi og entropi slik at vi som kjemikere kan forutse om prosessen vi ønsker å gjennomføre er _spontan_ eller ikke. Senere skal vi se på hva dette har å si for kjemiske reaksjoner, men for nå skal vi introdusere de to ulike typene fri energi dere må kjenne til i emnet.

### Helmholtz fri energi

Helmholtz fri energi benyttes når volumet i et system holdes konstant. Denne frie energien er vanlig å bruke i fysikk og statistisk mekanikk, men ikke like mye i kjemi. La oss først se på uttrykket for indre energi når volumet er konstant: 

$$U = q + w \leftrightarrow dU = dq + dw = dq + pdV$$

Når volumet er konstant, er dV = 0, og vi står igjen med:

$$ dU = dq_V $$

Ved konstant volum representerer altså den indre energien varmeoverføring, derfor er $U$ "energitermen" i uttrykket for Helmholtz frie energi F: 

$$F = U - TS$$

Helmholtz frie energi kan også defineres fra partisjonsfunksjonen Q. Det mikroskopiske uttrykket med Q og det makroskopiske uttrykket ovenfor er knyttet sammen matematisk i kapitellet om statistisk mekanikk. 

### Gibbs fri energi 

Gibbs fri energi benyttes når trykket i et system holdes konstant, og er derfor den frie energien vi benytter oss mest av i kjemifaget. Som vi så i delen om entalpi, er denne størrelsen et mål på varmeoverføring når trykket er konstant:

$$dH = dq_P$$

Og entalpi $H$ er derfor "energitermen" i uttrykket for Gibbs frie energi G:

$$G = H - TS$$

Den frie energien beskriver altså hvor mye energi vi har tilgjengelig for å drive en prosess fremover. Vi ser også at entropileddet er vektet av temperaturen. Samspillet mellom entalpi, entropi og temperatur vil hjelpe oss å forstå drivkreftene bak kjemiske reaksjoner senere, stay tuned!😎

## Termodynamiske drivkrefter matematisk og praktisk

Varme strømmer fra varmt til kaldt, gasser beveger seg fra høyt trykk til lavt trykk, stoffer diffunderer fra områder med høy konsentrasjon til områder med lav konsentrasjon. Dette virker kanskje intuitivt for oss, men hvorfor skjer disse prosessene egentlig? Hva har de til felles? Bak dem ligger de _termodynamiske drivkreftene_. Først skal vi se på hvordan vi matematisk kan definere de termodynamiske potensialene fra tilstandsfunksjonene. Dette gir opphav til _fundamentallikningene_, den matematiske definisjonen av de termodynamiske drivkreftene og _Maxwell-relasjonene_.  

### Fundamentallikningen for indre energi i et lukket system

La oss starte fra definisjonen av indre energi igjen: 

$$\Delta U = q + w$$

I stedet for definisjonen som inneholder stifunksjonene arbeid og varme ønsker vi nå å finne ut av hvordan vi kan uttrykke den indre energien ved hjelp av tilstandsfunksjoner som temperatur, trykk, volum og entropi. Vi benytter oss av noen uttrykk fra de forrige kapitlene og finner et nytt uttrykk for dU:

$$dS = \frac{dq}{T} \leftrightarrow dq = TdS$$
$$dw = -pdV$$
$$dU = TdS - pdV$$

Dette er en *fundamentallikning* for den indre energien og gjør indre energi til en funksjon av entropi og volum. Vi kan nå trekke ut temperatur og volum som partiellderiverte! Dersom volumet er konstant vil dV være lik 0 og vi står igjen med:

$$\delta U = T\delta S \leftrightarrow T = \left(\frac{\delta U}{\delta S}\right)_V$$

Temperatur viser altså hvor mye den indre energien må øke når vi får økning i entropi ved fast volum, for eksempel ved varmetilførsel. Vi sier at temperatur er _drivkraften_ for varmeoverføring. 

Dersom vi nå setter entropien konstant, vil dS være lik 0 og vi står igjen med:

$$\delta U = -p\delta V \leftrightarrow -p = \left(\frac{\delta U}{\delta V}\right)_S$$

Trykk viser altså hvor mye den indre energien må senkes når vi får økning i volum ved fast entropi. Vi sier at trykk er _drivkraften_ for volumendring.

### Fra indre energi til entalpi og fri energi

På samme måte som for den indre energien ønsker vi også å definere entalpi H og Helmholtz og Gibbs fri energi F og G ved hjelp av tilstandsfunksjoner. Dette blir enklere siden vi nå kan ta utgangspunkt i uttrykket vi fant for indre energi! Se på utledningen av fundamentalllikningen til entalpi først, samt hvordan vi kan finne de termodynamiske drivkreftene knyttet til entalpi. Deretter skal du få prøve deg på egenhånd for de frie energiene🔥

*Entalpi* 
Fra forrige kapittel vet vi at entalpi er definert som den indre energien pluss pV. Vi starter utledningen vår med å differensiere uttrykket og bytte ut $dU$ med uttrykket vi fant i første del av kapittelet:

$$H = U + pV$$
$$dH = dU + d(pV) = (TdS - pdV) + (pdV + Vdp)$$
$$dH = TdS + Vdp$$

Dersom vi setter trykket konstant, blir dp lik null, og vi står igjen med:

$$\delta H = T\delta S \longleftrightarrow T = \left(\frac{\partial H}{\partial S}\right)_p$$

Dersom vi setter entropien konstant, finner vi:

$$V = \left(\frac{\partial H}{\partial p}\right)_S$$

### Maxwell-relasjoner

Maxwell-relasjonene tar utgangspunkt i fundamentallikningene og beskriver noen viktige sammenhenger mellom de ulike tilstandsfunksjonene. Først må vi bruke noen matematiske triks, før vi kan trekke ut de nyttige relasjonene!

La oss se på noen egenskaper for en tilstandsstørrelse X, som kan ta to variable a og b:

$$X = X(a,b)$$

Vi kan skrive den totale differensialen for X på følgende måte, vi skal illustrere dette i praksis med fundamentallikningene straks:

$$dX = \left(\frac{\partial X}{\partial a}\right)_bda + \left(\frac{\partial X}{\partial b}\right)_adb$$

Dette kalles et _eksakt differensial_ fordi X er en tilstandsfunksjon av variablene a og b. For slike differensialer gjelder det at:

$$\frac{\partial}{\partial b}\left(\frac{\partial X}{\partial a}\right)_b = \frac{\partial}{\partial a}\left(\frac{\partial X}{\partial b}\right)_a$$

Vi sier at de _blandede andrederiverte_ er like. Denne sammenhengen er nøkkelen bak Maxwell-relasjonene! 


*Indre energi*

La oss begynne med indre energi. I forrige avsnitt fant vi følgende fundamentallikning for den indre energi: 

$$dU = TdS - pdV$$

Vi identifiserte også to uttrykk for T og -p:

$$T = \left(\frac{\partial U}{\partial S}\right)_V$$
$$-p = \left(\frac{\partial U}{\partial V}\right)_S$$

La oss prøve å sette de to uttrykkene inn i fundamentallikningen og dermed bytte ut T og -p med de partiellderiverte:

$$dU = \left(\frac{\partial U}{\partial S}\right)_VdS + \left(\frac{\partial U}{\partial V}\right)_SdV$$

Nå ser vi at vi har fått et differensial på samme form det eksakte differensialet vi definerte med X over! Da vet vi at vi kan bruke regelen om blandede deriverte. La oss først identifisere X, a og b: 

$$X = U, a = S, b = V$$

Vi skriver opp uttrykket for de blandede deriverte: 

$$\frac{\partial}{\partial V}\left(\frac{\partial U}{\partial S}\right)_V = \frac{\partial}{\partial S}\left(\frac{\partial U}{\partial V}\right)_S$$

Vi kjenner igjen uttrykkene for T og -p:

$$\frac{\partial}{\partial V}(T) = -\frac{\partial}{\partial S}(p)$$

La oss løse høyre side først. Når vi partiellderiverer T med hensyn på V, betyr det at S holdes konstant. Vi får derfor resultatet: 

$$\frac{\partial}{\partial V}(T) = \left(\frac{\partial T}{\partial V}\right)_S$$

På venstre side partiellderiverer vi p med hensyn på S, og V holdes derfor konstant. Vi står igjen med resultatet: 

$$-\frac{\partial}{\partial S}(p) = -\left(\frac{\partial p}{\partial S}\right)_V$$

Vi kan nå skrive opp den første Maxwell-relasjonen🤩

$$\left(\frac{\partial T}{\partial V}\right)_S = -\left(\frac{\partial p}{\partial S}\right)_V$$

Ved å bruke metoden vi viste over, burde du finne denne Maxwell-relasjonen: 

$$\left(\frac{\partial T}{\partial p}\right)_S = \left(\frac{\partial V}{\partial S}\right)_p$$


*Helmholtz fri energi*

For Helmholtz fri energi, skal du finne følgende fundamentallikning dersom du setter inn uttrykkene for -p og -S: 

$$dF = \left(\frac{\partial F}{\partial V}\right)_TdV + \left(\frac{\partial F}{\partial T}\right)_VdT$$

Ved å bruke metoden vi viste over, burde du finne denne Maxwell-relasjonen: 

$$\left(\frac{\partial p}{\partial T}\right)_V = \left(\frac{\partial S}{\partial V}\right)_T$$

*Gibbs fri energi*

For Gibbs fri energi, skal du finne følgende fundamentallikning dersom du setter inn uttrykkene for V og -S: 

$$dG = \left(\frac{\partial G}{\partial p}\right)_Tdp + \left(\frac{\partial G}{\partial T}\right)_pdT$$

Ved å bruke metoden vi viste over, burde du finne den siste Maxwell-relasjonen: 

$$\left(\frac{\partial V}{\partial T}\right)_p = -\left(\frac{\partial S}{\partial p}\right)_T$$
