## Introduksjon til viktige størrelser i klassisk termodynamikk

I forbindelse med den industrielle revolusjonen ble samspillet mellom gass, varme og arbeid særlig viktig. Vitenskapsmenn og oppfinnere fikk en ny forståelse av hvordan energi utveksles mellom disse parameterene, og kunne utnytte dette til å gjøre transport og produksjon av varer langt mer effektivt! I denne første delen skal vi introdusere disse størrelsene som senere blir viktige for å forstå klassiske termodynamiske uttrykk. 

### Varme i historien

Gjennom historien har menneskeheten forklart varme på ulike måter. På 1700-tallet trodde man at varme var en vektløs væske som flyter fra et varmt stoff til et kaldt stoff (Calorique-teorien). Dette ble senere kritisert da Joseph Black i 1750 oppdaget at vann rundt $100^\circ\mathrm{C}$ kunne ta opp og avgi varme uten at temperaturen ble endret. Dette kalte han latent varme. Ti år senere oppdaget Black at vann og kvikksølv som ble varmet over samme flamme hadde ulik temperatur. I dag forklarer vi disse observasjonene ved hjelp av _varmekapasitet_, og dette skal vi se nærmere på senere i kapitellet. I slutten av 1700-tallet bemerket Count Rumford en sammenheng mellom mekanisk energi og varme i forbindelse med boring av kanoner. Dette ble studert videre av Joule som var den første til å tolke temperaturøkning som følge av friksjon som en overføring av kinetisk mekanisk energi til varme. 

### Varme i dag

I moderne termodynamikk ser vi varme $q$ som en måte energi overføres på, ikke som en substans. Når et system mottar varme, øker energien i de mikroskopiske frihetsgradene – translasjon, rotasjon, vibrasjon og elektroniske tilstander. Husk hva dere lærte om disse på forrige side! I statistisk mekanikk betyr dette at varme er energi som endrer sannsynlighetsfordelingen over mikrotilstandene systemet kan befinne seg i. Flere mikrotilstander blir tilgjengelige, og systemet “flytter seg” til en mer sannsynlig fordeling. Varme overføres kun når det er temperaturforskjell mellom to systemer, og eksisterer derfor bare mens energi faktisk flyter. Etter overføringen snakker vi ikke om “varmeinnhold”, men om indre energi – den totale mikroskopiske energien som nå ligger lagret i systemets frihetsgrader. Kort sagt: varme er prosessen som endrer den mikroskopiske energien i et system, ikke en egen form for energi. Når varme er overført, er alt som gjenstår den indre energien som nå er fordelt på molekylenes frihetsgrader. 

### Arbeid og reversibilitet

Arbeid er den energien som overføres når et system "tvinges" til å endre en makroskopisk variabel. I fysikken kan dette være alt fra å komprimere en gass (endre volum) og trekke i en fær (endre lengde) til å flytte en ladning ved hjelp av et elektrisk felt (endring i posisjon). Matematisk er arbeid definert som produktet av en kraft F og forflytningen i kraftens retning s. 

$$w = -F \times s$$

Legg merke til konvensjonen om fortegn! Dersom $w<0$, forlater energi systemet, altså har systemet gjort arbeid på omgivelsene. Dersom $w>0$ får systemet energi, altså har omgivelsene gjort arbeid på systemet.

I kjemi er vi særlige interessert i tilfellet hvor vi ekspanderer en gass mot et konstant ytre trykk. Vi kan se for oss en gass i en sylinder som har et bevegelig stempel slik at gassen kan få mindre eller større volum. Stempelet er rundt og har et areal og gassen kan utgjøre et trykk p for å skyve stempelet utover. Kraften fra gassen på stempelet er lik $F = PA$. Videre kan vi se for oss at stempelet beveger seg med en liten avstand dx får vi følgende uttrykk for arbeidet:

$$dw = -F \times dx = PA \times dx$$

Vi kan også beskrive hvordan volumet tilgjengelig for gassen i sylinderen endrer seg. Volumet er gitt ved å gange sammen stempelets areal og lengden til sylinderen. For volumendringen får vi altså: $dV = Adx \leftrightarrow dx = \frac{dV}{A}$. Vi kan sette dette inn i uttrykket vårt for arbeid:

$$dw = PA \times dx \leftrightarrow dw = PA \times \frac{dV}{A} \leftrightarrow dw = P \times dV$$

For å beskrive arbeidet som utføres i løpet av prosessen kan vi integrere fra start- til sluttvolumet på denne måten: 

$$w = - \int_{V_i}^{V_f} PdV$$

Dersom vi har et konstant ytre trykk $P_{ex}$ kan vi trekke trykket utenfor integralet og står igjen med et enklere uttrykk:

$$w = - \int_{V_i}^{V_f} P_{ex}dV \leftrightarrow w = - P_{ex} \int_{V_i}^{V_f} dV = -P_{ex}\Delta V$$

Men hvordan kan vi utnytte en slik prosess på best mulig måte? Hvordan kan vi bruke 100% av energien vår til å forflytte stempelet og unngå å miste energi som varme på grunn av friksjon? Svaret på disse spørsmålene er at prosessen må være _reversibel_, kort forklart at den må gjøres uendelig sakte, det vil si deles inn i uendelig mange små deler! For å visualisere dette i praksis kan du tenke deg at vi har sand som ligger oppå stempelet og presser gassen sammen i sylinderen. For å få til en reversibel prosess kan du plukke av ett og ett sandkorn med en pinsett slik at stempelet beveger seg bare bittelitt mellom hvert av dem. På denne måten blir friksjonen mellom stempelet og sylinderen minimal og prosessen blir derfor tilnærmet reversibel. Du kan utforske dette eksempelet ved hjelp av simuleringen du finner nedenfor!

:class: tip dropdown

**a)** Reversibelt arbeid (volumene i forholdet trenger ikke gjøres om, siden $V_f/V_i$ er dimensjonsløst):

$w = -nRT\ln\frac{V_f}{V_i} = -1.0 \cdot 8.314 \cdot 298 \cdot \ln\frac{6.0}{2.0} = -2722\ \text{J}$

**b)** Vi gjør om volumene til SI: $V_i = 2.0 \times 10^{-3}\ \text{m}^3$ og $V_f = 6.0 \times 10^{-3}\ \text{m}^3$. Det ytre trykket er

$P_{ex} = \frac{nRT}{V_f} = \frac{1.0 \cdot 8.314 \cdot 298}{6.0 \times 10^{-3}} = 4.13 \times 10^{5}\ \text{Pa}$

Og det irreversible arbeidet:

$w = -P_{ex}(V_f - V_i) = -4.13 \times 10^{5} \cdot (6.0 - 2.0)\times 10^{-3} = -1652\ \text{J}$

**c)** Den reversible prosessen gir mest arbeid ut ($-2722$ mot $-1652\ \text{J}$, altså $1070\ \text{J}$ mer). Grunnen er at trykket ved reversibel ekspansjon holder seg så høyt som mulig hele veien - gassen "presser" mot et ytre trykk som til enhver tid er nesten like stort som gasstrykket. Ved irreversibel ekspansjon presser gassen bare mot et lavt, konstant ytre trykk, så den får utført mindre arbeid. Reversibel prosess gir alltid det maksimale arbeidet en gass kan yte mellom to tilstander - det er den ideelle grensen ekte prosesser aldri helt når opp til.

## Indre energi og termodynamikkens første lov

Nå som vi kan litt mer om varme og arbeid, kan vi definere størrelsen vi kaller indre energi! Indre energi betegnes ved bokstaven U, og kan defineres både fra et mikroskopisk perspektiv ved å benytte statistisk mekanikk og fra et makroskopisk perspektiv gjennom termodynamikkens første lov. Vi tar en kort recap av den mikroskopiske definisjonen fra kapitellet om statistisk mekanikk, før vi går løs på indre energi som en klassisk termodynamisk størrelse🥰

På det mikroskopiske plan defineres den indre energien som summen av alle de mikroskopiske energiene som finnes i systemet: Kinetisk energi fra molekylenes translasjon (når de flytter seg fra en posisjon med koordinater (x, y, z) til en annen), rotasjonsenergi, vibrasjonsenergi, elektronisk energi og potensiell energi fra intermolekylære krefter:

$$U = E_{trans} + E_{rot} + E_{vib} + E_{elek} + E_{pot}$$

Legg merke til vi benytter oss av stor bokstav E for å betegne energien! Dette gjør vi fordi vi snakker om en stor mengde partikler og ikke én og én slik vi gjerne gjør i statistisk mekanikk. Den totale energien $E$ er lik summen over energiene ${\varepsilon}_i$ fra de individuelle partiklene som utgjør systemet:

$$E = {\varepsilon}_1 +  {\varepsilon}_2 + ... +  {\varepsilon}_i$$

I stastisk mekanikk er den indre energien definert som den gjennomsnittlige energien over alle mikrotilstander, som beskrevet av Boltzmann fordelingen: 

$$U = ⟨E⟩$$

Dette kan du lese mer om i det foregående kapitellet om statistisk mekanikk. Der vises det også hvordan man matematisk kan gå fra den mikroskopiske definisjonen til den makroskopiske definisjonen!

På det makroskopiske plan er den indre energien definert ved hjelp av termodynamikkens første lov: Energien i et lukket system er bevart, altså kan ikke skapes eller forsvinne men kun skifte form. 

```{admonition} Termodynamikkens første lov
:class: important

Energien i et lukket system er bevart. Den kan ikke skapes eller forsvinne, men kan bytte form via varmeoverføring eller arbeid. Den indre energien til systemet har derfor uttrykket:

$$U = q + w$$
```

Deersom vi systemet går fra en tilstand A til en tilstand B, får vi en endring i indre energi som er lik er summen av varme og arbeid som har blitt tilført systemet i løpet av prosessen:

$$\Delta U = U_B - U_A = q + w$$

Selv om q og w er såkalte stifunksjoner og endrer seg avhengig av veien fra start- til sluttilstand, er den indre energien derimot en tilstandsfunksjon: Den er kun avhengig av hvilket trykk og temperatur systemet befinner seg i start- og sluttilstanden, og hvor mange molekyler det består av. Energien kan altså flyte fritt mellom de to formene arbeid og varme, men summen av dem vil alltid gi den samme økningen/senkingen av energien. Mer arbeid betyr mindre varme og vice versa. Et intuitivt eksempel kan være en fjelltur: Vi kan ta mange ulike stier opp et fjell (arbeid), og spise en variende mengde Kvikk Lunsj for å komme oss opp fjellet (varme), men til slutt vil man uansett ha kommet seg fra A til B og det er dette som representerer den indre energien. ⛰️

Vi kan også finne eksempler i kjemien! Dersom vi har en reaksjon hvor vi ønsker å danne et produkt fra to reaktanter er vi nødt til å få de to reagensene til å kollidere. Vi kan se for oss to måter å gjøre dette på: Vi kan mekanisk tvinge molekylene til sammen (arbeid) eller vi kan varme opp molekylene slik at de får økt bevegelse og større sjanse for å kollidere (varme). Selv om prosessene er svært ulike, vil de gi det samme ønskede produktet! Vi må allikevel huske på at dette er et svært enkelt eksempel, og at vi ofte må bruke en kombinasjon av mekaniske og termiske metoder for å få reaksjonene våre til å fungere🥼🧪

## Entalpi

I et kjemisk laboratorium skjer nesten alle reaksjoner ved konstant trykk: I åpne begerglass, kolber med en jevn strøm av inert gass eller i en reaktor. I en slik kontekst kan det kjemiske systemet kunne ta opp og gi fra seg varme, men også gjøre volumarbeid ved for eksempel dannelse av gass. Som sett i avsnittet om arbeid over kan arbeid ved konstant trykk uttrykkes som $w = -P\Delta V$. Når vi skal regne på ulike termokjemiske størrelser senere, kan det bli upraktisk å ta med dette leddet for arbeid hver gang vi ønsker å diskutere flyt av energi. Hvordan kan vi bli kvitt det? La oss se på uttrykket for størrelsen _entalpi_ H:

$$H = U + PV$$

Ved hjelp av litt algebra skjønner vi hvorfor entalpi er løsningen på problemet vårt:

$$H = U + PV$$
$$dH = dU + d(PV) = dU + VdP + PdV$$
$$dH = dq + dw + VdP + PdV = dq - PdV + VdP + PdV$$
$$dH = dq + VdP$$

Når trykket er konstant, er $dP=0$, og vi står igjen med: 

$$dH = dq_P$$

Med ord er entalpi varmen som overføres ved konstant trykk. Som vi diskuterte i starten av avsnittet foregår mange kjemiske prosesser ved konstant trykk og entalpi er derfor en svært nyttig størrelse! 

## Varmekapasitet og kalorimetri

### Varmekapasitet i klassisk termodynamikk

På forrige side lærte du om den statistiske definisjonen av varmekapasiteten $C$. I klassisk termodynamikk er varmekapasiteten definert som evnen et stoff har til å ta opp varme (energi) når temperaturen øker. To forskjellige stoffer kan få to ulike temperaturer selv om de varmes på samme måte, slik Joseph Black oppdaget i 1760 da han varmet vann og kvikksølv over samme flamme! Vi kan definere varmekapasitet ved fast volum eller ved fast trykk og benytter hos henholdsvis av den indre energien eller entalpien til systemet:

$$C_V = \left(\frac{\partial U}{\partial T}\right)_V$$
$$C_P = \left(\frac{\partial H}{\partial T}\right)_P$$

I praksis er varmekapasiteten et nyttig verktøy for å finne varmeoverføringen som er forbundet til ulike prosesser (kalorimetri). Varmekapasiteter til ulike stoffer har blitt målt og tabulert gjennom historien, og er derfor som regel kjente verdier. Varmekapasiteten bestemmes utifra kalibrering med et sett _standardstoffer_ som har en satt entalpiendring for sine prosesser. Dette lærer du mer om i labkurset, men foreløpig kan vi gå utifra at verdiene for varmekapasiteter er kjente!

### Kalorimetri

Eksperimentet kalorimetri er svært viktig i kjemi fordi det lar oss måle varmeoverføring! Kalorimetri er også utgangspunktet for den første laboppgaven i Fysikalsk Kjemi I. For å gjøre kalorimetri må vi vite hvor mye vi varmer opp systemet (temperatur) og og hvor "tungt" sytemet er å varme opp (varmekapasitet). Vi har to hovedtyper kalorimetri du må ha hørt om i emnet: 

**Åpent kalorimeter (Kaffekopp)**

I denne typen kalorimetri benyttes en åpen beholder, typisk en isopor- eller kaffekopp. Slik holdes trykket konstant (atmosfærisk trykk). To reaktanter blandes sammen og temperaturen måles før og etter. Varmeoverføring relateres til temperaturendringen gjennom varmekapasiteten som er kjent: 

$$q_P = \Delta T \times C_P$$

Men, som vi vet er varmeoverføringen ved konstant trykk lik entalpiendringen til systemet! Derfor måler vi altså entalpiendringer for reaksjonen mellom reagensene med kalorimetri-eksperimentet. Slike eksperimenter er grunnlaget for tabulerte verdier for entalpiendringer for mange ulike reaksjoner og dette lar oss gjøre viktige beregninger. Dette skal vi se videre på i neste kapitell om termokjemi🥰

**Bombekalorimetri**

I denne typen kalorimetri benyttes en lukket beholder, kalt en bombe. Slik holdes volumet konstant, men trykket kan øke. To reaktanter blandes sammen og temperaturen måles før og etter. Varmeoverføring relateres til temperaturendringen gjennom varmekapasiteten slik formelen ovenfor viser. Men, som vi vet er varmeoverføringen ved konstant volum lik endringen i indre enerigi i systemet! Derfor måler vi altså denne endringen for reaksjonen mellom reagensene med kalorimetri-eksperimentet. 

## Introduksjon til fri energi

Hvordan vet jeg om reaksjonen min kommer til å gå eller ikke? Hvor mye må jeg varme opp kolben min for at reaksjonen skal gå? Når vil likevekt innstilles? Nå har vi endelig alle verktøyene vi trenger for å diskutere fri energi! Fri energi er en størrelse som tar hensyn til både energi og entropi slik at vi som kjemikere kan forutse om prosessen vi ønsker å gjennomføre er _spontan_ eller ikke. Senere skal vi se på hva dette har å si for kjemiske reaksjoner, men for nå skal vi introdusere de to ulike typene fri energi dere må kjenne til i emnet.

### Helmholtz fri energi

Helmholtz fri energi benyttes når volumet i et system holdes konstant. Denne frie energien er vanlig å bruke i fysikk og statistisk mekanikk, men ikke like mye i kjemi. La oss først se på uttrykket for indre energi når volumet er konstant: 

$$U = q + w \leftrightarrow dU = dq + dw = dq + PdV$$

Når volumet er konstant, er dV = 0, og vi står igjen med:

$$ dU = dq_V $$

Ved konstant volum representerer altså den indre energien varmeoverføring, derfor er $U$ "energitermen" i uttrykket for Helmholtz frie energi F: 

$$F = U - TS$$

Helmholtz frie energi kan også defineres fra partisjonsfunksjonen Q. Det mikroskopiske uttrykket med Q og det makroskopiske uttrykket ovenfor er knyttet sammen matematisk i kapitellet om statistisk mekanikk. 

### Gibbs fri energi 

Gibbs fri energi benyttes når trykket i et system holdes konstant, og er derfor den frie energien vi benytter oss mest av i kjemifaget. Som vi så i delen om entalpi, er denne størrelsen et mål på varmeoverføring når trykket er konstant:

$$dH = dq_P$$

Og entalpi $H$ er derfor "energitermen" i uttrykket for Gibbs frie energi G:

$$G = H - TS$$

Den frie energien beskriver altså hvor mye energi vi har tilgjengelig for å drive en prosess fremover. Vi ser også at entropileddet er vektet av temperaturen. Samspillet mellom entalpi, entropi og temperatur vil hjelpe oss å forstå drivkreftene bak kjemiske reaksjoner senere, stay tuned!😎


:::{admonition} Underveisoppgave
:class: tip

Både Helmholtz-energien $F$ og Gibbs-energien $G$ kalles "frie energier". Forklar kort hva som er forskjellen på dem, og hvordan de begge til slutt uttrykker det samme som termodynamikkens andre lov.

:::


:::{admonition} Løsningsforslag
:class: tip dropdown

Begge er "frie energier" fordi de forteller hvor mye nyttig arbeid et system kan yte, og fordi et system spontant går mot lavest mulig verdi av dem. Forskjellen ligger i hvilke betingelser som holdes konstant: $F$ brukes ved konstant volum og temperatur, mens $G$ brukes ved konstant trykk og temperatur. Gibbs-energien er derfor den praktiske i kjemi, siden reaksjoner oftest skjer ved konstant trykk (åpen beholder).

De uttrykker begge andre lov, som sier at den totale entropien i universet alltid øker for en spontan prosess. I stedet for å regne på både systemet og omgivelsene, pakker $F$ og $G$ dette inn i én størrelse som bare avhenger av systemet: at $F$ eller $G$ synker, er nøyaktig det samme som at den totale entropien øker - bare uttrykt gjennom systemets egne variabler under de gitte betingelsene.

:::

## Termodynamiske drivkrefter matematisk og praktisk

Varme strømmer fra varmt til kaldt, gasser beveger seg fra høyt trykk til lavt trykk, stoffer diffunderer fra områder med høy konsentrasjon til områder med lav konsentrasjon. Dette virker kanskje intuitivt for oss, men hvorfor skjer disse prosessene egentlig? Hva har de til felles? Bak dem ligger de _termodynamiske drivkreftene_. Først skal vi se på hvordan vi matematisk kan definere de termodynamiske potensialene fra tilstandsfunksjonene. Dette gir opphav til _fundamentallikningene_, den matematiske definisjonen av de termodynamiske drivkreftene og _Maxwell-relasjonene_.  

### Fundamentallikningen for indre energi i et lukket system

La oss starte fra definisjonen av indre energi igjen: 

$$\Delta U = q + w$$

I stedet for definisjonen som inneholder stifunksjonene arbeid og varme ønsker vi nå å finne ut av hvordan vi kan uttrykke den indre energien ved hjelp av tilstandsfunksjoner som temperatur, trykk, volum og entropi. Vi benytter oss av noen uttrykk fra de forrige kapitlene og finner et nytt uttrykk for dU:

$$dS = \frac{dq}{T} \leftrightarrow dq = TdS$$
$$dw = -PdV$$
$$dU = TdS - PdV$$

Dette er en *fundamentallikning* for den indre energien og gjør indre energi til en funksjon av entropi og volum. Vi kan nå trekke ut temperatur og volum som partiellderiverte! Dersom volumet er konstant vil dV være lik 0 og vi står igjen med:

$$\delta U = T\delta S \leftrightarrow T = \left(\frac{\delta U}{\delta S}\right)_V$$

Temperatur viser altså hvor mye den indre energien må øke når vi får økning i entropi ved fast volum, for eksempel ved varmetilførsel. Vi sier at temperatur er _drivkraften_ for varmeoverføring.


:::{admonition} Temperatur som drivkraft - mikroskopisk utledning
:class: note

Vi har vært borti flere måter å beskrive temperatur allerede: endringen i entropi når den indre energien endres, og en størrelse som forteller oss når varmeutveksling vil skje. Objekter utveksler varme med hverandre for å oppnå termisk likevekt, og likevekt tilsvarer maksimal entropi. La oss se hvordan vi kan bruke det vi vet til å vise hvordan temperaturen driver termisk likevekt, dvs. la oss gå den mikroskopiske veien for å finne det vi observerer makroskopisk. 

![Populasjonsforhold for vibrasjonsnivåer](Simuleringer/gittermodell_oppsett_repeat.png)

Tenk deg to systemer, $A$ og $B$, som hver består av noen atomer. Atomene kan holde på energi i form av diskrete energipakker. $A$ og $B$ er i kontakt med hverandre, slik at energipakker kan flytte seg mellom dem, men de er isolert fra alt annet, ingen stoffmengde eller volum utveksles, og ingen energi lekker ut til omgivelsene. Det totale antallet energipakker er derfor konstant, selv om fordelingen mellom $A$ og $B$ kan endre seg over tid.

Entropien til hvert av de to delsystemene vil avhenge av den eneste frihetsgraden, som er indre energi $U_A$ og $U_B$. Den indre energien til hvert av objektene kan altså endre seg ved at varme overføres mellom dem, men *den totale* energien $U_{tot}=U_A+U_B$ er konstant, ettersom energi ikke utveksles med omgivelsene. Vi har at entropien til hvert system er: 

$$ S_A(U_A) = k_B \ln \Omega_A,$$

$$ S_B(U_B) = k_B \ln \Omega_B,$$

der vi nå har erstattet det diskrete "antall energipakker" med den kontinuerlige indre energien $U$, samme idé, bare glattet ut slik at vi kan derivere. $\Omega$ er antall måter energipakkene kan fordeles på, gitt hvor mange atomer det er. Den totale entropien er

$$ S_{tot}(U_A,U_B) = S_A + S_B.$$

Denne entropien kan variere, den er ikke konstant. Vi vet at likevekt tilsvarer maksimal entropi, altså vil vi finne likevektstilstanden når entropien ikke *endrer seg* mer, aka ved $dS_{tot}=0$. Først skriver vi opp hvordan en liten endring i $S_{tot}$ avhenger av små endringer i hver av de to energiene. Vi bruker det totale differensialet (se matte-modulen):

$$ dS_{tot}=dS_A+dS_B= \left( \frac{\partial S_A}{\partial U_A} \right)_{V,N} dU_A + \left( \frac{\partial S_B}{\partial U_B} \right)_{V,N} dU_B.$$

Her er $(\partial S_A/\partial U_A)$ *hvor mye entropien til $A$ øker per enhet energi $A$ mottar*, tenk på det som å gi atomene i $A$ enda en energipakke og se hvor mye $\ln\Omega_A$ endrer seg.

Så bruker vi bibetingelsen: siden $U_{tot}=U_A+U_B$ er konstant, er $dU_{tot}=dU_A+dU_B=0$, altså $dU_A=-dU_B$. Alt $B$ gir fra seg, mottar $A$. Vi bytter ut $dU_B$ med $-dU_A$:

$$ dS_{tot}= \left( \frac{\partial S_A}{\partial U_A} \right)_{V,N} dU_A - \left( \frac{\partial S_B}{\partial U_B} \right)_{V,N} dU_A = \left[ \left( \frac{\partial S_A}{\partial U_A} \right)_{V,N} - \left( \frac{\partial S_B}{\partial U_B} \right)_{V,N} \right] dU_A. $$

Nå setter vi $dS_{tot}=0$, for å se hva som skjer ved likevekt. For at dette skal holde matematisk, må parentesen være null, siden $dU_A$ generelt ikke er null: 

$$ \left( \frac{\partial S_A}{\partial U_A} \right)_{V,N} - \left( \frac{\partial S_B}{\partial U_B} \right)_{V,N} = 0 \: \rightarrow \left( \frac{\partial S_A}{\partial U_A} \right)_{V,N} = \left( \frac{\partial S_B}{\partial U_B} \right)_{V,N}. $$

Med andre ord: ved likevekt gir det like mye entropi å gi en liten energimengde til $A$ som til $B$, ingen av systemene "vinner" mer enn det andre på å motta energien.

Nå setter vi inn definisjonen $1/T \equiv (\partial S / \partial U)_{V,N}$ og får:

$$ \frac{1}{T_A} = \frac{1}{T_B} \quad\rightarrow\quad T_A = T_B. $$

Ved likevekt har altså $A$ og $B$ samme temperatur. Objekter med ulik temperatur vil utveksle varme fordi det driver det totale systemet mot maksimal entropi. Dette var kanskje kjent fra før, men nå har vi sett hvordan det kan *vises* ut fra det mikroskopiske ($\Omega$).

Enheten $1/T$ beskriver altså en *tendens til å overføre energi*: den forteller ikke hvor mye energi et objekt har, men hvor mye det har å *vinne* i entropi på å ta imot enda litt mer. Figuren under viser dette direkte for modellsystemet vårt: er det få energipakker i $A$, er $1/T_A$ større enn $1/T_B$, og energi strømmer inn i $A$. Er det mange energipakker i $A$, er $1/T_B$ størst, og energi strømmer ut av $A$. Bare i skjæringspunktet, der $1/T_A=1/T_B$, er det ingen nettostrøm igjen.

![Populasjonsforhold for vibrasjonsnivåer](Simuleringer/1_over_T_drivkraft.png)
:::

Dersom vi nå setter entropien konstant, vil dS være lik 0 og vi står igjen med:

$$\delta U = -P\delta V \leftrightarrow -P = \left(\frac{\delta U}{\delta V}\right)_S$$

Trykk viser altså hvor mye den indre energien må senkes når vi får økning i volum ved fast entropi. Vi sier at trykk er _drivkraften_ for volumendring. 


:::{admonition} Trykk som drivkraft - mikroskopisk utledning
:class: note

Vi har sett hvordan temperatur beskriver drivkraften for varmeoverføring. La oss bruke samme mikroskopiske tankegang for å vise at trykk er drivkraften for volumendring.

![Populasjonsforhold for vibrasjonsnivåer](Simuleringer/stempel_system_trykk.png)

Tenk deg en lukket sylinder delt i to av et bevegelig stempel, i delsystemene $A$ og $B$, som hver inneholder noen gassmolekyler. Stempelet kan gli fritt, slik at volumet til hvert delsystem endres, men det totale volumet $V_{tot}=V_A+V_B$ er konstant siden sylinderen er isolert fra omgivelsene. Veggene og stempelet leder varme godt, så $A$ og $B$ er hele tiden i termisk likevekt med samme temperatur $T$ (dette har vi allerede vist i forrige boks: $T_A=T_B$). Det eneste som gjenstår å finne ut av er altså hvordan volumet fordeler seg.

Entropien til hvert delsystem avhenger nå av volumet, $S_A(V_A)$ og $S_B(V_B)$. Igjen er total entropi summen $S_{tot}=S_A+S_B$, og likevekt tilsvarer maksimal entropi, altså $dS_{tot}=0$. Det totale differensialet er:

$$ dS_{tot}=dS_A+dS_B= \left( \frac{\partial S_A}{\partial V_A} \right)_{U} dV_A + \left( \frac{\partial S_B}{\partial V_B} \right)_{U} dV_B. $$

Her er $(\partial S_A/\partial V_A)$ *hvor mye entropien til $A$ øker per enhet volum $A$ mottar*, altså hva som skjer med $\ln\Omega_A$ når stempelet gir $A$ litt mer plass.

Bibetingelsen er at totalvolumet er konstant: $dV_{tot}=dV_A+dV_B=0$, altså $dV_A=-dV_B$. Det $B$ gir fra seg, mottar $A$. Vi setter inn:

$$ dS_{tot}= \left[ \left( \frac{\partial S_A}{\partial V_A} \right)_{U} - \left( \frac{\partial S_B}{\partial V_B} \right)_{U} \right] dV_A. $$

Ved likevekt, $dS_{tot}=0$, må klammeparentesen være null siden $dV_A$ generelt ikke er null:

$$ \left( \frac{\partial S_A}{\partial V_A} \right)_{U} = \left( \frac{\partial S_B}{\partial V_B} \right)_{U}. $$

Med andre ord: ved likevekt gir det like mye entropi å gi litt volum til $A$ som til $B$ - ingen av systemene "vinner" mer enn det andre på å motta volumet.

Fra fundamentallikningen for $dU$ har vi definisjonen $P/T \equiv (\partial S/\partial V)_U$ (du kan sjekke selv om du forstår overgangen!). Setter vi inn dette får vi:

$$ \frac{P_A}{T_A} = \frac{P_B}{T_B}. $$

Siden $T_A=T_B$ allerede fra termisk likevekt, forenkles dette til $P_A=P_B$: ved likevekt er trykket likt på begge sider av stempelet.

Størrelsen $P/T$ beskriver altså en *tendens til å gi fra seg volum*: den forteller ikke hvor mye plass et system har, men hvor mye det har å *vinne* i entropi på å gi litt av den fra seg. Figuren under viser dette direkte for sylinderen vår: er $V_A$ lite, er $P_A/T$ større enn $P_B/T$, og stempelet flytter seg slik at $A$ vokser. Er $V_A$ stort, er $P_B/T$ størst, og stempelet flytter seg motsatt vei. Bare i skjæringspunktet, der $P_A/T=P_B/T$, er det ingen nettobevegelse igjen.

![Populasjonsforhold for vibrasjonsnivåer](Simuleringer/p_over_T_drivkraft_graf.png)
:::

### Fra indre energi til entalpi og fri energi

På samme måte som for den indre energien ønsker vi også å definere entalpi H og Helmholtz og Gibbs fri energi F og G ved hjelp av tilstandsfunksjoner. Dette blir enklere siden vi nå kan ta utgangspunkt i uttrykket vi fant for indre energi! Se på utledningen av fundamentalllikningen til entalpi først, samt hvordan vi kan finne de termodynamiske drivkreftene knyttet til entalpi. Deretter skal du få prøve deg på egenhånd for de frie energiene🔥

*Entalpi* 
Fra forrige kapittel vet vi at entalpi er definert som den indre energien pluss pV. Vi starter utledningen vår med å differensiere uttrykket og bytte ut $dU$ med uttrykket vi fant i første del av kapittelet:

$$H = U + PV$$
$$dH = dU + d(PV) = (TdS - PdV) + (PdV + VdP)$$
$$dH = TdS + Vdp$$

Dersom vi setter trykket konstant, blir dP lik null, og vi står igjen med:

$$\delta H = T\delta S \longleftrightarrow T = \left(\frac{\partial H}{\partial S}\right)_P$$

Dersom vi setter entropien konstant, finner vi:

$$V = \left(\frac{\partial H}{\partial P}\right)_S$$

```{dropdown} Klikk for å se svaret!
*Helmholtz fri energi* 
Fra forrige kapittel vet vi at Helmholtz fri energi er definert som indre energi minus TS. Vi starter utledningen vår med å differensiere uttrykket og bytte ut $dU$ med uttrykket vi fant i første del av dette kapittelet:

$$F = U - TS$$
$$dF = dU - d(TS) = (TdS - PdV) - (TdS + SdT)$$
$$dF = -PdV - SdT$$

Dersom vi setter trykket konstant, blir dp lik null, og vi står igjen med:

$$\partial H = T\partial S \leftrightarrow T = \left(\frac{\partial H}{\partial S}\right)_P$$

Dersom vi setter temperaturen konstant, finner vi:

$$-p = \left(\frac{\partial F}{\partial V}\right)_T$$

Dersom vi setter volumet konstant, finner vi:

$$-S = \left(\frac{\partial F}{\partial T}\right)_V$$

*Gibbs fri energi*
Fra forrige kapittel vet vi at Gibbs fri energi er definert som entalpi minus TS. Vi starter utledningen vår med å differensiere uttrykket og bytte ut $dH$ med uttrykket vi fant i første del av dette kapittelet:

$$G = H - TS$$
$$dG = dH - d(TS) = (TdS + VdP) - (TdS + SdT)$$
$$dG = VdP - SdT$$

Dersom vi setter temperaturen konstant, finner vi:

$$V = \left(\frac{partial G}{\partial P}\right)_T$$

Dersom vi setter trykket konstant, finner vi:

$$-S = \left(\frac{\partial G}{\partial T}\right)_P$$

```

```{admonition} Fundamentallikningene
$$dH = TdS + VdP$$

$$dG = VdP - SdT$$
```

### Maxwell-relasjoner

Maxwell-relasjonene tar utgangspunkt i fundamentallikningene og beskriver noen viktige sammenhenger mellom de ulike tilstandsfunksjonene. Først må vi bruke noen matematiske triks, før vi kan trekke ut de nyttige relasjonene!

La oss se på noen egenskaper for en tilstandsstørrelse X, som kan ta to variable a og b:

$$X = X(a,b)$$

Vi kan skrive den totale differensialen for X på følgende måte, vi skal illustrere dette i praksis med fundamentallikningene straks:

$$dX = \left(\frac{\partial X}{\partial a}\right)_bda + \left(\frac{\partial X}{\partial b}\right)_adb$$

Dette kalles et _eksakt differensial_ fordi X er en tilstandsfunksjon av variablene a og b. For slike differensialer gjelder det at:

$$\frac{\partial}{\partial b}\left(\frac{\partial X}{\partial a}\right)_b = \frac{\partial}{\partial a}\left(\frac{\partial X}{\partial b}\right)_a$$

Vi sier at de _blandede andrederiverte_ er like. Denne sammenhengen er nøkkelen bak Maxwell-relasjonene! 


#### Indre energi

La oss begynne med indre energi. I forrige avsnitt fant vi følgende fundamentallikning for den indre energi: 

$$dU = TdS - PdV$$

Vi identifiserte også to uttrykk for T og -P:

$$T = \left(\frac{\partial U}{\partial S}\right)_V$$
$$-P = \left(\frac{\partial U}{\partial V}\right)_S$$

La oss prøve å sette de to uttrykkene inn i fundamentallikningen og dermed bytte ut T og -P med de partiellderiverte:

$$dU = \left(\frac{\partial U}{\partial S}\right)_VdS + \left(\frac{\partial U}{\partial V}\right)_SdV$$

Nå ser vi at vi har fått et differensial på samme form det eksakte differensialet vi definerte med X over! Da vet vi at vi kan bruke regelen om blandede deriverte. La oss først identifisere X, a og b: 

$$X = U, a = S, b = V$$

Vi skriver opp uttrykket for de blandede deriverte: 

$$\frac{\partial}{\partial V}\left(\frac{\partial U}{\partial S}\right)_V = \frac{\partial}{\partial S}\left(\frac{\partial U}{\partial V}\right)_S$$

Vi kjenner igjen uttrykkene for T og -P:

$$\frac{\partial}{\partial V}(T) = -\frac{\partial}{\partial S}(P)$$

La oss løse høyre side først. Når vi partiellderiverer T med hensyn på V, betyr det at S holdes konstant. Vi får derfor resultatet: 

$$\frac{\partial}{\partial V}(T) = \left(\frac{\partial T}{\partial V}\right)_S$$

På venstre side partiellderiverer vi P med hensyn på S, og V holdes derfor konstant. Vi står igjen med resultatet: 

$$-\frac{\partial}{\partial S}(P) = -\left(\frac{\partial P}{\partial S}\right)_V$$

Vi kan nå skrive opp den første Maxwell-relasjonen🤩

$$\left(\frac{\partial T}{\partial V}\right)_S = -\left(\frac{\partial P}{\partial S}\right)_V$$

## Kjemisk potensial

Hittil har du lært om temperatur som drivkraften til varmeoverføring og trykk som drivkraften til volumendring. Vi skal nå introdusere en tredje drivkraft som er svært viktig i kjemien, nemlig kjemisk potensial: Dette potensialet beskriver tendensen et system har til å utveksle partikler med omgivelsene, eller til å endre sin kjemiske sammensetning, ved for eksempel reaksjon, diffusjon eller faseoverganger. Helt konkret er det kjemiske potensialet $\mu$ energien som er knyttet til tilførelsen av ett mol av et stoff $i$ til et system ved konstant trykk og temperatur:

$${\mu}_i = \left(\frac{\partial G}{\partial n_i}\right)_{T,P}$$

I praksis vil partikler tendere til å bevege seg fra områder med høyt kjemisk potensial til områder med lavt kjemisk potensial. Hvordan ser dette ut i praksis? Prøv deg på oppgaven under! 

![Kjemisk potensial i et bilde](bilder/diffusjon_bilde.png)

```{admonition} Underveisoppgave
:class: tip

Se på bildene over: Konditorfarge blir dryppet i et vannglass og sprer seg utover i vannet helt uten røring! Hadde du forventet at dette skjer? Hvilken sammenheng tror du dette har med kjemisk potensial? 

```

Det kjemiske potensialet kan også beskrives med statistisk mekanikk! Vi kan se for oss at vi legger til én ny partikkel i et system: Feks en mini-dråpe av konditorfarge i vannglasset. Dette vil med en gang endre de tilgjengelige mikrotilstandene i vannglasset, og derfor er kjemisk potensial knyttet opp til systemets entropi og energi. Vi kan beskrive kjemisk potensial utifra en liten endring i antall partikler ved hjelp av [partisjonsfunksjonen](https://andreasdh.github.io/Bifrost/docs/statmek.html#partisjonsfunksjonen): 

$$\mu = -k_BT\left(\frac{\partial lnQ}{\partial N}\right)_{T, V}$$

Dersom $\mu$ er høy, har vi få tilgjengelige mikrotilstander og systemet vil helst ikke ta imot flere partikler $N$. Dersom $\mu$ er lav, har vi mange tilgjengelige mikrotilstander, og systemet tar gjerne imot flere partikler $N$. I glasset med konditorfarge har vi ikke rørt, varmet eller endret trykket: Det er kun den terodynamiske drivkraften $\mu$ som driver spredningen av fargen fremover! Områdene med mer farge vil helst bli kvitt fargen og "sender den" derfor til områder med lavere konsentrasjon av farge. Dette fortsetter helt til det er like mye farge overalt i glasset. Det kjemiske potensialet blir en viktig størrelse når dere senere skal lære om blandinger og faseoverganger🧪

Likningene kan utvides til åpne systemer ved å inkludere kjemisk potensial, mer spesifikt ved å legge til summen over det kjemiske potensialet for hver enkelt partikkel. Dette gir også opphav til nye Maxwell-relasjoner. Dersom du er interessert i å lese mer om dette snakker læreboka i emnet mer om det, men det finnes også gode ressurser på nettet. ## Oppsummering

På denne siden har du fått en introduksjon til **makroskopisk termodynamikk**🔥🚂 Den makroskopiske termodynamikken er et sett med størrelser, lover og formler som springer ut fra **empiriske observasjoner** gjort av oppdagere fra 1700-tallet av. Feltet ble særlig viktig i forbindelse med den industrielle revolusjonen, hvor forståelsen av samspillet mellom **varme** og **arbeid** ble sentralt for å øke produksjonen av varer og gjøre transport mer effektivt. Med utgangspunkt i dette ble **termodynamikkens første lov** definert: Energi kan ikke skapes eller forsvinne, men kan overføres mellom to former; varme og arbeid. Vi illustrerte også relevansen av denne loven i kjemien ved hjelp av eksempelet om ball-milling. Senere på siden gikk vi videre til å introdusere størrelsen **entalpi**, som beskriver varmeflyt ved konstant trykk, og derfor er en svært nyttig i kjemifaget. Deretter snakket vi kort om **varmekapasitet og kalorimetri** som er temaer i deres første labøvelse, før vi introduserte konseptet **fri energi**. Særlig viktig i kjemi er **Gibbs fri energi** som beskriver samspillet mellom entalpi, entropi og temperatur. Mot slutten av siden lærte du om **temperatur**, **trykk** og **kjemisk potensial** som termodynamiske drivkrefter for henholdsvis varmeutveksling, volumendring og materieutveksling. Du fikk også en innføring i utledningen av **fundamentallikningene** som kobler sammen sti- og tilstandsfunksjoner og de resulterende **Maxwell-relasjonene**. La oss gå videre til termokjemi!⚗️
