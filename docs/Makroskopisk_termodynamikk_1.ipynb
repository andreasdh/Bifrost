## Introduksjon til viktige størrelser i klassisk termodynamikk

I forbindelse med den industrielle revolusjonen ble samspillet mellom gass, varme og arbeid særlig viktig. Vitenskapsmenn og oppfinnere fikk en ny forståelse av hvordan energi utveksles mellom disse parameterene, og kunne utnytte dette til å gjøre transport og produksjon av varer langt mer effektivt! I denne første delen skal vi introdusere disse størrelsene som senere blir viktige for å forstå klassiske termodynamiske uttrykk. 

### Varme
Gjennom historien har menneskeheten forklart varme på ulike måter. På 1700-tallet trodde man at varme var en vektløs væske som flyter fra et varmt stoff til et kaldt stoff (Calorique-teorien). Dette ble senere kritisert da Joseph Black i 1750 oppdaget at vann rundt $100^\circ\mathrm{C}$ kunne ta opp og avgi varme uten at temperaturen ble endret. Dette kalte han latent varme. Ti år senere oppdaget Black at vann og kvikksølv som ble varmet over samme flamme hadde ulik temperatur. I dag forklarer vi disse observasjonene ved hjelp av varmekapasitet _varmekapasitet_, og dette skal vi se nærmere på straks. I slutten av 1700-tallet bemerket Count Rumford en sammenheng mellom mekanisk energi og varme i forbindelse med boring av kanoner. Dette ble studert videre av Joule som var den første til å tolke temperaturøkning som følge av friksjon som en overføring av kinetisk mekanisk energi til varme. I dag beskriver vi varme som en form for energi. 

### Arbeid og reversibilitet

Arbeid er den energien som overføres når en system "tvinges" til å endre en makroskopisk variabel. I fysikken kan dette være alt fra å komprimere en gass (endre volum) og trekke i en fær (endre lengde) til å flytte en ladning ved hjelp av et elektrisk felt (endring i posisjon). Matematisk er arbeid definert som produktet av en kraft F og forflytningen i kraftens retning s. 

$$w = -F \times s$$

Legg merke til konvensjonen om fortegn! Dersom $w<0$, forlater energi systemet, altså har systemet gjort arbeid på omgivelsene. Dersom $w>0$ får systemet energi, altså har omgivelsene gjort arbeid på systemet.

I kjemi er vi særlige interessert i tilfellet hvor vi ekspanderer en gass mot et konstant ytre trykk. Vi kan se for oss en gass i en sylinder som har et bevegelig stempel slik at gassen kan få mindre eller større volum. Stempelet er rundt og har et areal og gassen kan utgjøre et trykk p for å skyve stempelet utover. Kraften fra gassen på stempelet er lik $F = pA$. Videre kan vi se for oss at stempelet beveger seg med en liten avstand dx får vi følgende uttrykk for arbeidet:

$$dw = -F \times dx = pA \times dx$$

Vi kan også beskrive hvordan volumet tilgjengelig for gassen i sylinderen endrer seg. Volumet er gitt ved å gange sammen stempelets areal og lengden til sylinderen. For volumendringen får vi altså: $dV = Adx \leftrightarrow dx = \frac{dV}{A}$. Vi kan sette dette inn i uttrykket vårt for arbeid:

$$dw = pA \times dx \leftrightarrow dw = pA \times \frac{dV}{A} \leftrightarrow dw = p \times dV$$

For å beskrive arbeidet som utføres i løpet av prosessen kan vi integrere fra start- til sluttvolumet på denne måten: 

$$w = - \int_{V_i}^{V_f} pdV$$

Dersom vi har et konstant ytre trykk $p_{ex}$ kan vi trekke trykket utenfor integralet og står igjen med et enklere uttrykk:

$$w = - \int_{V_i}^{V_f} p_{ex}dV \leftrightarrow w = - p_{ex} \int_{V_i}^{V_f} dV = -p_{ex}\Delta V$$

Men hvordan kan vi utnytte en slik prosess på best mulig måte? Hvordan kan vi bruke 100% av energien vår til å forflytte stempelet og unngå å miste energi som varme på grunn av friksjon? Svaret på disse spørsmålene er at prosessen må være _reversibel_, kort forklart at den må gjøres uendelig sakte, det vil si deles inn i uendelig mange små deler! Dersom ekspansjonen skjer uendelig sakte, vil det alltid innstilles en likevekt mellom det indre og ytre trykket og vi vil aldri miste energi som varme. Du kan utforske dette eksempelet ved hjelp av simuleringen du finner nedenfor!

Deersom vi systemet går fra en tilstand A til en tilstand B, får vi en endring i indre energi som er lik er summen av varme og arbeid som har blitt tilført systemet i løpet av prosessen:

$$\Delta U = U_B - U_A = q + w$$

Selv om q og w er såkalte stifunksjoner og endrer seg avhengig av veien fra start- til sluttilstand, er den indre energien derimot en tilstandsfunksjon: Den er kun avhengig av hvilket trykk og temperatur systemet befinner seg i start- og sluttilstanden, og hvor mange molekyler det består av. Energien kan altså flyte fritt mellom de to formene arbeid og varme, men summen av dem vil alltid gi den samme økningen/senkingen av energien. Mer arbeid betyr mindre varme og vice versa. Et intuitivt eksempel kan være en fjelltur: Vi kan ta mange ulike stier opp et fjell (arbeid), og spise en variende mengde Kvikk Lunsj for å komme oss opp fjellet (varme), men til slutt vil man uansett ha kommet seg fra A til B og det er dette som representerer den indre energien. ⛰️

Vi kan også finne eksempler i kjemien! Dersom vi har en reaksjon hvor vi ønsker å danne et produkt fra to reagenser er vi nødt til å få de to reagensene til å kollidere. Vi kan se for oss to måter å gjøre dette på: Vi kan mekanisk tvinge molekylene til sammen (arbeid) eller vi kan varme opp molekylene slik at de får økt bevegelse og større sjanse for å kollidere (varme). Dette korresponderer til to ulike metoder i laben! For å tvinge frem kollisjonen mekanisk kan vi bruke for eksempel "ball-milling" hvor vi bruker keramiske kuler til å "banke" reagensene våre sammen, eller vi kan bruke en mer tradisjonell varmeplate og få reagensene våre til å reagere i en oppvarmet løsning. Selv om prosessene er svært ulike, vil de gi det samme ønskede produktet! Vi må allikevel huske på at dette er et svært enkelt eksempel, og at vi ofte må bruke en kombinasjon av mekaniske og termiske metoder for å få reaksjonene våre til å fungere🥼🧪


## Entalpi

I et kjemisk laboratorium skjer nesten alle reaksjoner ved konstant trykk: I åpne begerglass, kolber med en jevn strøm av inert gass eller i en reaktor. I en slik kontekst kan det kjemiske systemet kunne ta opp og gi fra seg varme, men også gjøre volumarbeid ved for eksempel dannelse av gass. Som sett i avsnittet om arbeid over kan arbeid ved konstant trykk uttrykkes som $w = -p\Delta V$. Når vi skal regne på ulike termokjemiske størrelser senere, kan det bli upraktisk å ta med dette leddet for arbeid hver gang vi ønsker å diskutere flyt av energi. Hvordan kan vi bli kvitt det? La oss se på uttrykket for størrelsen _entalpi_ H:

$$H = U + pV$$

Ved hjelp av litt algebra skjønner vi hvorfor entalpi er løsningen på problemet vårt:

$$H = U + pV$$
$$dH = dU + d(pV) = dU + Vdp + pdV$$
$$dH = dq + dw + Vdp + pdV = dq - pdV + Vdp + pdV$$
$$dH = dq + Vdp$$

Når trykket er konstant, er $dp=0$, og vi står igjen med: 

$$dH = dq_P$$

Med ord er entalpi varmen som overføres ved konstant trykk. Som vi diskuterte i starten av avsnittet foregår mange kjemiske prosesser ved konstant trykk og entalpi er derfor en svært nyttig størrelse! 

## Introduksjon til fri energi

Hvordan vet jeg om reaksjonen min kommer til å gå eller ikke? Hvor mye må jeg varme opp kolben min for at reaksjonen skal gå? Når vil likevekt innstilles? Nå har vi endelig alle verktøyene vi trenger for å diskutere fri energi! Fri energi er en størrelse som tar hensyn til både energi og entropi slik at vi som kjemikere kan forutse om prosessen vi ønsker å gjennomføre er _spontan_ eller ikke. Senere skal vi se på hva dette har å si for kjemiske reaksjoner, men for nå skal vi introdusere de to ulike typene fri energi dere må kjenne til i emnet.

### Helmholtz fri energi

Helmholtz fri energi benyttes når volumet i et system holdes konstant. Denne frie energien er vanlig å bruke i fysikk og statistisk mekanikk, men ikke like mye i kjemi. La oss først se på uttrykket for indre energi når volumet er konstant: 

$$U = q + w \leftrightarrow dU = dq + dw = dq + pdV$$

Når volumet er konstant, er dV = 0, og vi står igjen med:

$$ dU = dq_V $$

Ved konstant volum representerer altså den indre energien varmeoverføring, derfor er $U$ "energitermen" i uttrykket for Helmholtz frie energi F: 

$$F = U - TS$$

Helmholtz frie energi kan også defineres fra partisjonsfunksjonen Q. Det mikroskopiske uttrykket med Q og det makroskopiske uttrykket ovenfor er knyttet sammen matematisk i kapitellet om statistisk mekanikk. 

### Gibbs fri energi 

Gibbs fri energi benyttes når trykket i et system holdes konstant, og er derfor den frie energien vi benytter oss mest av i kjemifaget. Som vi så i delen om entalpi, er denne størrelsen et mål på varmeoverføring når trykket er konstant:

$$dH = dq_P$$

Og entalpi $H$ er derfor "energitermen" i uttrykket for Gibbs frie energi G:

$$G = H - TS$$

Gibbs fri energi beskriver altså hvor mye energi systemet har totalt (varme H) og hvor mye av denne som "brukes opp" på uorden (entropi). Den frie energien beskriver altså hvor mye energi vi har tilgjengelig for å drive en prosess fremover. Vi ser også at entropileddet er vektet av temperaturen. Samspillet mellom entalpi, entropi og temperatur vil hjelpe oss å forstå drivkreftene bak kjemiske reaksjoner senere, stay tuned!😎
