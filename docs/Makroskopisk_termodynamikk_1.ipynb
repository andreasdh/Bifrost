## Fra mikroskopisk til makroskopisk termodynamikk

I forbindelse med den industrielle revolusjonen ble samspillet mellom gass, varme og arbeid særlig viktig. Vitenskapsmenn og oppfinnere fikk en ny forståelse av hvordan energi utveksles mellom disse størrelsene, og kunne utnytte dette til å gjøre transport og produksjon av varer langt mer effektivt. Mange av begrepene i den klassiske termodynamikken ble derfor utviklet lenge før man hadde en moderne forståelse av atomer og molekyler.

I dag kjenner vi begge beskrivelsesnivåene. Mange av størrelsene vi trenger, har du allerede møtt: **indre energi** er en tilstandsfunksjon, mens **varme** og **arbeid** beskriver måter energi kan overføres over systemgrensen. Vi skal derfor ikke begynne på nytt, men bruke disse ideene som utgangspunkt for den makroskopiske beskrivelsen.

### Varme i historien

Gjennom historien har menneskeheten forklart varme på ulike måter. På 1700-tallet trodde man at varme var en vektløs væske som kunne flyte fra et varmt stoff til et kaldt stoff, den såkalte *calorique*-teorien. Joseph Black viste gjennom studier av blant annet vann at et stoff kan ta opp eller avgi varme uten at temperaturen nødvendigvis endres, det vi i dag forbinder med latent varme. Han observerte også at ulike stoffer får ulik temperaturendring når de tilføres samme mengde energi. I dag beskriver vi denne egenskapen ved hjelp av **varmekapasitet**.

I slutten av 1700-tallet bemerket Count Rumford en sammenheng mellom mekanisk energi og varme i forbindelse med boring av kanoner. Dette ble studert videre av James Joule, som gjennom eksperimenter viste at mekanisk arbeid og varme er ulike måter å overføre energi på.

### Varme i dag

I moderne termodynamikk ser vi varme $q$ som en måte energi overføres på, ikke som en substans eller noe systemet «inneholder». Varme overføres når det er en temperaturforskjell mellom system og omgivelser.

Når et system mottar energi som varme, kan den indre energien endres. På mikronivå kan energien fordeles mellom translasjon, rotasjon, vibrasjon og elektroniske tilstander, og den kan endre systemets potensielle energi gjennom endringer i molekylenes struktur og vekselvirkninger.

Når varmeoverføringen er ferdig, snakker vi derfor ikke om et «varmeinnhold». Energien inngår i systemets **indre energi**. Kort sagt: varme beskriver *hvordan energi overføres*, mens indre energi beskriver *energien systemet har*.


## Arbeid og reversible prosesser

Arbeid er den andre grunnleggende måten energi kan overføres over systemgrensen på. Arbeid skjer når en kraft virker gjennom en forflytning eller, mer generelt, når systemet endrer en makroskopisk variabel mot en ytre påvirkning. I fysikken kan dette være alt fra å komprimere en gass og trekke i en fjær til å flytte en ladning i et elektrisk felt.

I kjemisk termodynamikk er vi særlig interessert i **trykk–volum-arbeid**, for eksempel når en gass ekspanderer mot et ytre trykk.

Vi bruker den vanlige kjemiske fortegnskonvensjonen:

- $w>0$ når omgivelsene gjør arbeid **på** systemet
- $w<0$ når systemet gjør arbeid **på** omgivelsene

Vi kan se for oss en gass i en sylinder med et bevegelig stempel. Stempelet har areal $A$, og et ytre trykk $P_\mathrm{ext}$ presser på stempelet. Kraften som virker på stempelet er

$$F=P_\mathrm{ext}A$$

Dersom stempelet flyttes en liten avstand $dx$, endres volumet med

$$dV=A\,dx$$

Arbeidet som gjøres **på systemet** ved denne lille volumendringen blir derfor

$$\delta w=-F\,dx=-P_\mathrm{ext}A\,dx=-P_\mathrm{ext}\,dV$$

For en endelig prosess får vi

$$w=-\int_{V_i}^{V_f}P_\mathrm{ext}\,dV$$

Dersom det ytre trykket er konstant, kan vi trekke det utenfor integralet:

$$w=-P_\mathrm{ext}\int_{V_i}^{V_f}dV=-P_\mathrm{ext}\Delta V$$

Ved ekspansjon er $\Delta V>0$, og dermed er $w<0$: systemet gjør arbeid på omgivelsene.

### Hva betyr reversibel?

Men hvordan kan vi få mest mulig arbeid ut av en slik ekspansjon? Her kommer ideen om en **reversibel prosess** inn.

En reversibel prosess er en idealisert prosess som kan snus ved en uendelig liten endring i betingelsene. Systemet går gjennom en sammenhengende serie tilstander som ligger vilkårlig nær likevekt. For en gass som ekspanderer, betyr dette at det ytre trykket hele tiden er nesten lik gasstrykket, og at vi unngår dissipative effekter som friksjon.

Et klassisk bilde er å tenke seg sand som ligger oppå stempelet og presser gassen sammen. Dersom vi fjerner store mengder sand om gangen, endrer vi det ytre trykket brått og prosessen blir tydelig irreversibel. Dersom vi i stedet kunne fjerne ett og ett stadig mindre sandkorn, ville stempelet flytte seg bare litt mellom hvert trinn og systemet få tid til å innstille seg svært nær en ny likevekt.

En virkelig prosess kan aldri være perfekt reversibel. Den reversible prosessen er en idealisert **grenseverdi**, men en svært nyttig en: Ved ekspansjon gir den størst mulig arbeid fra systemet mellom to gitte tilstander.

## Indre energi og termodynamikkens første lov

Indre energi $U$ har vi allerede møtt fra et mikroskopisk perspektiv. Den beskriver den totale energien i systemet og kan inneholde bidrag fra molekylenes translasjon, rotasjon og vibrasjon, elektroniske tilstander og potensiell energi knyttet til struktur og vekselvirkninger.

Skjematisk kan vi skrive

$$U=E_\mathrm{trans}+E_\mathrm{rot}+E_\mathrm{vib}+E_\mathrm{elektronisk}+E_\mathrm{potensiell}$$

I statistisk termodynamikk beskrev vi den indre energien som den gjennomsnittlige energien over mikrotilstandene, vektet med sannsynligheten for hver mikrotilstand:

$$U=\langle E\rangle$$

På makroskopisk nivå trenger vi ikke kjenne de enkelte mikrotilstandene. Vi beskriver i stedet hvordan $U$ endres når energi krysser systemgrensen. Det kan skje ved varmeoverføring eller arbeid. Dette er termodynamikkens første lov:

:::{admonition} Termodynamikkens første lov
:class: important

For et lukket system kan den indre energien endres ved varmeoverføring eller arbeid:

$$\Delta U=q+w$$

Varme og arbeid beskriver energioverføring under en prosess, mens $U$ er en tilstandsfunksjon.
:::

Dersom systemet går fra tilstand A til tilstand B, er

$$\Delta U=U_B-U_A=q+w$$

Verdiene av $q$ og $w$ kan avhenge av veien mellom A og B, men summen deres må være den samme så lenge start- og sluttilstanden er de samme.

Dette er nettopp forskjellen mellom **tilstandsfunksjoner** og størrelser som avhenger av **prosessveien**. Indre energi $U$ er en tilstandsfunksjon: Når vi kjenner tilstanden til systemet, har $U$ en bestemt verdi uavhengig av hvordan systemet kom dit. Varme $q$ og arbeid $w$ er derimot ikke egenskaper ved selve tilstanden; de beskriver hva som skjer *på veien* mellom to tilstander.

Denne forskjellen blir viktig når vi nå skal introdusere de **termodynamiske potensialene**. Dette er tilstandsfunksjoner som er bygget opp fra $U$ og andre tilstandsvariabler. Vi skal se på **entalpi** $H$, **Helmholtz-energi** $F$ og **Gibbs-energi** $G$. De beskriver ikke nye typer energi som finnes «ved siden av» den indre energien. I stedet organiserer de den samme termodynamiske informasjonen på måter som er spesielt nyttige under ulike betingelser. For eksempel er entalpi særlig nyttig ved konstant trykk, mens Gibbs-energien blir sentral når både temperatur og trykk holdes konstante.

## Entalpi – nyttig ved konstant trykk

I et kjemisk laboratorium skjer svært mange reaksjoner ved tilnærmet konstant trykk: i åpne begerglass og kolber, eller i systemer som står i kontakt med atmosfæren. Et kjemisk system kan da ta opp og gi fra seg varme, men også gjøre volumarbeid dersom volumet endres, for eksempel når det dannes eller forbrukes gass.

Fra avsnittet om arbeid vet vi at trykk–volum-arbeid ved konstant ytre trykk er

$$w=-P_\mathrm{ext}\Delta V$$

Dersom vi hele tiden skulle bruke $U$ til å beskrive varmeutveksling ved konstant trykk, måtte vi samtidig holde styr på volumarbeidet. Det er upraktisk. Derfor innfører vi en ny tilstandsfunksjon, **entalpi**:

$$H=U+PV$$

La oss se hvorfor dette er nyttig. Vi differensierer uttrykket:

$$dH=dU+d(PV)$$

Produktregelen gir

$$dH=dU+P\,dV+V\,dP$$

For tilstandsfunksjoner som $U$ og $H$ bruker vi $d$ for en infinitesimal endring. Varme og arbeid er derimot avhengige av prosessveien og er ikke tilstandsfunksjoner. Derfor bruker vi ofte symbolet $\delta$ i stedet:

$$dU=\delta q+\delta w$$

For en prosess med bare trykk–volum-arbeid (som er et vanlig tilfelle i kjemiske reaksjoner og faseoverganger) er

$$\delta w=-P_\mathrm{ext}\,dV$$

slik at

$$dU=\delta q-P_\mathrm{ext}\,dV$$

Setter vi dette inn i uttrykket for $dH$, får vi

$$dH=\delta q-P_\mathrm{ext}\,dV+P\,dV+V\,dP$$

Vi kan samle volumleddene:

$$dH=\delta q+(P-P_\mathrm{ext})\,dV+V\,dP$$

Ved en reversibel prosess er det ytre trykket hele tiden tilnærmet lik systemets trykk, altså $P_\mathrm{ext}=P$. Da forsvinner det ene volumleddet:

$$dH=\delta q+V\,dP$$

Når trykket er konstant, er $dP=0$, og vi står igjen med

$$dH=\delta q_P$$

For en endelig prosess ved konstant trykk får vi derfor

$$\Delta H=q_P$$

Dette er grunnen til at entalpi er så nyttig i kjemien: Under vanlige laboratoriebetingelser kan vi ofte knytte den målbare varmeoverføringen direkte til en endring i en tilstandsfunksjon.

:::{admonition} Viktig
:class: important

Entalpi er **ikke varmeinnhold**. $H$ er en tilstandsfunksjon. Under betingelser med konstant trykk og bare trykk–volum-arbeid er endringen i entalpi lik varmen som overføres:

$$\Delta H=q_P$$
:::


## Varmekapasitet og kalorimetri

### Varmekapasitet i makroskopisk termodynamikk

To forskjellige stoffer kan få svært ulik temperaturendring selv om de tilføres samme mengde energi. Dette var blant observasjonene Joseph Black gjorde allerede på 1700-tallet. I dag beskriver vi dette ved hjelp av **varmekapasitet**.

Fra statistisk termodynamikk så vi hvordan varmekapasitet kan kobles til hvordan systemets energi endres med temperaturen. I makroskopisk termodynamikk skiller vi særlig mellom varmekapasitet ved konstant volum og ved konstant trykk:

$$C_V=\left(\frac{\partial U}{\partial T}\right)_V$$

$$C_P=\left(\frac{\partial H}{\partial T}\right)_P$$

Forskjellen har en klar fysisk betydning. Ved konstant volum kan systemet ikke ekspandere og utføre trykk–volum-arbeid. Ved konstant trykk kan en del av energiøkningen være knyttet til ekspansjon. For en gass er derfor $C_P$ vanligvis større enn $C_V$.

I praksis er varmekapasitet et viktig verktøy for å koble en målbar temperaturendring til energi som er overført som varme. Vi kan måle temperaturendringen $\Delta T$ og beregne varmen som er tilført eller avgitt ved hjelp av kalorimetri.

### Kalorimetri

**Kalorimetri** er en eksperimentell metode for å måle varmeoverføring. Dersom varmekapasiteten $C$ kan behandles som konstant over temperaturintervallet, har vi

$$q=C\Delta T$$

Det er nyttig å skille mellom to vanlige typer kalorimetri.

1. Kaffekoppkalorimetri

I et enkelt kaffekoppkalorimeter foregår prosessen ved omtrent konstant atmosfærisk trykk. Reaktanter blandes, og temperaturendringen måles. Siden trykket er konstant, er varmeoverføringen direkte knyttet til entalpiendringen:

$$q_P=\Delta H$$

Slike målinger er en viktig kilde til eksperimentelle reaksjonsentalpier.

2. Bombekalorimetri

I et bombekalorimeter foregår prosessen i en stiv, lukket beholder. Volumet er konstant, slik at trykk–volum-arbeidet er null. Da blir

$$q_V=\Delta U$$

Kalorimetri viser dermed svært konkret hvorfor både $U$ og $H$ er nyttige: Hvilken av dem som kobles direkte til den målte varmen, avhenger av hvilke betingelser eksperimentet utføres under.


## Hvorfor trenger vi flere termodynamiske potensialer?

Den indre energien $U$ er den mest grunnleggende energifunksjonen, men i laboratoriet er det ofte ikke entropi og volum som er de enkleste størrelsene å kontrollere. Vanligvis kontrollerer vi heller temperatur og trykk.

Derfor innfører vi ulike **termodynamiske potensialer**. De inneholder den samme termodynamiske informasjonen, men er tilpasset forskjellige betingelser.

| Potensial | Definisjon | Naturlige variabler | Særlig nyttig når |
|---|---|---|---|
| Indre energi $U$ | $U$ | $S,V$ | $S$ og $V$ er kontrollert |
| Entalpi $H$ | $H=U+PV$ | $S,P$ | $P$ er kontrollert |
| Helmholtz-energi $F$ | $F=U-TS$ | $T,V$ | $T$ og $V$ er kontrollert |
| Gibbs-energi $G$ | $G=H-TS$ | $T,P$ | $T$ og $P$ er kontrollert |

### Helmholtz-energi

Helmholtz-energien er definert som

$$F=U-TS$$

Ved konstant temperatur og volum går et lukket system spontant i retning av lavere $F$ fram mot likevekt.

### Gibbs-energi

Gibbs-energien er definert som

$$G=H-TS$$

Ved konstant temperatur og trykk går et lukket system spontant i retning av lavere $G$ fram mot likevekt.

Dette er grunnen til at Gibbs-energien blir særlig viktig i kjemi: temperatur og trykk er ofte de naturlige ytre betingelsene i laboratoriet.

:::{admonition} Termodynamikk, ikke kinetikk
:class: note

At $G$ avtar forteller hvilken retning som er termodynamisk gunstig ved konstant $T$ og $P$. Det sier ikke hvor raskt prosessen skjer. Reaksjonshastighet er et kinetisk spørsmål.
:::


## Fundamentallikningen – broen mellom mikro og makro

Nå skal vi samle flere av ideene fra de foregående kapitlene i én likning.

For en **reversibel** prosess gjelder

$$\delta q_\mathrm{rev}=T\,dS$$

og for reversibelt trykk–volum-arbeid

$$\delta w_\mathrm{rev}=-P\,dV$$

Setter vi dette inn i første lov,

$$dU=\delta q+\delta w$$

får vi

$$dU=T\,dS-P\,dV$$

:::{admonition} Fundamentallikningen for indre energi
:class: important

For et lukket system med konstant stoffmengde og bare trykk–volum-arbeid er

$$dU=T\,dS-P\,dV$$

Likningen viser at $U$ naturlig kan betraktes som en funksjon av $S$ og $V$:

$$U=U(S,V)$$
:::

### Temperaturen dukker opp igjen

Fra fundamentallikningen kan vi lese av

$$T=\left(\frac{\partial U}{\partial S}\right)_{V,N}$$

I kapitlet om entropi og Boltzmann fant vi den statistiske temperaturdefinisjonen

$$\frac{1}{T}=\left(\frac{\partial S}{\partial U}\right)_{V,N}$$

Dette er de samme to størrelsene skrevet som inverse deriverte. Her møtes den statistiske og den makroskopiske termodynamikken direkte.

På mikronivå fortalte $\left(\partial S/\partial U\right)_{V,N}$ hvor mye entropien øker når systemet mottar litt energi. På makronivå forteller den inverse deriverte hvor mye den indre energien må endres for en liten entropiendring.

### Trykket i fundamentallikningen

På samme måte kan vi lese av

$$-P=\left(\frac{\partial U}{\partial V}\right)_{S,N}$$

I gasskapitlet forklarte vi trykk mekanisk som resultatet av molekylenes kollisjoner med veggene. Her ser vi den samme størrelsen som en termodynamisk derivert.

Vi kan også skrive fundamentallikningen på entropiform:

$$dS=\frac{1}{T}\,dU+\frac{P}{T}\,dV$$

slik at

$$\frac{P}{T}=\left(\frac{\partial S}{\partial V}\right)_{U,N}$$

Dette viser enda en gang hvordan de makroskopiske størrelsene $T$ og $P$ kan kobles til entropiens respons på endringer i energi og volum.


## Fra indre energi til de andre potensialene

Når vi først kjenner

$$dU=T\,dS-P\,dV$$

kan vi finne differensialene til de andre termodynamiske potensialene ved vanlig differensiering.

### Entalpi

Fra

$$H=U+PV$$

får vi

$$dH=dU+P\,dV+V\,dP$$

og dermed

$$dH=T\,dS+V\,dP$$

Derfor er de naturlige variablene til entalpi $S$ og $P$:

$$T=\left(\frac{\partial H}{\partial S}\right)_P$$

$$V=\left(\frac{\partial H}{\partial P}\right)_S$$

### Helmholtz- og Gibbs-energi

:::{admonition} Underveisoppgave
:class: tip

Bruk $F=U-TS$ og $G=H-TS$ til å utlede uttrykkene for $dF$ og $dG$.
:::

:::{admonition} Løsningsforslag
:class: tip dropdown

For Helmholtz-energien:

$$F=U-TS$$

$$dF=dU-T\,dS-S\,dT$$

Setter vi inn $dU=T\,dS-P\,dV$, får vi

$$dF=-S\,dT-P\,dV$$

Dermed

$$S=-\left(\frac{\partial F}{\partial T}\right)_V$$

og

$$P=-\left(\frac{\partial F}{\partial V}\right)_T$$

For Gibbs-energien:

$$G=H-TS$$

$$dG=dH-T\,dS-S\,dT$$

Setter vi inn $dH=T\,dS+V\,dP$, får vi

$$dG=-S\,dT+V\,dP$$

Dermed

$$S=-\left(\frac{\partial G}{\partial T}\right)_P$$

og

$$V=\left(\frac{\partial G}{\partial P}\right)_T$$
:::

:::{admonition} Fundamentallikningene
:class: important

$$dU=T\,dS-P\,dV$$

$$dH=T\,dS+V\,dP$$

$$dF=-S\,dT-P\,dV$$

$$dG=-S\,dT+V\,dP$$
:::


## Maxwell-relasjonene

Fundamentallikningene inneholder mer informasjon enn det først kan se ut som. Fordi $U$, $H$, $F$ og $G$ er tilstandsfunksjoner, har de **eksakte differensialer**. Dette gir opphav til Maxwell-relasjonene.

For en generell tilstandsfunksjon $X(a,b)$ kan vi skrive

$$dX=\left(\frac{\partial X}{\partial a}\right)_b da+\left(\frac{\partial X}{\partial b}\right)_a db$$

Dersom funksjonen er tilstrekkelig glatt, er de blandede andrederiverte like:

$$\frac{\partial}{\partial b}\left(\frac{\partial X}{\partial a}\right)_b=
\frac{\partial}{\partial a}\left(\frac{\partial X}{\partial b}\right)_a$$

Dette er hele matematikken som ligger bak Maxwell-relasjonene.

:::{dropdown} Eksempel: Maxwell-relasjonen fra $U$

Fra fundamentallikningen

$$dU=T\,dS-P\,dV$$

kan vi lese av

$$T=\left(\frac{\partial U}{\partial S}\right)_V$$

og

$$-P=\left(\frac{\partial U}{\partial V}\right)_S$$

Likhet mellom blandede andrederiverte gir da

$$\left(\frac{\partial T}{\partial V}\right)_S
=
-\left(\frac{\partial P}{\partial S}\right)_V$$

Poenget er at en endring som involverer en vanskelig målbar størrelse kan kobles til en annen derivert som i enkelte sammenhenger er lettere å bruke.
:::

:::{dropdown} De fire Maxwell-relasjonene

Fra $U(S,V)$:

$$\left(\frac{\partial T}{\partial V}\right)_S
=
-\left(\frac{\partial P}{\partial S}\right)_V$$

Fra $H(S,P)$:

$$\left(\frac{\partial T}{\partial P}\right)_S
=
\left(\frac{\partial V}{\partial S}\right)_P$$

Fra $F(T,V)$:

$$\left(\frac{\partial P}{\partial T}\right)_V
=
\left(\frac{\partial S}{\partial V}\right)_T$$

Fra $G(T,P)$:

$$\left(\frac{\partial V}{\partial T}\right)_P
=
-\left(\frac{\partial S}{\partial P}\right)_T$$
:::

I dette kurset er det viktigst å forstå **hvor relasjonene kommer fra**: De følger av at de termodynamiske potensialene er tilstandsfunksjoner. Selve utledningen er nyttig matematisk trening, men relasjonene er først og fremst verktøy for å koble sammen termodynamiske egenskaper.


## Kjemisk potensial

Så langt har vi stort sett behandlet lukkede systemer der stoffmengden er konstant. I kjemi er vi imidlertid ofte interessert i systemer der sammensetningen kan endre seg: stoff kan diffundere, blandes, gå fra én fase til en annen eller reagere kjemisk.

Da trenger vi en ny termodynamisk størrelse: **kjemisk potensial**.

For komponent $i$ defineres det kjemiske potensialet ved konstant temperatur og trykk som

$$\mu_i=\left(\frac{\partial G}{\partial n_i}\right)_{T,P,n_{j\ne i}}$$

Uttrykket forteller hvor mye Gibbs-energien endres når vi tilfører en liten mengde av komponent $i$, mens temperaturen, trykket og mengden av de andre komponentene holdes konstante.

Kjemisk potensial kan derfor ses som en termodynamisk «tendens» til å flytte eller omfordele stoff. En forskjell i kjemisk potensial gir en drivkraft for materieoverføring, på samme måte som en temperaturforskjell gir en drivkraft for varmeoverføring.

![Kjemisk potensial i et bilde](bilder/diffusjon_bilde.png)

Tenk på en dråpe konditorfarge som dryppes i et glass vann. Selv uten omrøring sprer fargestoffet seg gradvis gjennom vannet. På mikronivå beveger molekylene seg tilfeldig, men makroskopisk ser vi en bestemt retning på prosessen: den opprinnelige konsentrasjonsforskjellen jevnes ut. Den mer generelle termodynamiske beskrivelsen er at stoffet omfordeles i retning av lavere Gibbs-energi, og ved likevekt er det kjemiske potensialet til komponenten det samme i hele systemet.

:::{admonition} Underveisoppgave
:class: tip

Konditorfarge dryppes i et glass vann og sprer seg utover uten omrøring. Hvorfor er det ikke tilstrekkelig å si at «molekylene bare beveger seg tilfeldig» dersom vi også ønsker å forklare den makroskopiske retningen på prosessen? Hvordan kan kjemisk potensial brukes i forklaringen?
:::

### Kobling tilbake til statistisk termodynamikk

Kjemisk potensial kan også kobles til partisjonsfunksjonen. For et kanonisk system er Helmholtz-energien

$$F=-k_\mathrm{B}T\ln Q_N$$

og

$$\mu=\left(\frac{\partial F}{\partial N}\right)_{T,V}$$

Dermed kan vi skrive

$$\mu=-k_\mathrm{B}T\left(\frac{\partial\ln Q_N}{\partial N}\right)_{T,V}$$

Dette er enda en bro mellom den statistiske og den makroskopiske beskrivelsen: Når antallet partikler endres, endres også de tilgjengelige mikrotilstandene og sannsynlighetsfordelingen. Den statistiske endringen kommer makroskopisk til uttrykk gjennom det kjemiske potensialet.

:::{admonition} Åpne systemer
:class: note

Når stoffmengden kan endres, utvides fundamentallikningen til

$$dU=T\,dS-P\,dV+\sum_i\mu_i\,dn_i$$

Leddet $\mu_i\,dn_i$ beskriver bidraget til endringen i indre energi som følger av at mengden av komponent $i$ endres.
:::


## Oppsummering

I dette kapitlet har vi gått fra den statistiske beskrivelsen tilbake til den klassiske, makroskopiske termodynamikken.

Vi startet med **varme og arbeid** som to måter energi kan krysse systemgrensen på, og brukte første lov

$$\Delta U=q+w$$

til å beskrive endringen i systemets indre energi.

For trykk–volum-arbeid fant vi

$$\delta w=-P_\mathrm{ext}\,dV$$

og så hvorfor en reversibel prosess gir en viktig idealisert grense for hvor mye arbeid et system kan utføre.

Deretter innførte vi **entalpi**, **Helmholtz-energi** og **Gibbs-energi**. De ulike potensialene er nyttige fordi forskjellige eksperimentelle betingelser gjør ulike variabler naturlige å kontrollere.

Den sentrale forbindelsen mellom termodynamiske størrelser er fundamentallikningen

$$dU=T\,dS-P\,dV$$

Her møtes også den statistiske og makroskopiske temperaturdefinisjonen direkte.

Til slutt så vi hvordan de eksakte differensialene til termodynamiske tilstandsfunksjoner gir **Maxwell-relasjonene**, og hvordan **kjemisk potensial** utvider beskrivelsen til systemer der sammensetningen kan endres.

I de neste kapitlene skal vi bruke disse verktøyene på kjemiske prosesser, faseoverganger, blandinger og likevekt.


## Oppgaver

### Oppgave 1 – Varme, arbeid og indre energi

Et lukket system går fra tilstand A til tilstand B. I én prosessvei tar systemet opp $500\ \mathrm{J}$ varme og gjør $200\ \mathrm{J}$ arbeid på omgivelsene.

a) Finn $\Delta U$.

b) En annen prosessvei mellom de samme tilstandene har $q=100\ \mathrm{J}$. Hva må $w$ være?

c) Forklar hvorfor $q$ og $w$ kan være forskjellige for de to prosessveiene selv om $\Delta U$ er den samme.

:::{dropdown} Løsningsforslag

a) Arbeid utført av systemet betyr $w=-200\ \mathrm{J}$:

$$\Delta U=q+w=500-200=300\ \mathrm{J}$$

b)

$$300=100+w$$

slik at

$$w=200\ \mathrm{J}$$

I denne prosessveien gjør omgivelsene netto arbeid på systemet.

c) Varme og arbeid avhenger av prosessveien. Indre energi er en tilstandsfunksjon og avhenger bare av start- og sluttilstanden.
:::


### Oppgave 2 – Entalpi og kalorimetri

En reaksjon foregår i et kaffekoppkalorimeter ved konstant trykk. Kalorimeteret og innholdet har samlet varmekapasitet $C_P=850\ \mathrm{J\,K^{-1}}$. Temperaturen øker med $4{,}2\ \mathrm{K}$.

a) Hvor mye varme tar kalorimeteret og innholdet opp?

b) Hva er reaksjonens entalpiendring dersom all denne varmen kommer fra reaksjonen?

c) Hvorfor ville et bombekalorimeter i stedet gi direkte informasjon om $\Delta U$?

:::{dropdown} Løsningsforslag

a)

$$q_\mathrm{kal}=C_P\Delta T=(850)(4{,}2)=3{,}57\ \mathrm{kJ}$$

b) Reaksjonen avgir den samme energimengden:

$$q_\mathrm{rxn}=-3{,}57\ \mathrm{kJ}$$

Ved konstant trykk er

$$\Delta H=q_P=-3{,}57\ \mathrm{kJ}$$

c) I et bombekalorimeter er volumet konstant, slik at trykk–volum-arbeidet er null. Da er $q_V=\Delta U$.
:::


### Oppgave 3 – Hvilket potensial?

For hver situasjon, velg det mest naturlige termodynamiske potensialet blant $U$, $H$, $F$ og $G$.

a) Et lukket system ved konstant temperatur og volum.

b) En kjemisk reaksjon i en åpen kolbe ved konstant temperatur og trykk.

c) Et system der entropi og volum er de naturlige uavhengige variablene.

d) Forklar hvorfor Gibbs-energien er spesielt sentral i kjemi.

:::{dropdown} Løsningsforslag

a) Helmholtz-energien $F$.

b) Gibbs-energien $G$.

c) Indre energi $U$.

d) Mange kjemiske prosesser studeres ved omtrent konstant temperatur og trykk. Under disse betingelsene går et lukket system spontant i retning av lavere $G$ fram mot likevekt.
:::


### Oppgave 4 – Fundamentallikningen

Ta utgangspunkt i

$$dU=T\,dS-P\,dV$$

a) Hvilke naturlige variabler har $U$?

b) Vis hvordan temperatur kan skrives som en partiell derivert av $U$.

c) Vis hvordan trykk kan skrives som en partiell derivert av $U$.

d) Forklar hvordan uttrykket for temperatur henger sammen med den statistiske definisjonen

$$\frac1T=\left(\frac{\partial S}{\partial U}\right)_{V,N}$$

:::{dropdown} Løsningsforslag

a) $U=U(S,V)$ for konstant stoffmengde.

b)

$$T=\left(\frac{\partial U}{\partial S}\right)_{V,N}$$

c)

$$-P=\left(\frac{\partial U}{\partial V}\right)_{S,N}$$

d) De to temperaturuttrykkene er inverse deriverte når de samme øvrige variablene holdes konstante:

$$\left(\frac{\partial U}{\partial S}\right)_{V,N}
=
\frac{1}{\left(\frac{\partial S}{\partial U}\right)_{V,N}}$$

Dermed beskriver den statistiske og makroskopiske formuleringen den samme temperaturen.
:::


### Oppgave 5 – Kjemisk potensial og likevekt

To områder A og B kan utveksle stoff av samme komponent. Temperaturen og trykket er de samme, men $\mu_A>\mu_B$.

a) I hvilken retning er materieoverføring termodynamisk gunstig?

b) Hva er betingelsen for likevekt med hensyn på materieoverføring?

c) Hvorfor er det utilstrekkelig å si at stoff alltid beveger seg «fra høy konsentrasjon til lav konsentrasjon»?

:::{dropdown} Løsningsforslag

a) Overføring fra A mot B er termodynamisk gunstig.

b) Ved likevekt er

$$\mu_A=\mu_B$$

c) Konsentrasjon påvirker kjemisk potensial, men kjemisk potensial avhenger generelt også av blant annet intermolekylære vekselvirkninger, trykk og sammensetning. Det er forskjellen i kjemisk potensial som er den generelle termodynamiske drivkraften.
:::
