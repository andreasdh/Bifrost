# Systemer og størrelser

I denne modulen introduserer vi noen sentrale begreper i termodynamikken. Dette er begreper som absolutt bør pugges for å forstå resten av kurset, og det er lurt å få det inn så tidlig som mulig. Noen av disse begrepene vil virke abstrakte nå, men det er uansett lurt å kunne definisjonene deres, for så å bygge intuisjonen for dem mens du arbeider med emnet. 

Blir du usikker senere, kan denne modulen også brukes som et "oppslagsverk" mens du leser. 

## Hva er et system?

Et **termodynamisk system** er en del av universet vi velger ut for å studere. Resten av universet er da **omgivelsene**. Systemet og omgivelsene er skilt fra hverandre av en fysisk eller imaginær barriere. 

## Ensembler

**Ensembler** er modeller som brukes i termodynamikken. De ulike ensemblene defineres ut fra hvilke størrelser systemet vårt *ikke* kan utveksle med omgivelsene:

**Mikrokanonisk ensemble:** isolert system med konstant energi $E$, volum $V$ og antall partikler $N$. Se for deg en termos med varmt vann; ingen energi og dermed ingen varme slipper ut, og vann kan ikke tilsettes eller fjernes. 

**Kanonisk ensemble:** lukket system med konstant temperatur $T$, volum $V$ og antall partikler $N$. Hvis termosen vår er åpen og står i et rom med samme temperatur som termosen, kan energi i form av varme fluktuere mellom termosen og luften i rommet, men det er fortsatt forbudt å tilsette eller fjerne vann. 

**Grand kanonisk ensemble:** åpent system med konstant temperatur $T$, volum $V$ og kjemisk potensiale $\mu$. Både energi og antallet partikler fluktuerer hele tiden. 

I dette emnet arbeider vi mest med det kanoniske ensemblet. 

## Mikroskopiske og Makroskopiske tilstander

**Mikroskopisk:**

- Mikroskopisk nivå: beskriver individuelle atomer eller molekyler.
- Mikroskopisk egenskap: energier, posisjoner og hastigheter til individuelle atomer eller molekyler.
- Mikroskopisk tilstand aka **mikrostilstand**: kartlagte energier til hvert enkelt atom eller molekyl i systemet. 

**Makroskopisk:**

- Makroskopisk nivå: beskriver systemet som helhet, slik vi kan observere og måle det. For eksempel egenskapene til en gass (bestående av flere enkelte gassmolekyler) i en beholder. 
- Makroskopisk egenskap: temperatur, trykk, volum, indre energi osv. Størrelser vi kan måle på laben. 
- Makroskopisk tilstand aka **makrotilstand**: en bestemt kombinasjon av målbare størrelser (f.eks. gitt temperatur, trykk og volum) som beskriver systemet som helhet. 

**Mikrotilstander tilsvarer makrotilstander**: 

Tenk deg at du kaster to terninger. Det du bryr deg om er summen av øynene, det er makrotilstanden. Men det finnes mange forskjellige kombinasjoner av terning 1 og terning 2 som gir akkurat samme sum. Hver slik kombinasjon er en egen mikrotilstand.

![Mikro og makro introduksjon](Simuleringer/terningkast_mikro_vs_makro.png)

Èn eller flere mikrotilstander kan altså tilsvare en makrotilstand, men du kan ikke gå baklengs. Hvis du bare vet at summen er 7, vet du ikke om det var (3,4), (1,6) eller (5,2) som ble kastet.  

## Intensive og Ekstensive størrelser

Vi deler de termodynamiske størrelsene inn i to kategorier, basert på om størrelsen avhenger av *mengden stoff*, altså antall atomer/molekyler, eller ikke. 

**Intensive størrelser:** Avhenger ikke av mengden stoff (f.eks. temperatur, trykk, tetthet).

**Ekstensive størrelser:** Avhenger av mengden stoff (f.eks. volum, masse, indre energi).

## Prosesser

Hva er egentlig en prosess? Vi bruker ordet hele tiden i kjemi og fysikk for å beskrive at *noe skjer*, men kan det spesifiseres litt mer i konteksten av kjemi?

En **prosess** er en endring i et systems tilstand, systemet går fra én tilstand (f.eks. gitt trykk, volum og temperatur) til en annen.

### Sti- og tilstandsfunksjoner

**Tilstandsfunksjoner:** Avhenger kun av start- og sluttilstand, ikke av veien dit (f.eks. indre energi, entalpi, entropi).

**Stifunksjoner:** Avhenger av selve veien mellom start- og sluttilstand (f.eks. varme og arbeid).

## Energi

**Kinetisk energi:** Bevegelsesenergi, energien et objekt har på grunn av sin hastighet (f.eks. molekyler i bevegelse).

**Potensiell energi:** Lagret energi på grunn av posisjon eller tilstand (f.eks. kjemisk bindingsenergi, energi i en strukket fjær).
