## Hva er et system?

Et **termodynamisk system** er en del av universet vi velger ut for å studere. Resten av universet er da **omgivelsene**. Systemet og omgivelsene er skilt fra hverandre av en fysisk eller imaginær barriere. Hvordan du definerer grensene til systemet ditt, avhenger av hva slags problem du vil løse. 

Et **delsystem** er en avgrenset del av et større system som vi velger å se isolert på, for å forenkle analysen. I praksis betyr det at du "tegner en grense" et sted i et system, og alt innenfor denne grensen kalles delsystemet, mens resten ligger utenfor og kalles da omgivelsene. Et enkelt eksempel: hvis du studerer ett spesifikt gassmolekyl i en beholder med mange molekyler, kan du behandle det ene molekylet som et delsystem, og resten av molekylene som omgivelsene det vekselvirker med.

## Ensembler

**Ensembler** er modeller som brukes i termodynamikken. De ulike ensemblene defineres ut fra hvilke størrelser systemet vårt *ikke* kan utveksle med omgivelsene:

**Mikrokanonisk ensemble:** isolert system med konstant energi $E$, volum $V$ og antall partikler $N$. Se for deg en termos med varmt vann; ingen energi og dermed ingen varme slipper ut, og vann kan ikke tilsettes eller fjernes. 

**Kanonisk ensemble:** lukket system med konstant temperatur $T$, volum $V$ og antall partikler $N$. Hvis termosen vår er åpen og står i et rom med samme temperatur som termosen, kan energi i form av varme fluktuere mellom termosen og luften i rommet, men det er fortsatt forbudt å tilsette eller fjerne vann. 

**Grand kanonisk ensemble:** (også kjent som makrokanonisk ensemble) åpent system med konstant temperatur $T$, volum $V$ og kjemisk potensiale $\mu$. Både energi og antallet partikler fluktuerer hele tiden. 

Når en størrelse holdes konstant for systemet, vises dette ofte i notasjonen. Det er bare for å gjøre det enkelt og oversiktlig å se hva som må til for at det gitte uttrykket vårt faktisk gjelder. Et eksempel er uttrykket for varmekapasiteten $C_V$, som vi skal tilbake til senere i kurset:   

$$ C_V = \left(\frac{\partial U}{\partial T}\right)_{V,N}.$$

Her er parentesen markert med $V$ og $N$. Dette betyr altså at uttrykket gjelder når vi har konstant volum og antall partikler. Du ser også at $C$ er markert med en $V$, altså konstant volum.

Alle tre ensemblene er bare modeller, eller antakelser, de beskriver ikke virkeligheten helt nøyaktig, men noen av dem er enklere å realisere enn andre. Det mikrokanoniske ensemblet er i praksis umulig, slike systemer finner vi ikke i naturen. Det grand kanoniske ensemblet er mulig å finne i naturen, men er ganske vanskelig å beskrive matematisk. I dette emnet arbeider vi mest med det kanoniske ensemblet. 

## Mikrotilstander og makrotilstander

En **makrotilstand** beskriver systemet med størrelser vi kan måle i laboratoriet: temperatur, trykk, volum, total energi. En **mikrotilstand** er én bestemt måte energien er fordelt på molekylnivå, konsistent med den observerte makrotilstanden. 

**Mikroskopisk:**

- Mikroskopisk nivå: når vi beskriver individuelle atomer eller molekyler.
- Mikroskopisk egenskap: energier, posisjoner og hastigheter til individuelle atomer eller molekyler.
- Mikroskopisk tilstand aka mikrostilstand: kartlagte energier til hvert enkelt atom eller molekyl i systemet. 

**Makroskopisk:**

- Makroskopisk nivå: når vi beskriver et system som helhet, slik vi kan observere og måle det. For eksempel egenskapene til en gass (bestående av flere enkelte gassmolekyler) i en beholder. 
- Makroskopisk egenskap: temperatur, trykk, volum, indre energi osv. Størrelser vi kan måle på laben. 
- Makroskopisk tilstand aka makrotilstand: en bestemt målbar størrelse (f.eks. energi, trykk, volum), som tilsvarer én eller flere mikrotilstander. 

La oss se på et enkelt eksempel utenfor molekylverdenen, der vi kaster fire mynter. Dette har vi illustrert nedenfor, der antallet mikrotilstander for makrotilstandene en «kron», «to kron» og «tre kron» er representert med de mulige makrotilstandene dette tilsvarer.

## Intensive og ekstensive størrelser

Vi deler de termodynamiske størrelsene inn i to kategorier, basert på om størrelsen avhenger av *mengden stoff*, altså antall atomer/molekyler, eller ikke. Hvis du kutter et system i to, vil verdien av størrelsen også halveres?

**Intensive størrelser:** Avhenger ikke av mengden stoff (f.eks. temperatur, trykk, tetthet, farge, smeltepunkt).

Å gange en intensiv størrelse med en ekstensiv størrelse, gir en ekstensiv størrelse. For eksempel er $\rho \times V = m$. 

**Ekstensive størrelser:** Avhenger av mengden stoff (f.eks. volum, masse, total energi, mol).

En ekstensiv størrelse gjøres intensiv ved å dele den på en annen ekstensiv størrelse. For eksempel masse $m$ delt på volum: $m / V = \rho$ gir tettheten til et stoff. 

## Prosesser

Hva er egentlig en prosess? Vi bruker ordet hele tiden i kjemi og fysikk for å beskrive at *noe skjer*, men kan det spesifiseres litt mer i konteksten av kjemi?

En **prosess** er en *endring i et systems tilstand*, systemet går fra én tilstand (f.eks. gitt trykk, volum og temperatur) til en annen.

### Sti- og tilstandsfunksjoner

**Tilstandsfunksjoner:** Avhenger kun av start- og sluttilstand, ikke av veien dit (f.eks. indre energi, entalpi, entropi).

**Stifunksjoner:** Avhenger av selve veien mellom start- og sluttilstand (f.eks. varme og arbeid).

Et intuitivt eksempel kan være en fjelltur: Vi kan ta mange ulike stier opp et fjell (arbeid), og spise en variende mengde Kvikk Lunsj for å komme oss opp fjellet (varme), men til slutt vil man uansett ha kommet seg fra bunn til topp og dette representerer den indre energien, som kun avhenger av start- og slutt-posisjon og ingen ting imellom⛰️

### Ulike typer prosesser

**Isoterm:** Temperaturen holdes konstant gjennom hele prosessen ($\Delta T = 0$).

**Isokor:** Volumet holdes konstant gjennom hele prosessen ($\Delta V = 0$), også kalt en isovolumetrisk prosess.

**Isobar:** Trykket holdes konstant gjennom hele prosessen ($\Delta P = 0$).

**Adiabatisk:** Ingen varme utveksles med omgivelsene ($q = 0$), all endring i indre energi skyldes da arbeid.

## Energi

**Kinetisk energi:** Bevegelsesenergi, energien et objekt har på grunn av sin hastighet (f.eks. molekyler i bevegelse). Klassisk uttrykkes den som

$$E_k = \frac{1}{2}mv^2$$

der $m$ er massen og $v$ er farten. Legg merke til at kinetisk energi øker med *kvadratet* av farten: dobler du farten, firedobler du den kinetiske energien.

**Potensiell energi:** Lagret energi på grunn av posisjon eller tilstand. To kjente eksempler fra klassisk mekanikk:

- Gravitasjonell potensielle energi: $E_p = mgh$ (høyde over bakken).
- Potensiell energi i en fjær: $E_p = \frac{1}{2}kx^2$ (hvor langt fjæren er strukket/komprimert).

I kjemi er det ikke høyde eller fjærer som bestemmer potensiell energi, men **avstanden mellom atomer/molekyler**. Dette kalles ofte **kjemisk energi**, og henger sammen med hvor sterkt atomene tiltrekker eller frastøter hverandre avhengig av hvor langt fra hverandre de er.

**Total energi** i et system er (i klassisk forstand) summen av disse to: $E = E_k + E_p$. 

## Empiriske vs. statistiske lover

**Empiriske lover** er matematiske uttrykk som har blitt satt opp til å matche observasjoner av makroskopisk oppførsel. En observasjon blir gjort $\rightarrow$ vi ønsker å beskrive det vi så matematisk $\rightarrow$ vi finner et uttrykk som beskriver observasjonen. 

**Statistiske lover** er matematiske uttrykk som er utledet fra grunnleggende prinsipper og sannsynlighetsbetraktninger om systemets mikroskopiske bestanddeler. I stedet for å tilpasses observasjoner i etterkant, følger loven direkte fra antakelser om hvordan de mange partiklene i systemet oppfører seg: vi bruker fysikkens antakelser og grunnleggende lover $\rightarrow$ vi summerer/gjennomsnitter over et enormt antall mikrotilstander som følger disse lovene $\rightarrow$ vi ender opp med et uttrykk som beskriver systemets makroskopiske oppførsel.

## Kontinuerlige vs. diskrete verdier

**Kontinuerlige verdier:** dette er den typen størrelse du alltid forholder deg til, og det er sånn vi kjenner verden. Du kan befinne deg nøyaktig hvor du vil i et rom, vannglasset ditt kan ha hvilken som helst mengde vann frem til det blir fult, bilen din kan kjøre hvilken som helst hastighet opp til en maksgrense. Men denne maksgrensen er ikke noe grunnleggende ved universet, det er bare en begrensning hos bilen. Kontinuerlig betyr at det ikke er noen avbrudd eller hopp, for eksempel når du går opp en bakke må du være på hver eneste posisjon fra bunnen til toppen, du kan ikke eksistere ved bunnen og plutselig på toppen uten å forflytte deg veien imellom. Klassisk fysikk, dvs. mest sannsynlig all fysikken du har vært borti tidligere, beskriver kontinuerlige størrelser. 

**Diskrete verdier:** en diskret verdi er en spesifikk, adskilt verdi. Tenk på trinnene i en stige. Det finnes ikke noe trinn nummer $1.5$ eller $\frac{1}{4}$, kun $0, 1, 2$ osv, og det er umulig å flyte i luften mellom trinnene. Dette er selvfølgelig bare en illustrasjon, i virkeligheten kommer diskrete energier fra kvantemekanikken, og vi sier at energien er **kvantisert**. Når vi snakker om diskret energi senere, snakker vi om helt bestemte verdier av en eller annen energi $E_0, E_1, E_2,...$ som et system kan ha, og det er ikke mulig for systemet å ha en energi med en verdi som ligger *mellom* disse. Dette er en iboende egenskap ved vårt univers, det er noe vi har oppdaget, ikke oppfunnet. Du har trolig vært borti det før når vi snakker om eksitasjon og deeksitasjon av elektroner i orbitaler.

## Termodynamikkens lover

Disse kommer vi tilbake til igjen og igjen og igjen. Her er en opplisting av konkrete formuleringer av lovene, forståelse for dem vil du bygge underveis i emnet. 

:::{admonition} Termodynamikkens 0. lov
:class: important

:::{admonition} Termodynamikkens 1. lov
:class: important
Energi kan ikke skapes eller ødelegges, men den kan endre form eller forflytte seg mellom system og omgivelser. Total energi i et lukket system endrer seg ikke. 

$$ \Delta U = q - w,$$

der $$\Delta U$$ er den indre energien i et system, $q$ er varmen og $w$ er arbeidet.

:class: important

$$ \Delta S_{univers} > 0.$$

## Grunnleggende antakelser i statistisk mekanikk

I dette kurset skal vi begynne å beskrive kjemien statistisk, det vil si at vi beskriver egenskaper til makroskopiske systemer basert på felles egenskaper til maaange individuelle mikroskopiske systemer. Dette er helt nytt, så før vi introduserer statistisk mekanikk skikkelig, går vi gjennom noen grunnleggende antakelser som vi bør ha med oss i bakhodet. 

**Kjemisk identiske molekyler beskrives av nøyaktig samme fysikk:** To molekyler av samme stoff følger de samme fysiske lovene og har de samme mulige energiene, det finnes *ingen* skjulte forskjeller mellom dem.

**Makroskopiske variable er kontinuerlige:** Selv om systemet egentlig består av et diskret (heltallig) antall partikler, er dette tallet så stort at makroskopiske størrelser (temperatur, trykk, volum osv.) med god tilnærming kan behandles som kontinuerlige variabler.

**Målte egenskaper reflekterer ensemblet's gjennomsnitt:** Det vi måler eksperimentelt er ikke egenskapen til ett enkelt molekyl, men et gjennomsnitt tatt over det enorme antallet mikrotilstander systemet befinner seg i (eller beveger seg gjennom).

### Kvantisert energi

**Kvantemekanikken** er en matematisk beskrivelse av oppførselen til enkelte molekyler, atomer eller partikler. Tidligere var det antatt at klassisk fysikk var nok til å beskrive disse systemene, men eksperimentelle funn (Ultrafiolett katastrofe, dobbeltspalteeksperimentet, fotoelektrisk effekt, Comptonspredning, Stern-Gerlach eksperimentet mm.) stemte ikke overens med fysikken slik vi kjente den, og en helt ny gren ble nødvendig for å forklare funnene. Felles for etterarbeidet med mange av disse eksperimentene var at **kvantisering** av energi ble forklaringen for å få observasjonene til å gi mening. 

Klassisk fysikk handler om kontinuerlige spektre av størrelser; kjører du oppover en bakke får du og bilen din (systemet) litt og litt mer potensiell energi mens du beveger deg oppover. Vår virkelighetsoppfattelse utfordres når dette ikke lenger stemmer i kvanteverdenen. Vi bruker her **Franck-Hertz eksperimentet** til å illustrere hva kvantisering er og det faktum at det er en iboende egenskap ved vårt univers. Eksperimentet selv er ikke del av pensum, men den konseptuelle forståelsen er det. Franck-Hertz eksperimentet ble utført i 1914 av James Franck og Gustav Hertz, og ga oss det første beviset på atomenes diskrete, kvantiserte energinivåer. 

I Franck-Hertz eksperimentet ble elektroner sendt gjennom en beholder fylt med kvikksølv ($\text{Hg}$) i gassform, dvs. kvikksølvatomer. Franck og Hertz kontrollerte hvor mye energi elektronene hadde før de kolliderte med kvikksølvatomene, $\varepsilon_i$, for så å måle hvor mye energi elektronene hadde igjen etter de hadde kollidert, $\varepsilon_f$. 

![Franck-Hertz oppsett](Simuleringer/franck_hertz_oppsett.png)

Basert på klassisk fysikk forventet de at elektronene skulle avgi energi til kvikksølvatomene på en gradvis måte; dersom elektronene fikk litt høyere startenergi, skulle de også miste litt mer energi i kollisjonene. De observerte derimot noe helt annet: 

1. Når elektronet hadde lavere enn én bestemt startenergi $\varepsilon_i < \Delta \varepsilon$, kom det uendret ut av kollisjonen. Verdien $\Delta \varepsilon$ representerer bare en energiverdi som ble målt, og du kan tenke på den som energiforskjellen mellom to underskall i atomet. 
2. Når elektronet hadde startenergi $\varepsilon_i=\Delta \varepsilon$ eller høyere, mistet det alltid *samme mengde* energi $\Delta \varepsilon$ under kollisjonen, og dersom startenergien ble økt litt mer, endret det ikke måleverdien. 
3. Først ved enda høyere startenergier kunne elektronet miste *to ganger* denne "minimumsenergien", altså $2\Delta \varepsilon$. 
4. Ved enda høyere startenergier enn dette igjen kunne elektronet miste $3\Delta \varepsilon$, $4\Delta \varepsilon$ osv.

Den grå og blå grafen under representerer henholdsvis forventede observasjoner og hva som faktisk ble observert.

Atomene kunne ikke absorbere vilkårlige energier, de absorberte helt bestemte, diskrete energimengder. De lovlige energitilstandene, eller nivåene, atomene kan eksistere i er dermed **kvantiserte**, dvs. de kan kun eksistere i tilstander av diskrete energier. ### Kvantetilstander

Vi har nå sett at atomer kun kan eksistere i bestemte energinivåer, men *hvor* i atomet ligger denne energien? Og når vi oppskalerer til molekyler, hvordan beskriver vi da alle de mulige energiene et system kan ha? For å svare på dette introduserer vi begrepet **kvantetilstand**. En kvantetilstand beskriver den fullstendige fysiske tilstanden til et kvantesystem (partikkel, atom eller molekyl). **Hver kvantetilstand har en bestemt energi**, og siden energien er kvantisert, finnes det bare bestemte kvantetilstander et system kan befinne seg i. De mulige energiene kan vi skrive som:

$$\varepsilon = \varepsilon_0,\varepsilon_1,\varepsilon_2,...\text{,}$$

$$ \text{eller} \, \varepsilon_n, \text{der} \, n=0,1,2,...$$

Hvert energinivå tilsvarer én eller flere kvantetilstander. Vi velger å bruke $\varepsilon$ i stedet for det vanlige symbolet for energi, $E$, fordi vi nå snakker om enkelte atomer eller molekyler. Dette er kun noe vi velger, vi kunne valgt en emoji i stedet🤠. Med tanke på at vi skal snakke om den totale energien til en samling av *flere* atomer eller molekyler senere, som vi kaller $E$, er det rett og slett bare mer ryddig å bestemme oss for å bruke $\varepsilon$ for veldig små systemer. Stor $E$, stort system. Liten $\varepsilon$, lite system.

For et atom er energien i stor grad knyttet til elektronenes tillatte tilstander rundt atomkjernen. Dette har du vært borti før, allerede i grunnleggende kjemi. Den laveste mulige energien kalles **grunntilstanden**, og betegnes her med $\varepsilon_0$. Grunntilstanden til et atom er når alle elektroner rundt atomkjernen er i sine laveste mulige energinivåer. Alle tilstander med høyere energi kalles **eksiterte tilstander** $\varepsilon_1, \varepsilon_2$ osv. For eksempel hvis et elektron eksiteres, sier vi at *elektronet* er eksitert og at *atomet* er i en eksitert *tilstand*. Dersom et system (atom) går fra en kvantetilstand med energi $\varepsilon_i$ til en annen kvantetilstand med energi $\varepsilon_f$, er energiforskjellen

$$\Delta \varepsilon=\varepsilon_f-\varepsilon_i\text{.}$$

Noen ganger kan to eller flere forskjellige kvantetilstander ha nøyaktig samme energi. Slike tilstander kalles **degenererte tilstander**. De er altså forskjellige fysiske tilstander, selv om energien er den samme. 

![Energier](Simuleringer/energi_stige_atom.png)

For molekyler blir situasjonen mer kompleks. I tillegg til elektronenes energi kan molekylet også bevege seg som en helhet, rotere, og vibrere. Disse uavhengige bevegelsesmåtene kalles **frihetsgrader**, og hver av dem representerer en måte molekylet kan lagre energi på. En frihetsgrad er altså en uavhengig type bevegelse eller dynamikk som bidrar til systemets totale energi $\varepsilon_{tot}$. Grunnen til at vi ikke snakker om frihetsgrader for enkeltatomer, er at et atom verken har noen kjemisk binding som kan vibrere eller noen molekylstruktur som kan rotere, det eneste et enkeltatom kan gjøre er å bevege seg som en helhet (translasjon) og ha elektronene sine i ulike tilstander.

En kvantetilstand for et molekyl må derfor beskrive alle disse bidragene samtidig:

- **Den elektroniske tilstanden** beskriver hvilken kvantetilstand elektronene rundt atomkjernene befinner seg i (tenk orbitalokkupasjon), akkurat det samme prinsippet som for et enkeltatom, bare nå fordelt på flere atomkjerner.
- **Vibrasjonstilstanden** beskriver hvordan atomene i molekylet oscillerer rundt sine likevektsposisjoner i bindingene. Et molekyl med flere atomer kan vibrere på flere uavhengige måter samtidig (kalt normalmoder), og hver av disse har sitt eget sett med kvantiserte energinivåer.
- **Rotasjonstilstanden** beskriver hvordan molekylet som en helhet roterer om en akse. Også rotasjonsenergien er kvantisert, molekylet kan bare rotere ved bestemte, tillatte "hastigheter".
- **Translasjonen** beskriver bevegelsen til hele molekylets massesenter gjennom rommet. 

Vi kan altså tenke på en kvantetilstand som en fullstendig beskrivelse av hvordan energien er fordelt mellom systemets ulike frihetsgrader. Den totale energien til et molekyl, $\varepsilon_n$, kan skrives som summen av bidragene fra hver frihetsgrad:

$$\varepsilon_n=\varepsilon_{tot}=\varepsilon_{elec}+\varepsilon_{trans}+\varepsilon_{rot}+\varepsilon_{vib}\text{.}$$

### Hva er *egentlig* en mikrotilstand?

Nå kan vi knytte kvantetilstander til et begrep vi allerede har møtt: mikrotilstander. En mikrotilstand er én spesifikk måte et system kan være organisert på. I kvantemekanikk tilsvarer en mikrotilstand én bestemt kvantetilstand. Når vi sier at vi kjenner systemets mikrotilstand, mener vi derfor at vi kjenner alle kvantetallene som beskriver systemets frihetsgrader. Dersom flere forskjellige kvantetilstander har samme energi (degenererte tilstander), finnes det altså flere mikrotilstander som tilsvarer den samme energien. Vi kan tenke oss sammenhengen slik: 

$$\text{Mikrotilstand} \longleftrightarrow \text{Kvantetilstand} \longrightarrow \text{Energinivå}$$

Flere forskjellige kvantetilstander kan altså peke mot det samme energinivået, tilstandene er da degenererte. 

Hvis energi ikke var kvantisert, ville hver frihetsgrad kunne anta kontinuerlig mange energier. Da ville antallet mulige kvantetilstander, og dermed mikrotilstander, vært uendelig selv innenfor et begrenset energiintervall. Kvantemekanikkens lover begrenser altså hvilke tilstander naturen faktisk tillater. Det er derfor mikrotilstander ikke bare er et matematisk hjelpemiddel. De er en del av naturen.

Videre skal vi se på mikrotilstanden for systemer av *flere partikler* og hvordan gjennomsnittene over disse gir opphav til størrelsene vi kan observere, som temperatur.
