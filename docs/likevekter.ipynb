## Kjemisk potensial 

Husk at to faser $\alpha$ og $\beta$ er i likevekt, når deres kjemiske potensialer er like, det vil si

$\mu_{\alpha} = \mu_{\beta}$

Det kjemiske potensialet til et stoff $i$ er endringen i Gibbs fri energi $G$ når vi legger til en liten mengde av stoff $i$, mens trykk $P$, temperatur $T$ og alle andre stoffmengder til stoffer $j$, $n_j$, holdes konstante. Når to kjemiske potensialer er like, vil det derfor bety at endringen i $G$ i forhold til hvor mye stoffmengde man tilfører lik, altså tjener ikke stoffet på å bytte fase. Med andre ord, vi kan ikke øke spontaniteten. 

Det kjemiske potensialet avhenger av hvor mye det er av det. Jo mer konsentrert stoffet er, jo høyere er potensialet, slik vi ser i formelen nedenfor.


:::{admonition}  Kjemisk potensial som funksjon av aktivitet
:class: note

Det kjemiske potensialet $\mu$ er gitt ved

$$\mu = \mu^\circ + RT \ln a$$

der $\mu^\circ$ er standardpotensialet, $R$ er gasskonstanten, $T$ er temperaturen og $a$ er aktiviteten.
:::

Vi skal la denne formelen ligge for nå, og kommer tilbake til dn og definisjonen av aktivitet senere. Før den tid trenger vi å introdusere **reaksjonens Gibbs-energi**

## Reaksjonens Gibbs-energi

Vi skal se på likevekt fra en annen side, og introdusere to nye størrelser - **ekstent $\xi$** og **reaksjonens Gibbs-energi** $\Delta_r G$.


### Ekstenten av en reaksjon

Ordet *extent* på englesk betyr *hvor mye* eller *i hvilken grad* noe strekker seg eller gjelder. På samme måte har vi en type ekstent for reaksjoner i kjemi. Se på den enkleste tenkelige reaksjonen

$A \rightleftharpoons B$

For å beskrive hvor langt den har kommet, innfører vi **ekstent av reaksjon** $\xi$ (den greske bokstaven «ksi»). Går reaksjonen et lite steg framover, får vi mer av stoffmengde $n_B$ ($dn_B$ øker), mens vi får mindre av stoffmengde $n_A$ ($dn_A$ avtar). 

I et isolert system er totalmengden $n$ er lik

$n = n_A + n_B \Rightarrow dn = dn_A + dn_B$

Systemet er isolert, og da er $dn = 0$, og vi får

$dn_B = -dn_A$

Når $n_B$ øker bittelitt med størrelse $dn_B$, avtar $n_A$ i samme størrelse, men i motsatt retning, det vil si $-dn_B$. Ekstenten er definert som denne endringen i stoffmengde, altså

$dn_A = -d\xi \qquad dn_B = +d\xi$

Ekstenten har enhet $[\text{mol}]$, og måler rett og slett hvor mange $\text{mol}$ reaksjonen har skredet fram.

Ved konstant temperatur $T$ og trykk $P$, er Gibbs-energi kun en funksjon av stoffmengdene $n_A$ og $n_B$, altså

$G = G(n_A, n_B)$

Husk fra Nyttige matematiske verktøy at for en vilkårlig funksjon $f(x,y)$, har vi

$df = \left( \frac{\partial f}{\partial x} \right)_y dx + \left( \frac{\partial f}{\partial y} \right)_x dy$

Vi får dermed

$dG = \left( \frac{\partial G}{\partial n_A} \right)_{T,P,n_B} dn_A + \left( \frac{\partial G}{\partial n_B} \right)_{T,P,n_A} dn_B$

Du kjenner kanskje igjen de partiellderiverte? Det er det kjemiske potensialet! Vi får dermed

$dG = \mu_A \ dn_A + \mu_B \ dn_B$

$dG = \mu_A \ (-d\xi) + \mu_B \ (d\xi) = (\mu_B -\mu_A) d\xi$

$\left( \frac{dG}{d\xi} \right)_{T,P} = \mu_B - \mu_A$

Hva er det vi har funnet, og hva sier det oss?

Vi vet at vet likevekt er $\mu_A = \mu_B$, så da er i så fall $$\left( \frac{\partial G}{\partial \xi} \right)_{T,P} = 0$ ved likevekt. Husk at vi skal minimere Gibbs energi, og ved likevekt er vi da på et punkt at hvis vi går et lite steg fram i reaksjonen, så øker Gibbs energien. Ha dette med videre i tankene ved neste definisjon.



### Reaksjonens Gibbs-energi

Størrelsen vi nettopp utledet ved bruk av ekstent har faktisk et navn - **reaksjonens Gibbs-energi**. Vi utledet den for reaksjonen $A \rightleftharpoons$ B, men den kan generaliseres til andre reaksjoner slik den er skrevet i definisjonen nedenfor for andre reaksjoner.


:::{admonition} Definisjon - Reaksjonens Gibbs-energi
:class: important

**Reaksjonens Gibbs-energi** er endringen i Gibbs-energi per enhet ekstent $\xi$ , ved konstant trykk $P$ og temperatur $T$:

$\Delta_r G = \left(\frac{\partial G}{\partial \xi}\right)_{T,P}$

For reaksjonen $A \rightleftharpoons B$ er den lik differansen i kjemisk potensial:

$\Delta_r G = \mu_B - \mu_A$

Vi kan også skrive

$\Delta_r G = \sum_J \nu_J \mu_J$

der $J$ benevner alle reaktantene og produktene, og $\nu_J$ er tilhørende støkiometriske tall.
:::

Reaksjonens Gibbs-energi er altså *stigningstallet* til $G$ når vi plotter den mot $\xi$. Det gir en direkte tolkning av hvilken vei reaksjonen går:

* Er $\Delta_r G < 0$ (kurven heller nedover), går reaksjonen spontant **forover**. Dette svarer til $\mu_A > \mu_B$ - stoff renner fra høyt til lavt kjemisk potensial, akkurat som før.
* Er $\Delta_r G > 0$, går reaksjonen spontant **bakover**.
* Er $\Delta_r G = 0$, er kurven flat: vi står i **likevekt**, og det er ingen netto drivkraft i noen retning.


:::{admonition} Definisjon - Likevektsbetingelsen
:class: important

Et system er i kjemisk likevekt når Gibbs-energien er i sitt minimum langs reaksjonsforløpet:

$\Delta_r G = \left(\frac{\partial G}{\partial \xi}\right)_{T,P} = 0$

For $A \rightleftharpoons B$ er dette ekvivalent med 

$\mu_A = \mu_B$
:::


Vi kan også definere eksergone og endergone reaksjoner ved bruk av reaksjonens Gibbs-energi, nemlig:

* $\Delta_r G < 0$: reaksjonen er **eksergon** og går spontant forover.
* $\Delta_r G > 0$: reaksjonen er **endergon** og går ikke forover av seg selv.

## Reaksjonskvotienten og aktiviteter

Vårt neste mål er å knytte aktivitet og støkiometriske tall til alle komponenter i en reaksjon, i ett og samme tall for å kunne si noe overordnet om hele reaksjonen.



### Aktivitet (effektiv konsentrasjon)

Tidligere introduserte vi en ny formel for $\mu$ der vi for første gang introduserte **aktiviteten** $a$ - men hva er det? **Aktiviteten** er et dimensjonsløst mål på *effektiv konsentrasjon*. 

Se for deg en løsning med mange ioner. Idealbildet er at hver partikkel er helt fri og uavhengig - da teller den fullt med. Men i virkeligheten trekker og dytter ionene på hverandre: et positivt ion omgir seg med negative naboer som "henger fast" i det og holder litt igjen. Da er ikke ionet fritt til å reagere så villig som antallet skulle tilsi. Resultatet er at stoffet deltar som om det var færre av dem enn det egentlig er. Den virkelige konsentrasjonen kan være $0.1 \text{ mol}/\text{L}$, men på grunn av all dyttingen "kjennes" det for reaksjonen ut som kanskje $0.08 \text{ mol}/\text{L}$. Det er da $0.08 \text{ mol}/\text{L}$ som er den effektive konsentrasjonen.

For en ideell gass er $a = p/p^\circ$, og for et løst stoff er $a \approx C/C^\circ$. 



### Støkiometriske tall

For en generell reaksjon skriver vi støkiometrien med **støkiometriske tall** $\nu_J$, som er positive for produkter og negative for reagenser. For $2A + B \rightleftharpoons 3C + D$ er for eksempel $\nu_A = -2$, $\nu_B = -1$, $\nu_C = +3$ og $\nu_D = +1$.



### Reaksjonskvotienten

I en reaksjon har vi flere aktiviteter og støkiometriske tall, og det er derfor hensiksmessig å samle dem i én og samme størrelse. Dette heter **reaksjonskvotienten** $Q$.


:::{admonition} Definisjon - Reaksjonskvotient
:class: important

**Reaksjonskvotienten** $Q$ er produktet av alle stoffenes aktiviteter, hver opphøyd i sitt støkiometriske tall:

$Q = \prod_J a_J^{\nu_J}$

Siden produktenes $\nu_J$ er positive og reagensenes er negative, havner produktene i telleren og reagensene i nevneren. For $aA + bB \rightleftharpoons cC + dD$ blir dette

$Q = \dfrac{a_C^{\,c}\,a_D^{\,d}}{a_A^{\,a}\,a_B^{\,b}}$
:::


## Sammenhengen mellom Gibbs-energi og sammensetning

Vi vil nå finne hvordan reaksjonens Gibbs-energi avhenger av sammensetningen $Q$. Nøkkelen er formelen for kjemisk potensial som en funksjon av aktiviteten, som vi introduserte tidligere.


:::{admonition} Utledning av reaksjonsisotermen
:class: note, dropdown

Reaksjonens Gibbs-energi er summen av de kjemiske potensialene vektet med de støkiometriske tallene:

$\Delta_r G = \sum_J \nu_J \mu_J$

Vi setter inn uttrykket $\mu_J = \mu_J^\circ + RT\ln a_J$ for hvert stoff:

$\Delta_r G = \sum_J \nu_J \mu_J^\circ + RT\sum_J \nu_J \ln a_J$

Den første summen er nettopp standard reaksjons-Gibbs-energi, $\Delta_r G^\circ = \sum_J \nu_J \mu_J^\circ$. I den andre summen bruker vi logaritmeregelen $\nu \ln a = \ln a^{\nu}$ og at en sum av logaritmer er logaritmen av produktet:

$\sum_J \nu_J \ln a_J = \sum_J \ln a_J^{\nu_J} = \ln \prod_J a_J^{\nu_J} = \ln Q$

Til sammen gir dette reaksjonsisotermen:

$\Delta_r G = \Delta_r G^\circ + RT\ln Q$
:::


:::{admonition} Reaksjonsisotermen
:class: note

$\Delta_r G = \Delta_r G^\circ + RT\ln Q$

Denne likningen forteller hvordan drivkraften $\Delta_r G$ avhenger av både reaksjonens iboende tendens ($\Delta_r G^\circ$) og den aktuelle sammensetningen ($Q$).
:::


:::{admonition} Definisjon - Standard reaksjons-Gibbs-energi
:class: important

**Standard reaksjons-Gibbs-energi** $\Delta_r G^\circ$ er reaksjonens Gibbs-energi når alle stoffer er i standardtilstand ($a_J = 1$). Den regnes ut fra tabulerte standard dannelses-Gibbs-energier med Hess' lov:

$\Delta_r G^\circ = \sum_J \nu_J \Delta_f G^\circ(J) = \sum \Delta_f G^\circ(\text{produkter}) - \sum \Delta_f G^\circ(\text{reagenser})$
:::

## Likevektskonstanten

Ved likevekt er $\Delta_r G = 0$, og reaksjonskvotienten har en bestemt verdi som vi kaller **likevektskonstanten** $K$. Setter vi $\Delta_r G = 0$ og $Q = K$ inn i reaksjonsisotermen,

$0 = \Delta_r G^\circ + RT\ln K$

og løser for $\Delta_r G^\circ$, får vi en av de viktigste likningene i kjemisk termodynamikk.


:::{admonition} Likevektskonstanten og Gibbs-energien
:class: note

$\Delta_r G^\circ = -RT\ln K$

der likevektskonstanten er reaksjonskvotienten evaluert ved likevekt:

$K = \left(\prod_J a_J^{\nu_J}\right)_{\text{likevekt}}$

Denne relasjonen kobler tabulerte termodynamiske data direkte til den målbare likevektssammensetningen.
:::


Sammenhengen er logaritmisk, og det forklarer hvor følsom likevekten er for $\Delta_r G^\circ$:

* Stor negativ $\Delta_r G^\circ$ gir stor $K$ - likevekten ligger langt mot produkt.
* Stor positiv $\Delta_r G^\circ$ gir liten $K$ - likevekten ligger mot reagens.
* $\Delta_r G^\circ \approx 0$ gir $K \approx 1$.

Fordi $\ln K$ inngår, gir selv en beskjeden endring i $\Delta_r G^\circ$ en dramatisk endring i $K$: en forskjell på noen få $\text{kJ/mol}$ kan flytte $K$ flere tierpotenser.


:::{card} 
**Eksempel**
^^^
For ammoniakksyntesen $N_2(g) + 3H_2(g) \rightleftharpoons 2NH_3(g)$ er $\Delta_r G^\circ = -32{.}9 \text{ kJ/mol}$ ved $298 \text{ K}$. Finn $K$.

Vi løser $\Delta_r G^\circ = -RT\ln K$ for $\ln K$:

$\ln K = -\frac{\Delta_r G^\circ}{RT} = -\frac{-32900 \text{ J/mol}}{(8{.}314 \text{ J/mol}\cdot\text{K})(298 \text{ K})} = 13{.}3$

Deretter opphøyer vi:

$K = e^{13{.}3} \approx 5{.}8 \times 10^5$

Den store verdien viser at likevekten ligger langt mot produktsiden - ammoniakk dannes i stor grad ved denne temperaturen.
:::


## Likevektskonstanten uttrykt ved entalpi og entropi

Siden $\Delta_r G^\circ = \Delta_r H^\circ - T\Delta_r S^\circ$, kan vi sette dette inn i $\Delta_r G^\circ = -RT\ln K$ og løse for $K$:

$K = e^{-\Delta_r H^\circ/RT}\,e^{\Delta_r S^\circ/R}$

Uttrykket viser tydelig de to bidragene. En **eksoterm** reaksjon ($\Delta_r H^\circ < 0$) gir et stort entalpiledd og trekker mot stor $K$. En **positiv entropiendring** trekker i samme retning gjennom det andre leddet. Er de uenige, avgjør temperaturen hvem som dominerer - den samme balansen mellom entalpi og entropi du så i termokjemien, nå uttrykt gjennom likevektskonstanten.

## Likevektens respons på ytre endringer

:::{admonition} Le Chateliers prinsipp
:class: important

Når et system i likevekt utsettes for en forstyrrelse, forskyver likevekten seg slik at forstyrrelsens virkning motvirkes.
:::


**Trykk.** Effekten av trykk avhenger av *hvordan* det påføres. Tilsetter vi en inert gass ved konstant volum, endres ikke partialtrykkene til de reagerende gassene - så lenge gassene er ideelle - og likevekten forskyves ikke.

Komprimerer vi derimot systemet, forskyves likevekten mot siden med færrest gassmolekyler, for da reduseres partikkeltallet og trykkøkningen dempes. For ammoniakksyntesen, der antallet gassmolekyler går fra fire til to, favoriserer økt trykk produktet. Det avgjørende poenget er at $K$ *ikke* endres av trykk - det er bare sammensetningen som flytter seg innenfor den samme $K$.

**Temperatur.** Temperatur er den eneste av disse som faktisk endrer $K$. Hvordan, uttrykkes av **van 't Hoffs likning**.


:::{admonition} Utledning av van 't Hoffs likning
:class: note, dropdown

Vi starter fra $\Delta_r G^\circ = -RT\ln K$, skrevet som

$\ln K = -\frac{\Delta_r G^\circ}{RT}$

Vi deriverer med hensyn på temperatur. For å håndtere $\Delta_r G^\circ / T$ bruker vi Gibbs–Helmholtz-relasjonen fra kapittelet om termodynamiske potensialer, $\frac{d(G/T)}{dT} = -\frac{H}{T^2}$, anvendt på standard reaksjonsstørrelsene:

$\frac{d(\Delta_r G^\circ / T)}{dT} = -\frac{\Delta_r H^\circ}{T^2}$

Siden $\ln K = -\frac{1}{R}\cdot\frac{\Delta_r G^\circ}{T}$, gir derivasjon av begge sider

$\frac{d\ln K}{dT} = -\frac{1}{R}\cdot\frac{d(\Delta_r G^\circ/T)}{dT} = \frac{\Delta_r H^\circ}{RT^2}$

Dette er van 't Hoffs likning. Antar vi at $\Delta_r H^\circ$ er konstant over temperaturintervallet, kan vi integrere fra $(T_1, K_1)$ til $(T_2, K_2)$ og få den praktiske toparts-formen.
:::

:::{admonition} Van 't Hoffs likning
:class: note

$\frac{d\ln K}{dT} = \frac{\Delta_r H^\circ}{RT^2}$

og på integrert form, med $\Delta_r H^\circ$ antatt konstant:

$\ln\frac{K_2}{K_1} = -\frac{\Delta_r H^\circ}{R}\left(\frac{1}{T_2} - \frac{1}{T_1}\right)$
:::

For en eksoterm reaksjon ($\Delta_r H^\circ < 0$) er $\frac{d\ln K}{dT} < 0$: $K$ synker når temperaturen øker, og likevekten forskyves bort fra produkt. For en endoterm reaksjon er det motsatt. Dette er nøyaktig hva Le Chateliers prinsipp forutsier hvis vi tenker på varme som henholdsvis et produkt eller et reagens.


:::{admonition} På laboratoriet
:class: tip

Van 't Hoffs likning er også et måleverktøy. Omformer vi den integrerte formen, ser vi at $\ln K$ er en lineær funksjon av $\frac{1}{T}$, med stigningstall $-\frac{\Delta_r H^\circ}{R}$.

Måler du derfor likevektskonstanten ved flere temperaturer og plotter $\ln K$ mot $\frac{1}{T}$, kan du lese av standard reaksjonsentalpi rett fra stigningstallet. Dette er en *ikke-kalorimetrisk* metode for å bestemme $\Delta_r H^\circ$ - et fint eksempel på hvordan likevektsmålinger gir termodynamiske data.
:::
