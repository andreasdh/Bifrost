```python
from rdkit import Chem
from rdkit.Chem import Draw
from IPython.display import display
from rdkit.Chem.Draw import IPythonConsole as ipc

# Define the SMILES string (e.g., caffeine)
smiles_string = "CN1C=NC2=C1C(=O)N(C(=O)N2C)C"

# Options
ipc.ipython_useSVG = True  # Use SVG for higher quality images
ipc.drawOptions.addAtomIndices = True  # Show atom indices
ipc.molSize = 250,250 # Set size of image

# Create a molecule object from the SMILES
molecule = Chem.MolFromSmiles(smiles_string)
molecule

# Optional: Save the image to a file
#display(Draw.MolToImage(molecule))
# Draw.MolToImageFile(molecule, 'caffeine.png')
```

```python
print(molecule.GetNumAtoms())
print(molecule.GetNumAtoms(onlyExplicit = False))
```

```python
molbonds =  molecule.GetBonds()

for bond in molbonds:
    start_atom = bond.GetBeginAtom()
    start_atom_i = bond.GetBeginAtomIdx()
    end_atom = bond.GetEndAtom()
    end_atom_i = bond.GetEndAtomIdx()
    bond_type = bond.GetBondType()
    print(f"Bond between {start_atom.GetSymbol()}{start_atom_i} and {end_atom.GetSymbol()}{end_atom_i} is a {bond_type} bond.")
```
