#### Oppgave 1

a) At energien til et atom er kvantisert betyr at atomet bare kan ha bestemte energiverdier - ikke alle mulige verdier imellom. Akkurat som en stige har trinn på faste høyder (trinn 0,1,2,...) og du ikke kan stå på trinn 1.5 eller sveve mellom trinnene, kan atomet bare "stå" på bestemte energinivåer ($\varepsilon_0, \varepsilon_1, \varepsilon_2, \dots$). 


b) Når et atom sender ut lys i form av et foton, hopper et elektron ned fra et høyere til et lavere energinivå, og lyset som sendes ut har en energi som er nøyaktig lik forskjellen mellom de to nivåene. Siden det bare finnes noen få bestemte, tillatte nivåer, finnes det bare noen få mulige energiforskjeller – og dermed bare noen få bestemte bølgelengder (farger) lyset kan ha. Derfor ser vi noen få skarpe linjer i stedet for et jevnt fargebånd. Et kontinuerlig fargebånd ville betydd at alle energiforskjeller var mulige, men det er de ikke.

c) En makrotilstand beskriver systemet med størrelser vi kan måle i laboratoriet: temperatur, trykk, volum, total energi. En mikrotilstand er én bestemt måte energien er fordelt på molekylnivå, konsistent med den observerte makrotilstanden.

En analogi er "en skog". På et makronivå kan flere ulike områder være skoger - de er samling av trær, busker, dyr osv. På et mikronivå er skogene forskjellige: trærne er posisjonert forskjellig, det er forskjellige antall trær osv.


d) En frihetsgrad er en uavhengig måte et molekyl kan bevege seg eller lagre energi på. Hver frihetsgrad er et eget "energilager" som bidrar til molekylets totale energi. Grunnen til at et molekyl har flere frihetsgrader enn et enkeltatom, er at et molekyl har bindinger som kan vibrere og en struktur som kan rotere – ting et enkeltatom ikke har.

Den totale energien til et molekyl er summen av bidragene fra hver frihetsgrad:

$$\varepsilon_{tot} = \varepsilon_{elec} + \varepsilon_{trans} + \varepsilon_{rot} + \varepsilon_{vib}$$

- Elektronisk ($\varepsilon_{elec}$): energien knyttet til hvilke tilstander elektronene rundt atomkjernene er i (tenk orbitaler).
- Translasjon ($\varepsilon_{trans}$): energien fra at hele molekylet beveger seg gjennom rommet.
- Rotasjon ($\varepsilon_{rot}$): energien fra at molekylet roterer om en akse.
- Vibrasjon ($\varepsilon_{vib}$): energien fra at atomene svinger fram og tilbake rundt bindingene sine.

Et enkeltatom har bare de to første (elektronisk og translasjon), siden det verken kan rotere eller vibrere.

En mikrotilstand er én bestemt, fullstendig beskrivelse av hvordan energien er fordelt på alle frihetsgradene samtidig – altså nøyaktig hvilken elektronisk, translasjons-, rotasjons- og vibrasjonstilstand molekylet er i.

En makrotilstand er derimot bare det vi kan måle utad, for eksempel den totale energien. Poenget er at mange forskjellige mikrotilstander kan gi samme makrotilstand: energien kan fordeles på frihetsgradene på mange ulike måter og likevel gi samme totale energi $\varepsilon_{tot}$.


e) Statistisk mekanikk er måten vi forklarer et systems målbare, makroskopiske egenskaper (som temperatur og trykk) ut fra hva de enormt mange små partiklene i systemet gjør, ved hjelp av sannsynlighet og gjennomsnitt over alle mikrotilstandene.

Forskjellen på de to lovtypene:

* En empirisk lov er satt opp for å passe til noe vi har observert. Vi måler noe først, og finner så en formel som beskriver målingene. Den forteller hva som skjer, men ikke hvorfor.
* En statistisk lov er utledet fra grunnleggende antakelser om hvordan partiklene oppfører seg. Den følger av teorien og forklarer hvorfor den makroskopiske oppførselen blir som den blir.

Kort sagt: den empiriske loven beskriver, den statistiske loven forklarer.

#### Oppgave 7

a) Vi starter med

$\Omega(q,N) = \frac{(q+N-1)!}{q!\,(N-1)!}$

Vi tar så logaritmen av uttrykket og bruker at $\ln(\frac{a}{b}) = \ln a - \ln b$:

$\ln\Omega = \ln(q+N-1)! - \ln q! - \ln(N-1)!$

For store $q$ og $N$ bruker vi $q + N - 1 \approx q + N$ og $N - 1 \approx N$:

$\ln\Omega \approx \ln(q+N)! - \ln q! - \ln N!$

Deretter bruker vi Stirlings approksimasjon $\ln n! \approx n\ln n - n$ på hvert ledd, slik at

$\ln(q+N)! \approx (q+N)\ln(q+N) - (q+N)$

$\ln q! \approx q\ln q - q$

$\ln N! \approx N\ln N - N$

Vi setter inn hvert av leddene og får

$\ln\Omega \approx \big[(q+N)\ln(q+N) - (q+N)\big] - \big[q\ln q - q\big] - \big[N\ln N - N\big] = (q+N)\ln(q+N) - q\ln q - N\ln N$



bVi har to systemer i kontakt med total energi 

$q_\text{tot} = q_A + q_B \Longrightarrow q_B = q_\text{tot} - q_A$

Den totale multiplisiteten er 

$\Omega_\text{tot} = \Omega_A \cdot \Omega_B$

slik at vi får

$\ln\Omega_\text{tot} = \ln\Omega_A(q_A) + \ln\Omega_B(q_B)$

Likevektsbetingelsen er der $\ln\Omega_\text{tot}$ er maksimal, altså der den deriverte med hensyn på $q_A$ er null

$\frac{d}{dq_A}\ln\Omega_{tot} = 0$

Vi deriverer først ved bruk av kjerneregelen

$\frac{d}{dq_A}\ln\Omega_{tot} = \frac{d}{dq_A} \left( \ln\Omega_A(q_A) + \ln\Omega_B(q_B) \right) = \frac{d}{dq_A} \ln\Omega_A(q_A) + \frac{d}{dq_A} \ln\Omega_B(q_B) = \frac{d\ln\Omega_A(q_A)}{dq_A} + \frac{d\ln\Omega_B(q_B)}{dq_B}\cdot\frac{dq_B}{dq_A}$

og krever at uttryket skal være lik null

$\frac{d\ln\Omega_A(q_A)}{dq_A} + \frac{d\ln\Omega_B(q_B)}{dq_B}\cdot\frac{dq_B}{dq_A} = 0$

Siden $q_B = q_\text{tot} - q_A$ er $dq_B/dq_A = -1$, så

$\frac{d\ln\Omega_A(q_A)}{dq_A} + \frac{d\ln\Omega_B(q_B)}{dq_B}\cdot(-1) = 0 \Longrightarrow \frac{d\ln\Omega_A(q_A)}{dq_A} = \frac{d\ln\Omega_B(q_B)}{dq_B}$

Nå kan vi bruke approksimasjonen for $\ln\Omega$ vi fant i a), og deriverer uttrykkene før vi setter dem inn:

$\frac{d\ln\Omega(q)}{dq} = \frac{d}{dq} \big[(q+N)\ln(q+N) - q\ln q - N\ln N \big] = \frac{d}{dq} \big[(q+N)\big] \ln(q+N) - \frac{d}{dq} \big[ q\ln q \big] - \frac{d}{dq} \big[ N \ln N \big]$

Vi kan derivere hvert ledd for seg

$\frac{d}{dq} \big[ (q+N) \ln(q+N) \big] = \frac{d}{dq} \big[q+N\big] \cdot \ln(q+N) + \frac{d}{dq} \big[ \ln(q+N) \big] \cdot (q+N) = \ln(q+N) + \frac{q+N}{q+N}= \ln (q+N) + 1$

$\frac{d}{dq} \big[ q\ln q \big] = \frac{d}{dq} \big[q \big] \cdot (\ln q) + \frac{d}{dq} \big[\ln q] \cdot q = \ln q + \frac{q}{q} = \ln q + 1$

$\frac{d}{dq} \big[ N \ln N \big] = 0$

og setter dem inn igjen

$\frac{d\ln\Omega(q)}{dq} = \frac{d}{dq} \big[(q+N)\ln(q+N) - q\ln q - N\ln N \big] = \ln (q+N) + 1 - (\ln q + 1) = \ln (q+N) - \ln q = \ln \left( \frac{q+N}{q} \right)$

Vi setter inn den deriverte for A og B:

$\ln\!\left(\frac{q_A+N_A}{q_A}\right) = \ln\!\left(\frac{q_B+N_B}{q_B}\right) \Longrightarrow \frac{q_A+N_A}{q_A} = \frac{q_B+N_B}{q_B} \Longrightarrow \frac{q_A+N_A}{q_A}-1 = \frac{q_B+N_B}{q_B}-1 \Longrightarrow \frac{q_A+N_A-q_A}{q_A} = \frac{q_B+N_B-q_B}{q_B} \Longrightarrow \frac{N_A}{q_A} = \frac{N_B}{q_B} \Longrightarrow \frac{q_A}{N_A} = \frac{q_B}{N_B}$

Med definisjonen $T = q/N$ får vi ved likevekt:

$T_A = T_B$

Venstre: $\Omega = \dfrac{5!}{5!\,0!\,0!\,0!} = 1$. Det finnes bare én måte å ha alle i grunntilstanden.

Høyre: $\Omega = \dfrac{5!}{2!\,2!\,1!\,0!} = \dfrac{120}{2 \cdot 2 \cdot 1} = 30$.


d) Entropien er $S = k_B \ln \Omega$.

Venstre: $S = k_B \ln 1 = 0$.

Høyre: $S = k_B \ln 30 \approx 3{,}40\,k_B \approx 4{,}70 \times 10^{-23} \text{ J}/\text{K}$ .

Venstretilstanden har lavest mulig energi ($U=0$), men også lavest mulig entropi ($S=0$), fordi det bare finnes én eneste måte å realisere den på - alle partiklene må ligge i grunntilstanden. Når energien øker, blir det flere måter å fordele den på mellom partiklene, og dermed flere mikrotilstander og høyere entropi.

Dette er kjernen i termodynamikken: et system trekkes ikke bare mot lav energi, men mot mange mikrotilstander. Grunntilstanden er energimessig gunstig, men entropisk fattig, og det er balansen mellom disse to som avgjør hvilken makrotilstand systemet faktisk havner i ved en gitt temperatur - akkurat den avveiningen Boltzmann-fordelingen beskriver.

1. I tilstandsfunksjoner kan veien mellom start- og sluttilstand velges fritt. Det å dele prosessen opp i delprosesser svarer til å endre veien men så lenge start- og sluttilstanden er de samme vil endringen i tilstandsfunksjonen _totalt_ være den samme. Hess lov gjelder for alle tilstandsfunksjoner, men uttrykkes for eksempel slik for entalpi: 

$$\Delta H = \sum \Delta H(produkter) - \sum \Delta^{\circ} _f H(reaktanter)$$

Dersom vi har delprosessene A, B, C, vil vi heller bruke loven slik: 

$$\Delta H = \Delta H(A) + \Delta H(B) + \Delta H(C)$$

Innad i disse delprosessene har allerede forskjellen i entalpi mellom start- og sluttilstand blitt beregnet eller målt i henhold til formelen over. I neste oppgave vil du se et eksempel på dette.

Vi kan også gå motsatt vei: I faseoverganger fra væske til gass bytter vi ut _sterke interaksjoner_ som gjør at væsken holdes sammen med _svake interaksjoner_ som lar gassen flyte fritt i rommet. I dette tilfellet må systemet ta opp varme fra omgivelsene som det bruker til å "tette" energigapet mellom tilstandene. Slik får vi **endoterme prosesser** hvor vi må tilføre varme til systemet.

6. En ikke-spontan prosess har $\Delta G > 0$ under de gitte betingelsene. Vi kan likevel få den til å skje ved å endre situasjonen slik at netto $\Delta G$ blir negativ. De vanligste strategiene er:

$$q_{vann} + q_{is} + q{isvann} = 0$$

Siden vi er interesserte i smelteentalpien er det $q_{is}$ vi vil finne da dette er likt $\Delta H_{is}$ fordi vi er ved konstant trykk!

$$\Delta H_{is} = q_{is} = -q_{vann} - q_{isvann} = - \Delta H_{vann} - \Delta H_{isvann}$$

$$\Delta H_{is} = 7.23 \, kJ - 1.95 \, kJ = 5.28 \, kJ$$

Dette er smelteentalpien for $25.8 \, g$ is! Vi ønsker å finne den molare smelteentalpien: 

Antall mol is er gitt ved: 

$$n = \frac{m_{is}}{M_{H_2O}} = \frac{25.8 \, g}{18.01 \, g/mol} = 1.43 \, mol$$

Vi kan nå finne den molare smelteentalpien: 

$$\Delta _mH_{fus} = \frac{\Delta H_{is}}{n} = \frac {5.28 \, kJ}{1.43 \, mol} = 3.69 \, kJ/mol$$

c) Vi ser at den beregnede verdien vår er betydelig lavere enn den teoretiske! Dette tyder på at ikke all varme som avgis fra vannet i koppen brukes til å varme opp isen og isvannet slik vi har antatt. I realiteten har vi varmeutveksling også med omgivelsene: Varmen kan lekke ut til bordet, luften, termometeret eller materiale i koppen. Mindre varme er tilgjengelig til å smelte isen enn det vi har antatt og derfor er den faktiske verdien for smelteentalpien høyere. Dersom isen kom rett fra fryseren kan det også hende den egentlig var kaldere enn $0^{\circ} C$, og da har ikke vi tatt hensyn til dette i beregningene våre. I tillegg har vi vanlige usikkerheter som lese- og kalibreringsfeil i termometeret, veieusikkerhet osv. 

2. Start med å finne dannelsesentalpiene du trenger fra læreboka eller et annet oppslagsverk. Her må vi bruke Hess lov:

Produktet her er de to ionene i løsning $Ag^+_{(aq)}$ og $Br^-_{(aq)}$, og reaktanten er sølvbromid-saltet $AgBr_{(s)}$:

**Drivkrefter og Gibbs fri energi**

1. 

a) Vi bruker Hess lov til å regne ut tilstandsfunksjonene: 

$$\Delta H = \sum \Delta H(produkter) - \sum \Delta H(reaktanter)$$

$$\Delta H_m^{\circ} = \Delta H_m^{\circ}(K^+_{(aq)}) + \Delta H_m^{\circ}(Cl^-_{(aq)}) - \Delta H_m^{\circ}(KCl_{(s)})$$

$$Delta H_m^{\circ} = (−252.38 + (–167.16) - (−436.75)) \, kJ/mol = 17.21 \, kJ/mol$$

$$\Delta S = \sum S(produkter) - \sum S(reaktanter)$$

$$\Delta S_m^{\circ} = S_m^{\circ}(K^+_{(aq)}) + S_m^{\circ}(Cl^-_{(aq)}) - S_m^{\circ}(KCl_{(s)})$$

$$\Delta S_m^{\circ} = (102.5 + 56.5 - 82.6) \, J.K^{−1}.mol^{−1} = 76.4 \, J.K^{−1}.mol^{−1}$$

Vi kan nå regne ut Gibbs fri energi: 

$$\Delta G = \Delta H - T\Delta S$$

$$\Delta G_m^{\circ} = Delta H_m^{\circ} - T\Delta S_m^{\circ}$$

$$\Delta G_m^{\circ} = 17.21 \, kJ/mol - 298.15 \, K \times (76.4 \times 10^{-3}) \, J.K^{−1}.mol^{−1} = -5.57 \, kJ/mol$$

b) Vi ser at reaksjonen er endoterm fra det positive fortegnet til entalpiendringen. Vi har derimot en positiv entropiendring! Det er denne økningen i entropi når saltet løses i vann som driver reaksjonen fremover. 

c) Dersom vi øker temperaturen vil $-T\Delta S$ leddet blitt mer negativt, altså vil reaksjonen bli _mer_ spontan (\Delta G blir mer negativ). 

2. Vi bruker igjen Hess lov til å regne ut tilstandsfunksjonene: 

$$\Delta H = \sum \Delta H(produkter) - \sum \Delta H(reaktanter)$$

$$\Delta _r H_m^{\circ} = \Delta H_m^{\circ}(CH_3COOH_{(l)}) - \Delta H_m^{\circ}(CO_{(g)}) - \Delta H_m^{\circ}(CH_3OH{(l)})$$

$$\Delta _r H_m^{\circ} = (-484.5 - (-238.66) - (-110.53)) \, kJ/mol = -135.3 \, kJ/mol$$

$$\Delta S = \sum S(produkter) - \sum S(reaktanter)$$

$$\Delta _r S_m^{\circ} = S_m^{\circ}(CH_3COOH{(l)}) - S_m^{\circ}(CO_{(g)}) - S_m^{\circ}(CH_3OH_{(l)})$$

$$\Delta _r S_m^{\circ} = (159.8 - 126.8 - 197.67) \, J/Kmol = -164.7 \, J/Kmol$$

Nå har vi det vi trenger for å regne ut Gibbs fri energi: 

$$\Delta _rG_m^{\circ} = \Delta _r H_m^{\circ} - T\Delta _r S_m^{\circ}$$

$$\Delta _rG_m^{\circ} = -135.3 \, kJ/mol - 298 \, K \times (-164.7 \times 10^{-3}) \, J/Kmol = -86.2 \, kJ/mol$$

3. Vi starter med å finne den differensiale formen av Gibbs fri energi: 

$$G = H - TS = U + pV - TS$$

La oss finne noen differensielle former:

Fra termodynamikkens første lov har vi: 

$$dU = dq + dw = dq - pdV$$

Fra relasjonen mellom entropi og varme har vi:

$$dS = \frac{dq_{rev}}{T} \leftrightarrow dq_{rev} = TdS$$

Dette gir $dU$:

$$dU = TdS - pdV$$

La oss nå finne differensialet av den frie energien: 

$$dG = dU + d(pV) - d(TS)$$

$$dG = TdS -pdV + pdV + Vdp - TdS - SdT$$

Vi forkorter og får:

$$dG = Vdp - SdT$$

Den molare formen vil se slik ut: 

$$dG_m = V_mdp - S_mdT$$

I temperaturintervallet oppgaven er interessert i er benzen en væske, så vi antar at trykket er konstant og får funksjonen: 

$$dG_m = - S_mdT$$

Nå har vi alt vi trenger for å plotte! 


````python
import numpy as np
import matplotlib.pyplot as plt 

def dG(T_slutt, T_start = 298): 
    Sm = 173.3 #J/molK
    return -Sm * (T_slutt - T_start)

T_start = 298
T_slutt = np.linspace (T_start, 318, 20)

plt.plot(T_slutt, dG(T_slutt), color = "hotpink")
plt.xlabel ("Sluttemperatur (K)")
plt.ylabel ("Endring i Gibbs fri energi (J/Kmol)")
plt.show
````

## Løsningsforslag - Intermolekylære interaksjoner

### Diskusjonsoppgave

#### Oppgave 1 — Lennard-Jones for metan

$$u(r) = 4\varepsilon\left[\left(\frac{\sigma}{r}\right)^{12} - \left(\frac{\sigma}{r}\right)^{6}\right], \qquad \sigma = 3{,}93\ \text{Å}, \quad \varepsilon/k_B = 148\ \text{K}$$

**a)** 
Kurven starter nær null når avstanden $(r)$ er stor. Da påvirker molekylene hverandre svært lite. Når molekylene kommer nærmere hverandre, synker energien og danner en «brønn». Den laveste energien finner vi ved $r \to 0$, som er den mest stabile avstanden mellom molekylene.

Videre øker energien igjen og blir lik null ved $r = \sigma$. På dette punktet er de attraktive og frastøtende kreftene akkurat like sterke og opphever hverandre. For avstander større enn $\sigma$ dominerer tiltrekningen, slik at den potensielle energien er negativ. For avstander mindre enn $\sigma$ blir frastøtningen svært sterk, og energien øker raskt til store positive verdier når molekylene presses enda nærmere hverandre.

Verdien $\sigma$ kan derfor tolkes som en effektiv «hard» molekylstørrelse, en grense der molekylene begynner å frastøte hverandre kraftig.


**b)** Likevektsavstanden ligger *utenfor* $\sigma$ fordi $\sigma$ bare er der den potensielle energien er lik null, ikke der *kreftene* balanserer. Det frastøtende $r^{-12}$-leddet avtar raskere enn det tiltrekende $r^{-6}$-leddet. Derfor vil tiltrekningen fortsatt dominere når vi beveger oss litt utover $\sigma$, og den potensielle energien fortsetter å synke.
Først ved likevektsavstanden $r_0$ er de tiltrekkende og frastøtende kreftene akkurat like store. Da er energien på sitt laveste, og systemet er mest stabilt.


**c)** Sett $r^* = 2^{1/6}\sigma$ inn: $(\sigma/r^*)^6 = (2^{-1/6})^6 = \tfrac12$, og $(\sigma/r^*)^{12} = \tfrac14$. Dermed

$$u(r^*) = 4\varepsilon\left[\tfrac14 - \tfrac12\right] = -\varepsilon.$$

Brønndybden er altså nettopp $\varepsilon$, tilsvarende 148 K, eller $\varepsilon = k_B \cdot 148 \approx 2{,}0\times10^{-21}\ \text{J}$. Til sammenligning er en kovalent C–H-binding rundt $7\times10^{-19}\ \text{J}$ — flere hundre ganger sterkere. Bindingen mellom to metanmolekyler er derfor *svak*: den rives lett fra hverandre av termisk bevegelse, og metan koker allerede ved 112 K.

**d)** Kraften er $f(r) = -\dfrac{du}{dr}$. Den er null i bunnen av brønnen ($r = r^*$) — et *stabilt* likevektspunkt. For $r < r^*$ er kraften frastøtende (peker utover), for $r > r^*$ er den tiltrekkende (peker innover). Enhver forskyvning fra $r^*$ gir en gjenopprettende kraft, derav stabiliteten.

![3e](bilder/LJ_3e.svg)

**e)** Å dele $\varepsilon$ på $k_B$ gir en temperaturskala:
* Når $k_B T \gg \varepsilon$ er den termiske energien langt større enn brønndybden, bevegelsen river molekylene fra hverandre, og stoffet er gass. 
* Når $k_B T \approx \varepsilon$ er de to sammenlignbare, og molekylene begynner å bli fanget i brønnen: kondensasjon setter inn. $\varepsilon/k_B$ er derfor grovt sett temperaturen der kondensasjon skjer, og metans kokepunkt (112 K) ligger nettopp i nærheten av 148 K.

#### Oppgave 2 — Ekte molekyler sammenlignet

| Molekyl | $\varepsilon/k_B$ (Kelvin) | $\sigma$ (Å) |
|---|---|---|
| Helium (He) | 10,22 | 2,58 |
| Etan (C₂H₆) | 205 | 4,23 |
| Benzen (C₆H₆) | 440 | 5,27 |

**a)** $\sigma$-kolonnen er størrelse, $\varepsilon$-kolonnen er bindingsstyrke. Benzen er både størst ($\sigma = 5{,}27$ Å) og binder sterkest ($\varepsilon/k_B = 440$ K); helium er minst og svakest.

**b)** Ingen av dem har permanent dipol, så Keesom-krefter (dipol–dipol) og Debye-krefter (dipol–indusert dipol) er utelukket. Tiltrekningen kommer fra **London-dispersjonskrefter**: øyeblikkelige fluktuasjoner i elektronskyen skaper flyktige dipoler, som induserer korrelerte dipoler i nabomolekylet.

**c)** Størrelsen på $\varepsilon$ øker med antall elektroner og hvor lett elektronskyen kan forskyves, altså **polariserbarheten**. Helium har bare to elektroner som er sterkt bundet til kjernen. Elektronskyen er derfor lite polariserbar, noe som gir svake tiltrekningskrefter og en potensialbrønn med liten dybde. Benzen har derimot mange elektroner, inkludert et lett polariserbart $\pi$-elektronsystem. Dette gjør at det lett dannes store øyeblikksdipoler, som gir sterkere tiltrekning mellom molekylene og dermed *en dypere potensialbrønn*.


**d)** Rangering etter $\varepsilon$: He < etan < benzen. Reelle kokepunkter: He $-269\,^\circ$C (4 K), etan $-89\,^\circ$C (184 K), benzen $+80\,^\circ$C (353 K). Rangeringen stemmer helt — dypere brønn gir høyere kokepunkt, og størrelsesordenen på $\varepsilon/k_B$ ligger nær de faktiske kokepunktene i kelvin.

**e)** Oppover alkanrekka øker både antall elektroner og kontaktflaten mellom molekylene. Det er den *samme* London-dispersjonen som virker hele veien, men den summerer seg opp langs kjeden: flere CH₂-ledd betyr flere bidrag, så $\varepsilon$ (og $\sigma$) vokser jevnt. Derfor er de korteste alkanene (metan–butan) gasser, mellomlange (ca. pentan–heksadekan) væsker, og de lengste voksaktige faste stoffer ved romtemperatur — stearinlys er nettopp lange alkaner holdt sammen av akkumulert dispersjon.


#### Oppgave 3 — Van der Waals a og b for ekte gasser

| Stoff | $a$ (L²·bar·mol⁻²) | $b$ (L·mol⁻¹) |
|---|---|---|
| Helium (He) | 0,035 | 0,0238 |
| Argon (Ar) | 1,355 | 0,0320 |
| Nitrogen (N₂) | 1,370 | 0,0387 |
| Metan (CH₄) | 2,30 | 0,0431 |
| Karbondioksid (CO₂) | 3,64 | 0,0427 |
| Benzen (C₆H₆) | 18,24 | 0,1193 |

**a)** $a$ hører til den **attraktive** delen av $u(r)$ (den $-(\sigma/r)^6$-halen), altså til $\varepsilon$: den senker trykket fordi molekylene trekkes svakt mot hverandre i stedet for mot beholderveggen. $b$ hører til den **repulsive** delen (den harde veggen ved liten $r$), altså til $\sigma$: molekylene har et endelig egenvolum, så det tilgjengelige volumet blir $V_m - b$.

**b)** 
* Etter størrelse ($b$): He < Ar < N₂ < CO₂ ≈ CH₄ < benzen. 
* Etter tiltrekning ($a$): He < Ar ≈ N₂ < CH₄ < CO₂ < benzen. 

Generelt følger de to parameterne hverandre: større molekyler har flere elektroner, høyere polariserbarhet og dermed sterkere dispersjonskrefter. Sammenhengen er likevel ikke perfekt. Et godt eksempel er CO₂ og metan, som har nesten samme verdi av $b$ (0,0427 mot 0,0431), men der CO₂ har en betydelig større verdi av $a$ (3,64 mot 2,30). Årsaken er at CO₂ i tillegg til dispersjonskreftene har et kvadrupolmoment og høy elektrontetthet i C=O-bindingene. Dette gir ekstra tiltrekning mellom molekylene. Parameteren $a$ avhenger derfor ikke bare av molekylstørrelsen, men også av hvordan elektronene er fordelt i molekylet.

**c)** 
* Stor $a$, liten $b$: tiltrekningsdominert, avviker mest ved **lav temperatur**, der tiltrekningen får molekylene til å samle seg og trykket faller under idealverdien (til slutt kondensasjon). 
* Stor $b$, liten $a$: egenvolum-dominert, avviker mest ved **høyt trykk / høy tetthet**, der molekylene presses tett og den endelige størrelsen skyver trykket over idealverdien. **Kort sagt: $a$ styrer avviket ved lav $T$, $b$ styrer det ved høyt $P$.**

**bonus** I hard-kule-modellen finnes ingen tiltrekning, så $a = 0$. Molekylene opptar fortsatt plass, så $b$ er endelig (proporsjonal med molekylvolumet). Vi sitter igjen med en gass av harde kuler som aldri tiltrekker hverandre — den avviker fra idealgassen kun gjennom egenvolumet, gir alltid *høyere* trykk enn idealgassen ved høy tetthet, og kondenserer aldri.


#### Oppgave 4 — Parkorrelasjonsfunksjonen g(r)

**a)** *Gass:* etter det korte forbudte området (der molekylene ikke kan overlappe) flater $g(r)$ raskt ut mot 1 — nesten ingen struktur. *Væske:* en tydelig første topp (nærmeste naboskall), etterfulgt av noen få, gradvis svakere topper — korttrekkende orden. *Fast stoff:* en serie skarpe topper som fortsetter over lange avstander — langtrekkende orden.

**b)** $g(r) = 0$ ved svært små $r$ speiler den repulsive delen av $u(r)$: den harde veggen gjør at to molekyler ikke kan komme nærmere enn omtrent $\sigma$. Sannsynligheten for å finne en nabo innenfor den avstanden er derfor null.

**c)** Posisjonen til første topp er den mest sannsynlige naboavstanden (nær $r^*$, bunnen av parpotensialet). At toppene *dør ut* på litt lengre avstand betyr at ordenen kun er lokal — hvert molekyl «kjenner» sine nærmeste naboer, men korrelasjonen forsvinner over noen få skall. Dette er signaturen til en væske.

**d)** *Fast* argon gir skarpe **Bragg-topper**: den periodiske ordenen gjør at konstruktiv interferens kun oppstår ved bestemte vinkler, gitt ved $n\lambda = 2d\sin\theta$. *Flytende* argon gir brede, diffuse ringer, siden bare korttrekkende orden finnes — ingen skarpe, veldefinerte plan å reflektere fra.

**e)** Sett inn verdiene:

| $r$ | $u_1 = -1/r$ | $u_2 = -1/r^6$ |
|---|---|---|
| 1 | $-1$ | $-1$ |
| 2 | $-0{,}5$ | $-1/64 \approx -0{,}016$ |

Ved $r = 1$ er de like sterke; ved $r = 2$ er Coulomb-potensialet $u_1$ omtrent 32 ganger sterkere. $1/r$ dør altså mye saktere med avstanden enn $1/r^6$. Derfor rekker monopol–monopol-vekselvirkningen (ione–ione) langt og gir **langtrekkende orden** — som i en NaCl-krystall, der hvert ion føler mange skall av naboer. Dispersjon ($1/r^6$) dør raskt og merkes bare mellom nærmeste naboer, så en edelgassvæske får bare **korttrekkende orden**. Dette illustrerer kapittelets poeng om at høyere multipolmomenter (dipol $\sim 1/r^3$, dispersjon $\sim 1/r^6$, …) gir stadig svakere og mer korttrekkende vekselvirkninger enn den langtrekkende monopolen.

#### Oppgave 5 — Energi mot entropi

$$F = U - TS$$

**a)** Ved **høy** temperatur er $T$ stor, så $-TS$-leddet veier tungt: entropien vinner, og systemet velger den fasen med flest tilgjengelige mikrotilstander — gassen. Ved **lav** temperatur er $-TS$-leddet lite, så $U$-leddet avgjør: systemet velger lavest potensiell energi — den kondenserte fasen. Temperaturen er «vekten» som avgjør hvor tungt entropien teller mot energien i $F = U - TS$.

**b)** Energien «vil» ha molekylene nær hverandre, i bunnen av den attraktive brønnen i $u(r)$ (lav $U$) — det er dette $a$-leddet i van der Waals-likningen fanger. Entropien «vil» spre dem utover (gass, mange mikrotilstander, høy $S$). Ved høy $T$ vinner entropien. Når du kjøler ned, blir $-TS$-leddet mindre viktig, og energigevinsten ved å ligge i brønnen blir avgjørende — molekylene samler seg, og gassen kondenserer.

**c)** *Sublimasjon* (tørris, fast CO₂ → gass): entropien favoriseres sterkt, og ved lavt trykk hoppes væskefasen over. *Smelting* (is → vann): entropien øker, og noe av hydrogenbindingsenergien gis opp. I begge tipper oppvarming balansen mot den fasen med høyest entropi, mens avkjøling tipper mot den med lavest energi / mest orden. (De fullstendige fasediagrammene kommer i et senere kapittel.)

**d)** At is er *mindre* tett enn vann viser at «lavest energi» ikke alltid betyr tettest pakking. De retningsbestemte hydrogenbindingene tvinger vannmolekylene inn i et åpent, «hullete» gitter som er energetisk gunstig, men romslig. Det er altså geometrien til bindingene — ikke bare styrken — som avgjør strukturen, og balansen energi/entropi kan gi et mindre tett fast stoff enn væsken.

#### Bonusoppgave ⭐ — Den hydrofobe effekten

**a)** Vannmolekyler binder sterkt til hverandre gjennom hydrogenbindinger. En upolar hydrokarbonhale kan ikke delta i dette nettverket, og «forstyrrer» det der den ligger. Halene trives derfor dårlig omgitt av vann.

**b)** Rundt hver eksponert hale må vannmolekylene ordne seg i et strukturert «bur» (klatrat-lignende) for å bevare flest mulig hydrogenbindinger seg imellom — det gir *lav* entropi for disse vannmolekylene. Skjuler man halene inni en micelle, frigjøres burvannet til å bevege seg fritt igjen. Vannets entropiøkning overgår da entropitapet ved å ordne halene, slik at **total entropi øker**. Selvorganiseringen er derfor entropidrevet — dette er den hydrofobe effekten.

**c)** For kondensasjon (Oppgave 5) er det $\Delta U < 0$ som driver: energien senkes, og entropitapet ($TS$) aksepteres fordi $T$ er lav. For den hydrofobe effekten er det $\Delta S_\text{total} > 0$ som driver, via det frigjorte vannet. Begge senker den frie energien $F = U - TS$, men fra hver sin ende av uttrykket. Poenget er fint: at ting «klumper seg sammen» er *ikke* alltid energidrevet — her er det entropien som gjør jobben.

**d)** Lipiddobbeltlag og flytende krystaller gir diffuse XRD-topper, ikke skarpe Bragg-topper. Det forteller at en membran har *flytende*, korttrekkende orden — nærmere en todimensjonal væske enn en krystall. Molekylene er ordnet lokalt (regelmessig avstand mellom hodegrupper, halene pakket sammen), men uten langtrekkende krystallinsk orden.

**e)** *(Åpen diskusjon)* Den samme balansen går igjen i proteinfolding (hydrofobe sidegrupper gjemmes i en kjerne), i cellemembraner, i DNA-basestabling, i emulsjoner og i at olje og vann ikke blandes. Det er rimelig å si at det er **samme fysikk** hele veien — svake krefter (dispersjon, dipol, hydrogenbinding) i konkurranse med entropi, styrt av $F = U - TS$ — ikke bare en analogi. Forskjellen fra tilfelle til tilfelle er hvilket ledd som dominerer, og om entropibidraget kommer fra molekylene selv eller fra omkringliggende vann.

Vi er klare for å bruke Gibbs faseregel:

- a) Kjemisk potensial er et mål på endring av fri energi på grunn av en endring i antall partikler. Kjemisk potensial er matematisk definert som den partielldervierte av den frie energien med hensyn på stoffmengde ved konstant volum ($C_V$)/trykk($C_p$) og temperatur: 

$$\mathrm{\mu }={\left(\frac{\partial G}{\partial N}\right)}_{V,T}$$

- b) 	For et én-komponent system har vi:

$$µ=G_{m}=\frac{H}{N}-\frac{TS}{N}=H_{m}-TS_{m}$$

Grafen viser hvordan µ varierer med temperatur. Vi finner stigningstallet ved å derivere med hensyn på T:  

$${\left(\frac{\partial \mu }{\partial T}\right)}_{p}={\left(\frac{\partial G_{m}}{\partial T}\right)}_{p}=-S_{m}$$ 

Så jo høyere entropi, jo større negativt stigningstall vil vises i et plot av kjemisk potensial mot temperaturen. Siden entropien er høyere for en gass (energien er mer spredt utover, flere antall frihetsgrader, tilgjengelige energinivåer osv) er stigningstallet for gassfasen størst. Vi ser noe av den samme trenden ved overgangen til væske fra fast stoff, men denne er svakere enn for overgangen fra væske til damp. 

- c) Fra fundamentallikningen for Gibbs fri energi vet vi at:

$$dG=Vdp-SdT={\left(\frac{\partial G}{\partial p}\right)}_{T}dp+{\left(\frac{\partial G}{\partial T}\right)}_{p}dT$$

Altså har vi:

$${\left(\frac{\partial G}{\partial p}\right)}_{T}=V slik at {{\left(\frac{\partial G_{m}}{\partial p}\right)}_{T}=\left(\frac{\partial \mu }{\partial p}\right)}_{T}=\ V_{m}$$

Gass har størst molart volum. Så stigningstallet ville vært størst for gassfasen. Men i dette tilfelle ville stigningen vært motsatt av forrige eksempel da 

$${\left(\frac{\partial \mu }{\partial p}\right)}_{T}=\ V_{m}$$

Væsker har stort sett et større molart volum enn fast fase (bortsett fra vann), så stigningstallet for denne ville i de fleste tilfeller vært nest størst. Så grafen ville sett noe liknende ut som vist under: 

![Kjemisk potensial vs. trykk](\bilder\kjemisk_potensial_trykk_faser.png)

- d) For én-komponent stoffer har vi som sagt: 

$$\mu =G_{m}=V_{m}dp-S_{m}dT$$

Avhengigheten av μ av trykk ved konstant temperatur er altså: 

$${\left(\frac{\partial G_{m}}{\partial p}\right)}_{T}={\left(\frac{\partial \mu }{\partial p}\right)}_{T}=\ V_{m}$$

Avhengigheten av μ av temperatur ved konstant trykk er: 

$${\left(\frac{\partial G_{m}}{\partial T}\right)}_{p}={\left(\frac{\partial \mu }{\partial T}\right)}_{p}=\ -S_{m}$$

- e) Figuren viser kjemisk potensial $\mu$ som funksjon av temperatur $T$ for fast og væske ved lavt og høyt trykk for to forskjellige stoffer. Helningen til hver kurve er gitt ved

$$\frac{d\mu}{dT} = -S$$

der $S$ er molarentropien. Væske har høyere entropi enn fast, og derfor blir kurven for væske brattere nedover enn kurven for fast.  

Smeltepunktet ved et gitt trykk er definert ved den temperaturen $T_f$ der fasene er i likevekt:

$$\mu_\text{fast}(T_f) = \mu_\text{væske}(T_f)$$

- i. Stoff (a) – «vanlig» oppførsel

I figur (a) ser vi at ved lavt trykk har vi et smeltepunkt $T_f$ der kurvene for fast og væske krysser. Ved høyt trykk ligger kurven for fast lavere enn kurven for væske (fast har lavere $\mu$ enn væske), og de to kurvene skjærer hverandre ved en høyere temperatur $T_f'$ slik at $T_f' > T_f$. Dette kan forklares ved at fastfasen har **mindre molart volum enn væsken**:

$$\Delta V_\text{fusjon} = V_\text{væske} - V_\text{fast} > 0$$

Endringen i kjemisk potensial med trykk kan skrives som:

$$d\mu \approx V\,dp$$

Når trykket øker, vil fasen med minst volum (her fast) få størst reduksjon i $\mu$ og dermed bli mest stabil. Økt trykk favoriserer derfor fastfasen, og man trenger høyere temperatur for at væsken skal bli like stabil som fast → smeltepunktet øker med trykk. Dette er typisk oppførsel for «vanlige» stoffer, som mange metaller og salter, f.eks. aluminium eller natriumklorid.

- ii. Stoff (b) – «anomal» oppførsel (vann-type)

I figur (b) har vi igjen et smeltepunkt $T_f$ ved lavt trykk, der $\mu_\text{fast} = \mu_\text{væske}$. Ved høyt trykk ser vi at kurven for væske nå ligger lavere enn kurven for fast, slik at væsken er mer stabil ved høyt trykk. Skjæringspunktet (smeltepunktet) flyttes til en lavere temperatur $T_f'$ slik at $T_f' < T_f$. Dette betyr at fastfasen har **større molart volum enn væsken**:

$$\Delta V_\text{fusjon} = V_\text{væske} - V_\text{fast} < 0$$

Siden $d\mu \approx V\,dp$, vil økt trykk senke det kjemiske potensialet mest for fasen med minst volum, altså væsken. Økende trykk favoriserer derfor væskefasen, og det blir «lettere» å smelte — smeltepunktet synker når trykket øker. Denne oppførselen er karakteristisk for vann, der is har lavere tetthet enn flytende vann (is har større volum enn vann). Derfor synker isens smeltepunkt med økende trykk.

**Oppgave 8**

Clapeyrons likning beskriver hvordan en faseovergang påvirkes av trykk:

$$\frac{dp}{dT}=\frac{∆_{trs}S_{m}}{∆_{trs}V_{m}}$$

```{admonition} Utledning
:class: important

Ved en faseovergang må de to fasene være i likevekt med hverandre, dvs. at deres kjemiske potensiale må være likt. Siden vi ser på avhengigheten av trykk og temperatur kan dette skrives som:

$${\mu }_{A}(p,T)={\mu }_{B}(p,T)$$

Det kjemiske potensiale svarer til den molare Gibbs fri energi: $\mu =G_{m}$

Den partiell-deriverte formen av Gibbs fri energi er:

$$dG_{m}=V_{m}dp-S_{m}dT$$

Setter vi at $dG_{m,A\ }=dG_{m,B}$ for de to fasene får vi videre:

$$V_{m,A}dp-S_{m,A}dT=V_{m,B}dp-S_{m,B}dT$$

$${(V}_{m,A\ }-V_{m,B\ })\frac{dp}{dT}=(S_{m,A\ }-S_{m,B\ })$$

Som igjen gir oss Clapeyron likningen!

$$\ \frac{dp}{dT}=\frac{S_{m,A\ }-S_{m,B\ }}{\ V_{m,A\ }-V_{m,B\ }\ }$$

```

Ved å bruke at $∆G = 0$ ved faseovergang kan denne gjøres om til:

$$\frac{dp}{dT}=\frac{∆_{trs}H_{m}}{∆_{trs}V_{m}\cdot T_{trs}}$$

Integrasjon gir videre:

$$p=p^{\ast }+\frac{∆_{trs}H_{m}}{∆_{trs}V_{m}}ln\left(\frac{T}{T^{\ast }}\right)\approx p^{\ast }+\frac{∆_{trs}H_{m}}{∆_{trs}V_{m}}\left(\frac{T-T^{\ast }}{T^{\ast }}\right)$$

Vi finner forskjellen i molart volum ved å bruke at:

$$\mathrm{\rho }=\frac{m}{V}=\frac{n\cdot M}{V}=\frac{M}{V_{m}}$$

Som gir: 

$$V_{m}=\frac{M}{\mathrm{\rho }}$$ 	

Slik at:

$∆V_{m}=M\left(\frac{1}{p_{s}}-\frac{1}{{\rho }_{l}}\right)=46.1\ g.mol^{-1}\cdot \left(\frac{1}{0.789 \, g.cm^{-3}}-\frac{1}{0.801 \, g.cm^{-3}}\right)=-0.8753 \, cm^{3}.mol^{-1}$

Siden det ikke er gitt noe annen informasjon antar vi at frysepunktet ved $-3.65 ^{\circ}C$ er gitt ved standard trykk = $1 \, atm$. Ved innsetting av verdiene i Clapeyrons likning fås da:

$1\ MPa=1 \, atm+\frac{8680 \, J.mol^{-1}}{-0.8753 \, cm^{3}.mol^{-1}}\cdot \left(\frac{T-269.5 \, K}{269.5 \, K}\right)$

$1\ MPa=0.101325 \, MPa+\frac{8680 \, J.mol^{-1}}{-0.8753 \, cm^{3}.mol^{-1}}\cdot \left(\frac{T-269.5 \, K}{269.5 \, K}\right)$

$0.898675\cdot {10}^{6} \, Pa\cdot \frac{-0.8753\cdot {10}^{-6} \, m^{3}.mol^{-1}}{8680 \, J.mol^{-1}}\cdot \ 269.5\ K\ +\ 269.5\ K\ =\ T$

$T = 269.48 \, K$

**Oppgave 11**

For å finne temperaturen der faseovergangen skjer er det hensiktsmessig plotte varmekapasitet som funksjon av temperatur. Varmekapasitet er definert som $C_{V}={\left(\frac{\partial U}{\partial T}\right)}_{V}$ og vi kan derivere dataen numerisk for å approksimere varmekapasiteten i dette systemet. 

Dette kan f.eks. implementeres i Python på følgende måte:

````python
import numpy as np
import matplotlib.pyplot as plt

T = [35, 40, 45, 50, 51, 51.5, 51.6, 55, 60]
U = [401, 435, 499, 582, 776, 887, 2012, 2020, 2031]
C = []

for i in range(len(T)-1):
    dT = T[i+1] - T[i]
    dU = U[i+1] - U[i]
    C.append(dU/dT)

plt.plot(T[0:-1], C)
plt.xlabel('Temperature (°C)')
plt.ylabel("Varmekapasitet")
plt.title('Varmekapasitet vs. Temperatur')
plt.show()
````

Varmekapasiteten har en tydelig topp rundt $51 ^{\circ}C$ og formen på toppen tyder på at det er en første ordens faseovergang som skjer her. Siden datamaskinen bruker diskrete tall vil vi ikke kunne observere at varmekapasiteten divergerer mot uendelig men toppen gir allikevel en god indikasjon på at faseovergangen skjer.
