# Løsningsforslag

## Uke 1 - Statistisk mekanikk, Entropi, Energi og Temperatur

### Diskusjonsoppgaver

#### Oppgave 1

**a)** Linjene er skarpe fordi energien i atomet/molekylet er kvantisert. Et foton kan bare absorberes eller sendes ut dersom energien dets er nøyaktig lik energiforskjellen $\Delta\epsilon$ mellom to tillatte nivåer. Siden bare bestemte $\Delta\epsilon$-verdier er lovlige, kan bare bestemte bølgelengder observeres, ikke et kontinuerlig spekter av farger.

**b)** For **ett atom**: den elektroniske kvantetilstanden (elektronenes fordeling i orbitaler) alene bestemmer mikrotilstanden, og all energi lagres der. For **ett molekyl**: mikrotilstanden er kombinasjonen av kvantetall fra *alle* frihetsgradene samtidig (elektronisk + vibrasjon $v$ + rotasjon $J$ + translasjon). Energien fordeles altså mellom flere uavhengige frihetsgrader. For **en samling av mange molekyler**: mikrotilstanden er den fullstendige kombinasjonen av *alle* molekylenes individuelle kvantetilstander samtidig, og energien fordeles i tillegg mellom de mange molekylene selv, ikke bare mellom frihetsgradene til ett molekyl.

#### Oppgave 2

**a)** En mikrotilstand for et system av mange molekyler beskriver hele systemets konfigurasjon samtidig (her: nøyaktig hvilke to plasser som har type 1-molekyler). Dette er *ikke* det samme som én mikrotilstand for ett enkelt atom, som bare beskriver tilstanden til akkurat det ene atomet, gitteret vårt sier noe om fire molekyler på én gang, ikke om ett.

**b)** En mikrotilstand er den fullstendige, detaljerte plasseringen. En makrotilstand er en observerbar egenskap ved hele systemet, her energien $E$. Flere ulike mikrotilstander gir samme makrotilstand, f.eks. gir plasseringene $\{A,B\}$, $\{A,C\}$, $\{B,D\}$ og $\{C,D\}$ alle $E=2\Delta\epsilon$.

**c)**

| Plassering (type 1 på) | Energi |
|---|---|
| $A,B$ | $2\Delta\epsilon$ |
| $A,C$ | $2\Delta\epsilon$ |
| $B,D$ | $2\Delta\epsilon$ |
| $C,D$ | $2\Delta\epsilon$ |
| $A,D$ | $4\Delta\epsilon$ |
| $B,C$ | $4\Delta\epsilon$ |

**d)** To ulike energiverdier ($2\Delta\epsilon$ og $4\Delta\epsilon$), altså to makrotilstander, blant seks mikrotilstander. Med $N_A\approx6\times10^{23}$ molekyler vokser antall mikrotilstander eksplosivt (kombinatorisk, som $\binom{N}{n}$), mens antall mulige energiverdier vokser mye saktere (bestemt av hva som er geometrisk mulig). Forholdet mikrotilstander-per-makrotilstand blir astronomisk stort. Dette er hvorfor vi trenger statistisk mekanikk: det er umulig å telle opp eller spore hver mikrotilstand for et reelt system, så vi må bruke sannsynligheter og gjennomsnitt over dem i stedet.

#### Oppgave 3

**a)** Mikrotilstanden endrer seg hele tiden fordi partiklene bytter tilstand med hverandre. Det som holder seg konstant er makrotilstanden, *antallet* $n$ partikler i eksitert tilstand, og dermed total energi $E=n\Delta\epsilon$, er uendret, selv om *hvilke* partikler som er eksiterte stadig endres. Det er nettopp dette som gjør systemet til en likevekt: en stabil makrotilstand bygget på en konstant strøm av mikroskopiske endringer.

**b)** Når $N$ øker, blir fordelingen $P(n)$ både høyere (mer sannsynlighet konsentrert nær gjennomsnittet) og *relativt* smalere, bredden $\sigma$ vokser bare som $\sqrt N$, mens gjennomsnittet vokser som $N$, så $\sigma/N\to0$. Dette er akkurat grunnen til at et system med $N_A\approx6\cdot10^{23}$ partikler alltid observeres i én bestemt, veldefinert makrotilstand: fordelingen er så ekstremt spiss at praktisk talt all sannsynlighet ligger på én eneste, smal energiverdi, selv om mikrotilstanden i prinsippet fluktuerer hele tiden.

**c)** Nei, det blir ikke alltid flere tilgjengelige mikrotilstander av mer energi. $\Omega(n)=\binom{N}{n}$ øker med $n$ bare opp til $n=N/2$; deretter *synker* den igjen etter hvert som stadig færre plasseringer er mulige (ved $n=N$, helt eksiterte system, er det igjen bare $\Omega(N)=1$ eneste mikrotilstand, akkurat som ved $n=0$). Multiplisiteten avhenger altså ikke bare av *hvor mye* energi som er tilgjengelig, men av *hvor mange forskjellige måter* den kan fordeles på.

**d)** $U$ er et sannsynlighetsveid *gjennomsnitt*, ikke nødvendigvis et heltall og ikke nødvendigvis toppunktet i fordelingen. Ved lav $N$ kan fordelingen være skjev nok til at den avrundede $U$-verdien og den mest sannsynlige energien (høyeste søyle) ikke er nøyaktig samme søyle. Når $N\to6\cdot10^{23}$, blir fordelingen så smal og symmetrisk at $U$ og den mest sannsynlige energien praktisk talt faller sammen,  forskjellen mellom dem blir umålelig liten.

#### Oppgave 4

**a)** Nær starten av kurven, lav $E$, stiger $S$ raskt. Stigningstallet $1/T$ er stort, som betyr at $T$ er lav. Mot toppunktet flater kurven gradvis ut; stigningstallet nærmer seg null, som betyr at $T$ vokser og til slutt går mot uendelig akkurat ved toppen.

**b)** Toppunktet er der $\Omega(n)$, og dermed sannsynligheten for makrotilstanden, er størst. Siden systemet statistisk sett driver mot den mest sannsynlige tilgjengelige makrotilstanden, er dette punktet nøyaktig likevektstilstanden, der er det simpelthen flest "veier dit".

**c)** Siden $n=0$ ligger langt fra toppunktet, forventer vi at $n$ øker over tid mot toppunktet ved $n=N/2$, drevet av at systemet beveger seg mot stadig mer sannsynlige (høyere entropi) makrotilstander. Grafen bekrefter dette: stigningstallet er stort og positivt ved $n=0$, som betyr at $S$ (og dermed sannsynligheten) øker kraftig for hver energienhet systemet tar opp, helt til toppunktet nås.

**d)** Nær $n=0$ er stigningstallet stort; nær toppunktet er det nesten null. Dette forteller oss at "drivkraften" mot likevekt er sterk når systemet er langt unna (store gevinster i sannsynlighet per energienhet som absorberes), men blir svakere og svakere jo nærmere likevekt systemet kommer, konsistent med hvordan systemer typisk nærmer seg likevekt raskt i starten og saktere etter hvert.

**e)** $U$ er et gjennomsnitt av de mikroskopiske energiene $\epsilon_n$, veid etter hvor sannsynlig hver er. $\Delta S$ bygger direkte på $\Omega$, antallet mikrotilstander. Begge er regnet ut fra mikroskopisk informasjon (individuelle energinivåer og hvor mange mikrotilstander som finnes), men resultatet er ett eneste tall vi kan måle med et vanlig laboratorieinstrument (kalorimeter, termometer).

#### Oppgave 5

**a)** Siden alle individuelle mikrotilstander (ved en gitt total energi i systemet) antas like sannsynlige, er sannsynligheten for å observere en bestemt makrotilstand proporsjonal med *hvor mange* mikrotilstander som gir opphav til den. En makrotilstand med flere mikrotilstander har rett og slett flere "veier dit", og observeres derfor oftere.

**b)** Fra oppgave 2c: $\Omega(2\Delta\epsilon)=4$ (plasseringene $\{A,B\}$, $\{A,C\}$, $\{B,D\}$, $\{C,D\}$) og $\Omega(4\Delta\epsilon)=2$ (plasseringene $\{A,D\}$, $\{B,C\}$).

**c)** $S(2\Delta\epsilon)=k_B\ln4\approx1{,}386\,k_B$. $S(4\Delta\epsilon)=k_B\ln2\approx0{,}693\,k_B$. Energien $2\Delta\epsilon$ har altså dobbelt så høy entropi som $4\Delta\epsilon$, siden den har dobbelt så mange mikrotilstander.

**d)** Grunntilstanden er $E_0=5\Delta\epsilon$ med $\Omega(E_0)=8$. Den mest sannsynlige energien er $E^*=15\Delta\epsilon$ med $\Omega=30\,216$ — over 3700 ganger større. Den mest sannsynlige energien er *ikke* grunntilstanden her, i motsetning til subsystemet i b) (der $2\Delta\epsilon$ tilfeldigvis både var lavest energi og hadde høyest $\Omega$). Årsaken: i det lille subsystemet er det for få mikrotilstander totalt til at forskjellen får utslag, mens i det store gitteret finnes det enormt mange flere måter å oppnå en middels energi på enn en ekstrem (enten svært lav eller svært høy) energi — multiplisiteten domineres av kombinatorikk, ikke av hvor "gunstig" energien er i seg selv.

**e)** $S(8\Delta\epsilon)=k_B\ln(402)\approx5{,}996\,k_B$. $S(9\Delta\epsilon)=k_B\ln(1112)\approx7{,}014\,k_B$.

$$\frac{\Delta S}{\Delta E}\approx\frac{7{,}014-5{,}996}{1}\,k_B/\Delta\epsilon\approx1{,}02\,k_B/\Delta\epsilon$$

Stigningstallet er positivt, som betyr $T>0$ — helt normalt, siden $8\Delta\epsilon$ og $9\Delta\epsilon$ begge ligger godt *under* toppunktet ($15\Delta\epsilon$) i fordelingen. Hadde stigningstallet vært negativt, ville det betydd at vi hadde beveget oss forbi toppunktet i entropikurven — inn i et område med "negativ temperatur", akkurat slik OBS-boksen i diskusjonsoppgave 4 nevnte.

#### Oppgave 6

**a)** $\Omega(n)$ forteller oss *hvor mange* mikrotilstander som gir en bestemt makrotilstand (energi $n\Delta\epsilon$) — et helt tall, en telling. $p(\epsilon)$ forteller oss sannsynligheten for å finne systemet i én bestemt, navngitt mikrotilstand med energi $\epsilon$ — et sannsynlighetstall mellom 0 og 1, ikke en telling.

**b)** Når $\Delta\epsilon/k_BT$ øker, blir eksponenten $-\Delta\epsilon/k_BT$ mer negativ, og $p(\epsilon_1)/p(\epsilon_0)=e^{-\Delta\epsilon/k_BT}$ blir mindre. Mikrotilstanden med høyere energi blir altså relativt mindre sannsynlig sammenlignet med grunntilstanden, jo større dette forholdet blir.

**c)**
- Avlest fra grafen: $N(2)/N(0)\approx0{,}14$.
- Analytisk ved $\Delta\epsilon/k_BT=1$: $N(2)/N(0)=e^{-2\cdot1}=e^{-2}\approx0{,}135$ — stemmer godt med avlesningen.
- Ved $\Delta\epsilon/k_BT=2$: $N(2)/N(0)=e^{-2\cdot2}=e^{-4}\approx0{,}018$. Forholdet ble mindre, som stemmer med b): et større $\Delta\epsilon/k_BT$ gir lavere relativ sannsynlighet for den høyere-energitilstanden.

At systemet "driver mot" makrotilstand 2 betyr ikke at noe tvinger det dit, men at det statistisk sett er langt mer sannsynlig å finne systemet der enn i makrotilstand 1, gitt tilstrekkelig tid og fri utveksling mellom de to.

## Termokjemi

**Hess lov** 

1. I tilstandsfunksjoner kan veien mellom start- og sluttilstand velges fritt. Det å dele prosessen opp i delprosesser svarer til å endre veien men så lenge start- og sluttilstanden er de samme vil endringen i tilstandsfunksjonen _totalt_ være den samme. Det å finne termodynamiske størrelser eksperimentelt kan være utfordrende. Istedenfor holder det å vite start og slutt-tilstandene. Veien mellom tilstandene kan da konstrueres ved å legge sammen kjente reaksjoner og man kan regne ut mange termodynamiske størrelser basert på tabeller over eksperimentelt målte verdier. Hess lov gjelder for alle tilstandsfunksjoner, men uttrykkes for eksempel slik for entalpi: 

$$\Delta H = \sum \Delta H(produkter) - \sum \Delta^{\circ} _f H(reaktanter)$$

Dersom vi har delprosessene A, B, C, vil vi heller bruke loven slik: 

$\Delta H = \Delta H(A) + \Delta H(B) + \Delta H(C)$ innad i disse delprosessene har allerede forskjellen i entalpi mellom start- og sluttilstand blitt beregnet eller målt i henhold til formelen over. I neste oppgave vil du se et eksempel på dette. 

2. La oss begynne med delprosess (1), vi ser at produktet vi ønsker å danne er på venstre (reaktant) side av likningen. Derfor må vi "snu" denne likningen for å få diboran som produkt.

$${B_2O_3}_{(s)} + 3\, {H_2O}_{(g)} \longrightarrow {B_2H_6}_{(g)} + 3\, {O_2}_{(g)}$$

Dette fører til at vi må bytte fortegn på entalpiendringen til delprosessen $\Delta H^{\circ}(1)$. Vi har laget den første delen av likningen vår:

$$\Delta _fH{\circ} = -\Delta H^{\circ}(1) ...$$

Vi ser at vi trenger ${B_2O_3}_{(s)}$ og $3\, {H_2O}_{(g)}$ for å gjøre delprosess (1). Dette kan vi få fra delprosess (2) og (3). Vi får ${B_2O_3}_{(s)}$ "direkte" fra prosess (2) og kan derfor legge entalpiendringen for prosessen rett inn i uttrykket vårt: 

$$\Delta _fH^{\circ} = -\Delta H^{\circ}(1) + \Delta H^{\circ}(2) ...$$

Vi får ${H_2O}_{(g)}$ "direkte" fra prosess (3), men kun ett vannmolekyl av gangen. Derfor må vi gange delprosessen med tre og vi får derfor også en tre ganger så høy entalpiendring: 

$$\Delta _fH^{\circ} = -\Delta H^{\circ}(1) + \Delta H^{\circ}(2) + 3 \times \Delta H^{\circ}(3)$$

Vi har funnet uttrykket vi trenger, og er klare for å regne ut dannelsesentalpien ved hjelp av tabellverdier. 


**Entalpi**

1. Entalpi har som matematisk uttrykk: 

$$\Delta H = \Delta U - pV$$

Dersom trykket holdes konstant, er entalpien lik: 

$$\Delta H = q_p$$

Det vil si varmeoverføringen ved konstant trykk. Utledningen av dette er forklart i detalj på "Makroskopisk termodynamikk" - siden. Entalpi er svært nyttig fordi vi "slipper" å tenke på arbeidsleddet til den indre energien. I kjemi er entalpi en særlig nyttig størrelse fordi de fleste kjemiske prosesser foregår ved konstant trykk. Entalpien lar oss dermed regne på varmeoverføringen som skjer i løpet av slike prosesser. 

2. En _endoterm_ prosess er en prosess som tar opp varme fra omgivelsene. For slike prosesser har vi som konvensjon at entalpiendringen $\Delta H$ har positivt fortegn. En _eksoterm_ prosess er en prosess som gir fra seg varme til omgivelsene. For slike prosesser har vi som konvensjon at entalpiendringen har negativt fortegn. 

3. Dannelsesentalpier er tabulerte verdier for mange vanlige kjemiske prosesser. De er enten direkte målt eksperimentelt eller beregnet fra eksperimentelle data. Vi har som utgangspunkt at dannelsesentalpien til grunnstoffer i sin mest stabile tilstand er lik 0. Dannelsesentalpier er verdiene vi benytter oss av når vi bruker Hess lov for å beregne entalpiendringer for ulike prosesser. Vi må bryte ned prosessene våre til delprosesser som har kjente, tabulerte dannelsesentalpier. 

4. I løpet av en kjemisk prosess ser vi ofte en endring i sammensetningen til systemet vi er interesserte i: I kjemiske reaksjoner forvandles reagenser til produkter ved at kjemiske bindinger brytes og dannes. I faseoverganger endres de _intermolekylære_ kreftene som holder molekylene i systemet sammen. Og i løsninger kan for eksempel salter _solvatiseres_ ved at ionene omringes av løsemiddel molekyler. For enkelte prosesser er det ikke sikkert vi har tilgang på dannelsesentalpiene vi trenger for å regne på $\Delta H$. Da kan det være nyttig å kunne resonnere seg frem til hva slags entalpiendring vi forventer for prosessen basert på de molekylære kreftene i systemet. 

Kort fortalt kan vi gjøre dette ved å vurdere den relative styrken til bindingene som brytes og dannes i løpet av prosessen! _Svake bindinger_ gir atomene mye frihet til å vibrere, rotere og bevege seg rundt og ligger derfor høyt i energi. Fordi et system alltid tenderer mot å minimere energien sin, byttes gjerne svake bindinger ut med _sterke bindinger_ som ligger lavere i energi da systemet blir mer rigid og atomene har mindre frihet til vibrasjon, rotasjon og translasjon. Energigapet mellom de to tilstandene frigjøres som varme, og det er slik vi ender opp med **eksoterme prosesser**. Vi skal se på et konkret eksempel på en eksoterm reaksjon i organisk kjemi i slutten av avsnittet. Vi kan også gå motsatt vei: I faseoverganger fra væske til gass bytter vi ut _sterke interaksjoner_ som gjør at væsken holdes sammen med _svake interaksjoner_ som lar gassen flyte fritt i rommet. I dette tilfellet må systemet ta opp varme fra omgivelsene som det bruker til å "tette" energigapet mellom tilstandene. Slik får vi **endoterme prosesser** hvor vi må tilføre varme til systemet.

5. Se tilbake i laboppgavene du har hatt for å finne eksempler! I alle reaksjoner hvor du har måttet kjøle ned eller varme en løsning eller blanding har du mest sannsynlig tatt hensyn til en entalpiendring. 

6. 

a) La oss se på bindingene som dannes og brytes i reaksjonen! Vi legger særlig merke til $$P=O$$ som er en veldig sterk binding. Også den nye $C=C$ bindingen er litt sterkere enn $C=O$ på reaktantsiden, men det er først og fremst dannelsen av $P=O$ bindingen som driver reaksjonen fremover. Fordi denne bindingen er sterkere og dermed mer stabil må energigapet mellom reaktanter og produktet tettes ved at systemet frigir energi på form av varme. Vi tror derfor at prosessen er _eksoterm_.

b) Siden reaksjonen er eksoterm er det tryggest og mest praktisk å gjøre reaksjonen ved lav temperatur for å kontrollere varmeutviklingen.


**Entropi**

1. _Faste stoffer_

Faste stoffer kjennetegnes ved at de er _ordnede_ systemer, i hvert fall når vi sammenlikner dem med væsker og gasser. Partiklene i et fast stoff sitter i gittere. Nesten alle posisjonene i gitteret er opptatt av partikler og derfor er det lite mulighet for hver enkelt partikkel å bevege seg rundt. Disse faste posisjonene gjør at systemet kan arrangeres på få måter, og derfor har det tilgang til få mikrotilstander for å realisere sin makroskopiske tilstand. Dette forklarer hvorfor faste stoffer generelt har lav entropi. Dette forklarer også de makroskopiske karakteristikkene til faste stoffer: De kan ofte være harde og flyter ikke utover slik væsker og gasser gjør. 

_Væsker og løsninger_

Væsker og løsninger eksisterer et sted mellom faste stoffer og gasser, det er verken fullstendig ordnede eller fullstendig uordnede. De flyter mye friere rundt enn et fast stoff gjør, men henger fremdeles sammen i motsetning til gasser som får spre seg i et åpent rom. I kjemi er vi særlig interesserte i løsninger. Et fast stoff som løses i en væske øker entropien sin, men løsemiddelet ordnes også i større grad enn før rundt det løste stoffet. Nyansene rundt entropi og løsninger forklarer vi nærmere senere i avsnittet. 

_Gasser_

Gasser kjennetegnes ved at de er _uordnede_ systemer i forhold til faste stoffer og væsker. I et åpent system får gass slippe ut og vil derfor oppta et stort volum, for eksempel et helt avtrekksskap. Dette gjør at gassen har utrolig mye plass å boltre seg på! Gasspartiklene har utallige plasser de kan være og derfor kan systemet arrangeres på svært mange måter. Med andre ord har det tilgang på mange mikrotilstander for å realisere sin makrotilstand og vil derfor ha en høy entropi! De fleste prosesser som innebærer dannelse av gass er entropisk drevne. 

2. Å observere aggregattilstandene til speciene på reagens- og produktside av en reaksjonslikning er en av de letteste måtene å vurdere entropiendringen i løpet av en prosess på. Aggregattilstandene fast stoff (subskript (s)), løst stoff (i vann subskrift (aq)), væske (subskrift (l)) og gass (subskrift (g)) har økende grad av uorden som beskrevet over. 

a) Når vann koker får vi dannelse av vanndamp. Siden gass har høyere entropi enn væske får vi en positiv entropiendring. 

b) Et løsemiddel som kokes vil gå over i gassfase. For at reaksjonsblandingen ikke koker tørt bruker vi en kjøler. Når løsemiddel-molekyler i gassfase når kjøleren, kjøles de ned og blir til væske igjen. Siden væske har lavere entropi enn gass har prosessen en negativ entropiendring. Bonus: Hvilken drivkraft tar over når temperaturen går ned i det gassen treffer kjøleren? Jo, ved denne lavere temperaturen blir i stedet entalpien dominerende: Det blir gunstig for systemet å gå tilbake i væskeform fordi sterkere bindinger kan dannes mellom molekylene. 

c) Et salt som løses i vann får økt entropi fordi ionene i løsningen har langt flere måter å arrangere seg på enn i saltstrukturen. Vi må allikevel passe på koordinerende egenskaper. Alle ladde partikler i vann vil i variende grad koordineres til vannmolekyler via ione-dipol bindinger. For noen kationer med høy ladningstetthet (små eller høy positiv ladning), er den koordinerde effekten på vann så kraftig at entropien faktisk går ned ved å forsøke å løse saltet i vann!

**Gibbs fri energi og spontanitet**

1. Vi vet at entalpien beskriver varmeoverføring ved konstant trykk. Siden entalpien inngår i uttrykket for Gibbs fri energi er det vanligst at vi bruker Gibbs fri energi når vi har en situasjon hvor trykket er konstant. Siden de fleste kjemiske prosesser foregår ved konstant trykk er Gibbs fri en nyttig størrelse for oss! Vi regner ut Gibbs fri energi ved en gitt temperatur T som også holdes konstant. 

2. En prosess som er spontan vil gå uten å behøve kontinuerlig påvirkning av ytre krefter som elektrisk arbeid, varme eller lys. To forskjellige situasjoner blir ofte omtalt som spontane prosesser, og dette kan virke litt forvirrende: En prosess er spontan dersom den går av seg selv når den først har startet, selv om det kanskje var behov for en ytre påvirkningskraft for å starte den. Et eksempel på dette kan være forbrenninger: De går av seg selv, men trenger ofte en gnist eller flamme for å kunne starte. Vi sier også spontan om en prosess som både starter og går helt av seg selv. Pass også på at vi snakker om spontanitet gitt en viss temperatur T. Ofte vil det være behov for å varme reaksjoner for at de skal skje, fordi de er spontane veden høyere temperatur enn romtemperatur. 

Eksergon: $\Delta G < 0$ — prosessen frigjør fri energi og er spontan.
Endergon: $\Delta G > 0$ — prosessen krever tilførsel av fri energi for å skje, altså ikke-spontan i den retningen.

Nei, spontane prosesser skjer ikke nødvendigvis raskt. Her må vi skille mellom _termodynamikken_ og _kinetikken_! Mens Gibbs fri energi kun måler om start- eller sluttilstanden til prosessen er mest gunstig (gir best stabilitet), bestemmes farten av _aktiveringsbarrieren_ mellom de to tilstandene. Hva slags _mellomtilstand_ må vi innom for å danne produkt? Et kjent eksempel er prosessen Diamant $\longrightarrow$ Grafitt. Prosessen er svært eksergon, men går så sakte at det er nesten umulig å spore endringen. Tenk på en diamantring i en skuff. En slik ring kan gå i arv i generasjoner uten å bli til grafitt. 

3. Dette handler om hvilket av leddene i $\Delta G = \Delta H - T\Delta S$ som dominerer og "trekker" $\Delta G$ i negativ retning:

Entalpisk drevet: $\Delta H$ er sterkt negativ og dominerer over $-T\Delta S$ (som kan være ubetydelig, eller til og med ugunstig). Typisk for reaksjoner med sterk binding/frigjøring av varme, f.eks. forbrenning.

Entropisk drevet: $-T\Delta S er sterkt negativ (altså $\Delta S$ er stor og positiv) og dominerer, selv om $\Delta H$ kan være tilnærmet null eller positiv. Typisk for blanding, oppløsning av ioner, og den hydrofobe effekt.

Med $\Delta H = -1368 \text{ kJ/mol}$ er entalpibidraget enormt, og reaksjonen er i all hovedsak entalpisk drevet — dette ene tallet er langt større i størrelse enn $T\Delta S$ vanligvis vil være ved romtemperatur for en slik reaksjon. I tillegg dannes det flere mol gass på produktsiden enn på reaktantsiden (fra 3 mol $O_2$​ til 2 mol $CO_2$​ + 3 mol $H_2O_{(g)}​ = 5 mol gass$), noe som gir et positivt entropibidrag. Begge termer trekker altså $\Delta G$ i negativ retning — reaksjonen er sterkt spontan!

At vi likevel ikke er redde for en åpen flaske etanol i romluft, handler om _kinetikk_, ikke _termodynamikk_: forbrenning har en betydelig aktiveringsenergi (man må tilføre en flamme eller gnist for å bryte O=O- og C–H-bindinger og starte radikalreaksjonene som skal til for forbrenningen). Uten denne "tenningsenergien" er reaksjonshastigheten ved romtemperatur neglisjerbar, selv om $\Delta G$ er svært negativ.

6. En ikke-spontan prosess har $\Delta G > 0$ under de gitte betingelsene. Vi kan likevel få den til å skje ved å endre situasjonen slik at netto $\Delta G$ blir negativ. De vanligste strategiene er:

- Koble prosessen til en sterkt eksergon prosess (energikobling). Hvis en ikke-spontan reaksjon ($\Delta G_1 > 0$) skjer samtidig med en sterkt spontan reaksjon ($\Delta G_2 < 0$, med $|\Delta G_2| > |\Delta G_1|$, blir summen $\Delta G_{tot} = \Delta G_1 + \Delta G_2 < 0$, og den koblede totalprosessen blir spontan. Dette er selve grunnprinsippet bak biologisk energiomsetning: hydrolyse av $ATP$ til $ADP + P_i$​ er sterkt eksergon ($\Delta G \approx -30 \text{ kJ/mol}$ og kobles til utallige ellers ikke-spontane prosesser i cellen: Muskelkontraksjon, aktiv transport over membraner, biosyntese av makromolekyler osv. Nøkkelen er at prosessene deler et felles mellomtrinn eller intermediat (f.eks. et fosforylert enzym), slik at de reelt er koblet og ikke bare skjer parallelt.

- Tilføre arbeid utenfra. Man kan drive en ikke-spontan prosess ved å tilføre energi i en annen form enn varme, for eksempel typisk elektrisk arbeid eller lys:

Elektrolyse er et klassisk eksempel: nedbryting av vann til $H_2$​ og $O_2$​ er sterkt ikke-spontant ($\Delta G > 0$), men ved å påtrykke en ytre spenning (tilføre elektrisk arbeid $w_{elek}$​) kan vi likevel drive reaksjonen. 

Fotosyntese er naturens variant: lysenergi driver den sterkt ikke-spontane omdannelsen av $CO_2$​ og $H_2O$ til glukose og $O_2$.

- Katalyse: En katalysator senker aktiveringsenergien og øker dermed reaksjonshastigheten — den endrer altså _kinetikken_, ikke termodynamikken. Siden G er en tilstandsfunksjon (avhenger kun av start- og sluttilstand), påvirker en katalysator verken $\Delta G$, $\Delta H$ eller $\Delta S$ for reaksjonen i det hele tatt. En katalysator senker aktiveringsenergien slik at det blir lettere for prosessen å gå mellom start- og sluttilstanden: Veien endres. I praksis skjer dette ofte ved at katalysatoren bidrar til å stabilisere overgangstilstanden.
