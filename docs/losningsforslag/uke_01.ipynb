## Diskusjonsoppgaver

### Oppgave 1

**a)** At energien til et atom er kvantisert betyr at atomet bare kan ha bestemte energiverdier - ikke alle mulige verdier imellom. Akkurat som en stige har trinn på faste høyder (trinn 0,1,2,...) og du ikke kan stå på trinn 1.5 eller sveve mellom trinnene, kan atomet bare "stå" på bestemte energinivåer ($\varepsilon_0, \varepsilon_1, \varepsilon_2, \dots$). 


**b)** Når et atom sender ut lys i form av et foton, hopper et elektron ned fra et høyere til et lavere energinivå, og lyset som sendes ut har en energi som er nøyaktig lik forskjellen mellom de to nivåene. Siden det bare finnes noen få bestemte, tillatte nivåer, finnes det bare noen få mulige energiforskjeller – og dermed bare noen få bestemte bølgelengder (farger) lyset kan ha. Derfor ser vi noen få skarpe linjer i stedet for et jevnt fargebånd. Et kontinuerlig fargebånd ville betydd at alle energiforskjeller var mulige, men det er de ikke.

**c)** En makrotilstand beskriver systemet med størrelser vi kan måle i laboratoriet: temperatur, trykk, volum, total energi. En mikrotilstand er én bestemt måte energien er fordelt på molekylnivå, konsistent med den observerte makrotilstanden.

En analogi er "en skog". På et makronivå kan flere ulike områder være skoger - de er samling av trær, busker, dyr osv. På et mikronivå er skogene forskjellige: trærne er posisjonert forskjellig, det er forskjellige antall trær osv.


**d)** En frihetsgrad er en uavhengig måte et molekyl kan bevege seg eller lagre energi på. Hver frihetsgrad er et eget "energilager" som bidrar til molekylets totale energi. Grunnen til at et molekyl har flere frihetsgrader enn et enkeltatom, er at et molekyl har bindinger som kan vibrere og en struktur som kan rotere – ting et enkeltatom ikke har.

Den totale energien til et molekyl er summen av bidragene fra hver frihetsgrad:

$$\varepsilon_{tot} = \varepsilon_{elec} + \varepsilon_{trans} + \varepsilon_{rot} + \varepsilon_{vib}$$

- Elektronisk ($\varepsilon_{elec}$): energien knyttet til hvilke tilstander elektronene rundt atomkjernene er i (tenk orbitaler).
- Translasjon ($\varepsilon_{trans}$): energien fra at hele molekylet beveger seg gjennom rommet.
- Rotasjon ($\varepsilon_{rot}$): energien fra at molekylet roterer om en akse.
- Vibrasjon ($\varepsilon_{vib}$): energien fra at atomene svinger fram og tilbake rundt bindingene sine.

Et enkeltatom har bare de to første (elektronisk og translasjon), siden det verken kan rotere eller vibrere.

En mikrotilstand er én bestemt, fullstendig beskrivelse av hvordan energien er fordelt på alle frihetsgradene samtidig – altså nøyaktig hvilken elektronisk, translasjons-, rotasjons- og vibrasjonstilstand molekylet er i.

En makrotilstand er derimot bare det vi kan måle utad, for eksempel den totale energien. Poenget er at mange forskjellige mikrotilstander kan gi samme makrotilstand: energien kan fordeles på frihetsgradene på mange ulike måter og likevel gi samme totale energi $\varepsilon_{tot}$.


**e)** Statistisk mekanikk er måten vi forklarer et systems målbare, makroskopiske egenskaper (som temperatur og trykk) ut fra hva de enormt mange små partiklene i systemet gjør, ved hjelp av sannsynlighet og gjennomsnitt over alle mikrotilstandene.

Forskjellen på de to lovtypene:

* En empirisk lov er satt opp for å passe til noe vi har observert. Vi måler noe først, og finner så en formel som beskriver målingene. Den forteller hva som skjer, men ikke hvorfor.
* En statistisk lov er utledet fra grunnleggende antakelser om hvordan partiklene oppfører seg. Den følger av teorien og forklarer hvorfor den makroskopiske oppførselen blir som den blir.

Kort sagt: den empiriske loven beskriver, den statistiske loven forklarer.

### Oppgave 2

**a)** Hver partikkel kan velge enten boks A eller boks B - dette er $2$ muligheter for hver av de $5$ partiklene som totalt gir

$2 \cdot 2 \cdot 2 \cdot 2 \cdot 2 = 2^5 = 32 \ \text{mikrotilstander}$


**b)** Makrotilstadnen er antall partikler i boks A, altså $n_A$. Dette kan være alt fra $0$ (ingen i A) til $5$ (alle i A):

$n_A = 0,1,2,3,4,5$

Altså; $6 \ \text{makrotilstander}$


**c)** Vi ønsker å regne ut hvor mange mikrotilstander som gir opphav til hver av makrotilstandene, altså multiplisiteten $\Omega (n_A)$ til hver $n_A$. Spørsmålet vi stiller oss er: hvor mange måter kan vi velge $n_A$ partikler fra totalt $N=5$ partikler?

Dette kan regnes ut ved binomialkoeffisienten:

$\binom{n_A}{5}=\frac{5!}{n_A!(5-n_A)!}$

Vi får dermed:

| $n_A$ | $0$ | $1$ | $2$ | $3$ | $4$ | $5$ |
| ----- | --- | --- | --- | --- | --- | --- |
| $\binom{5}{n_A}$ | $\binom{5}{0}=1$ | $\binom{5}{1}=5$ | $\binom{5}{2}=10$ | $\binom{5}{3}=10$ | $\binom{5}{4}=5$ | $\binom{5}{5}=1$ |


La oss se på $n_A=2$ og $n_A=5$, og vi nummererer de $5$ partiklene fra 1-5.

Hvordan kan vi velge $2$ partikler ut av totalt $5$ partikler? 

$(1,2), (1,3), (1,4), (1,5), (2,3), (2,4), (2,5), (3,4), (3,5), (4,5)$

Dette gir $\Omega (2) = 10$, som vi fant tidligere.
 
Hvordan kan vi velge $5$ partikler ut av totalt $5$ partikler?

$(1,2,3,4,5)$

Dette gir $\Omega (5) = 1$, som vi også fant tidligere.

Merk! For eksempel, $(1,2)$ er det samme som $(2,1)$. Det spiller ikke hvilken rekkefølge vi har valgt dem i.


**d)** Makrotilstandene $n_A=2$ og $n_A=3$ har samme, høyeste multiplisitet. Dette er som forventet fordi de jevneste fordelingene (omtrent like mange i hver boks) kan realiseres på flest måter, mens de ekstreme (alle i én boks) bare kan skje på én måte. Derfor "klumper" multiplisiteten seg i midten.


**e)** Sannsynligheter finner man generelt ved å dele antall måter å realisere en hendelse på, på antall måter totalt (gitt at alle måtene er like sannsynlige). For oss betyr det følgende:

$P(n_A)=\frac{\Omega (n_A)}{\text{\# mikrotilstander}}$

Derfor får vi

$P(5)=\frac{\Omega (5)}{32}=\frac{1}{32}$

$P(3)=\frac{\Omega (3)}{32}=\frac{10}{32}=\frac{5}{16}$

Vi ser at det er 10 ganger så sannsynlig at partiklene fordeler seg med 3 i A og 2 i B, enn at alle havner i boks A!


**f)** Når $N$ øker, blir fordelingen mer klokkeformet (en normalfordeling) og relativt smalere og skarpere rundt midten. Toppen ved den jevne fordelingen (omtrent halvparten i hver boks) blir stadig mer dominerende, mens de ekstreme utfallene (alle i én boks) blir forsvinnende usannsynlige.

Grunnen er at antall partikler i boks A er en sum av mange uavhengige "myntkast". Sentralgrenseteoremet sier at slike summer nærmer seg en normalfordeling, og den relative bredden krymper som $1/\sqrt{N}$ – så jo flere partikler, jo tettere presses sannsynligheten sammen rundt midten.


**g)** Med $N \approx 10^{23}$ partikler er toppen i fordelingen så ekstremt skarp at systemet i praksis *alltid* befinner seg i (eller uendelig nær) makrotilstanden med flest mikrotilstander – altså den jevneste fordelingen. Avvik fra denne er så usannsynlige at de aldri observeres. At all luften i et rom spontant samler seg i det ene hjørnet er ikke forbudt, bare så astronomisk usannsynlig (som "alle i boks A", men med $N \approx 10^{23}$) at det aldri skjer.

Dette er kjernen i termodynamikkens andre lov: systemet utvikler seg mot maksimal multiplisitet, og siden entropien er $S = k_B \ln \Omega$, betyr det at entropien øker mot et maksimum. Likevekt er rett og slett den fordelingen som kan realiseres på overveldende flest mikroskopiske måter.

### Oppgave 4

**a)** Det som holder seg noenlunde konstant er makrotilstanden, altså antall eksiterte partikler $n$ (og dermed energien $E = n\Delta\epsilon$), ikke selve mikrotilstanden. Systemet bytter hele tiden hvilke partikler som er eksiterte, men $n$ endrer seg lite fordi svært mange mikrotilstander gir samme $n$. Likevekt betyr derfor at de målbare størrelsene svinger rundt en stabil middelverdi, ikke at ingenting skjer på mikronivå.


**b)** Når $N$ øker, blir toppen på $P(n)$ høyere og skarpere. Den absolutte bredden vokser som $\sqrt{N}$, men den relative bredden (bredde delt på middelverdi) går som $\sqrt{N}/N = 1/\sqrt{N}$ og krymper mot null. For $N \approx 6\cdot10^{23}$ blir $1/\sqrt{N} \approx 10^{-12}$, så fordelingen er i praksis en uendelig smal nål ved én verdi, og det er derfor et makroskopisk system alltid ser ut til å ha én bestemt, veldefinert makrotilstand.


**c)** Når $k_BT/\Delta\epsilon$ øker, blir mer energi tilgjengelig og flere partikler eksiteres, så $n$ øker. Multiplisiteten $\Omega(n) = \binom{N}{n}$ er størst ved $n = N/2$ og synker mot 1 i begge ender. Så det blir ikke alltid flere mikrotilstander av mer energi: opp mot $n = N/2$ øker $\Omega$, men når nesten alle er eksiterte synker $\Omega$ igjen, og ved $n = N$ finnes bare én mikrotilstand — akkurat som ved $n = 0$.


**d)** Den røde søylen ($U = \langle E \rangle$, gjennomsnittet) og den høyeste søylen (mest sannsynlige $n$) faller bare sammen når fordelingen er symmetrisk. Ved liten $N$ eller energi langt fra $n = N/2$ blir $P(n)$ skjev, så gjennomsnittet trekkes mot den lange halen og lander litt ved siden av toppen. For $N \approx 6\cdot10^{23}$ blir fordelingen ekstremt smal og nærmest symmetrisk, så forskjellen mellom $U$ og mest sannsynlige energi blir forsvinnende liten.

### Oppgave 5

**a)** Når $E_i$ øker ved fast $T$, blir eksponenten $-E_i/k_BT$ mer negativ, så $P(E_i)$ synker. Høyere energitilstander er altså mindre sannsynlige.


**b)** Når $T$ øker ved fast $E_i$, går $E_i/k_BT$ mot null og $e^{-E_i/k_BT}$ mot 1, så tilstanden blir mer sannsynlig relativt til lavere tilstander. Fordelingen jevnes ut, slik at høyere energitilstander får en større andel.


**c)**

$\frac{P(E_2)}{P(E_1)} = e^{-(E_2-E_1)/k_BT}$ 

Når $T \to \infty$ går eksponenten mot 0 og forholdet mot 1, altså blir de to tilstandene like sannsynlige. Når $T \to 0$ går eksponenten mot $-\infty$ og forholdet mot 0, så den høyere tilstanden $E_2$ blir uendelig mye mindre sannsynlig enn $E_1$.


**d)** Ved svært lav temperatur befinner systemet seg så godt som alltid i grunntilstanden (lavest energi). Grunnen er at det knapt finnes termisk energi til å eksitere systemet, så alle høyere tilstander blir undertrykt av Boltzmann-faktoren.


**e)** Ved svært høy temperatur blir forskjellen i sannsynlighet mellom lave og høye energitilstander liten — alle tilgjengelige tilstander blir tilnærmet like sannsynlige. Den termiske energien er da så stor at energiforskjellene nesten ikke betyr noe.


**f)** Det er forholdet $E_i/k_BT$ som avgjør, fordi det er dimensjonsløst og sammenligner energien med den tilgjengelige termiske energien $k_BT$. En energi er bare "stor" eller "liten" i forhold til $k_BT$: samme $E_i$ kan være utilgjengelig ved lav $T$ og lett tilgjengelig ved høy $T$, så det er forholdet, ikke $E_i$ eller $T$ alene, som bestemmer sannsynligheten.

### Oppgave 7

**a)** Vi starter med

$\Omega(q,N) = \frac{(q+N-1)!}{q!\,(N-1)!}$

Vi tar så logaritmen av uttrykket og bruker at $\ln(\frac{a}{b}) = \ln a - \ln b$:

$\ln\Omega = \ln(q+N-1)! - \ln q! - \ln(N-1)!$

For store $q$ og $N$ bruker vi $q + N - 1 \approx q + N$ og $N - 1 \approx N$:

$\ln\Omega \approx \ln(q+N)! - \ln q! - \ln N!$

Deretter bruker vi Stirlings approksimasjon $\ln n! \approx n\ln n - n$ på hvert ledd, slik at

$\ln(q+N)! \approx (q+N)\ln(q+N) - (q+N)$

$\ln q! \approx q\ln q - q$

$\ln N! \approx N\ln N - N$

Vi setter inn hvert av leddene og får

$\ln\Omega \approx \big[(q+N)\ln(q+N) - (q+N)\big] - \big[q\ln q - q\big] - \big[N\ln N - N\big] = (q+N)\ln(q+N) - q\ln q - N\ln N$



bVi har to systemer i kontakt med total energi 

$q_\text{tot} = q_A + q_B \Longrightarrow q_B = q_\text{tot} - q_A$

Den totale multiplisiteten er 

$\Omega_\text{tot} = \Omega_A \cdot \Omega_B$

slik at vi får

$\ln\Omega_\text{tot} = \ln\Omega_A(q_A) + \ln\Omega_B(q_B)$

Likevektsbetingelsen er der $\ln\Omega_\text{tot}$ er maksimal, altså der den deriverte med hensyn på $q_A$ er null

$\frac{d}{dq_A}\ln\Omega_{tot} = 0$

Vi deriverer først ved bruk av kjerneregelen

$\frac{d}{dq_A}\ln\Omega_{tot} = \frac{d}{dq_A} \left( \ln\Omega_A(q_A) + \ln\Omega_B(q_B) \right) = \frac{d}{dq_A} \ln\Omega_A(q_A) + \frac{d}{dq_A} \ln\Omega_B(q_B) = \frac{d\ln\Omega_A(q_A)}{dq_A} + \frac{d\ln\Omega_B(q_B)}{dq_B}\cdot\frac{dq_B}{dq_A}$

og krever at uttryket skal være lik null

$\frac{d\ln\Omega_A(q_A)}{dq_A} + \frac{d\ln\Omega_B(q_B)}{dq_B}\cdot\frac{dq_B}{dq_A} = 0$

Siden $q_B = q_\text{tot} - q_A$ er $dq_B/dq_A = -1$, så

$\frac{d\ln\Omega_A(q_A)}{dq_A} + \frac{d\ln\Omega_B(q_B)}{dq_B}\cdot(-1) = 0 \Longrightarrow \frac{d\ln\Omega_A(q_A)}{dq_A} = \frac{d\ln\Omega_B(q_B)}{dq_B}$

Nå kan vi bruke approksimasjonen for $\ln\Omega$ vi fant i a), og deriverer uttrykkene før vi setter dem inn:

$\frac{d\ln\Omega(q)}{dq} = \frac{d}{dq} \big[(q+N)\ln(q+N) - q\ln q - N\ln N \big] = \frac{d}{dq} \big[(q+N)\big] \ln(q+N) - \frac{d}{dq} \big[ q\ln q \big] - \frac{d}{dq} \big[ N \ln N \big]$

Vi kan derivere hvert ledd for seg

$\frac{d}{dq} \big[ (q+N) \ln(q+N) \big] = \frac{d}{dq} \big[q+N\big] \cdot \ln(q+N) + \frac{d}{dq} \big[ \ln(q+N) \big] \cdot (q+N) = \ln(q+N) + \frac{q+N}{q+N}= \ln (q+N) + 1$

$\frac{d}{dq} \big[ q\ln q \big] = \frac{d}{dq} \big[q \big] \cdot (\ln q) + \frac{d}{dq} \big[\ln q] \cdot q = \ln q + \frac{q}{q} = \ln q + 1$

$\frac{d}{dq} \big[ N \ln N \big] = 0$

og setter dem inn igjen

$\frac{d\ln\Omega(q)}{dq} = \frac{d}{dq} \big[(q+N)\ln(q+N) - q\ln q - N\ln N \big] = \ln (q+N) + 1 - (\ln q + 1) = \ln (q+N) - \ln q = \ln \left( \frac{q+N}{q} \right)$

Vi setter inn den deriverte for A og B:

$\ln\!\left(\frac{q_A+N_A}{q_A}\right) = \ln\!\left(\frac{q_B+N_B}{q_B}\right) \Longrightarrow \frac{q_A+N_A}{q_A} = \frac{q_B+N_B}{q_B} \Longrightarrow \frac{q_A+N_A}{q_A}-1 = \frac{q_B+N_B}{q_B}-1 \Longrightarrow \frac{q_A+N_A-q_A}{q_A} = \frac{q_B+N_B-q_B}{q_B} \Longrightarrow \frac{N_A}{q_A} = \frac{N_B}{q_B} \Longrightarrow \frac{q_A}{N_A} = \frac{q_B}{N_B}$

Med definisjonen $T = q/N$ får vi ved likevekt:

$T_A = T_B$
