### Oppgave 1

**a)** At energien til et atom er kvantisert betyr at atomet bare kan ha bestemte energiverdier – ikke alle mulige verdier imellom. Vi kan tenke på energinivåene som trinn på en stige: atomet kan befinne seg på bestemte nivåer $E_0,E_1,E_2,\dots$, men ikke mellom dem.

**b)** Når et atom sender ut lys, går systemet fra et høyere til et lavere energinivå. Fotonet som sendes ut har en energi som er lik energiforskjellen mellom nivåene. **c)** En makrotilstand beskriver systemet med makroskopiske størrelser vi kan måle, for eksempel temperatur, trykk, volum eller total energi. En mikrotilstand er én bestemt, fullstendig mikroskopisk beskrivelse innenfor modellen. Mange forskjellige mikrotilstander kan gi samme makrotilstand.

**d)** En frihetsgrad er en uavhengig måte et system kan bevege seg eller fordele energi på. For molekyler skiller vi ofte mellom fire viktige **typer energibidrag**: translasjon, rotasjon, vibrasjon og elektronisk energi. De enkelte translasjons-, rotasjons- og vibrasjonsbevegelsene består igjen av én eller flere frihetsgrader.

Skjematisk kan vi skrive

$$E_\mathrm{tot}=E_\mathrm{trans}+E_\mathrm{rot}+E_\mathrm{vib}+E_\mathrm{elektronisk}$$

En mikrotilstand angir en bestemt kombinasjon av disse mikroskopiske tilstandene. Mange forskjellige kombinasjoner kan gi samme totale energi og dermed samme makrotilstand.

**e)** Statistisk mekanikk kobler et systems mikroskopiske tilstander til målbare, makroskopiske egenskaper ved hjelp av sannsynligheter og statistiske gjennomsnitt.

En empirisk lov er formulert på grunnlag av observerte sammenhenger. En statistisk lov følger av en statistisk beskrivelse av de underliggende mikroskopiske tilstandene. Det betyr ikke at empiriske lover bare beskriver og statistiske lover alltid forklarer alt, men statistisk mekanikk kan gi et mikroskopisk grunnlag for makroskopiske lover.

### Oppgave 7

**a)** Vi starter med

$$\Omega(q,N)=\frac{(q+N-1)!}{q!(N-1)!}$$

For store $q$ og $N$ bruker vi $q+N-1\approx q+N$ og $N-1\approx N$. Med Stirlings approksimasjon $\ln n!\approx n\ln n-n$ får vi

$$\ln\Omega\approx(q+N)\ln(q+N)-q\ln q-N\ln N$$

**b)** For to systemer med fast total energi er $q_B=q_\mathrm{tot}-q_A$, og

$$\ln\Omega_\mathrm{tot}=\ln\Omega_A+\ln\Omega_B$$

Ved maksimum er

$$\frac{d}{dq_A}\ln\Omega_\mathrm{tot}=0$$

Siden $dq_B/dq_A=-1$, får vi

$$\frac{d\ln\Omega_A}{dq_A}=\frac{d\ln\Omega_B}{dq_B}$$

Fra uttrykket i a) finner vi

$$\frac{d\ln\Omega}{dq}=\ln\left(\frac{q+N}{q}\right)$$

Dermed

$$\ln\left(\frac{q_A+N_A}{q_A}\right)=\ln\left(\frac{q_B+N_B}{q_B}\right)$$

som gir

$$\frac{q_A}{N_A}=\frac{q_B}{N_B}$$

I denne enkle modellen er $T$ proporsjonal med $q/N$, slik at likevekt innebærer $T_A=T_B$.

**c)** Med $N=8$ blir Stirlings tilnærming fortsatt ganske grov fordi $N$ ikke er særlig stor. Avviket i $\ln\Omega$ avtar når $q$ øker, men er fortsatt rundt 19 % ved $q=100$. Det illustrerer at Stirlings approksimasjon først blir virkelig god når fakultetene involverer store tall.
