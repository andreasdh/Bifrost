# Uke 10 – Likevekt

## Diskusjonsoppgaver

### Oppgave 1

**a)** Hvilke tre typer likevekter skiller vi mellom i dette kurset, og hva er drivkreftene bak dem?

**Svar** Vi skiller mellom termisk, mekanisk og kjemisk likevekt. Termisk likevekt drives av temperaturforskjeller, og det vil være en netto varmeoverføring frem til temperaturen er lik. Mekanisk likevekt drives av trykkforskjeller, og et bevegelig skille mellom to sider med ulikt trykk vil flytte seg til trykket er likt på begge sider. En kjemisk likevekt er drevet av forskjeller i kjemisk potensial. 

**b)** Fra et mikroskopisk perspektiv, forklar hvordan drivkreftene nevnt i a) fører til endring i systemet.

**Svar** 
1. Termisk likevekt: Temperatur er et mål på partiklenes gjennomsnittlige kinetiske energi. Varmestrøm er overføring av bevegelsesenergi gjennom kollisjoner.  

2. Trykk oppstår mikroskopisk fra partiklenes støt mot en vegg, som er summen av bevegelsesmengde som overføres per areal per tid. En trykkforskjell vil gi en større kraft i en retning på én side, så grensen vil skyves frem til kraften utlignes av den andre siden. 

3. Kjemisk potensial beskriver hvordan systemets energi endres når man endrer antallet av et specie. Det har både en energidel (bindinger og vekselvirkninger) og en entropidel (hvor mange måter partiklene kan fordeles på, altså konsentrasjon/blanding). Partikler har en tendens til å spre seg ut og til å søke lavere energi, begge deler er «bakt inn» i $\mu$. 

**c)** Et grunnleggende krav for innstilling av likevekt, som vi har sett på flere ganger, er maksimering av entropi. Når er ikke dette alene lenger nok til å definere likevekt?

**Forslag** Entropimaksimering av systemet alene er likevektskriteriet for et isolert system, der blant annet $U$, $V$ og stoffmengdene er konstante. Når systemet kan utveksle energi med omgivelsene, må vi i stedet betrakte entropien til system og omgivelser samlet. Ved konstant temperatur og volum er $\Delta S_{\mathrm{omg}}=-\Delta U_{\mathrm{sys}}/T$, og likevekten svarer til et minimum i Helmholtz-energien $A=U-TS$. Ved konstant temperatur og trykk er $\Delta S_{\mathrm{omg}}=-\Delta H_{\mathrm{sys}}/T$, og likevekten svarer til et minimum i Gibbs-energien $G=H-TS$. Mikroskopisk uttrykker dette konkurransen mellom energi og antallet tilgjengelige mikrotilstander.

**d)** Hvorfor kan vi si at alle de nevnte likevektene er dynamiske?

**Svar** Ved likevekt er det ingen netto endring i systemets makroskopiske egenskaper, men de mikroskopiske prosessene stopper ikke opp selv om likevekt er oppnådd, endringen skjer bare like raskt i begge retninger av likevekten.  

### Oppgave 2

Vi ser på en generell reaksjon $A_{(g)} \rightleftharpoons \ B_{(g)}$. Senere i oppgaven skal vi se på to ulike tilfeller. I tilfelle 1 er tettheten mellom energinivåene tilsvarende for A og B, noe vi f.eks. finner i reaksjoner hvor A og B er ulike cis/trans konfigurasjoner av samme molekyl. I tilfelle 2 ligger energinivåene i B langt tettere enn i A (et eksempel kan være at A er en stiv ringkonformasjon i likevekt med en mer fleksibel, ikke-planar konformasjon (B))
	
**a)** Hva kan vi si om ${\Delta }_{\mathrm{r}}G$ og likevektskonstanten, $K$, for reaksjoner: 
	
1.  Der reaksjonen $A_{(g)} \rightarrow \ B_{(g)}$ er spontan
	
2.  Der reaksjonen $B_{(g)} \rightarrow \ A_{(g)}$ er spontan 
	
3.  Ved likevekt

**Svar**

Sammenhengen mellom standard reaksjons-Gibbsenergi og likevektskonstanten er gitt ved følgende likninger:

$$\Delta_r G^\circ = -RT \ln K$$
$$K = e^{-\Delta_r G^\circ / RT}$$

Likevektskonstanten $K$ har en fast verdi ved en gitt temperatur (og ellers standardbetingelser), og forteller hvor likevekten ligger.

**Forslag** Det er viktig å skille mellom standard reaksjons-Gibbsenergi og reaksjonens Gibbsenergi ved den aktuelle sammensetningen. Sammenhengen $\Delta_rG^\circ=-RT\ln K$ sier hvor likevekten ligger, mens den aktuelle retningen bestemmes av $\Delta_rG=\Delta_rG^\circ+RT\ln Q$.

| Likevektsposisjon | $\Delta_r G^\circ$ | $K$ | Ved likevekt for en idealisert 1:1-reaksjon |
|---|:---:|:---:|---|
| Produktsiden favorisert | $< 0$ | $> 1$ | $a_B > a_A$ |
| Reaktantsiden favorisert | $> 0$ | $< 1$ | $a_B < a_A$ |
| $a_B=a_A$ ved likevekt | $= 0$ | $= 1$ | $a_B = a_A$ |

Ved en gitt sammensetning går reaksjonen spontant forover når $Q<K$ ($\Delta_rG<0$), spontant bakover når $Q>K$ ($\Delta_rG>0$), og er ved likevekt når $Q=K$ ($\Delta_rG=0$). For idealgasser kan aktiviteter uttrykkes ved dimensjonsløse partialtrykksforhold.


Om reaksjonen faktisk går forover i et gitt øyeblikk, bestemmes ikke av $K$ alene, men av reaksjonskvotienten, (Q).

$$\Delta_r G = \Delta_r G^\circ + RT\ln Q$$

altså av om $Q$ er mindre eller større enn $K$. For eksempel vil en reaksjon med $K < 1$ fortsatt danne produktet B en stund om man starter med ren $A$, siden $Q = 0 < K$ i starttidspunktet.


| Tilfelle | Hastigheter | Netto |
|---|---|---|
| Forover spontan | $v_\text{forover} > v_\text{bakover}$ | mot $B$ |
| Bakover spontan | $v_\text{bakover} > v_\text{forover}$ | mot $A$ |
| Likevekt | $v_\text{forover} = v_\text{bakover}$ | ingen netto endring |

Ved likevekt går reaksjonen fortsatt begge veier, like fort. Så konsentrasjonene står stille (dynamisk likevekt).

obs! merk at likevektskonstanten ikke er avhengig av trykk eller sammensetning, kun av temperatur. Endring i trykk eller sammensetning vil påvirke Q, ikke K. 


**b)** Hva er en reaksjonskvotient, og hvordan er den definert ved likevekt? 

**Svar**

$$Q = \frac{[B]}{[A]} \quad\text{(eventuelt med partialtrykk eller aktiviteter)}$$

$Q$ er forholdet mellom produkt og reaktant på et gitt tidspunkt. Ved å sammenligne $Q$ med $K$ ser man hvilken vei reaksjonen går:

- $Q < K \;\Rightarrow\;$ netto fremover reaksjon
- $Q > K \;\Rightarrow\;$ netto bakover reaksjon
- $Q = K \;\Rightarrow\;$ likevekt


**Forslag** La $q_i^{\mathrm{int}}$ være den interne partisjonsfunksjonen for art $i$, målt relativt til dens egen grunntilstand:

$$q_i^{\mathrm{int}}=\sum_j e^{-(\varepsilon_{j,i}-\varepsilon_{0,i})/kT}=1+e^{-(\varepsilon_{1,i}-\varepsilon_{0,i})/kT}+e^{-(\varepsilon_{2,i}-\varepsilon_{0,i})/kT}+\cdots$$

Partisjonsfunksjonen som også inkluderer grunntilstandsenergien er $q_i=e^{-\varepsilon_{0,i}/kT}q_i^{\mathrm{int}}$. I den forenklede modellen der øvrige bidrag er like for A og B, er kjemisk potensial:

$$\mu_i = kT \ln{\frac{N_i}{q_i}}+\text{(felles ledd)}$$

**c)** Uttrykk likevektskonstanten K som en funksjon av partisjonsfunksjonene for A og B.

**Forslag** Ved likevekt er $\mu_A=\mu_B$. Med uttrykket for kjemisk potensial fra del b får vi, når felles ledd kanselleres, $N_A/q_A=N_B/q_B$. Dermed blir likevektsforholdet

$$K=\frac{N_B}{N_A}=\frac{q_B}{q_A}=\frac{q_B^{\mathrm{int}}}{q_A^{\mathrm{int}}}e^{-(\varepsilon_{0,B}-\varepsilon_{0,A})/kT}$$

Her er $q_i^{\mathrm{int}}$ partisjonsfunksjonen relativt til grunntilstanden, mens $q_i$ også inkluderer grunntilstandsenergien. Uttrykket gjelder for den idealiserte modellen i oppgaven; i en fullstendig gassberegning må relevante translasjons- og standardtilstandsfaktorer også tas med.


**d)** Bruk svaret fra forrige oppgave til å forklare hvorfor specie A er favorisert dersom avstanden mellom energinivåene er omtrent like, men nivåene på grunntilstandene er ulike, $\varepsilon_{0,A} < \varepsilon_{0,B}$.

**Forslag** Når avstanden mellom energinivåene over grunntilstanden er omtrent lik for A og B, er $q_A^{\mathrm{int}}\approx q_B^{\mathrm{int}}$. Da blir $K\approx e^{-(\varepsilon_{0,B}-\varepsilon_{0,A})/kT}$. Siden $\varepsilon_{0,B}>\varepsilon_{0,A}$, gir Boltzmann-faktoren lavere populasjon av B enn A, og A favoriseres.

**e)** Hvorfor favoriseres specie B når tettheten av tilstander i B er langt større enn i A? (Diskuter ved å trekke inn partisjonsfunksjone - stor når mange tilstander er tilgjengelige!)
	
**Forslag** En større tetthet av energinivåer gir en større relativ partisjonsfunksjon, fordi flere tilstander bidrar til summen av Boltzmann-faktorene. Fra uttrykket i del c følger da at $q_B^{\mathrm{int}}/q_A^{\mathrm{int}}$ kan bli så stor at den oppveier energistraffen $e^{-(\varepsilon_{0,B}-\varepsilon_{0,A})/kT}$. B kan derfor bli favorisert selv om grunntilstanden ligger høyere i energi.

**Svar**

### Oppgave 3

**a)** Hva sier Le Chateliers prinsipp?

**Svar** Når et system i likevekt påføres en forstyrrelse – endring av temperatur, trykk/volum eller sammensetning – vil systemet svare ved å flytte likevekten i den retningen som **motvirker forstyrrelsen**, til en ny likevekt er innstilt.
En forstyrrelse vil gi  $\mu_A ≠ \mu_B$, og systemet vil svare ved å redusere ${\Delta_\mathrm{r}G}$ til null igjen (ved konstant trykk og temperatur).

**Forslag** Likevektskonstanten er definert ved valgte standardtilstander og er dimensjonsløs. Ved en gitt reaksjon og standardtilstand er det temperaturen som endrer $K$; endringer i sammensetning eller trykk endrer $Q$ og kan dermed flytte likevektsposisjonen, men endrer ikke $K$.


**b)** Hvordan kan vi benytte Le Chateliers prinsipp til å manipulere eksergone/endergone prosesser? 

**Forslag**

${\Delta_\mathrm{r}G} = {\Delta_\mathrm{r}G}^{\circ} + RT\ln Q$ viser hvordan ulike påvirkninger kan endre reaksjonsretningen eller likevektsposisjonen:

1. **Sammensetning:** Å fjerne produkt løpende (felling, destillering, Dean–Stark) eller å tilsette reaktant kan holde $Q<K$, slik at $\Delta_\mathrm{r}G<0$ og reaksjonen drives forover selv om $\Delta_\mathrm{r}G^\circ>0$. Tilsetting av produkt øker $Q$ og kan gjøre $Q>K$, slik at $\Delta_\mathrm{r}G>0$ og netto reaksjon går bakover. Sammensetningsendringer påvirker $Q$ og $\Delta_\mathrm{r}G$, ikke $\Delta_\mathrm{r}G^\circ$ eller $K$ ved konstant temperatur.

2. For gasser kan man endre trykk eller volum: Endring av $p$ eller $V$ endrer partialtrykkene og dermed $Q$ hvis $\Delta n_\mathrm{gass} \ne 0$; systemet svarer med en endring som favoriserer siden med færrest partikler ved komprimering (se e).

3. **Temperatur:** Fra van't Hoff-ligningen har vi $\dfrac{d\ln K}{dT} = \dfrac{\Delta_\mathrm{r}H^\circ}{RT^2}$. Oppvarming øker $K$ for en endoterm reaksjon ($\Delta_\mathrm{r}H^\circ>0$) og minker $K$ for en eksoterm reaksjon ($\Delta_\mathrm{r}H^\circ<0$).

**c)** I en esterifisering av karboksylsyrer (karboksylsyre + alkohol $\rightleftharpoons$ ester + $H_2O$) brukes ofte et Dean-Stark apparat til å kontinuerlig fjerne vanndamp. Forklar hvordan dette påvirker likevekten ved å holde prosessen eksergon. 

**Svar**  Uten fjerning av vann vil $\Delta_\mathrm{r}G$ gå mot null når $Q \to K$, og esterifiseringen stopper ved en endelig blanding (både syre og alkohol kan da være igjen i store mengder). Med Dean-Stark oppsett fjernes vand fortløpende og $ $Q \ll K$ → ${\Delta_\mathrm{r}G} = {\Delta_\mathrm{r}G}^{\circ} + RT\ln Q < 0$. Systemet vil i teorien da aldri oppnå likevekt og så lenge vannet fjernes og det er reaktanter tilstede vil esterifiseringen gå. I praksis vil man nå en grense hvor utbytte av å fortsette reaksjonen blir svært liten i forhold til hastigheten på reaksjonen. 


**d)** Hvordan ville reaksjonene i forrige oppgave blitt påvirket av en temperaturøkning?

**Forslag** I modellen i oppgaven ligger grunntilstanden til B høyere enn den til A, slik at reaksjonen A $\rightarrow$ B behandles som endoterm. Da vil en temperaturøkning forskyve likevekten mot B. Boltzmann-faktoren $e^{-(\varepsilon_{0,B}-\varepsilon_{0,A})/kT}$ øker når temperaturen øker. I scenario 2 bidrar også den større tettheten av tilstander i B til å favorisere B. Generelt bestemmes temperaturvirkningen av reaksjonsentalpien, slik van't Hoff-ligningen uttrykker.

**e)**  For en reaksjon $A \ \rightleftharpoons 2 \, B\ _{(g)}$, hvordan påvirkes likevekten av økt/senket trykk? Hvorfor er det sånn?

**Forslag** 

Fra Le Chateliers prinsipp har vi at komprimering (altså økt totaltrykk) presser likevekten mot **A** (færrest partikler), mens lavere trykk favoriserer **B**.

Dette kan man også se utifra likevekstkonstanten og reaksjonskvotienten:

$Q = \dfrac{(p_B/p^\circ)^2}{p_A/p^\circ}$. 
 
Med totaltrykk $P$: $p_B = x_B P$, $p_A = (1-x_B)P$, får vi 

$$Q = \dfrac{(x_B P/p^\circ)^2}{(1-x_B)P/p^\circ} =\dfrac{x_B^2 P}{(1-x_B) \cdot p^\circ} $$

$$\frac{x_B^2}{1-x_B} = \frac{K\,p^\circ}{P}$$

Ved konstant temperatur er $K$ uavhengig av trykket. Uttrykket over viser at når totaltrykket økes ved komprimering, må likevektsverdien av $x_B$ minke. Likevekten forskyves dermed mot venstre, siden venstresiden har færre mol gass.



**f)** Hvordan påvirker en katalysator eller et enzym en likevekt? Begrunn svaret deres.

**Svar**

Katalysatorer/enzymer påvirker ikke likevekten direkte. De senker *aktiveringsenergien* ved å f.eks. stabilisere en overgangstilstand,  og vil dermed påvirke reaksjonshastigheten, ikke likevekten.

### Oppgave 1 

Ved 3000 K i en industriell prosess er partialtrykkene ved likevekt for $CO_2$, $CO$, og $O_2$ henholdsvis $0.6$, $0.4$, og $0.2$ atm. Beregn $\Delta G_m^\circ$ (ved 3000 K) for reaksjonen

$$2 CO_2(g) \rightleftharpoons 2 CO (g) + O_2 (g)$$

**Svar**

Vi vet at standard trykk, $p_0$, er 1 bar. Vi må derfor konvertere partialtrykkene fra atm til bar med konverteringsfaktoren $1 atm = 1.01325 bar$. Likevektskonstanten kan vi da uttrykke ved partialtrykk på følgende måte:

$$K_p = \frac{(\frac{p_{CO}}{1})^2\cdot(\frac{p_{O_2}}{1})}{(\frac{p_{O_2}}{1})^2} = \frac{(0.4\cdot1.01325)^2\cdot (0.2\cdot1.01325)}{(0.6\cdot1.01325)^2} = 0.0901$$

Finner så $\Delta G_{m}^\circ = -RT \ln{K_p} = -8.31446 J K^{-1} mol^{-1} \cdot 3000 K \cdot \ln{0.091} = 60.0 kJ/mol $



### Oppgave 2

Beregn følgende størrelser for reaksjonen 

$$C_2H_4(g) + H_2(g) \rightleftharpoons C_2H_6(g)$$

**a)** $\Delta_rG^\circ$, $\Delta_rH^\circ$, $\Delta_rS^\circ$ ved standardbetingelser. Angi referansetilstand.

**Svar**

Vi ønsker å finne endringen i Gibbs energi, entalpi og entropi for reaksjonen ved standard betingelser - og kan derfor benytte oss av Hess' lov og tabulerte verdier. For referansetilstanden på 1 bar (standard trykk) finner vi:

$$\Delta_rG^\circ = \Delta_fG^\circ (C_2H_6(g)) - \Delta_fG^\circ (H_2(g)) - \Delta_fG^\circ (C_2H_4(g)) = (-32-0-68.4) kJ/mol = -100.4 kJ/mol$$

$$\Delta_rH^\circ = \Delta_fH^\circ (C_2H_6(g)) - \Delta_fH^\circ (H_2(g)) - \Delta_fH^\circ (C_2H_4(g)) = (-84.0-0-52.4) kJ/mol = -136.4 kJ/mol$$

$$\Delta_rG^\circ = \Delta_rH^\circ -T\Delta_rS^\circ \rightarrow \Delta_rS^\circ = \frac{\Delta_rH^\circ-\Delta_rG^\circ}{T} = \frac{(-136.4+100.4)kJ/mol}{298.15K} = -120.8 J/(mol\cdot K) $$


**b)** $K_p$ ved $298.15 K$.

**Svar**

Ved referansetilstand 1 bar har vi at $K_p≡K$. Dermed kan vi bruke endringen i Gibbs frie energi for å finne $K_p$:

$$\Delta_rG^\circ = -RT\ln{K_p}$$

$$\ln{K_p} = \frac{-\Delta_rG^\circ}{RT} = \frac{100.4 \cdot 10^3 J mol^{-1}}{8.31446 J mol^{-1}K^{-1}\cdot 298.15 K} = 40.501$$

$$K_p = e^{\ln{K_p}} = e^{40.501} = 3.89\cdot 10^{17}$$

Altså er reaksjonen drevet enormt mot produktet.

**c)** $K_c$ ved $298.15 K$, referansetilstand er 1M.

**Svar** 

For å finne likevektskonstanten ved referansetilstand 1 M kan vi bruke definisjonen for likevektskonstanten gitt ved aktivitetene (se Atkins side 210/211):

$$K = (\prod_j{a_j^{\nu_j}})_eq$$

Når vi jobber med konsentrasjoner kan vi tilnærme aktiviteten i likevektsuttrykket med med molar konsentrasjon:

$$a_j = \frac{[j]}{c^\circ}$$

Vi får da

$$K_c = \frac{\frac{[C_2H_6]}{c^\circ}}{\frac{[C_2H_4]}{c^\circ}\cdot \frac{[H_2]}{c^\circ}}$$

Dette kan vi relatere til likevektskonstanten K_p ved å anta ideell oppførsel (husk også at vi jobber med molar konsentrasjon!):

$$pV = n_jRT \rightarrow [j] = \frac{n_j}{V} = \frac{p}{RT}$$

$$K_c = \frac{\frac{p_{C_2H_6}}{RT}}{\frac{p_{C_2H_4}}{RT}\cdot \frac{p_{H_2}}{RT}}\cdot c^\circ = \frac{p_{C_2H_6}}{p_{C_2H_4}\cdot p_{H_2}}\cdot c^\circ \cdot RT$$

Setter inn for $K_p$

$$K_c = \frac{p_{C_2H_6}\cdot p^\circ}{p_{C_2H_4}\cdot p_{H_2}} \cdot \frac{RTc^\circ}{p^\circ} = K_p \frac{RTc^\circ}{p^\circ} $$

$$K_c = 3.89\cdot 10^{17}\cdot \frac{0.08315 bar \cdot dm^3\cdot K^{-1}\cdot mol^{-1}\cdot 298.15K \cdot 1 mol\cdot dm^{-3}}{1 bar } = 9.64 \cdot 10^{18}$$



**d)** $\Delta_rG^\circ$, referansetilstand er 1M.

**Svar**

Bruker igjen sammenhengen mellom Gibbs frie energi og likevektskonstanten:

$$\Delta_rG^\circ (ved 1M) = -RT\ln{K_c} = -8.31446 J K^{-1} mol^{-1} \cdot 298.15 K \cdot \ln{9.64 \cdot 10^{18}} = -108.4 kJ/mol$$



**e)** $K_p$ ved $373.15 K$ (anta at entalpi og entropi er temperatur-uavhengig.)


**Svar**

$$\Delta_rG^\circ (100^\circ C) = \Delta_rH^\circ(25^\circ C) -373.15K\cdot \Delta_rS^\circ(25^\circ C)$$

$$\Delta_rG^\circ (100^\circ C) = -136.4 kJ/mol -373.15K\cdot (-120.8 J/(K\cdot mol)) = 91.36 kJ/mol$$

Ved referansetilstand 1 bar:

$$\ln{K_p} = -\frac{\Delta_rG^\circ (100^\circ C)}{RT} = \frac{91.36 \cdot 10^3 J/mol}{8.31446 J/(K\cdot mol)\cdot 373.15 K} = 29.447$$

$$K_p = e^{\ln{K_p}} = 6.15\cdot 10^{12}$$

En temperaturøkning skyver likevekten noe mindre mot produktet, prosessen er eksoterm.

### Oppgave 3

Fosforpentaklorid, $PCl_5$, dissosierer i gassfase til fosfortriklorid, $PCl_3$, og klorgass, $Cl_2$:

$${PCl_5}{(g)}\rightleftharpoons {PCl_3}{(g)} + {Cl_2}{(g)}$$

Ved $250\,°C$ og trykket $1.00\,bar$ er gassens tetthet $2.66\,g\,dm^{-3}$. Finn dissosiasjonsgraden, og likevektskonstantene $K_c$, $K_p$, og $K_x$ ved disse betingelsene. I hvilken retning forskyves likevekten hvis trykket økes ved å tilsette $Ar{(g)}$?

**Svar**
$$\mathrm{PCl}_5(g) \;\rightleftharpoons\; \mathrm{PCl}_3(g) + \mathrm{Cl}_2(g)$$

Ved $250\ ^{\circ}\mathrm{C} = 523{.}15\ \mathrm{K}$ har vi $p = 1{.}00\ \mathrm{bar}, \qquad \rho = 2{.}66\ \mathrm{g/dm^3}$

Molmassene er $M_{\mathrm{PCl}_5} = 208{.}24$, $\ M_{\mathrm{PCl}_3} = 137{.}33$, $\ M_{\mathrm{Cl}_2} = 70{.}91\ \mathrm{g/mol}$.

Setter opp likevekten med dissosiasjonsgrad α:

Vi starter med $n_0$ mol PCl₅ og lar en andel, $\alpha$, dissosiere:

| | PCl₅ | PCl₃ | Cl₂ |
|---|---|---|---|
| Start | $n_0$ | $0$ | $0$ |
| Likevekt | $n_0(1-\alpha)$ | $n_0\alpha$ | $n_0\alpha$ |

Totalt antall mol er $\;n_\mathrm{tot} = n_0(1+\alpha)$. Merk at hvert PCl₅ som spaltes gir to molekyler, så gassmengden øker.

Finner så α fra tettheten:

Massen er bevart under spaltingen, så gjennomsnittlig molmasse blir

$$\bar{M} = \frac{m}{n_\mathrm{tot}} = \frac{n_0\,M_{\mathrm{PCl}_5}}{n_0(1+\alpha)} = \frac{M_{\mathrm{PCl}_5}}{1+\alpha}$$

Idealgassloven gir også ($\rho = m/V$ og $pV = nRT$):

$$\bar{M} = \frac{\rho RT}{p} = \frac{2{.}66 \cdot 0{.}08314 \cdot 523{.}15}{1{.}00} = 115{.}7\ \mathrm{g/mol}$$

Løser for α:

$$1+\alpha = \frac{M_{\mathrm{PCl}_5}}{\bar{M}} = \frac{208{.}24}{115{.}7} = 1{.}80 \quad\Rightarrow\quad {\alpha = 0{.}80}$$

Finner så molfraksjonene:

$$x_{\mathrm{PCl}_5} = \frac{1-\alpha}{1+\alpha} = \frac{0{.}20}{1{.}80} = 0{.}111$$

$$x_{\mathrm{PCl}_3} = x_{\mathrm{Cl}_2} = \frac{\alpha}{1+\alpha} = \frac{0{.}80}{1{.}80} = 0{.}444$$

Likevektskonstanten gitt ved molfraksjoner, Kₓ, blir da

$$K_x = \frac{x_{\mathrm{PCl}_3}\,x_{\mathrm{Cl}_2}}{x_{\mathrm{PCl}_5}} = \frac{\alpha^2}{1-\alpha^2} = \frac{0{.}80^2}{1-0{.}80^2} = \frac{0{.}64}{0{.}36} = {1{.}78}$$

Og gitt ved partialtrykkene (K_p): 

Med $p_i = x_i\,p$ og $p = 1{.}00\ \mathrm{bar}$:

$$p_{\mathrm{PCl}_3} = p_{\mathrm{Cl}_2} = 0{.}444\ \mathrm{bar}, \qquad p_{\mathrm{PCl}_5} = 0{.}111\ \mathrm{bar}$$

$$K_p = \frac{(p_{\mathrm{PCl}_3}/p^\circ)(p_{\mathrm{Cl}_2}/p^\circ)}{p_{\mathrm{PCl}_5}/p^\circ} = \frac{0{.}444^2}{0{.}111} = {1{.}78} \qquad (p^\circ = 1\ \mathrm{bar})$$


Total konsentrasjon er intensiv, så vi trenger ikke velge noe volum, og likevektskonstanten gitt ved konsentrasjoner blir da

$$c_\mathrm{tot} = \frac{p}{RT} = \frac{1{.}00}{0{.}08314 \cdot 523{.}15} = 0{.}0230\ \mathrm{mol/dm^3}$$

$$[\mathrm{PCl}_3] = [\mathrm{Cl}_2] = 0{.}444\,c_\mathrm{tot} = 0{.}01022\ \mathrm{mol/dm^3}, \qquad [\mathrm{PCl}_5] = 0{.}111\,c_\mathrm{tot} = 0{.}002554\ \mathrm{mol/dm^3}$$

$$K_c = \frac{([\mathrm{PCl}_3]/c^\circ)([\mathrm{Cl}_2]/c^\circ)}{[\mathrm{PCl}_5]/c^\circ} = \frac{0{.}01022^2}{0{.}002554} = {0{.}0409} \qquad (c^\circ = 1\ \mathrm{mol/dm^3})$$

Sjekker at likevektskonstantene ved konsentrasjon og trykk følger sin kjente sammenheng: $K_c = K_p\,\left(\dfrac{p^\circ}{c^\circ RT}\right)^{\Delta n} = 1{.}78 \cdot 0{.}0230 = 0{.}0409$ 

Hva skjer når man tilsetter Ar?

Når Ar tilsettes for å øke trykket, skjer det ved konstant volum og temperatur. Partialtrykkene til de reagerende gassene avhenger bare av deres egne stoffmengder og volumet:

$$p_i = \frac{n_i RT}{V}$$

Ar endrer verken $n_{\mathrm{PCl}_5}$, $n_{\mathrm{PCl}_3}$, $n_{\mathrm{Cl}_2}$ eller $V$, så alle partialtrykkene er uendret. Da er reaksjonskvotienten fortsatt lik $K_p$:

$$Q = \frac{(p_{\mathrm{PCl}_3}/p^\circ)(p_{\mathrm{Cl}_2}/p^\circ)}{p_{\mathrm{PCl}_5}/p^\circ} = K_p$$

Likevekten forskyves ikke, Ar øker det *totale* trykket, men ikke partialtrykkene til reaktantene/produktene.

$$\alpha = 0{.}80 \qquad K_c = 0{.}0409 \qquad K_p = 1{.}78 \qquad K_x = 1{.}78$$



### Oppgave 4

Nitrogendioksid, $NO2$, forekommer i likevekt med dinitrogentetraoksid, $N_2O_4$:

$${N_2O_4}{(g)}\rightleftharpoons 2 \, {NO_2}{(g)}$$

Ved $25 ^{\circ} C$ og $0.597 \, bar$ er gassens tetthet $1.477 \, g.dm^{-3}$. Finn dissosiasjonsgraden, og likevektskonstantene $K_c$, $K_P$, og $K_x$ ved disse betingelsene. I hvilken retning forskyves likevekten hvis trykket økes ved å tilsette $He{(g)}$?

**Svar** 

Setter først opp en oversikt over informasjonen gitt i teksten: 

$T = 25\ ^{\circ}\text{C} = 298{.}15\ \text{K}$, $\ p = 0{.}597\ \text{bar}$, $\ \rho = 1{.}477\ \text{g/dm}^3$
$M_{\text{N}_2\text{O}_4} = 92{.}02\ \text{g/mol}$, $\quad M_{\text{NO}_2} = 46{.}01\ \text{g/mol}$

Setter så opp likevekten med dissosiasjonsgrad α:

Vi starter med $n_0$ mol N₂O₄ og lar en andel $\alpha$ dissosiere:

| | N₂O₄ | NO₂ |
|---|---|---|
| Start | $n_0$ | $0$ |
| Endring | $-n_0\alpha$ | $+2n_0\alpha$ |
| Likevekt | $n_0(1-\alpha)$ | $2n_0\alpha$ |

Totalt antall mol: $\;n_\text{tot} =n_0(1-\alpha)+ 2n_0\alpha= n_0 + 2n_0\alpha - n_0\alpha = n_0(1+\alpha)$.

Finner α fra tettheten:

Massen er bevart når N₂O₄ spaltes, så gjennomsnittlig molmasse blir

$$\bar{M} = \frac{m}{n_\text{tot}} = \frac{n_0\,M_{\text{N}_2\text{O}_4}}{n_0(1+\alpha)} = \frac{M_{\text{N}_2\text{O}_4}}{1+\alpha}$$

Samtidig gir idealgassloven ($\rho = \tfrac{m}{V} = \frac{mp}{nRT} = \frac{Mp}{RT}$):

$$\bar{M} = \frac{\rho RT}{p} = \frac{1{.}477 \cdot 0{.}08314 \cdot 298{.}15}{0{.}597} = 61{.}3\ \text{g/mol}$$

Løser for α:

$$1+\alpha = \frac{M_{\text{N}_2\text{O}_4}}{\bar{M}} = \frac{92{.}02}{61{.}3} = 1{.}50 \quad\Rightarrow\quad {\alpha = 0{.}50}$$


Molfraksjonene blir da:

$$x_{\text{N}_2\text{O}_4} = \frac{1-\alpha}{1+\alpha} = \frac{0{.}5}{1{.}5} = \tfrac{1}{3} = 0{.}333$$

$$x_{\text{NO}_2} = \frac{2\alpha}{1+\alpha} = \frac{1{.}0}{1{.}5} = \tfrac{2}{3} = 0{.}667$$

Kₓ (fra molfraksjoner):

$$K_x = \frac{x_{\text{NO}_2}^2}{x_{\text{N}_2\text{O}_4}} = \frac{(2/3)^2}{1/3} = \frac{4}{3} = {1{.}33}$$

K_p (fra partialtrykk):

$$p_{\text{NO}_2} = x_{\text{NO}_2}\,p = 0{.}667 \cdot 0{.}597 = 0{.}398\ \text{bar}, \qquad p_{\text{N}_2\text{O}_4} = 0{.}333 \cdot 0{.}597 = 0{.}199\ \text{bar}$$

$$K_p = \frac{(p_{\text{NO}_2}/p^\circ)^2}{p_{\text{N}_2\text{O}_4}/p^\circ} = \frac{0{.}398^2}{0{.}199} = {0{.}80} \qquad (p^\circ = 1\ \text{bar})$$



K_c (fra konsentrasjoner):

$$c_\text{tot} = \frac{p}{RT} = \frac{0{.}597}{0{.}08314 \cdot 298{.}15} = 0{.}0241\ \text{mol/dm}^3$$

$$[\text{NO}_2] = x_{\text{NO}_2}\,c_\text{tot} = 0{.}01605\ \text{mol/dm}^3, \qquad [\text{N}_2\text{O}_4] = x_{\text{N}_2\text{O}_4}\,c_\text{tot} = 0{.}00803\ \text{mol/dm}^3$$

$$K_c = \frac{([\text{NO}_2]/c^\circ)^2}{[\text{N}_2\text{O}_4]/c^\circ} = \frac{0{.}01605^2}{0{.}00803} = {0{.}0321} \qquad (c^\circ = 1\ \text{mol/dm}^3)$$

Hva skjer når man tilsetter He?

Når He tilsettes for å øke trykket, skjer det ved konstant volum og temperatur. Partialtrykkene til de reagerende gassene avhenger bare av deres egne stoffmengder og volumet:

$$p_{\text{NO}_2} = \frac{n_{\text{NO}_2}RT}{V}, \qquad p_{\text{N}_2\text{O}_4} = \frac{n_{\text{N}_2\text{O}_4}RT}{V}$$

He endrer verken $n_{\text{NO}_2}$, $n_{\text{N}_2\text{O}_4}$ eller $V$, så disse partialtrykkene er uendret. Da er reaksjonskvotienten fortsatt lik $K_p$:

$$Q = \frac{(p_{\text{NO}_2}/p^\circ)^2}{p_{\text{N}_2\text{O}_4}/p^\circ} = K_p$$

Likevekten forskyves ikke, He øker det *totale* trykket, men ikke partialtrykkene til reaktantene/produktene. Igjen, husk at det kun er temperatur som kan endre likevektskonstanten K. 


$$\alpha = 0{.}50 \qquad K_c = 0{.}032 \qquad K_p = 0{.}80 \qquad K_x = 1{.}33$$




### Oppgave 5

Kvikksølv er ved romtemperatur en væske som står i likevekt med damp etter likningen $Hg{(l)} \rightleftharpoons Hg{(g)}$. Fordampningsentalpien $\Delta _{vap}H^{\circ}$ er $62.0 \, kJ\cdot mol^{-1}$ og fordampningsentropien $\Delta _{vap}S^{\circ}$ er $99.0 \, J\cdot mol^{-1}.K^{-1}$ ($25 ^{\circ}C$). 

**a)** Beregn (likevekts) damptrykket av kvikksølv ved $25 ^{\circ}C$.

**Svar**

Regner først ut standard Gibbs-energi for fordampningen:

$$\Delta_\text{vap}G^\circ = \Delta_\text{vap}H^\circ - T\,\Delta_\text{vap}S^\circ = 62{.}0\cdot 10^3 - 298{.}15\cdot 99{.}0 = +32{.}5\cdot 10^3\ \mathrm{J/mol}$$

At denne er positiv gir mening: ved standardtrykk (1 bar) er fordampning ikke spontan — Hg er væske ved romtemperatur.

Likevektskonstanten er $K_p = p_{\text{Hg}}/p^\circ$ (ren væske har aktivitet 1), så

$$\Delta_\text{vap}G^\circ = -RT\ln K_p \quad\Rightarrow\quad \ln\frac{p_{\text{Hg}}}{p^\circ} = -\frac{\Delta_\text{vap}G^\circ}{RT} = -\frac{32{.}5\cdot 10^3}{8{.}3145\cdot 298{.}15} = -13{.}10$$

$${p_{\text{Hg}} = e^{-13{.}10}\cdot p^\circ = 2{.}0\cdot 10^{-6}\ \mathrm{bar}}$$

(≈ 0,2 Pa — svært lavt, men nok til at kvikksølvdamp er en reell helsefare i lukkede rom.)

**b)** Regn ut tilnærmet kokepunkt for kvikksølv ved $1 \, atm$ lufttrykk. Anta at $\Delta_{vap}H^{\circ}$ og $\Delta_{vap}S^{\circ}$ er temperaturuavhengige fra $25 ^{\circ}C$ til kokepunktet.

$$\text{Hg}(l) \;\rightleftharpoons\; \text{Hg}(g)$$

Gitt ved $25\ ^{\circ}\mathrm{C} = 298{.}15\ \mathrm{K}$:
$$\Delta_\text{vap}H^\circ = 62{.}0\ \mathrm{kJ/mol}, \qquad \Delta_\text{vap}S^\circ = 99{.}0\ \mathrm{J\,K^{-1}mol^{-1}}$$


**Svar**

Ved kokepunktet står væske og damp i likevekt med $p_{\text{Hg}} = 1\ \mathrm{atm} \approx 1\ \mathrm{bar} = p^\circ$. Da er $K_p = 1$ og dermed

$$\Delta_\text{vap}G^\circ(T_\text{kp}) = -RT_\text{kp}\ln 1 = 0$$

Med antakelsen om at $\Delta_\text{vap}H^\circ$ og $\Delta_\text{vap}S^\circ$ er temperatur-uavhengige:

$$0 = \Delta_\text{vap}H^\circ - T_\text{kp}\,\Delta_\text{vap}S^\circ \quad\Rightarrow\quad T_\text{kp} = \frac{\Delta_\text{vap}H^\circ}{\Delta_\text{vap}S^\circ} = \frac{62{.}0\cdot 10^3}{99{.}0}$$

$${T_\text{kp} = 626{.}3\ \mathrm{K} \approx 353\ ^{\circ}\mathrm{C}}$$

Til sammenligning er det målte kokepunktet 357 °C. Med tanke på at vi gjorde en ganske grov antakelse med konstante $\Delta H$ og $\Delta S$ i hele temperaturintervallet, er estimasjonen av kokepunktet ganske god.

$$p_{\text{Hg}}(25\ ^{\circ}\mathrm{C}) = 2{.}0\cdot 10^{-6}\ \mathrm{bar}, \qquad T_\text{kp} = 626{.}3\ \mathrm{K}\ (353\ ^{\circ}\mathrm{C})$$
