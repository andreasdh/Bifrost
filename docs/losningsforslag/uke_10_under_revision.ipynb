**a)** Hvilke tre typer likevekter skiller vi mellom i dette kurset, og hva er drivkreftene bak dem?

**Svar** Vi skiller mellom termisk, mekanisk og kjemisk likevekt. Termisk likevekt drives av temperaturforskjeller, og det vil være en netto varmeoverføring frem til temperaturen er lik. Mekanisk likevekt drives av trykkforskjeller, og et bevegelig skille mellom to sider med ulikt trykk vil flytte seg til trykket er likt på begge sider. **a)** Hva kan vi si om ${\Delta }_{\mathrm{r}}G$ og likevektskonstanten, $K$, for reaksjoner: 
	
1)  Der reaksjonen $A_{(g)} \rightarrow \ B_{(g)}$ er spontan
	
2)  Der reaksjonen $B_{(g)} \rightarrow \ A_{(g)}$ er spontan 
	
3)  Ved likevekt
	
**b)** Hva er en reaksjonskvotient, og hvordan er den definert ved likevekt? 

Vi har tidligere sett på partisjonsfunksjonen, som vi kan definere som $q'=\sum_{j=0}^t{e^{-\varepsilon_j/kT}}$, hvor $\varepsilon_0$ er energien til grunntilstanden ($q'$ viser at grunntilstanden er eksplisitt definert i denne partisjonsfunksjonen da den kan være ulik mellom populasjon A og B). Ved å faktorere ut grunntilstanden ($q=e^{-\varepsilon_0(/kT)}q'$) kan vi skrive partisjonsfunksjonen som en funksjon av den relative avstanden mellom energinivåene $j≠0$ og grunntilstanden: 

$$q =e^{-\varepsilon_0(/kT)}q' = 1 + e^{-(\varepsilon_1-\varepsilon_0)/kT}+e^{-(\varepsilon_2-\varepsilon_0)/kT}+...+e^{-(\varepsilon_t-\varepsilon_0)/kT} $$

Og vi kan da skrive det kjemiske potensialet, uttrykt ved $q'$ som:

$$\mu_A = kT \ln{\frac{q'_A}{N_A}}$$

**c)** Uttrykk likevektskonstanten K som en funksjon av partisjonsfunksjonene for A og B.

**Svar**

$$\mu_A = \mu_B$$

$$K = \frac{N_B}{N_A} = \frac{q'_B}{q'_A} = \frac{q_B}{q_A}e^{-(\varepsilon_{0,B}-\varepsilon_{0,A})/kT}$$
