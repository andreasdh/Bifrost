## Diskusjonsoppgaver

### Oppgave 1

**a)**
Makroskopisk: Her beskriver vi likevekt som en tilstand der de målbare størrelsene ikke lenger endrer seg: trykk, temperatur, volum og konsentrasjoner holder seg konstante over tid. Utenfra ser systemet helt "stille" ut, hvor ingenting ser ut til å skjer. 

Mikroskopisk: Det er bildet er helt annet. Molekylene er i konstant, voldsom bevegelse, og systemet hopper hele tiden mellom enormt mange forskjellige mikrotilstander. Likevekt er ikke fravær av aktivitet, men tilstanden med flest tilgjengelige mikrotilstander (altså den mest sannsynlige makrotilstanden).

Én makrotilstand kan realiseres av mange mikrotilstander, og systemet tilbringer mest tid i den makrotilstanden som har flest mikrotilstander bak seg. Ved likevekt har systemet "funnet" dette maksimumet. Endringene på molekylnivå fortsetter (molekyler bindes og løsner, kolliderer og bytter energi hele tiden), men de skjer like ofte begge veier, så netto skjer ingenting målbart.

**b)**
Det er ingen fysisk kraft som skyver systemet, det er faktisk ren sannsynlighet som står bak. 
To terninger gir sum 7 oftere enn sum 2 fordi det finnes seks måter å få 7 på, men bare én for 2. Noe som gjelder det samme ger: en makrotilstand med mange mikrotilstander er mer sannsynlig, rett og slett fordi den kan realiseres på flere måter. Med et enormt antall molekyler blir froskjellen så ekstremt at lav entropi-tilstander praktisk talt aldri inntreffer. 

Hindrer noe den motsatte veien? Nei. Den er ikke forbudt, bare så vanvittig usannsynlig at den aldri observeres. Andre lov handler ikke om hva som er umulig, men om hva som er overveldende sannsynlig. 

**c)**
**Temperatur - Termisk likevekt**
Energien strømmer fra varmt til kaldt fordi det finnes flere mikrotilstander når energien er jevnt fordelt enn når den er samlet. Likevekten: lik temperatur overalt. 

**Trykk - Mekanisk likevekt**
Trykk er kraft per areal fra molekyler som dunker mot en flate. Har to områder ulikt trykk, vil volumet justere seg ved at den siden med høyere trykk utvider seg, helt til trykket er likt på begge sider. Da er systemet i mekanisk likevet. 

I begge tilfellene er den underliggende drivkraften den samme som i $b)$: systemet flytter seg mot den fordelingen som har flest tilgjengelige mikrotilstander. Temperatur styrer ujevning av energi, trykk styrer ujevning av volum/rom.

### Oppgave 2


**a)**
På nøyaktig to måter: ved å tilføre/fjerne **varme** ($q$), eller ved å utføre/motta **arbeid** ($w$). Det er hele innholdet i termodynamikkens første lov:

$$\Delta U = q + w$$

I simuleringen er dette akkurat de to knappene du har: varmeelementet ($\Delta q$) og rørepinna ($\Delta w$). Ingen andre kanaler finnes, vil du endre den indre energien, må det skje gjennom en av disse to.

**b)**
Varmen øker molekylenes gjennomsnittlige bevegelsesenergi. De begynner å bevege seg raskere. Siden temperatur er et mål på nettopp denne bevegelsesenergien, stiger temperaturen. Den indre energien øker.

I simuleringen ser du dette dobbelt: molekylene i venstre panel beveger seg raskere, og termometeret til høyre stiger.

**c)**
Resultatet er det samme som for varme: den indre energien øker, molekylene beveger seg raskere, og temperaturen stiger. Rørepinna som pisker gassen overfører mekanisk energi til molekylene, de blir dyttet og får høyere fart.

Poenget er at utfallet er identisk enten du tilfører energien som varme eller som arbeid. Systemet «merker» bare at den indre energien har økt; det holder ikke regnskap over *hvordan* energien kom inn.

**d)**
Oppsettet er en stiv, isolert beholder. Volumet er konstant hele tiden, så det gjøres ikke noe volumarbeid, og $\Delta U = q + w$ gjelder direkte. Du har to måter å tilføre energi på:

- **Varmeelementet** ($\Delta q$): tilfører (positiv) eller fjerner (negativ) varme.
- **Rørepinna** ($\Delta w$): gjør arbeid på gassen (positiv, som å piske) eller lar gassen drive noe utenfor (negativ).

For hvert steg du legger til, oppdaterer simuleringen tre tall: $q_\text{tot}$, $w_\text{tot}$ og summen $\Delta U_\text{tot} = q_\text{tot} + w_\text{tot}$. Termometeret viser temperaturen, som stiger eller synker i takt med $\Delta U_\text{tot}$, og molekylene i venstre panel beveger seg raskere jo høyere temperaturen er.

Kort sagt demonstrerer simuleringen første lov i praksis: den viser at varme og arbeid er to likeverdige måter å endre den indre energien på, og at det bare er summen som bestemmer temperaturen, ikke fordelingen mellom $q$ og $w$.

**e)**

Ja. Med bare positiv $\Delta q$ ser du at $q_\text{tot}$ blir positiv, $w_\text{tot}$ forblir 0, og $\Delta U_\text{tot}$ blir positiv og lik $q_\text{tot}$. Temperaturen stiger, og molekylene beveger seg raskere.

Dette bekrefter b): varme øker den indre energien og dermed temperaturen. Molekylene får høyere fart, akkurat som forutsagt.

**f)**

Nå ser du at $w_\text{tot}$ blir positiv, $q_\text{tot}$ forblir 0, og $\Delta U_\text{tot}$ blir positiv. Temperaturen stiger, på nøyaktig samme måte som da du bare tilførte varme.

Og her er det fine poenget: temperaturen øker like mye i begge tilfeller (så lenge energimengden er den samme), men *måten* energien kom inn på er fysisk forskjellig:

- I e) kom energien som **varme** — uordnet energioverføring fra et varmere element, molekyl for molekyl.
- I f) kom energien som **arbeid** — organisert, mekanisk dytting fra rørepinna.

> Systemet ender likevel i samme tilstand (samme $T$, samme indre energi), fordi den indre energien ikke «husker» om den ble fylt opp via varme eller arbeid. Dette er selve grunnen til at $U$ er en tilstandsfunksjon, mens $q$ og $w$ er veifunksjoner: to helt ulike veier (ren varme vs. rent arbeid) fører til samme endring i $U$.

**g)**
Termodynamikkens første lov er energibevaring for et termodynamisk system: den indre energien til et system kan bare endres ved varme eller arbeid, og energien forsvinner aldri, den bare flyttes.

$$\Delta U = q + w$$

Regnskapet i simuleringen er denne loven, satt opp som et bokholderi. Hver gang du legger til et steg, legges $\Delta q$ til $q_\text{tot}$ og $\Delta w$ til $w_\text{tot}$, og linja $\Delta U_\text{tot} = q_\text{tot} + w_\text{tot}$ regner ut summen. At temperaturen (indre energi) alltid følger denne summen (og aldri noe annet) er den direkte, synlige bekreftelsen på at første lov holder. Energien inn som varme pluss energien inn som arbeid er nøyaktig lik økningen i indre energi. Ikke mer, ikke mindre.

**h)**

Varmekapasiteten forteller hvor mye energi som skal til for å heve temperaturen med én grad. Ved konstant volum (som i denne stive beholderen) er den definert som

$$C_V = \left(\frac{\partial U}{\partial T}\right)_V$$

altså hvor mye den indre energien endrer seg per grad temperaturøkning. En stor $C_V$ betyr at systemet kan «svelge» mye energi uten at temperaturen stiger så mye; en liten $C_V$ betyr at temperaturen skyter raskt i været.

I forhold til indre energi: $C_V$ er rett og slett *stigningstallet* til $U$ som funksjon av $T$. Ved konstant volum er $\Delta U = C_V\,\Delta T$ , det er derfor du kan lese av temperaturøkningen direkte fra energien du tilfører. I simuleringen kan du velge gasstype (monoatomisk med $C_V = \tfrac{3}{2}nR$ osv.); en gass med høyere $C_V$ vil vise mindre temperaturøkning for samme tilførte energi, fordi energien fordeles på flere frihetsgrader.

**i)**

Prøv først å nå en viss temperatur med kun varme (ett stort $\Delta q$-steg). Noter $q_\text{tot}$ og $w_\text{tot}$. Reset, og nå samme temperatur med en blanding av flere steg med både $\Delta q$ og $\Delta w$, satt slik at summen $\Delta U_\text{tot}$ blir den samme.

Hva man ser:

- **$\Delta U_\text{tot}$ er lik i begge tilfeller** — nødvendigvis, siden du endte på samme temperatur.
- **Men $q_\text{tot}$ og $w_\text{tot}$ er helt forskjellige.** I det første tilfellet er all energien varme ($w_\text{tot} = 0$); i det andre er den delt mellom varme og arbeid.

Hva forteller dette om varmekapasitet og indre energi? At det er den totale indre energien ($\Delta U$) som bestemmer temperaturen — og dermed hvor varmekapasiteten «virker», Ikke hvor mye av energien som kom inn som varme. Varmekapasiteten kobler $\Delta U$ til $\Delta T$, ikke $q$ til $\Delta T$.

> Er $q$ alene et mål på indre energi?: Nei — og dette er hele poenget. $q$ er bare *én* av to bidragsytere til $\Delta U$. Du kan nå samme indre energi (samme temperatur) med mye varme og lite arbeid, eller lite varme og mye arbeid. Bare når $w = 0$ (som i en ren oppvarming ved konstant volum) faller $q$ tilfeldigvis sammen med $\Delta U$. Generelt er det summen $q + w$ som er indre energi-endringen, aldri varmen alene. $q$ og $w$ er veifunksjoner (de avhenger av hvordan du gikk fram) mens $\Delta U$ er en tilstandsfunksjon som bare bryr seg om start og slutt.

### Oppgave 3

**a)**

«Iso» betyr lik/samme, og «term» betyr varme/temperatur slik at en isoterm prosess er en prosess ved konstant temperatur. Temperaturen holdes den samme gjennom hele forløpet.

Setter du simuleringen til «Isoterm» og drar på $V_2$, ser du at $T_2$ aldri endrer seg. Den blir stående på 1,00 uansett hvilket sluttvolum du velger, akkurat lik $T_1$. Det er nettopp definisjonen: $T_1 = T_2$ hele veien.

**b)**
Sett prosessen til «Adiabatisk», velg $V_2 > V_1$, og kjør. Du ser at $T_2$ synker hvor gassen kjøles ned.

Dette gir mening fysisk: i en adiabatisk prosess tilføres ingen varme ($q = 0$). Når gassen ekspanderer, gjør den arbeid på omgivelsene (skyver stempelet opp). Den energien må komme fra et sted, og siden ingen varme kommer inn utenfra, må gassen bruke av sin *egen* indre energi. Lavere indre energi betyr lavere temperatur, gassen kjøler seg selv ned ved å arbeide. Og ja, dette stemmer med det du observerer: $T_2 < T_1$.

**c)**
Da stiger temperaturen. Uten arbeid ($w = 0$) er $\Delta U = q$, så all den tilførte varmen går rett til å øke den indre energien. Molekylene beveger seg raskere, og temperaturen øker. Dette er tilfellet med den stive beholderen fra forrige oppgave: varme inn, ingen volumendring, temperaturen opp.

**d)**
Man må tilføre varme i samme takt som gassen gjør arbeid. I en isoterm ekspansjon gjør gassen arbeid på omgivelsene og ville normalt kjøles ned (som i b), men hvis temperaturen skal holdes konstant, kan ikke den indre energien synke. Løsningen er å mate inn nøyaktig like mye varme som gassen gir fra seg som arbeid.

Matematisk: isotermt er $\Delta U = 0$ (siden $T$ er konstant), så første lov gir $q = -w$. Varmen inn balanserer arbeidet ut, ledd for ledd, slik at den indre energien (og dermed temperaturen) står stille.

**e)**
Kjør samme $V_1$ og $V_2$ med begge prosesstypene og se på $\Delta U$:

- **Isoterm:** $\Delta U = 0$.
- **Adiabatisk:** $\Delta U \neq 0$ (negativ ved ekspansjon).

Grunnen ligger i uttrykket $dU = C_V\,dT$. Her står $dT$ for endringen i temperatur. Den indre energien til en ideell gass avhenger *bare* av temperaturen, så $\Delta U$ er direkte proporsjonal med temperaturendringen.

- *Isotermt* er temperaturen konstant, altså $dT = 0$, og dermed $\Delta U = C_V \cdot 0 = 0$. Du ser det igjen i simuleringen ved at $T_1 = T_2$ — ingen temperaturendring, ingen energiendring.
- *Adiabatisk* endrer temperaturen seg ($T_2 \neq T_1$), så $dT \neq 0$, og dermed $\Delta U \neq 0$. Tallene viser $T_2 < T_1$ ved ekspansjon, som gir en negativ $\Delta U$.

Kort sagt: $\Delta U$ følger $\Delta T$, og du leser $\Delta T$ rett av ved å sammenligne $T_1$ og $T_2$ på skjermen.

**f)**
En adiabatisk endring er en prosess der det ikke utveksles varme med omgivelsene:

$$\boxed{q = 0}$$

Det skjer enten fordi beholderen er perfekt isolert, eller fordi prosessen går så raskt at varmen ikke rekker å strømme inn eller ut.

Sjekk i simuleringen: velg «Adiabatisk», prøv hvilke som helst $V_1$ og $V_2$, og du ser at $q$ alltid står på 0,00 uansett. Hele energiregnskapet går via arbeid alene: $\Delta U = w$. Det er dette som gjør at temperaturen *må* endre seg når gassen arbeider, siden det ikke finnes noen varme til å kompensere.

**g)**
For en reversibel adiabatisk prosess starter vi fra første lov med $q = 0$:

$$dU = dw \quad\Rightarrow\quad C_V\,dT = -p\,dV$$

Bruker vi $p = \dfrac{nRT}{V}$ og integrerer, kommer vi frem til adiabat-relasjonen $TV^{\gamma-1} = \text{konstant}$ (der $\gamma = C_p/C_V$). Arbeidet gjort *på* gassen er endringen i indre energi:

$$w = \Delta U = C_V(T_2 - T_1) = \frac{nR}{\gamma - 1}(T_2 - T_1)$$

Ekvivalent, uttrykt ved volumene via $p_1V_1^\gamma = p_2V_2^\gamma$:

$$\boxed{w = \frac{p_2V_2 - p_1V_1}{\gamma - 1}}$$

**Sjekk mot simuleringen:** velg konkrete verdier, f.eks. $V_1 = 1$, $V_2 = 2$, monoatomisk ($\gamma = 5/3$). Med reduserte enheter ($nR = 1$, $T_1 = 1$, så $p_1 = 1$):

$$T_2 = T_1\left(\frac{V_1}{V_2}\right)^{\gamma-1} = 1 \cdot \left(\frac{1}{2}\right)^{2/3} = 0{,}63$$

$$w = \frac{nR}{\gamma-1}(T_2 - T_1) = \frac{1}{2/3}(0{,}63 - 1) = 1{,}5 \times (-0{,}37) = -0{,}55$$

Negativt, fordi gassen gjør arbeid på omgivelsene under ekspansjonen. Sammenlign med $w$-tallet simuleringen viser og det skal stemme.

**h)**
Ja. «Adiabatisk» og «reversibel» er to uavhengige egenskaper, og en prosess kan godt være begge deler samtidig. Adiabatisk sier bare at $q = 0$; reversibel sier at prosessen skjer uendelig langsomt, i konstant likevekt. En *reversibel adiabatisk* prosess er faktisk idealtilfellet vi utledet formelen for i g).

**Er simuleringen en god modell for reversibilitet?** Ikke helt — og det er verdt å være ærlig om. Når du trykker «Kjør prosess», beveger stempelet seg fra $V_1$ til $V_2$ over et par sekunder, altså i endelig fart. En virkelig reversibel prosess må gå uendelig langsomt, slik at gassen er i likevekt i hvert øyeblikk og trykket er jevnt fordelt hele veien. Beveger stempelet seg raskt, rekker ikke gassen å utjevne seg, det oppstår trykkforskjeller og turbulens inne i sylinderen, og prosessen blir irreversibel.

> Hva ville kreves for at den skulle vært reversibel i praksis? At stempelet ble flyttet i uendelig mange, uendelig små steg, med en pause mellom hvert til gassen er i full likevekt igjen. I grensen av uendelig langsom bevegelse er det ytre trykket hele tiden nesten nøyaktig lik gassens eget trykk, ingen turbulens oppstår, og prosessen kan reverseres uten å etterlate spor. Simuleringen viser altså den idealiserte kurven (formelen fra g gjelder for den reversible adiabaten), men selve stempelbevegelsen på skjermen er en forenkling som en ekte reversibel prosess kan man strengt tatt aldri kjøre på «et par sekunder».

### Oppgave 4

**a)**

$dV$ står for en forsvinnende liten volumendring, ett lite trinn fra $V$ til $V + dV$ mens prosessen pågår. Integralet $\int$ legger sammen bidragene $-p\,dV$ fra hvert slikt trinn, fra begynnelse til slutt.

Vi må gå fram trinnvis nettopp fordi trykket kan variere underveis. Med varierende $p$ nytter det ikke å gange ett trykk med hele volumendringen på én gang; vi må dele opp, regne bit for bit, og summere.

**b)**
Holder $p$ seg fast, er den en konstant og kan trekkes ut av integralet:

$$w = -\int_{V_1}^{V_2} p\,dV = -p\int_{V_1}^{V_2} dV = -p(V_2 - V_1) = -p\,\Delta V$$

Da er arbeidet ganske enkelt trykk multiplisert med samlet volumendring. Fortegnet stemmer med intuisjonen: utvider gassen seg ($\Delta V > 0$), får vi $w < 0$, altså at systemet utfører arbeid på omgivelsene og gir fra seg energi.

**c)**
Her kan $p$ ikke flyttes ut av integralet, siden det forandrer seg gjennom hele forsøket. Slik går man fram:

1. **Uttrykk $p$ ved $V$.** I en reversibel prosess er systemet i likevekt hele tiden, så en tilstandslikning knytter trykket til volumet. For en ideell gass er dette idealgassloven:
$$p = \frac{nRT}{V}$$

2. **Bytt ut $p$ i integralet**, slik at hele uttrykket står i $V$:
$$w = -\int_{V_1}^{V_2} \frac{nRT}{V}\,dV$$

3. **Integrer** mellom $V_1$ og $V_2$. Er temperaturen konstant (isoterm), er $nRT$ konstant og kan settes utenfor; integralet av $1/V$ blir en logaritme:
$$w = -nRT\ln\frac{V_2}{V_1}$$

### Oppgave 5
**a)**
* Tilfelle 1 — reversibelt (isotermt)

Her varierer trykket, så vi integrerer:

$$w_1 = -nRT\ln\frac{V_2}{V_1} = -10{,}0\ \text{atm}\,\text{dm}^3 \times \ln\frac{10{,}0}{1{,}0} = -10{,}0 \times \ln 10 = -23{,}0\ \text{atm}\,\text{dm}^3$$

$$\boxed{w_1 = -23{,}0\ \text{atm}\,\text{dm}^3 = -2{,}33\ \text{kJ}}$$

$$w_2 = -p_\text{ytre}(V_2 - V_1) = -1{,}0\ \text{atm} \times (10{,}0 - 1{,}0)\ \text{dm}^3 = -9{,}0\ \text{atm}\,\text{dm}^3$$

| Prosess | $w$ (atm·dm³) | $w$ (kJ) |
|---|---|---|
| 1. Reversibelt | $-23{,}0$ | $-2{,}33$ |
| 2. Irreversibelt (1 steg) | $-9{,}0$ | $-0{,}91$ |
| 3. To trinn | $-13{,}0$ | $-1{,}32$ |

**b)**
Arbeidet (utført *av* gassen) er arealet under kurven i et $p$–$V$-diagram. Her er de tre tilfellene med det skraverte arealet:

**c)**

Det reversible arbeidet er maksimum: $|w_1| = 23{,}0\ \text{atm}\,\text{dm}^3 = 2{,}33\ \text{kJ}$.

Grunnen ser du direkte i diagrammene: for at gassen skal gjøre *mest mulig* arbeid, må den skyve mot et ytre trykk som er så høyt som mulig i hvert øyeblikk. I den reversible prosessen holdes det ytre trykket hele tiden *uendelig nær* gassens eget trykk — gassen presser mot maksimal motstand hele veien, og arealet under kurven blir størst mulig.

I de irreversible tilfellene «slipper» det ytre trykket brått ned til en lav verdi, så gassen skyver mot mye svakere motstand og får utrettet mindre. Jo flere trinn man deler prosessen i (som i tilfelle 3), desto tettere følger det ytre trykket gassens trykk, og desto mer nærmer arbeidet seg den reversible grensen. I grensen av uendelig mange, uendelig små trinn *er* prosessen reversibel.

Så: **reversibel ekspansjon gir maksimalt arbeid**, og ingen irreversibel vei mellom de samme to tilstandene kan overgå den. Dette er et helt sentralt resultat i termodynamikken — den reversible veien er idealgrensen alle virkelige prosesser måles mot.

### Oppgave 6

**Gitt:** $q = 187\ \text{J}$ tilført ved konstant trykk, $n = 1{,}9\ \text{mol}$, $\Delta T = 1{,}78\ \text{K}$.

**a)**

* Ved konstant trykk ($C_{p,m}$)

Ved konstant trykk er tilført varme lik entalpiendringen: $q_p = \Delta H = n\,C_{p,m}\,\Delta T$. Løser for $C_{p,m}$:

$$C_{p,m} = \frac{q_p}{n\,\Delta T} = \frac{187\ \text{J}}{1{,}9\ \text{mol} \times 1{,}78\ \text{K}}$$

$$\boxed{C_{p,m} = 55{,}3\ \text{J K}^{-1}\text{mol}^{-1}}$$

* Ved konstant volum ($C_{V,m}$)

Vi har ikke et eget forsøk ved konstant volum, men for en ideell gass gjelder relasjonen fra b): $C_{p,m} - C_{V,m} = R$. Derfor:

$$C_{V,m} = C_{p,m} - R = 55{,}3 - 8{,}314$$

$$\boxed{C_{V,m} = 47{,}0\ \text{J K}^{-1}\text{mol}^{-1}}$$

(Til orientering: disse verdiene er høyere enn for enatomige eller toatomige gasser, så dette er trolig en fleratomig gass med mange aktive frihetsgrader.)

**b)**
For en ideell gass er sammenhengen:

$$\boxed{C_{p,m} - C_{V,m} = R} \qquad (\approx 8{,}314\ \text{J K}^{-1}\text{mol}^{-1})$$

**Hvorfor er $C_p$ alltid større enn $C_V$?** Poenget er hva energien går til i de to tilfellene:

- **Ved konstant volum** kan ikke gassen utvide seg. Da går *all* den tilførte varmen til å øke den indre energien, altså molekylenes bevegelsesenergi, som er det temperaturen måler.
- **Ved konstant trykk** utvider gassen seg når den varmes opp. En del av den tilførte varmen «lekker ut» som ekspansjonsarbeid mot omgivelsene ($w = -p\,\Delta V$), i stedet for å heve temperaturen.

Derfor må du tilføre *mer* varme ved konstant trykk for å oppnå samme temperaturøkning, du betaler både for temperaturhevingen *og* for arbeidet. Nettopp derfor er $C_p > C_V$.

Størrelsen på differansen følger av idealgassloven: ved konstant trykk er ekspansjonsarbeidet per mol per grad akkurat $R$, som er nøyaktig det ekstra du må tilføre. Det er derfor differansen blir gasskonstanten $R$.

### Oppgave 7

**Gitt:** $n = 1{,}00\ \text{mol}$, $T = 27\ ^{\circ}\text{C} = 300{,}15\ \text{K}$ (isotermt), $p_1 = 3{,}00\ \text{atm} \to p_2 = 1{,}00\ \text{atm}$.

Siden prosessen er isoterm og gassen er ideell, gjelder to ting gjennom hele oppgaven:

$$\boxed{\Delta U = 0 \quad\text{og}\quad \Delta H = 0}$$

Den indre energien til en ideell gass avhenger bare av temperaturen, og den holdes konstant. Det samme gjelder entalpien. Dette er sant for **begge** veiene, siden $U$ og $H$ er tilstandsfunksjoner, de bryr seg bare om start og slutt, ikke veien.

 Ved konstant $T$ gir idealgassloven $V_2/V_1 = p_1/p_2 = 3{,}00$.

#### Vei 1 — reversibel isoterm ekspansjon

**Arbeid:** For en reversibel isoterm prosess:

$$w = -nRT\ln\frac{V_2}{V_1} = -nRT\ln\frac{p_1}{p_2} = -(1{,}00)(8{,}314)(300{,}15)\ln 3{,}00$$

$$\boxed{w = -2741\ \text{J} \approx -2{,}74\ \text{kJ}}$$

**Varme:** Fra første lov, $\Delta U = q + w = 0$, så:

$$q = -w = +2741\ \text{J} \approx +2{,}74\ \text{kJ}$$

All varmen gassen tar opp går rett ut igjen som arbeid.

**Entropi i systemet:** For en isoterm ideell gass:

$$\Delta S_\text{system} = nR\ln\frac{V_2}{V_1} = (1{,}00)(8{,}314)\ln 3{,}00 = +9{,}13\ \text{J K}^{-1}$$

**Entropi i omgivelsene:** Omgivelsene avgir den varmen systemet tok opp, reversibelt ved $T$:

$$\Delta S_\text{omgivelser} = \frac{-q_\text{rev}}{T} = \frac{-2741}{300{,}15} = -9{,}13\ \text{J K}^{-1}$$

**Total entropi:**

$$\Delta S_\text{tot} = 9{,}13 + (-9{,}13) = \boxed{0\ \text{J K}^{-1}}$$

Nøyaktig null, som gir kjennetegnet på en reversibel prosess.


#### Vei 2 — irreversibel ekspansjon mot konstant ytre trykk 1,00 atm

Start og slutt er de *samme* som i vei 1, så tilstandsfunksjonene er uendret:

$$\Delta U = 0, \qquad \Delta H = 0, \qquad \Delta S_\text{system} = +9{,}13\ \text{J K}^{-1}$$

$\Delta S_\text{system}$ er den samme fordi entropi er en tilstandsfunksjon — den bryr seg bare om start og slutt, ikke veien.

**Arbeid:** Nå skyver gassen mot et konstant ytre trykk $p_\text{ytre} = 1{,}00\ \text{atm}$:

$$w = -p_\text{ytre}(V_2 - V_1)$$

Jeg finner volumene: $V_1 = \dfrac{nRT}{p_1} = \dfrac{(1{,}00)(0{,}08206)(300{,}15)}{3{,}00} = 8{,}21\ \text{dm}^3$, og $V_2 = 3V_1 = 24{,}63\ \text{dm}^3$.

$$w = -(1{,}00\ \text{atm})(24{,}63 - 8{,}21)\ \text{dm}^3 = -16{,}4\ \text{atm}\,\text{dm}^3 \times 101{,}325\ \tfrac{\text{J}}{\text{atm}\,\text{dm}^3}$$

$$\boxed{w = -1663\ \text{J} \approx -1{,}66\ \text{kJ}}$$

Mindre arbeid enn den reversible veien, akkurat som forventet.

**Varme:** Igjen fra $\Delta U = q + w = 0$:

$$q = -w = +1663\ \text{J} \approx +1{,}66\ \text{kJ}$$

**Entropi i omgivelsene:** Omgivelsene avgir nå bare $1663\ \text{J}$ (den faktisk utvekslede varmen), ved $T$:

$$\Delta S_\text{omgivelser} = \frac{-q}{T} = \frac{-1663}{300{,}15} = -5{,}54\ \text{J K}^{-1}$$

**Total entropi:**

$$\Delta S_\text{tot} = 9{,}13 + (-5{,}54) = \boxed{+3{,}59\ \text{J K}^{-1}}$$

Positiv, som er kjennetegnet på en irreversibel (spontan) prosess.

#### Oppsummering

| Størrelse | Vei 1 (reversibel) | Vei 2 (irreversibel) |
|---|---|---|
| $\Delta U$ | $0$ | $0$ |
| $\Delta H$ | $0$ | $0$ |
| $w$ | $-2{,}74\ \text{kJ}$ | $-1{,}66\ \text{kJ}$ |
| $q$ | $+2{,}74\ \text{kJ}$ | $+1{,}66\ \text{kJ}$ |
| $\Delta S_\text{system}$ | $+9{,}13\ \text{J K}^{-1}$ | $+9{,}13\ \text{J K}^{-1}$ |
| $\Delta S_\text{omgivelser}$ | $-9{,}13\ \text{J K}^{-1}$ | $-5{,}54\ \text{J K}^{-1}$ |
| $\Delta S_\text{tot}$ | $0$ | $+3{,}59\ \text{J K}^{-1}$ |

#### Kommentar til ΔS-verdiene

**$\Delta S_\text{system}$ er lik for begge veiene** 
Entropi er en tilstandsfunksjon og den avhenger bare av start- og slutt-tilstand, som er identiske her. Systemets entropiøkning «vet» ingenting om *hvordan* gassen kom seg fra 3 atm til 1 atm.

**$\Delta S_\text{tot}$ er det som skiller de to** Her ligger forskjellen:

- **Reversibelt:** $\Delta S_\text{tot} = 0$. Systemet og omgivelsene endrer entropi med nøyaktig like mye i hver sin retning. En reversibel prosess kan i prinsippet gå «baklengs» uten å etterlate spor — den er den idealiserte grensen.
- **Irreversibelt:** $\Delta S_\text{tot} = +3{,}59\ \text{J K}^{-1} > 0$. Her skapes det netto entropi i universet. Dette er andre lov i praksis: enhver spontan, virkelig prosess øker den totale entropien.

**Hvorfor blir det slik?** Systemet tar opp *mindre* varme i den irreversible veien (1,66 kJ mot 2,74 kJ), så omgivelsene taper mindre entropi ($-5{,}54$ mot $-9{,}13$). Men systemet vinner *like mye* entropi uansett ($+9{,}13$). Denne ubalansen (full gevinst i systemet, men bare delvis tap i omgivelsene) er nøyaktig den netto entropien som produseres når vi «kaster bort» arbeidskapasitet ved å ekspandere mot et lavt trykk i stedet for reversibelt.

### Oppgave 8

#### Del 1 — to aluminiumsklosser

**Gitt:** to klosser, hver $n = 1{,}00\ \text{mol}$, $T_1 = 100\ ^{\circ}\text{C} = 373{,}15\ \text{K}$ og $T_2 = 0\ ^{\circ}\text{C} = 273{,}15\ \text{K}$, $C_{p,m} = 24{,}4\ \text{J K}^{-1}\text{mol}^{-1}$, termisk isolert.

**a) Slutt-temperaturen**

Systemet er isolert, så varmen den varme klossen avgir er nøyaktig den kalde tar opp: $q_\text{varm} + q_\text{kald} = 0$. Siden begge klossene har *samme* $n$ og $C_{p,m}$, forenkler dette seg til gjennomsnittet:

$$T_f = \frac{T_1 + T_2}{2} = \frac{373{,}15 + 273{,}15}{2}$$

$$\boxed{T_f = 323{,}15\ \text{K} = 50\ ^{\circ}\text{C}}$$

**b)
**
For oppvarming/avkjøling ved konstant trykk er $\Delta S = nC_{p,m}\ln\dfrac{T_f}{T_i}$.

**Den varme klossen** (avkjøles, $373{,}15 \to 323{,}15$):

$$\Delta S_\text{varm} = (1{,}00)(24{,}4)\ln\frac{323{,}15}{373{,}15} = 24{,}4 \times \ln(0{,}866) = -3{,}51\ \text{J K}^{-1}$$

**Den kalde klossen** (varmes opp, $273{,}15 \to 323{,}15$):

$$\Delta S_\text{kald} = (1{,}00)(24{,}4)\ln\frac{323{,}15}{273{,}15} = 24{,}4 \times \ln(1{,}183) = +4{,}10\ \text{J K}^{-1}$$

**Til sammen:**

$$\Delta S_\text{tot} = -3{,}51 + 4{,}10 = \boxed{+0{,}59\ \text{J K}^{-1}}$$

**Kommentar til fortegnene:**
- Den varme klossen mister entropi ($\Delta S < 0$) fordi den kjøles ned, molekylene mister bevegelsesenergi og systemet blir mer «ordnet».
- Den kalde klossen vinner entropi ($\Delta S > 0$) fordi den varmes opp.
- Totalen er **positiv**, selv om det utveksles like mye *varme* begge veier. Grunnen er at entropiutvekslingen skjer ved ulik temperatur: den kalde klossen tar opp varmen ved lavere temperatur, og entropibidraget $q/T$ er større når $T$ er liten. Derfor overgår gevinsten hos den kalde klossen tapet hos den varme. Varmestrøm fra varmt til kaldt er en spontan, irreversibel prosess, og $\Delta S_\text{tot} > 0$ er nettopp andre lov i praksis.

#### Del 2 — aluminiumskloss og jernbit

Nå er materialene ulike, og «ett mol» gjelder ikke lenger — vi må finne stoffmengdene fra volum og tetthet.

**Gitt:** begge $V = 50\ \text{cm}^3$; jern ved $T_\text{Fe} = 0\ ^{\circ}\text{C} = 273{,}15\ \text{K}$, aluminium ved $T_\text{Al} = 100\ ^{\circ}\text{C} = 373{,}15\ \text{K}$.
$C_{p,m}(\text{Al}) = 24{,}4$, $\ C_{p,m}(\text{Fe}) = 25{,}10\ \text{J K}^{-1}\text{mol}^{-1}$.
$\rho_\text{Al} = 2{,}698$, $\ \rho_\text{Fe} = 7{,}874\ \text{g cm}^{-3}$. Molmasser: $M_\text{Al} = 26{,}98$, $M_\text{Fe} = 55{,}85\ \text{g/mol}$.

**Først stoffmengdene** ($n = \rho V / M$):

$$n_\text{Al} = \frac{2{,}698 \times 50}{26{,}98} = 5{,}00\ \text{mol}, \qquad n_\text{Fe} = \frac{7{,}874 \times 50}{55{,}85} = 7{,}05\ \text{mol}$$

Det er verdt å legge merke til: samme volum, men jernbiten inneholder over 40 % flere mol. Det kommer til å trekke slutt-temperaturen mot jernets startverdi.

**c)**
Varmebalanse i et isolert system: $n_\text{Al}C_{p,\text{Al}}(T_f - T_\text{Al}) + n_\text{Fe}C_{p,\text{Fe}}(T_f - T_\text{Fe}) = 0$. Løser for $T_f$ (et varmekapasitetsveid gjennomsnitt):

$$T_f = \frac{n_\text{Al}C_{p,\text{Al}}\,T_\text{Al} + n_\text{Fe}C_{p,\text{Fe}}\,T_\text{Fe}}{n_\text{Al}C_{p,\text{Al}} + n_\text{Fe}C_{p,\text{Fe}}}$$

Regner ut de to varmekapasitetene (total, $nC_{p,m}$):

$$n_\text{Al}C_{p,\text{Al}} = 5{,}00 \times 24{,}4 = 122{,}0\ \text{J K}^{-1}, \qquad n_\text{Fe}C_{p,\text{Fe}} = 7{,}05 \times 25{,}10 = 176{,}9\ \text{J K}^{-1}$$

$$T_f = \frac{122{,}0 \times 373{,}15 + 176{,}9 \times 273{,}15}{122{,}0 + 176{,}9} = \frac{45524 + 48320}{298{,}9} = \frac{93844}{298{,}9}$$

$$\boxed{T_f = 313{,}9\ \text{K} = 40{,}8\ ^{\circ}\text{C}}$$

Slutt-temperaturen ligger *nærmere* jernets 0 °C enn midtpunktet (50 °C), akkurat som ventet - jernbiten har den største totale varmekapasiteten og «vinner» dragkampen om felles temperatur.

**d)**

**Aluminium** (avkjøles, $373{,}15 \to 313{,}9$):

$$\Delta S_\text{Al} = n_\text{Al}C_{p,\text{Al}}\ln\frac{T_f}{T_\text{Al}} = 122{,}0 \times \ln\frac{313{,}9}{373{,}15} = 122{,}0 \times \ln(0{,}8412) = -21{,}1\ \text{J K}^{-1}$$

**Jern** (varmes opp, $273{,}15 \to 313{,}9$):

$$\Delta S_\text{Fe} = n_\text{Fe}C_{p,\text{Fe}}\ln\frac{T_f}{T_\text{Fe}} = 176{,}9 \times \ln\frac{313{,}9}{273{,}15} = 176{,}9 \times \ln(1{,}1492) = +24{,}6\ \text{J K}^{-1}$$

**Til sammen:**

$$\Delta S_\text{tot} = -21{,}1 + 24{,}6 = \boxed{+3{,}5\ \text{J K}^{-1}}$$

**Kommentar til fortegnene:**
- Samme mønster som i del 1: aluminiumet (avkjøles) taper entropi, jernet (varmes opp) vinner entropi.
- Totalen er igjen **positiv**. Selv om aluminiumet og jernet utveksler nøyaktig like mye *varme*, tar jernet opp denne varmen ved lavere temperatur enn aluminiumet avgir den, og siden entropibidraget $q/T$ er størst der $T$ er lav, blir jernets gevinst større enn aluminiumets tap.
- $\Delta S_\text{tot} > 0$ bekrefter at varmeutjevningen er en spontan, irreversibel prosess. Så snart de to bitene har samme temperatur, stopper netto varmestrøm. systemet har nådd termisk likevekt, den tilstanden med flest tilgjengelige mikrotilstander.

### Oppgave 9

**Gitt:** $n = 1{,}00\ \text{mol}$, isotermt ved $T = 25\ ^{\circ}\text{C} = 298{,}15\ \text{K}$, volumøkning $\Delta V = +1{,}80\ \text{dm}^3$. Slutt-tilstand: $p_2 = 1{,}97\ \text{bar}$, $V_2 = 2{,}14\ \text{dm}^3$.

**Start-volum og -trykk:**

$$V_1 = V_2 - \Delta V = 2{,}14 - 1{,}80 = 0{,}34\ \text{dm}^3$$

Ved konstant temperatur gir Boyles lov $p_1 = \dfrac{p_2 V_2}{V_1} = \dfrac{1{,}97 \times 2{,}14}{0{,}34} = 12{,}4\ \text{bar}$.

**a)**
Skjer prosessen ved konstant trykk, er det sluttrykket $p_2 = 1{,}97\ \text{bar}$ som virker gjennom hele volumendringen. Arbeidet *på* systemet er $w = -p\,\Delta V$, og arbeidet *utført av* systemet er det motsatte:

$$w_\text{av system} = p\,\Delta V = 1{,}97\ \text{bar} \times 1{,}80\ \text{dm}^3 = 3{,}55\ \text{bar dm}^3$$

Gjør om til joule ($1\ \text{bar dm}^3 = 100\ \text{J}$):

$$\boxed{w_\text{av system} = 3{,}55\ \text{bar dm}^3 = 355\ \text{J}}$$

**b)**

At mekanisk likevekt opprettholdes hele veien betyr at prosessen er **reversibel**, det ytre trykket er hele tiden uendelig nær gassens eget trykk. For en reversibel isoterm ekspansjon:

$$w_\text{av system} = nRT\ln\frac{V_2}{V_1} = p_2V_2 \ln\frac{V_2}{V_1}$$

Bruker $nRT = p_2V_2 = 4{,}22\ \text{bar dm}^3$ (fra sluttilstanden oppgaven gir) for å være konsistent:

$$w_\text{av system} = 4{,}22\ \text{bar dm}^3 \times \ln\frac{2{,}14}{0{,}34} = 4{,}22 \times \ln(6{,}29) = 4{,}22 \times 1{,}839 = 7{,}76\ \text{bar dm}^3$$

$$\boxed{w_\text{av system} = 7{,}76\ \text{bar dm}^3 = 776\ \text{J}}$$

> Den reversible veien gir **mer enn dobbelt så mye arbeid** som ekspansjonen ved konstant (lavt) sluttrykk — 776 J mot 355 J. Det er samme prinsipp som i den forrige ekspansjonsoppgaven: ved konstant trykk skyver gassen bare mot det lave sluttrykket hele veien, mens den reversible prosessen lar gassen presse mot maksimal motstand i hvert øyeblikk. Reversibel ekspansjon gir alltid maksimalt arbeid mellom to gitte tilstander.
