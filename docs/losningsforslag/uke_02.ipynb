# Uke 2 – Statistisk mekanikk, energi, entropi og temperatur

## Diskusjonsoppgaver

### Oppgave 1

**a)** Termodynamikkens 2. lov kan formuleres på flere ekvivalente måter. For et **isolert system** gjelder

$$\Delta S\ge0$$

Entropien øker i en spontan prosess og er konstant i det reversible grensetilfellet. For et system som ikke er isolert, kan entropien i selve systemet avta så lenge den totale entropien til system og omgivelser øker.

**b)** For en reversibel varmeoverføring er den makroskopiske definisjonen

$$dS=\frac{\delta q_\mathrm{rev}}{T}$$

der $dS$ er en liten endring i entropi, $\delta q_\mathrm{rev}$ er en liten mengde varme som overføres reversibelt, og $T$ er temperaturen. Entropi er en tilstandsfunksjon, mens varme er energi som overføres under en prosess.

**c)** Den mikroskopiske definisjonen for like sannsynlige mikrotilstander er

$$S=k_B\ln\Omega$$

der $\Omega$ er multiplisiteten, altså antallet mikrotilstander som er forenlige med makrotilstanden.

**d)** Med 10 bokser og 3 identiske kuler finnes det langt flere måter å spre kulene over flere bokser enn å plassere alle tre i samme boks. Alle tre i én boks kan realiseres på 10 måter. To i én boks og én i en annen kan realiseres på $10\cdot9=90$ måter, mens én kule i hver av tre forskjellige bokser kan realiseres på $\binom{10}{3}=120$ måter. Dersom mikrotilstandene er like sannsynlige, er spredte fordelinger derfor langt mer sannsynlige.

**e)** Boksene er en modell for mulige mikroskopiske tilstander eller plasseringer. De kan for eksempel representere ulike områder i rommet, gitterplasser eller andre tillatte tilstander, avhengig av modellen.

**f)** Eksempler på frihetsgrader og energibidrag er translasjon, rotasjon og vibrasjon. Elektroniske tilstander og ulike romlige konfigurasjoner kan også øke antallet tilgjengelige mikrotilstander og dermed entropien.

**g)** Entropi $S$ er en **ekstensiv** størrelse: dobler vi mengden stoff i et ellers tilsvarende system, dobles entropien. Molar entropi $S_m=S/n$ er derimot **intensiv**.

**h)** Ved likevekt er den totale entropien maksimal under de gitte betingelsene. For et isolert system innebærer dette at systemet befinner seg nær den makrotilstanden som kan realiseres på overveldende flest mikroskopiske måter.

### Oppgave 2

**a)** Vi har tre plasser og tre skjelnbare kuler. Antallet mulige plasseringer er

$$3!=6$$

Hvis alle seks mikrotilstandene er like sannsynlige, er sannsynligheten for akkurat den viste konfigurasjonen $1/6$. Entropien er

$$S=k_B\ln6\approx2.47\times10^{-23}\ \mathrm{J/K}$$

**b)** For tre identiske kuler som skal okkupere tre av $N$ plasser, er antallet konfigurasjoner

$$\Omega(N,3)=\binom{N}{3}$$

| Alternativ | a | b | c | d |
|---|---:|---:|---:|---:|
| $N$ | 3 | 4 | 5 | 6 |
| $\Omega$ | 1 | 4 | 10 | 20 |
| $S$ [J/K] | 0 | $1.91\times10^{-23}$ | $3.18\times10^{-23}$ | $4.13\times10^{-23}$ |

**c)** Fra a) til d) er

$$\Delta S=S_d-S_a=4.13\times10^{-23}\ \mathrm{J/K}$$

**d)** I tilfelle d) finnes 20 mikrotilstander. De tre kulene ligger alle direkte ved siden av hverandre i fire av dem: $(1,2,3)$, $(2,3,4)$, $(3,4,5)$ og $(4,5,6)$. Dermed

$$P(\text{alle tre sammenhengende})=\frac{4}{20}=\frac15$$

Sannsynligheten for at de **ikke** alle tre ligger direkte ved siden av hverandre er derfor

$$1-\frac15=\frac45$$

**e)** Modellen viser at det ofte finnes mange flere mikrotilstander der partiklene er spredt utover enn der de er samlet. I makroskopiske systemer blir forskjellen i antall muligheter enorm, og det er den statistiske bakgrunnen for at gasser sprer seg og blandinger ofte dannes spontant når energi og vekselvirkninger ellers tillater det.

### Oppgave 3

**a)** Stigningstallet $\partial S/\partial E=1/T$ er stort og positivt i starten av kurven og avtar mot null når vi nærmer oss toppunktet. Entropien øker altså mye for en liten energiøkning ved lave energier, og mindre når vi nærmer oss toppen.

**b)** Ved toppunktet er $\Omega(n)$ størst, så denne energien kan realiseres på flest mikrotilstander i selve to-nivå-systemet. Men dette er **ikke alene nok** til å bestemme systemets likevektsenergi dersom det står i kontakt med omgivelser ved en gitt temperatur. Da må vi også ta hensyn til hvor mange mikrotilstander omgivelsene har når de har avgitt eller mottatt den tilsvarende energien.

**c)** System og omgivelser utgjør til sammen et større, isolert totalsystem. Når energi flyttes mellom dem, endres både $S_\mathrm{system}$ og $S_\mathrm{omgivelser}$. Den spontane retningen er den som øker

$$S_\mathrm{tot}=S_\mathrm{system}+S_\mathrm{omgivelser}$$

Likevekt nås når $S_\mathrm{tot}$ er maksimal. Det er dette som leder til betingelsen om lik temperatur mellom system og reservoar.

**d)** Nær $n=0$ er stigningstallet $1/T$ stort, og temperaturen er derfor lav. Mot toppunktet går stigningstallet mot null, slik at $1/T\to0$ og temperaturen blir svært høy. For det ideelle, begrensede to-nivå-systemet svarer selve toppunktet til grensen $T\to\infty$.

**e)** Mikrotilstandene har bestemte energier og bestemmer hvor mange mikroskopiske muligheter som finnes. Den makroskopiske indre energien $U$ er et statistisk gjennomsnitt av mikrotilstandenes energier, mens entropien knytter seg til hvor mange mikrotilstander som er tilgjengelige og hvordan sannsynligheten er fordelt mellom dem. $U$ og $S$ er derfor makroskopiske størrelser som oppsummerer informasjon om mikroverdenen.

### Oppgave 4

**a)** I modeller med faste plasser eller andre faste delsystemer kan vi skille delsystemene fra hverandre fordi plass A og plass B er ulike deler av modellen. Da kan tilstanden faktoriseres over disse delsystemene. Identiske, mobile partikler er derimot fundamentalt uskjelnbare: å bytte to identiske partikler gir ikke en ny fysisk mikrotilstand. Dette er viktig i partisjonsfunksjonen fordi feil telling ellers kan overtelle mikrotilstander, i den klassiske idealgassgrensen typisk med en faktor $N!$.

**b)** Partisjonsfunksjonen er

$$Q=\sum_i e^{-E_i/k_BT}$$

Summen går over alle mikrotilstander. $Q$ normaliserer Boltzmann-vektene slik at

$$p_i=\frac{e^{-E_i/k_BT}}{Q}$$

og brukes videre til å beregne makroskopiske størrelser som indre energi og entropi.

**c)** Når flere mikrotilstander har samme energi, kan vi gruppere dem. Her er $\Omega(\Delta E_j)$ antallet mikrotilstander med energien $\Delta E_j$, altså degenerasjonen til energinivået. Dermed

$$Q=\sum_j\Omega(\Delta E_j)e^{-\Delta E_j/k_BT}$$

For tabellen blir

$$Q=1+4e^{-\Delta\varepsilon/k_BT}+6e^{-2\Delta\varepsilon/k_BT}+4e^{-3\Delta\varepsilon/k_BT}$$

Ved $\Delta\varepsilon/(k_BT)=1$ får vi

$$Q=1+4e^{-1}+6e^{-2}+4e^{-3}\approx3.48$$

Sannsynligheten for et helt **energinivå** er

$$P(\Delta E_j)=\frac{\Omega(\Delta E_j)e^{-\Delta E_j/k_BT}}{Q}$$

Derfor er

$$\frac{P(\Delta E=\Delta\varepsilon)}{P(\Delta E=0)}=4e^{-1}\approx1.47$$

Selv om hver enkelt mikrotilstand ved $\Delta E=\Delta\varepsilon$ har lavere Boltzmann-vekt enn grunntilstanden, finnes det fire slike mikrotilstander. Hele energinivået blir derfor mer sannsynlig i dette eksempelet.

Når $\Delta\varepsilon/(k_BT)\to0$, går alle Boltzmann-faktorene mot 1, slik at

$$Q\to1+4+6+4=15$$

Når $\Delta\varepsilon/(k_BT)\to\infty$, går Boltzmann-faktorene for alle eksiterte energier mot 0, slik at

$$Q\to1$$

Den indre energien er

$$U=\frac{1}{Q}\sum_j\Omega(\Delta E_j)\Delta E_j e^{-\Delta E_j/k_BT}$$

Ved $\Delta\varepsilon/(k_BT)=1$ blir

$$U=\frac{4e^{-1}+12e^{-2}+12e^{-3}}{3.48}\Delta\varepsilon\approx1.06\Delta\varepsilon$$

Til slutt

$$\frac{S}{k_B}=\ln Q+\frac{U}{k_BT}$$

Siden $k_BT=\Delta\varepsilon$ i dette tilfellet, får vi

$$\frac{S}{k_B}=\ln(3.48)+1.06\approx2.31$$

### Oppgave 5

**a)** En makrotilstand er her bestemt av hvor mange partikler som befinner seg på hvert energinivå, uten at vi spesifiserer nøyaktig hvilken partikkel som er hvor. Vi kan angi den som $(n_0,n_1,n_2,n_3)$. Venstre figur har $(5,0,0,0)$ og høyre har $(2,2,1,0)$.

**b)** Energien til én partikkel på nivå $E$ er $Ek_BT$. Venstre figur gir derfor

$$U=0$$

mens høyre figur gir

$$U=(2\cdot0+2\cdot1+1\cdot2)k_BT=4k_BT$$

**c)** For fordelingen $(n_0,n_1,n_2,n_3)$ er antallet mikrotilstander

$$\Omega=\frac{N!}{n_0!n_1!n_2!n_3!}$$

Venstre:

$$\Omega=1$$

Høyre:

$$\Omega=\frac{5!}{2!2!1!0!}=30$$

**d)** Entropien er $S=k_B\ln\Omega$. Dermed

$$S_\mathrm{venstre}=0$$

og

$$S_\mathrm{høyre}=k_B\ln30\approx3.40k_B\approx4.70\times10^{-23}\ \mathrm{J/K}$$

**e)** I akkurat disse to viste makrotilstandene har tilstanden med høyere energi også høyere entropi, fordi energien kan fordeles på flere måter. Det er imidlertid ikke en generell regel at entropien alltid øker med energien for alle modeller og alle energier; to-nivå-systemet i oppgave 3 viser nettopp et moteksempel etter entropimaksimum. Poenget er at både energien og antallet tilgjengelige mikrotilstander er viktige.

### Oppgave 6

**a)**

$$\Delta U=U_2-U_1=150\ \mathrm{J}-100\ \mathrm{J}=50\ \mathrm{J}$$

$$\Delta S=S_2-S_1=2.3\ \mathrm{J/K}-2.0\ \mathrm{J/K}=0.30\ \mathrm{J/K}$$

**b)**

$$\frac{1}{T}\approx\frac{\Delta S}{\Delta U}$$

slik at

$$T\approx\frac{\Delta U}{\Delta S}=\frac{50\ \mathrm{J}}{0.30\ \mathrm{J/K}}\approx167\ \mathrm{K}$$

**c)** Makrotilstand 2 har høyest entropi og dermed størst multiplisitet. Hvis de to makrotilstandene er de eneste tilgjengelige tilstandene i det isolerte systemet, er makrotilstand 2 derfor den mest sannsynlige.
