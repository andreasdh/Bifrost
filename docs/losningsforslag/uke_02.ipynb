# Uke 2 – Statistisk mekanikk, energi, entropi og temperatur

## Diskusjonsoppgaver

### Oppgave 1

**a)** Under er termodynamikkens 2.lov formulert på flere ulike måter:

* Entropien i et isolert system avtar aldri ved en forandring
* Enhver prosess fører til at entropien til et system enten øker eller forblir konstant
* $\Delta S \ge 0$
* Prosesser som minker entropien til et isolert system er umulige
* Et isolert system vil gjennom en prosess enten forbli like ordnet eller bli mer uordnet


**b)** Den makroskopiske definisjonen av entropi er 

$dS=\frac{\partial q}{T}$

der $dS$ er en liten endring i entropi, $\partial q$ er endringen i varme (overført varme) og $T$ er temperatur.

Husk at $S$ er en tilstandsfunksjon - den avhenger kun av start-tilstand og slutt-tilstand, ikke veien mellom disse. På en annen side er $q$ en stifunksjon - endringen avhenger av veien fra start-tilstand til slutt-tilstand. Derfor skriver vi $d$ for entropi $S$, og $\partial$ for varme $q$.


**c)** Den mikroskopiske definisjonen av entropi er

$S=k_B \ln{\Omega}$

der $S$ er entropi, $k_B$ er Boltzmanns konstant og $\Omega$ er multiplisitet (antall mikrotilstand som gir en bestemt makrotilstand).

Du har sett flere eksempler på systemer som har mange mikrotilstander, men hvor flere mikrotilstander resulterer i én spesielt makrotilstand. Det er nettopp dette antallet $\Omega$ beskriver.


**d)** Vi har 10 bokser og 3 kuler. Vi har tre valg: 3 i én boks, 2 i én boks og 1 i en annen, og 1 i hver sin boks. 

Skal vi ha alle kulene i samme boks, er det 10 bokser å velge mellom, og vi får

$10 \ \text{muligheter}$

Skal vi ha 2 kuler i én boks er det 10 bokser å velge mellom, og den siste kulen har da 9 bokser å velge mellom, og vi får

$10 \cdot 9 = 90 \ \text{muligheter}$

Skal vi ha 1 i hver sin boks, er det antall måter å velge 3 kuler på 10 plasser gitt ved

$\binom{10}{3}=120 \ \text{muligheter}$

Hva sier dette oss? Det er langt flere måter å spre kulene ut enn å holde de samlet. Hvis alle kombinasjonene har lik sannsylighet, er det derfor mest sannsynlig spredning av kulene.


**e)** Flere typer bevegelse (frihetsgrader) bidrar til entropien: translasjon (bevegelse gjennom rommet), rotasjon, vibrasjon, og elektroniske tilstander. I tillegg bidrar konfigurasjonen (hvordan partikler eller energi er fordelt i rommet). 


**f)** Ved likevekt er entropien maksimal for et isolert system. Systemet utvikler seg spontant mot den makrotilstanden med flest mikrotilstander (høyest $\Omega$), og likevekt er nettopp der denne er størst og entropien ikke lenger kan øke.

### Oppgave 2

**a)** Vi har 3 plasser og 3 skjelnbare kuler. På første plass har vi 3 valg av kule, på andreplass har vi 2 valg og på siste har vi kun 1 valg. Dette gir

$3! = 3 \cdot 2 \cdot 1 = 6 \ \text{mikrotilstander}$

Dette er antall mikrotilstander for systemet, altså $\Omega=6$.

Hvis vi antar at alle mulighetene er like sannsynlige, er sannsynligheten for å finne akkurat den spesifikke konfigurasjonen på bildet lik:

$\frac{1}{6}$

Entropien til systemet blir

$S=k_B \ln{\Omega}=(1.38 \cdot 10^{-23} \ \text{J}/\text{K}) \cdot \ln{6}=2.47 \cdot 10^{-23} \ \text{J}/\text{K}$


**b)** Vi har nå 3 identiske kuler. Husk at vi kan finne antall måter å velge $n$ objekter fra $N$ objekter når rekkefølgen ikke spiller noen rolle ved binomialkoeffisienten, altså

$\Omega(N,n)=\binom{N}{n}=\frac{N!}{n!(N-n)!}$

I vårt tilfelle skal vi velge ut 3 plasser (til 3 kuler) av forskjellige antall plasser for hvert alternativ a)-d). Rekkefølgen spiller ikke noen rolle fordi kulene er identiske. 

Entropien finner vi ved $S=k_B \ln{(\Omega(N,n))}$

| Alternativ | a | b | c | d |
| ---------- | - | - | - | - |
| $N$ | $3$ | $4$ | $5$ | $6$ |
| $\Omega(N,3)=\binom{N}{3}$ | $\binom{3}{3}=1$ | $\binom{4}{3}=4$ | $\binom{5}{3}=10$ | $\binom{6}{3}=20$ |
| $S \ [\text{J}/\text{K}]$ | $0$ | $1.91 \cdot 10^{-23}$ | $3.18 \cdot 10^{-23}$ | $4.13 \cdot 10^{-23}$ |


**c)** Siden $S$ er en tilstandsfunksjon kan vi finne $\Delta S$ som differansen mellom slutt-entropi og start-entropi, det vil si

$\Delta S = S_{slutt}-S_{start}=4.13 \cdot 10^{-23} \ \text{J}/\text{K} - 0 \ \text{J}/\text{K} = 4.13 \cdot 10^{-23} \ \text{J}/\text{K}$


**d)** Vi har 6 posisjoner og ønsker å ha de 3 kulene ved siden av hverandre. Da kan vi ha følgende konfigurasjoner:

(1,2,3), (2,3,4), (3,4,5), (4,5,6)

Av de totalt $20$ mikrotilstandene, er det derfor $4$ av de som svarer til at kulene ligger ved siden av hverandre. Hvis vi antar at alle mikrotilstandene er like sannsynlige, får vi

$P(3 \ \text{kuler ved siden av hverandre})=4 \cdot \frac{1}{20}=\frac{1}{5}$

Den motsatte hendelsen er at kulene er spredt utover (minst 1 mellomrom mellom kulene), og vi får

$P(\text{kulene er spredt utover})=1-P(3 \ \text{kuler ved siden av hverandre})=1-\frac{1}{5}=\frac{4}{5}$


**e)** Flere av mikrotilstandene svarer til at kulene er blandet enn at de ligger ved siden av hverandre, og derfor er det overveiende mer sannsynlig at stoffer blander seg og fyller hele beholdere enn at de samles. Systemene vi ser på her er veldig små, så for reelle systemer med mange flere mikrotilstander blir sannsynligheten for blanding enda større.

### Oppgave 3

**a)** Stigningstallet $\partial S/\partial E = 1/T$ er stort og positivt i starten av kurven (bratt oppover), og synker gradvis mot null når vi nærmer oss toppunktet. Entropien øker altså raskt med energi i begynnelsen, men saktere og saktere jo nærmere toppen vi kommer.


**b)** Toppunktet er der $\Omega(n)$ er størst, altså der det finnes flest mikrotilstander. Siden systemet spontant søker mot makrotilstanden med flest mikrotilstander (høyest sannsynlighet), er dette nettopp likevektstilstanden — der entropien er maksimal og ikke lenger kan øke. Fra sannsynlighetenes perspektiv, er det ikke umulig å komme seg vekk fra denne makrotilstanden, men sannsynligheten for det er såpass liten at det praktisk talt er umulig.


**c)** Systemet starter i en svært usannsynlig tilstand med bare én mikrotilstand ($n = 0$, $\Omega = 1$). Når det får utveksle energi fritt, vil $n$ øke over tid fordi det finnes langt flere mikrotilstander ved høyere $n$. Grafen viser at systemet beveger seg oppover kurven mot toppunktet, der $\Omega$ og entropien er størst, og blir værende der.


**d)** Nær $n = 0$ er stigningstallet $1/T$ stort, mens det nær toppunktet er nesten null. Et stort stigningstall betyr høy "drivkraft" mot likevekt: langt fra likevekt drives systemet sterkt mot større entropi, mens det nesten ved likevekt knapt har noen netto drivkraft igjen. Systemet endrer seg altså raskt i starten og stadig saktere når det nærmer seg toppen.


**e)** De makroskopiske størrelsene $U$ (indre energi) og $\Delta S$ (entropiendring) er bygget opp fra mikroskopiske størrelser: $U$ er summen av energipakkene $\varepsilon_n$ over alle partikler, og $S = k_B \ln \Omega$ kommer direkte fra antall mikrotilstander $\Omega$. Det er riktig å si at de er "informasjon om mikroverdenen oversatt til noe vi kan måle", fordi $\Omega$ og $\varepsilon_n$ hører til den usynlige mikroskopiske skalaen, mens $U$ og $\Delta S$ er nettopp de målbare størrelsene i laboratoriet som disse mikroskopiske egenskapene summerer seg opp til.

### Oppgave 4

**a)** Skjelnbare partikler kan skilles fra hverandre (de har fast, identifiserbar posisjon, som atomene i et fast stoff), mens uskjelnbare partikler er identiske og kan ikke merkes fra hverandre (som molekyler i en gass). Vi skiller mellom dem fordi det påvirker tellingen av mikrotilstander: for uskjelnbare partikler gir det ingen ny mikrotilstand å bytte om på to like partikler, så vi må ikke telle slike ombytter som distinkte. Dette er viktig når vi setter opp partisjonsfunksjonen, fordi vi ellers overteller tilstandene (typisk med en faktor $N!$) og får feil entropi og fri energi.


**b)** Partisjonsfunksjonen 

$Q = \sum_i e^{-\epsilon_i / k_B T}$

er en sum over alle tilgjengelige mikrotilstander, vektet med Boltzmann-faktoren, og fungerer som en normeringsfaktor som sørger for at sannsynlighetene summerer til 1. Den brukes til å regne ut sannsynligheten for hver tilstand ($P_i = e^{-\epsilon_i/k_BT}/Q$) og til å utlede makroskopiske størrelser som indre energi, entropi og fri energi. Kort sagt inneholder $Q$ all statistisk informasjon om systemet, og er broen mellom mikrotilstandene og de målbare termodynamiske størrelsene. Husk at vi bruker $q$ for én partikkel, og $Q$ for systemer med mange partikler.


**c)**

* Vi ønsker å sette opp partisjonsfunksjonen $Q$ for systemet uttrykt ved energidifferanser, gitt verdiene i tabellen og at $\Delta \varepsilon / k_B T = 1$. 

Vi starter med definisjonen av $Q$ som en sum av Boltzmann-faktorer over **alle enkeltmikrotilstander** $i$:

$Q = \sum_{i} e^{-E_i / k_B T}$

Energidifferansen $\Delta E_i$ er differansen mellom mikrotilstandens energi $E_i$ og grunntilstandsenergien $E_0$. Vi finner et uttrykk for $E_i$:

$\Delta E_i = E_i - E_0 \Longrightarrow E_i = \Delta E_i + E_0$ 

For å forenkle uttrykket enda mer definerer $E_0 =0$, slik at

$\Delta E_i = E_i$

Vi setter uttrykket inn i $Q$ og får

$Q = \sum_{i}  e^{-\Delta E_i / k_B T}$

Husk nå at dette er summen over **alle mikrotilstander**. Mange mikrotilstander deler samme $\Delta E$. Vi grupperer derfor summen etter de distinkte energiverdiene og lar $\Omega(\Delta E)$ være antallet mikrotilstander med den energien:

$Q = \sum_{j} \Omega(\Delta E_j)\, e^{-\Delta E_j/k_BT}$

Her refererer $j$ til makrotilstanden. Uttrykkene vi får er dermed

$Q = \Omega(\Delta E_0)\, e^{-\Delta E_0 / k_B T} + \Omega(\Delta E_1)\, e^{-\Delta E_1 / k_B T} + \Omega(\Delta E_2)\, e^{-\Delta E_2 / k_B T} + \Omega(\Delta E_3)\, e^{-\Delta E_3 / k_B T}$

$Q = 1 + 4 e^{-\Delta \varepsilon / k_B T} + 6 e^{-2 \Delta \varepsilon / k_B T} + 4 e^{-3 \Delta \varepsilon / k_B T}$

Ved $\Delta \varepsilon / k_B T = 1$ blir verdien

$Q= 1 + 4 \cdot e^{-1} + 6 \cdot e^{-2} + 4 \cdot e^{-3}=3.48$



* Vi ønsker å finne $\frac{p(\Delta E_1)}{p(\Delta E_0)}$ fra partisjonsfunksjonen $Q$.

Fra Boltzmanns fordelingslov har vi at sannsynligheten for en **mikrotilstand** $i$ er gitt ved

$p(E_i) = \frac{1}{Q} \, e^{-E_i/k_B T} = \frac{1}{Q} \, e^{-\Delta E_i/k_B T} = p(\Delta E_i)$

Sannsynligheten for en **makrotilstand** (energi) $j$ finner vi ved å summere alle sannsynlighetene for mikrotilstanden som gir den energien, med andre ord, multiplisere med multiplisiteten $\Omega (E_j)$. Vi får dermed

$p(E_j) = \frac{1}{Q} \, \Omega(\Delta E_j) \, e^{-\Delta E_j/k_B T}$

Forholdet er dermed gitt ved

$\frac{p(\Delta E_1)}{p(\Delta E_0)} = \frac{\frac{1}{Q} \, \Omega(\Delta E_1) \, e^{-\Delta E_1/k_B T}}{\frac{1}{Q} \, \Omega(\Delta E_0) \, e^{-\Delta E_0/k_B T}}=\Omega(\Delta E_1) \, e^{-\Delta E_1/k_B T} = 4 \cdot e^{-\Delta \varepsilon / k_B T} = 4 \cdot e^{-1}$

$\frac{p(\Delta E_1)}{p(\Delta E_0)} = 1.47$



* Vi ønsker å se på $Q$ i grensetilfellene $\Delta \varepsilon / k_B T \to 0$ og $\Delta \varepsilon / k_B T \to \infty$.

Husk at uttrykkene vi fant for $Q$ er

$Q = \Omega(\Delta E_0)\, e^{-\Delta E_0 / k_B T} + \Omega(\Delta E_1)\, e^{-\Delta E_1 / k_B T} + \Omega(\Delta E_2)\, e^{-\Delta E_2 / k_B T} + \Omega(\Delta E_3)\, e^{-\Delta E_3 / k_B T} = 1 + 4 e^{-\Delta \varepsilon / k_B T} + 6 e^{-2 \Delta \varepsilon / k_B T} + 4 e^{-3 \Delta \varepsilon / k_B T}$

Når $\Delta \varepsilon / k_B T \to 0$ (høy $T$), vil $e^{\Delta E_i \varepsilon / k_B T} \to 1$ for alle $E_i$, slik at

$Q \to \Omega(\Delta E_0) + \Omega(\Delta E_1) + \Omega(\Delta E_2) + \Omega(\Delta E_3) = 1 + 4 + 6 + 4 = 15$

Dette er summen av alle mikrotilstandene, som betyr at ved høy $T$ blir alle mikrotilstander like tilgjengelige.

Når $\Delta \varepsilon / k_B T \to \infty$ (lav $T$), vil $e^{\Delta E_i \varepsilon / k_B T} \to 0$ for alle $E_i$, slik at

$Q \to \Omega(\Delta E_0) + \Omega(\Delta E_1) \cdot 0 + \Omega(\Delta E_2) \cdot 0 + \Omega(\Delta E_3) \cdot 0 = 1 + 0 + 0 + 0 = 1$

Dette er grunntilstanden, og bare denne er tilgjengelig ved lav $T$.


* Vi ønsker å finne indre energi $U$ og entropien $S$.

Vi har at 

$U = \langle E \rangle = \ = \langle \Delta E \rangle = \sum_{i} \Delta E_{i} \, p(\Delta E_i) = \sum_{j} \Delta E_{j} \left( \frac{1}{Q} \, \Omega(\Delta E_j) \, e^{-\Delta E_j/k_B T} \right) = \frac{1}{Q} \sum_{j} \Delta E_j \, \Omega(\Delta E_j) \, e^{-\Delta E_j/k_B T}$

Dermed får vi

$U = \frac{1}{3.48} \left( \Delta E_0 \, \Omega(\Delta E_0) \, e^{-\Delta E_0/k_B T} + \Delta E_1 \, \Omega(\Delta E_1) \, e^{-\Delta E_1/k_B T} + \Delta E_2 \, \Omega(\Delta E_2) \, e^{-\Delta E_2/k_B T} + \Delta E_3 \, \Omega(\Delta E_3) \, e^{-\Delta E_3/k_B T} \right)$

$U = \frac{1}{3.48} \left( 0 \Delta \varepsilon \cdot 0 \cdot 1 + \Delta \varepsilon \cdot 4 \cdot e^{-1} + 2\Delta \varepsilon \cdot 6 \cdot e^{-2} + 3\Delta \varepsilon \cdot 4 \cdot e^{-3} \right) = \frac{1}{3.48} \left( 3.69 \Delta \varepsilon \right)$

$U = 1.060 \, \Delta \varepsilon$

Videre har vi at

$S = k_B \ln Q + \frac{U}{T} \Longrightarrow \frac{S}{k_B} = \ln Q + \frac{U}{k_B T}$

Husk at vi har

$\Delta \varepsilon / k_B T = 1 \Longrightarrow k_B T = \Delta \varepsilon$

slik at vi får 

$S = \ln (3.48) + \frac{1.060 \, \Delta \varepsilon}{1 \, \Delta \varepsilon}$

$S = 2.31$

### Oppgave 5

**a)** En makrotilstand er den overordnede beskrivelsen av systemet: hvor mange partikler som befinner seg på hvert energinivå, uten å bry seg om nøyaktig hvilken partikkel som ligger hvor. Den angis ved fordelingen $(n_0, n_1, n_2, n_3)$, altså antall partikler i $E=0$, $E=1$, $E=2$ og $E=3$. Venstre figur har makrotilstanden $(5,0,0,0)$, mens høyre har $(2,2,1,0)$. En mikrotilstand er derimot den fullstendige beskrivelsen, der vi også vet hvilken boks hver enkelt partikkel ligger i.


**b)** Energien til én partikkel på nivå $E$ er $E k_B T$, så systemets indre energi er summen over alle partiklene, $U=\sum_i n_i \cdot E_i k_B T$.

Venstre: alle fem partikler ligger i $E=0$, så

$U = 5 \cdot 0 \cdot k_B T = 0$

Høyre: to i $E=0$, to i $E=1$, én i $E=2$:

$U = (2\cdot 0 + 2 \cdot 1 + 1 \cdot 2)\,k_B T = 4\,k_B T$


**c)** Antall mikrotilstander er antall måter å fordele de fem skjelnbare partiklene på slik at makrotilstanden bevares. Med fordelingen $(n_0,n_1,n_2,n_3)$ er dette

$\Omega = \frac{N!}{n_0!\,n_1!\,n_2!\,n_3!}$

som er nettopp det uordnede utvalget fra kombinatorikkdelen.

Venstre: $\Omega = \dfrac{5!}{5!\,0!\,0!\,0!} = 1$. Det finnes bare én måte å ha alle i grunntilstanden.

Høyre: $\Omega = \dfrac{5!}{2!\,2!\,1!\,0!} = \dfrac{120}{2 \cdot 2 \cdot 1} = 30$.


**d)** Entropien er $S = k_B \ln \Omega$.

Venstre: $S = k_B \ln 1 = 0$.

Høyre: $S = k_B \ln 30 \approx 3{,}40\,k_B \approx 4{,}70 \times 10^{-23} \text{ J}/\text{K}$ .


**e)** Sammenligningen viser at høyere indre energi henger sammen med høyere entropi. Venstretilstanden har lavest mulig energi ($U=0$), men også lavest mulig entropi ($S=0$), fordi det bare finnes én eneste måte å realisere den på - alle partiklene må ligge i grunntilstanden. Når energien øker, blir det flere måter å fordele den på mellom partiklene, og dermed flere mikrotilstander og høyere entropi.

Dette er kjernen i termodynamikken: et system trekkes ikke bare mot lav energi, men mot mange mikrotilstander. Grunntilstanden er energimessig gunstig, men entropisk fattig, og det er balansen mellom disse to som avgjør hvilken makrotilstand systemet faktisk havner i ved en gitt temperatur - akkurat den avveiningen Boltzmann-fordelingen beskriver.

### Oppgave 6

**a)** Både indre energi og entropi er tilstandsfunksjoner, slik at vi kan finne $\Delta U$ og $\Delta S$ ved å ta differansen mellom sluttverdien og startverdien. Vi får dermed

$\Delta U = U_2 - U_1 = 150 \text{ J}-100 \text{ J} = 50.0 \text{ J}$

$\Delta S = S_2 - S_1 = 2.3 \text{ J}/\text{K} - 2.0 \text{ J}/\text{K}=0.30 \text{ J}/\text{K}$


**b)** Vi skal estimere temperaturen $T$, og kan gjør dette ved å

$\frac{1}{T} \approx \frac{\Delta S}{\Delta U} \Longrightarrow T \approx \frac{\Delta U}{\Delta S} = \frac{50.0 \text{ J}}{0.30 \text{ J}/\text{K}} = 167 \text{ K}$


**c)** Siden makrotilstand 2 har høyere entropi, vil systemet mest sannsynlig befinne seg der. At entropien er høyere der betyr at det er flere mikrotilstander som svarer til den makrotilstanden.
