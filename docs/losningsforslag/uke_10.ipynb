# Uke 10 – Likevekt

## Diskusjonsoppgave

## Oppgave 1 

**a)**  Det finnes tre grunnleggende typer likevekt: 
* Termisk likevekt drives av temperaturforskjeller. Så lenge to systemer har ulik temperatur, strømmer det varme mellom dem, helt til temperaturen er lik. 
* Mekanisk likevekt drives av trykkforskjeller. Har to deler ulikt trykk, vil en bevegelig grense (f.eks. et stempel eller en membran) lfytter seg til trykket er utlignet. 
* Kjemisk (eller materiell) likevekt drives av forskjeller i kjemisk potensial mu. Stoffer strømmer, enten ved forflytning eller ved kjemisk reaksjon, fra høyt til lavt kjemisk potensial, til mi er likt. 
* Drivkreftene er altså temperatur (T), trykk (P) og kjemisk potensial (mi).

**b)**
Temperatur er et mål på partiklenes gjennomsnittlige kinetiske energi. Varmestrøm er rett og slett overføring av bevegelsesenergi gjennom kollisjoner: raske «varme» partikler gir energi til langsommere «kalde» til fordelingen jevner seg ut.  

Trykk oppstår mikroskopisk fra partiklenes støt mot en vegg, som er summer av bevegelsesmengde som overføres per areal per tid. En trykkforskjell betyr at det er flere/kraftigere støt på én side, så grensen dyttes. 

Kjemisk potensial beskriver hvor mye systemets (Gibbs-)energi endrer seg når man tilfører én partikkel. Det har både en energidel (bindinger og vekselvirkninger) og en entropidel (hvor mange måter partiklene kan fordeles på, altså konsentrasjon/blanding). Partikler har en tendens til å spre seg ut og til å søke lavere energi, begge deler er «bakt inn» i mu. Man kan tenke på mu som en rømningstendens.

**c)**
Da minimeres Gibbs fri energi. Kravet for likevekt er at G er på et minimum, altså dG = 0.

**d)**
Da minimeres Helmholtz fri energi F. Kravet er da dF = 0.

**e)**
Her maksimeres entropien S. Kravet er at S er på et maksimum, altsp dS = 0.

Dette står i likt linje med c) og d) ettersom alle tre beskriver samme lov: termodynamikkens 2.lov, som sier at den totale entropien (system + omgivelsen) skal være maksimal ved likevekt. Forskjellen er bare hva vi holder fast og hva systemet får utveksle med omgivelsene. For det isolerte systemet ser vi direkte på systemets egen entropi, fordi det ikke er noen omgivelser å utveksle med. 

Ved konstant T,V og konstant T,P «pakker» vi omgivelsenes entropiendring inn i F og G. Disse størrelsene er konstruert nettopp slik at å minimere dem er det samme som å maksimere den totale entropien under de gitte betingelsene. 

**f)**
Ved kjemisk likevekt er hastigheten til for- og bakover reaksjonen like. Reaksjonene stopper altså ikke. De går like fort begge veier, slik at konsentrasjonene ikke endrer seg netto. 

**g)**
Alle disse likevektene er dynamiske. På makronivå ser det ut som «ingenting skjer», men på mikronivå pågår prosessene hele tiden: partikler kolliderer og utveksler energi, treffer veggen fra begge sider, og molekyler dannes og brytes ned. Det er bare netto-effekten som er null. 

Fra f): reaksjonene fortsetter hele tiden i begge retninger med lik hastighet. Molekyler dannes og brytes ned kontinuerlig; det er bare nettoendringen som er null. 

Fra b): partiklene beveger seg, kolliderer og utveksler energi hele tiden. Ved termisk likevekt overføres det fortsatt energi i hvert støt, det balanserer bare i gjennomsnitt. Enkeltpartiklene skifter stadig tilstand; det er kun de makroskopiske gjennomsnittene (T, P, konsentrasjoner) som holder seg konstante.

Dette skiller seg fra en «ekte» statisk likevekt i mekanikken, som en pendel i hvile i sitt lavest punkt – der det virkelig ikke er noen bevegelse i det hele tatt. Termodynamisk likevekt er altså en balanse mellom prosesser som stadig pågår, ikke fravær av prosesser.

## Oppgave 2

**a)** Reaksjons-Gibbsenergi og likevektskonstanten $K$

Sammenhengen mellom standard reaksjons-Gibbsenergi og likevektskonstanten:

$$\Delta_r G^\circ = -RT \ln K \qquad\Longleftrightarrow\qquad K = e^{-\Delta_r G^\circ / RT}$$

$K$ er ett fast tall ved en gitt temperatur, og forteller hvor likevekten ligger.

| Tilfelle | $\Delta_r G^\circ$ | $K$ | Ved likevekt |
|---|:---:|:---:|---|
| Forover favorisert | $< 0$ | $> 1$ | $[B] > [A]$ (produkt dominerer) |
| Bakover favorisert | $> 0$ | $< 1$ | $[B] < [A]$ (reaktant dominerer) |
| Balansert | $= 0$ | $= 1$ | $[B] = [A]$ |

Om reaksjonen faktisk går forover akkurat nå, bestemmes ikke av $K$ alene, men av
$$\Delta_r G = \Delta_r G^\circ + RT\ln Q$$
altså av om $Q$ er mindre eller større enn $K$. Tabellen over beskriver hvor likevekten til slutt havner — ikke retningen fra et vilkårlig startpunkt. (En reaksjon med $K < 1$ kan fortsatt gå forover en stund om man starter med ren $A$, siden $Q = 0 < K$.)

Reaksjoner med piler (relativ hastighet forover/bakover):

| Tilfelle | Hastigheter | Netto |
|---|---|---|
| Forover spontan | $v_\text{forover} \gg v_\text{bakover}$ | mot $B$ |
| Bakover spontan | $v_\text{bakover} \gg v_\text{forover}$ | mot $A$ |
| Likevekt | $v_\text{forover} = v_\text{bakover}$ | ingen netto endring |

Ved likevekt går reaksjonen fortsatt begge veier, like fort. Så konsentrasjonene står stille (dynamisk likevekt).

**b)** Reaksjonskvotienten $Q$

$$Q = \frac{[B]}{[A]} \quad\text{(eventuelt med partialtrykk eller aktiviteter)}$$

$Q$ er forholdet mellom produkt og reaktant på et gitt tidspunkt. Ved å sammenligne $Q$ med $K$ ser man hvilken vei reaksjonen går:

- $Q < K \;\Rightarrow\;$ netto forover
- $Q > K \;\Rightarrow\;$ netto bakover
- $Q = K \;\Rightarrow\;$ likevekt

**c)** Molekylær tolkning av $K$

Likevektskonstanten kan skrives som et produkt av to faktorer:

$$K = \underbrace{e^{-\Delta_r H / RT}}_{\text{entalpi-faktor}} \cdot \underbrace{e^{\Delta_r S / R}}_{\text{entropi-faktor}}$$

- Entalpi-faktoren trekker mot siden med lavest energi.
- Entropi-faktoren trekker mot siden med flest tilgjengelige (tett-liggende) energinivåer, altså flest mikrotilstander.

I begge grafene ligger energinivåene til $B$ høyere enn til $A$, så reaksjonen $A \to B$ er endoterm: $\Delta_r H > 0$. Det gjør entalpi-faktoren $< 1$, slik at den favoriserer $A$. Da er det entropien(nivåtettheten) som avgjør utfallet.

*Graf (a): $A$ dominerer*

$A$ og $B$ har omtrent samme avstand mellom energinivåene → omtrent lik nivåtetthet → $\Delta_r S \approx 0$. Entropi-faktoren blir $\approx 1$ og gir ingen hjelp til $B$.

Entalpi-faktoren vinner:

$$\Delta_r H > 0,\quad \Delta_r S \approx 0 \;\Rightarrow\; \Delta_r G^\circ > 0 \;\Rightarrow\; K < 1$$

Likevekten er forskjøvet mot reaktanten $A$, og reaksjonen mot $B$ er ikke-spontan.

*Graf (b): $B$ dominerer*

$B$ har mye **tettere** energinivåer enn $A$ → mange flere tilgjengelige mikrotilstander for $B$ → stor positiv $\Delta_r S$. Entropi-faktoren blir så stor at den overgår den ugunstige entalpien:

$$\Delta_r H > 0,\quad \Delta_r S \gg 0 \;\Rightarrow\; \Delta_r G^\circ < 0 \;\Rightarrow\; K > 1$$

Selv om reaksjonen er endoterm, blir $B$ det dominerende molekylet ved likevekt, fordi entropien vinner.

Kort oppsummert

| Graf | Nivåtetthet | $\Delta_r S$ | Hvem vinner | Dominerer |
|:---:|---|:---:|---|:---:|
| (a) | lik for $A$ og $B$ | $\approx 0$ | entalpi | $A$ |
| (b) | mye høyere for $B$ | $\gg 0$ | entropi | $B$ |

## Oppgave 3

**a)** 
En endotermisk reaksjon om temperaturen økes:
- Når vi tilfører varme, forskyves likevekten mot produktsiden for å "forbruke" den tilførte varmen. Produktet favoriseres, og K øker. 

En eksotermisk reaksjon når temperaturen økes:
- Økt temperatur skyver likevekten mot reaktantsiden for å motvirke varmetilførselen. Reaktanten favoriseres, og K synker. 

Trykk og påvirking på likevekten:
- Svaret avhenger av differansen i antall mol gass mellom produktet- og reaktanten. Reduserer man volumet (øker trykket), forskyves likevekten mot siden med færrest gassmolekyler (reaktanten her), siden færre gasspartikler gir lavere trykk og dermed demper trykkøkningen. Øker man volumet (senker trykket), forskyves den mot siden med flest gassmolekyler (produktet her). Er antall mol gass likt på begge sider, skjer ingen forskyvning. 
*Det er endring i volum (og dermed partialtrykk/konsentrasjon) som teller. Å øke totaltrykket ved å tilsette en inert gass ved konstant volum endrer IKKE partialtrykkene til reaktantene og produktene, og gir derfor ingen forskyvning.*

Hvordan påvirker en katalysator eller et enzym en likevekt:
- En katalysator påvirker ikke likevekts posisjon, K forblir uendret. Begrunnelsen er at en katalysator senker aktiveringsenergien like mye for forover- og bakoverreaksjonen, så begge hastighetene øker med samme. faktor.
- Dette henger sammen med at K er termodynamisk bestemt, en tilstandsstørrelse som er uavhengig av reaksjonsveien.

:::{admonition} Mikroskopisk forklaring: Hvorfor virker temperatur motsatt for de to?
:class: important

Nøkkelen er at $K = k_{\text{forover}}/k_{\text{bakover}}$. Økt temperatur øker begge hastighetskonstantene, men ulikt, fordi de to retningene har ulik aktiveringsenergi:

$$\Delta H = E_a(\text{forover}) - E_a(\text{bakover})$$

Fra Boltzmann-fordelingen får den barrieren som er *høyest*, den største relative økningen når $T$ stiger. Det samme sier Arrhenius: $\dfrac{d(\ln k)}{dT} = \dfrac{E_a}{RT^2}$, altså vokser $k$ raskest for stor $E_a$.

- **Endoterm** ($\Delta H > 0$): forover har høyest barriere → $k_{\text{forover}}$ øker mest → $K$ øker → skift mot **produkt**.
- **Eksoterm** ($\Delta H < 0$): bakover har høyest barriere → $k_{\text{bakover}}$ øker mest → $K$ synker → skift mot **reaktant**.

Samme temperaturøkning gir altså motsatt forskyvning, fordi det alltid er retningen med den høyeste aktiveringsenergien som påvirkes mest.
:::

**b)**
Likevekten forskyves mot produktet: produktkonsentrasjonen stiger og reaktanten synker, før kurvene flater ut på en ny likevekt. Dette stemmer med teorien over, hvor for en endoterm reaksjon favoriserer økt temperatur produktsiden (K øker). Legg merker til at systemet finner en ny likevekt relativt raskt: forstyrrelsen dempes, men oppheves ikke.

**c)**
Nå går forskyvningen motsatt vei. Rett etter temperaturøkningen begynner konsentrasjonene å endre seg i motsatt retning av det endoterme tilfellet: produktet synker, og reaktanten stiger. Likevekten forskyves altså mot reaktantsiden (K synker), før systemet flater ut på en ny likevekt. 

**d)**
Ved likevekt er $Q = K$: reaksjonskvotienten har tatt igjen likevektskonstanten, og det er nettopp dette som *definerer* at systemet er i likevekt.

Reduserer du volumet (øker trykket), øker alle konsentrasjonene samtidig. Hvorvidt $Q$ havner over eller under $K$ rett etter endringen, avgjøres av støkiometrien, nærmere bestemt av differansen i antall mol gass mellom teller og nevner i $Q$-uttrykket. Er det flere gassmol på produktsiden, blir $Q$ større enn $K$; er det færre, blir $Q$ mindre.

I tilfellet du observerer blir $Q < K$ rett etter volumreduksjonen. Da må reaksjonen gå forover (mot produkt) for å øke $Q$ igjen, helt til $Q = K$ på nytt. Den generelle regelen er:

- $Q < K$ → reaksjonen går mot **produkt**
- $Q > K$ → reaksjonen går mot **reaktant**

Dette er konsistent med Le Chatelier: at reaksjonen går forover her, betyr nettopp at produktsiden har færrest gassmolekyler, og systemet forskyves derfor mot den siden som demper trykkøkningen.

:::{admonition} Fundamental forskjell mellom Q og K (makroskopisk vs. mikroskopisk)
:class: important

$Q$ og $K$ har samme matematiske form, men ulik rolle:

- $K$ er en fast størrelse ved gitt temperatur, termodynamisk bestemt gjennom $\Delta G^\circ = -RT \ln K$. Den forteller hvor likevekten ligger.
- $Q$ beregnes for konsentrasjonene slik de er i øyeblikket, uansett om systemet er i likevekt eller ikke. $Q$ er et «øyeblikksbilde» som beveger seg mot $K$.

Skillet mellom makroskopisk og mikroskopisk er verdt å trekke fram:

- **Makroskopisk** (det vi måler): ved likevekt er konsentrasjonene konstante, $Q = K$, og det ser ut som «ingenting skjer».
- **Mikroskopisk**: likevekten er *dynamisk*. Forover- og bakoverreaksjonen foregår fortsatt, men med like store hastigheter. Nettoendringen er null, men molekyler omdannes hele tiden i begge retninger.

Nettopp derfor kan vi skrive $K = k_{\text{forover}}/k_{\text{bakover}}$, som binder den makroskopiske likevektskonstanten sammen med de mikroskopiske hastighetskonstantene, og lukker sirkelen tilbake til temperaturforklaringen over.

:::

**e)**
Når man trykker «+A», hopper [A] opp momentant. Da synker Q = [B]/[A] under K. Systemet svarer med å kjøre foroverreaksjonen (A → B), slik at [B] stiger i tiden etterpå (og en del av den tilsatte A-en forbrukes), helt til Q = K igjen, en ny likevekt. 
	
Gjentar man fra en ny likevekt med «+B», skjer det motsatte: [B] hopper opp, Q blir større enn K, bakoverreaksjonen (B → A) kjører, og [A] stiger til Q = K på nytt.

Systemet skifter alltid mot den motsatte siden for delvis å forbruke det man tilsatte. Dette er Le Chatelier i konsentrasjonsform, og det kobler rett på Q vs. K bildet fra oppgaven over. Forstyrrelsen flytter Q vekk fra K, og reaksjonen går den veien som fører Q tilbake til K.


**f)**

Katalysatoren endrer bare hvor fort likevekten nås, ikke hvor den ligger. Slår man den på og gjentar for eksempel en temperaturendring. Ser du at systemet finner den nye likevekten mye raskere: uten katalysator får man et tydelig tilpasningsvindu der konsentrasjonene gradvis skifter mot den nye posisjonen, mens med katalysator kortes denne transienten kraftig ned, Q innhenter K nesten momentant. 

Selve likevektsposisjonen er derimot den samme. Forholdet mellom [A] og [B] som systemet ender på, er uendret enten katalysatoren er av eller på, og verdien for K i statuslinjen holder seg konstant mens man slår den av og på. Det stemmer med teorien: en katalysator senker aktiveringsenergien like mye begge veier, så $k_{\text{forover}}$, og $k_{\text{bakover}}$, øker med samme faktor og forholdet $K = k_{\text{forover}}/k_{\text{bakover}}$ blir uforandret.

## Oppgave 4

$$\text{N}_2\text{O}_4(g) \;\rightleftharpoons\; 2\,\text{NO}_2(g)$$

**Gitt:** $T = 25\ ^{\circ}\text{C} = 298{,}15\ \text{K}$, $\ p = 0{,}597\ \text{bar}$, $\ \rho = 1{,}477\ \text{g/dm}^3$
$M_{\text{N}_2\text{O}_4} = 92{,}02\ \text{g/mol}$, $\quad M_{\text{NO}_2} = 46{,}01\ \text{g/mol}$

1. Sett opp likevekten med dissosiasjonsgrad α

Vi starter med $n_0$ mol N₂O₄ og lar en andel $\alpha$ dissosiere:

| | N₂O₄ | NO₂ |
|---|---|---|
| Start | $n_0$ | $0$ |
| Endring | $-n_0\alpha$ | $+2n_0\alpha$ |
| Likevekt | $n_0(1-\alpha)$ | $2n_0\alpha$ |

Totalt antall mol: $\;n_\text{tot} = n_0(1+\alpha)$.

2. Finn α fra tettheten

Trikset: **massen er bevart** når N₂O₄ spaltes, så gjennomsnittlig molmasse blir

$$\bar{M} = \frac{m}{n_\text{tot}} = \frac{n_0\,M_{\text{N}_2\text{O}_4}}{n_0(1+\alpha)} = \frac{M_{\text{N}_2\text{O}_4}}{1+\alpha}$$

Samtidig gir idealgassloven ($\rho = \tfrac{m}{V}$ og $pV = nRT$):

$$\bar{M} = \frac{\rho RT}{p} = \frac{1{,}477 \cdot 0{,}08314 \cdot 298{,}15}{0{,}597} = 61{,}3\ \text{g/mol}$$

Setter de to uttrykkene like og løser for α:

$$1+\alpha = \frac{M_{\text{N}_2\text{O}_4}}{\bar{M}} = \frac{92{,}02}{61{,}3} = 1{,}50 \quad\Rightarrow\quad \boxed{\alpha = 0{,}50}$$

3. Molfraksjoner

$$x_{\text{N}_2\text{O}_4} = \frac{1-\alpha}{1+\alpha} = \frac{0{,}5}{1{,}5} = \tfrac{1}{3} = 0{,}333$$

$$x_{\text{NO}_2} = \frac{2\alpha}{1+\alpha} = \frac{1{,}0}{1{,}5} = \tfrac{2}{3} = 0{,}667$$

4. Kₓ (fra molfraksjoner)

$$K_x = \frac{x_{\text{NO}_2}^2}{x_{\text{N}_2\text{O}_4}} = \frac{(2/3)^2}{1/3} = \frac{4}{3} = \boxed{1{,}33}$$

5. K_p (fra partialtrykk)

$$p_{\text{NO}_2} = x_{\text{NO}_2}\,p = 0{,}667 \cdot 0{,}597 = 0{,}398\ \text{bar}, \qquad p_{\text{N}_2\text{O}_4} = 0{,}333 \cdot 0{,}597 = 0{,}199\ \text{bar}$$

$$K_p = \frac{(p_{\text{NO}_2}/p^\circ)^2}{p_{\text{N}_2\text{O}_4}/p^\circ} = \frac{0{,}398^2}{0{,}199} = \boxed{0{,}80} \qquad (p^\circ = 1\ \text{bar})$$

Sjekk: $K_p = K_x \cdot (p/p^\circ)^{\Delta n} = 1{,}33 \cdot 0{,}597 = 0{,}80$ ✓ (her er $\Delta n = 1$).

6. K_c (fra konsentrasjoner)

Total konsentrasjon er intensiv, så vi trenger ikke velge noe volum:

$$c_\text{tot} = \frac{p}{RT} = \frac{0{,}597}{0{,}08314 \cdot 298{,}15} = 0{,}0241\ \text{mol/dm}^3$$

$$[\text{NO}_2] = x_{\text{NO}_2}\,c_\text{tot} = 0{,}01605\ \text{mol/dm}^3, \qquad [\text{N}_2\text{O}_4] = x_{\text{N}_2\text{O}_4}\,c_\text{tot} = 0{,}00803\ \text{mol/dm}^3$$

$$K_c = \frac{([\text{NO}_2]/c^\circ)^2}{[\text{N}_2\text{O}_4]/c^\circ} = \frac{0{,}01605^2}{0{,}00803} = \boxed{0{,}0321} \qquad (c^\circ = 1\ \text{mol/dm}^3)$$

7. Hva skjer når man tilsetter He?

Når He tilsettes for å øke trykket, skjer det **ved konstant volum og temperatur**. Partialtrykkene til de reagerende gassene avhenger bare av deres egne stoffmengder og volumet:

$$p_{\text{NO}_2} = \frac{n_{\text{NO}_2}RT}{V}, \qquad p_{\text{N}_2\text{O}_4} = \frac{n_{\text{N}_2\text{O}_4}RT}{V}$$

He endrer verken $n_{\text{NO}_2}$, $n_{\text{N}_2\text{O}_4}$ eller $V$, så disse partialtrykkene er uendret. Da er reaksjonskvotienten fortsatt lik $K_p$:

$$Q = \frac{(p_{\text{NO}_2}/p^\circ)^2}{p_{\text{N}_2\text{O}_4}/p^\circ} = K_p$$

**Likevekten forskyves ikke.** He er en tilskuer — den øker det *totale* trykket, men ikke partialtrykkene til reaktantene/produktene.

> **Til sammenligning:** Hadde He blitt tilsatt ved *konstant totaltrykk* (slik at volumet må øke), ville gassene blitt fortynnet, og likevekten ville forskjøvet seg mot flere gassmolekyler — altså mot NO₂ (mer dissosiasjon). Det er verdt å nevne, siden forskjellen mellom «konstant $V$» og «konstant $p$» er en klassisk felle.


$$\alpha = 0{,}50 \qquad K_c = 0{,}032 \qquad K_p = 0{,}80 \qquad K_x = 1{,}33$$

Likevekten forskyves **ikke** når trykket økes ved å tilsette He (konstant volum).


## Oppgave 5

$$\text{Hg}(l) \;\rightleftharpoons\; \text{Hg}(g)$$

**Gitt** ved $25\ ^{\circ}\mathrm{C} = 298{,}15\ \mathrm{K}$:
$$\Delta_\text{vap}H^\circ = 62{,}0\ \mathrm{kJ/mol}, \qquad \Delta_\text{vap}S^\circ = 99{,}0\ \mathrm{J\,K^{-1}mol^{-1}}$$

**a)** Damptrykk ved 25 °C

Først standard Gibbs-energi for fordampningen:

$$\Delta_\text{vap}G^\circ = \Delta_\text{vap}H^\circ - T\,\Delta_\text{vap}S^\circ = 62{,}0\cdot 10^3 - 298{,}15\cdot 99{,}0 = +32{,}5\cdot 10^3\ \mathrm{J/mol}$$

At denne er **positiv** gir mening: ved standardtrykk (1 bar) er fordampning ikke spontan — Hg er tross alt væske ved romtemperatur.

Likevektskonstanten er $K_p = p_{\text{Hg}}/p^\circ$ (rein væske har aktivitet 1), så

$$\Delta_\text{vap}G^\circ = -RT\ln K_p \quad\Rightarrow\quad \ln\frac{p_{\text{Hg}}}{p^\circ} = -\frac{\Delta_\text{vap}G^\circ}{RT} = -\frac{32{,}5\cdot 10^3}{8{,}3145\cdot 298{,}15} = -13{,}10$$

$$\boxed{p_{\text{Hg}} = e^{-13{,}10}\cdot p^\circ = 2{,}0\cdot 10^{-6}\ \mathrm{bar}}$$

(≈ 0,2 Pa — svært lavt, men nok til at kvikksølvdamp er en reell helsefare i lukkede rom.)

**b)** Tilnærmet kokepunkt ved 1 atm

Ved kokepunktet står væske og damp i likevekt med $p_{\text{Hg}} = 1\ \mathrm{atm} \approx 1\ \mathrm{bar} = p^\circ$. Da er $K_p = 1$ og dermed

$$\Delta_\text{vap}G^\circ(T_\text{kp}) = -RT_\text{kp}\ln 1 = 0$$

Med antakelsen om at $\Delta_\text{vap}H^\circ$ og $\Delta_\text{vap}S^\circ$ er temperaturuavhengige:

$$0 = \Delta_\text{vap}H^\circ - T_\text{kp}\,\Delta_\text{vap}S^\circ \quad\Rightarrow\quad T_\text{kp} = \frac{\Delta_\text{vap}H^\circ}{\Delta_\text{vap}S^\circ} = \frac{62{,}0\cdot 10^3}{99{,}0}$$

$$\boxed{T_\text{kp} = 626{,}3\ \mathrm{K} \approx 353\ ^{\circ}\mathrm{C}}$$

Til sammenligning er det målte kokepunktet 357 °C — bra treff, med tanke på at vi antok konstante $\Delta H$ og $\Delta S$ over ~330 grader.

$$p_{\text{Hg}}(25\ ^{\circ}\mathrm{C}) = 2{,}0\cdot 10^{-6}\ \mathrm{bar}, \qquad T_\text{kp} = 626{,}3\ \mathrm{K}\ (353\ ^{\circ}\mathrm{C})$$




## Oppgave 6

$$\mathrm{PCl}_5(g) \;\rightleftharpoons\; \mathrm{PCl}_3(g) + \mathrm{Cl}_2(g)$$

**Gitt** ved $250\ ^{\circ}\mathrm{C} = 523{,}15\ \mathrm{K}$:
$$p = 1{,}00\ \mathrm{bar}, \qquad \rho = 2{,}66\ \mathrm{g/dm^3}$$

Molmasser: $M_{\mathrm{PCl}_5} = 208{,}24$, $\ M_{\mathrm{PCl}_3} = 137{,}33$, $\ M_{\mathrm{Cl}_2} = 70{,}91\ \mathrm{g/mol}$.

1. Sett opp likevekten med dissosiasjonsgrad α

Vi starter med $n_0$ mol PCl₅ og lar en andel $\alpha$ dissosiere:

| | PCl₅ | PCl₃ | Cl₂ |
|---|---|---|---|
| Start | $n_0$ | $0$ | $0$ |
| Likevekt | $n_0(1-\alpha)$ | $n_0\alpha$ | $n_0\alpha$ |

Totalt antall mol: $\;n_\mathrm{tot} = n_0(1+\alpha)$. Merk at hvert PCl₅ som spaltes gir **to** molekyler, så gassmengden øker.

2. Finn α fra tettheten

Massen er bevart under spaltingen, så gjennomsnittlig molmasse blir

$$\bar{M} = \frac{m}{n_\mathrm{tot}} = \frac{n_0\,M_{\mathrm{PCl}_5}}{n_0(1+\alpha)} = \frac{M_{\mathrm{PCl}_5}}{1+\alpha}$$

Idealgassloven gir også ($\rho = m/V$ og $pV = nRT$):

$$\bar{M} = \frac{\rho RT}{p} = \frac{2{,}66 \cdot 0{,}08314 \cdot 523{,}15}{1{,}00} = 115{,}7\ \mathrm{g/mol}$$

Setter uttrykkene like og løser for α:

$$1+\alpha = \frac{M_{\mathrm{PCl}_5}}{\bar{M}} = \frac{208{,}24}{115{,}7} = 1{,}80 \quad\Rightarrow\quad \boxed{\alpha = 0{,}80}$$

3. Molfraksjoner

$$x_{\mathrm{PCl}_5} = \frac{1-\alpha}{1+\alpha} = \frac{0{,}20}{1{,}80} = 0{,}111$$

$$x_{\mathrm{PCl}_3} = x_{\mathrm{Cl}_2} = \frac{\alpha}{1+\alpha} = \frac{0{,}80}{1{,}80} = 0{,}444$$

4. Kₓ (fra molfraksjoner)

$$K_x = \frac{x_{\mathrm{PCl}_3}\,x_{\mathrm{Cl}_2}}{x_{\mathrm{PCl}_5}} = \frac{\alpha^2}{1-\alpha^2} = \frac{0{,}80^2}{1-0{,}80^2} = \frac{0{,}64}{0{,}36} = \boxed{1{,}78}$$

5. K_p (fra partialtrykk)

Med $p_i = x_i\,p$ og $p = 1{,}00\ \mathrm{bar}$:

$$p_{\mathrm{PCl}_3} = p_{\mathrm{Cl}_2} = 0{,}444\ \mathrm{bar}, \qquad p_{\mathrm{PCl}_5} = 0{,}111\ \mathrm{bar}$$

$$K_p = \frac{(p_{\mathrm{PCl}_3}/p^\circ)(p_{\mathrm{Cl}_2}/p^\circ)}{p_{\mathrm{PCl}_5}/p^\circ} = \frac{0{,}444^2}{0{,}111} = \boxed{1{,}78} \qquad (p^\circ = 1\ \mathrm{bar})$$

Sjekk: $K_p = K_x\,(p/p^\circ)^{\Delta n} = 1{,}78 \cdot 1{,}00 = 1{,}78$ ✓ (her er $\Delta n = 1$, og siden $p = 1$ bar blir $K_p = K_x$ tallmessig).

6. K_c (fra konsentrasjoner)

Total konsentrasjon er intensiv, så vi trenger ikke velge noe volum:

$$c_\mathrm{tot} = \frac{p}{RT} = \frac{1{,}00}{0{,}08314 \cdot 523{,}15} = 0{,}0230\ \mathrm{mol/dm^3}$$

$$[\mathrm{PCl}_3] = [\mathrm{Cl}_2] = 0{,}444\,c_\mathrm{tot} = 0{,}01022\ \mathrm{mol/dm^3}, \qquad [\mathrm{PCl}_5] = 0{,}111\,c_\mathrm{tot} = 0{,}002554\ \mathrm{mol/dm^3}$$

$$K_c = \frac{([\mathrm{PCl}_3]/c^\circ)([\mathrm{Cl}_2]/c^\circ)}{[\mathrm{PCl}_5]/c^\circ} = \frac{0{,}01022^2}{0{,}002554} = \boxed{0{,}0409} \qquad (c^\circ = 1\ \mathrm{mol/dm^3})$$

Sjekk: $K_c = K_p\,\left(\dfrac{p^\circ}{c^\circ RT}\right)^{\Delta n} = 1{,}78 \cdot 0{,}0230 = 0{,}0409$ ✓

7. Hva skjer når man tilsetter Ar?

Når Ar tilsettes for å øke trykket, skjer det **ved konstant volum og temperatur**. Partialtrykkene til de reagerende gassene avhenger bare av deres egne stoffmengder og volumet:

$$p_i = \frac{n_i RT}{V}$$

Ar endrer verken $n_{\mathrm{PCl}_5}$, $n_{\mathrm{PCl}_3}$, $n_{\mathrm{Cl}_2}$ eller $V$, så alle partialtrykkene er uendret. Da er reaksjonskvotienten fortsatt lik $K_p$:

$$Q = \frac{(p_{\mathrm{PCl}_3}/p^\circ)(p_{\mathrm{Cl}_2}/p^\circ)}{p_{\mathrm{PCl}_5}/p^\circ} = K_p$$

**Likevekten forskyves ikke.** Ar er en tilskuer — den øker det *totale* trykket, men ikke partialtrykkene til reaktantene/produktene.

> **Til sammenligning:** Hadde Ar blitt tilsatt ved *konstant totaltrykk* (slik at volumet måtte øke), ville gassene blitt fortynnet. Da forskyves likevekten mot flere gassmolekyler — altså mot produktene PCl₃ + Cl₂, altså mer dissosiasjon. Skillet mellom «konstant $V$» og «konstant $p$» er en klassisk felle det er verdt å påpeke for studentene.


$$\alpha = 0{,}80 \qquad K_c = 0{,}0409 \qquad K_p = 1{,}78 \qquad K_x = 1{,}78$$

Likevekten forskyves **ikke** når trykket økes ved å tilsette Ar (konstant volum).


## Oppgave 7

**Modellen:** ligand $L$, reseptor $R$ og kompleks $C$ på et gitter med $V$ plasser, hver med $z$ orienteringer. Enpartikkel-partisjonsfunksjonene er

$$q_R = Vz, \qquad q_L = Vz, \qquad q_C = Vz\,e^{-\varepsilon_0/k_BT}$$

**a)** 

For ideelle (fortynnede) systemer der molekylene ikke konkurrerer om plass, gir statistisk mekanikk at likevektskonstanten uttrykt i partikkeltall er forholdet mellom enpartikkel-partisjonsfunksjonene:

$$K = \frac{N_C}{N_R N_L} = \frac{q_C}{q_R\,q_L}$$

Dette følger av at antallet molekyler av hver sort ved likevekt er proporsjonalt med partisjonsfunksjonen ($N_i \propto q_i$ i det fortynnede regimet). Setter vi inn:

$$K = \frac{q_C}{q_R\,q_L} = \frac{Vz\,e^{-\varepsilon_0/k_BT}}{(Vz)(Vz)} = \frac{Vz\,e^{-\varepsilon_0/k_BT}}{V^2 z^2}$$

$$\boxed{K = \frac{e^{-\varepsilon_0/k_BT}}{Vz}}$$

Legg merke til hva som overlever: kun **én** faktor $Vz$ i nevneren. To molekyler ($L$ og $R$) forsvinner og ett ($C$) dannes, så netto forsvinner én translasjonell frihetsgrad ($V$) og én rotasjonell ($z$) fra systemet. Det er nettopp disse to tapte frihetsgradene som gjør at $K$ avhenger av $V$ og $z$.

**b)**

$$K = \frac{e^{-\varepsilon_0/k_BT}}{Vz} \quad\Rightarrow\quad K \propto \frac{1}{V}$$

Så **$K$ synker når vi fortynner** (større $V$).

Fysisk mening: $K = N_C/(N_R N_L)$. Når $K$ synker med økende $V$, betyr det at forholdet mellom kompleks og frie molekyler faller — altså mindre kompleks ved likevekt. Dette gir god fysisk mening:

- For at $L$ og $R$ skal binde seg, må de først finne hverandre på gitteret. I et større volum er de i gjennomsnitt lenger fra hverandre, og sannsynligheten for at de havner på naboruter samtidig går ned.
- Entropisk uttrykt: den frie tilstanden (to separate molekyler) har mange flere måter å fordele seg på i et stort volum enn den bundne tilstanden (som «låser» dem sammen). Jo større $V$, jo mer favoriserer entropien de frie molekylene.

Dette er ganske enkelt Le Chateliers prinsipp: fortynning forskyver en likevekt der antall (oppløste) partikler minker, mot siden med flere frie partikler, altså mot dissosiasjon.

**c)**

$$K = \frac{e^{-\varepsilon_0/k_BT}}{Vz} \quad\Rightarrow\quad K \propto \frac{1}{z}$$

Så **$K$ synker når $z$ øker**.

Nøkkelen er at bindingen bare skjer for én av de $z$ orienteringene, men molekylet har fritt valg mellom alle $z$ når det er ubundet. Når $z$ vokser:

- Den frie liganden får flere tilgjengelige orienteringer (høyere orienteringsentropi i ubundet tilstand).
- Komplekset er derimot låst til den ene «riktige» orienteringen uansett hvor stor $z$ er.

>Å binde seg koster derfor mer entropi jo større $z$ er, molekylet må gi fra seg $\ln z$ orienteringsmuligheter for å låse seg inn i den ene bindende posisjonen. Flere «unyttige» orienteringer betyr en større entropisk straff for binding, og dermed lavere affinitet. Dette er kjernen i **orienteringskostnaden**

**d)**

**Energileddet driver bindingen.** Siden $\varepsilon_0 < 0$ er $\Delta h < 0$: å danne komplekset senker energien. Boltzmann-faktoren $e^{-\varepsilon_0/k_BT}$ blir da større enn 1 og trekker $K$ oppover — systemet «vil» ned i den gunstige lavenergi-tilstanden.

**Frihetsleddet motarbeider bindingen.** Faktoren $\tfrac{1}{Vz}$ er alltid mindre enn 1, så $\Delta s < 0$: binding koster frihet. Grunnen er at to frie molekyler har mange flere tilgjengelige tilstander enn ett bundet kompleks:
- **Posisjon ($\ln V$):** fri ligand og fri reseptor kan hver sitte hvor som helst på gitteret ($V \times V$ muligheter), mens komplekset er én enhet som bare kan plasseres på $V$ steder.
- **Orientering ($\ln z$):** en fri ligand kan peke i alle $z$ retninger, men i komplekset er den låst til den ene bindende retningen.

Dette er nøyaktig det dere så i **b)** (fortynning, $V$) og **c)** (orientering, $z$) — bare samlet under ett navn.

Frihetsleddet beskriver altså tapet av bevegelses- og orienteringsfrihet når to selvstendige molekyler smelter sammen til én bunden enhet. Binding er en dragkamp: energien vil ned, men friheten vil spres — og $K$ forteller hvem som vinner.


**e)**

*Konkrete kjemiske eksempler der «orienteringskostnad» er relevant:*

- **Enzym–substrat-binding:** et substrat må orienteres helt presist i det aktive setet for at katalysen skal skje. Denne orienteringskostnaden er nettopp grunnen til at enzymer utnytter *bindingsentalpi* (mange kontaktpunkter, hydrogenbindinger) for å «betale» for det store entropitapet. Fenomenet der bindingsenergi brukes til å tvinge substratet inn i riktig geometri kalles ofte **binding energy utilization** i enzymkinetikk.
- **Molekylær gjenkjenning generelt (lås-og-nøkkel / induced fit):** jo mer spesifikk gjenkjenningen er, desto høyere effektiv $z$ og desto større orienteringskostnad — og desto sterkere entalpisk «betaling» kreves for høy affinitet.
- **Kelateffekten:** en beslektet vri på samme prinsipp. En flertannet ligand er *allerede* delvis forhåndsorientert, så den betaler orienteringskostnaden bare én gang i stedet for flere ganger (som separate ligander ville gjort). Det gir en entropisk *gevinst* og forklarer hvorfor kelatkomplekser er så mye mer stabile.
- **Legemiddeldesign / rigidifisering:** man låser ofte en fleksibel medikamentmolekyl i sin bioaktive konformasjon (f.eks. ved å bygge inn ringer) nettopp for å redusere den effektive $z$ og slippe å betale orienteringskostnaden ved binding — dette hever affiniteten.

Fellesnevneren: **høyere geometrisk kravnivå → større entropisk bindingskostnad → lavere affinitet**, med mindre systemet kompenserer med sterkere entalpisk binding eller ved å forhåndsorientere molekylene.

**Bonus)**
$$\ln K = -\frac{\varepsilon_0}{k_BT} - \ln(Vz)$$

Deriverer med hensyn på $T$ (der $Vz$ er temperaturuavhengig):

$$\boxed{\frac{d\ln K}{dT} = \frac{\varepsilon_0}{k_BT^2}}$$

Stemmer med intuisjonen: binding er energetisk gunstig ($\varepsilon_0 < 0$) men entropisk ugunstig (den låser molekyler sammen). Ved høy temperatur veier entropien tyngre i $\Delta g = \Delta h - T\Delta s$, og termisk bevegelse river lettere komplekser fra hverandre. Dette er nettopp van't Hoffs likning for en «eksoterm» reaksjon: oppvarming forskyver mot dissosiasjon.
