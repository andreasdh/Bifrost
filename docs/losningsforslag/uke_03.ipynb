### Oppgave 4

**a)** Måleren står på grønt («i likevekt») fra start, og histogrammet passer med den oransje kurven umiddelbart. At gassen er i likevekt fra start betyr at den allerede har den fordelingen den ellers ville brukt tid på å finne — kollisjonene fortsetter, men de flytter bare molekyler *innenfor* samme fordeling, så formen står stille.

**b)** Helt i begynnelsen ser fordelingen helt annerledes ut enn den oransje kurven: nesten alle molekylene står i ro, så histogrammet er én høy søyle helt til venstre (fart ≈ 0), med ett molekyl langt ute til høyre. Konvergensmåleren står på rødt («langt unna») — dette er en ekstrem tilstand langt fra likevekt.

**c)** Etter hvert sprer fartene seg utover. Den ene raske partikkelen kolliderer med en i ro og gir den fart; de to treffer flere, som treffer flere igjen, og energien fordeles gradvis. Histogrammet flater ut fra den høye søylen og bygger seg opp til den klokkeformede kurven, helt til det legger seg oppå den oransje. Måleren beveger seg fra rød, via gul «nærmer seg», til grønn «i likevekt». Det tar en liten stund — energien må rekke å fordele seg gjennom mange kollisjoner.

**d)** Ja — denne ender på nøyaktig samme kurve som i c), selv om starten var helt annerledes. Det er hele poenget: **uansett hvor gassen starter, finner den samme fordeling.** Maxwell–Boltzmann er ikke bare én mulig fordeling, men den fordelingen et system av kolliderende molekyler alltid trekkes mot, fordi den er den desidert mest sannsynlige (den med flest tilgjengelige mikrotilstander). Startbetingelsene viskes ut, og bare likevektsfordelingen står igjen.

**e)** Når temperaturen øker, forskyver hele fordelingen seg mot **høyere farter** og blir samtidig **bredere og lavere**. Toppen flytter til høyre, og halen strekker seg lenger ut. $v_\text{rms}$ (og de to andre fartene) øker — molekylene beveger seg raskere i snitt. Høyere temperatur = mer bevegelsesenergi = raskere molekyler.

**f)** Ved lav temperatur er det motsatt: fordelingen er **smal og høy**, klemt sammen mot lave farter helt til venstre. Molekylene beveger seg tregt og med liten spredning i fart. Kurven er skarpere fordi det er mindre bevegelsesenergi å fordele.

**g)** Nei, $T$ endrer seg ikke mens histogrammet forandrer form (så lenge du ikke rører skyveren). Grunnen er at kollisjonene er **elastiske** — de flytter energi mellom molekyler, men skaper eller ødelegger aldri energi. Den totale bevegelsesenergien er bevart, og siden temperatur henger sammen med den, holder $T$ seg konstant. Energien bare *omfordeles* mellom molekylene mens fordelingen finner formen sin.

**h)** Når molekylene blir tyngre (høyere masse) ved samme temperatur, blir de **tregere** — fordelingen forskyver seg mot lavere farter, og alle tre fartene synker. Grunnen er at temperatur bestemmer den gjennomsnittlige *bevegelsesenergien* ($\sim \tfrac{1}{2}mv^2$), ikke farten direkte. Skal energien være den samme mens massen øker, må farten ned. Tunge molekyler beveger seg altså langsommere enn lette ved samme temperatur.

**i)** De lette hydrogenmolekylene (H₂) beveger seg klart raskest i snitt. Ved samme temperatur har begge gassene samme gjennomsnittlige bevegelsesenergi, men siden H₂ er mye lettere enn O₂, må H₂ ha høyere fart for å bære samme energi. Det er nøyaktig samme effekt som du så da du dro masse-skyveren opp i h) — bare med to ekte gasser i stedet.

**j)** De tre fartene er *ikke* like store — det er et fast mønster: den mest sannsynlige farten er alltid minst, gjennomsnittet ligger i midten, og rms-farten er størst.

$$v_\text{mp} < \bar{v} < v_\text{rms}$$

Mønsteret holder seg ved alle temperaturer — når du varmer opp gassen, øker alle tre fartene sammen, men rekkefølgen mellom dem er den samme. Grunnen er formen på fordelingen: den har en lang hale mot høye farter, som «drar» gjennomsnittet og særlig rms-farten (der de raske teller ekstra fordi fartene kvadreres) oppover, mens toppen (mest sannsynlig fart) ligger lengre til venstre. Så lenge fordelingen har samme karakteristiske klokkeform, er rangeringen den samme uansett temperatur.

### Oppgave 8

I oppgave 7 brukte dere Van der Waals likningen til å ta hensyn til intermolekylære krefter i en gass. Dette er en makroskopisk beskrivelse, men vi kan også se på tiltrekning og frastøtning mikroskopisk ved bruk av Lennard-Jones potensialet:

$$V(r) = 4\epsilon \left[ (\frac{\sigma}{r})^{12} - (\frac{\sigma}{r})^6 \right]$$

hvor $r$ er avstanden mellom partiklenes sentrum.

**a)** Tegn en kvalitativ skisse av potensialet, og forklar hva $\sigma$ og $\epsilon$ er. 


**Svar** Nedenfor vises et plott for Lennard-Potensialet til argon.  

```python
import numpy as np
import matplotlib.pyplot as plt

# Fysiske konstanter
k_B = 1.380649e-23  # J/K

# Lennard-Jones-parametere for argon
epsilon = 119.8 * k_B          # J
sigma   = 3.405e-10            # m  (3.405 Å)

def lennard_jones(r, epsilon, sigma):
    return 4*epsilon*((sigma/r)**12 - (sigma/r)**6)

# Avstander rundt sigma (i meter)
r = np.linspace(2.5e-10, 8e-10, 500)   # ca 2.5 Å til 8 Å

V_r = lennard_jones(r, epsilon, sigma)

plt.figure(figsize=(6,4))
plt.plot(r*1e10, V_r/epsilon, color='magenta', label='Lennard-Jones (Ar)')

plt.xlabel('distance $r$ (Å)')
plt.ylabel('potential $V(r)/\\epsilon$')
plt.title('Lennard-Jones potential for argon, $\\epsilon/k_B=119.8\\,\\text{K}$, $\\sigma=3.405\\,\\text{Å}$')

plt.grid(True)
plt.legend()

# Zoom inn så formen blir tydelig
plt.ylim(-2, 4)      # i enheter av epsilon
plt.xlim(2.5, 8.0)   # Å

plt.show()
```
