## Oppgave 1
**a)**	
- Makroskopisk: trykk og temperatur er størrelser vi måler direkte (manometer, termometer) – de beskriver makrotilstanden. 
- Mikroskopisk: trykk kommer fra molekylkollisjoner mot veggen, temperatur er gjennomsnittlig kinetisk energi til molekylene – dette er mikrotilstanden.
- Koblingen: mange ulike mikrotilstander gir samme makrotilstand; makrostørrelsene er gjennomsnitt over disse. 

**b)**	Energien som tilføres smeltende is går til å bryte bindingene i iskrystallen (potensiell energi), ikke til å øke bevegelseshastigheten. Derfor står temperaturen stille selv om energi tilføres hele tiden.

**c)**	En ekstensiv egenskap skalerer med systemets størrelse: dobler du mengden stoff (ved konstant tetthet og temperatur), dobles egenskapen. Volum V og antall molekyler N er typisk eksempler, to like beholdere satt samen gir dobbelt volum og dobbelt antall molekyler. En intensiv egenskap er uavhengig av systemets størrelse: den endres ikke om du dobler mengde stoff, så lenge tettheten og temperaturen er den samme. Sett to identiske beholdere med gass sammen (uten å endre trykk/temp), og trykket og temperaturen i det kombinerte systemet er fortsatt de samme som i hver enkelt beholder. 

**d)**	Fordi begge er definert som gjennomsnitt over svær mange molekyler. Ett molekyl gir ett støt, ikke et trykk; ett molekyl har en fart, ikke en «temperatur», temperatur er nettopp gjennomsnittet i en populasjon


## oppgave 2
**a)**	I en fortynnet gass er molekylene langt fra hverandre og vekselvirker sjelden, så vi kan se bort fra størrelse og krefter mellom dem. I væsker/faste stoffer er molekylene tett presset sammen med sterke, kontinuerlige vekselvirkninger/interaksjoner, som er mye vanskeligere å beskrive. 
**b)**	1) Molekylene har ikke eget volum (punktpartikler)
        2) Ingen krefter mellom dem utenom selve kollisjonene. 

**c)** Molekylenes eget volum trykke trykket opp: siden molekylene faktisk opptar plass, er det tilgjengelige «frie» volumet de kan bevege seg i mindre enn beholdervolumet V. Det betyr at de effektivt er mer sammenpresset enn idealgassloven antar, noe som gir flere kollisjoner per volumenhet og dermed høyere trykk enn PV=nRT skulle tilsi I guess. De intermolekylære tiltrekningskreftene trekker trykket ned: rett før et molekyl treffer veggen, blir den svakt trukket «innover» av nabomolekylene bak seg. Støtet mot veggen blir dermed litt svakere enn det ville vært uten denne tiltrekningen, og trykket blir lavere enn idealgassmodellen forutsier. 

**d)** Helium følger PV=nRT best ved romtemperatur og 1 atm: atomene er små (lite eget volum) og har svært svake tiltrekningskrefter, så begge avvikene er nesten neglisjerbare. CO2 har både større molekylvolum og sterkere tiltrekning, så avvikene fra ideelle gassloven er mer merkbare selv ved disse «normale» betingelsene. 

Komprimerer du begge til svært høyt trykk, blir molekylenes eget volum en stadig større brøkdel av det totale volumet (siden V blir lite), så volumeffekten (trykk opp) blir viktigere for begge, men mest for CO2, siden molekylene der er større.

Kjøler du begge kraftig ned, avtar den gjennomsnittlige kinetiske energien til molekylene, og tiltrekningskreftene (som er svakere sammenlignet med rask bevegelse) får relativt sett større innflytelse på banen til molekylene. Igjen rammer dette CO2 vil avvike mye tydeligere fra ideelle gassloven ved lav temperatur (og til slutt kondensere), mens helium forblir nær ideell langt lenger.

**e)** P (trykk) er, mikroskopisk sett, et mål på den kollektive kraften fra utallige molekylkollisjoner mot beholderveggen per arealenhet, som bestemt av hvor ofte molekylene treffer og hvor hardt hvert støt er. T (temperatur) er, mikroskopisk sett, et mål på den gjennomsnittlige kinetiske energien til molekylenes bevegelse, altså jo høyere temperatur, desto raskere beveger molekylene seg i gjennomsnitt. 

> Tilstandslikningen PV=nRT forteller oss hvordan disse størrelsene henger sammen for et gitt antall molekylet i et gitt volum, hvor den gir oss en pålitelig kobling mellom det vi kan måle i laben og den underliggende molekylære bevegelsen. Det den ikke forteller oss, er noe om systemets detaljerte mikrotilstand: den sier ingenting om posisjonen eller hastigheten til noe enkelt molekyl, om hvordan energien er fordelt mellom ulike molekyler, eller om systemets historie/dynamikk. Den beskriver bare et gjennomsnittlig, likevekts forhold mellom makroskopiske størrelser. 

## Oppgave 3
**a)**	Ja altså f(v) er en tetthet, ikke en sannsynlighet. Sannsynligheten for en eksakt fart er null (kontinuerlig variabel). Det er f(v)dv (tetthet x intervall) som gir en faktisk sannsynlighet. 

**b)**	Arealet under hele f(v) kurven må være nøyaktig 1. Dette kalles normalisering, og det følger av at et molekyl garantert har en eller annen fart – summerer (altså integrerer) man sannsynligheten over alle mulige farter fra 0 til uendelig, må man få 100% sikkerhet, altså 1. 

**c)**	Det er arealet under kurven mellom v1 og v2 som er relevant, ikke høyden i noe enkelt punkt. Høyden f(v) alene forteller bare om tettheten akkurat der; som er integralet av f(v) over intervallet [arealet av det skraverte området], som gir den faktiske andelen molekyler med fart i det intervallet, av samme grunn som a) og b).  

**d)**	Kurven ser likt ut fordi f(v) beskriver andelen molekyler ved en gitt fart, ikke antallet. Dobler man N og V sammen holdes tettheten N/V konstant, og siden temperaturen (og dermed fordelingen) er uendret, er kurven identisk. Dette viser at f(v), som temperatur, er intensiv.

## Oppgave 5

**a)** For nøyaktig 1 mol er massen $m = M$ og volumet $V = V_m$. Da blir $\rho = m/V = M/V_m$.

**b)** Med $n = m/M$:

$$PV = \frac{m}{M}RT \;\Rightarrow\; P = \frac{m}{V}\cdot\frac{RT}{M} = \frac{\rho RT}{M} \;\Rightarrow\; \boxed{\rho = \frac{PM}{RT}}.$$

Ved fast $P$ og $T$ er tettheten direkte proporsjonal med den molare massen: tunge gasser er tettere enn lette. Det er dette som gjør at man kan bestemme $M$ ved å måle tettheten (jf. Regneoppgave 2).

**c)**

$$\rho = \frac{PM}{RT} = \frac{101\,325 \cdot 0{,}02896}{8{,}314 \cdot 298{,}15} = \frac{2934}{2479} \approx 1{,}18\ \text{kg/m}^3.$$

Rimelig – lufttettheten ved rom­temperatur er om lag $1{,}2\ \text{kg/m}^3$.

## Oppgave 7

**a)**

$$n_{\text{N}_2} = \frac{100{,}00}{28{,}02} = 3{,}569\ \text{mol}, \qquad n_{\text{CO}_2} = \frac{100{,}00}{44{,}01} = 2{,}272\ \text{mol}, \qquad n_\text{tot} = 5{,}841\ \text{mol}.$$

$$P_{\text{CO}_2} = \frac{2{,}272 \cdot 8{,}314 \cdot 298{,}15}{2{,}50\times10^{-3}} = 2{,}25\times10^{6}\ \text{Pa} \approx 22{,}2\ \text{atm}.$$

**c)**

$$P_\text{tot} = P_{\text{N}_2} + P_{\text{CO}_2} = 34{,}9 + 22{,}2 = 57{,}1\ \text{atm} \;\;(\approx 5{,}79\times10^{6}\ \text{Pa}).$$

(Kontroll: $P_\text{tot} = n_\text{tot}RT/V$ gir samme svar.)

**d)** Ved ~57 atm er gassen langt fra fortynnet, og idealgassmodellen begynner å svikte. $\text{CO}_2$ avviker mest: molekylet er større og har sterkere intermolekylære tiltrekningskrefter (høyere kokepunkt, lettere å kondensere) enn $\text{N}_2$. Begge de reelle korreksjonene – eget volum og tiltrekning – er derfor større for $\text{CO}_2$.

## Oppgave 8

**a)** Konstant $V$ og $n$:

$$P_2 = P_1 \cdot \frac{T_2}{T_1} = 100\ \text{atm} \cdot \frac{500}{300} = 166{,}7\ \text{atm}.$$

**b)** Konstant $T$ og $n$ (Boyles lov):

$$P_3 = P_2 \cdot \frac{V_2}{V_3} = 166{,}7\ \text{atm} \cdot \frac{100}{10} = 1667\ \text{atm}.$$

**c)** Partialtrykket til en komponent er trykket den ville utøvd alene i hele beholderen ved samme temperatur. Fordi ideelle molekyler ikke merker hverandre, dytter hver gass på veggen uavhengig av de andre, og partialtrykkene summeres til totaltrykket (Daltons lov).

**d)** Antall mol $\text{N}_2$ fra starttilstanden:

$$n_{\text{N}_2} = \frac{P_1 V_1}{R T_1} = \frac{100\cdot101\,325 \cdot 100}{8{,}314 \cdot 300} \approx 4{,}06\times10^{5}\ \text{mol} \;(\approx 406\,000\ \text{mol}).$$

Partialtrykk av argon i $100\ \text{m}^3$ ved $500\ \text{K}$:

$$P_\text{Ar} = \frac{n_\text{Ar}RT}{V} = \frac{1000 \cdot 8{,}314 \cdot 500}{100} = 4{,}16\times10^{4}\ \text{Pa} \approx 0{,}41\ \text{atm}.$$

$$P_\text{tot} = 166{,}7 + 0{,}41 \approx 167{,}1\ \text{atm}.$$

Argon står for bare $0{,}41/167{,}1 \approx 0{,}25\ \%$ av trykket. Grunnen er at tanken allerede inneholder rundt $406\,000$ mol $\text{N}_2$ – de $1000$ mol argon er nesten forsvinnende i sammenligning. (Nettopp derfor er inertgass-oppbygging i resirkuleringssløyfa i ammoniakkanlegg noe man må jobbe aktivt for å lufte ut; det tar mye før effekten merkes.)

**e)** For en trykkøkning på $10\ \%$ må $P_\text{Ar} = 0{,}10 \cdot 166{,}7 = 16{,}7\ \text{atm}$:

$$n_\text{Ar} = \frac{P_\text{Ar} V}{RT} = \frac{16{,}7\cdot101\,325 \cdot 100}{8{,}314 \cdot 500} \approx 4{,}07\times10^{4}\ \text{mol} \;(\approx 40\,600\ \text{mol}).$$

Det er over 40 ganger mengden i **d)** – enda tydeligere hvor mange molekyler som allerede er til stede.
