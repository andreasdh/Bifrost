## Diskusjonsoppgaver

### Oppgave 1

**a)** 

En **tilstandsfunksjon** avhenger bare av systemets nåværende tilstand, ikke av hvordan systemet kom dit. Endringen mellom to tilstander er den samme uansett hvilken vei man tar. Eksempler: indre energi $U$, entalpi $H$, temperatur, trykk, volum, entropi.

En **veifunksjon** (stifunksjon) avhenger derimot av veien mellom start og slutt, ikke bare av endepunktene. Varme $q$ og arbeid $w$ er veifunksjoner, det så man direkte i ekspansjonsoppgavene: reversibel og irreversibel vei ga ulik $q$ og $w$, selv med samme start og slutt.

Det som skiller $H$ og $U$ fra $q$ og $w$ er nettopp dette:
* $U$ og $H$ er egenskaper *ved systemet*, de beskriver en tilstand systemet *er i*. 
* $q$ og $w$ er derimot *måter energi overføres på* under en prosess, de beskriver noe som *skjer*, ikke noe systemet *har*. 
Et system inneholder ikke varme eller arbeid; det inneholder energi, som kan endres ved å utveksle varme eller arbeid. Derfor skriver vi $\Delta U$ og $\Delta H$ (endring i noe systemet har), men bare $q$ og $w$ (overført mengde), aldri «$\Delta q$».

![tilstand](tilstand.svg)

> For en tilstandsfunksjon er endringen rundt en lukket sløyfe (tilbake til utgangspunktet) alltid null. For en veifunksjon trenger den ikke være det.



**b)**

Tetthet er en tilstandsfunksjon. Den avhenger bare av systemets nåværende tilstand, bestemt av temperatur og trykk men ikke av hvordan systemet kom dit.

**H₂O som eksempel:** Vann ved 25 °C og 1 atm har alltid en tetthet på ca. 0,997 g/cm³, uansett hvordan det kom til den tilstanden. Det spiller ingen rolle om vannet:
- ble varmet opp fra is som smeltet, eller
- ble kjølt ned fra damp som kondenserte.

Så lenge du ender på 25 °C og 1 atm, er tettheten den samme. Tettheten «husker» ikke veien, den avhenger bare av hvor systemet er nå. Det er nettopp kjennetegnet på en tilstandsfunksjon. (Tetthet er dessuten intensiv hvor den avhenger ikke av mengden vann, bare av tilstanden.)

**c)**

Vi varmer fra 0 °C til 100 °C langs to veier. La oss gå gjennom hver størrelse:

- $\Delta U$ og $\Delta H$ er *like* for begge prosessene. Begge er tilstandsfunksjoner, og for en ideell gass avhenger de bare av temperaturen. Siden start- (0 °C) og slutt-temperaturen (100 °C) er den samme uansett vei, blir $\Delta U$ og $\Delta H$ identiske. (At varmekapasiteten er temperaturavhengig endrer ikke dette, $\Delta U$ og $\Delta H$ er fortsatt bestemt kun av temperaturendringen, bare at man må integrere $C\,dT$ i stedet for å gange med en konstant.)

- $q$ og $w$ er *ulike* for de to veiene. Dette er veifunksjoner:
    - Ved **konstant volum** gjøres det *intet* volumarbeid ($\Delta V = 0 \Rightarrow w = 0$), så all varmen går til å øke indre energi: $q_V = \Delta U$.
    - Ved **konstant trykk** utvider gassen seg, så systemet gjør arbeid ($w \neq 0$), og det kreves *mer* varme: $q_p = \Delta H$.

Oppsummert: $\Delta U$ og $\Delta H$ er like; $q$ og $w$ er ulike. Nettopp fordi tilstandsfunksjoner ikke bryr seg om veien, mens veifunksjoner gjør det.

**d)**

Vi starter med definisjonen og deriverer:

$$H = U + pV \quad\Rightarrow\quad dH = dU + d(pV) = dU + p\,dV + V\,dp$$

Nå bruker vi de to forutsetningene:

1. **Første lov:** $dU = dq + dw$.
2. **Bare reversibelt volumarbeid:** $dw = -p\,dV$.

Setter vi inn (2) i (1): $dU = dq - p\,dV$. Og dette inn i uttrykket for $dH$:

$$dH = \underbrace{(dq - p\,dV)}_{dU} + p\,dV + V\,dp$$

De to $p\,dV$-leddene kansellerer:

$$\boxed{dH = dq + V\,dp}$$

**e)**

Ved **konstant trykk** er $dp = 0$, så leddet $V\,dp$ faller bort:

$$dH = dq_p$$

Dette betyr at entalpiendringen er nøyaktig lik varmen som overføres, så lenge trykket holdes konstant. Integrert: $\Delta H = q_p$.

Hvorfor er dette nyttig? Fordi $q$ egentlig er en veifunksjon. Den avhenger vanligvis av hvordan prosessen skjer. Men når vi låser trykket konstant, blir den utvekslede varmen lik en tilstandsfunksjon ($H$). Det betyr at varmen ved konstant trykk ikke lenger avhenger av veien, den er bestemt kun av start- og slutt-tilstanden. Entalpi er altså et smart konstruert begrep som «fanger» varmen ved konstant trykk og gjør den til en veluoppførende tilstandsfunksjon.

Tenk på det slik: ved konstant volum er varmen lik $\Delta U$ (fordi intet arbeid gjøres). Entalpi er den tilsvarende størrelsen for konstant *trykk* — den er skreddersydd slik at det innebygde $pV$-leddet automatisk tar hensyn til ekspansjonsarbeidet, så det som er igjen er ren varme.

**f)**

Fordi de aller fleste kjemiske reaksjoner skjer i åpne beholder (begerglass, reagensrør, reaktorer) som står åpne mot atmosfæren. Da er det trykket (atmosfæretrykket) som holdes konstant, mens volumet fritt kan endre seg når gasser dannes eller forbrukes.

Fra e) vet vi at ved konstant trykk er varmen lik entalpiendringen: $q_p = \Delta H$. Det betyr at:

- Når en kjemiker måler varmen som frigjøres eller tas opp i en reaksjon ved atmosfæretrykk, måler de *direkte* $\Delta H$ for reaksjonen.
- $\Delta H$ tar automatisk hensyn til ekspansjonsarbeidet som følger med hvis reaksjonen endrer gassvolumet trenger vi ikke regne på arbeidet separat.
- Siden $\Delta H$ er en tilstandsfunksjon, kan reaksjonsentalpier tabuleres og kombineres (Hess' lov) uten å bekymre seg om reaksjonsveien.

Hadde reaksjonene i stedet skjedd ved konstant volum (i lukkede, stive beholdere), ville $\Delta U$ vært den naturlige størrelsen. Men det er upraktisk når de fleste laboratorieforsøk foregår ved konstant trykk. Derfor er entalpi rett og slett den energistørrelsen som passer virkeligheten kjemikere jobber i, og det er grunnen til at reaksjonsvarmer nesten alltid oppgis som $\Delta H$.

### Oppgave 3

**a**

Makroskopisk defineres entropiendringen gjennom den reversibelt utvekslede varmen delt på temperaturen:

$$dS = \frac{dq_\text{rev}}{T} \qquad\text{eller integrert}\qquad \Delta S = \int \frac{dq_\text{rev}}{T}$$

Verdt å nevne:
- Det er **$q_\text{rev}$** som teller, varmen som ville blitt utvekslet dersom prosessen ble kjørt reversibelt mellom de samme to tilstandene. Selv om den virkelige prosessen er irreversibel, regner man $\Delta S$ langs en tenkt reversibel vei, siden $S$ er en tilstandsfunksjon og bare avhenger av start og slutt.
- Vi deler på **$T$**: den samme mengden varme gir større entropiendring ved lav temperatur enn ved høy. (Det var nettopp dette som gjorde at $\Delta S_\text{tot} > 0$ i klosse- og gassoppgavene.)

Entropi er en tilstandsfunksjon med enhet J K⁻¹.

**b)**
For en reversibel prosess må

$$\boxed{\Delta S_\text{tot} = 0}$$

**Hvorfor?** Den totale entropien (system + omgivelser) er den størrelsen som skiller de tre mulige utfallene:

- $\Delta S_\text{tot} > 0$ → prosessen er spontan/irreversibel
- $\Delta S_\text{tot} = 0$ → prosessen er reversibel (grensetilfellet)
- $\Delta S_\text{tot} < 0$ → prosessen er umulig (den motsatte veien er spontan)

En reversibel prosess er idealgrensen der prosessen kan gå «baklengs» uten å etterlate noe spor i universet. Skulle den skape netto entropi ($\Delta S_\text{tot} > 0$), kunne den ikke reverseres uten å måtte fjerne den entropien igjen, og det er umulig etter andre lov. Reversibilitet krever derfor at systemet og omgivelsene til enhver tid er i (nesten) perfekt balanse, slik at ingen netto entropi produseres. Det skjer bare når $\Delta S_\text{tot} = 0$.

**c)**

$\Delta S_\text{tot} = 0$ når prosessen er **reversibel** (fra b).

I dette tilfellet er systemets og omgivelsenes entropiendringer like store, men med motsatt fortegn:

$$\Delta S_\text{system} = -\Delta S_\text{omgivelser} \quad\Rightarrow\quad \Delta S_\text{tot} = \Delta S_\text{system} + \Delta S_\text{omgivelser} = 0$$

De trenger altså ikke være null hver for seg — de kan begge være store — men de balanserer hverandre nøyaktig. Et konkret eksempel dere allerede har regnet på: den **reversible isoterme ekspansjonen**, der $\Delta S_\text{system} = +9{,}13\ \text{J K}^{-1}$ og $\Delta S_\text{omgivelser} = -9{,}13\ \text{J K}^{-1}$, som summerte seg til nøyaktig null. Systemet vinner akkurat den entropien omgivelsene taper, uten at noe går tapt netto. Det er nettopp derfor ingen entropi «produseres» i en reversibel prosess.

**d)**

- **Makroskopisk** ($dS = dq_\text{rev}/T$): entropi målt utenfra, via varme og temperatur i laben. Dette er den termodynamiske, «bokførings»-orienterte definisjonen hvor den forteller *hvor mye* entropi som endres, men ikke *hva* entropi egentlig er.
- **Mikroskopisk** ($S = k_B\ln W$): entropi som et mål på antall tilgjengelige mikrotilstander $W$ bak en gitt makrotilstand hvor det statistiske bildet fra forrige uke. Dette forteller *hva* entropi er: et mål på hvor mange måter systemet kan realiseres på molekylnivå.

**Hvorfor beskriver de samme størrelse?** Fordi det å tilføre varme og det å øke antall mikrotilstander er to sider av samme sak:

Når du tilfører varme til et system, sprer du energien utover molekylene. Mer energi å fordele betyr flere måter å fordele den på — altså flere tilgjengelige mikrotilstander, og dermed større $W$. Å varme opp et system *er* å øke $W$.

Tenk spesifikt på gassekspansjon:

- **Mikroskopisk:** når volumet dobles, får hvert molekyl dobbelt så mange posisjoner å være i. For $N$ molekyler blir $W$ multiplisert med $2^N$, og $S = k_B\ln W$ øker med $k_B\ln(2^N) = Nk_B\ln 2 = nR\ln 2$.
- **Makroskopisk:** $\Delta S = \int dq_\text{rev}/T = nR\ln 2$ for den samme voldumdoblingen.

**Nøyaktig samme tall.** De to definisjonene møtes. Faktisk er det denne overensstemmelsen som fastsetter verdien av Boltzmanns konstant $k_B$, den er akkurat brolegemet som binder molekylenes verden ($\ln W$) til den målbare termodynamikken ($q/T$).

> Den makroskopiske definisjonen forteller deg hvordan du måler entropi; den mikroskopiske forteller deg hva den er. Varme delt på temperatur og «antall mikrotilstander» er samme fjell sett fra to sider: å spre energi som varme er nøyaktig det samme som å åpne opp flere mikrotilstander for systemet.

### Oppgave 5

**a)**
Vi har $dU = T\,dS - p\,dV$ for et lukket system.

Dette kalles den **fundamentale termodynamiske relasjonen** (eller den fundamentale likningen for den indre energien). Den kombinerer første og andre lov i én likning: første lov gir $dU = dq + dw$, andre lov (reversibelt) gir $dq_\text{rev} = T\,dS$, og volumarbeidet er $dw = -p\,dV$.

**Hvorfor er det nyttig å ha $U$ som funksjon av $S$ og $V$?** Fordi da er $S$ og $V$ de naturlige variablene til $U$. Ser du på likningen, står differensialene $dS$ og $dV$ på høyre side — det betyr at *hvis* du kjenner $U(S, V)$, kan du hente ut alt annet ved simpel derivasjon:

$$T = \left(\frac{\partial U}{\partial S}\right)_V, \qquad p = -\left(\frac{\partial U}{\partial V}\right)_S$$

En slik funksjon kalles en karakteristisk funksjon*: kjenner du $U$ uttrykt i sine naturlige variabler, har du *hele* den termodynamiske informasjonen om systemet — temperatur, trykk, og alt avledet følger av derivasjon alene. Uttrykker du derimot $U$ i «feil» variabler (f.eks. $U(T, V)$), mister du denne egenskapen og trenger tilleggsinformasjon for å rekonstruere systemet.

**b)**
Hver av de fire energifunksjonene har sitt eget naturlige variabelpar:

| Funksjon | Naturlige variabler | Fundamentallikning |
|---|---|---|
| $U$ | $S, V$ | $dU = T\,dS - p\,dV$ |
| $H$ | $S, p$ | $dH = T\,dS + V\,dp$ |
| $F$ (Helmholtz) | $T, V$ | $dF = -S\,dT - p\,dV$ |
| $G$ (Gibbs) | $T, p$ | $dG = -S\,dT + V\,dp$ |

**Fordelen** er at man velger den funksjonen hvis naturlige variabler er de man faktisk *kontrollerer* i eksperimentet:

- Et forsøk ved konstant $T$ og $p$ (typisk kjemi, åpen beholder) → bruk $G$, siden $T$ og $p$ er dens naturlige variabler.
- Et forsøk ved konstant $T$ og $V$ (lukket, stiv beholder) → bruk $F$.
- Konstant $S$ og $p$ → bruk $H$.

Da blir matematikken enklest mulig: den funksjonen «passer» direkte til de størrelsene du holder fast, og du slipper tungvinte omregninger. Dessuten forteller hver funksjon om spontanitet under nettopp *sine* betingelser (f.eks. $G$ minker for en spontan prosess ved konstant $T, p$), som er grunnen til at $G$ er kjemikerens favoritt.

**c)**
En Maxwell-relasjon oppstår fordi $U, H, F, G$ er tilstandsfunksjoner, og differensialene deres er **eksakte differensialer**.

Nøkkelen er en matematisk egenskap ved eksakte differensialer: rekkefølgen på blandede andrederiverte spiller ingen rolle. For en tilstandsfunksjon $U(S, V)$ gjelder

$$\frac{\partial^2 U}{\partial V\,\partial S} = \frac{\partial^2 U}{\partial S\,\partial V}$$

Fra $dU = T\,dS - p\,dV$ vet vi at $T = (\partial U/\partial S)_V$ og $-p = (\partial U/\partial V)_S$. Deriverer vi den første med $V$ og den andre med $S$, må de være like:

$$\left(\frac{\partial T}{\partial V}\right)_S = -\left(\frac{\partial p}{\partial S}\right)_V$$

Det er en Maxwell-relasjon. **Hvorfor krever dette en tilstandsfunksjon?** Fordi bare da er differensialet eksakt, og bare for et eksakt differensial gjelder likheten mellom de blandede andrederiverte. En veifunksjon (som $q$ eller $w$) har *ikke* et eksakt differensial — integralet avhenger av veien, endringen rundt en lukket sløyfe er ikke null, og «funksjonen» eksisterer ikke som en veldefinert størrelse man kan derivere to ganger. Da bryter hele konstruksjonen sammen. Det er altså nettopp tilstandsfunksjon-egenskapen som *garanterer* at slike relasjoner finnes.

**d)**
Fordi de lar deg **bytte ut en størrelse du nesten ikke kan måle med en du enkelt kan måle**.

Ta eksemplet i oppgaven: $(\partial S/\partial V)_T$ — hvordan entropien endrer seg med volumet ved konstant temperatur. Det finnes ikke noe «entropimeter» i laben; entropi er ikke en størrelse du leser av direkte, så dette er nesten umulig å måle.

Men en Maxwell-relasjon sier at

$$\left(\frac{\partial S}{\partial V}\right)_T = \left(\frac{\partial p}{\partial T}\right)_V$$

og høyresiden er *lett målbar*: du setter gassen i en beholder med fast volum, varmer den litt, og måler hvordan trykket stiger med temperaturen. Bare et manometer og et termometer.

Verdien ligger altså i at Maxwell-relasjonene **oversetter det umålbare til det målbare**. De lar oss regne ut entropiendringer, indre energier og andre abstrakte størrelser fra enkle mekaniske målinger av $p$, $V$ og $T$. Det er dette som gjør termodynamikken til et praktisk verktøy og ikke bare en samling definisjoner.

### Oppgave 6

**a)**
Det kjemiske potensialet $\mu$ forteller hvor mye systemets frie energi endrer seg når du tilsetter én ekstra partikkel (ett mol stoff), mens alt annet holdes fast. Matematisk defineres det oftest som den partielle deriverte av Gibbs fri energi med hensyn på stoffmengde:

$$\boxed{\mu = \left(\frac{\partial G}{\partial n}\right)_{T, p}}$$

Man kan definere det ut fra hvilken som helst av energifunksjonene, så lenge man holder de riktige variablene faste:

$$\mu = \left(\frac{\partial U}{\partial n}\right)_{S,V} = \left(\frac{\partial H}{\partial n}\right)_{S,p} = \left(\frac{\partial F}{\partial n}\right)_{T,V} = \left(\frac{\partial G}{\partial n}\right)_{T,p}$$

Kort sagt: $\mu$ er den frie energien «per partikkel» — energikostnaden ved å legge til litt mer stoff.

**b)**
De henger tett sammen, men er ikke det samme:

- **$G$ er en ekstensiv størrelse** — den avhenger av hvor mye stoff du har. Dobler du mengden, dobler du $G$.
- **$\mu$ er en intensiv størrelse** — den er $G$ *per mol*, og avhenger ikke av mengden. Dobler du mengden, er $\mu$ uendret.

For et rent stoff er sammenhengen rett og slett

$$\mu = \frac{G}{n}, \qquad\text{altså}\qquad G = n\mu$$

For en **blanding** er $\mu_i$ for hver komponent den partielle molare Gibbs-energien, $\mu_i = (\partial G/\partial n_i)_{T,p,n_{j\neq i}}$, og den totale Gibbs-energien er $G = \sum_i n_i\mu_i$. Poenget: $G$ er «total pengesum», $\mu$ er «pris per enhet».

**c)**
Sammenhengen går via entropien, som fra forrige uke er $S = k_B\ln W$ (der $W$ = antall tilgjengelige mikrotilstander).

Når du tilsetter en partikkel til et system, endrer du antallet mikrotilstander $W$. Et lavt kjemisk potensial svarer til et system der det å legge til en partikkel *øker* $W$ mye — det åpner opp mange nye måter å arrangere systemet på. Et høyt kjemisk potensial svarer til et system der en ekstra partikkel gir liten eller ingen økning i $W$ (systemet er allerede «trangt»).

Grovt sagt: $\mu$ måler hvor «villig» systemet er til å ta imot en partikkel, og den villigheten bestemmes av hvor mye den ekstra partikkelen bidrar til systemets samlede antall mikrotilstander. Lavt $\mu$ = mye å vinne (mange nye mikrotilstander) = systemet tar gjerne imot. Dette leder rett inn i d).

**d)**
Fordi et lavt kjemisk potensial betyr at systemet har **god plass til flere mikrotilstander** — når en partikkel kommer inn, øker antallet tilgjengelige mikrotilstander $W$ betydelig.

Og fra hele grunnlaget i statistisk mekanikk vet vi at systemer spontant beveger seg mot tilstander med *flere* mikrotilstander, rett og slett fordi de er overveldende mer sannsynlige (jf. terning-argumentet og andre lov). Så et system med lavt $\mu$ «vil gjerne» ta imot partikler, ikke fordi noe tvinger det, men fordi det å ta imot partikkelen fører systemet mot en tilstand med flere mikrotilstander — den mest sannsynlige veien.

Motsatt: et system med høyt $\mu$ er allerede «mettet» — en ekstra partikkel gir få nye mikrotilstander, så det er lite å vinne, og systemet er tilbøyelig til å *gi fra seg* partikler i stedet.

**e)**
Stoff beveger seg spontant **fra høyt kjemisk potensial til lavt kjemisk potensial**.

$$\boxed{\text{høy } \mu \;\longrightarrow\; \text{lav } \mu}$$

Dette er helt analogt med de andre drivkreftene:
- Varme strømmer fra høy til lav **temperatur**.
- Volum endres fra høyt til lavt **trykk** (gassen utvider seg mot lavere trykk).
- Partikler strømmer fra høyt til lavt **kjemisk potensial**.

I hvert tilfelle utjevnes forskjellen helt til den forsvinner, og da er systemet i likevekt. For kjemisk potensial betyr det at partikler flytter seg helt til $\mu$ er **likt overalt** — det er betingelsen for kjemisk likevekt og for likevekt ved partikkeltransport (f.eks. diffusjon eller faseovergang). Underliggende er det samme prinsippet som alltid: systemet søker den fordelingen som gir flest tilgjengelige mikrotilstander totalt.

Samme start og slutt, men helt ulikt arbeid — fordi arbeid er en veifunksjon.


### Oppgave 9

**Gitt:** $n = 2{,}0\ \text{mol}$, $p = 1{,}25\ \text{atm}$ (konstant), $T: 250 \to 277\ \text{K}$ (så $\Delta T = +27\ \text{K}$), $C_{p,m} = 37{,}11\ \text{J K}^{-1}\text{mol}^{-1}$.

**Varme = entalpi** (konstant trykk):
$$q_p = \Delta H = nC_{p,m}\Delta T = 2{,}0 \times 37{,}11 \times 27 = \boxed{2004\ \text{J} \approx 2{,}00\ \text{kJ}}$$

**Indre energi** (bruk $C_{V,m} = C_{p,m} - R = 37{,}11 - 8{,}314 = 28{,}80$):
$$\Delta U = nC_{V,m}\Delta T = 2{,}0 \times 28{,}80 \times 27 = \boxed{1555\ \text{J} \approx 1{,}56\ \text{kJ}}$$

$\Delta U$ er mindre enn $\Delta H$ fordi en del av varmen gikk til ekspansjonsarbeid da gassen utvidet seg.


### Oppagve 10

**Gitt:** $C_p = 20{,}17 + 0{,}4001\,T$ (per mol, antar 1 mol), $T: 0\ ^{\circ}\text{C} = 273{,}15\ \text{K} \to 100\ ^{\circ}\text{C} = 373{,}15\ \text{K}$.

Siden $C_p$ varierer med $T$, må vi integrere. La meg først regne ut $\Delta H$ og $\Delta U$, som er *like for begge veiene* (tilstandsfunksjoner, samme temperaturendring).

**$\Delta H = \int_{T_1}^{T_2} C_p\,dT$:**
$$\Delta H = \left[20{,}17\,T + \tfrac{0{,}4001}{2}T^2\right]_{273{,}15}^{373{,}15}$$
$$= 20{,}17(100) + 0{,}20005\,(373{,}15^2 - 273{,}15^2) = 2017 + 0{,}20005 \times 64630 = \boxed{14\,945\ \text{J} \approx 14{,}9\ \text{kJ}}$$

**$\Delta U = \int C_V\,dT$** med $C_V = C_p - R = 11{,}856 + 0{,}4001\,T$:
$$\Delta U = 11{,}856(100) + 0{,}20005\times64630 = 1186 + 12\,929 = \boxed{14\,115\ \text{J} \approx 14{,}1\ \text{kJ}}$$

Nå de veiavhengige størrelsene:

**Tilfelle 1 — konstant trykk:**
$$q_p = \Delta H = \boxed{14{,}9\ \text{kJ}}$$
$$w = \Delta U - q = 14\,115 - 14\,945 = \boxed{-830\ \text{J}}$$

**Tilfelle 2 — konstant volum:**
$$w = 0 \quad (\text{intet volumarbeid})$$
$$q_V = \Delta U = \boxed{14{,}1\ \text{kJ}}$$

$\Delta U$ og $\Delta H$ er like i begge tilfeller; det er $q$ og $w$ som skiller seg (veifunksjoner).


### Oppgave 11

**Reaksjonslikning:**
$$\text{Zn}(s) + 2\,\text{HCl}(aq) \rightarrow \text{ZnCl}_2(aq) + \text{H}_2(g)$$

**Gitt:** $m = 5{,}0\ \text{g}$ Zn ($M = 65{,}38\ \text{g/mol}$), $p = 1{,}1\ \text{atm}$, $T = 25\ ^{\circ}\text{C} = 298{,}15\ \text{K}$.

Stoffmengde Zn: $n_\text{Zn} = 5{,}0/65{,}38 = 0{,}0765\ \text{mol}$. Reaksjonen lager like mange mol H₂-gass: $n_{\text{H}_2} = 0{,}0765\ \text{mol}$.

Arbeidet kommer av at den nye gassen skyver atmosfæren unna ved konstant trykk. Enkleste vei er $pV = nRT$ for gassen:
$$w_\text{av system} = p\,\Delta V = n_{\text{H}_2}RT = 0{,}0765 \times 8{,}314 \times 298{,}15$$
$$\boxed{w_\text{av system} = 190\ \text{J}}$$

Systemet gjør altså ca. 190 J arbeid på omgivelsene når hydrogengassen dannes og presser mot atmosfæren. (Arbeidet *på* systemet er tilsvarende $-190$ J.)
