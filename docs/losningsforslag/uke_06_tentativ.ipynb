### Oppgave 1: Illustrasjon av Helmholtz fri energi, dimerer eller monomerer

Dette eksempelet viser hvordan vi kan bruke *minimering av fri energi* til å forutsi likevekt. For å gjøre dette skal vi se på en enkel mikroskopisk modell som kombinerer energi- og entropibidrag: et system med to partikler som enten er *assosiert* (en dimer) eller *dissosiert* (to frie monomerer). Vi kan teke på dette som at do atomer kan danne en binding eller være dissosiert som to enkeltatomer. 

**Gitt:** Vi har $N=2$ identiske partikler i en testrør modellert som et endimensjonalt gitter med $V$ plasser i rekkefølge, systemet er i et termisk likevekt med et varmebad med temperatur $T$:

- **Dimer-tilstand:** Når de to partiklene sitter på naboplasser, kan de danne enbinding med energi $\varepsilon>0$. I dette forenklede systemet med to partikler er dette eneste bidraget til den indre energien $U=-\varepsilon$

- **Monomer-tilstand:** Når de to partiklene ikke er ved siden av hverandre kan de ikke danne noe binding og dermed er den indre energien null, $U=0$. De plasseres på to

Vi ønsker å finne under hvilke betingelser systemet danner en dimer, og når det forblir som 2 separate partikler.

**a)** Diskuter følgende:

1) For dimeren, hvorfor er det det et minustegn i $U=-\varepsilon$?

2) Hvilke termodynamiske størrelser holder vi konstant her, og hvilket utrykk for den frie energien må vi bruke?

3) Hva forventer vi skjer med den frie energien ved likevekt?



**Svar** I et termisk bad er $T$ konstant; et stivt testrør gir konstant volum $V$; og $N=2$ er gitt og endres ikke. Med $(T,V,N)$ konstant er det **Helmholtz fri energi**

$$F=U-TS$$

som er på minimum ved likevekt.

**Hvorfor?** Kombiner de to lovene for systemet *og* omgivelsene. Omgivelsene (badet ved $T$) endrer entropi med $\Delta S_{\mathrm{omg}}=-q/T$ når de avgir varmen $q$ til systemet. Ved konstant volum gjøres intet volumarbeid, så første lov gir $\Delta U=q$. Total entropi må øke eller være konstant (andre lov):

$$\Delta S_{\mathrm{tot}}=\Delta S+\Delta S_{\mathrm{omg}}=\Delta S-\frac{\Delta U}{T}=-\frac{\Delta F}{T}\;\geq\;0$$

Altså minker $F$ for enhver spontan prosess, og er **minimum** ved likevekt ($\Delta S_{\mathrm{tot}}=0$). Fysisk betydning: $F$ balanserer interaksjonsenergien $U$ mot entropibidraget $-TS$, og temperaturen $T$ avgjør vektingen: ved høy $T$ dominerer entropileddet, ved lav $T$ dominerer energileddet.

**b)** Regn ut entropien og Helmholtz fri energi for **dimer-tilstanden**.

**Svar** Dimeren kan ligge med sin første partikkel på plass $1,2,\dots,V-1$ (fra plass $V$ ville den andre partikkelen falle utenfor gitteret). Merk at vi i det følgende bruker approksimasjonen $V-1\approx V$ (for et stort gitter). Med $S=k\ln W$:

$$S_{\mathrm{dimer}}=k\ln(V-1)\approx k\ln V$$

$$F_{\mathrm{dimer}}=U_{\mathrm{dimer}}-TS_{\mathrm{dimer}}=-\varepsilon-kT\ln V$$

Dette er bokens likning (8.7). Fra $dF/dT=-S$ ser vi at dimerkurven er «flat» (liten entropi, lite utslag).

**c)** Regn ut entropien og Helmholtz fri energi for **monomer-tilstanden**. Forklar hvorfor vi må trekke fra de $V-1$ naboplasseringene i tellingen.

**Svar** De to frie partiklene okkuperer to *forskjellige* plasser som ikke er naboer (alle naboplasseringer er dimer-tilstander). Antall måter å plassere dem på er binomialkoeffisienten minus nabotilfellene:

$$W_{\mathrm{monomer}}=\binom{V}{2}-(V-1)=\frac{V(V-1)}{2}-(V-1)=\frac{(V-2)(V-1)}{2}\approx\frac{V^2}{2}$$

Med $U_{\mathrm{monomer}}=0$:

$$F_{\mathrm{monomer}}=-TS_{\mathrm{monomer}}=-kT\ln\left(\frac{V^2}{2}\right)$$

Stigningstallet er her $dF/dT=-k\ln(V^2/2)$, som er **brattere** enn dimerens (siden $V^2/2\gg V$ for $V>2$): monomer-tilstanden har altså større entropi.

**d)** Finn krysstemperaturen $T_0$ der de to tilstandene er like stabile (lik fri energi). Hvorfor er dimeren den stableste tilstanden ved $T<T_0$, og monomeren ved $T>T_0$? Hvordan avhenger $T_0$ av $\varepsilon$ og $V$?

**Svar** Sett $F_{\mathrm{dimer}}=F_{\mathrm{monomer}}$ og løs for $T_0$ (bruk $\ln(V^2/2)=2\ln V-\ln 2$):

$$-\varepsilon-kT_0\ln V=-kT_0\ln\left(\frac{V^2}{2}\right)
\quad\Longrightarrow\quad
\varepsilon=kT_0\ln\left(\frac{V}{2}\right)
\quad\Longrightarrow\quad
\boxed{\,T_0=\frac{\varepsilon}{k\ln(V/2)}\,}$$

(bokens likning 8.8). Fysisk tolkning:

- Jo **sterkere tiltrekning** (større $\varepsilon$), jo **høyere** $T_0$ – det kreves mer termisk energi for å rive dimeren fra hverandre.
- Jo **større gitter** $V$, jo **lavere** $T_0$ – flere ledige plasser gjør den dissosierte tilstanden entropisk mer fordelaktig.
- For $T<T_0$ er $F_{\mathrm{dimer}}<F_{\mathrm{monomer}}$: **dimeren er stabilest** (energien dominerer). For $T>T_0$ er $F_{\mathrm{monomer}}<F_{\mathrm{dimer}}$: **monomeren dominerer** (entropien dominerer). Se figuren under.




```python
import numpy as np
import matplotlib.pyplot as plt

# Modellverdier (Eksempel 8.1), molare enheter for en enkel illustrasjon
V = 10.0      # antall gitterplasser (betingelsen V > 4 fra boken)
eps = 2.5     # bindingsenergi [kJ/mol]
R = 8.314e-3  # gasskonstant [kJ mol^-1 K^-1] (det molare motstykket til k)

T = np.linspace(0, 350, 400)
F_dimer = -eps - R * T * np.log(V)
F_monomer = -R * T * np.log(V**2 / 2)
T0 = eps / (R * np.log(V / 2))
print(f"Krysstemperatur T0 = {T0:.1f} K")

plt.figure(figsize=(8, 5))
plt.plot(T, F_dimer, color="C3", lw=2, label=r"Dimer: $F=-\varepsilon-kT\ln V$")
plt.plot(T, F_monomer, color="C2", lw=2, label=r"Monomerer: $F=-kT\ln(V^2/2)$")

plt.axvline(T0, color="gray", ls=":", lw=1.5)
F0 = -eps - R * T0 * np.log(V)
plt.plot([T0], [F0], "ko", ms=6, zorder=5)
plt.annotate(r"$T_0=\varepsilon/[R\ln(V/2)]$", (T0, F0),
             xytext=(T0 + 35, F0 + 1.2),
             arrowprops=dict(arrowstyle="->", color="k", lw=1))

# Hvilken tilstand som er stabilest (lavest F)
plt.fill_between(T, F_dimer, F_monomer, where=(F_dimer <= F_monomer),
                 color="C3", alpha=0.25, label="Dimeren er stabilest (lav T)")
plt.fill_between(T, F_dimer, F_monomer, where=(F_dimer > F_monomer),
                 color="C2", alpha=0.25, label="Monomerene er stabilest (høy T)")

plt.xlabel(r"Temperatur $T$ [K]")
plt.ylabel(r"Helmholtz fri energi $F$ [kJ mol$^{-1}$]")
plt.title("Frienergibalansen i dimeriseringsmodellen (Eksempel 8.1)")
plt.xlim(0, 350)
plt.ylim(-12.5, 1.0)
plt.grid(alpha=0.3)
plt.legend(loc="lower left")
plt.tight_layout()
plt.show()
```

**e)** Figuren over viser $F(T)$ for $V=10$ og $\varepsilon=2.5\ \mathrm{kJ\,mol^{-1}}$. Forklar med utgangspunkt i figuren hvorfor dimeren dominerer ved lav temperatur og monomerene ved høy temperatur, og hva stigningstallene representerer fysisk. (Boken påpeker at modellen bare er fysisk meningsfull for $V>4$.)

**Svar** Begge kurver *synker* med $T$, fordi $dF/dT=-S<0$. Dermed kan man lese entropiene rett av stigningstallene:

- **Dimerlinjen** er «flat» ($|dF/dT|=k\ln V$, lav entropi, få konfigurasjoner) og starter lavest, på $F=-\varepsilon$ ved $T=0$: bindingsenergien «kjøper» stabilitet.
- **Monomerlinjen** starter i $F=0$ og faller brattere ($|dF/dT|=k\ln(V^2/2)$): mange flere mikrotilstander ($W_{\mathrm{monomer}}\gg W_{\mathrm{dimer}}$) gjør at $F_{\mathrm{monomer}}$ faller fortere med $T$.

Konsekvens: til **venstre** for $T_0$ ligger dimerlinjen lavest, altså er dimeren den stabile tilstanden (skravert rødt). Til **høyre** for $T_0$ ligger monomerlinjen lavest, og systemet dissosierer (skravert grønt). I figuren er molare enheter brukt ($\varepsilon$ per mol og $R$ i stedet for $k$), slik at

$$T_0=\frac{\varepsilon}{R\ln(V/2)}=\frac{2500\ \mathrm{J\,mol^{-1}}}{8.314\ \mathrm{J\,K^{-1}\,mol^{-1}}\cdot\ln 5}\approx 187\ \mathrm{K}.$$

Merknaden fra figur 8.5 i boken: for at modellens fysiske tolkning («dimer mot monomerer») skal være meningsfull, kreves $V>4$; ellers blir tellingen av ikke-naboplasseringene uklar i forhold til de to tilstandene.

**f)** Hva ville vi konkludert hvis vi *kun* hadde maksimert entropien (valgt størst $W$)?

**Svar** $S_{\mathrm{dimer}}=k\ln V$ mens $S_{\mathrm{monomer}}=k\ln(V^2/2)$, og siden $V^2/2>V$ for $V\gtrsim 2$, ville ren entropimaksimering **alltid** valgt den dissosierte tilstanden – vi ville konkludert at dimerer aldri dannes, uansett temperatur. Det er feil, fordi vi da overser interaksjonsenergien $U=-\varepsilon$. Nøkkelen til $F=U-TS$ er at *begge* bidrag teller, og at $T$ bestemmer vektingen: kalde systemer blir styrt av lav energi, varme systemer av mange mikrotilstander.

**g)** Eksempel 8.1 trekker frem at modellen, til tross for enkelheten, fanger essensen i prosesser som koking, smelting og oppfolding. Diskuter kort hvorfor.

**Svar** Assosiasjon–dissosiasjon er en avveining mellom den *energetiske* tendensen for partikler å holde sammen (dominerer ved lav $T$) og den *entropiske* tendensen til å spre seg (dominerer ved høy $T$). Det er samme kompromiss som styrer:

- **koking** (molekyler i væsken frigjøres til gassfasen når $T$ overstiger en kritisk verdi),
- **smelting** (den ordnede krystallstrukturen brytes),
- **oppfolding/utfolging av polymerer og proteiner** (intramolekylær tiltrekning mot konformasjonsfrihet).

Alle disse viser en «ordnet» lav-temperatur-tilstand overfor en «spredt» høy-temperatur-tilstand med en krysstemperatur $T_0$ der prosessen skjer. Boken nevner at vi i kapittel 10 skal beregne de relative mengdene av monomerer og dimerer som funksjon av temperaturen.
