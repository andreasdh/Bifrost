**a)** Forrige uke jobbet vi med Gibbs frie energi og hvordan en minimering av den frie energien fører til en stabilisering av systemet. Diskuter, ved bruk an entropi-betraktninger ved konstant trykk og temperatur, hvorfor endringen i fri energi må være negativ, $\Delta G<0$, for at en prosess skal være spontan.

**svar** Gibbs energi er definert som $G = H-TS$. Siden både $H$, $T$ og $S$ alle er tilstandsfunksjoner vil også $G$ være det. For konstant temperatur vil vi derfor ha at 

$$\Delta G = \Delta H - T\Delta S$$

Når det gjelder entropi vet vi at følgende må gjelde:

$$\Delta S_{tot} = \Delta S_{sys} + \Delta S_{omg}$$

og 

$$dS =\frac{dq}{T} $$

For omgivelsene ved konstant trykk har vi at $\Delta S_{omg} = \frac{-q_{p,sys}}{T} = -\frac{\Delta H_{sys}}{T}$

Setter dette inn i definisjonen av total entropi og får at

$$\Delta S_{tot} = \Delta S_{sys} -\frac{\Delta H_{sys}}{T} \rightarrow T\Delta S_{tot} = T\Delta S_{sys} - \Delta H_{sys} \rightarrow -T\Delta S_{tot} = \Delta H-T\Delta S_{sys} = \Delta G$$

Fra termodynamikkens andre lov vet vi at den totale entropien kun kan øke (eller være lik 0 i en likevekt). Derfor må Gibbs energi minke for en spontan prosess, og være 0 i en likevekt. Obs! legg merke til at dette gjelder for et lukket system hvor T og p holdes konstant!


**b)** Hvordan definerer vi det kjemiske potensialet? Og hvordan kan vi tolke det?

**svar**

Fra fundamentallikningen $dG = Vdp-SdT+\sum_j{\mu dn_j}$ kan vi definere det kjemiske potensialet som $\mu = \left( \frac{\partial G}{\partial n_j} \right)_{p,T,n}$. Det tolkes ofte som drivkraften bak en overføring av partikler mellom faser, eller en kjemisk endring (kjemisk reaksjon). I mange lærebøker for kjemikere presenteres det ofte som "molar Gibbs fri energi", noe som strengt tatt bare er sant dersom man har et enkomponent-system. 


**c)** Bruk fundamentallikningen $dG = Vdp-SdT+\sum_j{\mu dn_j}$ til å utlede en definisjon av $d\mu$ for et enkomponent-system. Her kan det lønne seg å huske at det kjemiske potensialet er en intensiv størrelse, så $G=n_j\cdot \mu$. Hvorfor er denne definisjonen særlig nyttig når vi arbeider med faseoverganger?

**svar** 

Bruker at $G=n_j\cdot \mu$ og skriver på differensialform: 

$$dG = n_jd\mu + dn_j \mu \rightarrow n_jd\mu = dG - dn_j\mu$$

Setter inn fundamentallikningen for én komponent:

$$n_jd\mu = Vdp-SdT+ dn_j\mu - dn_j\mu = Vdp-SdT$$

deler på antallet mol av komponenten og får:

$$d\mu = V_mdp-S_mdT$$

Denne definisjonen er særlig nyttig når vi skal vurdere hvordan likevekter ved faseoverganger er avhengige av trykk og temperatur. Vi bruker denne definisjonen bl.a. til å utlede Clapeyrons likning som forteller oss hvordan trykket i en faseovergang endres hvis man endrer temperatur. Det er dette som gjør det mulig å tegne opp fasediagram som T-p "kart" over faseoverganger hvor $\mu_{\alpha} = \mu_{\beta}$.

**d)** Hvordan avhenger det kjemiske potensialet til et enkomponentsystem av trykk og temperatur?

Ved å bruke sammenhengen over kan vi nå se spesifikt på hvordan det kjemiske potensialet endres med de to størrelsene.

$$d\mu = \left(\frac{\partial \mu}{\partial p}\right)_Tdp+\left(\frac{\partial \mu}{\partial p}T\right)_pdT$$


Dette viser at 

$$ \left(\frac{\partial \mu}{\partial p}\right)_T = V_m \text{ og } \left(\frac{\partial \mu}{\partial p}T\right)_p = -S_m$$

$V_m >0$ (alltid), så det kjemiske potensialet øker altså når trykket øker. Hvor stor påvirkning trykket har på potensialet avhenger av det molare volumet. Gasser (stort $V_m$) vil ha en større endring i $\mu$ enn væsker (mindre $V_m$) vil ha ved samme endring i trykk. 

Det kjemiske potensialet minker når man øker temperaturen. En gass har høyere $S_m$ enn kondenserte faser (tenk tilbake til diskusjoner rundt mikrotilstander), og det kjemiske potensialet faller raskere i en gass enn i væsker og faste faser.

**c)** Hva betyr det i praksis at det kjemiske potensialet minker når temperaturen øker?

**Svar**

Ved høyere temperatur er antallet tilstander som er tilgjengelig for systemet større enn ved lavere temperaturer (husk hvordan energifordelingen ble bredere når T øker). Den energetiske kostnaden ved å "legge til" eller "fjerne" en partikkel til/fra et system med høy entropi er langt mindre enn for et lav-entropisk system, og dermed vil $\mu$ minke når temperaturen øker. ### Oppgave 2

Når en gitt væske med en molar masse på $46.1 \text{g mol}^{-1}$ fryser ved $-3.65^{\circ}C$ endres dens tetthet fra $0.789 \text{g cm}^{-3}$ til $0.801 \text{g cm}^{-3}$. Den molare smelteentalpien til væsken er $8.68 \text{kJ mol}^{-1}$. Estimer frysepunktet for væsken ved $1 \text{Mpa}$. 


**Svar**

Clapeyrons likning beskriver hvordan en faseovergang påvirkes av trykk:

$$\frac{dp}{dT}=\frac{∆_{trs}S_{m}}{∆_{trs}V_{m}}$$

:::{dropdown} Utledning:


Ved en faseovergang må de to fasene være i likevekt med hverandre, dvs. at deres kjemiske potensiale må være likt. Siden vi ser på avhengigheten av trykk og temperatur kan dette skrives som:

$${\mu }_{A}(p,T)={\mu }_{B}(p,T)$$

Det kjemiske potensiale svarer til den molare Gibbs fri energi: $\mu =G_{m}$, og den partiell-deriverte formen av Gibbs fri energi er:

$$dG_{m}=V_{m}dp-S_{m}dT$$

Setter vi at $dG_{m,A\ }=dG_{m,B}$ for de to fasene får vi videre:

$$V_{m,A}dp-S_{m,A}dT=V_{m,B}dp-S_{m,B}dT$$

$${(V}_{m,A\ }-V_{m,B\ })\frac{dp}{dT}=(S_{m,A\ }-S_{m,B\ })$$

Som igjen gir oss Clapeyron likningen!

$$\ \frac{dp}{dT}=\frac{S_{m,A\ }-S_{m,B\ }}{\ V_{m,A\ }-V_{m,B\ }\ }$$

:::


Ved å bruke at $∆G = 0$ ved faseovergang kan denne gjøres om til:

$$\frac{dp}{dT}=\frac{∆_{trs}H_{m}}{∆_{trs}V_{m}\cdot T_{trs}}$$

Integrasjon gir videre:

$$p=p^{\ast }+\frac{∆_{trs}H_{m}}{∆_{trs}V_{m}}ln\left(\frac{T}{T^{\ast }}\right)\approx p^{\ast }+\frac{∆_{trs}H_{m}}{∆_{trs}V_{m}}\left(\frac{T-T^{\ast }}{T^{\ast }}\right)$$

Vi finner forskjellen i molart volum ved å bruke at:

$$\mathrm{\rho }=\frac{m}{V}=\frac{n\cdot M}{V}=\frac{M}{V_{m}}$$

Som gir: 

$$V_m = \frac{M}{\rho}$$

Slik at:

$∆V_{m}=M\left(\frac{1}{\rho_{s}}-\frac{1}{{\rho }_{l}}\right)=46.1\ g.mol^{-1}\cdot \left(\frac{1}{0.789 \, g.cm^{-3}}-\frac{1}{0.801 \, g.cm^{-3}}\right)=-0.8753 \, cm^{3}.mol^{-1}$

Siden det ikke er gitt noe annen informasjon antar vi at frysepunktet ved $-3.65 ^{\circ}C$ er gitt ved standard trykk = $1 \, atm$. Ved innsetting av verdiene i Clapeyrons likning fås da:

$1\ MPa=1 \, atm+\frac{8680 \, J.mol^{-1}}{-0.8753 \, cm^{3}.mol^{-1}}\cdot \left(\frac{T-269.5 \, K}{269.5 \, K}\right)$

$1\ MPa=0.101325 \, MPa+\frac{8680 \, J.mol^{-1}}{-0.8753 \, cm^{3}.mol^{-1}}\cdot \left(\frac{T-269.5 \, K}{269.5 \, K}\right)$

$0.898675\cdot {10}^{6} \, Pa\cdot \frac{-0.8753\cdot {10}^{-6} \, m^{3}.mol^{-1}}{8680 \, J.mol^{-1}}\cdot \ 269.5\ K\ +\ 269.5\ K\ =\ T$

$T = 269.48 K$

Her ble det brukt at $Pa\cdot \ m^{3}\ =\ J$.  Selv om trykket altså øker med omtrent $1*10^6 \, mPa$, minker smeltepunktet minimalt. 

Til sammenlikning vet vi at forskjellen i atmosfærisk trykk mellom havnivå og toppen av Mt. Everest er ca $67 \, kPa$ ($0.067 \, MPa$), og at kokepunktet for vann da minker fra $100$ til $68 ^{\circ} C$. ### Oppgave 3

Molar fordampningsentalpi for vann ved $100^{\circ}C$ er $40.63 \text{kJ mol}^{-1}$. Finn arbeidet som utføres av systemet når $2.00 \text{mol}$ vann fordamper ved $100^{\circ}C$ og $1.00 \text{atm}$ trykk. Finn også forandringene i systemets indre energi $\Delta U$, Gibbs energi $\Delta G$ og entropi $\Delta S$. Tettheten for vann i væskeform ved $100^{\circ}C$ er $0.9584 \text{g cm}^{-3}$, og tettheten for vanndamp ved samme temperatur er $0.000596\text{g cm}^{-3}$.  


**Svar**

$$H_{2}O\left(l,\ 100\ °C,\ 1\ atm\right)\rightarrow \ \ H_{2}O(g,\ 100\mathrm{℃},\ 1\ atm)$$
 
Ved atmosfæretrykk: 

$$1\ atm\ =\ 1,01325\cdot {10}^{5}\ Pa\ (100\mathrm{℃})$$
 
Arbeidet som systemet gjør mot et konstant ytre trykk ($1 \, atm$) når gassen dannes er:

$w=\ -p_{ex}\cdot \Delta V=\ -p_{ex}\cdot n\cdot \Delta V_{m}$

Den molare endringen i volum finner vi, igjen, ved hjelp av tetthetene:

$$\Delta V_{m}=V_{m,vanndamp}-V_{m,vann}=\frac{M_{H_{2}O}}{{\rho }_{vanndamp}}-\frac{M_{H_{2}O}}{{\rho }_{vann}}=\frac{18.01\ gmol^{-1}}{0.000596\ g.cm^{-3}}-\frac{18.01\ g.mol^{-1}}{0.9584\ g.cm^{-3}}=3.0199\cdot {10}^{4} \, cm^{3}.mol^{-1}$$


$$w=\ -\left(1.01325\cdot {10}^{5} \, Pa\right)\cdot \left(2.00\ mol\right)\cdot \left(3.0199\cdot {10}^{4}\cdot {10}^{-6} \, m^{3}.mol^{-1}\right)=\ -6.12\cdot {10}^{3}J$$



Ettersom vi har oppgitt den molare fordampningsentalpien kan vi bruke definisjonen på entalpi for å finne endringen i indre energi:

$$\Delta H=\Delta U+p\Delta V\Rightarrow {\Delta }_{vap}U={\Delta }_{vap}H-p_{ex}\Delta V={\Delta }_{vap}H+w=n\cdot {\Delta }_{vap}H_{m}+w\ $$

$${\Delta }_{vap}U=\left(2.00\ mol\cdot 40.63\cdot {10}^{3}J\ mol^{-1}\right)-6.12\cdot {10}^{3}J=75.1\cdot {10}^{3}J$$



Siden vi er ved en faseovergang må systemet være i likevekt, og ${\Delta }_{\mathrm{vap}}G=0$.

Entropien finner vi ut fra definisjonen på entropi ved varmeoverføring:

$${\Delta }_{vap}S=\frac{q_{rev}}{T}=\frac{{\Delta }_{vap}H}{T_{kokepunkt}}=\frac{n\cdot {\Delta }_{vap}H_{m}}{T_{kokepunkt}}$$

$${\Delta }_{vap}S=\frac{2.00\ mol\cdot 40.63\cdot {10}^{3}J\ mol^{-1}}{373.15K}=218\ JK^{-1}$$

Oppsummert:

$$w=\ -6.12\cdot {10}^{3}J$$

$${\Delta }_{vap}U=75.1\cdot {10}^{3}J$$

$${\Delta }_{\mathrm{vap}}G=0$$

$${\Delta }_{vap}S=218\ JK^{-1}$$

### Oppgave 5


**svar**
For å finne temperaturen der faseovergangen skjer er det hensiktsmessig plotte varmekapasitet som funksjon av temperatur. Varmekapasitet er definert som $C_{V}={\left(\frac{\partial U}{\partial T}\right)}_{V}$ og vi kan derivere dataen numerisk for å approksimere varmekapasiteten i dette systemet. 

Dette kan f.eks. implementeres i Python på følgende måte:

```python
import numpy as np
import matplotlib.pyplot as plt

T = [35, 40, 45, 50, 51, 51.5, 51.6, 55, 60]
U = [401, 435, 499, 582, 776, 887, 2012, 2020, 2031]
C = []

for i in range(len(T)-1):
    dT = T[i+1] - T[i]
    dU = U[i+1] - U[i]
    C.append(dU/dT)

plt.plot(T[0:-1], C)
plt.xlabel('Temperature (°C)')
plt.ylabel("Varmekapasitet")
plt.title('Varmekapasitet vs. Temperatur')
plt.show()
```

Varmekapasiteten har en tydelig topp rundt $51 ^{\circ}C$ og formen på toppen tyder på at det er en første ordens faseovergang som skjer her. Siden datamaskinen bruker diskrete tall vil vi ikke kunne observere at varmekapasiteten divergerer mot uendelig men toppen gir allikevel en god indikasjon på at faseovergangen skjer.
