## Oppgave 1

**a)** En reaksjon er termodynamisk gunstig når den frie energien avtar, $\Delta G < 0$. Dette er en nødvendig betingelse for at reaksjonen skal kunne gå av seg selv i den angitte retningen ved konstant $T$ og $p$.

**b)** Nei. $\Delta G < 0$ sier bare at reaksjonen *kan* skje spontant sett fra et energiperspektiv, det sier ingenting om *hvor fort* den skjer. Alle reaksjoner vil skje om man venter lenge nok, men innenfor vår levealder og praktiske formål vil ikke alle reaksjoner skje. Disse reaksjonene kalles kinetisk hindret, og kommer fra at aktiveringsenergien for reaksjonen er for høy, slik at hastighetskonstanten blir veldig liten, og det vil ta veldig lang tid for reaksjonen til å skje.

**c)** Hastigheten til en reaksjon er hvor fort den skjer. Det kan for eksempel måles ved dannelse/forsvinning av gass, pH, absorbans (en eller annen karakteristisk egenskap ved enten reaktantene eller produktene). Man finner så hastigheten ved å analysere denne endringen over tid.

**d)**

$$v = -\frac{d[A]}{dt} = k[A]$$

Hastigheten er her avhengig av konsentrasjonen av $A$ (bortsett fra 0.ordens reaksjon) og hastighetskonstanten $k$, som igjen avhenger av temperatur og aktiveringsenergi/katalysator.

**e)**

$$k = A\exp\left(-\frac{E_a}{RT}\right)$$

Denne likningen forteller at hastighetskonstanten $k$ avhenger eksponentielt av forholdet mellom aktiveringsenergien $E_a$ og den termiske energien $RT$. $A$ er den preeksponentielle faktoren, som blant annet relateres til hvor ofte molekyler kolliderer i riktig orientering. $E_a$ er aktiveringsenergien, energibarrieren som må overvinnes for at reaksjonen skal skje.

- Økt $A$ gir proporsjonalt økt $k$ (lineær økning).
- Økt $E_a$ gir eksponentielt redusert $k$ (færre molekyler har nok energi).
- Økt $T$ gir eksponentielt økt $k$ (flere molekyler har nok energi til å overvinne barrieren).

Det er viktig å påpeke her at Arrhenius likningen kun forteller sammenhengen mellom aktiveringsenergi og hastighetskonstanten, og ikke den totale hastigheten til en reaksjon. Hastigheten til reaksjonen er også avhengig av konsentrasjonen av reaktantene. 

**f)** Det hastighetsbestemmende trinnet er det tregeste trinnet i en flertrinnsmekanisme, trinnet med høyest energibarriere eller lavest hastighetskonstant sammenlignet med de andre trinnene. Siden de andre trinnene er langt raskere, er det dette trinnet som setter farten for hele den observerte reaksjonen. Vi observerer kun det trege steget, da dette er det som begrenser reaksjonen. 

**g)**

_Profil a)_

Her er det ingen likevekter. Det hastighetsbestemmende trinnet er steg 1. Intermediatene vil ha nok energi til å fort gjennomføre steg 2. Siden det er så stor energibarriere fra produkter $\rightarrow$ intermediat så vil ikke dette bakover steget lett skje. Reaksjonen er også spontan siden $\Delta G < 0$, og energi vil blir frigitt fra reaksjonen.  

_Profil b)_

Reaksjonen er spontan siden $\Delta G < 0$, og energi vil blir frigjort. Siden energibarrieren for steg 1 og 2 er omtrent like stor, er det her ikke mulig å si hvilket steg som er hastighetsbestemmende.  

_Profil c)_

Denne reaksjonen er ikke spontan da $\Delta G > 0$ og energi må tilføres for at den skal kunne gå. Ellers er den lik som a).  

**h)** For denne reaksjonsprofilen er steg 1 det hastighetsbestemmende. Det er fordi dette steget er det med høyest energi for å overkomme, og dette blir den observerte $E_a$, som altså er også fra start-punktet, siden denne energien også kreves for å overkomme steg 1. Fordi $E_a$ i det første steget er veldig høy, så er konsentrasjonen av intermediater veldig lav. Dette er to faktorer som senker hastigheten til en reaksjon. Reaksjonen er i seg selv spontan, men vil kreve høy Tilførsel av energi for å overkomme energibarrieren 

**i)** Reaksjonshastigheten $v$ er definert slik at alle ledd blir like når de er korrigert for støkiometri:

$$v = -\frac{1}{2}\frac{d[A]}{dt} = -\frac{d[B]}{dt} = \frac{1}{4}\frac{d[C]}{dt} = \frac{d[D]}{dt}$$

Gitt $\dfrac{d[C]}{dt} = 3.2 \ \text{mol dm}^{-3}\text{s}^{-1}$:

$$v = \frac{1}{4}\times 3.2 = 0.8 \ \text{mol dm}^{-3}\text{s}^{-1}$$

Forbrukshastighet $A$:

$$-\frac{d[A]}{dt} = 2v = 2\times 0.8 = 1.6 \ \text{mol dm}^{-3}\text{s}^{-1}$$

Forbrukshastighet $B$:

$$-\frac{d[B]}{dt} = v = 0.8 \ \text{mol dm}^{-3}\text{s}^{-1}$$

Dannelseshastighet $D$:

$$\frac{d[D]}{dt} = v = 0.8 \ \text{mol dm}^{-3}\text{s}^{-1}$$

**j)** Approksimasjonen $\dfrac{d[I]}{dt} \approx 0$ betyr at konsentrasjonen av et reaktivt intermediat $I$ forblir tilnærmet konstant gjennom mesteparten av reaksjonsforløpet. Kjemisk grunnlag: intermediatet er svært reaktivt og forbrukes nesten like fort som det dannes, slik at det aldri bygger seg opp i noen betydelig mengde. Dannelseshastigheten og forbrukshastigheten til $I$ er tilnærmet like store til enhver tid, etter en kort initial oppbyggingsfase. Dette lar oss sette dannelses- og forbruksleddene for $I$ lik hverandre og løse for $[I]$, slik at intermediatet kan elimineres fra det totale hastighetsuttrykket.


## Oppgave 2

**a)** I simuleringen har hver partikkel en energi $E$. I molekylboksen fargelegges partiklene etter denne energien: **rød** betyr $E < E_a$ (ikke nok energi), **oransje** betyr $E \geq E_a$ (nok energi til å reagere ved kollisjon), og **grønn** betyr at partikkelen faktisk har reagert (kollidert med en annen partikkel der minst én hadde $E \geq E_a$). I mikrofordelingen tilsvarer x-aksen energien $E$, og den stiplede linjen ved $E_a$ markerer samme terskel som avgjør fargen i boksen: det skraverte området til høyre for $E_a$ i grafen representerer andelen partikler som ville vært oransje/grønne.

**b)** Stigningstallet blir omtrent halvert i tallverdi og kurven flater ut når $T$ dobles, siden $RT$ dobles. Fordelingen brer seg dermed utover mot høyere energier.

I boksen brukes en egen (skjematisk) energiskala som også øker med $T$ (fra $\approx 12$ ved 250 K til $40$ ved 500 K), så antall oransje molekyler i boksen øker tydelig etter hvert som $T$ trekkes oppover, i tråd med at en større andel av fordelingen nå ligger til høyre for $E_a$.

**c)** Simuleringen beregner andelen eksakt som $e^{-E_a/RT}$, og hastighetskonstanten som $k = A\cdot e^{-E_a/RT}$, derfor er $k$ direkte proporsjonal med andelen, med $A$ som proporsjonalitetskonstant. 

Andelen med nok energi øker kraftig med $T$, og $k$ følger nøyaktig samme relative økning, fordi denne mikroskopiske andelen *er* eksponentialleddet $e^{-E_a/RT}$ i Arrhenius-likningen $k=Ae^{-E_a/RT}$.

**d)** Eksempeltall å se på: 

| $E_a\,(\text{kJ/mol})$ | Andel $E>E_a$ | $k\,(\text{s}^{-1})$ |
|:---:|:---:|:---:|
| $20$ | $3{,}1\times10^{-4}$ | $3{,}1\times10^{8}$ |
| $50$ | $1{,}7\times10^{-9}$ | $1{,}7\times10^{3}$ |
| $80$ | $9{,}5\times10^{-15}$ | $9{,}5\times10^{-3}$ |
| $120$ | $9{,}2\times10^{-22}$ | $9{,}2\times10^{-10}$ |

Når $E_a$ øker fra $20$ til $120\,\text{kJ/mol}$, faller andelen molekyler med nok energi med over 17 størrelsesordener. I simuleringen ser man dette som at antall oransje molekyler i boksen nærmest forsvinner helt ved høy $E_a$, og at boksen fylles svært mye langsommere med grønne produktmolekyler. Reaksjonen går praktisk talt ikke ved høye $E_a$-verdier ved denne temperaturen.

**e)** Andelen molekyler med nok energi er allerede beskrevet fullstendig av eksponentialleddet $e^{-E_a/RT}$, det er dette leddet som endres når man drar i $T$- eller $E_a$-glidebryteren, ikke $A$. I simuleringen holdes $A=1{,}0\times10^{12}\,\text{s}^{-1}$ fast for å isolere hvordan $k$ endres utelukkende via den mikroskopiske energifordelingen. Siden $A$ ikke endres når andelen aktiverte molekyler endres, kan den fysisk ikke representere selve andelen, i stedet representerer $A$ hvor ofte molekyler kolliderer, og med hvilken sannsynlighet en kollisjon skjer i riktig retning/orientering for at reaksjonen skal kunne skje, altså kollisjonsfrekvensen og en eventuell steriske faktor, helt uavhengig av om energien er tilstrekkelig.

## Oppgave 3

Gjør om temperaturene til Kelvin:

$$T_1 = 19+273.15 = 292.15 \ \text{K}, \qquad T_2 = 37+273.15 = 310.15 \ \text{K}$$

Bruk to-punkts Arrhenius-formen for å finne $E_a$:

$$\ln\left(\frac{k_2}{k_1}\right) = -\frac{E_a}{R}\left(\frac{1}{T_2}-\frac{1}{T_1}\right)$$

Løs for $E_a$:

$$E_a = -R\frac{\ln(k_2/k_1)}{\left(\frac{1}{T_2}-\frac{1}{T_1}\right)} = -8.314\times \frac{\ln(3.38\times10^{-3}/2.78\times10^{-4})}{\left(\frac{1}{310.15}-\frac{1}{292.15}\right)}$$

$$E_a \approx 104.5 \ \text{kJ mol}^{-1}$$

Bruk deretter $E_a$ til å finne $A$ fra Arrhenius-likningen med ett av datapunktene:

$$A = \frac{k_1}{\exp(-E_a/RT_1)} = \frac{2.78\times10^{-4}}{\exp\left(-104546/(8.314\times292.15)\right)}$$

$$A \approx 1.37\times10^{15} \ \text{dm}^3\text{mol}^{-1}\text{s}^{-1}$$

Kontroll: samme $A$-verdi fås ved å bruke $k_2$ og $T_2$ i stedet.

## Oppgave 5

Mekanismen er:

$$\text{O}_3 \xrightarrow{k_a} \text{O}_2 + \text{O}$$
$$\text{O}_2 + \text{O} \xrightarrow{k_a'} \text{O}_3$$
$$\text{O} + \text{O}_3 \xrightarrow{k_b} 2\text{O}_2$$

Intermediatet er oksygenatomet $\text{O}$. Sett først opp hastighetsuttrykket for nedbrytning av ozon, basert på alle trinn som forbruker/danner $\text{O}_3$:

$$\frac{d[\text{O}_3]}{dt} = -k_a[\text{O}_3] + k_a'[\text{O}_2][\text{O}] - k_b[\text{O}][\text{O}_3]$$

Bruk steady-state-approksimasjonen på intermediatet $\text{O}$:

$$\frac{d[\text{O}]}{dt} = k_a[\text{O}_3] - k_a'[\text{O}_2][\text{O}] - k_b[\text{O}][\text{O}_3] \approx 0$$

Løs for $[\text{O}]$:

$$[\text{O}] = \frac{k_a[\text{O}_3]}{k_a'[\text{O}_2] + k_b[\text{O}_3]}$$

Sett dette uttrykket inn i hastighetsuttrykket for $\text{O}_3$ og forenkle:

$$\frac{d[\text{O}_3]}{dt} = -\frac{2k_ak_b[\text{O}_3]^2}{k_a'[\text{O}_2]+k_b[\text{O}_3]}$$

Dersom re-dannelsen av ozon (trinn 2) er langt raskere enn nedbrytningen i trinn 3 (dvs. $k_a'[\text{O}_2] \gg k_b[\text{O}_3]$, som er tilfellet i store deler av stratosfæren), forenkles uttrykket videre til:

$$\frac{d[\text{O}_3]}{dt} \approx -\frac{2k_ak_b}{k_a'}\cdot\frac{[\text{O}_3]^2}{[\text{O}_2]}$$
