# Uke 8 – Løsninger og blandinger

### Oppgave 1

**a)** Som repetisjon, hva er en ideell gass? Hva ville ${\Delta}_{mix}H$ vært for blanding av to ideelle gasser? 

**Svar** En ideell gass er en gass uten interaksjoner mellom molekylene, der vi regner hvert molekyl som et lite punkt uten utstrekning. $\Delta_{mix} H$ ville vært $0$ for en blanding av to ideelle gasser siden det ikke er noen intermolekylære interaksjoner mellom molekylene. 

**b)** Hva er en ideell væske? Hva er ${\Delta }_{mix}H$ for en ideell væske? Hva betyr dette for interaksjonene og strukturen i væsken? 

**Svar** Siden væsker har interaksjoner mellom seg, kan man ikke si at intermolekylære interaksjoner er lik $0$. Det man gjør isteden er å sette de til en gjennomsnittlig styrke. Det som menes med det er at interaksjonene mellom stoffene A og B er lik interaksjonene mellom A og A, og B og B. $\Delta_{mix} H$ er fortsatt $0$ når man blander to ideelle væsker. Det betyr at det ikke er noen forskjell i intermolekylære bindinger mellom molekylene i væsken. En homogen, helt ren væske vil derfor alltid være ideell, fordi alle interaksjonene mellom molekylene er de samme. 
	
**c)** Hva er drivkraften for miksing av to ideelle væsker?

**Svar** Drivkraften er entropisk: det er statistisk mye mer sannsynlig (flere mikrotilstander) at molekylene er tilfeldig blandet enn separert, noe som gir $\Delta S_{mix} > 0$ og dermed $\Delta G_{mix} = \Delta H_{mix} − T\Delta S_{mix} < 0 \rightarrow$ spontan blanding. Materie og energien blir mer spredt om væskene er mikset enn hver for seg.  

	
**d)** Man kan regne på egenskaper ved ideelle væsker ved både Henry og Raoults lov. Hva skiller disse fra hverandre? Når bruker man hvilken, og hvorfor kan ikke begge brukes på samme system? 

**Svar** Både Raoults og Henrys lov beskriver sammenhengen mellom damptrykk over en blanding og dens sammensetning. For en blanding av to ideelle væsker er denne sammenhengen lineær og dette beskrives av Raoults lov. I en blanding med høy konsentrasjon av én av væskene vil de fleste molekylene rundt et gitt molekyl være like som molekylet. Derfor vil blandingen oppføre seg nesten ideell og i det tilfelle kan den godt beskrives ved hjelp av Raoults lov. For ikke-ideelle blandinger vil sammenhengen mellom damptrykk og blandingsforhold være lineær kun ved lave konsentrasjoner. En slik situasjon beskrives av Henrys lov.  
	
**e)** Når sier man at en væske er ideell? Kom med to eksempler på systemer med to væsker som oppfører seg ideelt. (Evt. hva som skal til for at de mest sannsynlig vil oppføre seg ideelt.)

**Svar** En væske sies ideell om den følger Raoults lov, dvs når komponentene er så kjemisk/strukturelt like at A–A, B–B og A–B-interaksjoner er tilnærmet identiske. De to væskene bør være mest mulig like for at de mest sannsynlig oppfører seg ideelt. Eksempler er benzen og toluen, n-heptan og n-heksan, etyl-bromide og etyl-jodid, klorbenzen og brombenzen.  
	
**f)** Se for dere et stort volum med vann ved $25 ^{\circ} C$ ($10 \, L$ for eksempel). Når $1 \, mol$ vann videre tilsettes så øker volumet med $18 \, cm^3$, og man sier derfor at det molare volumet til vann er $18 \, cm^3$. Men når man tilsetter $1 \, mol$ vann til et stort volum med etanol ($10 \, L$ for eksempel) ved $25 ^{\circ} C$ øker volumet kun med $14 \, cm^3$. Hvorfor er det slik? Hva kommer dette av? 

**Svar** Det er fordi nettverket av hydrogenbindinger mellom vannmolekylene ikke kan opprettholdes sammen med etanol, og $\text{H}_2\text{O}$ molekylene bruker derfor mindre plass (de kommer nærmere hverandre, svakere krefter mellom dem) blandet med store mengder etanol.  
	
**g)** Er systemet i f) en ideell væske? 

**Svar** Nei, en blanding av etanol og vann er ikke en ideell væske. Interaksjonene mellom vann og vann, vann og etanol, og etanol og etanol er forskjellige.  Antall hydrogenbindinger som kan dannes imellom speciene varierer.  

**h)** Hva er partielt molart volum? Hvordan vil det være for en ideell væske?  

**Svar** Partielt molart volum er volumet som ett mol av en komponent i en blanding bidrar til det totale volumet. Partielt molart volum kan variere med sammensetning av blandingen. For en ideell væske vil det partielle molare volumet være lik som for de rene stoffene. I eksempelet med vann og etanol ville det vært $18 \text{cm}^3$.

**i)** Hvorfor brukes damptrykk og partialtrykk så mye for å se på egenskaper til væsker?  

**Svar** Her er det flere ting man kan komme innom, for eksempel:  

 - Enkelt å måle 
 - Gir utrykk for de molekylære egenskapene til systemet. Store forskjeller mellom molekyler, gir en god innsikt i hva som skjer, og hvordan forskjellige faktorer påvirker.  
 - Endres ved kolligative egenskaper 
 - Gir uttrykk for kinetisk energi og entropiske bidrag 


### Oppgave 2

**a)** Definisjonen av kjemisk potensial for en komponent j er ${\mu }_{j}={\left(\frac{\partial G}{\partial n_{j}}\right)}_{p,\ T,\ n^{´}}$. Forklar denne likningen og hva den betyr. 

**Svar** Den sier at kjemisk potensial er hvor mye Gibbs fri energi øker/avtar med endret mengde av komponent J. Så for ett stoff er kjemisk potensial det samme som molar Gibbs fri energi, men for flere komponenter er kjemisk potensial gitt ved stigningstallet til Gibbs fri energi når den ene komponenten endres.
	
**b)** Hva reduseres den fundamentale likningen for kjemisk termodynamikk seg til ved konstant trykk og temperatur (lab betingelser). Hva betyr det dere kommer frem til? 

**Svar** Den fundamentale likningen for kjemisk termodynamikk: 

$ \Delta G = VdP - SdT + \mu_Adn_A + \mu_Bdn_B. $
 
 Ved konstant trykk og temperatur reduseres denne til:  
 
 $ dG = \mu_Adn_A + \mu_Bdn_B.$
 
 Det betyr at endring i Gibbs fri energi kan komme fra endring i sammensetning i systemet, for eksempel endring i sammensetning når reaktanter blir produkter.

**c)** Ved konstant trykk og temperatur er Gibbs-Duhem likningen: $\sum_{j} n_{j}{d\mu }_{j}=0$. Hva betyr dette? 

**Svar** Gibbs-Duhem likningen forteller at det kjemiske potensialet av en komponent i systemet ikke kan endres uavhengig av de andre komponentene.  

**d)** Fra variasjonen i kjemisk potensial med trykk: $\mu ={\mu }^{0}+RTln\frac{p}{p^{0}}$, hvordan kommer man frem til Gibbs energi for blanding av de to gassene A og B. ${\Delta }_{mix}G=nRT\left(x_{A}\ln{x_{A}+x_{B}\ln{x_{B}}}\right)$? 

**Svar** Gibbs energien før blanding (rene gasser A og B hver for seg ved totaltrykk $p$) sammenlignes med Gibbs energien etter blanding (samme gasser blandet, hver med partialtrykk $x_Ap$ og $x_Bp$). Bruker man $\mu = \mu° + RT \ln(p/p°)$ for hver komponent før og etter, og summerer 

$G = \sum n_J \mu_J$, 

finner man at differansen blir

$\Delta_{mix}G = nRT(x_A \ln x_A + x_B \ln x_B)$

fordi 

$\ln(x_Jp) − \ln(p) = \ln x_J$

for hver komponent, og de øvrige leddene kansellerer.

**e)** Hva er ${\Delta }_{mix}S$ for en ideell væske? 

**Svar** Samme uttrykk som for ideelle gasser: $\Delta_{mix}S = −nR(x_A \ln x_A + x_B \ln x_B)$, siden entropien for ideell blanding kun avhenger av antall måter å fordele molekylene tilfeldig på, uavhengig av om det er gass eller væske.

### Oppgave 3

**a)** Kolligative egenskaper er intensive egenskaper som kun avhenger av andelen løst stoff per løsemiddel, ikke hva slags løsning det er. Egenskapene omfatter frysepunkt, kokepunkt, damptrykk og osmotisk trykk. Hvordan foregår slike endringer?

**Svar** Disse egenskapene endres fordi tilstedeværelsen av oppløst stoff senker det kjemiske potensialet til løsningsmiddelet, uavhengig av hva slags stoff det er, kun avhengig av antall partikler oppløst. Dette skjer fordi det oppløste stoffet "fortynner" løsningsmiddelet og øker løsningens entropi i flytende fase, noe som forskyver likevekten mellom faser.

**b)** Hvordan tror dere at de forskjellige egenskapene endres når mengde løst stoff endres? Blir de lavere eller høyere når andelen økes, og hvordan kan dere forklare dette? 
	- Damptrykk
	- Kokepunkt
	- Frysepunkt 

**Svar**

 - Damptrykk: senkes når mengde løst stoff øker, løsningsmiddelets kjemiske potensial/aktivitet reduseres.
 - Kokepunkt: heves. Siden damptrykket er lavere ved en gitt temperatur, må temperaturen økes mer for å nå damptrykk lik atmosfæretrykket.
 - Frysepunkt: senkes. Det oppløste stoffet stabiliserer flytende fase relativt til fast fase, slik at lavere temperatur kreves for å nå likevekt mellom fast og flytende fase.
 - Osmotisk trykk: økes. En økt konsentrasjon av løste partikler vil gi en sterkere drivkraft for vannmolekylene til å krysse den semipermeable membranen for å jevne ut det kjemiske potensialet.



### Oppgave 4
En blanding består av etyliodid og etylacetat. Damptrykket ($p_I$) av etyliodid(I) over blandingen som funksjon av molbrøken $x_I$ er ved $50.0 °C$ målt til:
| \($x_I$\) | \($p_I$\) / $torr$ |
|:-------:|:--------------:|
| 1.00 | 353 |
| 0.909 | 322 |
| 0.825 | 296 |
| 0.548 | 213 |
| 0.235 | 105 |
| 0.109 | 52.7 |
| 0.0579 | 28.0 |
| 0.0 | 0 |


**a)**	Tegn $p_I$ som funksjon av $x_I$. Følger denne damptrykkskurven Raoults lov? Hvorfor følger den, evt hvorfor følger den ikke? 


**Svar** 

$p_{I}(\text{Raoult}) = x_{I} \times 353 \ \text{Torr}$

| $x_{I}$ | Raoult-prediksjon $\text{[Torr]}$ | Faktisk $p_{I} \ \text{[Torr]}$ | Forhold |
|:---:|:---:|:---:|:---:|
| $1.00$ | $353$ | $353$ | $1.000$ |
| $0.909$ | $320.9$ | $322$ | $1.003$ |
| $0.825$ | $291.2$ | $296$ | $1.016$ |
| $0.548$ | $193.4$ | $213$ | $1.101$ |
| $0.235$ | $83.0$ | $105$ | $1.266$ |
| $0.109$ | $38.5$ | $52.7$ | $1.370$ |
| $0.0579$ | $20.4$ | $28.0$ | $1.370$ |

Ved høy $x_{I}$ (nær $1.00$) følger damptrykket Raoults lov godt (avvik $< 2\%$). Ved lav $x_{I}$ er avviket betydelig (opp mot $37\%$) — løsningsmiddelet følger Raoults lov nær $x_{I} = 1$, mens den fortynnede komponenten følger Henrys lov.



**b)**	Finn Henrys lov konstanten $K_I$ for etyliodid.

**Svar**

$K_{I} = \frac{p_{I}}{x_{I}}$

Ved lave konsentrasjoner:

$x_{I} = 0.0579: \quad K_{I} = \frac{28.0}{0.0579} \approx 483.6 \ \text{Torr}$

$x_{I} = 0.109: \quad K_{I} = \frac{52.7}{0.109} \approx 483.5 \ \text{Torr}$

## Regneoppgaver




### Oppgave 1

Hovedkomponentene i luft er ${N_2}_{(g)}$ og ${O_2}{(g)}$. 78 % ${N_2}_{(g)}$ og 22% $O_2$ blandes ved $298 \, K$, beregne:

**a)**	molar Gibbs blandingsenergi, $\Delta _{mix}G_m$

**Svar** 
$ x(\text{N}_2) = 0.78$, $x(\text{O}_2) = 0.22$, $T = 298 \text{K}$. 
$\Delta_{mix}G_m = RT[x_{N_2} \ln(x_{N_2}) + x_{O_2} \ln(x_{O_2})] = 8.314\times298(−0.5269) \approx −1305 \text{J/mol} \approx −1.31 \text{kJ/mol}$.

**b)**	molar Gibbs blandingsentalpi, $\Delta _{mix}H_m$

**Svar** $\Delta_{mix}H_m = 0$ (antar ideelle gasser, ingen interaksjonsenergi endres ved blanding).

**c)**	molar blandingsentropi, $\Delta _{mix}S_m$

Er blandingsprosessen spontan? Hva er drivkraften for at komponentene blandes?

**Svar** $\Delta_{mix}S_m = −R[x_{N_2} \ln(x_{N_2}) + x_{O_2}\ln(x_{O_2})] \approx 4.38 \text{J/(Kmol)}$

Test: $−T\Delta S = −298\times 4.38 = −1305 \text{J/mol} = \delta G$, siden $\Delta H = 0$.

Blandingen er spontan. Drivkraften er utelukkende entropi, siden entalpibidraget er null.


### Oppgave 2


**a)**	Finn konstanten i Henrys lov for en blanding som består av $75.0 \, g$ toluen (molar masse $92.1 g/mol$) (væske A) i en løsning med $1000 g$ av difenyleter (væske B). Væsken difenyleter (molar masse $170.2 \, g/mol$) har et damptrykk på $430 \, Torr$ over denne løsning og totaltrykket er $520 \, Torr$. Damptrykket for en ren væske A er $650 \, Torr$. 

**Svar** 

$$n_{\text{tol}} = \frac{75.0 g}{92.1 g/mol} = 0.8143 \ \text{mol}, \qquad n_{\text{dfe}} = \frac{1000 g}{170.2 g/mol} = 5.875 \ \text{mol}$$

$$x_{\text{tol}} = \frac{n_{\text{tol}}}{n_{\text{tol}} + n_{\text{dfe}}} = \frac{0.8143}{0.8143 + 5.875} = 0.1217, \qquad x_{\text{dfe}} = 1 - x_{\text{tol}} = 0.8783$$

$$p_{\text{dfe}} = 430 \ \text{Torr} \, \qquad p_{\text{tot}} = 520 \ \text{Torr}$$

$$p_{\text{tol}} = p_{\text{tot}} - p_{\text{dfe}} = 520 - 430 = 90 \ \text{Torr}$$

$K_{H} = \frac{p_{\text{tol}}}{x_{\text{tol}}} = \frac{90}{0.1217} \approx 739 \ \text{Torr}$

**b)**	Følger væske A Raoults lov? Forklar grunnen til svaret du får.  

**Svar** $p_{\text{tol}}(\text{Raoult}) = x_{\text{tol}} \cdot p_{\text{tol}}^{*} = 0.1217 \times 650 \approx 79.1 \ \text{Torr}$

Faktisk verdi er $90 \ \text{Torr}$, altså et **positivt avvik** fra Raoults lov. Toluen følger i stedet Henrys lov med sin egen empiriske konstant.

**c)**	Hva blir konstanten i Henrys lov om løsningen er ideell? 

**Svar** $K_{H} = p_{\text{tol}}^{*} = 650 \ \text{Torr}$.


### Oppgave 3

En $1.0 \, m$ løsning av $NaCl_{(s)}$ i vann gir en frysepunktnedsettelse på $3.7 \, K$. Hvordan kan man forklare denne observasjon? Frysepunktnedsettelseskoeffisienten for vann er $1.86 \, K.kg.mol^{-1}$.

**Svar** 

$\Delta T = K_{f} \cdot m = 1.86 \times 1.0 = 1.86 \ \text{K}$

Observert $\Delta T = 3.7 \ \text{K}$:

$\frac{3.7}{1.86} \approx 1.99 \approx 2$

Van't Hoff-faktoren $i \approx 2$, siden $\text{NaCl}$ dissosierer fullstendig i vann til $\text{Na}^{+}$ og $\text{Cl}^{-}$.

### Oppgave 4

I en blanding av aceton og kloroform er molbrøken av kloroform $0.290$.
De partielle molare volum av komponentene er bestemt:

$$V_{aceton}  =  73.78 \, cm^3/mol$$
$$V_{kloroform}  = 80.41 \, cm^3/mol$$

Beregn volumet av $1.000 \, kg$ blanding.


**Svar** 

$x_{\text{CHCl}_{3}} = 0.290, \qquad x_{\text{aceton}} = 1 - x_{\text{CHCl}_{3}} = 0.710$

$M_{\text{aceton}} = 58.08 \ \text{g/mol}, \qquad M_{\text{CHCl}_{3}} = 119.38 \ \text{g/mol}$

$M_{\text{avg}} = 0.710 \times 58.08 + 0.290 \times 119.38 \approx 75.86 \ \text{g/mol}$

$n_{\text{tot}} = \frac{1000 \ \text{g}}{75.86 \ \text{g/mol}} \approx 13.18 \ \text{mol}$

$n_{\text{aceton}} = 0.710 \times 13.18 \ \text{mol} \approx 9.36 \ \text{mol}, \qquad n_{\text{CHCl}_{3}} = 0.290 \times 13.18 \ \text{mol} \approx 3.82 \ \text{mol}$

$V = n_{\text{aceton}} \times V_{\text{aceton}} + n_{\text{CHCl}_{3}} \times V_{\text{CHCl}_{3}} = 9.36 \ \text{mol} \times 73.78 \ \text{cm}^{3} + 3.82 \ \text{mol} \times 80.41 \ \text{cm}^{3} \approx 998 \ \text{cm}^{3}$



### Oppgave 5

Beregne det osmotiske trykket som oppstår når $6.00 \, g$ urea, $(NH_2)_2CO$, løses i vann til $1.00 \, dm^3$ løsning ved $27.0 ^{\circ}C$. Bruk van’t Hoffs ligning.

**Svar**

### Oppgave 6

Beregn frysepunktet for $200 \, g$ vann som er tilsatt $2.50 \, g$ sukrose $C_{12}H_{22}O_{11}$. Kommenter svaret og gi en forklaring. 

**Svar**

Finn antall mol sukrose:

$n_{\text{sukrose}} = \frac{2.50}{342.3} \approx 0.00730 \ \text{mol}$

Regn om til molalitet:

$m = \frac{0.00730}{0.200 \ \text{kg}} \approx 0.0365 \ \text{mol/kg}$

Bruk frysepunktsnedsettelsesformelen:

$\Delta T_{f} = K_{f} \cdot m = 1.86 \times 0.0365 \approx 0.068 \ \text{K}$

Trekk fra normalt frysepunkt ($0°\text{C}$) for å finne nytt frysepunkt:

$T_{f} \approx -0.068\,°\text{C}$

**Kommentar:** Endringen er svært liten fordi konsentrasjonen av sukrose er lav, og sukrose ikke dissosierer i vann.
