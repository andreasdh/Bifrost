# Uke 9 – Løsninger og blandinger

## Diskusjonsoppgaver

## Oppgave 1

**a)**  I gittermodellen tenker vi oss et fast, regelmessig gitter der hver plass er okkupert av enten en A- eller B-partikkel. Ved å telle antall måter partiklene kan plasseres på (mikrotilstander, $\Omega$), kan vi bruke statistisk mekanikk ($S = k_B \ln \Omega$) til å finne blandingsentropien. Ved å telle nærmeste-nabo-kontakter mellom partikler og tilordne en kontaktenergi til hver type kontakt (AA, BB, AB), kan vi finne blandingsenergien. Disse to bidragene settes sammen til fri energi $F = U − TS$, og derfra kan vi utlede kjemisk potensial, aktivitet, fasediagram osv.

**b)** For å utlede et bestemt uttrykk, må vi først definere systemet vårt og hvilken prosess vi er interesserte i. Så må vi såklart basere utledningen på modellen vår, gittermodellen, der vi kjenner antallet stoffer, antallet plasser $N$ og har en størrelse for mengden av hver av stoffene $N_X$. Vi starter med uttrykket for løsningsentropien gitt ved Boltzmann entropi og multiplisiteten $\Omega$, gitt ved binomialkoeffisienten. Videre bruker vi diverse tilnærminger og metoder: Stirlings approksimasjon, logaritmeregler. Mye av prosessen handler om insetting og omskrivinger for å få et uttrykk som er lett leselig og enkelt å tolke fysisk: entropien øker fordi det finnes mange flere måter å blande A og B tilfeldig på enn å holde dem separert (kun én måte). Til slutt bytter vi ut antallet molekyler med relative konsentrasjoner i form av molfraksjoner.

**c)** I den enkleste versjonen av gittermodellen (ren blandingsentropi, ingen interaksjonsenergi mellom A og B, eller $w_{AA} = w_{BB} = w_{AB}$) er det ingen netto energiendring ved blanding ($\Delta U_{mix} = 0$). Da blir den frie energien $F = \Delta U_{mix} − T\Delta S_{sol} = −T\Delta S_{sol}$. Energibidraget forsvinner altså fordi vi antar at interaksjonene mellom like og ulike partikler er identiske, blandingen drives da utelukkende av entropi (statistisk sannsynlighet), ikke av energigevinst.

**d)** Termene betyr: 
 - $(zw_{AA}/2)N_A$: bidrag fra AA-kontakter. $z$ er koordinasjonstallet (antall naboer per plass), $w_AA$ er kontaktenergien for en AA-nabopar. Faktoren $1/2$ unngår dobbeltelling av par.
 - $(zw_{BB}/2)N_B$: tilsvarende bidrag fra BB-kontakter.
 - $m_{AB}(w_{AB} − (w_{AA}+w_{BB})/2)$: $m_{AB}$ er antall AB-kontakter (blandingskontakter). Leddet i parentes er "overskuddsenergien" ved å danne en AB-kontakt sammenlignet med gjennomsnittet av å ha en ren AA- og en ren BB-kontakt. Dette leddet er selve drivkraften (eller hindringen) for blanding på energisiden: er det negativt foretrekker systemet AB-kontakter (blanding er gunstig energetisk), er det positivt foretrekker systemet å holde like partikler sammen (fasesepareringstendens).

**e)** Antagelsen er at partiklene er tilfeldig fordelt på gitteret, slik at sannsynligheten for at en gitt nabo er A eller B kun avhenger av den gjennomsnittlige konsentrasjonen, uavhengig av hvor på gitteret man befinner seg. Dette lar oss erstatte det eksakte (og vanskelige) antall AB-kontakter $m_{AB}$ med et gjennomsnittsuttrykk basert på $x$. Tilnærmingen fungerer dårlig når det er sterk lokal orden/klumping, f.eks. nær fasesepareringspunktet, ved lave temperaturer der interaksjonene dominerer over tilfeldig termisk bevegelse, eller ved svært asymmetriske/sterke interaksjoner der partiklene ikke lenger er tilfeldig fordelt lokalt.

**f)** $\chi_{AB} = (z/k_BT)(w_{AB} − (w_{AA}+w_{BB})/2)$

Den fysiske tolkningen: $\chi$ måler styrken av den energetiske "ubalansen" mellom AB-kontakter og de rene AA/BB-kontaktene, relativt til termisk energi ($k_BT$). Stor positiv $\chi$ betyr at AB-kontakter er energetisk ugunstige sammenlignet med å holde like partikler sammen, jo større $\chi$, jo sterkere driver systemet mot fasesepareing. $\chi$ nær $0$ betyr tilnærmet ideell blanding.

**g)** Modellen antar:

- alle partikler/molekyler har lik størrelse og fyller gitterplassene perfekt (urealistisk for f.eks. polymerer eller store/små molekylkombinasjoner)
- kun nærmeste-nabo-interaksjoner telles
- tilfeldig (mean-field) fordeling selv når lokal orden faktisk oppstår
- ingen volumendring ved blanding
- interaksjonsenergiene $w_{ij}$ antas konstante og uavhengige av konsentrasjon og temperatur. Reelle løsninger har ofte volumendringer ved blanding, temperaturavhengige interaksjoner, og molekyler av ulik størrelse/form, modellen er derfor kun kvalitativt/semikvantitativt riktig for mange systemer.


## Oppgave 2

**a)** En ideell gass er en gass uten interaksjoner mellom molekylene, der vi regner hvert molekyl som et lite punkt uten utstrekning. $\Delta_{mix} H$ ville vært $0$ for en blanding av to ideelle gasser siden det ikke er noen intermolekylære interaksjoner mellom molekylene.  

**b)** Siden væsker har interaksjoner mellom seg, kan man ikke si at intermolekylære interaksjoner er lik $0$. Det man gjør isteden er å sette de til en gjennomsnittlig styrke. Det som menes med det er at interaksjonene mellom stoffene A og B er lik interaksjonene mellom A og A, og B og B. $\Delta_{mix} H$ er fortsatt $0$ når man blander to ideelle væsker. Det betyr at det ikke er noen forskjell i intermolekylære bindinger mellom molekylene i væsken. En homogen, helt ren væske vil derfor alltid være ideell, fordi alle interaksjonene mellom molekylene er de samme. 

**c)** Drivkraften er entropisk: det er statistisk mye mer sannsynlig (flere mikrotilstander) at molekylene er tilfeldig blandet enn separert, noe som gir $\Delta S_{mix} > 0$ og dermed $\Delta G_{mix} = \Delta H_{mix} − T\Delta S_{mix} < 0 \rightarrow$ spontan blanding. Materie og energien blir mer spredt om væskene er mikset enn hver for seg.  

**d)** Både Raoults og Henrys lov beskriver sammenhengen mellom damptrykk over en blanding og dens sammensetning. For en blanding av to ideelle væsker er denne sammenhengen lineær og dette beskrives av Raoults lov. I en blanding med høy konsentrasjon av én av væskene vil de fleste molekylene rundt et gitt molekyl være like som molekylet. Derfor vil blandingen oppføre seg nesten ideell og i det tilfelle kan den godt beskrives ved hjelp av Raoults lov. For ikke-ideelle blandinger vil sammenhengen mellom damptrykk og blandingsforhold være lineær kun ved lave konsentrasjoner. En slik situasjon beskrives av Henrys lov.  

**e)** En væske sies ideell om den følger Raoults lov, dvs når komponentene er så kjemisk/strukturelt like at A–A, B–B og A–B-interaksjoner er tilnærmet identiske. De to væskene bør være mest mulig like for at de mest sannsynlig oppfører seg ideelt. Eksempler er benzen og toluen, n-heptan og n-heksan, etyl-bromide og etyl-jodid, klorbenzen og brombenzen.  

**f)** Det er fordi nettverket av hydrogenbindinger mellom vannmolekylene ikke kan opprettholdes sammen med etanol, og $\text{H}_2\text{O}$ molekylene bruker derfor mindre plass (de kommer nærmere hverandre, svakere krefter mellom dem) blandet med store mengder etanol.  

**g)** Nei, en blanding av etanol og vann er ikke en ideell væske. Interaksjonene mellom vann og vann, vann og etanol, og etanol og etanol er forskjellige.  Antall hydrogenbindinger som kan dannes imellom speciene varierer.  

**h)** Partielt molart volum er volumet som ett mol av en komponent i en blanding bidrar til det totale volumet. Partielt molart volum kan variere med sammensetning av blandingen. For en ideell væske vil det partielle molare volumet være lik som for de rene stoffene. I eksempelet med vann og etanol ville det vært $18 \text{cm}^3$.

**i** Her er det flere ting man kan komme innom, for eksempel:  

 - Enkelt å måle 
 - Gir utrykk for de molekylære egenskapene til systemet. Store forskjeller mellom molekyler, gir en god innsikt i hva som skjer, og hvordan forskjellige faktorer påvirker.  
 - Endres ved kolligative egenskaper 
 - Gir uttrykk for kinetisk energi og entropiske bidrag 


## Oppgave 3

**a)** Den sier at kjemisk potensial er hvor mye Gibbs fri energi øker/avtar med endret mengde av komponent J. Så for ett stoff er kjemisk potensial det samme som molar Gibbs fri energi, men for flere komponenter er kjemisk potensial gitt ved stigningstallet til Gibbs fri energi når den ene komponenten endres.  

**b)** Den fundamentale likningen for kjemisk termodynamikk: 

$ \Delta G = VdP - SdT + \mu_Adn_A + \mu_Bdn_B. $
 
 Ved konstant trykk og temperatur reduseres denne til:  
 
 $ dG = \mu_Adn_A + \mu_Bdn_B.$
 
 Det betyr at endring i Gibbs fri energi kan komme fra endring i sammensetning i systemet, for eksempel endring i sammensetning når reaktanter blir produkter.  

**c)** Gibbs-Duhem likningen forteller at det kjemiske potensialet av en komponent i systemet ikke kan endres uavhengig av de andre komponentene.  

**d)** Gibbs energien før blanding (rene gasser A og B hver for seg ved totaltrykk $p$) sammenlignes med Gibbs energien etter blanding (samme gasser blandet, hver med partialtrykk $x_Ap$ og $x_Bp$). Bruker man $\mu = \mu° + RT \ln(p/p°)$ for hver komponent før og etter, og summerer 

$G = \sum n_J \mu_J$, 

finner man at differansen blir

$\Delta_{mix}G = nRT(x_A \ln x_A + x_B \ln x_B)$

fordi 

$\ln(x_Jp) − \ln(p) = \ln x_J$

for hver komponent, og de øvrige leddene kansellerer.

**e)** Samme uttrykk som for ideelle gasser: $\Delta_{mix}S = −nR(x_A \ln x_A + x_B \ln x_B)$, siden entropien for ideell blanding kun avhenger av antall måter å fordele molekylene tilfeldig på, uavhengig av om det er gass eller væske.


## Oppgave 4

**a)** Disse egenskapene endres fordi tilstedeværelsen av oppløst stoff senker det kjemiske potensialet til løsningsmiddelet, uavhengig av hva slags stoff det er, kun avhengig av antall partikler oppløst. Dette skjer fordi det oppløste stoffet "fortynner" løsningsmiddelet og øker løsningens entropi i flytende fase, noe som forskyver likevekten mellom faser.

**b)** 

 - Damptrykk: senkes når mengde løst stoff øker, løsningsmiddelets kjemiske potensial/aktivitet reduseres.
 - Kokepunkt: heves. Siden damptrykket er lavere ved en gitt temperatur, må temperaturen økes mer for å nå damptrykk lik atmosfæretrykket.
 - Frysepunkt: senkes. Det oppløste stoffet stabiliserer flytende fase relativt til fast fase, slik at lavere temperatur kreves for å nå likevekt mellom fast og flytende fase.


## Oppgave 5

**a)** Aktiviteten er definert som $a_A = p_A/p_A^*$, altså det faktiske damptrykket delt på damptrykket til den rene væsken. Det kjemiske potensialet uttrykkes da som $\mu_A = \mu_A^* + RT \ln a_A$. For en ideell løsning er $a_A = x_A$, siden Raoults lov da gjelder eksakt.

**b)** Aktiviteten nærmer seg molfraksjonen når konsentrasjonen av løsningsmiddelet er høy ($x_A \rightarrow 1$), altså i det fortynnede løsningsstoff-grensen der løsningsmiddelet oppfører seg tilnærmet ideelt, dette er Raoults-lov-grensen.

**c)** $γ_A$ i $a_A = γ_A x_A$ representerer avviket fra ideell oppførsel: 

$γ = 1$ betyr ideell oppførsel.

$γ > 1$ betyr positivt avvik: molekylene "flykter" lettere til damp enn ideelt forventet, ofte pga. svakere A–B enn A–A/B–B-interaksjoner. 

$γ < 1$ betyr negativt avvik: sterkere A–B-interaksjoner, f.eks. hydrogenbindinger, holder molekylene sterkere i løsning.

**d)** Molaritet/molalitet beskriver kun den nominelle konsentrasjonen, men i ikke-ideelle løsninger er det den "effektive" konsentrasjonen som styrer termodynamikken, ikke den nominelle, pga. interaksjoner mellom partiklene. Aktivitet korrigerer for dette avviket, slik at de termodynamiske likningene forblir eksakte selv i ikke-ideelle systemer.

**e)** Man kan ikke måle aktivitetskoeffisienten til en ionetype alene fordi man ikke kan ha en løsning med bare kationer eller bare anioner, elektronøytralitet krever begge. Derfor definerer man en gjennomsnittlig aktivitetskoeffisient $γ±$ som kombinerer bidragene fra begge ionetypene, og som er den eksperimentelt målbare/meningsfulle størrelsen.


## Oppgave 6

**a)** Fraksjonert destillering: En separasjonsteknikk der en væskeblanding gjentatte ganger fordampes og kondenseres via en kolonne med mange "trinn" for å utnytte at damp over en blanding er beriket på den mer flyktige komponenten, slik at komponentene gradvis separeres etter kokepunkt.

**b)** Azeotrop: Et sammensetningspunkt der væske- og dampfase har nøyaktig samme sammensetning, slik at videre destillering ikke kan separere komponentene ytterligere ved det punktet.

**c)** Critical solution temperature: Temperaturen der to delvis blandbare væsker akkurat blir fullstendig blandbare i alle proporsjoner. Over denne temperaturen forsvinner miscibility-gapet/fasesepareringen.

**d)** Faseseparering: Skjer når den frie blandingsenergien har to eller flere minima som funksjon av sammensetning i stedet for ett: systemet senker sin totale frie energi ved å dele seg i to faser med ulik sammensetning, hver ved sitt lokale minimum, i stedet for å forbli som én homogen blanding.

**e)** Osmotisk trykk oppstår fordi løsningsmiddelets kjemiske potensial er lavere på siden med oppløst stoff (pga. fortynningseffekten/entropien) enn på den rene løsningsmiddelsiden. For å gjenopprette likevekt gjennom en semipermeabel membran, må trykket økes på løsningssiden for å heve det kjemiske potensialet der tilbake til likevekt med den rene siden.

**a)** 

Tell antall blå (A) og røde (B) kuler i gitteret på bildet:

$$N_A = 8 \ (\text{blå}), \qquad N_B = 8 \ (\text{røde}), \qquad N = N_A + N_B = 16$$

Bruk formelen for antall mikrotilstander:

$W = \frac{N!}{N_A! \, N_B!} = \frac{16!}{8! \cdot 8!}$

$W = 12870$


**b)** 

$\Delta S_{sol} = −Nk_B[(N_A/N)\ln(N_A/N) + (N_B/N)\ln(N_B/N)] = −Nk_B[x \ln x + (1−x)\ln(1−x)],$

ved å definere $x = N_A/N$ og $(1−x) = N_B/N$.

**c)** 

$x_A = 7/10 = 0.70$, $x_B = 3/10 = 0.30$.

$\Delta S_{sol}/(Nk_B) = −[0.70\ln(0.70) + 0.30\ln(0.30)] = 0.6109$

Total blandingsentropi: $\Delta S_{sol} = 0.6109Nk_B = 10\times0.6109k_B \approx 6.11 k_B \approx 8.43\times10^{-23} \text{J/K}.$

**d)** 

Tell opp antall $\text{AA}$-, $\text{BB}$- og $\text{AB}$-kontakter (linjene i bildet) mellom nabokuler i gitteret:

$$m_{AA} = 10, \qquad m_{BB} = 1, \qquad m_{AB} = 13$$

(Kontroll: totalt antall kontakter i et $4\times4$-gitter er $2 \times 4 \times 3 = 24$, og $10+1+13 = 24$ ✓)

Sett inn i uttrykket for indre energi, med $w_{AA} = 1$, $w_{BB} = 1$, $w_{AB} = 2$:

$U = m_{AA}w_{AA} + m_{BB}w_{BB} + m_{AB}w_{AB} = 10 \times 1 + 1 \times 1 + 13 \times 2$

$U = 37$

**Kommentar:** Siden $w_{AB} > w_{AA}, w_{BB}$ og de fleste kontaktene er $\text{AB}$ (13 av 24), er dette systemet energetisk ugunstig blandet, det ville hatt lavere energi om flere like naboer lå sammen.


**e)** 

$\chi_{AB} = (z/k_BT)(w_{AB} − (w_{AA}+w_{BB})/2)$

Når $T\rightarrow\infty$: $\chi_AB \rightarrow 0$. Termisk energi dominerer fullstendig over interaksjonsforskjellene, og entropien vinner, systemet blandes uansett interaksjonsenergi.

Når $T\rightarrow 0$: $\chi_AB \rightarrow ± \infty$ avhengig av fortegn på $w_{AB} − (w_{AA}+w_{BB})/2$. Interaksjonsenergien dominerer fullstendig over entropien, og systemet vil enten fasesepareres sterkt hvis $\chi > 0$ eller danne en svært ordnet/foretrukket blanding hvis $\chi < 0$.

Dette stemmer med tolkningen av χ som forholdet mellom interaksjonsenergi og termisk energi ($k_BT$), $\chi$ vokser når termisk energi blir liten relativt til energiforskjellene.

**f)** Siden $\chi \approx 5$ for hydrokarbon/vann, er AB-kontakter (olje-vann) sterkt energetisk ugunstige sammenlignet med å holde like molekyler sammen, den frie energien har to separate minima, og systemet fasesepareres i stedet for å blande seg fullstendig. Ved svært lav konsentrasjon av olje er derimot entropileddet ($−x \ln x$) fortsatt stort nok til at en liten mengde olje likevel kan løses opp, dette er grunnen til at olje har en bitteliten løselighet i vann.

Finn total ionekonsentrasjon (Na⁺ og Cl⁻ til sammen):

$c_{\text{ioner}} = 0.50 + 0.50 = 1.00 \ \text{mol/dm}^{3}$

Finn mengde vann per liter løsning:

$n_{\text{H}_{2}\text{O}} = \frac{1000}{18.02} \approx 55.49 \ \text{mol/dm}^{3}$

Regn ut molfraksjonen av vann (resten er ioner):

$x_{\text{H}_{2}\text{O}} = \frac{n_{\text{H}_{2}\text{O}}}{n_{\text{H}_{2}\text{O}} + c_{\text{ioner}}} = \frac{55.49}{55.49 + 1.00} \approx 0.9823$

Bruk Raoults lov for å finne damptrykket over sjøvann:

Nedsettelsen blir da:

**Kommentar:** Nedsettelsen er liten fordi ionene utgjør en liten andel sammenlignet med den store mengden vann.


## Oppgave 16

Debye-Hückels grenselov gir aktivitetskoeffisienten ut fra ionestyrken $I$:

$\log \gamma_{\pm} = -A|z_{+}z_{-}|\sqrt{I}, \qquad I = \frac{1}{2}\sum_{i} c_{i}z_{i}^{2}$

**$\text{KCl}$ ($m = 5.0 \ \text{mmol/kg}$):** Begge ioner har $z = \pm 1$, så ionestyrken blir lik molaliteten:

$I = \frac{1}{2}(m \cdot 1^2 + m \cdot 1^2) = m = 5.0 \times 10^{-3} \ \text{mol/kg}$

Sett inn i grenseloven:

$\log \gamma_{\pm} = -0.509 \times 1 \times \sqrt{0.005} = -0.0360 \implies \gamma_{\pm} \approx 0.920$

**$\text{CaCl}_{2}$ ($m = 1.0 \ \text{mmol/kg}$):** Her er $\text{Ca}^{2+}$ ($z=2$) og to $\text{Cl}^{-}$ ($z=1$) per formelenhet, som gir høyere ionestyrke:

$I = \frac{1}{2}(m \cdot 2^{2} + 2m \cdot 1^{2}) = 3m = 3.0 \times 10^{-3} \ \text{mol/kg}$

Sett inn i grenseloven:

$\log \gamma_{\pm} = -0.509 \times 2 \times \sqrt{0.003} = -0.0558 \implies \gamma_{\pm} \approx 0.880$



## Oppgave 17

Molfraksjonene er gitt: $x_{T} = 0.330$, som gir $x_{B} = 0.670$.

Finn aktivitetene ved å dele faktisk partialtrykk på rent damptrykk:

$a_{B} = \frac{p_{B}}{p_{B}^{*}} = \frac{6.677}{9.657} \approx 0.691, \qquad a_{T} = \frac{p_{T}}{p_{T}^{*}} = \frac{1.214}{3.572} \approx 0.340$

$\gamma_{B} = \frac{a_{B}}{x_{B}} \approx \frac{0.691}{0.670} \approx 1.032, \qquad \gamma_{T} = \frac{a_{T}}{x_{T}} \approx \frac{0.340}{0.330} \approx 1.030$

**Kommentar:** Begge $\gamma$-verdier er nær $1$ med et lite positivt avvik — benzen/toluen er tilnærmet en ideell blanding.


## Oppgave 18

Molfraksjonen av vann finnes ved å trekke molfraksjonen til det løste stoffet fra 1:

## Oppgave 19

**a)** Finn molfraksjonen av oksygen løst i vann ved hjelp av Henrys lov:

$p_{\text{O}_{2}} = 0.21 \ \text{atm} \implies x_{\text{O}_{2}} = \frac{p_{\text{O}_{2}}}{K_{H}} = \frac{0.21}{40\,000} = 5.25 \times 10^{-6}$

Gjør om til konsentrasjon ved å gange med mengden vann per liter:

$n_{\text{O}_{2}} \approx x_{\text{O}_{2}} \times 55.5 \ \text{mol/L} \approx 2.91 \times 10^{-4} \ \text{mol/L}$

Gjør om til masse per liter:

$m_{\text{O}_{2}} \approx 9.3 \ \text{mg/L}$

**b)** Bruk van't Hoff-formen for å finne hvordan $K_H$ endrer seg med temperatur:

$\ln\left(\frac{K_{H,T_{2}}}{K_{H,T_{1}}}\right) = \frac{\Delta H_{\text{sol}}}{R}\left(\frac{1}{T_{2}} - \frac{1}{T_{1}}\right)$

Sett inn $T_{1} = 298.15 \ \text{K}$, $T_{2} = 299.05 \ \text{K}$ og $\Delta H_{\text{sol}} = -14.1 \ \text{kJ/mol}$:

$\ln\left(\frac{K_{2}}{K_{1}}\right) \approx 0.0171 \implies \frac{K_{2}}{K_{1}} \approx 1.017 \implies K_{H}(T_{2}) \approx 40\,700 \ \text{atm}$

**c)** 

Siden $\Delta H_{\text{sol}} < 0$, er oppløsningsprosessen **eksoterm**.

**d)** 

Løseligheten (molfraksjonen) er omvendt proporsjonal med $K_H$, så vi finner endringen ved å snu forholdet fra b):

$\frac{x_{2}}{x_{1}} = \frac{K_{1}}{K_{2}} \approx \frac{1}{1.017} \approx 0.983$

Dette gir en reduksjon på $\approx 1.7\%$ i løst $\text{O}_{2}$ ved $0.9\,°\text{C}$ oppvarming.

**e)** 

Gjentar samme beregning for $\Delta T = 2\,°\text{C}$, som gir en reduksjon på $\approx 3.7\%$. Redusert løselighet av $\text{O}_{2}$ i varmere hav øker risikoen for hypoksiske soner og legger et ekstra stressbidrag til marine økosystemer.
