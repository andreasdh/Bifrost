### Oppgave 1

**a)** Hva er forskjellen på en galvanisk og en elektrolytisk celle?

**Svar**

En galvanisk celle er en elektrokjemisk celle som produserer elektrisk arbeid. En elektrolytisk celle får tilført elektrisk arbeid og benytter det til å drive en ikke-spontan reaksjon. 

**b)**	Bruk Nernst likning til å forklare hvordan reaksjoner som ikke resulterer i en netto redoks reaksjon kan bli brukt til å generere strøm. 

**Svar**

Nernst likning: $E_{cell}=E_{cell}^0+\frac{RT}{νF}\cdot \ln{⁡Q}$

$E_{cell}$ er målt cellepotensial,$E_{cell}^0$ er cellepotensialet ved standard betingelser og ingen konsentrasjonsforskjell, R er gasskonstanten, T er temperatur, $ν$ er antall elektroner overført, F er Faraday konstanten og Q er reaksjonskvotienten, gitt av $Q=a_{halvcelle 1}/a_{halvcelle 2}$ . 

Leddet $\frac{RT}{νF}\cdot \ln{⁡Q}$ vil gi et bidrag til $E_{cell}$ så lenge det er en forskjell i aktivitet (f.eks. i ionekonsentrasjon) mellom de to halvcellene. Derfor kan reaksjoner som ikke er redoks reaksjoner gi strøm. Jo større aktivitetsforskjell (større konsentrasjonsforskjell) jo høyere strøm generes. Det er fordi det er en drivkraft for å jevne ut konsentrasjonsforskjellen mellom de to halvcellene. 

**c)** Med utgangspunkt i Nernst likning, hvordan kan man generere elektrisk arbeid

**Svar**

Nernst likning: $E_{cell}=E_{cell}^0-\frac{RT}{νF}\cdot \ln{⁡Q}$

$E_{cell}$ er målt cellepotensial, $E_{cell}^0$ er cellepotensialet ved standard betingelser og ingen konsentrasjonsforskjell, R er gasskonstanten, T er temperatur, $ν$ er antall elektroner overført, F er Faraday konstanten og Q er reaksjonskvotienten, gitt av $Q=a_{halvcelle 1}/a_{halvcelle 2}$ . 

Elektrisk arbeid kan dermed enten generes ved en konsentrasjonsforskjell, to halvceller med et redoks par, eller en kombinasjon av disse. 

**d)** Ved konstant trykk og temperatur, hvorfor er det maksimale elektriske arbeidet som kan gjøres lik Gibbs frie energi?

**Svar**

Gibbs frie energi gir det maksimale ikke-ekspansjonsarbeidet en kan få ut ved konstant trykk og temperatur, så det elektriske arbeidet er lik Gibbs fri energi $(w_{e,max}= \Delta_r G= -nFE_{cell})$.

**e)** Hvorfor er det en fordel ut utrykke Q i form av aktiviteter? Under hvilke betingelser kan konsentrasjoner brukes? 

**Svar**
Det er en fordel å bruke aktivitet da reelle bidrag (ikke-ideelle) blir tatt med i denne parameteren, og aktiviteten gir den «effektive konsentrasjonen». Ved lave mengde løst stoff, vil konsentrasjon være en god tilnærming fordi $a_i=γ_i  \frac{c}{c^0}$  så bidraget vil være veldig lite.


**f)** Forklar en metode for å bestemme standard celle potensial for et redoks par. 

**Svar — eksempel: galvanisk Zn/Cu-celle (uten referanseelektrode)**

Sett opp cellen $\mathrm{Zn}(s)|\mathrm{Zn}^{2+}(a_{\mathrm{Zn}^{2+}})||\mathrm{Cu}^{2+}(a_{\mathrm{Cu}^{2+}})|\mathrm{Cu}(s)$. Sink oksideres ved anoden, og kobber(II) reduseres ved katoden:

$$\mathrm{Zn}(s)\rightarrow\mathrm{Zn}^{2+}(aq)+2e^-,\qquad \mathrm{Cu}^{2+}(aq)+2e^-\rightarrow\mathrm{Cu}(s).$$

Nettoreaksjonen er $\mathrm{Zn}(s)+\mathrm{Cu}^{2+}(aq)\rightarrow\mathrm{Zn}^{2+}(aq)+\mathrm{Cu}(s)$, så $n=2$ og, siden faste stoffer ikke inngår i $Q$,

$$Q=\frac{a_{\mathrm{Zn}^{2+}}}{a_{\mathrm{Cu}^{2+}}}.$$

Nernst-ligningen ved $25^\circ\mathrm{C}$ blir

$$E_{\mathrm{cell}}=E^\circ_{\mathrm{cell}}-\frac{RT}{2F}\ln Q=E^\circ_{\mathrm{cell}}-(0{.}012846\,\mathrm{V})\ln Q.$$

I praksis lager man løsninger med kjente konsentrasjoner, men Nernst-ligningen bruker aktiviteter. For hvert ion er $a_i=\gamma_i c_i/c^\circ$, der $\gamma_i$ er aktivitetskoeffisienten og $c^\circ=1\,\mathrm{mol\,dm^{-3}}$. Derfor er

$$Q=\frac{a_{\mathrm{Zn}^{2+}}}{a_{\mathrm{Cu}^{2+}}}=\frac{\gamma_{\mathrm{Zn}^{2+}}}{\gamma_{\mathrm{Cu}^{2+}}}\frac{c_{\mathrm{Zn}^{2+}}}{c_{\mathrm{Cu}^{2+}}}.$$

Et konkret støtte-elektrolyttvalg er $0{.}10\,\mathrm{M}$ $KNO_3$ i begge halvcellene. Hold ionestyrken konstant; hvis konsentrasjonene av metallnitratene varierer mye, må støtte-elektrolyttmengden justeres tilsvarende. Dette gjør aktivitetskoeffisientene mer stabile, men setter dem ikke automatisk lik 1. I tilstrekkelig fortynnede løsninger kan man ofte bruke $a_i\approx c_i/c^\circ$.

Hvis man plotter mot konsentrasjonsforholdet uten å korrigere for aktivitetskoeffisientene, blir den tilsynelatende skjæringen $E^\circ_{\mathrm{cell}}-(RT/2F)\ln(\gamma_{\mathrm{Zn}^{2+}}/\gamma_{\mathrm{Cu}^{2+}})$ når dette koeffisientforholdet er konstant. For å finne det termodynamiske standard cellepotensialet må man derfor bruke aktiviteter eller korrigere for aktivitetskoeffisientene. Mål $E_{\mathrm{cell}}$ ved null strøm for flere kjente $Q$-verdier, og plott $E_{\mathrm{cell}}$ mot $\ln Q$: helningen skal være $-RT/(2F)$, og y-skjæringen ved $Q=1$ er $E^\circ_{\mathrm{cell}}$. Dette bestemmer standard cellepotensialet for Zn/Cu-cellen direkte, uten å bruke en standardelektrode. Metoden gir potensialforskjellen for hele cellen; enkeltstående standard reduksjonspotensialer må fortsatt defineres relativt til en referanse som SHE.

**g)** Hvorfor er det nødvendig å måle cellepotensiale under null strøms betingelser? 

**Svar**

Maksimalt arbeid er produsert når en reaksjon skjer reversibelt. Vi må derfor påse at reaksjonen/elektrokjemiske cellen skjer under reversible forhold. Vi vet at ΔG er relatert til Q, så vi må derfor påse at reaksjonen operer på en reversibel måte med konstant sammensetning. Begge disse forholdene er oppnådd når cellepotensialet er balansert med nøyaktig like store motsatt rettede potensial, slik at cellen driver reversibelt, sammensetningen er konstant, og ingen strøm går. Som er resultat er cellereaksjonen klar for endring, men endrer seg faktisk ikke. Den resulterende potensialdifferansen er det vi kaller cellepotensial $E_{cell}$ av cellen.

**h)** Forklar hvorfor det er et behov for en saltbru i galvaniske celler. 

**Svar**

Saltbruen opprettholder elektrisk nøytralitet mellom halvcellene, som gjør at cellen ikke kjører for raskt til likevekt.

### Oppgave 3

Vi tenker oss en generell galvanisk celle med de to halvreaksjonene:  

Anode: $A (s) \rightarrow A^{\nu+}(aq) + \nu {e^-} $

Katode $ B^{\nu+}(aq) + \nu {e^-} \rightarrow B(s)  $

Vi kan skrive opp endringen i raksjonens Gibbs frie energi ved "extent of reaction", $\xi$, på følgende måte: $\Delta_r G = (\frac{\partial G}{\partial \xi})_{p,T}$

Vis at dersom vi har en reaksjon $\nu$ elekroner overføres, vil endringen i fri energi være gitt ved $\Delta_r G= -\nu F E_{cell}$

**Svar**

Starter med $\Delta_r G = (\frac{\partial G}{\partial \xi})_{p,T}$ og skriver om for å få de to variablene på hver sin side:

$$dG = \Delta_r G d\xi$$

Vi vet at $dG = dw_{rev}$, altså det maksimale ikke-ekspansjonsarbeidet (i vårt tilfelle det elektriske arbeidet). Og, det elektriske arbeidet er gitt ved 

$$dw_{el} = -dq\cdot E_{cell}$$

der $q$ er den totale ladningen, gitt ved antallet ladninger overført, i dette tilfellet elektroner: $dq = - \nu e N_A d\xi$.

$$dG = dw_{el} =- \nu e N_A E_{cell} d\xi  $$

$$\Delta_r G =(\frac{\partial G}{\partial \xi})_{p,T} = \frac{\partial w_{el}}{\partial \xi} = - \nu e N_A E_{cell} $$

Setter inn for Faradaykonstanten ($F = eN_A$)

$$\Delta_r G = -\nu F E_{cell}$$


### Oppgave 4 

Cellepotensialet for cellen Ag(s)|AgI(s)|AgI(aq)|Ag(s) er $+0.9509V$ ved $25^\circ C$. Finn løselighetsproduktet, $K_s = K^-1$, for AgI(s)

**Svar**

Skriver opp høyre og venstre (R og L) side hver for seg:

R: $Ag^+ (aq)+e^-→Ag(s)$
L: $AgI (s)+e^-→Ag (s)+I^- (aq)$

Totalt (R-L): $Ag^+ (aq)+I^- (aq)→AgI (s)$, og $\nu = 1$

Finner først likevektskonstanten K, og deretter løselighetsproduktet:

$$\ln {K} = \frac{\nu E_{cell} F}{RT} = \frac{1\cdot 0.9509 V}{25.693 \cdot 10^-3 V}$$

Hvor $\frac{RT}{F} = 25.693 \cdot 10^-3 V$

$$K = e^{\frac{1\cdot 0.9509 V}{25.693 \cdot 10^-3 V}} = 1.184\cdot 10^{16}$$

$$K_s = K^{-1} = 8.45\cdot 10^{-17}$$

Løseligheten skrives som motsatt at likevektskonstanten:

$$K = \frac{[AgI(s)]}{[Ag^+(aq)][I^-(aq)]} \rightarrow K_s = \frac{[Ag^+(aq)][I^-(aq)]}{[AgI(s)]}$$



### Oppgave 4 revidert

For cellen $\mathrm{Ag}(s)|\mathrm{AgI}(s)|\mathrm{I}^-(aq)||\mathrm{Ag}^+(aq)|\mathrm{Ag}(s)$ er standard cellepotensialet $E^\circ_{\mathrm{cell}}=+0{.}9509\,\mathrm{V}$ ved $25^\circ\mathrm{C}$. Finn løselighetsproduktet $K_{s}$ for $\mathrm{AgI}(s)$ ved denne temperaturen.

**Svar**

Venstre elektrode er anoden, og høyre elektrode er katoden. Halvreaksjonene i reaksjonsretningen er

$$\begin{aligned}\text{Anode (oksidasjon): }&\mathrm{Ag}(s)+\mathrm{I}^-(aq)\rightarrow\mathrm{AgI}(s)+e^- \\ \text{Katode (reduksjon): }&\mathrm{Ag}^+(aq)+e^-\rightarrow\mathrm{Ag}(s).\end{aligned}$$

Når halvreaksjonene summeres, kanselleres elektronet og sølvmetallet:

$$\mathrm{Ag}^+(aq)+\mathrm{I}^-(aq)\rightarrow\mathrm{AgI}(s),\qquad \nu=1.$$

For denne utfellingsreaksjonen er $K_{\mathrm{utfelling}}=1/K_s

Koblingen mellom standard cellepotensial og standard reaksjons-Gibbsenergi er $\Delta_rG^\circ=-\nu FE^\circ_{\mathrm{cell}}$, mens $\Delta_rG^\circ=-RT\ln K_{\mathrm{utfelling}}$. Setter vi disse uttrykkene lik hverandre, får vi

$$\ln K_{\mathrm{utfelling}}=\frac{\nu FE^\circ_{\mathrm{cell}}}{RT}.$$

Ved $T=25^\circ\mathrm{C}=298{.}15\,\mathrm{K}$ er $RT/F=0{.}0256926\,\mathrm{V}$. Setter inn $\nu=1$ og det oppgitte standard cellepotensialet:

$$\ln K_{\mathrm{utfelling}}=\frac{(\nu)(0{.}9509\,\mathrm{V})}{0{.}0256926\,\mathrm{V}}=\frac{(1)(0{.}9509\,\mathrm{V})}{0{.}0256926\,\mathrm{V}}=37{.}01,$$

$$K_{\mathrm{utfelling}}=e^{37{.}01}=1{.}18\times10^{16}.$$

Derfor er løselighetsproduktet den inverse konstanten:

$$\boxed{K_{s}=\frac{1}{K_{\mathrm{utfelling}}}=e^{-37{.}01}=8{.}45\times10^{-17}.}$$
