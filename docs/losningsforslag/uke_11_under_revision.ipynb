# Uke 11 – Elektrokjemi

## Diskusjonsoppgaver

### Oppgave 1

**a)** Hva er forskjellen på en galvanisk og en elektrolytisk celle?

**Svar**

En galvanisk celle er en elektrokjemisk celle som produserer elektrisk arbeid. En elektrolytisk celle får tilført elektrisk arbeid og benytter det til å drive en ikke-spontan reaksjon. 

**b)**	Bruk Nernst likning til å forklare hvordan reaksjoner som ikke er redoks reaksjoner kan bli brukt til å generere strøm. 

**Svar**

Nernst likning: $E_{cell}=E_{cell}^0+\frac{RT}{νF}\cdot \ln{⁡Q}$

$E_{cell}$ er målt cellepotensial,$E_{cell}^0$ er cellepotensialet ved standard betingelser og ingen konsentrasjonsforskjell, R er gasskonstanten, T er temperatur, $ν$ er antall elektroner overført, F er Faraday konstanten og Q er reaksjonskvotienten, gitt av $Q=a_{halvcelle 1}/a_{halvcelle 2}$ . 

Leddet $\frac{RT}{νF}\cdot \ln{⁡Q}$ vil gi et bidrag til $E_{cell}$ så lenge det er en forskjell i aktivitet (f.eks. i ionekonsentrasjon) mellom de to halvcellene. Derfor kan reaksjoner som ikke er redoks reaksjoner gi strøm. Jo større aktivitetsforskjell (større konsentrasjonsforskjell) jo høyere strøm generes. Det er fordi det er en drivkraft for å jevne ut konsentrasjonsforskjellen mellom de to halvcellene. 

**c)** Med utgangspunkt i Nernst likning, hvordan kan man generere elektrisk arbeid

**Svar**

Nernst likning: $E_{cell}=E_{cell}^0-\frac{RT}{νF}\cdot \ln{⁡Q}$

$E_{cell}$ er målt cellepotensial, $E_{cell}^0$ er cellepotensialet ved standard betingelser og ingen konsentrasjonsforskjell, R er gasskonstanten, T er temperatur, $ν$ er antall elektroner overført, F er Faraday konstanten og Q er reaksjonskvotienten, gitt av $Q=a_{halvcelle 1}/a_{halvcelle 2}$ . 

Elektrisk arbeid kan dermed enten generes ved en konsentrasjonsforskjell, to halvceller med et redoks par, eller en kombinasjon av disse. 

**d)** Ved konstant trykk og temperatur, hvorfor er det maksimale elektriske arbeidet som kan gjøres lik Gibbs frie energi?

**Svar**

Gibbs frie energi gir det maksimale ikke-ekspansjonsarbeidet en kan få ut ved konstant trykk og temperatur, så det elektriske arbeidet er lik Gibbs fri energi $(w_{e,max}= \Delta_r G= -nFE_{cell})$.

**e)** Hvorfor er det en fordel ut utrykke Q i form av aktiviteter? Under hvilke betingelser kan konsentrasjoner brukes? 

**Svar**
Det er en fordel å bruke aktivitet da reelle bidrag (ikke-ideelle) blir tatt med i denne parameteren, og aktiviteten gir den «effektive konsentrasjonen». Ved lave mengde løst stoff, vil konsentrasjon være en god tilnærming fordi $a_i=γ_i  \frac{c}{c^0}$  så bidraget vil være veldig lite.


**f)** Forklar en metode for å bestemme standard celle potensial for et redoks par. 

**Svar**

Ved å sette opp et eksperiment hvor man måler cellepotensialet for flere forskjellige konsentrasjoner av den ene halvcellen i redoks-paret vil man kunne finne $E_{cell}^0$ ved å plotte $E_{cell}$ mot $\ln Q$. $E_{cell}^0$ vil da være krysningspunktet med y-aksen.

**g)** Hvorfor er det nødvendig å måle cellepotensiale under null strøms betingelser? 

**Svar**

Maksimalt arbeid er produsert når en reaksjon skjer reversibelt. Vi må derfor påse at reaksjonen/elektrokjemiske cellen skjer under reversible forhold. Vi vet at ΔG er relatert til Q, så vi må derfor påse at reaksjonen operer på en reversibel måte med konstant sammensetning. Begge disse forholdene er oppnådd når cellepotensialet er balansert med nøyaktig like store motsatt rettede potensial, slik at cellen driver reversibelt, sammensetningen er konstant, og ingen strøm går. Som er resultat er cellereaksjonen klar for endring, men endrer seg faktisk ikke. Den resulterende potensialdifferansen er det vi kaller cellepotensial $E_{cell}$ av cellen.

**h)** Forklar hvorfor det er et behov for en saltbru i galvaniske celler. 

**Svar**

Saltbruen opprettholder elektrisk nøytralitet mellom halvcellene, som gjør at cellen ikke kjører for raskt til likevekt.

### Oppgave 3

Vi kan skrive opp endringen i Gibbs frie energi ved "extent of reaction", $\xi$, på følgende måte: $\Delta_r G = (\frac{\partial G}{\partial \xi})_{p,T}$

Vis at dersom vi har en reaksjon hvor ladningen endres med $\nu$, vil endringen i fri energi være gitt ved $\Delta_r G= -\nu F E_{cell}$

**Svar**

Starter med $\Delta_r G = (\frac{\partial G}{\partial \xi})_{p,T}$ og skriver om for å få de to variablene på hver sin side:

$$dG = \Delta_r G d\xi$$

Vi vet at $dG = dw_{rev}$, altså det maksimale ikke-ekspansjonsarbeidet (i vårt tilfelle det elektriske arbeidet). Og, det elektriske arbeidet er gitt ved 

$$dw_{el} = -dq\cdot E_{cell}$$

der $q$ er den totale ladningen, gitt ved antallet ladninger overført, i dette tilfellet elektroner: $dq = - \nu e N_A d\xi$.

$$dG = dw_{el} =- \nu e N_A E_{cell} d\xi  $$

$$\Delta_r G =(\frac{\partial G}{\partial \xi})_{p,T} = \frac{\partial w_{el}}{\partial \xi} = - \nu e N_A E_{cell} $$

Setter inn for Faradaykonstanten ($F = eN_A$)

$$\Delta_r G = -\nu F E_{cell}$$


### Oppgave 4 

Cellepotensialet for cellen Ag(s)|AgI(s)|AgI(aq)|Ag(s) er $+0.9509V$ ved $25^\circ C$. Finn løselighetsproduktet, $K_s = K^-1$, for AgI(s)

**Svar**

Skriver opp høyre og venstre (R og L) side hver for seg:

R: $Ag^+ (aq)+e^-→Ag(s)$
L: $AgI (s)+e^-→Ag (s)+I^- (aq)$

Totalt (R-L): $Ag^+ (aq)+I^- (aq)→AgI (s)$, og $\nu = 1$

Finner først likevektskonstanten K, og deretter løselighetsproduktet:

$$\ln {K} = \frac{\nu E_{cell} F}{RT} = \frac{1\cdot 0.9509 V}{25.693 \cdot 10^-3 V}$$

Hvor $\frac{RT}{F} = 25.693 \cdot 10^-3 V$

$$K = e^{\frac{1\cdot 0.9509 V}{25.693 \cdot 10^-3 V}} = 1.184\cdot 10^{16}$$

$$K_s = K^{-1} = 8.45\cdot 10^{-17}$$

Løseligheten skrives som motsatt at likevektskonstanten:

$$K = \frac{[AgI(s)]}{[Ag^+(aq)][I^-(aq)]} \rightarrow K_s = \frac{[Ag^+(aq)][I^-(aq)]}{[AgI(s)]}$$


### Oppgave 5

Vi måler en verdi på $E_{cell} = 1.536 V$ i følgende elektrokjemiske celle:

$Zn(s)|Zn^{2+}(2.00M)||Ag^+(xM)|Ag(s)$

**a)** Lag en skisse av cellen

**b)** Hva er standard cellepotensial, $E_{cell}^0$?

**Svar**

Oksidasjon (R): $Zn(s) \rightarrow Zn^{2+}(aq)+2e^−$, $E^0 = -0.763V$
Reduksjon (L): $Ag^+(aq) + e^- \rightarrow Ag(s)$, $E^0 = 0.800V$

Nettoreaksjon: $Zn(s) + 2Ag^+(aq) \rightarrow Zn^{2+}(aq)+2Ag (s)$

$$E_{cell}^0 = -0.763V + 0.800V = 1.563V$$

obs! siden celle-potensiale er en intensiv-egenskap trenger vi ikke balansere antallet elektroner.

**c)** Beregn konsentrasjonen av $Ag^+$ ioner i cellen. 

**Svar**

Bruker Nernst likning for å finne konsentrasjonen av $Ag^+$ioner:

$$E_{cell}=E_{cell}^0-\frac{RT}{νF}\cdot \ln{⁡Q}$$

$$\ln{Q} =\frac{(E_{cell}^0-E_{cell})\nu F}{RT} = \frac{(1.563V - 1.536V)\cdot 2}{25.693 \cdot 10^{-3} V} = 2.1017 $$

Reaksjonskvotienten er gitt ved $Q = \frac{[Zn^{2+}(aq)]}{[Ag^+(aq)]^2} = \frac{2}{x^2}$

$$e^{\ln Q} = \frac{2}{x^2 } =e^{2.1017}=8.180$$


$$x= \sqrt{\frac{2}{8.180}} = 0.495 $$

$$[Ag^+(aq)] = 0.495 M $$

### Oppgave 3

I en elektrolyseprosess blir en kjemisk reaksjon indusert gjennom en elektrisk strøm. Gitt at $3.15 A$ strøm blir lagt på i 78 min, beregne antall gram metall som blir avsatt på katoden for løsninger med følgende kationer:


**a)** $Zn^{2+}$

**b)** $Sn^{2+}$

**c)** $Fe^{3+}$

**d)** $Ni^{2+}$


**Svar**

Antall mol elektroner er gitt ved $n_e= \frac{I \cdot t}{F}$
Hvor

I = elektrisk strøm (A) 
t = tid
F = faradays konstant (den elektriske ladningen til 1 mol med elektroner) 
$n_e=(I⋅t)/F$, sjekker at enheter gir mening: $\frac{[A⋅s]}{[C/mol]}=\frac{[A⋅s]}{[A⋅s]} [mol]$

Finner massen ved $m= \frac{M\cdot n_e}{\nu}$ hvor $\nu$ er oksidasjonstallet.

Antallet elektroner overført totalt er da: 

$$n_e= \frac{I \cdot t}{F} = \frac{3.15 A \cdot 78\cdot 60 s}{96485 C} = 0.153 mol$$

Dette gjelder i alle deloppgavene. Nå mangler vi bare å finne avsatt masse:

**a)** $m= \frac{(0.153 mol \cdot 65.41 g/mol)}{2}= 5.00 g$

**b)** $m= \frac{(0.153 mol \cdot 118.7 g/mol)}{2}= 9.07 g$

**c)** $m= \frac{(0.153 mol \cdot 55.85 g/mol)}{3}= 2.84 g$

**d)** $m= \frac{(0.153 mol \cdot 58.69 g/mol)}{2}= 4.49 g$
