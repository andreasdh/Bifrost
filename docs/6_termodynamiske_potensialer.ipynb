# Termodynamiske potensialer

:::{admonition} Læringsmål
:class: note

Etter å ha arbeidet med dette kapitlet skal du kunne:

- forklare hvorfor entalpi alene ikke kan avgjøre om en prosess er spontan
- forklare hvordan Gibbs-energien kobler termodynamikkens andre lov til et kriterium for spontanitet ved konstant temperatur og trykk
- regne ut endringer i Gibbs-energi for en prosess og bruke dette til å avgjøre om prosessen er spontan
- forklare hva et termodynamisk potensial er og hvorfor ulike potensialer har ulike naturlige variabler
- bruke fundamentallikningen $dU=T\,dS-P\,dV$ til å tolke temperatur og trykk som termodynamiske størrelser
- utlede differensialene til $H$, $F$ og $G$ fra fundamentallikningen
- forklare hva Maxwell-relasjonene uttrykker og hvorfor de er nyttige
- beskrive kjemisk potensial som drivkraft for materieoverføring og endring i sammensetning
- forklare koblingen mellom Helmholtz-energi, partisjonsfunksjonen og kjemisk potensial
:::

I forrige kapittel lærte vi å holde regnskap over energi som overføres som varme og arbeid. Ved konstant trykk fikk entalpien en særlig praktisk rolle fordi

$$\Delta H=q_P$$

Men dette svarer bare på hvor mye varme som utveksles. En kjemisk prosess kan være endoterm og likevel spontan, eller eksoterm uten å skje spontant under alle betingelser.

For å forstå **drivkraft og likevekt** må vi koble energiregnskapet til termodynamikkens andre lov. Det leder oss til de termodynamiske potensialene.


## Fra andre lov til Gibbs-energi

En spontan prosess må oppfylle termodynamikkens andre lov:

$$\Delta S_\mathrm{univers}=\Delta S_\mathrm{system}+\Delta S_\mathrm{omgivelser}>0$$

Ved konstant temperatur og trykk kan entropiendringen i omgivelsene knyttes til varmen systemet avgir eller tar opp. Fra forrige kapittel vet vi at $q_P=\Delta H$, og dermed

$$\Delta S_\mathrm{omgivelser}=-\frac{\Delta H}{T}$$

Kriteriet for spontanitet blir derfor

$$\Delta S_\mathrm{univers}=\Delta S_\mathrm{system}-\frac{\Delta H}{T}>0$$

Multipliserer vi med $-T$, får vi

$$\Delta H-T\Delta S_\mathrm{system}<0$$

Kombinasjonen på venstre side er så nyttig at vi definerer en egen tilstandsfunksjon:

$$G=H-TS$$

Dette er **Gibbs-energien**. Ved konstant temperatur og trykk gjelder

$$\Delta G=-T\Delta S_\mathrm{univers}$$

Dermed får vi et kriterium som bare krever at vi følger systemet:

$$\Delta G<0$$

for en spontan prosess i den angitte retningen

$$\Delta G=0$$

ved likevekt

Gibbs-energien «pakker» altså termodynamikkens andre lov inn i en størrelse som er definert for systemet.

Samtidig har $G$ en arbeidstolkning. Ved konstant temperatur og trykk er reduksjonen i Gibbs-energi lik det maksimale **ikke-ekspansjonsarbeidet** som kan hentes ut i en reversibel prosess. Elektrisk arbeid fra en elektrokjemisk celle er et viktig eksempel.

Dette er bakgrunnen for ordet **potensial**: størrelsen forteller noe om systemets mulighet/potensial til å drive en prosess under bestemte betingelser.


## Hvorfor trenger vi flere termodynamiske potensialer?

Gibbs-energien passer naturlig når temperatur og trykk er de uavhengige variablene. Men dette er ikke de eneste betingelsene vi kan arbeide under.

Vi har allerede møtt fire sentrale tilstandsfunksjoner:

$$U$$

$$H=U+PV$$

$$F=U-TS$$

$$G=H-TS$$

Disse funksjonene representerer ikke fire forskjellige «energityper». De organiserer den samme termodynamiske informasjonen på ulike måter.

For å forstå hvorfor akkurat disse kombinasjonene er nyttige, må vi se nærmere på hvordan en tilstandsfunksjon endres når de uavhengige variablene endres. Vi starter med den mest grunnleggende energifunksjonen, den indre energien $U$.

## Fundamentallikningen – broen mellom mikro og makro

Vi har nå definert flere termodynamiske potensialer. De beskriver det samme termodynamiske systemet, men bruker ulike størrelser som utgangspunkt for å beskrive endringer i systemet. Disse størrelsene kalles potensialets **naturlige variabler**. Så hvordan velger vi hvilke potensialer som er mest nyttige i en gitt situasjon?

For å svare på dette skal vi utlede en likning som beskriver hvordan den indre energien endres når entropien og volumet endres. Denne likningen kalles **fundamentallikningen for indre energi**, og den knytter sammen flere av størrelsene vi har møtt tidligere: energi, varme, arbeid, entropi, temperatur, trykk og volum.

Vi starter med en reversibel prosess. Da kan varmeutvekslingen skrives som (husk den makroskopiske definisjonen av entropi):

$$\delta q_\mathrm{rev}=T\,dS$$

og reversibelt trykk–volum-arbeid som

$$\delta w_\mathrm{rev}=-P\,dV$$

Setter vi dette inn i første lov,

$$dU=\delta q+\delta w$$

får vi

$$dU=T\,dS-P\,dV$$

:::{admonition} Fundamentallikningen for indre energi
:class: important

For et lukket system med konstant stoffmengde og bare trykk–volum-arbeid er

$$dU=T\,dS-P\,dV$$

Likningen viser at den indre energien naturlig kan betraktes som en funksjon av entropi og volum:

$$U=U(S,V)$$

:::

Vi brukte en reversibel prosess for å finne denne sammenhengen, fordi varme og arbeid da kan uttrykkes direkte ved $dS$ og $dV$. Men $U$, $S$ og $V$ er tilstandsfunksjoner. Fundamentallikningen beskriver derfor sammenhengen mellom systemets likevektstilstander og er ikke begrenset til prosesser som faktisk foregår reversibelt.

### Hva forteller fundamentallikningen?

Likningen

$$dU=T\,dS-P\,dV$$

forteller hvordan den indre energien reagerer på små endringer i entropi og volum.

Hvis volumet holdes konstant, er $dV=0$, og vi får

$$dU=T\,dS$$

Temperaturen forteller dermed hvor mye den indre energien endres når entropien endres litt ved konstant volum. Matematisk kan vi skrive dette som følgende uttrykk, der vi bruker den partiellderiverte fordi _U_ er en funksjon av både _S_ og _V_, og her deriverer vi med hensyn på _S_ mens vi holder _V_ konstant:

$$T=\left(\frac{\partial U}{\partial S}\right)_{V,N}$$

Hvis entropien i stedet holdes konstant, er $dS=0$, og vi får

$$dU=-P\,dV$$

Trykket forteller dermed hvor mye den indre energien endres når volumet endres litt ved konstant entropi:

$$-P=\left(\frac{\partial U}{\partial V}\right)_{S,N}$$

Minustegnet har en fysisk betydning. Når et system utvider seg, er $dV>0$. Systemet gjør da arbeid på omgivelsene, og den indre energien reduseres dersom entropien holdes konstant.

Fundamentallikningen viser altså at temperatur og trykk ikke bare er størrelser vi måler utenfra. De beskriver hvordan systemets energi responderer på endringer i entropi og volum.

### Temperaturen knytter mikro og makro sammen

I {doc}`kapitlet om entropi <2_entropi_boltzmann>` møtte vi temperaturen fra et statistisk utgangspunkt. Der fant vi

$$\frac{1}{T}=\left(\frac{\partial S}{\partial U}\right)_{V,N}$$

Denne likningen sier hvor mye entropien endres når systemets energi endres litt ved konstant volum.

Fra fundamentallikningen har vi nå funnet

$$T=\left(\frac{\partial U}{\partial S}\right)_{V,N}$$

Her stiller vi det motsatte spørsmålet: Hvor mye endres energien når entropien endres litt?

De to beskrivelsene er inverse av hverandre. I statistisk termodynamikk startet vi med entropien og undersøkte hvordan den avhenger av energi. I makroskopisk termodynamikk kan vi starte med energien og undersøke hvordan den avhenger av entropi.

Her møtes derfor den mikroskopiske og den makroskopiske beskrivelsen direkte: Temperaturen som vi måler makroskopisk, er den samme størrelsen som dukket opp da vi beskrev hvordan antallet tilgjengelige mikrotilstander endrer seg med energi.

### Trykket som termodynamisk størrelse

Vi har tidligere forklart trykk mekanisk som et resultat av molekylenes kollisjoner med veggene. Fundamentallikningen gir en annen, men ekvivalent beskrivelse:

$$-P=\left(\frac{\partial U}{\partial V}\right)_{S,N}$$

Trykket beskriver altså også hvordan energien reagerer på en liten volumendring når entropien holdes konstant.

Slik får både temperatur og trykk en termodynamisk betydning: De er størrelser som beskriver hvordan energien endres når vi beveger oss mellom nærliggende likevektstilstander.

:::{dropdown} Entropiformen av fundamentallikningen

Fundamentallikningen kan også løses med hensyn på $dS$:

$$dS=\frac{1}{T}\,dU+\frac{P}{T}\,dV$$

Da kan vi lese av

$$\frac{1}{T}=\left(\frac{\partial S}{\partial U}\right)_{V,N}$$

og

$$\frac{P}{T}=\left(\frac{\partial S}{\partial V}\right)_{U,N}$$

Den første av disse er nettopp temperaturdefinisjonen vi møtte i statistisk termodynamikk.

:::


## Fra indre energi til de andre potensialene

Vi har sett at fundamentallikningen for den indre energien er

$$dU=T\,dS-P\,dV$$

Denne likningen viser at $S$ og $V$ er de naturlige variablene til $U$. Nå kan vi se hva som skjer når vi omformer $U$ til de andre potensialene. Poenget er ikke å introdusere nye former for energi, men å bytte hvilke variabler som opptrer naturlig i differensialet.

### Entalpi

Entalpien er definert som

$$H=U+PV$$

Differensierer vi uttrykket, får vi

$$dH=dU+P\,dV+V\,dP$$

Setter vi inn fundamentallikningen

$$dU=T\,dS-P\,dV$$

får vi

$$dH=T\,dS-P\,dV+P\,dV+V\,dP$$

De to leddene med $dV$ kansellerer, slik at

$$dH=T\,dS+V\,dP$$

Nå ser vi direkte at entalpien naturlig beskrives ved entropi og trykk:

$$H=H(S,P)$$

Dette betyr at

$$T=\left(\frac{\partial H}{\partial S}\right)_P$$

og

$$V=\left(\frac{\partial H}{\partial P}\right)_S$$

Sammenliknet med $U$ har vi altså byttet ut volum $V$ med trykk $P$ som naturlig variabel. Dette er den matematiske grunnen til at entalpi passer naturlig i beskrivelser der trykket inngår som uavhengig variabel.

### Helmholtz- og Gibbs-energi

På samme måte kan vi finne de naturlige variablene til Helmholtz-energien og Gibbs-energien.

:::{admonition} Underveisoppgave
:class: tip

Bruk

$$F=U-TS$$

og

$$G=H-TS$$

til å utlede uttrykkene for $dF$ og $dG$. Hvilke naturlige variabler får de to potensialene?

:::

:::{dropdown} Løsningsforslag

For Helmholtz-energien starter vi med

$$F=U-TS$$

Differensiering gir

$$dF=dU-T\,dS-S\,dT$$

Setter vi inn

$$dU=T\,dS-P\,dV$$

får vi

$$dF=T\,dS-P\,dV-T\,dS-S\,dT$$

Dermed

$$dF=-S\,dT-P\,dV$$

Vi ser derfor at Helmholtz-energien har temperatur og volum som naturlige variabler:

$$F=F(T,V)$$

og

$$S=-\left(\frac{\partial F}{\partial T}\right)_V$$

$$P=-\left(\frac{\partial F}{\partial V}\right)_T$$

For Gibbs-energien starter vi med

$$G=H-TS$$

Differensiering gir

$$dG=dH-T\,dS-S\,dT$$

Setter vi inn

$$dH=T\,dS+V\,dP$$

får vi

$$dG=T\,dS+V\,dP-T\,dS-S\,dT$$

Dermed

$$dG=-S\,dT+V\,dP$$

Vi ser derfor at Gibbs-energien har temperatur og trykk som naturlige variabler:

$$G=G(T,P)$$

og

$$S=-\left(\frac{\partial G}{\partial T}\right)_P$$

$$V=\left(\frac{\partial G}{\partial P}\right)_T$$

:::

:::{admonition} Fundamentallikningene
:class: important

De fire termodynamiske potensialene kan oppsummeres som

$$dU=T\,dS-P\,dV$$

$$dH=T\,dS+V\,dP$$

$$dF=-S\,dT-P\,dV$$

$$dG=-S\,dT+V\,dP$$

Fra disse likningene kan vi lese av de naturlige variablene direkte:

| Potensial | Naturlige variabler |
|---|---|
| $U$ | $S,V$ |
| $H$ | $S,P$ |
| $F$ | $T,V$ |
| $G$ | $T,P$ |

:::

De termodynamiske potensialene er dermed ikke forskjellige typer energi. De er ulike måter å beskrive det samme termodynamiske systemet på, tilpasset hvilke størrelser vi ønsker å bruke som uavhengige variabler.

Vi kan se på overgangen fra ett potensial til et annet som et bytte av perspektiv. Fra $U$ til $H$ bytter vi ut volum med trykk. Fra $U$ til $F$ bytter vi ut entropi med temperatur. I $G$ har vi byttet ut begge, slik at temperatur og trykk blir de naturlige variablene.

Dette forklarer også hvorfor Gibbs-energien er så nyttig i kjemi. Mange kjemiske prosesser studeres nettopp ved omtrent konstant temperatur og trykk, og $G$ er potensialet som er naturlig tilpasset disse betingelsene.


## Maxwell-relasjonene – skjulte koblinger mellom termodynamiske størrelser

De fire fundamentallikningene viser hvordan de termodynamiske potensialene endres når de naturlige variablene endres. Men de inneholder også mer informasjon enn det vi hittil har brukt.

Dette er særlig nyttig fordi ikke alle termodynamiske størrelser er like enkle å måle. Temperatur, trykk og volum kan måles direkte i laboratoriet, mens entropiendringer ofte må bestemmes indirekte.

**Maxwell-relasjonene** lar oss koble slike størrelser til hverandre. En derivert som involverer entropi kan for eksempel erstattes av en derivert som bare involverer temperatur, trykk og volum.

La oss se hvordan dette følger direkte fra de termodynamiske potensialene.

### Et eksempel fra Gibbs-energien

For Gibbs-energien fant vi

$$dG=-S\,dT+V\,dP$$

Siden de naturlige variablene er $T$ og $P$, kan vi skrive

$$G=G(T,P)$$

Fra fundamentallikningen kan vi lese av

$$-S=\left(\frac{\partial G}{\partial T}\right)_P$$

og

$$V=\left(\frac{\partial G}{\partial P}\right)_T$$

Nå kommer en viktig egenskap ved tilstandsfunksjoner inn. Gibbs-energien har én bestemt verdi for hver tilstand, uavhengig av hvordan systemet kom dit. Matematisk betyr dette blant annet at de blandede andrederiverte av $G$ er like.

Dermed får vi

$$\left(\frac{\partial V}{\partial T}\right)_P=-\left(\frac{\partial S}{\partial P}\right)_T$$

Dette er én av **Maxwell-relasjonene**. Hva forteller den egentlig oss?

Venstre side beskriver hvordan volumet endres med temperaturen ved konstant trykk. Dette kan vi måle eksperimentelt. Høyre side beskriver hvordan entropien endres med trykket ved konstant temperatur, noe som er langt vanskeligere å måle direkte.

Maxwell-relasjonen forteller at disse to responsene egentlig er to sider av den samme termodynamiske informasjonen.

:::{admonition} Hvor kommer likheten mellom de deriverte fra?
:class: note dropdown

For en generell tilstandsfunksjon $X(a,b)$ kan vi skrive

$$dX=\left(\frac{\partial X}{\partial a}\right)_b da+\left(\frac{\partial X}{\partial b}\right)_a db$$

Dersom funksjonen er tilstrekkelig glatt, spiller det ingen rolle i hvilken rekkefølge vi deriverer med hensyn på $a$ og $b$:

$$\frac{\partial}{\partial b}\left(\frac{\partial X}{\partial a}\right)_b=\frac{\partial}{\partial a}\left(\frac{\partial X}{\partial b}\right)_a$$

Dette kalles likhet mellom blandede andrederiverte.

Det er denne matematiske egenskapen som gir opphav til Maxwell-relasjonene når $X$ er et termodynamisk potensial.

Se også {doc}`blandede partiellderiverte <matematikk/flervariabel_kalkulus>`.

:::

### Fire potensialer gir fire Maxwell-relasjoner

Hvert av de fire termodynamiske potensialene gir én Maxwell-relasjon.

Fra

$$dU=T\,dS-P\,dV$$

får vi

$$\left(\frac{\partial T}{\partial V}\right)_S=-\left(\frac{\partial P}{\partial S}\right)_V$$

Fra

$$dH=T\,dS+V\,dP$$

får vi

$$\left(\frac{\partial T}{\partial P}\right)_S=\left(\frac{\partial V}{\partial S}\right)_P$$

Fra

$$dF=-S\,dT-P\,dV$$

får vi

$$\left(\frac{\partial P}{\partial T}\right)_V=\left(\frac{\partial S}{\partial V}\right)_T$$

Fra

$$dG=-S\,dT+V\,dP$$

får vi

$$\left(\frac{\partial V}{\partial T}\right)_P=-\left(\frac{\partial S}{\partial P}\right)_T$$

:::{admonition} De fire Maxwell-relasjonene
:class: important dropdown

$$\left(\frac{\partial T}{\partial V}\right)_S=-\left(\frac{\partial P}{\partial S}\right)_V$$

$$\left(\frac{\partial T}{\partial P}\right)_S=\left(\frac{\partial V}{\partial S}\right)_P$$

$$\left(\frac{\partial P}{\partial T}\right)_V=\left(\frac{\partial S}{\partial V}\right)_T$$

$$\left(\frac{\partial V}{\partial T}\right)_P=-\left(\frac{\partial S}{\partial P}\right)_T$$

:::

Det viktigste er ikke å huske de fire Maxwell-relasjonene utenat. Poenget er å forstå hvor de kommer fra og hvorfor de er nyttige.

De termodynamiske potensialene er tilstandsfunksjoner, og derfor finnes det bestemte matematiske sammenhenger mellom hvordan ulike egenskaper responderer på endringer i temperatur, trykk, volum og entropi. Maxwell-relasjonene gjør disse sammenhengene eksplisitte.

De lar oss dermed bruke størrelser som er enkle å måle, som $P$, $V$ og $T$, til å få informasjon om størrelser som er vanskeligere å måle direkte, særlig entropi.

## Kjemisk potensial

Så langt har vi i hovedsak behandlet systemer med konstant stoffmengde. I kjemi er vi imidlertid ofte interessert i prosesser der sammensetningen kan endre seg: stoff kan diffundere, blandes, gå fra én fase til en annen eller reagere kjemisk.

Da trenger vi **kjemisk potensial**.

For komponent $i$ kan kjemisk potensial ved konstant temperatur og trykk defineres som

$$\mu_i=\left(\frac{\partial G}{\partial n_i}\right)_{T,P,n_{j\ne i}}$$

Uttrykket forteller hvor mye Gibbs-energien endres når vi tilfører en liten mengde av komponent $i$, mens temperaturen, trykket og mengden av de andre komponentene holdes konstante.

En forskjell i kjemisk potensial gir en drivkraft for materieoverføring. Det kan sammenliknes med hvordan en temperaturforskjell gir en drivkraft for varmeoverføring.

![Kjemisk potensial i et bilde](bilder/diffusjon_bilde.png)

Tenk på en dråpe konditorfarge som dryppes i et glass vann. På mikronivå beveger molekylene seg tilfeldig i alle retninger. Makroskopisk ser vi likevel en bestemt utvikling: konsentrasjonsforskjellen jevnes gradvis ut.

Den generelle termodynamiske beskrivelsen er at stoff omfordeles i retning av lavere Gibbs-energi. Ved likevekt finnes det ikke lenger noen netto drivkraft for overføring, og det kjemiske potensialet til komponenten er det samme i de delene av systemet som kan utveksle stoff.

:::{admonition} Underveisoppgave
:class: tip

Konditorfarge dryppes i et glass vann og sprer seg utover uten omrøring. Hvorfor er det ikke tilstrekkelig å si at «molekylene bare beveger seg tilfeldig» dersom vi også ønsker å forklare den makroskopiske retningen på prosessen? Hvordan kan kjemisk potensial brukes i forklaringen?

:::

### Når stoffmengden også kan endres

Fundamentallikningen

$$dU=T\,dS-P\,dV$$

gjaldt for konstant stoffmengde. Når sammensetningen også kan endres, utvides den til

$$dU=T\,dS-P\,dV+\sum_i\mu_i\,dn_i$$

Dermed får kjemisk potensial samme type rolle for stoffmengden som temperatur og trykk har for entropi og volum:

$$T=\left(\frac{\partial U}{\partial S}\right)_{V,n}$$

$$-P=\left(\frac{\partial U}{\partial V}\right)_{S,n}$$

$$\mu_i=\left(\frac{\partial U}{\partial n_i}\right)_{S,V,n_{j\ne i}}$$

Hvilket potensial vi bruker, avhenger av betingelsene. Ved konstant temperatur og trykk er Gibbs-energien mest naturlig:

$$\mu_i=\left(\frac{\partial G}{\partial n_i}\right)_{T,P,n_{j\ne i}}$$

### Kobling tilbake til statistisk termodynamikk

Kjemisk potensial kan også kobles tilbake til den statistiske beskrivelsen vi brukte tidligere i boka.

I det kanoniske ensemblet bestemmes sannsynligheten for en mikrotilstand av Boltzmann-faktoren

$$e^{-E_j/(k_\mathrm{B}T)}$$

og partisjonsfunksjonen summerer bidragene fra alle mikrotilstandene:

$$Q_N=\sum_j e^{-E_j/(k_\mathrm{B}T)}$$

Fra Boltzmann-fordelingen og Gibbs-entropien kan vi vise at Helmholtz-energien er knyttet til partisjonsfunksjonen ved

$$F=-k_\mathrm{B}T\ln Q_N$$

:::{dropdown} Hvor kommer uttrykket for Helmholtz-energien fra?

Boltzmann-fordelingen er

$$p_j=\frac{e^{-E_j/(k_\mathrm{B}T)}}{Q_N}$$

Tar vi logaritmen, får vi

$$\ln p_j=-\frac{E_j}{k_\mathrm{B}T}-\ln Q_N$$

Gibbs-entropien er

$$S=-k_\mathrm{B}\sum_j p_j\ln p_j$$

Setter vi inn uttrykket for $\ln p_j$, og bruker

$$U=\sum_jp_jE_j$$

og

$$\sum_jp_j=1$$

får vi

$$S=\frac{U}{T}+k_\mathrm{B}\ln Q_N$$

Dermed

$$U-TS=-k_\mathrm{B}T\ln Q_N$$

Siden

$$F=U-TS$$

får vi

$$F=-k_\mathrm{B}T\ln Q_N$$

:::

Ved konstant temperatur og volum er det kjemiske potensialet

$$\mu=\left(\frac{\partial F}{\partial N}\right)_{T,V}$$

Dermed får vi

$$\mu=-k_\mathrm{B}T\left(\frac{\partial\ln Q_N}{\partial N}\right)_{T,V}$$

Kjemisk potensial blir viktig igjen når vi senere skal studere faser, blandinger og kjemisk likevekt.


## Oppsummering

I dette kapitlet gikk vi fra energiregnskap til **termodynamisk drivkraft**.

Ved konstant temperatur og trykk kan termodynamikkens andre lov skrives som et kriterium for Gibbs-energien:

$$\Delta G=-T\Delta S_\mathrm{univers}$$

Dermed er en prosess spontan i den angitte retningen når

$$\Delta G<0$$

Vi så deretter at den indre energien har fundamentallikningen

$$dU=T\,dS-P\,dV$$

og at temperatur og trykk kan tolkes som mål på hvordan energien responderer på endringer i entropi og volum. Denne formuleringen kobler direkte tilbake til den statistiske definisjonen av temperatur.

Ved å omforme $U$ fikk vi

$$dH=T\,dS+V\,dP$$

$$dF=-S\,dT-P\,dV$$

$$dG=-S\,dT+V\,dP$$

De termodynamiske potensialene er dermed ulike beskrivelser av det samme systemet, tilpasset ulike sett av naturlige variabler.

Maxwell-relasjonene viste at disse tilstandsfunksjonene også inneholder skjulte koblinger mellom målbare termodynamiske responser.

Til slutt utvidet vi beskrivelsen til systemer med varierende sammensetning:

$$dU=T\,dS-P\,dV+\sum_i\mu_i\,dn_i$$

Kjemisk potensial gir den generelle drivkraften for materieoverføring, og gjennom

$$F=-k_\mathrm{B}T\ln Q_N$$

så vi igjen hvordan den makroskopiske og statistiske termodynamikken beskriver det samme systemet på ulike nivåer.

I neste kapittel skal vi bruke disse verktøyene på konkrete kjemiske prosesser.


## Oppgaver

### Oppgave 1 – Hvilket potensial passer til forsøket?

I termodynamikk velger vi potensial ut fra hvilke variabler som beskriver betingelsene naturlig.

For hver situasjon under, velg det mest naturlige potensialet blant $U$, $H$, $F$ og $G$, og begrunn kort.

a) En molekylsimulering av en væske kjøres ved fast temperatur og i en rigid boks med fast volum.

b) En kjemisk reaksjon foregår i en kolbe som står i et temperaturbad og er åpen mot atmosfæren.

### Oppgave 2 – Hvor mye elektrisk arbeid kan en brenselcelle gi?

I en hydrogenbrenselcelle skjer totalreaksjonen

$$\mathrm{H_2(g)+\frac{1}{2}O_2(g)\rightarrow H_2O(l)}$$

Ved $298\ \mathrm{K}$ og $1\ \mathrm{bar}$ er omtrent

$$\Delta H=-286\ \mathrm{kJ\,mol^{-1}}$$

$$\Delta G=-237\ \mathrm{kJ\,mol^{-1}}$$

a) Hvor mye ikke-ekspansjonsarbeid kan maksimalt hentes ut per mol hydrogen i en reversibel prosess ved konstant temperatur og trykk?

b) Hvorfor er denne verdien mindre i absoluttverdi enn $|\Delta H|$?

c) Hva skjer med resten av energien i en reversibel, isoterm prosess?

d) Forklar hvorfor $\Delta G$, ikke $\Delta H$, er den naturlige størrelsen når vi spør hvor mye elektrisk arbeid en brenselcelle maksimalt kan levere.

:::{dropdown} Løsningsforslag

a) Ved konstant temperatur og trykk er den maksimale mengden ikke-ekspansjonsarbeid gitt av reduksjonen i Gibbs-energi:

$$w_\mathrm{ikke-PV,max}=-\Delta G=237\ \mathrm{kJ\,mol^{-1}}$$

b) Ikke hele entalpiendringen kan omdannes til nyttig arbeid ved konstant temperatur. Termodynamikkens andre lov krever også entropioverføring til omgivelsene.

c) Forskjellen mellom $\Delta H$ og $\Delta G$ er knyttet til $T\Delta S$ og utveksles som varme med omgivelsene under de reversible betingelsene.

d) Gibbs-energien kombinerer første og andre lov under betingelser med konstant $T$ og $P$ og gir derfor direkte grensen for ikke-ekspansjonsarbeid.

:::


### Oppgave 3 – Maxwell-relasjoner og vann som utvider seg

For Gibbs-energien fant vi

$$dG=-S\,dT+V\,dP$$

Dette gir Maxwell-relasjonen

$$\left(\frac{\partial S}{\partial P}\right)_T=-\left(\frac{\partial V}{\partial T}\right)_P$$

For de fleste væsker øker volumet når temperaturen øker ved konstant trykk.

a) Hvilket fortegn har da

$$\left(\frac{\partial V}{\partial T}\right)_P$$

b) Hva forteller Maxwell-relasjonen om fortegnet til

$$\left(\frac{\partial S}{\partial P}\right)_T$$

c) Forklar med ord hva dette betyr for entropien når trykket økes ved konstant temperatur.

d) Vann har den uvanlige egenskapen at det mellom $0$ og omtrent $4\ ^\circ\mathrm{C}$ trekker seg sammen når temperaturen øker. Hva innebærer dette for fortegnet til $(\partial S/\partial P)_T$ i dette temperaturområdet?

:::{dropdown} Løsningsforslag

a) For en vanlig væske øker volumet med temperaturen:

$$\left(\frac{\partial V}{\partial T}\right)_P>0$$

b) Maxwell-relasjonen gir da

$$\left(\frac{\partial S}{\partial P}\right)_T<0$$

c) Entropien avtar når trykket økes ved konstant temperatur. Maxwell-relasjonen lar oss altså trekke en konklusjon om entropien ut fra en målbar volumendring.

d) Mellom $0$ og omtrent $4\ ^\circ\mathrm{C}$ gjelder for vann

$$\left(\frac{\partial V}{\partial T}\right)_P<0$$

og dermed

$$\left(\frac{\partial S}{\partial P}\right)_T>0$$

Dette er et eksempel på hvordan en uvanlig makroskopisk egenskap ved et stoff også får konsekvenser for andre termodynamiske størrelser gjennom Maxwell-relasjonene.

:::

### Oppgave 4 – Hvorfor fordeler jod seg mellom to væsker?

Jod, $\mathrm{I_2}$, kan løses både i vann og i et upolart organisk løsemiddel som sykloheksan. Hvis de to væskene blandes og deretter får skille seg i to faser, vil jod finnes i begge fasene, men vanligvis med svært forskjellige konsentrasjoner.

Anta først at

$$\mu_{\mathrm{I_2,vann}}>\mu_{\mathrm{I_2,sykloheksan}}$$

a) I hvilken retning er overføring av jod termodynamisk gunstig?

b) Hva er betingelsen for likevekt mellom de to fasene?

c) Må konsentrasjonen av jod være den samme i vann og sykloheksan ved likevekt? Begrunn svaret.

d) Hvorfor er utsagnet «stoff diffunderer alltid fra høy konsentrasjon til lav konsentrasjon» mindre generelt enn en forklaring basert på kjemisk potensial?

e) Væske–væske-ekstraksjon brukes mye i organisk kjemi for å flytte et stoff fra én væskefase til en annen. Forklar med kjemisk potensial hvorfor et stoff kan foretrekke den organiske fasen selv om det allerede finnes stoff der.

:::{dropdown} Løsningsforslag

a) Jod vil netto overføres fra vannfasen til sykloheksanfasen fordi dette senker Gibbs-energien.

b) Ved likevekt må det kjemiske potensialet til jod være det samme i de to fasene:

$$\mu_{\mathrm{I_2,vann}}=\mu_{\mathrm{I_2,sykloheksan}}$$

c) Nei. Det kjemiske potensialet avhenger ikke bare av konsentrasjonen, men også av hvordan stoffet vekselvirker med omgivelsene. Derfor kan samme kjemiske potensial svare til ulike konsentrasjoner i to forskjellige løsemidler.

d) Konsentrasjonsforskjeller er én mulig årsak til forskjeller i kjemisk potensial, men ikke den eneste. Kjemisk potensial er den mer generelle termodynamiske drivkraften.

e) Hvis vekselvirkningene mellom jod og det organiske løsemiddelet gjør det kjemiske potensialet lavere der, vil overføring til den organiske fasen være termodynamisk gunstig helt til kjemisk potensial er likt i begge fasene.

:::
