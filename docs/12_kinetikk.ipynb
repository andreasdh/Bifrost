## Reaksjonshastighet

I en reaksjon forbrukes reaktanter og produkter dannes. Endringene er ikke nødvendigvis like store. I reaksjonen

$A + 2B \rightarrow 3C$

forsvinner $B$ dobbelt så fort som $A$, og $C$ dannes tre ganger så fort. For å beskrive reaksjonen med én entydig hastighet deler vi derfor konsentrasjonsendringen til stoff $J$ på det støkiometriske tallet $\nu_J$.

:::{admonition} Definisjon - Reaksjonshastighet
:class: important

**Reaksjonshastigheten** til produkt/reaktant $J$ er endringen i konsentrasjon per tid $t$, delt på stoffets støkiometriske tall $\nu_J$:

$v = \frac{1}{\nu_J}\frac{d[J]}{dt}$
:::


:::{admonition} Underveisoppgave
:class: tip

Vi ser på syntesen av ammoniakk:

$N_2 + 3H_2 \rightarrow 2NH_3$

De støkiometriske tallene $\nu_J$ er negative for reaktanter og positive for produkter, så her er $\nu_{N_2} = -1$, $\nu_{H_2} = -3$ og $\nu_{NH_3} = +2$. Reaksjonshastigheten er den samme uansett hvilket stoff vi måler den ut fra: $v = \frac{1}{\nu_J}\frac{d[J]}{dt}$

**a)** I et forsøk måler vi at nitrogenet forbrukes med $\frac{d[N_2]}{dt} = -0.10\ \text{mol/(L s)}$. Bruk definisjonen til å finne reaksjonshastigheten $v$.

**b)** Bruk $v$ fra a) til å finne hvor raskt hydrogen forbrukes, $\frac{d[H_2]}{dt}$, og hvor raskt ammoniakk dannes, $\frac{d[NH_3]}{dt}$.

**c)** Hvorfor er det nyttig å dele på det støkiometriske tallet i definisjonen? Hva ville skjedd hvis vi bare brukte $\frac{d[J]}{dt}$ direkte som "hastigheten" for hvert stoff?

:::


:::{dropdown} Løsningsforslag

**a)** Vi bruker $v = \frac{1}{\nu_J}\frac{d[J]}{dt}$ på nitrogen, med $\nu_{N_2} = -1$:

$v = \frac{1}{-1}\cdot(-0.10) = 0.10\ \text{mol/(L s)}$

**b)** Vi løser definisjonen for hvert stoff, $\frac{d[J]}{dt} = \nu_J \cdot v$:

$\frac{d[H_2]}{dt} = \nu_{H_2}\cdot v = -3 \cdot 0.10 = -0.30\ \text{mol/(L s)}$

$\frac{d[NH_3]}{dt} = \nu_{NH_3}\cdot v = +2 \cdot 0.10 = +0.20\ \text{mol/(L s)}$

Hydrogen forbrukes tre ganger så raskt som nitrogen, og ammoniakk dannes dobbelt så raskt - akkurat som forholdstallene i reaksjonslikningen.

**c)** Uten å dele på det støkiometriske tallet ville hvert stoff hatt sin egen "hastighet", og de ville vært forskjellige ($-0.10$, $-0.30$, $+0.20$) selv om det er én og samme reaksjon. Ved å dele på $\nu_J$ får vi én entydig hastighet som er lik uansett hvilket stoff vi måler. Det gjør at reaksjonshastigheten blir en veldefinert egenskap ved reaksjonen, ikke ved det enkelte stoffet.

:::


Siden $\nu_J$ er negativ for reaktanter og positiv for produkter, blir hastigheten $v$ positiv uansett hvilket stoff vi følger. Nå går vi videre til hastighetslovene.

## Hastighetslover, orden og hastighetskonstant

Hvordan hastigheten avhenger av konsentrasjonene er ikke noe vi kan lese ut av reaksjonslikningen - det må bestemmes eksperimentelt. Resultatet uttrykkes som en **hastighetslov**.

:::{admonition} Definisjon - Hastighetslov
:class: important

En **hastighetslov** knytter reaksjonshastigheten til konsentrasjonene til reaktantene, ofte på formen

$v = k_r[A]^m[B]^n$

for en reaksjon

$\alpha A + \beta B \to \text{produkter}$
:::

Her er $k_r$ **hastighetskonstanten**, som er uavhengig av konsentrasjonene, men avhenger av temperaturen. Den skal vi se nærmere på litt senere. Potensene forteller **reaksjonens orden**.


:::{admonition} Definisjon - Reaksjonsorden
:class: important

**Ordenen** med hensyn på et bestemt stoff er potensen stoffet er opphøyd i hastighetsloven. **Den totale ordenen** er summen av alle potensene.
:::

Eksponentene $m$ og $n$ bestemmes eksperimentelt og trenger ikke å være lik de støkiometriske koeffisientene $\alpha$ og $\beta$. Den totale reaksjonsordenen er $m+n$. Vi skal i hovedsak se på reaksjoner av nullte, første og andre orden.


:::{admonition} Underveisoppgave
:class: tip

For reaksjonen mellom hydrogen og jod

$H_2 + I_2 \rightarrow 2HI$

er det eksperimentelt funnet at reaksjonen er første orden med hensyn på $H_2$ og første orden med hensyn på $I_2$.

**a)** Skriv opp hastighetsloven for denne reaksjonen.

**b)** Hva er den totale reaksjonsordenen?

:::


:::{dropdown} Løsningsforslag

**a)** Første orden i $H_2$ og første orden i $I_2$ gir eksponentene $a = 1$ og $b = 1$:

$v = k_r[H_2][I_2]$

**b)** Den totale reaksjonsordenen er summen av eksponentene: $1 + 1 = 2$, altså andre orden totalt.

:::


### Eksperimentell bestemmelse av reaksjonsorden

Det er ikke alltid like lett å bestemme ordenen med hensyn på ett stoff. I **isolasjonsmetoden** lar vi alle andre reaktanter være i stort overskudd. Da endrer konsentrasjonene deres seg lite underveis, og hastighetsloven forenkles til å avhenge av stoffet vi følger.

Er for eksempel $B$ i stort overskudd, kan $[B]$ regnes som konstant og slås sammen med hastighetskonstanten:

$k_{r,\text{eff}} = k_r[B]_0^n$

Da kan hastighetsloven skrives $v=k_{r,\text{eff}}[A]^m$. Hvis $m=1$, kaller vi reaksjonen *pseudoførsteordens*. Slik kan en sammensatt hastighetslov kartlegges ett stoff av gangen.

### Utledning av hastighetslovene

En hastighetslov er en differensiallikning: den forteller hvor fort konsentrasjonen endrer seg akkurat nå. Integrerer vi den, får vi konsentrasjonen som funksjon av tiden - som er det vi faktisk måler i laben. Du finner mer om differensiallikninger på nettstedet til [IN-KJM1900](https://andreasdh.github.io/programmering-i-kjemi/docs/modellering/differensiallikninger.html).

Vi skal utlede hastighetslovene til nullteordens, førsteordens og andreordens reaksjoner. Vi ser her på den enkleste typen reaksjoner: 

$A \rightarrow \text{produkter}$ 

Husk at for reaktanter er det støkiometriske tallet negativt, her $\nu_A = -1$, og vi får derfor

$v = \frac{1}{\nu_A} \frac{d[A]}{dt} = -\frac{d[A]}{dt}$

Samtidig har vi

$v = k_r [A]^a$

der $a$ er reaksjonsordenen. Til sammen gir begge likningene oss

$-\frac{d[A]}{dt} = k_r [A]^a$

som vi løser for $a=0,1,2$. Det betyr at løsningen er konsentrasjonen til A som en funksjon av tiden $t$, altså $[A](t)$

Vi lar konsentrasjonen ved tid $t=0$ være $[A]_0$, altså $[A](0)=[A]_0$.

Det anbefales å gå grundig gjennom hastighetslovene, og prøve å løse dem selv!


#### Nullteordens reaksjoner

Vi har

$-\frac{d[A]}{dt} = k_r [A]^0$

$-\frac{d[A]}{dt} = k_r$

$d[A] = -k_r \ dt$

$\int \ d[A] = \int -k_r \ dt$

$[A](t) = -k_r t + C$



:::{danger}
Husk at vi ved å integrere ubestemte integraler må ha med en [integreringskonstant](https://andreasdh.github.io/Bifrost/docs/matematikk/kalkulus.html#ubestemte-integraler-og-integrasjonskonstanten) $C$. Her, og i de neste eksemplene, har vi tatt en snarvei og har en integreringskonstant på én side av likningen, til tross for at vi har to integraler - ett på hver side. Vi har forenklet følgende:

$\int \ d[A] = \int -k_r \ dt$

$[A](t) + C_1 = -k_r t + C_2$

$[A](t) = -k_r t + (C_2-C_1)$

$[A](t) = -k_r t + C$

der $C = C_2 - C_1$

Derfor holder de å kun ha med en konstant på én side av likningen.
:::


Fra startbetingelsen får vi

$[A](0) = [A]_0 = -k_r \cdot 0 + C$

$C = [A]_0$

Hastighetsloven for en nullteordens reaksjon blir dermed

$[A](t) = -k_r t + [A]_0$


#### Førsteordens reaksjoner

Vi har

$-\frac{d[A]}{dt} = k_r [A]^1$

$-\frac{d[A]}{dt} = k_r [A]$

$\frac{1}{[A]} \ d[A] = -k_r \ dt$

$\int \frac{1}{[A]} \ d[A] = \int -k_r \ dt$

$\ln [A](t) = -k_r t + C$

$e^{\ln [A](t)} = e^{-k_r t + C}$

$[A](t) = e^{-k_r t} \cdot e^{C}$

La $D = e^C$, og vi får

$[A](t) = De^{-k_r t}$

Fra startbetingelsen får vi

$[A](0) = [A]_0 = De^{-k_r \cdot 0}$

$D = [A]_0$

Hastighetsloven for en førsteordens reaksjon blir dermed

$[A](t) = [A]_0e^{-k_r t}$


#### Andreordens reaksjoner

Vi har

$-\frac{d[A]}{dt} = k_r [A]^2$

$\frac{1}{[A]^2} \ d[A] = -k_r \ dt$

$\int \frac{1}{[A]^2} \ d[A] = \int -k_r \ dt$

$-\frac{1}{[A](t)} = -k_r t + C$

Vi ganger med $-1$:

$\frac{1}{[A](t)} = k_r t - C$

Fra startbetingelsen får vi

$[A](0) = [A]_0 \quad\Rightarrow\quad \frac{1}{[A]_0} = k_r \cdot 0 - C$

$-C = \frac{1}{[A]_0}$

Vi setter inn $-C = \frac{1}{[A]_0}$, og hastighetsloven for en andreordens reaksjon blir dermed

$\frac{1}{[A](t)} = k_r t + \frac{1}{[A]_0}$

Merk at $[A]_0$ her står i en nevner, så loven forutsetter $[A]_0 \neq 0$ - men det er uansett alltid oppfylt, siden $[A]_0$ er startkonsentrasjonen, og en reaksjon må starte med noe reaktant til stede.



:::{admonition} Underveisoppgave
:class: tip

Prøv å integrere og utlede hver hastighetslov selv. Bruk utledningene til hjelp hvis du står fast.

:::

:::{dropdown} Løsningsforslag

For nullte orden integrerer du $-d[A]/dt=k_r$ direkte. For første orden samler du konsentrasjonen på venstre side, $d[A]/[A]=-k_r\,dt$. For andre orden bruker du $d[A]/[A]^2=-k_r\,dt$. Sett til slutt inn startbetingelsen $[A](0)=[A]_0$ for å bestemme integrasjonskonstanten. Da får du de tre uttrykkene i tabellen nedenfor.
:::



### Halveringstid

Et intuitivt mål på hvor fort en reaksjon går, er tiden det tar før halvparten av reaktanten er borte. Dette kalles **halveringstiden**.

:::{admonition} Definisjon - Halveringstid
:class: important

**Halveringstiden** $t_{1/2}$ er tiden det tar for konsentrasjonen å synke til halvparten av utgangsverdien.
:::

Vi kan bruke de integrerte hastighetslovene for å bestemme halveringstidene for de ulike ordens reaksjonene. Likningen vi skal løse er

$[A](t_{1/2}) = \frac{1}{2} [A]_0$

og vi løser for $t_{1/2}$.


#### Halveringstiden for en nullteordensreaksjon

$[A](t_{1/2}) = \frac{1}{2} [A]_0$

$\frac{1}{2} [A]_0 = -k_r t_{1/2} + [A]_0$

$k_r t_{1/2} = \frac{1}{2} [A]_0$

$t_{1/2} = \frac{[A]_0}{2 k_r}$


#### Halveringstiden for en førsteordensreaksjon

$[A](t_{1/2}) = \frac{1}{2} [A]_0$

$\frac{1}{2} [A]_0 = [A]_0e^{-k_r t_{1/2}}$

$\frac{1}{2} = e^{-k_r t_{1/2}}$

$\ln \frac{1}{2} = \ln e^{-k_r t_{1/2}}$

$-\ln 2 = -k_r t_{1/2}$

$t_{1/2} = \frac{\ln 2}{k_r}$


#### Halveringstiden for en andreordensreaksjon

$\frac{1}{[A](t)} = k_r t + \frac{1}{[A]_0}$

$[A](t_{1/2}) = \frac{1}{2} [A]_0$

$\frac{1}{[A](t_{1/2})} = \frac{2}{[A]_0}$

$\frac{2}{[A]_0} = k_r t_{1/2} + \frac{1}{[A]_0}$

$\frac{1}{[A]_0} = k_r t_{1/2}$

$t_{1/2} = \frac{1}{k_r[A]_0}$


### Oppsummering av hastighetslovene og halveringstidene i tabell


| Orden | Hastighetslov | $t_{1/2}$ |
| ----- | ------------- | ------------- |
| $0$   | $[A](t)=-k_r t + [A]_0$ | $\frac{[A]_0}{2 k_r}$ |
| $1$   | $[A](t)=[A]_0e^{-k_r t}$ | $\frac{\ln 2}{k_r}$ |
| $2$   | $\frac{1}{[A](t)}= k_r t + \frac{1}{[A]_0}$ | $\frac{1}{k_r[A]_0}$|


:::{admonition} Underveisoppgave
:class: tip

**a)** En nullteordensreaksjon har $k_r = 0.020\ \text{mol/(L s)}$ og $[A]_0 = 0.80\ \text{mol/L}$. Regn ut halveringstiden.

**b)** En førsteordensreaksjon har $k_r = 0.035\ \text{s}^{-1}$. Regn ut halveringstiden.

**c)** En andreordensreaksjon har $k_r = 0.50\ \text{L/(mol s)}$ og $[A]_0 = 0.10\ \text{mol/L}$. Regn ut halveringstiden.

**d)** Se på de tre uttrykkene: hvordan avhenger halveringstiden av startkonsentrasjonen $[A]_0$ for hver orden? Hva skjer med $t_{1/2}$ hvis du dobler $[A]_0$ i hvert tilfelle?

:::


:::{dropdown} Løsningsforslag

**a)** $t_{1/2} = \frac{[A]_0}{2k_r} = \frac{0.80}{2 \cdot 0.020} = 20\ \text{s}$

**b)** $t_{1/2} = \frac{\ln 2}{k_r} = \frac{0.693}{0.035} = 20\ \text{s}$

**c)** $t_{1/2} = \frac{1}{k_r[A]_0} = \frac{1}{0.50 \cdot 0.10} = 20\ \text{s}$

**d)** De tre ordenene oppfører seg helt ulikt:

- **Nullte orden:** $t_{1/2} \propto [A]_0$. Dobler du $[A]_0$, blir halveringstiden dobbelt så lang.
- **Første orden:** $t_{1/2}$ er uavhengig av $[A]_0$. Dobler du $[A]_0$, endres ikke halveringstiden - den er alltid $\frac{\ln 2}{k_r}$.
- **Andre orden:** $t_{1/2} \propto \frac{1}{[A]_0}$. Dobler du $[A]_0$, blir halveringstiden halvert.

Startkonsentrasjonen virker altså motsatt vei for nullte og andre orden, mens den ikke spiller noen rolle for første orden.

:::

## Numerisk løsning av fartslover

For de enkle hastighetslovene ovenfor finner vi en analytisk løsning. Reaksjonsmekanismer med flere stoffer og koblede trinn lar seg ikke alltid løse like enkelt. Da kan vi integrere differensiallikningene numerisk.

Eksemplet nedenfor løser den førsteordens fartsloven

$\dfrac{d[A]}{dt}=-k_r[A]$

med **solve_ivp** fra SciPy. Vi oppgir fartsloven, tidsintervallet og startkonsentrasjonen. Programmet beregner så $[A](t)$ i de ønskede tidspunktene. Den stiplede kurven er den analytiske løsningen og viser at metodene gir samme resultat.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

k_r = 0.15       # s^-1
A_0 = 1.0        # mol/L

def fartslov(t, A):
    return -k_r*A

t = np.linspace(0, 30, 200)
losning = solve_ivp(fartslov, [0, 30], [A_0], t_eval=t)

plt.plot(losning.t, losning.y[0], label="numerisk")
plt.plot(t, A_0*np.exp(-k_r*t), "--", label="analytisk")
plt.xlabel("Tid (s)")
plt.ylabel("[A] (mol/L)")
plt.legend()
plt.show()
```

:::

:::{dropdown} Løsningsforslag

**a)** Fartsloven står i funksjonen **fartslov**. **b)** Startkonsentrasjonen er **A_0**, og sendes inn som **[A_0]**. **c)** Tidsintervallet er **[0, 30]**. Dobler vi $k_r$, avtar konsentrasjonen raskere, og halveringstiden blir halvert fordi $t_{1/2}=\ln 2/k_r$.
:::

## Arrhenius-likningen

Tidligere nevnte vi så vidt hastighetskonstanten for en reaksjon $k_r$, og at denne avhenger av temperaturen $T$. De fleste reaksjoner går fortere når temperaturen øker. Over et begrenset temperaturintervall finner man ofte at et plott av $\ln k_r$ mot $1/T$ gir en rett linje. Dette er innholdet i Arrhenius-likningen. Den matematiske formen ligner van 't Hoffs likning, men likningene beskriver ulike størrelser.


:::{admonition} Utledning av Arrhenius-likningen
:class: note, dropdown

Arrhenius-likningen ble opprinnelig funnet fra eksperimenter. En temperaturavhengig, tilsynelatende aktiveringsenergi kan uttrykkes ved

$\frac{d\ln k_r}{dT} = \frac{E_a}{RT^2}$

Legg merke til at dette har nøyaktig samme form som van 't Hoffs likning, bare med $k_r$ i stedet for $K$ og $E_a$ i stedet for $\Delta_r H^\circ$.

**Antakelse: aktiveringsenergien er konstant.** Vi antar at $E_a$ ikke endrer seg over temperaturintervallet vi ser på. Da kan vi integrere begge sider:

$\int d\ln k_r = \int \frac{E_a}{RT^2}\,dT$

Venstre side gir $\ln k_r$. På høyre side er $E_a/R$ en konstant, og den antideriverte av $1/T^2$ er $-1/T$, så vi får

$\ln k_r = -\frac{E_a}{RT} + C$

der $C$ er en integrasjonskonstant.

**Vi gir konstanten et navn.** Kaller vi $C = \ln A$, kan vi skrive

$\ln k_r = \ln A - \frac{E_a}{RT}$

Dette er den logaritmiske formen, som viser hvorfor et plott av $\ln k_r$ mot $1/T$ blir en rett linje med stigningstall $-E_a/R$ og skjæringspunkt $\ln A$.

**Vi løser for $k_r$** ved å opphøye $e$ på begge sider:

$k_r = A\,e^{-E_a/RT}$

som er Arrhenius-likningen på sin vanlige form.

**Toparts-formen.** Vil vi i stedet sammenligne to temperaturer direkte, integrerer vi den bestemte formen fra $(T_1, k_{r,1})$ til $(T_2, k_{r,2})$:

$\int_{k_{r,1}}^{k_{r,2}} d\ln k_r = \frac{E_a}{R}\int_{T_1}^{T_2} \frac{dT}{T^2}$

Dette gir

$\ln\frac{k_{r,2}}{k_{r,1}} = -\frac{E_a}{R}\left(\frac{1}{T_2} - \frac{1}{T_1}\right)$

som er formen man bruker når man kjenner hastighetskonstanten ved én temperatur og vil finne den ved en annen.
:::


:::{admonition} Arrhenius-likningen
:class: note

For hastighetskonstanten til en reaksjon $k_r$, har vi

$k_r = A\,e^{-E_a/RT} \qquad \Longleftrightarrow \qquad \ln k_r = \ln A - \frac{E_a}{RT}$

der $A$ er den preeksponensielle faktoren, $E_a$ er aktiveringsenergien for reaksjonen, $R$ er gasskonstanten og $T$ er temperaturen.
:::


:::{admonition} Definisjon - Aktiveringsenergi og frekvensfaktor
:class: important

**Aktiveringsenergien** $E_a$ beskriver energibarrieren langs reaksjonsveien. Når Arrhenius-likningen gjelder over temperaturintervallet, finnes den fra stigningstallet $-E_a/R$ i et Arrhenius-plott.

**Den preeksponensielle faktoren** $A$ samler blant annet bidrag fra kollisjonsfrekvens, orientering og hvordan den aktuelle reaksjonsmodellen er formulert.
:::


Kjenner vi hastighetskonstanten ved én temperatur, kan vi finne den ved en annen. Skriver vi Arrhenius-likningen for to temperaturer og trekker fra hverandre, får vi

$\ln\frac{k_{r,2}}{k_{r,1}} = -\frac{E_a}{R}\left(\frac{1}{T_2} - \frac{1}{T_1}\right)$


:::{admonition} På laboratoriet
:class: tip

Når vi tar logaritmen av Arrhenius-likningen, får vi en lineær funksjon på formen $y=ax+b$, der $y = \ln k_r$, $x=1/T$, $a=-E_a/R$ og $b=\ln A$. Måler vi $k_r$ ved flere temperaturer, kan vi derfor bestemme $E_a$ fra stigningstallet.
:::


:::{card} 
**Eksempel**
^^^
En reaksjon har aktiveringsenergi $E_a = 50 \text{ kJ/mol}$. Hvor mye raskere går den ved kroppstemperatur ($37^\circ\text{C} = 310 \text{ K}$) enn ved romtemperatur ($25^\circ \text{C} = 298 \text{ K}$)?

Vi bruker toparts-formen:

$\ln\frac{k_{r,2}}{k_{r,1}} = -\frac{50000}{8{.}314}\left(\frac{1}{310} - \frac{1}{298}\right) = 0{.}78$

Opphøyer vi, får vi $k_{r,2}/k_{r,1} = e^{0{.}78} \approx 2{.}2$. En temperaturøkning på bare $12^\circ\text{C}$ mer enn dobler hastigheten - et illustrerende eksempel på hvor følsomme reaksjoner er for temperatur.
:::


:::{admonition} Underveisoppgave
:class: tip

Arrhenius-likningen $k_r = A\,e^{-E_a/RT}$ beskriver hvordan hastighetskonstanten avhenger av temperaturen, der $A$ er den preeksponensielle faktoren og $E_a$ er aktiveringsenergien. Bruk $R = 8.314\ \text{J/(mol K)}$.

**a)** Se på likningen uten å regne. Hva skjer med $k_r$ når temperaturen $T$ øker? Og hva skjer med $k_r$ når aktiveringsenergien $E_a$ er stor sammenlignet med en reaksjon med liten $E_a$?

**b)** Aktiveringsenergien er den energibarrieren molekylene må over for å reagere. Forklar med egne ord hvorfor en høyere temperatur gjør at flere molekyler kommer over barrieren (tenk på Maxwell-Boltzmann-fordelingen fra tidligere).

**c)** En reaksjon har $A = 2.0 \times 10^{11}\ \text{s}^{-1}$ og $E_a = 75\ \text{kJ/mol}$. Regn ut hastighetskonstanten $k_r$ ved $T = 298\ \text{K}$.

**d)** Regn ut $k_r$ for samme reaksjon ved $T = 308\ \text{K}$ (altså 10 grader varmere), og finn forholdet $\frac{k_r(308)}{k_r(298)}$. Stemmer det med tommelfingerregelen om at reaksjonshastigheten omtrent dobles for hver tiende grad?

**e)** For en annen reaksjon er hastighetskonstanten målt ved to temperaturer: $k_r = 1.0 \times 10^{-3}\ \text{s}^{-1}$ ved $300\ \text{K}$ og $k_r = 4.5 \times 10^{-3}\ \text{s}^{-1}$ ved $320\ \text{K}$. Bruk den logaritmiske formen $\ln k_r = \ln A - \frac{E_a}{RT}$ til å finne aktiveringsenergien $E_a$.

:::


:::{dropdown} Løsningsforslag

**a)** Når $T$ øker, blir eksponenten $-\frac{E_a}{RT}$ mindre negativ, så $e^{-E_a/RT}$ vokser - altså øker $k_r$ med temperaturen. En stor $E_a$ gir en mer negativ eksponent, så $k_r$ blir mindre: reaksjoner med høy aktiveringsenergi går langsommere enn de med lav, ved samme temperatur.

**b)** Aktiveringsenergien er en terskel: bare molekyler med nok energi til å komme over barrieren kan reagere. Maxwell-Boltzmann-fordelingen viser at ved høyere temperatur har en større andel av molekylene høy energi (fordelingen strekker seg mot høyere farter/energier). Dermed er det flere molekyler som klarer å komme over barrieren, og reaksjonen går raskere.

**c)** Vi gjør om $E_a = 75\ \text{kJ/mol} = 75000\ \text{J/mol}$ og setter inn:

$k_r = 2.0 \times 10^{11} \cdot e^{-75000/(8.314 \cdot 298)} = 1.4 \times 10^{-2}\ \text{s}^{-1}$

**d)** Ved $308\ \text{K}$:

$k_r = 2.0 \times 10^{11} \cdot e^{-75000/(8.314 \cdot 308)} = 3.8 \times 10^{-2}\ \text{s}^{-1}$

Forholdet blir $\frac{3.8 \times 10^{-2}}{1.4 \times 10^{-2}} = 2.7$. Hastigheten nesten tredobles for 10 grader her, så tommelfingerregelen om en dobling stemmer omtrent - den eksakte faktoren avhenger av hvor stor $E_a$ er.

**e)** Vi bruker den logaritmiske formen i begge punktene og trekker dem fra hverandre. Differansen fjerner $\ln A$:

$\ln k_2 - \ln k_1 = -\frac{E_a}{R}\left(\frac{1}{T_2} - \frac{1}{T_1}\right)$

Løser vi for $E_a$:

$E_a = \frac{R\,\ln(k_2/k_1)}{\frac{1}{T_1} - \frac{1}{T_2}} = \frac{8.314 \cdot \ln(4.5)}{\frac{1}{300} - \frac{1}{320}} = 60\ \text{kJ/mol}$

:::


### Aktivert kompleks og overgangstilstand

Aktiveringsenergien $E_a$ har en anskuelig tolkning. Følger vi den potensielle energien mens reaktantene møtes, forvrenges og omdannes til produkter, stiger den til et maksimum før den faller mot produktenes nivå. Toppen kalles **overgangstilstanden**, og klyngen av atomer der kalles det **aktiverte komplekset**.

![Aktiveringsenergi](bilder/energiprofil.svg)

Aktiveringsenergien er høyden på denne barrieren. Molekylmøter som ikke kommer over barrieren, fører ikke til reaksjon. Det forklarer hvorfor høyere temperatur - som gir en større andel molekyler med høy energi - øker hastigheten, og hvorfor en høy barriere gjør reaksjonen mer temperaturfølsom.

## Reaksjonsmekanismer

Hittil har vi sett på reaksjoner på formen

$A + B \to \ \text{produkter}$

De fleste reaksjoner skjer imidlertid ikke i ett sprang, men gjennom en rekke enklere trinn - en **mekanisme**. Å finne mekanismen er et av kinetikkens hovedmål, fordi den beskriver hva som skjer mellom molekylene.

Et klassisk eksempel er nedbrytningen av ozon i atmosfæren. Totalreaksjonen er

$2\,\text{O}_3 \longrightarrow 3\,\text{O}_2$

men denne skjer ikke i ett sprang. Den foregår gjennom to trinn:

$\text{Trinn 1:} \quad \text{O}_3 \rightleftharpoons \text{O}_2 + \text{O}$

$\text{Trinn 2:} \quad \text{O} + \text{O}_3 \longrightarrow 2\,\text{O}_2$

Her er oksygenatomet $\text{O}$ et **mellomprodukt** - det dannes i trinn 1 og forbrukes i trinn 2, så det står ikke i totallikningen. Legger vi sammen de to trinnene, kansellerer $\text{O}$, og vi sitter igjen med totalreaksjonen $2\,\text{O}_3 \longrightarrow 3\,\text{O}_2$.


:::{admonition} Definisjon - Mellomprodukt
:class: important

Et **reaksjonsmellomprodukt** dannes og forbrukes underveis i mekanismen, men står ikke i totallikningen.
:::


### Det hastighetsbestemmende trinnet

I en mekanisme med flere trinn går ikke alle like fort, og det tregeste trinnet fungerer som en flaskehals som hele reaksjonen må presses gjennom - dette kaller vi **det hastighetsbestemmende trinnet**.

:::{admonition} Definisjon - Hastighetsbestemmende trinn
:class: important

Det **hastighetsbestemmende trinnet** er trinnet i mekanismen som styrer totalhastigheten - typisk det tregeste, som virker som en flaskehals produktene må gjennom.
:::

Vi omtaler ofte ett trinn som hastighetsbestemmende når det begrenser strømmen gjennom mekanismen under de aktuelle betingelsene. Det er en nyttig tilnærming, men ikke en regel vi kan lese av bare ved å sammenligne hastighetskonstantene: konsentrasjonene og eventuelle forlikevekter spiller også inn.

## Steady-state-tilnærmingen

Så snart en mekanisme har mellomprodukter og flere trinn, blir de eksakte likningene fort kompliserte. **Steady-state-tilnærmingen** er en antakelse vi bruker for å forenkle likningene.


:::{admonition} Definisjon - Steady-state-tilnærmingen
:class: important

Et reaktivt mellomprodukt dannes og forbrukes så raskt at konsentrasjonen holder seg tilnærmet konstant og lav. Da kan nettodannelsen settes lik null:

$\frac{d[I]}{dt} \approx 0$
:::

Fremgangsmåten er den samme: Skriv opp nettohastigheten for hvert mellomprodukt, sett den lik null, løs for mellomproduktets konsentrasjon, og sett dette inn i uttrykket for produktdannelsen. Da sitter vi igjen med en hastighetslov uttrykt gjennom reaktantene.


### Utledning med steady-state

:::{admonition} Steady-state på et to-trinns skjema
:class: note, dropdown

Se på skjemaet $A \xrightarrow{k_a} I \xrightarrow{k_b} P$, der $I$ er et mellomprodukt.

For å finne hvor fort mengden av $I$ endrer seg, må vi tenke på at $I$ er med i to trinn samtidig - ett som lager det, og ett som bruker det opp.

I det **første trinnet** dannes $I$ fra $A$. Dette trinnet *øker* mengden $I$, med en hastighet $k_a[A]$.

I det **andre trinnet** forbrukes $I$ når det blir til $P$. Dette trinnet *minker* mengden $I$, med en hastighet $k_b[I]$.

Nettoendringen i $[I]$ er rett og slett det som kommer inn minus det som går ut - på samme måte som vannstanden i et kar bestemmes av hvor fort det renner inn i toppen minus hvor fort det renner ut i bunnen. Derfor setter vi et plusstegn foran dannelsen og et minustegn foran forbruket:

$\frac{d[I]}{dt} = \underbrace{k_a[A]}_{\text{dannes}} - \underbrace{k_b[I]}_{\text{forbrukes}}$

Steady-state-tilnærmingen setter denne lik null:

$k_a[A] - k_b[I] \approx 0 \qquad \Longrightarrow \qquad [I] \approx \frac{k_a}{k_b}[A]$

Produktet dannes med hastigheten $\frac{d[P]}{dt} = k_b[I]$. Setter vi inn uttrykket for $[I]$:

$\frac{d[P]}{dt} = k_b \cdot \frac{k_a}{k_b}[A] = k_a[A]$

Etter den korte innledende fasen reduseres mekanismen dermed til en enkel førsteordens hastighetslov. Tilnærmingen passer når $I$ holder seg lav og forbrukes omtrent like raskt som det dannes. Hvis $I$ hoper seg opp, er ikke steady-state-antakelsen god.
:::


:::{admonition} Underveisoppgave
:class: tip

Steady-state-tilnærmingen sier at $\frac{d[I]}{dt} \approx 0$ for et mellomprodukt. Det betyr *ikke* at det ikke dannes eller forbrukes noe $I$ - bare at det dannes og forbrukes like fort.

Tenk på vannkar-bildet fra utledningen: vann renner inn i toppen og ut i bunnen. Hva må gjelde for at vannstanden skal holde seg konstant, selv om det hele tiden strømmer vann gjennom karet? Og hva sier dette om mengden mellomprodukt $I$ i reaksjonen?

:::


:::{dropdown} Løsningsforslag

Vannstanden holder seg konstant når det renner inn like fort som det renner ut - ikke når strømmen stopper. Det strømmer hele tiden vann gjennom karet, men nivået står stille fordi inn- og utstrømmen balanserer.

På samme måte betyr $\frac{d[I]}{dt} \approx 0$ at $I$ dannes like fort som det forbrukes ($k_a[A] \approx k_b[I]$), så mengden mellomprodukt holder seg tilnærmet konstant. Det betyr altså ikke at $I$ er fraværende eller uvirksomt - tvert imot strømmer det stadig $A$ gjennom $I$ og videre til $P$. Mellomproduktet er bare kortlevd: det brukes opp nesten like raskt som det lages, så konsentrasjonen rekker aldri å bygge seg opp.

:::


## Enzymkinetikk

En viktig anvendelse av steady-state er enzymreaksjoner. Enzymer er biologiske katalysatorer - de framskynder reaksjoner uten selv å forbrukes, og er så effektive og spesifikke at ett enzym kan øke hastigheten mange tierpotenser.

:::{admonition} Definisjon - Katalysator
:class: important

En **katalysator** øker reaksjonshastigheten ved å tilby en reaksjonsvei med lavere aktiveringsenergi, uten selv å forbrukes netto.
:::

:::{admonition} Definisjon - Enzym, substrat og aktivt sete
:class: important

Et **enzym** binder en reaktant - kalt **substratet** - i et bestemt område kalt det **aktive setet**, og omdanner det til produkt før det frigjøres og setet er klart igjen.
:::


### Michaelis–Menten-mekanismen

Enzymreaksjoner beskrives ofte av et to-trinns skjema: enzym $\text{E}$ og substrat $\text{S}$ danner først et kompleks $\text{ES}$ i en reversibel likevekt, og komplekset gir så produkt $\text{P}$:

$\text{Trinn 1:} \quad \text{E} + \text{S} \underset{k_a'}{\overset{k_a}{\rightleftharpoons}} \text{ES}$

$\text{Trinn 2:} \quad \text{ES} \xrightarrow{k_b} \text{E} + \text{P}$

Videre bruker vi indeks $0$ for å betegne startkonsentrasjoner. Vi innfører fire antakelser:

1. **Steady-state for komplekset**. Konsentrasjonen av enzym-substrat-komplekset $[\text{ES}]$ holder seg tilnærmet konstant - det dannes og forbrukes like fort:

$\frac{d[\text{ES}]}{dt} \approx 0$

2. **Substratet er i overskudd**. Substratet er i stort overskudd over enzymet. Da forbrukes det så lite substrat i kompleksdannelsen i det første trinnet, at vi kan anta at startkonsentrasjonen av substratet $[S]_0$, er tilnærmet lik konsentrasjonen av substrat underveis i reaksjonen, $[S]$:

$[S]_0 \approx [S]$

3. **Ingen tilbakereaksjon**. Vi ser på den *initielle* hastigheten mens det ennå ikke er dannet nevneverdig produkt. Da kan vi se bort fra at produktet går tilbake til komplekset, og trinn 2 regnes som enveis.

4. **Konstant konsentrasjon av enzym**. Enzymet er enten fritt eller bundet i kompleks, og totalmengden er konstant:

$[E]_0 = [E] + [ES]$


Hva kan vi så utlede fra dette?


:::{admonition} Advarsel!
:class: danger

Utledningen er lang. Legg mest vekt på antakelsene, de sentrale stegene og resultatet.

:::


Vi starter med det to-trinns skjemaet

$\text{E} + \text{S} \underset{k_a'}{\overset{k_a}{\rightleftharpoons}} \text{ES} \xrightarrow{k_b} \text{E} + \text{P}$

Vi skriver opp hvordan komplekset endrer seg. Komplekset $\text{ES}$ dannes i det første trinnet (med hastighet $k_a[\text{E}][\text{S}]$) og forbrukes på to måter - det kan falle tilbake til E og S (hastighet $k_a'[\text{ES}]$), eller gå videre til produkt (hastighet $k_b[\text{ES}]$). Nettoendringen er det som dannes minus det som forbrukes:

$\frac{d[\text{ES}]}{dt} = k_a[\text{E}][\text{S}] - k_a'[\text{ES}] - k_b[\text{ES}]$

**Antakelse 1 - steady-state.** Komplekset dannes og forbrukes like fort, så konsentrasjonen holder seg konstant og vi setter nettoendringen lik null:

$k_a[\text{E}][\text{S}] - k_a'[\text{ES}] - k_b[\text{ES}] = 0$

**Antakelse 2 - enzymbevaring.** Enzymet er enten fritt eller bundet i kompleks, og totalmengden $[\text{E}]_0$ er konstant:

$[\text{E}]_0 = [\text{E}] + [\text{ES}] \qquad\Rightarrow\qquad [\text{E}] = [\text{E}]_0 - [\text{ES}]$

**Antakelse 3 - substratoverskudd.** Siden det er mye mer substrat enn enzym, endrer substratkonsentrasjonen seg nesten ikke av bindingen, så vi bruker startverdien: 

$[\text{S}] \approx [\text{S}]_0$.

Vi setter inn fra antakelse 2 og 3, og løser for $[\text{ES}]$:

$k_a([\text{E}]_0 - [\text{ES}])[\text{S}]_0 - (k_a' + k_b)[\text{ES}] = 0$

Vi ganger ut og samler alle ledd med $[\text{ES}]$:

$k_a[\text{E}]_0[\text{S}]_0 = k_a[\text{S}]_0[\text{ES}] + (k_a' + k_b)[\text{ES}]$

$k_a[\text{E}]_0[\text{S}]_0 = \big(k_a[\text{S}]_0 + k_a' + k_b\big)[\text{ES}]$

som gir

$[\text{ES}] = \frac{k_a[\text{E}]_0[\text{S}]_0}{k_a[\text{S}]_0 + k_a' + k_b}$

Vi finner hastigheten til produktdannelsen, Produktet dannes bare i det siste trinnet, så 

$v = k_b[\text{ES}]$

$v = \frac{k_a k_b[\text{E}]_0[\text{S}]_0}{k_a[\text{S}]_0 + k_a' + k_b}$

Vi rydder så opp i uttrykket. Deler vi teller og nevner på $k_a[\text{S}]_0$, og innfører Michaelis-konstanten $K_M = \dfrac{k_a' + k_b}{k_a}$, får vi

$v = \frac{k_b[\text{E}]_0}{1 + K_M/[\text{S}]_0}$



:::{admonition} Definisjon - Michaelis-konstanten
:class: important

**Michaelis-konstanten** samler de tre hastighetskonstantene i mekanismen:

$K_M = \frac{k_a' + k_b}{k_a}$
:::


:::{admonition} Michaelis–Menten-likningen
:class: note

Med Michaelis–Menten-antakelsene er hastigheten for produktdannelse $v$ gitt ved

$v = \frac{k_b[E]_0}{1 + K_M/[S]_0}$

Likningen fanger to grenser. Ved lav substratkonsentrasjon er hastigheten tilnærmet førsteordens i $[S]$. Ved høy substratkonsentrasjon blir enzymet *mettet* - alle aktive seter er opptatt - og hastigheten nærmer seg et maksimum, $v_{max} = k_b[E]_0$, uansett hvor mye mer substrat man tilsetter.

Alternativt, kan vi skrive

$v = \frac{v_{max}}{1 + K_M/[S]_0} = \frac{v_{max}[S]_0}{K_M + [S]_0}$
:::

Legg spesielt merke til hva Michaelis-konstanten $K_M$ betyr fysisk: setter vi $[S]_0 = K_M$ inn i likningen, blir nevneren $1 + 1 = 2$, så $v = \frac{1}{2}v_{max}$. Michaelis-konstanten er altså nettopp den substratkonsentrasjonen som gir halve maksimalhastigheten.


![Michaelis-Menten plott](bilder/michaelis_menten.svg)

:::{admonition} Underveisoppgave
:class: tip

Michaelis-Menten-likningen gir hastigheten for en enzymreaksjon:

$v = \frac{v_{max}[S]_0}{K_M + [S]_0}$

der $v_{max} = k_b[E]_0$ er maksimalhastigheten og $K_M$ er Michaelis-konstanten. For et bestemt enzym er $k_b = 100\ \text{s}^{-1}$, $[E]_0 = 1.0 \times 10^{-6}\ \text{mol/L}$ og $K_M = 2.0 \times 10^{-3}\ \text{mol/L}$.

**a)** Regn ut maksimalhastigheten $v_{max}$.

**b)** Regn ut hastigheten $v$ når substratkonsentrasjonen er $[S]_0 = 0.50 \times 10^{-3}\ \text{mol/L}$ (altså godt under $K_M$). Hvor mange prosent av $v_{max}$ er dette?

**c)** Regn ut hastigheten når $[S]_0 = 20 \times 10^{-3}\ \text{mol/L}$ (godt over $K_M$). Hvor mange prosent av $v_{max}$ er dette nå?

**d)** Se på svarene i b) og c). Forklar med egne ord hva som skjer med enzymet ved høy substratkonsentrasjon, og hvorfor hastigheten ikke fortsetter å øke uansett hvor mye substrat vi tilsetter.

**e)** Vis ved regning at hastigheten blir nøyaktig $\frac{1}{2}v_{max}$ når $[S]_0 = K_M$. Hva forteller dette oss om den fysiske betydningen av Michaelis-konstanten?

:::


:::{dropdown} Løsningsforslag

**a)** Maksimalhastigheten er $v_{max} = k_b[E]_0$:

$v_{max} = 100 \cdot 1.0 \times 10^{-6} = 1.0 \times 10^{-4}\ \text{mol/(L s)}$

**b)** Vi setter inn i likningen:

$v = \frac{1.0 \times 10^{-4} \cdot 0.50 \times 10^{-3}}{2.0 \times 10^{-3} + 0.50 \times 10^{-3}} = 2.0 \times 10^{-5}\ \text{mol/(L s)}$

Dette er $\frac{2.0 \times 10^{-5}}{1.0 \times 10^{-4}} = 20\ \%$ av $v_{max}$.

**c)** Med $[S]_0 = 20 \times 10^{-3}\ \text{mol/L}$:

$v = \frac{1.0 \times 10^{-4} \cdot 20 \times 10^{-3}}{2.0 \times 10^{-3} + 20 \times 10^{-3}} = 9.1 \times 10^{-5}\ \text{mol/(L s)}$

Dette er omtrent $91\ \%$ av $v_{max}$.

**d)** Ved lav substratkonsentrasjon (b) er mange av enzymets aktive seter ledige, så mer substrat gir raskt mer produkt - hastigheten er nesten proporsjonal med $[S]_0$. Ved høy konsentrasjon (c) er nesten alle aktive seter opptatt hele tiden; enzymet er *mettet*. Da er det ikke mangel på substrat som begrenser hastigheten, men hvor fort enzymet klarer å omdanne kompleks til produkt. Derfor flater hastigheten ut mot $v_{max}$, og mer substrat hjelper nesten ingenting - enzymet jobber allerede så fort det kan.

**e)** Vi setter $[S]_0 = K_M$ inn i likningen:

$v = \frac{v_{max} \cdot K_M}{K_M + K_M} = \frac{v_{max} \cdot K_M}{2K_M} = \frac{v_{max}}{2}$

Michaelis-konstanten er altså den substratkonsentrasjonen som gir halve maksimalhastigheten. En liten $K_M$ betyr at enzymet når halv maksimalhastighet ved lav substratkonsentrasjon. $K_M$ kan bare tolkes direkte som et mål på bindingsaffinitet under bestemte kinetiske betingelser.

:::

### Lineweaver–Burk-plottet

Michaelis–Menten-likningen gir en krummet kurve som det kan være vanskelig å lese parametrene direkte fra. Tar vi den inverse på begge sider, får vi en lineær funksjon.

:::{admonition} Lineweaver–Burk-likningen
:class: note

$\frac{1}{v} = \frac{1}{v_{max}} + \frac{K_M}{v_{max}}\cdot\frac{1}{[S]_0}$
:::

Plotter vi $1/v$ mot $1/[S]_0$, får vi en rett linje. Skjæringspunktet med $y$-aksen er $1/v_{max}$, og stigningstallet er $K_M/v_{max}$. Fra disse to leser vi av både maksimalhastigheten og Michaelis-konstanten.


:::{admonition} På laboratoriet
:class: tip

Lineweaver–Burk-plottet gjør sammenhengen lett å tolke grafisk.

Framgangsmåten er slik: mål den initielle hastigheten $v$ ved noen forskjellige substratkonsentrasjoner $[S]_0$, regn ut de inverse verdiene $1/v$ og $1/[S]_0$, og legg en rett linje gjennom punktene. Fra linjen leser du av begge de karakteristiske størrelsene:

* **Maksimalhastigheten** finnes fra skjæringspunktet med $y$-aksen, som er $1/v_{max}$ - så $v_{max}$ er den inverse av skjæringspunktet,
* **Michaelis-konstanten** finnes fra stigningstallet, som er $K_M/v_{max}$ - så $K_M$ er stigningstallet ganget med $v_{max}$.

Plottet er særlig nyttig for å forstå hvordan parameterne påvirker kurven. I praksis forsterker den inverse transformasjonen målefeil ved lave hastigheter. Derfor tilpasses Michaelis–Menten-likningen vanligvis direkte til dataene når $v_{max}$ og $K_M$ skal bestemmes.
:::

:::{card} 
**Eksempel**
^^^
Enzymet karboanhydrase katalyserer hydrering av $\text{CO}_2$. Følgende data ble målt:

| $[S]$ (mmol/dm³) | 1.25 | 2.50 | 5.00 | 20.0 |
|---|---|---|---|---|
| $v$ (mmol/dm³/s) | 0.0278 | 0.0500 | 0.0833 | 0.167 |

Vi lager et Lineweaver-Burk-plott ved å regne ut de inverse verdiene og tilpasse en rett linje.
:::


```python
import numpy as np
import matplotlib.pyplot as plt

# Eksperimentelle data
S = np.array([1.25, 2.5, 5, 20])               # mmol/dm^3
v = np.array([0.0278, 0.0500, 0.0833, 0.167])  # mmol/dm^3/s

# Lineweaver-Burk: 1/v = (1/v_max) + (K_M/v_max)*(1/S)
x = 1/S
y = 1/v
a, b = np.polyfit(x, y, 1)
v_max = 1/b
K_M = a*v_max


# Printer verdiene
print(f"v_max = {v_max:.3f} mmol/(dm^3*s)")
print(f"K_M = {K_M:.3f} mmol/dm^3")

# Linje for plott
xline = np.linspace(-0.1, 0.85, 100)
yline = a*xline + b

# Plotter den best tilpassede linjen, målepunktene og skjæringen med y-aksen
plt.plot(xline, yline, color="#1f5f8b", lw=2, label="tilpasset linje")
plt.plot(x, y, "o", color="#c0392b", ms=8, label="målepunkter")
plt.plot(0, b, "s", color="#1a7a4c", ms=9, label=f"skjæringspunkt = {b:.2f}")

# x-og y-akse
plt.axhline(0, color="#6b6252", lw=0.8)
plt.axvline(0, color="#6b6252", lw=0.8)

# Tittel til x-og y-aksen
plt.xlabel("1/[S]  (dm³/mmol)")
plt.ylabel("1/v  (dm³·s/mmol)")

# Plot-tittel, rutenett og navn på figurdelene
plt.title("Lineweaver–Burk-plott")
plt.grid(alpha=0.2)
plt.legend()

plt.show()
```

## Kinetisk og termodynamisk kontroll

Til slutt knytter vi kinetikk og likevekt sammen. Når en reaksjon kan gi to ulike produkter, kan betingelsene avgjøre hvilket som dominerer.

For to parallelle, irreversible reaksjoner fra samme reaktant er produktforholdet tidlig i reaksjonen gitt av forholdet mellom hastighetskonstantene:

$\frac{[P_2]}{[P_1]} = \frac{k_{r,2}}{k_{r,1}}$

For mer sammensatte mekanismer kan uttrykket være annerledes, men hovedideen er den samme: under **kinetisk kontroll** dominerer produktet som dannes raskest. Hvis produktene kan omdannes tilbake og systemet når likevekt, bestemmes forholdet i stedet av likevektskonstantene og de relative Gibbs-energiene. Da har vi **termodynamisk kontroll**.

:::{admonition} Definisjon - Kinetisk og termodynamisk kontroll
:class: important

Ved **kinetisk kontroll** bestemmes produktfordelingen av reaksjonshastighetene. Ved **termodynamisk kontroll** bestemmes den av likevekten mellom produktene.
:::

Om et produkt dominerer, kan altså avhenge av temperatur, reaksjonstid og om reaksjonen er reversibel.

## Oppsummering

- Reaksjonshastigheten normaliseres med de støkiometriske tallene, slik at den får samme verdi uansett hvilket stoff vi følger.
- Hastighetsloven og reaksjonsordenen bestemmes eksperimentelt.
- En hastighetslov kan løses analytisk eller numerisk.
- Arrhenius-likningen beskriver hvordan hastighetskonstanten ofte avhenger av temperaturen.
- Reaksjonsmekanismer kan inneholde mellomprodukter. Steady-state-tilnærmingen gjør det mulig å fjerne dem fra den observerte hastighetsloven.
- Michaelis–Menten-likningen beskriver metning i en enkel enzymmekanisme.
- Kinetisk kontroll handler om hvor raskt produktene dannes; termodynamisk kontroll handler om likevekt.

## Oppgaver

:::{admonition} Oppgave 1: Hastighetslov og orden
:class: tip

En reaksjon har hastighetsloven $v=k_r[A]^2[B]$. Hva er ordenen med hensyn på hvert stoff, og hva er totalordenen? Hvordan endres hastigheten hvis $[A]$ dobles mens $[B]$ holdes konstant?
:::

:::{dropdown} Løsningsforslag

Reaksjonen er andre orden i $A$, første orden i $B$ og tredje orden totalt. En dobling av $[A]$ øker hastigheten med en faktor $2^2=4$.
:::

:::{admonition} Oppgave 2: Førsteordens reaksjon
:class: tip

For en førsteordens reaksjon er $k_r=0{,}020\ \text{s}^{-1}$ og $[A]_0=0{,}50\ \text{mol/L}$. Finn halveringstiden og konsentrasjonen etter $60\ \text{s}$.
:::

:::{dropdown} Løsningsforslag

$t_{1/2}=\ln 2/k_r=34{,}7\ \text{s}$. Videre er $[A](60)=[A]_0e^{-k_rt}=0{,}50e^{-0{,}020\cdot60}=0{,}151\ \text{mol/L}$.
:::

:::{admonition} Oppgave 3: Numerisk modell
:class: tip

Du skal løse to koblede fartslover numerisk. Hvilke tre opplysninger må minst inngå i modellen før du kan starte integrasjonen?
:::

:::{dropdown} Løsningsforslag

Du trenger differensiallikningene som beskriver farten, startverdiene til konsentrasjonene og tidsintervallet. I tillegg må alle parameterverdier, for eksempel hastighetskonstantene, være kjent.
:::

:::{admonition} Oppgave 4: Kinetikk og termodynamikk
:class: tip

Et produkt dannes raskest, mens et annet er mest stabilt ved likevekt. Forklar hvilket produkt som kan dominere ved henholdsvis kort reaksjonstid og etter at likevekt er nådd.
:::

:::{dropdown} Løsningsforslag

Ved kort reaksjonstid kan det raskest dannede produktet dominere: kinetisk kontroll. Hvis reaksjonen er reversibel og får nå likevekt, kan det mest stabile produktet dominere: termodynamisk kontroll.
:::
